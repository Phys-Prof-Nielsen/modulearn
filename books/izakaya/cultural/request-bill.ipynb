# Requesting the Bill
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Overview

In this lesson, we will focus on the important skill of requesting the bill at a Japanese restaurant. Knowing how to ask for the check politely is crucial in Japanese dining etiquette, and it shows respect towards the establishment and staff.

During this lesson, you will learn the common phrases and expressions used in Japan to request the bill. You will also understand the proper timing to ask for the check, as well as how to handle the payment process in a polite and respectful manner. By the end of this lesson, you will feel confident in ending your meal smoothly and respectfully at a Japanese restaurant.

Key Lesson Concepts:

* Learn common phrases for requesting the bill in Japanese
* Understand the proper timing to ask for the check
* Practice handling the payment process politely and respectfully

## Common Phrases for Requesting the Bill

The phrase most commonly used to request the bill in Japan is:

* お会計をお願いします。
(Okaikei o onegai shimasu.)
Please bring me the bill.

Alternatively, you can use:

* お会計をください。
(Okaikei o kudasai.)
The bill, please.

If you wish to be more casual (appropriate in a less formal setting like an izakaya), you can say:

* チェックをお願いします。
(Chekku o onegai shimasu.)
Check, please.
(Note: "check" is borrowed from English and is pronounced "chekku.")

When dining with a group and you would like to pay separately, you can ask:

* 別々でお願いします。
(Betsubetsu de onegai shimasu.)
We’d like to pay separately, please.

## Proper Timing to Ask for the Check

In Japanese restaurants, the check is not typically brought to the table automatically. Instead, diners are expected to ask for it when they are ready to leave. Here are some tips on timing:

### Wait until you are completely done eating: 

It’s customary to finish your meal before requesting the bill. Servers may interpret an early request as a sign that you are rushing to leave.


### Make eye contact or politely call for the server: 

If the server isn’t nearby, you can say:

* すみません。 (Sumimasen.) – Excuse me.

Follow this with your request:

* お会計をお願いします。
(Okaikei o onegai shimasu.)

## Handling the Payment Process

The payment process in Japan typically involves going to the cashier instead of paying at the table. Here's how to handle this politely:

### Steps for Payment:

1. Wait for the check to be brought to your table.
The check is often presented in a small tray or folder.

2. Take the check to the cashier.
Look for the cashier station near the entrance or exit.

3. Present the check and payment politely:
When handing over cash or a card, place it on the tray provided rather than handing it directly to the staff.
You can say:

    * これでお願いします。
(Kore de onegai shimasu.) – I’d like to pay with this, please.

4. Accept your change or receipt respectfully:
When receiving change or a receipt, bow slightly as a gesture of gratitude and say:

    * ありがとうございます。
(Arigatou gozaimasu.) – Thank you.


### Key Cultural Notes:

* No tipping: Tipping is not customary in Japan. The bill reflects the full cost of the meal, and excellent service is considered standard.

* Be discreet with money: Avoid counting change

## Practice Dialogue

Here’s a sample interaction to practice requesting the bill and paying:

Customer:
すみません、お会計をお願いします。
(Sumimasen, okaikei o onegai shimasu.)
Excuse me, the bill, please.

Server:
かしこまりました。少々お待ちください。
(Kashikomarimashita. Shōshō omachi kudasai.)
Certainly. Please wait a moment.

[The server brings the bill to the table.]

Customer:
ありがとうございます。
(Arigatou gozaimasu.)
Thank you.

[At the cashier station:]

Cashier:
お会計は３,０００円になります。
(Okaikei wa sanzen en ni narimasu.)
Your total is 3,000 yen.

Customer:
これでお願いします。
(Kore de onegai shimasu.)
I’d like to pay with this, please. (hands over cash or card)

Cashier:
おつりとレシートでございます。ありがとうございました。
(Otsuri to reshiito de gozaimasu. Arigatou gozaimashita.)
Here is your change and receipt. Thank you very much.

Customer:
ありがとうございます。
(Arigatou gozaimasu.)
Thank you.
