# Tips for Navigating a Japanese Menu
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Overview

When dining at a Japanese izakaya, understanding the menu is key to making the most of your experience. Izakaya menus are often packed with options, from small sharing plates to set menus and drink specials.

Firstly, familiarize yourself with common Japanese dishes and their ingredients. This will help you understand what to expect when browsing through the menu. Additionally, don't be afraid to ask your server for recommendations or explanations of specific dishes. Japanese cuisine is known for its unique flavors and ingredients, so don't hesitate to try something new.

Lastly, pay attention to the format of the menu. Japanese menus often feature a variety of sections such as appetizers, sushi, sashimi, and main dishes. Take your time to explore each section and choose a well-rounded meal that includes a mix of flavors and textures. Don't forget to indulge in some traditional Japanese desserts to complete your dining experience.

In this lesson, you’ll learn how to read and navigate a typical izakaya menu, understand common menu sections, and recognize terms like “all-you-can-eat” and “all-you-can-drink.”

Key Lesson Concepts:

* Familiarize yourself with common Japanese dishes and their ingredients
* Ask your server for recommendations or explanations of specific dishes
* Pay attention to the format of the menu and choose a well-rounded meal

## Familiarizing Yourself with Common Dishes

Understanding the basics of Japanese cuisine can make ordering easier. Here are some popular dish categories and their typical ingredients:

* 寿司 (すし / Sushi):
Vinegared rice paired with raw fish, seafood, or vegetables. Popular varieties include 握り (にぎり / Nigiri) and 巻き寿司 (まきずし / Makizushi) rolls.

    * Example: サーモン握り (さーもん にぎり / Sāmon nigiri) – Salmon sushi.

* 刺身 (さしみ / Sashimi):
Thinly sliced raw fish or seafood served with soy sauce and wasabi.

    * Example: まぐろ刺身 (まぐろ さしみ / Maguro sashimi) – Tuna sashimi.

* 天ぷら (てんぷら / Tempura):
Lightly battered and deep-fried vegetables or seafood.
   * Example: えび天ぷら (えび てんぷら / Ebi tenpura) – Shrimp tempura.

* 焼き物 (やきもの / Yakimono):
Grilled or broiled dishes, such as fish or skewers.

    * Example: 焼き鳥 (やきとり / Yakitori) – Grilled chicken skewers.

* 麺類 (めんるい / Menrui):
Noodle dishes such as ラーメン (Rāmen), そば (Soba), and うどん (Udon).

## Asking Your Server for Recommendations

If you’re unsure about a dish, don’t hesitate to ask the server for advice or explanations. Use these phrases to communicate:

* おすすめは何ですか？
(Osusume wa nan desu ka?)
What do you recommend?

* この料理は何が入っていますか？
(Kono ryōri wa nani ga haitte imasu ka?)
What ingredients are in this dish?

* 辛いですか？
(Karaidesu ka?)
Is it spicy?

* 初めてなので、教えてください。
(Hajimete nanode, oshiete kudasai.)
It’s my first time, so please explain.

Servers in Japan are often knowledgeable and eager to help, so don’t be shy about asking questions.

## Understanding Menu Format and Choosing a Well-Rounded Meal

Japanese menus are often divided into sections, making it easier to build a balanced meal. Here’s a typical structure:

### 前菜 (ぜんさい / Zensai): Appetizers
These are small dishes to start your meal.

##### Examples
* 枝豆 (えだまめ / Edamame) – Steamed and salted soybeans.
* 冷奴 (ひややっこ / Hiyayakko) – Chilled tofu with toppings like green onions and soy sauce.
* 酢の物 (すのもの / Sunomono) (vinegared dishes).

### 主菜 (しゅさい / Shusai): Main Dishes
Choose a centerpiece for your meal, such as sushi, sashimi, or grilled items.

* 焼き物 (やきもの / Yakimono): Grilled Dishes
Grilled skewers or items cooked over charcoal.

    Examples
    * 焼き鳥 (やきとり / Yakitori) – Grilled chicken skewers.
    * 魚の塩焼き (さかな の しおやき / Sakana no shioyaki) – Grilled fish with salt.

* 揚げ物 (あげもの / Agemono): Fried Dishes
Deep-fried foods, often crispy and flavorful.

    Examples
    * 唐揚げ (からあげ / Karaage) – Japanese-style fried chicken.
    * 天ぷら (てんぷら / Tempura) – Lightly battered and fried vegetables or seafood.

* 刺身 (さしみ / Sashimi): Raw Slices of Fish
Thinly sliced raw seafood served with soy sauce and wasabi.

    Example
    * まぐろ刺身 (まぐろ さしみ / Maguro sashimi) – Tuna sashimi.


### 副菜 (ふくさい / Fukusai): Side Dishes
Add balance with side dishes like 

   * 味噌汁 (みそしる / Misoshiru) (miso soup).  
   * 漬物 (つけもの / Tsukemono) (pickled vegetables).
    
   * サラダ (さらだ / Sarada): Salads
       Example: 
       * 海藻サラダ (かいそう さらだ / Kaisō sarada) – Seaweed salad.
        
### 飲み物 (のみもの / Nomimono): Drinks
Drink menus often include a wide range of alcohol and non-alcoholic options.

   Example: 
   * 生ビール (なま びーる / Nama bīru) – Draft beer.
   * ハイボール (はいぼーる / Haibōru) – Highball (whisky with soda).
   * 緑茶 (りょくちゃ / Ryokucha) (green tea).
   * 日本酒 (にほんしゅ / Nihonshu) (sake).

### デザート (でざーと / Dezāto): Desserts
End with something sweet, like 

   * 抹茶アイス (まっちゃ あいす / Matcha aisu) (green tea ice cream).

## Set Menus and Specials

### セットメニュー (せっとめにゅー / Setto menyū): Set Menus
A set menu usually includes a combination of dishes at a fixed price.

   Example: 
   * おつまみセット (おつまみ せっと / Otsumami setto) – A set of small appetizers.
   * 定食 (ていしょく / Teishoku) – A meal set with a main dish, rice, soup, and pickles.

### 食べ放題 (たべほうだい / Tabehōdai): All-You-Can-Eat
This option allows unlimited food for a set time and price.

   Example: 
   * 焼肉食べ放題 (やきにく たべほうだい / Yakiniku tabehōdai) – All-you-can-eat grilled meat.

### 飲み放題 (のみほうだい / Nomihōdai): All-You-Can-Drink
This is a popular option at izakayas, offering unlimited drinks for a set period.

   Example: 
   * ２時間飲み放題 (にじかん のみほうだい / Nijikan nomihōdai) – Two hours of all-you-can-drink.

## How to Recognize and Order from the Menu

### Key Terms to Know:

* 本日おすすめ (ほんじつ おすすめ / Honjitsu osusume): Today’s Specials
Check this section for the freshest and recommended dishes.
* 数量限定 (すうりょう げんてい / Sūryō gentei): Limited Quantity
These items may sell out quickly.

### Phrases for Ordering:
* これをお願いします。 (Kore o onegai shimasu.) – I’ll have this, please.
* 飲み放題はありますか？ (Nomihōdai wa arimasu ka?) – Do you have all-you-can-drink?
* 辛いものはありますか？ (Karaimono wa arimasu ka?) – Do you have spicy dishes?

## Practice Dialogue

### Scene 1

Customer:
おすすめは何ですか？
(Osusume wa nan desu ka?)
What do you recommend?

Server:
焼き魚定食がおすすめです。
(Yaki zakana teishoku ga osusume desu.)
I recommend the grilled fish set meal.

Customer:
その料理は何が入っていますか？
(Sono ryōri wa nani ga haitte imasu ka?)
What ingredients are in that dish?

Server:
魚、野菜、味噌汁、ご飯です。
(Sakana, yasai, misoshiru, gohan desu.)
Fish, vegetables, miso soup, and rice.

Customer:
それをお願いします。
(Sore o onegai shimasu.)
I’ll have that, please.

### Scene 2

Customer:
飲み放題はありますか？
(Nomihōdai wa arimasu ka?)
Do you have all-you-can-drink?

Server:
はい、２時間飲み放題があります。
(Hai, nijikan nomihōdai ga arimasu.)
Yes, we have a two-hour all-you-can-drink option.

Customer:
おすすめは何ですか？
(Osusume wa nan desu ka?)
What do you recommend?

Server:
唐揚げと焼き鳥がおすすめです。
(Karaage to yakitori ga osusume desu.)
The fried chicken and grilled chicken skewers are recommended.

Customer:
それと生ビールをお願いします。
(Sore to nama bīru o onegai shimasu.)
I’ll have those and a draft beer, please.

## Tips for Success

### Practice Reading Menus: 
Learn katakana and hiragana to recognize dish names. Many menus include photos to help you decide.

### Learn Ingredient Names: 
Familiarize yourself with common food vocabulary to recognize items on the menu.
Example: 鶏肉 (とりにく / Toriniku) – Chicken, 海老 (えび / Ebi) – Shrimp.

### Be Adventurous: 
Don’t be afraid to try something new, even if you don’t know it well. Japanese cuisine offers a variety of unique and delicious dishes.

### Politeness is Key: 
Always say ありがとうございます (Arigatou gozaimasu) (Thank you) to show your appreciation.
