```python
import tkinter as tk
import random

# JLPT N5 Nouns Dictionary (Japanese: English)
jlpt_n5_nouns = {
    '水': 'water', '食べ物': 'food', '本': 'book', '家': 'house',
    '猫': 'cat', '犬': 'dog', '学校': 'school', '先生': 'teacher',
    '時間': 'time', '車': 'car', '友達': 'friend', '電車': 'train',
    '子供': 'child', '山': 'mountain', '川': 'river', '空': 'sky',
    '電話': 'telephone', '日本': 'Japan', 'お金': 'money', '人': 'person',
    '道': 'road', '町': 'town', '木': 'tree', '雨': 'rain',
    '魚': 'fish', '月': 'moon', '花': 'flower', '部屋': 'room'
}

# Initialize tkinter window
root = tk.Tk()
root.title("JLPT N5 Quiz")
root.geometry("400x500")

# Function to generate a random quiz question
def generate_question():
    # Clear the previous question and options
    for widget in root.winfo_children():
        widget.destroy()

    # Select a random noun and its English meaning
    japanese_noun, correct_answer = random.choice(list(jlpt_n5_nouns.items()))
    
    # Create a canvas for the colored box
    canvas = tk.Canvas(root, width=400, height=150)
    canvas.pack()
    # Create a colored rectangle
    canvas.create_rectangle(50, 25, 350, 125, fill="lightblue", outline="black")
    # Add the large Japanese noun text inside the rectangle
    canvas.create_text(200, 75, text=japanese_noun, font=("Arial", 48), fill="black")

    # Generate answer options
    options = [correct_answer]
    while len(options) < 4:
        random_answer = random.choice(list(jlpt_n5_nouns.values()))
        if random_answer not in options:
            options.append(random_answer)
    random.shuffle(options)
    
    # Function to check the selected answer
    def check_answer(selected_option):
        if selected_option == correct_answer:
            result_label.config(text="Correct!", fg="green")
        else:
            result_label.config(text=f"Wrong! The correct answer is '{correct_answer}'", fg="red")
    
    # Display the options as buttons
    for i, option in enumerate(options):
        button = tk.Button(root, text=option, command=lambda opt=option: check_answer(opt), font=("Arial", 14))
        button.pack(pady=5)

    # Label for displaying whether the answer was correct or wrong
    global result_label
    result_label = tk.Label(root, text="", font=("Arial", 16))
    result_label.pack(pady=10)
    
    # Button to load the next question
    next_button = tk.Button(root, text="Next Question", command=generate_question, font=("Arial", 14))
    next_button.pack(pady=10)

# Start the quiz with the first question
generate_question()

# Run the tkinter main loop
root.mainloop()
```
