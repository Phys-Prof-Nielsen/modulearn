## Multiple Choice Quiz

```python
import json
import random

# Load quiz from JSON file
def load_quiz(file_name="greetings-quiz-mc.json"):
    with open(file_name, "r") as file:
        return json.load(file)["quiz"]

# Randomize options for a multiple-choice question
def randomize_options(options):
    randomized = options.copy()
    random.shuffle(randomized)
    return randomized

# Display a single question and collect user's answer
def ask_question(question):
    print(question["question"])
    options = randomize_options(question["options"])
    for idx, option in enumerate(options):
        print(f"{idx + 1}. {option}")
    while True:
        try:
            choice = int(input("Enter the number of your choice: ")) - 1
            if 0 <= choice < len(options):
                return options[choice]
        except ValueError:
            print("Invalid input. Please enter a number between 1 and 4.")

# Evaluate the quiz results
def evaluate_quiz(quiz, user_answers):
    score = 0
    print("\nQuiz Results:")
    for idx, question in enumerate(quiz):
        correct_answer = question["correct_answer"]
        user_answer = user_answers[idx]
        if user_answer.lower() == correct_answer.lower():
            print(f"Question {idx + 1}: Correct!")
            score += 1
        else:
            print(f"Question {idx + 1}: Incorrect. The correct answer was '{correct_answer}'.")
    return score

# Main function to run the quiz
def main():
    quiz = load_quiz()
    user_answers = []
    for question in quiz:
        user_answer = ask_question(question)
        user_answers.append(user_answer)
    total_score = evaluate_quiz(quiz, user_answers)
    print(f"\nYour total score: {total_score}/{len(quiz)}")

# Run the main function
if __name__ == "__main__":
    main()
```

## Type-out Quiz

For this quiz, your keyboard must be able to type hiragana and katakana. The quiz is NOT using kanji characters.

```python
import json

# Load quiz from JSON file
def load_quiz(file_name="greetings-quiz-write.json"):
    with open(file_name, "r") as file:
        return json.load(file)["quiz"]

# Display quiz and collect answers
def take_quiz(quiz):
    user_answers = {}
    print("Welcome to the Quiz!\n")
    for idx, question in enumerate(quiz):
        print(f"Q{idx + 1}: {question['question']}")
        user_answers[idx] = input("Your answer: ").strip()
    return user_answers

# Evaluate answers
def evaluate_quiz(quiz, user_answers):
    score = 0
    print("\nResults:\n")
    for idx, question in enumerate(quiz):
        correct_answer = question["correct_answer"].strip().lower()
        user_answer = user_answers.get(idx, "").strip().lower()
        if user_answer == correct_answer:
            print(f"Q{idx + 1}: Correct!")
            score += 1
        else:
            print(f"Q{idx + 1}: Incorrect. The correct answer was '{question['correct_answer']}'.")
    return score

# Main function to run the quiz
def run_quiz():
    quiz = load_quiz()
    user_answers = take_quiz(quiz)
    score = evaluate_quiz(quiz, user_answers)
    print(f"\nYour total score: {score}/{len(quiz)}")

# Run the quiz
run_quiz()
```
