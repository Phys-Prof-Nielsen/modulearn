# Casual vs. Formal Speech
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Overview

In Japanese culture, the way you speak to others can vary greatly depending on the social context and your relationship with the person you are speaking to. One of the key distinctions in Japanese language is the difference between casual and formal speech. Understanding when to use each form is essential when communicating in a restaurant setting.

Casual speech, known as "tamego," is used when speaking with friends, family, or anyone you have a close relationship with. It is more relaxed and informal, allowing for a sense of familiarity and closeness. On the other hand, formal speech, known as "keigo," is used when speaking to elders, customers, or individuals of higher rank. It shows respect and indicates a polite and professional demeanor.

Key Lesson Concepts:

* Casual speech (tamego) is used with friends and family.
* Formal speech (keigo) is used with elders and customers.
* Knowing when to use each form is important for effective communication in a restaurant setting.

## Tamego Overview

In Japanese, casual speech, often referred to as tamego (タメ語), is the informal register used in relaxed and personal settings. It is a cornerstone of communication among friends, peers, and family members, where maintaining a sense of closeness and equality is more important than formality or deference. Tamego contrasts sharply with polite and honorific forms, as it simplifies sentence structures and omits the layers of respect embedded in more formal speech.

### Characteristics of Tamego

Tamego is characterized by its straightforward and abbreviated structure, often omitting formal verb endings and particles. For example, instead of saying 食べます (tabemasu, polite form of "to eat"), one would simply say 食べる (taberu). Similarly, nouns and adjectives in tamego drop the です (desu) copula, so かわいいです (kawaii desu, "it's cute") becomes かわいい (kawaii). This form also frequently omits pronouns and particles altogether when context makes the meaning clear, as in 行く？ (Iku?, "Are you going?") instead of the more polite 行きますか？ (Ikimasu ka?).

Tamego lends itself to brevity and a sense of intimacy. By stripping away formalities, speakers can connect on a personal level without the barriers that politeness can sometimes create. It reflects the speaker's comfort and trust in their relationship with the listener.

### When to Use Tamego

Tamego is appropriate in specific contexts where there is mutual understanding and consent to speak casually. These situations include conversations among close friends, siblings, classmates, or coworkers of equal standing. It is also common in online communication and text messaging, where informality often prevails. However, it is essential to recognize the social dynamics at play; using tamego with someone of higher status or a stranger can come across as disrespectful or presumptuous.

One of the key social cues in transitioning to tamego is when the relationship becomes more familiar, often signaled by one party explicitly inviting the other to speak casually. For instance, someone might say, タメ語でいいよ (Tamego de ii yo, "It's okay to speak casually"). Until such permission is given or a clear rapport is established, defaulting to polite speech is generally safer.

### Advantages of Tamego

Tamego fosters a sense of equality and camaraderie, making it ideal for building stronger personal relationships. It eliminates the rigid hierarchies embedded in more formal speech, creating a conversational tone that feels natural and unguarded. This casualness can make communication more engaging and relaxed, allowing for freer expression of thoughts and emotions.

For learners of Japanese, tamego is often easier to grasp than keigo because of its simplicity. The conjugation rules are straightforward, and there is less vocabulary to memorize compared to the extensive honorific and humble forms. However, understanding when and where to use tamego requires cultural awareness and sensitivity.

### Pitfalls of Misusing Tamego

While tamego can strengthen relationships, using it inappropriately can lead to misunderstandings or even offense. Speaking casually to superiors, elders, or strangers without establishing the appropriate context risks being seen as rude or disrespectful. For example, addressing a teacher with 行く？ (Iku?) instead of 行きますか？ (Ikimasu ka?) would be considered a breach of social etiquette. In professional settings, sticking to polite forms is the norm to avoid undermining credibility or rapport.

In conclusion, tamego is a vital component of Japanese communication that reflects intimacy and equality in relationships. Mastering its usage is not only about linguistic fluency but also about understanding the social nuances that dictate when and with whom it is appropriate. Balancing tamego with more formal speech is a skill that enhances both personal and professional interactions in Japanese.

## Keigo Overview

Keigo (敬語), the honorific language in Japanese, is a highly structured system designed to elevate the listener or subject of the conversation. It is subdivided into three main categories: sonkeigo (尊敬語, respectful language), kenjōgo (謙譲語, humble language), and teineigo (丁寧語, polite language). Each serves a distinct purpose and conveys different nuances of respect.

### Sonkeigo (尊敬語): Respectful Language

Sonkeigo is used to elevate the status of the person being spoken about, demonstrating respect for their actions, status, or position. This form often replaces standard verbs with their honorific equivalents. For example, instead of saying 食べる (taberu, to eat), one would use 召し上がる (meshiagaru) when referring to someone else's eating. Similarly, いる (iru, to be) becomes いらっしゃる (irassharu). Sonkeigo is particularly common when addressing superiors, customers, or people in positions of authority. It highlights the speaker's deference and acknowledgment of the other person's elevated role in the social hierarchy.

### Kenjōgo (謙譲語): Humble Language

Kenjōgo, in contrast to sonkeigo, is used to humble the speaker or their actions, effectively elevating the listener. This form is often employed when the speaker is discussing their own actions or contributions in a way that minimizes their importance. For instance, 行く (iku, to go) becomes 参る (mairu) in kenjōgo, and 言う (iu, to say) becomes 申し上げる (mōshiageru). Kenjōgo is essential in professional environments, such as during business meetings or customer interactions, where maintaining humility reflects well on the speaker's respect and professionalism.

### Teineigo (丁寧語): Polite Language

Teineigo is the most commonly used form of keigo and serves as a general marker of politeness. Unlike sonkeigo and kenjōgo, which are situational, teineigo simply makes the speech more formal by adding -ます (-masu) or -です (-desu) endings to verbs and sentences. For instance, 食べる (taberu, to eat) becomes 食べます (tabemasu), and かわいい (kawaii, cute) becomes かわいいです (kawaii desu). Teineigo is versatile and used in a variety of contexts, from interactions with strangers to service situations, providing a respectful yet neutral tone.

### Practical Applications of Keigo

Understanding and using keigo appropriately is a critical skill in Japanese society. For example, when speaking to a boss about their actions, one would use sonkeigo to elevate them, such as 社長は会議にいらっしゃいます (Shachō wa kaigi ni irasshaimasu, "The president is attending the meeting"). When describing their own actions, kenjōgo would be used: 会議に参ります (Kaigi ni mairimasu, "I will attend the meeting"). For everyday polite conversation with a customer, teineigo suffices: こちらを確認します (Kochira o kakunin shimasu, "I will check this for you").

Mastering the distinctions between these forms requires practice and cultural sensitivity, as they are not interchangeable and depend on the speaker's relationship to the listener and the social setting. Misuse can lead to awkwardness or offense, making keigo an essential, though challenging, aspect of fluency in Japanese.

## Examples Often Heard at Izakaya

### Irasshaimase (いらっしゃいませ) 

This is an quintessential example of sonkeigo (尊敬語), the respectful form of keigo, used to elevate and honor the customer in a business setting. It is derived from the verb irassharu (いらっしゃる), which is a respectful way of saying "to come," "to go," or "to be." The addition of -mase (-ませ), a polite imperative ending, transforms the phrase into a warm and formal invitation, roughly translating to "Welcome" or "Please come in." This expression underscores the cultural emphasis on hospitality in Japan, known as omotenashi (おもてなし), where the customer is treated with the utmost respect and care. By using irasshaimase, shopkeepers, restaurant staff, and service providers not only greet patrons but also subtly elevate their status, creating an atmosphere of deference and professionalism. Its use is a staple in Japanese customer service, reflecting the deep-rooted societal value of prioritizing and respecting others.

### Itadakimasu (いただきます) 

This is a culturally rich expression in Japanese, rooted in kenjōgo (謙譲語), the humble form of keigo. Derived from the verb itadaku (いただく), which means "to humbly receive," it reflects the speaker's humility and gratitude when partaking in food or drink. The phrase is traditionally said before meals to express appreciation not only to the person who prepared the meal but also to everyone and everything involved in bringing the food to the table, including farmers, animals, and the natural world. As a form of kenjōgo, itadakimasu acknowledges the speaker's lower status in the act of receiving, symbolizing respect and thankfulness. Its use highlights Japan's deep-seated cultural values of mindfulness, respect for life, and interconnectedness, making it a central ritual in everyday life.

### Gochisōsama (ごちそうさま)

This is often said as gochisōsama deshita (ごちそうさまでした) after a meal, is a phrase deeply rooted in Japanese culture and reflects elements of kenjōgo (謙譲語), or humble language. The term gochisō (ご馳走) originally referred to the act of "running around" to gather ingredients or prepare a meal, emphasizing the effort and care that goes into providing food. By adding sama (さま), a respectful suffix, the phrase elevates the acknowledgment of this effort. When said after a meal, it expresses humility and gratitude not only toward the person who prepared or served the food but also toward the ingredients and the process involved. The phrase encapsulates the cultural values of appreciation, respect, and recognition of others' labor, embodying the spirit of interconnectedness and mindfulness central to Japanese etiquette.
