# Making Modifications to Orders

## Overview

In this lesson, we will explore how to make modifications to orders in a Japanese restaurant. When dining out, it is common to have specific dietary preferences or restrictions, and knowing how to make modifications to your order is essential to ensure a pleasant dining experience.

First, we will discuss how to politely request modifications to your dish. Whether you are looking to customize your order by adding or removing ingredients, it is important to communicate your preferences clearly and respectfully to the server. ## How to Politely Request Modifications

In Japanese, politeness is crucial when interacting with restaurant staff. To modify an order, you should first make your request clear, and then politely ask if the change is possible. Here are some useful phrases for requesting modifications:

### General Modifications

1. これを抜いてもらえますか？
(Kore o nuite moraemasu ka?)
Can you remove this?

これ (kore) = this (referring to the ingredient or item)
抜いて (nuite) = remove (the te-form of the verb nuku, meaning to take out or remove)
もらえますか？ (moraemasu ka?) = can you (politely asking for help)

2. これを追加できますか？
(Kore o tsuika dekimasu ka?)
Can you add this?

追加 (tsuika) = addition
できますか？ (dekimasu ka?) = can you do this?

3. この料理をベジタリアンにできますか？
(Kono ryōri o bejitarian ni dekimasu ka?)
Can you make this dish vegetarian?

ベジタリアン (bejitarian) = vegetarian
にできますか？ (ni dekimasu ka?) = can you make it?

4. この料理をグルテンフリーにできますか？
(Kono ryōri o gurutenfurī ni dekimasu ka?)
Can you make this dish gluten-free?

グルテンフリー (gurutenfurī) = gluten-free

5. 辛くしないでください。
(Karaku shinaide kudasai.)
Please make it not spicy.

辛く (karaku) = spicy
しないでください (shinaide kudasai) = please do not make it (polite request)

6. もっと少なくできますか？
(Motto sukunaku dekimasu ka?)
Can you make it less (amount)?

もっと (motto) = more
少なく (sukunaku) = less

### Modifying for Dietary Restrictions

1. 私はビーガンです。これをビーガンにできますか？
(Watashi wa bīgan desu. Kore o bīgan ni dekimasu ka?)
I am vegan. Can you make this vegan?

2. アレルギーがあります。これにアレルギーのあるものを抜いてもらえますか？
(Arerugī ga arimasu. Kore ni arerugī no aru mono o nuite moraemasu ka?)
I have an allergy. Can you remove the ingredient I am allergic to?

3. グルテンフリーのものをお願いできますか？
(Gurutenfurī no mono o onegaidekimasu ka?)
Can you offer a gluten-free option?
お願いできますか？ (onegaidekimasu ka?) = can you provide/offer?

## Common Modifications

Knowing how to communicate common dietary modifications will help you confidently navigate restaurant menus. Here are some common requests you might make:

### Vegan/Vegetarian Modifications

ベジタリアン (bejitarian) = vegetarian

ビーガン (bīgan) = vegan

肉を抜いてください (Niku o nuite kudasai) = Please remove the meat

卵を抜いてください (Tamago o nuite kudasai) = Please remove the eggs

乳製品を抜いてください (Nyūseihin o nuite kudasai) = Please remove dairy products

### Gluten-Free Modifications

グルテンフリー (gurutenfurī) = gluten-free

グルテンフリーの料理はありますか？
(Gurutenfurī no ryōri wa arimasu ka?)
Do you have any gluten-free dishes?

これに小麦は入っていますか？
(Kore ni komugi wa haitteimasu ka?)
Does this contain wheat?
(Note: Wheat is the most common gluten source in Japan.)

## Practice Dialogue

### Scene 1

Customer:
この料理をベジタリアンにできますか？
(Kono ryōri o bejitarian ni dekimasu ka?)
Can you make this dish vegetarian?

Waiter:
はい、できます。肉を抜いてもいいですか？
(Hai, dekimasu. Niku o nuite mo ii desu ka?)
Yes, we can. Is it okay to remove the meat?

Customer:
はい、肉を抜いてください。
(Hai, niku o nuite kudasai.)
Yes, please remove the meat.

### Scene 2

Customer:
私はグルテンフリーです。これをグルテンフリーにできますか？
(Watashi wa gurutenfurī desu. Kore o gurutenfurī ni dekimasu ka?)
I am gluten-free. Can you make this gluten-free?

Waiter:
はい、グルテンフリーのオプションがあります。
(Hai, gurutenfurī no opushon ga arimasu.)
Yes, we have a gluten-free option.

## Cultural Tips for Modifying Orders in Japan

* Politeness is Key: Always make sure to use polite forms when requesting modifications. Even if you have specific dietary restrictions, it’s important to be respectful when asking for changes.

* Be Specific: When explaining allergies or dietary preferences, be clear and specific. For example, if you cannot eat dairy, say 乳製品を食べられません (nyūseihin o taberaremasen), meaning "I cannot eat dairy products."

* Dietary Language: Japanese restaurants may not always have the same understanding of "vegan" or "gluten-free" as you might in your home country. It's a good idea to explain your dietary restrictions in more detail if needed.
