# Asking About Ingredients

## Overview

In this lesson, we will focus on learning how to ask about ingredients in a Japanese restaurant. Being able to communicate effectively about the ingredients in a dish is essential when dining out in Japan, as many dishes may contain allergens or ingredients that you prefer to avoid. By mastering the phrases and vocabulary related to inquiring about ingredients, you will be able to make informed choices and enjoy your dining experience to the fullest.

During this lesson, you will learn common phrases for asking about ingredients in Japanese, such as "Kore wa nanidesuka?" (What is this?) and "Kono ryouri wa nanidesuka?" (What are the ingredients in this dish?). You will also practice pronouncing the names of common ingredients in Japanese, so that you can recognize them when reading a menu or speaking with restaurant staff. Additionally, we will provide you with tips on how to express food preferences and dietary restrictions in Japanese, so that you can communicate your needs clearly and confidently.

Key Lesson Concepts:

* Learn phrases for asking about ingredients in Japanese
* Practice pronouncing the names of common ingredients
* Understand how to express food preferences and dietary restrictions

## Phrases for Asking About Ingredients

Here are some common phrases you can use when asking about the ingredients in a dish:

1. これには何が入っていますか？
(Kore ni wa nani ga haitteimasu ka?)
What ingredients are in this?

これ (kore) = this
には (ni wa) = in (this dish)
何が (nani ga) = what (ingredients)
入っていますか？ (haitteimasu ka?) = does it contain?

2. この料理には何が使われていますか？
(Kono ryōri ni wa nani ga tsukawareteimasu ka?)
What ingredients are used in this dish?

この料理 (kono ryōri) = this dish
使われていますか？ (tsukawareteimasu ka?) = are used?

3. これにはアレルギーのあるものが入っていますか？
(Kore ni wa arerugī no aru mono ga haitteimasu ka?)
Does this contain anything I'm allergic to?

アレルギー (arerugī) = allergy
のある (no aru) = that I have
もの (mono) = things/items (referring to ingredients)

4. この料理に肉は入っていますか？
(Kono ryōri ni niku wa haitteimasu ka?)
Does this dish contain meat?

肉 (niku) = meat

5. これには魚が入っていますか？
(Kore ni wa sakana ga haitteimasu ka?)
Does this contain fish?

魚 (sakana) = fish

6. この料理には卵が使われていますか？
(Kono ryōri ni wa tamago ga tsukawareteimasu ka?)
Does this dish use eggs?

この料理 (kono ryōri) = this dish
卵 (tamago) = eggs

6. この料理は辛いですか？
(Kono ryōri wa karai desu ka?)
Is this dish spicy?

辛い (karai) = spicy

7. アレルギーがあるので、この料理は食べられません。
(Arerugī ga aru node, kono ryōri wa taberaremasen.)
I have an allergy, so I can’t eat this dish.

アレルギー (arerugī) = allergy
ある (aru) = to have (or, there is or to exist)
ので (node) = because or since
食べられません (taberaremasen) = cannot eat in polite form

## Expressing Food Preferences and Dietary Restrictions

In addition to asking about ingredients, it’s also helpful to know how to express food preferences and dietary restrictions. Here are some useful phrases:

私はベジタリアンです。
(Watashi wa bejitarian desu.)
I am a vegetarian.

私はグルテンアレルギーがあります。
(Watashi wa guruten arerugī ga arimasu.)
I have a gluten allergy.

肉を食べません。
(Niku o tabemasu.)
I don’t eat meat.

卵を食べません。
(Tamago o tabemasu.)
I don’t eat eggs.

辛いものは苦手です。
(Karai mono wa nigate desu.)
I don’t like spicy food.

甘いものが好きです。
(Amai mono ga suki desu.)
I like sweet things.

## Dialogue Example

### Scene 1

Customer:
これに何が入っていますか？
(Kore ni nani ga haitteimasu ka?)
What ingredients are in this?

Waiter:
この料理には、肉と野菜が入っています。
(Kono ryōri ni wa, niku to yasai ga haitteimasu.)
This dish contains meat and vegetables.

Customer:
この料理には魚は使われていますか？
(Kono ryōri ni wa sakana wa tsukawareteimasu ka?)
Does this dish use fish?

Waiter:
いいえ、この料理には魚は使われていません。
(Iie, kono ryōri ni wa sakana wa tsukawareteimasen.)
No, this dish does not contain fish.

Customer:
アレルギーがあるので、この料理は食べられません。
(Arerugī ga aru node, kono ryōri wa taberaremasen.)
I have an allergy, so I can’t eat this dish.

Waiter:
わかりました。別の料理をお勧めします。
(Wakarimashita. Betsu no ryōri o osusume shimasu.)
Understood. I will recommend a different dish.

### Scene 2

Customer:
これには何が入っていますか？
(Kore ni wa nani ga haitteimasu ka?)
What ingredients are in this?

Waiter:
この料理には肉と野菜が入っています。
(Kono ryōri ni wa niku to yasai ga haitteimasu.)
This dish contains meat and vegetables.

Customer:
肉は食べません。
(Niku wa tabemasen.)
I don’t eat meat.

Waiter:
それなら、こちらの料理はどうですか？
(Sore nara, kochira no ryōri wa dō desu ka?)
In that case, how about this dish?

Customer:
これには卵が入っていますか？
(Kore ni wa tamago ga haitteimasu ka?)
Does this contain eggs?

Waiter:
はい、卵が入っています。
(Hai, tamago ga haitteimasu.)
Yes, it contains eggs.

Customer:
では、アレルギーがあるので、これを食べられません。
(Dewa, arerugī ga aru node, kore o taberaremasen.)
Well, I have an allergy, so I can’t eat this.

## Summary

In this lesson, we learned how to ask about ingredients in Japanese when dining at an izakaya. It’s important to be polite and clear, especially when you have dietary restrictions or need to inquire about specific ingredients like fish, eggs, or spice levels. Using -ます forms makes the conversation polite, and understanding key phrases like 入っていますか? and 使われていますか? will help you navigate ordering food at an izakaya with ease.

We learned how to ask about ingredients in Japanese and how to express food preferences and dietary restrictions. By using phrases like 「これには何が入っていますか？」 (What ingredients are in this?) and 「私はベジタリアンです。」 (I am a vegetarian), you can ensure that your dietary needs are understood and respected when dining in Japan. By practicing the vocabulary for common ingredients and dietary restrictions, you'll be able to confidently navigate restaurant menus and have clear communication with restaurant staff.
