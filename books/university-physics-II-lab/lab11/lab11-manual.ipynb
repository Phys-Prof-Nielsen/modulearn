## **Background Theory**

### **RC Charging**

When a capacitor charges through a resistor from a voltage source $V_0$, the voltage across the capacitor evolves as:

$$
V_C(t) = V_0 \left(1 - e^{-t/RC} \right)
$$

- Initially: $V_C(0) = 0$  
- Long time: $V_C \to V_0$

---

### **RC Discharging**

When the capacitor discharges through the resistor:

$$
V_C(t) = V_0 e^{-t/RC}
$$

- Initially: $V_C(0) = V_0$  
- Long time: $V_C \to 0$

---

### **Time Constant**

The time constant is defined as:

$$
\tau = RC
$$

- Charging: $V_C(\tau) \approx 0.63 V_0$  
- Discharging: $V_C(\tau) \approx 0.37 V_0$  

---

### **Linearization**

$$
\ln V_C(t) = \ln V_0 - \frac{t}{RC}
$$

👉 A plot of $\ln V_C$ vs. $t$ gives slope $-1/RC$

## **Experimental Procedure**

### **Part A — Charging/Discharging**

1. Upload Arduino code  
2. Record analog values over time  
3. Convert to voltage:

$$
V = \frac{\text{analogRead}}{1023} \times 5.0
$$

4. Start the Serial Monitor.

5. First, voltage increases until it stabilizes. The program is set to stop recording the charging process after 1 s.
6. Second, the voltage decrease until near zero. The program is set to stop recording the discharging process after 1 s.

7. This process continues through a loop on the arduino.

## **Arduino Code Template**

`const int chargePin = 8;`<br>
`const int readPin = A0;`<br>

`void setup() {`<br>
`  Serial.begin(9600);`<br>
`  pinMode(chargePin, OUTPUT);`<br>
`}`<br>

`void loop() {`<br>
`  digitalWrite(chargePin, HIGH);`<br>
`  for (int i = 0; i < 100; i++) {`<br>
`    int val = analogRead(readPin);`<br>
`    float voltage = val * (5.0 / 1023.0);`<br>
`    float time = i * 0.01;`<br>
`    Serial.print(time);`<br>
`    Serial.print("\t");`<br>
`    Serial.println(voltage);`<br>
`    delay(10);`<br>
`  }`<br>

`  digitalWrite(chargePin, LOW);`<br>
`  for (int i = 0; i < 100; i++) {`<br>
`    int val = analogRead(readPin);`<br>
`    float voltage = val * (5.0 / 1023.0);`<br>
`    float time = i * 0.01;`<br>
`    Serial.print(time);`<br>
`    Serial.print("\t");`<br>
`    Serial.println(voltage);`<br>
`    delay(10);`<br>
`  }`<br>
`  delay(3000);`<br>
`}`
