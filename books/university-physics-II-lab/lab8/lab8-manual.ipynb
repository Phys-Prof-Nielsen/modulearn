## 3. Background Theory

---

### 3.1 Distance Dependence of Electromagnetic Waves

When discussing how the electric field depends on distance, it is important to distinguish between:

- A **static electric field** (such as the field from a point charge), and  
- A **radiating electromagnetic wave** (such as microwaves from the transmitter)

These behave differently.

---

#### Static Electric Field (Coulomb Field)

For a point charge at rest, the electric field decreases as:

$$
E_{\text{static}} \propto \frac{1}{R^2}
$$

This is because the electric field lines spread out over the surface of a sphere of area:

$$
A = 4\pi R^2
$$

As the sphere gets larger, the same number of field lines spreads over more area, producing the familiar inverse-square law.

This field does **not** transport energy away from the source.

---

#### Radiating Electromagnetic Wave

The microwave transmitter emits electromagnetic radiation that **carries energy away from the source**.

The total power emitted by the source spreads over a spherical surface of area:

$$
A = 4\pi R^2
$$

Therefore, the **intensity** (power per unit area) decreases as:

$$
I \propto \frac{1}{R^2}
$$

However, for an electromagnetic wave:

$$
I \propto E^2
$$

Therefore:

$$
E^2 \propto \frac{1}{R^2}
$$

Taking the square root:

$$
E \propto \frac{1}{R}
$$

---

#### Summary

| Quantity | Distance Dependence |
|----------|---------------------|
| Static electric field | $1/R^2$ |
| Wave intensity | $1/R^2$ |
| Wave electric field amplitude | $1/R$ |

---

#### Important Note for This Experiment

The PASCO microwave transmitter is **not an ideal point source**. It produces:

- A directional beam
- Standing wave effects between transmitter and receiver
- Reflections from nearby surfaces

Therefore, you may not observe a perfect $1/R$ or $1/R^2$ relationship.

Your goal is to determine which relationship the receiver voltage most closely follows.

---

### 3.2 Malus’s Law and Detector Response

When a linearly polarized electromagnetic wave encounters a detector (or polarizer) that is oriented at an angle $\theta$ relative to the wave’s electric field direction, only the component of the electric field along the detector axis is measured.

If the incident electric field has amplitude $E_0$, then the component along the detector axis is:

$$
E_{\text{detected}} = E_0 \cos\theta
$$

This comes directly from vector projection.

---

#### From Electric Field to Intensity

The **intensity** of an electromagnetic wave is proportional to the square of the electric field amplitude:

$$
I \propto E^2
$$

Therefore, if the transmitted field amplitude is reduced by $\cos\theta$, the transmitted intensity becomes:

$$
I = I_0 \cos^2\theta
$$

This is known as **Malus’s Law**.

It describes how the *intensity* of linearly polarized radiation changes when passing through a polarizer.

---

#### What Does the Receiver Measure?

The microwave receiver outputs a voltage. But what is that voltage proportional to?

There are two possibilities:

1. **If the receiver responds linearly to electric field amplitude**, then:

$$
V = V_0 \cos\theta
$$

2. **If the receiver responds to intensity**, then:

$$
V = V_0 \cos^2\theta
$$

Real detectors often behave more like intensity detectors, but nonlinearity and circuit response can modify the relationship.

---

#### How We Will Test This

You will:

- Measure voltage as a function of angle $\theta$
- Plot $V$ vs $\theta$
- Compare your data to:
  - $V_0 \cos\theta$
  - $V_0 \cos^2\theta$

The curve that best matches your data reveals whether the detector is responding primarily to electric field amplitude or to intensity.

---

#### Conceptual Summary

1. The **cos θ** factor comes from projecting a vector.
2. The **cos² θ** factor comes from squaring the electric field to get intensity.
3. The experiment determines whether the detector measures field amplitude or intensity.
