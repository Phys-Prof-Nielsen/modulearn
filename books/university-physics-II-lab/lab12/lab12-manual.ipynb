## **Background Theory**

### **Magnetic Field of a Dipole**

A small permanent magnet behaves approximately like a **magnetic dipole** when observed at distances larger than its size.

Along the axis of the magnet, the magnetic field is approximately

$$
B(r) = \frac{\mu_0}{4\pi}\frac{2m}{r^3}
$$

where

- $B$ = magnetic field strength  
- $m$ = magnetic dipole moment of the magnet  
- $r$ = distance from the center of the magnet  
- $\mu_0$ = permeability of free space  

This equation predicts that the magnetic field decreases very rapidly with distance.

---

### **Power-Law Form**

The dipole field can also be written in a simplified form

$$
B(r) = \frac{C}{r^3}
$$

where $C$ is a constant depending on the magnet.

If this model is correct, then a **log–log plot** should produce a straight line:

$$
\ln B = \ln C - 3\ln r
$$

The slope of this line should be **$-3$**.

### **Hall Effect Sensor (SS495B)**

In this experiment we use the **SS495B linear Hall-effect sensor** to measure magnetic field strength. The sensor produces an output voltage that is **linearly proportional to the magnetic field** perpendicular to the sensing surface.

The full datasheet for the sensor can be found here:

https://www.honeywell.com/content/dam/honeywell-sensing-and-iot/en/documents/sps-siot-ss495a1-ss495a2-ss495a-linear-hall-effect-sensor-datasheet-005850-2-en.pdf

#### **How the Sensor Works**

The operation of a Hall sensor is based on the **Hall effect**, discovered by Edwin Hall in 1879. When an electric current flows through a conductor or semiconductor and a magnetic field is applied perpendicular to the direction of the current, the moving charge carriers experience a sideways force. This force causes the charges to accumulate on one side of the material, producing a measurable voltage difference across the conductor. This voltage is called the **Hall voltage**.

The sideways force responsible for this effect is the **magnetic component of the Lorentz force** acting on moving charges:

$$
\vec{F} = q \vec{v} \times \vec{B}
$$

where  

- $q$ is the charge of the carrier  
- $\vec{v}$ is the drift velocity of the charge carriers  
- $\vec{B}$ is the magnetic field  

Because the force acts perpendicular to both the current and the magnetic field, charge carriers are pushed toward one side of the sensor material. As charges accumulate, an electric field builds up that eventually balances the magnetic force. The resulting potential difference across the material is the **Hall voltage**, which is proportional to the magnetic field strength.

In a **Hall-effect sensor**, this voltage is measured and amplified using integrated electronics. The SS495B used in this experiment is a **linear Hall sensor**, meaning the output voltage varies approximately linearly with magnetic field. With no magnetic field present, the sensor outputs a voltage near **half the supply voltage**. When a magnetic field is applied, the output voltage increases or decreases depending on the field direction. This allows the sensor to measure both the **magnitude and direction** of the magnetic field.

When a magnetic field is present, the Hall sensor generates a voltage change proportional to the field:

$$
V_{\text{out}} = V_0 + S B
$$

where

- $V_{\text{out}}$ = sensor output voltage  
- $V_0$ = zero-field voltage (about half the supply voltage)  
- $S$ = sensor sensitivity  
- $B$ = magnetic field strength  

For the SS495B operating at $5\text{ V}$:

- Zero-field voltage is approximately

$$
V_0 \approx 2.5 \text{ V}
$$

- Sensitivity is approximately

$$
S \approx 3.125 \text{ mV/G}
$$

This means the output voltage increases or decreases depending on the magnetic field direction.

---

#### **Pin Identification**

The SS495B has **three legs**. When holding the sensor so that the **larger flat face of the package is facing you** and the legs point downward, the pins from **left to right** are:

| Pin | Function | Connection |
|----|----|----|
| 1 | $V_{cc}$ | Connect to **5 V** on the Arduino |
| 2 | GND | Connect to **Arduino ground** |
| 3 | Output | Connect to **analog input A0** |

---

#### **Sensor Orientation**

The sensor package has **two flat faces**:

- one **larger flat face**
- one **smaller flat face**

The **larger flat face is the sensing face** and should be considered the **front of the sensor**.

👉 **The magnet should face this larger flat side.**

The sensor measures the component of the magnetic field **perpendicular to this face**.

Correct orientation improves the strength and consistency of the measured signal.

---

#### **Important Experimental Note**

Before beginning measurements, record the sensor voltage when **no magnet is nearby**. This baseline voltage is

$$
V_0 \approx \frac{V_{cc}}{2}
$$

The magnetic field signal used in the analysis will be

$$
B_{\text{signal}} \propto V_{\text{out}} - V_0
$$

This removes the sensor offset and isolates the magnetic field contribution.

## **Arduino Code**

Use the following program:

`const int hallPin = A0;`

`void setup()`

`{`

`  Serial.begin(9600);`

`}`

`void loop()`

`{`

`  int sensorValue = analogRead(hallPin);`

`  float voltage = sensorValue * (5.0 / 1023.0);`

`  Serial.println(voltage);`

`  delay(500);`

`}`

This program outputs `voltage` every half second.
