# Lab 7 — Homopolar Motor

## Overview
In this lab you will build a **homopolar motor** using a **AA battery**, a **neodymium magnet**, and a **copper wire**.  
Despite its simplicity, this device demonstrates a core idea of electromagnetism: a **current-carrying conductor in a magnetic field experiences a force**. When the wire is shaped so that the force produces a **torque**, the wire begins to rotate continuously.

---

By the end of this lab, you should be able to:

2. Identify the direction of force using the **right-hand rule** (cross product).
3. Describe how **current, magnetic field strength, and geometry** influence torque and rotation.
4. Measure rotation rate and connect it to electrical and magnetic quantities.

## Background Theory

### 1) Magnetic force on a current-carrying wire
A wire segment of length vector $\vec{L}$ carrying current $I$ in a magnetic field $\vec{B}$ experiences a force
$$
\vec{F} = I\,\vec{L}\times \vec{B}.
$$

The magnitude is
$$
F = I L B \sin\theta,
$$
where $\theta$ is the angle between $\vec{L}$ and $\vec{B}$.

In a homopolar motor, the magnetic field is roughly axial (through the magnet), and the current in parts of the wire has a component perpendicular to $\vec{B}$, creating a non-zero force.

---

### 2) Torque and rotation
If the magnetic force acts at a lever arm distance $r$ from the axis of rotation, it produces a torque
$$
\vec{\tau} = \vec{r}\times \vec{F},
$$
with magnitude (in the simplest picture)
$$
\tau \approx rF.
$$

As long as the torque exceeds resistive effects (friction, contact losses, air drag), the wire rotates.

---

### 3) Current and electrical power
The motor is powered by the battery. A simple circuit model is
$$
I \approx \frac{V}{R_{\text{total}}} = \frac{V}{R_{\text{wire}} + R_{\text{contacts}} + r_{\text{internal}}}.
$$

Electrical power delivered is
$$
P_{\text{elec}} = IV.
$$

Only a fraction becomes mechanical power; much is dissipated as heat at the wire and contact points:
$$
P_{\text{heat}} = I^2 R_{\text{total}}.
$$

## Experimental Procedure

### Part A — Build a Working Homopolar Motor 

In this configuration, the **magnet will be placed on the positive terminal** of the AA battery.  
The small dent in the negative (flat) terminal provides a convenient pivot point for the wire.

---

#### 1. Attach the magnet
Place the **neodymium magnet on the positive (button) terminal** of the AA battery.

- Make sure it is centered.
- The magnet should sit flat and make good electrical contact.

Now the magnet is electrically connected to the positive terminal.

---

#### 2. Shape the copper wire

Bend the copper wire so that:

- The **top center of the wire rests in the small dent of the negative terminal** (this acts as a pivot point).
- The lower ends of the wire lightly touch the **magnet**.
- The wire is symmetric so that its mass is balanced around the battery.

A common and effective design is a symmetric loop or heart-shaped form.  
The key requirements are:

- The wire must complete the circuit:
- The bottom contact points must touch the magnet lightly (not clamp down).
- The wire must be free to rotate without scraping along the battery casing.

---

#### 3. Start the motor

Gently place the shaped wire onto the battery so:

- The top rests in the negative terminal dent.
- The bottom ends contact the magnet.

If the circuit is properly completed, the wire should begin to rotate.

If it does not rotate:

- Slightly adjust the contact points.
- Ensure the wire is not pressing too hard against the magnet.
- Improve symmetry of the wire.
- Lightly spin it to overcome static friction.

---

#### 4. Observe and record

Once rotating, observe and note:

- Direction of rotation.
- Stability of rotation (smooth or wobbling?).
- Changes in speed over time.
- Whether heating occurs at contact points.

>⚠ **Important:** Do not let the motor run continuously for long periods (>10-15 s).  
Large currents can heat the wire and battery quickly. Run in short intervals and allow cooling.

## Data Analysis: Short-Burst Rotation Rate (RPM) Measurements

Because the homopolar motor can draw a large current and warm up quickly, you should **not** run it continuously for long periods. Instead, you will measure rotation rate using **short, safe bursts** and repeat trials for consistency.

---

### Goal
Measure the motor’s rotation rate and quantify how it changes with a simple modification (wire shape, contact quality, or magnet stacking), using **short runs** and basic statistics.

---

### What you will measure
You will determine the rotation frequency $f$ and convert it to angular speed $\omega$ and RPM.

If the motor completes $N$ rotations in a time interval $\Delta t$, then:
$$
f = \frac{N}{\Delta t}.
$$

Angular speed:
$$
\omega = 2\pi f.
$$

Rotation rate in revolutions per minute:
$$
\text{RPM} = 60f.
$$

---

### Procedure (short-burst method)
1. **Test conditions**:
   - **Condition A:** baseline wire shape + one magnet  
   - **Condition B:** modified wire shape
   - **Condition C:** two stacked magnets instead of one  

2. For a single trial:
   - Run the motor for **10–15 s** maximum.
   - Use a phone video (slow motion if available) or a stopwatch.

3. During that short run, measure the time for a fixed number of rotations:
   - Measure the time $t_{10}$ for **10 rotations**.

   Then compute:
   $$
   f = \frac{10}{t_{10}},
   \qquad
   \text{RPM} = 60\left(\frac{10}{t_{10}}\right).
   $$

4. **Let the motor cool** (about 30–60 s), then repeat.

5. Collect at least **5 trials** for each condition.

---

### Analysis
Compute the mean RPM:
$$
\overline{\text{RPM}} = \frac{1}{n}\sum_{i=1}^{n} \text{RPM}_i.
$$

Compute the sample standard deviation:
$$
s_{\text{RPM}} =
\sqrt{\frac{1}{n-1}\sum_{i=1}^{n}\left(\text{RPM}_i - \overline{\text{RPM}}\right)^2 }.
$$

Uncertainty in the mean (standard error):
$$
u_{\overline{\text{RPM}}} = \frac{s_{\text{RPM}}}{\sqrt{n}}.
$$

---

### What to conclude
- Report $\overline{\text{RPM}} \pm u_{\overline{\text{RPM}}}$ for each condition.
- Compare conditions using a percent difference:
$$
\%\text{ difference} =
\frac{\left|\overline{\text{RPM}}_2-\overline{\text{RPM}}_1\right|}{\overline{\text{RPM}}_1}\times 100\%.
$$

Then answer: Which modification produced the highest rotation rate, and why?

Use scaling ideas to support your explanation (qualitatively is fine):
$$
F \sim I L B,
\qquad
\tau \sim r F \sim r I L B.
$$

This connects your measured RPM changes to current $I$, magnetic field strength $B$, and geometry (lever arm $r$).
