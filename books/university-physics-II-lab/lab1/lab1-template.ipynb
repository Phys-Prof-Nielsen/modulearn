### Resistor A: Fit & Plots

*Make short comments followed by code**

```python
import numpy as np

# --- Linear fit for Resistor A: V = R*I + b ---
x = A["current_A"].values
y = A["voltage_V"].values

# Unweighted linear regression with covariance for uncertainties
p, cov = np.polyfit(x, y, deg=1, cov=True)
R_A, b_A = p[0], p[1]
u_RA, u_bA = np.sqrt(np.diag(cov))

print(f"R_A = {R_A:.2f} ± {u_RA:.2f} Ω")
print(f"b_A = {b_A:.3f} ± {u_bA:.3f} V")
```

*Make short comments followed by code*

```python
import matplotlib.pyplot as plt
import matplotlib.pyplot as plt
import numpy as np

# Plot data + fit
x_line = np.linspace(x.min()*0.95, x.max()*1.05, 200)
y_line = R_A * x_line + b_A

plt.figure()
plt.scatter(x, y, label="A data")
plt.plot(x_line, y_line, label="A fit")
plt.xlabel("Current I (A)")
plt.ylabel("Voltage V (V)")
plt.title("Ohm's Law: Resistor A")
plt.legend()
plt.tight_layout()
plt.show()

# Residuals
y_fit = R_A * x + b_A
res = y - y_fit

plt.figure()
plt.axhline(0, linestyle="--")
plt.scatter(x, res)
plt.xlabel("Current I (A)")
plt.ylabel("Residual (V)")
plt.title("Residuals: Resistor A")
plt.tight_layout()
plt.show()
```

## Comparison to Manufacturer Specs

*Read in manufacturers spec file and print value for comparison*

```python
# Load manufacturer specifications
specs = pd.read_csv("/content/drive/MyDrive/Physics-II-Lab/resistor_specs.csv")

def compare_to_spec(resistor_label, R_meas, u_R):
    R_spec = float(
        specs.loc[specs["resistor"] == resistor_label, "R_nominal_ohm"].iloc[0]
    )
    tol = float(
        specs.loc[specs["resistor"] == resistor_label, "tolerance_percent"].iloc[0]
    )
    
    pct_diff = 100.0 * abs(R_meas - R_spec) / R_spec
    within = pct_diff <= tol
    
    print(f"Resistor {resistor_label}: Spec R = {R_spec:.1f} Ω (±{tol:.1f}%)")
    print(f"  Measured R = {R_meas:.2f} ± {u_R:.2f} Ω")
    print(f"  Percent difference = {pct_diff:.2f}%  ->  Within tolerance? {'Yes' if within else 'No'}")

# Compare results to specifications
compare_to_spec("A", R_A, u_RA)
# compare_to_spec("B", R_B, u_RB)
```
