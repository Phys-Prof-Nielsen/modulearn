Real power sources (batteries and power supplies) are **not ideal**. When a current is drawn, the terminal voltage decreases due to the source’s **internal resistance**.

## Background Theory

### Real Source Model
A real voltage source can be modeled as an ideal emf $\mathcal{E}$ in series with an internal resistance $r$.

When a load resistor $R$ is connected, the circuit current is

$$
I = \frac{\mathcal{E}}{R + r}.
$$

The **terminal voltage** across the load resistor is

$$
V = IR.
$$

Substituting for $I$ gives

$$
V = \mathcal{E} - Ir.
$$

This predicts a **linear relationship** between terminal voltage and current:

$$
V = \mathcal{E} - rI.
$$

From a linear fit to $V$ vs. $I$:
- the vertical intercept is $\mathcal{E}$,
- the slope is $-r$.

## Experimental Procedure

### Part 1 — Open-Circuit Voltage
1. Do **not** connect any load resistor to the source.
2. Measure the voltage across the source terminals.
3. Record this value as the **open-circuit voltage** $V_{\text{OC}}$.

---

### Part 2 — Load Measurements
For each load resistor $R$:

1. Connect the resistor to the source.
2. Measure the **current** $I$ through the circuit.
3. Measure the **terminal voltage** $V$ across the load resistor.
4. Record both values in your data table.
5. Disconnect the resistor before switching to the next value.

Use **at least five** different load resistors.

## Appendix — Manual Calculation of Fit Parameter Uncertainties

This appendix shows how the uncertainties in the intercept and slope of a linear fit can be
calculated directly from the data using standard least-squares methods. You are **not required**
to perform these calculations by hand for this lab, but the results are equivalent to the
uncertainties reported by Python or Excel fitting routines.

---

### Linear Model
The data are modeled by a straight line

$$
V = a + bI,
$$

where $a$ is the intercept and $b$ is the slope.

---

### Residuals and Data Scatter
After determining the best-fit values of $a$ and $b$, define the residuals

$$
\Delta V_i = V_i - (a + b I_i).
$$

The variance of the data about the fit is

$$
\sigma^2 = \frac{1}{N - 2} \sum_i (\Delta V_i)^2,
$$

where $N$ is the number of data points and the factor $(N-2)$ accounts for the two fitted parameters.

---

### Uncertainty in the Slope
The uncertainty in the slope $b$ is

$$
\delta b
= \sqrt{
\frac{\sigma^2}{\sum_i (I_i - \bar I)^2}
},
$$

where $\bar I$ is the mean of the current values.

---

### Uncertainty in the Intercept
The uncertainty in the intercept $a$ is

$$
\delta a
= \sqrt{
\sigma^2
\left(
\frac{1}{N}
+
\frac{\bar I^2}{\sum_i (I_i - \bar I)^2}
\right)
}.
$$

---

### Interpretation
These expressions show that:
- the uncertainty in the slope (and therefore in the internal resistance $r = -b$) decreases when the
  current values are spread over a wide range,
- clustering data points near a single current value leads to a poorly determined slope,
- the uncertainties reported by software fitting routines come directly from these formulas.

---

### Practical Note
In this lab, you should obtain $\delta a$ and $\delta b$ from the fitting routine used in Python or
Excel. The purpose of this appendix is to demonstrate that these uncertainties have a clear
mathematical and physical origin and are not arbitrary outputs of the software.
