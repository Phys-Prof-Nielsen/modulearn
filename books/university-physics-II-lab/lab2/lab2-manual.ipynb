## Background — LEDs, Forward Voltage, and Polarity

An **LED (Light-Emitting Diode)** is a semiconductor device that emits light when an electric current flows through it in the **forward direction**. Like all diodes, an LED allows current to flow easily in one direction and blocks current in the opposite direction. Understanding **polarity** and **forward voltage** is essential for this lab.

---

### LED Polarity: Anode vs Cathode
An LED has two terminals:

- **Anode (+)** — the terminal where current enters the LED  
- **Cathode (−)** — the terminal where current exits the LED  

Current flows through an LED **from anode to cathode** when the LED is forward biased.

You can identify the anode and cathode in several ways:

#### 1. Lead Length (most common)
- **Longer lead → Anode (+)**
- **Shorter lead → Cathode (−)**

#### 2. Flat Edge on the Case
- **Flat side → Cathode (−)**
- **Rounded side → Anode (+)**

#### 3. Internal Structure
Looking through the transparent casing:
- **Larger internal metal piece → Cathode (−)**
- **Smaller internal metal piece → Anode (+)**

> **Lab Tip:** Always identify the cathode first. If an LED does not turn on, the most common mistake is reversed polarity.

---

### Forward Bias and Light Emission
When the LED is **forward biased** (anode at a higher electric potential than the cathode):

- Charge carriers cross the semiconductor junction
- Energy is released as visible light
- The LED conducts current and turns on

If the LED is connected in reverse, essentially no current flows and the LED remains off (at the low voltages used in this lab).

---

### Forward Voltage $V_f$
The **forward voltage** $V_f$ is the **voltage drop across the LED when it is conducting current**.

In this lab, it is measured as:
$$
V_f = V_{\text{anode}} - V_{\text{cathode}}
$$

Important characteristics of $V_f$:
- $V_f$ depends on the **LED material and color**
- $V_f$ depends on the **current** through the LED
- $V_f$ depends weakly on **temperature**

Typical values:
- Red LED: $V_f \approx 1.8$–$2.1\ \text{V}$
- Green LED: $V_f \approx 2.0$–$2.4\ \text{V}$
- Blue/White LED: $V_f \approx 2.7$–$3.3\ \text{V}$

Even LEDs of the same color will not have exactly the same $V_f$ due to **manufacturing variation**.

---

### Why a Resistor Is Required
An LED is **not an ohmic device**. Once it turns on, a small increase in voltage can cause a large increase in current.

To limit the current and protect the LED, a **series resistor** is used:
$$
I \approx \frac{V_{\text{supply}} - V_f}{R}
$$

In this lab:
- The Arduino provides a fixed supply voltage
- The resistor keeps the current approximately constant
- This allows meaningful comparison of $V_f$ across different LEDs

---

### Why Forward Voltage Is the Focus of This Lab
The forward voltage $V_f$ is ideal for studying uncertainty and statistics because:

- The physical quantity is approximately constant during each run
- Repeated measurements should give the same value *in principle*
- Any observed scatter is due to **measurement uncertainty**
- Differences between LEDs reveal **device-to-device variation**

This makes $V_f$ a natural quantity for learning how to distinguish:
- **random error** vs **systematic error**
- **spread of data** vs **uncertainty in the mean**

## Circuit Setup (Read Carefully)

Correct wiring and polarity are essential for this experiment. The LED will only conduct and emit light when it is connected in the **forward direction**.

### Step-by-Step Wiring

1. **Digital supply**
   - Connect Arduino **digital pin D8** to **one end of the resistor**.
   - This pin provides a fixed DC voltage when set HIGH.

2. **Resistor to LED (current control)**
   - Connect the **other end of the resistor** to the **anode (+)** of the LED.
   - The **anode** is the **longer lead** (or the rounded side of the LED case).

3. **LED to ground**
   - Connect the **cathode (−)** of the LED to **GND** on the Arduino.
   - The **cathode** is the **shorter lead**, the **flat side** of the LED, or the side with the **larger internal metal piece**.

4. **Voltage measurement connections**
   - Connect **analog pin A0** to the **anode of the LED**  
     (the junction between the resistor and the LED).
   - Connect **analog pin A1** to the **cathode of the LED**  
     (the junction between the LED and ground).

---

### What Each Connection Does

- **D8 → resistor → LED → GND**  
  Sets a fixed current path through the LED when D8 is HIGH.

- **A0 (anode)** measures the electric potential *just before* current enters the LED.

- **A1 (cathode)** measures the electric potential *just after* current leaves the LED.

The LED forward voltage is the **difference** between these two potentials:
$$
V_f = V(\text{A0}) - V(\text{A1})
$$

This is a **differential voltage measurement**, which is how voltage drops across components are measured in professional laboratory setups.

---

### Important Notes

- The LED must be oriented so that **current flows from anode to cathode**.
- If the LED does not turn on, **check polarity first**.
- During each run, the LED remains **ON continuously** while the 20 measurements are collected.
  - This ensures a stable physical state.
  - Any variation in $V_f$ during a run is due to **measurement uncertainty**, not changes in the LED itself.

### Step 2 — Across LEDs (manufacturing variability)
Now treat the list of LED means as your dataset:
$$
\{\bar{V}_{f,1}, \bar{V}_{f,2}, \ldots, \bar{V}_{f,N}\}
$$

Compute:

1. Overall mean forward voltage (typical LED):
$$
\bar{V}_f = \frac{1}{N}\sum_{i=1}^{N}\bar{V}_{f,i}
$$

2. Standard deviation across LEDs:
$$
s_{\text{LED}} = \sqrt{\frac{1}{N-1}\sum_{i=1}^{N}\left(\bar{V}_{f,i}-\bar{V}_f\right)^2}
$$

3. Standard error of the mean (uncertainty in the typical LED voltage):
$$
\mathrm{SEM} = \frac{s_{\text{LED}}}{\sqrt{N}}
$$

## Final Reported Result
Report your final result as:
$$
\boxed{\bar{V}_f \pm \mathrm{SEM}}
$$

- $\bar{V}_f$ is the best estimate of the typical LED forward voltage for your batch.
- SEM is the uncertainty in that mean value.

## Common Pitfalls
- Mixing datasets: do not pool all $20N$ readings into one dataset when computing SEM.
- Confusing SD and SEM: SD describes spread; SEM describes uncertainty in the mean.
- Over-rounding: keep extra digits during calculations; round only the final reported value.
