### Uncertainty in the Mean
- Compute the standard error of the mean (SEM) using the standard deviation across LEDs and the number of LEDs measured.
- Clearly indicate which quantity the SEM applies to.

```python
# Example code
import numpy as np

# Mean forward voltage for each LED (in volts)
# Replace these example values with your measured LED means
Vf_means = np.array([
    1.98,   # LED 1
    2.01,   # LED 2
    1.99,   # LED 3
    2.00,   # LED 4
    1.97    # LED 5
])

# Number of LEDs
N = len(Vf_means)

# Overall mean forward voltage
Vf_mean = np.mean(Vf_means)

# Standard deviation across LEDs (sample standard deviation!)
Vf_std = np.std(Vf_means, ddof=1)

# Standard error of the mean
SEM = Vf_std / np.sqrt(N)

print(f"Number of LEDs: N = {N}")
print(f"Overall mean forward voltage: {Vf_mean:.3f} V")
print(f"Standard deviation across LEDs: {Vf_std:.4f} V")
print(f"Standard error of the mean (SEM): {SEM:.4f} V")
```
