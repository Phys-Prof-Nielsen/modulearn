## Physics background (short)
Ohm’s Law describes the relationship between the electric potential difference across a resistive element and the resulting current:

$$
V = IR,
$$

where  
$V$ is the voltage (V),  
$I$ is the current (A), and  
$R$ is the resistance (Ω).

For an ideal resistor, voltage and current are **linearly related**, with the resistance equal to the slope of a $V$ vs. $I$ graph. Real measurements include noise due to instrument resolution, electrical fluctuations, and contact resistance. The provided dataset includes realistic measurement scatter and a small number of imperfect data points to emulate real laboratory conditions.

In this lab, you will determine the resistance of one or more resistors by fitting a linear model to measured voltage–current data and evaluating how well the model describes the system.
