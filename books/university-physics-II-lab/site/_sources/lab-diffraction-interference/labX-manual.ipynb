## **Background Theory**

### **1. Principle of Superposition**

Light behaves as a wave. When two or more light waves overlap, the total electric field is the sum:

$$
E_{\text{total}} = E_1 + E_2 + \cdots
$$

The intensity seen on the screen is proportional to the square of the field amplitude:

$$
I \propto |E|^2
$$

---

### **2. Single-Slit Diffraction**

A slit of width $a$ causes diffraction because different parts of the slit act as sources of wavelets. Destructive interference (dark fringes) occurs at angles satisfying:

$$
a\sin\theta = m\lambda \qquad (m = 1,2,3,\dots)
$$

For small angles, we can use:

$$
\sin\theta \approx \tan\theta \approx \frac{y}{L}
$$

where:
- $y$ is the distance on the screen from the central maximum
- $L$ is the slit-to-screen distance

So the minima locations satisfy:

$$
a\frac{y_m}{L} = m\lambda
\quad \Rightarrow \quad
y_m = \frac{m\lambda L}{a}
$$

**Key features of a single-slit pattern**
- A wide bright **central maximum**
- Dimmer side maxima
- Smooth intensity decrease away from center

---

### **3. Double-Slit Interference**

Two slits separated by distance $d$ create two coherent sources. The path difference between light from the two slits is:

$$
\Delta r = d\sin\theta
$$

Constructive interference (bright fringes) occurs when:

$$
d\sin\theta = m\lambda \qquad (m = 0,\pm1,\pm2,\dots)
$$

Using the small-angle approximation:

$$
d\frac{y_m}{L} = m\lambda
$$

The spacing between adjacent bright fringes is:

$$
\Delta y = \frac{\lambda L}{d}
$$

**Key features of a double-slit pattern**
- Bright fringes that are **evenly spaced**
- Dark fringes between them
- Much narrower bright regions than the single-slit central maximum

---

### **4. Diffraction Envelope on a Double-Slit Pattern**

Real slits have finite width $a$, so each slit diffracts. The observed double-slit pattern is:

- **Rapid interference fringes** (from slit separation $d$)
- **Modulated by a diffraction envelope** (from slit width $a$)

Missing fringes can occur when an interference maximum lands on a diffraction minimum.

---

### **5. Multiple Slits (Diffraction Grating Behavior)**

For $N$ equally spaced slits (slit separation $d$), the principal maxima still satisfy:

$$
d\sin\theta = m\lambda
$$

As $N$ increases:
- Maxima become **sharper and narrower**
- Peak intensity increases
- Minima become darker

This is why diffraction gratings provide high resolution in spectroscopy.

## **Experimental Procedure**

### **Part A: Setup and Alignment**

1. Mount the laser securely on the optics track.
2. Place the slit wheel holder in the beam path.
3. Place the viewing screen a distance $L$ from the slit wheel.
4. Adjust alignment so the beam passes cleanly through the selected slit and the pattern is centered on the screen.
5. Measure and record the slit-to-screen distance $L$.

---

### **Part B: Single-Slit Diffraction**

1. Rotate the wheel to select a **single slit**.
2. Observe the diffraction pattern and sketch it.
3. Measure the distance from the center to the first minimum on both sides:
   - $y_{1,\text{left}}$, $y_{1,\text{right}}$
4. Repeat for the second minimum:
   - $y_{2,\text{left}}$, $y_{2,\text{right}}$
5. Compute average values:
   $$
   y_1 = \frac{|y_{1,\text{left}}| + |y_{1,\text{right}}|}{2}, 
   \qquad
   y_2 = \frac{|y_{2,\text{left}}| + |y_{2,\text{right}}|}{2}
   $$

---

### **Part C: Double-Slit Interference**

1. Rotate the wheel to select a **double slit**.
2. Observe the interference pattern and sketch it.
3. To reduce uncertainty, measure the distance across several bright fringes:
   - Measure from bright fringe #0 to bright fringe #$n$
   - Call this total width $W$
4. Compute average fringe spacing:
   $$
   \Delta y = \frac{W}{n}
   $$

---

### **Part D: Multiple Slits**

1. Rotate the wheel to select a **multiple-slit** pattern (3 slits or grating).
2. Observe and sketch the pattern.
3. Compare to the double slit:
   - Are maxima sharper?
   - Are minima darker?
   - Is fringe spacing the same or different?

## **Data Analysis**

### **1. Single Slit: Solve for Slit Width $a$ (or Wavelength $\lambda$)**

From the minima condition:

$$
a = \frac{m\lambda L}{y_m}
$$

Compute $a$ using $m=1$ and $m=2$ and compare.

---

### **2. Double Slit: Solve for Wavelength $\lambda$ (or Slit Separation $d$)**

From fringe spacing:

$$
\Delta y = \frac{\lambda L}{d}
\quad \Rightarrow \quad
\lambda = \frac{d\Delta y}{L}
$$

If the laser wavelength is known, you can instead solve for $d$:

$$
d = \frac{\lambda L}{\Delta y}
$$

---

### **3. Percent Error (if comparing to expected wavelength)**

$$
\% \text{ error} = 
\frac{|\lambda_{\text{measured}} - \lambda_{\text{expected}}|}
{\lambda_{\text{expected}}}
\times 100
$$
