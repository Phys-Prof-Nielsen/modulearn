# Katakana: **r-row** (ラ行) + **ワ／ヲ／ン**

Katakana’s **r-row** uses a quick flap (somewhere between English *r* and *l*): **ラ・リ・ル・レ・ロ**. We also meet the **w** kana (**ワ**, historical **ヰ/ヱ**, **ヲ**) and the syllabic **ン**.  
- **ヲ**: mostly the object particle, pronounced **“o.”**  
- **ン**: the syllabic *n*; before **b/p/m** it sounds closer to **m** (e.g., ラ**ン**プ “lamp”).  
- Watch for **ッ** (sokuon, doubles next consonant timing) and **ー** (long vowel).

## Character Tables
**r-row (ラ行)**  
| a | i | u | e | o |
|---|---|---|---|---|
| **ラ** | **リ** | **ル** | **レ** | **ロ** |

**w + n**  
| wa | (wi) | (we) | wo | n |
|----|------|------|----|---|
| **ワ** | ヰ* | ヱ* | **ヲ** | **ン** |

\* ヰ/ヱ are historical; modern Japanese uses **イ／エ** instead.
