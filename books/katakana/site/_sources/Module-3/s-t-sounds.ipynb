# Katakana: The **s** (サ行) and **t** (タ行) Sounds

This module focuses on the **S-row** (サ・シ・ス・セ・ソ) and the **T-row** (タ・チ・ツ・テ・ト), plus the common companions you’ll actually see with them in katakana words:
- **Sokuon (促音)**: small **ッ** for double consonants (e.g., サッカー).
- **Chōonpu (ー)**: long vowel mark (e.g., セーター, スキー).
- **Yōon (拗音)**: small **ャ/ュ/ョ** after **シ/チ** for contracted sounds (シャ/シュ/ショ, チャ/チュ/チョ).
- **Dakuten (゛)**: voiced S→Z (ザ・ジ・ズ・ゼ・ゾ) and T→D (ダ・ヂ・ヅ・デ・ド) to cover real loanwords like **チーズ**, **ドア**.

---

## Character Tables

**S-row and Z-row (dakuten)**

| Row | a  | i  | u  | e  | o  |
|-----|----|----|----|----|----|
| S   | サ | シ | ス | セ | ソ |
| Z   | ザ | ジ | ズ | ゼ | ゾ |

**T-row and D-row (dakuten)**

| Row | a  | i  | u  | e  | o  |
|-----|----|----|----|----|----|
| T   | タ | チ | ツ | テ | ト |
| D   | ダ | ヂ | ヅ | デ | ド |

**Yōon with シ / チ**  
- シ + ャ/ュ/ョ → **シャ / シュ / ショ** (sha / shu / sho)  
- チ + ャ/ュ/ョ → **チャ / チュ / チョ** (cha / chu / cho)

**Add-ons used in this module:** **ッ**, **ー**, **ャ/ュ/ョ**

## Vocabulary Practice

> We’ll mostly target S/T words, but allow necessary vowels, ッ, ー, ャ/ュ/ョ, and dakuten Z/D to match authentic Japanese spellings.

| Katakana       | Romaji        | English              | Notes |
|----------------|---------------|----------------------|-------|
| ドア           | doa           | door 🚪              | D-row (voiced T) |
| カード         | kādo          | card 💳              | long vowel ー |
| スキー         | sukī          | ski 🎿               | long vowel ー |
| ドキドキ       | dokidoki      | heartbeat ❤️         | onomatopoeia |
| セーター       | sētā          | sweater 🧶           | long vowels |
| ソース         | sōsu          | sauce 🍝             | long vowel ー |
| ソックス       | sokkusu       | socks 🧦             | ッ + S-row |
| タクシー       | takushī       | taxi 🚖              | long vowel ー |
| チーズ         | chīzu         | cheese 🧀            | Z-row (ズ), long vowel |
| シャツ         | shatsu        | shirt 👕             | yōon (シャ) |
| サッカー       | sakkā         | soccer ⚽            | ッ + long vowel |
| テスト         | tesuto        | test 📝              | classic S/T mix |
| セット         | setto         | set                  | ッ |
| ソーダ         | sōda          | soda 🥤              | long vowel |
| チケット       | chiketto      | ticket 🎟️           | ッ |

**Pronunciation tips**
- **ッ** marks a short pause before the consonant: ソックス (so**k**-kusu), サッカー (sa**k**-kā).  
- **ー** lengthens the preceding vowel: セーター (se-e), スキー (su-ki-i).  
- **シャ/シュ/ショ・チャ/チュ/チョ** are single syllables (not *si-ya*, *ti-yo*).
