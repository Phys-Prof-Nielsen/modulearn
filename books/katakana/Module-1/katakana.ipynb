# カタカナ (Katakana) Overview

Katakana (カタカナ) is one of the three writing systems used in Japanese, alongside hiragana and kanji. Its origins trace back to the early 9th century during the Heian period, when Buddhist monks developed it as a form of shorthand. They simplified fragments of complex Chinese characters (kanji) to create a more compact script. Unlike hiragana, which emerged from cursive-style simplifications of entire characters, katakana was derived from angular components, which is why it looks more rigid and geometric. Originally, katakana was used mainly by scholars and priests to annotate Chinese texts, providing phonetic readings and clarifications. Over time, its role expanded beyond religious and scholarly use, becoming an integral part of the Japanese writing system.

In modern Japanese, katakana serves specific and practical purposes that distinguish it from hiragana and kanji. It is primarily used for:  
- **Loanwords and foreign names**: words like コンピュータ (konpyūta = computer), テレビ (terebi = television), and personal names like ジョン (Jon).  
- **Onomatopoeia and sound effects**: e.g., ワンワン (wanwan = dog’s bark), ドキドキ (dokidoki = heartbeat).  
- **Scientific, technical, and brand names**: species names, medical terms, company names, and product branding often use katakana for emphasis and clarity.  
- **Emphasis**: similar to italics in English, katakana can highlight words in advertising, packaging, or informal writing.  

The importance of katakana lies in how it bridges Japanese with the rest of the world. As Japan modernized and absorbed global influences, katakana became the natural tool for incorporating foreign vocabulary into the language. It reflects cultural exchange, modernization, and Japan’s adaptation of international ideas into its own linguistic system. Moreover, in education, katakana is essential for learners of Japanese because it provides the foundation for reading menus, maps, product labels, and even manga sound effects. While it may appear secondary to hiragana and kanji, katakana holds a crucial role: it signals foreignness, highlights nuance, and ensures that Japanese writing remains flexible and adaptable in an increasingly globalized world.

## The Long Vowel Mark in Katakana (ー)

In Katakana, you will often see a horizontal dash-like mark: **「ー」**.  
This is called the **長音符（ちょうおんぷ）**, or *long vowel mark*.  

---

### What it means

- 「ー」 tells you to **hold (stretch) the vowel sound before it**.  
- Examples:  
  - コーヒー (*kōhī*) → coffee  
  - スーパー (*sūpā*) → supermarket  
  - ケーキ (*kēki*) → cake  
  - ネーム (*nēmu*) → name  

So **コ** (*ko*) + 「ー」 becomes **コー** (*kō*).

---

### Rules

1. It only lengthens the **immediately preceding vowel**.  
   - シー (*shī*) → long “i”  
   - ツー (*tsū*) → long “u”  
   - ネーム (*nēmu*) → long “e”  

2. The mark follows the orientation of text:  
   - Horizontal writing → horizontal line (ー)  
   - Vertical writing → vertical line (｜)

---

### Is it only used in Katakana?

✅ **Mostly yes.**  
- In **Katakana words** (loanwords, onomatopoeia, foreign names), 「ー」 is standard.  
- In **Hiragana and Kanji**, long vowels are written differently:  
  - おかあさん (*okaasan*, mother) → long あ  
  - とうきょう (*Tōkyō*, Tokyo) → long お/う  
  - おにいさん (*oniisan*, older brother) → long い  

So while Katakana uses 「ー」 to mark long vowels, Hiragana and Kanji rely on extra kana or spelling rules.

---

✅ **Summary**:  
The **「ー」** mark is unique to Katakana. It is an essential part of writing loanwords and gives them their natural pronunciation in Japanese.

## The Small ッ (Sokuon) for Double Consonants

In katakana, a **double consonant** (like the *kk* in “ko**kk**ou”) is written with a **small ッ** called **sokuon** (small **tsu**). It marks a brief pause (glottal stop) before the next consonant and **lengthens timing by one mora**.

**Key points**
- Write the pause with **small ッ** before the consonant you’re doubling: **カッコ (ka-kko)**, **コッコウ (ko-k-kō)**.  
- Sokuon commonly appears before **K / S / T / P** rows.
- In **rōmaji**, you’ll see the doubled consonant: *kakko, kakkō, kokkō*.  
- Don’t confuse ッ (double consonant) with **ー** (long vowel). You can have both: **コッコー** (kokkō).  
- Timing: each of ア/イ/ウ/エ/オ, ッ, and ー counts as **one mora** (beat).

---

### Examples using Module-1 characters (A/K row + ッ + ー)

| Romaji     | Katakana     | Notes |
|------------|--------------|-------|
| **kakko**  | **カッコ**   | “( )” brackets; often seen as a teaching/typography term; katakana acceptable for practice |
| **kakkō**  | **カッコウ** | “cuckoo” (bird); species names frequently appear in katakana |
| **kokkō**  | **コッコウ** | Phonetic example (also appears in compounds/names) |
| **kokkō** (onomatopoeia) | **コッコー** | Crowing/pecking-like sfx; compare with rooster **コケコッコー** |
| **kekka**  | **ケッカ**    | “result/outcome” (normally written with kanji 結果; katakana form appears in signage/UI/testing contexts) |

> **Compare:**  
> - **ココ** (koko) vs **コッコ** (kokko) → the small **ッ** inserts a pause and doubles **k**.  
> - **ケキ** (keki) vs **ケーキ** (kēki “cake”) → here the **ー** lengthens the vowel **e**, no consonant doubling.  
> - **ココウ** (kokō) vs **コッコウ** (kokkō) → the second has a doubled **k** (sokuon).

```python
# Stroke Order (katagana) — fully offline: inline SVGs loaded from ./svg_stroke by ROMAJI
from IPython.display import HTML, display
from textwrap import dedent
from string import Template
import uuid, json, os, glob

uid = "hira_" + uuid.uuid4().hex[:8]  # unique suffix so multiple cells don't collide

headers = ["", "a", "i", "u", "e", "o"]
rows = [
    ("a", ["ア","イ","ウ","エ","オ"]),
    ("k", ["カ","キ","ク","ケ","コ"]),
    ("s", ["サ","シ","ス","セ","ソ"]),
    ("t", ["タ","チ","ツ","テ","ト"]),
    ("n", ["ナ","ニ","ヌ","ネ","ノ"]),
    ("h", ["ハ","ヒ","フ","ヘ","ホ"]),
    ("m", ["マ","ミ","ム","メ","モ"]),
    ("y", ["ヤ","","ユ","","ヨ"]),
    ("r", ["ラ","リ","ル","レ","ロ"]),
    ("w", ["ワ","","","","ヲ"]),
    ("",  ["ン","","","",""]),
]

# --- Kana -> romaji (basic) ---
kana_to_romaji = {
    "ア":"a","イ":"i","ウ":"u","エ":"e","オ":"o",
    "カ":"ka","キ":"ki","ク":"ku","ケ":"ke","コ":"ko",
    "サ":"sa","シ":"shi","ス":"su","セ":"se","ソ":"so",
    "タ":"ta","チ":"chi","ツ":"tsu","テ":"te","ト":"to",
    "ナ":"na","ニ":"ni","ヌ":"nu","ネ":"ne","ノ":"no",
    "ハ":"ha","ヒ":"hi","フ":"fu","ヘ":"he","ホ":"ho",
    "マ":"ma","ミ":"mi","ム":"mu","メ":"me","モ":"mo",
    "ヤ":"ya","ユ":"yu","ヨ":"yo",
    "ラ":"ra","リ":"ri","ル":"ru","レ":"re","ロ":"ro",
    "ワ":"wa","ヲ":"o","ン":"n"
}

# --- Map each katakana to a local SVG filename (by romaji) ---
# Populate the ones you already downloaded; add more as needed.
svg_map = {
    "ア":"a","イ":"i","ウ":"u","エ":"e","オ":"o",
    "カ":"ka","キ":"ki","ク":"ku","ケ":"ke","コ":"ko",
    "シ":"shi","チ":"chi","ツ":"tsu","フ":"fu",
    "ヤ":"ya","ユ":"yu","ヨ":"yo",
    "ラ":"ra","リ":"ri","ル":"ru","レ":"re","ロ":"ro",
    "ワ":"wa","ヲ":"wo","ン":"n"
    # Add others as you place files, e.g. "サ":"sa","ス":"su", ...
}

# ---- Load all local SVGs into memory (romaji -> svg text) ----
base_dir = "svg_stroke"
svg_map = {}  # romaji -> svg text
for path in glob.glob(os.path.join(base_dir, "*.svg")):
    name = os.path.splitext(os.path.basename(path))[0]  # "a", "fu", "wo"
    try:
        with open(path, "r", encoding="utf-8") as f:
            svg_map[name] = f.read()
    except Exception as e:
        print(f"Failed to read {path}: {e}")

# Optional: quick report for missing files the chart will need
needed = sorted({kana_to_romaji[k] for _, cells in rows for k in cells if k})
missing = [r for r in needed if r not in svg_map]
#print("CWD:", os.getcwd())
#print("Found in svg_stroke/:", sorted(os.listdir(base_dir)) if os.path.isdir(base_dir) else "folder not found")
if missing:
    print("Missing svg_stroke files:", ", ".join(missing))

# ---- Build table HTML ----
header_html = "<tr>" + "".join(f"<th>{h}</th>" for h in headers) + "</tr>"
body_rows = []
for label, cells in rows:
    left = "&nbsp;" if label == "" else f"{label}-row"
    tds = []
    for k in cells:
        if not k:
            tds.append("<td class='empty'>—</td>")
        else:
            tds.append(f"<td><button class='kana' data-char='{k}'>{k}</button></td>")
    body_rows.append(f"<tr><th class='rowhead'>{left}</th>{''.join(tds)}</tr>")
table_html = header_html + "".join(body_rows)

# ---- Unique IDs ----
root_id   = f"stroke-root-{uid}"
area_id   = f"stroke-area-{uid}"
table_id  = f"stroke-table-{uid}"
label_id  = f"stroke-label-{uid}"
replay_id = f"stroke-replay-{uid}"
status_id = f"stroke-status-{uid}"

tpl = Template(dedent("""
<div id="$root_id" style="font-family: system-ui, -apple-system, Segoe UI, Roboto, Helvetica, Arial;">
  <style>
    #$root_id .wrap { display: grid; grid-template-columns: 1fr 280px; gap: 16px; align-items: start; }
    #$root_id table { border-collapse: collapse; width: 100%; }
    #$root_id th, #$root_id td { border-bottom: 1px solid #ddd; text-align: center; padding: 6px; }
    #$root_id th.rowhead { width: 90px; text-align: left; }
    #$root_id td.empty { color: #aaa; }
    #$root_id .kana { width: 56px; height: 48px; font-size: 1.4rem; border-radius: 6px; cursor: pointer; border: 1px solid #999; background: #fff; }
    #$root_id .panel { border: 1px solid #ddd; border-radius: 8px; padding: 8px; }
    #$root_id .svgbox { width: 240px; height: 240px; border: 1px solid #ccc; border-radius: 6px; overflow: hidden; display:flex; align-items:center; justify-content:center; }
    #$root_id .svgbox svg { width: 240px; height: 240px; }
    #$root_id .controls { display:flex; gap:8px; align-items:center; margin-top:6px; }
    #$root_id .note { font-size: 0.9rem; color: #444; margin-top: 8px; }
    #$root_id .warn { color:#b00020; }
    #$root_id code { background:#f6f8fa; padding:2px 4px; border-radius:4px; }
  </style>

  <h3>Stroke Order (書き順) – Katakana</h3>
  
  <div class="wrap">
    <div class="panel">
      <table id="$table_id">$table_html</table>
    </div>

    <div class="panel">
      <div id="$label_id" style="font-weight:600; margin:4px 0 8px;">Click a kana…</div>
      <div id="$area_id" class="svgbox"><div style="color:#666;">SVG will load here</div></div>
      <div class="controls">
        <button id="$replay_id" disabled>Replay</button>
        <span id="$status_id" class="note"></span>
      </div>
    </div>
  </div>

  <script>
  (function(){
    const table   = document.getElementById('$table_id');
    const area    = document.getElementById('$area_id');
    const labelEl = document.getElementById('$label_id');
    const replay  = document.getElementById('$replay_id');
    const status  = document.getElementById('$status_id');

    const kanaToRomaji = $kana_to_romaji_json;
    const svgMap = $svg_map_json; // romaji -> raw SVG text

    let lastSVG = null;

    function inlineSVGfor(ch){
      const r = kanaToRomaji[ch];
      return r ? svgMap[r] || null : null;
    }

    function loadInlineSVG(ch){
      const svgText = inlineSVGfor(ch);
      if (!svgText) {
        area.innerHTML = '<div class="warn">No SVG found for <code>' + (kanaToRomaji[ch] || '?') + '.svg</code> in <code>svg_stroke/</code>.</div>';
        replay.disabled = true;
        status.textContent = 'Missing file.';
        return;
      }
      status.textContent = 'Loaded inline.';
      area.innerHTML = svgText;
      const svg = area.querySelector('svg');
      if (svg) {
        svg.setAttribute('width','240');
        svg.setAttribute('height','240');
        const clone = svg.cloneNode(true); // reset SMIL timeline
        area.innerHTML = '';
        area.appendChild(clone);
        lastSVG = clone;
        replay.disabled = false;
      } else {
        area.innerHTML = '<div class="warn">SVG file did not contain a <code>&lt;svg&gt;</code> root.</div>';
        replay.disabled = true;
      }
    }

    function replayAnim(){
      if (!lastSVG) return;
      const fresh = lastSVG.cloneNode(true);
      area.innerHTML = '';
      area.appendChild(fresh);
      lastSVG = fresh;
    }

    table.addEventListener('click', (e) => {
      const btn = e.target.closest('.kana');
      if (!btn) return;
      const ch = btn.getAttribute('data-char');
      labelEl.textContent = 'Stroke Animation: ' + ch + ' (' + (kanaToRomaji[ch] || '?') + ')';
      loadInlineSVG(ch);
    });

    replay.addEventListener('click', replayAnim);
  })();
  </script>
</div>
"""))

# Use safe_substitute so unknown placeholders don’t raise KeyError.
html = tpl.safe_substitute(
    root_id=root_id, table_id=table_id, area_id=area_id,
    label_id=label_id, replay_id=replay_id, status_id=status_id,
    table_html=table_html
)

# Now inject JSON blobs
html = html.replace("$kana_to_romaji_json", json.dumps(kana_to_romaji, ensure_ascii=False))
html = html.replace("$svg_map_json", json.dumps(svg_map, ensure_ascii=False))

display(HTML(html))
```
