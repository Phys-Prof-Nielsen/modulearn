# Katakana: The **n** (ナ行) and **h** (ハ行) Sounds

This module focuses on the **n-row** (ナ・ニ・ヌ・ネ・ノ) and **h-row** (ハ・ヒ・フ・ヘ・ホ), plus their **voiced/half-voiced** partners from the H-row:
- **Dakuten (゛)** → **b-row**: バ・ビ・ブ・ベ・ボ  
- **Handakuten (゜)** → **p-row**: パ・ピ・プ・ペ・ポ  
We’ll also use: **small ッ** (sokuon) for doubled consonants and **ー** (chōonpu) for long vowels.

### Character Tables

**N-row**  
| a | i | u | e | o |
|---|---|---|---|---|
| ナ | ニ | ヌ | ネ | ノ |

**H / B / P rows**  
| Row | a | i | u | e | o |
|-----|---|---|---|---|---|
| H   | ハ | ヒ | フ | ヘ | ホ |
| B   | バ | ビ | ブ | ベ | ボ |
| P   | パ | ピ | プ | ペ | ポ |

> Notes:  
> • **フ** is typically pronounced like **fu** (bilabial [ɸɯ]).  
> • **ッ** marks a short pause before the next consonant: コ**ッ**プ (koppu).  
> • **ー** lengthens the preceding vowel: コーヒー (kōhī).

**Pronunciation tips**  
- **ッ** inserts a brief stop: コップ (ko**p**-pu), フック (fu**k**-ku).  
- **ー** stretches the vowel: コーヒー (ko-o-hī).  
- B/P rows are just **H-row with dakuten/handakuten**: ハ→バ/パ, ヒ→ビ/ピ, etc.
