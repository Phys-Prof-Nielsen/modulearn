# Katakana: The **m** (マ行) and **y** (ヤ行) Sounds  

## Character Tables

**m-row**  
| a | i | u | e | o |
|---|---|---|---|---|
| マ | ミ | ム | メ | モ |

**y-row**  
| a |   | u |   | o |
|---|---|---|---|---|
| ヤ |   | ユ |   | ヨ |

```python
# Module 5 — Pronunciation Player (M & Y)
# Click ▶ to play the JP word. Prefers Google ja-JP voice if available.

from IPython.display import HTML, display
from textwrap import dedent
from string import Template
import uuid, json

uid = "kata_mod5_filtered2_" + uuid.uuid4().hex[:8]

VOCAB = [
    {"jp":"マスク","romaji":"masuku","en":"mask"},
    {"jp":"トマト","romaji":"tomato","en":"tomato"},
    {"jp":"マイク","romaji":"maiku","en":"microphone"},
    {"jp":"ジャム","romaji":"jamu","en":"jam"},
    {"jp":"モップ","romaji":"moppu","en":"mop"},
    {"jp":"タイヤ","romaji":"taiya","en":"tire"},
    {"jp":"ユニーク","romaji":"yunīku","en":"unique"},
    {"jp":"ユーフォー","romaji":"yūfō","en":"UFO"},
    {"jp":"ユニフォーム","romaji":"yunifōmu","en":"uniform"},
    {"jp":"ヨーヨー","romaji":"yōyō","en":"yoyo"},
    {"jp":"ヨット","romaji":"yotto","en":"yacht"},
    {"jp":"マップ","romaji":"mappu","en":"map"},
    {"jp":"メモ","romaji":"memo","en":"memo / note"},
    {"jp":"ユーチューブ","romaji":"yūchūbu","en":"YouTube"},
    {"jp":"モーター","romaji":"mōtā","en":"motor"},
    {"jp":"ヤッホー","romaji":"yahhō","en":"yoo-hoo! (call)"},
    {"jp":"ピアノ","romaji":"piano","en":"piano"},
    {"jp":"マヨネーズ","romaji":"mayonēzu","en":"mayonnaise"},
    {"jp":"ムズムズ","romaji":"muzumuzu","en":"tingling (sfx)"},
]

root_id   = f"mod5-root-{uid}"
rate_id   = f"mod5-rate-{uid}"
rateval_id= f"mod5-rateval-{uid}"
reload_id = f"mod5-reload-{uid}"
voiceinfo_id = f"mod5-voiceinfo-{uid}"
table_id  = f"mod5-table-{uid}"
search_id = f"mod5-search-{uid}"

tpl = Template(dedent("""
<div id="$root_id" style="font-family: system-ui, -apple-system, Segoe UI, Roboto, Helvetica, Arial;">
  <style>
    #$root_id h2 { margin: 8px 0 6px; }
    #$root_id .controls { display:flex; gap:12px; align-items:center; flex-wrap:wrap; margin: 8px 0 12px; }
    #$root_id .panel { border:1px solid #e2e2e2; border-radius:10px; padding:10px; }
    #$root_id table { border-collapse:collapse; width:100%; }
    #$root_id th, #$root_id td { border-bottom:1px solid #eee; text-align:left; padding:8px; vertical-align:top; }
    #$root_id th.jp, #$root_id td.jp { font-size:1.1rem; }
    #$root_id td.romaji { color:#555; white-space:nowrap; }
    #$root_id td.en { color:#333; }
    #$root_id button.play { cursor:pointer; padding:4px 8px; border-radius:8px; border:1px solid #ccc; }
    #$root_id .tiny { font-size:.9rem; color:#555; }
    #$root_id .voiceinfo { font-size:.9rem; color:#444; }
    #$root_id .search { padding:6px 8px; border-radius:8px; border:1px solid #ccc; min-width:240px; }
    #$root_id .ratebox { display:flex; align-items:center; gap:6px; }
    #$root_id input[type=range] { padding:0; }
  </style>

  <h2>▶ Module 5 — m & y Words (Audio)</h2>
  <div class="controls">
    <input id="$search_id" class="search" type="text" placeholder="Filter (Japanese / romaji / English)…">
    <div class="voiceinfo" id="$voiceinfo_id">Voice: (detecting Google Japanese…)</div>
    <div class="ratebox">
      <label for="$rate_id"><b>Speed:</b></label>
      <input type="range" id="$rate_id" min="0.6" max="1.6" step="0.05" value="1.00">
      <span id="$rateval_id">1.00×</span>
    </div>
    <button id="$reload_id" title="Re-detect voice">Re-detect</button>
  </div>

  <div class="panel">
    <table id="$table_id">
      <thead>
        <tr>
          <th></th>
          <th class="jp">カタカナ</th>
          <th>Romaji</th>
          <th>English</th>
        </tr>
      </thead>
      <tbody></tbody>
    </table>
  </div>

  <script>
  (function(){
    const DATA = $data_json;

    const rate = document.getElementById('$rate_id');
    const rateVal = document.getElementById('$rateval_id');
    const voiceInfo = document.getElementById('$voiceinfo_id');
    const reloadBtn = document.getElementById('$reload_id');
    const search = document.getElementById('$search_id');
    const tbody = document.querySelector('#$table_id tbody');

    let currentVoice = null;

    function updateRateLabel(){ const r = parseFloat(rate.value)||1.0; rateVal.textContent = r.toFixed(2)+'×'; }
    rate.addEventListener('input', updateRateLabel);
    rate.addEventListener('change', updateRateLabel);
    updateRateLabel();

    function pickGoogleJa(list){
      const lower = s => (s||'').toLowerCase();
      let v = list.find(v => lower(v.name).includes('google') && v.lang && v.lang.toLowerCase().startsWith('ja'));
      if (v) return v;
      v = list.find(v => v.lang && v.lang.toLowerCase().startsWith('ja'));
      return v || list[0] || null;
    }
    function describeVoice(v){ return v ? 'Voice: '+(v.name||'?')+' — '+(v.lang||'?') : 'Voice: (none)'; }

    function tryLoadVoicesOnce(){
      const list = speechSynthesis.getVoices() || [];
      if (list.length){
        currentVoice = pickGoogleJa(list);
        voiceInfo.textContent = describeVoice(currentVoice);
        return true;
      }
      return false;
    }
    function ensureVoice(){
      if (tryLoadVoicesOnce()) return;
      try { const warm = new SpeechSynthesisUtterance(' '); warm.volume=0; speechSynthesis.speak(warm);}catch(e){}
      let tries=0;
      const t=setInterval(()=>{ tries++; if (tryLoadVoicesOnce()||tries>12) clearInterval(t); },200);
    }
    if (speechSynthesis.addEventListener){ speechSynthesis.addEventListener('voiceschanged', tryLoadVoicesOnce); }
    reloadBtn.addEventListener('click', ensureVoice);
    ensureVoice();

    function currentRate(){ const r = parseFloat(rate.value); return Math.max(0.6, Math.min(1.6, isNaN(r)?1.0:r)); }
    function speak(text){
      const u = new SpeechSynthesisUtterance((text||'').trim());
      u.lang='ja-JP'; if (currentVoice) u.voice=currentVoice; u.rate=currentRate();
      try{ speechSynthesis.cancel(); speechSynthesis.speak(u);}catch(e){ speechSynthesis.speak(new SpeechSynthesisUtterance(text)); }
    }

    function rowHTML(rec){
      return `
        <tr>
          <td><button class="play" data-text="${rec.jp}">▶</button></td>
          <td class="jp">${rec.jp}</td>
          <td class="romaji">${rec.romaji}</td>
          <td class="en">${rec.en}</td>
        </tr>
      `;
    }

    function render(filterText=''){
      const q = (filterText||'').toLowerCase();
      const rows = DATA.filter(rec=>{
        if (!q) return true;
        const hay = [rec.jp||'', rec.romaji||'', rec.en||''].join(' ').toLowerCase();
        return hay.includes(q);
      }).map(rowHTML).join('');
      tbody.innerHTML = rows || `<tr><td colspan="4" class="tiny">No results.</td></tr>`;
    }

    tbody.addEventListener('click', e=>{
      const btn = e.target.closest('button.play'); if(!btn) return;
      speak(btn.getAttribute('data-text'));
    });

    search.addEventListener('input', ()=>render(search.value));
    render();
  })();
  </script>
</div>
"""))

html = tpl.safe_substitute(
    root_id=root_id, uid=uid,
    rate_id=rate_id, rateval_id=rateval_id,
    reload_id=reload_id, voiceinfo_id=voiceinfo_id,
    search_id=search_id, table_id=table_id,
    data_json=json.dumps(VOCAB, ensure_ascii=False)
)

display(HTML(html))
```
