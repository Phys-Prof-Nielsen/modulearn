# Faculty Implementation Guide

## Purpose

This guide provides practical recommendations for implementing AI-supported project-based learning within introductory STEM and physics courses.

The framework was developed specifically for computational physics projects that integrate:
- Python programming,
- Jupyter Notebooks,
- numerical methods,
- uncertainty analysis,
- Monte Carlo simulations,
- and responsible AI-supported workflows.

The central goal is to create learning environments where students engage in authentic scientific reasoning while using AI responsibly as a support tool rather than as a replacement for intellectual work.

---

# Core Philosophy

The framework is built around several key principles:

1. Students should remain the intellectual owners of their work.

2. AI should support scientific reasoning rather than replace it.

3. Assessment should emphasize process, reasoning, and synthesis rather than polished final products alone.

4. Computational workflows should reflect authentic scientific and engineering practices.

5. Students should learn to critically evaluate AI-generated output rather than blindly trust it.

---

# Why Project-Based Computational Physics?

Traditional physics assessments often emphasize:
- isolated problem solving,
- short analytical exercises,
- and highly constrained questions.

While these remain valuable, they do not fully capture modern scientific workflows.

Project-based computational physics allows students to:
- investigate open-ended systems,
- work with simulations and data,
- perform uncertainty analysis,
- develop computational reasoning,
- and engage in scientific communication.

These projects more closely resemble real scientific and engineering practice.

---

# Why Integrate AI?

AI tools are increasingly becoming part of:
- scientific computing,
- software development,
- engineering workflows,
- data analysis,
- and technical communication.

Ignoring AI in STEM education is therefore unrealistic.

Instead of prohibiting AI entirely, this framework emphasizes:
- transparent AI usage,
- responsible computational workflows,
- verification of results,
- and preservation of intellectual ownership.

Students should learn:
- what AI is useful for,
- what AI is unreliable at,
- and how to critically evaluate AI-supported workflows.

# Recommended Project Structure

A staged workflow is strongly recommended.

Breaking the project into phases helps:
- reduce student overwhelm,
- encourage iteration,
- improve feedback opportunities,
- and discourage last-minute AI outsourcing.

A suggested structure is:

1. Topic Exploration  
2. Research Question Development  
3. Proposal Development  
4. Computational Model Development  
5. Data Analysis & Visualization  
6. Monte Carlo Uncertainty Analysis  
7. Scientific Interpretation  
8. Final Communication Product  
9. Responsible AI Reflection  

Each stage should include:
- checkpoints,
- feedback opportunities,
- and reflection.

# Recommended Computational Environment

The framework was designed around:
- Python,
- Jupyter Notebook,
- and Google Colab.

These environments are recommended because they:
- support iterative computational workflows,
- combine code with scientific documentation,
- allow visualization,
- encourage reproducibility,
- and are widely used in STEM fields.

Recommended libraries include:
- numpy,
- matplotlib,
- scipy,
- pandas,
- and sympy.

# Recommended AI Workflow

Students should be required to use a constrained AI “research mentor” rather than unrestricted AI prompting.

The AI mentor should:
- guide thinking,
- ask questions,
- critique reasoning,
- identify missing components,
- and encourage reflection.

The AI mentor should NOT:
- write proposals,
- choose methodologies,
- select variables,
- generate polished submissions,
- or make scientific decisions for students.

This distinction is essential for preserving:
- scientific reasoning,
- intellectual ownership,
- and assessment integrity.

---

# Preventing AI Overreach

One of the largest implementation challenges is preventing AI from taking over the intellectual work.

Early testing showed that unconstrained AI systems tend to:
- generate proposal sections,
- choose research questions,
- select methodologies,
- and effectively complete projects for students.

This undermines the educational goals of the project.

To prevent this, the AI mentor should include:
- non-delegation rules,
- refusal behavior,
- question-driven interaction,
- and strict writing limitations.

Students should remain responsible for:
- all major scientific decisions,
- methodology selection,
- interpretation,
- and final writing.

---

# Recommended Assessment Philosophy

Assessment should emphasize:
- reasoning,
- process,
- interpretation,
- uncertainty analysis,
- and scientific synthesis.

Assessment should not primarily reward:
- polished writing,
- complexity for its own sake,
- or AI-generated sophistication.

Projects should be evaluated based on:
- justification of decisions,
- interpretation of evidence,
- understanding of limitations,
- and integration of physics with computation.

---

# Oral Assessment Components

Short oral questioning can be extremely valuable in AI-supported environments.

Possible oral components include:
- explaining methodology,
- describing numerical methods,
- interpreting plots,
- explaining uncertainty analysis,
- or discussing debugging decisions.

These discussions help distinguish:
- genuine understanding,
- from superficial or AI-generated work.

Oral components also encourage:
- accountability,
- conceptual understanding,
- and scientific communication skills.

---

# Responsible AI Reflection

Students should submit a reflection discussing:
- how AI was used,
- what decisions remained student-made,
- how correctness was verified,
- and how intellectual ownership was maintained.

The reflection should not be treated as a formality.

It should encourage students to think critically about:
- strengths of AI support,
- limitations of AI,
- and risks of overreliance.

# Equity & Accessibility Considerations

Several implementation choices help improve accessibility:

- Google Colab removes the need for local software installation.
- Scaffolded project stages reduce barriers for novice programmers.
- AI-supported debugging can help students overcome programming frustration.
- Jupyter notebooks allow integration of code, explanation, and visualization in one environment.

However, instructors should also discuss:
- risks of unequal AI familiarity,
- overreliance on AI,
- and verification of AI-generated content.

Students should understand that:
- AI fluency does not replace scientific understanding.

---

# Common Implementation Challenges

## 1. Students Asking AI To Complete the Work

This is extremely common.

Students may attempt to:
- outsource proposal writing,
- delegate methodology choices,
- or copy AI-generated code.

The constrained AI mentor framework is specifically designed to reduce this behavior.

---

## 2. Students Trusting Incorrect AI Output

AI systems frequently:
- hallucinate,
- generate incorrect physics,
- misuse terminology,
- or produce unstable code.

Students must be repeatedly reminded:
- verification is their responsibility.

---

## 3. Overemphasis on Coding

Students may focus excessively on coding complexity rather than scientific reasoning.

Projects should emphasize:
- interpretation,
- uncertainty,
- assumptions,
- and methodology.

The goal is scientific synthesis, not software engineering.

---

# Suggested Instructor Workflow

A possible instructor workflow is:

1. Introduce the AI mentor framework  
2. Explain responsible AI use expectations  
3. Introduce the project stages  
4. Require checkpoint submissions  
5. Provide feedback at each stage  
6. Require AI reflection documentation  
7. Include oral questioning or presentation  
8. Evaluate scientific reasoning and synthesis  

This staged structure helps maintain:
- student ownership,
- accountability,
- and manageable project development.
