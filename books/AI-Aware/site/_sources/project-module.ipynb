# Project Overview

In this Phase A project, you will design and investigate your own computational physics research project.

This project is designed to simulate authentic scientific and engineering workflows. Rather than solving a standard textbook problem, you will:
- develop a research question,
- construct computational models,
- analyze numerical methods,
- investigate uncertainty,
- interpret evidence critically,
- and communicate scientific conclusions professionally.

Your project should integrate:
- physics theory,
- computational modeling,
- numerical methods,
- uncertainty analysis,
- and scientific reasoning

into a coherent scientific investigation.

The project emphasizes:
- scientific synthesis,
- computational reasoning,
- scientific methodology,
- and responsible use of AI.

---

# AI-Supported Scientific Workflow (“Vibe Coding”)

This project intentionally incorporates responsible AI-supported computational work, sometimes informally referred to as “vibe coding.”

In professional scientific and engineering environments, researchers increasingly use AI tools to:
- brainstorm ideas,
- debug code,
- explain concepts,
- organize workflows,
- suggest computational approaches,
- and accelerate development.

However, AI should support scientific thinking — not replace it.

For this project:
- YOU must remain the scientific decision maker.
- YOU must understand the physics, methodology, assumptions, and computational work.
- YOU must verify correctness and interpret results.

AI is a tool, not a substitute for scientific reasoning.

# Example Project Topic

One example project is computational orbital dynamics.

For example, students might investigate:
- numerical stability of orbital simulations,
- conservation of energy and angular momentum,
- timestep sensitivity,
- uncertainty propagation,
- or comparison of numerical integration methods.

For a two-body gravitational system:

$$
F = \frac{GMm}{r^2}
$$

Newton’s second law gives:

$$
m\vec{a} = \vec{F}
$$

which leads to:

$$
\vec{a}
=
-\frac{GM}{r^3}\vec{r}
$$

This is only ONE example project direction.

You are encouraged to develop your own scientific investigation.

# Project Structure

The project is divided into several stages.

At each stage, you are expected to make the scientific decisions.

The AI mentor will guide your thinking, but it will not complete the project for you.

---

# Stage 1 — Research Question Development

You will develop a research question related to your chosen physical system and computational investigation.

Possible broad themes include:
- numerical stability,
- conservation laws,
- uncertainty propagation,
- model limitations,
- parameter sensitivity,
- comparison of numerical methods,
- or model-to-data comparison.

You should think about:
- What specifically are you investigating?
- What quantities will you compare?
- What evidence will answer your question?
- What makes the question testable?

You should expect the AI mentor to ask you many questions during this stage.

---

# Stage 2 — Computational Model Development

You will develop a computational model using Python.

Possible numerical methods may include:
- Euler,
- Euler-Cromer,
- Verlet,
- RK4,
- finite-difference methods,
- Monte Carlo methods,
- or other methods you choose and justify.

You must justify:
- your numerical method,
- assumptions,
- parameter choices,
- timestep choices,
- and computational approach.

The AI mentor may help you evaluate your reasoning, but you must make the decisions.

---

# Stage 3 — Data Analysis

You will decide:
- what quantities to track,
- what graphs to create,
- what comparisons to make,
- and how to evaluate your model.

Depending on your project, possible quantities might include:
- energy,
- momentum,
- position,
- velocity,
- angular momentum,
- temperature,
- oscillation amplitude,
- probability distributions,
- or numerical error.

You should focus on scientific interpretation, not simply generating plots.

---

# Stage 4 — Monte Carlo Uncertainty Analysis

You are required to include advanced uncertainty analysis using Monte Carlo methods.

You should investigate:
- sensitivity to parameters,
- propagation of uncertainty,
- robustness of conclusions,
- and limitations of the model.

For example:

$$
x \rightarrow x + \delta x
$$

$$
v \rightarrow v + \delta v
$$

The goal is not simply to “run simulations,” but to interpret uncertainty scientifically.

---

# Stage 5 — Scientific Synthesis

Your project should integrate:
- physics theory,

- and scientific interpretation

into one coherent scientific investigation.

This project is NOT simply a coding assignment.

The emphasis is on:
