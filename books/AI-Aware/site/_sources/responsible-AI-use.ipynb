# Responsible AI Use Guidelines

## Purpose

Artificial intelligence is becoming an increasingly important part of scientific and engineering workflows. Researchers now use AI-assisted tools for:
- coding,
- debugging,
- brainstorming,
- data analysis,
- visualization,
- organization,
- and communication.

The goal of this project is not to avoid AI.

Instead, the goal is to learn how to use AI responsibly while maintaining:
- scientific reasoning,
- intellectual ownership,
- methodological understanding,
- and academic integrity.

These guidelines explain what constitutes appropriate and inappropriate AI use within the Phase A computational physics project.

---

# Central Principle

AI should support scientific thinking — not replace scientific thinking.

You are expected to remain the:
- scientific decision maker,
- computational designer,
- methodological owner,
- and interpreter of the results.

If AI performs the central intellectual work for you, then you are no longer demonstrating scientific synthesis.

---

# What “Responsible AI Use” Means

Responsible AI use means:
- using AI as a support tool,
- critically evaluating AI output,
- verifying correctness,
- maintaining intellectual ownership,
- and understanding the work you submit.

Responsible AI use requires active thinking.

Passive copying is not responsible AI use.

---

# AI as a Scientific Tool

You may use AI similarly to how a scientist or engineer might use:
- a calculator,
- a coding assistant,
- documentation,
- a tutor,
- or a brainstorming partner.

However, AI is not:
- your research advisor,
- your proposal writer,
- your scientific decision maker,
- or your substitute for understanding.

# Acceptable AI Use

The following are examples of acceptable AI use.

---

## Conceptual Support

You may ask AI to:
- explain physics concepts,
- explain numerical methods,
- explain uncertainty propagation,
- explain Monte Carlo methods,
- clarify terminology,
- or help interpret equations.

Example:

> “Can you explain the difference between Euler and Euler-Cromer integration?”

---

## Coding Assistance

You may use AI to:
- help debug code,
- explain Python syntax,
- explain error messages,
- suggest ways to organize code,
- or explain library functions.

Example:

> “Why is my simulation becoming unstable?”

However, you must still:
- understand the code,
- verify correctness,
- and explain your implementation choices.

---

## Brainstorming

You may use AI to:
- brainstorm possible directions,
- identify possible limitations,
- suggest things to consider,
- or help organize thoughts.

However, the final scientific decisions must remain yours.

---

## Revision & Feedback

You may use AI to:
- critique drafts,
- identify unclear explanations,
- improve organization,
- or identify missing components.

However, the original ideas and reasoning must remain student-generated.

---

# Unacceptable AI Use

The following are NOT acceptable.

---

## Writing the Proposal

You may not ask AI to:
- write proposal sections,
- generate the methodology,
- create the research question,
- write conclusions,
- or produce ready-to-submit content.

Examples:
- “Write my methodology section.”
- “Create my proposal.”
- “Write my AI reflection.”

These violate the purpose of the project.

---

## Delegating Scientific Decisions

You may not ask AI to decide:
- what project to do,
- which method to use,
- which variables to analyze,
- what timestep to choose,
- what uncertainty method to use,
- or how to interpret the results.

You are responsible for those decisions.

---

## Blind Copying

You may not:
- copy AI-generated code without understanding it,
- submit AI-generated explanations you cannot explain,
- use AI-generated graphs without understanding them,
- or present AI-generated reasoning as your own.

If you cannot explain it, you do not understand it sufficiently for submission.

---

# Verification Responsibility

You are responsible for verifying:
- the physics,
- the mathematics,
- the computational results,
- the interpretation,
- and the correctness of the code.

AI systems can:
- hallucinate,
- generate incorrect physics,
- misuse terminology,
- produce unstable code,
- and provide misleading explanations.

You are responsible for checking correctness.

---

# Intellectual Ownership

You must remain the intellectual owner of:
- the scientific question,
- methodology,
- computational design,
- uncertainty analysis,
- and interpretation of results.

You should be able to explain:
- why decisions were made,
- why methods were chosen,
- what assumptions were made,
- and what the results mean.

If you cannot explain the project independently, then you do not sufficiently own the work.

---

# AI Reflection Requirement

Your final project must include a reflection discussing:
- how AI was used,
- what assistance AI provided,
- what decisions remained student-made,
- how correctness was verified,
- and how intellectual ownership was maintained.

Your reflection should be honest and specific.

Generic statements such as:
> “I used AI to help with the project.”

are insufficient.

---

# Examples of Strong Reflection Questions

You should think about:
- What did AI help you do efficiently?
- Where did AI give poor suggestions?
- Which AI suggestions did you reject?
- How did you verify the physics?
- How did you verify the code?
- Where did you maintain control of the scientific reasoning?
- What are the risks of overreliance on AI in scientific work?

---

# Why These Rules Exist

The purpose of this project is not simply to produce a polished final product.

The purpose is to help you develop:
- scientific reasoning,
- computational thinking,
- methodological judgment,
- uncertainty analysis skills,
- and scientific communication abilities.

These skills are essential in real scientific and engineering practice.

AI can accelerate workflows, but it cannot replace genuine scientific understanding.

---

# Scientific Integrity in the AI Era

Scientific integrity now includes:
- transparency about AI use,
- verification of AI-generated content,
- critical evaluation of AI suggestions,
- and maintaining intellectual ownership.

Responsible scientists do not blindly trust computational tools.

AI should be treated the same way:
- as a useful tool,
- but one that requires verification, skepticism, and understanding.

---

# Final Perspective

The goal of this project is not to avoid AI.

The goal is to learn how to:
- think scientifically,
- work computationally,
- evaluate uncertainty,
- and use modern AI-supported workflows responsibly.

A strong scientist is not someone who avoids tools.

A strong scientist is someone who understands the tools, verifies the results, and remains responsible for the science.
