## ✅ 3. Common Reactions & Interjections

| Japanese | English |
|----------|---------|
| ほんとう？ | Really? |
| そうですか。 | I see. |
| そうですね。 | Right… / Let me think… |
| いいですね！ | That’s nice! |
| すごい！ | Wow! / Amazing! |
| ええ〜！ | Whaat?! (surprised or doubtful) |
| たしかに。 | True / I agree |
| ちょっと... | Hmm… / Not really... (soft refusal) |
| だいじょうぶです。 | I’m okay. / It’s fine. |

---

## ✅ 4. Grammar Notes

### Polite vs Casual Adjective Endings  
- Polite:  
　たのしいです。 → It’s fun.  
　さびしいです。 → I’m lonely.  

- Casual (omit です):  
　たのしい〜！ → So fun!  
　うれしい〜！ → I’m happy!

## ✅ 5. Mini Conversations

**Conversation 1 – Good News**  
A: にほんにいきます！

**Conversation 2 – Tired**  
A: ねむいです…  
B: だいじょうぶ？  
A: うん、だいじょうぶ。

**Conversation 3 – Compliment**  
A: このケーキ、おいしいですね！  
B: そうですね。わたしもすきです。

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 7 Sentences – Feelings, Reactions, and Mini Conversations
day7_sentences = [
    # --- Emotions ---
    "うれしい",  # happy
    "たのしい",  # fun
    "さびしい",  # lonely
    "かなしい",  # sad
    "おいしい",  # delicious
    "まずい",    # bad-tasting
    "こわい",    # scary
    "すごい",    # amazing
    "たいへん",  # hard/difficult
    "ねむい",    # sleepy

    # --- Common Reactions ---
    "ほんとう？",
    "そうですか。",
    "そうですね。",
    "いいですね！",
    "すごい！",
    "ええ〜！",
    "たしかに。",
    "ちょっと...",
    "だいじょうぶです。",

    # --- Mini Conversation 1 ---
    "にほんにいきます！",
    "ほんとう？すごい！たのしんでね！",

    # --- Mini Conversation 2 ---
    "ねむいです…",
    "だいじょうぶ？",
    "うん、だいじょうぶ。",

    # --- Mini Conversation 3 ---
    "このケーキ、おいしいですね！",
    "そうですね。わたしもすきです。"
]

# ✅ Generate audio for each and display play button
for i, sentence in enumerate(day7_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
