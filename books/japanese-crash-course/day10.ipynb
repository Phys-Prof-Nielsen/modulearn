## ✅ 2. Key Restaurant Expressions

| Japanese | English |
|----------|---------|
| メニューをください。 | Menu, please. |
| おすすめはなんですか？ | What do you recommend? |
| これをください。 | I’ll take this, please. |
| 〜をください。 | [item], please. |
| 〜をおねがいします。 | [item], please (polite). |
| 〜はありますか？ | Do you have ~? |
| すみません。 | Excuse me (to call server). |
| おみずをください。 | Water, please. |
| べつべつでおねがいします。 | Separate checks, please. |
| いっしょでおねがいします。 | All together, please. |
| ごちそうさまでした。 | Thank you for the meal. |

## ✅ 4. Simple Restaurant Dialogue

**1. Ordering**  
A: すみません。メニューをください。  
B: はい、どうぞ。  
A: てんぷらをおねがいします。  
B: はい。おのみものは？  
A: みずをください。

**2. Asking for Recommendation**  
A: おすすめはなんですか？  
B: こちらのカレーがにんきです。  
A: じゃあ、それをください。

**3. Paying & Exiting**  
A: ごちそうさまでした。  
B: ありがとうございました！

## ✅ 9. Cultural Tip  

Always say **「ごちそうさまでした」** when leaving after a meal — it shows respect and gratitude for the food and service.  
Even if you didn’t like it, say it with a smile!

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 10 – Restaurant and Food Phrases
day10_sentences = [
    # --- Key Restaurant Expressions ---
    "メニューをください。",
    "おすすめはなんですか？",
    "これをください。",
    "〜をください。",
    "〜をおねがいします。",
    "〜はありますか？",
    "すみません。",
    "おみずをください。",
    "べつべつでおねがいします。",
    "いっしょでおねがいします。",
    "ごちそうさまでした。",

    # --- Useful Food Vocabulary ---
    "ごはん",
    "さかな",
    "にく",
    "やさい",
    "くだもの",
    "てんぷら",
    "おすし",
    "みそしる",
    "カレーライス",
    "おちゃ",
    "みず",
    "ビール",

    # --- Conversation 1: Ordering ---
    "すみません。メニューをください。",
    "はい、どうぞ。",
    "てんぷらをおねがいします。",
    "はい。おのみものは？",
    "みずをください。",

    # --- Conversation 2: Recommendation ---
    "おすすめはなんですか？",
    "こちらのカレーがにんきです。",
    "じゃあ、それをください。",

    # --- Conversation 3: Paying and Leaving ---
    "ごちそうさまでした。",
    "ありがとうございました！"
]

# ✅ Generate and display play buttons
for i, sentence in enumerate(day10_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence.replace("〜", ""), lang='ja')  # remove 〜 for clean speech
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
