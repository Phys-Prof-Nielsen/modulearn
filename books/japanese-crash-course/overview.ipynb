# 🇯🇵 Japanese Crash Course: 2-Week Survival Communication Plan

### Goal
**Get comfortable with standard daily Japanese communication**  
- Focus: High-frequency verbs, adjectives, survival vocabulary, and polite conversational patterns  
- Target: Speaking with fiancée, family, and handling daily situations in Japan  
- **Time Commitment:** 60–90 minutes per day (can split into 2 sessions)

---

## ✅ WEEK 1: Vocabulary Foundation & Survival Phrases

### **Day 1 – Polite Conversation Basics & Core Greetings**
**Grammar & Set Phrases**
- です・ます, ください, 〜たいです, 〜てください
- Greetings:  
  - おはようございます, こんにちは, こんばんは  
  - はじめまして, よろしくお願いします  
  - ありがとうございます, すみません, ごめんなさい
- Survival phrases:  
  - はい、わかりました。(**Yes, I understand.**)  
  - もういちどおねがいします。(**One more time, please.**)  
  - ゆっくり話してください。(**Please speak slowly.**)
- **Homework:** Practice greetings aloud 10× each; listen & shadow.

### **Day 2 – High-Frequency Verbs (Group 1)**
**10 Most Common Verbs (Polite form & 〜てください):**  
1. します (**do**)  
2. いきます (**go**)  
3. きます (**come**)  
4. みます (**see/watch**)  
5. たべます (**eat**)  
6. のみます (**drink**)  
7. いいます (**say**)  
8. あります (**exist – objects**)  
9. います (**exist – people/animals**)  
10. かいます (**buy**)  

**Patterns:**  
- 〜たいです (I want to ___): たべたいです, のみたいです  
- 〜てください (Please ___): まってください, みてください  

**Mini Dialogue:**  
A: どこにいきますか？ (Where are you going?)  
B: レストランにいきます。 (I’m going to a restaurant.)  

A: なにをたべますか？ (What will you eat?)  
B: さかなをたべます。 (I’ll eat fish.)

### **Day 3 – Core Adjectives**
**10 Daily Use Adjectives:**  
1. たのしい (**fun**)  
2. おいしい (**delicious**)  
3. すごい (**amazing**)  
4. かわいい (**cute**)  
5. きれい (**pretty/clean**)  
6. たいへん (**tough**)  
7. あつい (**hot**) / さむい (**cold**)  
8. いい (**good**) / よくない (**not good**)  

**Examples:**  
- このりょうりはおいしいです。  
- あなたのシャツはかわいいですね！

### **Day 4 – Daily Life Nouns (Food & Shopping)**
**Food & Drinks:**  
ごはん, パン, にく, さかな, やさい, くだもの  
みず, おちゃ, コーヒー, ビール  

**Shopping Phrases:**  
- これ/それ/あれ (**this/that/that over there**)  
- いくらですか？ (**How much is it?**)  
- これをください。 (**I’ll take this, please.**)

### **Day 5 – Family, Directions & Time**
**Family Words:**  
おかあさん, おとうさん, おねえさん, おにいさん, むすめさん, むすこさん  

**Directions:**  
みぎ (**right**), ひだり (**left**), まっすぐ (**straight**)  
ここ/そこ/あそこ (**here/there/over there**)  

**Time Phrases:**  
いま、なんじですか？ (**What time is it now?**)  

**Mini Dialogue:**  
A: えきはどこですか？ (Where is the station?)  
B: まっすぐいって、みぎです。 (Go straight, then right.)

### **Day 6 – Self-Intro & Polite Conversation Practice**
**5 Self-Intro Sentences:**  
1. わたしはキムです。  
2. デンマークからきました。  
3. にほんごがすこしわかります。  
4. よろしくおねがいします。  
5. (Optional) にほんがだいすきです。  

**Mini Role Play:**  
A: おはようございます！きょうはなにをしますか？  
B: こうえんにいきます。

### **Day 7 – Listening & Shadowing Review**
- Review all learned words.  
- Listen to slow Japanese dialogues for 10 min.  
- Shadow aloud; record & compare.

## ✅ WEEK 2: Conversation Expansion & Daily Situations

### **Day 8 – Restaurants & Home Conversations**
**Ordering Phrases:**  
- メニューをください。 (**Menu, please.**)  
- 〜をおねがいします。 (**___, please.**)  
- おいしいですね！ (**It’s delicious!**)  

**Mini Dialogue:**  
A: おなかすいた？ (Are you hungry?)  
B: はい、すごくすいた！ (Yes, very hungry!)

### **Day 9 – Feelings & Opinions**
**Useful Words:**  
うれしい (**happy**), たのしみ (**excited**), かなしい (**sad**), だいじょうぶ (**okay**)  

**Example:**  
きょうはたのしかったです！ (**Today was fun!**)

### **Day 10 – Travel & Shopping Situations**
**Essential Words:**  
チケット, ホテル, バス, タクシー, でんしゃ, きっぷ  

**Mini Dialogue:**  
A: このでんしゃはヒロシマにいきますか？  
B: はい、いきます。

### **Day 11 – Casual Talk with Fiancée & Her Family**
**Compliments:**  
- きれいですね。 (**You look nice.**)  
- りょうりがとてもおいしいです！ (**The food is very delicious!**)  

**Polite Questions:**  
- しごとはたいへんですか？ (**Is work tough?**)  
- こどもさんはなんさいですか？ (**How old is your child?**)

### **Day 12 – Review & Speed Drills**
- Flashcard/Anki review (verbs, adjectives, nouns).  
- Make 10 random sentences.

### **Day 13 – Listening & Shadowing (Longer Dialogues)**
- 15 min of daily-life Japanese audio; repeat 2–3 times.

### **Day 14 – Simulated Day in Japan**
- Morning: Self-introduction & greetings  
- Afternoon: Ordering food & asking directions  
- Evening: Talking with fiancée & complimenting
