## ✅ 3. Conversation Templates to Review  

### Asking Questions  
- なにをたべますか？ (What will you eat?)  
- どこにいきますか？ (Where are you going?)  
- おにいさんはなんさいですか？ (How old is your older brother?)  
- いま、なんじですか？ (What time is it now?)  
- これはいくらですか？ (How much is this?)

### Giving Responses  
- さかなをたべます。  
- レストランにいきます。  
- じゅうごさいです。  
- ごごよじです。  
- それはちょっとたかいです。

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 6 Phrases: Self-Intro + Conversation Review
day6_sentences = [
    # --- Self-Introduction ---
    "はじめまして。",
    "わたしはキムです。",
    "デンマークからきました。",
    "にほんごがすこしわかります。",
    "だいがくでぶつりをおしえています。",
    "にほんがだいすきです。",
    "よろしくおねがいします。",

    # --- Review Questions ---
    "なにをたべますか？",
    "どこにいきますか？",
    "おにいさんはなんさいですか？",
    "いま、なんじですか？",
    "これはいくらですか？",

    # --- Review Answers ---
    "さかなをたべます。",
    "レストランにいきます。",
    "じゅうごさいです。",
    "ごごよじです。",
    "それはちょっとたかいです。",

    # --- Mini Conversation 1: Self-Intro ---
    "はじめまして。",
    "よろしくおねがいします。",
    "デンマークからきました。にほんがすきです。",

    # --- Mini Conversation 2: Restaurant ---
    "このカレーをください。",
    "はい、いくらですか？",
    "ごひゃくえんです。",

    # --- Mini Conversation 3: Directions ---
    "こうえんはどこですか？",
    "まっすぐいって、ひだりです。"
]

# ✅ Generate and display audio play buttons
for i, sentence in enumerate(day6_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
