## ✅ 2. Core Adjectives

| Japanese | Meaning | Example Sentence |
|----------|---------|------------------|
| たのしい | Fun | パーティーはたのしいです (The party is fun) |
| おいしい | Delicious | このラーメンはおいしいです (This ramen is delicious) |
| すごい | Amazing / great | すごいですね！ (That’s amazing!) |
| かわいい | Cute | ねこはかわいいです (Cats are cute) |
| きれい | Pretty / clean | サナエさんはきれいです (Sanae is pretty) |
| たいへん | Tough / hard | しごとはたいへんです (Work is hard) |
| あつい | Hot (weather or touch) | きょうはあついです (Today is hot) |
| さむい | Cold (weather) | きょうはさむいです (Today is cold) |
| いい | Good | それはいいです (That is good) |
| よくない | Not good | それはよくないです (That’s not good) |

---

## ✅ 3. Using Adjectives in Sentences

### **Basic Polite Form**
- [Adjective] + です → **Polite statement**  
  - このケーキはおいしいです (**This cake is delicious**)  
  - きょうはさむいです (**It’s cold today**)

### **Reaction/Agreement**
- Add **ですね** (softens / agrees politely):  
  - すごいですね！ (**That’s amazing, isn’t it!**)  
  - おいしいですね！ (**It’s delicious, isn’t it!**)

**Dialogue 2 – Weather**

B: そうですね、たいへんです。  

**Dialogue 3 – Compliment**  
A: サナエさんはきれいですね。  
B: ありがとうございます。

```python
%reset -f 

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 3 – Core Adjectives & Dialogues
day3_sentences = [
    # --- Core Adjectives ---
    "たのしいです。",
    "おいしいです。",
    "すごいですね。",
    "かわいいです。",
    "きれいです。",
    "たいへんです。",
    "あついです。",
    "さむいです。",
    "いいです。",
    "よくないです。",
    
    # --- Example Sentences ---
    "このケーキはおいしいです。",
    "きょうはさむいです。",
    "すごいですね！",
    "おいしいですね！",
    "サナエさんはきれいですね。",
    
    # --- Mini Dialogues ---
    "このカレーはおいしいですね。",
    "はい、とてもおいしいです。",
    "きょうはあついですね。",
    "そうですね、たいへんです。",
    "サナエさんはきれいですね。",
    "ありがとうございます。"
]

# ✅ Generate and display play buttons
for i, sentence in enumerate(day3_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
