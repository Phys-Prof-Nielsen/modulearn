## ✅ 2. Useful Polite Instructions

| Japanese | English |
|----------|---------|
| まってください。 | Please wait. |
| きてください。 | Please come. |
| すわってください。 | Please sit down. |
| たってください。 | Please stand up. |
| はいってください。 | Please come in. |
| みてください。 | Please look. |
| ちょっとまってください。 | Wait a moment, please. |
| おねがいします。 | Please (general polite request) |
| きをつけて。 | Take care. |
| だいじょうぶですよ。 | It’s okay. Don’t worry. |

---

## ✅ 3. Giving Instructions Casually  
(Close friends, family, or children)

| Casual | Polite |
|--------|--------|
| まって。 | まってください。 |
| きて。 | きてください。 |
| すわって。 | すわってください。 |
| はいって。 | はいってください。 |

Note: Use casual forms only with people close to you or younger than you. Stick with **〜てください** in polite situations.

---

## ✅ 4. Common Instructions You Might Hear

| Japanese | English |
|----------|---------|
| こちらへどうぞ。 | This way, please. |
| どうぞおはいりください。 | Please come in. |
| おかけください。 | Please have a seat. |
| おまちください。 | Please wait. |
| おきをつけて。 | Take care. |
| ごちゅういください。 | Please be careful. |

These are often used by store staff, hosts, or receptionists — very polite and formal.

## ✅ 5. Mini Conversations  

**1. Visiting someone's home**  
A: こんにちは。  
B: どうぞおはいりください。  
A: おじゃまします。  

**2. At a restaurant**  
A: こちらへどうぞ。  
B: はい。  
A: おすわりください。  

**3. Leaving a place**  
A: きをつけてくださいね。  
B: はい、ありがとうございます。

## ✅ 9. Bonus Vocabulary (Softening & Politeness)  

| Japanese | English |
|----------|---------|
| 〜ください | please (used after verb stem) |
| 〜ね | softens or invites listener (いいですね = sounds nice, right?) |
| お〜ください | very formal polite command (e.g., おまちください) |
| おじゃまします | “Sorry for intruding” (used when entering someone’s home) |

Repeat the following 3× each:

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 8 Phrases: Simple Instructions & Commands
day8_sentences = [
    # --- Polite Instructions ---
    "まってください。",
    "きてください。",
    "すわってください。",
    "たってください。",
    "はいってください。",
    "みてください。",
    "ちょっとまってください。",
    "おねがいします。",
    "きをつけて。",
    "だいじょうぶですよ。",

    # --- Casual Forms (compare to polite) ---
    "まって。",
    "きて。",
    "すわって。",
    "はいって。",

    # --- Phrases you might hear in public/formal settings ---
    "こちらへどうぞ。",
    "どうぞおはいりください。",
    "おかけください。",
    "おまちください。",
    "おきをつけて。",
    "ごちゅういください。",

    # --- Mini Conversation 1: Visiting someone’s home ---
    "こんにちは。",
    "どうぞおはいりください。",
    "おじゃまします。",

    # --- Mini Conversation 2: At a restaurant ---
    "こちらへどうぞ。",
    "はい。",
    "おすわりください。",

    # --- Mini Conversation 3: Saying goodbye politely ---
    "きをつけてくださいね。",
    "はい、ありがとうございます。"
]

# ✅ Generate audio for each sentence with play button
for i, sentence in enumerate(day8_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
