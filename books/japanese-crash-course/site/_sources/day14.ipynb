## ✅ 2. Key Expressions for Scheduling

| Japanese | English |
|----------|---------|
| よてい | schedule / plan |
| じかんがあります | I have time / I'm available |
| じかんがありません | I don’t have time / I’m not free |
| あいています | It's open / available |
| あいていません | It's not open / unavailable |
| よやくします | to make a reservation / appointment |
| あしたのごごはあいていますか？ | Are you free tomorrow afternoon? |
| なんじがいいですか？ | What time works for you? |
| いつがいいですか？ | When is good for you? |
| 〜はどうですか？ | How about ~? |

---

## ✅ 4. Example Sentences

- あしたのごご、あいていますか？  
→ Are you free tomorrow afternoon?

- すみません、そのじかんはちょっと…  
→ Sorry, that time doesn't work for me.

- なんじがいいですか？  
→ What time is good?

- よていがあります。  
→ I have plans.

- らいしゅうのげつようびはあいています。  
→ I’m free next Monday.

- ごごにじにあいましょう。  
→ Let’s meet at 2 p.m.

## ✅ 6. Mini Conversations

**1. Making an Appointment**  
A: あしたのごご、あいていますか？  
B: はい、だいじょうぶです。なんじがいいですか？  
A: ごごにじはどうですか？  
B: はい、にじにあいましょう。

**2. Soft Refusal**  
A: きんようびのよるは？  
B: すみません、よていがあります。

## ✅ 11. Cultural Tip

In Japanese, confirming plans is often done very politely. It’s common to use soft phrases like:
- ちょっと…そのじかんはむずかしいです  
- 〜はどうですか？ (as a gentle suggestion)

Politeness and flexibility are valued when scheduling!

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 14 – Appointments & Scheduling
day14_sentences = [
    # --- Key scheduling expressions ---
    "よてい",  # schedule / plan
    "じかんがあります。",
    "じかんがありません。",
    "あいています。",
    "あいていません。",
    "よやくします。",
    "あしたのごごはあいていますか？",
    "なんじがいいですか？",
    "いつがいいですか？",
    "〜はどうですか？",

    # --- Time & date vocabulary ---
    "あさ", "ごご", "よる", "いま", "あとで", "あした", "きょう",
    "らいしゅう", "まいにち", "しゅうまつ",

    # --- Days of the week ---
    "にちようび", "げつようび", "かようび", "すいようび", "もくようび", "きんようび", "どようび",

    # --- Example sentences ---
    "あしたのごご、あいていますか？",
    "すみません、そのじかんはちょっと…",
    "なんじがいいですか？",
    "よていがあります。",
    "らいしゅうのげつようびはあいています。",
    "ごごにじにあいましょう。",

    # --- Mini conversation 1 ---
    "あしたのごご、あいていますか？",
    "はい、だいじょうぶです。なんじがいいですか？",
    "ごごにじはどうですか？",
    "はい、にじにあいましょう。",

    # --- Mini conversation 2 ---
    "きんようびのよるは？",
    "すみません、よていがあります。",

    # --- Bonus polite scheduling phrases ---
    "かくにんします。",
    "けっていします。",
    "へんこうします。",
    "つぎ",
    "さいしょ",
    "さいご",
    "たぶん"
]

# ✅ Generate and play each sentence with a button
for i, sentence in enumerate(day14_sentences):
    clean_text = sentence.replace("〜", "")  # remove placeholder
    print(f"Sentence {i+1}: {clean_text}")
    tts = gTTS(text=clean_text, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
