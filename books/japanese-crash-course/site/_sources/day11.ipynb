---

## ✅ 2. Key Grammar & Phrases

| Japanese | English |
|----------|---------|
| 〜で | at/in (place of action) |
| 〜に | at/on (point in time or destination) |
| 〜ましょう | let’s (polite invitation) |
| いきましょう | let’s go |
| たべましょう | let’s eat |
| 〜ませんか？ | shall we...? / would you like to...? |
| はい、いいですね。 | Yes, sounds good. |
| ちょっと… | Umm... (soft refusal) |
| なんじにあいますか？ | What time shall we meet? |
| どこであいますか？ | Where shall we meet? |

---

## ✅ 3. Examples Using Time + Place + Action

- ごごごじにえきであいましょう。  
→ Let’s meet at the station at 5 p.m.

- にちようびにレストランでひるごはんをたべましょう。  
→ Let’s have lunch at a restaurant on Sunday.

- あした、こうえんでジョギングしませんか？  
→ Shall we go jogging at the park tomorrow?

- どようびにともだちのうちにいきます。  
→ I’m going to a friend’s house on Saturday.

| あした | tomorrow |
| きょう | today |

---

## ✅ 5. Mini Conversations

**1. Planning a Meeting**  
A: どこであいますか？  
B: えきであいましょう。  
A: なんじに？  
B: ごごごじに。

**2. Casual Invitation**  
A: あした、こうえんでジョギングしませんか？  
B: はい、いいですね！

**3. Soft Refusal**  
A: にちようびにえいがをみましょう。  
B: ちょっと…にちようびはようじがあります。

## ✅ 8. Bonus Grammar Tip

- Use **に** for time: ごごごじに (at 5 p.m.)  
- Use **で** for place: レストランで (at the restaurant)  
- Order is flexible but natural: [Time] に [Place] で [Action]

## ✅ 10. Cultural Note

Japanese people often meet **on time or early**, especially for dates or plans.  
Use **〜ましょう** for polite invitations, and **ちょっと…** for gentle refusals — no need to say “no” directly!

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 11 – Making Plans and Invitations
day11_sentences = [
    # --- Core expressions ---
    "〜で",  # at / in (place of action)
    "〜に",  # at / on (time or destination)
    "〜ましょう",  # let’s (polite)
    "〜ませんか？",  # would you like to...?
    "いきましょう。",
    "たべましょう。",
    "はい、いいですね。",
    "ちょっと…",
    "なんじにあいますか？",
    "どこであいますか？",

    # --- Full example sentences ---
    "ごごごじにえきであいましょう。",
    "にちようびにレストランでひるごはんをたべましょう。",
    "あした、こうえんでジョギングしませんか？",
    "どようびにともだちのうちにいきます。",

    # --- Time and Place vocab ---
    "えき", "こうえん", "レストラン", "ともだちのうち",
    "ごごごじ", "ごぜんじゅうじ", "にちようび", "どようび", "あした", "きょう",

    # --- Mini Conversation 1 ---
    "どこであいますか？",
    "えきであいましょう。",
    "なんじに？",
    "ごごごじに。",

    # --- Mini Conversation 2 ---
    "あした、こうえんでジョギングしませんか？",
    "はい、いいですね！",

    # --- Mini Conversation 3 ---
    "にちようびにえいがをみましょう。",
    "ちょっと…にちようびはようじがあります。"
]

# ✅ Generate audio and display play buttons
for i, sentence in enumerate(day11_sentences):
    clean_text = sentence.replace("〜", "")  # remove placeholder marks
    print(f"Sentence {i+1}: {clean_text}")
    tts = gTTS(text=clean_text, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
