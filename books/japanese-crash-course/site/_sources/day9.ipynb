## ✅ 2. Key Shopping Expressions

| Japanese | English |
|----------|---------|
| これ、ください。 | I’ll take this, please. |
| いくらですか？ | How much is it? |
| たかいですね。 | It’s expensive, huh. |
| やすいですね！ | That’s cheap! |
| ちょっとたかいです。 | It’s a bit expensive. |
| べつのをみせてください。 | Please show me a different one. |
| もうすこしおおきいのはありますか？ | Do you have a slightly bigger one? |
| これをください。 | I’ll take this one. |
| カードでおねがいします。 | I’ll pay by card, please. |
| げんきんでおねがいします。 | I’ll pay in cash, please. |

---

## ✅ 5. Mini Conversations

**1. Asking for price**  
A: これはいくらですか？  
B: ごひゃくえんです。  
A: ちょっとたかいですね…

**2. Buying something**  
A: これをください。  
B: はい、おかいけいはせんにひゃくえんです。  
A: カードでおねがいします。

**3. Asking for something else**  
A: もうすこしおおきいのはありますか？  
B: はい、こちらです。  
A: これ、ください。

## ✅ 10. Cultural Note  

It’s polite to **hand and receive money or cards with two hands** in Japan, especially at convenience stores.  
You can also say **ありがとうございます** after paying or receiving change.

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 9 Shopping & Payment Sentences
day9_sentences = [
    # --- Shopping Expressions ---
    "これ、ください。",
    "いくらですか？",
    "たかいですね。",
    "やすいですね！",
    "ちょっとたかいです。",
    "べつのをみせてください。",
    "もうすこしおおきいのはありますか？",
    "これをください。",
    "カードでおねがいします。",
    "げんきんでおねがいします。",

    # --- Useful Nouns & Counters ---
    "ひとつ", "ふたつ", "みっつ", "よっつ",
    "このへん", "おかいけい", "レジ", "おつり", "しょうひぜい",
    "クレジットカード", "げんきん",

    # --- Numbers & Prices ---
    "ひゃくえん",
    "ごひゃくえん",
    "せんえん",
    "にせんえん",
    "いちまんえん",

    # --- Mini Conversation 1 ---
    "これはいくらですか？",
    "ごひゃくえんです。",
    "ちょっとたかいですね…",

    # --- Mini Conversation 2 ---
    "これをください。",
    "はい、おかいけいはせんにひゃくえんです。",
    "カードでおねがいします。",

    # --- Mini Conversation 3 ---
    "もうすこしおおきいのはありますか？",
    "はい、こちらです。",
    "これ、ください。"
]

# ✅ Generate audio and display play buttons
for i, sentence in enumerate(day9_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
