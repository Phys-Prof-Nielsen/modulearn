## ✅ 4. Time Phrases

| Japanese | Meaning |
|----------|---------|
| いま | now |
| なんじですか？ | What time is it? |
| いちじです | It’s 1 o’clock |
| はちじはんです | It’s 8:30 |
| ごぜん | A.M. |
| ごご | P.M. |

---

## ✅ 5. Grammar & Usage Patterns

### Asking Time
- いま、なんじですか？ → What time is it now?  
- ごぜん or ごご can be added before the time:  
  - ごぜんいちじ → 1 A.M.  
  - ごごさんじ → 3 P.M.

### Giving Directions
- 〜はどこですか？ → Where is ~?  
  - えきはどこですか？ → Where is the station?  
- Combine location words:  
  - まっすぐいって、みぎです → Go straight, then right.

---

## ✅ 6. Mini-Dialogues

**Dialogue 1 – Asking for Directions**  
A: えきはどこですか？  
B: まっすぐいって、みぎです。  

**Dialogue 2 – Asking the Time**  
A: いま、なんじですか？  
B: はちじはんです。  

**Dialogue 3 – Talking About Family**  
A: おにいさんはなんさいですか？  
B: じゅうごさいです。

```python
%reset -f 

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 5: Family, Directions, and Time Phrases
day5_sentences = [
    # --- Family Vocabulary ---
    "おかあさん。",  # mother
    "おとうさん。",  # father
    "おねえさん。",  # older sister
    "おにいさん。",  # older brother
    "むすめさん。",  # daughter
    "むすこさん。",  # son

    # --- Direction Vocabulary ---
    "みぎ。",       # right
    "ひだり。",     # left
    "まっすぐ。",   # straight
    "ここ。",       # here
    "そこ。",       # there
    "あそこ。",     # over there

    # --- Time Vocabulary & Phrases ---
    "いま。",           # now
    "なんじですか？",   # What time is it?
    "いちじです。",     # It is 1 o'clock
    "はちじはんです。", # It's 8:30
    "ごぜんですか？",   # Is it A.M.?
    "ごごです。",       # It's P.M.

    # --- Mini Dialogue 1 – Asking for Directions ---
    "えきはどこですか？",
    "まっすぐいって、みぎです。",

    # --- Mini Dialogue 2 – Asking Time ---
    "いま、なんじですか？",
    "はちじはんです。",

    # --- Mini Dialogue 3 – Talking about Family ---
    "おにいさんはなんさいですか？",
    "じゅうごさいです。"
]

# ✅ Generate audio and embed play buttons
for i, sentence in enumerate(day5_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
