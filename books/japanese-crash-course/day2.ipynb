## ✅ 2. Core Verbs (Polite Forms)

| Verb | Japanese | Meaning | Example |
|------|----------|---------|---------|
| 1 | します | do | スポーツをします (I play sports) |
| 2 | いきます | go | レストランにいきます (I go to a restaurant) |
| 3 | きます | come | うちにきます (Come to the house) |
| 4 | みます | see/watch | テレビをみます (I watch TV) |
| 5 | たべます | eat | パンをたべます (I eat bread) |
| 6 | のみます | drink | おちゃをのみます (I drink tea) |
| 7 | いいます | say | こんにちはといいます (Say “hello”) |
| 8 | あります | exist (objects) | つくえがあります (There is a desk) |
| 9 | います | exist (people/animals) | ねこがいます (There is a cat) |
| 10 | かいます | buy | ほんをかいます (I buy a book) |

---

## ✅ 3. Essential Patterns

### **1) 〜たいです (I want to ___)**
- たべたいです (**I want to eat**)  
- のみたいです (**I want to drink**)  

### **2) 〜てください (Please ___)**
- まってください (**Please wait**)  
- みてください (**Please look**)  

---

## ✅ 4. Mini-Dialogues (Practice Speaking)

**Dialogue 1 – Going Somewhere**  
A: どこにいきますか？ (Where are you going?)  
B: レストランにいきます。 (I’m going to a restaurant.)  

**Dialogue 2 – Eating Something**  
A: なにをたべますか？ (What will you eat?)  
B: さかなをたべます。 (I’ll eat fish.)  

**Dialogue 3 – Request**  
A: まってください。 (Please wait.)  
B: はい、わかりました。 (Yes, I understand.)

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io


# ✅ Day 2 – High-Frequency Verbs & Mini Dialogues
day2_sentences = [
    # --- Verbs ---
    "します。",       # do
    "いきます。",     # go
    "きます。",       # come
    "みます。",       # see, watch
    "たべます。",     # eat
    "のみます。",     # drink
    "いいます。",     # say
    "あります。",     # exist (objects)
    "います。",       # exist (people)
    "かいます。",     # buy
    
    # --- Example Patterns ---
    "たべたいです。", # I want to eat
    "のみたいです。", # I want to drink
    "まってください。", # Please wait
    "みてください。",   # Please look

    # --- Mini Dialogues ---
    "どこにいきますか？",    # Where are you going?
    "レストランにいきます。", # I’m going to a restaurant
    "なにをたべますか？",    # What will you eat?
    "さかなをたべます。"     # I will eat fish
    "まってください。",
    "はい、わかりました。"
]

# ✅ Generate and play each sentence
for i, sentence in enumerate(day2_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
