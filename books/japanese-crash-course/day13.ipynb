## ✅ 2. Essential Grammar & Structures

| Japanese | English |
|----------|---------|
| 〜はどこですか？ | Where is ~? |
| どうやっていきますか？ | How do I get there? |
| まっすぐいってください。 | Go straight. |
| みぎにまがってください。 | Turn right. |
| ひだりにまがってください。 | Turn left. |
| となり | next to |
| まえ | in front of |
| うしろ | behind |
| ちかく | nearby / close |
| とおい | far |
| 〜のとなりにあります。 | It's next to ~. |
| 〜のまえにあります。 | It's in front of ~. |

---

## ✅ 3. Key Vocabulary

| Japanese | English |
|----------|---------|
| コンビニ | convenience store |
| えき | station |
| ぎんこう | bank |
| ゆうびんきょく | post office |
| こうえん | park |
| としょかん | library |
| びょういん | hospital |
| みせ | shop |
| みち | street / road |
| こうさてん | intersection |
| しんごう | traffic light |

---

## ✅ 4. Sample Questions and Answers

**1. Asking a location**  
A: すみません、ゆうびんきょくはどこですか？  
B: あのこうさてんをひだりにまがってください。

**2. Asking for directions**  
A: ぎんこうはどうやっていきますか？  
B: まっすぐいって、しんごうをみぎにまがってください。

**3. Simple landmark response**  
A: としょかんは？  
B: コンビニのとなりにあります。

## ✅ 9. Cultural Tip

In Japan, strangers are often very willing to help with directions. Even if they don’t speak much English, many will walk with you or point clearly.  
Just start politely with:  
**「すみません、ちょっときいてもいいですか？」**

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 13 – Directions and Places
day13_sentences = [
    # --- Core Grammar & Expressions ---
    "〜はどこですか？",
    "どうやっていきますか？",
    "まっすぐいってください。",
    "みぎにまがってください。",
    "ひだりにまがってください。",
    "〜のとなりにあります。",
    "〜のまえにあります。",
    "ちかく", "とおい",
    "まえ", "うしろ", "となり",

    # --- Places Vocabulary ---
    "コンビニ", "えき", "ぎんこう", "ゆうびんきょく",
    "こうえん", "としょかん", "びょういん", "みせ", "みち", "こうさてん", "しんごう",

    # --- Mini Conversation 1 ---
    "すみません、ゆうびんきょくはどこですか？",
    "あのこうさてんをひだりにまがってください。",

    # --- Mini Conversation 2 ---
    "ぎんこうはどうやっていきますか？",
    "まっすぐいって、しんごうをみぎにまがってください。",

    # --- Mini Conversation 3 ---
    "としょかんは？",
    "コンビニのとなりにあります。",

    # --- Bonus Phrases ---
    "すみません、ちょっときいてもいいですか？",
    "わかりました。",
    "わかりません。",
    "ありがとうございます！",
    "〜にいきたいです。"
]

# ✅ Generate and display audio for each phrase
for i, sentence in enumerate(day13_sentences):
    clean_text = sentence.replace("〜", "")  # Remove placeholder
    print(f"Sentence {i+1}: {clean_text}")
    tts = gTTS(text=clean_text, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
