## ✅ Goal for Today
Learn & practice **basic polite expressions**. Be able to greet, thank, and make simple polite requests. Practice **listening & shadowing** for natural pronunciation.

---

## ✅ Grammar Review (Polite Basics)
- **です / ます forms** – Standard polite speech  
  - わたしはキムです。(**I am Kim.**)  
  - にほんごがすこしわかります。(**I understand a little Japanese.**)  
- **ください (Please...)** – Polite requests  
  - ゆっくり話してください。(**Please speak slowly.**)  
- **〜たいです (I want to...)** – Expressing desire  
  - コーヒーをのみたいです。(**I want to drink coffee.**)

---

## ✅ Core Greetings (Learn & Repeat)
| Japanese | Meaning | When to Use |
|----------|---------|------------|
| おはようございます | Good morning | Morning (formal) |
| こんにちは | Hello/Good afternoon | Daytime |
| こんばんは | Good evening | Evening |
| はじめまして | Nice to meet you (first time) | Introductions |
| よろしくお願いします | Please treat me well / Nice to meet you | After introduction |
| ありがとうございます | Thank you (polite) | Anytime |
| すみません | Excuse me / Sorry | Getting attention / apologizing |
| ごめんなさい | I’m sorry | Apologizing (casual but polite enough) |

---

## ✅ Useful Survival Phrases
- はい、わかりました。(**Yes, I understand.**)  
- もういちどおねがいします。(**One more time, please.**)  
- ゆっくり話してください。(**Please speak slowly.**)  

---

## ✅ Mini-Dialogues (Practice Speaking)
**Dialogue 1 – First Meeting**  
A: はじめまして。キムです。  
B: はじめまして。サナエです。よろしくおねがいします。  
A: よろしくおねがいします。  

**Dialogue 2 – Asking to Repeat**  
A: すみません、もういちどおねがいします。  
B: はい、ゆっくり話しますね。  

**Dialogue 3 – Daily Greeting**  
A: おはようございます！  
B: おはようございます、キムさん。

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 1: Greetings, Polite Phrases, and Mini Dialogues
day1_sentences = [
    # --- Core Greetings ---
    "おはようございます。",
    "こんにちは。",
    "こんばんは。",
    "はじめまして。よろしくおねがいします。",
    "ありがとうございます。",
    "すみません。",
    "ごめんなさい。",

    # --- Survival Phrases ---
    "はい、わかりました。",
    "もういちどおねがいします。",
    "ゆっくり話してください。",

    # --- Mini Dialogue 1 – First Meeting ---
    "はじめまして。キムです。",
    "はじめまして。サナエです。よろしくおねがいします。",
    "よろしくおねがいします。",

    # --- Mini Dialogue 2 – Asking to Repeat ---
    "すみません、もういちどおねがいします。",
    "はい、ゆっくり話しますね。",

    # --- Mini Dialogue 3 – Daily Greeting ---
    "おはようございます！",
    "おはようございます、キムさん。"
]

# ✅ Loop through and generate a play button for each sentence
for i, sentence in enumerate(day1_sentences):
    print(f"Sentence {i+1}: {sentence}")
    tts = gTTS(text=sentence, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
