## ✅ 2. Expressing Likes & Dislikes

| Japanese | English |
|----------|---------|
| 〜がすきです。 | I like ~. |
| 〜がきらいです。 | I dislike ~. |
| 〜がだいすきです。 | I love ~. |
| 〜がにがてです。 | I’m not good with ~. |
| なにがすきですか？ | What do you like? |
| すきな〜はなんですか？ | What kind of ~ do you like? |

> Use **が** after the thing you like/dislike.

---

## ✅ 3. Talking About Hobbies

| Japanese | English |
|----------|---------|
| しゅみは〜です。 | My hobby is ~. |
| ひまなとき、〜をします。 | In my free time, I ~. |
| よく〜をします。 | I often ~. |
| たまに〜をします。 | I sometimes ~. |
| 〜にきょうみがあります。 | I’m interested in ~. |

---

## ✅ 4. Example Sentences

- サッカーがすきです。  
→ I like soccer.

- おんがくをきくのがすきです。  
→ I like listening to music.

- ホラーえいががにがてです。  
→ I’m not good with horror movies.

- しゅみはどくしょです。  
→ My hobby is reading.

- にほんごにきょうみがあります。  
→ I’m interested in Japanese.

## ✅ 6. Mini Conversations

**1. Asking what someone likes**  
A: なにがすきですか？  
B: サッカーがすきです。あなたは？  
A: おんがくがすきです。

**2. Talking about hobbies**  
A: しゅみはなんですか？  
B: どくしょです。あなたは？  
A: アニメをみるのがすきです。

**3. Expressing dislike politely**  
A: ホラーえいががすきですか？  
B: ちょっと…にがてです。

## ✅ 11. Cultural Tip

In Japanese, even if you dislike something, it’s common to soften your tone using **ちょっと...** or **にがて** instead of saying **きらい** directly — it’s more polite and humble.

```python
%reset -f

from gtts import gTTS
from IPython.display import display, Audio
import io

# ✅ Day 12 – Likes, Dislikes, and Hobbies
day12_sentences = [
    # --- Expressing likes/dislikes ---
    "〜がすきです。",
    "〜がきらいです。",
    "〜がだいすきです。",
    "〜がにがてです。",
    "なにがすきですか？",
    "すきな〜はなんですか？",

    # --- Talking about hobbies ---
    "しゅみは〜です。",
    "ひまなとき、〜をします。",
    "よく〜をします。",
    "たまに〜をします。",
    "〜にきょうみがあります。",

    # --- Example Sentences ---
    "サッカーがすきです。",
    "おんがくをきくのがすきです。",
    "ホラーえいががにがてです。",
    "しゅみはどくしょです。",
    "にほんごにきょうみがあります。",

    # --- Hobbies & Interests Vocabulary ---
    "えいが", "おんがく", "どくしょ", "りょこう", "スポーツ", "サッカー", "やきゅう",
    "アニメ", "えをかくこと", "ギター", "りょうり", "にほんご",

    # --- Mini Conversation 1 ---
    "なにがすきですか？",
    "サッカーがすきです。あなたは？",
    "おんがくがすきです。",

    # --- Mini Conversation 2 ---
    "しゅみはなんですか？",
    "どくしょです。あなたは？",
    "アニメをみるのがすきです。",

    # --- Mini Conversation 3 ---
    "ホラーえいががすきですか？",
    "ちょっと…にがてです。"
]

# ✅ Generate audio and display play buttons
for i, sentence in enumerate(day12_sentences):
    clean_text = sentence.replace("〜", "")  # Remove placeholder
    print(f"Sentence {i+1}: {clean_text}")
    tts = gTTS(text=clean_text, lang='ja')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    display(Audio(mp3_fp.read(), autoplay=False))
```
