## ✅ 3. Useful Shopping Vocabulary

| Japanese | Meaning |
|----------|---------|
| メニュー | menu |
| これ / それ / あれ | this / that / that over there |
| いくらですか？ | How much is it? |
| 〜をください | Please give me ~ |
| ありますか？ | Do you have ~? |
| ちょっとたかいです | It's a little expensive |
| じゃあ、けっこうです | Then, no thank you |

---

## ✅ 4. Grammar & Phrase Patterns

### **1) Pointing to Items**
- これをください。(**I'll take this, please.**)  
- それはちょっとたかいです。(**That’s a little expensive.**)

### **2) Asking for Availability**
- メニューはありますか？(**Do you have a menu?**)  
- おちゃはありますか？(**Do you have tea?**)

### **3) Asking for Price**
- これはいくらですか？(**How much is this?**)  
- コーヒーはいくらですか？(**How much is the coffee?**)

---

## ✅ 5. Mini-Dialogues (Practice Speaking)

**Dialogue 1 – Ordering Food**  
A: メニューをください。  
B: はい、どうぞ。  
A: このカレーをください。

**Dialogue 2 – Asking Price**  
A: これはいくらですか？  
B: 500えんです。

**Dialogue 3 – Polite Decline**  
A: このケーキはちょっとたかいですね。  
B: はい、すこし。  
A: じゃあ、けっこうです。
