## A5.4.1 Motivation: Why Symmetries Matter

In particle physics, symmetries are more than just elegant patterns—they are **fundamental to the structure of physical laws**. Through **Noether’s theorem**, we know that every continuous symmetry in nature corresponds to a **conserved quantity**.

Understanding symmetries allows us to:
- Predict which particle processes are allowed or forbidden
- Identify conservation laws that apply at every interaction vertex
- Understand the underlying structure of the Standard Model

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A5.4.2 Noether’s Theorem (Qualitative)

Noether’s theorem states that:

> For every continuous symmetry of the laws of physics, there exists a corresponding conserved quantity.

Examples:
- **Time translation symmetry** → **Energy conservation**
- **Space translation symmetry** → **Momentum conservation**
- **Rotational symmetry** → **Angular momentum conservation**

In quantum field theory, internal symmetries (e.g., gauge symmetries) lead to conservation of quantities like **electric charge** and **lepton number**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A5.4.3 Key Conservation Laws in the Standard Model

At every interaction (i.e., each vertex in a Feynman diagram), certain quantities must be conserved. These conservation laws come from internal symmetries of the Standard Model and are essential for determining which processes are physically allowed.

Below we describe the most important conservation laws and the particles they apply to.

---

### Electric Charge Conservation

This is one of the most fundamental and intuitive conservation laws.

- Every particle has an **electric charge**, measured in units of the proton charge:
  - Electron ($e^-$): $-1$
  - Positron ($e^+$): $+1$ (antiparticle of the electron)
  - Muon ($\mu^-$): $-1$
  - Quarks: fractional charges
    - Up, charm, top ($u, c, t$): $+2/3$
    - Down, strange, bottom ($d, s, b$): $-1/3$
  - Photon ($\gamma$): $0$

At any interaction vertex:
$$
\text{Total charge before} = \text{Total charge after}
$$

Example:
$$
\mu^- \rightarrow e^- + \bar{\nu}_e + \nu_\mu
$$
$$
(-1) = (-1) + 0 + 0 \quad \text{(✓)}
$$

---

### Lepton Number Conservation

Leptons include:
- **Electron family**: $e^-$, $\nu_e$
- **Muon family**: $\mu^-$, $\nu_\mu$
- **Tau family**: $\tau^-$, $\nu_\tau$
- Each has a corresponding **antiparticle** with opposite lepton number

We define:
- $L_e = +1$ for $e^-$ and $\nu_e$  
- $L_\mu = +1$ for $\mu^-$ and $\nu_\mu$  
- $L_\tau = +1$ for $\tau^-$ and $\nu_\tau$  
- Their **antiparticles** carry $-1$  
- All other particles (quarks, photons, etc.) have **lepton number = 0**

These lepton numbers are **conserved separately** in the Standard Model:
- You cannot convert a muon into an electron **without also producing neutrinos to balance flavor**.

Example:
$$
\mu^- \rightarrow e^- + \bar{\nu}_e + \nu_\mu
$$

Lepton flavor check:
- Muon number: $L_\mu = 1 \rightarrow 0 + 0 + 1$ ✓  
- Electron number: $L_e = 0 \rightarrow 1 + (-1) + 0$ ✓

---

### Baryon Number Conservation

Baryons are particles made of **three quarks**:
- Proton ($uud$): $B = 1$
- Neutron ($udd$): $B = 1$
- Antibaryons (e.g., antiproton): $B = -1$
- Mesons (quark-antiquark pairs) and leptons: $B = 0$

Each **quark** has baryon number $+1/3$, and each **antiquark** has $-1/3$.

Baryon number is conserved in all known interactions:
- You cannot turn a neutron into an electron without also producing a proton to preserve baryon number.

Example: neutron beta decay
$$
n \rightarrow p + e^- + \bar{\nu}_e 
$$
$$
B: 1 \rightarrow 1 + 0 + 0 \quad \text{(✓)}
$$

---

> These three conservation laws—electric charge, lepton number, and baryon number—are obeyed at **every vertex** of a valid Feynman diagram in the Standard Model.

We’ll now apply these rules to determine which decays are allowed and which are forbidden.

## A5.4.4 Energy and Momentum Conservation

These are always conserved in every particle process. In Feynman diagrams:
- Time and space translation symmetry imply:
  - **Energy** is conserved at each vertex
  - **Momentum** is conserved at each vertex

While energy and momentum are not obvious in simple schematic diagrams, they are enforced mathematically in the calculation of scattering amplitudes.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A5.4.5 Example

### Forbidden decay

$$
\mu^- \rightarrow e^- + \gamma
$$

Check:
- Charge: $-1 \rightarrow -1 + 0$ ✓  
- Lepton number:
  - $L_\mu: 1 \rightarrow 0$ ✗  
  - $L_e: 0 \rightarrow 1$ ✗  

❌ This decay violates **lepton flavor conservation** and is **forbidden** in the Standard Model.

(Note: this decay may occur in physics **beyond the Standard Model** with neutrino mixing or lepton flavor violation, but it is highly suppressed.)

### Notes:
- Lepton number is **flavor-specific**: $L_e$, $L_\mu$, and $L_\tau$ are tracked separately.
- Baryon number is conserved in **all known interactions**.
- Only quarks and baryons carry nonzero baryon number.
- Photons and gauge bosons do **not** carry lepton or baryon number.

## A5.4.7 Summary

- Symmetries give rise to **conservation laws** via Noether’s theorem.
- The Standard Model conserves:
  - Electric charge
  - Lepton flavor numbers
  - Baryon number
  - Energy and momentum

These conservation laws allow us to:
- Interpret Feynman diagrams
- Identify valid decay and scattering processes
- Understand why some processes are forbidden
