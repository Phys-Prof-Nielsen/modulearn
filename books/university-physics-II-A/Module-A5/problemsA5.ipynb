Justify your answer based on the number and types of quarks they contain.

| Symbol       | Name                | Quark Content      |

| $n$          | Neutron             | $udd$              |

## Problem 2: Kaon$^+$  Decay

Determine whether the following decay is allowed in the Standard Model:

$$
K^+ \rightarrow \mu^+ + \nu_\mu
$$

Check for conservation of:

- Electric charge  
- Lepton number  
- Baryon number  


```{admonition} Problem 2: Solution
:class: tip, dropdown

We examine each conservation law:

---

**1. Electric charge:**

- Initial: $K^+$ has charge $+1$
- Final: $\mu^+$ is $+1$, $\nu_\mu$ is neutral → sum = $+1$

✅ Charge is conserved.

---

**2. Lepton number:**

- Initial: $K^+$ is a meson → no lepton content → $L = 0$
- Final:
  - $\mu^+$ is an antilepton → $L = -1$
  - $\nu_\mu$ is a lepton → $L = +1$
  - Total $L = -1 + 1 = 0$

✅ Lepton number is conserved.

---

**3. Baryon number:**

- Initial: $K^+$ is a meson → $B = 0$
- Final:
  - $\mu^+$ and $\nu_\mu$ are leptons → $B = 0$

✅ Baryon number is conserved.

---

✅ **Conclusion**:

The decay

$$
K^+ \rightarrow \mu^+ + \nu_\mu
$$

is **allowed** in the Standard Model and is one of the dominant decay channels of the charged kaon, mediated by the **weak interaction** (via $W^+$ exchange).

```

## Problem 3: Tau$^-$ decay

Is the following decay allowed in the Standard Model?

$$
\tau^- \rightarrow \mu^- + \bar{\nu}_\mu + \nu_\tau
$$

Analyze the decay using:

- Electric charge  
- Lepton number and flavor  
- Baryon number  


```{admonition} Problem 3: Solution
:class: tip, dropdown

**1. Electric charge:**

- Initial: $\tau^-$ has charge $-1$
- Final: $\mu^-$ is $-1$, the neutrinos are neutral → total = $-1$

✅ Charge is conserved.

---

**2. Lepton number and flavor:**

- Initial: $\tau^-$ → $L_\tau = +1$, $L_\mu = 0$, $L_e = 0$
- Final:
  - $\mu^-$ → $L_\mu = +1$
  - $\bar{\nu}_\mu$ → $L_\mu = -1$
  - $\nu_\tau$ → $L_\tau = +1$

Check each flavor:

- $L_\tau: 1 \rightarrow 0 + 0 + 1$ → ✓  
- $L_\mu: 0 \rightarrow 1 + (-1) + 0$ → ✓  
- $L_e: 0 \rightarrow 0$ → ✓

✅ Lepton flavor is conserved.

---

**3. Baryon number:**

- All particles are leptons → $B = 0$ throughout.

✅ Baryon number is conserved.

---

✅ **Conclusion**:

The decay

$$
\tau^- \rightarrow \mu^- + \bar{\nu}_\mu + \nu_\tau
$$

is **fully allowed** in the Standard Model via the **weak interaction**, and is a common decay channel of the tau lepton.

```

## Problem 4: Quark-Antiquark Annihilation

An up quark ($u$) and an anti-down quark ($\bar{d}$) interact via the weak force ($W^+$ boson) and produce a lepton pair. One of the final particles is known to be a **positron ($e^+$)** (also known as an anti-electron).

**Tasks:**

1. Using **charge conservation**, determine the charge of the second final-state particle.
2. Using **lepton number conservation**, identify whether the second particle is an **electron neutrino** or an **antineutrino**.
4. Based on your answer above, **draw the Feynman diagram** for this process, labeling all particles and the intermediate boson.


```{admonition} Problem 4: Solution
:class: tip, dropdown

**1. Charge Conservation**:
- Initial state: $u$ (+2/3) and $\bar{d}$ (+1/3) → total charge = +1
- One final-state particle: $e^+$ (charge +1)
- The second particle must be neutral → charge = 0

**2. Lepton Number Conservation**:
- $e^+$ has lepton number −1
- To conserve total lepton number (initial was 0), the second particle must have lepton number +1 ⇒ **$\nu_e$**

**4. Feynman Diagram**:

You should now draw a Feynman diagram with:
- Incoming $u$ and $\bar{d}$
- A $W^+$ boson as the mediator (dashed line)
- Outgoing $e^+$ and $\nu_e$

This is a **charged-current weak interaction**.

Diagram is below.

```

````python
import matplotlib.pyplot as plt

def draw_corrected_quark_feynman_diagram(ax):
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.axis('off')

    # Vertex position
    vertex_x, vertex_y = 3.5, 5

    # Incoming up quark
    ax.annotate("", xy=(vertex_x, vertex_y), xytext=(0.5, 9),
                arrowprops=dict(arrowstyle="->", color="blue"))
    ax.text(0.2, 9.2, r'$u$', fontsize=14)

    # Incoming anti-down quark
    ax.annotate("", xy=(vertex_x, vertex_y), xytext=(0.5, 1),
                arrowprops=dict(arrowstyle="<-", color="red"))
    ax.text(0.2, 0.5, r'$\bar{d}$', fontsize=14)

    # W+ mediator (dashed line)
    ax.annotate("", xy=(7, 5), xytext=(vertex_x, vertex_y),
                arrowprops=dict(arrowstyle="->", linestyle="dashed", color="green"))
    ax.text(5.1, 5.4, r'$W^+$', fontsize=14, color="green")

    # Outgoing neutrino
    ax.annotate("", xy=(9.5, 8), xytext=(7, 5),
                arrowprops=dict(arrowstyle="->", color="black"))
    ax.text(9.6, 7.9, r'$\nu_e$', fontsize=14)

    # Outgoing positron
    ax.annotate("", xy=(9.5, 1), xytext=(7, 5),
                arrowprops=dict(arrowstyle="<-", color="red"))
    ax.text(9.6, 0.9, r'$e^+$', fontsize=14)

# Plot corrected diagram
fig, ax = plt.subplots(figsize=(9, 6))
draw_corrected_quark_feynman_diagram(ax)
plt.title("Corrected Quark-Antiquark Annihilation via $W^+$")
plt.show()
````

## Problem 5: Feynman Diagram for Neutron Beta Decay

A free neutron decays via the weak interaction according to the reaction:

$$
n \rightarrow p + e^- + \bar{\nu}_e
$$

This decay occurs at the **quark level**, where a neutron (made of $udd$ quarks) transforms into a proton (made of $uud$ quarks) through a $W^-$ boson.

**Tasks:**

1. Identify which quark in the neutron changes flavor, and what it changes into.
2. Use **charge conservation** and **lepton number conservation** to explain why the final products must include an **electron** and an **electron antineutrino**.
3. Draw the **Feynman diagram** for this process at the quark level, labeling:
   - The initial and final **quarks**
   - The **intermediate boson**
   - The **electron** and **antineutrino**
   - Time direction (left to right)


```{admonition} Problem 5: Solution
:class: tip, dropdown

**1. Quark Flavor Change**:
- A **neutron** is made of $udd$
- A **proton** is $uud$
- So:  
  One **down quark** changes into an **up quark**

**2. Conservation Laws**:

- **Charge**:
  - Down quark: $−\tfrac{1}{3}$
  - Up quark: $+\tfrac{2}{3}$
  - Change: $+\tfrac{1}{3} + \tfrac{2}{3} = +1$
  - To conserve charge, the emitted particle must carry **−1** → an **electron**. It could in principle also be a muon, however, the energy difference between a neutron and proton does not allow for a muon.

- **Lepton number**:
  - Initially: 0 (no leptons)
  - Electron: lepton number +1  
  - To balance: must emit a particle with lepton number −1 → **electron antineutrino**, $\bar{\nu}_e$

✅ So the decay is:
$$
d \rightarrow u + W^- \quad \text{followed by} \quad W^- \rightarrow e^- + \bar{\nu}_e
$$

**3. Feynman Diagram**:

- A down quark line emits a $W^-$ boson and turns into an up quark
- The $W^-$ decays into an electron and an electron antineutrino
- Time flows from **left to right**

Diagram is below.

```

````python
from matplotlib import pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(12, 6))
ax.set_xlim(0, 12)
ax.set_ylim(0, 6)
ax.axis('off')

# Spectator quarks at the top
ax.plot([1, 8], [5.5, 5.5], 'k')
ax.annotate('', xy=(8, 5.5), xytext=(1, 5.5),
            arrowprops=dict(arrowstyle='->', color='black'))
ax.text(0.3, 5.3, 'u (spectator)', fontsize=12, va='center')
ax.text(8.6, 5.3, 'u', fontsize=12, va='center')

ax.plot([1, 8], [4.5, 4.5], 'k')
ax.annotate('', xy=(8, 4.5), xytext=(1, 4.5),
            arrowprops=dict(arrowstyle='->', color='black'))
ax.text(0.3, 4.3, 'd (spectator)', fontsize=12, va='center')
ax.text(8.6, 4.3, 'd', fontsize=12, va='center')

# Decaying down quark
ax.plot([1, 3], [3.5, 3.5], 'k')
ax.annotate('', xy=(3, 3.5), xytext=(1, 3.5),
            arrowprops=dict(arrowstyle='->', color='black'))
ax.text(0.3, 3.3, 'd (decays)', fontsize=12, va='center')

# W- emission
ax.plot([3, 4.5], [3.5, 2], linestyle='dashed', color='blue')
ax.text(4, 2.6, 'W⁻', fontsize=12, color='blue')

# Final up quark from decay
ax.plot([4.5, 8], [3.5, 3.5], 'k')
ax.annotate('', xy=(8, 3.5), xytext=(4.5, 3.5),
            arrowprops=dict(arrowstyle='->', color='black'))
ax.text(8.6, 3.3, 'u', fontsize=12, va='center')

# W- decay products: electron and antineutrino
ax.plot([4.5, 6.5], [2, 1], 'k')
ax.annotate('', xy=(6.5, 1), xytext=(4.5, 2),
            arrowprops=dict(arrowstyle='->', color='black'))
ax.text(6.7, 1, 'e⁻', fontsize=12, va='center')

ax.plot([4.5, 6.5], [2, 0], 'k')
ax.annotate('', xy=(6.5, 0), xytext=(4.5, 2),
            arrowprops=dict(arrowstyle='<-', color='black'))
ax.text(6.7, 0, r'$\bar{\nu}_e$', fontsize=12, va='center')

# Interaction vertices
ax.plot(3, 3.5, 'ko')      # d → u + W⁻
ax.plot(4.5, 2, 'ko')      # W⁻ → e⁻ + antinu

# Neutron and proton labels
ax.text(2, 5.7, 'Neutron (udd)', fontsize=12, ha='center')
ax.text(7, 5.7, 'Proton (uud)', fontsize=12, ha='center')

# Time arrow
ax.arrow(1, -0.3, 2, 0, head_width=0.2, head_length=0.3, fc='gray', ec='gray')
ax.text(3.2, -0.3, 'time', fontsize=12, va='center')

plt.title("Neutron Beta Decay", fontsize=14)
plt.tight_layout()
plt.show()
````
