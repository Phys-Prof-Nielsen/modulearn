## A1.4.1 Introduction and Motivation

Electromagnetic (EM) waves **carry energy and momentum**.  
To understand energy flow and radiation pressure, we need to quantify:

1. **Energy density** stored in electric and magnetic fields.
2. **Energy flux** (rate of energy transport per unit area) → the **Poynting vector**.

This lecture derives these concepts directly from Maxwell’s equations.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.4.2 Energy Density in Electromagnetic Fields

### Work and Energy in the Electric Field

The energy stored in the electric field comes from the **work required to build it up** gradually.

For example, in a capacitor:

- Capacitance: $C = \varepsilon_0 A / d$
- Stored energy: $U = \frac12 C V^2$

Using $E = V/d$ and volume $= Ad$:

$$
u_E = \frac{U}{\text{Volume}} = \frac12 \varepsilon_0 E^2
$$

✅ **Why the factor \(1/2\)?**  
As the capacitor charges, $E$ grows from $0$ to $E_{\text{final}}$, so the **average field during charging** is $E_{\text{final}}/2$.  
The work done is proportional to $qV$, and since $V\propto q$, $W=\frac12 qV$.

---

### Energy in the Magnetic Field

Similarly, the energy stored in the magnetic field comes from building up the current in an inductor:

$$
u_B = \frac12 \frac{B^2}{\mu_0}
$$

This also originates from the gradual buildup: $U = \frac12 L I^2$ (we have not covered this, so we hit the **I believe** button).

---

### Total Electromagnetic Energy Density

In general:

$$
u = u_E + u_B = \frac12 \varepsilon_0 E^2 + \frac12 \frac{B^2}{\mu_0}
$$

---

### Special Case: Plane Electromagnetic Waves

For a plane wave in free space, $B = E/c$ and $c^2 = 1/(\mu_0 \varepsilon_0)$, giving:

$$
u_B = \frac12 \frac{B^2}{\mu_0} = \frac12 \varepsilon_0 E^2
$$

Thus, **electric and magnetic fields store equal amounts of energy on average**, and the **total energy density** becomes:

$$
u = u_E + u_B = \varepsilon_0 E^2
$$

**Key Takeaway:**  
- Each field contributes half the total energy density.

## A1.4.3 Energy Transport – The Poynting Vector

### Definition

The **Poynting vector** gives the rate of energy transfer per unit area:

$$
\mathbf{S} = \frac{1}{\mu_0} (\mathbf{E}\times \mathbf{B})
$$

- Units: W/m² (power per unit area)
- Direction: **energy flow direction** (same as wave propagation $\mathbf{k}$)
- Magnitude: **instantaneous energy flux**

For plane waves:

$$
|\mathbf{S}| = \frac{1}{\mu_0} E B = \frac{E^2}{\mu_0 c} = c \varepsilon_0 E^2
$$

(using $B = E/c$).

---

### Average Intensity

Because $E$ oscillates, we use a **time average** over one cycle:

$$
\langle S \rangle = I = c \varepsilon_0 E_{\text{rms}}^2
$$

where:

- $I$ = intensity (W/m²)
- $E_{\text{rms}} = E_0/\sqrt{2}$

Thus:

$$
I = \frac{1}{2} c \varepsilon_0 E_0^2 = \frac{1}{2} \frac{c}{\mu_0} B_0^2
$$

## A1.4.4 Momentum and Radiation Pressure

EM waves carry **momentum** as well:

- **Momentum density**: $\mathbf{g} = \frac{\mathbf{S}}{c^2}$ (we will see in next module why that is)  
- **Radiation pressure** on an absorbing surface:
   
$$
P = \frac{I}{c}
$$

For perfect reflection: $P = \frac{2I}{c}$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.4.5 Physical Picture

✅ **Direction:**  
$\mathbf{E}$, $\mathbf{B}$, and $\mathbf{S}$ form a **right-handed triad**:

$$
\mathbf{S} \propto \mathbf{E} \times \mathbf{B}
$$

✅ **Energy Oscillation:**  
At any instant, $u_E$ and $u_B$ oscillate, exchanging energy, but **on average** they are equal.

✅ **Light Intensity:**  
The brightness of light corresponds directly to $I = \langle S \rangle$.

## A1.4.6 Summary Table

| Quantity | Expression | Interpretation |
|----------|------------|----------------|
| **Electric Energy Density** | $u_E = \frac12 \varepsilon_0 E^2$ | Energy stored in $E$ field |
| **Magnetic Energy Density** | $u_B = \frac{1}{2\mu_0} B^2$ | Energy stored in $B$ field |
| **Total Energy Density** | $u = u_E + u_B = \varepsilon_0 E^2$ (plane wave) | Total EM energy per unit volume |
| **Poynting Vector** | $\mathbf{S} = \frac{1}{\mu_0}(\mathbf{E}\times\mathbf{B})$ | Energy flux (W/m²), direction of propagation |
| **Average Intensity** | $I = \langle S \rangle = \frac12 c \varepsilon_0 E_0^2$ | Average energy flow per unit area |
