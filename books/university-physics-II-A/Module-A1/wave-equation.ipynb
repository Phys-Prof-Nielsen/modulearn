# A1.2 Deriving the Electromagnetic Wave Equation from Maxwell’s Equations
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.2.1 Introduction and Motivation

Maxwell’s greatest achievement was realizing that the modified Ampère’s law (with displacement current) implied that **electric and magnetic fields can sustain each other** in space, even without charges or currents.  

This leads naturally to the prediction of **electromagnetic waves**—light being one example.

We will now derive the **wave equation** for electromagnetic fields in free space.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.2.2 Starting Point: Maxwell’s Equations in Vacuum

In a charge-free, current-free region ($\rho = 0$, $\mathbf{J} = 0$):

**Gauss’s Law (Electric):**  

$$
\nabla \cdot \mathbf{E} = 0
$$

**Gauss’s Law (Magnetic):** 

$$
\nabla \cdot \mathbf{B} = 0
$$

**Faraday’s Law:**  

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}
$$

**Ampère–Maxwell Law:** 

$$
\nabla \times \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}
$$

These are valid in vacuum.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.2.3 Vector Identity We Need

A useful vector identity:

$$
\nabla \times (\nabla \times \mathbf{F}) = \nabla(\nabla \cdot \mathbf{F}) - \nabla^2 \mathbf{F}
$$

where $ \nabla^2 = \nabla \cdot \nabla $ is the Laplacian operator.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.2.4 Deriving the Wave Equation for the Electric Field

### Take the Curl of Faraday’s Law

Start with:

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}
$$

Take the curl on both sides:

$$
\nabla \times (\nabla \times \mathbf{E}) = -\frac{\partial}{\partial t} (\nabla \times \mathbf{B})
$$

---

### Substitute Ampère–Maxwell Law

From Ampère–Maxwell law:

$$
\nabla \times \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}
$$

So:

$$
\nabla \times (\nabla \times \mathbf{E}) = -\mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{E}}{\partial t^2}
$$

---

### Apply the Vector Identity

Use:

$$
\nabla \times (\nabla \times \mathbf{E}) = \nabla(\nabla \cdot \mathbf{E}) - \nabla^2 \mathbf{E}
$$

But in free space $ \nabla \cdot \mathbf{E} = 0$, so:

$$
-\nabla^2 \mathbf{E} = -\mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{E}}{\partial t^2}
$$

Thus:

$$
\nabla^2 \mathbf{E} = \mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{E}}{\partial t^2}
$$

This is the **wave equation** for the electric field.
