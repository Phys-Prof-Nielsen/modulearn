<h2>Problem 1</h2>
    </header>

Show that the magnetic field at a distance $r>R$ from the axis of two circular parallel plates of radii $R$, produced by placing charge $Q(t)$ on the plates is

$$ B_{ind} = \frac{\mu_0}{2\pi r}\frac{dQ}{dt} $$

[This problem is a slightly modified version from OpenStax. Access for free here](https://openstax.org/books/university-physics-volume-2/pages/16-problems)

```{admonition} Problem 1: Solution
:class: tip, dropdown

**Ampère–Maxwell law**:

$$
\oint \mathbf{B}\cdot d\mathbf{l} = \mu_0 \left(I_{\text{cond}} + \varepsilon_0 \frac{d\Phi_E}{dt}\right)
$$

Inside the capacitor, there is **no conduction current** ($I_{\text{cond}}=0$), only **displacement current**:

$$
I_d = \varepsilon_0 \frac{d\Phi_E}{dt}
$$

So:

$$
\oint \mathbf{B}\cdot d\mathbf{l} = \mu_0 \varepsilon_0 \frac{d\Phi_E}{dt}
$$

---

**Choose Symmetry and Path**

- Symmetry: The magnetic field is azimuthal ($\mathbf{B}$ circles the axis).  
- Choose a circular Amperian loop of radius $r$ concentric with the plates.

Thus:

$$
B(2\pi r) = \mu_0 \varepsilon_0 \frac{d\Phi_E}{dt}
$$

---

**Electric Flux Through the Amperian Surface**

The electric field between the plates is uniform:

$$
E = \frac{Q}{\varepsilon_0 A}, \quad A = \pi R^2
$$

The flux through the surface of radius $r$ (the Amperian surface) is:

$$
\Phi_E = EA_{encl} = E(\pi r^2) = \frac{Q}{\varepsilon_0 \pi R^2}(\pi r^2)
$$

Differentiate:

$$
\frac{d\Phi_E}{dt} = \frac{1}{\varepsilon_0}\frac{dQ}{dt}\frac{r^2}{R^2} = \frac{Q'(t) r^2}{\varepsilon_0 R^2}
$$

---

**Substitute into Ampère–Maxwell Law**

Ampère–Maxwell:

$$
B(2\pi r) = \mu_0 \varepsilon_0 \frac{d\Phi_E}{dt}
$$

Substitute:

$$
B(2\pi r) = \mu_0 \varepsilon_0 \left(\frac{Q'(t) r^2}{\varepsilon_0 R^2}\right)
$$

Simplify:

$$
B = \frac{\mu_0 Q'(t)\, r}{2\pi R^2}
$$

✅ **As required.**

---

**Comments**

- **Inside the plates ($r < R$):** Field increases linearly with $r$.  
- **Outside the plates ($r > R$):** The enclosed displacement current is capped at $I_d = Q'(t)$, so:

$$
B = \frac{\mu_0 Q'(t)}{2\pi r}
$$

(similar to a current-carrying wire).
```

## Problem 4 – Displacement Current in a Charging Capacitor

A parallel-plate capacitor with plate area  
$A = 0.01 \, \text{m}^2$ and plate separation $d = 2.0 \, \text{mm}$ is being charged by a current of  
$I = 2.0 \, \text{A}$.  

**(a)** What is the rate of change of the electric field between the plates?  
**(b)** What is the displacement current between the plates?  
**(c)** What magnetic field is produced at a distance $r = 3.0 \, \text{cm}$ from the central axis between the plates (assume $r$ is much less than the plate radius)?

```{admonition} Problem 4: Solution
:class: tip, dropdown

**(a) Rate of change of the electric field**

The displacement current is related to the rate of change of electric flux:

$$
I = \varepsilon_0 \frac{d\Phi_E}{dt} = \varepsilon_0 A \frac{dE}{dt}
$$

Solve for $\frac{dE}{dt}$:

$$
\frac{dE}{dt} = \frac{I}{\varepsilon_0 A}
$$

Substitute:

- $I = 2.0 \, \text{A}$  
- $\varepsilon_0 = 8.85\times10^{-12} \, \text{F/m}$  
- $A = 0.01 \, \text{m}^2$

$$
\frac{dE}{dt} = \frac{2.0}{(8.85\times10^{-12})(0.01)} \approx 2.26\times10^{13}\,\text{V/(m·s)}
$$

---

**(b) Displacement current**

By definition:

$$
I_d = \varepsilon_0 A \frac{dE}{dt} = I
$$

So:

$$
I_d = 2.0\,\text{A}
$$

(consistent—Maxwell’s correction ensures continuity of current.)

---

**(c) Magnetic field at $r=0.03 \, \text{m}$**

Use **Ampère–Maxwell law** in integral form:

$$
\oint \mathbf{B}\cdot d\mathbf{l} = \mu_0 I_d^{\text{(enc)}}
$$

For a circular path of radius $r$:

$$
B(2\pi r) = \mu_0 I_d^{\text{(enc)}} \quad \Rightarrow \quad B = \frac{\mu_0 I_d^{\text{(enc)}}}{2\pi r}
$$

But only a fraction of $I_d$ is enclosed:

$$
I_d^{\text{(enc)}} = I_d\frac{r^2}{R^2} \quad (\text{uniform } E)
$$

Plate radius:

$$
R = \sqrt{\frac{A}{\pi}} = \sqrt{\frac{0.01}{\pi}} \approx 0.0564\,\text{m}
$$

Substitute:

- $I_d = 2.0 \, \text{A}$
- $\mu_0 = 4\pi\times10^{-7}$
- $r = 0.03 \, \text{m}$

$$
B = \frac{(4\pi\times10^{-7})(2.0)(0.03^2 / 0.0564^2)}{2\pi(0.03)}
$$

Simplify:

$$
B \approx 2.26\times10^{-6} \, \text{T}
$$

```

## Problem 5 – Electromagnetic Wave from Maxwell’s Equations

An electromagnetic wave in free space has an electric field amplitude  
$E_0 = 200 \, \text{V/m}$.

**(a)** Find the corresponding magnetic field amplitude $B_0$.  
**(b)** Calculate the average energy density of the wave.  
**(c)** Find the average intensity of the wave.  
**(d)** Determine the radiation pressure on a perfectly absorbing surface.

```{admonition} Problem 5: Solution
:class: tip, dropdown

**(a) Magnetic field amplitude**

Use:

$$
B_0 = \frac{E_0}{c}
$$

Substitute:

- $E_0 = 200 \, \text{V/m}$
- $c = 3.00\times10^8\,\text{m/s}$

$$
B_0 = \frac{200}{3.0\times10^8} \approx 6.67\times10^{-7}\,\text{T}
$$

---

**(b) Average energy density**

Total energy density:

$$
u = \varepsilon_0 E^2 = \varepsilon_0 E_{\text{rms}}^2 + \varepsilon_0 E_{\text{rms}}^2 = \varepsilon_0 E_0^2
$$

Actually, the **average** energy density is:

$$
\langle u \rangle = \varepsilon_0 E_{\text{rms}}^2 + \frac{1}{2\mu_0} B_{\text{rms}}^2 = \varepsilon_0 E_{\text{rms}}^2 + \varepsilon_0 E_{\text{rms}}^2 = \varepsilon_0 E_0^2
$$

with $E_{\text{rms}} = E_0/\sqrt{2}$:

$$
\langle u \rangle = \varepsilon_0 \frac{E_0^2}{2} + \frac{1}{2\mu_0}\frac{B_0^2}{2} = \varepsilon_0 E_0^2
$$

For a plane wave: **electric and magnetic energy densities are equal on average**.

Substitute:

- $\varepsilon_0 = 8.85\times10^{-12}$
- $E_0 = 200$

$$
\langle u \rangle = \varepsilon_0 E_0^2 = (8.85\times10^{-12})(200)^2 \approx 3.54\times10^{-7}\,\text{J/m}^3
$$

---

**(c) Average intensity**

$$
I = \langle S \rangle = \frac12 c \varepsilon_0 E_0^2
$$

Substitute:

$$
I = 0.5 (3.0\times10^8)(8.85\times10^{-12})(200)^2 \approx 53.1 \,\text{W/m}^2
$$

---

**(d) Radiation pressure**

For a perfectly absorbing surface:

$$
P = \frac{I}{c} = \frac{53.1}{3.0\times10^8} \approx 1.77\times10^{-7}\,\text{N/m}^2
$$

If perfectly reflecting, $P_{\text{ref}} = 2I/c \approx 3.54\times10^{-7}\,\text{N/m}^2$.

```

## Problem 6 – Verifying a Proposed Wave Solution

A plane electromagnetic wave in free space is described by:

$$
\mathbf{E}(x,t) = E_0 \hat{y}\cos(kx - \omega t)
$$

with $E_0 = 50\,\text{V/m}$.

**(a)** Verify that this electric field satisfies the electromagnetic wave equation:

$$
\nabla^2 \mathbf{E} = \mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{E}}{\partial t^2}
$$

**(b)** Show that $v = \omega/k$ equals $1/\sqrt{\mu_0\varepsilon_0}$.

**(c)** Determine the corresponding magnetic field $\mathbf{B}(x,t)$ (magnitude and direction).


```{admonition} Problem 6: Solution
:class: tip, dropdown

**(a) Verify the wave equation**

The proposed wave:

$$
\mathbf{E}(x,t) = E_0 \hat{y}\cos(kx - \omega t)
$$

Compute:

1. **Spatial Laplacian:**

Only $x$-dependence:

$$
\nabla^2 \mathbf{E} = \frac{\partial^2 \mathbf{E}}{\partial x^2} = -k^2 E_0 \hat{y}\cos(kx - \omega t)
$$

2. **Time derivative:**

$$
\frac{\partial^2 \mathbf{E}}{\partial t^2} = -\omega^2 E_0 \hat{y}\cos(kx - \omega t)
$$

Substitute into the wave equation:

$$
-k^2 E_0\cos(...) = \mu_0 \varepsilon_0 (-\omega^2 E_0\cos(...))
$$

Thus:

$$
k^2 = \mu_0 \varepsilon_0 \omega^2
$$

✅ Satisfied if $\omega/k = 1/\sqrt{\mu_0\varepsilon_0}$.

---

**(b) Wave speed**

From $k^2 = \mu_0\varepsilon_0 \omega^2$:

$$
v = \frac{\omega}{k} = \frac{1}{\sqrt{\mu_0\varepsilon_0}} = c
$$

Thus the wave propagates at the speed of light.

---

**(c) Magnetic field**

Use:

$$
\mathbf{B} = \frac{1}{c} \hat{k} \times \mathbf{E}
$$

- Direction of propagation: $\hat{k} = \hat{x}$  
- $\mathbf{E} \parallel \hat{y}$

Thus:

$$
\mathbf{B}(x,t) = \frac{E_0}{c}\hat{z}\cos(kx - \omega t)
$$

(sign consistent with right-handed rule: $\hat{x} \times \hat{y} = \hat{z}$)

Substitute:

$$
B_0 = \frac{E_0}{c} = \frac{50}{3\times10^8} \approx 1.67\times10^{-7}\,\text{T}
$$

```

## Problem 7 – Determining Wave Parameters

An electromagnetic wave in free space satisfies the wave equation. The electric field is given by:

$$
\mathbf{E}(z,t) = E_0 \hat{x}\cos(kz - \omega t)
$$

with intensity $I = 200\,\text{W/m}^2$.

**(a)** Determine $E_0$.  
**(b)** Determine $B_0$.  
**(c)** Find the wavelength $\lambda$ if the frequency is $f = 6.0\times10^{14}\,\text{Hz}$ (visible light).  
**(d)** State the direction of wave propagation and the orientation of $\mathbf{B}$.

```{admonition} Problem 7: Solution
:class: tip, dropdown

**(a) Electric field amplitude**

Use:

$$
I = \frac12 c \varepsilon_0 E_0^2
$$

Solve for $E_0$:

$$
E_0 = \sqrt{\frac{2I}{c\varepsilon_0}}
$$

Substitute:

- $I = 200$
- $c = 3.0\times10^8$
- $\varepsilon_0 = 8.85\times10^{-12}$

$$
E_0 = \sqrt{\frac{400}{(3.0\times10^8)(8.85\times10^{-12})}} \approx 547\,\text{V/m}
$$

---

**(b) Magnetic field amplitude**

$$
B_0 = \frac{E_0}{c} = \frac{547}{3.0\times10^8} \approx 1.82\times10^{-6}\,\text{T}
$$

---

**(c) Wavelength**

$$
\lambda = \frac{v}{f} = \frac{c}{f} = \frac{3.0\times10^8}{6.0\times10^{14}} \approx 5.0\times10^{-7}\,\text{m} = 500\,\text{nm}
$$

(Visible green light.)

---

**(d) Propagation direction and $\mathbf{B}$ orientation**

- Given $\mathbf{E}(z,t) = E_0 \hat{x}\cos(kz - \omega t)$ → propagates in $+z$ direction.  
- $\mathbf{B}$ must satisfy right-handed rule:

$$
\hat{k} \times \hat{E} = \hat{B} \quad \Rightarrow \quad \hat{z}\times\hat{x} = \hat{y}
$$

So:

$$
\mathbf{B}(z,t) = B_0 \hat{y}\cos(kz - \omega t)
$$

```

## Problem 8 – Energy Transport in an Electromagnetic Wave

A plane electromagnetic wave in free space has an electric field amplitude  
$E_0 = 120\,\text{V/m}$.

**(a)** Find the magnitude of the corresponding Poynting vector as a function of time.  
**(b)** Find the **average intensity** $\langle S\rangle$.  
**(c)** How much energy passes through a $5.0\,\text{cm}^2$ area in $10\,\text{s}$ if the wave is normally incident?

```{admonition} Problem 8: Solution
:class: tip, dropdown

**(a) Instantaneous Poynting vector**

For a plane wave:

$$
S(t) = \frac{1}{\mu_0} E(t)B(t) = \frac{E^2(t)}{\mu_0 c}
$$

Using $E(t) = E_0\cos(\omega t)$ and $B = E/c$:

$$
S(t) = \frac{E_0^2}{\mu_0 c} \cos^2(\omega t) = c \varepsilon_0 E_0^2 \cos^2(\omega t)
$$

Substitute:

- $c = 3.0\times10^8$
- $\varepsilon_0 = 8.85\times10^{-12}$
- $E_0 = 120$

$$
S(t) \approx (3.0\times10^8)(8.85\times10^{-12})(120)^2 \cos^2(\omega t)
$$

Numerically:

$$
S(t) \approx 38.3 \cos^2(\omega t)\,\text{W/m}^2
$$

---

**(b) Average intensity**

Average of $\cos^2$ is $1/2$:

$$
\langle S\rangle = \frac12 c \varepsilon_0 E_0^2 \approx \frac{38.3}{2} \approx 19.2\,\text{W/m}^2
$$

---

### (c) Total energy passing through area

Energy = (intensity)(area)(time):

- $A = 5.0\,\text{cm}^2 = 5.0\times10^{-4}\,\text{m}^2$
- $t = 10\,\text{s}$

$$
E_{\text{total}} = \langle S\rangle A t = 19.2 (5.0\times10^{-4})(10) \approx 0.096\,\text{J}
$$

```

## Problem 9 – Direction and Orientation of the Poynting Vector

A plane electromagnetic wave is described by:

$$
\mathbf{E}(x,t) = E_0 \hat{y}\cos(kx - \omega t), \\
\mathbf{B}(x,t) = B_0 \hat{z}\cos(kx - \omega t)
$$

**(a)** Determine the direction of wave propagation.  
**(b)** Write the Poynting vector $\mathbf{S}(x,t)$.  
**(c)** Determine the average energy flow direction and verify the right-handed orientation of $\mathbf{E},\mathbf{B},\mathbf{S}$.

```{admonition} Problem 9: Solution
:class: tip, dropdown

**(a) Direction of propagation**

Given:

- $\mathbf{E} \parallel \hat{y}$
- $\mathbf{B} \parallel \hat{z}$

Use $\mathbf{S} \propto \mathbf{E}\times \mathbf{B}$:

$$
\hat{y}\times \hat{z} = \hat{x}
$$

Thus the wave propagates in the **+x direction**.

---

**(b) Poynting vector expression**

$$
\mathbf{S}(x,t) = \frac{1}{\mu_0} (\mathbf{E}\times \mathbf{B}) = \frac{1}{\mu_0} E B \, \hat{x}
$$

Substitute $B = E/c$:

$$
\mathbf{S}(x,t) = \frac{E^2}{\mu_0 c} \hat{x} = c \varepsilon_0 E_0^2\cos^2(kx - \omega t)\,\hat{x}
$$

---

**(c) Average energy flow and orientation**

Average:

$$
\langle \mathbf{S}\rangle = \frac12 c \varepsilon_0 E_0^2 \, \hat{x}
$$

✅ **Right-handed orientation:**

- $\mathbf{E}$ along $+y$,  
- $\mathbf{B}$ along $+z$,  
- $\mathbf{S}$ (and $\mathbf{k}$) along $+x$.

So:

$$
\mathbf{k} \propto \mathbf{E}\times \mathbf{B}
$$
```

## Problem 10 – Solar Sail Acceleration

A solar sail of area $A = 100\,\text{m}^2$ and mass $m = 200\,\text{kg}$ is fully absorbing sunlight.

The solar intensity at Earth’s distance is approximately $I = 1360\,\text{W/m}^2$.

**(a)** Find the radiation pressure on the sail.  
**(b)** Determine the force exerted on the sail.  
**(c)** Calculate the acceleration of the solar sail.

```{admonition} Problem 10: Solution
:class: tip, dropdown

**(a) Radiation pressure (absorbing surface)**

For an absorbing surface:

$$
P_{\text{abs}} = \frac{I}{c}
$$

Substitute:

- $I = 1360$
- $c = 3.0\times10^8$

$$
P_{\text{abs}} = \frac{1360}{3.0\times10^8} \approx 4.53\times10^{-6}\,\text{N/m}^2
$$

---

**(b) Force on the sail**

$$
F = P A = (4.53\times10^{-6})(100) \approx 4.53\times10^{-4}\,\text{N}
$$

---

**(c) Acceleration**

$$
a = \frac{F}{m} = \frac{4.53\times10^{-4}}{200} \approx 2.27\times10^{-6}\,\text{m/s}^2
$$

Although tiny, the acceleration is continuous and accumulates over time.
```

## Problem 11 – Laser Beam Radiation Pressure

A continuous laser beam with power $P = 5.0\,\text{W}$ and cross-sectional area $A = 1.0\,\text{cm}^2$ strikes a perfectly reflecting mirror.

**(a)** Find the intensity of the beam.  
**(b)** Calculate the radiation pressure on the mirror.  
**(c)** Determine the momentum transferred to the mirror in $10\,\text{s}$.

```{admonition} Problem 11: Solution
:class: tip, dropdown

**(a) Intensity of the beam**

Intensity is power per unit area:

- $A = 1.0\,\text{cm}^2 = 1.0\times10^{-4}\,\text{m}^2$

$$
I = \frac{P}{A} = \frac{5.0}{1.0\times10^{-4}} = 5.0\times10^4\,\text{W/m}^2
$$

---

**(b) Radiation pressure (reflecting surface)**

For a perfectly reflecting surface:

$$
P_{\text{ref}} = \frac{2I}{c}
$$

Substitute:

$$
P_{\text{ref}} = \frac{2(5.0\times10^4)}{3.0\times10^8} \approx 3.33\times10^{-4}\,\text{N/m}^2
$$

---

**(c) Momentum transferred in 10 s**

The total force:

$$
F = P_{\text{ref}} A = (3.33\times10^{-4})(1.0\times10^{-4}) \approx 3.33\times10^{-8}\,\text{N}
$$

Momentum change over time:

$$
\Delta p = F t = (3.33\times10^{-8})(10) \approx 3.33\times10^{-7}\,\text{kg·m/s}
$$

```
