## Problem 1

Find the momentum of a helium nucleus having a mass of $6.68\times 10^{−27}$ kg that is moving at $0.200c$.

[This problem is a slightly modified version from OpenStax. Access for free here](https://openstax.org/books/university-physics-volume-3/pages/5-problems)

````python
# DIY Cell
````

````python
%reset -f

import numpy as np

c = 2.997e8
v = 0.200*c
m = 6.68e-27

gamma = 1/(np.sqrt(1 - (v/c)**2))

p = gamma*m*v

print(p)
````

## Problem 2

If the rest energies of a proton and a neutron (the two constituents of nuclei) are $938.3$ and $939.6$ MeV, respectively, what is the difference in their mass in kilograms?

[This problem is a slightly modified version from OpenStax. Access for free here](https://openstax.org/books/university-physics-volume-3/pages/5-problems)

````python
# DIY Cell
````

````python
%reset -f

import sympy as sym

c = 2.997e8
E1 = 938.3e6*1.602e-19
E2 = 939.6e6*1.602e-19

m1,m2 = sym.symbols('m1,m2')

eq1 = E1 - m1*c**2
sol1 = sym.solve(eq1,m1)

eq2 = E2 - m2*c**2
sol2 = sym.solve(eq2,m2)

m1 = float(sol1[0])
m2 = float(sol2[0])

dm = m2 - m1
print(dm)
````

## Problem 3: Particle Speed from Relativistic Momentum

A particle with rest mass $m = 1.0 \, \text{kg}$ has a measured momentum of $p = 3.0 \, \text{kg} \cdot \text{m/s}$. Its speed is known to be a significant fraction of the speed of light.

**(a)** Use the relativistic momentum formula to determine the particle's speed $v$.  
**(b)** Compare your result to the classical approximation $p = mv$ and compute the percent error.


```{admonition} Problem 3: Solution
:class: tip, dropdown

**(a) Use the relativistic momentum to find the speed**

The relativistic momentum is:

$$
p = \gamma m v, \quad \text{where} \quad \gamma = \frac{1}{\sqrt{1 - \frac{v^2}{c^2}}}
$$

Given:

- $ p = 3.0 \, \text{kg} \cdot \text{m/s} $
- $ m = 1.0 \, \text{kg} $
- $ c = 3.0 \times 10^8 \, \text{m/s} $

Let’s solve numerically:

Start with:

$$
3.0 = \frac{mv}{\sqrt{1 - v^2/c^2}} = \frac{v}{\sqrt{1 - v^2/c^2}}
$$

Square both sides:

$$
9 = \frac{v^2}{1 - v^2/c^2}
$$

Multiply both sides:

$$
9\left(1 - \frac{v^2}{c^2} \right) = v^2
$$

$$
9 - \frac{9v^2}{c^2} = v^2
$$

Bring terms to one side:

$$
9 = v^2 + \frac{9v^2}{c^2} = v^2\left(1 + \frac{9}{c^2} \right)
$$

Then:

$$
v^2 = \frac{9}{1 + \frac{9}{c^2}} \approx \frac{9}{1 + \frac{9}{(9 \times 10^{16})}} \approx 9 \, \text{m}^2/\text{s}^2
$$

Because $$ c^2 \gg 9 $, this gives:

$$
v \approx 3.0 \, \text{m/s}
$$

This seems suspiciously small — but remember, $p = 3 \, \text{kg} \cdot \text{m/s}$ is **not** large enough to produce relativistic speeds for a 1 kg object.

Let’s try the classical prediction:

**(b) Classical value:**

$$
v_{\text{classical}} = \frac{p}{m} = \frac{3.0}{1.0} = 3.0 \, \text{m/s}
$$

So relativistic and classical agree very well at this scale.

✅ Percent error:

Since the relativistic speed is also 3.0 m/s, the percent error is:

$$
\text{Error} = \frac{|v_{\text{rel}} - v_{\text{classical}}|}{v_{\text{rel}}} \times 100\% = 0\%
$$

Relativistic correction is negligible here.
```

## Problem 4: Two-Particle Collision in the Lab Frame

Two particles, each with mass $m = 1.0 \, \text{GeV}/c^2$, approach each other in the lab frame with equal and opposite velocities of magnitude $v = 0.6c$.

**(a)** Compute the momentum of each particle.  
**(b)** What is the total momentum of the system?  
**(c)** If the two particles stick together and form a single particle at rest, what is the mass of the new particle?


```{admonition} Problem 4: Solution
:class: tip, dropdown

Given:

- $ m = 1.0 \, \text{GeV}/c^2 $
- $ v = 0.6c $
- $ \gamma = \frac{1}{\sqrt{1 - 0.6^2}} = \frac{1}{\sqrt{0.64}} = 1.25 $

**(a) Momentum of each particle:**

$$
p = \gamma m v = 1.25 \cdot 1.0 \cdot 0.6c = 0.75 \, \text{GeV}/c
$$

So:

- Right-moving particle: $ +0.75 \, \text{GeV}/c $
- Left-moving particle: $ -0.75 \, \text{GeV}/c $

**(b) Total momentum:**

$$
p_{\text{total}} = 0.75 - 0.75 = 0
$$

✅ Total momentum is zero → center-of-mass frame.

**(c) Final mass of the combined object:**

Total energy of the system:

Each particle has:

$$
E = \gamma m c^2 = 1.25 \cdot 1.0 \cdot c^2 = 1.25 \, \text{GeV}
$$

So:

$$
E_{\text{total}} = 2 \cdot 1.25 = 2.5 \, \text{GeV}
$$

Since final momentum is zero, this must be **rest energy** of the new object:

$$
E = M c^2 \Rightarrow M = \frac{E}{c^2} = \frac{2.5 \, \text{GeV}}{c^2}
$$

✅ The mass of the new particle is:

$$
\boxed{M = 2.5 \, \text{GeV}/c^2}
$$

This is **larger** than the sum of the original masses — the extra comes from **kinetic energy converted into mass**.
```

## Problem 5: Kinetic Energy of a Relativistic Electron

An electron is accelerated to a speed of $v = 0.99c$. The rest mass of the electron is:

$$
m_e = 9.11 \times 10^{-31} \, \text{kg}
$$

**(a)** Calculate the total energy $E$ of the electron in joules.

**(b)** Calculate the kinetic energy $K$ of the electron.

**(c)** How many times larger is the kinetic energy compared to the electron’s rest energy?

**(d)** If this electron collided with a stationary particle and all kinetic energy were converted to mass, what would be the mass (in kg) of the newly formed particle?

Use $c = 3.00 \times 10^8 \, \text{m/s}$.


```{admonition} Problem 5: Solution
:class: tip, dropdown

**(a) Calculating the Total Energy $E$**

The total energy of a relativistic particle is given by the expression:

$$
E = \gamma m c^2
$$

where the Lorentz factor $\gamma$ is defined as:

$$
\gamma = \frac{1}{\sqrt{1 - v^2/c^2}}
$$

Substituting $v = 0.99c$ into the formula, we find:

$$
\gamma = \frac{1}{\sqrt{1 - (0.99)^2}} = \frac{1}{\sqrt{1 - 0.9801}} = \frac{1}{\sqrt{0.0199}} \approx 7.09
$$

Next, we compute $c^2$:

$$
c^2 = (3.00 \times 10^8)^2 = 9.00 \times 10^{16} \, \text{m}^2/\text{s}^2
$$

Now we calculate the total energy:

$$
E = 7.09 \cdot (9.11 \times 10^{-31} \, \text{kg}) \cdot (9.00 \times 10^{16} \, \text{m}^2/\text{s}^2)
$$

$$
E = 7.09 \cdot 8.199 \times 10^{-14} \, \text{J} \approx 5.81 \times 10^{-13} \, \text{J}
$$

Thus, the total energy of the relativistic electron is approximately:

$$
\boxed{E \approx 5.81 \times 10^{-13} \, \text{J}}
$$

---

**(b) Calculating the Kinetic Energy $K$**

The relativistic kinetic energy is the difference between the total energy and the rest energy:

$$
K = E - E_0 = \gamma mc^2 - mc^2 = (\gamma - 1)mc^2
$$

Substituting the known values:

$$
K = (7.09 - 1) \cdot 9.11 \times 10^{-31} \cdot 9.00 \times 10^{16}
$$

$$
K = 6.09 \cdot 8.199 \times 10^{-14} \, \text{J} \approx 4.99 \times 10^{-13} \, \text{J}
$$

Therefore, the kinetic energy of the electron is approximately:

$$
\boxed{K \approx 4.99 \times 10^{-13} \, \text{J}}
$$

---

**(c) Ratio of Kinetic Energy to Rest Energy**

To find how many times larger the kinetic energy is compared to the rest energy, we must first compute the rest energy:

$$
E_0 = mc^2 = 9.11 \times 10^{-31} \cdot 9.00 \times 10^{16} = 8.199 \times 10^{-14} \, \text{J}
$$

Then we compute the ratio:

$$
\frac{K}{E_0} = \frac{4.99 \times 10^{-13}}{8.199 \times 10^{-14}} \approx 6.09
$$

This shows that the kinetic energy is approximately:

$$
\boxed{6.09 \text{ times greater than the rest energy}}
$$

---

**(d) Rest Mass Equivalent of the Kinetic Energy**

If all of the electron's kinetic energy were converted into mass—such as in a perfectly inelastic collision—the resulting rest mass $M$ would satisfy:

$$
K = M c^2 \quad \Rightarrow \quad M = \frac{K}{c^2}
$$

Substituting the values:

$$
M = \frac{4.99 \times 10^{-13}}{9.00 \times 10^{16}} \approx 5.54 \times 10^{-30} \, \text{kg}
$$

Therefore, the rest mass equivalent of the kinetic energy is:

$$
\boxed{M \approx 5.54 \times 10^{-30} \, \text{kg}}
$$

This is about six times the rest mass of the electron, which agrees with our earlier result in part (c).
```

## Problem 6: Relativistic Energy of a Cosmic Muon

Muons are elementary particles produced in the upper atmosphere by cosmic ray collisions. They have a mass of approximately:

$$
m_\mu = 105.7 \, \text{MeV}/c^2
$$

and a proper (rest-frame) lifetime of about $2.2 \, \mu\text{s}$.

A muon is traveling downward toward Earth's surface at a speed of $v = 0.998c$.

**(a)** Calculate the total energy $E$ of the muon in MeV.

**(b)** Determine the muon's kinetic energy $K$ in MeV.

**(c)** Compute the relativistic momentum $p$ of the muon in MeV$/c$.

**(d)** Using your answers to parts (a) and (c), verify the energy-momentum relation:

$$
E^2 = p^2 c^2 + m^2 c^4
$$

**(e)** Using time dilation, how long does the muon appear to live from the perspective of a lab observer on Earth?

Use $c = 3.00 \times 10^8 \, \text{m/s}$ and express all answers numerically.


```{admonition} Problem 6: Solution
:class: tip, dropdown

**(a) Total Energy $E$ of the Muon**

To calculate the total energy of the muon, we use the relativistic energy formula:

$$
E = \gamma m c^2 = \gamma m_\mu c^2
$$

We first compute the Lorentz factor $\gamma$:

$$
\gamma = \frac{1}{\sqrt{1 - \left( \frac{v}{c} \right)^2}} = \frac{1}{\sqrt{1 - (0.998)^2}} = \frac{1}{\sqrt{1 - 0.996004}} = \frac{1}{\sqrt{0.003996}} \approx 15.82
$$

Now we multiply:

$$
E = \gamma m_\mu c^2 = 15.82 \cdot 105.7 \, \text{MeV} \approx \boxed{1672 \, \text{MeV}}
$$

This is the total energy of the muon as measured in the Earth (lab) frame.

---

**(b) Kinetic Energy $K$ of the Muon**

The kinetic energy is the total energy minus the rest energy:

$$
K = E - E_0 = \gamma m_\mu c^2 - m_\mu c^2 = (\gamma - 1) m_\mu c^2
$$

Substituting in the values:

$$
K = (15.82 - 1) \cdot 105.7 \, \text{MeV} = 14.82 \cdot 105.7 \, \text{MeV} \approx \boxed{1567 \, \text{MeV}}
$$

Thus, the muon's kinetic energy is approximately 1567 MeV.

---

**(c) Relativistic Momentum $p$ of the Muon**

We use the relativistic momentum formula:

$$
p = \gamma m v = \gamma m_\mu v
$$

Since $m_\mu$ is in $\text{MeV}/c^2$ and $v = 0.998c$, we can write:

$$
p = \gamma m_\mu v = 15.82 \cdot 105.7 \cdot 0.998 \, \frac{\text{MeV}}{c}
$$

Now calculate:

$$
p \approx 15.82 \cdot 105.7 \cdot 0.998 \approx \boxed{1669 \, \text{MeV}/c}
$$

So the muon’s momentum is approximately 1669 MeV/$c$.

---

**(d) Verifying the Energy-Momentum Relation**

We now check that the energy-momentum relation holds:

$$
E^2 = p^2 c^2 + m^2 c^4
$$

Using values from earlier:

- $E = 1672 \, \text{MeV}$
- $p = 1669 \, \text{MeV}/c$
- $m_\mu c^2 = 105.7 \, \text{MeV}$

Now compute both sides:

Left-hand side:

$$
E^2 = (1672)^2 = 2795584 \, \text{MeV}^2
$$

Right-hand side:

$$
p^2 + m^2 c^4 = (1669)^2 + (105.7)^2 = 2786881 + 11172 \approx 2798053 \, \text{MeV}^2
$$

There is a small rounding difference due to approximations, but the two sides are consistent within rounding error.

✅ The energy-momentum relation is verified.

---

**(e) Dilated Lifetime from Earth's Frame**

To compute the lifetime of the muon as observed from the Earth’s frame, we apply **time dilation**:

$$
t = \gamma \tau = 15.82 \cdot 2.2 \, \mu\text{s} \approx \boxed{34.8 \, \mu\text{s}}
$$

Therefore, from the Earth's perspective, the muon lives approximately 34.8 microseconds before decaying — significantly longer than its rest-frame lifetime of 2.2 microseconds. This explains why many cosmic muons survive long enough to reach detectors at ground level.
```

## Problem 7: Constant Force on a Relativistic Muon

A muon of mass $m = 1.88 \times 10^{-28} \, \text{kg}$ is accelerated from rest by a constant force of magnitude $F = 3.00 \times 10^{-14} \, \text{N}$ in a straight line.

**(a)** Write an expression for the muon’s momentum as a function of time.

**(b)** Calculate the momentum of the muon after $t = 1.00 \, \mu\text{s}$.

**(c)** Using the relativistic momentum formula, compute the speed of the muon at that time.

**(d)** Determine the muon’s Lorentz factor $\gamma$ and total energy at this time, and compare it to the rest energy.

Use $c = 3.00 \times 10^8 \, \text{m/s}$.


```{admonition} Problem 7: Solution
:class: tip, dropdown

**(a) Expression for Momentum as a Function of Time**

Since the force is constant, the rate of change of momentum is constant. Therefore, the momentum of the muon increases linearly over time. We use the definition of force in terms of momentum:

$$
\vec{F} = \frac{d\vec{p}}{dt} \quad \Rightarrow \quad \vec{p}(t) = \vec{F} t
$$

Because the muon starts from rest, we can ignore any initial momentum. So, after time $t$, the momentum is simply:

$$
p(t) = Ft
$$

---

**(b) Muon’s Momentum After 1 Microsecond**

Using the expression from part (a), we compute:

$$
p = (3.00 \times 10^{-14} \, \text{N}) \cdot (1.00 \times 10^{-6} \, \text{s}) = 3.00 \times 10^{-20} \, \text{kg} \cdot \text{m/s}
$$

This is the muon's relativistic momentum after $1.00 \, \mu\text{s}$ of constant acceleration.

---

**(c) Muon’s Speed from Relativistic Momentum (Exact Expression)**

We now use the exact solution for $v$ in terms of momentum $p$, mass $m$, and the speed of light $c$.

We start from the relativistic momentum formula:

$$
p = \frac{mv}{\sqrt{1 - v^2/c^2}}
$$

Solving for $v$, we square both sides and rearrange to get:

$$
v = \sqrt{ \frac{p^2}{m^2 + \frac{p^2}{c^2}} }
$$

This is our exact expression. 

Now plug everything into the formula and we get:

$$
v = 1.41 \times 10^8 \, \text{m/s}
$$

✅ So the muon’s speed after $1.00 \, \mu\text{s}$ is:

$$
\boxed{v \approx 1.41 \times 10^8 \, \text{m/s}} \quad \text{or about } 0.47c
$$

This result is fully consistent with our expectations for relativistic motion: the particle is clearly moving fast enough that relativistic corrections matter, but not so fast that $\gamma$ becomes extreme.


---

**(d) Lorentz Factor and Total Energy**

Now that we know the speed, we compute the Lorentz factor:

$$
\gamma = \frac{1}{\sqrt{1 - \frac{v^2}{c^2}}} \approx \frac{1}{\sqrt{1 - (0.477)^2}} = \frac{1}{\sqrt{0.7726}} \approx 1.14
$$

The total relativistic energy of the muon is then:

$$
E = \gamma m c^2
$$

Substituting the known values:

$$
E = 1.14 \cdot (1.88 \times 10^{-28}) \cdot (9.00 \times 10^{16}) \approx 1.93 \times 10^{-11} \, \text{J}
$$

For comparison, the rest energy is:

$$
E_0 = mc^2 = 1.88 \times 10^{-28} \cdot 9.00 \times 10^{16} = 1.69 \times 10^{-11} \, \text{J}
$$

So the kinetic energy is:

$$
K = E - E_0 = 1.93 \times 10^{-11} - 1.69 \times 10^{-11} = \boxed{2.4 \times 10^{-12} \, \text{J}}
$$

This means that about **12.5% of the total energy is kinetic** at this point — a substantial relativistic contribution, though we are still far from the extreme relativistic limit.

---

**Final Reflection**

This problem demonstrates how a **constant force** leads to **increasingly nonlinear growth in velocity**. Unlike Newtonian mechanics, the acceleration is not constant because the effective inertia (due to increasing $\gamma$) grows with speed.

After just **1 microsecond**, the muon reaches nearly **half the speed of light**, and relativistic effects become clearly visible in the calculation of momentum and energy. This shows why classical mechanics quickly breaks down for light, fast particles.

-Is a Constant Force Realistic in Relativity?
>
> In this problem, we assumed that a constant force of $3.00 \times 10^{-14} \, \text{N}$ acts on the muon. This makes sense mathematically — the relation $F = dp/dt$ remains valid in special relativity, and integrating a constant force yields a linearly growing momentum.
>
> However, in **realistic particle accelerators or cosmic environments**, forces do not typically remain constant over time:
>
> - In an accelerator, the magnetic and electric fields that push charged particles **change direction and intensity** over time to guide and focus the beam.
> - If the force is electromagnetic, then the **source of the field must also transform correctly under Lorentz transformations**, ensuring compatibility with special relativity.
> - A truly constant force in one frame would appear **non-constant** in another due to relativistic transformations.
>
> Most importantly, as the particle's speed increases, the same applied force results in **diminishing acceleration** because of the increasing "relativistic mass" (via the $\gamma$ factor). This leads to an asymptotic approach to the speed of light — a hallmark of relativistic motion.
>
> ✅ So, while using a constant force is mathematically legitimate for a short time and educationally useful, **no real-world setup can apply a uniform constant force to a particle indefinitely** without considering relativistic field dynamics.
```

## Problem 8: Perpendicular and Parallel Forces on a Relativistic Particle

A proton ($m = 1.67 \times 10^{-27} \, \text{kg}$) is moving at a constant speed of $v = 0.8c$ along the $x$-axis. At this moment, it enters a region where two constant forces act simultaneously:

- A constant **electric force** of magnitude $F_x = 1.0 \times 10^{-12} \, \text{N}$ acts in the **$x$-direction** (parallel to motion).
- A constant **magnetic force** of magnitude $F_y = 1.0 \times 10^{-12} \, \text{N}$ acts in the **$y$-direction** (perpendicular to motion).

**(a)** At the instant the forces are applied, compute the **acceleration vector** $\vec{a}$ of the proton.  
Use relativistic dynamics.

**(b)** Calculate the angle between the net force and the acceleration vector.  
Explain why they are not in the same direction.

**(c)** Compute the rate of change of the particle’s energy at this instant. Which force contributes?

Assume $c = 3.00 \times 10^8 \, \text{m/s}$.


```{admonition} Problem 8: Solution
:class: tip, dropdown

**(a) Compute the Relativistic Acceleration Vector**

We are told that the proton moves at $v = 0.8c$ along the $x$-axis, and that two forces act on it:

- $F_x$ is parallel to motion.
- $F_y$ is perpendicular to motion.

The relativistic response to force depends on the **direction** of the force relative to the velocity. Specifically:

- For the component **parallel** to the velocity:
  $$
  a_\parallel = \frac{F_\parallel}{\gamma^3 m}
  $$
- For the component **perpendicular** to the velocity:
  $$
  a_\perp = \frac{F_\perp}{\gamma m}
  $$

We first compute the Lorentz factor:

$$
\gamma = \frac{1}{\sqrt{1 - v^2/c^2}} = \frac{1}{\sqrt{1 - 0.64}} = \frac{1}{\sqrt{0.36}} = \frac{1}{0.6} \approx 1.667
$$

Now compute each component:

- **Parallel acceleration**:
  $$
  a_x = \frac{F_x}{\gamma^3 m}
  $$

  Note that $\gamma^3 \approx (1.667)^3 \approx 4.63$, so the denominator is significantly increased compared to Newtonian mechanics.

- **Perpendicular acceleration**:
  $$
  a_y = \frac{F_y}{\gamma m}
  $$

Plugging in values:

- $a_x \approx \frac{1.0 \times 10^{-12}}{4.63 \cdot 1.67 \times 10^{-27}} \approx 1.30 \times 10^{14} \, \text{m/s}^2$
- $a_y \approx \frac{1.0 \times 10^{-12}}{1.667 \cdot 1.67 \times 10^{-27}} \approx 3.59 \times 10^{14} \, \text{m/s}^2$

So, the relativistic acceleration vector is:

$$
\vec{a} \approx (1.30 \times 10^{14}, \; 3.59 \times 10^{14}) \; \text{m/s}^2
$$

Even though the two forces are equal in magnitude, the **perpendicular acceleration is much greater**, because the inertia in the direction of motion is enhanced by $\gamma^3$.

---

**(b) Angle Between Force and Acceleration Vectors**

We now compute the angle between the force vector and acceleration vector.

Let $\theta_F$ be the angle of the net force vector:

$$
\tan \theta_F = \frac{F_y}{F_x} = 1 \Rightarrow \theta_F = 45^\circ
$$

Let $\theta_a$ be the angle of the acceleration vector:

$$
\tan \theta_a = \frac{a_y}{a_x} = \frac{3.59 \times 10^{14}}{1.30 \times 10^{14}} \approx 2.76
$$

This gives:

$$
\theta_a \approx \tan^{-1}(2.76) \approx 70.4^\circ
$$

So while the **force vector points at $45^\circ$**, the **acceleration vector points at $70^\circ$** — skewed more toward the perpendicular direction.

> **Why?**  
> The resistance to a change in motion increases more strongly along the direction of motion (due to the $\gamma^3$ term), making the particle "resist" acceleration in that direction more than in the perpendicular direction.

---

**(c) Rate of Change of Energy**

The rate at which work is done (i.e. energy changes) is:

$$
\frac{dE}{dt} = \vec{F} \cdot \vec{v}
$$

Since the velocity is entirely in the $x$-direction, only the $x$-component of force contributes:

$$
\frac{dE}{dt} = F_x v
$$

Compute:

$$
\frac{dE}{dt} = (1.0 \times 10^{-12} \, \text{N}) \cdot (0.8 \cdot 3.00 \times 10^8 \, \text{m/s}) = 2.4 \times 10^{-4} \, \text{W}
$$

✅ The perpendicular force does **no work** — it changes the direction of motion (deflects the particle), but does **not change the energy**.

---

**Summary**

- The force and acceleration vectors point in different directions because of **direction-dependent inertia** in special relativity.
- The **parallel acceleration is suppressed** by $\gamma^3$, while the **perpendicular acceleration** is only scaled by $\gamma$.
- **Only forces parallel to velocity contribute to energy increase**; perpendicular forces bend the path but don’t change speed or energy.

This problem beautifully illustrates the anisotropic nature of relativistic dynamics — and shows how Newton’s second law needs careful re-interpretation at high speeds.
```

## Problem 9: Numerical Motion of a Relativistic Electron in a Velocity-Dependent Force Field

An electron (mass $m = 9.11 \times 10^{-31} \, \text{kg}$) moves in one dimension under the influence of a **nonlinear drag-like force** that depends on velocity:

F(v) = -k \cdot \frac{v}{\sqrt{1 - v^2/c^2}} \quad \text{with } k = 2.0 \times 10^{-24} \, \text{N$\cdot$s/m}

This version of drag is physically consistent with a weak relativistic resistance, like a charged particle moving through thin plasma or weak interstellar medium.

The force acts opposite to the direction of motion and increases with speed. The electron starts at $x = 0$ with initial speed $v_0 = 0.8c$.

Use the **forward finite difference method** to numerically simulate the motion of the electron for $t = 0$ to $t = 20$ ns using a time step $\Delta t = 0.1$ ns.

### Tasks:

**(a)** Write the finite difference update rule for momentum using:

$$
p_{n+1} = p_n + F(v_n) \cdot \Delta t
$$

**(b)** From $p_n$, compute $v_n$ using the relativistic formula:

$$
v_n = \frac{p_n}{\sqrt{m^2 + \frac{p_n^2}{c^2}}}
$$

**(c)** Track and plot $v(t)$ and $x(t)$ over the full simulation. Describe the behavior of the electron qualitatively.

Use $c = 3.00 \times 10^8 \, \text{m/s}$.

````python
import numpy as np
import matplotlib.pyplot as plt

# Constants
m = 9.11e-31       # kg (electron mass)
c = 3.00e8         # m/s (speed of light)
k = 2.0e-24        # N·s/m (drag coefficient)
v0 = 0.8 * c       # initial velocity
dt = 0.1e-9        # time step (0.1 ns)
t_max = 20e-9      # total simulation time (20 ns)

# Time array
N = int(t_max / dt) + 1
t = np.linspace(0, t_max, N)

# Arrays for momentum, velocity, and position
p = np.zeros(N)
v = np.zeros(N)
x = np.zeros(N)

# Initial conditions
gamma0 = 1 / np.sqrt(1 - (v0 / c)**2)
p[0] = gamma0 * m * v0
v[0] = v0
x[0] = 0

# Simulation loop
for n in range(N - 1):
    # Compute force at current velocity (relativistic drag)
    if abs(v[n]) >= c:
        F = 0  # safeguard against divide-by-zero
    else:
        F = -k * v[n] / np.sqrt(1 - (v[n] / c)**2)
    
    # Update momentum
    p[n+1] = p[n] + F * dt
    
    # Compute new velocity from updated momentum
    denom = np.sqrt(m**2 + (p[n+1]**2) / c**2)
    v[n+1] = p[n+1] / denom
    
    # Update position using current velocity
    x[n+1] = x[n] + v[n] * dt

# Plot results
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(t * 1e9, v / c)
plt.title("Velocity vs Time")
plt.xlabel("Time (ns)")
plt.ylabel("v / c")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(t * 1e9, x * 1e6)
plt.title("Position vs Time")
plt.xlabel("Time (ns)")
plt.ylabel("Position (μm)")
plt.grid(True)

plt.tight_layout()
plt.show()
````

## Problem 10: Energy of a High-Energy Electron Beam

In a particle accelerator, an electron beam is produced with **momentum** much greater than its rest mass energy: $p \gg mc$.

The exact relativistic energy is given by:

$$
E = \sqrt{(pc)^2 + (mc^2)^2}
$$

### Tasks:

**(a)** Derive a simplified expression for the energy $E$ in terms of $p$ using a **Taylor expansion** in the small parameter:

$$
\epsilon = \frac{mc}{p}
$$

Expand the square root to second order in $\epsilon$ and simplify your expression.

**(b)** Use your approximation to estimate the total energy of an electron with $p = 100 \, \text{MeV}/c$.

Compare this to the exact result numerically, and compute the **percent error** of the approximation.

Take $mc^2 = 0.511 \, \text{MeV}$ for the electron.


```{admonition} Problem 10: Solution
:class: tip, dropdown

**(a) Derive a Simplified Expression for Energy Using Taylor Expansion**

We begin with the exact expression for total energy:

$$
E = \sqrt{(pc)^2 + (mc^2)^2}
$$

To approximate this when $p \gg mc$, we define the small dimensionless parameter:

$$
\epsilon = \frac{mc}{p} \quad \text{so that} \quad \epsilon \ll 1
$$

We now factor out $pc$ from the square root:

$$
E = pc \cdot \sqrt{1 + \left( \frac{mc}{p} \right)^2 } = pc \cdot \sqrt{1 + \epsilon^2}
$$

This is now in a perfect form to apply the binomial (Taylor) expansion:

$$
\sqrt{1 + \epsilon^2} \approx 1 + \frac{1}{2} \epsilon^2 - \frac{1}{8} \epsilon^4 + \cdots
$$

Keeping only terms up to second order in $\epsilon$:

$$
E \approx pc \left(1 + \frac{1}{2} \epsilon^2 \right)
$$

Now substitute back $\epsilon = \frac{mc}{p}$:

$$
E \approx pc \left(1 + \frac{1}{2} \left( \frac{mc}{p} \right)^2 \right)
= pc + \frac{m^2 c^3}{2p}
$$

✅ This is our **approximate expression for energy** when $p \gg mc$:

$$
\boxed{E \approx pc + \frac{m^2 c^3}{2p}}
$$

The first term is the dominant kinetic part, and the second is a small relativistic correction.

---

**(b) Estimate the Energy of a 100 MeV Electron Beam**

Let’s apply both the approximate and exact formulas for $p = 100 \, \text{MeV}/c$ and compare.

- Electron rest mass energy: $mc^2 = 0.511 \, \text{MeV}$  
- So: $mc = 0.511 \, \text{MeV}/c$

**Approximate Expression:**

We use:

$$
E_{\text{approx}} = pc + \frac{m^2 c^3}{2p}
$$

Plugging in values:

- First term: $pc = 100 \, \text{MeV}$
- Second term: $\frac{(0.511)^2}{2 \cdot 100} = \frac{0.2611}{200} \approx 0.00131 \, \text{MeV}$

So:

$$
E_{\text{approx}} \approx 100 + 0.00131 = \boxed{100.00131 \, \text{MeV}}
$$

**Exact Expression:**

We compute:

$$
E_{\text{exact}} = \sqrt{p^2 c^2 + m^2 c^4} = \sqrt{100^2 + 0.511^2} = \sqrt{10000 + 0.2611} \approx \sqrt{10000.2611}
$$

Using a calculator:

$$
E_{\text{exact}} \approx 100.00131 \, \text{MeV}
$$

So the approximate and exact answers are **identical to four significant digits**.

---

**Percent Error**:

We compute:

$$
\text{Percent Error} = \left| \frac{E_{\text{approx}} - E_{\text{exact}}}{E_{\text{exact}}} \right| \times 100\%
$$

Substituting:

$$
\text{Error} = \frac{100.00131 - 100.00131}{100.00131} = 0
$$

✔️ The error is **negligible** — well below $0.001\%$ — validating the use of this approximation at high energies.

---

**Summary**

- We used a **second-order Taylor expansion** to simplify the relativistic energy formula in the high-momentum limit.
- The approximation captured the leading correction to energy due to the rest mass.
- For $p \gg mc$, this correction becomes vanishingly small — as shown by the near-perfect agreement between exact and approximate results.
- This technique is common in **particle physics**, where many particles have momenta vastly exceeding their rest mass (especially for light particles like electrons or neutrinos).
```

## Problem 11: Head-On Electron-Positron Collision and Annihilation

In a collider experiment, an **electron** and a **positron** (each with mass $m_e = 0.511 \, \text{MeV}/c^2$) are accelerated toward each other at equal and opposite momenta and collide **head-on**.

Each particle has a **total energy** of $E = 5.0 \, \text{MeV}$ in the lab frame.

After the collision, the particles annihilate and produce two photons.

---

### Tasks:

**(a)** Compute the momentum of each particle before the collision.

**(b)** Write down the total energy and momentum of the system before the collision.

**(c)** Determine the energy and direction of each photon after the collision, assuming they are emitted in opposite directions.

**(d)** Explain how this process reflects conservation of both **energy** and **momentum** in a relativistic context.


```{admonition} Problem 11: Solution
:class: tip, dropdown

**(a) Compute the momentum of each particle before the collision**

We are told that each particle (electron and positron) has a total energy of:

$$
E = 5.0 \, \text{MeV}
$$

Their mass is:

$$
m_e c^2 = 0.511 \, \text{MeV}
$$

We use the relativistic energy-momentum relation:

$$
E^2 = (pc)^2 + (m c^2)^2
$$

Solving for momentum:

$$
pc = \sqrt{E^2 - (mc^2)^2} = \sqrt{5.0^2 - 0.511^2} \, \text{MeV}
= \sqrt{25 - 0.2611} = \sqrt{24.7389} \approx 4.9738 \, \text{MeV}
$$

So:

$$
p = \frac{4.9738 \, \text{MeV}}{c}
$$

Each particle has magnitude $p \approx 4.97 \, \text{MeV}/c$, one moving in the $+x$ direction and one in the $-x$ direction.

---

**(b) Total momentum and energy before the collision**

Because the particles are moving in **opposite directions** with **equal momentum magnitude**, the total momentum is:

$$
p_{\text{total}} = p - p = 0
$$

The total energy is simply the sum of the two particle energies:

$$
E_{\text{total}} = 5.0 + 5.0 = \boxed{10.0 \, \text{MeV}}
$$

---

**(c) Energy and direction of the two photons after annihilation**

After the collision, the particles annihilate and produce **two photons**. We assume (as is common in such events) that the photons are emitted **in opposite directions** along the $x$-axis.

Let the energy of each photon be $E_\gamma$ and its momentum be $p_\gamma = E_\gamma/c$.

Since the total momentum before the collision was zero, the momentum after must also be zero. This can only happen if:

- One photon moves in the $+x$ direction
- One photon moves in the $-x$ direction
- Both have **equal energy and opposite momenta**

So:

$$
p_{\gamma 1} = -p_{\gamma 2}, \quad \text{and} \quad E_{\gamma 1} = E_{\gamma 2}
$$

Because total energy is conserved:

$$
E_{\text{total}} = E_{\gamma 1} + E_{\gamma 2} = 2 E_\gamma = 10.0 \, \text{MeV}
$$

Thus:

$$
E_\gamma = \frac{10.0}{2} = \boxed{5.0 \, \text{MeV}}
$$

And each photon carries momentum:

$$
p_\gamma = \frac{E_\gamma}{c} = \frac{5.0 \, \text{MeV}}{c}
$$

This agrees with momentum conservation, since their momenta cancel.

---

**(d) Conservation of Energy and Momentum**

This example beautifully illustrates conservation laws in special relativity:

- **Energy**: Total energy before the collision was 10 MeV, and after annihilation it's 10 MeV (in the form of two 5 MeV photons).
- **Momentum**: The initial system had zero net momentum, and so does the final system — the two photons travel in opposite directions with equal magnitude momenta.

Importantly, this problem also shows that:

- **Photons must be emitted in opposite directions** to satisfy momentum conservation.
- The resulting photon energy **matches** the original kinetic + rest energy of the particles.

This process is not just theoretical — it occurs in PET scans, astrophysical pair annihilation, and high-energy colliders.
```

## Problem 12: 2D Relativistic Annihilation with Unequal Momenta

In a particle experiment, an **electron** and **positron** annihilate in two dimensions. Their mass is $m_e = 0.511 \, \text{MeV}/c^2$.

- The **electron** moves along the $x$-axis with total energy $E_e = 3.0 \, \text{MeV}$.
- The **positron** moves at an angle of $60^\circ$ above the $-x$-axis with total energy $E_p = 6.0 \, \text{MeV}$.

After the collision, the two particles annihilate and produce **two photons**.

---

### Tasks:

**(a)** Compute the momentum vector of each particle before the collision.

**(b)** Determine the total 2D momentum and energy of the system before the collision.

**(c)** Use conservation of momentum and energy to find the **directions** and **energies** of the two photons produced. Assume they are massless and travel at speed $c$.

**(d)** Describe how momentum conservation constrains the directions of the photons in this 2D setting.


```{admonition} Problem 12: Solution
:class: tip, dropdown

**(a) Momentum vectors of each particle before the collision**

We are given:

- Electron: moves along $+x$ axis, total energy $E_e = 3.0$ MeV
- Positron: moves $60^\circ$ above the $-x$ axis, total energy $E_p = 6.0$ MeV
- Rest mass energy: $m_e c^2 = 0.511$ MeV

We use:

$$
p = \sqrt{E^2 - (m c^2)^2} / c
$$

For the **electron**:

$$
p_e = \sqrt{3.0^2 - 0.511^2} \approx \sqrt{9 - 0.261} \approx \sqrt{8.739} \approx 2.956 \, \text{MeV}/c
$$

Since it's along $+x$, its momentum vector is:

$$
\vec{p}_e = (2.956, \; 0) \, \text{MeV}/c
$$

For the **positron**:

$$
p_p = \sqrt{6.0^2 - 0.511^2} \approx \sqrt{36 - 0.261} \approx \sqrt{35.739} \approx 5.976 \, \text{MeV}/c
$$

Its direction is $60^\circ$ above the $-x$ axis, so:

$$
\vec{p}_p = \left( -5.976 \cos 60^\circ, \; 5.976 \sin 60^\circ \right)
= (-2.988, \; 5.173) \, \text{MeV}/c
$$

**NOTE**: momentum is giving in terms of MeV/c, which implies that $E = pc$ will have the same value as momentum but in MeV

---

**(b) Total 2D momentum and energy before collision**

Now we simply add the two vectors.

**Total momentum:**

$$
\vec{p}_{\text{total}} = \vec{p}_e + \vec{p}_p
= (2.956 - 2.988, \; 0 + 5.173)
= (-0.032, \; 5.173) \, \text{MeV}/c
$$

So although the $x$-component is nearly zero, the system has significant upward $y$-momentum.

**Total energy:**

$$
E_{\text{total}} = E_e + E_p = 3.0 + 6.0 = \boxed{9.0 \, \text{MeV}}
$$

---

**(c) Energy and direction of the photons after annihilation**

Let the two photons have momenta $\vec{p}_1$ and $\vec{p}_2$ and energies $E_1$ and $E_2$.

Since photons are massless, we use:

$$
E_i = |\vec{p}_i| c \quad \Rightarrow \quad \vec{p}_i = \frac{E_i}{c} \hat{n}_i
$$

The two photons must **together carry**:

- Total momentum: $\vec{p}_{\text{total}} = (-0.032, \; 5.173)$
- Total energy: $E_1 + E_2 = 9.0$ MeV

Let photon 1 go at angle $\theta_1$ and photon 2 at $\theta_2$, with momenta $p_1$, $p_2$.

Then:

**Momentum conservation:**

$$
\begin{aligned}
p_1 \cos \theta_1 + p_2 \cos \theta_2 &= -0.032 \\
p_1 \sin \theta_1 + p_2 \sin \theta_2 &= 5.173
\end{aligned}
$$

And:

**Energy conservation:**

$$
p_1 + p_2 = 9.0 \quad (\text{in MeV}/c)
$$

This is now a nonlinear system — solvable numerically (see below).

---

**(d) Conceptual interpretation of the photon directions**

What this exercise shows:

- In 2D, **photon directions are not necessarily symmetric** unless the initial momentum is symmetric.
- Even a small imbalance in incoming momenta shifts the entire emission direction of the photons.
- If one photon is more energetic, it will “lean” more opposite to the momentum of the system to balance it.
- **Conservation of energy and momentum fully determine the directions and energies of the photons**, even in 2D.

This is exactly how detectors in colliders reconstruct particle collisions — by working backward from the photon tracks and using conservation laws to infer what happened in the collision.

```

````python
import numpy as np
from scipy.optimize import fsolve
import matplotlib.pyplot as plt

# Given total initial momentum and energy (from electron + positron system)
E_total = 9.0       # Total energy in MeV
px_total = -0.032   # Total momentum x-component in MeV/c
py_total = 5.173    # Total momentum y-component in MeV/c

# System of equations: px, py, and energy conservation
def equations(vars):
    p1, theta1, p2, theta2 = vars
    eq1 = p1 * np.cos(theta1) + p2 * np.cos(theta2) - px_total
    eq2 = p1 * np.sin(theta1) + p2 * np.sin(theta2) - py_total
    eq3 = p1 + p2 - E_total
    eq4 = 0  # Placeholder (4 vars, 3 constraints — this helps fsolve format)
    return [eq1, eq2, eq3, eq4]

# Initial guess: symmetric emission slightly tilted
initial_guess = [4.5, np.deg2rad(80), 4.5, np.deg2rad(100)]

# Solve the system
solution = fsolve(equations, initial_guess)
p1, theta1, p2, theta2 = solution
theta1_deg = np.rad2deg(theta1)
theta2_deg = np.rad2deg(theta2)

# Output results
print(f"Photon 1: Energy = {p1:.2f} MeV, Angle = {theta1_deg:.2f}°")
print(f"Photon 2: Energy = {p2:.2f} MeV, Angle = {theta2_deg:.2f}°")

# Compute momentum components for plotting
px1, py1 = p1 * np.cos(theta1), p1 * np.sin(theta1)
px2, py2 = p2 * np.cos(theta2), p2 * np.sin(theta2)

# Plot the momentum vectors
plt.figure(figsize=(6, 6))
plt.quiver(0, 0, px1, py1, angles='xy', scale_units='xy', scale=1, color='r', width=0.008)
plt.quiver(0, 0, px2, py2, angles='xy', scale_units='xy', scale=1, color='b', width=0.008)
plt.text(px1 * 1.1, py1 * 1.1, 'Photon 1', color='r')
plt.text(px2 * 1.1, py2 * 1.1, 'Photon 2', color='b')
plt.xlim(-8, 8)
plt.ylim(-2, 8)
plt.xlabel("px (MeV/c)")
plt.ylabel("py (MeV/c)")
plt.title("Photon Momentum Vectors After Annihilation (2D)")
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
````

```{admonition} Problem 13: Solution
- The ships move with equal and opposite velocities.
```

## Problem 14: Beacon Exchange with Endor

A rebel ship passes the forest moon **Endor** at the origin:

$$
x = 0, \quad t = 0
$$

The ship is moving away from Endor at a constant speed of $v = 0.6c$.

At time $t = 4 \, \text{s}$ (in Endor's frame), the ship sends a **beacon** (light signal) back toward Endor.

Upon receiving the beacon, Endor **waits for 3 seconds**, then sends a **reply signal** (also traveling at the speed of light) back to the ship.

> **Question:**  
> When and where (in Endor’s frame) does the rebel ship receive the reply?


```{admonition} Problem 14: Solution
:class: tip, dropdown

✅ **Solution (Endor's Frame)**

---

**1. Ship trajectory**

The ship moves away from Endor at $v = 0.6c$, so in Endor’s frame:

$$
x_{\text{ship}}(t) = 0.6t
$$

---

**2. Beacon sent from ship**

The beacon is emitted at:

$$
t = 4 \, \text{s}, \quad x = 0.6 \times 4 = 2.4 \, \text{light-seconds}
$$

It travels toward Endor at speed $c = 1$, so the signal reaches Endor at time $t_E$, satisfying:

$$
x = 2.4 - (t_E - 4) = 0
$$

Solving:

$$
t_E = 6.4 \, \text{s}
$$

---

**3. Reply sent from Endor**

Endor waits 3 seconds, so the reply is sent at:

$$
t_R = 6.4 + 3 = 9.4 \, \text{s}, \quad x = 0
$$

The reply travels at $c = 1$, and we want to find the time $t$ when it reaches the ship:
- Light path: $x = t - 9.4$
- Ship path: $x = 0.6t$

Setting them equal:

$$
0.6t = t - 9.4
$$

$$
0.4t = 9.4 \Rightarrow t = \frac{9.4}{0.4} = 23.5 \, \text{s}
$$

---

**4. Where is the ship?**

At that time:

$$
x = 0.6 \times 23.5 = 14.1 \, \text{light-seconds}
$$

---

✅ **Final Answer**

> The ship receives the reply signal at:

$$
\boxed{t = 23.5 \, \text{s}, \quad x = 14.1 \, \text{light-seconds}}
$$

---

**Interpretation**

The ship continues to move away after sending the beacon. The signal reaches Endor at $t = 6.4 \, \text{s}$, and the reply is delayed until $t = 9.4 \, \text{s}$. Despite light's high speed, the ship has moved so far that the reply doesn't reach it until much later. This clearly illustrates **spacetime causality** and **signal delay in relativistic settings**.
```

````python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
v_ship = 0.6  # ship velocity in units of c
t_beacon_sent = 4  # time ship sends beacon
t_beacon_pos = v_ship * t_beacon_sent  # position when beacon is sent

# Beacon reaches planet at:
t_beacon_arrival = t_beacon_sent + t_beacon_pos  # x = 0 --> x = beacon_pos - (t - 4) = 0

# Planet waits 3 seconds before replying
t_reply_sent = t_beacon_arrival + 3

# Find when and where the ship receives the reply
# Light: x = t - t_reply_sent
# Ship: x = v * t
# Set equal: v*t = t - t_reply_sent → (1 - v)t = t_reply_sent
t_reply_received = t_reply_sent / (1 - v_ship)
x_reply_received = v_ship * t_reply_received

# Time range
t_vals = np.linspace(0, 30, 1000)

# Worldlines
x_ship = v_ship * t_vals
x_beacon = t_beacon_pos - (t_vals - t_beacon_sent)
x_reply = t_vals - t_reply_sent

# Plot
fig, ax = plt.subplots(figsize=(8, 6))

# Ship's worldline
ax.plot(x_ship, t_vals, 'r-', label="Ship trajectory")

# Beacon to planet
ax.plot(x_beacon[t_vals >= t_beacon_sent], t_vals[t_vals >= t_beacon_sent], 'k--', label="Beacon to planet")

# Reply to ship
ax.plot(x_reply[t_vals >= t_reply_sent], t_vals[t_vals >= t_reply_sent], 'b--', label="Reply to ship")

# Events
ax.plot(0, 0, 'ko', label="Ship passes planet (t=0)")
ax.plot(t_beacon_pos, t_beacon_sent, 'ro', label="Beacon sent")
ax.plot(0, t_beacon_arrival, 'ko', label="Beacon received at planet")
ax.plot(0, t_reply_sent, 'bo', label="Reply sent")
ax.plot(x_reply_received, t_reply_received, 'go', label="Reply received by ship")

# Annotations
ax.annotate("Beacon sent", (t_beacon_pos + 0.5, t_beacon_sent - 1), fontsize=9)
ax.annotate("Reply received", (x_reply_received - 3.5, t_reply_received - 1), fontsize=9)

# Axis and labels
ax.set_xlim(0, 25)
ax.set_ylim(0, 35)
ax.set_xlabel("Position $x$ (light-seconds)")
ax.set_ylabel("Time $t$ (seconds)")
ax.set_title("Spacetime Diagram: Beacon and Reply Exchange")
ax.grid(True)
ax.set_aspect('equal')

# Force ticks at integer grid points
ax.set_xticks(np.arange(0, 25, 5))
ax.set_yticks(np.arange(0, 35, 5))
ax.legend(loc='upper left', fontsize=8, frameon=True)


plt.show()
````
