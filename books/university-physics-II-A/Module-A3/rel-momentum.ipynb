## A3.2.1 Motivation: Why Classical Momentum Fails

In classical mechanics, momentum is:

$$
\vec{p} = m\vec{v}
$$

But this definition breaks down at **relativistic speeds**, i.e., when $v \sim c$.

### Key Inconsistencies:
- Momentum conservation fails in different frames under Lorentz transformations.
- No speed limit exists in Newtonian mechanics.
- Newton’s laws violate the principle of relativity when using Galilean transforms.

**We need a new definition** that:
- Reduces to $m\vec{v}$ at low speeds.
- Obeys conservation laws in all inertial frames.
- Is consistent with Lorentz transformations.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.2.2 Derivation of Relativistic Momentum

Assume momentum must be conserved in **all inertial frames**.

To preserve this in special relativity, we define:

$$
\boxed{\vec{p} = \gamma m \vec{v}}
$$

Where:

- $\gamma = \frac{1}{\sqrt{1 - v^2/c^2}}$
- $m$ is the mass
- $\vec{v}$ is the particle velocity
- $\vec{p}$ is the **relativistic momentum**

**Box 1: Low-Speed Limit ($v \ll c$)**

Show that 

$$
\vec{p} \approx m\vec{v}
$$

that is, the relativistic momentum **reduces to classical momentum** in the low-speed limit.

</div>

---

### High-Speed Behavior ($v \to c$)

As $v \to c$:

$$
\gamma \to \infty \quad \Rightarrow \quad \vec{p} \to \infty
$$

→ Requires **infinite momentum** to reach light speed  
→ No massive object can reach or exceed $c$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.2.4 Directional Behavior

Relativistic momentum always points in the **same direction** as velocity, but its **magnitude increases nonlinearly** as $v \to c$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.2.5 Relativistic Collisions

### Momentum Conservation Still Holds:

In all inertial frames:

$$
\sum \vec{p}_{\text{before}} = \sum \vec{p}_{\text{after}}
$$

But use:

$$
\vec{p} = \gamma m\vec{v}
$$

for each particle.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.2.6 Example Problem 1: Compare Classical and Relativistic Momentum

A particle of mass $m = 1$ kg moves at $v = 0.8c$.

### Classical Momentum:

$$
p_{\text{classical}} = mv = 1 \cdot 0.8c = 0.8c
$$

### Relativistic Momentum:

$$
\gamma = \frac{1}{\sqrt{1 - (0.8)^2}} = \frac{1}{\sqrt{0.36}} = \frac{5}{3}
$$

$$
p_{\text{rel}} = \gamma mv = \frac{5}{3} \cdot 1 \cdot 0.8c = \frac{4}{3}c \approx 1.33c
$$

**Difference**: Relativistic momentum is **66% larger**!

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.2.7 Example Problem 2: Two-Body Collision

Two identical particles of rest mass $m$ approach each other with equal speeds $v = 0.6c$ in the lab frame and collide elastically.

### Before:

Total momentum:

$$
\vec{p}_{\text{total}} = \gamma m v\hat{v} - \gamma m v\hat{v} = 0
$$

After: particles must fly apart with **equal and opposite velocities**, preserving $\vec{p} = 0$.
