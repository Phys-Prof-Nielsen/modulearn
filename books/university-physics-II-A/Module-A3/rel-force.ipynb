# A3.4 Relativistic Force

In this lecture, we define and interpret **force in special relativity** using only 3-vector dynamics. We build on our previous discussions of relativistic momentum and energy.

We will cover:

- Why Newton’s $ \vec{F} = m\vec{a} $ fails at high speeds  
- Force from momentum: $ \vec{F} = \frac{d\vec{p}}{dt} $  
- Relativistic acceleration  
- Force directions: parallel vs perpendicular  
- Work and energy connection  
- Example problems

---

## A3.4.1 Why Not $ \vec{F} = m\vec{a} $?

In classical physics:

$$
\vec{F} = m \vec{a}, \quad \vec{p} = m\vec{v}
$$

But in **special relativity**, momentum is:

$$
\vec{p} = \gamma m \vec{v}, \quad \text{where} \quad \gamma = \frac{1}{\sqrt{1 - v^2/c^2}}
$$

This gives:

$$
\vec{F} = \frac{d\vec{p}}{dt} = \frac{d}{dt}(\gamma m \vec{v})
$$

Which is more complex than $ m\vec{a} $, especially at high velocities.

---

## A3.4.2 Computing $ \vec{F} = \frac{d\vec{p}}{dt} $

From relativistic momentum:

$$
\vec{p} = \gamma m \vec{v}
$$

We compute:

$$
\vec{F} = \frac{d\vec{p}}{dt} = \frac{d}{dt}(\gamma m \vec{v}) = m \left( \gamma \frac{d\vec{v}}{dt} + \vec{v} \frac{d\gamma}{dt} \right)
$$

To proceed, we need the time derivative of $\gamma$.

> #### Derivation Box: Time Derivative of $ \gamma $
> Start with:
>
> $$
 \gamma = \frac{1}{\sqrt{1 - \frac{v^2}{c^2}}}
 $$
>
> Let $v^2 = \vec{v} \cdot \vec{v}$. Then use the chain rule:
>
> $$
 \frac{d\gamma}{dt} = \frac{d\gamma}{dv^2} \cdot \frac{dv^2}{dt}
 $$
>
> First,
> 
> $$
 \frac{d\gamma}{dv^2} = \frac{1}{2} \left(1 - \frac{v^2}{c^2}\right)^{-3/2} \cdot \frac{1}{c^2} = \frac{\gamma^3}{2c^2}
 $$
>
> And
> 
> $$
 \frac{d(v^2)}{dt} = \frac{d}{dt}(\vec{v} \cdot \vec{v}) = 2 \vec{v} \cdot \vec{a}
 $$
>
> Therefore,
> 
> $$
 \frac{d\gamma}{dt} = \frac{\gamma^3}{2c^2} \cdot 2 \vec{v} \cdot \vec{a} = \frac{\gamma^3}{c^2} \vec{v} \cdot \vec{a}
 $$

---

Substitute into the force equation:

$$
\vec{F} = m \left[ \gamma \vec{a} + \gamma^3 \frac{\vec{v} \cdot \vec{a}}{c^2} \vec{v} \right]
$$

This is the **relativistic force law** in full vector form. It reduces to $ \vec{F} = m \vec{a} $ only when $ v \ll c $ and $\gamma \approx 1$.

---

## A3.4.3 Force Directions: Parallel vs Perpendicular

We decompose the acceleration into **parallel** and **perpendicular** components relative to $ \vec{v} $.

From the full expression:

$$
\vec{F} = m \left[ \gamma \vec{a} + \gamma^3 \frac{\vec{v} \cdot \vec{a}}{c^2} \vec{v} \right]
$$

we see that the **direction of the force depends** on how $\vec{a}$ aligns with $\vec{v}$.

> #### Are Force and Acceleration Always Aligned?
>
> Not in special relativity.
>
> The force contains **two components**:
>
> 1. $ \gamma \vec{a} ~$: points in the direction of acceleration  
> 2. $ \gamma^3 \frac{\vec{v} \cdot \vec{a}}{c^2} \vec{v} ~$: points in the direction of velocity  
>
> These components are generally **not parallel** unless $ \vec{a} $ is:
>
> - Entirely **parallel** to $ \vec{v} $ (then both terms align)
> - Entirely **perpendicular** to $ \vec{v} $ (then the second term vanishes)
>
> So in **general**, the force and acceleration vectors **do not point in the same direction**.

---

### ➤ If $ \vec{a} \parallel \vec{v} $:

$$
\vec{F}_\parallel = m \gamma^3 \vec{a}
$$

It becomes increasingly difficult to speed up a particle along its direction of motion as $ v \to c $.

---

### ➤ If $ \vec{a} \perp \vec{v} $:

$$
\vec{F}_\perp = m \gamma \vec{a}
$$

It’s easier to change **direction** than **speed** at high velocities.


---

## A3.4.4 Power and Work

Relativistic energy is:

$$
E = \gamma m c^2
$$

The **rate of work** done by a force is still:

$$
\frac{dE}{dt} = \vec{F} \cdot \vec{v}
$$

The **work-energy theorem still holds** in relativity.

---

## A3.4.5 Example Problem 1: Acceleration at $0.8c$

A $1\, \text{kg}$ particle moves at $v = 0.8c$ and is accelerated forward at $a = 1 \, \text{m/s}^2$. What is the required force?

### Solution:

$$
\gamma = \frac{1}{\sqrt{1 - (0.8)^2}} = \frac{1}{\sqrt{0.36}} = \frac{5}{3}
$$

Then:

$$
F_\parallel = m \gamma^3 a = 1 \cdot \left( \frac{5}{3} \right)^3 \cdot 1 = \frac{125}{27} \approx 4.63 \, \text{N}
$$

✅ Much larger than the classical result $F = ma = 1 \, \text{N}$.

---

## A3.4.6 Example Problem 2: Constant Force Over Time

Suppose a constant force $F = 1 \, \text{N}$ is applied to a $1\, \text{kg}$ object starting from rest. How does the velocity evolve?

This is governed by:

$$
F = \frac{d}{dt}(\gamma m v)
$$

Since $m = 1$, we write:

$$
F = \frac{d}{dt}(\gamma v)
$$

This is **nonlinear**, and $v(t)$ approaches $c$ **asymptotically** — never reaching it, no matter how long the force is applied.

---

## A3.4.7 Summary

| Quantity                        | Expression                                     |
|--------------------------------|------------------------------------------------|
| Momentum                       | $ \vec{p} = \gamma m \vec{v} $                |
| Relativistic Force (general)   | $ \vec{F} = m \left[ \gamma \vec{a} + \gamma^3 \frac{\vec{v} \cdot \vec{a}}{c^2} \vec{v} \right] $ |
| Force (parallel to motion)     | $ \vec{F}_\parallel = m \gamma^3 \vec{a} $     |
| Force (perpendicular to motion)| $ \vec{F}_\perp = m \gamma \vec{a} $           |
| Power                          | $ \vec{F} \cdot \vec{v} = \frac{dE}{dt} $       |

---

## Final Thoughts

- The classical form $ \vec{F} = m\vec{a} $ only holds at low speeds.
- In relativity, **acceleration and force are not proportional**.
- The direction of force matters: parallel and perpendicular motion scale differently with $ \gamma $.
- Relativistic mechanics maintains consistency with conservation of energy and momentum — at the cost of simplicity.
