## Why this section matters

In the previous section we introduced **diffraction** and emphasized its connection to the **interference ideas you studied in Phase C**. Diffraction patterns arise because many parts of a wavefront pass through an aperture and **interfere with each other**.

In this section we develop the **quantitative description of diffraction** for the most important case: a **single slit**.

Studying single-slit diffraction is important because:

- it reveals the **mathematical structure of diffraction patterns**
- it explains why waves spread after passing through apertures
- it leads directly to the **diffraction limits of optical systems**

We will focus on the regime known as **Fraunhofer diffraction**, where the diffraction pattern depends only on the **angle relative to the aperture**.

Finally, we extend the idea from a **slit aperture** to a **circular aperture**, which produces the famous **Airy disk** — the fundamental diffraction pattern governing the resolution of telescopes, microscopes, and cameras.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Fraunhofer Diffraction**

**Fraunhofer diffraction** occurs when the light source and observation screen are sufficiently far from the aperture that incoming and outgoing waves can be treated as **parallel plane waves**.

In this regime the diffraction pattern depends only on the **angle relative to the optical axis**, making the mathematics simpler and the patterns easier to interpret.

</div>

---

## Single-Slit Diffraction

Consider monochromatic light of wavelength $\lambda$ passing through a slit of width $a$.

Using **Huygens' principle**, every point across the slit emits secondary waves. These waves travel to the screen and interfere with one another.

The resulting intensity pattern is

$$
I(\theta) = I_0 \left(\frac{\sin \beta}{\beta}\right)^2
$$

where

$$
\beta = \frac{\pi a}{\lambda}\sin\theta
$$

This expression predicts the **distribution of light intensity as a function of angle**.

Important features of the pattern:

- a **wide central maximum**
- weaker **secondary maxima**
- dark **minima** separating the maxima

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Diffraction Minima**

For a slit of width $a$, destructive interference occurs when

$$
a\sin\theta = m\lambda
$$

where

- $m = 1,2,3,\dots$

These angles correspond to **dark fringes** in the diffraction pattern.

</div>

---

## Structure of the Diffraction Pattern

The **central maximum** occurs at $\theta = 0$.

The first dark fringes occur when

$$
a\sin\theta = \lambda
$$

These minima define the width of the central diffraction peak.

A key property of the pattern is that

**the central maximum is twice as wide as the other maxima.**

Thus the diffraction pattern consists of

- one **dominant central peak**
- progressively weaker **side peaks**

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Width of the Central Diffraction Maximum**

A laser with wavelength $ \lambda = 600\,\text{nm} $ passes through a slit of width $ a = 0.10\,\text{mm} $.

The first minimum occurs when

$$
a\sin\theta = \lambda
$$

Thus

$$
\sin\theta = \frac{\lambda}{a}
$$

Substituting values:

$$
\sin\theta = \frac{600\times10^{-9}}{1.0\times10^{-4}}
$$

$$
\sin\theta = 6.0\times10^{-3}
$$

For small angles,

$$
\theta \approx 0.34^\circ
$$

The central maximum therefore extends approximately

$$
\pm 0.34^\circ
$$

around the optical axis.

</div>

---

## Fraunhofer vs. Fresnel Diffraction

Diffraction patterns depend on the **distance between the aperture and the observation screen**.

Two main regimes exist.

### Fraunhofer Diffraction (Far Field)

- source and screen effectively far from the aperture
- waves are nearly **parallel**
- pattern depends only on **angle**
- mathematical description is simpler

This is the regime most commonly studied in optics experiments.

---

### Fresnel Diffraction (Near Field)

- source or screen is relatively close to the aperture
- wavefront curvature becomes important
- patterns become more complex

In this course we primarily focus on **Fraunhofer diffraction**, which captures the essential physics while remaining mathematically manageable.

---

## Circular Apertures and the Airy Disk

Most optical systems — such as **telescopes, microscopes, and cameras** — use **circular apertures** rather than slits.

Diffraction from a circular aperture produces a characteristic pattern called the **Airy pattern**.

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Airy Disk**

The **Airy disk** is the bright central region of the diffraction pattern produced by a circular aperture.

It is surrounded by a series of faint concentric rings.

</div>

---

The angular radius of the first dark ring is approximately

$$
\theta = 1.22 \frac{\lambda}{D}
$$

where

- $D$ = aperture diameter  
- $\lambda$ = wavelength  

This result plays a crucial role in determining the **resolution of optical systems**.

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Diffraction from a Telescope**

A telescope has an aperture diameter

$$
D = 2.4\,\text{m}
$$

and observes light of wavelength

$$
\lambda = 550\,\text{nm}
$$

The angular radius of the Airy disk is

$$
\theta = 1.22\frac{\lambda}{D}
$$

Substituting values:

$$
\theta = 1.22\frac{550\times10^{-9}}{2.4}
$$

$$
\theta \approx 2.8\times10^{-7}\,\text{rad}
$$

This corresponds to approximately

$$
0.06 \text{ arcseconds}
$$

which determines the telescope's theoretical **diffraction-limited resolution**.

</div>

---

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Comparing Telescope Apertures**

Two telescopes observe the same star.

- Telescope A: $D = 0.20\,\text{m}$
- Telescope B: $D = 2.0\,\text{m}$

Both observe light of wavelength $500\,\text{nm}$.

**Tasks**

1. Which telescope produces the **smaller Airy disk**?
2. Use

$$
\theta = 1.22\frac{\lambda}{D}
$$

to estimate the angular resolution for each telescope.
3. Explain physically why astronomers prefer **large aperture telescopes**.

<details>
<summary>Solution</summary>

Because the diffraction angle scales as

$$
\theta \propto \frac{1}{D}
$$

the larger aperture produces the smaller Airy disk.

Therefore telescope B provides **better angular resolution**, allowing astronomers to distinguish objects that are closer together in the sky.

</details>

</div>

## Key Physical Insight

Diffraction reveals a fundamental property of waves:

**no optical system can focus light into a perfect point.**

Even a perfect lens produces a diffraction pattern determined by the wavelength of light and the aperture size.

This limitation leads directly to the **diffraction limit of imaging systems**, which we explore next through the **Rayleigh criterion for optical resolution**.
