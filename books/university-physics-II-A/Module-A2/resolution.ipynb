# A2.3 Optical Resolution and the Rayleigh Criterion
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this section matters

In the previous section we learned that light passing through an aperture produces a **diffraction pattern** rather than a perfect point. Even the best optical systems — telescopes, microscopes, cameras, or the human eye — cannot focus light into an infinitely sharp spot.

Instead, a point source forms an **Airy disk**, the central bright region of a diffraction pattern produced by a circular aperture.

This leads to a fundamental question:

**When can two nearby objects still be distinguished as separate?**

This question defines the concept of **optical resolution**.

The answer is determined not by imperfections in lenses, but by the **wave nature of light itself**. Diffraction imposes a fundamental limit on the resolving power of any optical instrument.

In this section we explore:

- how diffraction limits image sharpness,
- how the **Rayleigh criterion** defines when two objects can be resolved,
- and why larger optical instruments provide better resolution.

These ideas explain why astronomers build **large telescopes**, why microscopes require **short wavelengths**, and why optical systems always have **finite resolving power**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Optical Resolution**

**Optical resolution** is the ability of an imaging system to distinguish two nearby objects as separate.

Higher resolution means the system can resolve **smaller angular or spatial separations** between objects.

</div>

---

## Images of Point Sources

Consider a single distant point source observed through a circular aperture.

Due to diffraction, the image formed is not a point but an **Airy pattern** consisting of:

- a bright **central Airy disk**
- surrounding **concentric rings**

Thus every point source spreads into a small diffraction pattern on the image plane.

When two sources are present, their diffraction patterns **overlap**.

If the overlap is too large, the two sources appear as **one blurred object**.

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Diffraction-Limited Imaging**

An optical system is said to be **diffraction limited** when its resolution is determined solely by diffraction effects rather than imperfections in the optics.

</div>

---

## The Rayleigh Criterion

To determine when two objects can be distinguished, Lord Rayleigh proposed a practical rule.

Two point sources are considered **just resolvable** when the central maximum of one Airy pattern coincides with the first minimum of the other.

This condition produces a visible dip between the two peaks of intensity.

The corresponding minimum resolvable angular separation is

$$
\theta_{\text{min}} = 1.22 \frac{\lambda}{D}
$$

where

- $\lambda$ = wavelength of light  
- $D$ = diameter of the aperture  

This is known as the **Rayleigh criterion**.

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Resolving Two Stars**

A telescope has an aperture diameter

$$
D = 1.0\,\text{m}
$$

and observes light with wavelength

$$
\lambda = 500\,\text{nm}
$$

The Rayleigh criterion gives

$$
\theta_{\text{min}} = 1.22\frac{\lambda}{D}
$$

Substituting values:

$$
\theta_{\text{min}} = 1.22\frac{500\times10^{-9}}{1.0}
$$

$$
\theta_{\text{min}} = 6.1\times10^{-7}\,\text{rad}
$$

Converting to arcseconds:

$$
\theta_{\text{min}} \approx 0.13 \text{ arcseconds}
$$

Two stars separated by a smaller angle would appear **unresolved**.

</div>

---

## Improving Optical Resolution

The Rayleigh criterion reveals two key ways to improve resolution.

### Increase the aperture size

Since

$$
\theta_{\text{min}} \propto \frac{1}{D}
$$

larger apertures produce **smaller diffraction angles**.

This is why telescopes are built with **large mirrors**.

---

### Use shorter wavelengths

Because

$$
\theta_{\text{min}} \propto \lambda
$$

shorter wavelengths improve resolution.

This principle explains why microscopes sometimes use:

- **ultraviolet light**
- **electron beams**

to achieve higher resolving power.

---

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Curiosity — Why Space Telescopes Are Powerful**

Even very large telescopes on Earth suffer from **atmospheric turbulence**, which blurs images.

Space telescopes such as the **Hubble Space Telescope** avoid this problem and operate close to the **diffraction limit** of their mirrors.

</div>

**Box Activity 1 — Comparing Telescope Resolution**

Two telescopes observe a distant binary star system.

- Telescope A: $D = 0.5\,\text{m}$
- Telescope B: $D = 5.0\,\text{m}$

Both observe light with wavelength $600\,\text{nm}$.

**Tasks**

1. Use

$$
\theta_{\text{min}} = 1.22\frac{\lambda}{D}
$$

to estimate the angular resolution of each telescope.

2. Which telescope can resolve smaller angular separations?

3. By what factor does the resolution improve?

<details>
<summary>Solution</summary>

The resolution scales as

$$
\theta_{\text{min}} \propto \frac{1}{D}
$$

Since telescope B has ten times the diameter, its minimum resolvable angle is **ten times smaller**.

Thus telescope B can distinguish objects separated by angles ten times closer together.

</details>

## Key Physical Insight

The Rayleigh criterion shows that the **wave nature of light imposes a fundamental limit on imaging systems**.

No optical system can resolve arbitrarily small details, because diffraction spreads light into finite patterns.

This concept connects optics to many areas of physics, including

- telescope design
- microscopy
- astronomy
- imaging technology
