# A4.6 — The Quantum Harmonic Oscillator

## Motivation

In the previous modules we explored quantum systems such as the **infinite square well** and **finite potential barriers**, which illustrated how boundary conditions and potentials lead to **quantized energy levels** and phenomena such as **tunneling**.

Another extremely important system in physics is the **harmonic oscillator**.

In classical mechanics, many physical systems behave approximately like harmonic oscillators when they are displaced slightly from equilibrium. Examples include

- masses attached to springs
- atoms vibrating in a solid
- molecules vibrating about equilibrium positions

Near equilibrium, many potentials can be approximated by a **quadratic potential**, leading to simple harmonic motion.

In quantum mechanics this leads to the **quantum harmonic oscillator**, one of the most important exactly solvable quantum systems.

The quantum oscillator appears throughout physics, including

- **molecular vibrations**
- **lattice vibrations in solids**
- **quantized electromagnetic fields**
- **quantum field theory**

Studying this system reveals several key quantum ideas:

- **discrete energy levels**
- **zero-point energy**
- **wavefunctions with increasing numbers of nodes**

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Harmonic Oscillator Potential

The potential energy for a harmonic oscillator is

$$
V(x) = \frac{1}{2}kx^2
$$

where

- $k$ is the spring constant
- $x$ is the displacement from equilibrium.

The time-independent Schrödinger equation becomes

$$
-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2}
+ \frac{1}{2}kx^2 \psi
= E\psi
$$

Solving this equation leads to a discrete set of allowed energies.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Energy Levels of the Quantum Harmonic Oscillator**

The allowed energies of the quantum harmonic oscillator are

$$
E_n = \left(n + \frac{1}{2}\right)\hbar \omega
$$

where

- $n = 0,1,2,3,\dots$
- $\omega = \sqrt{\frac{k}{m}}$ is the angular frequency.

These energy levels are **equally spaced**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Comparing with the Infinite Square Well

Notice the differences between the two systems.

For the **particle in a box**

$$
E_n \propto n^2
$$

so the energy levels become farther apart as $n$ increases.

For the **harmonic oscillator**

$$
E_n \propto n
$$

so the energy levels are **equally spaced**.

This difference leads to very different physical behavior.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Insight — Zero-Point Energy**

For the harmonic oscillator, the lowest energy corresponds to

$$
n = 0
$$

giving

$$
E_0 = \frac{1}{2}\hbar\omega
$$

This means the oscillator **cannot have zero energy**.

Even in its lowest energy state the particle still possesses energy.

This minimum energy is called the **zero-point energy** and arises from the **uncertainty principle**.

If the particle had zero energy it would have both perfectly defined position and momentum, which quantum mechanics forbids.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Wavefunctions

The wavefunctions of the quantum harmonic oscillator are more complicated than those of the infinite square well.

They involve **Gaussian functions multiplied by polynomials** known as **Hermite polynomials**.

Although the full mathematical solution is beyond the scope of this course, the qualitative features are important.

- The **ground state** has no nodes.
- Each higher state adds **one additional node**.
- The wavefunctions extend beyond the classical turning points but decay rapidly.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — The Classical Turning Points**

In classical mechanics a particle in a harmonic oscillator moves only between positions where

$$
E = V(x)
$$

These are called **turning points**.

In quantum mechanics the wavefunction does not abruptly stop at these points. Instead it extends slightly beyond them.

Thus there is a small probability of finding the particle **outside the classically allowed region**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Energy Spacing**

For a harmonic oscillator with angular frequency $\omega$, the energy difference between adjacent levels is

$$
E_{n+1} - E_n = \hbar\omega
$$

Thus every transition between neighboring levels corresponds to the **same energy difference**.

This property plays an important role in systems such as **molecular vibrations** and **quantized electromagnetic fields**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Energy Level Pattern**

The harmonic oscillator energies are

$$
E_n = \left(n+\frac{1}{2}\right)\hbar\omega
$$

1. Calculate $E_0$, $E_1$, and $E_2$.
2. Compute the differences $E_1 - E_0$ and $E_2 - E_1$.
3. What does this tell you about the spacing of the energy levels?

<details>
<summary>Solution</summary>

$$
E_0 = \frac{1}{2}\hbar\omega
$$

$$
E_1 = \frac{3}{2}\hbar\omega
$$

$$
E_2 = \frac{5}{2}\hbar\omega
$$

Thus

$$
E_1 - E_0 = \hbar\omega
$$

and

$$
E_2 - E_1 = \hbar\omega
$$

So the energy levels are **equally spaced**.

</details>
