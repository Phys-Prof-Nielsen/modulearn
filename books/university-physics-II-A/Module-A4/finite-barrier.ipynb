# A4.5 — Finite Potential Barriers and Quantum Tunneling
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Motivation

In the previous module we studied the **infinite square well**, where the particle was perfectly confined by walls of infinite potential. In that case the wavefunction was exactly zero outside the box, meaning the particle could never exist beyond the boundaries.

Real physical systems, however, rarely have **infinite potential barriers**. Instead, particles encounter **finite potential barriers**, which they may sometimes cross even when classical physics says they should not.

This leads to one of the most remarkable phenomena in quantum mechanics:

>**quantum tunneling**.

Tunneling occurs because the wavefunction describing a particle does not abruptly stop at a barrier. Instead, it **penetrates into the classically forbidden region**, allowing a finite probability that the particle may appear on the other side of the barrier.

This effect has no classical analogue but plays a crucial role in many real systems, including

- **alpha decay in nuclear physics**
- **electron tunneling in semiconductor devices**
- **scanning tunneling microscopes**
- **fusion reactions in stars**

Understanding tunneling therefore provides insight into both **fundamental quantum behavior and important real-world technologies**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Finite Potential Barrier

Consider a potential barrier of height $V_0$:

$$
V(x) =
\begin{cases}
0 & x < 0 \\
V_0 & 0 \le x \le a \\
0 & x > a
\end{cases}
$$

This represents a region where the particle encounters a **finite obstacle**.

In classical physics:

- if $E > V_0$, the particle crosses the barrier
- if $E < V_0$, the particle is completely reflected

Quantum mechanics predicts a different outcome.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Quantum Tunneling**

**Quantum tunneling** occurs when a particle has a **nonzero probability of passing through a potential barrier even when its energy is less than the barrier height**.

This happens because the wavefunction penetrates into the classically forbidden region.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Schrödinger Equation in the Barrier

Inside the barrier where $V(x) = V_0$, the time-independent Schrödinger equation becomes

$$
-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2} + V_0\psi = E\psi
$$

Rearranging gives

$$
\frac{d^2\psi}{dx^2} = \frac{2m(V_0 - E)}{\hbar^2}\psi
$$

Define

$$
\kappa = \sqrt{\frac{2m(V_0 - E)}{\hbar^2}}
$$

so the equation becomes

$$
\frac{d^2\psi}{dx^2} = \kappa^2 \psi
$$

The solutions are **exponential functions**

$$
\psi(x) = Ae^{\kappa x} + Be^{-\kappa x}
$$

Unlike oscillatory solutions, these **decay exponentially** inside the barrier.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Insight — Wavefunction Penetration**

Even when the particle's energy is less than the barrier height, the wavefunction does not abruptly stop at the barrier.

Instead it **decays exponentially** inside the forbidden region.

Because the wavefunction extends beyond the barrier, there is a **finite probability** that the particle may appear on the other side.

This is the physical origin of **quantum tunneling**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — Energy Is Not Lost During Tunneling**

It may seem as though the particle must "borrow energy" to cross the barrier.

However, tunneling **does not violate energy conservation**.

The particle's energy remains constant throughout the process.

The key idea is that quantum mechanics describes particles using **wavefunctions**, which can extend into classically forbidden regions.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Transmission Probability

When a particle encounters a barrier, the wavefunction partially reflects and partially transmits.

This leads to two important probabilities:

- **Reflection probability** $R$
- **Transmission probability** $T$

with

$$
R + T = 1
$$

For a barrier where $E < V_0$, the transmission probability decreases rapidly with barrier width:

$$
T \sim e^{-2\kappa a}
$$

This shows that tunneling becomes less likely when

- the barrier becomes **thicker**
- the barrier becomes **higher**

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Alpha Decay**

In certain radioactive nuclei, an **alpha particle** is trapped inside a potential barrier created by nuclear forces.

Classically the particle should remain trapped because its energy is less than the barrier height.

However, quantum tunneling allows the particle to **escape the nucleus**, producing **alpha decay**.

This was one of the earliest successful applications of quantum mechanics.

</div>

**Box Activity 1 — Barrier Penetration**

Consider a particle approaching a potential barrier where $E < V_0$.

1. Describe how the wavefunction behaves **inside the barrier**.
2. What happens to the probability density as the particle moves deeper into the barrier?
3. How does increasing the **barrier width** affect the transmission probability?

<details>
<summary>Solution</summary>

Inside the barrier the wavefunction becomes an **exponentially decaying function** rather than an oscillatory one.

Thus the probability density decreases rapidly as the particle penetrates the barrier.

If the barrier becomes thicker, the exponential decay continues for a longer distance, making it much less likely that the particle will appear on the other side.

Therefore the transmission probability decreases rapidly as the **barrier width increases**.

</details>

## Looking Ahead

Quantum tunneling demonstrates that particles described by wavefunctions can penetrate regions that classical physics forbids.

This phenomenon plays a key role in many areas of physics and technology.
