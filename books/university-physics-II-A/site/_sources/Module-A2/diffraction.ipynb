# A2.1 Diffraction — When Waves Spread and Interfere

## Why this section matters

In **Phase C** we studied **interference** — the phenomenon where two or more waves combine to produce regions of constructive and destructive intensity. Classic examples included the **double-slit experiment**, thin-film interference, and other situations where waves from multiple paths overlap.

However, interference does not require multiple slits or separate sources. Even a **single opening** can produce an interference pattern.

This occurs because light passing through an aperture does not travel as a single ray. Instead, **every point across the opening acts like a source of waves**, and these waves interfere with one another. The resulting pattern is called **diffraction**.

Diffraction and interference are therefore **not separate phenomena** — they are closely related manifestations of **wave superposition**.

- **Interference** usually refers to patterns produced by a small number of distinct wave sources.
- **Diffraction** refers to interference arising from **many points across a continuous wavefront**.

In reality, most optical situations involve **both simultaneously**. For example:

- A **double slit experiment** shows interference fringes.
- But each slit also produces diffraction.
- The observed pattern is actually **interference fringes modulated by a diffraction envelope**.

Understanding diffraction therefore reveals something fundamental: **wave behavior sets limits on how light propagates and how sharply images can form**.

**Definition — Diffraction**

**Diffraction** is the spreading of waves when they pass through an aperture or around an obstacle whose size is comparable to the wavelength.

Diffraction occurs because different portions of the wavefront **interfere with each other after passing through the opening**.

</div>

## Diffraction from a Single Slit

Consider light passing through a narrow slit of width $a$.

Using **Huygens' principle**, every point across the slit acts as a source of secondary waves. These waves propagate forward and interfere with each other.

Because waves from different parts of the slit travel slightly different distances to a point on the screen, they can arrive **in phase** or **out of phase**.

This produces an intensity pattern consisting of:

- a **bright central maximum**
- weaker **secondary maxima**
- dark regions called **minima**

The angles at which minima occur satisfy

$$
a\sin\theta = m\lambda \qquad (m = 1,2,3,\dots)
$$

where

- $a$ = slit width  
- $\lambda$ = wavelength of light  
- $\theta$ = angle from the central axis  

This equation predicts where **destructive interference** occurs.

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Predicting the First Diffraction Minimum**

A laser with wavelength $ \lambda = 650\,\text{nm} $ passes through a slit of width $ a = 0.20\,\text{mm} $.

Where is the **first diffraction minimum**?

Using

$$
a\sin\theta = \lambda
$$

we obtain

$$
\sin\theta = \frac{\lambda}{a}
$$

Substituting values:

$$
\sin\theta = \frac{650\times10^{-9}}{2.0\times10^{-4}}
$$

$$
\sin\theta = 3.25\times10^{-3}
$$

Since the angle is small,

$$
\theta \approx 0.19^\circ
$$

This means the first dark fringe appears about **0.19° from the center of the pattern**.

</div>

## Interference vs. Diffraction

Because we previously studied **interference** in Phase C, it is helpful to clarify how the two ideas relate.

| Feature | Interference | Diffraction |
|---|---|---|
| Typical source | Two or more distinct sources | Continuous wavefront |
| Number of interfering waves | Few | Many |
| Example | Double slit | Single slit |
| Underlying mechanism | Superposition | Superposition |

The key point:

**Diffraction is simply interference involving many wave sources across an aperture.**

Thus the concepts are **not fundamentally different** — diffraction is really a **special case of interference with a continuous distribution of sources**.

---

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Curiosity — Do Pure Interference Patterns Exist?**

In most real experiments, diffraction and interference occur **together**.

For example, in a **double-slit experiment**:

- each slit produces its own diffraction pattern
- the two slits then interfere

The observed intensity pattern is therefore:

**interference fringes within a diffraction envelope**

This means the brightness of the interference fringes decreases away from the center due to diffraction.

</div>

## Why Narrow Openings Cause More Spreading

A key qualitative result of diffraction is:

- **wide aperture → small angular spreading**
- **narrow aperture → large angular spreading**

This happens because waves originating from different points across the slit have larger path differences when the slit is narrow.

Thus diffraction becomes most noticeable when

$$
a \sim \lambda
$$

If the aperture is much larger than the wavelength, diffraction effects become small and **ray optics becomes a good approximation**.

**Box Activity 1 — Comparing Aperture Sizes**

Two identical lasers pass through different slits:

- Slit A: $a = 0.5\,\text{mm}$
- Slit B: $a = 0.05\,\text{mm}$

Both lasers have wavelength $650\,\text{nm}$.

**Tasks**

1. Which slit produces the **larger diffraction angle**?
2. Estimate the first minimum angle for each slit using  

$$
a\sin\theta = \lambda
$$

3. Explain physically why the narrower slit produces a wider diffraction pattern.

<details>
<summary>Solution</summary>

The diffraction angle satisfies

$$
\sin\theta = \frac{\lambda}{a}
$$

Since slit B has a **smaller width**, the ratio $ \lambda/a $ is larger.

Therefore slit B produces the **larger diffraction angle**.

Physically, a smaller aperture restricts the wavefront more strongly, causing the wave to spread more after passing through the opening.

</details>

## Key Physical Insight

Diffraction reveals something fundamental about waves:

**waves cannot be perfectly localized in space without spreading in angle.**

This idea appears again in many areas of physics:

- quantum mechanics (wave-particle duality)
- signal processing (Fourier transforms)
- optical imaging limits

Thus diffraction is not merely an optical curiosity — it reflects a **deep property of wave phenomena**.
