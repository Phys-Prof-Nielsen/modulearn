# A2.4 Coherence of Light

## Why this section matters

In **Phase C** we studied **interference**, where waves combine to produce alternating regions of constructive and destructive intensity. In earlier modules we often assumed that light waves maintain a stable phase relationship so that clear interference patterns can form.

In reality, this is **not always true**.

Many light sources — such as light bulbs or the Sun — emit waves whose phases fluctuate randomly. When waves do not maintain a stable phase relationship, interference patterns become unstable and wash out.

The property that determines whether interference patterns remain stable is called **coherence**.

Understanding coherence is important because it explains:

- why interference experiments often require **lasers or narrow apertures**
- why extended light sources tend to destroy interference patterns
- why certain optical instruments require highly coherent light

Coherence therefore determines **when interference and diffraction patterns can actually be observed**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Coherence**

**Coherence** describes the stability of the phase relationship between waves.

When waves maintain a predictable phase relationship, stable interference patterns can form.

</div>

---

## Why Interference Requires Coherence

To produce a stable interference pattern, the phase difference between waves must remain **constant in time**.

If the phase fluctuates randomly, the bright and dark fringes move rapidly and the pattern averages out.

Thus interference experiments require light sources with **sufficient coherence**.

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Light Bulb vs Laser**

A **laser** produces light with a very narrow range of wavelengths and highly ordered phases. This gives lasers extremely strong coherence, allowing clear interference patterns.

A **light bulb**, by contrast, emits light over a wide range of wavelengths with rapidly changing phases. As a result, interference patterns from ordinary light sources are usually weak or short-lived.

</div>

---

## Temporal Coherence

One important type of coherence is **temporal coherence**, which describes how long a wave maintains a predictable phase relationship.

This property is related to the **range of wavelengths** emitted by a light source.

A source with a narrow spread of wavelengths maintains phase relationships longer.

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Coherence Time and Coherence Length**

The **coherence time** $\tau_c$ is the time over which a wave maintains a predictable phase relationship.

The corresponding **coherence length** is

$$
L_c = c\tau_c
$$

where $c$ is the speed of light.

Coherence length represents the maximum path difference over which interference can still occur.

</div>

---

Sources with small wavelength spread (such as lasers) have **long coherence lengths**, while broadband sources have **very short coherence lengths**.

---

## Spatial Coherence

Another important form of coherence is **spatial coherence**, which describes how consistent the phase is across different points of a wavefront.

Large extended sources tend to produce waves arriving from many different directions, reducing spatial coherence.

This is why interference experiments often use:

- small apertures
- point-like sources
- lasers

to increase spatial coherence.

---

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Curiosity — Why Young’s Double-Slit Experiment Uses a Single Source**

If two completely independent light sources illuminate a double slit, the phases fluctuate independently and the interference pattern disappears.

Instead, the experiment uses **one source whose light is split into two paths**, ensuring the waves remain coherent.

</div>

---

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Coherence and Path Difference**

A light source has a coherence length

$$
L_c = 2\,\text{mm}
$$

An interferometer splits the beam into two paths with a path difference of

$$
3\,\text{mm}
$$

**Tasks**

1. Will a stable interference pattern be observed?
2. Explain your reasoning using the concept of coherence length.

<details>
<summary>Solution</summary>

Interference requires the path difference to be smaller than the coherence length.

Since the path difference ($3\,\text{mm}$) is larger than the coherence length ($2\,\text{mm}$), the waves lose their phase relationship before recombining.

Therefore **no stable interference pattern will appear**.

</details>

</div>

## Key Physical Insight

Coherence determines whether the wave nature of light becomes **observable in practice**.

Even though light always behaves as a wave, interference and diffraction patterns only appear when the light maintains a stable phase relationship.

Thus coherence acts as the **practical condition that allows wave phenomena to be seen in experiments**.

In the next section we explore how coherence enables **interferometry**, one of the most powerful tools for precision measurement in physics.
