# A4.1 — Operators in Quantum Mechanics

## Motivation

In Phase B, we introduced the idea that particles are described by **wavefunctions** and that measurements in quantum mechanics are fundamentally probabilistic. We also saw that wave-like behavior leads naturally to ideas such as **uncertainty**, **momentum spread**, and **probability density**.

To move further into quantum mechanics, we now need a mathematical language that tells us how to extract physical information from a wavefunction.

That language is built around **operators**.

In classical mechanics, physical quantities such as position, momentum, and energy are treated as numbers attached to a particle. In quantum mechanics, these measurable quantities are represented by **operators** that act on the wavefunction. This is one of the most important conceptual shifts in the subject.

In this section, we will introduce:

- what an **operator** is,
- how operators act on functions,
- why **position** and **momentum** are represented by different kinds of operators,
- and how operators lead naturally to **eigenvalue problems**, which will appear throughout this phase.

This section is therefore the mathematical doorway into the rest of the quantum module.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Operator**

An **operator** is a mathematical rule that acts on a function and produces another function.

If $\hat{A}$ is an operator and $f(x)$ is a function, then

$$
\hat{A}f(x)
$$

means that the operator $\hat{A}$ acts on $f(x)$.

Examples:

- Multiplication by $x$
- Taking a derivative
- Taking a second derivative

In quantum mechanics, measurable quantities such as **position**, **momentum**, and **energy** are represented by operators.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Simple Examples of Operators

Let us look at a few simple examples before discussing quantum operators.

### Multiplication operator

Suppose the operator $\hat{A}$ multiplies a function by $x$:

$$
\hat{A}f(x) = x f(x)
$$

If

$$
f(x) = x^2
$$

then

$$
\hat{A}f(x) = x(x^2) = x^3
$$

### Derivative operator

Suppose the operator $\hat{B}$ differentiates with respect to $x$:

$$
\hat{B}f(x) = \frac{d}{dx}f(x)
$$

If

$$
f(x) = x^2
$$

then

$$
\hat{B}f(x) = 2x
$$

### Second-derivative operator

Suppose the operator $\hat{C}$ takes the second derivative:

$$
\hat{C}f(x) = \frac{d^2}{dx^2}f(x)
$$

For the same function,

$$
f(x) = x^2
$$

we get

$$
\hat{C}f(x) = 2
$$

These examples show that an operator is not just a number. It is an **action**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Insight — Why Operators Matter in Quantum Mechanics**

In classical physics, if we know a particle's state, we simply assign numbers like position $x$, momentum $p$, and energy $E$.

In quantum mechanics, the state is described by a **wavefunction** $\psi(x)$.

So how do we extract information about measurable quantities from $\psi(x)$?

We do this using **operators**.

Quantum mechanics tells us that observables are not inserted as ordinary numbers. Instead, they are represented by mathematical operations acting on the wavefunction.

This is one of the central structural ideas of the entire subject.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Operators Acting on Wavefunctions

A wavefunction $\psi(x)$ contains the information about the state of a particle in one dimension.

An operator acts on that wavefunction to produce new information.

For example:

- the **position operator** tells us how position is represented,
- the **momentum operator** tells us how momentum is represented,
- and later the **Hamiltonian operator** will represent the total energy.

Thus quantum mechanics replaces the classical question

> “What number is the momentum?”

with the quantum question

> “How does the momentum operator act on the wavefunction?”

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Position Operator**

In one dimension, the **position operator** is

$$
\hat{x} = x
$$

This means that the operator acts by multiplying the wavefunction by $x$:

$$
\hat{x}\psi(x) = x\psi(x)
$$

So position is represented by a **multiplication operator**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Applying the Position Operator**

Let

$$
\psi(x) = e^{-x^2}
$$

Then applying the position operator gives

$$
\hat{x}\psi(x) = x e^{-x^2}
$$

The operator does not return a single number. It returns a **new function**.

This is typical of quantum operators.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Momentum Operator**

In one dimension, the **momentum operator** is

$$
\hat{p} = -i\hbar \frac{d}{dx}
$$

This means that the operator acts on a wavefunction as

$$
\hat{p}\psi(x) = -i\hbar \frac{d\psi}{dx}
$$

So momentum is represented by a **derivative operator**, not by simple multiplication.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why Does Momentum Involve a Derivative?

This result is deeply connected to the wave nature of matter.

Recall from Phase B that a wave with wavelength $\lambda$ has momentum

$$
p = \frac{h}{\lambda}
$$

A sinusoidal or exponential wave changes in space at a rate determined by its wavelength.

- Shorter wavelength $\rightarrow$ more rapid variation in space
- More rapid variation $\rightarrow$ larger derivative
- Larger momentum $\rightarrow$ shorter wavelength

So the derivative naturally captures the spatial variation associated with momentum.

This is why momentum is represented by a differential operator.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Applying the Momentum Operator**

Let

$$
\psi(x) = Ae^{ikx}
$$

Apply the momentum operator:

$$
\hat{p}\psi(x) = -i\hbar \frac{d}{dx}\left(Ae^{ikx}\right)
$$

Since

$$
\frac{d}{dx}\left(Ae^{ikx}\right) = ikAe^{ikx}
$$

we obtain

$$
\hat{p}\psi(x) = -i\hbar (ik)Ae^{ikx}
$$

so

$$
\hat{p}\psi(x) = \hbar k \psi(x)
$$

This is an extremely important result:

$$
\hat{p}\psi = p\psi
\qquad \text{with} \qquad
p = \hbar k
$$

So the wavefunction $Ae^{ikx}$ is an **eigenfunction of the momentum operator**, and the corresponding eigenvalue is the momentum $\hbar k$.

</div>

**Interpreting the Result**

The result

$$
\hat{p}\psi = \hbar k \psi
$$

shows that the wavefunction $Ae^{ikx}$ is an eigenfunction of the momentum operator with eigenvalue $p=\hbar k$.

In fact, we could reverse this logic and use it to **define the momentum operator**. From de Broglie's relation we know that a plane wave

$$
\psi(x)=Ae^{ikx}
$$

represents a state with definite momentum

$$
p=\hbar k .
$$

If we demand that this wavefunction satisfy an eigenvalue equation of the form

$$
\hat{p}\psi = p\psi ,
$$

then the operator that accomplishes this must be

$$
\hat{p} = -i\hbar\frac{d}{dx}.
$$

Thus the derivative form of the momentum operator is not arbitrary—it is precisely the operator that returns the correct momentum eigenvalue for a matter wave.

---

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Deriving the Kinetic Energy Operator**

In classical mechanics, the kinetic energy of a particle is

$$
K = \frac{p^2}{2m}.
$$

In quantum mechanics, momentum is represented by the **momentum operator**

$$
\hat{p} = -i\hbar\frac{d}{dx}.
$$

1. Substitute the momentum operator into the classical kinetic energy expression to construct the **kinetic energy operator**.
2. Apply the resulting operator to a wavefunction $\psi(x)$.
3. Simplify the expression to obtain the final form of the kinetic energy operator.

<details>
<summary>Solution</summary>

Start with the classical kinetic energy expression

$$
K = \frac{p^2}{2m}.
$$

In quantum mechanics we replace the momentum $p$ with the **momentum operator**

$$
\hat{p} = -i\hbar\frac{d}{dx}.
$$

Thus the kinetic energy operator becomes

$$
\hat{K} = \frac{\hat{p}^2}{2m}.
$$

Now substitute the momentum operator:

$$
\hat{K} = \frac{1}{2m}\left(-i\hbar\frac{d}{dx}\right)^2.
$$

Apply this operator to a wavefunction $\psi(x)$:

$$
\hat{K}\psi(x) = \frac{1}{2m}\left(-i\hbar\frac{d}{dx}\right)^2 \psi(x).
$$

Squaring the operator gives

$$
(-i\hbar)^2 = -\hbar^2.
$$

Thus

$$
\hat{K}\psi(x) =
-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2}.
$$

Therefore the **kinetic energy operator** is

$$
\boxed{\hat{K} = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2}}
$$

This operator will appear in the **Hamiltonian operator**, which represents the total energy of a quantum system.

</details>

</div>

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — Operators Are Not Always Ordinary Multiplication**

A common mistake is to think that every observable in quantum mechanics acts like multiplication by a number.

That is true for the position operator:

$$
\hat{x}\psi = x\psi
$$

but not for momentum:

$$
\hat{p}\psi = -i\hbar \frac{d\psi}{dx}
$$

Different physical observables are represented by different kinds of operators.

This is one reason quantum mechanics feels so different from classical mechanics.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Linear Operators

Most quantum operators are **linear**.

An operator $\hat{A}$ is linear if

$$
\hat{A}\left[c_1 f(x) + c_2 g(x)\right]
=
c_1 \hat{A}f(x) + c_2 \hat{A}g(x)
$$

where $c_1$ and $c_2$ are constants.

This matters because wavefunctions can be added together to form **superpositions**, and quantum mechanics depends heavily on that structure.

For example, the derivative operator is linear:

$$
\frac{d}{dx}\left[c_1 f(x) + c_2 g(x)\right]
=
c_1 \frac{df}{dx} + c_2 \frac{dg}{dx}
$$

The position operator is also linear:

$$
x[c_1 f(x) + c_2 g(x)] = c_1 x f(x) + c_2 x g(x)
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Cool Information — Matrices Also Represent Operators**

In this phase, we will mainly use operators as **differential operators acting on wavefunctions**.

However, in more advanced quantum mechanics, the same ideas can also be expressed using **matrices acting on vectors**.

So when we later talk about **eigenvalues** and **eigenvectors** in linear algebra, we are seeing a mathematical structure closely related to what we are doing here.

For now, the differential-operator viewpoint is enough for solving the problems we need in this course.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Eigenvalue Equations

One of the most important ideas in quantum mechanics is the **eigenvalue equation**

$$
\hat{A}\psi = a\psi
$$

This means:

- $\hat{A}$ is an operator
- $\psi$ is a function that the operator acts on
- $a$ is a number called the **eigenvalue**

If applying the operator to a function gives back the **same function times a constant**, then that function is called an **eigenfunction** of the operator.

The constant is the **eigenvalue**.

This is crucial in quantum mechanics because measurement outcomes are associated with operator eigenvalues.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Eigenfunction and Eigenvalue**

If an operator $\hat{A}$ satisfies

$$
\hat{A}\psi = a\psi
$$

then:

- $\psi$ is an **eigenfunction** of $\hat{A}$
- $a$ is the corresponding **eigenvalue**

In quantum mechanics, eigenvalues correspond to the **allowed measured values** of the observable associated with the operator.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — A Simple Eigenvalue Problem**

Consider the derivative operator

$$
\hat{D} = \frac{d}{dx}
$$

and the function

$$
f(x) = e^{3x}
$$

Apply the operator:

$$
\hat{D}f(x) = \frac{d}{dx}(e^{3x}) = 3e^{3x}
$$

Thus

$$
\hat{D}f = 3f
$$

So $e^{3x}$ is an eigenfunction of the derivative operator, with eigenvalue

$$
3
$$

This is the same mathematical structure that appears in quantum mechanics.


</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why Eigenvalue Problems Matter in Quantum Mechanics

Quantum mechanics uses eigenvalue equations everywhere.

Examples:

- **Momentum eigenvalue equation**

$$
\hat{p}\psi = p\psi
$$

- **Energy eigenvalue equation**

$$
\hat{H}\psi = E\psi
$$

- later, for angular momentum and spin, we will encounter similar equations again.

These equations tell us the **allowed values** of measurable quantities and the wavefunctions associated with them.

In particular, the **time-independent Schrödinger equation** is an eigenvalue problem for energy.

That is why operators and eigenvalue equations must come before solving quantum systems.

**Box Activity 2 — Identifying Operators**

For each of the following, determine what the operator does to the function.

1. If

$$
\hat{A} = x
\qquad \text{and} \qquad
f(x) = x^2 + 1
$$

find $\hat{A}f(x)$.

2. If

$$
\hat{B} = \frac{d}{dx}
\qquad \text{and} \qquad
g(x) = x^3
$$

find $\hat{B}g(x)$.

3. Which of the two operators above acts by **multiplication**, and which acts by **differentiation**?

<details>
<summary>Solution</summary>

1. Since $\hat{A}$ multiplies by $x$,

$$
\hat{A}f(x) = x(x^2+1)=x^3+x
$$

2. Since $\hat{B}$ differentiates,

$$
\hat{B}g(x)=\frac{d}{dx}(x^3)=3x^2
$$

3. The operator $\hat{A}=x$ is a multiplication operator, while $\hat{B}=\frac{d}{dx}$ is a derivative operator.

</details>

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 3 — Is This an Eigenfunction?**

Consider the operator

$$
\hat{A} = \frac{d^2}{dx^2}
$$

and the function

$$
f(x) = \sin(2x).
$$

1. Apply the operator $\hat{A}$ to the function $f(x)$.
2. Determine whether $f(x)$ is an **eigenfunction** of the operator.
3. If it is an eigenfunction, determine the **eigenvalue**.
4. What other functions might also be eigenfunctions of this operator?

<details>
<summary>Solution</summary>

First apply the operator:

$$
\hat{A}f(x) = \frac{d^2}{dx^2}(\sin(2x)).
$$

Take the derivatives step by step.

First derivative:

$$
\frac{d}{dx}(\sin(2x)) = 2\cos(2x).
$$

Second derivative:

$$
\frac{d^2}{dx^2}(\sin(2x)) = -4\sin(2x).
$$

Thus

$$
\hat{A}f(x) = -4\sin(2x).
$$

This can be written as

$$
\hat{A}f(x) = -4f(x).
$$

Since the result is the **original function multiplied by a constant**, the function is an **eigenfunction** of the operator.

The **eigenvalue** is

$$
-4.
$$

Other functions such as $\cos(2x)$ are also eigenfunctions of the same operator, because

$$
\frac{d^2}{dx^2}(\cos(2x)) = -4\cos(2x).
$$

This is important in physics because **sinusoidal functions frequently appear as eigenfunctions of differential operators**, especially in wave problems and quantum systems.

</details>

## Summary

In this section we introduced the mathematical language that we will use throughout the rest of the quantum module.

The key ideas are:

- An **operator** acts on a function and produces another function.
- In quantum mechanics, observables are represented by **operators**.
- The **position operator** acts by multiplication:

$$
\hat{x}\psi = x\psi
$$

- The **momentum operator** acts by differentiation:

$$
\hat{p}\psi = -i\hbar\frac{d\psi}{dx}
$$

- **Eigenvalue equations** have the form

$$
\hat{A}\psi = a\psi
$$

and play a central role in quantum mechanics.

In the next section, we will build directly on this idea by looking more carefully at **eigenvalue problems** and then applying them to the **Schrödinger equation**.
