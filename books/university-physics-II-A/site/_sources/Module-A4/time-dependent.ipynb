# A4.2 — The Time-Dependent Schrödinger Equation
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Motivation

In the previous module we introduced **operators** and **eigenvalue equations**, which form the mathematical language of quantum mechanics. Observables such as position and momentum are represented by operators that act on the **wavefunction**. Combined with Phase B, wehere we learned that particles behave like **waves** and that physical quantities such as position and momentum cannot both be known with arbitrary precision due to the **uncertainty principle**, the next step is to ask a fundamental question:

**How does the wavefunction evolve and what determines the allowed quantum states of a system?**

The answer is provided by the **Schrödinger equation**, the central equation of non-relativistic quantum mechanics. It plays a role similar to **Newton's second law in classical mechanics**: given the forces (or potential energy) acting on a system, the equation determines the behavior of the system.

We now introduce the equation that governs how the wavefunction evolves in time: the **time-dependent Schrödinger equation**.

Just as **Newton’s second law**

$$
F = m a
$$

determines how the motion of a particle evolves in classical mechanics, the Schrödinger equation determines how the **quantum state of a system evolves in time**. 

>For systems whose potential does not change with time, the Schrödinger equation reduces to a particularly useful form known as the **time-independent Schrödinger equation**. This equation allows us to determine the **stationary states** and **allowed energies** of a quantum system.

The equation connects several ideas we have already encountered:

- wavefunctions describing quantum states  
- operators representing physical observables  
- eigenvalue problems determining allowed measurement outcomes  

In particular, the Schrödinger equation introduces the **Hamiltonian operator**, which represents the **total energy of the system**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Time-Dependent Schrödinger Equation**

For a particle of mass $m$ moving in a potential $V(x,t)$, the **time-dependent Schrödinger equation** is

$$
i\hbar \frac{\partial \psi(x,t)}{\partial t}
=
\hat{H}\psi(x,t)
$$

where the **Hamiltonian operator** is

$$
\hat{H}
=
-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}
+
V(x,t)
$$

Here

- $\psi(x,t)$ is the **wavefunction**
- $\hat{H}$ is the **Hamiltonian operator**
- $i$ is the imaginary unit
- $\hbar$ is the reduced Planck constant

This equation determines how the quantum state **changes with time**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Interpreting the Equation

The Schrödinger equation can be interpreted as a **dynamical equation for the wavefunction**.

The left side

$$
i\hbar \frac{\partial \psi}{\partial t}
$$

describes how the wavefunction **changes in time**.

The right side

$$
\hat{H}\psi
$$

represents the action of the **Hamiltonian operator**, which corresponds to the **total energy of the system**.

Thus the equation states that the **time evolution of the quantum state is governed by the system’s energy**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Insight — The Hamiltonian**

The Hamiltonian operator is the quantum-mechanical version of the **total energy**.

In classical mechanics we write

$$
E = K + V
$$

where

- $K$ is kinetic energy
- $V$ is potential energy.

In quantum mechanics this becomes

$$
\hat{H}
=
\hat{K} + \hat{V}
$$

where the kinetic energy operator is

$$
\hat{K}
=
-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}.
$$

Thus the Schrödinger equation is essentially the quantum version of the classical **energy equation**, expressed using operators.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Probability and the Wavefunction

Recall from Phase B that the wavefunction itself is **not directly observable**.

The measurable quantity is the **probability density**

$$
P(x,t) = |\psi(x,t)|^2
$$

which gives the probability of finding the particle near position $x$ at time $t$.

The Schrödinger equation therefore determines how the **probability distribution evolves in time**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — The Schrödinger Equation Is Not a Classical Wave Equation**

Although the wavefunction behaves like a wave mathematically, it is **not a physical oscillation in space** like a water wave or sound wave.

The wavefunction encodes **probability amplitudes**, not physical displacement.

Because of this:

- $\psi(x,t)$ itself is not measurable.
- Only $|\psi(x,t)|^2$ has direct physical meaning.

This distinction is one of the key conceptual differences between **classical waves and quantum waves**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Time-Independent Potentials

Many important quantum systems have a potential energy that **does not depend on time**:

$$
V(x,t) = V(x).
$$

When this happens, the Schrödinger equation allows solutions that separate into **space and time parts**.

These solutions take the form

$$
\psi(x,t) = \psi(x)e^{-iEt/\hbar}.
$$

Substituting this form into the Schrödinger equation leads to a new equation:

the **time-independent Schrödinger equation**.

This equation will allow us to determine the **allowed energies and stationary states** of a quantum system.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Free Particle**

Consider a particle moving in a region where the potential is zero:

$$
V(x) = 0.
$$

The Schrödinger equation becomes

$$
i\hbar \frac{\partial \psi}{\partial t}
=
-\frac{\hbar^2}{2m}
\frac{\partial^2 \psi}{\partial x^2}.
$$

A solution to this equation is the plane wave

$$
\psi(x,t)
=
Ae^{i(kx-\omega t)}.
$$

Using the Schrödinger equation, one can show that the angular frequency must satisfy

$$
\omega = \frac{\hbar k^2}{2m}.
$$

Thus the Schrödinger equation connects the **wave properties of matter** with the **energy and momentum of the particle**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Identifying the Hamiltonian**

The time-dependent Schrödinger equation can be written as

$$
i\hbar \frac{\partial \psi}{\partial t}
=
\left(
-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}
+
V(x)
\right)\psi.
$$

1. Identify which term corresponds to the **kinetic energy operator**.
2. Identify which term corresponds to the **potential energy**.
3. Explain why the Hamiltonian represents the **total energy of the system**.

<details>
<summary>Solution</summary>

The Hamiltonian operator is

$$
\hat{H}
=
-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}
+
V(x).
$$

The first term

$$
-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}
$$

comes from replacing momentum in the classical kinetic energy formula

$$
K = \frac{p^2}{2m}
$$

with the momentum operator

$$
\hat{p} = -i\hbar\frac{d}{dx}.
$$

Squaring this operator gives

$$
\hat{p}^2 = -\hbar^2\frac{d^2}{dx^2}.
$$

Thus the first term represents the **kinetic energy operator**.

The second term

$$
V(x)
$$

represents the **potential energy** of the particle.

Therefore the Hamiltonian operator is simply the quantum version of

$$
E = K + V,
$$

which explains why it represents the **total energy of the system**.

</details>

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 2 — Time Evolution of a Wavefunction**

Suppose a wavefunction has the form

$$
\psi(x,t) = \psi(x)e^{-iEt/\hbar}.
$$

1. What part of the wavefunction depends only on **space**?
2. What part depends only on **time**?
3. What does the constant $E$ represent physically?

<details>
<summary>Solution</summary>

The wavefunction is written as a product of a spatial part and a time part.

The spatial part is

$$
\psi(x),
$$

which determines the **shape of the wavefunction in space**.

The time-dependent part is

$$
e^{-iEt/\hbar}.
$$

This factor describes how the **phase of the wavefunction evolves in time**.

The constant $E$ represents the **energy eigenvalue** associated with the stationary state.

Thus this form of the solution describes a **state with definite energy**.

</details>

</div>

## Looking Ahead

The time-dependent Schrödinger equation describes the **full time evolution of a quantum system**.

However, solving it directly can be difficult. Fortunately, when the potential is time-independent we can reduce the problem to the **time-independent Schrödinger equation**, which takes the form of an **eigenvalue problem**.

This will allow us to determine the **allowed energies and stationary states** of quantum systems such as

- particles confined in a box,
- particles encountering potential barriers,
- and harmonic oscillators.

These systems will reveal how **quantization naturally emerges from the Schrödinger equation**.
