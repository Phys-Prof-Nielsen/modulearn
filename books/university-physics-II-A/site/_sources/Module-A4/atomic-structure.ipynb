# A4.9 — Quantum Numbers and Atomic Structure
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Motivation

Throughout this module we have developed the key ingredients of quantum mechanics:

- the **Schrödinger equation**
- **quantized energy levels**
- **angular momentum**
- **spin**

These ideas come together when we study **electrons in atoms**.

Unlike the simple systems we studied earlier, atoms involve electrons moving in a **three-dimensional potential created by the nucleus**. Solving the Schrödinger equation for atoms leads to a set of **quantum numbers** that uniquely describe the allowed states of an electron.

These quantum numbers determine

- the **energy of the electron**
- the **shape of the orbital**
- the **orientation of the orbital**
- the **spin state of the electron**

Together they explain the structure of **atomic orbitals** and ultimately the organization of the **periodic table**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — The Four Quantum Numbers**

The state of an electron in an atom is described by four quantum numbers.

**Principal quantum number**

$$
n = 1,2,3,\dots
$$

Determines the **energy level** and approximate **size of the orbital**.

---

**Orbital angular momentum quantum number**

$$
\ell = 0,1,2,\dots,n-1
$$

Determines the **shape of the orbital**.

---

**Magnetic quantum number**

$$
m_\ell = -\ell,-\ell+1,\dots,\ell
$$

Determines the **orientation of the orbital in space**.

---

**Spin magnetic quantum number**

$$
m_s = +\frac{1}{2},\;-\frac{1}{2}
$$

Determines the **spin orientation of the electron**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Orbital Labels

The values of $\ell$ correspond to familiar orbital labels used in atomic physics:

| $\ell$ | Orbital label |
|------|---------------|
| 0 | s |
| 1 | p |
| 2 | d |
| 3 | f |

Thus

- **s orbitals** correspond to $\ell=0$
- **p orbitals** correspond to $\ell=1$
- **d orbitals** correspond to $\ell=2$
- **f orbitals** correspond to $\ell=3$

These orbitals represent **different spatial patterns of the electron wavefunction**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Number of Orbitals

For a given value of $\ell$, the number of orbitals is

$$
2\ell + 1.
$$

Thus:

| Orbital type | $\ell$ | Number of orbitals |
|---------------|------|--------------------|
| s | 0 | 1 |
| p | 1 | 3 |
| d | 2 | 5 |
| f | 3 | 7 |

Each orbital corresponds to a different value of $m_\ell$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Insight — Why Orbitals Have These Shapes**

The shapes of atomic orbitals arise from the **angular momentum properties of the electron wavefunction**.

Solving the Schrödinger equation in spherical coordinates produces solutions that depend on the angular momentum quantum numbers $\ell$ and $m_\ell$.

These solutions determine the **spatial patterns of the electron probability distribution**, which appear as the familiar **s, p, d, and f orbitals**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## The Pauli Exclusion Principle

Electrons are **fermions**, which means that no two electrons in an atom can occupy the same quantum state.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Pauli Exclusion Principle**

No two electrons in an atom can have the **same set of four quantum numbers**.

</div>

This principle determines how electrons fill atomic orbitals and ultimately explains the structure of the **periodic table**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — Orbitals Are Not Planetary Orbits**

Early models of the atom pictured electrons orbiting the nucleus like planets around the Sun.

In quantum mechanics, electrons do **not follow definite paths**.

Instead, orbitals represent **probability distributions** describing where an electron is likely to be found.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Orbitals in the $n=2$ Shell**

For $n=2$, the allowed values of $\ell$ are

$$
\ell = 0,1.
$$

Thus the orbitals are

- **2s** ($\ell=0$)
- **2p** ($\ell=1$)

For $\ell=1$, the magnetic quantum numbers are

$$
m_\ell = -1,0,1
$$

so there are **three 2p orbitals**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Counting States**

Consider the shell with principal quantum number

$$
n=3.
$$

1. List the allowed values of $\ell$.
2. Determine how many orbitals exist for each value of $\ell$.
3. How many **total orbitals** exist in the $n=3$ shell?
4. Since each orbital can hold two electrons, how many **electrons** can occupy the $n=3$ shell?

<details>
<summary>Solution</summary>

For $n=3$,

$$
\ell = 0,1,2.
$$

Thus the orbitals are

- $3s$ ($\ell=0$)
- $3p$ ($\ell=1$)
- $3d$ ($\ell=2$)

Number of orbitals:

- $s$ → 1
- $p$ → 3
- $d$ → 5

Total orbitals:

$$
1 + 3 + 5 = 9.
$$

Since each orbital can hold two electrons, the shell can hold

$$
18 \text{ electrons}.
$$

</details>

## Describing Atoms Using Orbital Notation

The four quantum numbers describe the **allowed states of electrons in atoms**, but chemists and physicists usually summarize these states using a more compact notation called the **electron configuration**.

Instead of listing all four quantum numbers for every electron, we group electrons according to the **orbitals they occupy**. Each orbital is labeled using the **principal quantum number $n$** and the **orbital type** determined by the angular momentum quantum number $\ell$.

For example:

- $\ell = 0$ corresponds to **s orbitals**
- $\ell = 1$ corresponds to **p orbitals**
- $\ell = 2$ corresponds to **d orbitals**
- $\ell = 3$ corresponds to **f orbitals**

Using this notation, we can quickly describe how electrons fill the available orbitals in an atom.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Electron Configuration Notation**

Atoms are commonly described using **electron configuration notation**, which indicates how electrons occupy atomic orbitals.

An orbital is labeled using the **principal quantum number $n$** and the **orbital type** determined by $\ell$.

Examples of orbital labels include

- $1s$
- $2s$
- $2p$
- $3d$

The number written as a superscript indicates how many electrons occupy that set of orbitals.

Examples:

- $1s^2$ → two electrons in the $1s$ orbital (the maximum it can have)
- $2p^3$ → three electrons distributed among the three $2p$ orbitals  (maximum is 6 electrons)
- $3d^{10}$ → ten electrons filling the five $3d$ orbitals (maximum is 10 electrons)

A full electron configuration lists the orbitals in order of increasing energy. For example, the electron configuration of carbon is

$$
1s^2 \; 2s^2 \; 2p^2
$$

This notation provides a compact way to describe how electrons occupy the allowed quantum states of an atom.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Filling Atomic Orbitals

Once we know the available orbitals in an atom, we must determine **how electrons occupy these orbitals**. Electrons do not fill orbitals randomly. Instead, their arrangement follows a set of rules that lead to the **lowest-energy configuration of the atom**.

Three key principles determine how electrons fill orbitals:

- the **Aufbau principle**
- the **Pauli exclusion principle**
- **Hund’s rule**

Together these rules determine the electron configurations of atoms and explain the structure of the **periodic table**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Aufbau Principle**

The **Aufbau principle** states that electrons occupy the **lowest-energy orbitals first** before filling orbitals with higher energy.

For example, the lowest orbitals fill in the order

$$
1s \rightarrow 2s \rightarrow 2p \rightarrow 3s \rightarrow 3p \rightarrow 4s \rightarrow 3d \rightarrow 4p \rightarrow \dots
$$

This ordering produces the **lowest possible total energy** for the atom.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Pauli Exclusion Principle**

No two electrons in an atom can have the **same set of four quantum numbers**.

As a result, each orbital can contain **at most two electrons**, and these electrons must have **opposite spins**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Hund’s Rule**

When electrons occupy **degenerate orbitals** (orbitals within the same subshell), they fill the orbitals **singly first**, with **parallel spins**, before any pairing occurs.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Insight — Why Hund’s Rule Works**

Electrons repel each other because they carry electric charge. When electrons spread out among different orbitals, they tend to remain **farther apart**, which reduces electron–electron repulsion and lowers the energy of the atom.

As a result, atoms prefer configurations where electrons occupy separate orbitals whenever possible.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Warning — Orbital Energy Overlap**

Although the Aufbau principle provides a useful guideline for filling orbitals, the actual **energy ordering of orbitals is not always perfectly sequential**.

In multi-electron atoms, interactions between electrons can cause **energy overlap between subshells**. As a result, an orbital with a higher value of $\ell$ can sometimes have **lower energy** than an orbital with a smaller $\ell$ in a higher shell.

For example,

$$
4s
$$

is typically filled **before**

$$
3d.
$$

Thus the filling order becomes

$$
\ldots \rightarrow 3p \rightarrow 4s \rightarrow 3d \rightarrow 4p \rightarrow \dots
$$

rather than strictly increasing with $n$.

Because of this overlap, the Aufbau principle should be understood as a **general guideline based on orbital energies**, not simply an ordering by the principal quantum number $n$.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Electron Configuration of Carbon**

Carbon has **6 electrons**.

To determine its electron configuration, we fill the available orbitals in order of increasing energy.

The lowest-energy orbitals are filled first:

$$
1s^2
$$

This accounts for **2 electrons**.

The next orbital is

$$
2s^2
$$

which adds **2 more electrons**.

We now have **4 electrons placed**, leaving **2 electrons remaining**.

The next available subshell is **2p**, which contains **three orbitals** and can hold up to **6 electrons**. The remaining two electrons occupy this subshell:

$$
2p^2
$$

Thus the full electron configuration of carbon is

$$
1s^2 \; 2s^2 \; 2p^2.
$$

According to **Hund’s rule**, electrons fill degenerate orbitals singly before pairing. Therefore the two electrons in the $2p$ subshell occupy **two different p orbitals** with parallel spins.

</div>

**Box Activity 2 — Electron Configuration of Oxygen**

Oxygen has **8 electrons**.

Use the **Aufbau principle**, **Pauli exclusion principle**, and **Hund’s rule** to determine how these electrons fill the available orbitals.

1. Write the **electron configuration** for oxygen.
2. Draw the **orbital diagram** showing how the electrons occupy the $2p$ orbitals.
3. How many **unpaired electrons** does oxygen have?

<details>
<summary>Solution</summary>

Oxygen has **8 electrons**.

Fill the orbitals in order of increasing energy.

First fill the $1s$ orbital:

$$
1s^2
$$

This accounts for **2 electrons**.

Next fill the $2s$ orbital:

$$
2s^2
$$

We now have **4 electrons placed**, leaving **4 electrons remaining**.

The next subshell is **$2p$**, which contains **three orbitals** and can hold up to **6 electrons**.

Thus we place the remaining electrons in the $2p$ subshell:

$$
2p^4
$$

The full electron configuration is

$$
1s^2 \; 2s^2 \; 2p^4.
$$

Using **Hund’s rule**, the first three electrons occupy the three $2p$ orbitals singly with parallel spins:

$$
\uparrow \quad \uparrow \quad \uparrow
$$

The fourth electron must then pair with one of them:

$$
\uparrow\downarrow \quad \uparrow \quad \uparrow
$$

Therefore oxygen has **two unpaired electrons**.

</details>

## Summary

Quantum mechanics describes electrons in atoms using **four quantum numbers**.

These quantum numbers determine

- the **energy level**
- the **shape of the orbital**
- the **orientation of the orbital**
- the **spin state of the electron**

Together they explain the structure of **atomic orbitals** and the organization of the **periodic table**.
