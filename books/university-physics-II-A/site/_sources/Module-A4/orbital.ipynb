# A4.7 — Orbital Angular Momentum in Quantum Mechanics
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Motivation

So far in this module we have studied quantum systems in **one spatial dimension**, such as the infinite square well and the harmonic oscillator. These systems helped us understand how the Schrödinger equation produces **quantized energy levels** and structured wavefunctions.

Many important physical systems, however, involve **rotational motion in three dimensions**. Examples include

- electrons moving in atoms,
- rotating molecules,
- orbital motion in central forces.

In classical mechanics, orbitalk angular momentum is defined by

$$
\mathbf{L} = \mathbf{r} \times \mathbf{p}.
$$

Angular momentum can point in **any direction** and its magnitude can take **any value**.

Quantum mechanics introduces a very different picture.

In quantum systems:

- the **magnitude of angular momentum is quantized**
- only certain **components of angular momentum** can be measured simultaneously
- angular momentum vectors cannot point in completely arbitrary directions.

Understanding angular momentum is essential for describing **atomic structure**, **spectroscopy**, and the organization of electrons in atoms.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Angular Momentum Operators

In quantum mechanics, angular momentum is represented by **operators**.

The three components are

$$
\hat{L}_x, \quad \hat{L}_y, \quad \hat{L}_z.
$$

These operators act on the wavefunction and correspond to measurements of angular momentum along the three coordinate axes.

In spherical coordinates, the operator for the $z$ component is

$$
\hat{L}_z = -i\hbar \frac{\partial}{\partial \phi}.
$$

This operator will play an important role when solving problems with **rotational symmetry**, such as the hydrogen atom.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Quantized Angular Momentum**

Angular momentum in quantum mechanics is described by two quantum numbers.

The **orbital angular momentum quantum number**

$$
\ell = 0,1,2,3,\dots
$$

determines the **magnitude of the angular momentum**

$$
L = \sqrt{\ell(\ell+1)}\,\hbar.
$$

The **magnetic quantum number**

$$
m_\ell = -\ell, -\ell+1, \dots , \ell-1, \ell
$$

determines the **component of angular momentum along the $z$-axis**

$$
L_z = m_\ell \hbar.
$$

For a given value of $\ell$, there are

$$
2\ell + 1
$$

possible orientations of the angular momentum vector.

Thus both the **magnitude** and **projection** of angular momentum are **quantized**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Allowed Orientations

For a given value of $\ell$, the quantum number $m_\ell$ determines the possible orientations of the angular momentum vector relative to the $z$ axis.

There are

$$
2\ell + 1
$$

possible orientations.

For example:

| $\ell$ | Possible $m_\ell$ values | Number of orientations |
|------|--------------------------|------------------------|
| 0 | 0 | 1 |
| 1 | -1, 0, 1 | 3 |
| 2 | -2, -1, 0, 1, 2 | 5 |

Thus the angular momentum vector can only take certain **discrete orientations in space**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Insight — Space Quantization**

In classical mechanics, angular momentum vectors can point in **any direction**.

Quantum mechanics restricts the allowed orientations.

This phenomenon is called **space quantization**.

The angular momentum vector still has a definite magnitude, but its projection along the $z$ axis can only take certain discrete values.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Measuring Angular Momentum

A famous experiment demonstrating the quantization of angular momentum is the **Stern–Gerlach experiment**.

In this experiment:

- a beam of particles passes through a **nonuniform magnetic field**
- the magnetic moment interacts with the field
- the beam splits into **discrete components**.

Instead of forming a continuous distribution, the particles appear in **separate spots**, corresponding to the allowed values of angular momentum projection.

This was one of the earliest experimental demonstrations of **quantization in nature**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — Angular Momentum Is Not a Classical Vector**

It is tempting to picture the angular momentum vector pointing in a precise direction with a well-defined magnitude and components.

In quantum mechanics, however, the components of angular momentum **cannot all be known simultaneously**.

Only the magnitude $L$ and **one component**, usually $L_z$, can be measured at the same time.

This is a consequence of the **non-commuting nature of the angular momentum operators**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Angular Momentum for $\ell = 1$**

For $\ell = 1$:

$$
L = \sqrt{2}\hbar
$$

The possible $L_z$ values are

$$
-\hbar, \quad 0, \quad \hbar.
$$

Thus there are **three possible orientations** of the angular momentum vector relative to the $z$ axis.

</div>

**Box Activity 1 — Allowed Angular Momentum States**

Consider a quantum state with angular momentum quantum number $\ell = 2$.

1. Calculate the magnitude of the angular momentum.
2. List all possible values of $m_\ell$.
3. How many possible orientations of the angular momentum vector exist?

<details>
<summary>Solution</summary>

The magnitude of the angular momentum is

$$
L = \sqrt{\ell(\ell+1)}\hbar.
$$

For $\ell = 2$:

$$
L = \sqrt{6}\hbar.
$$

The allowed $m_\ell$ values are

$$
-2, -1, 0, 1, 2.
$$

Thus there are

$$
2\ell + 1 = 5
$$

possible orientations of the angular momentum vector.
