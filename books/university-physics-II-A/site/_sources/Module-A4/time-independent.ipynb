# A4.3 — The Time-Independent Schrödinger Equation
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Motivation

In the previous module we introduced **operators** and **eigenvalue equations**, which form the mathematical language of quantum mechanics. Physical observables such as **position**, **momentum**, and **energy** are represented by operators acting on the **wavefunction**.

A central question in quantum mechanics is therefore:

**How does the wavefunction evolve in time, and how do we determine the allowed energy states of a system?**

The time evolution of a quantum system is governed by the **time-dependent Schrödinger equation**

$$
i\hbar \frac{\partial \psi(x,t)}{\partial t} = \hat{H}\psi(x,t)
$$

where $\hat{H}$ is the **Hamiltonian operator**, representing the total energy of the system.

For a particle moving in a potential $V(x)$, the Hamiltonian operator is

$$
\hat{H} = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2} + V(x).
$$

When the potential does **not depend on time**, the Schrödinger equation allows solutions that separate into spatial and temporal parts. These solutions take the form

$$
\psi(x,t) = \psi(x)e^{-iEt/\hbar}.
$$

Substituting this form into the time-dependent Schrödinger equation leads to an **eigenvalue equation for the Hamiltonian operator**. This equation is known as the **time-independent Schrödinger equation** and determines the allowed **stationary states** and **energy levels** of the system.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Time-Independent Schrödinger Equation**

For a particle of mass $m$ moving in a potential energy $V(x)$, the time-independent Schrödinger equation is

$$
-\frac{\hbar^2}{2m}\frac{d^2\psi(x)}{dx^2} + V(x)\psi(x) = E\psi(x)
$$

where

- $\psi(x)$ is the **wavefunction**
- $V(x)$ is the **potential energy**
- $E$ is the **energy eigenvalue**
- $\hbar = \frac{h}{2\pi}$ is the **reduced Planck constant**

Solutions $\psi(x)$ that satisfy this equation correspond to **stationary quantum states**, and the allowed values of $E$ are the **quantized energies** of the system.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Interpreting the Equation

The time-independent Schrödinger equation can be written compactly using operators as

$$
\hat{H}\psi = E\psi
$$

This is an **eigenvalue equation** for the Hamiltonian operator.

- $\hat{H}$ is the **energy operator**
- $\psi$ is an **energy eigenfunction**
- $E$ is the corresponding **energy eigenvalue**

Thus solving the Schrödinger equation means finding the wavefunctions that satisfy this equation and determining the corresponding allowed energies.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Insight — Why Quantization Appears**

In classical physics, a particle can have **any energy**.

In quantum mechanics, the Schrödinger equation combined with **boundary conditions** often restricts the allowed solutions. Only certain wave patterns "fit" into the system.

This is similar to **standing waves on a string**: only specific wavelengths are allowed between two fixed ends.

Because wavelength and momentum are related by the **de Broglie relation**

$$
p = \frac{h}{\lambda}
$$

only certain momenta and therefore **only certain energies** are possible.

Quantization therefore emerges naturally from **wave behavior and spatial constraints**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — The Wavefunction is Not a Physical Wave**

It is tempting to think of $\psi(x)$ as a literal physical wave like a water wave.

However, the wavefunction itself is **not directly observable**.

Instead, the measurable quantity is the **probability density**

$$
P(x) = |\psi(x)|^2
$$

which gives the probability of finding the particle at position $x$.

The wavefunction is therefore best understood as a **mathematical object encoding probability information** about the system.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Boundary Conditions

Not every mathematical solution of the Schrödinger equation is physically acceptable.

A valid wavefunction must satisfy several conditions:

1. **Finite everywhere**  
2. **Single-valued**  
3. **Continuous**  
4. **Normalizable**

The normalization condition requires that the total probability of finding the particle somewhere in space is equal to 1:

$$
\int |\psi(x)|^2 dx = 1
$$

These requirements eliminate many mathematical solutions and leave only those corresponding to **physically meaningful quantum states**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Free Particle**

Consider a particle moving in a region where the potential energy is zero:

$$
V(x) = 0
$$

The Schrödinger equation becomes

$$
-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2} = E\psi
$$

which can be rewritten as

$$
\frac{d^2\psi}{dx^2} + k^2\psi = 0
$$

where

$$
k = \frac{\sqrt{2mE}}{\hbar}.
$$

The solutions are traveling waves

$$
\psi(x) = Ae^{ikx} + Be^{-ikx}.
$$

These correspond to **matter waves** moving in opposite directions.

For a free particle there are **no boundary conditions restricting the wavelength**, so the energy is **not quantized**.

Quantization appears only when the particle is **confined**.

</div>

**Box Activity 1 — Interpreting the Schrödinger Equation**

Consider the time-independent Schrödinger equation

$$
-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2} + V(x)\psi = E\psi.
$$

Discuss the following questions with your group:

1. Which term in the equation represents the **kinetic energy contribution**?
2. Which term represents the **potential energy contribution**?
3. Why might the **curvature of the wavefunction** be related to the kinetic energy of the particle?

<details>
<summary>Solution</summary>

We analyze the time-independent Schrödinger equation

$$
-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2} + V(x)\psi = E\psi
$$

term by term.

**1. Kinetic energy term**

The kinetic energy contribution is

$$
-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2}.
$$

This term originates from replacing momentum in the classical kinetic energy formula

$$
K = \frac{p^2}{2m}
$$

with the momentum operator

$$
\hat{p} = -i\hbar\frac{d}{dx}.
$$

Squaring this operator gives

$$
\hat{p}^2 = -\hbar^2\frac{d^2}{dx^2}.
$$

Thus the first term represents the **kinetic energy operator**.

**2. Potential energy term**

The term

$$
V(x)\psi
$$

represents the **potential energy** contribution.

**3. Total energy**

The right-hand side

$$
E\psi
$$

represents the **total energy of the stationary state**.

Thus the Schrödinger equation is the quantum version of

$$
E = K + V.
$$

**4. Curvature and kinetic energy**

The kinetic energy depends on the **second derivative of the wavefunction**, which measures the **curvature**.

- large curvature → short wavelength  
- short wavelength → large momentum  
- large momentum → large kinetic energy

Thus the **shape of the wavefunction** directly reflects the energy of the particle.

</details>
