# A3.3 Relativistic Energy
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.1 Motivation: Classical Energy is Incomplete

In classical mechanics:

$$
K = \frac{1}{2}mv^2
$$

But this:

- Fails at high speeds ($ v \to c $)  
- Is not conserved in relativistic collisions  
- Does not match observed particle behavior (e.g. particle accelerators)

We need a **relativistically consistent** definition of energy.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.2 Deriving Total Relativistic Energy

We define **total energy** based on work and momentum:

Start with:

$$
\vec{F} = \frac{d\vec{p}}{dt} = \frac{d(\gamma m \vec{v})}{dt}
$$

The work done:

$$
dW = \vec{F} \cdot d\vec{r} = \frac{d\vec{p}}{dt} \cdot \vec{v} dt = \vec{v} \cdot d\vec{p}
$$

So the **total energy** is:

$$
E = \int \vec{v} \cdot d\vec{p}
$$

With some calculus (covered in advanced texts), this evaluates to:

$$
\boxed{E = \gamma m c^2}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.3 Rest Energy

When the object is **at rest** ($ v = 0 $, so $ \gamma = 1 $):

$$
\boxed{E_0 = mc^2}
$$

This is the **rest energy** — a fundamental insight of special relativity:

> Mass is a form of energy.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.4 Kinetic Energy in Special Relativity

We define **kinetic energy** as:

$$
K = E - E_0 = \gamma m c^2 - mc^2 = (\gamma - 1)mc^2
$$

<div class="alert alert-warning">
    
**Box 1: Kinetic Energy - Low-speed Limit**

Use the relativistic energy to show that the low-speed case reduces to the classical kinetic energy:

$$
K \approx \frac{1}{2}mv^2
$$

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.5 Energy-Momentum Relation

In special relativity, **energy** and **momentum** are not separate ideas — they are deeply connected through the structure of spacetime. Combining energy and momentum gives:

<div class="alert alert-warning">
    
**Box 2: Energy-Momentum Relation**

Prove the Energy-Momentum Relation:

$$
E^2 = p^2 c^2 + m^2 c^4
$$

</div>

```{admonition} Box 2: Solution
:class: tip, dropdown

The total relativistic energy is:

$$
E = \gamma m c^2
$$

The relativistic momentum is:

$$
p = \gamma m v
$$

To **combine energy and momentum into one equation**, we use the identity:

$$
\begin{aligned}
E^2 &= (\gamma m c^2)^2 = \gamma^2 m^2 c^4 \\
p^2 &= (\gamma m v)^2 = \gamma^2 m^2 v^2
\end{aligned}
$$

Multiply the momentum term by $ c^2 $:

$$
p^2 c^2 = \gamma^2 m^2 v^2 c^2
$$

Subtract:

$$
\begin{aligned}
E^2 - p^2 c^2 &= \gamma^2 m^2 c^4 - \gamma^2 m^2 v^2 c^2 \\
&= \gamma^2 m^2 c^2 (c^2 - v^2)
\end{aligned}
$$

Now recall:

$$
\gamma^2 = \frac{1}{1 - v^2/c^2} \quad \Rightarrow \quad \gamma^2 (c^2 - v^2) = c^2
$$

So:

$$
E^2 - p^2 c^2 = m^2 c^4
$$

---

### ✅ Final Result:

$$
\boxed{E^2 = p^2 c^2 + m^2 c^4}
$$

This is the **energy-momentum relation**, valid for all particles, massive or massless.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.6 Massless Particles (Photons)

For particles with **zero rest mass** (like photons):

$$
m = 0 \quad \Rightarrow \quad E = pc
$$

→ Photons **carry energy and momentum** despite having no mass.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.7 Example Problem 1: Kinetic Energy of a Fast Particle

A proton ($m = 938 \, \text{MeV}/c^2$) is moving at $v = 0.9c$. What is its kinetic energy?

### Solution:

Compute $ \gamma $:

$$
\gamma = \frac{1}{\sqrt{1 - v^2/c^2}} = \frac{1}{\sqrt{1 - 0.81}} = \frac{1}{\sqrt{0.19}} \approx 2.294
$$

Then:

$$
K = (\gamma - 1)mc^2 = (2.294 - 1)(938 \, \text{MeV}) \approx 1.294 \cdot 938 \approx 1,213 \, \text{MeV}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.8 Example Problem 2: Photon Momentum

A gamma-ray photon has energy $ E = 2.0 \, \text{MeV} $. What is its momentum?

### Solution:

Use:

$$
E = pc \Rightarrow p = \frac{E}{c}
$$

So:

$$
p = \frac{2.0 \, \text{MeV}}{c}
$$

(This is already in natural units — often written as $ 2.0 \, \text{MeV}/c $)

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A3.3.9 Summary

| Quantity        | Expression                     | Interpretation                       |
|----------------|----------------------------------|--------------------------------------|
| Total Energy    | $ E = \gamma mc^2 $           | Includes rest and kinetic energy     |
| Rest Energy     | $ E_0 = mc^2 $                | Intrinsic energy of mass             |
| Kinetic Energy  | $ K = (\gamma - 1) mc^2 $     | Increases rapidly as $ v \to c $   |
| Relation        | $ E^2 = p^2c^2 + m^2c^4 $     | Holds for all particles              |
| Photon Energy   | $ E = pc $                    | Massless particle limit              |

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Final Thoughts

- **Mass and energy are deeply related** — mass is a concentrated form of energy.
- The **energy-momentum relation** is central to particle physics and quantum field theory.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">
