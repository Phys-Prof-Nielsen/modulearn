## A1.3.1 Introduction

In the previous note, we derived the electromagnetic wave equation:

$$
\nabla^2 \mathbf{E} = \mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{E}}{\partial t^2}, \qquad
\nabla^2 \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{B}}{\partial t^2}
$$

The natural next step is to look for solutions.  
One of the simplest and most important solutions is the **plane wave**, which models light and radio waves in free space.

We will show:

✅ The **form** of the plane-wave solution.  
✅ The **transverse nature**: $\mathbf{E}$ and $\mathbf{B}$ are perpendicular to the direction of propagation.  
✅ The relationship between $\mathbf{E}$ and $\mathbf{B}$ magnitudes.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.3.2 The Plane-Wave Solution

A standard solution to the wave equation is of the form:

$$
\mathbf{E}(\mathbf{r},t) = \mathbf{E}_0 \cos(\mathbf{k}\cdot \mathbf{r} - \omega t + \phi)
$$

where:

- $\mathbf{E}_0$: Amplitude vector (constant)
- $\mathbf{k}$: **Wave vector**, direction of propagation, $|\mathbf{k}| = k = \frac{2\pi}{\lambda}$
- $\omega$: Angular frequency, $\omega = 2\pi f$
- $\phi$: Phase constant  

Similarly for $\mathbf{B}$:

$$
\mathbf{B}(\mathbf{r},t) = \mathbf{B}_0 \cos(\mathbf{k}\cdot \mathbf{r} - \omega t + \phi)
$$

In vacuum, the dispersion relation is:

$$
\omega = c k, \qquad c = \frac{1}{\sqrt{\mu_0 \varepsilon_0}}
$$

## A1.3.3 Transverse Nature of Electromagnetic Waves

<div class="alert alert-warning">
    
**Box 1: Divergence Conditions**

From Maxwell’s equations in free space:

**Gauss (Electric):** 

$$
\nabla \cdot \mathbf{E} = 0
$$

**Gauss (Magnetic):**

$$
\nabla \cdot \mathbf{B} = 0
$$

Show that

$$
\mathbf{k}\cdot \mathbf{E}_0 = 0
$$

and:

$$
\mathbf{k}\cdot \mathbf{B}_0 = 0
$$

</div>

### ✅ Interpretation

- **$\mathbf{E}$ and $\mathbf{B}$ are perpendicular to the direction of propagation** ($\mathbf{k}$).  
- Electromagnetic waves are **transverse waves**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.3.4 Relationship Between $\mathbf{E}$ and $\mathbf{B}$

<div class="alert alert-warning">
    
**Box 2**

Use Faraday's Law:

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}
$$

and use plane-wave forms: $e^{i(\mathbf{k}\cdot\mathbf{r} - \omega t)}$

to show that

$$
\mathbf{k}\times \mathbf{E}_0 = \omega \mathbf{B}_0
$$

</div>

### ✅ Interpretation

- $\mathbf{B}_0$ is **perpendicular to both $\mathbf{k}$ and $\mathbf{E}_0$**.

---

### Magnitude Relationship

Taking magnitudes:

$$
|\mathbf{B}_0| = \frac{k}{\omega} |\mathbf{E}_0| = \frac{1}{v} |\mathbf{E}_0| \quad \text{(since $\omega = v k$)}
$$

In vacuum $v = c$, so:

$$
B_0 = \frac{E_0}{c}
$$

Thus, the electric field dominates in magnitude, and the magnetic field is weaker by a factor of $c$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.3.5 Geometry of the Fields

- $\mathbf{E}$ ⟂ $\mathbf{B}$ ⟂ $\mathbf{k}$
- They form a **right-handed triad**:

$$
\mathbf{k} \propto \mathbf{E}\times \mathbf{B}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.3.6 Summary

1. Plane-wave solutions satisfy the electromagnetic wave equation.  
2. Maxwell’s equations enforce **transverse waves**:  
   - $\mathbf{k}\cdot\mathbf{E}=0$ and $\mathbf{k}\cdot\mathbf{B}=0$.  
3. $\mathbf{E}$, $\mathbf{B}$, and $\mathbf{k}$ are **mutually perpendicular**.  
4. In vacuum, $B_0 = E_0 / c$.
