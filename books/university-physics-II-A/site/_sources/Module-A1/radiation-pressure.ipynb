# A1.5 Radiation Pressure and Momentum Transfer in Electromagnetic Waves
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.5.1 Introduction and Motivation

Electromagnetic (EM) waves carry **momentum** in addition to energy.  
When EM waves strike a surface, they exert a **radiation pressure**.  

This effect is small but measurable, and it has important applications:

- **Solar sails** in space propulsion  
- **Optical tweezers** trapping microscopic particles  
- **Laser cooling of atoms**

In this lecture, we derive radiation pressure and relate it to the Poynting vector.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.5.2 Momentum Density of Electromagnetic Waves

### Energy-Momentum Relationship

Special relativity tells us that for any energy flow:

$$
\text{momentum density:} \; \mathbf{g} = \frac{\text{energy flux}}{c^2}
$$

For EM waves, using the **Poynting vector**:

$$
\mathbf{g} = \frac{\mathbf{S}}{c^2}
$$

with $\mathbf{S} = \frac{1}{\mu_0} (\mathbf{E}\times \mathbf{B})$.

- Direction of $\mathbf{g}$: **same as wave propagation**.  
- Units: kg/(m²·s).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.5.3 Radiation Pressure on Surfaces

Radiation pressure is **momentum transfer per unit time and unit area**:

$$
P = \frac{\Delta p}{\Delta t \, A} = \frac{\text{momentum flux}}{\text{area}}
$$

But momentum flux = energy flux / $c$ = $I/c$ (where $I = \langle S \rangle$ is intensity).

### Absorbing Surface

If the surface **fully absorbs** the radiation:

$$
P_{\text{abs}} = \frac{I}{c}
$$

---

### Perfectly Reflecting Surface

If the wave is **perfectly reflected**, the momentum change doubles:

$$
P_{\text{ref}} = \frac{2I}{c}
$$

(Photon momentum reverses direction, so $\Delta p = 2p$.)

---

### Mixed Reflection and Absorption

For partial reflection:

$$
P = \frac{(1+R)I}{c}
$$

where $R$ is the reflectivity ($0 \le R \le 1$).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.5.4 Photon Picture (Optional Insight)

For a single photon:

- Energy: $E_\gamma = h f$  
- Momentum: $p_\gamma = \frac{E_\gamma}{c} = \frac{h}{\lambda}$

Thus, intensity $I$ corresponds to the flux of photons hitting the surface.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.5.5 Example Order of Magnitude

**Solar Radiation Pressure:**

- Solar intensity near Earth: $I \approx 1360 \, \text{W/m}^2$  
- Absorbing surface:

$$
P_{\text{abs}} = \frac{I}{c} \approx \frac{1360}{3\times10^8} \approx 4.5 \times 10^{-6} \, \text{N/m}^2
$$

Though tiny, over large surfaces in space (solar sails), this can produce continuous acceleration.

## A1.5.6 Summary Table

| Surface Type | Radiation Pressure |
|--------------|--------------------|
| **Absorbing** | $P = \dfrac{I}{c}$ |
| **Perfectly Reflecting** | $P = \dfrac{2I}{c}$ |
| **Partial Reflection** | $P = \dfrac{(1+R)I}{c}$ |

## A1.5.7 Applications

✅ **Space Propulsion:** Solar sails use radiation pressure for thrust.  
✅ **Optical Tweezers:** Lasers trap and manipulate microscopic objects.  
✅ **Laser Cooling:** Momentum transfer from photons slows atomic motion.
