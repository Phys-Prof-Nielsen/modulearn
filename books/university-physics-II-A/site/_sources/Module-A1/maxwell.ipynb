## A1.1.1 Introduction and Motivation

Maxwell’s equations form the foundation of classical electromagnetism, unifying electricity, magnetism, and optics into one consistent theory.  

Before Maxwell, we had:

- **Gauss’s law for electricity** (Coulomb’s law reformulated)
- **Gauss’s law for magnetism** (no magnetic monopoles)
- **Faraday’s law of induction** (changing magnetic flux induces an electric field)
- **Ampère’s law** (currents create magnetic fields)

Maxwell’s crucial insight was the **displacement current**, which fixed an inconsistency in Ampère’s law and revealed that changing electric fields produce magnetic fields. This led to the prediction of **electromagnetic waves**—light itself.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A1.1.2 Review of the Fundamental Laws

### Gauss's Law for Electricity

The total electric flux through a closed surface is proportional to the enclosed electric charge:

$$
\oint \mathbf{E} \cdot d\mathbf{A} = \frac{Q_{\text{enc}}}{\varepsilon_0}
$$

- $ \mathbf{E} $: Electric field  
- $ d\mathbf{A} $: Outward-oriented surface element  
- $ Q_{\text{enc}} $: Total enclosed charge  
- $ \varepsilon_0 $: Vacuum permittivity  

---

### Gauss's Law for Magnetism

The total magnetic flux through a closed surface is always zero—there are no isolated magnetic charges (monopoles):

$$
\oint \mathbf{B} \cdot d\mathbf{A} = 0
$$

This expresses that magnetic field lines form closed loops.

---

### Faraday’s Law of Induction

A time-varying magnetic flux through a surface induces an electromotive force (EMF):

$$
\oint \mathbf{E} \cdot d\mathbf{l} = -\frac{d}{dt} \int \mathbf{B} \cdot d\mathbf{A}
$$

---

### Ampère’s Law (Pre-Maxwell)

Ampère originally stated that the circulation of $\mathbf{B}$ is proportional to the current passing through the surface:

$$
\oint \mathbf{B} \cdot d\mathbf{l} = \mu_0 I_{\text{enc}}
$$

## A1.1.3 Maxwell’s Displacement Current

### The Problem

Ampère’s law fails for **time-dependent situations**.  

Example: **Charging a capacitor**  
- Current flows in wires but not through the capacitor gap.  
- Yet $\mathbf{B}$ exists between the plates.  

How can Ampère’s law work if there is no real conduction current in the gap?

---

### Maxwell’s Insight

Maxwell proposed the existence of a **displacement current**:

$$
I_{\text{d}} = \varepsilon_0 \frac{d\Phi_E}{dt} \quad \text{where} \quad \Phi_E = \int \mathbf{E}\cdot d\mathbf{A}
$$

Adding this term, Ampère’s law becomes:

$$
\oint \mathbf{B} \cdot d\mathbf{l} = \mu_0 \left( I_{\text{enc}} + \varepsilon_0 \frac{d}{dt} \int \mathbf{E}\cdot d\mathbf{A} \right)
$$

---

### Why It Matters

This additional term is revolutionary:

- **Symmetry**: Changing $E$ fields produce $B$ fields (Faraday already showed changing $B$ produce $E$).
- **Electromagnetic waves**: Oscillating $E$ and $B$ can sustain each other, propagating through space at speed

$$
c = \frac{1}{\sqrt{\mu_0 \varepsilon_0}}
$$

Thus, **light is an electromagnetic wave**.

## A1.1.4 Maxwell’s Equations (Global / Integral Form)

Summarizing all four laws:

**Gauss (Electric):**
  
$$
\oint \mathbf{E} \cdot d\mathbf{A} = \frac{Q_{\text{enc}}}{\varepsilon_0}
$$

**Gauss (Magnetic):**

$$
\oint \mathbf{B} \cdot d\mathbf{A} = 0
$$

**Faraday:**

$$
\oint \mathbf{E} \cdot d\mathbf{l} = - \frac{d}{dt}\int \mathbf{B}\cdot d\mathbf{A}
$$

**Ampère–Maxwell:**

$$
\oint \mathbf{B} \cdot d\mathbf{l} = \mu_0 I_{\text{enc}} + \mu_0 \varepsilon_0 \frac{d}{dt}\int\mathbf{E}\cdot d\mathbf{A}
$$

## A1.1.5 Stokes’ Theorem and the Differential (Local) Form

### Stokes’ Theorem

Stokes’ theorem connects the integral of a vector field’s curl over a surface to its circulation along the boundary:

$$
\oint \mathbf{F}\cdot d\mathbf{l} = \iint (\nabla \times \mathbf{F})\cdot d\mathbf{A}
$$

Similarly, the divergence theorem converts surface integrals to volume integrals:

$$
\oint \mathbf{F}\cdot d\mathbf{A} = \iiint (\nabla\cdot \mathbf{F})\, dV
$$

---

## A1.1.6 Maxwell’s Equations (Local / Differential Form)

Using these theorems, Maxwell’s equations become:

**Gauss (Electric):**

$$
\nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0}
$$

**Gauss (Magnetic):**

$$
\nabla \cdot \mathbf{B} = 0
$$

**Faraday:**

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}
$$

**Ampère–Maxwell:**

$$
\nabla \times \mathbf{B} = \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}
$$
