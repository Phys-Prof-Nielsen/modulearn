## B1.5.2 Motivation: Why Do We Need Euler’s Equations?

When analyzing a **rotating rigid body**, Newton's second law for rotation is:

$$
\frac{d\vec{L}}{dt} = \vec{N}
$$

where:
- $ \vec{L} $ is angular momentum
- $ \vec{N} $ is torque
- The derivative is taken **in the inertial (space) frame**

But in most real problems (like analyzing a rotating spacecraft or top), it’s far more convenient to work in a **body-fixed coordinate system** — one that rotates with the object.

❗ The complication: in a rotating frame, the derivative of a vector isn’t just $ \frac{d}{dt} $, but includes the effect of the rotation of the frame itself.

Euler’s equations provide a way to write the dynamics in the **body frame** using the **principal axes**, where the inertia tensor is diagonal.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.5.3 Setup and Assumptions

- The body is rotating with angular velocity $ \vec{\omega} $ in its **own (body-fixed) frame**.
- The coordinate axes are aligned with the **principal axes** of the rigid body.
- The inertia tensor is **diagonal**:

$$
\mathbf{I} =
\begin{bmatrix}
I_1 & 0 & 0 \\
0 & I_2 & 0 \\
0 & 0 & I_3
\end{bmatrix}
$$

- Torque $ \vec{N} $ is expressed in the body frame.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.5.4 Derivation of Euler’s Equations

We start with:

$$
\left( \frac{d\vec{L}}{dt} \right)_{\text{space}} = \vec{N}
$$

But in the rotating (body) frame:

$$
\left( \frac{d\vec{L}}{dt} \right)_{\text{space}} =
\left( \frac{d\vec{L}}{dt} \right)_{\text{body}} + \vec{\omega} \times \vec{L}
$$

So:

$$
\left( \frac{d\vec{L}}{dt} \right)_{\text{body}} = \vec{N} - \vec{\omega} \times \vec{L}
$$

Since $ \vec{L} = \mathbf{I} \vec{\omega} $ and $ \mathbf{I} $ is diagonal in the principal frame:

$$
\begin{aligned}
L_1 &= I_1 \omega_1 \\
L_2 &= I_2 \omega_2 \\
L_3 &= I_3 \omega_3
\end{aligned}
$$

Then compute componentwise:

$$
\begin{aligned}
I_1 \dot{\omega}_1 &= N_1 + (I_2 - I_3)\omega_2 \omega_3 \\
I_2 \dot{\omega}_2 &= N_2 + (I_3 - I_1)\omega_3 \omega_1 \\
I_3 \dot{\omega}_3 &= N_3 + (I_1 - I_2)\omega_1 \omega_2
\end{aligned}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.5.5 Euler’s Equations (Final Form)

$$
\boxed{
\begin{aligned}
I_1 \dot{\omega}_1 &= N_1 + (I_2 - I_3) \omega_2 \omega_3 \\
I_2 \dot{\omega}_2 &= N_2 + (I_3 - I_1) \omega_3 \omega_1 \\
I_3 \dot{\omega}_3 &= N_3 + (I_1 - I_2) \omega_1 \omega_2
\end{aligned}
}
$$

These are **Euler’s equations of motion**, describing how the angular velocity vector evolves in the **body frame**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.5.6 Special Case: Torque-Free Motion

If $ \vec{N} = 0 $, then:

$$
\begin{aligned}
I_1 \dot{\omega}_1 &= (I_2 - I_3) \omega_2 \omega_3 \\
I_2 \dot{\omega}_2 &= (I_3 - I_1) \omega_3 \omega_1 \\
I_3 \dot{\omega}_3 &= (I_1 - I_2) \omega_1 \omega_2
\end{aligned}
$$

- The angular velocity components are **non-constant**, even though no external torque is acting.
- The motion can be **periodic**, **chaotic**, or **stable**, depending on the body's symmetry.

This is the situation for a freely spinning satellite or an isolated asteroid.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.5.7 Example: Symmetric Top

If $ I_1 = I_2 \neq I_3 $, then:

$$
\begin{aligned}
I_1 \dot{\omega}_1 &= (I_1 - I_3) \omega_2 \omega_3 \\
I_1 \dot{\omega}_2 &= (I_3 - I_1) \omega_1 \omega_3 \\
I_3 \dot{\omega}_3 &= 0
\end{aligned}
$$

- $ \omega_3 $ is constant.
- $ \omega_1 $ and $ \omega_2 $ oscillate.
- This leads to **precession** of the angular velocity vector about the symmetry axis.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">


## B1.5.8 Summary

| Concept                 | Expression |
|-------------------------|------------|
| Angular momentum        | $ \vec{L} = \mathbf{I} \vec{\omega} $ |
| Time derivative in body frame | $ \left( \frac{d\vec{L}}{dt} \right)_{\text{space}} = \left( \frac{d\vec{L}}{dt} \right)_{\text{body}} + \vec{\omega} \times \vec{L} $ |
| Euler’s equations       | $ I_i \dot{\omega}_i = N_i + (I_j - I_k) \omega_j \omega_k $ |
| Torque-free case        | Euler’s equations with \( \vec{N} = 0 \) |
| Stability insight       | Rotation about largest or smallest inertia axis is stable |

### Box 1

#### Step 1: Euler’s Equations

Euler’s equations for torque-free motion are:

$$
\begin{aligned}
I_1 \dot{\omega}_1 &= (I_2 - I_3) \omega_2 \omega_3 \\
I_2 \dot{\omega}_2 &= (I_3 - I_1) \omega_3 \omega_1 \\
I_3 \dot{\omega}_3 &= (I_1 - I_2) \omega_1 \omega_2
\end{aligned}
$$

---

#### Step 2: Plug in $I_1 = I_2 = I_3 = I$

Substitute into each equation:

$$
\begin{aligned}
I \dot{\omega}_1 &= (I - I) \omega_2 \omega_3 = 0 \\
I \dot{\omega}_2 &= (I - I) \omega_3 \omega_1 = 0 \\
I \dot{\omega}_3 &= (I - I) \omega_1 \omega_2 = 0
\end{aligned}
$$

This simplifies to:

$$
\dot{\omega}_1 = \dot{\omega}_2 = \dot{\omega}_3 = 0
$$

---

#### Step 3: Interpretation

Since all derivatives are zero:

- The components of $\vec{\omega}$ in the **body frame** are constant:

  $$
  \vec{\omega}_\text{body}(t) = \vec{\omega}_\text{body}(0)
  $$

- But if $\vec{\omega}$ is **constant in the body frame**, and the body is rotating, then the vector must also point in a **fixed direction** in space.

Hence, $\vec{\omega}$ is also constant in the **space (inertial) frame**:

- This means the **rotation axis does not precess**.
- The body spins uniformly about a fixed axis in space — like a **perfect sphere** or symmetric

### Box 2

#### Step 1: Euler’s Equations (Torque-Free)

For a rigid body with no external torque, Euler’s equations are:

$$
\begin{aligned}
I_1 \dot{\omega}_1 &= (I_2 - I_3) \omega_2 \omega_3 \\
I_2 \dot{\omega}_2 &= (I_3 - I_1) \omega_3 \omega_1 \\
I_3 \dot{\omega}_3 &= (I_1 - I_2) \omega_1 \omega_2
\end{aligned}
$$

---

#### Step 2: Plug in Given Values

Substitute values:

- $I_1 = 2$, $I_2 = 3$, $I_3 = 4$
- $\omega_1 = 1$, $\omega_2 = 0$, $\omega_3 = 2$

**Equation 1**:

$$
2 \dot{\omega}_1 = (3 - 4)(0)(2) = (-1)(0) = 0 \quad \Rightarrow \quad \dot{\omega}_1 = 0
$$

**Equation 2**:

$$
3 \dot{\omega}_2 = (4 - 2)(2)(1) = (2)(2)(1) = 4 \quad \Rightarrow \quad \dot{\omega}_2 = \frac{4}{3}
$$

**Equation 3**:

$$
4 \dot{\omega}_3 = (2 - 3)(1)(0) = (-1)(1)(0) = 0 \quad \Rightarrow \quad \dot{\omega}_3 = 0
$$

---

#### Final Answer

At $t = 0$:

$$
\dot{\omega}_1 = 0, \quad \dot{\omega}_2 = \frac{4}{3}, \quad \dot{\omega}_3 = 0
$$

- The body is momentarily gaining angular velocity about axis 2.
- Axes 1 and 3 show no instantaneous acceleration at this instant.

### Box 3

### Conceptual Background

Let a rigid body have three distinct principal moments of inertia:

$$
I_1 < I_2 < I_3
$$

with corresponding principal axes.

Euler's equations (torque-free) in the body frame are:

$$
\begin{aligned}
I_1 \dot{\omega}_1 &= (I_2 - I_3) \omega_2 \omega_3 \\
I_2 \dot{\omega}_2 &= (I_3 - I_1) \omega_3 \omega_1 \\
I_3 \dot{\omega}_3 &= (I_1 - I_2) \omega_1 \omega_2
\end{aligned}
$$

We examine the **stability** of pure rotation about each axis by **linearizing** these equations under small perturbations.

---

### Step-by-Step Stability Argument

#### Detailed Linearization: Case A — Rotation About the Minimum Axis $I_1$

Let’s carefully analyze the **stability of rotation about the principal axis with the smallest moment of inertia**, $I_1$.

We assume the body is **initially spinning only about axis 1**, and that the other two angular velocity components are **small perturbations**:

$$
\omega_1(t) = \Omega \quad (\text{constant}), \qquad \omega_2(t),\, \omega_3(t) \ll 1
$$

---

### Step 1: Euler’s Equations (Torque-Free)

Recall:

$$
\begin{aligned}
I_1 \dot{\omega}_1 &= (I_2 - I_3) \omega_2 \omega_3 \\
I_2 \dot{\omega}_2 &= (I_3 - I_1) \omega_3 \omega_1 \\
I_3 \dot{\omega}_3 &= (I_1 - I_2) \omega_1 \omega_2
\end{aligned}
$$

---

### Step 2: Plug in Assumptions

Since $\omega_1 = \Omega$ is constant and $\omega_2, \omega_3$ are small:

- $\omega_2 \omega_3$ is second-order small, so we **neglect** it in the first equation:
  
  $$
  \dot{\omega}_1 \approx 0
  $$

- The other two equations become:

  $$
  \begin{aligned}
  I_2 \dot{\omega}_2 &= (I_3 - I_1) \Omega \omega_3 \\
  I_3 \dot{\omega}_3 &= (I_1 - I_2) \Omega \omega_2
  \end{aligned}
  $$

---

### Step 3: Derive a Second-Order Equation

Differentiate the first equation:

$$
I_2 \ddot{\omega}_2 = (I_3 - I_1) \Omega \dot{\omega}_3
$$

Now substitute $\dot{\omega}_3$ from the second equation:

$$
\dot{\omega}_3 = \frac{(I_1 - I_2) \Omega}{I_3} \omega_2
$$

Plug into the equation for $\ddot{\omega}_2$:

$$
I_2 \ddot{\omega}_2 = (I_3 - I_1) \Omega \cdot \frac{(I_1 - I_2) \Omega}{I_3} \omega_2
$$

Simplify:

$$
\ddot{\omega}_2 = \left[ \frac{(I_3 - I_1)(I_1 - I_2)}{I_2 I_3} \Omega^2 \right] \omega_2
= \lambda^2 \omega_2
$$

---

### Step 4: Determine Stability

Now examine the sign of:

$$
\lambda^2 = \frac{(I_3 - I_1)(I_1 - I_2)}{I_2 I_3} \Omega^2
$$

We assume $I_1 < I_2 < I_3$:

- $(I_3 - I_1) > 0$
- $(I_1 - I_2) < 0$

So the numerator is negative ⇒

$$
\lambda^2 < 0
$$

This leads to the differential equation:

$$
\ddot{\omega}_2 = -|\lambda^2| \omega_2
$$

✅ **Solution is oscillatory** ⇒ the perturbation $\omega_2$ stays bounded.

✅ **Conclusion**: rotation about axis 1 is **stable**.

---

### Summary of Procedure

This is the general method:

1. Assume pure rotation about one axis, perturb the other two.
2. Linearize Euler’s equations by dropping second-order small terms.
3. Derive a second-order differential equation for the perturbation.
4. Analyze the sign of $\lambda^2$:
   - $\lambda^2 < 0$ ⇒ stable
   - $\lambda^2 > 0$ ⇒ unstable

The same steps can be applied to Cases B and C (rotation about axes $I_2$ and $I_3$).

---

#### Case B: Rotation about the **maximum inertia axis** ($I_3$)

Assume:

- $\omega_3 = \Omega$
- $\omega_1, \omega_2 \approx 0$

Similarly:

$$
\lambda^2 = \frac{(I_2 - I_1)(I_3 - I_2)}{I_1 I_2} \Omega^2
$$

Here both terms are **positive** ⇒ $\lambda^2 < 0$ again ⇒ **stable**

---

#### Case C: Rotation about the **intermediate axis** ($I_2$)

Assume:

- $\omega_2 = \Omega$
- $\omega_1, \omega_3 \approx 0$

Then:

$$
\lambda^2 = \frac{(I_2 - I_3)(I_1 - I_2)}{I_1 I_3} \Omega^2
$$

Now both terms in the numerator are **negative**:

- $I_2 - I_3 < 0$
- $I_1 - I_2 < 0$

Product is **positive** ⇒ $\lambda^2 > 0$ ⇒ **exponential growth** ⇒ **unstable**

---

### Interpretation: The Tennis Racket Theorem

- An object thrown spinning about its **intermediate axis** will flip — not because of external forces, but because that axis is **dynamically unstable**.
- Even tiny perturbations in the other components grow, causing a dramatic shift in the orientation of the body.
- This is **not chaos**, but a **geometrically constrained instability**.

In contrast:

- Rotating about the **shortest or longest axis** leads to bounded, periodic deviations — these axes are **dynamically stable**.

---

### Final Insight

| Axis              | Moment of Inertia | Stability   | Behavior                     |
|-------------------|--------------------|-------------|------------------------------|
| $I_1$ (short)      | Minimum            | ✅ Stable    | Oscillatory, bounded         |
| $I_2$ (intermediate) | Intermediate     | ❌ Unstable  | Exponential divergence → flips |
| $I_3$ (long)       | Maximum            | ✅ Stable    | Oscillatory, bounded         |

**Conclusion**:  
Stability arises when energy and angular momentum surfaces intersect in closed loops. The intermediate axis corresponds to a **saddle point** — perturbations grow because the angular velocity vector $\vec{\omega}$ is geometrically “pushed away” from that configuration.
