## B1.3.1 Motivation: Why Do We Need Euler Angles?

In classical mechanics and aerospace engineering, understanding how a **rigid body rotates in 3D** is crucial. Imagine a satellite, a robot arm, or even a spinning top — to fully describe how the body **is oriented in space**, we need more than just angular velocity:

- Vectors like angular velocity $ \vec{\omega} $ or angular momentum $ \vec{L} $ describe **rates of change**, not orientation.
- The **rotation matrix** tells us how the body is oriented relative to a fixed coordinate system, but it's often too abstract or difficult to interpret physically.
- We need a **minimal, interpretable way** to describe orientation that aligns with real-world intuition (e.g. "pitch," "yaw," "roll").

**Euler angles** provide exactly this:  
A systematic way to express the orientation of a rigid body using just **three angles**, corresponding to **three sequential rotations**.

## B1.3.3 What Are Euler Angles?

Euler angles describe the orientation of a rigid body using **three sequential rotations** about specified axes. There are several conventions, but the most common (used in physics) is the **ZXZ convention**:

1. **$ \phi $** (phi): rotate about the **Z-axis** (precession)
2. **$ \theta $** (theta): rotate about the **new X-axis** (nutation)
3. **$ \psi $** (psi): rotate about the **new Z-axis** (spin)

Each of these angles corresponds to a physically meaningful motion:

- $ \phi $: Orientation of the body around the fixed vertical axis.
- $ \theta $: Tilt angle from the vertical (zenith angle).
- $ \psi $: Rotation about the body's own symmetry axis.

## B1.3.4 Euler Rotation Sequence (ZXZ Convention)

Let’s define each step of the rotation:

1. **Rotate by $ \phi $ about the fixed Z-axis**:

$$
R_z(\phi) =
\begin{bmatrix}
\cos \phi & -\sin \phi & 0 \\
\sin \phi & \cos \phi & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

2. **Rotate by $ \theta $ about the new X-axis**:

$$
R_x(\theta) =
\begin{bmatrix}
1 & 0 & 0 \\
0 & \cos \theta & -\sin \theta \\
0 & \sin \theta & \cos \theta
\end{bmatrix}
$$

3. **Rotate by $ \psi $ about the new Z-axis**:

$$
R_z(\psi) =
\begin{bmatrix}
\cos \psi & -\sin \psi & 0 \\
\sin \psi & \cos \psi & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

The total rotation matrix is:

$$
R = R_z(\phi) \cdot R_x(\theta) \cdot R_z(\psi)
$$

## B1.3.5 Clarifying the Z-Axes in Euler Angles (ZXZ Convention)

Euler angles can be confusing — especially because in the **ZXZ** convention, we use the **Z-axis twice**: once as a fixed axis and once as a body axis.

| Step | Axis Rotated About | Type of Axis        | Angle     |
|------|---------------------|---------------------|-----------|
| 1    | Fixed Z             | Space frame axis    | $ \phi $ (precession) |
| 2    | New X               | Intermediate frame  | $ \theta $ (tilt)     |
| 3    | New Z (body Z)      | Body frame axis     | $ \psi $ (spin)       |



### What’s Actually Happening?

Euler angles describe how the **coordinate system rotates**. So at each step, the rotation axis may belong to a different frame:

- **$ \phi $**: Rotation about the **fixed Z-axis** in the space frame.
- **$ \theta $**: Rotation about the **new X-axis** (after the first rotation).
- **$ \psi $**: Rotation about the **new Z-axis**, which is now aligned with the **body's own axis**.

While these axes are all labeled "Z" or "X", they **represent different directions** at different stages of rotation.



### Why Reuse “Z”?

We reuse the letter "Z" because:

- In **step 1**, the Z-axis is part of the **inertial space frame**.
- In **step 3**, the Z-axis is part of the **rotated body frame**.
- The axes share a name because they are **both aligned with Z at the start**, but diverge after rotation.


### Tip for Clarity

To avoid confusion:

- Explicitly label the axes: $ Z_{\text{space}} $, $ X_{\text{intermediate}} $, $ Z_{\text{body}} $
- Visualize the process step-by-step:
  - Start with the fixed (space) frame.
  - Apply each rotation and track how the axes move.
- Emphasize that **the axes themselves rotate** in intrinsic conventions.

This distinction is crucial for understanding angular velocity decomposition and preventing mistakes in coordinate transformations.

## B1.3.6 Interpreting the Euler Angles Physically

Euler angles provide a way to describe **attitude**:

- A **gyroscope** spinning on its own axis exhibits all three:
  - $ \phi $: how the base spins on the table (precession)
  - $ \theta $: the lean angle
  - $ \psi $: spin around its own shaft

In spacecraft dynamics:
- $ \phi $: yaw (azimuth)
- $ \theta $: pitch (elevation)
- $ \psi $: roll (axial spin)

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.3.7 Gimbal Lock

**Gimbal lock** occurs when two axes in the rotation sequence align, causing a loss of one degree of freedom.

- For ZXZ, this happens when $ \theta = 0 $ or $ \pi $: the first and third rotations happen about the same axis.
- Physically, this means we cannot distinguish between changes in $ \phi $ and $ \psi $.

➡️ In such cases, alternative representations like **quaternions** are used in robotics and aerospace.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.3.8 Angular Velocity in Terms of Euler Angles

In the body-fixed frame, the angular velocity vector $ \vec{\omega} $ can be written as:

$$
\vec{\omega} = \dot{\phi} \, \hat{z} + \dot{\theta} \, \hat{x}' + \dot{\psi} \, \hat{z}''
$$

More precisely (in body coordinates):

$$
\vec{\omega} =
\begin{bmatrix}
\dot{\theta} \cos \psi + \dot{\phi} \sin \theta \sin \psi \\
\dot{\theta} \sin \psi - \dot{\phi} \sin \theta \cos \psi \\
\dot{\psi} + \dot{\phi} \cos \theta
\end{bmatrix}
$$

This is crucial for expressing **rotational kinetic energy** and deriving **Euler’s equations of motion** in rigid body dynamics.

## B1.3.9 Summary

| Symbol | Meaning in ZXZ Convention                  |
|--------|---------------------------------------------|

| $ \theta $ | Rotation about new **X-axis** (tilt/nutation)     |
| $ \psi $   | Rotation about new **Z-axis** (body spin)         |
| $ R $      | Total rotation matrix from Euler angles           |
| $ \vec{\omega} $ | Angular velocity in terms of $ \phi, \theta, \psi $ |

### Box 1

**Solution**:

In the **ZXZ Euler angle convention**, the orientation of a rigid body is described by three successive rotations:

1. **$ \phi $ — Precession**:
   - This is a rotation about the **fixed Z-axis**.
   - It describes how the body turns horizontally in space — for example, the circular motion of a spinning top's axis.
   - Often associated with "yaw" in aerospace.

2. **$ \theta $ — Nutation (Tilt)**:
   - This is a rotation about the **new X-axis** after the first rotation.
   - It sets the tilt angle between the body’s symmetry axis and the vertical.
   - Often interpreted as "pitch" — it tilts the axis away from vertical.

3. **$ \psi $ — Spin**:
   - This is a rotation about the **new Z-axis** (now aligned with the body's axis).
   - It represents the body's own rotation about its axis — for instance, how a top spins on itself.
   - In aerospace, this is typically associated with "roll."

**Summary**:

| Angle   | Axis Rotated About        | Physical Interpretation |
|---------|---------------------------|--------------------------|
| $ \phi $ | Fixed Z-axis               | Precession (azimuth)     |
| $ \theta $ | New X-axis after $ \phi $ | Tilt/Nutation (inclination) |
| $ \psi $ | New Z-axis after $ \theta $ | Spin (axial rotation)    |

### Box 2

**Solution**:

We apply three successive rotations in the **ZXZ** convention:

#### Step 1: Rotation by $ \phi $ about the fixed Z-axis

$$
R_z(\phi) =
\begin{bmatrix}
\cos \phi & -\sin \phi & 0 \\
\sin \phi & \cos \phi & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

#### Step 2: Rotation by $ \theta $ about the new X-axis

$$
R_x(\theta) =
\begin{bmatrix}
1 & 0 & 0 \\
0 & \cos \theta & -\sin \theta \\
0 & \sin \theta & \cos \theta
\end{bmatrix}
$$

#### Step 3: Rotation by $ \psi $ about the new Z-axis

$$
R_z(\psi) =
\begin{bmatrix}
\cos \psi & -\sin \psi & 0 \\
\sin \psi & \cos \psi & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

The total rotation matrix is:

$$
R = R_z(\phi) \cdot R_x(\theta) \cdot R_z(\psi)
$$

Performing the matrix multiplication gives:

$$
R =
\begin{bmatrix}
\cos \phi \cos \psi - \sin \phi \cos \theta \sin \psi &
- \cos \phi \sin \psi - \sin \phi \cos \theta \cos \psi &
\sin \phi \sin \theta \\
\sin \phi \cos \psi + \cos \phi \cos \theta \sin \psi &
- \sin \phi \sin \psi + \cos \phi \cos \theta \cos \psi &
- \cos \phi \sin \theta \\
\sin \theta \sin \psi & \sin \theta \cos \psi & \cos \theta
\end{bmatrix}
$$

**This matrix** transforms coordinates from the **body frame to the fixed space frame** using ZXZ Euler angles.

### Box 3

**Solution**:

We analyze the motion of a spinning top:

1. **Spin about its own axis**:
   - This corresponds to **$ \psi $**.
   - It describes how the top rotates around its own symmetry axis, which is not necessarily vertical.

2. **Cone-shaped precession**:
   - This corresponds to **$ \phi $**.
   - The top’s axis traces a cone as it rotates around the vertical (fixed Z-axis) — this is **precession**.

3. **Lean angle of the axis**:
   - This is the **tilt** of the top away from vertical and corresponds to **$ \theta $**.
   - It measures the angle between the symmetry axis of the top and the vertical axis.

**Angle Interpretations**:

| Description                     | Euler Angle | Axis Rotated About |
|---------------------------------|-------------|---------------------|
| Spin of the top                | $ \psi $   | Body Z-axis         |
| Precession (conical motion)     | $ \phi $   | Fixed Z-axis        |
| Lean or tilt from vertical      | $ \theta $ | New X-axis after $ \phi $ |
