# B1.4 Body and Space Frames

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.4.1 Motivation

When describing the motion of rigid bodies—such as spinning tops, gyroscopes, or satellites—we must be able to express how the object **rotates** and **moves** through space. However, describing this motion using a single coordinate system quickly becomes cumbersome, since the body itself is rotating relative to the observer.  

To simplify analysis, physicists introduce two coordinate systems:

1. **Space Frame (Inertial Frame)** — fixed in space, usually associated with the laboratory or an external observer.
2. **Body Frame (Rotating Frame)** — attached to the rigid body itself, moving and rotating with it.

By using both frames, we can describe the orientation and motion of the body relative to both its own structure and the external world. This dual viewpoint becomes essential when we formulate **Euler’s equations** for rotational motion, since those equations are naturally expressed in the **body-fixed frame**, while energy and angular momentum are often evaluated in the **space frame**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.4.2 Space (Inertial) Frame

The **space frame**, sometimes called the **laboratory frame**, is a fixed, non-accelerating reference frame.  
In this frame, Newton’s laws take their standard form. We denote it by axes $(X, Y, Z)$, and all quantities measured in this frame are denoted with respect to these axes.

For instance, the angular velocity vector $\boldsymbol{\omega}$ of a rotating body, when expressed in the space frame, is written as:

$$
\boldsymbol{\omega} = \omega_X \hat{X} + \omega_Y \hat{Y} + \omega_Z \hat{Z}.
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.4.3 Body (Rotating) Frame

The **body frame** is rigidly attached to the rotating body itself. Its axes $(x, y, z)$ are fixed relative to the body — meaning that any point on the body has fixed coordinates in this frame.  

However, the **orientation** of the body frame changes with time relative to the space frame. For example, a gyroscope spinning about its own axis has a constant orientation in its **body frame**, but in the **space frame**, its spin axis traces out a cone due to precession.

A key advantage of the body frame is that the **moment of inertia tensor** is often constant and diagonal in this frame:

$$
\mathbf{I}_{\text{body}} =
\begin{pmatrix}
I_1 & 0 & 0 \\
0 & I_2 & 0 \\
0 & 0 & I_3
\end{pmatrix}.
$$

This makes the equations of motion much simpler to express and solve.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.4.4 Transformation Between Frames

The orientation of the body frame relative to the space frame can be represented by a **rotation matrix** $\mathbf{R}$, which transforms a vector expressed in body coordinates to its representation in the space frame:

$$
\mathbf{v}_{\text{space}} = \mathbf{R} \, \mathbf{v}_{\text{body}}.
$$

The matrix $\mathbf{R}$ is an **orthogonal** matrix with determinant +1 (a member of the rotation group $SO(3)$), satisfying:

$$
\mathbf{R}^T \mathbf{R} = \mathbf{I}, \qquad \det(\mathbf{R}) = 1.
$$

As the body rotates, $\mathbf{R}$ changes with time. Its time derivative is related to the angular velocity vector $\boldsymbol{\omega}$ through:

$$
\dot{\mathbf{R}} = \boldsymbol{\Omega} \, \mathbf{R},
$$

where $\boldsymbol{\Omega}$ is the **skew-symmetric matrix** associated with $\boldsymbol{\omega}$:

$$
\boldsymbol{\Omega} =
\begin{pmatrix}
0 & -\omega_z & \omega_y \\
\omega_z & 0 & -\omega_x \\
-\omega_y & \omega_x & 0
\end{pmatrix}.
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.4.5 Time Derivative of a Vector in a Rotating Frame

If a vector $\mathbf{A}$ is expressed in the body frame, its time derivative observed from the space frame is given by the important relation:

$$
\left( \frac{d\mathbf{A}}{dt} \right)_{\text{space}} =
\left( \frac{d\mathbf{A}}{dt} \right)_{\text{body}} + \boldsymbol{\omega} \times \mathbf{A}.
$$

This result plays a central role in deriving Euler’s equations for rigid-body rotation.  
It tells us that even if a vector $\mathbf{A}$ is constant in the **body frame**, it may still change in the **space frame** due to the rotation of the body itself.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.4.6 Example: Angular Momentum in Both Frames

The angular momentum of a rigid body is given by:

$$
\mathbf{L} = \mathbf{I} \boldsymbol{\omega}.
$$

In the **space frame**, both $\mathbf{I}$ and $\boldsymbol{\omega}$ can change with time, making $\mathbf{L}$ difficult to compute.  
However, in the **body frame**, the inertia tensor $\mathbf{I}_{\text{body}}$ is constant, and only $\boldsymbol{\omega}$ changes with time.

To relate the two:

$$
\mathbf{L}_{\text{space}} = \mathbf{R} \, \mathbf{L}_{\text{body}}, \qquad
\mathbf{I}_{\text{space}} = \mathbf{R} \, \mathbf{I}_{\text{body}} \, \mathbf{R}^T.
$$

## B1.4.7  Physical Interpretation

The distinction between **space** and **body** frames allows us to separate:
- the **external motion** of the body in space (translation and orientation), and  
- the **internal motion** relative to its own axes (spinning, wobbling, nutation).

This dual framework forms the foundation for describing complex rotational behaviors such as **precession** and **nutation**, which we will later analyze using **Euler angles** and **Euler’s equations**.
