## B3.6.1 Motivation

The spring pendulum is one of the richest systems in classical mechanics. It consists of a mass attached to a spring, where the mass is free both to stretch the spring and to swing like a pendulum.

At first glance, this system appears simple. It is just a mass, a spring, and gravity. However, its motion can exhibit an extraordinary range of behavior:

- nearly independent spring and pendulum motion,
- energy exchange between radial and angular motion,
- internal resonance,
- autoparametric resonance,
- phase locking between modes,
- complicated trajectories in the $(x,y)$ plane,
- Poincaré sections,
- and chaotic motion.

In this capstone project, we will use the spring pendulum as a miniature laboratory for nonlinear dynamics. The goal is not simply to solve one problem, but to investigate a system the way a physicist would: by deriving a model, simulating the equations, visualizing the motion, and interpreting the results.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Spring Pendulum</strong>

A spring pendulum is a system in which a mass attached to a spring can move both radially and angularly.

The radial coordinate describes stretching of the spring, while the angular coordinate describes swinging motion.

Because these two motions are coupled, the system can exhibit nonlinear energy exchange, resonance, and chaos.

## B3.6.2 Building the Model

To analyze the spring pendulum, we must first choose a set of generalized coordinates that completely specify the position of the mass.

Because the mass can both stretch the spring and swing about the pivot, two coordinates are required. Our goal is to derive the Lagrangian

$$
L=T-U
$$

for this system.

## B3.6.3 Equations of Motion

Having constructed the Lagrangian, our next goal is to determine the equations governing the motion of the spring pendulum.

Because the coordinates

$$
r
\qquad\text{and}\qquad
\theta
$$

are coupled, we expect to obtain two nonlinear differential equations. These equations form the mathematical model that will be used throughout the remainder of this project.

\frac{d}{dt}

Unlike the circular orbit of a simple pendulum, the spring pendulum exhibits a much richer motion because both the radius and the angle vary with time.

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Spring pendulum simulation
#
# State vector:
# y = [r, r_dot, theta, theta_dot]
# ------------------------------------------------------------

# Physical parameters
m = 0.50          # kg
k = 20.0          # N/m
L0 = 0.30         # m
g = 9.81          # m/s^2

# Time settings
dt = 0.001
t_max = 20
t = np.arange(0, t_max, dt)

# Initial conditions
r0 = 0.40
r_dot0 = 0.0
theta0 = np.radians(20.0)
theta_dot0 = 0.0

def derivatives(state):
    r, r_dot, theta, theta_dot = state

    r_ddot = (
        r*theta_dot**2
        - (k/m)*(r - L0)
        + g*np.cos(theta)
    )

    theta_ddot = (
        -2*r_dot*theta_dot/r
        - (g/r)*np.sin(theta)
    )

    return np.array([r_dot, r_ddot, theta_dot, theta_ddot])

def rk4_step(state, dt):
    k1 = derivatives(state)
    k2 = derivatives(state + 0.5*dt*k1)
    k3 = derivatives(state + 0.5*dt*k2)
    k4 = derivatives(state + dt*k3)

    return state + (dt/6)*(k1 + 2*k2 + 2*k3 + k4)

# Arrays
r = np.zeros(len(t))
r_dot = np.zeros(len(t))
theta = np.zeros(len(t))
theta_dot = np.zeros(len(t))

# Initial state
state = np.array([r0, r_dot0, theta0, theta_dot0])

# Numerical integration
for i in range(len(t)):
    r[i] = state[0]
    r_dot[i] = state[1]
    theta[i] = state[2]
    theta_dot[i] = state[3]

    state = rk4_step(state, dt)

# ------------------------------------------------------------
# Plot r(t)
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(t, r)

plt.xlabel("Time (s)")
plt.ylabel(r"$r(t)$ (m)")
plt.title("Spring Pendulum: Radial Motion")
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# Plot theta(t)
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(t, np.degrees(theta))

plt.xlabel("Time (s)")
plt.ylabel(r"$\theta(t)$ (degrees)")
plt.title("Spring Pendulum: Angular Motion")
plt.grid(True)

plt.show()
```

```python
# ------------------------------------------------------------
# Motion in physical space
#
# Convert from polar-like coordinates to Cartesian coordinates:
#
# x = r sin(theta)
# y = -r cos(theta)
# ------------------------------------------------------------

x = r*np.sin(theta)
y = -r*np.cos(theta)

# ------------------------------------------------------------
# Plot x(t) and y(t)
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(t, x, label=r"$x(t)$")
plt.plot(t, y, label=r"$y(t)$")

plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Spring Pendulum: Cartesian Coordinates")
plt.legend(loc="upper right")
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# Plot trajectory in the x-y plane
# ------------------------------------------------------------

plt.figure(figsize=(7, 7))

plt.plot(x, y)

plt.xlabel(r"$x$ (m)")
plt.ylabel(r"$y$ (m)")
plt.title("Spring Pendulum Trajectory in Physical Space")
plt.axis("equal")
plt.grid(True)

plt.show()
```

## B3.6.5 Energy Exchange Between Radial and Angular Motion

The spring pendulum has two types of motion:

- radial motion, associated with stretching and compressing the spring,
- angular motion, associated with swinging back and forth.

Because these motions are coupled, energy can flow between them. A motion that begins primarily as spring oscillation may gradually excite pendulum motion, and a motion that begins primarily as pendulum motion may excite radial oscillation.

This exchange of energy is one of the most important signatures of nonlinear coupling.

```python
# ------------------------------------------------------------
# Energy exchange in the spring pendulum
#
# This code assumes the previous simulation cell has already created:
#
# t, r, r_dot, theta, theta_dot, m, k, L0, g
# ------------------------------------------------------------

# Radial kinetic energy
T_r = 0.5*m*r_dot**2

# Angular kinetic energy
T_theta = 0.5*m*r**2*theta_dot**2

# Spring potential energy
U_s = 0.5*k*(r - L0)**2

# Gravitational potential energy
U_g = -m*g*r*np.cos(theta)

# Total energy
E = T_r + T_theta + U_s + U_g

# ------------------------------------------------------------
# Plot energy components
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(t, T_r, label=r"$T_r$")
plt.plot(t, T_theta, label=r"$T_\theta$")
plt.plot(t, U_s, label=r"$U_s$")
plt.plot(t, U_g, label=r"$U_g$")
plt.plot(t, E, "k--", linewidth=2, label=r"$E_{\mathrm{total}}$")

plt.xlabel("Time (s)")
plt.ylabel("Energy (J)")
plt.title("Energy Exchange in the Spring Pendulum")
plt.legend(loc="upper right")
plt.grid(True)

plt.show()
```

## B3.6.6 Internal Resonance and Autoparametric Resonance

The spring pendulum becomes especially interesting when the radial and angular motions influence one another strongly.

For small oscillations, the system has two approximate frequencies:

$$
\omega_r
\approx
\sqrt{\frac{k}{m}},
$$

for radial spring motion, and

$$
\omega_\theta
\approx
\sqrt{\frac{g}{r_{\mathrm{eq}}}},
$$

for angular pendulum motion, where

$$
r_{\mathrm{eq}}
=
\ell_0+\frac{mg}{k}.
$$

A particularly strong energy exchange occurs when these frequencies satisfy approximately

$$
\omega_r
\approx
2\omega_\theta.
$$

This condition is known as an **internal resonance** because the resonance occurs between two modes of the same system.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Autoparametric Resonance</strong>

Autoparametric resonance occurs when one mode of a system parametrically excites another mode of the same system.

For the spring pendulum, strong energy exchange occurs when the radial oscillation frequency is approximately twice the angular oscillation frequency:

$$
\omega_r
\approx
2\omega_\theta.
$$

</div>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The resonance condition is

$$
\omega_r=2\omega_\theta.
$$

Substituting the approximate frequencies gives

$$
\sqrt{\frac{k}{m}}
=
2\sqrt{\frac{g}{r_{\mathrm{eq}}}}.
$$

Squaring both sides,

$$
\frac{k}{m}
=
\frac{4g}{r_{\mathrm{eq}}}.
$$

Using

$$
r_{\mathrm{eq}}
=
\ell_0+\frac{mg}{k},
$$

we obtain

$$
\frac{k}{m}
=
\frac{4g}{\ell_0+\frac{mg}{k}}.
$$

Multiplying both sides by

$$
\ell_0+\frac{mg}{k}
$$

gives

$$
\frac{k\ell_0}{m}
+
g
=
4g.
$$

Therefore,

$$
\frac{k\ell_0}{m}
=
3g,
$$

so

$$
\boxed{
k
=
\frac{3mg}{\ell_0}
}
$$

Using the given values,

$$
k
=
\frac{3(0.50)(9.81)}{0.30}
=
49.05~\mathrm{N/m}.
$$

Thus, a spring constant near

$$
k\approx49~\mathrm{N/m}
$$

should place the system close to the internal resonance condition.

Away from resonance, the radial and angular motions are still coupled, but the energy exchange is relatively weak. The spring length may oscillate, but the angular motion remains comparatively regular.

Near

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Spring pendulum: away from and near autoparametric resonance
#
# State vector:
# y = [r, r_dot, theta, theta_dot]
# ------------------------------------------------------------

# Physical parameters
m = 0.50
L0 = 0.30
g = 9.81

# Time settings
dt = 0.001
t_max = 40
t = np.arange(0, t_max, dt)

# Initial angular displacement
theta0 = np.radians(20.0)
theta_dot0 = 0.0
r_dot0 = 0.0

def derivatives(state, k):
    r, r_dot, theta, theta_dot = state

    r_ddot = (
        r*theta_dot**2
        - (k/m)*(r - L0)
        + g*np.cos(theta)
    )

    theta_ddot = (
        -2*r_dot*theta_dot/r
        - (g/r)*np.sin(theta)
    )

    return np.array([r_dot, r_ddot, theta_dot, theta_ddot])

def rk4_step(state, dt, k):
    k1 = derivatives(state, k)
    k2 = derivatives(state + 0.5*dt*k1, k)
    k3 = derivatives(state + 0.5*dt*k2, k)
    k4 = derivatives(state + dt*k3, k)

    return state + (dt/6)*(k1 + 2*k2 + 2*k3 + k4)

def simulate_spring_pendulum(k):
    r_eq = L0 + m*g/k

    state = np.array([
        r_eq,
        r_dot0,
        theta0,
        theta_dot0
    ])

    r = np.zeros(len(t))
    r_dot = np.zeros(len(t))
    theta = np.zeros(len(t))
    theta_dot = np.zeros(len(t))

    for i in range(len(t)):
        r[i] = state[0]
        r_dot[i] = state[1]
        theta[i] = state[2]
        theta_dot[i] = state[3]

        state = rk4_step(state, dt, k)

    x = r*np.sin(theta)
    y = -r*np.cos(theta)

    return r, r_dot, theta, theta_dot, x, y, r_eq

# ------------------------------------------------------------
# Compare two spring constants
# ------------------------------------------------------------

k_values = [
    (20.0, "Away from resonance"),
    (49.0, "Near autoparametric resonance")
]

results = {}

for k, label in k_values:
    results[label] = simulate_spring_pendulum(k)

# ------------------------------------------------------------
# Plot r(t)
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

for k, label in k_values:
    r, r_dot, theta, theta_dot, x, y, r_eq = results[label]
    plt.plot(t, r, label=rf"{label}, $k={k}$ N/m")

plt.xlabel("Time (s)")
plt.ylabel(r"$r(t)$ (m)")
plt.title("Radial Motion")
plt.legend(loc="upper right")
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# Plot theta(t)
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

for k, label in k_values:
    r, r_dot, theta, theta_dot, x, y, r_eq = results[label]
    plt.plot(t, np.degrees(theta), label=rf"{label}, $k={k}$ N/m")

plt.xlabel("Time (s)")
plt.ylabel(r"$\theta(t)$ (degrees)")
plt.title("Angular Motion")
plt.legend(loc="upper right")
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# Plot trajectories in physical space
# ------------------------------------------------------------

for k, label in k_values:
    r, r_dot, theta, theta_dot, x, y, r_eq = results[label]

    plt.figure(figsize=(7, 7))

    plt.plot(x, y)

    plt.xlabel(r"$x$ (m)")
    plt.ylabel(r"$y$ (m)")
    plt.title(rf"$(x,y)$ Trajectory: {label}, $k={k}$ N/m")
    plt.axis("equal")
    plt.grid(True)

    plt.show()
```

## B3.6.10 Poincaré Sections of the Spring Pendulum

As the motion of the spring pendulum becomes increasingly complicated, ordinary trajectories and phase portraits may become difficult to interpret.

A powerful way to simplify the dynamics is to construct a **Poincaré section**. Rather than displaying the entire trajectory, we record the state of the system whenever a specified condition is satisfied.

For the spring pendulum, one convenient choice is to record the radial state

$$
(r,\dot r)
$$

whenever

$$
\theta=0
$$

and

$$
\dot\theta>0.
$$

Thus, each point corresponds to a crossing of the downward vertical with the mass moving in the same direction.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Poincaré Section of the Spring Pendulum</strong>

A Poincaré section for the spring pendulum may be constructed by recording

$$
(r,\dot r)
$$

whenever

$$
\theta=0
$$

and

$$
\dot\theta>0.
$$

The resulting pattern provides a compact description of the long-term dynamics.

</div>

Because the spring pendulum is not externally driven, there is no natural driving period. Instead, the section is defined geometrically by choosing a surface in phase space.

<div style="background-color:#e8f5e9; border-left:6px solid #006633; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Box Activity — Constructing a Poincaré Section</h3>

For the parameter values used here, the Poincaré section typically consists of a relatively simple collection of points or smooth curves rather than a complicated cloud.

This indicates that the motion remains largely regular, even though the system is highly nonlinear and exhibits strong energy exchange between radial and angular motion.

A truly chaotic motion would produce a much more complicated structure containing a broad cloud of points rather than smooth curves.

This illustrates an important lesson of nonlinear dynamics:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Nonlinear Does Not Necessarily Mean Chaotic</strong>

Nonlinearity can produce complicated trajectories, internal resonance, and strong coupling without leading to chaos.

Chaos depends not only on the form of the equations but also on the parameters, energy, and initial conditions.

Thus, the Poincaré section provides a powerful tool for distinguishing regular motion from chaotic motion.

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Poincaré section of the spring pendulum
#
# Record (r, r_dot) whenever:
#
#     theta crosses 0 upward
#     theta_dot > 0
# ------------------------------------------------------------

# Physical parameters
m = 0.50
L0 = 0.30
g = 9.81
k = 49.0

# Equilibrium length
r_eq = L0 + m*g/k

# Time settings
dt = 0.001
t_max = 200
t = np.arange(0, t_max, dt)

# Initial conditions
r0 = r_eq
r_dot0 = 0.0
theta0 = np.radians(20.0)
theta_dot0 = 0.0

def derivatives(state):
    r, r_dot, theta, theta_dot = state

    r_ddot = (
        r*theta_dot**2
        - (k/m)*(r - L0)
        + g*np.cos(theta)
    )

    theta_ddot = (
        -2*r_dot*theta_dot/r
        - (g/r)*np.sin(theta)
    )

    return np.array([r_dot, r_ddot, theta_dot, theta_ddot])

def rk4_step(state, dt):
    k1 = derivatives(state)
    k2 = derivatives(state + 0.5*dt*k1)
    k3 = derivatives(state + 0.5*dt*k2)
    k4 = derivatives(state + dt*k3)

    return state + (dt/6)*(k1 + 2*k2 + 2*k3 + k4)

# ------------------------------------------------------------
# Numerical integration
# ------------------------------------------------------------

r = np.zeros(len(t))
r_dot = np.zeros(len(t))
theta = np.zeros(len(t))
theta_dot = np.zeros(len(t))

state = np.array([r0, r_dot0, theta0, theta_dot0])

for i in range(len(t)):
    r[i] = state[0]
    r_dot[i] = state[1]
    theta[i] = state[2]
    theta_dot[i] = state[3]

    state = rk4_step(state, dt)

# ------------------------------------------------------------
# Construct Poincaré section
# ------------------------------------------------------------

r_section = []
r_dot_section = []

for i in range(len(t)-1):

    # Upward crossing of theta = 0
    if theta[i] < 0 and theta[i+1] >= 0 and theta_dot[i] > 0:

        # Linear interpolation for a cleaner crossing estimate
        frac = -theta[i] / (theta[i+1] - theta[i])

        r_cross = r[i] + frac*(r[i+1] - r[i])
        r_dot_cross = r_dot[i] + frac*(r_dot[i+1] - r_dot[i])

        r_section.append(r_cross)
        r_dot_section.append(r_dot_cross)

r_section = np.array(r_section)
r_dot_section = np.array(r_dot_section)

# ------------------------------------------------------------
# Plot radial phase portrait
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.plot(r, r_dot, linewidth=0.7)

plt.xlabel(r"$r$ (m)")
plt.ylabel(r"$\dot r$ (m/s)")
plt.title("Spring Pendulum: Radial Phase Portrait")
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# Plot Poincaré section
# ------------------------------------------------------------

plt.figure(figsize=(7, 7))

plt.plot(r_section, r_dot_section, ".", markersize=4)

plt.xlabel(r"$r$ (m)")
plt.ylabel(r"$\dot r$ (m/s)")
plt.title(r"Poincaré Section: $(r,\dot r)$ at $\theta=0$, $\dot\theta>0$")
plt.grid(True)

plt.show()
```

The study of classical mechanics often begins with idealized systems and simple equations. Yet, as we have seen, even Newton's laws can lead to an astonishing richness of behavior. From simple harmonic motion to synchronization and chaos, classical mechanics provides a gateway to some of the most beautiful and profound ideas in all of physics.
