## B3.1.1 Motivation

Up to this point, we have studied oscillatory systems whose equations of motion are linear. Examples include the simple harmonic oscillator, coupled oscillators, normal modes, and the wave equation.

Linear systems possess several remarkable properties:

- superposition holds,
- frequencies are independent of amplitude,
- solutions are sinusoidal,
- and the future motion is relatively simple to predict.

However, most physical systems are only approximately linear. As amplitudes become larger or interactions become stronger, nonlinear effects begin to appear. These effects can produce entirely new phenomena:

- amplitude-dependent frequencies,
- multiple equilibrium positions,
- self-sustained oscillations,
- synchronization,
- period doubling,
- and ultimately chaos.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Curiosity</strong>

Although linear oscillators occupy much of introductory physics, truly linear systems are rare in nature. Nonlinear behavior appears in systems as diverse as heart cells, lasers, electrical circuits, planetary motion, and weather systems.

</div>

Thus, nonlinear oscillations represent not merely a correction to simple harmonic motion but an entirely new world of physical behavior.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.1.2 Linear Versus Nonlinear Oscillators

In a linear oscillator, the restoring force is proportional to the displacement and Newton's second law becomes

$$
m\ddot{x}+kx=0.
$$

Its solution is

$$
x(t)=A\cos(\omega_0t+\phi),
$$

where

$$
\omega_0=\sqrt{\frac{k}{m}}
$$

is independent of amplitude.

Because the equation is linear, superposition holds:

$$
x_1(t)+x_2(t)
$$

is also a solution whenever $x_1(t)$ and $x_2(t)$ are solutions.

---

In contrast, nonlinear oscillators contain terms that are not proportional to displacement or velocity.

Examples include

$$
F=-kx-\beta x^3,
$$

or

$$
F=-mg\sin\theta.
$$

These lead to nonlinear equations such as

$$
m\ddot{x}+kx+\beta x^3=0,
$$

and

$$
\ddot{\theta}+\frac{g}{L}\sin\theta=0.
$$

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Linear Oscillator</strong>

A linear oscillator is one whose equation of motion is linear in the dependent variable and its derivatives. Such systems obey the principle of superposition.

</div>

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Nonlinear Oscillator</strong>

A nonlinear oscillator is one whose equation of motion contains nonlinear terms. In such systems, superposition generally fails and entirely new behaviors may emerge.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.1.3 Characteristics of Nonlinear Systems

Nonlinear oscillators exhibit several behaviors that are absent from simple harmonic motion.

Some of the most important characteristics of nonlinear systems include:

- **Amplitude-dependent frequencies**
  
  The oscillation frequency depend on the amplitude. For example, the period of a pendulum increases for sufficiently large angles.

- **Breakdown of superposition**
  
  In general,

  $$
  x_1(t)+x_2(t)
  $$

  is not a solution, even if both

  $$
  x_1(t)
  \quad\text{and}\quad
  x_2(t)
  $$

  are solutions individually.

- **Multiple equilibria**
  
  Nonlinear systems may possess several equilibrium positions, some stable and some unstable.

- **Self-sustained oscillations**
  
  Some systems are capable of maintaining oscillations without periodic forcing.

  Examples include:

  - heart cells,
  - neurons,
  - lasers,
  - and electronic oscillators.

- **Chaotic motion**
  
  Under suitable conditions, nonlinear systems may become chaotic, displaying extreme sensitivity to initial conditions.

<div style="background-color:#ffe6e6; border-left:6px solid #cc0000; padding:14px; border-radius:4px;">
<strong>Danger:</strong><br>

Chaos should not be confused with randomness.

Chaotic systems are governed by completely deterministic equations. Their apparent unpredictability arises because tiny differences in initial conditions can grow exponentially with time.

</div>

## B3.1.4 The Pendulum Revisited

One of the simplest examples of a nonlinear oscillator is the pendulum.

Its exact equation of motion is

$$
\ddot{\theta}
+
\frac{g}{L}\sin\theta
=
0.
$$

For sufficiently small angles,

$$
\sin\theta\approx\theta,
$$

giving

$$
\ddot{\theta}
+
\frac{g}{L}\theta
=
0,
$$

which is the familiar simple harmonic oscillator.

However, for larger amplitudes the approximation

$$
\sin\theta\approx\theta
$$

becomes inaccurate, and the oscillation period depends on amplitude.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Large-Angle Pendulum</h3>

A pendulum released from a small angle oscillates with approximately constant period independent of amplitude.

As the amplitude increases, the exact restoring torque

$$
\tau=-mgL\sin\theta
$$

deviates from the linear approximation

$$
\tau\approx -mgL\theta.
$$

Consequently, the oscillation period becomes longer and the motion is no longer perfectly sinusoidal.

Thus, the simple harmonic oscillator is merely the small-amplitude approximation of a fundamentally nonlinear system.

</div>

For small initial angles, the approximation

is very accurate. The nonlinear and linear models should remain nearly identical.

For larger initial angles, the approximation becomes less accurate. The nonlinear pendulum has a longer period than the linear prediction. This means the nonlinear solution gradually lags behind the simple harmonic oscillator solution.

The main observation is:

$$
\boxed{
\text{The pendulum period increases with amplitude.}
}
$$

This amplitude-dependent period is one of the simplest signatures of nonlinear oscillatory motion.

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Large-Angle Pendulum: Linear vs. Nonlinear Motion
# ---------------------------------------------------------

# Constants
g = 9.81          # m/s^2
L = 1.0           # m
dt = 0.01         # s
t_max = 20        # s

# Time array
t = np.arange(0, t_max, dt)

# Initial angle (change to 5, 60, 120 degrees)
theta0_deg = 60
theta0 = np.radians(theta0_deg)

# Initial angular velocity
omega0 = 0.0

# Arrays for nonlinear pendulum
theta_nl = np.zeros(len(t))
omega_nl = np.zeros(len(t))

# Arrays for linear pendulum
theta_lin = np.zeros(len(t))
omega_lin = np.zeros(len(t))

# Initial conditions
theta_nl[0] = theta0
theta_lin[0] = theta0

omega_nl[0] = omega0
omega_lin[0] = omega0

# Euler-Cromer method
for i in range(len(t)-1):

    # Nonlinear pendulum
    alpha_nl = -(g/L)*np.sin(theta_nl[i])
    omega_nl[i+1] = omega_nl[i] + alpha_nl*dt
    theta_nl[i+1] = theta_nl[i] + omega_nl[i+1]*dt

    # Linear approximation
    alpha_lin = -(g/L)*theta_lin[i]
    omega_lin[i+1] = omega_lin[i] + alpha_lin*dt
    theta_lin[i+1] = theta_lin[i] + omega_lin[i+1]*dt

# ---------------------------------------------------------
# Plot
# ---------------------------------------------------------

plt.figure(figsize=(9,5))

plt.plot(t,
         np.degrees(theta_nl),
         label="Nonlinear pendulum")

plt.plot(t,
         np.degrees(theta_lin),
         "--",
         label="Small-angle approximation")

plt.xlabel("Time (s)")
plt.ylabel(r"$\theta$ (degrees)")
plt.title(f"Large-Angle Pendulum ($\\theta_0={theta0_deg}^\\circ$)")
plt.legend(loc="upper right")
plt.grid()

plt.show()
```
