# B4.2 Discrete Symmetries in Physics

## B4.2.1 What Are Discrete Symmetries?

- **Definition:**  
  A **discrete symmetry** is a transformation that leaves the physical laws (or action/Lagrangian) invariant, but the transformation can only take values from a **finite or countable set**.

- **Contrast with continuous symmetries:**  
  Continuous → parameter changes smoothly (translations, rotations).  
  Discrete → transformations are "all or nothing" (mirror flips, reversal of time).

- **Mathematical groups:**  
  - Typical discrete symmetry groups:  
    - **Parity (P):** $Z_2$ (two elements: $+1$ and $-1$).  
    - **Time reversal (T):** $Z_2$.  
    - **Charge conjugation (C):** $Z_2$.  
  - Some systems have larger discrete groups (e.g., crystal lattices).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.2.2 Why Are Discrete Symmetries Important?

1. **Simplify physical problems** (mirror symmetry reduces calculation domain).  
2. **Classify states** (even/odd wavefunctions).  
3. **Predict selection rules** (forbidden transitions in atomic spectra).  
4. **Reveal new physics when broken** (e.g., CP violation in weak interactions → Nobel prizes).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.2.3 Historical Context

- **Parity (P):**  
  - Assumed universal until **1956**, when **Chien-Shiung Wu** showed **beta decay** violates parity.  

- **Time Reversal (T):**  
  - Classical mechanics largely T-invariant; quantum mechanics treats T as an **anti-unitary operator** (but you don’t need that detail yet).

- **Charge Conjugation (C) & CPT theorem:**  
  - Discrete symmetries became central in quantum field theory; **CPT** (combined) is fundamental.

## B4.2.4 Examples in Physics I & II (Where You Already Saw It!)

Even without calling it "discrete symmetry," you have used it:

- **Mechanics:**  
  - A mass-spring system: even/odd oscillations under $x \to -x$.  
  - Projectile motion in a uniform field: symmetric about the vertical axis.

- **Electromagnetism:**  
  - Electrostatics: mirror symmetry used in **method of images**.  
  - Magnetic fields around straight wires: mirror symmetry across a plane.

## B4.2.5 The Three Main Discrete Symmetries

### Parity (P)

#### Definition:
- Parity (spatial inversion):
   
  $$
  \mathbf{r} \longrightarrow -\,\mathbf{r}, \quad
  t \longrightarrow t
  $$
  
- In 1D: $x \to -x$.  
- In 3D: $(x, y, z) \to (-x, -y, -z)$.

#### Physical Interpretation:
- Reflects the system through the origin (mirror inversion in all axes).

#### Invariance of the Lagrangian:
For a Lagrangian $L(\mathbf{r}, \dot{\mathbf{r}})$:
- If $L(-\mathbf{r}, -\dot{\mathbf{r}}) = L(\mathbf{r}, \dot{\mathbf{r}})$ → **P-symmetric**.

#### Example – Simple Harmonic Oscillator:

$$
L = \frac12 m \dot{x}^2 - \frac12 k x^2
$$

- Apply $x \to -x$, $\dot{x} \to -\dot{x}$:
  
$$
L' = \frac12 m (-\dot{x})^2 - \frac12 k (-x)^2 = L
$$

→ **Parity conserved**.

---

### Time Reversal (T)

#### Definition:
- Time reversal:
  
  $$
  t \longrightarrow -t, \quad
  \mathbf{r} \longrightarrow \mathbf{r}, \quad
  \dot{\mathbf{r}} \longrightarrow -\dot{\mathbf{r}}
  $$

#### Physical Interpretation:
- Reverse the direction of motion (velocities change sign, positions unchanged).

<header>
<h3>Example – Free Particle</h3>
    </header>

$$
L = \frac12 m \dot{x}^2
$$

- Under T: $\dot{x}\to -\dot{x}$ → $L' = \frac12 m(-\dot{x})^2 = L$.  
→ **T-symmetric**.

**Breaking T-Symmetry**:
- Add damping: $L \sim -b x \dot{x}$ changes sign → **not T-invariant**.

</div>

---

### Charge Conjugation (C)

- Swaps particles ↔ antiparticles (e.g., electron ↔ positron).  
- Classical analogy: $q \to -q$ in electromagnetism.  
- Maxwell’s equations in vacuum are C-symmetric.

---

### Combined Symmetries & CPT

- **CPT theorem (QFT):** All local Lorentz-invariant theories are invariant under the combined transformation of $CPT$.

## B4.2.6 Worked Example

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;"> 
<header>
<h3>Example – Check P and T Symmetry</h3>
    </header>
    
**Problem:**  
Check whether the Lagrangian

$$
L = \frac12 m \dot{x}^2 - \frac12 k x^2 + \lambda x^3
$$

is invariant under **parity (P)** and **time reversal (T)**.

---

**Solution:**

1. **Parity: $x \to -x$, $\dot{x} \to -\dot{x}$**

Substitute:

$$
L' = \frac12 m (-\dot{x})^2 - \frac12 k (-x)^2 + \lambda (-x)^3 \\
= \frac12 m \dot{x}^2 - \frac12 k x^2 - \lambda x^3
$$

- $L' = L$ **only if** $\lambda = 0$.  
- → **Parity broken** when $\lambda \neq 0$.

---

2. **Time Reversal: $t \to -t$, $x \to x$, $\dot{x} \to -\dot{x}$**

Substitute:

$$
L' = \frac12 m (-\dot{x})^2 - \frac12 k x^2 + \lambda x^3 = L
$$

- → **T-symmetric**, even with $\lambda \neq 0$.

---

**Interpretation:**
- A cubic potential breaks parity but keeps time-reversal invariance.
  
</div>
