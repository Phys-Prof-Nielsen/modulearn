# B4.4 Noether’s Theorem – Symmetries and Conservation Laws
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.4.1 Motivation

- In Lagrangian mechanics, the dynamics are derived from a **single scalar function** $L(q, \dot{q}, t)$.
- **Emmy Noether (1918)** proved a deep connection:

$$
\textbf{Every continuous symmetry of the action corresponds to a conserved quantity.}
$$

This unifies and generalizes what we already know:

| **Symmetry**                   | **Conserved Quantity**         |
|----------------------------------|--------------------------------|
| Time translation $t \to t+\epsilon$ | Energy $E$                     |
| Spatial translation $q \to q+\epsilon$ | Linear momentum $\mathbf{p}$  |
| Rotational symmetry              | Angular momentum $\mathbf{L}$  |

Noether’s theorem is central to modern physics — from classical mechanics to quantum field theory.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.4.2 Preliminaries

We start with the **action**:

$$
S[q(t)] = \int_{t_1}^{t_2} L(q, \dot{q}, t)\, dt
$$

The actual path satisfies the **Euler–Lagrange equations**:

$$
\frac{d}{dt}\left( \frac{\partial L}{\partial \dot{q}_i} \right) - \frac{\partial L}{\partial q_i} = 0
$$

---

### Definition: Continuous Symmetry

A **continuous symmetry** of a Lagrangian system is a smooth transformation of the form:

$$
\begin{aligned}
q_i &\;\to\; q_i' \;=\; q_i \;+\; \epsilon\, \Delta q_i(q, \dot{q}, t) \\
t   &\;\to\; t'   \;=\; t   \;+\; \epsilon\, \Delta t(q, \dot{q}, t)
\end{aligned}
$$

where:

- $\epsilon$ is an **infinitesimal parameter** (the “size” of the transformation).  
- $\Delta q_i$ and $\Delta t$ are smooth functions describing how coordinates and possibly time change under the symmetry.

The transformation is a **symmetry** if the Lagrangian changes at most by a total time derivative:

$$
L(q', \dot{q}', t') \;=\; L(q, \dot{q}, t) \;+\; \epsilon \frac{dF}{dt}
$$

Here, $F(q, t)$ is some function whose total derivative does not affect the equations of motion (since total derivatives add a constant to the action after integration).

---

### Key Points

1. **Continuous vs. Discrete:**
   - **Continuous symmetry**: can be parametrized by a small continuous parameter $\epsilon$ (e.g., translation by a tiny amount, rotation by a small angle).
   - **Discrete symmetry**: finite jumps (e.g., parity $x \to -x$, time reversal $t \to -t$).  
     → Noether’s theorem applies **only to continuous** symmetries.

2. **Active vs. Passive View:**
   - We can think of actually moving the system (“active”), or simply relabeling coordinates (“passive”).  
     The mathematics is the same — it’s a transformation of the description, not the physical reality.

3. **Why Total Derivative Matters:**
   - Adding $\frac{dF}{dt}$ to $L$ changes the action by a boundary term:
   
     $$
     S = \int L\, dt \;\to\; \int \left(L + \frac{dF}{dt}\right) dt = S + F\Big|_{t_1}^{t_2}
     $$
     
     which does **not** affect the Euler–Lagrange equations.

---

### Geometric Interpretation

- A continuous symmetry corresponds to a **direction in configuration space** along which the action does not change (or changes only by a harmless boundary term).
- Noether’s theorem tells us there is a corresponding conserved “momentum” **conjugate to that direction**.

---

### Examples of Continuous Symmetries

1. **Spatial Translation (Free Particle):**  
   $q \to q + \epsilon$  
   ($\Delta q = 1$, $\Delta t = 0$)

2. **Time Translation (Any time-independent $L$):**  
   $t \to t + \epsilon$  
   ($\Delta q = 0$, $\Delta t = 1$)

3. **Rotation in 2D:**  
   $\phi \to \phi + \epsilon$  
   ($\Delta \phi = 1$, $\Delta t = 0$)

4. **Gauge-like Transformations:**  
   Adding a total derivative to $L$ corresponds to a “gauge” symmetry in mechanics (precursor to gauge theories in field physics).

---

### Infinitesimal vs. Finite Transformations

- In practice, we consider only **infinitesimal transformations** ($\epsilon$ very small).  
- Finite transformations (like $x \to x + a$) can be built by applying infinitesimal ones repeatedly.

## B4.4.3 Noether’s Theorem – Statement

### Theorem (Simplest Form)

If the Lagrangian is invariant (up to a total time derivative) under a continuous transformation, then:

$$
\boxed{
\frac{d}{dt} \left[ \sum_i \frac{\partial L}{\partial \dot{q}_i}\, \Delta q_i \;-\; \left( L\,\Delta t - F \right) \right] = 0
}
$$

Thus, the quantity in brackets is **conserved** (constant in time).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.4.4 Noether’s Theorem – Detailed Derivation

We start from the **action**:

$$
S[q(t)] = \int_{t_1}^{t_2} L\big(q, \dot{q}, t\big)\, dt
$$

and assume that the system satisfies the Euler–Lagrange equations.

---

### Step 1

We have the infinitesimal transformation:

$$
\begin{aligned}
t &\to t' = t + \epsilon\,\Delta t(q,\dot{q},t), \\
q_i &\to q_i' = q_i + \epsilon\,\Delta q_i(q,\dot{q},t).
\end{aligned}
$$

The new velocity is:

$$
\dot{q}_i' \equiv \frac{dq_i'}{dt'} = \frac{dq_i + \epsilon\, d(\Delta q_i)}{dt + \epsilon\, d(\Delta t)}.
$$

---

#### Step 1.1 – Express Differentials

- Since $dq_i = \dot{q}_i\, dt$,  
  and $\;d(\Delta q_i) = \frac{d(\Delta q_i)}{dt}\, dt$,

we can write the numerator as:

$$
dq_i' = (\dot{q}_i\, dt) + \epsilon\,\frac{d(\Delta q_i)}{dt}\, dt
= \left(\dot{q}_i + \epsilon\,\frac{d}{dt}(\Delta q_i)\right)\, dt
$$

Similarly for the denominator:

$$
dt' = dt + \epsilon\, d(\Delta t) = \left(1 + \epsilon\,\frac{d}{dt}(\Delta t)\right)\, dt
$$

---

#### Step 1.2 – Divide

Now substitute into:

$$
\dot{q}_i' = \frac{dq_i'}{dt'}
= \frac{\left(\dot{q}_i + \epsilon\,\frac{d}{dt}(\Delta q_i)\right)\, dt}
{\left(1 + \epsilon\,\frac{d}{dt}(\Delta t)\right)\, dt}
$$

Cancel $dt$:

$$
\dot{q}_i' = \frac{\dot{q}_i + \epsilon\,\frac{d}{dt}(\Delta q_i)}
{1 + \epsilon\,\frac{d}{dt}(\Delta t)}
$$

Expand the denominator using $(1+x)^{-1} \approx 1-x$:

$$
\dot{q}_i' \approx \left(\dot{q}_i + \epsilon\,\frac{d}{dt}(\Delta q_i)\right)
\left(1 - \epsilon\,\frac{d}{dt}(\Delta t)\right)
$$

---

#### Step 1.3 – Expand to First Order

Multiplying and ignoring $\mathcal{O}(\epsilon^2)$:

$$
\dot{q}_i' \approx \dot{q}_i
+ \epsilon\,\frac{d}{dt}(\Delta q_i)
- \epsilon\,\dot{q}_i\,\frac{d}{dt}(\Delta t)
$$

Thus:

$$
\boxed{
\delta \dot{q}_i = \dot{q}_i' - \dot{q}_i
= \epsilon\left[\frac{d}{dt}(\Delta q_i) - \dot{q}_i\,\frac{d}{dt}(\Delta t)\right]
}
$$

---

#### Key Point

The step begins with **differentials** $dq_i'$ and $dt'$, not derivatives.  
Only after writing everything in terms of differentials do we divide to obtain derivatives.

This is the cleanest and most rigorous way to justify the formula for $\dot{q}_i'$.

---

### Step 2: Expanding $L$ to First Order in $\epsilon$

We impose the symmetry condition:

$$
L(q', \dot{q}', t') = L(q, \dot{q}, t) + \epsilon\,\frac{dF}{dt}
$$

where $F$ is some function of $(q,t)$ whose total derivative does not affect the equations of motion.

---

#### Step 2.1 – Taylor Expand $L$

Treat $L(q,\dot{q},t)$ as a function of three variables $(q_i, \dot{q}_i, t)$ and expand to first order in $\epsilon$:

$$
\begin{aligned}
L(q', \dot{q}', t') &= 
L(q, \dot{q}, t)
+ \frac{\partial L}{\partial q_i}(q_i' - q_i)
+ \frac{\partial L}{\partial \dot{q}_i}(\dot{q}_i' - \dot{q}_i)
+ \frac{\partial L}{\partial t}(t' - t)
+ \mathcal{O}(\epsilon^2)
\end{aligned}
$$

Substitute the infinitesimal transformations:

- $q_i' - q_i = \epsilon\,\Delta q_i$
- $\dot{q}_i' - \dot{q}_i = \delta \dot{q}_i$ (already expanded in Step 1)
- $t' - t = \epsilon\,\Delta t$

Thus:

$$
L(q', \dot{q}', t') = L(q, \dot{q}, t)
+ \epsilon\left[
\frac{\partial L}{\partial q_i}\Delta q_i
+ \frac{\partial L}{\partial \dot{q}_i}\,\delta \dot{q}_i
+ \frac{\partial L}{\partial t}\Delta t
\right]
+ \mathcal{O}(\epsilon^2)
$$

The **first-order variation of the Lagrangian** is then:

$$
\boxed{
\delta L \equiv L(q', \dot{q}', t') - L(q, \dot{q}, t)
= \epsilon\left[
\frac{\partial L}{\partial q_i}\Delta q_i
+ \frac{\partial L}{\partial \dot{q}_i}\,\delta \dot{q}_i
+ \frac{\partial L}{\partial t}\Delta t
\right]
}
$$

and by the symmetry condition:

$$
\delta L = \epsilon\,\frac{dF}{dt}
$$

---

#### Step 2.2 – Substitute $\delta \dot{q}_i$

From **Step 1**, recall:

$$
\delta \dot{q}_i = \epsilon\left[
\frac{d}{dt}(\Delta q_i) - \dot{q}_i\,\frac{d}{dt}(\Delta t)
\right]
$$

Substitute (dropping the explicit $\epsilon$ factor for now, since we work only to first order):

$$
\delta L =
\frac{\partial L}{\partial q_i}\Delta q_i
+ \frac{\partial L}{\partial \dot{q}_i}
\left(\frac{d}{dt}(\Delta q_i) - \dot{q}_i\,\frac{d}{dt}(\Delta t)\right)
+ \frac{\partial L}{\partial t}\Delta t
$$

This is the fully expanded first-order variation of $L$.

---

#### **Key Point**

- We only keep **terms linear in $\epsilon$** because Noether’s theorem concerns **infinitesimal** transformations.  
- Higher-order terms $\mathcal{O}(\epsilon^2)$ vanish in the $\epsilon \to 0$ limit.

---

### Step 3: Rearranging Terms

Substitute the Euler–Lagrange equations:

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\right) - \frac{\partial L}{\partial q_i} = 0
\quad\Rightarrow\quad
\frac{\partial L}{\partial q_i} = \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\right)
$$

Replace in the $\delta L$ expression:

$$
\begin{aligned}
\delta L &= \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\right)\Delta q_i
+ \frac{\partial L}{\partial \dot{q}_i}\left(\frac{d}{dt}(\Delta q_i) - \dot{q}_i\,\frac{d}{dt}(\Delta t)\right)
+ \frac{\partial L}{\partial t}\,\Delta t
\end{aligned}
$$

---

### Step 4: Combine into Total Derivatives

Expand $\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\Delta q_i\right)$:

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\Delta q_i\right)
=
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\right)\Delta q_i
+ \frac{\partial L}{\partial \dot{q}_i}\frac{d}{dt}(\Delta q_i)
$$

Substitute back:

$$
\delta L =
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\Delta q_i\right)
- \frac{\partial L}{\partial \dot{q}_i}\,\dot{q}_i\,\frac{d}{dt}(\Delta t)
+ \frac{\partial L}{\partial t}\,\Delta t
$$

But $L$ itself has explicit time dependence:

$$
\frac{d}{dt}(L\,\Delta t) = \frac{dL}{dt}\,\Delta t + L\,\frac{d}{dt}(\Delta t)
= \left(\frac{\partial L}{\partial q_i}\dot{q}_i + \frac{\partial L}{\partial \dot{q}_i}\ddot{q}_i + \frac{\partial L}{\partial t}\right)\Delta t + L\,\frac{d}{dt}(\Delta t)
$$

However, along the equations of motion we can rewrite $ \frac{\partial L}{\partial q_i}\dot{q}_i + \frac{\partial L}{\partial \dot{q}_i}\ddot{q}_i = \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\right)$.

So:

$$
\frac{d}{dt}(L\,\Delta t) = \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\right)\Delta t + \frac{\partial L}{\partial t}\Delta t + L\frac{d}{dt}(\Delta t)
$$

We need to subtract this carefully.

---

### **Step 5: Subtracting the Total Time Derivative**

At this stage, we have the first-order variation of $L$:

$$
\delta L =
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\Delta q_i\right)
\;-\;
\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\,\frac{d}{dt}(\Delta t)
\;+\;
\frac{\partial L}{\partial t}\Delta t
$$

and the symmetry condition:

$$
\delta L = \frac{dF}{dt}.
$$

We need to rearrange this into a single total time derivative.

---

#### **Step 5.1 – Why Add and Subtract $d(L\,\Delta t)/dt$?**

The term we *want* in the final conserved quantity is:

$$
-\left(L\,\Delta t - F\right)
$$

Thus, we need to rewrite the middle terms in a way that produces $-\,d(L\,\Delta t)/dt$.

We can always add and subtract the same total derivative:

$$
0 \;=\; +\frac{d}{dt}(L\,\Delta t) \;-\;\frac{d}{dt}(L\,\Delta t)
$$

and insert it into the equation.

The trick is to combine:

$$
-\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\,\frac{d}{dt}(\Delta t)
+ \frac{\partial L}{\partial t}\Delta t
\;-\;\frac{d}{dt}(L\,\Delta t)
$$

into something that vanishes when using the chain rule for $dL/dt$.

---

#### **Step 5.2 – Expand $d(L\,\Delta t)/dt$**

Use the product rule:

$$
\frac{d}{dt}(L\,\Delta t)
= \frac{dL}{dt}\,\Delta t + L\,\frac{d}{dt}(\Delta t)
$$

and also expand $dL/dt$:

$$
\frac{dL}{dt} =
\frac{\partial L}{\partial q_i}\dot{q}_i
+ \frac{\partial L}{\partial \dot{q}_i}\ddot{q}_i
+ \frac{\partial L}{\partial t}
$$

Substitute into $d(L\,\Delta t)/dt$:

$$
\frac{d}{dt}(L\,\Delta t)
= \left(\frac{\partial L}{\partial q_i}\dot{q}_i + \frac{\partial L}{\partial \dot{q}_i}\ddot{q}_i + \frac{\partial L}{\partial t}\right)\Delta t
+ L\,\frac{d}{dt}(\Delta t)
$$

---

#### **Step 5.3 – Replace $\partial L/\partial q_i$**

Using the Euler–Lagrange equations:

$$
\frac{\partial L}{\partial q_i} = \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\right)
$$

so:

$$
\frac{\partial L}{\partial q_i}\dot{q}_i
= \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\right)
- \frac{\partial L}{\partial \dot{q}_i}\ddot{q}_i
$$

Substitute back into $d(L\,\Delta t)/dt$:

$$
\frac{d}{dt}(L\,\Delta t)
= \left[\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\right)
- \frac{\partial L}{\partial \dot{q}_i}\ddot{q}_i
+ \frac{\partial L}{\partial \dot{q}_i}\ddot{q}_i
+ \frac{\partial L}{\partial t}\right]\Delta t
+ L\,\frac{d}{dt}(\Delta t)
$$

The $\ddot{q}_i$ terms cancel:

$$
\frac{d}{dt}(L\,\Delta t)
= \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\right)\Delta t
+ \frac{\partial L}{\partial t}\Delta t
+ L\,\frac{d}{dt}(\Delta t)
$$

---

#### **Step 5.4 – Explicitly Performing the Add-and-Subtract**

We have the expanded first-order variation of $L$:

$$
\delta L =
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\Delta q_i\right)
-\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\,\frac{d}{dt}(\Delta t)
+\frac{\partial L}{\partial t}\Delta t
$$

and the symmetry condition:

$$
\delta L = \frac{dF}{dt}.
$$

Our goal is to rearrange this into:

$$
\frac{d}{dt}\left[\frac{\partial L}{\partial \dot{q}_i}\Delta q_i \;-\; (L\,\Delta t - F)\right] = 0
$$

---

##### **1) Insert the "zero" cleverly**

Add and subtract $\dfrac{d}{dt}(L\,\Delta t)$:

$$
\delta L =
\underbrace{\left[\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\Delta q_i\right)
-\frac{d}{dt}(L\,\Delta t)\right]}_{\text{we want this as the final conserved total derivative}}
\;+\;
\left[-\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\,\frac{d}{dt}(\Delta t)
+\frac{\partial L}{\partial t}\Delta t
+\frac{d}{dt}(L\,\Delta t)\right]
$$

---

##### **2) Show that the remaining bracket is zero**

Expand the added $d(L\,\Delta t)/dt$ (from Step 5.3):

$$
\frac{d}{dt}(L\,\Delta t)
= \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\right)\Delta t
+ \frac{\partial L}{\partial t}\Delta t
+ L\,\frac{d}{dt}(\Delta t)
$$

Focus on the terms proportional to $d(\Delta t)/dt$:

- From $L\,d(\Delta t)/dt$ we write:

$$
L = \frac{\partial L}{\partial \dot{q}_i}\dot{q}_i - \left(\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i - L\right)
$$

so:

$$
L\,\frac{d}{dt}(\Delta t)
= \frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\,\frac{d}{dt}(\Delta t)
-\left(\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i - L\right)\frac{d}{dt}(\Delta t)
$$

The first term exactly **cancels** $-\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\,\frac{d}{dt}(\Delta t)$ in $\delta L$.

The remaining pieces combine with $\frac{\partial L}{\partial t}\Delta t$ to form the full $d(L\,\Delta t)/dt$ we already added.

Thus the entire second bracket:

$$
-\frac{\partial L}{\partial \dot{q}_i}\dot{q}_i\,\frac{d}{dt}(\Delta t)
+\frac{\partial L}{\partial t}\Delta t
+\frac{d}{dt}(L\,\Delta t)
= 0
$$

---

##### **3) The Final Form**

We are left only with:

$$
\delta L =
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\Delta q_i\right)
-\frac{d}{dt}(L\,\Delta t)
=
\frac{d}{dt}\left[\frac{\partial L}{\partial \dot{q}_i}\Delta q_i - L\,\Delta t\right]
$$

But by the symmetry condition $\delta L = dF/dt$, so:

$$
\frac{d}{dt}\left[\frac{\partial L}{\partial \dot{q}_i}\Delta q_i - (L\,\Delta t - F)\right] = 0
$$

Thus the quantity in brackets is **conserved**.

---

#### **Key Intuition**

The **add-and-subtract** step is just a clever way of recognizing that the remaining terms can be written as a **total time derivative** of $L\,\Delta t$.  
By inserting $0 = +d(L\,\Delta t)/dt - d(L\,\Delta t)/dt$, we can combine all terms into a single total derivative, which is exactly what Noether’s theorem guarantees is conserved.

---

### Step 6: Conserved Quantity

Since the total derivative vanishes:

$$
\boxed{
\frac{d}{dt}\left[\sum_i \frac{\partial L}{\partial \dot{q}_i}\,\Delta q_i \;-\;\big(L\,\Delta t - F\big)\right] = 0
}
$$

Thus the quantity in brackets is **conserved**.

---

### Simplified Case

If:

- $\Delta t = 0$ (no time transformation),
- $F=0$ (Lagrangian strictly invariant),

then:

$$
Q = \sum_i \frac{\partial L}{\partial \dot{q}_i}\,\Delta q_i
\quad \text{is constant.}
$$

## B4.4.5 Canonical Form (No Time Transformation)

If $\Delta t = 0$ and $F=0$ (pure configuration-space symmetry):

$$
\boxed{Q = \sum_i \frac{\partial L}{\partial \dot{q}_i}\, \Delta q_i \quad \text{is conserved.}}
$$

This is often sufficient for mechanical systems.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.4.6 Examples

### Example 1: Spatial Translation → Linear Momentum

For a free particle in 1D:

- $L = \tfrac{1}{2}m\dot{x}^2$

A small translation: $x \to x + \epsilon$, so:

- $\Delta x = 1$, $\Delta t = 0$

Conserved quantity:

$$
Q = \frac{\partial L}{\partial \dot{x}}\, \Delta x = m \dot{x} = p \quad (\text{linear momentum})
$$

---

### Example 2: Time Translation → Energy Conservation

**Problem:**  
Show that if the Lagrangian does not depend explicitly on time ($\partial L/\partial t = 0$), energy is conserved.

---

#### Step 1: Identify the Symmetry Transformation

A time translation means:

$$
t' = t + \epsilon
$$

Comparing to the general infinitesimal form $t' = t + \epsilon\,\Delta t$:

- **$\Delta t = 1$** (the generator is a unit shift in time per unit $\epsilon$).

The generalized coordinates $q_i$ do **not** change under a pure time translation:

$$
\Delta q_i = 0
$$

($q_i$ evaluated at a shifted time $t+\epsilon$ changes only indirectly via the dynamics, not as an explicit transformation).

---

#### Step 2: Noether’s Conserved Quantity (Naive approach)

The general Noether conserved quantity (full form) is:

$$
Q = \sum_i \frac{\partial L}{\partial \dot{q}_i}\,\Delta q_i \;-\;\left(L\,\Delta t - F\right)
$$

For this symmetry:

- $\Delta q_i = 0$
- $\Delta t = 1$
- $F = 0$ (no extra total derivative)

So:

$$
Q = -\left(L\,\Delta t\right) = -L
$$

>Because Noether’s theorem requires the active infinitesimal transformation of every dynamical variable. For time translations, the field shape doesn’t change — but the argument of the function shifts:
>
>$$ q(t) \rightarrow q(t+\epsilon)$$
>
>This induces a genuine change in q:
>
>$$\Delta q = \epsilon\dot{q}$$
>
>Failing to include this induced change leads directly to the “wrong” Noether charge.
>
> **If we naively treat a time translation as having $\Delta q_i = 0$, Noether’s theorem gives $Q = -L$.  
> But this ignores the induced variation $\Delta q_i = \dot{q}_i\,\Delta t$ that arises when the time argument shifts.  
> Including this full transformation yields the correct conserved quantity  
> $\displaystyle Q = \sum_i p_i \dot{q}_i - L$.**


---

#### Step 3: From Noether’s General Form to the Hamiltonian

The general conserved quantity from Noether’s theorem is:

$$
Q = \sum_i p_i\,\Delta q_i \;-\;\big(L\,\Delta t - F\big),
$$

with $p_i = \dfrac{\partial L}{\partial \dot{q}_i}$.

For a **time translation**:

1. **The full transformation affects both $t$ and $q_i$:**

   Because we shift time $t \to t + \epsilon$,
   each generalized coordinate also shifts:

   $$
   q_i(t) \to q_i(t+\epsilon)
   \approx q_i(t) + \epsilon\,\dot{q}_i
   $$

   So:

   $$
   \Delta q_i = \dot{q}_i\,\Delta t, \qquad (\Delta t = 1)
   $$

2. **Substitute into $Q$:**

   $$
   Q = \sum_i p_i(\dot{q}_i\,\Delta t) \;-\; L\,\Delta t
   $$

3. **Factor out $\Delta t$:**

   $$
   Q = \left(\sum_i p_i\,\dot{q}_i \;-\; L\right)\Delta t
   $$

4. **Choose $\Delta t = 1$:**

   $$
   \boxed{E \equiv Q = \sum_i p_i\,\dot{q}_i \;-\; L}
   $$

Thus the conserved Noether charge for time-translation invariance is exactly the **Hamiltonian** $H$.

---

#### Step 4: Physical Interpretation

- The symmetry means the system looks the same at all moments of time — the Lagrangian does not depend explicitly on $t$.  
- This invariance implies that the **energy of the system cannot change with time**.

---

### Example 3: Rotational Symmetry → Angular Momentum

In 2D polar coordinates with a central potential:

- $L = \tfrac{1}{2}m\left(\dot{r}^2 + r^2\dot{\phi}^2\right) - V(r)$

A small rotation: $\phi \to \phi + \epsilon$:

- $\Delta \phi = 1$, $\Delta t = 0$

Conserved quantity:

$$
Q = \frac{\partial L}{\partial \dot{\phi}} \Delta \phi = m r^2 \dot{\phi} = \ell \quad (\text{angular momentum})
$$

---

### Summary Table – Symmetries and Conserved Quantities

| **Symmetry**            | **Transformation**                  | **Conserved Quantity** |
|--------------------------|--------------------------------------|-------------------------|
| Time translation        | $t \to t + \epsilon$                | $E = \sum_i \dot{q}_i\,p_i - L$ |
| Spatial translation     | $q_i \to q_i + \epsilon$            | $p_i = \dfrac{\partial L}{\partial \dot{q}_i}$ |
| Rotational symmetry     | $\phi \to \phi + \epsilon$          | $L_z = \dfrac{\partial L}{\partial \dot{\phi}} = m r^2 \dot{\phi}$ |

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.4.7 Why Noether’s Theorem Leads to Group Theory

Noether’s theorem reveals a **deep and universal link**:

$$
\text{Continuous symmetry} \;\Longleftrightarrow\; \text{Conserved quantity.}
$$

But **symmetries are naturally described by groups**:

---

### 1) Symmetries Form Groups

- A set of symmetry transformations, when combined, follows group axioms:
  - **Closure**: performing two symmetry operations is equivalent to a third.
  - **Identity**: doing nothing is a symmetry.
  - **Inverse**: every symmetry has an opposite transformation.
  - **Associativity**: the order of combining operations respects group rules.

- Examples in mechanics:
  - **Time translations** form the additive group $\mathbb{R}$.
  - **Spatial rotations** form the Lie group $SO(3)$.
  - **Translations and rotations together** form the Euclidean group $E(3)$.

---

### 2) Lie Groups and Generators

- Continuous symmetry groups are **Lie groups**.  
- Infinitesimal transformations (used in Noether’s theorem) are generated by the **Lie algebra** of the group.
  
For example:
- Rotations in 3D are generated by the angular momentum operators $L_x, L_y, L_z$ (the Lie algebra of $SO(3)$).
- Momentum operators generate translations.

Noether’s theorem essentially identifies **conserved quantities as the generators of continuous symmetries**.

---

### 3) Transition to Quantum Mechanics

- In quantum mechanics, observables correspond to operators, and **conserved quantities are represented by Hermitian operators** that generate symmetry transformations via commutators.

Examples:
- Time translation → Hamiltonian $H$ (energy conservation).
- Spatial translation → Momentum operators $p_i$.
- Rotation → Angular momentum operators $\hat{L}_i$.

The commutation relations of these operators mirror the Lie algebra of the symmetry group (e.g., $[L_x, L_y] = i\hbar L_z$ for $SO(3)$).

---

### 4) Why We Study Group Theory

- **Noether’s theorem** gives the first taste of this structure: **symmetry → conserved quantity**.  
- **Group theory** formalizes and classifies these symmetries systematically, both in classical and quantum systems.
- Mastering group theory lets us:
  - Predict conserved quantities for any symmetry group.
  - Classify particle states in quantum mechanics.
  - Understand selection rules, degeneracies, and deeper invariance principles.

## B4.4.8 Conclusion

Noether’s theorem is not just a result in mechanics — it is the **gateway** to a symmetry-based view of physics.  
The next natural step is to study **Lie groups and Lie algebras**, which provide the mathematical language to fully exploit these symmetry principles in both **classical mechanics** and **quantum mechanics**.
