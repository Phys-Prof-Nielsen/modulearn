# C7.5 Extension to Generalized Coordinates and Coordinate-Dependent Matrices
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.5.1 Motivation

Until now, we have expressed our systems in **Cartesian coordinates**, where the kinetic energy takes a simple form and the **mass matrix** $M$ is diagonal and constant.  
However, most physical systems—pendulums, molecules, rotating bodies, or mechanical linkages—are not best described by straight-line coordinates.  
In such systems, the kinetic energy depends on the **geometry** and **constraints**, and the entries of $M$ become functions of the generalized coordinates.  

This section introduces the **generalized-coordinate formulation**, where $M(\mathbf{q})$ may vary with position and can include off-diagonal terms representing **inertial coupling** between coordinates.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.5.2 Generalized coordinates and Lagrangian form

Let $\mathbf{q} = (q_1, q_2, \dots, q_n)$ denote a set of **generalized coordinates** that uniquely specify the configuration of a system.  
The kinetic and potential energies are written in the general form

$$
T = \tfrac{1}{2}\dot{\mathbf{q}}^T M(\mathbf{q}) \dot{\mathbf{q}},
\qquad
U = U(\mathbf{q}).
$$

The **Lagrangian** is then

$$
L(\mathbf{q}, \dot{\mathbf{q}}) = T - U
= \tfrac{1}{2}\dot{\mathbf{q}}^T M(\mathbf{q}) \dot{\mathbf{q}} - U(\mathbf{q}).
$$

The Euler–Lagrange equations yield:

$$
\frac{d}{dt}\!\left(\frac{\partial L}{\partial \dot{q}_i}\right)
- \frac{\partial L}{\partial q_i} = 0.
$$

Carrying out the derivatives gives

$$
M(\mathbf{q})\,\ddot{\mathbf{q}}
+ C(\mathbf{q},\dot{\mathbf{q}})\,\dot{\mathbf{q}}
+ \nabla U(\mathbf{q}) = 0,

$$
where the matrix $C$ collects velocity-dependent terms arising from $\dot{M}(\mathbf{q})$.  
For small oscillations about equilibrium, the dependence on $\mathbf{q}$ can be neglected and $M$ is treated as constant.

## C7.5.3 Example 1

<div style="background-color:#d9d9d9; padding:15px; border-radius:10px;">

### Simple pendulum

Let the generalized coordinate be the angle $\theta$ measured from the vertical.

- Kinetic energy:

  $$
  T = \tfrac{1}{2} m (L\dot{\theta})^2 = \tfrac{1}{2} m L^2 \dot{\theta}^2
  \Rightarrow M = [\,mL^2\,].
  $$
  
- Potential energy (relative to the lowest point):

  $$
  U = mgL(1 - \cos\theta)
  \approx \tfrac{1}{2} mgL\,\theta^2
  \Rightarrow K = [\,mgL\,].
  $$

Equation of motion:

$$
mL^2 \ddot{\theta} + mgL\,\theta = 0,
\qquad
\text{or}
\qquad
\ddot{\theta} + \frac{g}{L}\theta = 0.
$$

Here, $M$ is **not just the mass**—it represents the **rotational inertia** about the pivot.
</div>

## C7.5.4 Example 2

<div style="background-color:#d9d9d9; padding:15px; border-radius:10px;">

Coupled double pendulum (outline)

For a planar double pendulum with angles $\theta_1$ and $\theta_2$:

- The kinetic energy contains cross terms such as $\dot{\theta}_1 \dot{\theta}_2$.
- This leads to a **non-diagonal mass matrix**:

  $$
  M(\mathbf{q}) =
  \begin{pmatrix}
  m_1 L_1^2 + m_2(L_1^2 + 2L_1L_2\cos\theta_2 + L_2^2) & m_2(L_1L_2\cos\theta_2 + L_2^2) \\[6pt]
  m_2(L_1L_2\cos\theta_2 + L_2^2) & m_2L_2^2
  \end{pmatrix}.
  $$
  
- The off-diagonal elements represent **inertial coupling**: motion in one coordinate affects the inertia of the other.

This example illustrates that $M(\mathbf{q})$ can depend explicitly on the configuration angles, and its symmetry guarantees that kinetic energy is always real and positive.

</div>

## C7.5.5 Geometric interpretation

The kinetic energy

$$
T = \tfrac{1}{2}\dot{\mathbf{q}}^T M(\mathbf{q}) \dot{\mathbf{q}}
$$

can be viewed as defining a **metric tensor** on the configuration space.  
Thus, $M(\mathbf{q})$ plays the same role as the metric $g_{ij}$ in differential geometry:

$$
ds^2 = \sum_{i,j} g_{ij} dq_i dq_j.
$$

This geometric view links mechanics with geometry:
- The **shape** of $M(\mathbf{q})$ determines how kinetic energy depends on direction in configuration space.
- The **curvature** of $U(\mathbf{q})$ determines the restoring forces near equilibrium.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.5.6 Linearization near equilibrium

At an equilibrium $\mathbf{q}_0$, expand:

$$
\mathbf{q} = \mathbf{q}_0 + \boldsymbol{\xi}, \quad
M(\mathbf{q}) \approx M(\mathbf{q}_0), \quad
U(\mathbf{q}) \approx U(\mathbf{q}_0) + \tfrac{1}{2}\boldsymbol{\xi}^T K \boldsymbol{\xi}.
$$

The linearized equations of motion are then

$$
M(\mathbf{q}_0)\ddot{\boldsymbol{\xi}} + K\,\boldsymbol{\xi} = 0,
$$

identical in form to our earlier Cartesian result, but with $M$ and $K$ now representing **geometric and energetic curvatures** at equilibrium.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.5.7 Summary

| Concept | Newtonian / Cartesian | Generalized coordinates |
|----------|----------------------|--------------------------|
| Mass matrix $M$ | Constant, often diagonal | Coordinate-dependent, may have off-diagonal terms |
| Stiffness matrix $K$ | Constant near equilibrium | Derived from curvature of $U(\mathbf{q})$ |
| Coupling | From connecting springs | From geometry or constraints |
| Example | Coupled masses on a line | Double pendulum, rotating linkages |

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.5.8 Transition

We are now equipped to handle **small oscillations in arbitrary coordinate systems**.  
The matrices $M(\mathbf{q}_0)$ and $K$ evaluated at equilibrium fully characterize the linearized dynamics:

$$
M(\mathbf{q}_0)\ddot{\boldsymbol{\xi}} + K\,\boldsymbol{\xi} = 0,
\qquad
(K - \omega^2 M)\mathbf{A} = 0.
$$

In the next section, we use these same matrices to analyze the **stability** of equilibria, showing how the curvature of $U$ and the definiteness of $M$ determine whether small perturbations grow, decay, or oscillate about equilibrium.
