# B2.6 Stability and the Hessian
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.1 Motivation

Not every equilibrium point of a mechanical system is stable.  
A ball at the bottom of a bowl will return to equilibrium after a small displacement,  
while a ball balanced on top of a hill will roll away at the slightest disturbance.  

The question of **stability** asks:  
> “If we slightly perturb a system from its equilibrium position, will it return, remain nearby, or move away?”

Stability analysis builds directly on our small-oscillation framework.  
By expanding the potential energy near equilibrium and examining its **curvature**, we can determine whether motion is oscillatory (stable) or divergent (unstable).  
The mathematical tool for this analysis is the **Hessian matrix** of second derivatives of the potential energy.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.2 Equilibrium and small perturbations

Consider a conservative system with generalized coordinates $\mathbf{q} = (q_1, q_2, \dots, q_n)$ and potential energy $U(\mathbf{q})$.  
An equilibrium point $\mathbf{q}_0$ satisfies:

$$
\left.\frac{\partial U}{\partial q_i}\right|_{\mathbf{q}_0} = 0.
$$

To study small motions near $\mathbf{q}_0$, define the small displacement

$$
\boldsymbol{\xi} = \mathbf{q} - \mathbf{q}_0.
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.3 Expansion of potential energy: The Hessian

Perform a Taylor expansion of $U(\mathbf{q})$ about $\mathbf{q}_0$:

$$
U(\mathbf{q}) = U(\mathbf{q}_0)
+ \tfrac{1}{2}\,\boldsymbol{\xi}^T K\,\boldsymbol{\xi}
+ \mathcal{O}(\xi^3),

$$
where the **Hessian matrix** $K$ is defined as

$$
K_{ij} = 
\left.\frac{\partial^2 U}{\partial q_i \partial q_j}\right|_{\mathbf{q}_0}.
$$

This quadratic term describes the local “shape” of the potential surface.  
In one dimension, $K = U''(q_0)$ represents the curvature of $U$ at equilibrium.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.4 Equation of motion near equilibrium

The Lagrangian near $\mathbf{q}_0$ can be written as

$$
L = \tfrac{1}{2}\dot{\boldsymbol{\xi}}^T M(\mathbf{q}_0)\dot{\boldsymbol{\xi}}
    - \tfrac{1}{2}\boldsymbol{\xi}^T K \boldsymbol{\xi}.
$$

The Euler–Lagrange equations give

$$
M(\mathbf{q}_0)\ddot{\boldsymbol{\xi}} + K\,\boldsymbol{\xi} = 0,
$$

which leads to the eigenvalue problem

$$
(K - \omega^2 M)\mathbf{A} = 0.
$$

The signs of the eigenvalues of $M^{-1}K$ determine the nature of equilibrium.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.5 Stability criterion

For a conservative system:

- **Stable equilibrium** → all eigenvalues of $M^{-1}K$ are **positive**  
  ⇒ all $\omega^2 > 0$ → small oscillations are sinusoidal and bounded.
- **Unstable equilibrium** → at least one eigenvalue of $M^{-1}K$ is **negative**  
  ⇒ some $\omega^2 < 0$ → motion grows exponentially in time.
- **Neutral equilibrium** → at least one eigenvalue is **zero**  
  ⇒ motion is undetermined along that direction (flat potential valley).

The stability thus depends on the **definiteness** of the Hessian matrix $K$ at equilibrium.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.6 Geometric interpretation

The potential $U(\mathbf{q})$ defines a surface in configuration space.  
The Hessian $K$ measures the local curvature of this surface:

| Shape of $U$ near $\mathbf{q}_0$ | Eigenvalues of $K$ | Stability type | Description |
|----------------------------------|--------------------|----------------|--------------|
| Local minimum (bowl) | all positive | Stable | Restoring forces pull system back |
| Local maximum (hill) | all negative | Unstable | Forces push system further away |
| Saddle point | mixed signs | Unstable | Stable in some directions, unstable in others |
| Flat direction | one or more zero | Neutral | No restoring force in that direction |

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.6 Examples

<div style="background-color:#d9d9d9; padding:15px; border-radius:10px;">

**Simple pendulum**

For a pendulum with potential $U = mgL(1 - \cos\theta)$:

- At $\theta = 0$ (bottom):  
  $U''(0) = mgL > 0$ → **stable** small oscillations.

- At $\theta = \pi$ (top):  
  $U''(\pi) = -mgL < 0$ → **unstable** equilibrium.

</div>

<br>

<div style="background-color:#d9d9d9; padding:15px; border-radius:10px;">

**Two-dimensional potential surface**

Let

$$
U(x,y) = \tfrac{1}{2}(4x^2 + y^2 - 2xy).
$$

Then

$$
K =
\begin{pmatrix}
4 & -1 \\[4pt]
-1 & 1
\end{pmatrix}.
$$

Compute the eigenvalues:

$$
\lambda_{1,2} = \frac{5 \pm \sqrt{5^2 - 4(3)}}{2} = \frac{5 \pm \sqrt{13}}{2} > 0.
$$

Since both are positive, the equilibrium at $(x,y) = (0,0)$ is **stable**.

If one eigenvalue had been negative, it would have indicated a **saddle point**, unstable along that direction.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.7 Energy viewpoint

The total energy near equilibrium is approximately quadratic:

$$
E = \tfrac{1}{2}\dot{\boldsymbol{\xi}}^T M \dot{\boldsymbol{\xi}}
  + \tfrac{1}{2}\boldsymbol{\xi}^T K \boldsymbol{\xi}.
$$

If both $M$ and $K$ are **positive definite**, then $E$ is positive definite —  
the energy increases with displacement, so the system remains bounded.  
If $K$ is not positive definite, $E$ can decrease indefinitely in some direction → instability.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.8 Practical stability test

1. Compute the equilibrium positions by solving $\nabla U = 0$.  
2. Evaluate the Hessian $K$ at each equilibrium.  
3. Determine the signs of its eigenvalues:  
   - All positive → stable  
   - Any negative → unstable  
   - Zero → neutral (check higher-order terms)

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.9 Connection to normal modes

Each positive eigenvalue $\lambda_i$ of $M^{-1}K$ corresponds to a **normal mode** with frequency

$$
\omega_i = \sqrt{\lambda_i}.
$$

Thus, the stability problem and the oscillation problem are **two aspects of the same linearized dynamics**:
- Stability → sign of $\lambda_i$
- Oscillation → magnitude of $\lambda_i$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.10 Summary

| Concept | Mathematical condition | Physical meaning |
|----------|------------------------|------------------|
| Stable equilibrium | $K$ positive definite | All small perturbations oscillate and remain bounded |
| Unstable equilibrium | $K$ indefinite (mixed signs) | Some perturbations grow exponentially |
| Neutral equilibrium | $K$ semidefinite | Flat directions; no restoring force |
| Small oscillations | $\omega_i^2 = \text{eigenvalues}(M^{-1}K)$ | Frequencies of bounded motion |

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.6.11 Concluding remarks

The **Hessian** provides a local map of how potential energy curves in configuration space.  
Its **positive definiteness** ensures small perturbations produce oscillatory motion—  
the mathematical signature of stability.

If time allows, we can extend this analysis to include **damping** and **driving**,  
which modify the stability of equilibria in realistic systems.
