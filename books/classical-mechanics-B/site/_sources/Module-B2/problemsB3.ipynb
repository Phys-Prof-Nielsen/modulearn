### Problem 1: Frequency Ratio Identification
A Lissajous figure shows **three horizontal lobes** and **two vertical lobes**.  
The curve is stable and closed.

1. What is the frequency ratio $\omega_x : \omega_y$?  
2. If the phase difference increases, what geometric change will occur to the curve?

<details>
<summary><strong>Solution 1</strong></summary>

**1. Frequency ratio**

Horizontal lobes correspond to the numerator, vertical lobes to the denominator:

$$
\omega_x : \omega_y = 3 : 2.
$$

**2. Effect of increasing $\Delta\phi$**

Changing $\Delta\phi$:

- does **not** change the number of lobes  
- **does** rotate, twist, or skew the figure  
- separates overlapping arcs, making the structure appear “unfolded”

Thus the curve changes **orientation**, not its topological structure.

</details>

### Problem 4: Rational vs. Irrational Ratio – Plotting

Consider:

$$
x(t)=\cos(3t), \qquad y(t)=\cos(4t + \tfrac{\pi}{4}),
$$

and

$$
x(t)=\cos(t), \qquad y(t)=\cos(\sqrt{2}\,t + \tfrac{\pi}{4}).
$$

1. Which motion produces a **closed** curve?  
2. Which produces a **quasi-periodic** curve?  
3. Produce a Python script showing **both plots side-by-side**.


<details>
<summary><strong>Solution 4</strong></summary>

**1. Closed curve**

The first motion has ratio $3:4$ → **closed**.

---

**2. Non-closed**

The second motion uses an irrational ratio $1:\sqrt{2}$ → **quasi-periodic**, never repeating.

---

**3. Python script**

</details>


```python
%reset -f
import numpy as np
import matplotlib.pyplot as plt

t1 = np.linspace(0, 2*np.pi, 3000)
t2 = np.linspace(0, 40*np.pi, 8000)

# closed case: 3:4
x1 = np.cos(3*t1)
y1 = np.cos(4*t1 + np.pi/4)

# irrational case: 1:sqrt(2)
x2 = np.cos(t2)
y2 = np.cos(np.sqrt(2)*t2 + np.pi/4)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,5))

ax1.plot(x1, y1)
ax1.set_title("Closed Lissajous (3:4)")
ax1.set_aspect("equal", "box")
ax1.grid(True)

ax2.plot(x2, y2, linewidth=0.5)
ax2.set_title("Non-Closed (1 : √2)")
ax2.set_aspect("equal", "box")
ax2.grid(True)

plt.tight_layout()
plt.show()
```

### Problem 5: Determining Phase From Oscilloscope Data

A Lissajous figure on an oscilloscope shows an **ellipse**, tilted upward to the right.  
Both signals have the same frequency.

1. Is $\Delta\phi$ positive or negative?  
2. If the ellipse collapsed into a straight line with **positive slope**, what would $\Delta\phi$ be?  
3. Create a Python script plotting an equal-frequency ellipse with $\Delta\phi=\pi/3$.


<details>
<summary><strong>Solution 5</strong></summary>

**1. Sign of $\Delta\phi$**

A right-tilted ellipse means $y(t)$ **leads** $x(t)$:

$$
\Delta\phi > 0.
$$

---

**2. Straight line of positive slope**

This corresponds to:

$$
\Delta\phi = 0.
$$

---

**3. Python script**

See next cell.

</details>


```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

A = 1.0
omega = 1.0

phi_x = 0.0
phi_y = np.pi/3   # 60° lead

t = np.linspace(0, 2*np.pi, 2000)

x = A * np.cos(omega*t + phi_x)
y = A * np.cos(omega*t + phi_y)

plt.figure(figsize=(5,5))
plt.plot(x, y)
plt.title(r"Equal-Frequency Ellipse: $\Delta\phi=\pi/3$")
plt.xlabel("$x(t)$")
plt.ylabel("$y(t)$")
plt.gca().set_aspect("equal", "box")
plt.grid(True)
plt.show()
```

### Problem 6: Two Given Input Signals → Determine Lissajous Geometry

Two signals are fed into the horizontal and vertical inputs of an oscilloscope:

$$
x(t) = 2\cos\!\left(5t - \frac{\pi}{6}\right),
$$

$$
y(t) = 3\cos\!\left(3t + \frac{\pi}{3}\right).
$$

1. Determine the **frequency ratio** $\omega_x : \omega_y$.  
2. Compute the **phase difference** $\Delta\phi = \phi_y - \phi_x$.  
3. Based on the frequency ratio, will the Lissajous figure be **closed** or **quasi-periodic**?  
4. Without plotting, predict the **number of horizontal and vertical lobes**.  
5. Write a **Python script** to plot the resulting Lissajous figure.


<details>
<summary><strong>Solution 6</strong></summary>

### **1. Frequency ratio**

From the expressions:

- $x(t)$ uses frequency $\omega_x = 5$
- $y(t)$ uses frequency $\omega_y = 3$

Thus:

$$
\omega_x : \omega_y = 5 : 3.
$$

---

### **2. Phase difference**

Extract phases:

- $\phi_x = -\frac{\pi}{6}$
- $\phi_y = +\frac{\pi}{3}$

Thus:

$$
\Delta\phi = \phi_y - \phi_x
           = \frac{\pi}{3} - \left(-\frac{\pi}{6}\right)
           = \frac{\pi}{3} + \frac{\pi}{6}
           = \frac{\pi}{2}.
$$

The $y$-signal **leads** the $x$-signal by $90^\circ$.

---

### **3. Closed or quasi-periodic?**

The ratio $5/3$ is rational, so the Lissajous curve will be **closed**.

---

### **4. Number of horizontal and vertical lobes**

For a rational ratio $m:n$:

- **Horizontal lobes** = $m$ (from $x(t)$)
- **Vertical lobes** = $n$ (from $y(t)$)

Thus:

- **5 horizontal lobes**  
- **3 vertical lobes**

---

### **5. Expected appearance**

A 5:3 figure tends to resemble a symmetric woven shape with 5 “wrappings” across horizontally and 3 vertically.  
The phase difference of $\pi/2$ will **tilt** and **skew** the pattern.

</details>


```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Given signals
t = np.linspace(0, 2*np.pi, 4000)

x = 2 * np.cos(5*t - np.pi/6)
y = 3 * np.cos(3*t + np.pi/3)

plt.figure(figsize=(6,6))
plt.plot(x, y, linewidth=1.0)

plt.title(r"Lissajous Figure for $x(t)=2\cos(5t-\pi/6)$ and $y(t)=3\cos(3t+\pi/3)$")
plt.xlabel("$x(t)$")
plt.ylabel("$y(t)$")

plt.grid(True)
plt.gca().set_aspect("equal", "box")
plt.show()
```

  $$
  \theta_{\text{knee, osc}}(t)

### 2. Frequency ratio

Hip frequency:

\omega_{\text{hip}} = 2

\omega_{\text{hip}} : \omega_{\text{knee}}

\phi_{\text{arm}} = \frac{\pi}{6}

\Delta\phi = \phi_{\text{arm}} - \phi_{\text{COM}}

### Problem 9: Advanced Biomechanics – Healthy vs. Parkinsonian Gait

During a gait analysis study, sensors are placed on the **left** and **right** shins to record the forward–backward swing of each leg.  

Two simplified models are used:

---

#### Healthy Gait (phase-locked)
The left and right leg oscillations are modeled as:

$$
x_L(t) = 18\cos\!\left(4t - \frac{\pi}{6}\right), \qquad
x_R(t) = 18\cos\!\left(4t + \frac{5\pi}{6}\right).
$$

#### Parkinsonian Gait (loss of phase-locking)
The frequency of the right leg oscillation becomes slightly irregular:

$$
x_L(t) = 18\cos\!\left(4t - \frac{\pi}{6}\right), \qquad
x_R(t) = 18\cos\!\left((4 + 0.05)t + \frac{5\pi}{6}\right).
$$

---

#### Tasks

1. For the healthy gait:
   - Determine the phase difference $\Delta\phi = \phi_R - \phi_L$.
   - Will the Lissajous figure (left vs. right) be **closed** or **non-closed**?
   - Sketch/predict the qualitative appearance of the curve.

2. For the Parkinsonian gait:
   - Explain why the slight frequency mismatch leads to a **drifting, non-closed** curve.
   - Describe what this drift indicates about neural coordination.

3. Using Python, plot **both** Lissajous figures (healthy and Parkinsonian) in **side-by-side subplots**.

4. Compare the two plots and describe how clinicians can use such patterns to detect early gait impairment.


<details>
<summary><strong>Solution 9</strong></summary>

### **1. Healthy gait**

Left leg:
$$
\phi_L = -\frac{\pi}{6}, \qquad \omega_L = 4.
$$

Right leg:
$$
\phi_R = \frac{5\pi}{6}, \qquad \omega_R = 4.
$$

Phase difference:
$$
\Delta\phi = \phi_R - \phi_L
= \frac{5\pi}{6} - \left(-\frac{\pi}{6}\right)
= \pi.
$$

Thus the legs are in **perfect anti-phase**, as expected for normal walking.

Since the frequencies are equal:
$$
\omega_L = \omega_R,
$$
the Lissajous figure is a **closed** shape (an ellipse or line).

Because $\Delta\phi = \pi$:
- The pattern collapses into a **straight line with negative slope**,  
- meaning perfect left/right alternation.

This represents **highly coordinated gait**.

---

### **2. Parkinsonian gait**

Now the right-leg frequency is:
$$
\omega_R = 4.05, \qquad \omega_L = 4.0.
$$

The ratio:
$$
\frac{\omega_L}{\omega_R} = \frac{4}{4.05} \approx 0.9877
$$
is **irrational** for all practical purposes.

Thus the Lissajous figure is **not closed** and slowly drifts.

This drift comes from **phase desynchronization**:

$$
\Delta\phi(t) 
= (\omega_R - \omega_L)t + \text{constant} 
= 0.05 t + C.
$$

Meaning:

- The right leg slowly gets **out of phase** with the left,
- The pattern becomes **quasi-periodic**, filling an area,
- This behavior matches the **loss of rhythm and coordination** seen in Parkinson’s disease.

---

### **3. Clinical interpretation**

The closed (healthy) vs. drifting (Parkinsonian) Lissajous figures show:

- **Healthy gait** → stable, phase-locked coordination  
- **Parkinsonian gait** → deterioration in left–right timing, visible as a **non-closed, wandering curve**

Researchers and clinicians use this type of analysis to quantify:

- gait stability,  
- stride-to-stride variability,  
- freezing-of-gait onset,  
- early neurological impairment.

</details>


```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Time interval long enough to show drift in Parkinsonian gait
t = np.linspace(0, 20*np.pi, 8000)

# Healthy gait (equal frequencies)
xL_healthy = 18*np.cos(4*t - np.pi/6)
xR_healthy = 18*np.cos(4*t + 5*np.pi/6)

# Parkinsonian gait (slight frequency mismatch)
xL_PD = 18*np.cos(4*t - np.pi/6)
xR_PD = 18*np.cos((4 + 0.05)*t + 5*np.pi/6)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,6))

# Plot healthy
ax1.plot(xL_healthy, xR_healthy, linewidth=1.0)
ax1.set_title("Healthy Gait: Closed L-R Lissajous (Phase-Locked)")
ax1.set_xlabel("Left Leg Motion")
ax1.set_ylabel("Right Leg Motion")
ax1.grid(True)
ax1.set_aspect("equal", "box")

# Plot Parkinsonian gait
ax2.plot(xL_PD, xR_PD, linewidth=0.5)
ax2.set_title("Parkinsonian Gait: Drifting L-R Lissajous (Loss of Coordination)")
ax2.set_xlabel("Left Leg Motion")
ax2.set_ylabel("Right Leg Motion")
ax2.grid(True)
ax2.set_aspect("equal", "box")

plt.tight_layout()
plt.show()
```

### Problem 10: Lissajous Art – Two-Pendulum Harmonograph

A “harmonograph” is an art device based on two perpendicular pendulums.  
As each pendulum swings, its motion controls either the **x-position** or **y-position** of a pen.  
The resulting drawings are classic examples of **damped Lissajous figures**.

A simplified physical model is:

$$
x(t) = A_x\,e^{-d_x t}\cos(\omega_x t + \phi_x),
$$

$$
y(t) = A_y\,e^{-d_y t}\cos(\omega_y t + \phi_y).
$$

A particular harmonograph uses:

- $A_x = 1.0$, $A_y = 1.0$
- $\omega_x = 3.0$, $\omega_y = 4.0$
- $d_x = d_y = 0.02$
- $\phi_x = 0$, $\phi_y = \frac{\pi}{2}$

**Tasks**

1. Explain how this system relates to ordinary (undamped) Lissajous figures.  
2. Will the resulting drawing be **closed** or **spiral inward**? Why?  
3. Predict the number of “lobes” or main loops.  
4. Describe qualitatively how damping changes the pattern compared to a standard Lissajous figure.  
5. Write a Python script that generates the resulting harmonograph artwork.


<details>
<summary><strong>Solution 10</strong></summary>

### 1. Relation to standard Lissajous figures

If $d_x = d_y = 0$, the system reduces to:

$$
x(t) = A\cos(\omega_x t + \phi_x), \qquad
y(t) = A\cos(\omega_y t + \phi_y),
$$

which is a **pure Lissajous curve**.  
So a harmonograph is simply a Lissajous system with **damping applied to both coordinates**.

---

### 2. Closed or inward spiral?

Because damping factors $d_x$ and $d_y$ are nonzero:

$$
e^{-d t}
$$

shrinks the amplitude with time.  
Thus the trajectory **spirals inward**, eventually collapsing toward the origin.

Even though the frequency ratio is rational ($3:4$), the curve does *not* close because the amplitude is continually changing.

---

### 3. Number of lobes

Ignoring damping, a $3{:}4$ curve has:

- **3 horizontal lobes**
- **4 vertical lobes**

Damping preserves this lobe count but causes the pattern to **wind inward** through the same regions repeatedly.

---

### 4. Effect of damping

Damping:

- compresses the outer loops inward,
- creates a **spiral or rosette** pattern,
- introduces an artistic “fade” toward the center,
- prevents the curve from ever repeating exactly.

This is why harmonograph drawings look like intricate flowers or rosettes.

---

### 5. Qualitative description

Initially the curve looks like a clean $3{:}4$ Lissajous, but the envelope shrinks slowly, producing:

- sweeping arcs,
- repeated crossings,
- gradually decreasing amplitude,
- an overall rosette structure.

This is exactly the style seen in Lissajous-based kinetic art.

</details>


```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Parameters for the harmonograph
A_x, A_y = 1.0, 1.0
omega_x, omega_y = 3.0, 4.0
d_x, d_y = 0.02, 0.02
phi_x, phi_y = 0.0, np.pi/2

t = np.linspace(0, 40*np.pi, 8000)

x = A_x * np.exp(-d_x*t) * np.cos(omega_x*t + phi_x)
y = A_y * np.exp(-d_y*t) * np.cos(omega_y*t + phi_y)

plt.figure(figsize=(7,7))
plt.plot(x, y, linewidth=0.8)

plt.title("Harmonograph Art (Damped 3:4 Lissajous)")
plt.xlabel("x(t)")
plt.ylabel("y(t)")
plt.axis("equal")
plt.grid(False)

# Aesthetics
plt.gca().set_facecolor("white")
plt.show()
```

### **(B) Spirograph (Hypotrochoid)**
A spirograph curve is defined by:
$$

<details>
<summary><strong>Solution 11A</strong></summary>

### 1. Why the curve is closed when $\delta = 0$

If $\delta = 0$, then

$$
y(t) = \cos(3t + \frac{\pi}{4}),
$$

so both $x(t)$ and $y(t)$ oscillate with the **same frequency** $\omega = 3$.  
Their relative phase is constant, so the parametric plot $(x(t), y(t))$ repeats exactly after each period:

$$
T = \frac{2\pi}{3}.
$$

Thus the curve is a **closed Lissajous figure**, typically an ellipse.

---

### 2. Phase drift when $\delta \neq 0$

With a small frequency mismatch:

$$
y(t) = \cos\!\left((3+\delta)t + \frac{\pi}{4}\right),
$$

the phase of the $y$–signal is

$$
\phi_y(t) = (3+\delta)t + \frac{\pi}{4},
$$

while the $x$–signal has phase

$$
\phi_x(t) = 3t.
$$

The instantaneous phase difference is

$$
\Delta\phi(t) = \phi_y(t) - \phi_x(t)
= \delta t + \frac{\pi}{4}.
$$

Thus the two oscillations **continuously drift out of phase** at a constant rate proportional to $\delta$.

---

### 3. Qualitative appearance of the phase-drift pattern

Because the phase difference is continually changing:

- The curve **does not close**, even after very long times.  
- It slowly morphs through a family of Lissajous shapes.  
- The pattern looks like a **shape that rotates, stretches, and slides** across the plane.  
- Over long intervals, it fills a region in a quasi-periodic way, but never entirely.  
- The visual effect resembles an oscilloscope tuned between two slightly mismatched audio tones.

The result is not a flower or a rosette —  
it is a **continuously evolving ribbon-like pattern**.

---

### 4. Effect of changing $\delta$

- **Smaller $\delta$** → slower phase drift → pattern evolves gently and appears more structured.  
- **Larger $\delta$** → faster drift → more rapid shape changes and a more complex, less stable pattern.  
- **Very large $\delta$** → the curve becomes highly irregular and visually noisy.

Thus $\delta$ controls how quickly the Lissajous figure “slides” through its family of shapes.

The ratio $5:4$ gives a closed Lissajous figure with:

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Base frequencies
omega_x = 3
delta = 0.02               # small drift
omega_y = omega_x + delta

# Phase offset
phi_y = np.pi / 4

# Time interval: long enough to see drift, not long enough to become dense
t = np.linspace(0, 30*np.pi, 6000)

# Signals
x = np.cos(omega_x * t)
y = np.cos(omega_y * t + phi_y)

plt.figure(figsize=(7,7))
plt.plot(x, y, linewidth=0.5, color="blue")

plt.title("Phase-Drift Lissajous Pattern (3 : 3.02)")
plt.xlabel("x(t)")
plt.ylabel("y(t)")

plt.gca().set_aspect("equal", "box")
plt.grid(False)

plt.show()
```

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

R = 5
r = 3
d = 5

t = np.linspace(0, 40*np.pi, 4000)

x = (R - r)*np.cos(t) + d*np.cos((R - r)/r * t)
y = (R - r)*np.sin(t) - d*np.sin((R - r)/r * t)

plt.figure(figsize=(6,6))
plt.plot(x, y, linewidth=1.0)
plt.title("Spirograph / Hypotrochoid ($R=5$, $r=3$, $d=5$)")
plt.xlabel("x(t)")
plt.ylabel("y(t)")
plt.gca().set_aspect("equal", "box")
plt.grid(False)
plt.show()
```
