# B2.4 Lagrangian Formulation and the Origin of the Mass Matrix

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.4.1 Motivation

In the previous section, we obtained the coupled equations of motion using **Newton’s second law** and discovered that they can be expressed compactly in the **matrix form**

$$
M\,\ddot{\mathbf{x}} + K\,\mathbf{x} = 0.
$$

Here, $M$ and $K$ were simply assembled by inspection of the Newtonian equations.  
But where do these matrices *come from* fundamentally?  

The **Lagrangian formalism** provides a powerful and systematic way to derive them directly from the system’s **kinetic** and **potential energy** expressions.  
This approach generalizes effortlessly to many degrees of freedom, and to systems with **non-Cartesian coordinates**, where $M$ may depend on position.  
Understanding the origin of the mass matrix is therefore a key step toward the general theory of oscillations and stability.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.4.2 System setup

We revisit the same two-mass system with possibly unequal spring constants:

$$
| \text{Wall} | - k_1 - m_1 - k_2 - m_2 - k_3 - | \text{Wall} |.
$$

The generalized coordinates are the displacements from equilibrium:

$$
\mathbf{x} =
\begin{pmatrix}
x_1 \\[4pt] x_2
\end{pmatrix}.
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

### Step 1: Kinetic energy

<div style="background-color:#fff1e6; border-left:8px solid #ff6f3c; padding:16px; border-radius:10px;">

#### Box Activity 1: Verifying the Kinetic Energy Matrix Form

The total kinetic energy is

$$
T = \tfrac{1}{2}m_1\dot{x}_1^2 + \tfrac{1}{2}m_2\dot{x}_2^2.
$$

In matrix form:

$$
T = \tfrac{1}{2}\dot{\mathbf{x}}^T
\begin{pmatrix}
m_1 & 0 \\[4pt]
0 & m_2
\end{pmatrix}
\dot{\mathbf{x}},
\quad
\text{where} \quad
\dot{\mathbf{x}} =
\begin{pmatrix}
\dot{x}_1 \\[4pt] \dot{x}_2
\end{pmatrix}.
$$

Thus, the **mass matrix** $M$ arises as the coefficient matrix in the quadratic form of the kinetic energy.

Your task: **verify explicitly** that this matrix expression reproduces the scalar form above.

---

1) Write out the matrix multiplication step-by-step:

$$
\dot{\mathbf{x}}^T
\begin{pmatrix}
m_1 & 0 \\[4pt]
0 & m_2
\end{pmatrix}
\dot{\mathbf{x}}
= 
\begin{pmatrix}
\dot{x}_1 & \dot{x}_2
\end{pmatrix}
\begin{pmatrix}
m_1\dot{x}_1 \\[4pt]
m_2\dot{x}_2
\end{pmatrix}.
$$

2) Simplify the result to obtain:

$$
m_1\dot{x}_1^2 + m_2\dot{x}_2^2.
$$

3) Multiply by $\tfrac{1}{2}$ to recover the total kinetic energy:

$$
T = \tfrac{1}{2}m_1\dot{x}_1^2 + \tfrac{1}{2}m_2\dot{x}_2^2.
$$

---

✅ **Conclusion:**  
The matrix form is simply a compact way of writing the sum of all kinetic energy terms.  
The diagonal entries of the mass matrix correspond to the coefficients of $\dot{x}_1^2$ and $\dot{x}_2^2$, and off-diagonal entries (if present) would represent **coupled kinetic terms** like $\dot{x}_1\dot{x}_2$ in systems where the motion of one coordinate affects the inertia of another.

<hr style="height:1px;border-width:0;background-color:#ff6f3c;opacity:0.4">

<details>
<summary><em>Hint (click to expand)</em></summary>

Matrix multiplication works row-by-column:

$$
\dot{\mathbf{x}}^T M \dot{\mathbf{x}}
= \sum_{i,j} \dot{x}_i M_{ij} \dot{x}_j.
$$

When $M$ is diagonal, this becomes $\sum_i m_i \dot{x}_i^2$ — exactly the scalar kinetic energy expression.

### Step 2: Potential energy

<div style="background-color:#e6f2ff; border-left:8px solid #007acc; padding:16px; border-radius:10px;">

#### Box Activity 2: Verifying the Potential Energy Matrix Form

Each spring stores potential energy proportional to its extension from equilibrium.  
Using Hooke’s law, the total potential energy is

$$
U = \tfrac{1}{2}k_1x_1^2 + \tfrac{1}{2}k_2(x_2 - x_1)^2 + \tfrac{1}{2}k_3x_2^2.
$$

Expand and collect terms:
$$
U = \tfrac{1}{2}(k_1 + k_2)x_1^2 + \tfrac{1}{2}(k_2 + k_3)x_2^2 - k_2x_1x_2.
$$

In matrix form:

$$
U = \tfrac{1}{2}\mathbf{x}^T
\begin{pmatrix}
k_1 + k_2 & -k_2 \\[4pt]
 -k_2 & k_2 + k_3
\end{pmatrix}
\mathbf{x},
\quad
\text{where} \quad
\mathbf{x} =
\begin{pmatrix}
x_1 \\[4pt] x_2
\end{pmatrix}.
$$

Thus, the **stiffness matrix** $K$ originates from the quadratic expansion of the potential energy about equilibrium.

Your task: **expand this matrix expression** step-by-step to verify that it matches the scalar form above.

---

1) Write out the first multiplication:

$$
\mathbf{x}^T
\begin{pmatrix}
k_1 + k_2 & -k_2 \\[4pt]
 -k_2 & k_2 + k_3
\end{pmatrix}
=
\begin{pmatrix}
x_1 & x_2
\end{pmatrix}
\begin{pmatrix}
(k_1 + k_2)x_1 - k_2x_2 \\[4pt]
 -k_2x_1 + (k_2 + k_3)x_2
\end{pmatrix}.
$$

2) Carry out the multiplication:

$$
x_1[(k_1 + k_2)x_1 - k_2x_2]
+ x_2[-k_2x_1 + (k_2 + k_3)x_2].
$$

3) Simplify:

$$
(k_1 + k_2)x_1^2 + (k_2 + k_3)x_2^2 - 2k_2x_1x_2.
$$

4) Finally, include the prefactor $\tfrac{1}{2}$:

$$
U = \tfrac{1}{2}(k_1 + k_2)x_1^2 + \tfrac{1}{2}(k_2 + k_3)x_2^2 - k_2x_1x_2.
$$

---

✅ **Conclusion:**  
The matrix form correctly reproduces the scalar expression for the potential energy.  
The diagonal terms of $K$ give the coefficients of $x_1^2$ and $x_2^2$, while the off-diagonal terms represent the **coupling energy** between the two masses through the shared spring.

<hr style="height:1px;border-width:0;background-color:#007acc;opacity:0.4">

<summary><em>Hint (click to expand)</em></summary>

Matrix multiplication rule:

$$
\mathbf{x}^T K \mathbf{x} = \sum_{i,j} x_i K_{ij} x_j.
$$
Because $K_{12} = K_{21} = -k_2$, the cross term $x_1x_2$ appears with coefficient $2K_{12} = -2k_2$, which becomes $-k_2x_1x_2$ after multiplying by $\tfrac{1}{2}$.
</details>

### Step 3: The Lagrangian and equations of motion

The Lagrangian is

$$
L = T - U
= \tfrac{1}{2}\dot{\mathbf{x}}^T M \dot{\mathbf{x}}
 - \tfrac{1}{2}\mathbf{x}^T K \mathbf{x}.
$$

The Euler–Lagrange equations for each generalized coordinate $x_i$ are

$$
\frac{d}{dt}\!\left(\frac{\partial L}{\partial \dot{x}_i}\right)
- \frac{\partial L}{\partial x_i} = 0.
$$

Applying these to $L$ gives:

$$
M\,\ddot{\mathbf{x}} + K\,\mathbf{x} = 0,
$$
which is **identical** to the matrix equation obtained from Newton’s laws—but now derived from a **single scalar function** $L$.

---

### Step 4: Interpretation

The Lagrangian approach automatically yields the matrices that govern motion:

| Term | Mathematical form | Physical meaning |
|------|--------------------|------------------|
| $M$ | from $T = \tfrac{1}{2}\dot{\mathbf{x}}^T M \dot{\mathbf{x}}$ | Describes how inertia is distributed and coupled between coordinates |
| $K$ | from $U = \tfrac{1}{2}\mathbf{x}^T K \mathbf{x}$ | Describes how restoring forces depend on displacement |

Unlike the Newtonian method, no component-wise force analysis is required—the structure of the matrices follows directly from energy expressions.

---

### Step 5: The generalized eigenvalue problem

The Lagrangian naturally leads to the **normal mode condition**:

$$
(K - \omega^2 M)\mathbf{A} = 0,
$$

obtained by substituting the trial solution $\mathbf{x}(t) = \mathbf{A}\cos(\omega t)$ into the linearized equations.

Nontrivial motion requires

$$
\det(K - \omega^2 M) = 0,
$$

yielding the **normal frequencies** $\omega_1$, $\omega_2$, and corresponding **mode shapes** $\mathbf{A}_1$, $\mathbf{A}_2$.

---

### Step 6: Advantages of the Lagrangian formulation

1. **Coordinate independence:**  
   The Lagrangian works in any generalized coordinate system; $M$ and $K$ automatically reflect the geometry of the system.
2. **Scalability:**  
   Extends easily to systems with many degrees of freedom or continuous media.
3. **Energy-based clarity:**  
   Kinetic and potential energy clearly identify where coupling arises—through cross terms in $T$ or $U$.
4. **Gateway to stability and nonlinear dynamics:**  
   The same $M$ and $K$ matrices form the foundation for analyzing equilibrium stability and small oscillations in arbitrary systems.

## B2.4.3 Transition

We now understand that both the **mass matrix** and **stiffness matrix** originate from the quadratic structure of the system’s energy near equilibrium.  
This insight prepares us to move beyond Cartesian coordinates: in many physical systems, the generalized coordinates describe **angles, distances, or other curvilinear quantities**, making $M$ itself coordinate-dependent.

In the next section, we extend this formalism to **generalized coordinates**, where  

$$
T = \tfrac{1}{2}\dot{\mathbf{q}}^T M(\mathbf{q}) \dot{\mathbf{q}}, \qquad U = U(\mathbf{q}),
$$

and the same principles yield the equations of motion and stability conditions for systems of arbitrary geometry.
