# B4.6 SO(3) and SU(2): Rotations, Lie Algebras, and Spin Representations

## B4.6.1 The Group SO(3)

- $SO(3)$ is the **special orthogonal group** in 3D:
  
  $$
  SO(3) = \{ R \in \mathbb{R}^{3 \times 3} \mid R^T R = I, \ \det(R) = 1 \}
  $$
  
- These matrices represent **real, proper rotations** in 3D Euclidean space.
- Elements act on vectors: $\vec{v} \to R \vec{v}$.

---

### Lie Algebra $\mathfrak{so}(3)$

- It consists of **real $3 \times 3$ antisymmetric matrices**.
- A general element:
  
  $$
  A = \theta_1 J_x + \theta_2 J_y + \theta_3 J_z
  $$
  
  with generators:

  $$
  \begin{align*}
  J_x &= 
  \begin{pmatrix}
  0 & 0 & 0 \\
  0 & 0 & -1 \\
  0 & 1 & 0
  \end{pmatrix}, \quad
  J_y = 
  \begin{pmatrix}
  0 & 0 & 1 \\
  0 & 0 & 0 \\
  -1 & 0 & 0
  \end{pmatrix}, \quad
  J_z = 
  \begin{pmatrix}
  0 & -1 & 0 \\
  1 & 0 & 0 \\
  0 & 0 & 0
  \end{pmatrix}
  \end{align*}
  $$
  
- Commutation relations:
  
  $$
  [J_i, J_j] = \varepsilon_{ijk} J_k
  $$

## B4.6.2 The Group SU(2)

- $SU(2)$ is the group of **unitary $2 \times 2$ complex matrices** with determinant 1:

  $$
  SU(2) = \{ U \in \mathbb{C}^{2 \times 2} \mid U^\dagger U = I, \ \det U = 1 \}
  $$

- This group arises naturally in **quantum mechanics** as the symmetry group of **spin-½ systems**.

---

### Lie Algebra $\mathfrak{su}(2)$

- The Lie algebra $\mathfrak{su}(2)$ consists of **traceless anti-Hermitian $2 \times 2$ matrices**.
- A standard basis (up to normalization) is given by the **Pauli matrices**:

$$
  \begin{align*}
  \sigma_x &= 
  \begin{pmatrix}
  0 & 1 \\
  1 & 0
  \end{pmatrix}, \quad
  \sigma_y = 
  \begin{pmatrix}
  0 & -i \\
  i & 0
  \end{pmatrix}, \quad
  \sigma_z = 
  \begin{pmatrix}
  1 & 0 \\
  0 & -1
  \end{pmatrix}
  \end{align*}
$$

Define:

$$
T_i = \frac{1}{2} \sigma_i
$$

Then the commutation relations are:

$$
[T_i, T_j] = i \varepsilon_{ijk} T_k
$$

So $\mathfrak{su}(2)$ and $\mathfrak{so}(3)$ share **the same structure constants**.

## B4.6.3 Understanding the Relationship Between SU(2) and SO(3)

### What Are We Comparing?

- $SO(3)$ is the group of all **rotations in 3D space**.
  - It acts on real 3D vectors.
  - Used in **classical mechanics** (e.g., rotating a rigid body).
  
- $SU(2)$ is the group of **special unitary $2 \times 2$ complex matrices**.
  - It acts on complex vectors (spinors).
  - Used in **quantum mechanics** (e.g., spin-½ particles like electrons).

So at first glance, they seem very different:
- $SO(3)$ is real, $3 \times 3$, and geometric.
- $SU(2)$ is complex, $2 \times 2$, and more abstract.

But — surprisingly — they are deeply connected.

---

### Same Local Structure: Isomorphic Lie Algebras

- The **Lie algebras** $\mathfrak{so}(3)$ and $\mathfrak{su}(2)$ are **isomorphic**.
  - Both are 3-dimensional vector spaces.
  - Both have the **same commutation relations**:
    
    $$
    [T_i, T_j] = \varepsilon_{ijk} T_k
    $$
    
  - So locally, they describe the **same infinitesimal symmetries**.

✅ This means they are **algebraically identical** at the infinitesimal level.

---

### But Not the Same Globally: SU(2) is a Double Cover of SO(3)

- $SU(2)$ and $SO(3)$ are **not isomorphic as groups**.
- There is a special map:
  
  $$
  f: SU(2) \longrightarrow SO(3)
  $$
  
  that is:
  - A **group homomorphism** (it respects multiplication),
  - **Onto** (every rotation in $SO(3)$ comes from some element in $SU(2)$),
  - But **not one-to-one**:
    
    $$
    f(U) = f(-U)
    $$

- This means:  
  **Two distinct elements** in $SU(2)$ correspond to **the same rotation** in $SO(3)$.

This makes $SU(2)$ a **double cover** of $SO(3)$.

---

### Why This Matters Physically

- In classical mechanics (rotating a ball), $SO(3)$ is enough.
- In quantum mechanics (spinors), $SU(2)$ is required.

For example:
- A spin-½ particle **does not return to its original state after a $2\pi$ rotation**.
  - Instead, it **picks up a minus sign**.
  - Only after a full $4\pi$ rotation does it return to its original state.

This can only be described by $SU(2)$, not $SO(3)$. Thus $SU(2)$ is the **true rotation group** for quantum mechanics.

---

### Summary

| Concept | $SO(3)$ | $SU(2)$ |
|--------|--------|--------|
| Type | Rotation group | Unitary group |
| Action | On real vectors | On spinors |
| Matrix size | $3 \times 3$ real | $2 \times 2$ complex |
| Lie algebra | $\mathfrak{so}(3)$ | $\mathfrak{su}(2)$ |
| Algebraic structure | Same (isomorphic) | Same (isomorphic) |
| Global structure | Simple (no sign ambiguity) | Double cover of $SO(3)$ |
| Physical use | Classical rotation | Quantum spin-½ |

---

### Final Picture

- **Infinitesimally**: $SU(2)$ and $SO(3)$ look the same.
- **Globally**: $SU(2)$ "wraps around" $SO(3)$ twice.
- In quantum mechanics, the "double-valued" nature of spinor states makes $SU(2)$ the right group.

## B4.6.4 Matrix Representations and Exponentiation

- Finite rotations can be built from the Lie algebra by **exponentiating**:

  For $SO(3)$:

  
  $$
  R(\hat{n}, \theta) = e^{\theta \hat{n} \cdot \vec{J}}
  $$

  For $SU(2)$:

  
  $$
  U(\hat{n}, \theta) = e^{i \theta \hat{n} \cdot \vec{T}} = \cos\left(\frac{\theta}{2}\right) I + i \sin\left(\frac{\theta}{2}\right) (\hat{n} \cdot \vec{\sigma})
  $$

- Note the **factor of ½** in the $SU(2)$ version — this is why spinors "come back" only after $4\pi$.
