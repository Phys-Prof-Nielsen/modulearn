## **Problem 1 – Discrete Symmetry Check (Parity & Time Reversal)**

Consider the Lagrangian:

$$
L = \frac12 m\dot{x}^2 - \frac12 k x^2 + \alpha x^4
$$

1. Check whether this Lagrangian is invariant under **parity (P)**.  
2. Check whether it is invariant under **time reversal (T)**.  
3. State whether energy conservation depends on either of these symmetries.

```{admonition} Problem 1: Solution
:class: tip, dropdown

**Given:**

$$
L = \frac12 m\dot{x}^2 - \frac12 k x^2 + \alpha x^4
$$

---

**1. Parity (P)**

- **Transformation:** $x \to -x$, $\dot{x}\to -\dot{x}$.

Substitute:

$$
L' = \frac12 m(-\dot{x})^2 - \frac12 k (-x)^2 + \alpha (-x)^4 \\
= \frac12 m\dot{x}^2 - \frac12 k x^2 + \alpha x^4 = L
$$

✅ **Parity conserved**.

---

**2. Time Reversal (T)**

- **Transformation:** $t \to -t$, $x\to x$, $\dot{x}\to -\dot{x}$.

Substitute:

$$
L' = \frac12 m(-\dot{x})^2 - \frac12 k x^2 + \alpha x^4 = L
$$

✅ **Time reversal symmetry preserved**.

---

**3. Energy Conservation – Apply Noether’s Theorem**

Even though $P$ and $T$ are discrete, **energy conservation** is tied to **time translation symmetry**, a continuous symmetry.

---

**Step 3.1 – Check Time Translation Symmetry**

- The Lagrangian $L(x, \dot{x})$ has **no explicit $t$ dependence** → time translation symmetry exists.

---

**Step 3.2 – Noether Energy Expression**

For time translation symmetry, the Noether conserved quantity is (we have $\Delta t = 1$):

$$
E = \dot{x} \frac{\partial L}{\partial \dot{x}} - L
$$

---

**Step 3.3 – Compute Generalized Momentum**

$$
\frac{\partial L}{\partial \dot{x}} = m\dot{x}
$$

---

**Step 3.4 – Construct the Energy**

Substitute:

$$
E = \dot{x}(m\dot{x}) - \left(\frac12 m\dot{x}^2 - \frac12 k x^2 + \alpha x^4\right)
$$

Simplify:

$$
E = \frac12 m\dot{x}^2 + \frac12 k x^2 - \alpha x^4
$$

Since $L$ has no explicit $t$ dependence:

$$
\frac{dE}{dt} = 0
$$

✅ **Energy conserved**.

---

**4. Physical Meaning**

- The system is **parity and time-reversal symmetric**.  
- Because of **time translation symmetry**, total mechanical energy is conserved, including the quartic potential energy term.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## **Problem 2 – Parity of a Central Potential**

The potential energy for an electric dipole in an electric field is:

$$
U(\mathbf{r}) = \frac{\mathbf{p}\cdot \mathbf{r}}{4\pi \epsilon_0 r^3}
$$

1. Determine whether this potential is invariant under parity transformation $(\mathbf{r} \to -\mathbf{r})$.  
2. Explain physically why this result makes sense.

```{admonition} Problem 2: Solution
:class: tip, dropdown

**1. Parity Transformation**

Substitute $\mathbf{r} \to -\mathbf{r}$:

$$
U' = \frac{\mathbf{p}\cdot(-\mathbf{r})}{4\pi\epsilon_0(-r)^3}
$$

Since $r = |\mathbf{r}| \to |- \mathbf{r}| = r$:

$$
U' = -\frac{\mathbf{p}\cdot\mathbf{r}}{4\pi\epsilon_0 r^3} = -U
$$

❌ **Parity is not conserved** (potential changes sign).

---

**2. Physical Meaning**

A dipole has a preferred direction. A mirror inversion flips this direction → odd under parity.

```

## **Problem 3 – Continuous Symmetry & Conserved Quantity**

The Lagrangian for a free particle in 2D polar coordinates is:

$$
L = \frac12 m(\dot{r}^2 + r^2\dot{\theta}^2)
$$

1. Identify which coordinates correspond to a continuous symmetry.  
2. Use Noether’s theorem to find the conserved quantity.  
3. Explain its physical meaning.

```{admonition} Problem 3: Solution
:class: tip, dropdown

**Given:**

$$
L = \frac12 m(\dot{r}^2 + r^2\dot{\theta}^2)
$$

---

**1. Identify Symmetry**

- The Lagrangian does not depend explicitly on $\theta$ → **rotational symmetry** in $\theta$.
- The continuous transformation is:

$$
\theta \longrightarrow \theta + \epsilon, \qquad r \longrightarrow r
$$

with $\epsilon$ an infinitesimal constant parameter.

---

**2. Apply Noether’s Theorem**

For a continuous symmetry parameterized by $\epsilon$, Noether’s conserved quantity is:

$$
Q = \sum_i \frac{\partial L}{\partial \dot{q}_i} \, \delta q_i
$$

where $\delta q_i$ is the change in each coordinate due to the symmetry.

---

**Step 2.1 – Identify the coordinates and variations**

- Generalized coordinates: $q_1 = r$, $q_2 = \theta$.
- Variations:
  - $\delta r = 0$ (radial coordinate unchanged under rotation).
  - $\delta \theta = \epsilon$ (small constant shift).

---

**Step 2.2 – Compute the generalized momenta**

- For $r$:
  
$$
\frac{\partial L}{\partial \dot{r}} = m\dot{r}, \qquad \delta r = 0
$$

- For $\theta$:

$$
\frac{\partial L}{\partial \dot{\theta}} = m r^2 \dot{\theta}, \qquad \delta \theta = \epsilon
$$

---

**Step 2.3 – Construct Noether’s conserved quantity**

Substitute into $Q$:

$$
Q = \frac{\partial L}{\partial \dot{r}}\,\delta r + \frac{\partial L}{\partial \dot{\theta}}\,\delta \theta \\
Q = (m\dot{r})(0) + (m r^2\dot{\theta})(\epsilon)
$$

Since $\epsilon$ is constant, we can drop it (Noether’s charges are defined up to a constant factor):

$$
\boxed{Q = m r^2 \dot{\theta} = \text{constant}}
$$

---

**3. Physical Meaning**

- $Q$ is the **angular momentum** about the origin:
  
$$
L_z = m r^2 \dot{\theta}
$$

Rotational symmetry → **angular momentum is conserved**.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## **Problem 4 – Time Translation Symmetry**

A particle moves in a potential:

$$
L = \frac12 m\dot{x}^2 - \frac12 k x^2 - F x \cos(\omega t)
$$

1. Does this system have time translation symmetry?  
2. Is energy conserved?  
3. If not, explain why.

```{admonition} Problem 4: Solution
:class: tip, dropdown

**Given:**

$$
L = \frac12 m\dot{x}^2 - \frac12 k x^2 - F x \cos(\omega t)
$$

---

**1. Identify Symmetry**

- A system has **time translation symmetry** if:

$$
t \to t + \epsilon
$$

does **not** change the Lagrangian.

- Here, $L$ depends explicitly on $t$ through the term $F x \cos(\omega t)$.

Thus:

$$
L(x, \dot{x}, t + \epsilon) \neq L(x, \dot{x}, t)
$$

❌ **Time translation symmetry is broken**.

---

**2. Apply Noether’s Theorem**

For time translation symmetry, the conserved quantity (Noether charge) is the **energy** (we have $\Delta t = 1$):

$$
E = \dot{x} \frac{\partial L}{\partial \dot{x}} - L
$$

If the symmetry is broken, $E$ is **not constant**.

Let’s still compute $E$ to see its time dependence.

---

**Step 2.1 – Compute the generalized momentum**

$$
\frac{\partial L}{\partial \dot{x}} = m \dot{x}
$$

---

**Step 2.2 – Construct the Noether charge**

Substitute into the energy expression:

$$
E = \dot{x}(m\dot{x}) - \left(\frac12 m\dot{x}^2 - \frac12 k x^2 - F x \cos(\omega t)\right)
$$

Simplify:

$$
E = \frac12 m\dot{x}^2 + \frac12 k x^2 + F x \cos(\omega t)
$$

Unlike a conserved energy, this explicitly depends on $t$.

---

**3. Physical Meaning**

- The driving force $F x \cos(\omega t)$ continuously exchanges energy with the particle.  
- The time dependence of $E$ shows that energy is **not conserved** because time translation symmetry is broken.

```

## **Problem 5 – Mixed Symmetry Reasoning**

A particle moves in 3D under a potential:

$$
V(x, y, z) = \frac12 k(x^2 + y^2) + mgz
$$

1. List all continuous symmetries.  
2. List any discrete symmetries.  
3. For each symmetry, state the corresponding conserved quantity.

```{admonition} Problem 5: Solution
:class: tip, dropdown

**Given:**

$$
V(x, y, z) = \frac12 k(x^2 + y^2) + mgz
$$

and the corresponding Lagrangian:

$$
L = T - V = \frac12 m(\dot{x}^2 + \dot{y}^2 + \dot{z}^2) - \frac12 k(x^2 + y^2) - mgz
$$

---

**1. Identify Continuous Symmetries**

**1.1 Rotational Symmetry in the $x$–$y$ Plane**

- $L$ depends only on $x^2 + y^2$, not on the angular coordinate $\theta$ in the $x$–$y$ plane.
- Thus, **rotational symmetry** in the $x$–$y$ plane → Noether’s theorem applies.

---

**1.2 Translational Symmetry**

- $L$ explicitly depends on $x$, $y$, and $z$ → **no translational symmetry**.

---

**2. Noether’s Theorem for Rotational Symmetry**

Use polar coordinates in the $x$–$y$ plane:

- Define $r = \sqrt{x^2 + y^2}$, $\theta = \arctan(y/x)$.
- Rewrite the Lagrangian in these coordinates:

$$
L = \frac12 m(\dot{r}^2 + r^2\dot{\theta}^2 + \dot{z}^2) - \frac12 k r^2 - mg z
$$

Here, $\theta$ is a **cyclic coordinate** (no explicit $\theta$ dependence).

---

**Step 2.1 – Noether Charge Expression**

For a continuous symmetry parameterized by $\epsilon$ ($\Delta t = 0$ since we consider a spatial symmetry):

$$
Q = \sum_i \frac{\partial L}{\partial \dot{q}_i}\,\delta q_i
$$

For rotations, $\delta \theta = \epsilon$, $\delta r = 0$, $\delta z = 0$.

So:

$$
Q = \frac{\partial L}{\partial \dot{\theta}}\,\delta \theta
$$

We drop the constant $\epsilon$ (Noether charges are defined up to a multiplicative constant):

$$
Q = \frac{\partial L}{\partial \dot{\theta}}
$$

---

**Step 2.2 – Compute the Generalized Momentum**

$$
\frac{\partial L}{\partial \dot{\theta}} = m r^2 \dot{\theta}
$$

Thus:

$$
\boxed{L_z = m r^2 \dot{\theta} = \text{constant}}
$$

✅ **Angular momentum about the $z$-axis is conserved**.

---

**3. Discrete Symmetries**

Check each coordinate under parity:

- **$x$:** $x \to -x$ → $x^2$ unchanged → even parity.  
- **$y$:** $y \to -y$ → $y^2$ unchanged → even parity.  
- **$z$:** $z \to -z$ → $mgz \to -mgz$ → parity broken along $z$.

---

**4. Summary Table**

| Symmetry | Present? | Conserved Quantity / Meaning |
|----------|----------|------------------------------|
| **Rotation ($x$–$y$ plane)** | ✅ | $L_z = m r^2 \dot{\theta}$ (Noether) |
| **Translation (x, y, z)** | ❌ | None |
| **Parity (x, y)** | ✅ | Even/odd wavefunction classification (QM) |
| **Parity (z)** | ❌ | None |

---

**5. Physical Interpretation**

- The harmonic oscillator in the $x$–$y$ plane behaves like a 2D central potential → rotational symmetry keeps $L_z$ constant.  
- Gravity breaks both $z$-parity and $z$-translation symmetry, so no conservation of vertical momentum or symmetry in $z$.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 6: Lie Brackets in $\mathfrak{so}(3)$ and $\mathfrak{su}(2)$

Let the matrices:

$$
J_x = 
\begin{pmatrix}
0 & 0 & 0 \\
0 & 0 & -1 \\
0 & 1 & 0
\end{pmatrix}, \quad
J_y = 
\begin{pmatrix}
0 & 0 & 1 \\
0 & 0 & 0 \\
-1 & 0 & 0
\end{pmatrix}, \quad
J_z = 
\begin{pmatrix}
0 & -1 & 0 \\
1 & 0 & 0 \\
0 & 0 & 0
\end{pmatrix}
$$

be a basis for $\mathfrak{so}(3)$ (real antisymmetric matrices), and define:

$$
T_x = \frac{1}{2} \sigma_x, \quad
T_y = \frac{1}{2} \sigma_y, \quad
T_z = \frac{1}{2} \sigma_z
$$

where:

\begin{align*}
\sigma_x &= 
\begin{pmatrix}
0 & 1 \\
1 & 0
\end{pmatrix}, \quad
\sigma_y = 
\begin{pmatrix}
0 & -i \\
i & 0
\end{pmatrix}, \quad
\sigma_z = 
\begin{pmatrix}
1 & 0 \\
0 & -1
\end{pmatrix}
\end{align*}

### Tasks:

**(a)** Compute the commutator $[J_x, J_y]$ and show that it equals $J_z$.

**(b)** Compute the commutator $[T_x, T_y]$ and simplify your result.

**(c)** What do your results say about the relationship between $\mathfrak{so}(3)$ and $\mathfrak{su}(2)$?

```{admonition} Problem  6: Solution
:class: tip, dropdown

**(a) Compute $[J_x, J_y]$**

We compute:

$$
\begin{align*}
[J_x, J_y] &= J_x J_y - J_y J_x \\
&=
\begin{pmatrix}
0 & 0 & 0 \\
0 & 0 & -1 \\
0 & 1 & 0
\end{pmatrix}
\begin{pmatrix}
0 & 0 & 1 \\
0 & 0 & 0 \\
-1 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
0 & 0 & 1 \\
0 & 0 & 0 \\
-1 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
0 & 0 & 0 \\
0 & 0 & -1 \\
0 & 1 & 0
\end{pmatrix}
\end{align*}
$$

Multiplying the matrices:

First product:

$$
J_x J_y = 
\begin{pmatrix}
0 & 0 & 0 \\
1 & 0 & 0 \\
0 & 0 & -1
\end{pmatrix}
$$

Second product:

$$
J_y J_x = 
\begin{pmatrix}
0 & -1 & 0 \\
0 & 0 & 0 \\
0 & 0 & -1
\end{pmatrix}
$$

Subtract:

$$
[J_x, J_y] = 
\begin{pmatrix}
0 & 1 & 0 \\
1 & 0 & 0 \\
0 & 0 & 0
\end{pmatrix}
= J_z
$$

✅ So, $[J_x, J_y] = J_z$.

---

**(b) Compute $[T_x, T_y]$**

Recall:

$$
T_x = \frac{1}{2}
\begin{pmatrix}
0 & 1 \\
1 & 0
\end{pmatrix}, \quad
T_y = \frac{1}{2}
\begin{pmatrix}
0 & -i \\
i & 0
\end{pmatrix}
$$

Then:

$$
\begin{align*}
[T_x, T_y] &= T_x T_y - T_y T_x \\
&= \frac{1}{4}
\left(
\begin{pmatrix}
0 & 1 \\
1 & 0
\end{pmatrix}
\begin{pmatrix}
0 & -i \\
i & 0
\end{pmatrix}
\begin{pmatrix}
0 & -i \\
i & 0
\end{pmatrix}
\begin{pmatrix}
0 & 1 \\
1 & 0
\end{pmatrix}
\right)
\end{align*}
$$

Compute each product:

First:

$$
T_x T_y = \frac{1}{4}
\begin{pmatrix}
i & 0 \\
0 & -i
\end{pmatrix}
= \frac{i}{4} \sigma_z
$$

Second:

$$
T_y T_x = \frac{1}{4}
\begin{pmatrix}
-i & 0 \\
0 & i
\end{pmatrix}
= -\frac{i}{4} \sigma_z
$$

So the commutator is:

$$
[T_x, T_y] = \frac{i}{2} T_z
$$

✅ Thus:

$$
[T_x, T_y] = i T_z
$$

---

**(c) Interpretation**

- Both sets of generators satisfy the same **Lie bracket structure**:
 
  $$
  [T_i, T_j] = i \varepsilon_{ijk} T_k, \quad
  [J_i, J_j] = \varepsilon_{ijk} J_k
  $$

- Therefore, the **Lie algebras $\mathfrak{su}(2)$ and $\mathfrak{so}(3)$ are isomorphic** — they have the same algebraic structure.

However:

- The factor of $i$ in $\mathfrak{su}(2)$ arises from working over complex numbers and using Hermitian matrices.
- $\mathfrak{so}(3)$ uses real antisymmetric matrices.

But the underlying structure (i.e., how the generators combine) is **identical**.

✅ **Conclusion**:  
$\mathfrak{so}(3)$ and $\mathfrak{su}(2)$ are different realizations of the **same Lie algebra structure** — one real and geometric, the other complex and quantum-mechanical.

```

## Problem 7: Do These Matrices Form a Lie Algebra?

Let the following $2 \times 2$ real matrices be defined:

$$
A = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}, \quad
B = \begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}, \quad
C = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}
$$

### Tasks:

**(a)** Compute all three commutators: $[A, B]$, $[C, A]$, and $[C, B]$.

**(b)** Check whether the set $\{A, B, C\}$ is closed under the commutator.

**(c)** Based on your results, does this set form a Lie algebra?


```{admonition} Problem 7: Solution
:class: tip, dropdown

**(a) Compute the commutators**

We compute the Lie brackets (commutators) pairwise:

---

**1.** $[A, B] = AB - BA$

$$
AB = 
\begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}
\begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}
= 
\begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}
$$

$$
BA = 
\begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}
\begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}
= 
\begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix}
$$

So:

$$
[A, B] = AB - BA = 
\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} = C
$$

---

**2.** $[C, A] = CA - AC$

$$
CA = 
\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}
\begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}
= 
\begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} = A
$$

$$
AC = 
\begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}
\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}
= 
\begin{pmatrix} 0 & -1 \\ 0 & 0 \end{pmatrix} = -A
$$

So:

$$
[C, A] = A - (-A) = 2A
$$

---

**3.** $[C, B] = CB - BC$

$$
CB = 
\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}
\begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}
= 
\begin{pmatrix} 0 & 0 \\ -1 & 0 \end{pmatrix} = -B
$$

$$
BC = 
\begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}
\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}
= 
\begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix} = B
$$

So:

$$
[C, B] = -B - B = -2B
$$

---

**(b) Is the set $\{A, B, C\}$ closed under the bracket?**

Let’s summarize:

- $[A, B] = C$
- $[C, A] = 2A$
- $[C, B] = -2B$

Each result is a **linear combination** of $A$, $B$, and $C$.

✅ So the set is **closed under commutators**.

---

**(c) Does this form a Lie algebra?**

Check the axioms:

1. **Closure** — ✅ Yes  
2. **Bilinearity** — ✅ Holds for matrix commutators  
3. **Antisymmetry** — ✅ $[X, Y] = -[Y, X]$  
4. **Jacobi identity** — ✅ Holds automatically for matrices

✅ Therefore, the set $\{A, B, C\}$ forms a **3-dimensional Lie algebra**.

---

**Interpretation**

This algebra is **isomorphic to $\mathfrak{sl}(2, \mathbb{R})$**, the set of all $2 \times 2$ real traceless matrices under commutator.

It shares the **same dimension** and **structure** as $\mathfrak{su}(2)$ and $\mathfrak{so}(3)$, but is over $\mathbb{R}$ and not unitary.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 8: Spinor Rotation About the z-Axis

Let $\chi$ be a two-component spinor:

$$
\chi = \begin{pmatrix} 1 \\ 0 \end{pmatrix}
$$

Consider a rotation about the $z$-axis by angle $\theta$. The spinor transforms under $SU(2)$ as:

$$
\chi' = U_z(\theta) \chi
$$

where the SU(2) rotation matrix is given by:

$$
U_z(\theta) = \exp\left(-i \frac{\theta}{2} \sigma_z\right), \quad \text{with } \sigma_z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}
$$

---

### Tasks:

**(a)** Compute the matrix $U_z(\theta)$ explicitly using the exponential of a diagonal matrix.

**(b)** Apply $U_z(\theta)$ to the spinor $\chi = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$ and simplify the result.

**(c)** What happens when $\theta = 2\pi$? What about $\theta = 4\pi$?

**(d)** Interpret the result in terms of **$SU(2)$ vs. $SO(3)$** rotation behavior.


```{admonition} Problem 8: Solution
:class: tip, dropdown

**(a) Compute $U_z(\theta)$**

Since $\sigma_z$ is diagonal, we can exponentiate it easily:

$$
U_z(\theta) = \exp\left(-i \frac{\theta}{2} \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}\right)
= 
\begin{pmatrix}
e^{-i \theta/2} & 0 \\
0 & e^{i \theta/2}
\end{pmatrix}
$$

✅ So:

$$
U_z(\theta) = 
\begin{pmatrix}
e^{-i \theta/2} & 0 \\
0 & e^{i \theta/2}
\end{pmatrix}
$$

---

**(b) Apply $U_z(\theta)$ to $\chi$**

Recall $\chi = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$. Then:

$$
\chi' = U_z(\theta) \chi
= 
\begin{pmatrix}
e^{-i \theta/2} & 0 \\
0 & e^{i \theta/2}
\end{pmatrix}
\begin{pmatrix} 1 \\ 0 \end{pmatrix}
=
\begin{pmatrix} e^{-i \theta/2} \\ 0 \end{pmatrix}
$$

✅ So the transformed spinor is:

$$
\chi' = e^{-i \theta/2} \begin{pmatrix} 1 \\ 0 \end{pmatrix}
$$

---

**(c) Special Cases**

- If $\theta = 2\pi$:

  $$
  \chi' = e^{-i \pi} \chi = (-1)\chi
  $$

- If $\theta = 4\pi$:

  $$
  \chi' = e^{-i 2\pi} \chi = \chi
  $$

✅ The spinor **changes sign** under a $2\pi$ rotation, but **returns to itself** under a $4\pi$ rotation.

---

**(d) Interpretation**

This is the classic behavior of a **spin-½ object**:

- A $2\pi$ rotation does **not** bring it back to the original state.
- Instead, it gains a **minus sign** — it is **double-valued** under rotation.
- Only after a $4\pi$ rotation does the spinor return to itself.

This behavior **cannot** be captured by $SO(3)$, which treats $2\pi$ as the identity.

✅ This shows why quantum spinors **require $SU(2)$**: it covers $SO(3)$ twice and allows spin-½ behavior.

```

## Problem 9: Eigenvalues and Ladder Operators for Angular Momentum

In quantum mechanics, the angular momentum operators $L_x$, $L_y$, and $L_z$ satisfy the commutation relations:

$$
[L_i, L_j] = i\hbar \varepsilon_{ijk} L_k
$$

Define the **ladder operators**:

$$
L_+ = L_x + i L_y, \quad L_- = L_x - i L_y
$$

Let $| \ell, m \rangle$ be a simultaneous eigenstate of $L^2$ and $L_z$, satisfying:

$$
L^2 | \ell, m \rangle = \hbar^2 \ell(\ell + 1) | \ell, m \rangle
$$
$$
L_z | \ell, m \rangle = \hbar m | \ell, m \rangle
$$

---

### Tasks:

**(a)** Show that:

$$
[L_z, L_\pm] = \pm \hbar L_\pm
$$

**(b)** Using this, prove that $L_\pm | \ell, m \rangle$ is also an eigenstate of $L_z$, and find its eigenvalue.


```{admonition} Problem 9: Solution
:class: tip, dropdown

**(a) Commutator $[L_z, L_\pm]$**

Since $L_\pm = L_x \pm i L_y$, we use linearity and the standard commutators:

$$
\begin{align*}
[L_z, L_\pm] &= [L_z, L_x \pm i L_y] \\
&= [L_z, L_x] \pm i [L_z, L_y] \\
&= i\hbar L_y \pm i(-i\hbar L_x) \\
&= \pm \hbar (L_x \pm i L_y) = \pm \hbar L_\pm
\end{align*}
$$

✅ So:

$$
[L_z, L_\pm] = \pm \hbar L_\pm
$$

---

**(b) Action on $| \ell, m \rangle$**

Let’s compute:

$$
\begin{align*}
L_z (L_\pm | \ell, m \rangle) 
&= ([L_z, L_\pm] + L_\pm L_z) | \ell, m \rangle \\
&= (\pm \hbar L_\pm + \hbar m L_\pm) | \ell, m \rangle \\
&= \hbar (m \pm 1) L_\pm | \ell, m \rangle
\end{align*}
$$

✅ So $L_\pm | \ell, m \rangle$ is an eigenstate of $L_z$ with eigenvalue $\hbar (m \pm 1)$.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 10: Angular Momentum Algebra in Classical and Quantum Systems

Consider a particle moving in 3D space with position $\vec{r} = (x, y, z)$ and momentum $\vec{p} = (p_x, p_y, p_z)$. The **classical angular momentum** is defined as:

$$
\vec{L} = \vec{r} \times \vec{p}
$$

In quantum mechanics, the angular momentum components $L_x$, $L_y$, $L_z$ become operators that satisfy:

$$
[L_x, L_y] = i \hbar L_z \quad \text{(and cyclic permutations)}
$$

---

### Tasks:

**(a)** Compute the classical **Poisson bracket** $\{L_x, L_y\}$ using:

$$
\{f, g\} = \sum_{i} \left( \frac{\partial f}{\partial x_i} \frac{\partial g}{\partial p_i} - \frac{\partial f}{\partial p_i} \frac{\partial g}{\partial x_i} \right)
$$

**(b)** Show that $\{L_x, L_y\} = L_z$ and verify the cyclic identities.

```{admonition} Problem 10: Solution
:class: tip, dropdown

**(a) Compute $\{L_x, L_y\}$**

Start with:

- $L_x = y p_z - z p_y$
- $L_y = z p_x - x p_z$
- $L_z = x p_y - y p_x$

Use the Poisson bracket definition:

$$
\{L_x, L_y\} = \sum_i \left( \frac{\partial L_x}{\partial x_i} \frac{\partial L_y}{\partial p_i} - \frac{\partial L_x}{\partial p_i} \frac{\partial L_y}{\partial x_i} \right)
$$

We compute the relevant terms:

- $\frac{\partial L_x}{\partial y} = p_z$, $\frac{\partial L_y}{\partial p_y} = 0$
- $\frac{\partial L_x}{\partial z} = -p_y$, $\frac{\partial L_y}{\partial p_z} = -x$
- $\frac{\partial L_x}{\partial p_y} = -z$, $\frac{\partial L_y}{\partial x} = -p_z$
- $\frac{\partial L_x}{\partial p_z} = y$, $\frac{\partial L_y}{\partial x} = -p_z$

Only two terms are non-zero:

\begin{align*}
\{L_x, L_y\} 
&= \left( \frac{\partial L_x}{\partial p_y} \frac{\partial L_y}{\partial y} - \frac{\partial L_x}{\partial y} \frac{\partial L_y}{\partial p_y} \right) 
+ \left( \frac{\partial L_x}{\partial p_z} \frac{\partial L_y}{\partial z} - \frac{\partial L_x}{\partial z} \frac{\partial L_y}{\partial p_z} \right) \\
&= 0 + (y \cdot 0 - (-p_y)(-x)) = -x p_y + y p_x
= L_z
\end{align*}

✅ So:

$$
\{L_x, L_y\} = L_z
$$

By symmetry, you can verify (do it!):

$$
\{L_y, L_z\} = L_x, \quad \{L_z, L_x\} = L_y
$$

---

**(b) Lie Algebra of Classical Angular Momentum**

From part (a), we showed explicitly:

$$
\{L_x, L_y\} = L_z
$$

We now want to verify that the **classical angular momentum components** form a **Lie algebra**, meaning:

$$
\{L_i, L_j\} = \varepsilon_{ijk} L_k
$$

where $\varepsilon_{ijk}$ is the Levi-Civita symbol.

---

**By symmetry and cyclic structure**:

We expect:

- $\{L_y, L_z\} = L_x$
- $\{L_z, L_x\} = L_y$

Since you did your dilligent job in Part (a), we conclude that the components close under the bracket and form the **Lie algebra $\mathfrak{so}(3)$** — the same algebra that governs rotations in 3D space.

---

**Key Insight**

This structure exactly mirrors the quantum case:

$$
[L_i, L_j] = i\hbar \varepsilon_{ijk} L_k
$$

So the classical and quantum angular momentum share the **same algebraic backbone**, just with Poisson brackets replaced by commutators and scaled by $\hbar$.

```
