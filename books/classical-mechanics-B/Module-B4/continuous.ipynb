# B4.3 Continuous Symmetries in Physics
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.3.1 What Are Continuous Symmetries?

- **Definition:**  
  A **continuous symmetry** is a transformation that depends smoothly on one or more continuous parameters and leaves the physical laws (or action/Lagrangian) invariant.

- **Examples:**  
  - **Translation in space:** parameter = distance $a$.  
  - **Rotation in space:** parameter = angle $\theta$.  
  - **Time translation:** parameter = time shift $t_0$.  
  - **Phase transformation in QM:** $\psi \to e^{i\alpha}\psi$, parameter = $\alpha$.

- **Contrast with discrete symmetry:**  
  - Discrete = "on or off" (finite operations, e.g., $Z_2$).  
  - Continuous = infinitely many transformations connected smoothly.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.3.2 Why Are Continuous Symmetries Important?

1. **Conserved Quantities (Noether’s theorem):**  
   - Every continuous symmetry ↔ a conserved quantity.  
     - Space translation → momentum conservation.  
     - Time translation → energy conservation.  
     - Rotational symmetry → angular momentum conservation.

2. **Simplifies Problem Solving:**  
   - Adapt coordinates to symmetry → drastically easier equations.  
     - Central potentials → spherical coordinates.  
     - Uniform fields → choose axes aligned with symmetry.

3. **Classifies Physical Systems:**  
   - Atomic orbitals labeled by angular momentum ($l, m$ quantum numbers ~ rotational symmetry).  
   - Particle physics: particles organized by representations of continuous groups (*SO(3), SU(2), SU(3)*).

4. **Predicts Forbidden Processes:**  
   - Symmetries restrict allowed transitions (selection rules).

## B4.3.3 Historical Context

- **Classical Mechanics:**  
  Newton and Lagrange exploited symmetry implicitly (e.g., spherical symmetry of gravity → inverse-square law).

- **Noether (1915–1918):**  
  First to **prove** the link between continuous symmetries and conserved quantities.

- **Modern Physics:**  
  - Relativity = built entirely on Lorentz symmetry.  
  - Quantum Field Theory = gauge symmetries.

## B4.4.3 Where Have You Seen Continuous Symmetries Before?

You used them all through **Physics I & II**—even if you didn’t call them "symmetries":

### Mechanics
- Conservation of momentum → translation symmetry.  
- Conservation of angular momentum → rotational symmetry.  
- Work-energy conservation → time translation symmetry.

### Electromagnetism
- Gauss’s law: assumed spherical or cylindrical symmetry to simplify flux integrals.  
- Ampère’s law: rotational symmetry → choose circular loops.  
- EM waves: translational symmetry in uniform media.

## B4.3.5 Key Continuous Symmetries in Physics

---

### Space Translation

#### Definition:
$$
x \longrightarrow x + a, \quad t \longrightarrow t
$$

The parameter $a$ is continuous.

#### Lagrangian Invariance:
For a free particle:


$$
L = \frac12 m \dot{x}^2
$$

Since $L$ depends only on $\dot{x}$ (not on $x$), shifting $x$ by a constant leaves $L$ unchanged.

#### Consequence:
- **Noether’s theorem → conserved momentum:**

$$
p = \frac{\partial L}{\partial \dot{x}} = m\dot{x} = \text{constant}
$$

---

### Time Translation

#### Definition:

$$
t \longrightarrow t + t_0, \quad x \longrightarrow x
$$

#### Example:
For any Lagrangian $L(x, \dot{x})$ that does not depend explicitly on $t$:
- **Noether → conserved energy (Hamiltonian):**

$$
E = \dot{x}\frac{\partial L}{\partial \dot{x}} - L = \text{constant}
$$

---

### Rotational Symmetry

#### Definition (2D example):

$$
\begin{aligned}
x &\longrightarrow x\cos\theta - y\sin\theta \\
y &\longrightarrow x\sin\theta + y\cos\theta
\end{aligned}
$$

Parameter: rotation angle $\theta$.

#### Example – Central Potential:

$$
L = \frac12 m(\dot{r}^2 + r^2\dot{\theta}^2) - V(r)
$$
- $L$ independent of $\theta$ → **angular momentum conserved**:
  
$$
L_z = \frac{\partial L}{\partial \dot{\theta}} = m r^2 \dot{\theta} = \text{constant}
$$

#### Physical Meaning:
No preferred direction in space → rotational symmetry → angular momentum conservation.

---

### Internal (Gauge-Like) Symmetry

- Example: **Phase symmetry** in QM:
  
$$
\psi \to e^{i\alpha}\psi
$$

Noether → **charge conservation**.

This will lead to **U(1), SU(2), SU(3)** groups.

## B4.3.6 Worked Example

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;"> 
<header>
<h3>Example – Time Translation Symmetry</h3>
    </header>
    
**Problem:**  
Check whether the Lagrangian

$$
L = \frac12 m\dot{x}^2 - \frac12 k x^2
$$

is invariant under time translation, and find the conserved quantity.

---

### **Solution:**

1. **Check explicit $t$-dependence:**  
   $L$ depends on $x$ and $\dot{x}$, but not explicitly on $t$ → **time translation symmetry present**.

2. **Noether’s conserved quantity:**  
   Use the general energy expression:

   
$$
E = \dot{x}\frac{\partial L}{\partial \dot{x}} - L
$$

Substitute:

$$
E = \dot{x}(m\dot{x}) - \left(\frac12 m\dot{x}^2 - \frac12 k x^2\right) \\
E = \frac12 m \dot{x}^2 + \frac12 k x^2
$$

→ **Total mechanical energy conserved**.

</div>
