## B4.5.1 Motivation and Background

In the previous module, we learned how **continuous symmetries** of the action give rise to **conserved quantities** via Noether's theorem.

These symmetries are often represented by **Lie groups** — continuous groups parameterized by real variables (e.g., rotations, translations, boosts).

However, **to study the local structure of a Lie group**, especially near the identity, we introduce the concept of a **Lie algebra**.

This algebraic structure captures how infinitesimal transformations behave and combine.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B4.5.2 What Is a Lie Algebra?

A **Lie algebra** is a vector space $\mathfrak{g}$ over $\mathbb{R}$ or $\mathbb{C}$ equipped with a bilinear operation:

$$
[ \cdot , \cdot ] : \mathfrak{g} \times \mathfrak{g} \to \mathfrak{g}
$$

called the **Lie bracket**, satisfying:

1. **Bilinearity:**
 
   $$
   [aX + bY, Z] = a[X, Z] + b[Y, Z] \quad \text{and similarly in second argument}
   $$
   
2. **Antisymmetry:**
 
   $$
   [X, Y] = -[Y, X]
   $$
   
3. **Jacobi identity:**
 
   $$
   [X, [Y, Z]] + [Y, [Z, X]] + [Z, [X, Y]] = 0
   $$

## B4.5.3 Example

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;"> 

**$\mathbb{R}^2$ as a Lie Algebra**

Let’s define a Lie algebra on the vector space $\mathbb{R}^2$ — that is, on ordinary 2D vectors:

$$
\vec{v} = \begin{pmatrix} x \\ y \end{pmatrix}
$$

We define a **Lie bracket** between two vectors $\vec{v}_1 = (x_1, y_1)$ and $\vec{v}_2 = (x_2, y_2)$ as:

$$
[\vec{v}_1, \vec{v}_2] = \begin{pmatrix} 0 \\ x_1 y_2 - x_2 y_1 \end{pmatrix}
$$

This is a **custom rule** — not matrix multiplication, not cross product — but we will check that it satisfies the three Lie algebra properties.

---

### Check 1: Bilinearity of the Lie Bracket

Let $\vec{u} = \begin{pmatrix} u_1 \\ u_2 \end{pmatrix}$, $\vec{v} = \begin{pmatrix} v_1 \\ v_2 \end{pmatrix}$, $\vec{w} = \begin{pmatrix} w_1 \\ w_2 \end{pmatrix}$, and $a, b \in \mathbb{R}$.

We defined the bracket as:

$$
[\vec{x}, \vec{y}] = 
\begin{pmatrix} 
0 \\ x_1 y_2 - x_2 y_1 
\end{pmatrix}
$$

---

#### ➤ First Argument: Linearity in the First Slot

We want to verify:

$$
[a\vec{u} + b\vec{v}, \vec{w}] = a[\vec{u}, \vec{w}] + b[\vec{v}, \vec{w}]
$$

Compute the left-hand side:

$$
\begin{align*}
[a\vec{u} + b\vec{v}, \vec{w}] 
&= \left[
\begin{pmatrix} a u_1 + b v_1 \\ a u_2 + b v_2 \end{pmatrix},
\begin{pmatrix} w_1 \\ w_2 \end{pmatrix}
\right] \\
&= \begin{pmatrix} 
0 \\
(a u_1 + b v_1) w_2 - w_1 (a u_2 + b v_2)
\end{pmatrix} \\
&= \begin{pmatrix}
0 \\
a(u_1 w_2 - u_2 w_1) + b(v_1 w_2 - v_2 w_1)
\end{pmatrix} \\
&= a[\vec{u}, \vec{w}] + b[\vec{v}, \vec{w}]
\end{align*}
$$

So the bracket is linear in the first argument.

---

#### ➤ Second Argument: Linearity in the Second Slot

We want to verify:

$$
[\vec{w}, a\vec{u} + b\vec{v}] = a[\vec{w}, \vec{u}] + b[\vec{w}, \vec{v}]
$$

Compute the left-hand side:

$$
\begin{align*}
[\vec{w}, a\vec{u} + b\vec{v}] 
&= \left[
\begin{pmatrix} w_1 \\ w_2 \end{pmatrix},
\begin{pmatrix} a u_1 + b v_1 \\ a u_2 + b v_2 \end{pmatrix}
\right] \\
&= \begin{pmatrix}
0 \\
w_1 (a u_2 + b v_2) - (a u_1 + b v_1) w_2
\end{pmatrix} \\
&= \begin{pmatrix}
0 \\
a(w_1 u_2 - u_1 w_2) + b(w_1 v_2 - v_1 w_2)
\end{pmatrix} \\
&= a[\vec{w}, \vec{u}] + b[\vec{w}, \vec{v}]
\end{align*}
$$

So the bracket is linear in the second argument as well.

---

#### ✅ Conclusion

The bracket

$$
[\vec{v}_1, \vec{v}_2] = 
\begin{pmatrix} 0 \\ x_1 y_2 - x_2 y_1 \end{pmatrix}
$$

is **bilinear in both arguments**, satisfying the first required property of a Lie bracket.


---

### Check 2: Antisymmetry

Swap the inputs:

$$
[\vec{v}_2, \vec{v}_1] = \begin{pmatrix} 0 \\ x_2 y_1 - x_1 y_2 \end{pmatrix} = -[\vec{v}_1, \vec{v}_2]
$$

So it is antisymmetric ✅.

---

### Check 3: Jacobi Identity

Let $\vec{v}_1 = (x_1, y_1), \vec{v}_2 = (x_2, y_2), \vec{v}_3 = (x_3, y_3)$.

Compute:

$$
[\vec{v}_1, [\vec{v}_2, \vec{v}_3]] + [\vec{v}_2, [\vec{v}_3, \vec{v}_1]] + [\vec{v}_3, [\vec{v}_1, \vec{v}_2]]
$$

Each of these is of the form $\begin{pmatrix} 0 \\ \cdot \end{pmatrix}$, and when you compute the full sum, the second components cancel out.

✅ The Jacobi identity holds!

---

### Interpretation

This is a Lie algebra where:
- The "bracket" operation is just a custom rule on pairs of 2D vectors.
- The result always lies along the $y$-axis.
- The output depends on the **signed area** of the parallelogram formed by the vectors — it's proportional to $x_1 y_2 - x_2 y_1$.

We should understand:
- The Lie bracket is **defined**, not assumed.
- The structure is in the **rules**, not the objects.

## B4.5.4 The Commutator as a Lie Bracket on Operators

Up to now, we have treated the **Lie bracket** $[X,Y]$ as an abstract operation that merely satisfies three axioms.

In physics, however, Lie algebras are most often realized by **linear operators** (or matrices).  
In that setting, the natural Lie bracket is the **commutator**:

$$
[A,B] \equiv AB - BA.
$$

This commutator turns the space of linear operators into a Lie algebra.

---

### Commutator as a Lie Bracket

Let $\mathcal{L}(V)$ be the vector space of all linear maps $A : V \to V$.  
Define the bracket:

$$
[A,B] = AB - BA.
$$

We now verify that this satisfies the three Lie algebra axioms.

---

### 1. Bilinearity

For scalars $a,b$ and operators $A,B,C$:

- **First slot:**

$$
[aA + bB,\, C] = (aA + bB)C - C(aA + bB)
= a(AC - CA) + b(BC - CB)
= a[A,C] + b[B,C].
$$

- **Second slot:**

$$
[A,\, aB + bC] = A(aB + bC) - (aB + bC)A
= a(AB - BA) + b(AC - CA)
= a[A,B] + b[A,C].
$$

Thus the commutator is bilinear.

---

### 2. Antisymmetry

$$
[B,A] = BA - AB = - (AB - BA) = -[A,B].
$$

---

### 3. Jacobi Identity (Idea)

The Jacobi identity requires:

$$
[A,[B,C]] + [B,[C,A]] + [C,[A,B]] = 0.
$$

Expanding each bracket as a difference of products shows that every term cancels pairwise.  
Thus the commutator obeys the Jacobi identity.

Therefore, the commutator is a **valid Lie bracket**.

---

### Why This Matters in Physics

Many Lie algebras you encounter arise from operators:

- **Classical rotations:**  

  $$[J_i, J_j] = J_i J_j - J_j J_i.$$

- **Quantum observables:**  

  $$[\hat{A},\hat{B}] = \hat{A}\hat{B} - \hat{B}\hat{A}.$$

In these settings, the Lie bracket *is* the commutator.

---

### Simple $2\times 2$ Example

Let

$$
A =
\begin{pmatrix}
0 & 1 \\
0 & 0
\end{pmatrix},
\qquad
B =
\begin{pmatrix}
0 & 0 \\
1 & 0
\end{pmatrix}.
$$

Compute:

$$
AB =
\begin{pmatrix}
1 & 0 \\
0 & 0
\end{pmatrix},
\qquad
BA =
\begin{pmatrix}
0 & 0 \\
0 & 1
\end{pmatrix}.
$$

Thus,

$$
[A,B] = AB - BA =
\begin{pmatrix}
1 & 0 \\
0 & -1
\end{pmatrix}.
$$

This shows:

- $A$ and $B$ do **not** commute.
- The commutator $[A,B]$ is another operator in the same space.
- The set of matrices is **closed** under the commutator.

---

### Bridge to $\mathfrak{so}(3)$ and Angular Momentum

In the next section, we apply this to infinitesimal rotation generators $J_x, J_y, J_z$, with:

$$
[J_i, J_j] = \varepsilon_{ijk} J_k.
$$

Here the Lie bracket is **exactly** the matrix commutator. This connects the abstract axioms of Lie algebras with physically meaningful symmetry generators.

## B4.5.5 Physical Example: Rotations in 3D and the Lie Algebra $\mathfrak{so}(3)$


Let’s explore a very important and physically motivated Lie algebra: **rotations in 3D space**.

---

### The Lie Group $SO(3)$

- The group $SO(3)$ consists of all **real $3 \times 3$ orthogonal matrices with determinant 1**:

  $$
  R^T R = I, \quad \det(R) = 1
  $$

- These represent **proper rotations** of vectors in $\mathbb{R}^3$ that preserve length and orientation.

- If $\vec{v}$ is a position vector in 3D, then applying a rotation $R \in SO(3)$ transforms it as:

  $$
  \vec{v} \to R \vec{v}
  $$

---

### Infinitesimal Rotations and the Lie Algebra $\mathfrak{so}(3)$

We now look **very close to the identity** matrix and consider infinitesimal rotations:

$$
R(\boldsymbol{\theta}) \approx I + \varepsilon A
$$

where $\varepsilon \ll 1$ and $A$ is a real $3 \times 3$ matrix.

To stay within $SO(3)$, $R$ must satisfy:

$$
R^T R = I \quad \Rightarrow \quad (I + \varepsilon A)^T (I + \varepsilon A) = I + \varepsilon (A + A^T) + \mathcal{O}(\varepsilon^2)
$$

So we must have:

$$
A^T = -A
$$

That is, $A$ must be **antisymmetric**.

---

### The Lie Algebra $\mathfrak{so}(3)$

So the **Lie algebra $\mathfrak{so}(3)$** consists of all real $3 \times 3$ **antisymmetric** matrices.

A general element looks like:

$$
X = 
\begin{pmatrix}
0 & -\theta_3 & \theta_2 \\
\theta_3 & 0 & -\theta_1 \\
-\theta_2 & \theta_1 & 0
\end{pmatrix}
$$

where the components $\theta_1, \theta_2, \theta_3$ represent the **infinitesimal rotation angles** around the $x$, $y$, and $z$ axes, respectively.

We define the **basis generators** of $\mathfrak{so}(3)$ as:

$$
\begin{align*}
J_x &= 
\begin{pmatrix}
0 & 0 & 0 \\
0 & 0 & -1 \\
0 & 1 & 0
\end{pmatrix}, \quad
J_y = 
\begin{pmatrix}
0 & 0 & 1 \\
0 & 0 & 0 \\
-1 & 0 & 0
\end{pmatrix}, \quad
J_z = 
\begin{pmatrix}
0 & -1 & 0 \\
1 & 0 & 0 \\
0 & 0 & 0
\end{pmatrix}
\end{align*}
$$

These satisfy the commutation relations:

$$
[J_x, J_y] = J_z, \quad [J_y, J_z] = J_x, \quad [J_z, J_x] = J_y
$$

Or more generally:

$$
[J_i, J_j] = \varepsilon_{ijk} J_k
$$

These are the **structure constants** of the algebra $\mathfrak{so}(3)$.

---

### Physical Meaning

- The $J_i$ matrices **generate infinitesimal rotations** about the coordinate axes.
- A finite rotation about axis $\hat{n}$ by angle $\theta$ is:

  $$
  R(\theta \hat{n}) = e^{\theta (n_1 J_x + n_2 J_y + n_3 J_z)}
  $$
- The Lie algebra encodes the **local, infinitesimal behavior**, and the Lie group is built by exponentiating these generators.

---

### From Classical to Quantum

In **quantum mechanics**, the same commutation relations reappear:

$$
[L_i, L_j] = i\hbar \varepsilon_{ijk} L_k
$$

where $L_i$ are angular momentum operators.

This shows that $\mathfrak{so}(3)$ is deeply connected to the quantum theory of angular momentum.

Later, we will see how **$\mathfrak{su}(2)$**, which has the same Lie algebra structure, plays an even more fundamental role due to its richer representation theory (e.g., spin-½ particles).

---

### ✅ Summary

- The group $SO(3)$ describes finite 3D rotations.
- Its Lie algebra $\mathfrak{so}(3)$ consists of **antisymmetric $3 \times 3$ matrices**.
- The generators $J_x, J_y, J_z$ satisfy the Lie bracket:

  $$
  [J_i, J_j] = \varepsilon_{ijk} J_k
  $$
- These same commutation rules govern **angular momentum** in classical and quantum mechanics.

## B4.5.6 General Structure Constants in a Lie Algebra

A **Lie algebra** is not just a vector space with a bracket — it has an internal structure defined by how its elements "interact" via the Lie bracket.

To understand this structure more clearly, we introduce the concept of **structure constants**.

---

### Basis for a Lie Algebra

Suppose a Lie algebra $\mathfrak{g}$ has a **finite dimension** $n$, and let $\{T_1, T_2, \dots, T_n\}$ be a basis for $\mathfrak{g}$.

Any element $X \in \mathfrak{g}$ can be written as a linear combination:

$$
X = \sum_{a=1}^n x^a T_a
$$

To understand the bracket $[X, Y]$, we only need to know the brackets between the **basis elements** $T_a$.

We define:

$$
[T_a, T_b] = \sum_{c=1}^n f_{abc} T_c
$$

The real (or complex) numbers $f_{abc}$ are called the **structure constants** of the Lie algebra.

They encode **everything** about the bracket structure.

---

### Properties of Structure Constants

- **Antisymmetry** in the first two indices:
  $$
  f_{abc} = -f_{bac}
  $$

- **Jacobi identity** imposes algebraic constraints:
  $$
  \sum_d \left( f_{abd} f_{dce} + f_{bcd} f_{dae} + f_{cad} f_{dbe} \right) = 0
  $$

These are not random constants — they are **tightly constrained** by the Lie algebra axioms.

---

### Example: $\mathfrak{so}(3)$

The Lie algebra $\mathfrak{so}(3)$ has generators $J_1, J_2, J_3$ (or $J_x, J_y, J_z$), and the commutators:

$$
[J_i, J_j] = \varepsilon_{ijk} J_k
$$

This means the **structure constants** are:

$$
f_{ijk} = \varepsilon_{ijk}
$$

The Levi-Civita symbol $\varepsilon_{ijk}$ is totally antisymmetric and defined by:

- $\varepsilon_{123} = 1$
- $\varepsilon_{ijk} = -\varepsilon_{jik}$
- $\varepsilon_{ijk} = 0$ if any two indices are equal

Thus, for example:

- $[J_1, J_2] = J_3 \Rightarrow f_{123} = 1$
- $[J_2, J_1] = -J_3 \Rightarrow f_{213} = -1$
- $[J_1, J_1] = 0 \Rightarrow f_{11c} = 0$

The **entire structure of $\mathfrak{so}(3)$** is captured by these constants.

---

### Physical Meaning

In physics:

- The **generators** $T_a$ represent **infinitesimal symmetries** (e.g., rotations, translations).
- The **structure constants** tell you how these generators **combine and interact**.

In quantum mechanics, the structure constants determine how observables like angular momentum or isospin obey specific commutation relations.

---

### Summary

- A Lie algebra $\mathfrak{g}$ has a basis $\{T_a\}$ and structure constants $f_{abc}$ defined by:
  $$
  [T_a, T_b] = f_{abc} T_c
  $$

- These constants obey:
  - Antisymmetry: $f_{abc} = -f_{bac}$
  - Jacobi identity constraints

- In $\mathfrak{so}(3)$, the structure constants are $\varepsilon_{ijk}$, encoding the geometry of 3D space.

## B4.5.7 Classical Mechanics Application: Poisson Brackets

In classical mechanics, the Poisson bracket plays the role of the Lie bracket.

For example, angular momentum components $L_x, L_y, L_z$ satisfy:

$$
\{L_x, L_y\} = L_z \quad \text{(cyclic)}
$$

The **Poisson bracket algebra** of observables can be a **classical Lie algebra**, with the bracket $\{f, g\}$ instead of the commutator.
