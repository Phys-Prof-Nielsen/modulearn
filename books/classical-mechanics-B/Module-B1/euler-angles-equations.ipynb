## B1.6.3 Why Use Euler Angles?

Euler angles $(\phi, \theta, \psi)$ are often used as **generalized coordinates** in **Lagrangian mechanics** for describing rotational motion. They provide a **minimal and intuitive way** to represent the orientation of a rigid body, especially when using:

- **Configuration space modeling**
- **Kinetic energy** in rotational systems
- **Symbolic dynamics** and **analytical mechanics**

They allow us to compute the **angular velocity vector $\vec{\omega}$** as a function of time derivatives:

$$
\vec{\omega} = f(\phi, \theta, \psi, \dot{\phi}, \dot{\theta}, \dot{\psi})
$$

That angular velocity vector can then be used in:

- Computing **kinetic energy**:  
  $$
  T = \frac{1}{2} \vec{\omega}^T \mathbf{I} \vec{\omega}
  $$
- Deriving the **equations of motion** via Euler–Lagrange equations
- Applying **Euler’s equations** directly in the body frame

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.6.4 Key Connection: Angular Velocity

To connect Euler angles to Euler’s equations, you must express the **angular velocity vector $\vec{\omega}$** — which appears in Euler’s equations — **in terms of the Euler angles and their time derivatives**.

In the **ZXZ convention**, the angular velocity in the **body-fixed frame** is:

$$
\vec{\omega}_{\text{body}} =
\begin{bmatrix}
\dot{\theta} \cos \psi + \dot{\phi} \sin \theta \sin \psi \\
\dot{\theta} \sin \psi - \dot{\phi} \sin \theta \cos \psi \\
\dot{\psi} + \dot{\phi} \cos \theta
\end{bmatrix}
$$

So the angular velocity components used in Euler’s equations are **functions of the Euler angles and their derivatives**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B1.6.5 Euler’s Equations in Principal Axes

In the body frame, Euler’s equations for a rigid body with diagonal inertia tensor $\mathbf{I} = \text{diag}(I_1, I_2, I_3)$ are:

$$
\begin{aligned}
I_1 \dot{\omega}_1 &= N_1 + (I_2 - I_3) \omega_2 \omega_3 \\
I_2 \dot{\omega}_2 &= N_2 + (I_3 - I_1) \omega_3 \omega_1 \\
I_3 \dot{\omega}_3 &= N_3 + (I_1 - I_2) \omega_1 \omega_2
\end{aligned}
$$

If there’s no external torque $(\vec{N} = 0)$, the equations simplify to describe **free rigid body rotation**.

## B1.6.6 Worked Example: Euler Angles and Euler’s Equations in a Symmetric Top

### Problem Setup

A rigid body has:

- Principal moments of inertia:  
  $I_1 = I_2 = I$, $I_3 \neq I$
- Euler angles (ZXZ convention): $\phi(t), \theta(t), \psi(t)$
- No external torque: $\vec{N} = 0$

### Objective

- Compute angular velocity $\vec{\omega}$ from Euler angles  
- Use Euler’s equations to describe the motion  
- Analyze precession and spin

---

### Step 1: Angular Velocity from Euler Angles

$$
\begin{aligned}
\omega_1 &= \dot{\theta} \cos \psi + \dot{\phi} \sin \theta \sin \psi \\
\omega_2 &= \dot{\theta} \sin \psi - \dot{\phi} \sin \theta \cos \psi \\
\omega_3 &= \dot{\psi} + \dot{\phi} \cos \theta
\end{aligned}
$$

---

### Step 2: Apply Euler’s Equations

With $I_1 = I_2 = I$, Euler’s equations simplify:

$$
\begin{aligned}
I \dot{\omega}_1 &= (I - I_3) \omega_2 \omega_3 \\
I \dot{\omega}_2 &= (I_3 - I) \omega_1 \omega_3 \\
I_3 \dot{\omega}_3 &= 0
\end{aligned}
$$

From the last equation:  
✅ $\omega_3 = \text{constant}$

---

### Step 3: Example with Constant $\theta$

Let:

- $\theta = \frac{\pi}{4}$ (constant)
- $\phi(t) = \omega_p t$, $\psi(t) = \omega_s t$

Then:

$$
\begin{aligned}
\dot{\phi} &= \omega_p, \quad \dot{\psi} = \omega_s \\
\omega_1 &= \omega_p \sin \theta \sin (\omega_s t) \\
\omega_2 &= -\omega_p \sin \theta \cos (\omega_s t) \\
\omega_3 &= \omega_s + \omega_p \cos \theta
\end{aligned}
$$

- $\omega_3 = \text{constant}$
- $\omega_1, \omega_2$ oscillate sinusoidally
- This is **regular precession** with constant spin

## B1.6.8 Final Insight

Euler angles are best thought of as **generalized coordinates** for describing orientation.  
Euler’s equations describe the **dynamics** — how the angular velocity vector evolves over time.

They connect via the angular velocity:

$$
\boxed{
\text{Euler angles } (\phi, \theta, \psi)
\ \xrightarrow{\,\text{time derivative}\,}\
\vec{\omega}
\ \xrightarrow{\,\text{Euler’s equations}\,}\
\dot{\vec{\omega}}
}
$$

This full chain allows you to describe and simulate rotational motion from configuration to dynamics — and back.
