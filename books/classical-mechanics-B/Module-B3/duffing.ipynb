## B3.2.1 Motivation

Real springs obey Hooke's law only approximately. For sufficiently small displacements, the restoring force is proportional to the displacement:

$$
F=-kx,
$$

which leads to simple harmonic motion with a frequency independent of amplitude.

However, as the oscillation amplitude becomes larger, real systems often deviate from this ideal behavior. The restoring force may become stronger or weaker than predicted by Hooke's law, producing nonlinear effects that cannot be explained using the theory of simple harmonic motion.

Examples of such nonlinear oscillators include:

- large-amplitude pendulums,
- stretched strings,
- vibrating beams,
- MEMS devices,
- nanomechanical resonators,
- Josephson junctions,
- and many electronic circuits.

One of the simplest and most important models describing these effects is the **Duffing oscillator**, whose restoring force contains an additional cubic term.

This system exhibits several phenomena that are impossible in linear oscillators:

- amplitude-dependent frequencies,
- hardening and softening behavior,
- multiple equilibrium positions,
- bistability,
- hysteresis,
- and, under sufficiently strong driving, chaotic motion.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Curiosity</strong>

The Duffing oscillator was introduced by the German engineer Georg Duffing in 1918 while studying vibrating mechanical systems. More than a century later, the same equation appears in subjects ranging from structural engineering and microelectromechanical systems (MEMS) to quantum circuits and nonlinear optics.

</div>

The Duffing oscillator serves as one of the simplest examples of how nonlinear systems can produce rich and sometimes surprising behavior. It provides an important stepping stone between the familiar world of simple harmonic motion and the onset of chaos that we will encounter later in this module.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.2.2 The Duffing Equation

The Duffing oscillator is obtained by extending Hooke's law to include a nonlinear restoring force. Instead of

$$
F=-kx,
$$

we assume

$$
F=-kx-\beta x^3,
$$

where $\beta$ is a constant that determines the strength of the nonlinearity.

Applying Newton's second law gives

$$
m\ddot{x}+kx+\beta x^3=0.
$$

More generally, the Duffing oscillator may include damping and an external driving force:

$$
m\ddot{x}
+b\dot{x}
+kx
+\beta x^3
=
F_0\cos(\omega t).
$$

Dividing by the mass, this equation is often written in the form

$$
\ddot{x}
+\delta\dot{x}
+\alpha x
+\beta x^3
=
\gamma\cos(\omega t),
$$

where

- $\delta$ represents damping,
- $\alpha$ determines the linear restoring force,
- $\beta$ determines the nonlinear restoring force,
- $\gamma$ is the driving amplitude,
- and $\omega$ is the driving frequency.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Duffing Oscillator</strong>

The Duffing oscillator is a nonlinear oscillator whose restoring force contains both a linear term and a cubic term.

Its equation of motion is

$$
\ddot{x}
+\delta\dot{x}
+\alpha x
+\beta x^3
=
\gamma\cos(\omega t).
$$

</div>

Notice that the familiar damped-driven harmonic oscillator is recovered when

$$
\beta=0,
$$

giving

$$
\ddot{x}
+\delta\dot{x}
+\alpha x
=
\gamma\cos(\omega t).
$$

Thus, the Duffing oscillator may be viewed as a natural extension of the harmonic oscillator that incorporates nonlinear restoring forces.

<div style="background-color:#ffe6e6; border-left:6px solid #cc0000; padding:14px; border-radius:4px;">
<strong>Danger:</strong><br>

Because the equation contains the nonlinear term

$$
\beta x^3,
$$

the principle of superposition no longer applies. If two functions are solutions individually, their sum is generally <em>not</em> a solution.

</div>

In the next section, we will investigate how the sign of $\beta$ affects the potential energy and gives rise to single-well and double-well oscillators.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.2.3 Potential Energy and Equilibrium Points

To better understand the behavior of the Duffing oscillator, it is useful to consider its potential energy.

For the moment, we consider the undamped and unforced Duffing equation

$$
m\ddot{x}
+
\alpha x
+
\beta x^3
=
0,
$$

whose restoring force is

$$
F
=
-\alpha x
-
\beta x^3.
$$

Since this force depends only on position, it is conservative and can therefore be written in terms of a potential energy function:

$$
F
=
-\frac{dU}{dx}.
$$

Integrating gives

$$
U(x)
=
\frac12\alpha x^2
+
\frac14\beta x^4.
$$

Thus, the qualitative behavior of the Duffing oscillator is determined by the shape of its potential energy.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Potential Energy of the Duffing Oscillator</strong>

For the conservative Duffing oscillator,

$$
F=-\alpha x-\beta x^3,
$$

the corresponding potential energy is

$$
U(x)
=
\frac12\alpha x^2
+
\frac14\beta x^4.
$$

Equilibrium points occur wherever

$$
\frac{dU}{dx}=0,
$$

or equivalently where the net force vanishes.

</div>

---

When

$$
\alpha>0
\qquad\text{and}\qquad
\beta>0,
$$

the potential energy has a single minimum at

$$
x=0.
$$

The oscillator behaves similarly to a simple harmonic oscillator, although its frequency becomes amplitude dependent.

---

More interesting behavior occurs when

$$
\alpha<0
\qquad\text{and}\qquad
\beta>0.
$$

The potential energy remains

$$
U(x)
=
\frac12\alpha x^2
+
\frac14\beta x^4,
$$

but because the quadratic coefficient is negative, the origin becomes a local maximum and two minima emerge, producing the characteristic **double-well potential**.

The equilibrium points are found from

$$
\frac{dU}{dx}=0,
$$

which gives

$$
\alpha x+\beta x^3=0.
$$

Factoring out $x$,

$$
x\left(\alpha+\beta x^2\right)=0,
$$

yielding three equilibrium points:

$$
x=0,
$$

and

$$
x=\pm\sqrt{-\frac{\alpha}{\beta}}.
$$

The equilibrium point

$$
x=0
$$

corresponds to a local maximum of the potential energy and is therefore unstable.

The two outer equilibrium points

$$
x=\pm\sqrt{-\frac{\alpha}{\beta}}
$$

correspond to minima of the potential energy and are stable.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Physical Meaning of $\alpha$</strong>

Unlike the spring constant in Hooke's law,

$$
F=-kx,
\qquad k>0,
$$

the parameter

$$
\alpha
$$

is simply a coefficient that determines the curvature of the potential energy near the origin.

When

$$
\alpha>0,
$$

the origin is a stable equilibrium point.

When

$$
\alpha<0,
$$

the origin becomes unstable and two new stable equilibrium positions emerge.

Thus, a negative value of

$$
\alpha
$$

does not imply a negative spring constant. Instead, it indicates that the system possesses a double-well potential.

</div>

---

<div style="background-color:#ffe6e6; border-left:6px solid #cc0000; padding:14px; border-radius:4px;">
<strong>What Does k<0 Mean?</strong><br>

For an ordinary spring obeying Hooke's law,

$$
F=-kx,
$$

the spring constant satisfies

$$
k>0,
$$

by definition. A literal spring with

$$
k<0
$$

would be unstable and therefore does not represent a physically realizable spring.

However, many systems that are <em>not</em> springs can behave <em>as if</em> they possess an effective negative spring constant near certain equilibrium configurations. In such cases, the restoring force may be approximated by

$$
F\approx -kx,
$$

with

$$
k<0,
$$

indicating that the equilibrium point is unstable rather than stable.

Examples include:

- **Buckled beams and compressed columns:** after buckling, the straight configuration becomes unstable and two new stable configurations appear.

- **Magnetic systems near phase transitions:** the symmetric state may become unstable, producing two preferred orientations.

- **Molecular inversion potentials:** molecules such as ammonia possess double-well potentials with two equivalent configurations.

- **Superconducting circuits and Josephson junctions:** effective double-well potentials arise naturally and are used in quantum computing.

- **Landau theory of phase transitions:** the coefficient of the quadratic term may change sign as temperature changes, signaling the onset of spontaneous symmetry breaking.

Thus, a negative coefficient in the Duffing potential should not be interpreted as a negative spring constant of a literal spring. Rather, it indicates that the system has undergone a change in stability, causing the original equilibrium point to become unstable and new stable equilibria to emerge.

</div>

---

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Double-Well Potential</h3>

Suppose

$$
\alpha=-1
\qquad\text{and}\qquad
\beta=1.
$$

Then

$$
U(x)
=
-\frac12x^2
+
\frac14x^4.
$$

The equilibrium points are

$$
x=0,
\qquad
x=\pm1.
$$

The point

$$
x=0
$$

is unstable, while

$$
x=\pm1
$$

are stable equilibrium positions.

A particle with sufficiently low energy will oscillate within one of the wells. If its energy becomes large enough, it may cross the central barrier and move between the two wells.

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Duffing potential energy
# U(x) = (1/2) alpha x^2 + (1/4) beta x^4
# ------------------------------------------------------------

def U(x, alpha, beta):
    return 0.5*alpha*x**2 + 0.25*beta*x**4

# Position range
x = np.linspace(-2.0, 2.0, 1000)

# Parameters for double-well case
alpha = -1.0
beta = 1.0

# Potential energy
Ux = U(x, alpha, beta)

# Equilibrium points
x_unstable = 0.0
x_stable_1 = -np.sqrt(-alpha/beta)
x_stable_2 = np.sqrt(-alpha/beta)

U_unstable = U(x_unstable, alpha, beta)
U_stable_1 = U(x_stable_1, alpha, beta)
U_stable_2 = U(x_stable_2, alpha, beta)

# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

plt.plot(x, Ux, label=r"$U(x)=\frac{1}{2}\alpha x^2+\frac{1}{4}\beta x^4$")

# Mark equilibrium points
plt.plot(x_unstable, U_unstable, "o", markersize=8, label="Unstable equilibrium")
plt.plot([x_stable_1, x_stable_2],
         [U_stable_1, U_stable_2],
         "o",
         markersize=8,
         label="Stable equilibria")

# Labels
plt.annotate("unstable",
             xy=(x_unstable, U_unstable),
             xytext=(0.15, U_unstable + 0.25),
             arrowprops=dict(arrowstyle="->"))

plt.annotate("stable",
             xy=(x_stable_1, U_stable_1),
             xytext=(-1.75, U_stable_1 + 0.35),
             arrowprops=dict(arrowstyle="->"))

plt.annotate("stable",
             xy=(x_stable_2, U_stable_2),
             xytext=(1.25, U_stable_2 + 0.35),
             arrowprops=dict(arrowstyle="->"))

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.xlabel(r"Position $x$")
plt.ylabel(r"Potential energy $U(x)$")
plt.title(r"Duffing Double-Well Potential: $\alpha=-1$, $\beta=1$")
plt.legend(loc="upper right")
plt.grid(True)

plt.show()
```

## B3.2.4 Hardening and Softening Springs

One of the most important consequences of nonlinear restoring forces is that the oscillation frequency is no longer constant. Instead, the frequency may depend on the amplitude of the motion.

For the conservative Duffing oscillator,

$$
m\ddot{x}+kx+\beta x^3=0,
$$

the cubic term becomes increasingly important as the amplitude grows.

Depending on the sign of

$$
\beta,
$$

the oscillator may become effectively stiffer or softer.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Hardening and Softening Springs</strong>

The sign of the nonlinear coefficient

$$
\beta
$$

determines how the oscillation frequency changes with amplitude.

- If

  $$
  \beta>0,
  $$

  the oscillator becomes progressively stiffer and the frequency increases with amplitude.

- If

  $$
  \beta<0,
  $$

  the oscillator becomes progressively softer and the frequency decreases with amplitude.

</div>

---

When

$$
\beta>0,
$$

the restoring force

$$
F=-kx-\beta x^3
$$

grows faster than Hooke's law predicts.

Consequently, the system behaves like an increasingly stiff spring for large displacements.

This type of behavior is called a **hardening spring**.

---

When

$$
\beta<0,
$$

the cubic term partially opposes the linear restoring force,

$$
F=-kx+|\beta|x^3,
$$

causing the system to become less stiff as the amplitude increases.

Such systems are known as **softening springs**.

---

Unlike simple harmonic motion, where

$$
\omega_0=\sqrt{\frac{k}{m}}
$$

is independent of amplitude, the Duffing oscillator exhibits amplitude-dependent frequencies:

- larger amplitudes correspond to higher frequencies for hardening systems,
- larger amplitudes correspond to lower frequencies for softening systems.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Hardening Spring</h3>

Suppose

$$
k=1
\qquad\text{and}\qquad
\beta=0.5.
$$

For small oscillations, the cubic term is negligible and the motion resembles simple harmonic motion.

As the amplitude increases, however, the term

$$
\beta x^3
$$

becomes increasingly important, making the restoring force stronger than predicted by Hooke's law.

Consequently, larger-amplitude oscillations occur at higher frequencies.

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Duffing oscillator: amplitude-dependent frequency
# ------------------------------------------------------------

# Parameters
m = 1.0
k = 1.0
beta = 1.0

# Time settings
dt = 0.001
t_max = 30
t = np.arange(0, t_max, dt)

# Initial amplitudes to compare
x0_values = [0.2, 1.0, 2.0]
v0 = 0.0

plt.figure(figsize=(10, 5))

for x0 in x0_values:

    x = np.zeros(len(t))
    v = np.zeros(len(t))

    x[0] = x0
    v[0] = v0

    # Euler-Cromer integration
    for i in range(len(t)-1):

        # m x'' + kx + beta x^3 = 0
        a = -(k*x[i] + beta*x[i]**3)/m

        v[i+1] = v[i] + a*dt
        x[i+1] = x[i] + v[i+1]*dt

    plt.plot(t, x, label=rf"$x_0={x0}$")

plt.xlabel("Time")
plt.ylabel("Position $x(t)$")
plt.title("Amplitude-Dependent Frequency in the Duffing Oscillator")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

## B3.2.5 Phase Space and Phase Portraits

In University Physics I, we described oscillatory motion by plotting the displacement as a function of time. While such plots are useful, they often hide important information about the underlying dynamics.

A more powerful approach, as briefly mentioned in Phase D, is to examine the motion in **phase space**, where each point represents the instantaneous state of the system.

For a one-dimensional oscillator, the phase-space coordinates are

$$
(x,\dot{x}),
$$

where

- $x$ is the displacement,
- $\dot{x}$ is the velocity.

As the oscillator evolves, the point

$$
(x,\dot{x})
$$

moves through phase space, tracing out a curve known as a **phase portrait**.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Phase Portrait</strong>

A phase portrait is a graph of velocity versus displacement,

$$
(\dot{x}\text{ versus }x),
$$

that provides a geometric representation of the motion of a dynamical system.

</div>

---

For the simple harmonic oscillator,

$$
m\ddot{x}+kx=0,
$$

the total energy

$$
E
=
\frac12m\dot{x}^2
+
\frac12kx^2
$$

is constant.

Consequently, the phase portrait consists of closed ellipses centered on the origin.

Every initial condition corresponds to a different ellipse, but all trajectories are periodic.

---

For the Duffing oscillator,

$$
m\ddot{x}
+
kx
+
\beta x^3
=
0,
$$

the phase portraits are still closed curves because the system is conservative.

However, unlike the harmonic oscillator, the trajectories are generally no longer perfect ellipses.

Their shapes depend on:

- the oscillation amplitude,
- the sign of

  $$
  \beta,
  $$

- and the total energy of the system.

---

When

$$
\beta>0,
$$

the phase curves become increasingly distorted as the amplitude grows.

Large-amplitude trajectories differ significantly from the ellipses of simple harmonic motion, reflecting the amplitude dependence of the oscillation frequency.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Comparing SHM and the Duffing Oscillator</h3>

For a harmonic oscillator, the phase portrait consists of concentric ellipses.

For a Duffing oscillator with

$$
\beta>0,
$$

the phase curves become progressively distorted as the amplitude increases.

Thus, phase space immediately reveals that the motion is nonlinear, even though the displacement-versus-time plots may still appear nearly sinusoidal.

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Phase portraits: SHM vs Duffing oscillator
# ------------------------------------------------------------

m = 1.0
k = 1.0
beta = 1.0

dt = 0.01
t_max = 30
t = np.arange(0, t_max, dt)

x0_values = [0.5, 1.0, 2.0]
v0 = 0.0

plt.figure(figsize=(8, 6))

for x0 in x0_values:

    # SHM arrays
    x_shm = np.zeros(len(t))
    v_shm = np.zeros(len(t))

    # Duffing arrays
    x_duff = np.zeros(len(t))
    v_duff = np.zeros(len(t))

    x_shm[0] = x0
    x_duff[0] = x0
    v_shm[0] = v0
    v_duff[0] = v0

    for i in range(len(t)-1):

        # SHM: m x'' + kx = 0
        a_shm = -(k*x_shm[i])/m
        v_shm[i+1] = v_shm[i] + a_shm*dt
        x_shm[i+1] = x_shm[i] + v_shm[i+1]*dt

        # Duffing: m x'' + kx + beta x^3 = 0
        a_duff = -(k*x_duff[i] + beta*x_duff[i]**3)/m
        v_duff[i+1] = v_duff[i] + a_duff*dt
        x_duff[i+1] = x_duff[i] + v_duff[i+1]*dt

    plt.plot(x_shm, v_shm, label=rf"SHM, $x_0={x0}$")
    plt.plot(x_duff, v_duff, label=rf"Duffing, $x_0={x0}$")

plt.xlabel(r"Position $x$")
plt.ylabel(r"Velocity $\dot{x}$")
plt.title("Phase Portraits: SHM vs Duffing Oscillator")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

## B3.2.6 The Driven Duffing Oscillator

So far, we have considered the conservative Duffing oscillator, in which energy is conserved and the motion is confined to closed curves in phase space.

Real systems, however, are often subject to both damping and external forcing. Including these effects leads to the **driven Duffing oscillator**,

$$
m\ddot{x}
+
b\dot{x}
+
kx
+
\beta x^3
=
F_0\cos(\omega t),
$$

or, after dividing by the mass,

$$
\ddot{x}
+
\delta\dot{x}
+
\alpha x
+
\beta x^3
=
\gamma\cos(\omega t).
$$

This equation combines several competing effects:

- the restoring force,
- nonlinear corrections,
- energy dissipation through damping,
- and continuous energy input from the driving force.

The resulting behavior can be extraordinarily rich.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Driven Duffing Oscillator</strong>

The driven Duffing oscillator is governed by

$$
\ddot{x}
+
\delta\dot{x}
+
\alpha x
+
\beta x^3
=
\gamma\cos(\omega t).
$$

The competition between damping, driving, and nonlinear restoring forces gives rise to phenomena such as

- amplitude-dependent resonance,
- multiple steady states,
- hysteresis,
- period doubling,
- and chaos.

</div>

---

Unlike the simple harmonic oscillator, the response of the Duffing oscillator is not uniquely determined by the driving frequency.

For certain parameter values, the system may possess several possible steady-state oscillations.

Consequently, two oscillators with identical parameters but different initial conditions may settle into entirely different motions.

>This phenomenon is known as **bistability**.

---

Because of the nonlinear restoring force, the resonance frequency depends on amplitude.

As the driving frequency is slowly varied, the oscillation amplitude may change smoothly for a while and then suddenly jump to a completely different value.

>This phenomenon is called **jump resonance**.

---

If the driving frequency is subsequently swept in the opposite direction, the transition occurs at a different frequency.

As a result, the system follows a different path than before.

>This memory effect is known as **hysteresis**.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Jump Resonance</h3>

Suppose the driving frequency of a hardening Duffing oscillator is gradually increased.

Initially, the oscillation amplitude increases smoothly.

However, at a critical frequency the system suddenly jumps to a much smaller amplitude branch.

If the frequency is then decreased, the reverse transition occurs at a different frequency.

Thus, the amplitude-versus-frequency curve forms a loop rather than a single-valued function.

</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Curiosity</strong>

Hysteresis appears throughout physics.

Examples include

- ferromagnets,
- superconductors,
- structural instabilities,
- lasers,
- and phase transitions.

In all these systems, the current state depends not only on the present conditions but also on the history of the system.

</div>

As the driving strength increases further, the motion may become increasingly complicated. Period doubling, irregular oscillations, and eventually chaotic motion may appear.

Thus, the driven Duffing oscillator serves as one of the simplest systems capable of exhibiting deterministic chaos.

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Bistability in the driven Duffing oscillator
# ------------------------------------------------------------

# Duffing parameters
alpha = -1.0
beta = 1.0
delta = 0.20
gamma = 0.30
omega_drive = 1.20

# Time settings
dt = 0.01
t_max = 300
t = np.arange(0, t_max, dt)

# Two different initial conditions
initial_conditions = [
    (-1.0, 0.0, "Initial condition near left well"),
    ( 1.0, 0.0, "Initial condition near right well")
]

plt.figure(figsize=(10, 5))

for x0, v0, label in initial_conditions:

    x = np.zeros(len(t))
    v = np.zeros(len(t))

    x[0] = x0
    v[0] = v0

    for i in range(len(t)-1):

        # x'' + delta x' + alpha x + beta x^3 = gamma cos(omega t)
        a = (
            gamma*np.cos(omega_drive*t[i])
            - delta*v[i]
            - alpha*x[i]
            - beta*x[i]**3
        )

        v[i+1] = v[i] + a*dt
        x[i+1] = x[i] + v[i+1]*dt

    plt.plot(t, x, label=label)

plt.xlabel("Time")
plt.ylabel("Position $x(t)$")
plt.title("Bistability in the Driven Duffing Oscillator")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Phase-space view of bistability
# ------------------------------------------------------------

# Duffing parameters
alpha = -1.0
beta = 1.0
delta = 0.20
gamma = 0.30
omega_drive = 1.20

# Time settings
dt = 0.01
t_max = 300
t = np.arange(0, t_max, dt)

# Ignore early transient behavior
transient_time = 100
transient_index = int(transient_time/dt)

# Two different initial conditions
initial_conditions = [
    (-1.0, 0.0, "Attractor from left well"),
    ( 1.0, 0.0, "Attractor from right well")
]

plt.figure(figsize=(7, 6))

for x0, v0, label in initial_conditions:

    x = np.zeros(len(t))
    v = np.zeros(len(t))

    x[0] = x0
    v[0] = v0

    for i in range(len(t)-1):

        # x'' + delta x' + alpha x + beta x^3 = gamma cos(omega t)
        a = (
            gamma*np.cos(omega_drive*t[i])
            - delta*v[i]
            - alpha*x[i]
            - beta*x[i]**3
        )

        v[i+1] = v[i] + a*dt
        x[i+1] = x[i] + v[i+1]*dt

    plt.plot(
        x[transient_index:],
        v[transient_index:],
        label=label
    )

plt.xlabel("Position $x$")
plt.ylabel("Velocity $\dot{x}$")
plt.title("Phase Portrait Showing Bistability")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

## Interpretation

The phase-space plot reveals an important feature of nonlinear systems: **the long-term behavior may depend on the initial conditions**.

In the present example, the two trajectories begin in different wells of the potential and eventually settle into different steady motions, even though the governing equation and all parameter values are identical.

Thus, the system possesses two distinct stable long-term states.

This phenomenon is known as **bistability**.

In contrast, a damped simple harmonic oscillator has only one stable equilibrium point. Regardless of the initial conditions, every trajectory eventually approaches the same final state.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Attractors</strong>

An <strong>attractor</strong> is a state or set of states toward which a system naturally evolves after sufficient time has passed.

Different initial conditions may lead to different attractors.

Examples include:

<ul>
<li>a stable equilibrium point,</li>
<li>a periodic orbit (limit cycle),</li>
<li>quasi-periodic motion,</li>
<li>or even a chaotic attractor.</li>
</ul>

Once transients have died away, the system remains close to the attractor unless it is subjected to a sufficiently large disturbance.

</div>

In the driven Duffing oscillator, the phase-space trajectories do not wander randomly. Instead, they approach one of two stable attractors.

Consequently, two systems having identical parameters may exhibit different long-term motions simply because they started from different initial conditions.

This sensitivity to initial conditions and the existence of multiple attractors are hallmarks of nonlinear dynamics and have no counterpart in the simple harmonic oscillator.

## B3.2.7 Applications of the Duffing Oscillator

Although the Duffing oscillator was originally introduced as a model for vibrating mechanical systems, its applications extend throughout modern science and engineering.

Whenever restoring forces become nonlinear, the Duffing equation often provides an excellent approximation to the motion.

---

### Mechanical Structures

Large-amplitude vibrations of beams, plates, and suspension systems often exhibit nonlinear behavior.

Examples include:

- bridges,
- aircraft wings,
- skyscrapers,
- and mechanical resonators.

Understanding these nonlinear oscillations is essential for avoiding structural instabilities and mechanical failures.

---

### MEMS and Nanomechanical Devices

Microelectromechanical systems (MEMS) and nanomechanical resonators frequently operate in regimes where the restoring force is nonlinear.

Examples include:

- accelerometers,
- gyroscopes,
- pressure sensors,
- and resonators used in communication devices.

In these systems, Duffing nonlinearities can shift resonance frequencies and produce hysteresis.

---

### Nonlinear Electrical Circuits

Electronic circuits containing nonlinear components may behave like Duffing oscillators.

Examples include:

- nonlinear RLC circuits,
- Josephson junctions,
- and superconducting devices.

Such systems may exhibit multiple stable states, making them useful for memory devices and quantum technologies.

---

### Optical Systems and Lasers

Nonlinear optics provides another important application.

Duffing-type equations arise in:

- laser cavities,
- optical resonators,
- and nonlinear waveguides.

These systems may display bistability, synchronization, and chaotic behavior.

---

### Plasma Physics

Nonlinear oscillations are common in plasma systems.

Examples include:

- charged particles trapped in electromagnetic fields,
- nonlinear plasma waves,
- and instabilities in fusion devices.

Understanding these oscillations is crucial for controlled nuclear fusion and space plasma physics.

---

### Molecular and Condensed Matter Physics

Duffing-like potentials appear in:

- molecular vibrations,
- superconductors,
- magnetic materials,
- and phase transitions.

Double-well potentials are particularly important in describing spontaneous symmetry breaking and quantum tunneling phenomena.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — A Guitar String</h3>

A vibrating guitar string behaves approximately like a simple harmonic oscillator for small amplitudes.

However, at larger amplitudes the tension in the string increases slightly, causing the restoring force to become nonlinear.

As a result, the oscillation frequency increases with amplitude, producing hardening behavior that is well described by the Duffing oscillator.

</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Curiosity</strong>

The Duffing oscillator is often called the "hydrogen atom of nonlinear dynamics."

Just as the hydrogen atom provides one of the simplest systems for understanding quantum mechanics, the Duffing oscillator provides one of the simplest systems for exploring nonlinear phenomena, including bistability, hysteresis, and chaos.

</div>

Because of its simplicity and extraordinary richness, the Duffing oscillator has become one of the most important models in nonlinear dynamics.
