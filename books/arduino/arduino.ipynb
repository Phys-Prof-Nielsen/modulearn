# Arduino Quickstart Manual (New Users)

---
 
**Purpose:** This guide introduces Arduino for first-time users: what it is, why it is useful, and how to use it. You will learn how to install the Arduino IDE, connect and select the board (Arduino IDE 2), verify and upload code, use the **Serial Monitor** and **Serial Plotter**, and run a test program that demonstrates all of these features.

---

## 1. What Is an Arduino?

An **Arduino** is a small, programmable microcontroller board. It can:

- read electrical inputs (voltages, button states, sensors),
- perform calculations and timing,
- control outputs (LEDs, motors, relays),
- communicate with a computer using a USB connection.

In physics laboratories, Arduino is useful because it can function as:
- a **data acquisition device**,
- a **timing system**,
- a **controller** for repeatable experiments,
- a bridge between abstract physics concepts and real instrumentation.

---

## 2. Why We Use Arduino in This Course

Arduino allows you to:
- automate measurements,
- collect large datasets reliably,
- directly interface with real electrical systems,
- export data for analysis in Python/Jupyter.

Later labs will use Arduino to measure voltages, control circuits, and stream data for statistical analysis.

---

## 3. Installing the Arduino IDE (If Needed)

1. Download **Arduino IDE 2** from the official Arduino website.
2. Install it like any standard application.
3. Launch the Arduino IDE.

If your operating system asks for permissions (especially on macOS), allow access to USB devices.

---

## 4. Connecting the Arduino (Do This First)

**Plug the Arduino into your computer before selecting the board.**

1. Connect the Arduino to your computer using a USB data cable.
2. A small power LED on the board should turn on.

If no lights appear:
- try a different USB cable,
- try a different USB port,
- ask for help before proceeding.

---

## 5. Selecting the Board and Port (Arduino IDE 2)

Arduino IDE 2 combines **Board** and **Port** into a single selector.

1. Look at the **top center dropdown menu** of the Arduino IDE window.
2. Click the label (it may say *Select Board*).
3. Choose your device, for example:
   - `Arduino Nano on COM3` (Windows)
   - `Arduino Nano on /dev/cu.usbmodem...` (macOS)

Once selected, the IDE remembers this choice.

---

## 6. Verifying the Code

Before uploading code, always verify it.

- Click the **Verify** button (✓).
- The code will compile.
- If there are errors, they appear in red text.

Do not upload until the code verifies successfully.

---

## 7. Uploading the Code

Uploading sends your program (“sketch”) to the Arduino.

- Click the **Upload** button (→).
- Wait for **“Done uploading”**.
- The Arduino will reset and begin running the program.

---

## 8. Serial Communication: Monitor vs Plotter

Arduino can send data to your computer using **Serial communication**.

### Serial Monitor
- Displays text and numbers sent from Arduino.
- Used for messages, debugging, and data tables.

### Serial Plotter
- Graphs numeric values in real time.
- Used to visualize signals and trends.

Both require:
- `Serial.begin(9600);` in the code,
- the baud rate set to **9600** in the IDE.

---

## 9. Test Program Overview

The test program below will:
- blink the onboard LED,
- print status messages and numeric data,
- send data suitable for plotting,
- demonstrate timing using `millis()`.

---

## 10. Arduino Test Code  
*(Copy and paste into the Arduino IDE — each line shown explicitly)*

`/*`  
`Arduino Quickstart Test`  
`- Blinks onboard LED`  
`- Prints messages to Serial Monitor`  
`- Outputs numeric data for Serial Plotter`  
 
`Serial output format:`  
`time_s, led_state, sawtooth`  
`*/`
 
`const int LED_PIN = LED_BUILTIN;`  
`const unsigned long BLINK_MS = 500;`
 
`unsigned long lastToggle = 0;`  
`bool ledState = false;`
  
`void setup() {`  
`  pinMode(LED_PIN, OUTPUT);`

`  Serial.begin(9600);`  
`  while (!Serial) { }`
 
`  Serial.println("Arduino Quickstart Test Running...");`  
`  Serial.println("Columns: time_s, led_state, sawtooth");`  
`}`
  
`void loop() {`  
`  unsigned long t_ms = millis();`
  
`  if (t_ms - lastToggle >= BLINK_MS) {`  
`    lastToggle = t_ms;`  
`    ledState = !ledState;`  
`    digitalWrite(LED_PIN, ledState ? HIGH : LOW);`  
`  }`
  
`  float t_s = t_ms / 1000.0;`
  
`  int sawtooth = (t_ms / 100) % 101;`
  
`  Serial.print(t_s, 2);`  
`  Serial.print(", ");`  
`  Serial.print(ledState ? 1 : 0);`  
`  Serial.print(", ");`  
`  Serial.println(sawtooth);`
 
`  delay(100);`  
`}`

## 11. Using the Serial Monitor

1. Open the Serial Monitor:
   - click the **Serial Monitor icon**.
2. Set the baud rate to **9600**.
3. You should see text and continuously updating data.

If nothing appears:
- confirm the baud rate,
- press the reset button on the Arduino.

---

## 12. Using the Serial Plotter

1. Open the Serial Plotter:
   - click the **Serial Plotter icon**.
2. Set the baud rate to **9600**.
3. The plot should show changing values over time.
   - The sawtooth signal should rise and reset repeatedly.

The plotter automatically graphs all numeric columns it detects.

---

## 13. Common Issues and Fixes

- **Upload fails:** check board/port selection.
- **No serial output:** check baud rate and reset the board.
- **Flat plot:** wait a few seconds for autoscaling.
- **Garbled text:** baud rate mismatch.

---

## 14. What You Should Be Able to Do Now

After completing this guide, you should be able to:
- connect and configure an Arduino,
- verify and upload code,
- use the Serial Monitor for data output,
- use the Serial Plotter for visualization.
