## Problem B5.1

A $45^\circ$ wedge is pushed along a horizontal table with constant acceleration $A$. A block of mass $m$ slides without friction on the incline part of the wedge. Optimus is a stationary observer on the floor and see the wedge moving away from him while Optimus Prime is at rest at the bottom of the wedge. 

1. Find the block’s acceleration as observed by Optimus.
2. Find the block’s acceleration as observed by Optimus Prime.

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
import numpy as np

theta = np.radians(45)

AX, ax, ay, fn, m, g = sp.symbols('AX, ax, ay, fn, m, g')

#-----------------------

# PART 1
# N2 in x-direction
eq_x = m*ax - fn*sp.sin(theta)
eq_y = m*ay - fn*sp.cos(theta) + m*g

#constraint equation
#con = (H-y) - sp.tan(theta)*(x - X)
con_ddot = -ay - sp.tan(theta)*(ax - AX)

sol = sp.solve((eq_x, eq_y, con_ddot),(ax,ay,fn))
print(sol)

#-----------------------

# PART 2

ax_prime, ay_prime, fn, m, g = sp.symbols('ax_prime, ay_prime, fn, m, g')
f_fict_x = -m*AX
f_fict_y = 0
# N2 in x_prime-direction
eq_x_prime = m*ax_prime - fn*sp.sin(theta) - f_fict_x
eq_y_prime = m*ay_prime - fn*sp.cos(theta) + m*g - 0

#constraint equation
#con = (H-y_prime) - sp.tan(theta)*x_prime
con_ddot = -ay_prime - sp.tan(theta)*ax_prime

sol = sp.solve((eq_x_prime, eq_y_prime, con_ddot),(ax_prime,ay_prime,fn))
print(sol)
```

## Problem B5.2

A small weight of mass $m$ hangs from a string in an automobile that
accelerates at rate $A$ along a horizontal road. What is the static angle of the string from the
vertical, and what is the tension in the string?

Solve it for both an observer stationary on the road (Optimus)  and an observer at rest in the car (Optimus Prime).

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
import numpy as np

m, A, theta, ft, g = sp.symbols('m, A, theta, ft, g')

# Optimus
#--------

eqx = m*A - ft*sp.sin(theta) 
eqy = 0 - ft*sp.cos(theta) + m*g

sol = sp.solve((eqx, eqy),(theta,ft))
print(sol)
print(' ')
print('NOTICE: python does not simpliy expression: solution is: tan(theta) = A/g and ft = sqrt(g^2 + A^2)')

# Optimus-prime
#--------
f_fict = -m*A
eqx_prime = m*A - ft*sp.sin(theta) 
eqy_prime = 0 - ft*sp.cos(theta) + m*g

sol = sp.solve((eqx_prime, eqy_prime),(theta,ft))
print(' ')
print(sol)
print(' ')
print('NOTICE: python does not simpliy expression: solution is: tan(theta) = A/g and ft = sqrt(g^2 + A^2)')
```
