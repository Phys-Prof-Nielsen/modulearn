## B2.2.1 Motivation
When collecting and analyzing data, errors are unavoidable. Understanding how to quantify and interpret errors ensures that our results are reliable and meaningful. Error analysis helps identify sources of uncertainty, estimate their impact, and determine confidence in the measurements.

In kinematics, common data types include position, velocity, and acceleration as functions of time. Measurements of these quantities involve various sources of error that must be carefully considered. The primary goals of error analysis include:

- Identifying the types of errors present in the data.
- Understanding how errors propagate through calculations.
- Using statistical measures to assess data reliability.
- Quantifying confidence in measurements and conclusions.

## B2.2.2 Measures of Central Tendency
The mean is the most commonly used measure of central tendency and is calculated by summing all data points and dividing by the number of points:

$$
\mu = \frac{1}{N} \sum_{i=1}^{N} x_i
$$

where:
- $ \mu $ is the mean,
- $ N $ is the total number of data points,
- $ x_i $ represents each individual data point.

**Example:**
Given the dataset $\{2, 4, 6, 8, 10\} $, the mean is:

$$
\mu = \frac{2+4+6+8+10}{5} = 6
$$

The mean is sensitive to extreme values (outliers), which can distort the representation of the dataset.

### Median
The median is the middle value when data points are arranged in ascending order. If $ N $ is odd, the median is the middle number. If $ N $ is even, the median is the average of the two middle numbers.

**Example:**
- Dataset: $ \{2, 4, 6, 8, 10\} $ → Median = 6 (middle value)
- Dataset: $ \{2, 4, 6, 8\}$ → Median = $ \frac{4+6}{2} = 5 $

The median is useful when the data has outliers, as it is less sensitive to extreme values than the mean.

### Mode
The mode is the most frequently occurring value in a dataset. A dataset can have one mode (unimodal), multiple modes (multimodal), or no mode at all.

**Example:**
- Dataset: $ \{1, 2, 2, 3, 4\}$ → Mode = 2
- Dataset: $ \{1, 1, 2, 2, 3, 4\} $ → Modes = 1 and 2 (bimodal)

The mode is useful in categorical data analysis and identifying trends in repeated measurements.

## B2.2.3 Measures of Variability

Variability describes how spread out data points are, providing insight into data consistency and reliability. Common measures include:

### Range
The range is the simplest measure of variability, defined as the difference between the highest and lowest values:

$$
\text{Range} = x_{max} - x_{min}
$$

**Example:**
For $ \{2, 4, 6, 8, 10\} $, the range is:

$$
10 - 2 = 8
$$

The range gives a quick sense of dispersion but is highly sensitive to outliers.

### Variance
Variance quantifies the **overall spread of data points from the mean**:

$$
\sigma^2 = \frac{1}{N} \sum_{i=1}^{N} (x_i - \mu)^2
$$

where $ \sigma^2 $ is the variance. A higher variance indicates greater data spread.

### Standard Deviation (SD)
The standard deviation is the square root of variance and provides a **measure of dispersion in the same units as the data**:

$$
s = \sqrt{\sigma^2}
$$

A smaller standard deviation indicates that data points are closely clustered around the mean.

## B2.2.4 Measures of Confidence

When working with sample data, we estimate how well the sample represents the population. Two key measures are the standard error of the mean (SEM) and confidence intervals (CI).

### Standard Error of the Mean (SEM)
SEM quantifies how much the sample mean deviates from the true population mean:

$$
s_{\bar{x}} = \frac{\sigma}{\sqrt{N}}
$$

where:
- $ \sigma $ is the sample standard deviation,
- $ N $ is the sample size.

A smaller SEM suggests a more precise estimate of the population mean $\bar{x}$.

>### Reporting with Standard Deviation vs. Standard Error
>
>When reporting results, it is important to distinguish between **variability of the data** and **uncertainty in the mean**. The **sample standard deviation** $s$ describes how widely the individual measurements are spread around the mean; it is most appropriate when you want to show the natural scatter in the raw data (for example, in a histogram or when comparing how consistent repeated trials are). The **standard error of the mean** $s_{\bar{x}} = s/\sqrt{n}$, on the other hand, reflects how precisely the mean itself has been determined from the limited number of trials. In scientific reporting, the convention is usually to quote the **mean with its standard error** (or with a confidence interval derived from it), because this communicates the reliability of the reported central value. However, including both $s$ and $s_{\bar{x}}$ is often helpful: $s$ shows the spread of the measurements, while $s_{\bar{x}}$ shows the confidence we have in the mean.

### Confidence Intervals (CI)
A confidence interval provides a range within which the true population mean is likely to fall. A $95\%$ confidence interval is calculated as:

$$
CI = \mu \pm z \times SEM
$$

where $ z $ is the critical value from the standard normal distribution ($1.96$ for $95\%$ confidence).

**Example:**
For a dataset with $ \mu = 6 $, $ \sigma = 2 $, and $ N = 25 $:

$$
SEM = \frac{2}{\sqrt{25}} = 0.4
$$

$$
CI = 6 \pm (1.96 \times 0.4)
$$

$$
CI = (5.22, 6.78)
$$

This means we are $95\%$ confident that the population mean lies between $5.22$ and $6.78$.

### Common $ z$-Values for Confidence Intervals

| Confidence Level ($\%$) | $ z$-Value |
|----------------------|------------|
| $80\%$                 | $1.28$       |
| $85\%$                 | $1.44$       |
| $90\%$                 | $1.645$      |
| $95\%$                 | $1.96$       |
| $98\%$                | $2.33$       |
| $99\%$                | $2.576$      |

These $z$-values correspond to the number of standard deviations from the mean required to capture the specified confidence level in a normal distribution.

## B2.2.5. Summary
Error analysis is crucial for assessing the reliability of experimental results. By quantifying measures of central tendency, variability, and confidence, we ensure that conclusions drawn from data are valid. Understanding these principles allows for better experimental design, improved measurement techniques, and more reliable interpretations of physical phenomena. Through careful analysis, scientists and engineers can make more informed decisions based on measured data.
