## **Problem 1: Working with Data and Basic Error Analysis**  
A student measures the time it takes for a ball to roll down a ramp in five trials:  

| Trial | Time (s) |
|-------|---------|
| 1     | 2.1     |
| 2     | 2.3     |
| 3     | 2.2     |
| 4     | 2.4     |
| 5     | 2.2     |

1. Calculate the mean, median, and mode of the data.  
2. Compute the variance and standard deviation of the time measurements.  
3. Calculate the confidence intervals for the mean time using 95$\%$ and 99$\%$ confidence level. Assume the sample follows a normal distribution.  
4. Interpret what the standard deviation and confidence interval tell you about the consistency and reliability of the measurements.

## **Problem 2: Linear Regression and Best Fit Line**  
A toy car moves along a straight track, and its position at different times is recorded:  

| Time (s) | Position (m) |
|----------|-------------|
| 0        | 0           |
| 1        | 2.1         |
| 2        | 4.0         |
| 3        | 6.2         |
| 4        | 8.1         |

1. Plot the position-time data on a graph and determine if the motion is linear.  
2. Use linear regression to determine the best fit equation for the data.  
3. Interpret the physical meaning of the slope and y-intercept in the context of the motion.
