# B12.2 The Simple Pendulum

## B12.2.1 Motivation

A **pendulum** is one of the most familiar and useful oscillating systems — from clock mechanisms and seismographs to sensing instruments and even biological motion. Despite its apparent simplicity, it provides deep insight into **rotational dynamics**, **restoring forces**, and the general nature of **periodic motion**.

Pendulums appear across disciplines:
- In **engineering**, they are used in vibration control systems, timing devices, accelerometers, and stability analysis of structures.  
- In **physics and chemistry**, the pendulum models small-angle oscillations and serves as an analog for molecular vibrations and potential wells.  
- In **biology and medicine**, pendulum-like motion occurs in walking and running gait cycles, limb movement, cardiac rhythm, and neural oscillations.

By understanding the simple pendulum, we gain not only a physical intuition for oscillatory motion but also a mathematical framework that connects natural phenomena across science and engineering.

## B12.2.2 Derivation of the Equation of Motion

The pendulum moves along a circular arc, so it is most natural to describe its motion using **polar coordinates** rather than Cartesian coordinates.

Let the bob of mass $m$ be attached to a string of fixed length $L$.  
The position of the bob is specified by the polar coordinates $(r = L,\, \theta)$, where $\theta$ is the angle measured from the vertical.

Because the length of the pendulum is constant, $r = L$ is fixed, so $\dot{r} = \ddot{r} = 0$.  
The acceleration in polar coordinates is given by

$$
\vec{a} = (\ddot{r} - r\dot{\theta}^2)\,\hat{r} + (r\ddot{\theta} + 2\dot{r}\dot{\theta})\,\hat{\theta}
$$

Substituting $r = L$ and $\dot{r} = 0$, the acceleration reduces to

$$
\vec{a} = (-L\dot{\theta}^2)\,\hat{r} + (L\ddot{\theta})\,\hat{\theta}
$$

The forces acting on the bob are:
1. **Tension** $\vec{T}$ directed along $-\hat{r}$  
2. **Gravity** $\vec{F}_g = -mg\,\hat{y}$

Decomposing the gravitational force into components along $\hat{r}$ and $\hat{\theta}$ gives

$$
\vec{F}_g = -mg\cos\theta\,\hat{r} - mg\sin\theta\,\hat{\theta}
$$

Applying Newton’s second law $\vec{F} = m\vec{a}$ in each direction:

- **Radial component:**

  $$
  -T - mg\cos\theta = -mL\dot{\theta}^2
  $$

  which determines the tension (not needed for the angular motion directly).

- **Tangential component:**

  $$
  -mg\sin\theta = mL\ddot{\theta}
  $$

Simplifying the tangentional motion, we obtain the **equation of motion**:

$$
\frac{d^2\theta}{dt^2} + \frac{g}{L}\sin\theta = 0
$$

This is the exact nonlinear equation describing the motion of a simple pendulum.  
As before, it becomes identical in form to the SHM equation when we use the **small-angle approximation**, $\sin\theta \approx \theta$.

### Small-Angle Approximation

A smooth function $f(\theta)$ can be expanded about $\theta=0$ using a **Taylor series**:

$$
f(\theta)=f(0)+f'(0)\,\theta+\frac{f''(0)}{2!}\theta^2+\frac{f^{(3)}(0)}{3!}\theta^3+\cdots
$$

Applying this to $\sin\theta$:

$$
\sin\theta=\theta-\frac{\theta^3}{3!}+\frac{\theta^5}{5!}-\cdots
$$

For **small angles** (e.g., $\theta\lesssim 0.1\ \text{rad}\approx 6^\circ$), higher-order terms like $\theta^3/6$ are negligible, so we take

$$
\sin\theta \approx \theta.
$$

Substituting $\sin\theta \approx \theta$ into the equation of motion gives

$$
\boxed{\displaystyle \frac{d^2\theta}{dt^2} + \frac{g}{L}\theta = 0}
$$

We should immediately recognize this as the **standard ODE form for simple harmonic motion**,

$$
\frac{d^2x}{dt^2} + \omega^2 x = 0,
$$

from which we can identify, **by inspection**,

$$
\boxed{\displaystyle \omega = \sqrt{\frac{g}{L}}.}
$$

### Solution

The angular displacement is therefore:

$$
\theta(t) = \Theta_0 \cos(\omega t + \phi)
$$

where $\Theta_0$ is the maximum angular displacement (amplitude).  

The **period** and **frequency** follow as:

$$
T = 2\pi \sqrt{\frac{L}{g}}, \qquad f = \frac{1}{T} = \frac{1}{2\pi}\sqrt{\frac{g}{L}}
$$

These expressions show that the period depends only on the pendulum length $L$ and gravitational acceleration $g$ — **not** on the mass or amplitude (as long as the small-angle condition holds).

## B12.2.3 Example

<div style="background-color:#d9d9d9; padding:15px; border-radius:10px;">

**Determining the Period**

A simple pendulum of length $L = 0.50\ \text{m}$ oscillates with small amplitude near Earth’s surface ($g = 9.81\ \text{m/s}^2$).  
Find its period and frequency.

**Solution**

$$
T = 2\pi \sqrt{\frac{L}{g}} = 2\pi \sqrt{\frac{0.50}{9.81}} = 1.42\ \text{s}
$$

$$
f = \frac{1}{T} = 0.70\ \text{Hz}
$$

Thus, the pendulum completes one full swing every 1.42 seconds.

## B12.2.4 Energy Considerations

As with the mass–spring system, energy oscillates between **potential** and **kinetic** forms.

- The **potential energy** (relative to the lowest point) is  

  $$
  U = mgL(1 - \cos\theta)
  $$

- The **kinetic energy** is  

  $$
  K = \tfrac{1}{2} m (L\dot{\theta})^2
  $$

For small angles ($\cos\theta \approx 1 - \tfrac{1}{2}\theta^2$):

$$
U \approx \tfrac{1}{2} mgL\theta^2
$$

so total energy is approximately constant:

$$
E = \tfrac{1}{2} m L^2 \dot{\theta}^2 + \tfrac{1}{2} mgL \theta^2 = \tfrac{1}{2} mgL \Theta_0^2
$$

This is identical in form to the energy of a linear SHM system ($E = \tfrac{1}{2} kA^2$), confirming that small-angle motion of a pendulum is **simple harmonic**.

## B12.2.5 Physical Interpretation

- For **small angles**, the restoring torque is proportional to displacement, giving SHM.
- For **larger angles**, $\sin\theta$ deviates from $\theta$, and the motion slows — the period increases slightly.
- The motion is **independent of mass**, a reflection of the universality of free fall.

## B12.2.6 Summary Box

$$
\boxed{
\begin{aligned}
&\text{Equation of Motion:} &\quad \frac{d^2\theta}{dt^2} + \frac{g}{L}\sin\theta = 0 \\
&\text{Small-Angle SHM Form:} &\quad \frac{d^2\theta}{dt^2} + \frac{g}{L}\theta = 0 \\
&\text{Angular Frequency:} &\quad \omega = \sqrt{\frac{g}{L}} \\
&\text{Period:} &\quad T = 2\pi \sqrt{\frac{L}{g}}
\end{aligned}
}
$$

## B12.2.7 Connection to the Spring Oscillator

Both systems share the same mathematical structure for SHM:

$$
\frac{d^2(\text{displacement})}{dt^2} + \omega^2 (\text{displacement}) = 0
$$

Only the **source of the restoring force** differs:
- For the spring: $F = -kx$
- For the pendulum: $F = -mg\sin\theta \approx -mg\theta$

This shared form explains why their energy, phase, and frequency relationships are nearly identical — they are all **manifestations of the same underlying harmonic dynamics**.
