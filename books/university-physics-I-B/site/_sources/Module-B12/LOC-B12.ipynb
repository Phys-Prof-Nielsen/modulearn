# B12.0 Learning Outcomes - Oscillations

## Motivation

Many systems in nature and engineering oscillate about an equilibrium position — from atoms vibrating in molecules to bridges swaying in the wind, from the motion of a pendulum to the vibrations of a guitar string. **Simple Harmonic Motion (SHM)** provides a powerful and elegant mathematical framework to describe these oscillations.  

In this module, we explore how linear restoring forces lead to sinusoidal motion, why small oscillations around stable equilibrium positions are approximately harmonic, and how this connects to broader ideas of **stability** in physical systems. Understanding SHM not only builds a foundation for more complex oscillations (damped and driven motion in the next phase) but also for waves, resonance, and energy exchange in both mechanical and electromagnetic systems.

## Learning Outcomes

By the end of this module, you should be able to:

### Conceptual Understanding
1. Explain the physical meaning of **simple harmonic motion** and identify systems that exhibit it.  
2. Distinguish between **stable**, **unstable**, and **neutral** equilibrium using potential energy diagrams.  
3. Describe how SHM arises naturally from a **linear restoring force** near a stable equilibrium.  
4. Explain the physical differences between a **simple pendulum** and a **physical pendulum**, and discuss conditions under which each approximates SHM.

### Analytical and Mathematical Skills
5. Derive the **equation of motion** for a mass–spring system using Newton’s second law and calculus.  
6. Solve the differential equation of SHM and express position, velocity, and acceleration as functions of time.  
7. Show that total **mechanical energy** in SHM is constant and alternates between potential and kinetic forms.  
8. Linearize the equations of motion for small oscillations to connect **stability** with harmonic motion.  
9. Derive and interpret the **period** and **frequency** for simple and physical pendulums, including small-angle approximations.

### Experimental and Computational Skills
10. Analyze experimental or simulated data to determine the period, frequency, and amplitude of an oscillator.  
11. Plot potential energy curves and identify equilibrium points and their stability.
