# B12.3 The Physical Pendulum
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B12.3.1 Motivation

Unlike the *simple pendulum*, which assumes all the mass is concentrated at a single point, most real objects have **distributed mass**. When such an extended body oscillates about a pivot under gravity, we call it a **physical pendulum**.

Examples include:
- A **swinging door** about its hinges,  
- A **rod or disk** pivoted at one end,  
- A **person balancing a leg or arm**, or  
- **Clock pendulums** and **sensors** where shape and mass distribution matter.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B12.3.2 Derivation of the Equation of Motion

Consider a rigid body of total mass $m$, pivoted at a point $P$ that is a distance $d$ from its center of mass (CM).  
When displaced by a small angle $\theta$ from equilibrium, gravity exerts a restoring torque about the pivot.

The **torque** due to gravity is:

$$
\tau = -mgd \sin\theta
$$

Applying **Newton’s Second Law for Rotation**:

$$
\sum \tau = I_P \frac{d^2\theta}{dt^2}
$$

where $I_P$ is the **moment of inertia** about the pivot.

Thus,

$$
I_P \frac{d^2\theta}{dt^2} = -mgd \sin\theta
$$

---

### Small-Angle Approximation

For small oscillations ($\sin\theta \approx \theta$), this becomes:

$$
I_P \frac{d^2\theta}{dt^2} = -mgd \theta
$$

or

$$
\boxed{\displaystyle \frac{d^2\theta}{dt^2} + \frac{mgd}{I_P}\theta = 0}
$$

---

### Identifying the Angular Frequency

We can now compare this to the general SHM form:

$$
\frac{d^2x}{dt^2} + \omega^2 x = 0
$$

By inspection,

$$
\boxed{\displaystyle \omega = \sqrt{\frac{mgd}{I_P}}}
$$

and the **period of oscillation** is

$$
\boxed{\displaystyle T = 2\pi \sqrt{\frac{I_P}{mgd}}}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B12.3.3 Example

<div style="background-color:#d9d9d9; padding:15px; border-radius:10px;">

**Uniform Rod Pivoted at One End**

For a uniform rod of length $L$ and mass $m$, pivoted at one end:

$$
I_P = \frac{1}{3} mL^2, \qquad d = \frac{L}{2}
$$

Then

$$
T = 2\pi \sqrt{\frac{I_P}{mgd}} = 2\pi \sqrt{\frac{(1/3)mL^2}{mg(L/2)}} = 2\pi \sqrt{\frac{2L}{3g}}
$$

If $L = 1.0\ \text{m}$, the period is:

$$
T = 2\pi \sqrt{\frac{2(1.0)}{3(9.81)}} = 1.64\ \text{s}
$$

</div>

## B12.3.4 Energy Considerations

The potential energy (taking $\theta = 0$ as the lowest point) is:

$$
U = mgd(1 - \cos\theta)
$$

and the kinetic energy associated with rotation is:

$$
K = \tfrac{1}{2} I_P \dot{\theta}^2
$$

For small angles ($\cos\theta \approx 1 - \tfrac{1}{2}\theta^2$):

$$
U \approx \tfrac{1}{2} mgd\,\theta^2
$$

so total energy is approximately constant:

$$
E = \tfrac{1}{2} I_P \dot{\theta}^2 + \tfrac{1}{2} mgd\,\theta^2 = \tfrac{1}{2} mgd\,\Theta_0^2
$$

This quadratic potential energy again confirms that small oscillations of a physical pendulum are **simple harmonic**.

## B12.3.5 Physical Interpretation

- The term $mgd$ acts as the **restoring torque constant**, analogous to $k$ in a spring.
- The **moment of inertia $I_P$** plays the role of **mass** — it determines how resistant the system is to rotational acceleration.
- The system oscillates faster when the pivot is close to the center of mass (small $d$) or when the body is less rotationally massive (small $I_P$).
- Any rigid body can act as a pendulum, and by measuring its period, we can experimentally determine its **moment of inertia** or even **gravitational acceleration**.
  
In more complex oscillating systems — such as multiple connected rigid bodies or coupled pendulums — the quantities analogous to $m$ and $k$ are no longer single numbers.  
Instead, they become **matrices**:

- The **mass (or inertia) matrix** $\mathbf{M}$ generalizes $m$ or $I_P$, describing how mass is distributed and how the motion of one coordinate affects another.  
- The **stiffness matrix** $\mathbf{K}$ generalizes $k$ or $mgd$, encoding how the system’s restoring torques or forces depend on each displacement coordinate.

The general equation of motion then takes the matrix form

$$
\boxed{\mathbf{M}\,\ddot{\boldsymbol{\theta}} + \mathbf{K}\,\boldsymbol{\theta} = 0,}
$$

which reduces to the familiar single-variable SHM equation when $\mathbf{M}$ and $\mathbf{K}$ are scalars.  
Solving this system yields **normal modes** — collective oscillations where all parts move sinusoidally with the same frequency but possible different amplitudes and phases.

## B12.3.6 Connection to the Simple Pendulum

If all the mass were concentrated at a single point a distance $L$ from the pivot, then

$$
I_P = mL^2, \quad d = L
$$

Substituting into the period expression gives

$$
T = 2\pi \sqrt{\frac{I_P}{mgd}} = 2\pi \sqrt{\frac{mL^2}{mgL}} = 2\pi \sqrt{\frac{L}{g}},
$$

which is precisely the result for the **simple pendulum** — confirming that the simple pendulum is a *special case* of the physical pendulum.
