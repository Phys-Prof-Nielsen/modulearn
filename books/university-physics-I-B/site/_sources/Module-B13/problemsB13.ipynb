## Problem 1 — Reading Wave Parameters and Evaluating the Field

A transverse mechanical wave on a string is described by  

$$
y(x,t) = 0.080\,\sin(5.0x - 20.0t + \tfrac{\pi}{6}).
$$

1. Identify the **amplitude** $A$, **wavenumber** $k$, and **angular frequency** $\omega$.  
2. Compute the **wavelength** $\lambda = \tfrac{2\pi}{k}$, **frequency** $f = \tfrac{\omega}{2\pi}$, and **phase speed** $v = \tfrac{\omega}{k}$.  
3. State the **direction of propagation**.  
4. Evaluate the **displacement** $y$ at $x = 0.60\,\text{m}$ and $t = 0.25\,\text{s}$.


<div style="background-color:#e8f3ec; border-left:8px solid #275D38; padding:16px; border-radius:10px;">

<details>
<summary><em><strong>Solution to Problem 1</strong></em></summary>

We are given the transverse wave

$$
y(x,t) = 0.080\,\sin\!\big(5.0x - 20.0t + \tfrac{\pi}{6}\big).
$$

We compare this with the standard traveling-wave form

$$
y(x,t) = A \sin(kx - \omega t + \phi).
$$

---

**1. Identify $A$, $k$, and $\omega$**

By direct comparison:

- Amplitude:

  $$
  A = 0.080\ \text{m}
  $$
  
- Wavenumber:

  $$
  k = 5.0\ \text{rad/m}
  $$

  
- Angular frequency:

  $$
  \omega = 20.0\ \text{rad/s}
  $$

  
- Phase constant:

  $$
  \phi = \frac{\pi}{6}
  $$

---

**2. Compute $\lambda$, $f$, and $v$**

**Wavelength**:

$$
\lambda = \frac{2\pi}{k}
= \frac{2\pi}{5.0}
\ \text{m}.
$$

**Frequency**:

$$
f = \frac{\omega}{2\pi}
= \frac{20.0}{2\pi}
= \frac{10}{\pi}
\ \text{Hz}.
$$

**Phase speed**:

$$
v = \frac{\omega}{k}
= \frac{20.0}{5.0}
= 4.0\ \text{m/s}.
$$

So numerically,
- $\lambda \approx 1.257\ \text{m}$
- $f \approx 3.183\ \text{Hz}$
- $v = 4.0\ \text{m/s}$

---

**3. Direction of propagation**

The wave has the argument

$$
5.0x - 20.0t + \frac{\pi}{6} = kx - \omega t + \phi,
$$

with a **minus** sign in front of $\omega t$.

A wave of the form $y(x,t) = A\sin(kx - \omega t + \phi)$ travels in the **$+x$ direction** (to the right).  
Therefore, this wave propagates in the **positive $x$ direction**.

---

**4. Displacement at $x = 0.60\ \text{m}$ and $t = 0.25\ \text{s}$**

First compute the phase:

$$
\theta = kx - \omega t + \phi
= (5.0)(0.60) - (20.0)(0.25) + \frac{\pi}{6}.
$$

Step by step:

$$
(5.0)(0.60) = 3.0,
$$

$$
(20.0)(0.25) = 5.0,
$$

so

$$
\theta = 3.0 - 5.0 + \frac{\pi}{6}
= -2.0 + \frac{\pi}{6}.
$$

Thus the displacement is

$$
y(0.60, 0.25)
= 0.080\,\sin\!\Big(-2.0 + \frac{\pi}{6}\Big)\ \text{m}.
$$

Numerically, this evaluates to approximately

$$
y(0.60, 0.25) \approx 0.080\,\sin(-1.4768)\ \text{m}
\approx -0.079\ \text{m},
$$

so the string is displaced about $7.9\ \text{cm}$ **below** equilibrium at that point.

</details>

</div>


```python
%reset -f

import numpy as np

# Given parameters from the wave
A = 0.080          # m
k = 5.0            # rad/m
omega = 20.0       # rad/s
phi = np.pi / 6    # rad

# 1) Derived quantities
lam = 2 * np.pi / k        # wavelength λ
f = omega / (2 * np.pi)    # frequency f
v = omega / k              # phase speed v

# 2) Evaluate y at x = 0.60 m, t = 0.25 s
x = 0.60
t = 0.25
theta = k * x - omega * t + phi
y_xt = A * np.sin(theta)

print("Derived parameters:")
print(f"λ = {lam:.4f} m")
print(f"f = {f:.4f} Hz")
print(f"v = {v:.4f} m/s")

print("\nPhase at (x,t) = (0.60 m, 0.25 s):")
print(f"θ = {theta:.4f} rad")

print("\nDisplacement:")
print(f"y(0.60 m, 0.25 s) = {y_xt:.5f} m")
```

## Problem 2 — Superposition with a Phase Difference

Two waves of equal amplitude travel in the same direction:

$$
y_1 = A \sin(kx - \omega t), \qquad
y_2 = A \sin(kx - \omega t + \phi),
$$

with $A = 0.050\,\text{m}$, $\lambda = 1.20\,\text{m}$, $f = 3.0\,\text{Hz}$, and $\phi = 120^\circ$.

1. Derive the resultant wave $y = y_1 + y_2$ in the form  

$$
y = A_{\text{eff}} \sin(kx - \omega t + \phi_{\text{eff}}),
$$

and compute $A_{\text{eff}}$.  
2. Evaluate $y$ at $x = 0.40\,\text{m}$, $t = 0.10\,\text{s}$.


<div style="background-color:#e8f3ec; border-left:8px solid #275D38; padding:16px; border-radius:10px;">

<details>
<summary><em><strong>Solution to Problem 2</strong></em></summary>

We are given two waves of equal amplitude traveling in the same direction:

$$
y_1 = A \sin(kx - \omega t), \qquad
y_2 = A \sin(kx - \omega t + \phi),
$$

with  
$A = 0.050\,\text{m}$, $\lambda = 1.20\,\text{m}$, $f = 3.0\,\text{Hz}$, and $\phi = 120^\circ$.

We want the resultant wave

$$
y = y_1 + y_2
$$

written in the form

$$
y = A_{\text{eff}} \sin(kx - \omega t + \phi_{\text{eff}}),
$$

and then evaluate $y$ at $x = 0.40\,\text{m}$, $t = 0.10\,\text{s}$.

---

**1. Derive the resultant wave and find $A_{\text{eff}}$**

Start with

$$
y = y_1 + y_2
= A\sin(kx - \omega t) + A\sin(kx - \omega t + \phi).
$$

Let

$$
\alpha = kx - \omega t, \qquad
\beta = kx - \omega t + \phi.
$$

Then

$$
y = A\sin\alpha + A\sin\beta.
$$

Use the trigonometric identity

$$
\sin\alpha + \sin\beta =
2\sin\!\left(\frac{\alpha + \beta}{2}\right)
 \cos\!\left(\frac{\alpha - \beta}{2}\right).
$$

Compute the half-sum and half-difference:

$$
\frac{\alpha + \beta}{2}
= \frac{(kx - \omega t) + (kx - \omega t + \phi)}{2}
= kx - \omega t + \frac{\phi}{2},
$$

$$
\frac{\alpha - \beta}{2}
= \frac{(kx - \omega t) - (kx - \omega t + \phi)}{2}
= -\,\frac{\phi}{2}.
$$

Thus

$$
y = 2A
\sin\!\left(kx - \omega t + \frac{\phi}{2}\right)
\cos\!\left(-\frac{\phi}{2}\right).
$$

Since $\cos(-\theta) = \cos\theta$, this simplifies to

$$
y = 2A\cos\!\left(\frac{\phi}{2}\right)
\sin\!\left(kx - \omega t + \frac{\phi}{2}\right).
$$

Comparing with the desired form

$$
y = A_{\text{eff}} \sin(kx - \omega t + \phi_{\text{eff}}),
$$

we identify

$$
A_{\text{eff}} = 2A\cos\!\left(\frac{\phi}{2}\right),
\qquad
\phi_{\text{eff}} = \frac{\phi}{2}.
$$

Now plug in the numbers. We have  

- $A = 0.050\ \text{m}$  
- $\phi = 120^\circ \Rightarrow \dfrac{\phi}{2} = 60^\circ$  

So

$$
A_{\text{eff}} = 2(0.050)\cos(60^\circ)
= 0.100 \times 0.5
= 0.050\ \text{m}.
$$

Interestingly, for a phase difference of $120^\circ$, the effective amplitude equals the original amplitude $A$.

---

**2. Evaluate $y$ at $x = 0.40\ \text{m}$, $t = 0.10\ \text{s}$**

First compute $k$ and $\omega$ from $\lambda$ and $f$:

$$
k = \frac{2\pi}{\lambda}
= \frac{2\pi}{1.20}\ \text{rad/m},
$$

$$
\omega = 2\pi f
= 2\pi(3.0)
= 6\pi\ \text{rad/s}.
$$

Using the **original superposition** form:

$$
y(x,t) = A\sin(kx - \omega t)
       + A\sin(kx - \omega t + \phi).
$$

At $x = 0.40\ \text{m}$ and $t = 0.10\ \text{s}$:

1. Compute the phase $kx - \omega t$:

   $$
   kx = \frac{2\pi}{1.20}(0.40),
   \qquad
   \omega t = (6\pi)(0.10).
   $$

2. Evaluate each sine term:

   $$
   y_1 = A\sin(kx - \omega t),
   \qquad
   y_2 = A\sin(kx - \omega t + \phi),
   $$

3. Add to get $y(x,t)$.

Alternatively, using the **compact** form:

$$
y(x,t) = A_{\text{eff}}
\sin\!\left(kx - \omega t + \frac{\phi}{2}\right),
$$

with $A_{\text{eff}} = 0.050\ \text{m}$ and $\dfrac{\phi}{2} = 60^\circ$, we evaluate

$$
y(0.40, 0.10)
= 0.050\,\sin\!\left(k(0.40) - \omega(0.10) + 60^\circ\right).
$$

In either case, the numerical value will match (up to rounding).  
We will let Python carry out the numerical evaluation in the code cell below.

</details>

</div>


```python
%reset -f

import numpy as np

# Given parameters
A = 0.050        # m
lam = 1.20       # m
f = 3.0          # Hz
phi_deg = 120.0  # degrees
phi = np.deg2rad(phi_deg)  # radians

# Wave numbers and frequencies
k = 2 * np.pi / lam        # rad/m
omega = 2 * np.pi * f      # rad/s

# Effective amplitude and phase shift from superposition identity
A_eff = 2 * A * np.cos(phi / 2)
phi_eff = phi / 2

# Point where we evaluate
x = 0.40   # m
t = 0.10   # s

# Direct superposition
y1 = A * np.sin(k * x - omega * t)
y2 = A * np.sin(k * x - omega * t + phi)
y_total = y1 + y2

# Compact effective form
y_compact = A_eff * np.sin(k * x - omega * t + phi_eff)

print("Wave parameters:")
print(f"k = {k:.4f} rad/m")
print(f"ω = {omega:.4f} rad/s")

print("\nEffective parameters from superposition:")
print(f"A_eff = {A_eff:.5f} m")
print(f"phi_eff = {phi_eff:.5f} rad ({np.rad2deg(phi_eff):.2f}°)")

print("\nEvaluation at x = 0.40 m, t = 0.10 s:")
print(f"y1 = {y1:.6f} m")
print(f"y2 = {y2:.6f} m")
print(f"y_total (from y1 + y2) = {y_total:.6f} m")
print(f"y_compact (from A_eff, phi_eff) = {y_compact:.6f} m")
```

## Problem 3 — Pressure Amplitude in a Sound Wave

A longitudinal sound wave in air has displacement amplitude $s_m = 1.0\,\mu\text{m}$ and frequency $f = 1000\,\text{Hz}$.  
Assume air density $\rho = 1.21\,\text{kg/m}^3$ and speed of sound $v = 343\,\text{m/s}$.

1. Compute the wavelength $\lambda$, wavenumber $k$, and angular frequency $\omega$.  
2. Using  

$$
p_{\text{amp}} = \rho v \omega s_m,
$$  

find the **pressure amplitude** $p_{\text{amp}}$.


<div style="background-color:#e8f3ec; border-left:8px solid #275D38; padding:16px; border-radius:10px;">

<details>
<summary><strong>Solution to Problem 3</strong></summary>

We are given a longitudinal **sound wave in air** with:

- Displacement amplitude:  

  $$
  s_m = 1.0\,\mu\text{m} = 1.0 \times 10^{-6}\ \text{m}
  $$

- Frequency:  

  $$
  f = 1000\ \text{Hz}
  $$

- Air density:  

  $$
  \rho = 1.21\ \text{kg/m}^3
  $$

- Speed of sound:  

  $$
  v = 343\ \text{m/s}
  $$

---

**1. Compute $\lambda$, $k$, and $\omega$**

**Wavelength**

Using

$$
\lambda = \frac{v}{f},
$$

we get

$$
\lambda = \frac{343}{1000}
= 0.343\ \text{m}.
$$

---

**Wavenumber**

$$
k = \frac{2\pi}{\lambda}
= \frac{2\pi}{0.343}\ \text{rad/m}.
$$

---

**Angular frequency**

$$
\omega = 2\pi f
= 2\pi (1000)
= 2000\pi\ \text{rad/s}.
$$

---

**Compute the **pressure amplitude****

For a plane sound wave, the pressure amplitude is:

$$
p_{\text{amp}} = \rho\,v\,\omega\,s_m.
$$

Substituting:

$$
p_{\text{amp}}
= (1.21)(343)(2000\pi)(1.0\times10^{-6})\ \text{Pa}.
$$

This gives a pressure amplitude on the order of **1 pascal**, typical for moderately loud sound.

---

</details>

</div>


```python
%reset -f

import numpy as np

# Given values
s_m = 1.0e-6      # m
f = 1000.0        # Hz
rho = 1.21        # kg/m^3
v = 343.0         # m/s

# Calculations
lam = v / f
k = 2 * np.pi / lam
omega = 2 * np.pi * f
p_amp = rho * v * omega * s_m

print("Computed quantities:")
print(f"λ = {lam:.4f} m")
print(f"k = {k:.4f} rad/m")
print(f"ω = {omega:.2f} rad/s")
print(f"Pressure amplitude p_amp = {p_amp:.6f} Pa")
```

## Problem 4 — String Wave Speed from Tension

A wave travels leftward on a string with linear mass density $\mu = 5.0\times10^{-3}\,\text{kg/m}$ and tension $T = 120\,\text{N}$.  
The wavelength is $\lambda = 0.80\,\text{m}$.

1. Compute the wave speed $v = \sqrt{T / \mu}$.  
2. Find the frequency $f = v / \lambda$ and period $T_p = 1 / f$.  
3. Write the left-traveling wave equation  

$$
y(x,t) = A \sin(kx + \omega t + \phi),
$$  

with $A = 0.020\,\text{m}$, $\phi = 0$, and verify that $\omega / k = v$.


<div style="background-color:#e8f3ec; border-left:8px solid #275D38; padding:16px; border-radius:10px;">

<details>
<summary><strong>Solution to Problem 4</strong></summary>

We have a wave traveling **leftward** on a string with:

- Linear mass density  

  $$
  \mu = 5.0\times10^{-3}\,\text{kg/m}
  $$
  
- Tension  

  $$
  T = 120\,\text{N}
  $$

  
- Wavelength  

  $$
  \lambda = 0.80\,\text{m}
  $$

We will:

1. Compute the wave speed $v = \sqrt{T/\mu}$.  
2. Find the frequency $f$ and period $T_p$.  
3. Write a left-traveling wave equation and verify $\omega/k = v$.

---

**1. Wave speed $v = \sqrt{T/\mu}$**

For a wave on a tensioned string, the speed is

$$
v = \sqrt{\frac{T}{\mu}}.
$$

Insert the numbers:

$$
v = \sqrt{\frac{120}{5.0\times10^{-3}}}
  = \sqrt{24000}.
$$

Numerically,

$$
v \approx 155\ \text{m/s}.
$$

(We will compute this more precisely in Python below.)

---

**2. Frequency $f$ and period $T_p$**

We use the relation between speed, frequency, and wavelength:

$$
v = f \lambda \quad\Rightarrow\quad f = \frac{v}{\lambda}.
$$

So

$$
f = \frac{v}{0.80}.
$$

Once we know $f$, the period is

$$
T_p = \frac{1}{f}.
$$

---

**3. Wave equation and check that $\omega/k = v$**

For a sinusoidal wave, the wavenumber and angular frequency are

$$
k = \frac{2\pi}{\lambda}, \qquad
\omega = 2\pi f.
$$

Using $f = v/\lambda$:

$$
\omega = 2\pi \frac{v}{\lambda}.
$$

Thus,

$$
\frac{\omega}{k}
= \frac{2\pi (v/\lambda)}{2\pi/\lambda}
= v,
$$

so the ratio $\omega/k$ equals the wave speed, as expected.

Because the wave travels **leftward**, a convenient form for the wave is

$$
y(x,t) = A \sin(kx + \omega t + \phi),
$$

with

- $A = 0.020\,\text{m}$  
- $\phi = 0$

so explicitly:

$$
y(x,t) = 0.020\,\sin(kx + \omega t).
$$

Here the **plus** sign in front of $\omega t$ corresponds to propagation in the **negative $x$ direction** (left).

We will compute $v$, $f$, $T_p$, $k$, $\omega$, and verify $\omega/k$ numerically in the Python cell below.

</details>

</div>


```python
%reset -f

import numpy as np

# Given values
mu = 5.0e-3   # kg/m
T = 120.0     # N
lam = 0.80    # m
A = 0.020     # m
phi = 0.0     # rad

# 1) Wave speed
v = np.sqrt(T / mu)

# 2) Frequency and period
f = v / lam
T_p = 1.0 / f

# 3) k, omega, and check omega/k
k = 2 * np.pi / lam
omega = 2 * np.pi * f
ratio = omega / k

print("Wave and string parameters:")
print(f"v = sqrt(T/μ) = {v:.4f} m/s")
print(f"f = v/λ       = {f:.44f} Hz")
print(f"T_p = 1/f     = {T_p:.6f} s")

print("\nWave numbers:")
print(f"k = 2π/λ      = {k:.4f} rad/m")
print(f"ω = 2π f      = {omega:.4f} rad/s")
print(f"ω/k           = {ratio:.4f} m/s (should equal v)")
print("\nLeft-traveling wave form:")
print("y(x,t) = 0.020 * sin(k x + ω t)")
```

```python
%reset -f

# Possible solution

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Given constants
v = 3.0      # m/s (wave speed)
lam = 2.0    # m (wavelength)
A = 0.10     # m (amplitude)
t0 = 0.50    # s (snapshot time)

# -----------------------------------------
# 1) Use SymPy to solve v = f * λ for f
# -----------------------------------------

# Define the symbol for frequency
f = sp.symbols('f', real=True)

# TODO: write the equation v = f * lam using SymPy's Eq
equation = sp.Eq(v, f * lam)

# TODO: solve the equation for f
solution_for_f = sp.solve(equation, f)[0]

# Convert symbolic result to float
f_value = float(solution_for_f)

print(f"Solved frequency: f = {f_value:.3f} Hz")

# -----------------------------------------
# 2) Compute omega and k
# -----------------------------------------

omega = 2 * np.pi * f_value          # TODO: angular frequency
k = 2 * np.pi / lam                  # TODO: wavenumber

print(f"ω = {omega:.3f} rad/s")
print(f"k = {k:.3f} rad/m")

# -----------------------------------------
# 3) Define y(x, t0) and create data for plotting
# -----------------------------------------

def y_xt(x, t):
    # TODO: implement y(x,t) = A * sin(k x - ω t)
    return A * np.sin(k * x - omega * t)

# Create x array
x_vals = np.linspace(0, 10, 400)     # from 0 to 10 m
y_vals = y_xt(x_vals, t0)

# -----------------------------------------
# 4) Plot y(x, t0) vs x
# -----------------------------------------

plt.figure(figsize=(8, 4))
plt.plot(x_vals, y_vals, label=r"$y(x, t_0)$ at $t_0 = 0.50\,\text{s}$")
plt.axhline(0, color='gray', lw=1, linestyle='--', alpha=0.7)

plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Snapshot of a Traveling Wave on a String")
plt.legend(loc="upper right")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
%reset -f

# Possible solution

import numpy as np

x_min = 0.0
x_max = 10.0
N = 1000     # number of points

# -----------------------------------------
# 1) Define the function y(x)
# -----------------------------------------
def y(x):
    # TODO: return the correct expression 0.10*sin(2x + 1)
    return 0.10 * np.sin(2*x + 1)


# -----------------------------------------
# 2) Create the x-array
# -----------------------------------------
# TODO: make N evenly spaced points between x_min and x_max
xs = np.linspace(x_min, x_max, N)

ys = y(xs)

# -----------------------------------------
# 3) Find zero crossings
# -----------------------------------------
zero_positions = []

for i in range(N-1):
    # A zero crossing occurs if y changes sign
    if ys[i] == 0:
        zero_positions.append(xs[i])
    elif ys[i] * ys[i+1] < 0:
        # store the midpoint between the two x-values
        zero_positions.append((xs[i] + xs[i+1]) / 2)


# -----------------------------------------
# 4) Print results
# -----------------------------------------
print("Approximate zero crossings:")
for zp in zero_positions:
    print(f"{zp:.3f}")
```
