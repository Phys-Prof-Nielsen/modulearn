## Motivation  

Many phenomena in nature — from the ripples on a pond to the vibration of a guitar string, to the sounds we hear every day — are examples of *mechanical waves*: disturbances that travel through a medium, carrying energy and information without permanently transporting matter.  

Understanding mechanical waves is essential across science and engineering.  
- In **biology**, waves describe how sound travels through the ear and how heartbeats generate pressure pulses in blood vessels.  
- In **engineering**, vibrations in bridges, airplane wings, and machinery are modeled as waves.  
- In **earth science**, seismic waves reveal the internal structure of Earth.  
- In **physics**, studying waves leads directly to understanding **sound**, **light**, and later, **quantum waves** that describe particles themselves.  

By mastering the language and mathematics of waves, we gain a universal framework to describe oscillations that propagate — a cornerstone for modern science and technology.

## Learning Outcomes (LOC)

After completing this module, students will be able to:

**1. Conceptual Understanding**  
- Distinguish between **mechanical waves** and other types of waves (e.g., electromagnetic).  
- Describe how energy and momentum are transmitted through a medium by a wave without net transport of matter.  
- Explain the relationships between **wavelength**, **frequency**, **period**, and **wave speed**.  

**2. Traveling Waves**  
- Interpret the general wave function  

  $$
  y(x,t) = A \sin(kx - \omega t + \phi)
  $$
  
  and relate its parameters to physical quantities.  
- Identify the direction of propagation from the wave equation’s sign convention.  
- Determine the speed, frequency, and wavelength of a traveling wave from experimental or graphical data.  

**3. Superposition and Interference**  
- Apply the **principle of superposition** to predict the resultant displacement from two or more waves.  
- Explain constructive and destructive interference qualitatively and quantitatively.  
- Use superposition to analyze **beats** and **wave envelopes** (introduced qualitatively here).  

**4. Application: Sound Waves**  
- Describe sound as a **longitudinal wave** in air and other media.  
- Explain qualitatively how sound waves relate to pressure variations and particle motion.  
- Interpret simple sound wave graphs (pressure vs. time, displacement vs. time).  

**5. Quantitative Skills and Modeling**  
- Calculate wave parameters from given data using relationships such as $v = f\lambda$.  
- Represent waves graphically and through simple Python models.  
- Analyze data to identify wave patterns and interference behavior.
