# B1.0 Learning Outcomes

By the end of this module, students will be able to:

## 1. Understand the Concept of Derivatives
- Explain the meaning of a derivative in the context of physics.
- Interpret derivatives as rates of change and their applications in kinematics.

## 2. Apply Differentiation Rules
- Use standard differentiation rules (sum, product, quotient, and chain rule) to compute derivatives of common functions.
- Differentiate polynomials, exponentials, trigonometric functions, and other common mathematical expressions relevant to physics.

## 3. Compute Derivatives Symbolically Using Python
- Utilize SymPy (Symbolic Python) to calculate derivatives of functions.
- Verify analytical results using symbolic computation.

## 4. Analyze Instantaneous Velocity and Acceleration
- Define and compute instantaneous velocity and acceleration using derivatives.
- Apply differentiation techniques to solve kinematic problems.

## 5. Understand Integrals as Antiderivatives and Infinite Sums
- Interpret integrals as the reverse operation of differentiation.
- Explain the concept of integration as an infinite sum.

## 6. Compute Definite and Indefinite Integrals
- Evaluate indefinite integrals of common functions.
- Calculate definite integrals and interpret their physical meaning in physics problems.

## 7. Define and Apply the Concept of Differentials
- Explain the meaning of a differential in calculus and physics.
- Use differentials in problem-solving.

## 8. Perform Integrals Symbolically Using Python
- Use SymPy to compute integrals of functions.
- Verify analytical integration results using computational tools.
