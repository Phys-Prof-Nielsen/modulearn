## Problem 1: Sunglasses and Reflected Glare

You're standing near a lake on a sunny day. Light reflects off the water surface and causes glare. Your polarized sunglasses are designed to reduce this glare.

**a)** What direction is the reflected light from the water mostly polarized in?

**b)** What orientation should your sunglasses' polarizing axis be to reduce the glare effectively?

**c)** If your sunglasses transmit only 10% of the reflected light, estimate the angle between the polarization of the reflected light and the polarizer axis.

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
import numpy as np

print('(a) When light reflects off a **horizontal surface** like water, the reflected light becomes partially or fully polarized horizontally — meaning the electric field oscillates parallel to the surface.')
print(' ')
print('(b) To block this horizontally polarized glare, the sunglasses should have a vertical transmission axis, i.e., they only pass vertically polarized light.')
print(' ')

theta = sp.Symbol('theta')

# I = 0.1*I0 = I0*cos^2(theta)
eq = 0.1 - (sp.cos(theta))**2

sol = sp.solve(eq,theta)
print('(c) Angle from sunglass axis: '+str(sol[0]*180/np.pi)+' degrees')
```

## Problem 2: Two Polarizers at an Angle

A beam of unpolarized light with intensity $ I_0 = 100 \, \text{W/m}^2 $ passes through two ideal linear polarizers.

- The **first** polarizer aligns the light along the vertical.
- The **second** polarizer is rotated **30°** relative to the vertical.

**a)** What is the intensity after the first polarizer?

**b)** What is the final transmitted intensity after the second polarizer?

**c)** What would the transmitted intensity be if a **third polarizer** at 15° were inserted between the first two?

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
import numpy as np

I0 = 100

# Part a
print('(a) Unpolarized through linear polarizer cuts intensity in half: 50 W/m\u00b2')
print(' ')

# Part b

I1 = 100/2

#apply Malus' law
theta = 30*np.pi/180

I2 = I1*(np.cos(theta)**2)

print('(b) '+str(I2)+' W/m\u00b2')
print(' ')
# Part c
theta1 = 15*np.pi/180
I2 = I1*(np.cos(theta1)**2)

theta2 = (30-15)*np.pi/180
I3 = I2*(np.cos(theta2)**2)

print('(c) '+str(I3)+' W/m\u00b2')
print(' ')
print('The final intensity is higher than without the third polarizer. Counterintuitive fact: Adding a polarizer between two crossed ones can increase transmission if it bridges their polarization angles.')
```

## Problem 3: Wavelength from Interference Pattern

A monochromatic laser beam passes through two narrow slits that are 0.25 mm apart. The interference pattern is projected onto a screen 2.0 meters away. The distance between adjacent bright fringes is measured to be 5.2 mm.

**Questions:**

a) What is the wavelength of the light?

b) What type (or approximate color) of visible light is this?

```python
# DIY Cell
```

```python
%reset -f

d = 0.25E-3
L = 2.0
ym = 5.2E-3

wavel = (ym*d)/L

print('(a) The wavelength is: '+str(wavel)+' m')
print(' ')
print('(b) The is visible light in the red part of the spectrum.')
```

## Problem 4: Finding the Slit Separation

A physics student performs a double-slit experiment using a laser with a known wavelength of 600.0 nm. The interference pattern is projected onto a screen 1.8 meters away. The distance between the central maximum and the 4th-order bright fringe is measured to be 14.4 mm.

**Questions:**

a) What is the distance between the two slits?

b) What would the fringe spacing be if the screen were moved to 3.6 meters?

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

wavel = 600E-9
L = 1.8
y4 = 14.4E-3
m = 4

# Part a
#-------
d = (m*wavel*L)/y4

answer = "{:.1e}".format(d)
print('(a) distance between slits: '+str(answer)+' m')
print(' ')

# Part b
#-------
L = 3.6

#pick a random order m
m = sp.Symbol('m')
#unknown is dy
dy = sp.Symbol('dy')

#dy = ((m+1)*wavel*L)/d - ((m)*wavel*L)/d
eq = dy - ((m+1)*wavel*L)/d + ((m)*wavel*L)/d

sol = sp.solve(eq,dy)

answer = "{:.1e}".format(sol[0])
print('(b) fringe spacing: '+str(answer)+' m')
```

## Problem 6: Soap Bubble Colors

A soap bubble (index of refraction $n = 1.33$) is illuminated with white light. The film is viewed in reflection at normal incidence. The film thickness is such that **destructive interference** occurs for **wavelength 600.0 nm** (in air).

**Questions:**

a) What is the minimum thickness of the bubble wall that causes this cancellation?

b) Which colors (shorter or longer wavelengths) will be enhanced instead?

```python
# DIY Cell
```

```python
%reset -f

n_film = 1.33
wavel_air = 600.0E-9

n_air = 1.00

# Part a
#-------
# we only have one phase shift (relfection from air to film) --> destructive is given by: 2*n_f*t = m*lambda

# now solve for t with m = 1
m = 1
t = (m*wavel_air)/(2*n_film)

answer = "{:.2e}".format(t)
print('(a) Minimum thickness: '+str(answer)+' m')
print(' ')

print('(b) Shorter wavelengths (e.g., blue, violet) will undergo constructive interference, so the bubble will appear bluish.')
```

## Problem 7: Designing an Anti-Reflection Coating

A camera lens has a glass surface ($n = 1.50$). You are asked to design a single-layer anti-reflection coating for light at wavelength 550.0 nm. The coating material has refractive index $n = 1.22$.

**Question:**
What is the minimum thickness of the coating that minimizes reflection at this wavelength?

```python
# DIY Cell
```

```python
%reset -f

n_glass = 1.50
wavel_air = 550.0E-9
n_film = 1.22

# Part a
#-------

# air-film-glass: n_air < n_film < n_glass...so two phase shift
# minimize reflection = deconstructive
# minimum: m = 0
m = 0
t = ((m+0.5)*wavel_air)/(2*n_film)

answer = "{:.2e}".format(t)
print('Minimum thickness: '+str(answer)+' m')
```

## Problem 10: Non-Minimal Thickness in Thin Film Interference

A microscope slide is coated with a transparent film of refractive index $n = 1.40$ and thickness $t = 490$ nm. When illuminated at near-normal incidence with white light, **constructive interference** is observed in reflection at a wavelength $\lambda_{\text{air}} = 560$ nm. Assume glass has index of refraction less than the film.

**Questions:**

a) What is the **interference order** $m$ that satisfies the constructive condition at this thickness?

b) Would the same thickness produce constructive interference for any shorter wavelengths?

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

n_film = 1.40
t = 490E-9
wavel_air = 560E-9

# Part a
#-------

m = sp.Symbol('m')

eq = 2*n_film*t - (m+0.5)*wavel_air

sol = sp.solve(eq,m)

m = round(sol[0])

print('Interference order: '+str(m))
print(' ')

# Part b
#-------
print('Yes — at this thickness, constructive interference will also occur at shorter wavelengths for higher interference orders m = 3, 4, ...:')
print('Thus, you might also see constructive interference at violet or blue wavelengths for higher m.')
```
