## C5.4.1 Motivation: What Happens When Waves Meet?

In ray optics, light rays just pass through each other.  
But in wave optics, **waves can combine** — and this leads to amazing effects:

- **Fringes** of bright and dark bands
- **Color patterns** in soap films or oil slicks
- Laser interference used in measuring atomic distances and gravitational waves

This is called **interference**, and it arises naturally when **waves overlap in space and time**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.4.2 What Is Interference?

> Interference is the phenomenon where two or more waves **superpose** to form a new wave pattern.

This is a **core feature of all wave phenomena**.

### Principle of Superposition

If two waves have electric fields:

$$
E_1(t) = E_0 \cos(\omega t), \quad E_2(t) = E_0 \cos(\omega t + \phi)
$$

Then the total field is:

$$
E_{\text{total}}(t) = E_1 + E_2 = 2E_0 \cos\left( \frac{\phi}{2} \right) \cos\left( \omega t + \frac{\phi}{2} \right)
$$

And the intensity (which is proportional to the **square of the amplitude**) becomes:

$$
I \propto \left( 2E_0 \cos\left( \frac{\phi}{2} \right) \right)^2 = 4E_0^2 \cos^2\left( \frac{\phi}{2} \right)
$$

### Conditions for Interference

To observe stable interference patterns, the sources must be:

- **Coherent** (constant phase difference)
- **Monochromatic** (same frequency)
- **Same polarization direction**

## C5.4.3 Young’s Double-Slit Experiment (1801)

### Historical Context

In the early 1800s, scientists debated the nature of light:
- **Isaac Newton** believed light was made of particles ("corpuscles").
- **Christiaan Huygens** had proposed in the 1600s that light behaves as a **wave**.

But Huygens’ idea had largely been set aside due to Newton’s influence — until **Thomas Young** changed everything.

In **1801**, Young performed a groundbreaking experiment to test the wave theory of light.

### The Experiment

Young allowed sunlight to pass through a narrow slit and then through **two closely spaced pinholes (or slits)** in an opaque screen. On a screen behind the slits, he observed:

- A series of **bright and dark fringes** — a pattern **impossible** to explain using particles alone.
- The only plausible explanation: **interference** of waves.

This confirmed that light exhibits **wave behavior**, supporting Huygens’ principle.

Young allowed sunlight to pass through a narrow **single slit** before it reached the two slits.  
**Why?**

> The single slit ensures that the light reaching both slits is **spatially coherent** — that is, it comes from the same point on the wavefront and has a well-defined phase relationship.

Without this step:
- The light hitting each slit could originate from different parts of the Sun’s extended surface.
- This would lead to **incoherence**, washing out the interference pattern.

By using the single slit first, Young effectively created a **coherent source** for both slits — an essential requirement for stable and visible interference fringes.

### Why It Mattered

Young’s result was **revolutionary**:
- It revived the **wave theory of light**.
- It provided the **first accurate measurement** of the **wavelength** of visible light.
- It laid the groundwork for **modern wave optics** and helped overturn Newton’s corpuscular theory.

## C5.4.4 Mathematical Model of the Double-Slit Interference

Let:
- $d$ = distance between the slits  
- $L$ = distance to the screen  
- $x$ = position on the screen  
- $\lambda$ = wavelength of light  

Under the small-angle approximation, the **path difference** $\Delta r$ between light from the two slits at a point $x$ is:

$$
\Delta r = \frac{xd}{L}
$$

This path difference leads to a **phase difference** $\phi$:

$$
\phi = \frac{2\pi}{\lambda} \cdot \frac{xd}{L}
$$

### Constructive and Destructive Interference

- **Constructive (bright) fringes**: $\Delta r = m\lambda \Rightarrow x_m = \frac{m\lambda L}{d}$
- **Destructive (dark) fringes**: $\Delta r = \left(m + \frac{1}{2}\right)\lambda \Rightarrow x_m = \frac{(m + 0.5)\lambda L}{d}$

where $m = 0, \pm1, \pm2, \ldots$

Below is an animation illustrating the interferece of the wavelets using Huygen's principle.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

# Parameters
wavelength = 500e-9  # Not used directly; for reference
slit_separation = 1.0      # distance between slits (arbitrary units)
screen_distance = 5.0      # distance from slits to screen
num_wavefronts = 40        # number of wavelets emitted
slit_y = 0                 # y-coordinate of slits

# Slit positions
slit1 = np.array([-slit_separation / 2, slit_y])
slit2 = np.array([slit_separation / 2, slit_y])

# Screen x-coordinates
screen_x = np.linspace(-3, 3, 300)
screen_y = np.full_like(screen_x, screen_distance)

# Create figure
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlim(-4, 4)
ax.set_ylim(-1, 6)
ax.set_aspect('equal')
ax.set_title("Huygens' Wavelets from Double Slits to Screen")
ax.set_xlabel("x (arbitrary units)")
ax.set_ylabel("y (arbitrary units)")

# Draw slits and screen
ax.plot([-slit_separation / 2, slit_separation / 2], [0, 0], 'ko', markersize=8, label='Slits')
ax.plot(screen_x, screen_y, 'k-', lw=1, label='Screen')

# Initialize circular wavelets
wavelet_circles_1 = [plt.Circle(slit1, 0, color='blue', fill=False, lw=0.5) for _ in range(num_wavefronts)]
wavelet_circles_2 = [plt.Circle(slit2, 0, color='green', fill=False, lw=0.5) for _ in range(num_wavefronts)]
for circle in wavelet_circles_1 + wavelet_circles_2:
    ax.add_patch(circle)

def init():
    for circle in wavelet_circles_1 + wavelet_circles_2:
        circle.set_radius(0)
    return wavelet_circles_1 + wavelet_circles_2

def animate(i):
    for j, circle in enumerate(wavelet_circles_1):
        radius = (i - j) * 0.1
        circle.set_radius(radius if radius > 0 else 0)
    for j, circle in enumerate(wavelet_circles_2):
        radius = (i - j) * 0.1
        circle.set_radius(radius if radius > 0 else 0)
    return wavelet_circles_1 + wavelet_circles_2

ani = animation.FuncAnimation(fig, animate, frames=num_wavefronts + 10,
                              init_func=init, blit=True, interval=100)

HTML(ani.to_jshtml())
```

### Intensity Distribution on the Screen

Let each slit emit a wave:

$$
E_1(t) = E_0 \cos(\omega t), \quad E_2(t) = E_0 \cos(\omega t + \phi)
$$

Using a trig identity, the superposed electric field is:

$$
E(t) = E_1 + E_2 = 2E_0 \cos\left(\frac{\phi}{2}\right) \cos\left(\omega t + \frac{\phi}{2}\right)
$$

The time-averaged **intensity** is proportional to the square of the amplitude:

$$
I(x) = 4I_0 \cos^2\left(\frac{\pi d x}{\lambda L}\right)
$$

This creates the familiar **fringe pattern** on the screen, with peak spacing:

$$
\Delta x = \frac{\lambda L}{d}
$$

- Larger $\lambda$ → wider fringes  
- Larger $L$ → wider fringes  
- Larger $d$ → **narrower** fringes

Below is an animation illustrating the dependency of the intensity pattern as a function of wavelength.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

# Parameters
d = 0.5e-3           # slit separation (0.5 mm)
L = 1.0              # distance to screen (1 m)
I0 = 1.0             # max intensity
x = np.linspace(-0.02, 0.02, 1000)  # screen range: +/- 2 cm

# Interference pattern function
def interference_intensity(x, wavelength):
    delta_r = d * x / L
    phi = 2 * np.pi * delta_r / wavelength
    return I0 * np.cos(phi / 2)**2

# Set up the plot
fig, ax = plt.subplots(figsize=(8, 4))
line, = ax.plot([], [], lw=2)
ax.set_xlim(x.min(), x.max())
ax.set_ylim(0, 1.1 * I0)
ax.set_xlabel("Position on screen (m)")
ax.set_ylabel("Intensity (arbitrary units)")

# Wavelengths to animate (400–700 nm)
wavelengths = np.linspace(400e-9, 700e-9, 60)

def init():
    line.set_data([], [])
    return line,

def animate(i):
    wl = wavelengths[i]
    I = interference_intensity(x, wl)
    line.set_data(x, I)
    ax.set_title(f"Double-Slit Interference: Wavelength = {int(wl * 1e9)} nm")
    return line,

ani = animation.FuncAnimation(fig, animate, frames=len(wavelengths),
                              init_func=init, blit=True, interval=100)

HTML(ani.to_jshtml())
```

## C5.4.5 Connection to Quantum Mechanics

> The double-slit experiment is not just a classic wave-optics demonstration — it lies at the heart of modern quantum theory.

- In the 20th century, scientists found that **electrons**, **atoms**, and even **molecules** show the same interference pattern if sent through a double slit.
- Remarkably, this happens even when **particles are sent one at a time**:
  - Over time, an interference pattern **emerges**.
  - Each particle behaves like a **wave that interferes with itself** — even if no one watches.

This experiment with single electrons or photons is known as the **quantum double-slit experiment**, and it challenges classical ideas of particles and measurement.

### Key Implications

- Light and matter have **wave-particle duality**.
- The pattern forms only if the **which-path information** is unknown.
- This led to foundational principles in **quantum mechanics**, including the role of measurement and superposition.

### What is "Which-Path Information"?

In the **double-slit experiment**, we say we have *which-path information* if we can, even in principle, determine **which slit** a particle (like a photon or electron) passed through.

- If a detector is placed near one of the slits and can tell you “The particle went through slit A,” then you **have** which-path information.
- If there's **no way to tell** which slit it went through — the paths are **indistinguishable** — then you **do not** have which-path information.

#### Why It Matters

- **No which-path information → Interference pattern appears.**  
  The particle behaves like a **delocalized wave**, going through **both slits at once**, and interfering with itself.

- **Which-path information available → No interference pattern.**  
  As soon as you can tell which slit it passed through, the wave-like behavior **disappears**, and you just get two bright blobs (as expected from particles).

#### In Quantum Mechanics Terms

The **wavefunction** of the particle includes both paths:
$$
\Psi = \Psi_A + \Psi_B
$$

The corresponding probability distribution is:
$$
|\Psi|^2 = |\Psi_A + \Psi_B|^2 = |\Psi_A|^2 + |\Psi_B|^2 + 2\text{Re}(\Psi_A^* \Psi_B)
$$

which includes an **interference term**.

But if we measure which slit it went through, the wavefunction **collapses** to just one path:
$$
|\Psi|^2 = |\Psi_A|^2 \quad \text{or} \quad |\Psi_B|^2
$$

→ **No interference**.

#### Takeaway

> **Which-path information destroys interference.**  
Even the *possibility* of obtaining that information — without actually measuring — is enough to erase the interference.

This is one of the most profound insights of quantum mechanics:  
**What can be known changes what is.**
