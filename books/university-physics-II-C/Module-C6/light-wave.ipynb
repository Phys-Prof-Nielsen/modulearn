## C6.1.1 Motivation: Why Go Beyond Ray Optics?

In **Phase D**, we treated light as rays — straight lines that refract and reflect, letting us model lenses, mirrors, and image formation.

But some phenomena cannot be explained by ray optics:

- Why do polarized sunglasses work?
- Why do soap bubbles show colorful patterns?
- Why do CDs and DVDs shimmer with rainbow-like colors?
- Why does light bend around edges, producing shadows that aren’t sharp?

These questions lead us to **wave optics**, a deeper understanding of light's nature that goes beyond the straight-line approximation.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.2 Historical Motivation

Historically, light was long thought to be a particle (Newton's corpuscular theory). But:

- Thomas Young’s double-slit experiment (1801) showed **interference** patterns, possible only if light is a wave.
- Augustin-Jean Fresnel developed a wave theory of diffraction.
- James Clerk Maxwell later showed that **light is an electromagnetic wave** — a solution to Maxwell’s equations.

Thus, the **wave nature of light** explains behavior that ray optics cannot.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.3 Light as an Electromagnetic Wave

Light is a transverse electromagnetic wave. This means:

- The electric field $\vec{E}$ and magnetic field $\vec{B}$ oscillate perpendicular to each other and to the direction of propagation.
- The speed of propagation in vacuum is:

  $$
  c = \frac{1}{\sqrt{\mu_0 \varepsilon_0}} \approx 3.00 \times 10^8 \ \text{m/s}
  $$

Let’s consider a linearly polarized plane wave moving in the $+z$ direction:
- $\vec{E}$ oscillates in $x$-direction: $\vec{E}(z,t) = E_0 \cos(kz - \omega t)\hat{x}$
- $\vec{B}$ oscillates in $y$-direction: $\vec{B}(z,t) = B_0 \cos(kz - \omega t)\hat{y}$

Here:
- $k = \frac{2\pi}{\lambda}$ is the **wave number**
- $\omega = 2\pi f$ is the **angular frequency**
- $c = \lambda f$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.4 Wave Properties of Light

Let’s recall some key wave properties that will become central:

### 1. **Interference**
When two waves overlap in space, their electric fields add:
- **Constructive interference**: amplitudes reinforce
- **Destructive interference**: amplitudes cancel

This gives rise to **fringes**, patterns of alternating brightness.

### 2. **Diffraction**
Waves bend around edges and spread after passing through narrow slits. This is deeply tied to **Huygens’ principle**, which we’ll explore later.

### 3. **Polarization**
Only transverse waves can be polarized — another indicator that light behaves like a wave. We can control the **direction of the electric field**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.5 Mathematical Description

A general form for a **monochromatic plane wave** traveling in the $z$-direction:

$$
\vec{E}(z,t) = E_0 \cos(kz - \omega t + \phi) \hat{x}
$$

We often use the complex representation for convenience:

$$
\vec{E}(z,t) = \text{Re}\left[ \vec{E}_0 e^{i(kz - \omega t)} \right]
$$

Key relationships:
- $c = \lambda f$
- $k = \frac{2\pi}{\lambda}$
- $\omega = 2\pi f$

This will let us derive **phase differences**, **path differences**, and **intensity patterns** — the mathematical core of interference and diffraction.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.6 Energy and Intensity of Light Waves

The **intensity** of a wave is proportional to the square of the electric field amplitude:

$$
I \propto E_0^2
$$

For time-averaged intensity of a sinusoidal wave:

$$
I = \frac{1}{2} \varepsilon_0 c E_0^2
$$

This concept of intensity helps quantify brightness in interference and diffraction patterns.

## C6.1.8 Summary

- Light is an electromagnetic wave with transverse electric and magnetic fields.
- Wave optics explains interference, diffraction, and polarization — phenomena ray optics cannot.
- Understanding light as a wave allows us to model and predict new behaviors that are essential in modern technology and physics.
