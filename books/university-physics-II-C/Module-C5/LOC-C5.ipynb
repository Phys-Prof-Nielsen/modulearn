# C5.0 Learning Outcomes — Magnetic Force on a Current-Carrying Conductor
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this module matters

In Phase D, we studied the magnetic force on a **single moving charge**. In this module, we extend that idea to systems containing **many moving charges simultaneously** — that is, electric currents.

A current is simply organized motion of charge. When a current flows through a conductor placed in a magnetic field, each charge experiences a magnetic force. The combined microscopic forces result in a measurable **macroscopic force on the wire itself**.

This principle is foundational for:

- electric motors,
- loudspeakers,
- magnetic actuators,
- rail guns,
- and many other electromagnetic devices.

At the C-level, the emphasis is on:

- connecting microscopic Lorentz force to macroscopic conductor force,
- expressing magnetic force on conductors in compact vector form,
- and analyzing forces on straight and curved wires in magnetic fields.

This module builds the conceptual bridge from particle-level magnetism to current-based electromagnetic systems.

By the end of this module, you will be able to:

---

### A. From Moving Charges to Current  
*Builds the microscopic-to-macroscopic connection.*

- **A1.** Explain how the magnetic force on individual moving charges leads to a net force on a current-carrying conductor.
- **A2.** Define electric current as  

  $$
  I = nqAv_d,
  $$

  and interpret each quantity physically.
- **A3.** Relate drift velocity and charge density to the total magnetic force on a conductor.
- **A4.** Demonstrate explicitly how the Lorentz force on charges leads to a force proportional to current.

---

### B. Magnetic Force on a Straight Current-Carrying Wire  
*Develops the standard force relationship.*

- **B1.** Write the magnitude of the magnetic force on a straight wire segment of length $L$ in a uniform magnetic field:

  $$
  F = I L B \sin\theta,
  $$

  where $\theta$ is the angle between the wire direction and $\vec{B}$.
- **B2.** Express the force in full vector form:

  $$
  \vec{F} = I \vec{L} \times \vec{B}.
  $$

- **B3.** Use the right-hand rule to determine the direction of the force on a current.
- **B4.** Recognize limiting cases:
  - maximum force when perpendicular,
  - zero force when parallel.

---

### C. Magnetic Force on Curved Conductors  
*Generalizes the force law.*

- **C1.** Write the differential magnetic force on a current element:

  $$
  d\vec{F} = I\, d\vec{\ell} \times \vec{B}.
  $$

- **C2.** Set up integrals to determine the net force on a curved wire in a magnetic field.
- **C3.** Recognize that the net force on a closed loop in a **uniform** magnetic field is zero.
- **C4.** Interpret qualitatively how non-uniform fields can produce net forces on loops.

---

### D. Conceptual Connections and Physical Interpretation  
*Strengthens structural understanding.*

- **D1.** Clearly distinguish between:
  - magnetic force on a single charge,
  - magnetic force on a current-carrying conductor.
- **D2.** Identify the structural similarity between  

  $$
  \vec{F} = q \vec{v} \times \vec{B}
  \quad \text{and} \quad
  \vec{F} = I \vec{L} \times \vec{B}.
  $$

- **D3.** Explain physically why magnetic forces can produce mechanical motion in conductors.
- **D4.** Recognize how this force law forms the basis for electromagnetic devices studied later.

---

### Performance Criteria

- The macroscopic force law is explicitly derived from microscopic Lorentz forces.
- Vector cross products are used correctly for magnitude and direction.
- Integration for curved conductors is set up clearly and physically.
- Conceptual distinctions between charge motion and current motion are precise.
- Physical interpretations connect mathematics to real devices.
