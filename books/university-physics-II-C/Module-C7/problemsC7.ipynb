## Problem 1: When and Where Did It Happen?

**Scenario**:  
Two inertial frames:
- Frame S (lab frame)
- Frame S', moving to the right at speed $v = 0.6c$ relative to S.

An explosion occurs in frame S' at position $x' = 900\ \text{m}$ and time $t' = 4\ \mu\text{s}$.

**Question**:  

Determine the coordinates $(x, t)$ of the event in the lab frame S.

```{admonition} Problem 1: Solution
:class: tip, dropdown

**Solution**:

**Given**:
- $v = 0.6c \Rightarrow \beta = 0.6$
- $\gamma = \frac{1}{\sqrt{1 - \beta^2}} = \frac{1}{\sqrt{0.64}} = 1.25$

Use the inverse Lorentz transformations:

$$
\begin{aligned}
x &= \gamma (x' + v t') \\
t &= \gamma \left(t' + \frac{v x'}{c^2} \right)
\end{aligned}
$$

Convert time to seconds:
- $t' = 4\ \mu\text{s} = 4 \times 10^{-6}\ \text{s}$

---

**Step 1: Position in S**

$$
v t' = 0.6 \cdot 3 \times 10^8 \cdot 4 \times 10^{-6} = 720\ \text{m}
$$

$$
x = 1.25 (900 + 720) = 1.25 \cdot 1620 = 2025\ \text{m}
$$

---

**Step 2: Time in S**

$$
t = 1.25 \left(4 \times 10^{-6} + \frac{(0.6c)(900)}{c^2} \right)
= 1.25 \left(4 \times 10^{-6} + \frac{540}{3 \times 10^8} \right)
= 1.25 \left(4 + 1.8 \right)\ \mu\text{s}
= 7.25\ \mu\text{s}
$$

---

✅ **Answer**:  
The event occurs at $x = 2025$ m, $t = 7.25\ \mu$s in the lab frame.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 2: Are They Simultaneous?

**Scenario**:  
In the S frame (lab), two events happen:

- **Event A**: at $x = 0$, $t = 0$
- **Event B**: at $x = 300$ m, $t = 0$

Frame S' moves at $v = 0.8c$ to the right.

**Question**:  

What are the times of the two events in S'? Are they simultaneous in that frame?

```{admonition} Problem 2: Solution
:class: tip, dropdown

**Solution**:

Given:
- $\beta = 0.8$
- $\gamma = \frac{1}{\sqrt{1 - 0.64}} = 1.6667$

Lorentz time transformation:

$$
t' = \gamma \left(t - \frac{v x}{c^2} \right)
$$

---

**Event A ($x = 0,\ t = 0$)**:

$$
t'_A = \gamma (0 - 0) = 0
$$

---

**Event B ($x = 300\ \text{m},\ t = 0$)**:

$$
t'_B = -\gamma \cdot \frac{v x}{c^2}
= -1.6667 \cdot \frac{0.8 \cdot 300}{3 \times 10^8}
= -1.6667 \cdot 8 \times 10^{-7}
= -1.33\ \mu\text{s}
$$

---

✅ **Answer**:  
- $t'_A = 0$  
- $t'_B = -1.33\ \mu$s  
⟶ The events are **not simultaneous** in S′.

```

## Problem 3: When Did the Laser Hit?

Luke Skywalker is aboard a stationary X-Wing, watching a **TIE fighter** zoom past at $v = 0.9c$ along the $x$-axis.  
The moment their ships pass each other ($x = x' = 0$) is defined as $t = t' = 0$.

From the **TIE fighter’s rest frame (S')**, a laser is fired **from the rear of the fighter** at $x' = -270\ \text{m}$ at time $t' = 2\ \mu\text{s}$. It travels toward the front to hit Luke.

From Luke's frame (**S**), when and where was the laser fired?

```{admonition} Problem 3: Solution
:class: tip, dropdown

**Solution**

**Given**:

- $v = 0.9c \Rightarrow \beta = 0.9$
- $\gamma = \frac{1}{\sqrt{1 - \beta^2}} = \frac{1}{\sqrt{1 - 0.81}} = \frac{1}{\sqrt{0.19}} \approx 2.294$
- Event in S':
  - $x' = -270\ \text{m}$
  - $t' = 2\ \mu\text{s} = 2 \times 10^{-6}\ \text{s}$

---

**Lorentz Coordinate Transformation (S′ → S)**

We use:

$$
\begin{aligned}
x &= \gamma (x' + v t') \\
t &= \gamma \left(t' + \frac{v x'}{c^2} \right)
\end{aligned}
$$

---

**Step 1: Position in Luke's Frame**

$$
x = 2.294 \cdot \left(-270 + 0.9c \cdot 2 \times 10^{-6} \right)
$$

Compute $0.9c \cdot 2 \times 10^{-6}$:

$$
0.9 \cdot 3 \times 10^8 \cdot 2 \times 10^{-6} = 540\ \text{m}
$$

$$
x = 2.294 \cdot (-270 + 540) = 2.294 \cdot 270 = 619.4\ \text{m}
$$

---
**Step 2: Time in Luke's Frame**

$$
t = 2.294 \cdot \left(2 \times 10^{-6} + \frac{0.9 \cdot (-270)}{3 \times 10^8} \right)
$$

$$
\frac{-243}{3 \times 10^8} = -8.1 \times 10^{-7}
$$

$$
t = 2.294 \cdot (2 \times 10^{-6} - 8.1 \times 10^{-7}) = 2.294 \cdot (1.19 \times 10^{-6}) \approx 2.73\ \mu\text{s}
$$

---

**✅ Final Answer**:

- The laser was fired at position $x = 619.4$ m  
- and time $t = 2.73\ \mu\text{s}$ in Luke's frame (S)

---

**Interpretation**

To Luke, the laser was fired **ahead of him**, not behind, because the rear of the TIE fighter was **already moving forward** when the laser was released.

📌 This is a powerful illustration of:
- How events can shift in space and time depending on the observer.
- Why simultaneity and spatial order are not absolute in special relativity.
```

## Problem 5: The TIE Fighter Chase

A **TIE fighter** is escaping to the right at $v = 0.9c$ relative to a Rebel X-Wing.

Inside the TIE fighter, a **targeting beam** is fired **backward** (toward the left) at speed $u'_x = -0.3c$ relative to the TIE fighter.

What is the speed and direction of the beam as observed by the **Rebel pilot** in frame S?

```{admonition} Problem 5: Solution
:class: tip, dropdown

**Solution**:

Apply the 1D velocity transformation:

$$
u_x = \frac{u'_x + v}{1 + \frac{u'_x v}{c^2}} = \frac{-0.3c + 0.9c}{1 + \frac{(-0.3)(0.9)c^2}{c^2}} = \frac{0.6c}{1 - 0.27} = \frac{0.6c}{0.73} \approx 0.822c
$$

---

✅ **Answer**:  

The Rebel pilot sees the beam moving forward (in the $+x$ direction) at approximately $0.822c$ — **even though it was fired backward** in the TIE fighter’s frame.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">
