## C7.7.1 Motivation

Nuclear energy is one of the most powerful and efficient sources of energy known to humanity. Unlike chemical reactions, which involve electrons in atoms, **nuclear reactions** release energy by rearranging the nucleons (protons and neutrons) inside atomic nuclei. This can result in **millions of times more energy** per reaction compared to chemical processes like combustion.

Understanding how nuclear energy works allows us to:
- Harness power in nuclear reactors
- Develop medical imaging and cancer treatments
- Explain the energy output of stars
- Explore clean alternatives to fossil fuels
- Grasp the physics of nuclear weapons

The energy comes not from the particles themselves, but from their **binding energy**, and ultimately from the most famous equation in physics:

$$
E = mc^2
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.7.2 Mass-Energy Equivalence

Einstein’s equation tells us that **mass and energy are interchangeable**:

- In nuclear reactions, the total mass of the products is **less** than that of the reactants.
- This “missing mass” is converted into energy:
  
  $$
  \Delta E = \Delta m \, c^2
  $$

Even a tiny mass difference yields huge energy due to the factor $c^2$.

## C7.7.3 Nuclear Fission

### What is Fission?

**Fission** is the splitting of a heavy nucleus into two lighter nuclei, along with free neutrons and a large release of energy.

Example:

$$
{}^{235}_{92}\text{U} + n \rightarrow {}^{141}_{56}\text{Ba} + {}^{92}_{36}\text{Kr} + 3n + \text{energy}
$$

- Triggered by a neutron absorption
- Releases $\sim 200 \, \text{MeV}$ per fission event
- Neutrons can induce **chain reactions**

### Fission Chain Reaction

Each fission releases neutrons that can trigger more fissions, potentially causing an exponential energy release.

- **Controlled** chain reaction → nuclear reactor
- **Uncontrolled** chain reaction → nuclear bomb

### Fission Energy per Kilogram

1 kg of Uranium-235 yields roughly:

$$
E = 8.2 \times 10^{13} \, \text{J}
$$

This is millions of times more than burning 1 kg of coal.

## C7.7.4 Nuclear Fusion

### What is Fusion?

**Fusion** is the process where two light nuclei **combine** to form a heavier nucleus, releasing energy.

Example (in stars):

$$
{}^1_1\text{H} + {}^1_1\text{H} \rightarrow {}^2_1\text{H} + e^+ + \nu_e
$$

Eventually forming:

$$
{}^4_2\text{He} + \text{energy}
$$

### Why Fusion Releases Energy

- The products are **more tightly bound** than the reactants
- Mass is lost and converted into energy

Fusion of hydrogen into helium releases:

- About $26.7 \, \text{MeV}$ per event
- About 0.7% of mass is converted into energy

### Fusion in the Sun

In the Sun:
- 600 million tons of hydrogen are converted each second
- The mass loss of 4 million tons per second appears as radiant energy

## C7.7.5 Fission vs Fusion

| Feature             | Fission                          | Fusion                           |
|---------------------|----------------------------------|----------------------------------|
| Nuclei              | Heavy nuclei (e.g., U-235)       | Light nuclei (e.g., H, D, T)     |
| Process             | Splitting                        | Combining                        |
| Conditions          | Neutron absorption               | High temperature & pressure      |
| Energy per kg       | Very high                        | Even higher                      |
| Byproducts          | Radioactive waste                | Helium (non-toxic)               |
| Control             | Achieved in reactors             | Still experimental (on Earth)    |

## C7.7.6 Applications of Nuclear Energy

### Nuclear Reactors

- Use **controlled fission** to generate heat
- Heat boils water → steam spins turbines → electricity

### Nuclear Weapons

- Use **uncontrolled fission** or **fusion** for massive explosions
- Fusion weapons (“hydrogen bombs”) use fission as a trigger

### Medical Applications

- Radiation therapy for cancer
- Isotope tracers for imaging (e.g., PET scans)

### Space & Research

- RTGs (radioisotope thermoelectric generators)
- Particle physics experiments

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.7.6 Safety and Challenges

### Fission Challenges

- Long-lived **radioactive waste**
- Potential for **meltdowns** (e.g., Chernobyl, Fukushima)
- Weapon proliferation

### Fusion Challenges

- Requires **extreme temperature and confinement**
- No sustained net-energy reactors yet, but progress is ongoing (e.g., ITER)
