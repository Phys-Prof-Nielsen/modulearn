## C6.1.1 Galilean Relativity: The Classical Worldview

In the 17th century, **Galileo Galilei** introduced a principle that would underpin classical physics:  
> The laws of mechanics are the same in all inertial (non-accelerating) frames of reference.

This principle was formalized through **Galilean transformations**, which relate the coordinates of two observers moving at a constant velocity relative to each other:

$$
x' = x - vt, \quad y' = y, \quad z' = z, \quad t' = t
$$

These transformations assume:
- **Time is absolute** — the same for all observers.
- **Lengths and durations are invariant**.
- **Simultaneity is universal**.

For over two centuries, this framework worked seamlessly with **Newtonian mechanics**. But as the 19th century unfolded, cracks began to show — particularly in the newly unified field of **electromagnetism**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.2 Enter Electromagnetism: Maxwell’s Equations Change the Game

In the mid-1800s, **James Clerk Maxwell** synthesized decades of experimental results into four elegant equations describing electric and magnetic fields. These equations predicted something astonishing:

> Light is an electromagnetic wave that travels through space at a fixed speed 

$$
 c = \frac{1}{\sqrt{\mu_0 \varepsilon_0}} \approx 3.00 \times 10^8 \ \text{m/s}
 $$

This result was revolutionary — but also problematic:
- **Maxwell’s equations predicted a fixed speed of light,** regardless of the motion of the source or observer.
- This directly **contradicted Galilean relativity**, which allowed velocities to simply add: 

  $$ c' = c \pm v $$

To reconcile this, physicists hypothesized the existence of a medium called the **luminiferous aether** — a stationary "substance" through which light propagated, like air for sound.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.3 The Michelson-Morley Experiment (1887): Aether on Trial

**Albert Michelson** and **Edward Morley** designed a brilliant interferometric experiment to detect the Earth's motion through the aether by measuring the speed of light in different directions.

### Prediction:
- Light moving along the direction of Earth's motion should travel at a different speed than light moving perpendicular to it, due to the "aether wind."

### Result:
- **Null result.** No variation in the speed of light was detected — no matter the time of day or year.

This result stunned the scientific community. If light’s speed didn’t depend on Earth’s motion:
- Was the aether undetectable?
- Or did the aether **not exist at all**?

And if light’s speed is always **constant**, then Galilean transformations — and thus Newtonian mechanics — must fail.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.4 The E&M Paradox Deepens

Another inconsistency came from trying to apply **Galilean transformations** to Maxwell’s equations:

- When transformed to a moving frame using Galilean rules, **Maxwell’s equations don’t retain their form**.
- This violated the principle that physical laws should be the same in all inertial frames.

In contrast, **Newton’s laws** *do* remain valid under Galilean transformations. So:
- Something had to give — and it was the old notion of absolute time and space.

This mathematical incompatibility was as damning as the experimental failures.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.5 Lorentz’s Attempt to Save the Aether

To explain the Michelson-Morley result while preserving the aether theory, **Hendrik Lorentz** (building on ideas from **George FitzGerald**) proposed:

- **Length contraction**: objects physically contract in the direction of motion relative to the aether.
- **Time dilation (in disguise)**: moving clocks slow down — though Lorentz still believed in an absolute time underneath it all.

He developed the **Lorentz transformations** to ensure that Maxwell’s equations would stay invariant between inertial frames — but he still clung to the existence of the aether.

So while Lorentz’s math worked, his **interpretation was conservative**: he treated the transformations as physical effects caused by motion through a real, undetectable medium.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.6 Einstein’s Radical Leap (1905): A New Foundation

At age 26, **Albert Einstein** published a paper titled *“[On the Electrodynamics of Moving Bodies](https://archive.org/details/einstein-1905-relativity/page/24/mode/2up).”* He proposed a bold, elegant solution:

### Einstein’s Two Postulates of Special Relativity:
1. **The laws of physics are the same in all inertial frames.**
2. **The speed of light in vacuum is constant in all inertial frames, independent of the motion of the source or observer.**

Einstein **eliminated the aether entirely**. He viewed Lorentz’s transformations not as a trick to preserve Maxwell’s equations, but as a deep symmetry of nature — a consequence of the structure of spacetime itself.

This new viewpoint yielded:
- **Time dilation**
- **Length contraction**
- **Relativity of simultaneity**
- **Mass-energy equivalence**: $ E = mc^2 $

Importantly, **Einstein’s interpretation was physical**. These weren’t distortions due to motion through an invisible medium — they were properties of space and time themselves.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.8 Conclusion

The development of **electromagnetism** played a central role in the collapse of Newtonian ideas of space and time. It was the **incompatibility between Maxwell’s equations and Galilean relativity**, alongside the **null result of the Michelson-Morley experiment**, that forced physicists to reconsider the foundations of physics.

Lorentz provided a critical stepping stone with his transformations, but it was Einstein who reinterpreted them and built a new vision of reality — one in which space and time are intertwined, and light sets the ultimate speed limit of the universe.
