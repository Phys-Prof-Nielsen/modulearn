# C7.3 Beyond Bohr: What the Bohr Model Gets Right — and What It Misses

## C7.3.1 Recap: What Bohr's Model Achieved

The Bohr model was a bold and revolutionary step toward understanding atomic structure. With just a few postulates, it explained:

- Why atoms are stable
- Why hydrogen emits discrete spectral lines
- Why the Rydberg formula works
- How energy levels depend on a principal quantum number $n$
- The concept of quantum jumps and photon emission

But Bohr’s model is only the beginning of quantum atomic theory. It successfully models **hydrogen**, but fails or lacks depth in many areas. Let’s look at what it leaves out — many of which we’ll tackle in **Phase B**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.2 What the Bohr Model Leaves Out

Despite its successes, Bohr's model has **serious limitations**. It’s not a complete quantum theory — more like a hybrid between classical orbits and quantum rules. Here’s what it **doesn’t** include:

### ❌ 1. No Explanation of Transition Probabilities

Bohr’s model tells us *which* energy levels exist and *how much* energy a photon has during a transition — but not **how likely** the transition is or **how fast** it occurs.

In full quantum mechanics, we define:

### Einstein Coefficients

- **$A_{21}$**: Probability per unit time of **spontaneous emission** (from level 2 to 1)
- **$B_{12}$**: Probability of **absorption** due to an external photon
- **$B_{21}$**: Probability of **stimulated emission**, where a photon triggers emission of a second identical photon

These coefficients are essential for:

- Understanding the **intensity** of spectral lines
- Describing **population inversion** and **laser physics**
- Explaining why **not all transitions are equally probable**

Bohr’s model predicts photon energy, but **not the dynamics of emission**.

---

### ❌ No Selection Rules

Bohr’s model implies that an electron can jump between any two levels $n_i \rightarrow n_f$. But in real atoms, **symmetry and conservation laws** restrict transitions.

In quantum mechanics, **electric dipole transitions** obey strict **selection rules**:

- Change in orbital angular momentum: $\Delta \ell = \pm 1$
- Change in magnetic quantum number: $\Delta m = 0, \pm 1$
- Transitions with $\Delta \ell = 0$ or greater than $\pm 1$ are **forbidden** or **highly suppressed**

These rules come from **angular momentum algebra**, **parity**, and **group theory** — all absent in the Bohr framework.

---

### ❌ No Spin or Fine Structure

Bohr's model only uses $n$, the principal quantum number. But in reality:

- **Electron spin** is intrinsic angular momentum ($s = \pm \frac{1}{2}$)
- **Spin–orbit coupling** causes **fine structure splitting** — closely spaced levels due to interaction between electron’s magnetic moment and its orbital motion
- This explains **doublets** in spectra (e.g. sodium D-lines)

Fine structure energy corrections arise from **relativistic effects**, **Dirac theory**, and magnetic interactions — beyond the Bohr picture.

---

### ❌ No Account of Line Widths or Lifetimes

Spectral lines have **finite width** due to quantum uncertainty. Bohr’s model predicts **energy differences**, but doesn’t address:

- **State lifetimes** — how long electrons remain in excited states
- **Natural linewidth** — due to time–energy uncertainty: $\Delta E \cdot \Delta t \sim \hbar$
- **Broadening mechanisms**: Doppler broadening, collisional broadening, pressure effects

This connects directly to **probabilistic decay** and **wave packet spreading**.

---

### ❌ No Multi-Electron Physics: Shielding, Exchange, Pauli Principle

Bohr’s model is exact for **hydrogen**, and approximately works for **hydrogen-like ions** (He⁺, Li²⁺). It fails completely for atoms with more than one electron.

In multi-electron atoms:

### Electron Shielding

- Inner electrons **partially block** the nuclear charge felt by outer electrons
- Leads to **effective nuclear charge** ($Z_{\text{eff}}$) that depends on both $n$ and $\ell$

### Pauli Exclusion and Exchange Interaction

- No two electrons can occupy the same quantum state
- Introduces **shell structure**, **Hund’s rules**, and **term splitting**

### Orbital Penetration and Ordering

- $4s$ fills before $3d$ due to penetration effects
- Shell filling can’t be explained by Bohr’s model — needs **quantum numbers** $n$, $\ell$, $m$, and $s$

---

### ❌ No Response to External Fields: Stark and Zeeman Effects

Atoms in **external electric or magnetic fields** show energy shifts and spectral line splittings:

### Zeeman Effect

- Splitting of energy levels in a **magnetic field**
- Arises from interaction between **magnetic dipole moment** and the field

### Stark Effect

- Shifting and splitting of levels in an **electric field**
- Especially important in high-field environments (e.g., plasma spectroscopy)

These effects involve **degenerate perturbation theory** and angular momentum coupling — completely outside the Bohr model.

---

### ❌ No Probabilities, Uncertainty, or Wavefunctions

Bohr’s electrons move in perfect orbits — their position and momentum are precisely known. But in quantum mechanics:

- Position and momentum **cannot be known simultaneously** (Heisenberg uncertainty)
- Electrons are described by **wavefunctions**, not orbits
- Orbitals are **probability clouds**, not definite paths
- Transitions arise from **evolving wave amplitudes**, not classical jumps

These foundational concepts are introduced fully in **Phase B**, where we move to **Schrödinger’s equation**, **probability densities**, and **measurement**.

### Summary Table: What Bohr Misses

| Topic | Description | Missing from Bohr? |
|-------|-------------|---------------------|
| Transition rates | How fast transitions occur | ✅ |
| Einstein coefficients | Absorption and stimulated emission | ✅ |
| Selection rules | Which transitions are allowed | ✅ |
| Lifetimes & linewidths | Time-energy uncertainty | ✅ |
| Spin & fine structure | Doublets, relativistic corrections | ✅ |
| Multi-electron behavior | Shielding, shell structure | ✅ |
| External field effects | Stark & Zeeman | ✅ |
| Probability & uncertainty | Core of modern QM | ✅ |
