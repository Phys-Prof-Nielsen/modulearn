## Problem 1: Radiocarbon Dating an Ancient Artifact

Archaeologists uncover a wooden spear tip in a peat bog. Laboratory analysis finds that the ratio of $^{14}\text{C}$ to $^{12}\text{C}$ in the sample is only **25%** of what is found in living organisms.

**Questions:**

- (a) How many half-lives have passed since the death of the organism?
- (b) Estimate the age of the artifact. Use the half-life of $^{14}\text{C}$:
  
  $$
  T_{1/2} = 5730 \, \text{years}
  $$


```{admonition} Problem 1: Solution
:class: tip, dropdown

We are told that only 25% of the original $^{14}\text{C}$ remains. This corresponds to:

$$
\frac{1}{2^n} = 0.25 \Rightarrow n = 2 \, \text{half-lives}
$$

With a half-life of:

$$
T_{1/2} = 5730 \, \text{years}
$$

The age of the sample is:

$$
\text{Age} = 2 \cdot 5730 = \boxed{11,\!460 \, \text{years}}
$$

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 2: Energy Yield from Uranium Fission

Each fission of a $^{235}\text{U}$ nucleus releases approximately **200 MeV** of energy. A nuclear power plant uses **1.00 kg** of $^{235}\text{U}$ fuel.

**Questions:**

- (a) How many uranium atoms are in 1.00 kg of $^{235}\text{U}$?  
  Use Avogadro's number: $N_A = 6.022 \times 10^{23} \, \text{mol}^{-1}$.

- (b) Estimate the total energy released in joules.

- (c) Compare this to the energy from burning 1.00 kg of coal  
  (approximately $2.5 \times 10^7 \, \text{J}$).


```{admonition} Problem 2: Solution
:class: tip, dropdown

**(a) Number of atoms in 1 kg of $^{235}$U:**

$$
n = \frac{1000 \, \text{g}}{235 \, \text{g/mol}} = 4.26 \, \text{mol}
$$

$$
N = n \cdot N_A = 4.26 \cdot 6.022 \times 10^{23} = \boxed{2.56 \times 10^{24}}
$$

**(b) Energy released per fission:**

$$
E_{\text{fission}} = 200 \, \text{MeV} = 3.2 \times 10^{-11} \, \text{J}
$$

$$
E_{\text{total}} = N \cdot E_{\text{fission}} = 2.56 \times 10^{24} \cdot 3.2 \times 10^{-11} = \boxed{8.2 \times 10^{13} \, \text{J}}
$$

**(c) Compare to coal:**

$$
E_{\text{coal}} \approx 2.5 \times 10^7 \, \text{J/kg}
$$

$$
\frac{8.2 \times 10^{13}}{2.5 \times 10^7} = \boxed{3.28 \times 10^6 \, \text{kg coal}}
$$

```

## Problem 3: Fusion in the Sun

In the Sun, four protons fuse into one helium nucleus, and **0.7%** of the total mass involved is converted to energy.

**Questions:**

- (a) Calculate the energy released in one fusion cycle.  
  Assume proton mass $m_p = 1.67 \times 10^{-27} \, \text{kg}$ and use

  $$
  E = \Delta m c^2
  $$

  with $c = 3.00 \times 10^8 \, \text{m/s}$.

- (b) If the Sun fuses about **600 million tons of hydrogen per second**, estimate its power output in watts.

```{admonition} Problem 3: Solution
:class: tip, dropdown

Mass of 4 protons: $4 \cdot 1.67 \times 10^{-27} = 6.68 \times 10^{-27}$ kg  
Mass loss: $0.007 \cdot 6.68 \times 10^{-27} = 4.68 \times 10^{-29}$ kg

$$
E = \Delta m \cdot c^2 = 4.68 \times 10^{-29} \cdot (3 \times 10^8)^2 = \boxed{4.21 \times 10^{-12} \, \text{J}}
$$

**(b) Solar power output:**

600 million tons = $6 \times 10^{14}$ kg of H per second

$$
\text{Reactions per second} = \frac{6 \times 10^{14}}{4 \cdot 1.67 \times 10^{-27}} \approx 9 \times 10^{40}
$$

$$
\text{Total power} = 9 \times 10^{40} \cdot 4.21 \times 10^{-12} = \boxed{3.79 \times 10^{29} \, \text{W}}
$$

(Note: This overestimates; true solar output is about $3.85 \times 10^{26} \, \text{W}$)

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 4: Iodine-131 in Medical Imaging

Iodine-131 is commonly used in thyroid diagnostics. It has a half-life of $T_{1/2} = 8.0 \, \text{days}$. A patient receives a dose with an initial activity of **10 MBq**.

**Questions:**

- (a) What is the activity of the sample after **24 days**?

- (b) After how many days will the activity drop below **1 MBq**?

Use the decay law:

$$
A(t) = A_0 e^{-\lambda t}, \quad \lambda = \frac{\ln 2}{T_{1/2}}
$$

```{admonition} Problem 4: Solution
:class: tip, dropdown

**(a) Activity after 24 days:**

$$
\lambda = \frac{\ln 2}{8.0} \approx 0.0866 \, \text{day}^{-1}
$$

$$
A = A_0 e^{-\lambda t} = 10 \cdot e^{-0.0866 \cdot 24} \approx \boxed{1.25 \, \text{MBq}}
$$

**(b) Time when activity is 1 MBq:**

$$
1 = 10 e^{-\lambda t} \Rightarrow t = \frac{\ln 10}{\lambda} = \frac{2.303}{0.0866} \approx \boxed{26.6 \, \text{days}}
$$

```

## Problem 5: Fusion Reactor Fuel Efficiency

A prototype fusion reactor is testing deuterium-tritium fusion:

$$
{}^2_1\text{D} + {}^3_1\text{T} \rightarrow {}^4_2\text{He} + n + 17.6 \, \text{MeV}
$$

Suppose 1 gram of a 50/50 D-T mixture is used.

**Questions:**

- (a) How many fusion reactions occur?

- (b) Estimate the total energy released (in joules).

- (c) Compare this to the energy required to heat **1 ton of water by $50^\circ$C**.  
  (Use $c = 4186 \, \text{J/kg} \cdot \text{C}^\circ$ for water.)

```{admonition} Problem 5: Solution
:class: tip, dropdown

**(a) Reactions from 1 g D-T mix (1:1 ratio):**

0.5 g D + 0.5 g T = 0.5 mol of fuel pairs

$$
n = 0.5 \cdot \frac{1}{2.5} \cdot 6.022 \times 10^{23} = \boxed{1.2 \times 10^{23}}
$$

**(b) Total energy:**

$$
E = 1.2 \times 10^{23} \cdot 2.82 \times 10^{-12} = \boxed{3.38 \times 10^{12} \, \text{J}}
$$

**(c) Water heating comparison:**

Energy to heat 1 ton of water by $50^\circ$C:

$$
Q = mc\Delta T = 1000 \cdot 4186 \cdot 50 = 2.1 \times 10^8 \, \text{J}
$$

$$
\frac{3.38 \times 10^{12}}{2.1 \times 10^8} \approx \boxed{16.1 \, \text{tons of water}}
$$

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 6: The Little Boy Bomb (Hiroshima)

The “Little Boy” bomb dropped on Hiroshima used about **64 kg** of uranium-235, but only about **1 kg** underwent fission. Each fission releases **200 MeV** of energy.

**Questions:**

- (a) Estimate the number of fissions that occurred.

- (b) Calculate the total energy released in joules and in kilotons of TNT.  
  (1 kiloton of TNT = $4.184 \times 10^{12} \, \text{J}$)

- (c) How does this compare to the total potential energy from the full 64 kg?


```{admonition} Problem 6: Solution
:class: tip, dropdown

**(a) Fissions from 1 kg of $^{235}$U:**

$$
N = \frac{1000}{235} \cdot 6.022 \times 10^{23} = \boxed{2.56 \times 10^{24}}
$$

**(b) Total energy released:**

$$
E = 2.56 \times 10^{24} \cdot 3.2 \times 10^{-11} = 8.2 \times 10^{13} \, \text{J}
$$

Convert to kilotons TNT:

$$
\frac{8.2 \times 10^{13}}{4.184 \times 10^{12}} \approx \boxed{19.6 \, \text{kt}}
$$

**(c) Full potential of 64 kg:**

$$
64 \cdot 19.6 = \boxed{1254.4 \, \text{kt}}
$$

```

## Problem 7: Castle Bravo and Fusion Chain Yield

The Castle Bravo test (1954) was a U.S. hydrogen bomb based on deuterium-tritium fusion and a lithium-6 fusion-fission chain. It yielded **15 megatons** of TNT.

**Questions:**

- (a) Convert the energy yield to joules.  
  (1 megaton TNT = $4.184 \times 10^{15} \, \text{J}$)

- (b) How many D-T fusion reactions would be required to produce this energy  
  if each releases **17.6 MeV**?

- (c) Compare the mass of D-T fuel required to a common object (e.g., a watermelon).


```{admonition} Problem 7: Solution
:class: tip, dropdown

**(a) Convert 15 megatons to joules:**

$$
15 \cdot 4.184 \times 10^{15} = \boxed{6.28 \times 10^{16} \, \text{J}}
$$

**(b) Reactions needed:**

$$
\frac{6.28 \times 10^{16}}{2.82 \times 10^{-12}} = \boxed{2.23 \times 10^{28}}
$$

**(c) Fuel mass (1 reaction = 5 amu = $8.3 \times 10^{-27}$ kg):**

$$
m = 2.23 \times 10^{28} \cdot 8.3 \times 10^{-27} = \boxed{0.827 \, \text{kg}}
$$

A fusion bomb that yielded 15 megatons was powered by less than 1 kg of fuel—demonstrating the immense energy density of nuclear fusion.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">
