## Problem 1: Two Point Charges on a Line

Two point charges are located along the x-axis:

- $ q_1 = +2.0\,\mu\text{C} $ at $ x = 0\,\text{m} $
- $ q_2 = -3.0\,\mu\text{C} $ at $ x = 0.30\,\text{m} $

**(a)** Find the electric potential $ V $ at point $ x = 0.10\,\text{m} $.

**(b)** Using the 1D approximation (this is known as a central difference method)  

$$
E(x) \approx -\frac{V(x+\Delta x) - V(x-\Delta x)}{2\Delta x}
$$  

estimate the electric field at $ x = 0.10\,\text{m} $ using $ \Delta x = 0.010\,\text{m} $.


```python
#DIY Cell
```

```python
%reset -f
import numpy as num

q1 = 2.0E-6
q2 = -3.0E-6

x1 = 0.0
x2 = 0.30

x = 0.10

k = 8.99E9

# Part a
V1 = k*q1/num.abs((x - x1))
V2 = k*q2/num.abs((x - x2))

V = V1 + V2

print(V)

# Part b
dx = 0.01
V_forward = k*q1/num.abs(x - x1 + dx) + k*q2/num.abs(x - x2 + dx)
V_backward = k*q1/num.abs(x - x1 - dx) + k*q2/num.abs(x - x2 - dx)

E = -(V_forward - V_backward)/(2*dx)

print(E)
```

## Problem 2: Non-Uniform Line Charge

A thin rod lies along the x-axis from $ x = 0 $ to $ x = 2.0\,\text{m} $. The **linear charge density varies with position** as:

$$
\lambda(x) = \lambda_0 \left(1 + \frac{x}{L} \right), \quad \text{with} \quad \lambda_0 = 2.0 \times 10^{-6} \,\text{C/m}, \quad L = 2.0\,\text{m}
$$

A point $ P $ is located at $ x = 3.0\,\text{m} $.

**(a)** Set up and evaluate the integral for the electric potential at point $ P $:

**(b)** Use a finite-difference estimate to approximate the electric field at $ x = 3.0\,\text{m} $.

```python
#DIY Cell
```

```python
%reset -f

import sympy as sym
import numpy as num
from scipy.integrate import quad

lambda0 = 2.0E-6
L = 2.0
Px = 3.0

k = 8.99E9


# Part a
#------------------------
# Specify boundaries of integration
#------------------------
x1 = 0.0
x2 = L

#------------------------
# define unknown parameters
#------------------------
x = sym.Symbol('x')

#------------------------
# define desnity function 
#------------------------
density = lambda0*(1 + x/L)

#------------------------
# define the electric potential integrand function
dV = k*density/(sym.Abs(Px - x))

#------------------------
# find the integral 
#------------------------
V = sym.integrate(dV,(x,x1,x2))  

#------------------------
#print the result
#------------------------
print('The electric potential is: '+str(V)+' Volts')


# Part b

Px = 3.0
dx = 0.01
x_forward = Px + dx
x_backward = Px - dx

# define the electric potential integrand function
dV = k*density/(sym.Abs(x_forward - x))
V_forward = sym.integrate(dV,(x,x1,x2))

dV = k*density/(sym.Abs(x_backward - x))
V_backward = sym.integrate(dV,(x,x1,x2))

# use the same difference method as in problem 1
E = -(V_forward - V_backward)/(2*dx)

print(E)
```

## Problem 4: Electric Potential from a Charged Ring (Cylindrical Coordinates)

A **thin ring** of radius $ R = 0.2\,\text{m} $ lies in the $xy$-plane and is centered at the origin. The ring carries a **uniform line charge density** of $ \lambda = 6.0 \times 10^{-6} \,\text{C/m} $.

You are asked to find the **electric potential** at a point along the $z$-axis at $ z = 0.3\,\text{m} $.

**(a)** Set up an integral in cylindrical coordinates to compute the electric potential at the point $ (0, 0, z) $. Use the fact that every element of charge $ dq = \lambda R d\theta $ on the ring is the same distance from the point on the $z$-axis.

**(b)** Evaluate the integral and find the electric potential at $ z = 0.3\,\text{m} $.

*Hint: Use the symmetry of the ring — all $dq$ elements are the same distance from the point on the $z$-axis.*

*Useful formula: For a point on the $z$-axis, the distance from any point on the ring is constant:*

$$
r = \sqrt{R^2 + z^2}
$$

You should find that the potential is:

$$
V(z) = k \int \frac{dq}{\sqrt{R^2 + z^2}} = k \frac{Q}{\sqrt{R^2 + z^2}}
$$

Where $ Q = \lambda \cdot 2\pi R $ is the total charge on the ring.
