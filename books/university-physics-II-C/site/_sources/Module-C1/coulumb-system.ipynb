# C1.1 Electric Force - System of Charges
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C1.1.1 Motivation: Beyond Two-Particle Interactions

In **Phase D**, we studied electric forces between **two point charges** using **Coulomb’s law**. This laid the foundation for understanding how individual charges interact — attraction vs. repulsion, the role of distance, and the vector nature of the force.

But real-world systems rarely involve just two charges. Think of:
- electrons in an atom,
- ions in a salt crystal,
- or static charges on a comb near your hair.

In all of these, a single charge is influenced by **multiple other charges at once**. So how do we find the **net force** on a charge from many sources?

That's where the **principle of superposition** becomes essential.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C1.1.2 Review: Coulomb’s Law for Two Charges

The electric force between two point charges is:

$$
\vec{F} = k \frac{q_1 q_2}{r^2} \hat{r}
$$

Where:
- $\vec{F}$ is the electric force vector (Newtons)
- $k = 8.988 \times 10^9 \, \text{N·m}^2/\text{C}^2$ is Coulomb's constant
- $q_1$ and $q_2$ are the magnitudes of the two charges (C)
- $r$ is the distance between them (m)
- $\hat{r}$ is the unit vector from the source to the target

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C1.1.3 Superposition: Multiple Source Charges

If a test charge $q_0$ is placed near **multiple point charges** $q_1, q_2, \dots, q_n$, the total force on $q_0$ is the **vector sum** of the individual forces:

$$
\vec{F}_{\text{net}} = \vec{F}_1 + \vec{F}_2 + \dots + \vec{F}_n
$$

Each individual force is given by:

$$
\vec{F}_i = k \frac{q_i q_0}{r_i^2} \hat{r}_i
$$

So the full expression becomes:

$$
\vec{F}_{\text{net}} = \sum_{i=1}^{n} k \frac{q_i q_0}{r_i^2} \hat{r}_i
$$

This is a **vector addition** process. You must take both magnitude and direction into account — forces from each $q_i$ can point in different directions depending on charge signs and positions.

If we make the source/test locations more specific, we can write it as

<div class="alert alert-success">
    
$$ \vec{F} = \sum kqq^{'}_i\frac{\vec{r} - \vec{r}_{i}'}{|\vec{r} - \vec{r}_{i}'|^3} $$
</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">


## C1.1.4 Key Takeaways

- **Coulomb’s law** applies to each individual pair of charges.
- Use the **superposition principle**: add all electric force vectors to get the total force.
- **Direction matters**: use vectors carefully.
- Forces from **like charges** are repulsive; from **opposite charges**, attractive.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C1.1.5 Coulomb’s Force from a Continuous Charge Distribution

When we deal with **many point charges** close together — as in a charged rod, surface, or volume — it's no longer practical to sum over individual particles.

Instead, we treat the charge as being **spread out continuously** in space. This requires us to **replace the summation with an integral**.

This generalization leads us toward concepts like the **electric field**, **Gauss’s law**, and **electrostatics of materials**.

The infinitesimal force on charge $q$ due to an infinitesimal source charge $dq'$ is

$$ d\vec{F} = kqdq^{'}\frac{\vec{r} - \vec{r}'}{|\vec{r} - \vec{r}'|^3} $$

and the **Generalized Coulomb’s Law** is

<div class="alert alert-success">
    
$$ \vec{F} = \int kqdq^{'}\frac{\vec{r} - \vec{r}'}{|\vec{r} - \vec{r}'|^3} $$
</div>

Sometimes it may be more convenient to make a "u-substitution" such that we integrate over a spatial dimension instead of charge. The differentials are related through the **charge densities** as

- **Linear charge density** (1D objects):  
  $$dq = \lambda dx$$ 

- **Surface charge density** (2D objects):  
  $$dq = \sigma dA $$ 

- **Volume charge density** (3D objects):  
  $$ dq = \rho dV $$ 



### Notes

- This is a **vector integral**, so you must take into account the **direction** of each contribution.
- This formulation is valid for **any static distribution** of charge (electrostatics).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C1.1.7 Key Takeaways

- Continuous charge distributions require **integrals** instead of summations.
- The **force** on a point charge becomes a **superposition of infinitesimal forces**.
- Mastering this leads naturally into the concept of the **electric field** as a property of space.

   $$
