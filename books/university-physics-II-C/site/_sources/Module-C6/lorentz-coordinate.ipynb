## C6.2.1 From Classical Intuition to Relativistic Reality

In classical physics, we are taught that the laws of motion are the same in all inertial frames. This idea, called **Galilean invariance**, forms the backbone of Newtonian mechanics. It allows us to switch between observers moving at constant velocity relative to each other using simple equations:

$$
x' = x - vt, \quad t' = t
$$

These transformations assume that **time is absolute** and **simultaneity is universal** — that events occur at the same time for all observers, regardless of their motion. This made perfect sense... until light got involved.

The **Michelson-Morley experiment** (1887) attempted to detect the Earth's motion through the "luminiferous aether" by measuring variations in the speed of light. The result was surprising:

> The speed of light was found to be the same in all directions — regardless of the Earth's motion.

This result contradicted Galilean relativity and implied something profound:
- The speed of light is constant **in all inertial frames**, not just one.

Einstein took this result seriously. He proposed two foundational postulates for his theory of **Special Relativity**:
1. The laws of physics are the same in all inertial frames.
2. The speed of light in vacuum is the same for all observers, regardless of their relative motion.

These two postulates **cannot** coexist with Galilean transformations. Something had to change — and it was our conception of time and space.

To reconcile these postulates, we must abandon the idea that time is absolute. We must adopt a new framework where:
- **Time and space are not independent** but are interwoven.
- **Simultaneity becomes relative** — different observers may disagree on whether two events happened at the same time.
- **Distances and durations depend on the observer’s motion.**

This leads us to the **Lorentz transformations** — the mathematical structure that replaces the Galilean transformations in relativistic settings.

Lorentz transformations are the key to understanding:
- **Time dilation**: moving clocks run slower.
- **Length contraction**: moving objects appear shorter.
- **Relativity of simultaneity**
- **Four-dimensional spacetime geometry**
- **Velocity addition in relativity**
- **Transformation of electric and magnetic fields**

They form the **mathematical foundation of special relativity**, underlying everything from GPS satellites and particle accelerators to modern field theories and cosmology.

We study Lorentz transformations because:

1. **Experiments demand them** — they explain results classical physics cannot.
2. **They reshape our understanding of space and time**, making them observer-dependent.
3. **They are foundational to all modern physics**, from electromagnetism to quantum field theory.

In the next section, we will derive the Lorentz transformations from Einstein’s postulates — preserving the constancy of the speed of light and revealing the deep symmetry of spacetime.

## C6.2.2 Deriving the Lorentz Coordinate Transformation: Reexamining Position

### Revisiting the Galilean Coordinate Transformation

In classical mechanics, when dealing with two inertial frames in relative motion, we used the **Galilean transformation** to relate position measurements between frames:

$$
x = x' + ut
$$

Here:
- $x$ is the position of the object in frame **S** (the "stationary" frame),
- $x'$ is the position of the same object in frame **S′**, moving at velocity $u$ relative to S,
- $t$ is the time, assumed to be the same in both frames.

To be precise about **which observer** is making which measurement, we now write:

$$
x_S = x'_{S'} + u t_S \tag{1}
$$

This equation states:  
> The position of the object in frame S is equal to the position measured in frame S′, plus the displacement of the S′ origin (which moves at speed $u$) from the perspective of frame S.

This is **fine in classical physics**, where:
- All observers agree on time ($t_S = t_{S'}$),
- All observers agree on lengths ($x'_{S'} = x'_{S}$).

But this breaks down in the relativistic world.

---

### The Breakdown at High Speeds

At speeds approaching that of light $c$, the assumptions behind Equation (1) are no longer valid. Two serious problems arise:

1. **Simultaneity is no longer absolute.** Events that are simultaneous in one frame may not be in another.
2. **Length is no longer invariant.** A rod of proper length $L_0$ in its rest frame appears contracted to $L = L_0/\gamma$ in a frame moving relative to it.

So if $x'_{S'}$ is the position of the object measured **in the moving frame S′**, we cannot simply substitute it into a position formula meant to describe what the stationary observer in **S** sees.

This discrepancy means that Equation (1) mixes measurements made by two different observers — a fatal flaw in galilean relativity.

---

### What Does "Position" Really Mean?

Let’s slow down and ask: **what is position?**

- Position is a **directed length measurement** from the origin to the location of the object.
- In operational terms, this requires:
  - The observer knows where the origin is,
  - The observer knows where the object is **at the same time**.

Thus, position is a **simultaneous length comparison**: "How far is the object from my origin right now?"

In classical physics, observers agree on “right now,” so no problem arises. But in special relativity, they do **not** agree on simultaneity. Hence:

- $x'_{S'}$ and $x'_{S}$ are **not the same quantity**.
- For Equation (1) to make sense, we must express all quantities **from the perspective of a single observer**.

---

### Step Toward Consistency: Express Everything in S

Let’s write the coordinate transformation entirely in terms of measurements made in **frame S**:

$$
x_S = x'_{S} + u t_S \tag{2}
$$

Here, $x'_{S}$ means:  
> "What is the position of the object *in S′’s coordinate system* — but **measured by an observer in S**?"

This is different from $x'_{S'}$, which is the same coordinate as measured **by an observer in S′**.

So how do we relate these?

---

### Length Contraction Bridges the Gap

The key to relating $x'_{S'}$ and $x'_{S}$ is **length contraction**.

If an object has a length $L_{S′}$ in its own rest frame, an observer in S sees it as:

$$
L_S = \frac{L_{S′}}{\gamma}, \quad \text{where } \gamma = \frac{1}{\sqrt{1 - \frac{u^2}{c^2}}}
$$

Applying this to our case:

$$
x'_{S} = \frac{x'_{S′}}{\gamma} \tag{3}
$$

So now, combining (2) and (3):

$$
x_S = \frac{x'_{S′}}{\gamma} + u t_S \tag{4}
$$

This is a **hybrid transformation** — we are now halfway toward the full Lorentz transformation. We have:

- Started from a **classical perspective**,
- Identified the **problem** of mixing measurements across frames,
- Introduced the **first relativistic correction** via length contraction.

But we're not done yet. We must also account for the fact that **time is frame-dependent**. The next step would be to incorporate **time dilation** and the **relativity of simultaneity** to arrive at a fully consistent, symmetric transformation between frames. In the following, we will not do that. Instead, we will take a more rigorous approach that:

* builds on Einstein's postulates of relativity
* builds on a symmetric transformation

We will see that this simple starting point will reveal time dilation and length contraction without using them in the derivation: that is, those two relativistic effects are consequences of the Lorentz transformation.

## C6.2.3 Deriving the Lorentz Coordinate Transformation: Symmetry Approach

### Building a Symmetric Transformation

We now seek a consistent and **linear** relationship between the coordinates $(x, t)$ in S and $(x', t')$ in S′. Based on symmetry and linearity, we assume the transformation takes the form:

$$
\begin{aligned}
x' &= A(x - ut) \\
t' &= B(t - \frac{u}{c^2} x)
\end{aligned}
$$

Here $A$ and $B$ are constants to be determined. We’ll soon determine $A$ and $B$ by requiring that the **speed of light is the same** in both frames. But first, let's understand the meaning of the term $t' = B(t - \frac{u}{c^2} x)$.

### Why Must the Time Transformation Include $ux/c^2$?

In our derivation of the Lorentz transformation, we proposed the general linear form:

$$
\begin{aligned}
x' &= A(x - ut) \\
t' &= B\left(t - \frac{u}{c^2}x\right)
\end{aligned}
$$

But why do we need that specific term `ux/c^2` in the time transformation? Why not just write $t' = B(t - \alpha x)$ for some constant $\alpha$?

Let’s unpack this choice carefully. We'll use a blend of **symmetry**, **units**, and **physical motivation**.

---

#### 1. Requirement of Linearity

Special relativity respects the **principle of relativity**: no inertial frame is preferred. That means the transformation between frames must be:

- **Linear**: straight-line motion stays straight-line motion, and uniform velocity maps to uniform velocity.

So the most general form for a time transformation must be linear in both $x$ and $t$:

$$
t' = B(t + \text{something} \times x)
$$

---

#### 2. Units Argument: What Can Be Added to Time?

The term added to time must also have **units of time**, since we are adding it to $t$.

- $t$ has units of seconds.
- $x$ has units of meters.

To convert $x$ into something with units of time, we must **divide by a velocity**.

So we expect:

$$
t' = B\left(t - \alpha \frac{x}{v}\right)
$$

for some velocity $v$ and dimensionless constant $\alpha$.

But what velocity should that be?

---

#### 3. The Only Universal Velocity: The Speed of Light

We are dealing with a transformation that must:

- Reduce to the Galilean case when $u \ll c$,
- Respect the fact that the **speed of light $c$ is invariant** in all inertial frames,
- Be **symmetric** — switching $u \leftrightarrow -u$ should yield the inverse transformation.

Hence, the only velocity that can appear in the time transformation — aside from the relative velocity $u$ — is the **speed of light**, $c$.

So we refine our guess to:

$$
t' = B\left(t - \frac{u}{c^2}x\right)
$$

---

#### 4. Relativity of Simultaneity: The Physical Reason

This term is essential because **events that are simultaneous in one frame are not necessarily simultaneous in another**.

To see this, suppose we didn’t include the $x$-dependent term:

> Then $t' = Bt$ — meaning all observers would agree on simultaneity, which contradicts experimental and theoretical results.

By including a term like $ux/c^2$, we ensure that:
- Observers in S and S′ will **disagree about the time ordering** of spatially separated events,
- And that this disagreement depends **linearly on distance and velocity**, as observed in nature.

---

#### 5. Symmetry Under Velocity Reversal

The transformation should be symmetric: if S′ moves at velocity $+u$ relative to S, then S moves at $-u$ relative to S′.

This requires that when we reverse the velocity, the time transformation becomes:

$$
t = B\left(t' + \frac{u}{c^2}x'\right)
$$

This symmetry is preserved **only if the structure of the $x$-term is $ux/c^2$** — no more, no less.

---

### Summary

We use `t' = B(t - ux/c^2)` because:

- ✅ It's **linear** (required by inertial frame transformations),
- ✅ The added term has **units of time**,
- ✅ It involves only **universal constants** ($u$ and $c$),
- ✅ It is necessary to model the **relativity of simultaneity**,
- ✅ It guarantees **symmetry between frames**,
- ✅ It ensures **light speed invariance** when we plug in $x = ct$ later.

This specific structure is not arbitrary — it is the **simplest and only consistent** form that satisfies the foundational principles of special relativity.


---

### Next Step: Determining the Coefficients

We now have a general form for the coordinate transformation between S and S′:

$$
\begin{aligned}
x' &= A(x - ut) \\
t' &= B(t - \frac{u}{c^2} x)
\end{aligned}
$$

In the next step, we’ll determine the values of $A$ and $B$ by applying the requirement that **a light pulse moving at speed $c$ in one frame must also move at speed $c$ in the other**.

That condition — the **invariance of the speed of light** — will give us the full Lorentz transformation.

## C6.2.4 Deriving the Lorentz Coordinate Transformation: Step 3 — Enforcing Light Speed Invariance

### Goal

We now determine the constants $A$ and $B$ in the transformation equations:

$$
\begin{aligned}
x' &= A(x - ut) \\
t' &= B(t - \frac{u}{c^2} x)
\end{aligned}
$$

These equations are linear, time-reversible, and respect the symmetry between frames. The final constraint comes from the **second postulate of special relativity**:

> The speed of light is the same in all inertial frames, regardless of the motion of the source or observer.

---

### Applying the Light Propagation Condition

Let’s consider a **light pulse** that starts at the origin ($x = 0$ and $t = 0$) and travels in the $+x$ direction.

In frame **S**, the light pulse follows:

$$
x = ct \tag{1}
$$

If the Lorentz transformation is correct, then in frame **S′** the same light pulse must follow:

$$
x' = ct' \tag{2}
$$

Let’s plug Equation (1) into the transformation equations and enforce Equation (2).

---

### Step-by-Step Substitution

#### Step 1: Use $x = ct$ in the first transformation

From:

$$
x' = A(x - ut)
$$

we substitute $x = ct$:

$$
x' = A(ct - ut) = A(c - u)t \tag{3}
$$

#### Step 2: Use $x = ct$ in the second transformation

From:

$$
t' = B\left(t - \frac{u}{c^2} x\right)
$$

substitute $x = ct$:

$$
t' = B\left(t - \frac{u}{c^2}(ct)\right) = B(1 - \frac{u}{c})t \tag{4}
$$

#### Step 3: Set $x' = ct'$ and substitute (3) and (4)

From Equation (2):

$$
A(c - u)t = c B(1 - \frac{u}{c})t
$$

Cancel $t$ from both sides:

$$
A(c - u) = c B(1 - \frac{u}{c}) \tag{5}
$$

Simplify right-hand side:

$$
B(1 - \frac{u}{c}) = B\left(\frac{c - u}{c}\right)
$$

So:

$$
A(c - u) = c \cdot B \cdot \left(\frac{c - u}{c}\right)
$$

Cancel $c - u$ (assuming $c \ne u$):

$$
A = B \tag{6}
$$

---

### Apply the Same Trick in the Opposite Direction

Now repeat the same process for a light pulse moving in the **negative x-direction**, so in frame S:

$$
x = -ct
$$

and require again:

$$
x' = -ct'
$$

Proceed as before.

From:

$$
x' = A(x - ut) = A(-ct - ut) = -A(c + u)t
$$

and

$$
t' = B\left(t - \frac{u}{c^2}x\right) = B\left(t + \frac{u}{c}t\right) = B(1 + \frac{u}{c})t
$$

Then:

$$
x' = -ct' \quad \Rightarrow \quad -A(c + u)t = -c B(1 + \frac{u}{c})t
$$

Cancel $-t$ from both sides:

$$
A(c + u) = c B(1 + \frac{u}{c})
$$

As before, this simplifies to:

$$
A = B \tag{7}
$$

---

### Solving for $A = B = \gamma$

We now impose a final condition: the **inverse transformation** should have the same form, with $u \to -u$. This symmetry requires:

$$
\begin{aligned}
x &= A(x' + ut') \\
t &= A(t' + \frac{u}{c^2} x')
\end{aligned}
$$

Let’s solve for $A$ (and $B$) by plugging back into the light propagation equation one more time.

From our assumed transformations:

$$
x' = A(x - ut), \quad t' = A(t - \frac{u}{c^2}x)
$$

Now compute:

$$
x'^2 - c^2 t'^2 = A^2[(x - ut)^2 - c^2(t - \frac{u}{c^2}x)^2]
$$

We require that:

$$
x'^2 - c^2 t'^2 = x^2 - c^2 t^2 \quad \text{(invariant spacetime interval)}
$$

This happens only when:

$$
A = \frac{1}{\sqrt{1 - \frac{u^2}{c^2}}} = \gamma
$$

Thus:

$$
A = B = \gamma = \frac{1}{\sqrt{1 - \frac{u^2}{c^2}}} \tag{8}
$$

---

### Final Lorentz Transformations

We now have the full Lorentz transformation:

$$
\begin{aligned}
x' &= \gamma(x - ut) \\
t' &= \gamma\left(t - \frac{u}{c^2}x\right)
\end{aligned}
$$

And the inverse transformation is:

$$
\begin{aligned}
x &= \gamma(x' + ut') \\
t &= \gamma\left(t' + \frac{u}{c^2}x'\right)
\end{aligned}
$$

---

### Summary of Step 3

| Element | Value |
|--------|-------|
| Requirement | Speed of light is constant in all frames |
| Key substitution | $x = ct \Rightarrow x' = ct'$ |
| Solution | $A = B = \gamma$ |
| Final result | Lorentz transformation equations |

---

In the next section, we will explore the **consequences** of these transformations: time dilation, length contraction, and the relativity of simultaneity — all encoded in the structure of spacetime.

## C6.2.5 Physical Consequences of Lorentz Transformations 

Now that we have derived the full Lorentz transformations:

$$
\begin{aligned}
x' &= \gamma(x - ut) \\
t' &= \gamma\left(t - \frac{u}{c^2}x\right)
\end{aligned}
$$

we can use them to explore three of the most surprising and experimentally verified consequences of special relativity.

---

### 1. Time Dilation

**Time dilation** means that a clock moving relative to an observer ticks more slowly than one at rest in that observer’s frame.

#### Thought Experiment: Moving Clock

Imagine a clock at rest in frame **S′**. Since it’s at rest, its position does not change: $x' = 0$.

Let’s use the **inverse Lorentz transformation** to compute the time between two ticks as measured in frame **S**:

$$
t = \gamma\left(t' + \frac{u}{c^2}x'\right)
$$

Since $x' = 0$, this simplifies to:

$$
t = \gamma t'
$$

So the time between two ticks in frame **S** is **longer** than in **S′**:

$$
\Delta t = \gamma \Delta t' \tag{1}
$$

#### Interpretation:

- $\Delta t'$ is the **proper time** — time measured in the clock’s own rest frame.
- $\Delta t$ is the **dilated time** — measured by a stationary observer watching the moving clock.

> A moving clock runs slow. Time intervals appear stretched in the moving frame.

---

### 2. Length Contraction

**Length contraction** means that objects moving relative to an observer appear shorter along the direction of motion.

#### Thought Experiment: Moving Rod

Suppose a rod is **at rest in frame S′** and has length $L_0$, measured from $x'_1$ to $x'_2$. The rod moves at speed $u$ relative to frame **S**.

To measure the length **in S**, an observer must measure the two ends **simultaneously** — that is, at the same time $t$.

Let’s compute the positions $x'_1$ and $x'_2$ using the **forward Lorentz transformation**:

$$
x' = \gamma(x - ut)
$$

So:

$$
\begin{aligned}
x'_1 &= \gamma(x_1 - ut) \\
x'_2 &= \gamma(x_2 - ut)
\end{aligned}
$$

Subtracting:

$$
x'_2 - x'_1 = \gamma(x_2 - x_1)
$$

Thus:

$$
L_0 = \gamma L \Rightarrow L = \frac{L_0}{\gamma} \tag{2}
$$

#### Interpretation:

- $L_0$ is the **proper length** — the length of the object in its own rest frame.
- $L$ is the **contracted length** — what a moving observer sees.

> A moving object appears shorter along the direction of motion.

---

### 3. Relativity of Simultaneity

**Simultaneity is not absolute.** Two events that happen at the same time in one frame may occur at different times in another.

#### Thought Experiment: Two Simultaneous Flashes

Suppose two light flashes occur simultaneously in frame **S** at $t = 0$, one at $x = -d$ and one at $x = +d$.

In frame **S′**, what are the times of these two events?

Using:

$$
t' = \gamma\left(t - \frac{u}{c^2}x\right)
$$

and $t = 0$, we get:

$$
\begin{aligned}
t'_1 &= -\gamma \frac{u}{c^2}(-d) = \gamma \frac{ud}{c^2} \\
t'_2 &= -\gamma \frac{u}{c^2}(+d) = -\gamma \frac{ud}{c^2}
\end{aligned}
$$

So in S′:
- The light flash at $x = -d$ occurs **after** the flash at $x = +d$.

> What is simultaneous in one frame is **not** simultaneous in another.

This is not just an illusion — it's a physical, measurable effect.

---

### Summary of Consequences

| Effect                   | Description                 | Equation                          |
|--------------------------|-----------------------------|-----------------------------------|
| Time dilation            | Moving clocks run slow      | `Δt = γ Δt′`                      |
| Length contraction       | Moving rods appear shorter  | `L = L₀ / γ`                      |
| Relativity of simultaneity | Simultaneity depends on observer | `t′ = γ (t - ux / c²)`        |


---

These phenomena have been confirmed experimentally:
- Muons reaching Earth’s surface due to time dilation
- Particle accelerator lifetimes
- GPS satellite time corrections

## C6.2.6 GPS and Relativity: Quantifying the Position Error Without Velocity Correction

### Overview

GPS satellites orbit Earth at an altitude of ~20,200 km and move at speeds of about **3.87 km/s**.

Their onboard atomic clocks are affected by:
- **Special relativity**: due to their high orbital speed.
- **General relativity**: due to being farther from Earth's gravitational field.

In this example, we focus on the **special relativistic effect** — specifically, **how velocity affects time** and leads to positioning errors if not corrected.

---

### Known Values

- Speed of satellite: $v = 3.87 \times 10^3\ \text{m/s}$
- Speed of light: $c = 3.00 \times 10^8\ \text{m/s}$
- One day = $86,400\ \text{s}$
- Signal timing accuracy required: $\pm 20\ \text{ns}$ for meter-level precision

---

### Step 1: Compute Time Dilation from Special Relativity

From special relativity, a moving clock ticks slower by:

$$
\Delta t' = \frac{\Delta t}{\gamma}, \quad \text{where } \gamma = \frac{1}{\sqrt{1 - \frac{v^2}{c^2}}}
$$

First, compute $\gamma$:

$$
\frac{v^2}{c^2} = \left(\frac{3.87 \times 10^3}{3.00 \times 10^8}\right)^2 \approx 1.67 \times 10^{-10}
$$

$$
\gamma \approx 1 + \frac{1}{2} \cdot \frac{v^2}{c^2} \approx 1 + 0.835 \times 10^{-10}
$$

(This approximation is valid because $v \ll c$.)

---

### Step 2: Time Loss per Day

The difference in clock rate is:

$$
\Delta t - \Delta t' \approx \Delta t \left(1 - \frac{1}{\gamma}\right) \approx \Delta t \cdot \left(\frac{1}{2} \cdot \frac{v^2}{c^2}\right)
$$

Over one day:

$$
\Delta t = 86,400\ \text{s}, \quad \Delta t - \Delta t' \approx 86,400 \cdot 0.835 \times 10^{-10} \approx 7.21\ \mu\text{s}
$$

So the satellite’s clock **lags behind Earth clocks by about 7.2 microseconds per day** due to special relativity.

---

### Step 3: Convert Time Error to Distance Error

GPS positioning depends on accurate timing of signals traveling at the speed of light:

$$
\text{Distance error} = c \cdot \Delta t = 3.00 \times 10^8\ \text{m/s} \cdot 7.21 \times 10^{-6}\ \text{s} \approx 2.16\ \text{km}
$$

> ❌ If special relativity weren't corrected for, GPS would be off by over **2 kilometers every day**.

---

### Real GPS Systems

In real GPS:
- The satellite clocks are **pre-corrected** for special and general relativistic effects.
- Ground-based systems apply additional **relativistic corrections** to signal timing, such as:
  - **Velocity of the receiver** (e.g., airplane or rocket),
  - **Doppler shift**,
  - **Sagnac effect** (due to Earth’s rotation).

---

### Summary

| Effect                       | Value                       |
|-----------------------------|-----------------------------|
| Satellite speed             | `3.87 km/s`                 |
| Relativistic time lag (SR)  | `~7.2 μs/day`               |
| Position error per day      | `~2.16 km`                  |

This is why GPS **must use special relativity** — without it, navigation would rapidly drift into uselessness.

## C6.2.7 Star Wars & the Tilted Stick: Relativity of Simultaneity in Action

### Scenario: Lightsaber Training on a Moving Star Destroyer

Luke Skywalker is training with a **vertical training droid staff** — a rigid pole, 2 meters long, standing upright inside a training room **on a stationary Star Destroyer**, as seen from the Star Destroyer's rest frame (**S′**).

- The stick is **perfectly vertical** in the Star Destroyer frame.
- The **bottom** is at position: $(x = 0,\ y = 0,\ t = 0)$
- The **top** is at: $(x = 0,\ y = 2\ \text{m},\ t = 0)$

Now imagine a **Rebel scout ship** flies past the Star Destroyer at **$v = 0.8c$** in the $x$-direction and takes a photo of the stick.

> How does the stick appear in the **Rebel frame (S)**?

---

### The Key Idea: Relativity of Simultaneity

To get a photograph in the Rebel frame (S), we must determine **where the ends of the stick are at the same time in S**.

But since the stick is at rest in S′, both ends are at $x = 0$ in S′ at time $t' = 0$.

To find how this appears in the **Rebel frame**, we use the **inverse Lorentz transformations** to compute the positions of both ends at **$t = 0$ in S**:

---

### Lorentz Transformations (Inverse)

Let’s find the $x$-coordinates of the **top and bottom** of the stick in S at $t = 0$.

Recall the **inverse Lorentz transformation** for space:

$$
x = \gamma (x' + v t')
$$

And for time:

$$
t = \gamma \left(t' + \frac{v x'}{c^2} \right)
$$

But we’ll solve this **in reverse**: given $t = 0$ in S, find where each event (bottom and top of the stick) must be in S.

---

#### Step 1: Lorentz Parameters

- $v = 0.8c$
- $\gamma = \frac{1}{\sqrt{1 - v^2/c^2}} = \frac{1}{\sqrt{1 - 0.64}} = \frac{1}{\sqrt{0.36}} = 1.6667$

---

#### Step 2: Use Time Transformation to Find $t'$ at Each $y$ (in S′)

We need to find when (in the Star Destroyer frame S′) the **Rebel observer** takes the photo.

Let’s calculate $t'$ for the bottom and top **assuming the Rebel takes a snapshot at $t = 0$ in S**:

Using:

$$
t' = \gamma \left( t - \frac{v x}{c^2} \right)
$$

For the **bottom of the stick** (at $x = 0$ in S′), its $x$ in S is:

$$
x_{\text{bottom}} = \gamma (x' + v t') = \gamma \cdot (0 + v t') \rightarrow x \text{ depends on } t'
$$

But we want to find $x_{\text{bottom}}$ and $x_{\text{top}}$ at the **same $t = 0$** in S.

So we invert the **space transformation**:

$$
x' = \gamma (x - v t)
$$

From that:

$$
x' = \gamma x \quad \text{since } t = 0
\Rightarrow x = \frac{x'}{\gamma}
$$

Let’s now evaluate $x$ in the Rebel frame for both ends of the stick **at $t = 0$**:

---

#### Bottom of the Stick (at $x' = 0$ in S′)

$$
x_{\text{bottom}} = \frac{0}{\gamma} = 0
$$

---

#### Top of the Stick (at $x' = 0$ too — but at $y' = 2$ m in S′)

Yes, both ends have $x' = 0$ (since the stick is vertical in S′), **but to appear at the same time in S**, we must "catch" the top at an **earlier time** in S′.

From:

$$
t = \gamma \left(t' + \frac{v x'}{c^2} \right)
\Rightarrow t = \gamma t'
\Rightarrow t' = \frac{t}{\gamma}
$$

So to be seen **at the same $t = 0$ in S**, the **top** must be observed **at an earlier $t'$** in S′ so that it "moves into position."

Now, to get the **apparent $x$ positions** of both ends in S at the same $t = 0$, use:

$$
x = \gamma (x' + v t')
$$

We set:
- For bottom: $x' = 0,\ t' = 0 \Rightarrow x_{\text{bottom}} = 0$
- For top: $x' = 0,\ t' = -\Delta t'$

Let’s say light travels vertically in the Star Destroyer frame. Then to observe both ends at once in S, the **top must be seen earlier**.

Choose $\Delta t' = \frac{v \Delta y}{c^2}$ so the top reaches the correct position in S. Let’s plug in:

$$
\Delta t' = \frac{v \Delta y}{c^2} = \frac{(0.8c)(2)}{c^2} = \frac{1.6}{c}
$$

Then:

$$
x_{\text{top}} = \gamma (0 + v (-\Delta t')) = \gamma (-v \cdot \Delta t') = \gamma \left(-v \cdot \frac{v \Delta y}{c^2} \right)
= -\gamma \frac{v^2 \Delta y}{c^2}
$$

Now insert numbers:

- $\gamma = 1.6667$
- $v = 0.8c$
- $\Delta y = 2\ \text{m}$

Then:

$$
x_{\text{top}} = -1.6667 \cdot \frac{(0.8)^2 \cdot 2}{1} = -1.6667 \cdot 1.28 = -2.133\ \text{m}
$$

---

#### Result: Apparent Tilt

- Bottom of stick: $x = 0$
- Top of stick: $x \approx -2.13$ m

So the stick appears **tilted backward**, with the **top lagging behind** in the $x$ direction.

You can compute the tilt angle:

$$
\theta = \tan^{-1}\left( \frac{2}{2.133} \right) \approx 43.4^\circ
$$

---

### Takeaway

| Feature                | Star Destroyer Frame (S′) | Rebel Frame (S)              |
|------------------------|----------------------------|-------------------------------|
| Stick orientation      | Vertical                   | Tilted backward (~43.4°)      |
| Cause of tilt          | None                       | Relativity of simultaneity    |
| Contraction direction  | None (perpendicular)       | None (length unchanged)       |

- ✅ No length contraction occurs **perpendicular** to motion.  
- 🌀 But due to different **simultaneity slices**, the stick appears **rotated in spacetime**.

---

### Why It Matters

This is a vivid consequence of special relativity:

- Objects that are upright in one frame may **not appear upright** in another.
- **Spacetime is not Euclidean** — it “tilts” under Lorentz transformations.

So next time you see a perfectly vertical antenna on a speeding TIE fighter — remember:

> To an outside observer, it might be leaning slightly...  
> **Thanks to Einstein.**

## C6.2.8 Star Wars & Lorentz "Rotations": The Relativity of Simultaneity

### Scenario: The Battle Above Hoth

Two X-wing pilots — **Wedge** and **Poe** — are in formation above Hoth. From the Rebel base’s perspective (frame S), they fire laser blasts **simultaneously** at a Star Destroyer.

- Wedge is at position $x = -1000$ m,
- Poe is at $x = +1000$ m,
- Both fire their lasers at $t = 0$ in the base frame.

So in frame S (the base frame), the event coordinates are:

- Wedge's shot: $(x_1 = -1000,\ t_1 = 0)$
- Poe's shot: $(x_2 = +1000,\ t_2 = 0)$

> The Star Destroyer gets hit **from both sides at once**.

---

### But What If the Star Destroyer Is Moving?

Let’s say the Star Destroyer is moving **rightward** at $u = 0.6c$ relative to the Rebel base. We'll use a Lorentz transformation to see how the shots appear in the **Star Destroyer frame (S′)**.

The Lorentz time transformation is:

$$
t' = \gamma \left(t - \frac{u x}{c^2} \right), \quad \text{where } \gamma = \frac{1}{\sqrt{1 - u^2/c^2}}
$$

At $u = 0.6c$, we get:

$$
\gamma = \frac{1}{\sqrt{1 - 0.36}} = \frac{1}{\sqrt{0.64}} = 1.25
$$

Now compute the firing times **in the Star Destroyer’s frame**.

---

#### Wedge’s Shot (left side):

$$
t'_1 = \gamma \left(0 - \frac{(0.6c)(-1000)}{c^2} \right) = 1.25 \cdot \left(\frac{600}{c} \right) = \frac{750}{c}
$$

---

#### Poe’s Shot (right side):

$$
t'_2 = \gamma \left(0 - \frac{(0.6c)(+1000)}{c^2} \right) = 1.25 \cdot \left(-\frac{600}{c} \right) = -\frac{750}{c}
$$

---

#### What Just Happened?

In the **Star Destroyer frame**:

- **Poe’s shot** (from the front) happened **earlier**.
- **Wedge’s shot** (from the rear) happened **later**.

> To the Empire crew, it looked like **they were hit in the front first**... and then shot from behind.

---

### What This Means: Lorentz as a Spacetime "Rotation"

The Lorentz transformation mixes space and time like a **rotation**, but instead of rotating in Euclidean space, it rotates in **Minkowski spacetime** — which distorts **simultaneity**:

- Events that are simultaneous in one frame ($\Delta t = 0$) **are not** simultaneous in another frame.
- The "plane of simultaneity" **tilts** with relative velocity.

> You could say: Darth Vader sees the galaxy “tilted.”

---

### Summary

| Frame         | Wedge Fires | Poe Fires | Simultaneity |
|---------------|-------------|-----------|---------------|
| Rebel Base (S) | $t = 0$     | $t = 0$   | Simultaneous  |
| Star Destroyer (S′) | $t'_1 = +750/c$ | $t'_2 = -750/c$ | Not simultaneous |

In Star Wars, it may look like a perfect synchronized strike.  
But from Vader’s bridge, it looks like a **deliberately staggered ambush**.

---

### Takeaway

Lorentz transformations "rotate" the axes of space and time.  
They preserve the **speed of light**, but **twist simultaneity** — meaning **your “now” isn’t my “now”** in a galaxy far, far away.
