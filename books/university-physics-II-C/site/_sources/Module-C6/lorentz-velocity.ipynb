# C6.3 Velocity Transformation in Special Relativity

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.3.1 Objective

We’ve seen how **space and time coordinates** transform between inertial frames using the Lorentz transformations. Now we want to understand:

> How do **velocities** transform between frames moving relative to each other?

In classical (Galilean) physics, we’d simply add or subtract the relative speed:

$$
v = v' + u
$$

But this **fails in special relativity**, especially when $v'$ or $u$ approach the speed of light $c$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.3.2 Recap: Lorentz Coordinate Transformations

For two frames **S** and **S′** where S′ moves at speed $u$ relative to S in the $+x$ direction, we have:

$$
\begin{aligned}
x' &= \gamma(x - ut) \\
t' &= \gamma\left(t - \frac{u}{c^2}x\right)
\end{aligned}
$$

Our goal is to compute $v_x = dx/dt$ in frame S, given $v'_x = dx'/dt'$ in S′.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.3.3 Deriving the 1D Velocity Transformation

Let an object move in the $x$-direction. Let us write down the differential position and time:

$$
\begin{aligned}
dx' &= \gamma(dx - u\,dt) \\
dt' &= \gamma\left(dt - \frac{u}{c^2} dx\right)
\end{aligned}
$$

Now divide both:

$$
v'_x = \frac{dx'}{dt'} = \frac{\gamma(dx - u\,dt)}{\gamma\left(dt - \frac{u}{c^2} dx\right)} = \frac{dx - u\,dt}{dt - \frac{u}{c^2} dx}
$$

Divide numerator and denominator by $dt$:

$$
v'_x = \frac{v_x - u}{1 - \frac{u v_x}{c^2}} \tag{1}
$$

This is the **relativistic velocity addition formula**.

> It tells us how the velocity of an object in S transforms into the velocity in S′.

We can invert this to find $v_x$ in terms of $v'_x$:

$$
v_x = \frac{v'_x + u}{1 + \frac{u v'_x}{c^2}} \tag{2}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.3.4 Key Features of the Velocity Transformation

1. If $v'_x = c$, then $v_x = c$ — light still travels at $c$ in all frames ✅
2. If both $v'_x$ and $u$ are much less than $c$, the denominator ≈ 1, and we recover the classical result:  
   $v_x \approx v'_x + u$
3. **No matter how fast you move**, the result never exceeds $c$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.3.5 Example: Approaching Light Speed

Suppose in S′ an object moves at $v'_x = 0.9c$, and S′ moves at $u = 0.8c$ relative to S. Then in S:

$$
v_x = \frac{0.9c + 0.8c}{1 + \frac{(0.9c)(0.8c)}{c^2}} = \frac{1.7c}{1 + 0.72} = \frac{1.7c}{1.72} \approx 0.988c
$$

✅ Still less than $c$!

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.3.6 Transformation of Perpendicular Velocities

Let’s now consider motion **perpendicular** to the direction of relative motion (e.g. in the $y$-direction). This is a bit subtler.

Start again from Lorentz transformations:

$$
\begin{aligned}
y' &= y \\
t' &= \gamma\left(t - \frac{u x}{c^2}\right)
\end{aligned}
$$

Ratio of differentials:

$$
v'_y = \frac{dy'}{dt'} = \frac{dy}{\gamma\left(dt - \frac{u}{c^2} dx\right)} = \frac{v_y}{\gamma\left(1 - \frac{u v_x}{c^2}\right)} \tag{3}
$$

So the perpendicular velocity transforms as:

$$
v'_y = \frac{v_y \sqrt{1 - \frac{u^2}{c^2}}}{1 - \frac{u v_x}{c^2}} \tag{4}
$$

Likewise for $v'_z$:

$$
v'_z = \frac{v_z \sqrt{1 - \frac{u^2}{c^2}}}{1 - \frac{u v_x}{c^2}} \tag{5}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.3.7 Summary of Velocity Transformation Formulas

| Direction | Velocity in S | Velocity in S′ |
|----------|----------------|----------------|
| Parallel ($x$) | $v_x$ | `v'_x = (v_x - u) / (1 - uv_x/c²)` |
| Perpendicular ($y$) | $v_y$ | `v'_y = v_y √(1 - u²/c²) / (1 - uv_x/c²)` |
| Perpendicular ($z$) | $v_z$ | `v'_z = v_z √(1 - u²/c²) / (1 - uv_x/c²)` |

> All components are affected, even perpendicular ones!

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.3.8 Why Is This Important?

These velocity transformation rules are critical for:

- Understanding motion near the speed of light
- Predicting particle paths in accelerators
- Modeling high-energy astrophysical phenomena
- Explaining phenomena like **aberration of light** and **Doppler shifts**

And most importantly:  
> They protect the speed of light as the **ultimate speed limit** — no matter how hard you try to "add velocities," $v < c$ always.

In the next section, we’ll examine some practical **applications** and work through examples involving relative motion and light propagation in different frames.

## C3.6.8 Star Wars Relativity: Chasing the Millennium Falcon at Relativistic Speed


### Scenario

Darth Vader is piloting an Imperial Star Destroyer in pursuit of the **Millennium Falcon**, which is trying to escape at high speed.

- The **Falcon** is traveling at $v_F = 0.9c$ relative to the planet Endor (frame S).
- A **TIE fighter** is launched from the Star Destroyer and moves forward at $v_T' = 0.4c$ relative to the **Star Destroyer** (frame S′), which itself is moving at $u = 0.8c$ relative to Endor.

> What is the **velocity of the TIE fighter relative to Endor**, and how fast is it approaching the Falcon?

---

### Step 1: Velocity of the TIE Fighter in Endor's Frame (S)

We apply the **relativistic velocity addition formula**:

$$
v_T = \frac{v_T' + u}{1 + \frac{v_T' u}{c^2}} = \frac{0.4c + 0.8c}{1 + \frac{(0.4)(0.8)}{1}} = \frac{1.2c}{1.32} \approx 0.9091c
$$

So, the TIE fighter is moving at approximately **$0.91c$** relative to Endor.

✅ Even though classically you'd expect $1.2c$, **relativity keeps the velocity below $c$**.

---

### Step 2: Velocity of the TIE Fighter Relative to the Millennium Falcon

Now, the TIE fighter is at $0.9091c$ and the Falcon at $0.9c$, both relative to Endor.  
We want the **relative velocity** as measured in the **Falcon’s frame**.

Use:

$$
v_{\text{TIE relative to Falcon}} = \frac{v_T - v_F}{1 - \frac{v_T v_F}{c^2}} = \frac{0.9091c - 0.9c}{1 - (0.9091)(0.9)} = \frac{0.0091c}{1 - 0.8182} = \frac{0.0091c}{0.1818} \approx 0.05c
$$

### Interpretation:

To the crew of the Millennium Falcon, the TIE fighter is **slowly gaining** at just **5% the speed of light** — not the huge closing speed you'd expect from simple subtraction.

---

### What About the Reverse?

From the TIE fighter’s perspective, the Falcon is moving away at:

$$
v_{\text{Falcon relative to TIE}} = \frac{v_F - v_T}{1 - \frac{v_F v_T}{c^2}} = \frac{0.9c - 0.9091c}{1 - 0.9 \cdot 0.9091} = \frac{-0.0091c}{0.1818} \approx -0.05c
$$

✅ Same result, opposite sign — relativity is symmetric and self-consistent!

---

### Summary

| Frame              | Measured Velocity         |
|-------------------|---------------------------|
| Endor (S)         | Falcon: `0.9c`, TIE: `0.91c` |
| Falcon’s frame     | TIE approaching at `0.05c` |
| TIE’s frame        | Falcon escaping at `0.05c`  |

---

### Moral of the Story

In Star Wars, ships often appear to **zip past each other instantly**, but in real relativity:

- Velocities **don’t simply add**,
- Even tiny relative speeds near $c$ require **careful transformation**,
- And **no ship, not even the Falcon, can escape physics**.

May the **force of Lorentz invariance** be with you.

## C6.3.9 Real-Life Example: Muon Decay and Relativity in Earth's Atmosphere

High-energy **muons** are created in the upper atmosphere (around 10 km above sea level) when cosmic rays collide with air molecules. These muons are unstable particles with a **proper lifetime** (in their own rest frame) of:

$ \tau_0 \approx 2.2\ \mu\text{s} $

They travel toward the Earth at a speed of:

$ v \approx 0.998c $

---

### 🤔 Classical Expectation

In the Earth's frame (lab frame), the time it takes for a muon to reach the surface is:

$$
t = \frac{L_0}{v} = \frac{10^4\ \text{m}}{0.998c} \approx 33.4\ \mu\text{s}
$$

But the muon’s proper lifetime is only $2.2\ \mu\text{s}$. 

**Classical prediction**: the muons should decay long before reaching Earth.  
**Reality**: many muons are detected at sea level.

This discrepancy is resolved by **special relativity**.

---

### ✅ Earth Frame (S): Time Dilation

From the Earth’s perspective, the muons are moving at $v = 0.998c$. Therefore, their internal clocks tick slower by a factor of:

$$
\gamma = \frac{1}{\sqrt{1 - (0.998)^2}} \approx 15.8
$$

So the **dilated lifetime** is:

$$
\tau = \gamma \tau_0 = 15.8 \times 2.2\ \mu\text{s} \approx 34.8\ \mu\text{s}
$$

The distance the muon can travel before decaying:

$$
x = v \tau \approx 0.998c \times 34.8\ \mu\text{s} \approx 10.4\ \text{km}
$$

✅ Enough for most muons to reach the ground.

---

### Muon Frame (S′): Length Contraction

In the muon's frame, it is at rest and the Earth (along with the atmosphere) is rushing toward it at $v = 0.998c$. Due to **length contraction**, the thickness of the atmosphere is reduced:

$$
L = \frac{L_0}{\gamma} = \frac{10\ \text{km}}{15.8} \approx 0.63\ \text{km}
$$

The time it takes Earth to reach the muon:

$$
t' = \frac{L}{v} = \frac{0.63\ \text{km}}{0.998c} \approx 2.1\ \mu\text{s}
$$

✅ This matches the muon's proper lifetime. In its own frame, the atmosphere is thinner — so it survives the trip.

---

### 🚀 Rocket Frame (S″): Velocity Transformation

Now suppose a spacecraft is rising upward at $u = 0.6c$. What is the velocity of the muon in this frame?

We use the **relativistic velocity transformation**:

$$
v' = \frac{v - u}{1 - \frac{uv}{c^2}} = \frac{0.998c - 0.6c}{1 - 0.998 \cdot 0.6} = \frac{0.398c}{0.4012} \approx 0.993c
$$

✅ Even in this frame, the muon is still moving at an extremely high relativistic speed.

---

### Summary

| Frame       | Observer               | Relativistic Effect     | Interpretation                          |
|-------------|------------------------|--------------------------|------------------------------------------|
| Earth (S)   | Lab detector           | Time dilation            | Muons live longer in the lab frame       |
| Muon (S′)   | Rest frame of muon     | Length contraction       | Atmosphere is thinner; Earth arrives in time |
| Rocket (S″) | Rising observer at 0.6c | Velocity transformation  | Muons move at $v' \approx 0.993c$        |

---

### Why This Matters

This is not just theory — we detect these muons at sea level every day. Their survival is direct **experimental evidence** of special relativity.

Without:
- **Time dilation**
- **Length contraction**
- **Velocity transformation**

we couldn’t explain their presence on the ground. The muon decay example is a textbook case of relativity in action — and it confirms that our universe obeys the principles of Einstein’s theory.

## C6.3.10 Real Experiments in Space Involving Relativistic Velocity Transformation

Special relativity isn’t just for theoretical physicists — **real particle experiments on satellites and spacecraft** rely on relativistic **velocity transformations** to interpret data correctly.

Below are key systems and experiments where the **relativistic addition of velocities** plays an essential role.

---

### 1. GPS Satellite Systems — Timekeeping and Signal Accuracy

- GPS satellites orbit Earth at approximately **3.9 km/s** ($\approx 1.3 \times 10^{-5} c$).
- Each satellite carries an **atomic clock**.
- To synchronize satellite clocks with receivers on Earth or in motion (e.g., airplanes), **relativistic corrections are applied**, including:

  - **Time dilation** (due to satellite speed),
  - **Gravitational time shift** (general relativity),
  - **Velocity transformation** for moving receivers or aircraft.

> Even small errors in velocity transformation (on the order of $v^2/c^2$) lead to meter-level positioning errors.

---

### 2. AMS-02 (Alpha Magnetic Spectrometer) on the ISS

- AMS-02 is a high-energy particle physics detector on the **International Space Station**, which orbits Earth at about **7.8 km/s** ($\approx 2.6 \times 10^{-5} c$).
- It measures cosmic ray particles: electrons, positrons, protons, antiprotons, nuclei.

#### Why velocity transformation matters:

- Particles are observed in the **ISS frame**, which is not the Earth’s rest frame.
- To compare data across detectors and times, velocities and momenta are transformed into:
  - The **Earth frame**,
  - Or the **solar rest frame** (e.g., for solar-origin particles).

This transformation affects:
- **Arrival direction** and **angular distribution**,
- **Energy spectra interpretation**,
- **Source identification** for cosmic particles.

---

### 3. PAMELA Detector — Cosmic Ray Observation

- PAMELA (Payload for Antimatter Matter Exploration and Light-nuclei Astrophysics) flew aboard a satellite in a near-polar orbit (~700 km altitude).
- It observed **antiprotons, positrons, electrons**, and light nuclei in cosmic radiation.

#### Velocity transformation relevance:

- Velocities measured in the **satellite frame** must be transformed into:
  - **Earth frame**,
  - **Heliocentric frame** (for solar particles),
  - Or **galactic rest frame** (for cosmic ray origin studies).

Velocity transformations are used to:
- Reconstruct particle **trajectories**,
- Understand flux **anisotropies**,
- Match observed distributions with theoretical models.

---

### 4. Relativistic Doppler Shift and Particle Angle Transformation

Space-based instruments observing incoming relativistic particles or radiation experience:

- **Relativistic Doppler shifts** in frequency and energy,
- **Angle transformation (aberration of light)**,
- **Apparent velocity changes** due to frame switching.

These effects require using the **Lorentz velocity transformation** when:
- Comparing observations from moving spacecraft to ground-based data,
- Modeling particle paths and directions in high-energy astrophysics.

---

### Summary Table

| System / Experiment      | Role of Velocity Transformation                            |
|--------------------------|-------------------------------------------------------------|
| **GPS satellites**        | Time synchronization, signal Doppler shift, moving receivers |
| **AMS-02 on ISS**         | Particle velocity/momentum in Earth's frame                |
| **PAMELA satellite**      | Transformation to solar/galactic rest frames                |
| **Cosmic ray telescopes** | Angular and energy spectra from moving observers            |

---

### Why It Matters

Modern satellite and orbital physics requires more than time dilation — it needs **full relativistic kinematics**. The **velocity transformation** ensures:

- Light and particles obey the same $c$ limit in all frames,
- High-speed data is interpreted consistently,
- Global navigation and particle astrophysics remain accurate and aligned with special relativity.
