## C5.2.1 Motivation: What Is Polarization?

When wearing polarized sunglasses, you may notice:

- Glare is reduced when looking at water or roads.
- The sky's color seems to change at certain angles.
- LCD screens become black when rotated.

Why? Because **light can be polarized** — and that reveals something fundamental:

> Light is a **transverse wave**. Its electric field oscillates **perpendicular** to its direction of travel.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.2.2 What Does Polarization Mean?

A wave is **polarized** when the direction of its electric field is well-defined and consistent.

### Transverse vs Longitudinal
- Longitudinal waves (like sound) can't be polarized — their oscillations are in the direction of travel.
- Transverse waves (like light) can — their oscillations are **orthogonal** to the travel direction.

For a light wave moving in the $+z$ direction:
- The electric field oscillates in the **x-y plane**
- Polarization describes the **direction** of that oscillation

<hr style="height:2px;border-width:0;color:gray;background-color:gray">


## C5.2.3 Types of Polarization

Light can exhibit various types of polarization depending on how the electric field vector behaves as the wave propagates. The most common types are:

### ▶ 1. **Linear Polarization**

**Definition:**  
The electric field oscillates in a single fixed direction, perpendicular to the direction of wave propagation.

**Mathematical Form:**

If electric field is oscillating along the $x$-axis:

$$
\vec{E}(z,t) = E_0 \cos(kz - \omega t) \, \hat{x}
$$

or at an angle:

$$
\vec{E}(z,t) = E_0 \cos(kz - \omega t) (\cos\theta\, \hat{x} + \sin\theta\, \hat{y})
$$

**Example Use Cases:**
- Light after passing through a polarizing filter
- Laser beams are typically linearly polarized
- LCD screens emit linearly polarized light

**Visualization:**  
If you observe the electric field head-on, it oscillates back and forth along a line.

### ▶ 2. **Circular Polarization**

**Definition:**  
The tip of the electric field vector traces out a **circle** in the plane perpendicular to the direction of propagation. The magnitude of the field is constant, but its direction rotates uniformly.

**Mathematical Form:**

$$
\vec{E}(z,t) = E_0 \cos(kz - \omega t)\, \hat{x} + E_0 \sin(kz - \omega t)\, \hat{y}
$$
- **Right circular**: $\hat{x}$ leads $\hat{y}$ by 90° phase
- **Left circular**: $\hat{y}$ leads $\hat{x}$ by 90°

**Example Use Cases:**
- 3D movie glasses (each lens filters right or left circular polarization)
- Radar systems use circular polarization to distinguish targets
- Circularly polarized light occurs in some biological systems (like beetles' shells)

**Visualization:**  
From the front, the electric field vector spins like a propeller.

### ▶ 3. **Elliptical Polarization**

**Definition:**  
A general case of polarization where the electric field vector traces an **ellipse**. This happens when the $x$ and $y$ components of the electric field have different amplitudes or arbitrary phase shifts.

**Mathematical Form:**

$$
\vec{E}(z,t) = E_x \cos(kz - \omega t)\, \hat{x} + E_y \cos(kz - \omega t + \phi)\, \hat{y}
$$

- If $E_x = E_y$ and $\phi = \pm \pi/2$, → **circular**
- If $\phi = 0$ or $\pi$, → **linear**
- Otherwise → **elliptical**

**Example Use Cases:**
- Found naturally in optical activity (e.g., sugar solutions or biological molecules)
- Occurs in birefringent crystals (like quartz or calcite)
- Used in advanced microscopy techniques

**Visualization:**  
The electric field vector traces a tilted oval (ellipse) as it rotates.

### ▶ 4. **Unpolarized Light**

**Definition:**  
The electric field oscillates randomly in all perpendicular directions over time — no preferred orientation.

**Example Use Cases:**
- Most natural light sources (sunlight, incandescent bulbs)
- Becomes partially polarized after scattering or reflection

**Key Fact:**  
Passing unpolarized light through a linear polarizer produces **linearly polarized light with half the original intensity.** 

Unpolarized light consists of many electromagnetic waves whose electric field vectors point in **random directions**, all perpendicular to the direction of travel.

#### Why Does a Polarizer Transmit Only Half the Intensity of Unpolarized Light?

#####  Assumptions

- Incoming light is **unpolarized**: its electric field vectors point in **random directions** perpendicular to the direction of travel.
- A **linear polarizer** transmits only the component of the electric field **along its transmission axis**.
- Intensity is proportional to the **square of the electric field amplitude**:  

  $$
  I \propto E^2
  $$
  
##### Step 1: Electric Field Component Through the Polarizer

For a single light wave whose electric field is at angle $\theta$ relative to the polarizer's axis, only the component along the axis is transmitted:

$$
E_{\text{trans}} = E_0 \cos\theta
$$

So the transmitted intensity for that wave is:

$$
I(\theta) \propto E_{\text{trans}}^2 = E_0^2 \cos^2\theta
$$




##### Step 2: Average Over All Polarization Angles

For unpolarized light, $\theta$ is uniformly distributed between $0$ and $2\pi$.

We calculate the **average value** of $\cos^2\theta$ over all angles:

$$
\langle \cos^2\theta \rangle = \frac{1}{2\pi} \int_0^{2\pi} \cos^2\theta \, d\theta
$$

Use the trigonometric identity:

$$
\cos^2\theta = \frac{1 + \cos(2\theta)}{2}
$$

Substitute into the integral:

$$
\langle \cos^2\theta \rangle = \frac{1}{2\pi} \int_0^{2\pi} \frac{1 + \cos(2\theta)}{2} \, d\theta
= \frac{1}{2\pi} \cdot \frac{1}{2} \int_0^{2\pi} (1 + \cos(2\theta)) \, d\theta
$$

Now evaluate the integral:

$$
\int_0^{2\pi} 1 \, d\theta = 2\pi, \quad \int_0^{2\pi} \cos(2\theta) \, d\theta = 0
$$

So:

$$
\langle \cos^2\theta \rangle = \frac{1}{2}
$$

##### Final Result

The **average transmitted intensity** is:

$$
I_{\text{avg}} = I_0 \cdot \langle \cos^2\theta \rangle = I_0 \cdot \frac{1}{2}
$$

> 🔹 A linear polarizer transmits **half** the intensity of incoming unpolarized light — not because of absorption or loss, but because only **half the randomly oriented field components** align enough to contribute to the transmitted wave.

### Summary Table

| Type               | Field Behavior                            | Real-World Examples                              |
|--------------------|--------------------------------------------|--------------------------------------------------|
| **Linear**         | Oscillates in one direction                | Lasers, polarized sunglasses, LCDs               |
| **Circular**       | Constant magnitude, rotating direction     | 3D movies, radar, beetle shells                  |
| **Elliptical**     | Rotating, varying amplitude                | Birefringent materials, chiral molecules         |
| **Unpolarized**    | Random direction, varies over time         | Sunlight, lamps, reflections become polarized    |

## C5.2.4 How Do We Polarize Light?

Polarizing light means manipulating it so that the electric field oscillates in a **specific, preferred manner**. There are several physical mechanisms that can polarize light:

### ▶ 1. **Polarizing Filters (Absorptive Polarizers)**

A **polarizing filter** consists of long-chain molecules or microstructured grids that absorb electric field components **perpendicular** to the filter’s transmission axis. Only the component **aligned with the axis** is transmitted.

#### ➤ What Happens:
- Unpolarized light becomes **linearly polarized** after passing through the filter.
- Linearly polarized light at angle $\theta$ to the axis has only a component $E \cos\theta$ transmitted.


#### Malus’ Law: Intensity After a Polarizer

When linearly polarized light passes through a polarizer at an angle $\theta$ to its transmission axis, the transmitted **intensity** is:

$$
\boxed{I = I_0 \cos^2 \theta}
$$

Where:
- $I_0$ is the initial intensity (when $\theta = 0$)
- $\theta$ is the angle between the light's polarization and the polarizer axis

#### Derivation (Conceptual)

Let $E_0$ be the amplitude of the incoming electric field.

- Only the **component along the transmission axis** passes:

  $$
  E = E_0 \cos \theta
  $$

- Intensity is proportional to the square of the electric field:

  $$
  I \propto E^2 = E_0^2 \cos^2 \theta \Rightarrow I = I_0 \cos^2 \theta
  $$

#### Real-World Applications

- **Sunglasses** reduce glare by blocking horizontally polarized light.
- **LCD screens** rely on polarizing filters to control brightness.
- **Photography** uses polarizing filters to cut reflections and enhance contrast.

#### Two Polarizers in Sequence (Unpolarized Light Input)

When **unpolarized light** passes through two polarizers:

1. The **first polarizer** (aligner) transmits only the component of light aligned with its transmission axis:

   $$
   I_1 = \frac{1}{2} I_0
   $$
   
   This light is now **linearly polarized** along that axis.

2. The **second polarizer** (analyzer), placed at angle $\theta$ relative to the first, transmits only the component of the electric field aligned with its own axis:

   $$
   I_2 = I_1 \cos^2 \theta = \frac{1}{2} I_0 \cos^2 \theta
   $$

#### Final Result for Two Polarizers:

$$
\boxed{I = \frac{1}{2} I_0 \cos^2 \theta}
$$

This shows that the first polarizer cuts the intensity in half (from unpolarized to polarized), and the second projects the polarized light onto its axis.


### ▶ 2. **Reflection (Polarization by Reflection)**

When light reflects off a surface (like water, glass, or asphalt), the **reflected wave becomes partially polarized**, and at a specific angle, it can be **completely polarized**.

#### ➤ Brewster’s Angle:

At **Brewster’s angle**, the reflected light is fully polarized **perpendicular to the plane of incidence**.

$$
\tan \theta_B = \frac{n_2}{n_1}
$$

Where:
- $n_1$ is the refractive index of the incident medium
- $n_2$ is the refractive index of the reflecting medium

At this angle, the reflected and refracted rays are **90° apart**.


#### Real-World Example:
- Glare from roads or water is mostly horizontally polarized.
- Polarized sunglasses block this horizontally polarized glare.

### ▶ 3. **Scattering (Polarization by Scattering)**

When light is scattered by small particles (like air molecules), the scattered light becomes **partially polarized**, especially at **90°** to the original light path.

#### ➤ Physical Basis:
- Oscillating electrons re-emit light preferentially perpendicular to their motion.
- Scattered light at 90° is mostly polarized perpendicular to the original direction.

#### Real-World Example:
- The **blue sky** is partially polarized.
- **Bees and ants** can detect this pattern and use it for navigation.
- **Photographers** use polarizers to darken blue skies by enhancing the polarized component.


### Other Techniques

| Method               | Description                                     |
|----------------------|-------------------------------------------------|
| **Birefringence**    | Materials like calcite split light into two orthogonal polarizations with different speeds (used in wave plates) |
| **Dichroism**        | Some crystals absorb one polarization more than the other |
| **Wave plates**      | Convert linear ↔ circular/elliptical polarization by introducing a phase shift between orthogonal components |


### Summary Table

| Method                 | Type of Polarization Produced | Typical Use Case                      |
|------------------------|-------------------------------|---------------------------------------|
| **Polarizing filter**  | Linear                        | Sunglasses, LCDs, cameras             |
| **Reflection**         | Linear (at Brewster’s angle)  | Glare control, optical analysis       |
| **Scattering**         | Partial linear                | Sky, navigation, polarimetric sensors |
| **Wave plates**        | Elliptical / Circular         | Advanced optics and laser systems     |

> Each method exploits a physical asymmetry — absorption, angle, or material response — to filter or reshape light’s electric field.

## C5.2.5 Applications of Polarization

- **Sunglasses** – reduce glare by filtering horizontally polarized light
- **LCD screens** – rely on controlling polarization
- **3D movies** – use orthogonal polarizations for each eye
- **Stress analysis** – birefringent materials change polarization under stress
- **Astronomy** – polarized light gives information about interstellar dust and magnetic fields

## C5.2.6 Summary

- **Polarization** confirms that light is a **transverse wave**.
- It describes the **direction** of electric field oscillations.
- Polarizers **filter** light based on polarization, reducing glare or modifying intensity.
- Polarization arises from **filters**, **reflection**, and **scattering**.
- Applications are everywhere from sunglasses to astronomy.
