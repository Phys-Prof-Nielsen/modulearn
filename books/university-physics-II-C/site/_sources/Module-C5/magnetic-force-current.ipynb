# C5.1 — Magnetic Force on a Straight Current-Carrying Wire
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.1.1 Motivation

In Phase D, we learned that a magnetic field can exert a force on a **single moving charge**. But in real electrical systems we rarely deal with one charge at a time. Instead, we work with **currents**: many charges drifting through a conductor.

When a current-carrying wire is placed in an external magnetic field, the wire can experience a **macroscopic force** large enough to lift, rotate, or vibrate the conductor. This is the core physical mechanism behind:

- electric motors (force on current loops creates rotation),
- loudspeakers (force on a coil drives a cone),
- galvanometers and analog meters (force produces deflection),
- magnetic actuators and relays.

This section develops the force law for a **straight wire segment** in a uniform magnetic field and shows how it follows directly from the Lorentz force on individual charges.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.1.2 Historical Note 

During the early 1800s, scientists began to recognize a deep connection between **electric currents** and **magnetism**. In 1820, Hans Christian Ørsted discovered that a compass needle deflects near a current-carrying wire, demonstrating that currents create magnetic effects.

While Ørsted’s 1820 discovery showed that electric currents **create magnetic fields**, an equally important development followed almost immediately: scientists began investigating whether magnetic fields could, in turn, exert forces **on currents**.

In the years after Ørsted’s experiment, **André-Marie Ampère** carefully studied the mechanical forces between current-carrying wires. He demonstrated experimentally that:

- Two parallel wires carrying currents exert forces on one another.
- The force depends on the magnitude of the currents.
- The direction of the force depends on whether the currents flow in the same or opposite directions.

These observations established something profound:  
>**currents experience mechanical forces due to magnetic interactions.**

This was not just a curiosity. It implied that electrical energy could be converted into mechanical motion. By the 1830s and 1840s, inventors and physicists began constructing early electromagnetic devices that used this force principle. The foundations of the electric motor were laid during this period.

Later, the force between parallel current-carrying wires became so fundamental that it was used to define the SI unit of current, the ampere. The definition was historically based on the force per unit length between two long, parallel wires carrying equal currents.

Thus, the magnetic force on a current-carrying conductor is not just a theoretical extension of the Lorentz force — it is one of the earliest experimentally verified manifestations of electromagnetism and one of the first demonstrations that electricity can produce controlled mechanical motion.

In this module, we connect that historical experimental observation to its microscopic explanation:  
the magnetic force on a wire arises from the Lorentz force acting on the moving charges inside the conductor.

## C5.1.3 Quick Review from Phase D: Magnetic Force on a Single Charge

From experiments, we found that a magnetic field exerts a force on a moving charge given by:

$$
\vec{F}_B = q\,\vec{v}\times\vec{B}.
$$

Key ideas to carry forward:

- The force is **zero** if the charge is not moving ($\vec{v}=\vec{0}$).
- The force is **perpendicular** to both $\vec{v}$ and $\vec{B}$.
- The magnitude is

$$
F_B = |q|\,v\,B\,\sin\theta,
$$

where $\theta$ is the angle between $\vec{v}$ and $\vec{B}$.

In this module, we will apply this same law to the charges inside a wire that collectively produce a current.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.1.4 Derivation: Force on a Straight Current-Carrying Wire Segment

### Step 1: Model the wire’s moving charges

Consider a straight wire segment of length $L$ carrying a steady current $I$ through a region where the magnetic field is uniform $\vec{B}$.

Inside the metal, charge carriers (electrons) drift with an average drift velocity $\vec{v}_d$.  
If the wire has cross-sectional area $A$ and number density of mobile charge carriers $n$ (carriers per volume), then the number of carriers in a segment of length $L$ is:

$$
N = nAL.
$$

Each carrier has charge $q$ (for electrons, $q=-e$).

---

### Step 2: Force on one carrier

Each carrier experiences a magnetic force:

$$
\vec{F} = q\,\vec{v}_d \times \vec{B}.
$$

---

### Step 3: Net force on all carriers in the segment

If the wire is in steady-state, the magnetic forces on the moving charges are transmitted to the lattice (through collisions and interactions), producing a net force on the wire segment.

Assuming the field is uniform over the segment, the total force on all carriers is:

$$
\vec{F}_{\text{net}} = N\,\vec{F}
= (nAL)\,q\,\vec{v}_d\times\vec{B}.
$$

Rearrange:

$$
\vec{F}_{\text{net}} = (nAq\vec{v}_d)\,L \times \vec{B}.
$$

At this point, we recognize the current relation.

---

### Step 4: Replace microscopic quantities with the current $I$

The current in a conductor is related to drift velocity by:

$$
I = nqAv_d.
$$

In vector form, the direction of current is along the wire, so we can write the vector length element as

$$
\vec{L} = L\,\hat{\ell},
$$

where $\hat{\ell}$ points along the direction of conventional current.

Using the current relation, the net force becomes:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Magnetic Force on a Straight Current-carrying Wire</h3>

$$
\vec{F} = I\,\vec{L}\times\vec{B}.
$$

</div>

This is the standard magnetic force law for a straight wire segment in a uniform magnetic field.

---

### Magnitude Form

If $\theta$ is the angle between the wire direction (along $\vec{L}$) and $\vec{B}$, then

$$
F = ILB\sin\theta.
$$

---

### Direction (Right-Hand Rule)

The direction of $\vec{F}$ is determined by the cross product:

- Point your fingers along $\vec{L}$ (direction of conventional current).
- Rotate toward $\vec{B}$.
- Your thumb points in the direction of $\vec{F}$.

If the charge carriers are electrons, their drift velocity is opposite the current direction, but the negative charge reverses the force direction — the final wire force is still correctly given by:

$$
\vec{F} = I\,\vec{L}\times\vec{B}.
$$

## C5.1.5 Result to Remember

For a straight wire segment of length $L$ carrying current $I$ in a uniform magnetic field $\vec{B}$:

$$
\boxed{\vec{F} = I\,\vec{L}\times\vec{B}}
$$

and

$$
\boxed{F = ILB\sin\theta.}
$$

<h3 style="margin-top:0; color:#000000;">Example — Magnetic Force on a Straight Current-Carrying Wire</h3>

A straight wire segment of length $L = 0.40~\text{m}$ lies along the **positive $x$-axis** and carries a steady current of $I = 6.0~\text{A}$ in the $+x$ direction.

The wire is placed in a uniform magnetic field of magnitude $B = 0.50~\text{T}$ that lies in the $xy$-plane and makes an angle of $\theta = 60^\circ$ **counterclockwise from the positive $x$-axis**.

Answer the following:

1. Determine the **magnitude** of the magnetic force on the wire segment.  
2. Determine the **direction** of the force using the right-hand rule.  

---

**Step 1: Identify the relevant relationship**

The magnetic force on a straight current-carrying wire is

$$
\vec{F} = I\,\vec{L} \times \vec{B}.
$$

The magnitude is

$$
F = I L B \sin\theta.
$$

Here, $\theta$ is the angle from $\vec{L}$ (along $+x$) to $\vec{B}$ measured counterclockwise in the $xy$-plane.

---

**Step 2: Compute the magnitude**

$$
F = (6.0~\text{A})
    (0.40~\text{m})
    (0.50~\text{T})
    \sin(60^\circ).
$$

Since $\sin(60^\circ) = 0.866$,

$$
F = (6.0)(0.40)(0.50)(0.866).
$$

$$
F \approx 1.0~\text{N}.
$$

---

**Step 3: Determine the direction (Right-Hand Rule)**

- $\vec{L}$ points along $+x$.
- $\vec{B}$ lies in the $xy$-plane at $60^\circ$ counterclockwise from $+x$.
  This means $\vec{B}$ has positive $x$ and positive $y$ components.

Now apply the cross product:

$$
\vec{F} = I\,\vec{L} \times \vec{B}.
$$

Using the right-hand rule:

- Point your fingers along $+x$ (current direction).
- Rotate your fingers toward $\vec{B}$ (which has a positive $y$ component).
- Your thumb points along the **positive $z$-axis**.

Therefore, the magnetic force on the wire points **out of the page** (along $+z$).

---

**Step 4: Physical interpretation**

- Only the component of $\vec{B}$ perpendicular to the wire contributes to the force.
- The force is perpendicular to both the current direction ($+x$) and the magnetic field.
- This perpendicular force allows magnetic fields to produce mechanical motion without changing electrical current magnitude.

If either the current direction or magnetic field direction were reversed, the force direction would also reverse.

<h3 style="margin-top:0; color:#000000;">Box Activity 1 — Magnetic Force on a Vertical Wire</h3>

A straight wire segment of length $L = 0.50~\text{m}$ hangs vertically and carries a steady current of $I = 5.0~\text{A}$.

- The wire lies along the **positive $y$-axis**.
- The current flows **upward** (in the $+y$ direction).
- A uniform magnetic field has magnitude $B = 0.40~\text{T}$.
- The magnetic field points along the **positive $x$-axis**.

---

Work through the following:

1. **Angle and geometry**  
   Determine the angle $\theta$ between the current direction and the magnetic field.

2. **Magnitude of the magnetic force**  
   Use

   $$
   F = I\,L\,B\,\sin\theta
   $$

   to compute the force magnitude.

3. **Direction of the magnetic force (Right-Hand Rule)**  
   Determine whether the force points:
   - out of the page ($+z$),
   - into the page ($-z$),
   - upward,
   - downward,
   - left,
   - or right.

4. **Conceptual extension**  
   Suppose the current were downward instead.  
   How would the direction of the magnetic force change?

---

A Possible Solution Guide

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1: Angle between $\vec{L}$ and $\vec{B}$**

- $\vec{L}$ points along $+y$
- $\vec{B}$ points along $+x$

Since the $x$ and $y$ directions are perpendicular,

$$
\theta = 90^\circ.
$$

---

**Step 2: Magnitude of the magnetic force**

Because $\sin(90^\circ) = 1$,

$$
F = (5.0~\text{A})(0.50~\text{m})(0.40~\text{T})(1).
$$

$$
F = 1.0~\text{N}.
$$

---

**Step 3: Direction using the right-hand rule**

We use

$$
\vec{F} = I\,\vec{L}\times\vec{B}.
$$

- Point your fingers upward along $+y$ (current direction)
- Curl them toward $+x$ (magnetic field direction)

Using the right-hand rule carefully:

$$
\hat{y} \times \hat{x} = -\hat{z}.
$$

Therefore, the force points **into the page ($-z$ direction)**.

---

**Step 4: If the current reverses**

If the current flows downward ($-y$ direction),

then

$$
(-\hat{y}) \times \hat{x} = +\hat{z}.
$$

So the force would point **out of the page ($+z$ direction)**.

---

**Physical interpretation**

- The magnetic force is perpendicular to both the current direction and the magnetic field.
- When the current is vertical and the magnetic field is horizontal, the force must point perpendicular to the page.
- Reversing the current reverses the force direction — a key feature exploited in electric motors.
