# C5.2 — Magnetic Force on Curved Current-Carrying Conductors
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.2.1 Motivation

In C5.1, we derived the magnetic force on a **straight wire segment** in a uniform magnetic field:

$$
\vec{F} = I\,\vec{L}\times\vec{B}.
$$

But real conductors are not always straight.  
Wires can bend, curve, loop, or form complex geometries — especially in motors, coils, and electromagnetic devices.

When the direction of the conductor changes along its length, we can no longer treat the entire wire as a single vector $\vec{L}$. Instead, we must:

- Break the wire into **infinitesimal segments**,
- Compute the force on each segment,
- Integrate to determine the **net force**.

This section develops the differential force law and shows how integration naturally extends the straight-wire result.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.2.2 From Finite Segments to Differential Elements

For a small current element $d\vec{\ell}$ carrying current $I$ in a magnetic field $\vec{B}$, the magnetic force is:

$$
d\vec{F} = I\, d\vec{\ell} \times \vec{B}.
$$

This is simply the straight-wire formula applied to an infinitesimal piece of the conductor.

To determine the total force on a curved wire, we integrate:

$$
\vec{F}_{\text{net}} = I \int d\vec{\ell} \times \vec{B}.
$$

---

## C5.2.3 Important Observations

### Uniform Magnetic Field

If $\vec{B}$ is constant in space, it can be taken outside the integral:

$$
\vec{F}_{\text{net}} = I \left( \int d\vec{\ell} \right) \times \vec{B}.
$$

But

$$
\int d\vec{\ell}
$$

is simply the vector displacement from the starting point to the ending point of the wire.

This leads to an important result:

- For a **closed loop** in a uniform magnetic field,

$$
\int d\vec{\ell} = 0,
$$

so

$$
\vec{F}_{\text{net}} = 0.
$$

A closed loop in a uniform magnetic field experiences **no net force** (though it may experience torque).

---

### Non-Uniform Magnetic Field

If $\vec{B}$ varies with position, it cannot be taken outside the integral:

$$
\vec{F}_{\text{net}} = I \int d\vec{\ell} \times \vec{B}(\vec{r}).
$$

In that case, curved conductors can experience a nonzero net force — this is the basis of many electromagnetic devices.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.2.4 Strategy for Solving Curved-Wire Problems

1. Parameterize the wire geometry.
2. Write $d\vec{\ell}$ in component form.
3. Compute $d\vec{F} = I\, d\vec{\ell} \times \vec{B}$.
4. Integrate over the full conductor.
5. Interpret direction and symmetry carefully.

<h3 style="margin-top:0; color:#000000;">Example — Semicircular Wire in a Uniform Magnetic Field</h3>

A wire is bent into a semicircle of radius $R = 0.20~\text{m}$ lying in the $xy$-plane.

- The center of the semicircle is at the origin.
- The semicircle extends from $\theta = 0$ to $\theta = \pi$.
- The current $I = 3.0~\text{A}$ flows counterclockwise.
- A uniform magnetic field $\vec{B} = 0.60~\text{T}\,\hat{x}$.

Find the **net magnetic force** on the semicircular wire.

---

**Step 1: Parameterize the geometry**

For a circle in the $xy$-plane:

$$
d\vec{\ell} = R\,d\theta\,\hat{\theta}.
$$

In Cartesian components:

$$
d\vec{\ell} = (-R\sin\theta\,\hat{x} + R\cos\theta\,\hat{y})\,d\theta.
$$

---

**Step 2: Compute $d\vec{F}$**

$$
d\vec{F} = I\, d\vec{\ell} \times \vec{B}.
$$

Since $\vec{B} = B\hat{x}$,

only the $\hat{y}$ component of $d\vec{\ell}$ contributes.

After computing the cross product:

$$
d\vec{F} = IBR\cos\theta\,d\theta\,(-\hat{z}).
$$

---

**Step 3: Integrate**

$$
\vec{F}_{\text{net}} = -IBR \int_0^\pi \cos\theta\, d\theta\,\hat{z}.
$$

But

$$
\int_0^\pi \cos\theta\, d\theta = 0.
$$

Therefore,

$$
\vec{F}_{\text{net}} = 0.
$$

---

**Conclusion**

The semicircular wire experiences **zero net force** in a uniform magnetic field.

However, the force distribution is not zero everywhere — different parts of the wire feel forces in opposite directions that cancel.

<h3 style="margin-top:0; color:#000000;">Box Activity 1 — Quarter Circle Wire</h3>

A wire is bent into a quarter circle of radius $R = 0.25~\text{m}$ lying in the $xy$-plane.

- The arc extends from $\theta = 0$ to $\theta = \frac{\pi}{2}$.
- The current is counterclockwise.
- A uniform magnetic field $\vec{B} = 0.80~\text{T}\,\hat{x}$.
- The current is $I = 2.0~\text{A}$.

---

1. Write $d\vec{\ell}$ in Cartesian components.
2. Compute $d\vec{F}$.
3. Integrate to find the net force.
4. Determine the direction of the force.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Following the same method as the example:

$$
\vec{F}_{\text{net}} = -IBR \int_0^{\pi/2} \cos\theta\, d\theta\,\hat{z}.
$$

Since

$$
\int_0^{\pi/2} \cos\theta\, d\theta = 1,
$$

we obtain

$$
\vec{F}_{\text{net}} = -IBR\,\hat{z}.
$$

Substitute values:

$$
F = (2.0)(0.80)(0.25) = 0.40~\text{N}.
$$

So the force is

$$
\vec{F}_{\text{net}} = -0.40~\text{N}\,\hat{z}.
$$

The force points **into the page**.
