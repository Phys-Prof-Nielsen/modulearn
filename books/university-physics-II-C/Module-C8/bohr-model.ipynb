# C7.2 The Bohr Model of the Atom

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.2.1 Historical Context (Expanded)

At the dawn of the 20th century, physicists had made major advances in understanding matter and light — but a full model of the atom remained elusive. Classical physics could not explain why atoms are stable or why they emit discrete spectral lines. Several experimental discoveries and theoretical crises converged to set the stage for Bohr's revolutionary model.

---

### Rutherford’s Nuclear Atom (1911)

In 1909, **Ernest Rutherford**, along with Geiger and Marsden, conducted the famous **gold foil experiment**. They fired alpha particles (helium nuclei) at a thin sheet of gold. Classical expectations predicted that the particles would pass through with minor deflections. However, a small fraction were deflected at large angles — some even bounced back.

> **Conclusion**: Atoms must consist of a tiny, dense, positively charged nucleus surrounded by mostly empty space.

This led to the **Rutherford model** of the atom — a miniature solar system with electrons orbiting the nucleus. But it had a fatal flaw...

---

### The Radiation Problem

According to **Maxwell’s equations**, an accelerating charge (like an electron in circular motion) should continuously emit electromagnetic radiation, losing energy as it spirals inward. This implies that:

- Electrons should collapse into the nucleus in **~10⁻¹¹ seconds**.
- Atoms should be **unstable**, yet the world around us is clearly stable.

> ❗ Classical physics predicts that atoms should not exist!

---

### The Mystery of Atomic Spectra

In the 19th century, scientists like Balmer and Rydberg cataloged **spectral lines** of hydrogen — sharp, regular wavelengths of emitted light. These lines formed distinct series (Lyman, Balmer, Paschen…), but no one could explain *why* only specific wavelengths were allowed.

- **Balmer (1885)**: Found an empirical formula for visible hydrogen lines.
- **Rydberg (1890)**: Generalized it to all hydrogen series:

  $$
  \frac{1}{\lambda} = R_H \left( \frac{1}{n_f^2} - \frac{1}{n_i^2} \right)
  $$

> But where did this formula come from? What physical process was behind it?

---

### Planck and Quantization (1900)

**Max Planck** resolved the blackbody radiation problem by introducing the idea that energy is quantized:

$$
E = h f
$$

This was radical: energy could only be emitted in discrete "quanta", not continuously.

---

### Einstein and the Photoelectric Effect (1905)

**Albert Einstein** extended Planck’s idea to light itself. He proposed that light comes in packets — **photons** — with energy $E = h f$. This explained why light below a certain frequency couldn’t eject electrons from metal, no matter the intensity.

> This strengthened the idea that **energy levels** might also be quantized in atoms.

---

### Bohr’s Leap (1913)

Niels Bohr built on all of this — Rutherford’s nucleus, Planck’s quantization, and Einstein’s photons — to create a **new atomic model**. He postulated that:

1. Electrons occupy **quantized orbits**.
2. These orbits are **stable** and don’t radiate.
3. Light is emitted or absorbed only when electrons **jump between levels**, with energy given by

   $$
   \Delta E = E_i - E_f = h f
   $$

Bohr’s model:

- Explained hydrogen’s spectral lines **quantitatively**.
- Introduced a working picture of **quantum energy levels**.
- Set the stage for the full development of **quantum mechanics**.

> **Impact**: Bohr’s model was the first to correctly describe atomic structure and spectra — at least for hydrogen — and marked the beginning of quantum theory applied to atoms.

## C7.2.2 Bohr’s Postulates

To resolve the contradictions between classical physics and atomic stability, Niels Bohr introduced **three foundational postulates** in 1913 that redefined our understanding of the atom.

---

### Postulate 1: Quantized Orbits (Orbital Angular Momentum is Discrete)

Bohr proposed that electrons can occupy only certain stable circular orbits without radiating energy, and that these orbits correspond to discrete values of orbital angular momentum:

$$
L = m_e v r = n \hbar, \quad \text{where } n = 1, 2, 3, \ldots
$$

- $m_e$ is the electron mass  
- $v$ is the orbital speed  
- $r$ is the radius of the orbit  
- $\hbar = \frac{h}{2\pi}$ is the reduced Planck constant  
- $n$ is the **principal quantum number**

This quantization condition was inspired by Planck’s and Einstein’s earlier work on quantized energy, but it was the first time angular momentum itself was assumed to be quantized.

> **Interpretation**: Not all orbits are allowed — only those where the electron’s angular momentum is an integer multiple of $\hbar$.

---

### Postulate 2: Stationary States (No Radiation in Allowed Orbits)

While in one of the allowed orbits (called a **stationary state**), the electron does **not emit electromagnetic radiation**, even though it's accelerating (which would normally violate classical physics).

> **Key Insight**: The laws of classical electrodynamics do not apply at the atomic scale in their usual form.

This postulate ensured that atoms are **stable**, avoiding the classical prediction that electrons should spiral into the nucleus. Energy is conserved unless the electron transitions between levels.

---

### Postulate 3: Quantum Jumps (Photons Carry the Energy Difference)

An electron can **jump** between energy levels — up or down — by absorbing or emitting a **quantum of light** (photon). The frequency of the emitted or absorbed light is determined by the **energy difference** between the two levels:

$$
\Delta E = E_i - E_f = h f
$$

or equivalently, for photon wavelength:

$$
\lambda = \frac{c}{f} = \frac{hc}{E_i - E_f}
$$

- $E_i$ is the initial (higher) energy level  
- $E_f$ is the final (lower) energy level  
- $f$ is the frequency of the emitted/absorbed photon  
- $h$ is Planck’s constant  
- $c$ is the speed of light

> **Explains**: The discrete lines seen in atomic spectra are due to specific allowed transitions between quantized energy levels.

---

### Summary of Conceptual Shifts

| Classical Atom | Bohr Atom |
|----------------|-----------|
| Continuous orbits | Discrete orbits ($n = 1, 2, 3, \dots$) |
| Electrons radiate continuously | No radiation in stationary states |
| Collapse into nucleus expected | Stable quantized orbits |
| No explanation for spectra | Spectral lines from energy level transitions |

---


Bohr’s postulates provided a radically new foundation for atomic theory — one that would eventually lead to the full development of **quantum mechanics**.

## C7.2.3 Why Does Planck’s Constant Appear in Angular Momentum?

Bohr didn’t derive angular momentum quantization — he **postulated** it:

$$
L = m v r = n \hbar
$$

This assumption was inspired by:

- The quantization of energy (via Planck and Einstein),
- The need to explain **discrete spectral lines** in atoms,
- The failure of classical physics to produce stable orbits.

By assuming angular momentum is an integer multiple of $\hbar$, Bohr could explain why only certain electron orbits — and hence certain atomic energies — are allowed.

Though Bohr introduced quantized orbits in 1913, a deeper explanation came later with **Louis de Broglie**’s wave-particle hypothesis:

- Electrons behave like waves with wavelength:

  $$
  \lambda = \frac{h}{p} = \frac{h}{m v}
  $$

- A stable orbit must form a **standing wave**, requiring:

  $$
  2\pi r = n \lambda \quad \Rightarrow \quad m v r = n \frac{h}{2\pi} = n \hbar
  $$

This provides a **physical justification** for Bohr’s postulate: it ensures the electron’s wavefunction fits neatly around the nucleus without destructive interference.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.2.4 Planck's Constant on a Deeper Level

Planck’s constant can be understood as a natural and fundamental parameter that emerges from the structure of quantum physics, especially when we describe systems using the Lagrangian formulation and the principle of action.

In classical mechanics, the motion of a system is determined by the principle of least action: the system follows the path that makes a quantity called the action stationary. This framework is elegant and powerful in its own right, but it takes on a deeper meaning in quantum mechanics. In the quantum world, particles don’t follow a single trajectory — instead, they explore all possible paths between two points. According to Feynman’s path integral formulation, each of these paths contributes to the system’s overall behavior, and the amount of contribution depends on the action along that path. This is where Planck’s constant comes in: it acts as a scaling factor that determines how sensitive the quantum system is to differences in action. If the action is much larger than Planck’s constant, quantum effects average out and classical behavior emerges. But when the action is comparable to Planck’s constant, interference between different paths becomes significant, leading to distinctly quantum effects like superposition and quantization. 

This is why Planck’s constant appears in every foundational part of quantum mechanics — in energy levels, wavelengths, angular momentum, and uncertainty relations. It is not just a numerical constant that showed up in early experiments; it reflects something deep about the granularity of nature at small scales. It marks the boundary between classical smoothness and quantum discreteness. In this sense, Planck’s constant is not just a historical artifact of the blackbody radiation problem — it is the natural unit of quantum action, woven into the very framework of physical law.

> Planck’s constant sets the “resolution” of the universe — how finely you can slice action before physics breaks down.

## C7.2.4 Derivation of Energy Levels in the Bohr Model 

The Bohr model successfully explains the discrete energy levels of the hydrogen atom by applying classical mechanics with a key quantum assumption: **angular momentum is quantized**.

Let’s derive the energy levels of the hydrogen atom step-by-step.

---

### Step 1: Electrostatic Force as Net Radial Force

An electron of mass $m_e$ and charge $-e$ orbits a nucleus (proton) of charge $+e$. The attractive Coulomb force provides the **centripetal acceleration** that keeps the electron in a circular orbit:

$$
F_{\text{Coulomb}} = \frac{1}{4\pi \varepsilon_0} \cdot \frac{e^2}{r^2}
$$

This must equal the **net radial force** needed to keep the electron in orbit:

$$
\sum F_r = \frac{m_e v^2}{r}
$$

Equating the two:

$$
\frac{1}{4\pi \varepsilon_0} \cdot \frac{e^2}{r^2} = \frac{m_e v^2}{r}
\quad \Rightarrow \quad
\frac{1}{4\pi \varepsilon_0} \cdot \frac{e^2}{r} = m_e v^2 \quad \text{(1)}
$$

---

### Step 2: Angular Momentum is Quantized

Bohr’s quantum assumption is that the **orbital angular momentum** of the orbiting electron is quantized in units of $\hbar$:

$$
L = m_e v r = n\hbar, \quad \text{with } n = 1, 2, 3, \dots \quad \text{(2)}
$$

Solve for velocity $v$:

$$
v = \frac{n\hbar}{m_e r} \quad \text{(3)}
$$

---

### Step 3: Solve for Orbit Radius $r_n$

Substitute equation (3) into equation (1):

Start with:

$$
\frac{1}{4\pi \varepsilon_0} \cdot \frac{e^2}{r} = m_e v^2
$$

Now plug in $v$ from (3):

$$
\frac{1}{4\pi \varepsilon_0} \cdot \frac{e^2}{r} = m_e \left( \frac{n\hbar}{m_e r} \right)^2
= \frac{n^2 \hbar^2}{m_e r^2}
$$

Now solve for $r$:

$$
r_n = \frac{4\pi \varepsilon_0 \hbar^2}{m_e e^2} \cdot n^2
$$

Define the **Bohr radius**:

$$
a_0 = \frac{4\pi \varepsilon_0 \hbar^2}{m_e e^2} \approx 0.529 \times 10^{-10} \, \text{m}
$$

Thus, the radius of the $n$-th orbit is:

$$
r_n = a_0 n^2
$$

---

### Step 4: Total Energy of the Electron

The total energy is the sum of kinetic and potential energy:

#### Kinetic Energy:
From equation (1):

$$
K = \frac{1}{2} m_e v^2 = \frac{1}{2} \cdot \frac{1}{4\pi \varepsilon_0} \cdot \frac{e^2}{r}
= \frac{e^2}{8\pi \varepsilon_0 r}
$$

#### Potential Energy:
The potential energy of an electron in a Coulomb field is:

$$
U = - \frac{1}{4\pi \varepsilon_0} \cdot \frac{e^2}{r}
$$

#### Total Energy:

$$
E = K + U = \frac{e^2}{8\pi \varepsilon_0 r} - \frac{e^2}{4\pi \varepsilon_0 r} = - \frac{e^2}{8\pi \varepsilon_0 r}
$$

Now substitute the expression for $r_n$:

$$
E_n = - \frac{e^2}{8\pi \varepsilon_0 a_0 n^2}
$$

But recall from earlier that $a_0 = \frac{4\pi \varepsilon_0 \hbar^2}{m_e e^2}$. Substituting this gives:

$$
E_n = - \frac{m_e e^4}{8 \varepsilon_0^2 h^2} \cdot \frac{1}{n^2}
$$

Or, numerically:

$$
E_n = -13.6 \, \text{eV} \cdot \frac{1}{n^2}
$$

This is the famous result: the electron in a hydrogen atom can occupy **only certain energy levels**, and they depend on $1/n^2$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.2.5 Interpretation

- The lowest energy state ($n = 1$) is the **ground state**.
- Higher $n$ values correspond to **excited states**.
- The electron cannot exist between these levels.
- The energy levels are **negative** because the electron is **bound** to the nucleus.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.2.6 Photon Transitions

When an electron jumps from a higher to a lower energy level, it emits a **photon** with energy:

$$
E_{\text{photon}} = E_i - E_f = 13.6 \, \text{eV} \left( \frac{1}{n_f^2} - \frac{1}{n_i^2} \right)
$$

### Explaining the Hydrogen Spectrum

When an electron transitions from a higher to a lower energy level, a photon is emitted:

$$
E_\gamma = h f = E_i - E_f = 13.6 \, \text{eV} \left( \frac{1}{n_f^2} - \frac{1}{n_i^2} \right)
$$

The **wavelength** of the emitted photon is:

$$
\frac{1}{\lambda} = R_H \left( \frac{1}{n_f^2} - \frac{1}{n_i^2} \right)
\quad \text{with} \quad R_H = 1.097 \times 10^7 \, \text{m}^{-1}
$$

### Series in Hydrogen Spectrum:

| Series      | Final Level $ n_f $ | Region        |
|-------------|-----------------------|---------------|
| **Lyman**   | 1                     | Ultraviolet   |
| **Balmer**  | 2                     | Visible       |
| **Paschen** | 3                     | Infrared      |

## C7.2.7 Summary

- The Bohr model assumes circular motion and quantized angular momentum.
- Using Newton’s laws and Coulomb’s force, we solve for allowed radii and energies.
- Energy levels scale as $1/n^2$ and are negative — representing bound states.
- The model explains hydrogen’s spectral lines and was the first quantum model to make accurate predictions.


### ⚠️ Limitations of the Bohr Model

- Only works accurately for **hydrogen-like atoms**.
- Cannot explain **fine structure** or **Zeeman effect**.
- Doesn’t account for **wave-particle duality** or **electron spin**.
- Violates principles of **quantum mechanics** (uses classical orbits).
