## C8.1.1 Why Study the Structure of Matter?

> Every bit of matter around you — from a grain of salt to a star — is built from the same fundamental stuff. But how deep does that structure go? And how did we figure it out?

The journey from the indivisible “atomos” of the ancient Greeks to today’s Standard Model is one of the most thrilling sagas in science. In this module, we take a guided tour into the **architecture of matter** — peeling back the layers from atoms to nuclei to the subatomic particles that bind the universe together.

## C8.1.2 Historical Context: Seeing the Unseen

### *From Light to Insight*  
In the early 20th century, mysterious spectral lines in hydrogen’s emission spectrum hinted at a hidden structure within atoms. These fingerprints of light couldn’t be explained by classical physics.

### *Rutherford’s Scattering*  
A few years earlier, alpha particles were being fired at thin gold foil. Most passed through — but some bounced back! Why? Because atoms are mostly empty space with a **dense central nucleus**.

### ⚛️ *Bohr’s Leap*  
Niels Bohr bridged the gap between classical intuition and quantum logic: he proposed electrons orbit in **discrete energy levels**, and light is emitted or absorbed when they jump between them. It explained hydrogen’s spectrum and laid the foundation for quantum theory.

## C8.1.4 Why It Matters

Understanding the structure of matter is foundational for:

- **Physics**: Quantum mechanics, nuclear physics, and particle physics  
- **Engineering**: Semiconductors, nuclear reactors, and radiation shielding  
- **Medicine**: MRI, PET scans, and radiation therapy  
- **Astronomy**: Stellar spectra, fusion, cosmic evolution
