## C8.4.1 Motivation: Why Introduce Matter Waves?

### The Story So Far

By the time of Bohr’s model, scientists had started to make sense of atomic spectra using quantized energy levels. The model explained:

- Why hydrogen emits specific spectral lines
- Why atoms are stable (electrons don't spiral into the nucleus)
- How energy relates to photon emission through discrete transitions

But the **quantization** in the Bohr model felt arbitrary — it was imposed, not derived. Why should angular momentum be restricted to $n\hbar$? Why only certain orbits?

These questions hinted at a **deeper principle** beneath Bohr’s rules.

---

### A Bigger Picture Emerges: Light Behaves Like Particles

In the early 20th century, several experiments revealed that **light — classically a wave — behaves like a particle** under certain conditions:

- **Photoelectric Effect (Einstein, 1905):** Light ejects electrons from metal surfaces only if its frequency exceeds a threshold, regardless of intensity.
- **Compton Scattering (1923):** Photons collide with electrons like billiard balls, transferring energy and momentum in quantized chunks.

These results showed that **light exhibits both wave and particle properties** — a phenomenon now known as **wave–particle duality**.

If waves can behave like particles, what about the reverse?

---

### A Bold Reversal: What if Particles Are Waves?

In 1924, **Louis de Broglie** proposed a revolutionary idea:

> If light waves can act like particles, then perhaps particles — like electrons — can act like waves.

He introduced the concept of the **matter wave**: every particle has a wavelength, inversely proportional to its momentum:

$$
\lambda = \frac{h}{p}
$$

This wasn’t just poetic symmetry — it suggested that electrons are not just “tiny balls,” but have intrinsic **wave-like behavior** that could influence how they move and interact.

---

### Why It Matters

If electrons have wavelength, then:

- **They can interfere and diffract**, like light and sound waves.
- **Only certain standing wave patterns** would fit around the nucleus — leading naturally to quantized orbits.
- **Bohr’s angular momentum rule** could be explained by **constructive interference of electron waves**.

This shifts the entire understanding of atomic structure:

- From **quantization as a rule**  
- To **quantization as a consequence of wave interference**

---

### The Payoff: Experimental Confirmation of Matter Waves

The idea that particles like electrons could behave as waves was bold and unprecedented. But within just a few years of de Broglie’s proposal, two landmark experiments confirmed it — each revealing a different facet of the quantum world.

---

#### 1. Davisson–Germer Experiment (1927)

In this experiment, electrons were accelerated and directed at a **nickel crystal**. If electrons were particles, you’d expect random scattering. Instead, Davisson and Germer observed **intensity peaks** at specific angles — just like what occurs when **X-rays diffract** off crystal planes.

This was direct, quantitative evidence of **electron diffraction** — a phenomenon possible **only if electrons behave like waves**.

##### ✅ What It Proved:

- Electrons have a **measurable wavelength**.
- The diffraction angles matched those predicted by **Bragg's law** and **de Broglie's formula** $\lambda = h/p$.
- Matter waves are **not just an idea** — they have real, testable consequences.

This experiment earned Davisson the Nobel Prize and transformed the status of the wave–particle duality from a hypothesis to a confirmed principle.

---

#### 2. Electron Double-Slit Experiment

In this now-famous variation of the classic light interference experiment, electrons are fired at a barrier with **two narrow slits**, then detected on a screen.

Even when **electrons are sent one at a time**, the resulting pattern on the screen gradually forms a **wave-like interference pattern**.

##### ✅ What It Revealed:

- **Individual electrons** exhibit interference — they seem to go through **both slits at once**.
- The **interference pattern spacing** corresponds exactly to the **de Broglie wavelength**.
- If you place a detector to observe which slit the electron passes through, the **interference pattern disappears**. The act of measurement collapses the wave behavior.

This experiment illustrates the **fundamental strangeness** of quantum mechanics: electrons behave like waves when unobserved, and like particles when measured.

#### A Surprising Truth: The Double-Slit Experiment Came *After* Quantum Theory

It often comes as a surprise to students — and even to many physicists — that the famous electron double-slit experiment, frequently cited in textbooks, documentaries, and popular science as the quintessential demonstration of quantum strangeness, wasn’t actually performed until **1961**. Given how central it is to our understanding of **wave–particle duality** and **quantum measurement**, one might assume it was one of the foundational experiments that shaped quantum mechanics. But in reality, physicists like **Bohr, Heisenberg, and Schrödinger** in the 1920s had **never seen** electrons form an interference pattern through slits — they only **theorized** that it could happen.

At the time, the technology required to produce slits narrow enough, beams coherent enough, and detectors sensitive enough simply did not exist. Instead, early confirmation of de Broglie’s hypothesis came from **electron diffraction off crystals**, such as in the **Davisson–Germer** and **G.P. Thomson** experiments in **1927**.

> The electron double-slit experiment remained a powerful **thought experiment**, and none used it more famously than **Richard Feynman**, who described it in his lectures as the experiment that “**contains the only mystery**” of quantum mechanics. He imagined sending electrons one at a time through a double slit and seeing an interference pattern build up — a mystery that was **only confirmed experimentally decades later**.

That it was finally performed by **Claus Jönsson** in 1961 — and later refined to show interference even when electrons are sent **one at a time** — only added to its mystique. It remains one of the most vivid and philosophically profound experiments in all of physics.
 

---

Together, these experiments **confirmed de Broglie’s theory** and **shattered the classical idea** of particles as rigid, deterministic entities. Matter, it turns out, has **wave nature**, and that wave nature governs its behavior at small scales.

---

> “The electron is both a particle and a wave — and neither concept alone is sufficient to describe it.”

Matter waves are not just a theoretical tool — they are **real**, **observable**, and **essential** to understanding atoms, molecules, and ultimately the structure of matter itself.

---

### Bridging to Quantum Mechanics

The idea of matter waves:

- Provides a **physical justification** for Bohr’s quantized orbits
- Explains **why only certain energy levels exist**
- Leads directly to the idea of a **wavefunction** and the **Schrödinger equation**
- Lays the foundation for understanding **probability**, **uncertainty**, and **orbital shapes**

In short: **matter waves complete the bridge from classical particles to quantum mechanics.**

---

### Final Thought

> “Electrons are not particles circling a nucleus like planets around the sun — they are waves confined in a potential well, and only certain wave patterns are allowed.”  

By introducing matter waves, we move one step closer to a truly quantum picture of nature.

## C8.4.2 The de Broglie Hypothesis: Where Does It Come From?

In 1924, Louis de Broglie made a bold proposal: if light — traditionally considered a wave — can behave like a particle (as seen in the photoelectric effect), then perhaps matter — traditionally thought of as particles — can behave like waves.

This was a revolutionary shift, but it wasn’t just philosophical. De Broglie used known physics to derive a concrete mathematical expression for the **wavelength** associated with a particle.

Let’s see how he did it.

---

### 🔹 Step 1: Planck's Relation for Light

Planck’s work on blackbody radiation introduced the idea that light energy is **quantized**:

$$
E = h f
$$

where:
- $E$ is the energy of a photon,
- $f$ is the frequency,
- $h$ is Planck’s constant.

Since the speed of light $c = \lambda f$, we can also write:

$$
E = \frac{h c}{\lambda}
$$

---

### 🔹 Step 2: Einstein’s Photon Momentum

Einstein, in explaining the **photoelectric effect**, associated a **momentum** with light — even though light has no rest mass. He used the energy–momentum relation for massless particles (which he had theorized as part of his theory of relativity):

$$
E = pc
\quad \Rightarrow \quad
p = \frac{E}{c}
$$

Substitute Planck’s relation $E = h f$:

$$
p = \frac{h f}{c} = \frac{h}{\lambda}
$$

So for photons, momentum and wavelength are related by:

$$
\lambda = \frac{h}{p}
$$

This was already well known for light. De Broglie asked: **what if this relation also holds for material particles like electrons?**

---

### 🔹 Step 3: De Broglie's Leap

De Broglie proposed that **any particle** with momentum $p$ has a wavelength:

$$
\lambda = \frac{h}{p}
$$

This applied not only to photons, but to electrons, protons, and anything with momentum.

For a classical particle:
- $p = mv$ (non-relativistic momentum)
- So $\lambda = h / mv$

This meant that **even solid matter** has wave properties — just with very short wavelengths unless the mass is small or velocity is low.

---

### 🔹 Step 4: What It Means Physically

If a particle has a wavelength, it can:
- **Interfere and diffract** like a wave
- Only fit into certain bound systems as **standing waves**
- Explain why only **discrete energy levels** are allowed in atoms (like in the Bohr model)

This gave a **wave-based explanation** for atomic structure and laid the foundation for **quantum mechanics**.

---

### Later Confirmation

The de Broglie wavelength was confirmed just a few years later by:
- **Davisson–Germer** (1927): electron diffraction off a crystal
- **G.P. Thomson** (1927): diffraction using thin metal foils

Both showed that the measured interference pattern matched the wavelength predicted by:

$$
\lambda = \frac{h}{p}
$$

## C8.4.3 Energy and Momentum of Matter Waves

### Kinetic Energy and Momentum (Non-relativistic)

For a particle of mass $m$ and speed $v$:

- Momentum: $p = mv$
- Kinetic energy: $K = \frac{1}{2}mv^2$

We can also express energy in terms of momentum:

$$
K = \frac{p^2}{2m}
$$

This is useful because it lets us relate the de Broglie wavelength to kinetic energy:

$$
\lambda = \frac{h}{\sqrt{2mK}}
$$

So the **faster** a particle moves, the **shorter** its wavelength.

## C8.4.4 Standing Waves in Circular Orbits

De Broglie’s insight offers a powerful justification for **Bohr’s quantization rule**.

Imagine the electron as a **wave** traveling around a circular orbit. For the wave to be stable, it must **fit perfectly** around the orbit — just like a standing wave on a string:

$$
2\pi r = n\lambda, \quad n = 1, 2, 3, \dots
$$

Using de Broglie’s formula for $\lambda$, we get:

$$
2\pi r = n \frac{h}{p} \Rightarrow p r = n \hbar
$$

And since $L = p r$, this gives:

$$
L = n\hbar
$$

➡️ This is **Bohr’s quantization condition**, now derived from wave behavior rather than assumed arbitrarily.

Below is an animation of an electron standing wave around the nucleus. Feel free to change the quantum number $n$ in the code to explore various "orbits".

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Set matplotlib interactive mode for Jupyter
#%matplotlib notebook

# Parameters
n = 2  # number of wave crests around the orbit
radius = 1.0
amplitude = 0.15
num_points = 1000

theta = np.linspace(0, 2 * np.pi, num_points)
base_x = radius * np.cos(theta)
base_y = radius * np.sin(theta)
wave = amplitude * np.sin(n * theta)

# Set up the figure
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_aspect('equal')
ax.set_xlim(-1.3, 1.3)
ax.set_ylim(-1.3, 1.3)
ax.axis('off')

# Draw the nucleus
ax.plot(0, 0, 'ro')  # red dot at center

# Create a line object for the standing wave
wave_line, = ax.plot([], [], 'b', lw=2)

# Initialization
def init():
    wave_line.set_data([], [])
    return wave_line,

# Animation
def animate(i):
    phase = 2 * np.pi * i / 100
    radial = radius + wave * np.cos(phase)
    x = radial * np.cos(theta)
    y = radial * np.sin(theta)
    wave_line.set_data(x, y)
    return wave_line,

# Animate
ani = FuncAnimation(fig, animate, init_func=init, frames=100, interval=50, blit=True)

HTML(ani.to_jshtml())
```
