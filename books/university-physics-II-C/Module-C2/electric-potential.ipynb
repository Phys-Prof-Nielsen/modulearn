# C2.1 Electric Potential for Multiple Point Charges and Continuous Charge Distributions

## C2.1.1 Motivation

So far, we've explored electric **force** using Coulomb’s law and electric **field** of system of charges through the **superposition principle**.

But forces and electric fields are **vectors** — direction matters, and calculating them often requires trigonometry or component addition. As we saw in Phase D, working with energy and potential is simpler, and that is often the stronger case when dealing with systems of particles. 

Electric potential lets us describe electric effects using **scalar values** — easier to work with and deeply connected to **energy**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C2.1.2 Electric Potential from a Single Point Charge

We found in Phase D that the electric potential $V$ at a point in space due to a **single point charge** $q$ is:

$$
V = k\frac{q}{r}
$$

Where:
- $V$ is the electric potential (volts)
- $r$ is the distance from the charge to the point where $V$ is measured
- $k$ is the Coulomb constant

⚠️ **Note:** Electric potential is **not a vector**. It has **magnitude only**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C2.1.3 Superposition: Multiple Point Charges

Electric potential **adds algebraically** (since it's a scalar). For multiple point charges $q_1, q_2, \dots, q_n$:

$$
V_{\text{net}} = \sum_{i=1}^{n} k\frac{q_i}{r_i}
$$

Where $r_i$ is the distance from charge $q_i$ to the point of interest.

No vectors needed! This is much simpler than computing force directly.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C2.1.4 Electric Potential from a Continuous Charge Distribution

For a continuous distribution of charge, the sum becomes an integral:

$$
V(\vec{r}) = k\int \frac{dq}{|\vec{r} - \vec{r}'|}
$$

Where:
- $\vec{r}$ is the position where you're calculating potential
- $\vec{r}'$ is the location of a charge element $dq$
- The denominator $|\vec{r} - \vec{r}'|$ is the distance between source and field point

Depending on the geometry:
- Line charge: $dq = \lambda \, dl$
- Surface charge: $dq = \sigma \, dA$
- Volume charge: $dq = \rho \, dV$

So:

- For a **line** distribution:

  $$
  V = k\int \frac{\lambda(l) \, dl}{|\vec{r} - \vec{r}'|}
  $$

- For a **surface** distribution:

  $$
  V = k\int \frac{\sigma(\vec{r}') \, dA'}{|\vec{r} - \vec{r}'|}
  $$

- For a **volume** distribution:

  $$
  V = k\int \frac{\rho(\vec{r}') \, dV'}{|\vec{r} - \vec{r}'|}
  $$

## C2.1.5 Example: Potential from a Uniform Line of Charge

A rod of length $L$ has total charge $Q$, uniformly distributed. Find the electric potential at a point $P$ located a distance $d$ from the end closest the point and along the axis of the rod.

**Steps**:
1. Let $\lambda = Q/L$
2. Choose a coordinate system: place the rod from $x = 0$ to $x = L$
3. The distance from an element at position $x$ to point $P$ is $r = d + x$
4. The potential is:

   $$
   V = k\int_0^L \frac{\lambda \, dx}{d + x}
   $$

5. Solve the integral:

   $$
   V = k\lambda\ln\left(\frac{d + L}{d}\right)
   $$

This gives a **scalar** expression — much simpler than finding the force directly!

## C2.1.6 Key Takeaways

- Electric potential is a **scalar** quantity — no direction.
- It simplifies problems involving multiple charges or extended distributions.
- Superposition still applies: **potentials add**.
- The relationship between **potential** and **electric field** is (in 1D):

  $$
  \vec{E} = -\frac{dV}{dx}\hat{i}
  $$

  (The electric field points **in the direction of decreasing potential**.)
