# C6.1 Linear Work: Point Particle
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.1 Review Linear Work in 1D
We conceptually defined work:

<div class="alert alert-success">
Work is the process of transferring energy.
</div>

and quantified it as a force acting through a directed distance:

<div class="alert alert-success">
<b>Linear Work on Point Particle</b>
    
$$ w = f_{x}\Delta x$$
</div>

The net work done on a particle is the sum of individual work contributions:  

<div class="alert alert-success">
<b>Net Linear Work on Point Particle</b>
    
$$ w_{net} = w_1 + w_2 + \dots + w_N = \sum_i^N w_i$$
</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C6.1.2 Linear Work in 2D and 3D

Since the directed distance is simply what we call the displacement, we observe that the 1D work is the multiplication between two vector components yielding a non-vector component (work is not a vector). This is suggestive of a relationship involving multiplication between two vectors. We have seen such an operation: the dot product. In fact, the energy transfer (or work) for a **constant force** is defined as    

<div class="alert alert-success">
<b>Linear Work on Point Particle</b>
    
$$ w = \vec{f}\cdot\Delta \vec{r}$$
</div>

The net work can be obtained by adding the work done by each individual force:
<div class="alert alert-success">
<b>Net Linear Work on Point Particle</b>
    
$$ w_{net} = w_1 + w_2 + \dots + w_N = \sum_i^N w_i$$
</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;">
<header>
<h3> Example 1</h3>
    </header>

<b>Problem</b>
    <br>
A strap inclined upward at a 42.0 degree angle pulls a suitcase through the airport. The tension in the strap is 20.0 N. How much work does the tension do if the suitcase is pulled 100.0 m along the horizontal floor? 
    
<b>Methodology</b>
    <br>
We are giving geometrically properties of the force and displacement vectors: length and relative angle. Hence, we will use the geometrical approach to calculate the dot product:
  
$$ w = f\Delta r \cos\theta $$
    
where $f$, $\Delta r$ are the <b>magnitudes</b> of the vectors and $\theta$ is the angle <b>between</b> the vectors when drawn tail to tail.

<b>Solution</b>
    <br>
In this case, the angle to be used in the equation above is the angle given. Letting the positive x-direction be in direction of motion and plugging in our values, we get
    
$$ w = (+20.0)(+100.0)\cos(42.0) = +1.49\times 10^3~\textrm{J} $$

<b>Evaluation</b>
    <br>
    
The net work is positive so energy is transferred into the suitcase from the person pulling it.  
</div>
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;">
<header>
<h3> Example 2</h3>
    </header>

<b>Problem</b>
    <br>
A car is being pushed with a force and displacement given by 

$$ \vec{f} = \begin{bmatrix} +160.0~\textrm{N}  \\ +40.0~\textrm{N}  \end{bmatrix} ~~,~~\Delta\vec{r} = \begin{bmatrix} +14.0~\textrm{m}  \\ -11.0~\textrm{m}  \end{bmatrix} $$
    
How much work is done pushing the car by this applied force?
    
<b>Methodology</b>
    <br>
We are giving analytical properties of the force and displacement vectors: components. Hence, we will use the analytical approach to calculate the dot product:
  
$$ w = f_{x}\Delta x + f_{y}\Delta y $$

<b>Solution</b>
    <br>
We can read off the components from the given vectors to obtain
    
$$ w = (+160.0)(+14.0) + (+40.0)(-11.0) = +1.80\times 10^3~\textrm{J} $$

<b>Evaluation</b>
    <br>
    
The net work is positive so energy is transferred into the car.  
</div>
