## **Problem 1: Work Done by a Constant Force in 2D**  
A force of $\vec{f} = (6.0 \, \hat{i} + 8.0 \, \hat{j}) \, \text{N}$ acts on a particle that moves along a straight path from point $A(1.0, 2.0) \, \text{m} $ to point $ B(4.0, 6.0) \, \text{m} $.  
1. Calculate the displacement vector $\Delta\vec{r} $ of the particle from $ A $ to $ B $.  
2. Find the work done by the force $ \vec{f} $ as the particle moves from $ A $ to $ B $.

## **Problem 2: Conservative and Non-Conservative Forces**  
A block slides down a frictionless ramp inclined at $30.0^\circ$ to the horizontal. The block has a mass of $2.0$ kg and starts from rest at the top of the ramp, which is $5.0$ meters high.  
1. Calculate the work done by the gravitational force as the block moves from the top to the bottom of the ramp.  
2. Determine the speed of the block at the bottom of the ramp.  
3. What is the role of friction in this problem if it were present, and how would it affect the total mechanical energy of the system?

## **Problem 3: Potential Energy Function for a Constant Force**  
A constant force of $ \vec{f} = (10.0 \, \hat{i}) \, \text{N} $ acts on an object moving along the $x$-axis.  
1. Calculate the potential energy function $u(x)$ associated with this force, assuming $u(0) = 0$ at the origin.  
2. Calculate the change in potential energy when the object moves from $x = 0$ to $x = 5.0 \, \text{m}$.

## **Problem 4: Gravitational Potential Energy**  
A 1.0 kg ball is lifted to a height of 10.0 meters above the ground.  
1. Calculate the gravitational potential energy of the ball at this height.  
2. If the ball is released and falls to the ground, what is the speed of the ball just before it hits the ground? Assume no air resistance.

## **Problem 5: Conservation of Mechanical Energy**  
A 0.5 kg object is dropped from rest from a height of 20.0 meters. The object is subject only to gravity.  
1. Calculate the object's speed just before it hits the ground.  
2. Using the principle of conservation of mechanical energy, determine the work done by non-conservative forces if the object loses 10.0% of its mechanical energy during the fall. What is the final speed of the object when it reaches the ground?
