## Problem C5.1 — Galilean Position Transformation

A shinkansen train moves eastward (the $+x$ direction) with a constant speed of  

$$
u = 90.0\ \text{m/s}.
$$

Define the ground (station) frame as $S$ and the train frame as $S'$. The origins coincide and clocks are synchronized when the train passes the station, so $x=x'=0$ and $t=t'=0$ at that instant.

At time  

$$
t = 8.0\ \text{s},
$$

a worker standing on the ground measures the position of a maintenance cart on the tracks to be

$$
x_S = +520\ \text{m}.
$$

Answer the following:

1. Using the **primed-observer form** of the Galilean transformation,

   $$
   x' = x + u'\,\Delta t',
   $$

   find the coordinate of the cart in the train frame, $x'_{S'}$, at this same instant.

2. Interpret the sign of your answer (what does it mean physically in the train frame?).

Be careful with **units**, **signs**, and **significant figures**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equation**

We use the primed observer’s viewpoint:

$$
x' = x + u'\,\Delta t'
$$

Here:

- $x$ is the coordinate of the event measured in frame $S$  
- $x'$ is the coordinate of the same event measured in frame $S'$  
- $u'$ is the velocity of frame $S$ **as measured by** $S'$  
- $\Delta t'$ is the elapsed time measured in $S'$

---

**Step 1 — Identify $u'$**

In the ground frame, the train moves at $+u$.  
Therefore, in the train frame, the ground moves backward at

$$
u' = -u = -90.0\ \text{m/s}.
$$

---

**Step 2 — Identify $\Delta t'$**

Because clocks are synchronized at the meeting event ($t=t'=0$) and Galilean relativity assumes absolute time, the elapsed time is the same in both frames:

$$
\Delta t' = \Delta t = 8.0\ \text{s}.
$$

---

**Step 3 — Compute $x'_{S'}$**

Use

$$
x'_{S'} = x_S + u'\,\Delta t'.
$$

Substitute:

$$
x'_{S'} = 520 + (-90.0)(8.0).
$$

$$
x'_{S'} = 520 - 720 = -200\ \text{m}.
$$

With proper significant figures:

$$
\boxed{x'_{S'} = -2.0\times 10^2\ \text{m}}
$$

---

**Interpretation**

The negative sign means that, in the train frame, the cart is located **behind** the train’s origin at that instant.  

Even though the cart is ahead of the station in the ground frame, the train has moved farther east than the cart’s location by $t=8.0$ s.

</div>
</details>

```python
%reset -f

import sympy as sp

# ============================================================
# Problem C5.1 — Galilean Position Transformation (Rigorous Form)
# ============================================================
# We use the primed-observer form:
#
#   x' = x + u' * Δt'
#
# where:
#   x   = coordinate of the event measured in frame S  (ground/station)
#   x'  = coordinate of the same event measured in frame S' (train)
#   u'  = velocity of frame S as measured by S'
#   Δt' = elapsed time measured in S'
#
# Galilean assumptions:
#   u' = -u
#   Δt' = Δt  (absolute time in classical mechanics)
# ============================================================

# -----------------------------
# Given values (SI units)
# -----------------------------
u = 90.0      # m/s   (train speed eastward as measured in S)
t = 8.0       # s     (time after passing the station)
x_S = 520.0   # m     (cart position measured in S at time t)

# -----------------------------
# Symbols (for clean math display / rigor)
# -----------------------------
u_sym, t_sym, x_sym = sp.symbols('u t x', real=True)
uprime_sym = sp.Symbol("u'", real=True)
tprime_sym = sp.Symbol("t'", real=True)

# Rigorous transformation: x' = x + u' * t'
xprime_expr = x_sym + uprime_sym * tprime_sym

# Apply Galilean identifications: u' = -u and t' = t
xprime_galilean = sp.simplify(xprime_expr.subs({
    uprime_sym: -u_sym,
    tprime_sym: t_sym
}))

# Numeric evaluation
xprime_num = float(xprime_galilean.subs({u_sym: u, t_sym: t, x_sym: x_S}))

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Problem C5.1 — Galilean Position Transformation\n")

print("Given (frame S / ground):")
print(f"  Train speed:        u = {u:.3g} m/s  (eastward, +x)")
print(f"  Time after pass:    t = {t:.3g} s")
print(f"  Cart position:   x_S = {x_S:.3g} m\n")

print("Rigorous transformation (primed observer form):")
print("  x' = x + u' t'\n")

print("Frame relations in Galilean relativity:")
print("• If S' moves at +u relative to S, then S moves at u' = −u relative to S'.")
print("• Galilean time is absolute, so elapsed time is the same: t' = t.\n")

print("Compute u' and substitute:")
u_prime = -u
print(f"  u' = -u = {u_prime:.3g} m/s")
print(f"  t' = t  = {t:.3g} s\n")

print("Result:")
print(f"  x' = x_S + u' t' = {x_S:.3g} + ({u_prime:.3g})({t:.3g}) = {xprime_num:.3g} m\n")

print("Interpretation:")
print("• A negative x' means the cart is behind the train-origin (as defined at t'=0) in the train frame.")
print("• Even though the cart is ahead of the station (+x in S), the train has moved farther in +x by this time.")
```

## Problem C5.2 — Galilean Position Transformation (3D)

A drone is being tracked near a test range. Define the ground (lab) reference frame as $S$ and a moving truck’s reference frame as $S'$. At the instant the truck passes the lab origin, the coordinate systems are aligned and the clocks are synchronized so that

- $O$ and $O'$ coincide at $t=t'=0$  
- the $x$, $y$, and $z$ axes of $S$ and $S'$ are parallel

The truck moves with constant velocity (as measured in $S$)

$$
\vec{u} = (12.0\,\hat{i} - 4.0\,\hat{j} + 0\,\hat{k})\ \text{m/s}.
$$

At time

$$
t = 6.00\ \text{s},
$$

the lab observer measures the drone’s position to be

$$
\vec{r}_S = (150\,\hat{i} + 90.0\,\hat{j} + 40.0\,\hat{k})\ \text{m}.
$$

Answer the following:

1. Using the Galilean position transformation,

   $$
   \vec{r}' = \vec{r} + \vec{u}'\,\Delta t',
   $$

   determine the drone’s position vector $\vec{r}'_{S'}$ as measured in the truck frame.

2. State the value of $\vec{u}'$ and explain (in one sentence) why it has that sign.

Assume Galilean time so that $\Delta t'=\Delta t$.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant idea**

We use the primed observer’s transformation:

$$
\vec{r}' = \vec{r} + \vec{u}'\,\Delta t'.
$$

Here $\vec{u}'$ is the velocity of frame $S$ as measured by $S'$.  
Since $S'$ moves with velocity $\vec{u}$ relative to $S$, the ground frame moves with the opposite velocity relative to the truck:

$$
\vec{u}' = -\vec{u}.
$$

Also, clocks were synchronized at $t=t'=0$ and Galilean relativity assumes absolute time, so

$$
\Delta t'=\Delta t = t.
$$

---

**Step 1: Compute $\vec{u}'$ and $\Delta t'$**

$$
\vec{u}' = -(12.0\,\hat{i} - 4.0\,\hat{j} + 0\,\hat{k})
= (-12.0\,\hat{i} + 4.0\,\hat{j} + 0\,\hat{k})\ \text{m/s}.
$$

$$
\Delta t' = 6.00\ \text{s}.
$$

---

**Step 2: Compute the frame displacement term**

$$
\vec{u}'\,\Delta t' =
(-12.0\,\hat{i} + 4.0\,\hat{j} + 0\,\hat{k})(6.00)
= (-72.0\,\hat{i} + 24.0\,hat{j} + 0\,\hat{k})\ \text{m}.
$$

---

**Step 3: Apply the transformation**

$$
\vec{r}'_{S'} = \vec{r}_S + \vec{u}'\,\Delta t'
$$

$$
\vec{r}'_{S'} =
(150\,\hat{i} + 90.0\,\hat{j} + 40.0\,\hat{k})
+
(-72.0\,\hat{i} + 24.0\,\hat{j} + 0\,\hat{k}).
$$

So

$$
\vec{r}'_{S'} = (78.0\,\hat{i} + 114\,\hat{j} + 40.0\,\hat{k})\ \text{m}.
$$

---

**Final Answer**

$$
\boxed{\vec{r}'_{S'} = (78.0\,\hat{i} + 114\,\hat{j} + 40.0\,\hat{k})\ \text{m}}
$$

and

$$
\boxed{\vec{u}' = (-12.0\,\hat{i} + 4.0\,\hat{j} + 0\,\hat{k})\ \text{m/s}}.
$$

**One-sentence explanation:** The velocity changes sign because each observer measures the other frame moving in the opposite direction.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# ============================================================
# Problem C5.2 — Galilean Position Transformation (3D)
#
# Rigorous primed-observer form:
#     r' = r + u' Δt'
#
# where:
#   r   = position measured in frame S (ground)
#   r'  = position measured in frame S' (truck)
#   u'  = velocity of frame S as measured by S'
#   Δt' = elapsed time measured in S'
#
# Galilean assumptions:
#   u' = -u
#   Δt' = Δt   (absolute time)
# ============================================================

# -----------------------------
# Given values (SI units)
# -----------------------------
u = np.array([12.0, -4.0, 0.0])        # truck velocity in S (m/s)
t = 6.00                                # time (s)
r_S = np.array([150.0, 90.0, 40.0])     # drone position in S (m)

# -----------------------------
# Apply Galilean relations
# -----------------------------
u_prime = -u            # velocity of ground as measured in truck
t_prime = t             # absolute time assumption

# Compute transformed position
r_prime = r_S + u_prime * t_prime

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Problem C5.2 — Galilean Position Transformation (3D)\n")

print("Given (ground frame S):")
print(f"  r_S  = {r_S} m")
print(f"  u    = {u} m/s")
print(f"  t    = {t:.3g} s\n")

print("Galilean relations:")
print(f"  u'   = -u = {u_prime} m/s")
print(f"  t'   = t  = {t_prime:.3g} s\n")

print("Transformation:")
print("  r' = r + u' t'\n")

print("Compute frame displacement term:")
print(f"  u' t' = {u_prime * t_prime} m\n")

print("Final result (truck frame S'):")
print(f"  r' = {r_prime} m\n")

print("Component form:")
print(f"  x' = {r_prime[0]:.3f} m")
print(f"  y' = {r_prime[1]:.3f} m")
print(f"  z' = {r_prime[2]:.3f} m\n")

print("Physical interpretation:")
print("• Each component transforms independently.")
print("• Only components aligned with the relative motion are shifted.")
print("• The sign change in u' reflects reciprocal motion between frames.")
```

## Problem C5.3 — Length Measurement and Simultaneity

A cart moves along the $+x$ direction in the ground frame $S$ with constant velocity  

$$
v = +5.0\ \text{m/s}.
$$

The cart has a true length of  

$$
L = 2.0\ \text{m}.
$$

At time $t = 0$, the back end of the cart is located at  

$$
x_B(0) = 0.
$$

Thus, at $t=0$ in the ground frame:

- back end: $x_B(0) = 0$  
- front end: $x_F(0) = 2.0\ \text{m}$  

---

### Part A — Simultaneous Measurement

1. Using simultaneous measurements at $t=0$, compute the length of the cart in the ground frame.
2. State clearly why this measurement represents the true length in Galilean relativity.

---

### Part B — Non-Simultaneous Measurement

A student attempts to measure the length of the moving cart but does not measure both endpoints at the same time. Instead:

- The back end is recorded at  

  $$
  t_1 = 0
  $$

- The front end is recorded at  

  $$
  t_2 = 0.40\ \text{s}.
  $$

1. Determine the position of the front end at $t_2$.
2. Compute the “length” the student would report:

   $$
   L_{\text{measured}} = x_F(t_2) - x_B(t_1).
   $$

3. Compare this result to the true length $L$.
4. Explain in one or two sentences why the student’s measurement does not represent the true length of the cart.

Be careful with **units**, **signs**, and the meaning of “simultaneous.”

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Part A**

At $t=0$:

$$
L = x_F(0) - x_B(0) = 2.0 - 0 = 2.0\ \text{m}.
$$

This is valid because both endpoints were measured at the same time in the same frame.

---

**Part B**

Displacement of the cart during $t_2=0.40$ s:

$$
\Delta x = v t_2 = (5.0)(0.40) = 2.0\ \text{m}.
$$

Position of the front end at $t_2$:

$$
x_F(t_2) = 2.0 + 2.0 = 4.0\ \text{m}.
$$

Back end at $t_1=0$:

$$
x_B(t_1) = 0.
$$

Student’s reported length:

$$
L_{\text{measured}} = 4.0 - 0 = 4.0\ \text{m}.
$$

Since

$$
L_{\text{measured}} \neq L,
$$

the student’s measurement is incorrect.

**Explanation:**  
A valid length measurement must compare the positions of both endpoints at the same time in the chosen reference frame. Because the cart moved between the two measurements, the reported value includes additional displacement and does not represent the true length.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# ============================================================
# Problem C5.3 — Length Measurement and Simultaneity
#
# A cart moves with constant velocity.
# We compare:
#   (A) Correct simultaneous length measurement
#   (B) Incorrect non-simultaneous measurement
# ============================================================

# -----------------------------
# Given values
# -----------------------------
v = 5.0              # m/s   (cart velocity)
L_true = 2.0         # m     (true cart length)
t1 = 0.0             # s     (time back end measured)
t2 = 0.40            # s     (time front end measured)

# At t=0:
x_back_0 = 0.0
x_front_0 = L_true

# ============================================================
# Part A — Simultaneous measurement (t = 0)
# ============================================================

L_simultaneous = x_front_0 - x_back_0

# ============================================================
# Part B — Non-simultaneous measurement
# ============================================================

# Position of front end at time t2
x_front_t2 = x_front_0 + v * t2

# Back end measured at time t1
x_back_t1 = x_back_0

# Reported length (incorrect)
L_measured = x_front_t2 - x_back_t1

# ============================================================
# Output
# ============================================================

print("Problem C5.3 — Length Measurement and Simultaneity\n")

print("Given:")
print(f"  Cart velocity:       v = {v:.3g} m/s")
print(f"  True length:         L = {L_true:.3g} m")
print(f"  Back measured at:    t1 = {t1:.3g} s")
print(f"  Front measured at:   t2 = {t2:.3g} s\n")

print("Part A — Simultaneous measurement:")
print(f"  L_simultaneous = {L_simultaneous:.3g} m\n")

print("Part B — Non-simultaneous measurement:")
print(f"  Front position at t2:  x_F(t2) = {x_front_t2:.3g} m")
print(f"  Back position at t1:   x_B(t1) = {x_back_t1:.3g} m")
print(f"  Reported length:       L_measured = {L_measured:.3g} m\n")

print("Conclusion:")
if L_measured != L_true:
    print("  The measured value is NOT equal to the true length.")
    print("  Because the cart moved between measurements,")
    print("  the result includes extra displacement.")
else:
    print("  The measurement matches the true length.")

print("\nPhysical Insight:")
print("• A valid length measurement requires simultaneous")
print("  endpoint measurements in the chosen frame.")
print("• Otherwise, motion contaminates the result.")
```

## Problem C5.4 — Galilean Velocity Transformation

A research boat moves due east (the $+x$ direction) with constant velocity  

$$
u = +6.0\ \text{m/s}
$$

as measured from the shore (ground frame $S$). A drone launches from the boat and flies with constant velocity relative to the boat (boat frame $S'$)

$$
\vec{v}'_{S'} = ( +4.0\,\hat{i} + 3.0\,\hat{j})\ \text{m/s},
$$

where $+y$ points north.

The boat and shore origins coincide and clocks are synchronized at $t=t'=0$.

Answer the following:

1. Using the Galilean velocity transformation,

   $$
   \vec{v}' = \vec{v} + \vec{u}',
   $$

   determine the drone’s velocity as observed from the shore, $\vec{v}_S$.

2. State the value of $\vec{u}'$ and explain (in one sentence) why it has that sign.

3. Find the **speed** of the drone as measured from the shore.

Be careful with **units**, **signs**, and **vector components**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1: Identify the frames and quantities**

- $S$: shore (ground) frame  
- $S'$: boat frame  

Given in the boat frame:

$$
\vec{v}' = (4.0\,\hat{i} + 3.0\,\hat{j})\ \text{m/s}.
$$

The boat moves relative to the shore at

$$
\vec{u} = (6.0\,\hat{i})\ \text{m/s}.
$$

From the boat’s perspective, the shore moves backward with

$$
\vec{u}' = -\vec{u} = (-6.0\,\hat{i})\ \text{m/s}.
$$

---

**Step 2: Use the primed-observer velocity transformation**

The primed observer relation is

$$
\vec{v}' = \vec{v} + \vec{u}'.
$$

Solve for the shore-observed velocity:

$$
\vec{v} = \vec{v}' - \vec{u}'.
$$

Substitute $\vec{u}' = -\vec{u}$:

$$
\vec{v} = \vec{v}' + \vec{u}.
$$

Now compute:

$$
\vec{v} = (4.0\,\hat{i} + 3.0\,\hat{j}) + (6.0\,\hat{i})
= (10.0\,\hat{i} + 3.0\,\hat{j})\ \text{m/s}.
$$

So the drone’s velocity in the shore frame is

$$
\boxed{\vec{v}_S = (10.0\,\hat{i} + 3.0\,\hat{j})\ \text{m/s}}.
$$

---

**Step 3: Speed in the shore frame**

$$
|\vec{v}_S| = \sqrt{10.0^2 + 3.0^2}
= \sqrt{109}
= 10.4\ \text{m/s}.
$$

With appropriate significant figures:

$$
\boxed{|\vec{v}_S| \approx 10\ \text{m/s}}.
$$

**Interpretation:**  
The $x$-component increases because the boat already carries the drone eastward, while the northward component is unchanged because the boat has no $y$-velocity.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# ============================================================
# Problem C5.4 — Galilean Velocity Transformation (3D)
#
# Rigorous primed-observer form:
#     v' = v + u'
#
# where:
#   v'  = velocity measured in moving frame S' (boat)
#   v   = velocity measured in ground frame S (shore)
#   u'  = velocity of S as measured by S'
#
# Galilean relations:
#   u' = -u
#   v  = v' - u' = v' + u
# ============================================================

# -----------------------------
# Given values (SI units)
# -----------------------------
u = np.array([6.0, 0.0])          # boat velocity in S (m/s)
v_prime = np.array([4.0, 3.0])    # drone velocity in boat frame S' (m/s)

# -----------------------------
# Apply Galilean relations
# -----------------------------
u_prime = -u                     # ground velocity as measured in boat frame

# Solve v' = v + u'  →  v = v' - u'
v = v_prime - u_prime            # equivalent to v = v' + u

# Compute speed in ground frame
speed = np.linalg.norm(v)

# -----------------------------
# Output
# -----------------------------
print("Problem C5.4 — Galilean Velocity Transformation (3D)\n")

print("Given:")
print(f"  Boat velocity (ground frame):      u  = {u} m/s")
print(f"  Drone velocity (boat frame):       v' = {v_prime} m/s\n")

print("Galilean relations:")
print(f"  u' = -u = {u_prime} m/s")
print("  v  = v' - u'  (rigorous form)\n")

print("Computed velocity in ground frame:")
print(f"  v  = {v} m/s\n")

print("Component form:")
print(f"  v_x = {v[0]:.3f} m/s")
print(f"  v_y = {v[1]:.3f} m/s\n")

print("Speed in ground frame:")
print(f"  |v| = {speed:.3f} m/s\n")

print("Physical interpretation:")
print("• The x-component increases because the boat already moves east.")
print("• The y-component is unchanged since the boat has no northward motion.")
print("• This illustrates classical Galilean velocity addition.")
```

## Problem C5.5 — Wind Correction for Straight-Line Flight (2D Galilean Velocity Transformation)

A small airplane is trying to fly **due east** (straight ahead along the $+x$ direction) relative to the ground.

- The plane’s airspeed (speed relative to the air) is  

  $$
  v' = 180\ \text{km/hr}.
  $$
  
- A steady wind blows **toward the north** (along $+y$) with speed  

  $$
  u = 50.0\ \text{km/hr}.
  $$

Assume $+x$ is east and $+y$ is north.

Answer the following:

1. What direction (angle) must the pilot aim the plane **relative to the air** so that the plane’s **ground velocity** is due east (no north–south drift)?
2. What is the resulting **ground speed** of the airplane?

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Start with the Galilean velocity transformation**

$$
\vec{v} = \vec{v}' + \vec{u}.
$$

Now identify each quantity:

- $\vec{v}$ is the airplane’s velocity **measured in the ground frame** (ground velocity).  
- $\vec{v}'$ is the airplane’s velocity **measured relative to the air** (airspeed vector).  
- $\vec{u}$ is the velocity of the air **relative to the ground** (the wind velocity).

So the ground velocity is “airspeed + wind.”

---

**Step 1: Write the wind vector $\vec{u}$**

Wind is toward north:

$$
\vec{u} = 0\,\hat{i} + 50.0\,\hat{j}\ \text{km/hr}.
$$

---

**Step 2: Express the plane’s air-velocity vector $\vec{v}'$**

The pilot chooses the direction of $\vec{v}'$, but its magnitude is fixed:

$$
|\vec{v}'| = 180\ \text{km/hr}.
$$

To end up moving due east, the plane must aim **south of east** so that its $y$-component cancels the wind.

Let $\theta$ be the heading angle **south of east**. Then

$$
\vec{v}' = (180\cos\theta)\,\hat{i} + (-180\sin\theta)\,\hat{j}\ \text{km/hr}.
$$

---

**Step 3: Impose “move due east” in the ground frame**

“Due east” means the $y$-component of the ground velocity is zero:

$$
(\vec{v})_y = 0.
$$

From $\vec{v}=\vec{v}'+\vec{u}$, the $y$-component is:

$$
(\vec{v})_y = (\vec{v}')_y + (\vec{u})_y
= (-180\sin\theta) + 50.0.
$$

Set this equal to zero:

$$
-180\sin\theta + 50.0 = 0
$$

$$
180\sin\theta = 50.0
$$

$$
\sin\theta = \frac{50.0}{180} = 0.2778.
$$

So

$$
\theta = \sin^{-1}(0.2778) \approx 16.1^\circ.
$$

**Answer (heading):**

$$
\boxed{\text{Aim }16.1^\circ\ \text{south of east}.}
$$

---

**Step 4: Find the ground speed**

Since the ground motion is due east, the ground speed is the $x$-component of $\vec{v}$:

$$
(\vec{v})_x = (\vec{v}')_x + (\vec{u})_x
= (180\cos\theta) + 0.
$$

So

$$
v_{\text{ground}} = 180\cos(16.1^\circ) \approx 173\ \text{km/hr}.
$$

**Answer (ground speed):**

$$
\boxed{v_{\text{ground}} \approx 1.73\times 10^2\ \text{km/hr}.}
$$

---

**Interpretation**

The wind adds a northward velocity component in the ground frame. The pilot must therefore choose a southward component in $\vec{v}'$ so that, when added using $\vec{v}=\vec{v}'+\vec{u}$, the net $y$-component becomes zero.

```python
%reset -f

import numpy as np

# ============================================================
# Problem C5.5 — Wind Correction (Galilean Velocity Addition)
#
# Start from the Galilean velocity transformation:
#
#     v = v' + u
#
# where:
#   v   = ground velocity (what we want to control)
#   v'  = airplane velocity relative to air (airspeed vector)
#   u   = wind velocity (air relative to ground)
#
# Goal: Make ground velocity purely eastward (zero y-component).
# ============================================================

# -----------------------------
# Given values
# -----------------------------
v_air_mag = 180.0      # km/hr (magnitude of v')
u = np.array([0.0, 50.0])   # wind vector (km/hr), northward

# ------------------------------------------------------------
# Solve for required heading angle theta (south of east)
# ------------------------------------------------------------
# Condition for straight east motion:
#     v_y = v'_y + u_y = 0
#
# Let:
#     v'_y = -v_air_mag * sin(theta)
#
# Then:
#     -v_air_mag sin(theta) + u_y = 0
#     sin(theta) = u_y / v_air_mag
# ------------------------------------------------------------

sin_theta = u[1] / v_air_mag
theta_rad = np.arcsin(sin_theta)
theta_deg = np.degrees(theta_rad)

# ------------------------------------------------------------
# Compute resulting ground velocity
# ------------------------------------------------------------

v_prime = np.array([
    v_air_mag * np.cos(theta_rad),
    -v_air_mag * np.sin(theta_rad)
])

v_ground = v_prime + u
ground_speed = np.linalg.norm(v_ground)

# -----------------------------
# Output
# -----------------------------

print("Problem C5.5 — Wind Correction (Galilean Velocity Transformation)\n")

print("Given:")
print(f"  Airspeed magnitude (v') = {v_air_mag:.3g} km/hr")
print(f"  Wind vector (u)         = {u} km/hr\n")

print("Galilean relation:")
print("  v = v' + u\n")

print("Required heading:")
print(f"  theta = {theta_deg:.3f} degrees south of east\n")

print("Velocity vectors:")
print(f"  v' (air frame)   = {v_prime} km/hr")
print(f"  v  (ground)      = {v_ground} km/hr\n")

print("Ground speed:")
print(f"  |v| = {ground_speed:.3f} km/hr\n")

print("Check:")
print(f"  Ground y-component = {v_ground[1]:.6f} km/hr (should be ~0)")
print("\nPhysical interpretation:")
print("• The pilot aims slightly south so that the southward air-velocity")
print("  cancels the northward wind.")
print("• The ground velocity becomes purely eastward.")
```
