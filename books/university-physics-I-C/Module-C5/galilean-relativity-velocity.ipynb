## C5.4.1 Introduction and Motivation – Galilean Velocity Transformation

So far, we have learned how to relate **positions** measured in two inertial reference frames using the Galilean coordinate transformation. But in mechanics, position is rarely the final goal — we are usually interested in **velocity** and **acceleration**.

Since velocity is defined as the rate of change of position with respect to time, it is natural to ask:

> If two observers disagree on position, how do their velocity measurements compare?

This question arises constantly in everyday life.

- A passenger walking forward inside a moving train has one speed relative to the train.
- The same passenger has a different speed relative to the ground.
- A bird flying alongside a train appears slow to someone on the platform, but may appear to move backward to someone on the train.

Velocity is therefore **frame-dependent**. Unlike length (in Galilean relativity), velocity is not invariant.

In classical mechanics, the relationship between velocities in two inertial frames follows directly from the position transformation. If frame $S'$ moves with velocity $\vec{u}$ relative to frame $S$, then the velocities measured in the two frames are related by simple addition:

$$
\vec{v} = \vec{v}' + \vec{u}.
$$

This rule is known as the **Galilean velocity transformation**.

It formalizes something we already use intuitively:

>Speed of object relative to ground = Speed of object relative to train + Speed of train relative to ground”

This transformation works perfectly for everyday speeds and forms the backbone of Newtonian mechanics.

However, as we will later discover in University Physics II, this simple addition rule does not hold when speeds become comparable to the speed of light. The failure of this velocity transformation is one of the key clues that leads to Special Relativity.

For now, our goal is to understand how and why velocity transforms the way it does in classical mechanics.

<h3 style="margin-top:0; color:#000000;">Box Activity 1 – The Bird’s Speed: Station vs. Train (Common-Sense Relativity)</h3>

In this activity, you will use **everyday intuition** about relative motion to describe how two observers can measure different velocities for the same bird.

**Scenario:**

Your teacher was waiting at Maibara Station when the Nozomi shinkansen blasted through at

$$
u = 350.0\ \text{km/hr}
$$

in the $+x$ direction.

Exactly

$$
t_1 = 3.00\ \text{s}
$$

later, a bird was seen tumbling due to the train’s passage:

- $30.0\ \text{m}$ down the tracks (in the direction of train motion) as observed from the station  
- at a height of $20.0\ \text{m}$ above the tracks

Fortunately, the bird recovered and flew in a straight line down the track in the direction of the train while maintaining its altitude. After an additional

$$
\Delta t = 5.00\ \text{s},
$$

the bird was observed to be

$$
80.0\ \text{m}
$$

down the tracks (still measured from the station).

---

**Tasks**

1. Based on the station observations, what was the bird’s velocity as seen by your teacher?  
2. Without using any formal transformation equation, use common sense to determine the bird’s velocity as seen by a passenger on the train.

Be clear about direction (sign).

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

**1) Bird’s velocity as seen from the station**

From the station:

- Initial position: $x_1 = 30.0\ \text{m}$
- Final position: $x_2 = 80.0\ \text{m}$
- Time interval: $\Delta t = 5.00\ \text{s}$

So the bird’s average velocity (along the track) is:

$$
v_{\text{station}} = \frac{x_2 - x_1}{\Delta t}
= \frac{80.0 - 30.0}{5.00}
= 10.0\ \text{m/s}.
$$

Direction is $+x$ (east), so:

$$
\boxed{v_{\text{station}} = +10.0\ \text{m/s}}
$$

---

**2) Bird’s velocity as seen from the train (intuition)**

Now think like a passenger on the train:

- The train is moving **very fast forward**.
- The bird is moving **forward too**, but much more slowly (only $10.0\ \text{m/s}$ as seen from the ground).
- From the train, almost everything outside appears to move **backward**.

So the passenger should see the bird moving backward at roughly:

> “train speed minus bird speed” (in the forward direction)

First convert the train speed:

$$
350.0\ \text{km/hr} = 97.22\ \text{m/s}.
$$

Then compare speeds:

- Train: $+97.22\ \text{m/s}$
- Bird (ground): $+10.0\ \text{m/s}$

From the train, the bird appears to drift backward at:

$$
v_{\text{train}} \approx 10.0 - 97.22 = -87.22\ \text{m/s}.
$$

So:

$$
\boxed{v_{\text{train}} \approx -87.2\ \text{m/s}}
$$

---

**Interpretation**

- From the ground, the bird moves slowly forward.
- From the train, the bird moves quickly backward.

Both observations are “correct” because velocity depends on the observer’s frame.

## C5.4.2 Velocity Transformation

The velocity transformation is required when an object is moving relative to an observer and a second observer is moving relative to the first observer. While this sounds more complicated, the transformation between velocities are straightforward when we have established the displacement transformation. Even better, as per Example 1 in <i>Galilean Relativity: Coordinate Transformation</i> notes, we have great intuitive understanding of how to transform these velocities. 

Let us consider Equation (1) from the notes on the displacement transformation:

$$ \Delta\vec{r}_{S} = \Delta\vec{r}'_{S'} + \vec{u}\Delta t $$

If we divide through by $\Delta t$ we get:

$$ \frac{\Delta\vec{r}_{S}}{\Delta t} = \frac{\Delta\vec{r}'_{S'}}{\Delta t} + \vec{u} $$

Since $\Delta t = \Delta t'$, we have:

$$ \frac{\Delta\vec{r}_{S}}{\Delta t} = \frac{\Delta\vec{r}'_{S'}}{\Delta t'} + \vec{u} $$

and we have:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Galilean velocity transformation</h3>

    
$$ \vec{v}_{S} = \vec{v}'_{S'} + \vec{u} $$
</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example – Using the Galilean Velocity Transformation Directly</h3>

In Box 1, you may have determined the bird’s velocity in the train frame by computing its change in position relative to the train.

A more direct approach is to apply the **Galilean velocity transformation**.

From the station frame ($S$), the bird’s velocity was found to be

$$
v_S = +10.0\ \text{m/s}.
$$

The relative velocity between the station observer and the train observer is

$$
u = +97.22\ \text{m/s}.
$$

Using the Galilean velocity transformation,

$$
v' = v - u,
$$

we obtain

$$
v' = 10.0 - 97.22 = -87.2\ \text{m/s}.
$$

---

**Interpretation**

- The bird moves slowly forward relative to the station.  
- The train moves much faster forward relative to the station.  
- Therefore, from the train’s perspective, the bird appears to move backward.

The negative sign indicates motion opposite the direction of the train’s travel.

<h3 style="margin-top:0; color:#000000;">Box Activity 2 – Head-On Collision and Momentum Conservation</h3>

In this activity, you will apply the definition of momentum and the principle of **conservation of momentum** in one dimension.

**Scenario:**

In an introductory laboratory experiment, students collide two objects head-on along the $x$-axis.

- Object 1:

  $$
  m_1 = 3.0\ \text{kg}, \qquad v_{1,i} = +4.0\ \text{m/s}
  $$

- Object 2:

  $$
  m_2 = 1.0\ \text{kg}, \qquad v_{2,i} = -3.0\ \text{m/s}
  $$

After the collision, object 2 has velocity

$$
v_{2,f} = +3.0\ \text{m/s}
$$

as measured in the laboratory frame.

---

**Tasks**

1. The momentum of an object is defined as

   $$
   \vec{p} = m\vec{v}.
   $$

   
   What is the **initial momentum of the system** in the laboratory frame?

2. Assuming total momentum remains constant, what is the velocity of object 1 after the collision as measured in the laboratory frame?

Use proper signs throughout.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

**1) Initial momentum of the system**

Momentum of object 1:

$$
p_{1,i} = m_1 v_{1,i} = (3.0)(+4.0) = +12.0\ \text{kg·m/s}
$$

Momentum of object 2:

$$
p_{2,i} = m_2 v_{2,i} = (1.0)(-3.0) = -3.0\ \text{kg·m/s}
$$

Total initial momentum:

$$
p_{\text{tot},i} = 12.0 - 3.0 = +9.0\ \text{kg·m/s}
$$

---

**2) Final velocity of object 1**

Since momentum is conserved:

$$
p_{\text{tot},f} = p_{\text{tot},i} = +9.0\ \text{kg·m/s}.
$$

Final momentum of object 2:

$$
p_{2,f} = m_2 v_{2,f} = (1.0)(+3.0) = +3.0\ \text{kg·m/s}.
$$

Let $v_{1,f}$ be the final velocity of object 1.

Total final momentum:

$$
m_1 v_{1,f} + 3.0 = 9.0.
$$

Solve:

$$
3.0\,v_{1,f} = 6.0
$$

$$
v_{1,f} = +2.0\ \text{m/s}.
$$

---

**Final Answer**

Initial system momentum:

$$
\boxed{p_{\text{tot}} = +9.0\ \text{kg·m/s}}
$$

Final velocity of object 1:

$$
\boxed{v_{1,f} = +2.0\ \text{m/s}}
$$

</details>

<h3 style="margin-top:0; color:#000000;">Box Activity 3 – Momentum Conservation in a Moving Frame</h3>

In this activity, you will revisit the collision from the previous box — but now from the perspective of a **moving observer**.

**Scenario:**

Refer to the collision described in the previous box:

- $m_1 = 3.0\ \text{kg}, \quad v_{1,i} = +4.0\ \text{m/s}$
- $m_2 = 1.0\ \text{kg}, \quad v_{2,i} = -3.0\ \text{m/s}$
- After the collision, $v_{2,f} = +3.0\ \text{m/s}$
- From the previous box, we found $v_{1,f} = +2.0\ \text{m/s}$

Now consider a student walking in the $+x$ direction at

$$
v_{\text{student}} = +2.0\ \text{m/s}
$$

relative to the laboratory frame.

---

**Tasks**

1. What are the **initial velocities** of both objects as observed by the student?  
2. What is the **total initial momentum** of the system as observed by the student?  
3. What are the **final velocities** of both objects as observed by the student?  
4. What is the **total final momentum** of the system as observed by the student?  
5. Compare conservation of momentum in this moving frame to the laboratory frame.

Use proper signs throughout.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

**Step 1: Transform velocities into the student’s frame**

Using everyday intuition (subtract the student’s speed from all lab-frame velocities):

Initial velocities:

$$
v'_{1,i} = 4.0 - 2.0 = +2.0\ \text{m/s}
$$

$$
v'_{2,i} = -3.0 - 2.0 = -5.0\ \text{m/s}
$$

---

**Step 2: Initial momentum in the student’s frame**

$$
p'_{1,i} = (3.0)(2.0) = 6.0\ \text{kg·m/s}
$$

$$
p'_{2,i} = (1.0)(-5.0) = -5.0\ \text{kg·m/s}
$$

Total initial momentum:

$$
p'_{\text{tot},i} = 6.0 - 5.0 = +1.0\ \text{kg·m/s}
$$

---

**Step 3: Final velocities in the student’s frame**

Final lab velocities:

- $v_{1,f} = +2.0\ \text{m/s}$
- $v_{2,f} = +3.0\ \text{m/s}$

Subtract student velocity:

$$
v'_{1,f} = 2.0 - 2.0 = 0
$$

$$
v'_{2,f} = 3.0 - 2.0 = +1.0\ \text{m/s}
$$

---

**Step 4: Final momentum in the student’s frame**

$$
p'_{1,f} = (3.0)(0) = 0
$$

$$
p'_{2,f} = (1.0)(1.0) = +1.0\ \text{kg·m/s}
$$

Total final momentum:

$$
p'_{\text{tot},f} = +1.0\ \text{kg·m/s}
$$

---

**Conclusion**

Initial momentum in student frame:

$$
p'_{\text{tot},i} = +1.0\ \text{kg·m/s}
$$

Final momentum in student frame:

$$
p'_{\text{tot},f} = +1.0\ \text{kg·m/s}
$$

Momentum is conserved in the student’s frame just as it was in the laboratory frame.

Although the **numerical value** of the total momentum changes between frames, the principle of **conservation of momentum holds in all inertial frames**.

This is a powerful illustration of Galilean invariance: **laws of physics MUST be the same for all inertial observers**.

</details>

<h3 style="margin-top:0; color:#000000;">Box Activity 4 – When Galilean Velocity Addition Breaks Down</h3>

In this activity, you will carefully apply the **Galilean velocity transformation** to a high-speed situation and evaluate the result.

**Scenario:**

During a beta decay, a radioactive material ejects two electrons in opposite directions.

From the laboratory frame $S$:

- Electron A has velocity

  $$
  v_A = +0.6c
  $$

- Electron B has velocity

  $$
  v_B = -0.7c
  $$

Here, $c$ is the speed of light in vacuum.

We now ask:

> What is the velocity of electron B as observed from electron A, using Galilean relativity?

---

**Tasks**

1. Begin with the Galilean velocity transformation written in its proper form:

   $$
   v' = v + u'
   $$

   
   Clearly identify what $v$ and $u'$ represent in this problem.

2. Compute the velocity of electron B as observed from electron A.

3. Comment on your result.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

**Step 1: Identify the quantities**

We want the velocity of electron B as measured in the frame moving with electron A.

Using

$$
v' = v + u',
$$

- $v$ is the velocity of electron B measured in the laboratory frame:

  $$
  v = v_B = -0.7c.
  $$

- $u'$ is the velocity of the laboratory frame as measured from electron A’s frame.

Since electron A moves at $+0.6c$ relative to the lab, the lab moves at

$$
u' = -0.6c
$$

relative to electron A.

---

**Step 2: Apply the transformation**

$$
v' = (-0.7c) + (-0.6c).
$$

$$
v' = -1.3c.
$$

So according to Galilean relativity,

$$
\boxed{v' = -1.3c}.
$$

---

**Step 3: Interpretation**

This result predicts that electron B moves at **1.3 times the speed of light** relative to electron A.

This is physically impossible.

No massive object can exceed the speed of light.

---

**Conclusion**

The key insight:

- Galilean velocity addition works for everyday speeds.
- It completely fails when velocities approach the speed of light.

This contradiction is one of the fundamental reasons we must replace Galilean relativity with **Special Relativity** at high velocities.

</details>
