# C5.0 Learning Outcomes — Galilean Transformations
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this module matters

In Phase D, motion was described in **one dimension**. In Phase C, we have now expanded to **two and three dimensions**, introducing vectors as the natural language of motion.

Now we take the next conceptual step:

> Motion is always described **relative to a reference frame**.

A train moving at 20 m/s relative to the road has a different velocity relative to a passenger walking inside the train. A ball thrown straight up inside a train follows a vertical path for one observer—but a different path for someone standing outside.

These are not contradictions. They are **different descriptions of the same physical event**, viewed from different inertial frames.

In this module, we introduce **Galilean transformations** as the mathematical rules that connect measurements made in different inertial frames moving at constant velocity relative to one another.

We focus on:
- How **position, velocity, and acceleration transform**
- How to reason with **vector relative motion**
- What quantities remain **invariant**
- The physical meaning of **classical relativity**

We do **not** yet consider light or high speeds—that belongs to University Physics II (special relativity). Here, we remain entirely within **classical mechanics**.

This module strengthens conceptual clarity and prepares the foundation for later studies of:
- Newton’s Laws in different frames  
- Non-inertial frames  
- Relativity

By the end of this module, you will be able to:

---

### **A. Reference Frames and Relative Motion**
*Recognizes motion as frame-dependent.*

- **A1.** Define a **reference frame** as a coordinate system used to measure position and time.  
- **A2.** Distinguish between **inertial** and **non-inertial** reference frames.  
- **A3.** Explain that velocity is always measured **relative to a chosen frame**.  
- **A4.** Describe everyday examples of relative motion.  

---

### **B. Galilean Position Transformation**
*Connects positions measured in two inertial frames.*

- **B1.** State that if frame $S'$ moves with constant velocity $\vec{v}$ relative to frame $S$, then positions are related by:

  $$
  \vec{r}' = \vec{r} - \vec{v}\Delta t
  $$

- **B2.** Interpret each term physically:
  - $\vec{r}$ = position in frame $S$  
  - $\vec{r}'$ = position in frame $S'$  
  - $\vec{v}$ = velocity of $S'$ relative to $S$  
- **B3.** Apply the transformation in 1D, 2D, and 3D vector form.  

---

### **C. Galilean Velocity Transformation**
*Describes how velocities add.*

- **C1.** Obtain the velocity transformation by differentiating the position relation.  
- **C2.** State the velocity transformation:

  $$
  \vec{u}' = \vec{u} - \vec{v}
  $$

- **C3.** Solve relative-velocity problems using vector addition.  
- **C4.** Interpret classical velocity addition in physical scenarios (boats in rivers, airplanes in wind, walking on trains).  

---

### **D. Acceleration and Invariance**
*Identifies what does not change.*

- **D1.** Show that acceleration is invariant under Galilean transformations:

  $$
  \vec{a}' = \vec{a}
  $$

- **D2.** Explain why Newton’s Second Law has the same form in all inertial frames.  
- **D3.** Recognize acceleration as a **frame-independent quantity** in classical mechanics.  

---

### **E. Graphical and Conceptual Interpretation**
*Develops geometric understanding of frame changes.*

- **E1.** Interpret motion graphs from different frames of reference.  
- **E2.** Explain how the same physical trajectory can appear different in different frames.  
- **E3.** Use vector diagrams to represent relative velocities.  
- **E4.** Distinguish between:
  - Motion of an object  
  - Motion of the coordinate system  

---

### **F. Scope and Preparation for Later Phases**
*Clarifies the limits of classical transformations.*

- **F1.** State that Galilean transformations assume **absolute time measurements**:

  $$
  \Delta t' = \Delta t
  $$

- **F2.** Recognize that these transformations apply only when relative speeds are much less than the speed of light.  
- **F3.** Identify that inconsistencies at high speeds lead to the need for **Special Relativity**, introduced in later course.  
- **F4.** Describe Galilean relativity as a statement that the laws of mechanics are the same in all inertial frames.

---

### **Performance Criteria**

- Relative motion problems are solved using **clear vector reasoning**, not memorized shortcuts.  
- Transformations are interpreted **physically**, not treated as abstract algebra.  
- Students clearly identify:
  - Which frame is being used  
  - The velocity of one frame relative to another  
  - The quantity being transformed  
- Explanations distinguish between:
  - Absolute vs relative descriptions  
  - Object motion vs frame motion  
- Acceleration invariance is understood conceptually, not just symbolically.
