## C5.2.1 Motivation

We may not put much effort into thinking about how to make a length measurements. However, over-thinking this problem will put us at an advantage when digging into special relativity. In physics, we actually have a very clear definition of what a length measurement is:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Proper Length</h3>
    
The length of an object at rest with respect to an observer is determined by measuring the difference between the spatial coordinates between the two end points. Since the object is at rest, we can perform the two measurements at **any** time. We call this length measurement the **proper lengt**.
    
In contrast, an observer moving relatively to the object must measure the positions of the two end points **simultaneously** in order to determine the correct length.
</div>

## C5.2.2 Lenght Measurement Illustrated

In the discussion below, we will label the two end points of the object with $A$ and $B$ and drop the subscripts $S$ and $S'$ as the primed coordinates give away which frame the measurements are observed. Instead, subscripts indicate which end points are measured. We are no longer concerning synchronizing the clocks, but we will consider the same orientation of reference frames.

<img src="C5-fig8.png" width="300">

From the above vector diagram (blue vectors), we have the following:

* Observer $S'$: $L' = x'_{B} - x'_{A}$, where the two positions can be measured at **any time** → proper length.
* Observer $S$: $L = x_{B} - x_{A}$, where the two positions **must** be taken **simultaneously** → contracted length. **In Galilean relativity, the proper length and contracted length are the same**.

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Box Activity 1 – Length Invariance in Galilean Relativity</h3>

In this activity, you will show that **length is invariant** under a Galilean transformation when measured in two inertial frames. Be careful with frame notation, the goal is to keep the notation *rigorous* so it transfers cleanly to Special Relativity later.

**Setup:**

An object lies along the common $x$-axis with endpoints $A$ and $B$.

- Observer $S$ measures the endpoints **at the same time in $S$** and defines

  $$
  L = x_B - x_A.
  $$

- Observer $S'$ measures the endpoints **at the same time in $S'$** and defines

  $$
  L' = x'_B - x'_A.
  $$

Assume the frames were synchronized so that the origins coincide at $t_0=t'_0=0$. Use the primed-observer form of the Galilean transformation:

$$
x' = x + u' \Delta t'
$$

where $u'$ is the velocity of frame $S$ **as measured by** $S'$.

You may also use the Galilean assumptions:

- Relative velocities are opposite:

  $$
  u' = -u
  $$
  
- Elapsed time is the same in both frames:

  $$
  \Delta t' = \Delta t.
  $$

---

### Task

Based on the above, show that the two observers measure the same length:

$$
L' = L.
$$

**HINT:** Express $x'_A$ in terms of $x_A$ and $u'\Delta t'$, and similarly for $x'_B$.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

**Step 1: Transform each endpoint using the primed observer’s equation**

Apply

$$
x' = x + u'\Delta t'
$$

to endpoint $A$:

$$
x'_A = x_A + u'\Delta t'.
$$

Apply it to endpoint $B$:

$$
x'_B = x_B + u'\Delta t'.
$$

---

**Step 2: Form the length in $S'$**

By definition,

$$
L' = x'_B - x'_A.
$$

Substitute the expressions above:

$$
L' = (x_B + u'\Delta t') - (x_A + u'\Delta t').
$$

---

**Step 3: Simplify**

The $u'\Delta t'$ terms cancel:

$$
L' = x_B - x_A.
$$

But $x_B - x_A$ is exactly $L$, so

$$
L' = L.
$$

---

**Interpretation (why this works in Galilean relativity)**

Both endpoints receive the *same* frame-displacement term $u'\Delta t'$, so the separation between them is unchanged. This relies on the Galilean assumptions that time is absolute and simultaneity is shared in a simple way between inertial frames—an assumption that will be modified in Special Relativity.
