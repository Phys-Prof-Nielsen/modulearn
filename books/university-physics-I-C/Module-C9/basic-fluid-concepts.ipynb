## C9.1.1 Motivation: Why Study Fluids?

Fluids are everywhere — in nature, in technology, and even within our own bodies. Understanding how fluids behave allows us to explain and predict a vast range of physical phenomena and engineering applications.

When we studied **Newton’s laws**, **momentum**, and **energy**, we treated matter as discrete particles or rigid bodies. However, many real-world systems cannot be described this way. Air, water, blood, and even the molten rock beneath Earth’s crust are all **continuous materials** that **flow** and **deform** continuously under applied forces. To describe these systems, we must extend the tools of mechanics to **continuous media**, leading us to **fluid dynamics**.

Fluids play a central role in both natural and technological systems:
- The **air we breathe** exerts pressure that enables weather, flight, and sound propagation.  
- The **water cycle** drives climate and life.  
- **Hydraulic systems** use fluid pressure to lift heavy machinery.  
- **Blood flow** delivers oxygen and nutrients through the circulatory system.  
- **Buoyancy and drag** determine how ships float and how fish or submarines move.  
- Even the **forces on airplanes** and the **lift of a bird’s wing** are governed by fluid behavior.

By understanding basic fluid properties like **density** and **pressure**, we establish the foundation for explaining:
- Why pressure increases with depth underwater.  
- How hydraulics multiply force (Pascal’s principle).  
- Why some objects float and others sink (Archimedes’ principle).  
- How pressure differences create motion — from ocean currents to human breathing.

In short, **fluid mechanics connects the microscopic world of molecules to the macroscopic world of motion and forces**. It bridges physics, biology, engineering, and environmental science — making it one of the most broadly applicable topics in classical mechanics.

## C9.1.2 Density

The **density** of a substance is defined as the **mass per unit volume**:

$$
\rho = \frac{m}{V}
$$

where  
- $\rho$ = density (kg/m$^3$),  
- $m$ = mass (kg),  
- $V$ = volume (m$^3$).

For a uniform material, density is constant throughout. 

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C9.1.3 Specific Weight and Specific Gravity

The **specific weight** (also called **weight density**) is the weight per unit volume:

$$
\gamma = \frac{W}{V} = \rho g
$$

where $g$ is the acceleration due to gravity.

The **specific gravity (SG)** is the ratio of the density of a substance to the density of water at $4^\circ$C (approximately $1000 \text{ kg/m}^3$):

$$
\text{SG} = \frac{\rho_\text{substance}}{\rho_\text{water}}
$$

This dimensionless number indicates whether an object will float or sink in water.

## C9.1.4 Pressure in a Fluid

**Pressure** is the **normal force per unit area** exerted by a fluid:

$$
P = \frac{F_\perp}{A}
$$

where  
- $P$ = pressure (Pa or N/m$^2$),  
- $F_\perp$ = force acting perpendicular to the surface,  
- $A$ = surface area.

Pressure is a scalar quantity because it represents the normal component of the stress tensor in a fluid. In a fluid at rest, the stress tensor is isotropic, so the normal stress — the pressure — has the same value for any orientation of the surface element.

## C9.1.5  Absolute and Gauge Pressure

Pressure can be measured in two main ways, depending on the reference point chosen:

---

### Absolute Pressure ($P_\text{abs}$)

Absolute pressure is measured **relative to a perfect vacuum** (zero pressure).  
It represents the **total pressure** exerted by the fluid, including the effects of the surrounding atmosphere.

$$
P_\text{abs} = P_\text{atm} + P_\text{gauge}
$$

where  
- $P_\text{atm}$ = atmospheric pressure (~101.3 kPa at sea level)  
- $P_\text{gauge}$ = gauge pressure (measured relative to the atmosphere)

---

### Gauge Pressure ($P_\text{gauge}$)

Gauge pressure is measured **relative to the local atmospheric pressure** rather than a vacuum.  
This is the type of pressure most commonly used in everyday applications.

$$
P_\text{gauge} = P_\text{abs} - P_\text{atm}
$$

If the absolute pressure is **greater** than atmospheric pressure, the **gauge pressure is positive**.  
If it is **less**, the gauge pressure is **negative**, which is sometimes referred to as a **partial vacuum**.

---

### Examples of Gauge Pressure in Practice

- **Automobile and Bicycle Tires:**  
  When a tire gauge reads “32 psi,” it means **32 psi above atmospheric pressure**.  
  The absolute pressure inside the tire is approximately:

  $$
  P_\text{abs} = 32 \, \text{psi} + 14.7 \, \text{psi} = 46.7 \, \text{psi}
  $$

  
  since 14.7 psi is the average atmospheric pressure at sea level.

- **Hydraulic Systems (e.g., car brakes, lifts):**  
  Hydraulic gauges display the pressure inside the fluid relative to the surrounding air.  
  These pressures can reach thousands of kilopascals, depending on the load being lifted.

- **Medical Applications:**  
  - **Blood Pressure:**  
    A blood pressure reading of 120/80 mmHg is a **gauge pressure** relative to atmospheric pressure.  
    The higher number (systolic) corresponds to the peak pressure when the heart contracts, and the lower number (diastolic) when it relaxes.  
    In absolute terms, the systolic pressure would be:

    $$
    P_\text{abs} = 101.3 \, \text{kPa} + (120 \, \text{mmHg}) = 117.3 \, \text{kPa}
    $$
    
    since $1 \, \text{mmHg} \approx 133.3 \, \text{Pa}$.
  - **Intracranial or Intraocular Pressure:**  
    These are also measured relative to atmospheric pressure to assess internal fluid pressure within the body.

- **Industrial and Laboratory Equipment:**  
  Pressure gauges on pumps, boilers, and gas cylinders all measure **gauge pressure**, allowing operators to quickly see whether pressure is above or below ambient conditions.

- **Vacuum Systems:**  
  A negative gauge pressure indicates a vacuum. For example, a reading of $-50$ kPa gauge means the pressure inside is $50$ kPa **below** atmospheric.

---

### Why Gauge Pressure is Common

Atmospheric pressure is always present and constant under most everyday conditions.  
Because most systems operate relative to this environment, it is more practical to measure **pressure differences** rather than total pressure.

Gauge pressure simplifies design and measurement — what matters for the integrity of a container or the function of a device is **how much greater (or less)** the internal pressure is compared to the outside.
