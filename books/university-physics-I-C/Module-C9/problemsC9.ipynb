## Problem C9.1
A storage tank holds oil with a density of $920\,\text{kg/m}^3$.  
If the tank contains $3.0\,\text{m}^3$ of oil, what is the **mass** and **weight** of the oil?

```{admonition} Problem C8.1
:class: tip, dropdown

**Given:** $\rho = 920\,\text{kg/m}^3,\; V = 3.0\,\text{m}^3,\; g = 9.81\,\text{m/s}^2$

**Mass**

$$
m = \rho V = (920)\,(3.0) = 2{,}760\;\text{kg}.
$$

**Weight**

$$
W = mg = (2{,}760)(9.81) = 2.71\times 10^4\;\text{N}\;\;(\text{more precisely }2.71\times 10^4\text{ N}).
$$

**Answer:** $m = 2.76\times 10^3\ \text{kg},\quad W \approx 2.71\times 10^4\ \text{N}.$
```

## Problem C9.2
The average density of human body tissue is about $985\,\text{kg/m}^3$.  
If a diver has a total body volume of $0.075\,\text{m}^3$, what is the diver’s **mass**?  
Would this diver naturally **sink or float** in freshwater ($\rho = 1000\,\text{kg/m}^3$)?

```{admonition} Problem C8.2
:class: tip, dropdown

**Given:** $\rho_{\text{body}} = 985\,\text{kg/m}^3,\; V_{\text{body}} = 0.075\,\text{m}^3,\; \rho_{\text{freshwater}} = 1000\,\text{kg/m}^3$

**Mass**

$$
m = \rho_{\text{body}} V_{\text{body}} = (985)(0.075) = 73.875\ \text{kg} \approx 73.9\ \text{kg}.
$$

**Sink or float?** Since $\rho_{\text{body}} < \rho_{\text{freshwater}}$, the diver’s average density is slightly lower than water, so the diver tends to **float** gently at the surface (neutral-ish buoyancy).

**Answer:** $m \approx 73.9\ \text{kg}$, diver tends to **float** in freshwater.

```

## Problem C9.3
A submarine is submerged in seawater ($\rho = 1030\,\text{kg/m}^3$) at a depth of $250\,\text{m}$.  
Find the **absolute pressure** on the submarine’s hull if the atmospheric pressure at the surface is $1.01\times10^5\,\text{Pa}$.

```{admonition} Problem C8.3
:class: tip, dropdown

**Given:** $\rho = 1030\,\text{kg/m}^3,\; h = 250\,\text{m},\; g = 9.81\,\text{m/s}^2,\; P_{\text{atm}} = 1.01\times 10^5\,\text{Pa}$

**Hydrostatic increase**

$$
\Delta P = \rho g h = (1030)(9.81)(250) \approx 2.53\times 10^6\ \text{Pa}.
$$

**Absolute pressure at depth**

$$
P_{\text{abs}} = P_{\text{atm}} + \Delta P 
= 1.01\times 10^5 + 2.53\times 10^6 
\approx 2.63\times 10^6\ \text{Pa}.
$$

**Answer:** $P_{\text{abs}} \approx 2.63\ \text{MPa}$ (about $26$ atmospheres).
```

## Problem C9.5
A car hydraulic lift has a small piston with an area of $0.010\,\text{m}^2$ and a large piston with an area of $0.50\,\text{m}^2$.  
If a mechanic applies a force of $200\,\text{N}$ to the small piston, what **force** is exerted on the car by the large piston?  
Ignore any height differences.

```{admonition} Problem C8.5
:class: tip, dropdown

**Given:** $A_1 = 0.010\,\text{m}^2,\; A_2 = 0.50\,\text{m}^2,\; F_1 = 200\,\text{N}$

**Pascal’s Principle applies to the *change* in pressure**

$$
\Delta P_1 = \Delta P_2 \;\;\Rightarrow\;\; \frac{F_1}{A_1} = \frac{F_2}{A_2}.
$$

**Solve for $F_2$**

$$
F_2 = F_1 \frac{A_2}{A_1} 
= (200)\,\frac{0.50}{0.010} 
= 10{,}000\ \text{N}.
$$

**Energy consistency (conceptual):** $F_1 d_1 = F_2 d_2$ so the large piston moves a shorter distance.

**Answer:** $F_2 = 1.0\times 10^4\ \text{N}$.

```

## Problem C9.6 
A patient’s **blood pressure cuff (sphygmomanometer)** inflates until it applies a pressure of $130\,\text{mmHg}$ uniformly around the arm.  
(a) Convert this to pascals.  
(b) If this pressure acts over an area of $50\,\text{cm}^2$, what total **force** is applied on that section of the arm?

```{admonition} Problem C8.6
:class: tip, dropdown

**Given:** Cuff pressure $=130\,\text{mmHg}$, $1\,\text{mmHg}\approx 133.3\,\text{Pa}$, area $A = 50\,\text{cm}^2 = 0.005\,\text{m}^2$

**Convert pressure**

$$
P = (130)(133.3) = 1.733\times 10^4\ \text{Pa}.
$$

**Force from uniform pressure on area**

$$
F = P A = (1.733\times 10^4)(0.005) \approx 86.6\ \text{N}.
$$

**Answer:** The cuff applies about **$8.66\times 10^1$ N** on that section of the arm.

```

## Problem C9.7
A solid aluminum block ($\rho_{\text{Al}} = 2700\,\text{kg/m}^3$) is completely submerged in water.  
If its volume is $0.004\,\text{m}^3$, determine the **buoyant force** and the **apparent weight** of the block underwater.

```{admonition} Problem C8.7
:class: tip, dropdown

**Given:** $\rho_{\text{water}}=1000\,\text{kg/m}^3,\; \rho_{\text{Al}}=2700\,\text{kg/m}^3,\; V_{\text{block}}=0.004\,\text{m}^3,\; g=9.81\,\text{m/s}^2$

**Buoyant force**

$$
F_B = \rho_{\text{water}} g V = (1000)(9.81)(0.004) = 39.24\ \text{N}.
$$

**Weight of block in air**

$$
W = \rho_{\text{Al}} V g = (2700)(0.004)(9.81) = 105.95\ \text{N}\;(\approx 1.06\times 10^2\ \text{N}).
$$

**Apparent weight underwater**

$$
W_{\text{app}} = W - F_B = 105.95 - 39.24 = 66.71\ \text{N}.
$$

**Answer:** $F_B \approx 39.2\ \text{N}$, apparent weight $\approx 66.7\ \text{N}$.

```

## Problem C9.8 
A person performing hydrostatic body composition testing is submerged in water.  
Their mass in air is $70.0\,\text{kg}$, and their apparent mass underwater is $3.0\,\text{kg}$ (buoyant effect from water, $\rho = 1000\,\text{kg/m}^3$).  
(a) Determine the person’s **body volume**.  
(b) Calculate their **average body density** and **body fat percentage** using the rule of thumb:

$$
\text{Body Fat (\%)} = 495 / \rho_{\text{body}} - 450
$$

where $\rho_{\text{body}}$ is in kg/L.

```{admonition} Problem C8.8
:class: tip, dropdown

**Given:** $m_{\text{air}}=70.0\,\text{kg}$, apparent mass underwater $=3.0\,\text{kg}$, $\rho_{\text{water}}=1000\,\text{kg/m}^3,\; g=9.81\,\text{m/s}^2$

**Buoyant force equals the *loss of weight***

$$
F_B = (m_{\text{air}} - m_{\text{underwater}})g = (70.0 - 3.0)(9.81) = 67(9.81) = 657.3\ \text{N}.
$$

**Body volume from displaced water**

$$
V_{\text{body}} = \frac{F_B}{\rho_{\text{water}} g} = \frac{657.3}{(1000)(9.81)} \approx 0.0670\ \text{m}^3.
$$

**Average body density**

$$
\rho_{\text{body}} = \frac{m}{V} = \frac{70.0}{0.0670} \approx 1{,}044.8\ \text{kg/m}^3 = 1.0448\ \text{kg/L}.
$$

**Body fat percentage (Siri equation)**

$$
\text{BF\%} = \frac{495}{\rho_{\text{body}}(\text{kg/L})} - 450 
= \frac{495}{1.0448} - 450 \approx 23.8\%.
$$

**Answer:** $V \approx 0.0670\ \text{m}^3$, $\rho_{\text{body}} \approx 1.045\ \text{kg/L}$, body fat $\approx 23.8\%$.

```
