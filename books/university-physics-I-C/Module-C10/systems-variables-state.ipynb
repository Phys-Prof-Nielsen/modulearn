## Why this section matters

In Phase D, we focused on **heat and energy transfer** — tracking how energy flows into and out of a system.

Now we make an important conceptual shift:

> Thermodynamics is fundamentally a theory of **states**, not just energy transfer.

A thermodynamic system is completely described by a small set of measurable quantities. Once those are known, the system’s equilibrium condition is determined.

In this section, we formalize:

- What a **thermodynamic system** is  
- What qualifies as a **state variable**  
- The difference between **state functions** and **path-dependent quantities**  
- The **Ideal Gas Law** as an equation connecting equilibrium states  

This is the foundation for all PV diagrams and thermodynamic processes that follow.

## 1. Thermodynamic Systems

A **thermodynamic system** is a defined portion of the universe chosen for analysis.

Everything outside the system is called the **surroundings**.

The surface separating them is the **boundary**.

---

### Types of Systems

**Open System**
- Exchanges energy and matter with surroundings.

**Closed System**
- Exchanges energy, but not matter.

**Isolated System**
- Exchanges neither energy nor matter.

---

In this course, we will primarily analyze **closed systems** containing a gas.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## 2. Equilibrium and State

A thermodynamic system is in **equilibrium** when its measurable properties are uniform and not changing with time.

When in equilibrium, the system can be described by a small set of measurable quantities called:

> **State variables**

Common state variables include:

- Pressure $P$  
- Volume $V$  
- Temperature $T$  
- Internal energy $U$  

Once these are specified, the system’s macroscopic state is fixed.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## 3. State Functions vs Path-Dependent Quantities

This distinction is central to Phase C thermodynamics.

### State Functions

A **state function** depends only on the current state of the system, not on how the system reached that state.

Examples:

- $U$ (internal energy)  
- $P$  
- $V$  
- $T$  

For a change between two states:

$$
\Delta U = U_f - U_i
$$

The value depends only on the initial and final states.

---

### Path-Dependent Quantities

Some quantities depend on the *process* taken between states.

Examples:

- Heat $Q$  
- Work $W$  

Two different paths between the same initial and final states can yield:

- Different values of $Q$
- Different values of $W$

But the same value of $\Delta U$.

This leads to the First Law:

$$
\Delta U = Q - W
$$

The change in internal energy is fixed by the endpoints — even though $Q$ and $W$ individually depend on the path.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example – Heat Depends on the Path</h3>

Consider a closed container holding an ideal gas.

Initial state:
- Temperature $T_1$
- Internal energy $U_1$

Final state:
- Temperature $T_2$
- Internal energy $U_2$

For an ideal gas:

$$
\Delta U = U_2 - U_1
$$

and internal energy depends only on temperature.

So if $T_2 - T_1$ is fixed, then $\Delta U$ is fixed.

---

**Path A – Gas is Heated While Free to Expand**

As the gas warms:

- It expands.
- It pushes outward on its surroundings.
- It does work on the environment.

Because energy leaves the system as work, extra heat must enter to reach the same final temperature.

So for this path:

$$
Q_A > \Delta U
$$

---

**Path B – Gas is Heated in a Rigid Container**

Now the container does not allow expansion.

- The gas cannot change volume.
- No work is done.

From the First Law:

$$
\Delta U = Q_B
$$

---

**Comparison**

Both paths have the same:

$$
\Delta U
$$

But:

$$
Q_A > Q_B
$$

The heat added depends on what happens during the heating.

---

**Key Idea**

Internal energy change depends only on temperature (state).

Heat depends on how the system is allowed to change (path).

This is why:

- $U$ is a **state function**
- $Q$ is **not**
</div>

## 4. The Ideal Gas Law — Equation of State

Up to this point, we have introduced **state variables** such as $P$, $V$, $T$, and $U$.  

But an important question arises:

> Are these variables independent?

For many real systems, the answer is **no**. The measurable state variables are connected by a relationship called an:

> **Equation of state: An equation of state is a mathematical relationship between the state variables of a system in equilibrium**.

---

### The Ideal Gas Law

The simplest and most important equation of state for gases is:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">The Ideal Gas Law</h3>

$$
PV = nRT
$$

</div>

where:

- $P$ = pressure  
- $V$ = volume  
- $n$ = number of moles  
- $R$ = universal gas constant  
- $T$ = absolute temperature  

This is the form most commonly used in chemistry and introductory thermodynamics.

---

### The Ideal Gas Law (Microscopic / Physics Form)

In physics, we often use a more fundamental version:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">The Ideal Gas Law</h3>
    
$$
PV = NkT
$$

</div>

where:

- $N$ = number of particles  
- $k$ = Boltzmann constant  
- $T$ = absolute temperature  

This form connects macroscopic measurements directly to the number of microscopic particles.

The two forms are completely equivalent because:

$$
nR = Nk
$$

since:

- $n = \frac{N}{N_A}$
- $R = N_A k$

where $N_A$ is Avogadro’s number.

---

### Why Physicists Often Prefer $NkT$

The $NkT$ form is especially useful because:

- It connects thermodynamics to microscopic physics.
- It emphasizes that temperature relates to particle motion.
- It prepares the groundwork for statistical mechanics (Phase A).

---

### What Do We Mean by "Ideal"?

An ideal gas is a simplified model of a gas in which:

- The particles occupy negligible volume.
- There are no intermolecular forces (no attraction or repulsion).
- Collisions between particles are perfectly elastic.
- The gas is sufficiently dilute.

Under many ordinary laboratory conditions (low pressure, moderate temperature), real gases behave approximately like ideal gases.

---

### Why It Is Called the Simplest Gas Law

The Ideal Gas Law assumes:

- No particle interactions  
- No finite particle size  
- Perfectly random motion  

Because of these simplifying assumptions, it has a very simple algebraic form and is easy to use.

It is the **starting point** for thermodynamic analysis in this course and just as the point particle was the starting point in Newtonian mechanics.

---

### Real Gases and More Advanced Models

Real gases deviate from ideal behavior at:

- High pressures  
- Low temperatures  
- Conditions near condensation  

More refined equations of state account for these effects. For example:

- **Van der Waals equation** — Corrects for finite molecular size and intermolecular attraction.  
- **Redlich–Kwong equation** — Improves accuracy at higher pressures.  
- **Virial equation of state** — Expresses deviations from ideal behavior in a power series form.  

These models are more accurate but also more mathematically complex.

In this course, we will primarily use the Ideal Gas Law because:

- It captures essential thermodynamic behavior.
- It keeps our focus on core principles.
- It allows us to develop geometric and conceptual reasoning before introducing deeper complexity.

---

### Conceptual Takeaway

The Ideal Gas Law teaches us something fundamental:

State variables are not independent — they are connected.

Once three of the variables are specified, the fourth is determined.

This relationship allows us to:

- Predict how a system responds to changes.
- Connect measurable quantities.
- Prepare for graphical representations in PV diagrams.

The Ideal Gas Law is therefore not just an equation — it is the structural backbone of Phase C thermodynamics.

## 5. Internal Energy of an Ideal Gas

Experiments show that for an ideal gas, internal energy depends only on temperature.

For a monatomic ideal gas:

$$
U = \frac{3}{2} nRT
$$

or equivalently,

$$
U = \frac{3}{2} NkT
$$

Therefore,

$$
\Delta U = \frac{3}{2} nR \Delta T
$$

---

### Why Are We Allowed to Use This?

At this stage, we will **not derive** this result.

For now, you may treat it as an experimentally established property of ideal gases:

- Internal energy is proportional to temperature.
- It does not depend explicitly on pressure or volume.

This allows us to analyze thermodynamic processes using only macroscopic reasoning.

---

### Where Does This Actually Come From?

The factor $\frac{3}{2}$ is not arbitrary.

It comes from microscopic physics:

- Gas particles possess kinetic energy.
- Temperature is related to average kinetic energy.
- Each degree of freedom contributes $\frac{1}{2}kT$ per particle.

The full derivation requires **statistical mechanics**, which we will develop in Phase A.

At that point, this equation will no longer be something to memorize — it will emerge naturally from particle motion.

---

### Why Include It Now?

We include this equation in Phase C because:

- It allows us to compute internal energy changes.
- It connects temperature to stored energy.
- It prepares the bridge between macroscopic thermodynamics and microscopic physics.

For now, treat it as a powerful empirical result.

Later, we will uncover its deeper origin.
