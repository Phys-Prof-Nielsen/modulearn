## Why this module matters

In Phase D, we studied **temperature, heat, heat transfer, and the First Law of Thermodynamics**. Energy was tracked carefully as it moved into and out of systems.

Now we take the next conceptual step:

> Thermodynamics is not just about energy transfer — it is about **state and process**.

A gas can move from one state to another in many different ways. The heat transferred may differ. The work done may differ. Yet the change in internal energy is the same.

These are not contradictions. They reveal something deeper:

- Some quantities depend only on the **state**.
- Others depend on the **path taken**.

In this module, we formalize thermodynamics as a theory of:

- **State variables**
- **Thermodynamic processes**
- **PV diagrams**
- **Work as geometric area**
- **Cyclic processes and engines**

We do **not** yet use calculus or statistical mechanics — those belong to Phase B and Phase A. Here, we develop geometric reasoning and system-level thinking.

This module prepares the foundation for:

- Differential thermodynamics  
- Entropy and the Second Law  
- Statistical mechanics

## Learning Outcomes

By the end of this module, you will be able to:

---

### **A. Thermodynamic Systems and State Variables**
*Recognizes thermodynamics as a theory of state.*

- **A1.** Define a **thermodynamic system**, surroundings, and boundary.  
- **A2.** Distinguish between **open, closed, and isolated systems**.  
- **A3.** Identify common **state variables**:
  - Pressure $P$  
  - Volume $V$  
  - Temperature $T$  
  - Internal energy $U$  
- **A4.** Distinguish between:
  - **State functions** (e.g., $U$)  
  - **Path-dependent quantities** (e.g., $Q$, $W$)  
- **A5.** Explain why $\Delta U$ depends only on initial and final states.

---

### **B. Ideal Gas Law as an Equation of State**
*Uses relationships between thermodynamic variables.*

- **B1.** Apply the ideal gas law:

  $$
  PV = nRT
  $$

- **B2.** Solve multi-step state changes using algebraic reasoning.  
- **B3.** Determine unknown state variables given partial information.  
- **B4.** Recognize the ideal gas law as a relationship between equilibrium states.

---

### **C. Work and PV Diagrams**
*Develops geometric reasoning in thermodynamics.*

- **C1.** Interpret a **PV diagram** as a representation of thermodynamic states.  
- **C2.** State that work done by a gas equals the area under the curve:

  $$
  W = \text{area under } P(V)
  $$

- **C3.** Compute work for:
  - Isobaric processes  
  - Linear processes  
  - Simple multi-step paths  
- **C4.** Compare work done along different paths between the same two states.  
- **C5.** Interpret the area enclosed in a cycle as **net work**.

---

### **D. Special Thermodynamic Processes**
*Analyzes energy flow in common processes.*

- **D1.** Describe and analyze:
  - Isobaric processes  
  - Isochoric processes  
  - Isothermal processes  
  - Adiabatic processes (qualitative)  
- **D2.** Apply the First Law:

  $$
  \Delta U = Q - W
  $$

- **D3.** Determine which quantities are zero in each process.  
- **D4.** Compare how energy flows differ between processes.

---

### **E. Heat Capacities and Energy Flow**
*Connects temperature change to internal energy.*

- **E1.** Use constant-volume and constant-pressure heat relations:

  $$
  Q = nC_V \Delta T
  $$

  $$
  Q = nC_P \Delta T
  $$

- **E2.** Explain physically why $C_P > C_V$.  
- **E3.** Relate heat capacity to internal energy changes.  
- **E4.** Analyze temperature change in multi-step processes.

---

### **F. Cyclic Processes and Heat Engines (Introductory)**
*Connects thermodynamics to real-world systems.*

- **F1.** Identify a cyclic process on a PV diagram.  
- **F2.** Define heat engine efficiency:

  $$
  e = \frac{W_{\text{out}}}{Q_{\text{in}}}
  $$

- **F3.** Interpret energy flow diagrams for engines.  
- **F4.** Explain why no engine can convert all heat into work (conceptually).

---

### **G. Entropy — Conceptual Introduction**
*Introduces directionality in thermodynamics.*

- **G1.** Describe entropy qualitatively as a measure of irreversibility.  
- **G2.** Identify processes in which entropy increases.  
- **G3.** Explain why spontaneous processes have a preferred direction.  
- **G4.** Recognize entropy as a state function (without calculus).

---

### **Performance Criteria**

- PV diagrams are interpreted **geometrically and physically**, not memorized.  
- Work is computed through **area reasoning**, not calculus.  
- Students clearly distinguish between:
  - State variables  
  - Path-dependent quantities  
- Multi-step thermodynamic problems are solved with:
  - Clear identification of states  
  - Proper use of the First Law  
  - Logical energy accounting  
- Efficiency calculations are physically interpreted, not treated as pure algebra.  
- Explanations demonstrate understanding of:
  - Why work depends on path  
  - Why internal energy does not  
  - Why cyclic processes produce net work
