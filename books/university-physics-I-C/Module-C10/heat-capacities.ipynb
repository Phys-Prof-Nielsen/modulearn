# C10.4 Heat Capacities at Constant Pressure and Constant Volume
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this section matters

In C10.3, we studied special thermodynamic processes and how energy flows differently depending on what is held constant.

Now we focus on an important quantitative question:

> How much heat is required to change the temperature of a system?

The answer depends on:

- The type of substance  
- The amount of substance  
- The process the system undergoes  

In thermodynamics, the relevant quantities are called **heat capacities**.

Heat capacities allow us to connect:

- Heat transfer $Q$  
- Temperature change $\Delta T$  
- Internal energy change $\Delta U$  

They also help explain why heating at constant pressure differs from heating at constant volume.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## 1. Definition of Heat Capacity

The heat capacity of a system describes how much heat is required to change its temperature.

For $n$ moles of a substance:

$$
Q = nC \Delta T
$$

where:

- $C$ = molar heat capacity  
- $\Delta T$ = change in temperature  

However, in thermodynamics, heat capacity depends on the **process**.

We therefore define two important heat capacities:

- $C_V$ → heat capacity at constant volume  
- $C_P$ → heat capacity at constant pressure  

---

## 2. Heat Capacity at Constant Volume ($C_V$)

In a constant-volume process:

$$
V = \text{constant}
$$

From C10.3:

$$
W = 0
$$

From the First Law:

$$
\Delta U = Q
$$

Therefore, at constant volume:

$$
Q = nC_V \Delta T
$$

Since all heat goes into changing internal energy:

$$
\Delta U = nC_V \Delta T
$$

This makes $C_V$ directly related to how internal energy depends on temperature.

---

## 3. Heat Capacity at Constant Pressure ($C_P$)

In a constant-pressure process:

$$
P = \text{constant}
$$

The gas is allowed to expand as it heats.

That means:

- The system does work.
- Not all the heat increases internal energy.

We define:

$$
Q = nC_P \Delta T
$$

But from the First Law:

$$
\Delta U = Q - W
$$

Since $W \neq 0$ at constant pressure, more heat is required to produce the same temperature change.

Therefore:

$$
C_P > C_V
$$

---

## 4. Why Is $C_P$ Larger Than $C_V$?

This is one of the most important conceptual results in thermodynamics.

At constant volume:

- No work is done.
- All heat increases internal energy.

At constant pressure:

- The gas expands.
- Some heat becomes work.
- Additional heat is required to achieve the same $\Delta T$.

Therefore:

$$
C_P = C_V + R
$$

for an ideal gas.

We will derive this relationship more formally in Phase B.

For now, treat it as a standard result for ideal gases.

---

## 5. Physical Interpretation

Heat capacity tells us how "resistant" a substance is to temperature change.

Large heat capacity → large heat required for small $\Delta T$  
Small heat capacity → temperature changes easily  

The distinction between $C_P$ and $C_V$ reminds us:

> The amount of heat required depends not only on the material, but on how the system is allowed to change.
