# C10.3 Special Thermodynamic Processes
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this section matters

In C10.2, we learned to represent thermodynamic states and changes on **PV diagrams** and to interpret work as geometric area.

Now we take the next step:

> We classify common types of thermodynamic paths.

Certain processes occur so frequently in physics and engineering that they have special names. These processes are defined by which state variable is held constant during the change.

Understanding these processes allows us to:

- Predict how $P$, $V$, and $T$ change relative to one another.
- Determine whether work is done.
- Apply the First Law systematically.
- Compare different energy-transfer mechanisms.

This section builds structured reasoning before we move to more mathematical treatments in Phase B.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## 1. Isobaric Process (Constant Pressure)

An **isobaric process** occurs when pressure remains constant:

$$
P = \text{constant}
$$

On a PV diagram, this appears as a **horizontal line**.

If volume changes from $V_1$ to $V_2$, the work done is:

$$
W = P (V_2 - V_1)
$$

If $V_2 > V_1$:
- The gas expands.
- $W > 0$.

From the First Law:

$$
\Delta U = Q - W
$$

Heat must both:
- Increase internal energy, and
- Supply the work done during expansion.

---

## 2. Isochoric Process (Constant Volume)

An **isochoric process** occurs when volume remains constant:

$$
V = \text{constant}
$$

On a PV diagram, this appears as a **vertical line**.

Since volume does not change:

$$
W = 0
$$

Therefore, from the First Law:

$$
\Delta U = Q
$$

All heat added changes the internal energy.

This process is conceptually simple:
- No mechanical work is performed.
- Energy transfer occurs purely as heat.

---

## 3. Isothermal Process (Constant Temperature)

An **isothermal process** occurs when temperature remains constant:

$$
T = \text{constant}
$$

For an ideal gas:

$$
PV = \text{constant}
$$

On a PV diagram, this produces a **curved path**.

---

### Internal Energy

For an ideal gas, internal energy depends only on temperature.

Since temperature does not change:

$$
\Delta U = 0
$$

From the First Law:

$$
\Delta U = Q - W
$$

Therefore:

$$
Q = W
$$

Any heat added to the system is entirely converted into work.

---

### Work Done During an Isothermal Expansion

For an ideal gas undergoing an isothermal change from $V_1$ to $V_2$, the work done is:

$$
W = nRT \ln\!\left(\frac{V_2}{V_1}\right)
$$

This equation allows us to calculate work directly without using geometry.

---

### Why Are We Allowed to Use This?

At this stage, we will not derive this expression.

It comes from summing many small contributions of work during continuous expansion — a calculation that requires calculus.

We will derive this formula carefully in **Phase B**, when we introduce:

$$
W = \int P\, dV
$$

For now, treat this as an established result for ideal gases.

---

### Physical Interpretation

- If $V_2 > V_1$, then:

  $$
  W > 0
  $$

  
  The gas expands and does work.

- If $V_2 < V_1$, then:

  $$
  W < 0
  $$

  Work is done on the gas.

This result shows clearly that even though temperature does not change, mechanical work can still be produced — provided heat flows into the system.

---

## 4. Adiabatic Process (No Heat Transfer)

An **adiabatic process** occurs when no heat enters or leaves the system:

$$
Q = 0
$$

From the First Law:

$$
\Delta U = -W
$$

If the gas expands:
- $W > 0$
- $\Delta U < 0$
- Temperature decreases.

If the gas is compressed:
- $W < 0$
- $\Delta U > 0$
- Temperature increases.

Adiabatic processes involve energy transfer only through mechanical work.

A deeper mathematical relationship between $P$ and $V$ in adiabatic processes will be developed in Phase B.

---

## 5. Comparing the Four Processes

Each process is defined by what remains constant:

- Isobaric → constant $P$
- Isochoric → constant $V$
- Isothermal → constant $T$
- Adiabatic → $Q = 0$

Each has a distinct energy-flow pattern:

- Isobaric: heat splits between $\Delta U$ and $W$
- Isochoric: all heat changes internal energy
- Isothermal: all heat becomes work
- Adiabatic: work changes internal energy

These distinctions are essential for understanding engines and cycles.
