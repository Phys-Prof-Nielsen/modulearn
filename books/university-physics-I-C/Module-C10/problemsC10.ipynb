## Problem C10.1 — Thermodynamic State Variables and the Ideal Gas Law

A sealed cylinder contains an ideal gas. Treat the gas as the **system**. The gas is initially in equilibrium with:

$$
P_1 = 1.20\times 10^5\ \text{Pa}, \qquad
V_1 = 4.00\times 10^{-3}\ \text{m}^3, \qquad
T_1 = 300\ \text{K}.
$$

The gas is later found in a new equilibrium state with:

$$
P_2 = 2.40\times 10^5\ \text{Pa}, \qquad
T_2 = 360\ \text{K}.
$$

Assume the amount of gas does not change and the gas remains ideal.

Answer the following:

1. Use the ideal gas law to determine the number of moles $n$ in the cylinder.
2. Find the final volume $V_2$.
3. List two examples of **state variables** in this problem and two examples of **path-dependent quantities** in thermodynamics.
4. Based only on the information given, can you determine the values of $Q$ and $W$ for the change from state 1 to state 2? Explain briefly.

Be careful with **units**, **scientific notation**, and **significant figures**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equation**

For an ideal gas:

$$
PV = nRT
$$

with $R = 8.314\ \text{J/(mol·K)}$.

---

**1) Solve for the number of moles $n$**

Use the initial state:

$$
n = \frac{P_1V_1}{RT_1}.
$$

Compute:

$$
P_1V_1 = (1.20\times 10^5)(4.00\times 10^{-3})
= 4.80\times 10^2\ \text{J}.
$$

Then:

$$
n = \frac{4.80\times 10^2}{(8.314)(300)}
= \frac{480}{2494.2}
= 0.192\ \text{mol}.
$$

So:

$$
\boxed{n \approx 0.192\ \text{mol}}
$$

---

**2) Solve for the final volume $V_2$**

Use the final state:

$$
V_2 = \frac{nRT_2}{P_2}.
$$

Substitute:

$$
V_2 = \frac{(0.192)(8.314)(360)}{2.40\times 10^5}.
$$

First compute the numerator:

$$
(0.192)(8.314)(360) \approx 5.75\times 10^2\ \text{J}.
$$

Then:

$$
V_2 = \frac{5.75\times 10^2}{2.40\times 10^5}
= 2.40\times 10^{-3}\ \text{m}^3.
$$

So:

$$
\boxed{V_2 \approx 2.40\times 10^{-3}\ \text{m}^3}
$$

---

**3) State variables vs path-dependent quantities**

State variables in this problem include (any two):

- $P$ (pressure)
- $V$ (volume)
- $T$ (temperature)

Path-dependent quantities in thermodynamics include (any two):

- Heat $Q$
- Work $W$

---

**4) Can we determine $Q$ and $W$?**

No.

Even though we know the initial and final states (so we can determine things like $P$, $V$, and $T$ at each state), the quantities $Q$ and $W$ depend on the **path** taken between the states.

Two different processes could connect the same state 1 and state 2 and produce different values of $Q$ and $W$, while still satisfying the First Law:

$$
\Delta U = Q - W.
$$

So we can say:

- The endpoints determine the state change (and therefore $\Delta U$ for an ideal gas once we later connect $U$ to $T$),
- But $Q$ and $W$ are not determined without knowing *how* the change occurred.

</details>

```python
%reset -f

import sympy as sp

# ============================================================
# Problem C10.1 — Thermodynamic State Variables & Ideal Gas Law
# ============================================================
# We use the ideal gas law:
#
#   P V = n R T
#
# Step 1: Use state 1 to solve for n:
#
#   n = (P1 V1) / (R T1)
#
# Step 2: Use state 2 to solve for V2:
#
#   V2 = (n R T2) / P2
#
# Notes (Phase C framing):
# - P, V, T are state variables.
# - Q and W are path-dependent quantities, so they cannot be determined
#   from endpoint states alone.
# ============================================================

# -----------------------------
# Given values (SI units)
# -----------------------------
P1 = 1.20e5      # Pa
V1 = 4.00e-3     # m^3
T1 = 300.0       # K

P2 = 2.40e5      # Pa
T2 = 360.0       # K

R = 8.314        # J/(mol·K)

# -----------------------------
# Symbols (for clean math display / rigor)
# -----------------------------
P1_sym, V1_sym, T1_sym, P2_sym, T2_sym, R_sym = sp.symbols('P1 V1 T1 P2 T2 R', positive=True, real=True)
n_sym, V2_sym = sp.symbols('n V2', positive=True, real=True)

# -----------------------------
# Step 1: Solve for n from state 1
# -----------------------------
n_expr = sp.simplify((P1_sym * V1_sym) / (R_sym * T1_sym))

# Numeric evaluation
n_num = float(n_expr.subs({P1_sym: P1, V1_sym: V1, R_sym: R, T1_sym: T1}))

# -----------------------------
# Step 2: Solve for V2 from state 2
# -----------------------------
V2_expr = sp.simplify((n_sym * R_sym * T2_sym) / P2_sym)

# Numeric evaluation (substitute n from step 1)
V2_num = float(V2_expr.subs({n_sym: n_num, R_sym: R, T2_sym: T2, P2_sym: P2}))

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Problem C10.1 — Thermodynamic State Variables & Ideal Gas Law\n")

print("Given (State 1):")
print(f"  P1 = {P1:.3g} Pa")
print(f"  V1 = {V1:.3g} m^3")
print(f"  T1 = {T1:.3g} K\n")

print("Given (State 2):")
print(f"  P2 = {P2:.3g} Pa")
print(f"  T2 = {T2:.3g} K\n")

print(f"Constant:")
print(f"  R  = {R:.4g} J/(mol·K)\n")

print("Step 1 — Solve for n using PV = nRT:")
print("  n = (P1 V1)/(R T1)\n")

print(f"  n = ({P1:.3g})({V1:.3g}) / ({R:.4g})({T1:.3g})")
print(f"    = {n_num:.3g} mol\n")

print("Step 2 — Solve for V2 using PV = nRT:")
print("  V2 = (n R T2)/P2\n")

print(f"  V2 = ({n_num:.3g})({R:.4g})({T2:.3g}) / ({P2:.3g})")
print(f"     = {V2_num:.3g} m^3\n")

print("State variables vs path-dependent quantities:")
print("• State variables here include: P, V, T (and U).")
print("• Path-dependent quantities include: Q (heat) and W (work).\n")

print("Key idea:")
print("• Knowing only the initial and final states does NOT determine Q or W,")
print("  because Q and W depend on the path taken between the states.")
```

## Problem C10.2 — PV Diagram Cycle and Net Work

An ideal gas is taken through the following **four-step cyclic process** on a $P$–$V$ diagram. All states are equilibrium states.

- **State 1:**

  $$
  V_1 = 1.00\ \text{m}^3,\qquad P_1 = 3.00\times 10^5\ \text{Pa}
  $$

- **Step 1 → 2 (constant pressure):** The gas expands at constant pressure to  

  $$
  V_2 = 3.00\ \text{m}^3
  $$

  so

  $$
  P_2 = P_1.
  $$

- **Step 2 → 3 (constant volume):** The gas is cooled at constant volume until the pressure reaches  

  $$
  P_3 = 1.00\times 10^5\ \text{Pa}
  $$

  so

  $$
  V_3 = V_2.
  $$

- **Step 3 → 4 (constant pressure):** The gas is compressed at constant pressure $P_3$ until the volume returns to  

  $$
  V_4 = V_1
  $$

  so

  $$
  P_4 = P_3.
  $$

- **Step 4 → 1 (constant volume):** The gas is returned to State 1 at constant volume $V_1$ (pressure increases back to $P_1$).

---

Answer the following:

1. Make a **PV diagram** of the full cycle.  
   - Label the axes, the four states (1,2,3,4), and show the direction of the cycle.
2. Calculate the work done by the gas during each step: $W_{12}$, $W_{23}$, $W_{34}$, and $W_{41}$.  
3. Find the net work for the full cycle, $W_{\text{net}}$.  
4. Is the cycle **clockwise** or **counterclockwise**? What does this imply about the sign of $W_{\text{net}}$?

Be careful with **units**, **signs**, and **scientific notation**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Key idea**

On a PV diagram, the work done by the gas is the signed area under the path:

- Expansion ($\Delta V > 0$) gives $W>0$
- Compression ($\Delta V < 0$) gives $W<0$
- Constant volume gives $W=0$

---

**1) PV diagram description**

- $1\to2$: horizontal line at $P_1$ from $V_1$ to $V_2$  
- $2\to3$: vertical line at $V_2$ down from $P_1$ to $P_3$  
- $3\to4$: horizontal line at $P_3$ from $V_2$ back to $V_1$  
- $4\to1$: vertical line at $V_1$ up from $P_3$ to $P_1$

This is a closed rectangular loop.

---

**2) Work in each step**

**Step $1\to2$ (constant pressure $P_1$)**

$$
W_{12} = P_1 (V_2 - V_1)
$$

$$
W_{12} = (3.00\times 10^5)(3.00-1.00)
= (3.00\times 10^5)(2.00)
= 6.00\times 10^5\ \text{J}
$$

$$
\boxed{W_{12}=+6.00\times 10^5\ \text{J}}
$$

---

**Step $2\to3$ (constant volume)**

$$
\boxed{W_{23}=0}
$$

---

**Step $3\to4$ (constant pressure $P_3$)**

$$
W_{34} = P_3 (V_4 - V_3)
$$

Here $V_4=V_1$ and $V_3=V_2$, so the gas is being compressed and $(V_4-V_3)<0$:

$$
W_{34} = (1.00\times 10^5)(1.00-3.00)
= (1.00\times 10^5)(-2.00)
= -2.00\times 10^5\ \text{J}
$$

$$
\boxed{W_{34}=-2.00\times 10^5\ \text{J}}
$$

---

**Step $4\to1$ (constant volume)**

$$
\boxed{W_{41}=0}
$$

---

**3) Net work**

$$
W_{\text{net}} = W_{12}+W_{23}+W_{34}+W_{41}
$$

$$
W_{\text{net}} = (6.00\times 10^5) + 0 + (-2.00\times 10^5) + 0
= 4.00\times 10^5\ \text{J}
$$

So:

$$
\boxed{W_{\text{net}}=+4.00\times 10^5\ \text{J}}
$$

---

**4) Direction and sign**

The cycle goes:

- Rightward at high pressure (large positive work)  
- Leftward at lower pressure (smaller negative work)  

This is a **clockwise** loop on the PV diagram, so:

$$
W_{\text{net}} > 0
$$

**Interpretation**

The gas produces net work over one cycle. The net work equals the area of the rectangle:

$$
W_{\text{net}} = (P_1 - P_3)(V_2 - V_1)
$$

which matches the numerical result:

$$
(3.00\times 10^5 - 1.00\times 10^5)(3.00-1.00)
= (2.00\times 10^5)(2.00)
= 4.00\times 10^5\ \text{J}.
$$

</div>
</details>

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Problem C10.2 — PV Diagram Cycle and Net Work 
# ============================================================
# Four-step cyclic process (rectangular PV loop):
#
#   State 1: (V1, P1)
#   1 -> 2: constant pressure P1, expand from V1 to V2
#   2 -> 3: constant volume V2, pressure drops from P1 to P3
#   3 -> 4: constant pressure P3, compress from V2 to V1
#   4 -> 1: constant volume V1, pressure rises from P3 to P1
#
# Work done by the gas:
#   - Constant pressure: W = P * (V_f - V_i)
#   - Constant volume:   W = 0
#
# Net work for a cycle:
#   W_net = (P1 - P3) * (V2 - V1)  (area of rectangle)
# ============================================================

# -----------------------------
# Given values (SI units)
# -----------------------------
V1 = 1.00        # m^3
P1 = 3.00e5      # Pa

V2 = 3.00        # m^3
P3 = 1.00e5      # Pa

# Derived states
P2 = P1
V3 = V2
V4 = V1
P4 = P3

# -----------------------------
# Work calculations
# -----------------------------
W12 = P1 * (V2 - V1)      # expansion at high pressure
W23 = 0.0                 # constant volume
W34 = P3 * (V4 - V3)      # compression at low pressure (negative)
W41 = 0.0                 # constant volume

Wnet = W12 + W23 + W34 + W41

# -----------------------------
# Print results
# -----------------------------
print("Problem C10.2 — PV Diagram Cycle and Net Work\n")

print("States:")
print(f"  State 1: V1 = {V1:.3g} m^3,  P1 = {P1:.3g} Pa")
print(f"  State 2: V2 = {V2:.3g} m^3,  P2 = {P2:.3g} Pa")
print(f"  State 3: V3 = {V3:.3g} m^3,  P3 = {P3:.3g} Pa")
print(f"  State 4: V4 = {V4:.3g} m^3,  P4 = {P4:.3g} Pa\n")

print("Work by segment (work done by the gas):")
print(f"  W12 = P1 (V2 - V1) = {W12:.3g} J")
print(f"  W23 = 0            = {W23:.3g} J")
print(f"  W34 = P3 (V4 - V3) = {W34:.3g} J")
print(f"  W41 = 0            = {W41:.3g} J\n")

print("Net work for the cycle:")
print(f"  W_net = {Wnet:.3g} J\n")

# Check with rectangle area formula
W_area = (P1 - P3) * (V2 - V1)
print("Area check:")
print(f"  (P1 - P3)(V2 - V1) = {W_area:.3g} J\n")

print("Interpretation:")
print("  The loop is clockwise, so W_net > 0 (net work produced by the gas).")

# -----------------------------
# Build PV paths for plotting
# -----------------------------
# 1 -> 2: horizontal at P1
V_12 = np.linspace(V1, V2, 200)
P_12 = np.full_like(V_12, P1)

# 2 -> 3: vertical at V2
P_23 = np.linspace(P2, P3, 200)
V_23 = np.full_like(P_23, V2)

# 3 -> 4: horizontal at P3
V_34 = np.linspace(V3, V4, 200)
P_34 = np.full_like(V_34, P3)

# 4 -> 1: vertical at V1
P_41 = np.linspace(P4, P1, 200)
V_41 = np.full_like(P_41, V1)

# -----------------------------
# Plot PV diagram
# -----------------------------
plt.figure(figsize=(7, 5))

# Plot segments
plt.plot(V_12, P_12, linewidth=2)
plt.plot(V_23, P_23, linewidth=2)
plt.plot(V_34, P_34, linewidth=2)
plt.plot(V_41, P_41, linewidth=2)

# Shade net work area
V_fill = np.linspace(V1, V2, 300)
plt.fill_between(V_fill, P3, P1, alpha=0.3)

# Mark and label states
states = [("State 1", V1, P1),
          ("State 2", V2, P2),
          ("State 3", V3, P3),
          ("State 4", V4, P4)]

for label, vx, px in states:
    plt.plot(vx, px, marker='o')
    plt.text(vx, px, "  " + label)

# Direction markers embedded in each segment
plt.plot((V1+V2)/2, P1, marker='>', markersize=9)             # 1->2
plt.plot(V2, (P2+P3)/2, marker='v', markersize=9)             # 2->3
plt.plot((V3+V4)/2, P3, marker='<', markersize=9)             # 3->4
plt.plot(V1, (P4+P1)/2, marker='^', markersize=9)             # 4->1

plt.xlabel("Volume (V) [m$^3$]")
plt.ylabel("Pressure (P) [Pa]")
plt.title("PV Cycle (Net Work Shaded)")

plt.xlim(0.5, 3.5)
plt.ylim(0.5e5, 3.5e5)

plt.show()
```

## Problem C10.3 — Isothermal Expansion of an Ideal Gas

An ideal gas undergoes a slow **isothermal expansion** at constant temperature

$$
T = 300\ \text{K}.
$$

The amount of gas is

$$
n = 0.500\ \text{mol}.
$$

The gas expands from an initial volume

$$
V_1 = 2.00\times 10^{-3}\ \text{m}^3
$$

to a final volume

$$
V_2 = 6.00\times 10^{-3}\ \text{m}^3.
$$

Assume ideal-gas behavior.

---

### Tasks

1. Calculate the work done by the gas, $W$.  
2. Calculate the heat transferred to the gas, $Q$.  
3. Sketch a $P$–$V$ diagram of the process and indicate the area representing the work.

Use:

$$
P = \frac{nRT}{V}
$$

and (for an isothermal ideal-gas process):

$$
W = nRT \ln\!\left(\frac{V_2}{V_1}\right).
$$

Be careful with **units**, **significant figures**, and **sign conventions**.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

**Step 1 — Endpoint Pressures**

Use the ideal gas law in the form:

$$
P = \frac{nRT}{V}
$$

At State 1:

$$
P_1 = \frac{(0.500)(8.314)(300)}{2.00\times 10^{-3}}
$$

First compute the numerator:

$$
nRT = (0.500)(8.314)(300) = 1247\ \text{J}
$$

Then:

$$
P_1 = \frac{1247}{2.00\times 10^{-3}}
= 6.24\times 10^5\ \text{Pa}
$$

At State 2:

$$
P_2 = \frac{1247}{6.00\times 10^{-3}}
= 2.08\times 10^5\ \text{Pa}
$$

---

**Step 2 — Work Done by the Gas**

For an isothermal ideal-gas process:

$$
W = nRT \ln\!\left(\frac{V_2}{V_1}\right)
$$

Substitute:

$$
W = (1247)\ln\!\left(\frac{6.00\times 10^{-3}}{2.00\times 10^{-3}}\right)
$$

$$
W = (1247)\ln(3)
$$

Since:

$$
\ln(3) = 1.099
$$

$$
W = (1247)(1.099)
= 1.37\times 10^3\ \text{J}
$$

So:

$$
\boxed{W = +1.37\times 10^3\ \text{J}}
$$

The positive sign indicates work done **by** the gas (expansion).

---

**Step 3 — Heat Transferred**

For an ideal gas, internal energy depends only on temperature.

Since the process is isothermal:

$$
\Delta T = 0
$$

Therefore:

$$
\Delta U = 0
$$

From the First Law:

$$
\Delta U = Q - W
$$

$$
0 = Q - W
$$

So:

$$
Q = W
$$

$$
\boxed{Q = +1.37\times 10^3\ \text{J}}
$$

Heat must flow **into** the gas to compensate for the work done during expansion.

---

**Step 4 — PV Diagram Interpretation**

- The curve is a decreasing hyperbola: $P = \dfrac{nRT}{V}$.
- The shaded area under the curve between $V_1$ and $V_2$ represents the work done by the gas.
- The path moves rightward (increasing volume), so it represents expansion.
- The curve is not a straight line; pressure decreases as volume increases.

---

**Physical Interpretation**

- The gas expands and does work on the surroundings.
- Its temperature does not change.
- Therefore, heat must flow into the gas to replace the energy lost as work.
- The First Law is satisfied:
  
  $$
  \Delta U = 0, \qquad Q = W.
  $$

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.markers import MarkerStyle

# ============================================================
# Problem C10.3 — Isothermal Expansion 
# ============================================================
# Isothermal ideal gas:
#   P(V) = nRT / V
#
# Work done by the gas (given/used in Phase C):
#   W = nRT ln(V2/V1)
#
# For an ideal gas, U depends only on T.
# Isothermal => ΔT = 0 => ΔU = 0
# First Law: ΔU = Q - W  =>  Q = W
# ============================================================

# -----------------------------
# Given
# -----------------------------
n = 0.500                 # mol
T = 300.0                 # K
V1 = 2.00e-3              # m^3
V2 = 6.00e-3              # m^3
R = 8.314                 # J/(mol·K)

# -----------------------------
# Calculations
# -----------------------------
W = n * R * T * np.log(V2 / V1)   # J
Q = W                             # J (since ΔU = 0 for isothermal ideal gas)

# Pressures at endpoints
P1 = n * R * T / V1
P2 = n * R * T / V2

print("Problem C10.3 — Isothermal Expansion\n")
print(f"Given:")
print(f"  n  = {n:.3g} mol")
print(f"  T  = {T:.3g} K")
print(f"  V1 = {V1:.3g} m^3")
print(f"  V2 = {V2:.3g} m^3\n")

print("Endpoint pressures (from P = nRT/V):")
print(f"  P1 = {P1:.3g} Pa")
print(f"  P2 = {P2:.3g} Pa\n")

print("Work done by the gas (isothermal ideal gas):")
print(f"  W = nRT ln(V2/V1) = {W:.3g} J\n")

print("Heat transferred to the gas:")
print("  Isothermal ideal gas => ΔU = 0, so Q = W")
print(f"  Q = {Q:.3g} J\n")

# -----------------------------
# PV Plot
# -----------------------------
V = np.linspace(V1, V2, 500)
P = n * R * T / V

plt.figure(figsize=(7, 5))
plt.plot(V, P, linewidth=2)

# Shade area under curve (work for expansion)
plt.fill_between(V, 0, P, alpha=0.25)

# Mark states
plt.plot(V1, P1, marker='o')
plt.plot(V2, P2, marker='o')
plt.text(V1, P1, "  State 1")
plt.text(V2, P2, "  State 2")

# Direction marker aligned with curve (tangent direction)
mid = len(V) // 2
x0, y0 = V[mid], P[mid]

# Numerical slope dP/dV near the midpoint
dP_dV = np.gradient(P, V)[mid]
angle_deg = np.degrees(np.arctan(dP_dV))

marker = MarkerStyle(">")
marker._transform = marker.get_transform().rotate_deg(angle_deg)
plt.plot(x0, y0, marker=marker, markersize=10)

plt.xlabel("Volume (V) [m$^3$]")
plt.ylabel("Pressure (P) [Pa]")
plt.title("Isothermal Expansion on a PV Diagram (Work Shaded)")

plt.xlim(0.0, 1.15 * V2)
plt.ylim(0.0, 1.15 * P1)

plt.show()
```

## Problem C10.4 — Heat Capacities at Constant Volume and Constant Pressure

An ideal monatomic gas is initially at temperature

$$
T_1 = 300\ \text{K}.
$$

The amount of gas is

$$
n = 1.20\ \text{mol}.
$$

The gas is heated to a final temperature

$$
T_2 = 500\ \text{K}.
$$

Answer the following:

1. If the heating occurs at **constant volume**, calculate:
   - The heat added, $Q_V$  
   - The change in internal energy, $\Delta U$  
   - The work done, $W_V$

2. If the heating occurs at **constant pressure**, calculate:
   - The heat added, $Q_P$  
   - The change in internal energy, $\Delta U$  
   - The work done, $W_P$

Use for a monatomic ideal gas:

$$
C_V = \frac{3}{2}R,
\qquad
C_P = \frac{5}{2}R
$$

Be careful with **units**, **sign conventions**, and **interpretation**.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

**Step 1 — Temperature Change**

$$
\Delta T = T_2 - T_1 = 500 - 300 = 200\ \text{K}
$$

---

**Part A — Constant Volume**

At constant volume:

$$
W_V = 0
$$

Heat added:

$$
Q_V = n C_V \Delta T
$$

Since

$$
C_V = \frac{3}{2}R
$$

$$
Q_V = n \left(\frac{3}{2}R\right)\Delta T
$$

Substitute values:

$$
Q_V = (1.20)\left(\frac{3}{2}\right)(8.314)(200)
$$

$$
Q_V = (1.20)(1.5)(8.314)(200)
$$

$$
Q_V = 2.99\times 10^3\ \text{J}
$$

So:

$$
\boxed{Q_V = 2.99\times 10^3\ \text{J}}
$$

Since $W_V=0$, from the First Law:

$$
\Delta U = Q_V
$$

$$
\boxed{\Delta U = 2.99\times 10^3\ \text{J}}
$$

---

**Part B — Constant Pressure**

At constant pressure:

$$
Q_P = n C_P \Delta T
$$

Since

$$
C_P = \frac{5}{2}R
$$

$$
Q_P = n \left(\frac{5}{2}R\right)\Delta T
$$

Substitute:

$$
Q_P = (1.20)(2.5)(8.314)(200)
$$

$$
Q_P = 4.99\times 10^3\ \text{J}
$$

So:

$$
\boxed{Q_P = 4.99\times 10^3\ \text{J}}
$$

---

**Internal Energy Change**

Internal energy depends only on temperature for an ideal gas:

$$
\Delta U = n C_V \Delta T
$$

So:

$$
\boxed{\Delta U = 2.99\times 10^3\ \text{J}}
$$

This is the same as in the constant-volume case.

---

**Work at Constant Pressure**

From the First Law:

$$
\Delta U = Q_P - W_P
$$

Solve for work:

$$
W_P = Q_P - \Delta U
$$

$$
W_P = (4.99\times 10^3) - (2.99\times 10^3)
$$

$$
W_P = 2.00\times 10^3\ \text{J}
$$

So:

$$
\boxed{W_P = 2.00\times 10^3\ \text{J}}
$$

---

**Physical Interpretation**

- At constant volume, all heat increases internal energy.
- At constant pressure, some heat increases internal energy and some becomes work.
- Therefore:

$$
C_P > C_V
$$

because extra heat is required to allow expansion.

</details>

```python
%reset -f

import numpy as np

# ============================================================
# Problem C10.4 — Heat Capacities 
# ============================================================
# Monatomic ideal gas:
#   C_V = (3/2) R
#   C_P = (5/2) R
#
# Internal energy change:
#   ΔU = n C_V ΔT
#
# Constant volume:
#   W_V = 0
#   Q_V = ΔU
#
# Constant pressure:
#   Q_P = n C_P ΔT
#   W_P = Q_P - ΔU
# ============================================================

# -----------------------------
# Given values
# -----------------------------
n = 1.20          # mol
T1 = 300.0        # K
T2 = 500.0        # K
R = 8.314         # J/(mol·K)

# Temperature change
dT = T2 - T1

# Heat capacities
C_V = (3/2) * R
C_P = (5/2) * R

# -----------------------------
# Constant Volume
# -----------------------------
dU = n * C_V * dT
W_V = 0.0
Q_V = dU

# -----------------------------
# Constant Pressure
# -----------------------------
Q_P = n * C_P * dT
W_P = Q_P - dU

# -----------------------------
# Output
# -----------------------------
print("Problem C10.4 — Heat Capacities\n")

print("Given:")
print(f"  n  = {n:.3g} mol")
print(f"  T1 = {T1:.3g} K")
print(f"  T2 = {T2:.3g} K")
print(f"  ΔT = {dT:.3g} K\n")

print("Heat capacities (monatomic ideal gas):")
print(f"  C_V = (3/2)R = {C_V:.3g} J/(mol·K)")
print(f"  C_P = (5/2)R = {C_P:.3g} J/(mol·K)\n")

print("Constant Volume Process:")
print(f"  ΔU  = n C_V ΔT = {dU:.3g} J")
print(f"  W_V = {W_V:.3g} J")
print(f"  Q_V = {Q_V:.3g} J\n")

print("Constant Pressure Process:")
print(f"  ΔU  = {dU:.3g} J  (same as constant volume)")
print(f"  Q_P = n C_P ΔT = {Q_P:.3g} J")
print(f"  W_P = Q_P - ΔU = {W_P:.3g} J\n")

print("Interpretation:")
print("  At constant volume, all heat increases internal energy.")
print("  At constant pressure, additional heat is required to do expansion work.")
print("  Therefore C_P > C_V.")
```

## Problem C10.5 — Heat Engine Efficiency and Energy Flow

A heat engine operates in a cycle. During each cycle, it absorbs heat from a hot reservoir:

$$
Q_H = 1.50\times 10^3\ \text{J},
$$

and it rejects heat to a cold reservoir:

$$
Q_C = 9.00\times 10^2\ \text{J}.
$$

Answer the following:

1. Calculate the net work produced per cycle, $W_{\text{out}}$.  
2. Calculate the thermal efficiency, $e$.  
3. If the engine completes $120$ cycles each minute, compute the average power output in watts.  
4. State whether this engine is acting as a heat engine or a refrigerator, and explain briefly.

Be careful with **units**, **sign conventions**, and **interpretation**.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

**Step 1 — Work per cycle**

Energy conservation for a cyclic engine gives:

$$
Q_H = W_{\text{out}} + Q_C
$$

So:

$$
W_{\text{out}} = Q_H - Q_C
$$

Substitute:

$$
W_{\text{out}} = (1.50\times 10^3) - (9.00\times 10^2)
= 6.00\times 10^2\ \text{J}
$$

$$
\boxed{W_{\text{out}} = 6.00\times 10^2\ \text{J}}
$$

---

**Step 2 — Thermal efficiency**

By definition:

$$
e = \frac{W_{\text{out}}}{Q_H}
$$

So:

$$
e = \frac{6.00\times 10^2}{1.50\times 10^3}
= 0.400
$$

$$
\boxed{e = 0.400 \ \text{(40.0\%)}}
$$

---

**Step 3 — Average power output**

The engine completes:

$$
120\ \text{cycles/min} = \frac{120}{60} = 2.00\ \text{cycles/s}
$$

Power is work per time:

$$
P = W_{\text{out}} \times (\text{cycles per second})
$$

So:

$$
P = (6.00\times 10^2)(2.00)
= 1.20\times 10^3\ \text{W}
$$

$$
\boxed{P = 1.20\times 10^3\ \text{W}}
$$

---

**Step 4 — Engine or refrigerator?**

Because:

- $Q_H$ flows *into* the engine from the hot reservoir
- The engine produces positive net work ($W_{\text{out}} > 0$)
- Heat is rejected to the cold reservoir ($Q_C$)

this device is operating as a **heat engine**.

A refrigerator would require work input and would move heat from cold to hot.

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

# ============================================================
# Problem C10.5 — Heat Engine 
# ============================================================
# Given:
#   Q_H = heat absorbed from hot reservoir
#   Q_C = heat rejected to cold reservoir
#
# For a cyclic heat engine:
#   W_out = Q_H - Q_C
#
# Thermal efficiency:
#   e = W_out / Q_H
#
# Power:
#   P = (work per cycle) × (cycles per second)
# ============================================================

# -----------------------------
# Given values
# -----------------------------
Q_H = 1.50e3        # J
Q_C = 9.00e2        # J
cycles_per_min = 120.0

# -----------------------------
# Step 1 — Work per cycle
# -----------------------------
W_out = Q_H - Q_C

# -----------------------------
# Step 2 — Efficiency
# -----------------------------
efficiency = W_out / Q_H

# -----------------------------
# Step 3 — Power output
# -----------------------------
cycles_per_sec = cycles_per_min / 60.0
power = W_out * cycles_per_sec

# -----------------------------
# Output
# -----------------------------
print("Problem C10.5 — Heat Engine\n")

print("Given:")
print(f"  Q_H = {Q_H:.3g} J")
print(f"  Q_C = {Q_C:.3g} J")
print(f"  Cycles per minute = {cycles_per_min:.3g}\n")

print("Work per cycle:")
print(f"  W_out = Q_H - Q_C = {W_out:.3g} J\n")

print("Thermal efficiency:")
print(f"  e = W_out / Q_H = {efficiency:.3g}")
print(f"  e = {efficiency*100:.3g}%\n")

print("Power output:")
print(f"  Cycles per second = {cycles_per_sec:.3g} s^-1")
print(f"  P = W_out × (cycles/s) = {power:.3g} W\n")

print("Interpretation:")
print("  Since W_out > 0 and heat flows from hot to cold,")
print("  the device operates as a heat engine.")
```

## Problem C10.6 — Carnot Engine and Maximum Efficiency

A Carnot engine operates between two thermal reservoirs:

$$
T_H = 600\ \text{K}
$$

$$
T_C = 300\ \text{K}
$$

During each cycle, the engine absorbs

$$
Q_H = 2.40\times 10^3\ \text{J}
$$

from the hot reservoir.

Answer the following:

1. What is the maximum possible efficiency of this engine?
2. How much work does the engine produce per cycle?
3. How much heat is rejected to the cold reservoir?
4. If a real engine operating between the same reservoirs has efficiency 35%, is it possible for it to outperform the Carnot engine? Explain.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

**Step 1 — Carnot Efficiency**

For a Carnot engine:

$$
e_{\text{Carnot}} = 1 - \frac{T_C}{T_H}
$$

Substitute:

$$
e_{\text{Carnot}} = 1 - \frac{300}{600}
$$

$$
e_{\text{Carnot}} = 1 - 0.500
$$

$$
\boxed{e_{\text{Carnot}} = 0.500 \text{ (50.0\%)}}
$$

---

**Step 2 — Work per Cycle**

Efficiency is defined as:

$$
e = \frac{W_{\text{out}}}{Q_H}
$$

So:

$$
W_{\text{out}} = e Q_H
$$

$$
W_{\text{out}} = (0.500)(2.40\times 10^3)
$$

$$
\boxed{W_{\text{out}} = 1.20\times 10^3\ \text{J}}
$$

---

**Step 3 — Heat Rejected**

From energy conservation:

$$
Q_H = W_{\text{out}} + Q_C
$$

So:

$$
Q_C = Q_H - W_{\text{out}}
$$

$$
Q_C = (2.40\times 10^3) - (1.20\times 10^3)
$$

$$
\boxed{Q_C = 1.20\times 10^3\ \text{J}}
$$

---

**Step 4 — Can a Real Engine Do Better?**

The Carnot engine represents the maximum possible efficiency for engines operating between these two temperatures.

A real engine with efficiency:

$$
e = 0.35 \text{ (35\%)}
$$

is below the Carnot limit of 50%.

No real engine can exceed:

$$
e_{\text{Carnot}} = 50\%.
$$

Therefore:

> It is impossible for a real engine operating between 600 K and 300 K to outperform the Carnot engine.

The Carnot efficiency is a fundamental upper limit set by the Second Law of Thermodynamics.

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

# ============================================================
# Problem C10.6 — Carnot Engine
# ============================================================
# Carnot efficiency:
#   e_c = 1 - Tc/Th
#
# Work per cycle:
#   W_out = e_c * Q_H
#
# Heat rejected:
#   Q_C = Q_H - W_out
# ============================================================

# -----------------------------
# Given values
# -----------------------------
T_H = 600.0          # K
T_C = 300.0          # K
Q_H = 2.40e3         # J

# Real engine comparison
e_real = 0.35        # 35%

# -----------------------------
# Calculations
# -----------------------------
e_carnot = 1.0 - (T_C / T_H)
W_out = e_carnot * Q_H
Q_C = Q_H - W_out

# -----------------------------
# Output
# -----------------------------
print("Problem C10.6 — Carnot Engine\n")

print("Given:")
print(f"  T_H = {T_H:.3g} K")
print(f"  T_C = {T_C:.3g} K")
print(f"  Q_H = {Q_H:.3g} J\n")

print("Step 1 — Carnot efficiency:")
print(f"  e_Carnot = 1 - T_C/T_H = 1 - ({T_C:.3g}/{T_H:.3g}) = {e_carnot:.3g}")
print(f"  e_Carnot = {e_carnot*100:.3g}%\n")

print("Step 2 — Work per cycle:")
print(f"  W_out = e_Carnot * Q_H = ({e_carnot:.3g})({Q_H:.3g}) = {W_out:.3g} J\n")

print("Step 3 — Heat rejected:")
print(f"  Q_C = Q_H - W_out = {Q_H:.3g} - {W_out:.3g} = {Q_C:.3g} J\n")

print("Step 4 — Real engine comparison:")
print(f"  e_real = {e_real:.3g} ({e_real*100:.3g}%)")
print(f"  e_Carnot = {e_carnot:.3g} ({e_carnot*100:.3g}%)")

if e_real > e_carnot:
    print("\nConclusion:")
    print("  This would exceed the Carnot limit, so it is NOT physically possible.")
else:
    print("\nConclusion:")
    print("  This is below the Carnot limit, so it is physically possible (though not guaranteed).")
```
