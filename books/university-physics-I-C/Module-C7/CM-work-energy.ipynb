# C7.3 Pseudo Work-Energy Theorem - Center of Mass
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.1 Review: Work–Energy Theorem for a Single Particle

For a single particle of mass $m$, Newton’s second law states that

$$
\vec{f}_{\text{net}} = m \vec{a}.
$$

Since acceleration is the change in velocity over time,

$$
\vec{a} = \frac{\Delta \vec{v}}{\Delta t}.
$$

Substituting this into Newton’s second law gives

$$
\vec{f}_{\text{net}} = m \frac{\Delta \vec{v}}{\Delta t}.
$$

---

### Multiply Both Sides by the Displacement

During the same time interval $\Delta t$, the displacement of the particle is approximately

$$
\Delta \vec{r} = \vec{v}_{\text{avg}} \, \Delta t,
$$

where $\vec{v}_{\text{avg}} = \tfrac{1}{2} (\vec{v}_i + \vec{v}_f)$ is the average velocity over that interval.

Taking the dot product of both sides of Newton’s law with $\Delta \vec{r}$:

$$
\vec{f}_{\text{net}} \cdot \Delta \vec{r}
= m \frac{\Delta \vec{v}}{\Delta t} \cdot \Delta \vec{r}.
$$

Substitute $\Delta \vec{r} = \vec{v}_{\text{avg}} \, \Delta t$ on the right side:

$$
\vec{f}_{\text{net}} \cdot \Delta \vec{r}
= m \frac{\Delta \vec{v}}{\Delta t} \cdot \vec{v}_{\text{avg}} \, \Delta t
= m \, \vec{v}_{\text{avg}} \cdot \Delta \vec{v}.
$$

---

### Relate to Kinetic Energy

The average velocity can be written as $\vec{v}_{\text{avg}} = \tfrac{1}{2} (\vec{v}_f + \vec{v}_i)$, so

$$
m \, \vec{v}_{\text{avg}} \cdot \Delta \vec{v}
= \frac{m}{2} (\vec{v}_f + \vec{v}_i) \cdot (\vec{v}_f - \vec{v}_i).
$$

Simplify the dot product:

$$
(\vec{v}_f + \vec{v}_i) \cdot (\vec{v}_f - \vec{v}_i)
= v_f^2 - v_i^2.
$$

Therefore,

$$
\vec{f}_{\text{net}} \cdot \Delta \vec{r}
= \frac{1}{2} m (v_f^2 - v_i^2).
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.2 Interpretation

The left-hand side, $\vec{f}_{\text{net}} \cdot \Delta \vec{r}$, is the **work** done by the net force over the displacement.  
The right-hand side is the **change in kinetic energy**:

$$
\boxed{w_{\text{net}} = \Delta k = \frac{1}{2} m v_f^2 - \frac{1}{2} m v_i^2.}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.3 Extending to a System of Particles

Consider now a system of $N$ particles with total mass $M = \sum_i m_i$.  
Each particle obeys

$$
\vec{f}_i^{\text{(ext)}} + \vec{f}_i^{\text{(int)}} = m_i \vec{a}_i,
$$

where $\vec{f}_i^{\text{(ext)}}$ are external forces and $\vec{f}_i^{\text{(int)}}$ are internal forces from other particles in the system.

Summing over all particles, the internal forces cancel in pairs (by Newton’s third law), giving

$$
\sum_i \vec{f}_i^{\text{(ext)}} = \sum_i m_i \vec{a}_i = M \vec{A}_{\text{cm}}.
$$

This shows that the **center of mass moves as if all external forces acted on a single particle of mass $M$**:

$$
\boxed{\vec{F}_{\text{ext}} = M \vec{A}_{\text{cm}}.}
$$

---

### Kinetic Energy Decomposition

The total kinetic energy of the system can be written as

$$
K_{\text{tot}} = \sum_i \frac{1}{2} m_i v_i^2.
$$

Each particle’s velocity can be expressed as the sum of the CM velocity and the velocity relative to the CM:

$$
\vec{v}_i = \vec{V}_{\text{cm}} + \vec{v}_i',
$$

where $\vec{v}_i'$ is the velocity in the CM frame and satisfies $\sum_i m_i \vec{v}_i' = 0$.

Substituting gives

$$
K_{\text{tot}} 
= \sum_i \frac{1}{2} m_i (\vec{V}_{\text{cm}} + \vec{v}_i')^2
= \frac{1}{2} M V_{\text{cm}}^2 + \sum_i \frac{1}{2} m_i v_i'^2.
$$

Thus,

$$
\boxed{K_{\text{tot}} = K_{\text{cm}} + K_{\text{int}}.}
$$

The total kinetic energy is the sum of  
- $K_{\text{cm}} = \tfrac{1}{2} M V_{\text{cm}}^2$, the kinetic energy associated with CM motion, and  
- $K_{\text{int}} = \sum_i \tfrac{1}{2} m_i v_i'^2$, the kinetic energy of internal motion relative to the CM.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.4 The Pseudo Work–Energy Theorem

Similarly to the point particle case, we can take the dot product with $\Delta \vec{R}$ to obtain the **pseudo work-energy theorem**:

$$
\boxed{W_{\text{ext}}^{(\text{cm})} = \Delta K_{\text{cm}}.}
$$

This result *resembles* the work–energy theorem, but it involves only the **work done by the resultant external force acting at the CM** and the **kinetic energy of the CM**.  
It does *not* account for internal motions or potential energies within the system.

Hence, this relation is called the **pseudo work–energy theorem**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.5 Physical Interpretation

- $W_{\text{ext}}^{(\text{cm})}$ does **not** represent the total mechanical work done on the entire system, since individual external forces can also cause internal deformation, rotation, or changes in internal structure.

- The *true* work–energy theorem for the entire system can be written schematically as

  $$
  W_{\text{ext, real}} = \Delta K_{\text{tot}} + \Delta U_{\text{int}},
  $$

  where:
  - $W_{\text{ext, real}}$ is the total work done by all **external forces** (both conservative and non-conservative),
  - $\Delta K_{\text{tot}}$ is the total change in kinetic energy of all particles, and
  - $\Delta U_{\text{int}}$ is the change in **internal potential energy** due to **internal conservative forces** (such as stretching or compressing springs, or molecular interactions within the system).

- In many macroscopic problems, $\Delta U_{\text{int}}$ is either negligible (for rigid systems) or modeled separately as an internal energy change (e.g., heating, deformation, or rotation).

- The pseudo theorem, on the other hand,

  $$
  W_{\text{ext}}^{(\text{cm})} = \Delta K_{\text{cm}},
  $$

isolates only the **translational motion** of the system’s center of mass. It describes how the *net external force* affects the CM’s kinetic energy, but it **does not** account for internal motions, rotations, or deformations.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.6 Summary Comparison

| Quantity | Describes | Includes | Excludes |
|-----------|------------|-----------|-----------|
| $W_{\text{ext}}^{(\text{cm})} = \Delta K_{\text{cm}}$ | **Pseudo work–energy theorem** (motion of CM) | Resultant external force acting at CM | Internal and rotational energies |
| $W_{\text{ext, real}} = \Delta K_{\text{tot}} + \Delta U_{\text{int}}$ | **Full energy theorem** (entire system) | All external forces (both conservative and non-conservative) | None — internal work is included in $\Delta U_{\text{int}}$ |

Hence, the *pseudo* work–energy theorem is a **translational approximation**, useful for predicting the motion of the center of mass, while the *full* work–energy theorem describes the **complete energy accounting** of the system, including internal transformations.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.7 Example Illustration

Imagine throwing a ball that deforms slightly in flight.  
The net external force (gravity and air resistance) determines the motion of the center of mass through $ \vec{F}_{\text{ext}} = M \vec{a}_{\text{cm}} $.  
However, internal vibrations and deformations change the total energy of the ball, even though the CM motion alone still satisfies

$$
W_{\text{ext}}^{(\text{cm})} = \Delta K_{\text{cm}}.
$$

Thus, the pseudo work–energy theorem provides a **useful but limited** description of how external forces affect the *translation* of the system as a whole, while omitting internal and rotational effects.

## C7.3.8 Comparison with the Impulse-Momentum Theorem

| Theorem | Includes Internal Effects? | Exact for System? | Comments |
|----------|-----------------------------|-------------------|-----------|
| **Impulse–Momentum:** $\vec{J}_{\text{ext}} = \Delta (M \vec{V}_{\text{cm}})$ | No — internal forces cancel | ✅ Yes | Describes CM motion exactly |
| **Work–Energy (pseudo):** $W_{\text{ext}}^{(\text{cm})} = \Delta K_{\text{cm}}$ | No — internal work ignored | ⚠️ Approximate | Describes CM translation only |
