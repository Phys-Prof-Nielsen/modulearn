## C1.4.1 Approach

We can scale a vector $\vec{A}$ by multiplying it by a scalar $c$ (where do you think the name "scalar" comes from?). The scaled vector is a new vector $\vec{B} = c\vec{A}$. If $c$ is positive, then $\vec{A}$ and $\vec{B}$ are in the same directions. If $c$ is negative then the two vectors are in opposite directions.

<h3 style="margin-top:0; color:#000000;">Example — Multiplying a vector by a scalar</h3>
    
In this example (see illustration below), we are multiplying the vector by $c = 1.5$ and $c = -1$. The scalar multiplication scales the length of the vector, and the sign determines if it remains in the same direction or opposite direction.
</div>

```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Original vector A
# -----------------------------
A = np.array([3, 2])

# Scalar multiples
A_pos = 1.5 * A
A_neg = -1.0 * A

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(5,5))

# Original vector
plt.quiver(1, 2, A[0], A[1],
           angles='xy', scale_units='xy', scale=1)
plt.text(2.5, 3.2, r"$\vec{A}$")

# Scaled by +1.5
plt.quiver(1,3, A_pos[0], A_pos[1],
           angles='xy', scale_units='xy', scale=1)
plt.text(2.7, 4.8, r"$1.5\,\vec{A}$")

# Scaled by −1
plt.quiver(4,2.5, A_neg[0], A_neg[1],
           angles='xy', scale_units='xy', scale=1)
plt.text(2.5, 1.9, r"$-\,\vec{A}$")

# Axes and formatting
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')

plt.xlim(0, 8)
plt.ylim(0, 8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Geometric Scalar Multiplication of a Vector")

plt.show()
```
