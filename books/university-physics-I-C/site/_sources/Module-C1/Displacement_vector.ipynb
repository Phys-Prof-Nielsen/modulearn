## C1.5.1 Displacement Vector in 1D

We will introduce the displacement vector in 1D first and use it to illustrate geometrical addition of vectors in the next section. After that, we will consider the displacement in 2D and 3D.

Consider an object located at x$_A$ = 3.0 m with position vector $\vec{x}_A$.
Now imagine that the object moves to a new position x$_B$ = 4.5 m with a new position vector $\vec{x}_B$ by moving a distance $\Delta$x = 1.5 m in the positive direction (NOTE: distance changed + direction = vector : $\Delta\vec{x}$). This can be illustrated as

```python
import matplotlib.pyplot as plt

# -----------------------------
# Positions
# -----------------------------
x_A = 3.0
x_B = 6.0   # choose so Δx is visible and clean

# Vertical offsets for clarity
y_axis = 0.0
y_A = 0.8
y_B = 1.4

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(9,3))

# x-axis
plt.axhline(y_axis, color='black', linewidth=1.5)

# Axis arrows and labels
plt.annotate("", xy=(7.5, y_axis), xytext=(-1.6, y_axis),
             arrowprops=dict(arrowstyle="<->", linewidth=1.5))
plt.text(7.6, y_axis, r"$+x$", va='center')
plt.text(-1.85, y_axis, r"$-x$", va='center')

# Mark points
plt.plot([0, x_A,x_B], [y_axis, y_axis,y_axis], 'ko')
plt.text(0, y_axis - 0.35, r"$0$", ha='center')
plt.text(x_A, y_axis - 0.35, r"$3$", ha='center')
plt.text(x_B, y_axis - 0.35, r"$4.5$", ha='center')

# -----------------------------
# Position vectors
# -----------------------------
# r_A
plt.quiver(0, y_A, x_A, 0,
           angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=3)
plt.text(x_A/2, y_A + 0.15, r"$\vec{x}_A$", ha='center', color='blue')

# r_B
plt.quiver(0, y_B, x_B, 0,
           angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=3)
plt.text(x_B*0.5, y_B + 0.15, r"$\vec{x}_B$", ha='center', color='blue')

# -----------------------------
# Displacement vector Δx
# -----------------------------
plt.quiver(x_A, y_A, x_B - x_A, 0,
           angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=3)
plt.text(x_A + (x_B - x_A)/2, y_A + 0.15,
         r"$\Delta\vec{x}$", ha='center', color='blue')

# -----------------------------
# Formatting
# -----------------------------
plt.xlim(-1.5, 7.5)
plt.ylim(-0.8, 2.2)
plt.axis('off')

plt.title("1D Position Vectors and Displacement", pad=10)

plt.show()
```


>Note: the final position vector can be visualized by adding the two arrows representing $\vec{x}_A$ and $\Delta\vec{x}$ together. In Layman's terms, we will say that a starting position plus a change results in a new position. In the mathematical language,
we would say:

$$ \vec{x}_B = \vec{x}_A + \Delta\vec{x}$$

The change in position vector, $\Delta\vec{x}$, is known as the <span style="color:red">displacement vector</span>:

$$ \Delta\vec{x} = \vec{x}_B - \vec{x}_A$$

which in words is stated as: <span style="color:red">the directed line segment from starting position to final position</span>. We can state that the displacement vector is describing a change (magnitude and/or direction) in the position of an object.

<h3 style="margin-top:0; color:#000000;">Example — Displacement Vector</h3>

Consider the object located at x$_A$ = 3.0 with position vector $\vec{x}_A$. Now imagine that object moves to a new
position x$_B$ = 1.5 with a new position vector $\vec{x}_B$ by moving a distance $\Delta $x = 1.5 in the negative direction. We can illustrate this change in position through the following vector diagram
</div>

```python
import matplotlib.pyplot as plt

# -----------------------------
# Positions
# -----------------------------
x_A = 3.0
x_B = 1.5   # choose so Δx is visible and clean

# Vertical offsets for clarity
y_axis = 0.0
y_A = 0.8
y_B = 1.4

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(9,3))

# x-axis
plt.axhline(y_axis, color='black', linewidth=1.5)

# Axis arrows and labels
plt.annotate("", xy=(7.5, y_axis), xytext=(-1.6, y_axis),
             arrowprops=dict(arrowstyle="<->", linewidth=1.5))
plt.text(7.6, y_axis, r"$+x$", va='center')
plt.text(-1.85, y_axis, r"$-x$", va='center')

# Mark points
plt.plot([0, x_A,x_B], [y_axis, y_axis,y_axis], 'ko')
plt.text(0, y_axis - 0.35, r"$0$", ha='center')
plt.text(x_A, y_axis - 0.35, r"$3$", ha='center')
plt.text(x_B, y_axis - 0.35, r"$1.5$", ha='center')

# -----------------------------
# Position vectors
# -----------------------------
# r_A
plt.quiver(0, y_A, x_A, 0,
           angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=3)
plt.text(x_A/2, y_A + 0.15, r"$\vec{x}_A$", ha='center', color='blue')

# r_B
plt.quiver(0, y_B, x_B, 0,
           angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=3)
plt.text(x_B*0.5, y_B + 0.15, r"$\vec{x}_B$", ha='center', color='blue')

# -----------------------------
# Displacement vector Δx
# -----------------------------
plt.quiver(x_A, y_B, x_B - x_A, 0,
           angles='xy', scale_units='xy', scale=1,
           color='blue', linewidth=3)
plt.text(x_A + (x_B - x_A)/2, y_B + 0.15,
         r"$\Delta\vec{x}$", ha='center', color='blue')

# -----------------------------
# Formatting
# -----------------------------
plt.xlim(-1.5, 7.5)
plt.ylim(-0.8, 2.2)
plt.axis('off')

plt.title("1D Position Vectors and Displacement", pad=10)

plt.show()
```

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Displacement Vector — Head-To-Tail Approach</h3>

In this example, we may conclude the following: if we follow the initial position vector $\vec{x}_A$ from tail to head, then continue along the displacement $\Delta\vec{x}$, we end up at the tip of the final position vector $\vec{x}_B$. This is a general rule for geometric addition of vectors, which we will cover in the next section.
