## **Problem 2: Representing Vectors**  
A vector $\vec{A}$ has a magnitude of $10.0$ units and is directed at an angle of $37^\circ$ above the positive $x$-axis.  
1. Find the $x$- and $y$-components of $\vec{A}$.  
2. Represent $\vec{A}$ in component form as $\vec{A} = A_x\hat{i} + A_y\hat{j}$.

## **Problem 3: Vector Addition and Subtraction**  
Two vectors are given as:  
- $\vec{B} = (4.0 \, \hat{i} - 3.0 \, \hat{j}) \, \text{units}$,  
- $\vec{C} = (-2.0 \, \hat{i} + 5.0 \, \hat{j}) \, \text{units}$.  

1. Compute the resultant vector $\vec{R} = \vec{B} + \vec{C}$.  
2. Find the magnitude and direction (angle with the $+x$-axis) of $\vec{R}$.  
3. Calculate the vector $\vec{D} = \vec{B} - \vec{C}$.

## **Problem 4: Dot Product and Unit Vectors**  
Two vectors are given as:  
- $\vec{P} = (6.0 \, \hat{i} + 2.0 \, \hat{j}) \, \text{units}$,  
- $\vec{Q} = (3.0 \, \hat{i} - 4.0 \, \hat{j}) \, \text{units}$.  

1. Compute the dot product $\vec{P} \cdot \vec{Q}$.  
2. Interpret the physical significance of a positive or negative dot product.  
3. Determine the unit vector in the direction of $\vec{P}$.

## **Problem 5: Position and Displacement Vectors**  
A particle moves from point $A(2, 3)$ to point $B(8, 7)$ in a 2D coordinate system.  
1. Write the position vectors $\vec{r}_A$ and $\vec{r}_B$.  
2. Calculate the displacement vector $\vec{d} = \vec{r}_B - \vec{r}_A $.  
3. Find the magnitude and direction of the displacement vector.
