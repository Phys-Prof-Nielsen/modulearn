## **Problem 1: Velocity and Acceleration in 2D**  
A particle moves along a straight path in the $x-y$ plane, and its velocity at any time $t$ is given by the vector $\vec{v}(t) = (4.0 \, \hat{i} - 3.0 \, \hat{j}) + (2.0 \, \hat{i} + 5.0 \, \hat{j}) t$ m/s.  
1. Calculate the velocity vector at $t = 2.0 \, \text{s} $.  
2. Find the avearage acceleration between $ t = 2.0 \, \text{s} $ and $ t = 5.0 \, \text{s} $.  
3. Determine the object's speed at $ t = 2.0 \, \text{s} $.

## **Problem 2: Kinematic Equations in 2D**

1. What is the car's velocity after 5.0 seconds?  
2. How far has the car traveled during this time?  
3. What is the car's acceleration in vector form, assuming no motion in the $y$-direction?

## **Problem 4: 3D Motion with Constant Acceleration**

1. After 5.0 seconds, calculate the velocity of the particle.

## **Problem 5: Application of Kinematic Principles in 2D and 3D**  
A drone is flying in a straight line at an initial velocity of $\vec{v}_0 = (5.0 \, \hat{i} + 3.0 \, \hat{j} + 2.0 \, \hat{k}) \, \text{m/s}$ at a height of 10.0 meters above the ground. The drone accelerates with a constant acceleration of $\vec{a} = (1.0 \, \hat{i} - 2.0 \, \hat{j} + 0.5 \, \hat{k}) \, \text{m/s}^2$ for 5.0 seconds. After this period, the drone changes direction and begins moving with a new velocity.

1. **Velocity After 5 Seconds:**
   - Calculate the velocity vector of the drone after 5.0 seconds of motion.

2. **Displacement After 5 Seconds:**
   - Determine the displacement vector of the drone during the 5.0 seconds.

3. **Position After 5 Seconds:**
   - Find the position of the drone relative to its starting point after 5.0 seconds.

4. **Conceptual Analysis:**
   - Interpret the physical meaning of the displacement and velocity vectors in the context of this motion.
   - How do the components of the velocity and acceleration vectors relate to the drone's overall motion in 3D space?
