# C2.0 Learning Outcomes

By the end of this module, students will be able to:  

## **Kinematics in Two and Three Dimensions**  
1. Define and distinguish between the concepts of speed, velocity, and acceleration in two and three dimensions.  
2. Represent velocity and acceleration vectors graphically and algebraically, understanding their components in each dimension.  
3. Apply kinematic equations to solve problems involving motion in two and three dimensions with constant acceleration.  
4. Analyze motion along different coordinate axes (e.g., Cartesian coordinates), breaking vectors into their components for simpler analysis.  

## **Speed and Velocity**  
5. Calculate the average velocity of an object in two and three dimensions.  
6. Interpret the velocity vector in both magnitude and direction, and distinguish between the concepts of speed (scalar) and velocity (vector).  

## **Acceleration and Its Relationships**  
7. Define acceleration in vector form and relate it to changes in velocity over time.  
8. Solve problems involving constant acceleration in multiple dimensions, using appropriate kinematic equations.  

## **Projectile Motion**  
9. Analyze projectile motion in two dimensions, treating horizontal and vertical components separately.  
10. Solve problems involving the motion of a projectile without air drag, calculating key quantities such as range, maximum height, time of flight, and velocity components.  
11. Use the kinematic equations to model and solve projectile motion problems, applying the principles of independence of horizontal and vertical motion.    

## **Problem-Solving Skills**  
12. Develop a systematic approach to solving kinematic problems in two and three dimensions, applying vector notation, components, and algebraic equations.  
13. Interpret the results of kinematic problems within the context of real-world physical systems, including analyzing the motion of projectiles in various environments and comparing descriptions from different reference frames.
