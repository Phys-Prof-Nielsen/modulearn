By the end of this module, students will be able to:

1. **Define and Distinguish Fluids**
   - Describe the differences between solids, liquids, and gases based on their response to applied shear forces.  
   - Explain why fluids flow and how their inability to resist shear differentiates them from solids.

2. **Understand and Calculate Density**
   - Define density ($\rho = \frac{m}{V}$) and specific weight ($\gamma = \rho g$).  
   - Explain the concept of specific gravity and apply it to determine whether substances float or sink in water.  
   - Perform calculations involving density, mass, and volume for uniform and non-uniform materials.

3. **Define and Interpret Pressure**
   - Define pressure as force per unit area ($P = \frac{F}{A}$) and understand its units and physical meaning.  
   - Distinguish between normal and tangential stresses in fluids and explain why only normal stresses exist at rest.  
   - Explain the isotropy of pressure in a fluid at rest (pressure acts equally in all directions).

4. **Apply the Hydrostatic Pressure Relationship**
   - Derive and apply the relationship between pressure and depth in a static fluid:
     
     $$
     P = P_0 + \rho g h
     $$
     
   - Explain how this relationship arises from a force balance on a fluid element.  
   - Solve problems involving pressure differences between two points in a fluid.

5. **Differentiate Types of Pressure**
   - Distinguish between absolute, gauge, and atmospheric pressure.  
   - Convert between gauge and absolute pressures using:
      
     $$
     P_\text{abs} = P_\text{atm} + P_\text{gauge}
     $$

6. **Interpret Fluids in Equilibrium**
   - State that pressure is the same at all points along a horizontal plane in a static fluid.  
   - Explain that pressure variations depend only on vertical position and fluid density, not container shape.  
   - Connect microscopic particle collisions to macroscopic pressure and density.

7. **Connect to Physical and Engineering Contexts**
   - Recognize how pressure and density relate to real-world applications such as hydraulic systems, buoyancy, and atmospheric pressure variation.  
   - Explain how these foundational concepts set the stage for Pascal’s and Archimedes’ principles.
