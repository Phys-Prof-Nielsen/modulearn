# C9.4 Archimedes’ Principle
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C9.4.1 Motivation: Why Do Some Objects Float While Others Sink?

When you drop a rock into water, it sinks.  
When you drop a piece of wood, it floats.  
A heavy steel ship, on the other hand, floats easily — even though steel is much denser than water.  
What determines whether something floats, sinks, or hovers in between?

This everyday observation puzzled ancient scientists for centuries until **Archimedes of Syracuse** (3rd century BCE) discovered the underlying principle.  
While studying how gold could be distinguished from silver by density, Archimedes realized that **an object immersed in a fluid experiences an upward force** — now known as the **buoyant force** — equal to the **weight of the fluid displaced** by the object.

Archimedes’ principle not only explains floating and sinking but also underlies technologies from **ship design** and **submarines** to **hot air balloons** and **blood flow measurements** in medicine.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C9.4.2 Statement of Archimedes’ Principle

> *Any object wholly or partially submerged in a fluid experiences an upward buoyant force equal in magnitude to the weight of the fluid displaced by the object.*

Mathematically:

$$
F_B = \rho_\text{fluid} g V_\text{displaced}
$$

where  
- $F_B$ = buoyant force (N),  
- $\rho_\text{fluid}$ = density of the surrounding fluid (kg/m³),  
- $g$ = gravitational acceleration (m/s²),  
- $V_\text{displaced}$ = volume of fluid displaced (m³).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C9.4.3 Derivation

Consider a rectangular object of cross-sectional area $A$ and height $h$ submerged in a fluid.

- Pressure at the **top** surface:  
  $P_\text{top} = P_0 + \rho g y_\text{top}$

- Pressure at the **bottom** surface:  
  $P_\text{bottom} = P_0 + \rho g y_\text{bottom}$

Because the bottom is deeper by a distance $h$:

$$
P_\text{bottom} - P_\text{top} = \rho g h
$$

The **upward force** on the bottom and **downward force** on the top are:

$$
F_\text{bottom} = P_\text{bottom} A, \quad F_\text{top} = P_\text{top} A
$$

So the **net upward (buoyant) force** is:

$$
F_B = F_\text{bottom} - F_\text{top} = (P_\text{bottom} - P_\text{top}) A = \rho g h A
$$

But $A h$ is the **volume of the displaced fluid**, $V_\text{displaced}$. Therefore,

$$
\boxed{F_B = \rho_\text{fluid} g V_\text{displaced}}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C9.4.4 Conditions for Floating, Sinking, or Equilibrium

Let $W$ be the **weight of the object** ($W = \rho_\text{object} g V_\text{object}$).

| Condition | Description | Mathematical Relation |
|:--|:--|:--|
| **Sinks** | Object weight > buoyant force | $\rho_\text{object} > \rho_\text{fluid}$ |
| **Floats** | Object weight < buoyant force | $\rho_\text{object} < \rho_\text{fluid}$ |
| **Neutral buoyancy** | Object weight = buoyant force | $\rho_\text{object} = \rho_\text{fluid}$ |

When floating, only part of the object is submerged so that the weight displaced fluid corresponds exactly to the weight of the object.

## C9.4.5 Applications

- **Ships and Submarines:**  
  A ship floats because its **average density** (including air spaces) is less than water.  
  Submarines control their depth by adjusting the amount of water in their ballast tanks, changing their average density.

- **Hot-Air Balloons:**  
  The buoyant force comes from the **displaced air**, and the air inside the balloon is heated to reduce its density.  
  When $\rho_\text{balloon} < \rho_\text{air}$, it rises.

- **Hydrometers:**  
  Instruments that measure fluid density by floating at different depths depending on the buoyant force.

- **Medical Applications:**  
  - **Body density measurements:** Used to estimate body fat percentage via water displacement (hydrostatic weighing).  
  - **Blood flow and pressure studies:** Buoyant effects on suspended particles in plasma can help determine density variations and flow characteristics.

- **Engineering and Geophysics:**  
  - **Icebergs** float with about 90% of their volume submerged because $\rho_\text{ice} / \rho_\text{water} \approx 0.9$.  
  - **Isostasy** in geology — the “floating” of continental crust on the denser mantle — is an application of Archimedes’ principle on a planetary scale.

## C9.4.6 Conceptual Summary

| Concept | Expression | Description |
|:--|:--|:--|
| Buoyant Force | $F_B = \rho_\text{fluid} g V_\text{displaced}$ | Upward force from displaced fluid |
| Floating Condition | $F_B = W$ | Object floats when weight equals buoyant force |
| Density Relation | $\rho_\text{object} < \rho_\text{fluid}$ | Required for floating |
| Pressure Basis | $F_B = (P_\text{bottom} - P_\text{top})A$ | Arises from pressure difference with depth |

## C9.4.7 Key Insights

- The buoyant force arises **entirely from pressure differences** within the fluid.  
- It depends only on the **fluid’s density** and **displaced volume**, not on the object’s material or shape (as long as the volume displaced is the same).  
- Archimedes’ principle provides a direct connection between **fluid statics** and **Newtonian mechanics**, linking microscopic pressure forces to macroscopic buoyancy behavior.
