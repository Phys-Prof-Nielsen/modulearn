# C5.5 Acceleration and Newton’s Laws in Galilean Relativity

## C5.5.1 Motivation — Why Study Acceleration?

So far, we have learned how **position** and **velocity** transform between inertial reference frames:

$$
\vec{r} = \vec{r}' + \vec{u}\Delta t,
$$

$$
\vec{v} = \vec{v}' + \vec{u}.
$$

These relations show that different observers moving at constant relative velocity will generally disagree about positions and velocities.

But Newton’s laws are not written in terms of position or velocity alone — they are written in terms of **acceleration**:

$$
\vec{F} = m\vec{a}.
$$

If the laws of physics are to have the same form in all inertial frames (the principle of relativity), then acceleration must transform in a very special way.

This section shows that under Galilean relativity,

$$
\boxed{\vec{a} = \vec{a}'}.
$$

This simple result is the foundation of classical mechanics.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.5.2 Acceleration Transformation (Algebraic Approach)

We begin with the velocity transformation:

$$
\vec{v} = \vec{v}' + \vec{u},
$$

where $\vec{u}$ is the constant relative velocity between two inertial frames $S$ and $S'$.

---

### Step 1 — Consider Two Times

Suppose a particle has velocities at two different times:

At time $t_1$:
$$
\vec{v}_1 = \vec{v}'_1 + \vec{u}
$$

At time $t_2$:
$$
\vec{v}_2 = \vec{v}'_2 + \vec{u}
$$

Because the frames move at constant relative velocity, $\vec{u}$ is the same at both times.

---

### Step 2 — Subtract

Subtract the first equation from the second:

$$
\vec{v}_2 - \vec{v}_1
=
(\vec{v}'_2 + \vec{u}) - (\vec{v}'_1 + \vec{u}).
$$

The $\vec{u}$ terms cancel:

$$
\vec{v}_2 - \vec{v}_1
=
\vec{v}'_2 - \vec{v}'_1.
$$

---

### Step 3 — Divide by the Same Time Interval

Acceleration is defined algebraically as

$$
\vec{a} = \frac{\vec{v}_2 - \vec{v}_1}{\Delta t},
\qquad
\vec{a}' = \frac{\vec{v}'_2 - \vec{v}'_1}{\Delta t'}.
$$

Under Galilean relativity:

- Time is absolute: $\Delta t = \Delta t'$.
- Relative velocity $\vec{u}$ is constant.

Therefore,

$$
\boxed{\vec{a} = \vec{a}'}.
$$

Acceleration is **invariant** under Galilean transformations.

### Calculus Approach

An easier approach is using calculus on the velocity transformation. Since $\vec{u}$ is constant, we readily obtain the same answer.

## C5.5.3 Consequences for Newton’s Laws

Newton’s Second Law states:

$$
\vec{F}_{net} = m\vec{a}.
$$

Since:

- $\vec{a} = \vec{a}'$,
- mass is invariant for a point particle,

it follows immediately that

$$
\vec{F}_{net} = \vec{F}'_{net}.
$$

Thus, Newton’s Second Law has the **same mathematical form** in all inertial reference frames.

This property is called **Galilean invariance**.

---

### Physical Meaning

Adding a constant velocity to every object in the universe:

- Changes positions.
- Changes velocities.
- Does **not** change accelerations.

Because acceleration measures *change* in velocity, and constant offsets cancel out.

That is why:

- A passenger in a smoothly moving train
- A person standing on the ground

will measure the same acceleration for a falling object.

## C5.5.4 Important Limitation

The result $\vec{a} = \vec{a}'$ depends on two assumptions:

1. The relative velocity between frames is constant (inertial frames).
2. Time is absolute ($t = t'$).

If either assumption fails:

- In accelerating (non-inertial) frames → fictitious forces appear.
- In special relativity → time is not absolute, and acceleration transformation becomes more subtle.

This marks the boundary of validity of Galilean relativity.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">


## Summary

Under Galilean transformations:

- Position depends on frame motion.
- Velocity depends on frame motion.
- Acceleration does **not** depend on frame motion.

This invariance of acceleration ensures that

$$
\vec{F} = m\vec{a}
$$

has the same form in all inertial frames, which is why classical mechanics is internally consistent.
