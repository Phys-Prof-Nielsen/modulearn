## **Problem 1: Newton's Laws and Free-Body Diagrams**  
A 5.0 kg box is placed on an inclined plane that makes a 30° angle with the horizontal. The coefficient of kinetic friction between the box and the plane is 0.2.  
1. Draw a free-body diagram for the box.  
2. Determine the acceleration of the box as it slides down the incline.  
3. Calculate the normal force acting on the box.

## **Problem 2: Motion in Two Dimensions with Constraints**  
Two masses, $m_1 = 3.0~\textrm{kg}$ and $m_2 = 5.0~\textrm{kg}$, are connected by a light, inextensible rope that passes over a frictionless pulley. Mass $m_1$ is initially at rests on a horizontal table with a coefficient of kinetic friction of 0.15, while $m_2$ hangs vertically. At $t = 0$s the system is released. 
1. Draw free-body diagrams for both masses.  
2. Using a suitable reference frame of your choice, specify the equation(s) of constraint(s).
3. Write the equations of motion for the objects in system.  
4. Determine the acceleration of the objects in the system and the tension in the rope right after the system is released.

## **Problem 3: Forces in Three Dimensions**  
A particle is acted upon by three forces in three-dimensional space:  
- $\vec{F}_1 = (3 \, \hat{i} + 2 \, \hat{j} - 4 \, \hat{k}) \, \text{N} $,  
- $ \vec{F}_2 = (-5 \, \hat{i} + 4 \, \hat{j} + 3 \, \hat{k}) \, \text{N} $,  
- $ \vec{F}_3 = (2 \, \hat{i} - 6 \, \hat{j} + 1 \, \hat{k}) \, \text{N} $.  

1. Determine the net force acting on the particle.  
2. If the particle has a mass of $2.0$ kg, calculate its acceleration.

## **Problem 4: Friction and Equilibrium**  
A block of mass $10.0$ kg is placed on a rough horizontal surface. A horizontal force $\vec{F}$ is applied to the block. The coefficient of static friction is 0.4, and the coefficient of kinetic friction is 0.3.  
1. Determine the minimum force required to start moving the block.  
2. Once the block starts moving, calculate the acceleration if the applied force is $60.0$ N.

## **Problem 5: Systems with Pulleys and Forces**  
A system consists of two objects connected by a rope passing over a pulley. A $10.0$ kg block is on a frictionless incline at an angle of $35^\circ$ to the horizontal, and it is connected to a $15.0$ kg block hanging vertically.  
1. Draw free-body diagrams for both blocks.  
2. Write the equations of motion for objects in the system.  
3. Determine the acceleration of the objects in the system and the tension in the rope.
