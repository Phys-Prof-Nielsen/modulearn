# C4.0 Learning Outcomes

By the end of this module, students will be able to:  

## **Newton's Laws of Motion**  
1. State and explain Newton's three laws of motion.  
2. Represent forces and motion in two and three dimensions using vector and components notations.  
3. Apply Newton’s laws to analyze the motion of a particle under various force conditions.  

## **Common Forces**  
4. Identify and describe common forces, including gravitational force, normal force, frictional force, tension, and applied forces.  
5. Represent these forces as vectors and include them in free-body diagrams.  

## **Equations of Motion and Constraints**  
6. Formulate equations of motion for systems involving multiple forces in two and three dimensions.  
7. Apply equations of constraints, such as relationships between connected objects or systems with pulleys, to simplify and solve problems.  

## **Problem-Solving Methodology**  
8. Construct free-body diagrams to identify all forces acting on a system.  
9. Develop a systematic approach to solving problems involving forces, including isolating components and applying vector operations.  
10. Solve problems involving forces and motion in two and three dimensions, emphasizing both analytical and conceptual understanding.  
11. Analyze and solve problems involving forces in systems with constraints, such as objects connected by ropes or surfaces with friction.  

## **Practical Applications**  
12. Apply Newton’s laws and vector-based methodologies to real-world problems, including understanding equilibrium, acceleration, and system dynamics.
