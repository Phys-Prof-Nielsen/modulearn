# C7.2 Impulse-Momentum Theorem - Center of Mass
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.2.1 Point Particle vs. Center of Mass

The CM is the position vector of an object where the object will **behave like a point particle** with a mass equal to that of the object.

The implications of this is: **the center of mass of an object will follow the same laws of physics as a point particle and everything we have discussed for a point particle can be applied to the CM concept of an object**.

<div class="alert alert-danger">
    
In this process, we can only describe the dynamics of the CM location of the object. We can not say anything about the specific dynamics of the object itself. Is it rotating or not? Is it stretching or contracting? We can not tell from the linear CM dynamics.
    
</div>

This later statement has great impact on the work-energy theorem as we will discover.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.2.2 Impulse–Momentum Theorem for a Single Particle

For a single particle of mass $m$, Newton’s second law states that

$$
\vec{f}_{\text{net}} = m \vec{a}.
$$

If the net force acts over a short time interval $\Delta t$, the acceleration produces a change in velocity $\Delta \vec{v}$, so

$$
\vec{f}_{\text{net}} = m \frac{\Delta \vec{v}}{\Delta t}.
$$

Rearranging gives

$$
\vec{f}_{\text{net}} \, \Delta t = m \, \Delta \vec{v}.
$$

The quantity $\vec{F}_{\text{net}} \, \Delta t$ is the **impulse**, the effect of a net force acting over a period of time.  
The quantity $m \vec{v}$ is the **momentum** of the particle.

Thus,

$$
\boxed{\vec{i}_{\text{net}} = \Delta \vec{p}},
$$

where $\vec{i}_{\text{net}} = \vec{F}_{\text{net}} \, \Delta t$ and $\vec{p} = m\vec{v}$.

This is the **impulse–momentum theorem** for a single particle.

## C7.2.3 System of Particles

Now consider a system of $N$ particles.  
Each particle satisfies

$$
\vec{f}_i^{(\text{ext})} + \vec{f}_i^{(\text{int})} = m_i \frac{\Delta \vec{v}_i}{\Delta t}.
$$

Multiplying by $\Delta t$ and summing over all particles gives

$$
\sum_i \left( \vec{f}_i^{(\text{ext})} + \vec{f}_i^{(\text{int})} \right) \Delta t
= \sum_i m_i \, \Delta \vec{v}_i.
$$

---

### Cancellation of Internal Forces

The internal forces come in equal and opposite pairs that act along the same line (Newton’s third law).  
When we add them all up, they cancel:

$$
\sum_i \vec{f}_i^{(\text{int})} = 0.
$$

Thus, only the **external forces** contribute to the total change in momentum:

$$
\sum_i \vec{f}_i^{(\text{ext})} \, \Delta t
= \sum_i m_i \, \Delta \vec{v}_i.
$$

---

### Define Total Momentum and CM Motion

The total momentum of the system is

$$
\vec{P} = \sum_i m_i \vec{v}_i = M\vec{V}_{\text{cm}}.
$$

Hence, the change in total momentum over the interval is

$$
\Delta \vec{P} = \sum_i m_i \, \Delta \vec{v}_i = M \, \Delta \vec{V}_{\text{cm}}.
$$

Expressing in terms of impulse, we get

$$
\boxed{\vec{I}_{\text{ext}} = \Delta \vec{P}},
$$

where $\vec{I}_{\text{ext}} = \sum_i \vec{F}_i^{(\text{ext})} \, \Delta t$ is the **total external impulse** on the system.

Because $\vec{P} = M \vec{V}_{\text{cm}}$, this can also be written as

$$
\boxed{\vec{I}_{\text{ext}} = M \, \Delta \vec{V}_{\text{cm}}.}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.2.4 Interpretation

- The **total external impulse** on a system equals the **change in the momentum of its center of mass**.
- Internal forces cannot change the system’s total momentum, since they act in equal and opposite pairs.
- Therefore, the center of mass moves as if all external forces acted on a single particle of mass $M$:

  $$
  \vec{F}_{\text{ext}} = M \, \vec{A}_{\text{cm}}.
  $$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">


## C7.2.5 Physical Summary

- Momentum analysis for a system is **exact** as long as mass is constant and Newton’s third law holds for internal forces.  
- Energy analysis requires more care, since internal forces can store or release potential energy.  
- The impulse–momentum theorem is therefore the most general and reliable way to analyze **translational motion**.
