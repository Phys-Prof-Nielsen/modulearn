# C6.0 Learning Outcomes

By the end of this module, students will be able to:

## **Work with Constant Force**
1. Define and calculate the work done by a constant force in one, two, and three dimensions using vector notation.
2. Understand and apply the concept of work as the dot product of force and displacement vectors.
3. Calculate the work done by a force when the angle between the force and displacement vectors is given.

## **Conservative and Non-Conservative Forces**
4. Define conservative forces and explain their key properties, including the independence of work from the path taken.
5. Describe non-conservative forces, such as friction, and understand their effect on the total mechanical energy of a system.
6. Apply the concepts of conservative and non-conservative forces to analyze physical systems and solve related problems.

## **Potential Energy Function for a Constant Force**
7. Derive and apply the potential energy function associated with a constant force.
8. Solve problems to find the potential energy of a system under the influence of a constant force.
9. Understand the relationship between force and potential energy, including how to calculate force from the gradient of the potential energy function.

## **Gravitational Potential Energy**
10. Define gravitational potential energy and understand its role in mechanical systems.
11. Calculate the gravitational potential energy of objects near Earth’s surface.

## **Work-Energy Theorem**
12. State and apply the work-energy theorem, relating the work done on an object to its change in kinetic energy.
13. Solve problems involving the work-energy theorem to calculate changes in kinetic energy resulting from external forces.

## **Conservation of Mechanical Energy**
14. State and apply the principle of conservation of mechanical energy in systems where only conservative forces are acting.
15. Analyze systems where both conservative and non-conservative forces are present, and apply the work-energy theorem and conservation of mechanical energy to determine changes in energy.
16. Solve problems involving the conservation of mechanical energy, accounting for kinetic energy, potential energy, and work done by non-conservative forces.

## **Problem-Solving and Application**
17. Develop systematic strategies for solving problems involving work, energy, and forces in various mechanical systems.
18. Interpret and analyze the results of energy calculations within the context of physical systems, understanding the interplay between work, energy, and force.
