# C1.0 Learning Outcomes

By the end of this module, students will be able to:  

## **Mathematical Foundations**  
1. Demonstrate proficiency in basic trigonometry, including identifying and using trigonometric functions to solve problems related to angles and right triangles.  
2. Apply trigonometric identities and the Pythagorean theorem in physics contexts.  

## **Understanding Vectors**  
3. Define and distinguish between scalar and vector quantities.  
4. Represent vectors geometrically and in component form using Cartesian coordinates.  
5. Convert between the geometric (magnitude and direction) and component forms of vectors.  

## **Vector Operations**  
6. Perform vector addition and subtraction both graphically and analytically.  
7. Calculate the dot product of two vectors, interpret its physical significance, and apply it in physics problems (e.g., work and projections).  
8. Normalize vectors to compute unit vectors and use them to express vector direction.  

## **Position and Displacement Vectors**  
9. Differentiate between position and displacement vectors in physical systems.  
10. Analyze motion in one or more dimensions using position and displacement vectors.  
11. Solve problems involving relative motion using displacement and position vectors.  

## **Problem-Solving and Applications**  
12. Apply vector concepts to solve real-world problems in physics, including motion, forces, and energy.  
13. Use a systematic approach to analyze problems involving multiple vectors, ensuring accurate and precise results.
