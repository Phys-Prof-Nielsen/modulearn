## C1.3.2 Vector Components
Consider a vector $\vec{A}$ in a 2D reference frame as shown in the figure below. If we project the two end points of the vector (or arrow) onto the x and y-axis, respectively, then the directed line segments of the projections are the <span style="color:red">components</span> of $\vec{A}$. We can write them as $A_x$ and $A_y$.

> NOTE: these directed line segments are exactly what we encountered in Phase D.

```python
import matplotlib.pyplot as plt

# -----------------------------
# Vector A endpoints
# -----------------------------
x0, y0 = 1, 1
xA, yA = 4, 5

# Vector components
Ax = xA - x0
Ay = yA - y0

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(5,5))

# Main vector A
plt.quiver(x0, y0, Ax, Ay,
           angles='xy', scale_units='xy', scale=1)

# Projections
# x-projection
plt.quiver(x0, 0.1, Ax, 0,
           angles='xy', scale_units='xy', scale=1, color='r')

# y-projection
plt.quiver(0.1, y0, 0, Ay,
           angles='xy', scale_units='xy', scale=1, color='r')

# Dashed guide lines
plt.plot([xA, xA], [0.1, yA], 'k--', linewidth=1)
plt.plot([0.1, xA], [yA, yA], 'k--', linewidth=1)
plt.plot([0.1, x0], [x0, y0], 'k--', linewidth=1)
plt.plot([1,1], [0.1, 1], 'k--', linewidth=1)

# Labels
plt.text(xA + 0.1, yA + 0.1, r"$\vec{A}$", fontsize=12)
plt.text(x0 + Ax/2, 0.3, r"$A_x$", ha='center')
plt.text(0.3, y0 + Ay/2, r"$A_y$", va='center')

# Axes and formatting
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')

plt.xlim(0, 6)
plt.ylim(0, 6)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Vector A and Its Projections")

plt.show()
```

We can now describe the vector $\vec{A}$ through its components (or in component form):

$$ \vec{A} = \begin{bmatrix} A_x  \\ A_y  \end{bmatrix} $$

Other ways of writing the same thing:

$$ \vec{A} = ~< A_x , A_y > $$

or

$$ \vec{A} = ( A_x, A_y ) $$

When we discuss unit vectors, we will see one more way of writing a vector.

## C1.3.3 Magnitude of a Vector
Since the two projected components and the length of the vector itself form a right-angle triangle, we calculate the
magnitude of the vector by applying the Pythagorean theorem:

$$ |\vec{A}| = \sqrt{ A^{2}_{x} + A^{2}_{y} } $$

It can be extended beyond 2D, so in a 3D coordinate
system the magnitude of a the vector would be

$$ |\vec{A}| = \sqrt{ A^{2}_{x} + A^{2}_{y} + A^{2}_{z} } $$

## C1.3.4 Direction of a Vector
Above, we saw how we can find the magnitude of a vector if we have knowledge of the vector components. In order to fully describe the vector, we still need to find its direction. The direction can be provided through an angle measured with respect to some reference line. If we draw the vector and its components, we can identify a right-angle triangle and apply trigonometric functions to specify an angle.

```python
import matplotlib.pyplot as plt

# -----------------------------
# Vector A endpoints
# -----------------------------
x0, y0 = 1, 1
xA, yA = 4, 5

# Vector components
Ax = xA - x0
Ay = yA - y0

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(5,5))

# Main vector A
plt.quiver(x0, y0, Ax, Ay,
           angles='xy', scale_units='xy', scale=1)

# Projections
# x-projection
plt.quiver(x0, x0, Ax, 0,
           angles='xy', scale_units='xy', scale=1, color='r')

# y-projection
plt.quiver(xA, y0, 0, Ay,
           angles='xy', scale_units='xy', scale=1, color='r')

# Labels
plt.text(xA + 0.1, yA + 0.1, r"$\vec{A}$", fontsize=12)
plt.text(x0 + Ax/2, 0.7, r"$A_x$", ha='center')
plt.text(4.2, y0 + Ay/2, r"$A_y$", va='center')
plt.text(1.3, 1.2, r"$\theta$", va='center')

# Axes and formatting
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')

plt.xlim(0, 6)
plt.ylim(0, 6)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Direction of a Vector")

plt.show()
```

The relation between the angle $\theta$ and the components $A_x$ (adjacent side) and $A_y$ (opposite side) is

$$ \tan\theta = \frac{A_y}{A_x} $$

or

$$ \theta = \tan^{-1}\left(\frac{A_y}{A_x}\right) $$

Similar relations exists if we include the magnitude of the vector itself (hypotenuse):

$$ \sin\theta = \frac{A_y}{|\vec{A}|} $$

and

$$ \cos\theta = \frac{A_x}{|\vec{A}|} $$

<div style="background-color:#ffe6e6; border-left:6px solid #990000; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">⚠️ Important Warning — Association of Trigonometric functions with Directions</h3>
Here, we identified x- and y-components based on a standard Cartesion coordinate system for the given triangle. One should avoid associating cosine with x-component and sine with y-component as it depends on choice and orientation of the coordinate axes. Instead, it is advisable to consider each case by itself and go back to basic trigonometric function definition using adjacent and opposite side terminology.</div>

<h3 style="margin-top:0; color:#000000;">Box Activity 4 — Components from Magnitude and Direction</h3>

**Task:**

A position vector $\vec{r}$ has a **magnitude** of $8.0\ \text{m}$ and is directed  
$30.0^\circ$ counterclockwise from the positive $x$-axis.

1. Sketch the vector $\vec{r}$ on an $x$–$y$ coordinate system.  
2. Determine the $x$-component $r_x$ and the $y$-component $r_y$.  
3. State whether each component is positive or negative, and explain why.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

When a vector of magnitude $r$ makes an angle $\theta$ with the positive $x$-axis, its components are

$$
r_x = r\cos\theta
\qquad
r_y = r\sin\theta.
$$

Substituting $r = 8.0\ \text{m}$ and $\theta = 30.0^\circ$:

$$
r_x = (8.0)\cos(30.0^\circ) \approx 6.9\ \text{m}
$$

$$
r_y = (8.0)\sin(30.0^\circ) = 4.0\ \text{m}.
$$

Both components are positive because the vector points into the **first quadrant**, where $x>0$ and $y>0$.
