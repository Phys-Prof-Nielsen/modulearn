By the end of this module, students will be able to:  

## **Linear Momentum**  
1. Define linear momentum as a vector quantity and describe its physical significance.  
2. Represent the linear momentum of a point particle in two and three dimensions using vector notation.  
3. Analyze the relationship between mass, velocity, and linear momentum in various physical systems.  

## **Impulse**  
4. Define impulse as the change in linear momentum of a point particle in vector form.  
5. Calculate impulse using the difference in momentum vectors.  
6. Explain the relationship between impulse and external constant forces acting on a particle over a time interval.  

## **Vector Applications in Momentum and Impulse**  
7. Solve problems involving linear momentum and impulse in two and three dimensions using vector operations.  
8. Interpret the geometric and component forms of vectors in momentum and impulse problems.  
9. Apply conservation of linear momentum to analyze physical systems involving collisions and interactions in multiple dimensions.  

## **Problem-Solving Skills**  
10. Use a systematic and logical approach to analyze scenarios involving momentum and impulse, ensuring clarity and accuracy in vector calculations.
