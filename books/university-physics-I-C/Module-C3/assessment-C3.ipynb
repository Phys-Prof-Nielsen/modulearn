## **Problem 5: Conservation of Momentum**  
Two particles collide in two dimensions. Before the collision:  
- Particle 1 (mass = 2.0 kg) has a velocity of $\vec{v}_1 = (3.0 \, \hat{i} + 2.0 \, \hat{j}) \, \text{m/s}$.  
- Particle 2 (mass = 4.0 kg) has a velocity of $\vec{v}_2 = (-1.0 \, \hat{i} + 1.0 \, \hat{j}) \, \text{m/s}$.  

After the collision:  
- Particle 2 has a velocity of $\vec{v}_2' = (0.0 \, \hat{i} + 1.5 \, \hat{j}) \, \text{m/s}$.  

1. Using the principle of conservation of momentum, calculate the final velocity of particle 1.
2. Verify that the total momentum before and after the collision is conserved.
