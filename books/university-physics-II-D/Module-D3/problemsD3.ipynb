## Problem D3.1 — Parallel-Plate Capacitance, Voltage, Field, and Stored Energy

A parallel-plate capacitor has plate area

$$
A = 0.0150\ \text{m}^2
$$

and plate separation

$$
d = 1.50\ \text{mm}.
$$

Assume the space between the plates is air (treat as vacuum), so use

$$
\varepsilon_0 = 8.85\times 10^{-12}\ \text{F/m}.
$$

The capacitor is charged so that the magnitude of charge on each plate is

$$
Q = 6.00\times 10^{-6}\ \text{C}.
$$

Answer the following:

1. Compute the capacitance $C$ from geometry.
2. Compute the potential difference $\Delta V$ between the plates.
3. Compute the electric field magnitude $E$ between the plates (assume it is uniform).
4. Compute the change in stored electric potential energy $\Delta U$.

Be careful with units and provide a short physical interpretation of your results.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Parallel-plate capacitance:

$$
C = \varepsilon_0 \frac{A}{d}
$$

Capacitance definition:

$$
\Delta V = \frac{Q}{C}
$$

Uniform field between plates:

$$
E = \frac{\Delta V}{d}
$$

Energy stored in a capacitor:

$$
\Delta U = \frac{1}{2}Q\,\Delta V
$$

---

**Given**

- $A = 0.0150\ \text{m}^2$
- $d = 1.50\times 10^{-3}\ \text{m}$
- $\varepsilon_0 = 8.85\times 10^{-12}\ \text{F/m}$
- $Q = 6.00\times 10^{-6}\ \text{C}$

---

**1) Capacitance**

$$
C = \varepsilon_0\frac{A}{d}
= (8.85\times 10^{-12})\frac{0.0150}{1.50\times 10^{-3}}
= 8.85\times 10^{-11}\ \text{F}
$$

$$
\boxed{C = 8.85\times 10^{-11}\ \text{F}}
$$

---

**2) Potential difference**

$$
\Delta V = \frac{Q}{C}
= \frac{6.00\times 10^{-6}}{8.85\times 10^{-11}}
= 6.78\times 10^{4}\ \text{V}
$$

$$
\boxed{\Delta V = 6.78\times 10^{4}\ \text{V}}
$$

---

**3) Electric field magnitude**

$$
E = \frac{\Delta V}{d}
= \frac{6.78\times 10^{4}}{1.50\times 10^{-3}}
= 4.52\times 10^{7}\ \text{V/m}
$$

$$
\boxed{E = 4.52\times 10^{7}\ \text{V/m}}
$$

---

**4) Stored energy**

$$
\Delta U = \frac{1}{2}Q\,\Delta V
= \frac{1}{2}(6.00\times 10^{-6})(6.78\times 10^{4})
= 2.03\times 10^{-1}\ \text{J}
$$

$$
\boxed{\Delta U = 2.03\times 10^{-1}\ \text{J}}
$$

---

**Physical interpretation**

- The capacitance is small because vacuum/air capacitors require large plate area or very small separation to store significant charge.
- The large voltage and strong electric field reflect that storing microcoulombs on an air-gap capacitor can require very high potentials.
- The stored energy is modest here because $C$ is small, even though the voltage is large.

</div>
</details>

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Given values
# -----------------------------
eps0 = 8.85e-12    # F/m
A = 0.0150         # m^2
d = 1.50e-3        # m
Q = 6.00e-6        # C

# -----------------------------
# Physics:
# Parallel-plate capacitor:
# C = eps0 * A / d
#
# Capacitance definition:
# ΔV = Q / C
#
# Uniform field between plates:
# E = ΔV / d
#
# Stored energy (change from uncharged state):
# ΔU = (1/2) Q ΔV
# -----------------------------
C = eps0 * A / d
dV = Q / C
E = dV / d
dU = 0.5 * Q * dV

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Parallel-plate capacitor: capacitance, voltage, field, and stored energy\n")
print("Given:")
print(f"  Plate area:        A = {A:.3e} m^2")
print(f"  Plate separation:  d = {d:.3e} m")
print(f"  Vacuum permittivity: eps0 = {eps0:.3e} F/m")
print(f"  Stored charge:     Q = {Q:.3e} C\n")

print("Results (3 significant figures):")
print(f"  Capacitance:       C = {C:.3e} F")
print(f"  Potential diff.:  ΔV = {dV:.3e} V")
print(f"  Electric field:    E = {E:.3e} V/m")
print(f"  Stored energy:    ΔU = {dU:.3e} J\n")

print("Physical interpretation:")
print("• C depends only on geometry (A and d) for an air/vacuum capacitor.")
print("• For fixed Q, a small capacitance implies a large voltage: ΔV = Q/C.")
print("• The field between plates is E = ΔV/d (approximately uniform).")
print("• ΔU is the energy added while charging from Q = 0 to Q; it equals the work done by an external agent.")
```

## Problem D3.2 — Battery Connected vs Disconnected: What Changes When Plate Spacing Changes?

A parallel-plate capacitor has plate area

$$
A = 0.0200\ \text{m}^2
$$

and an initial plate separation

$$
d_0 = 1.00\ \text{mm}.
$$

Assume air (treat as vacuum), so

$$
\varepsilon_0 = 8.85\times 10^{-12}\ \text{F/m}.
$$

The capacitor is initially connected to a battery that maintains a constant potential difference

$$
\Delta V_0 = 500\ \text{V}.
$$

The plate separation is then slowly increased to

$$
d_1 = 3.00\ \text{mm}.
$$

Answer the following for **each** scenario:

### Scenario A — Battery stays connected
1. Find $C_0$ and $C_1$.
2. Find $Q_0$ and $Q_1$.
3. Find $E_0$ and $E_1$.
4. Find $\Delta U_0$ and $\Delta U_1$.

### Scenario B — Battery is disconnected before the plates move
The capacitor is first charged to the initial state, then the battery is removed so that charge cannot change.
1. Find $C_0$ and $C_1$.
2. Find $Q_0$ and $Q_1$.
3. Find $\Delta V_0$ and $\Delta V_1$.
4. Find $E_0$ and $E_1$.
5. Find $\Delta U_0$ and $\Delta U_1$.

Finally, in one or two sentences: **Which quantities are held fixed in each scenario, and why?**

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Parallel-plate capacitance:

$$
C = \varepsilon_0 \frac{A}{d}
$$

Battery connected:

$$
Q = C\,\Delta V \quad (\Delta V\ \text{fixed})
$$

Battery disconnected:

$$
\Delta V = \frac{Q}{C} \quad (Q\ \text{fixed})
$$

Uniform field between plates:

$$
E = \frac{\Delta V}{d}
$$

Energy stored:

$$
\Delta U = \frac{1}{2}C(\Delta V)^2 = \frac{Q^2}{2C}
$$

---

**Given**

- $A = 0.0200\ \text{m}^2$
- $d_0 = 1.00\times 10^{-3}\ \text{m}$
- $d_1 = 3.00\times 10^{-3}\ \text{m}$
- $\Delta V_0 = 500\ \text{V}$
- $\varepsilon_0 = 8.85\times 10^{-12}\ \text{F/m}$

---

**Step 1: Capacitances**

$$
C_0 = \varepsilon_0\frac{A}{d_0}
= (8.85\times 10^{-12})\frac{0.0200}{1.00\times 10^{-3}}
= 1.77\times 10^{-10}\ \text{F}
$$

$$
C_1 = \varepsilon_0\frac{A}{d_1}
= (8.85\times 10^{-12})\frac{0.0200}{3.00\times 10^{-3}}
= 5.90\times 10^{-11}\ \text{F}
$$

So $C_1 = \frac{1}{3}C_0$.

---

**Scenario A — Battery connected ($\Delta V$ fixed)**

**Charge**

$$
Q_0 = C_0 \Delta V_0 = (1.77\times 10^{-10})(500)= 8.85\times 10^{-8}\ \text{C}
$$

$$
Q_1 = C_1 \Delta V_0 = (5.90\times 10^{-11})(500)= 2.95\times 10^{-8}\ \text{C}
$$

**Electric field**

$$
E_0 = \frac{\Delta V_0}{d_0} = \frac{500}{1.00\times 10^{-3}} = 5.00\times 10^{5}\ \text{V/m}
$$

$$
E_1 = \frac{\Delta V_0}{d_1} = \frac{500}{3.00\times 10^{-3}} = 1.67\times 10^{5}\ \text{V/m}
$$

**Energy**

$$
\Delta U_0 = \frac{1}{2}C_0(\Delta V_0)^2
= \frac{1}{2}(1.77\times 10^{-10})(500)^2
= 2.21\times 10^{-5}\ \text{J}
$$

$$
\Delta U_1 = \frac{1}{2}C_1(\Delta V_0)^2
= \frac{1}{2}(5.90\times 10^{-11})(500)^2
= 7.38\times 10^{-6}\ \text{J}
$$

---

**Scenario B — Battery disconnected ($Q$ fixed)**

Here the charge stays the same:

$$
Q_1 = Q_0 = 8.85\times 10^{-8}\ \text{C}.
$$

**Voltage**

$$
\Delta V_1 = \frac{Q_0}{C_1}
= \frac{8.85\times 10^{-8}}{5.90\times 10^{-11}}
= 1.50\times 10^{3}\ \text{V}
$$

So $\Delta V_1 = 3\Delta V_0$.

**Electric field**

$$
E_0 = \frac{\Delta V_0}{d_0} = 5.00\times 10^{5}\ \text{V/m}
$$

$$
E_1 = \frac{\Delta V_1}{d_1} = \frac{1.50\times 10^{3}}{3.00\times 10^{-3}}
= 5.00\times 10^{5}\ \text{V/m}
$$

So the electric field remains unchanged.

**Energy**

Using $\Delta U = \frac{Q^2}{2C}$ (best form when $Q$ is fixed):

$$
\Delta U_0 = \frac{Q_0^2}{2C_0}
= \frac{(8.85\times 10^{-8})^2}{2(1.77\times 10^{-10})}
= 2.21\times 10^{-5}\ \text{J}
$$

$$
\Delta U_1 = \frac{Q_0^2}{2C_1}
= \frac{(8.85\times 10^{-8})^2}{2(5.90\times 10^{-11})}
= 6.64\times 10^{-5}\ \text{J}
$$

---

**Final interpretation**

- **Battery connected:** $\Delta V$ is fixed, so $Q$ decreases when $d$ increases, and the field decreases.
- **Battery disconnected:** $Q$ is fixed, so $\Delta V$ increases when $d$ increases, and the field stays the same.

</div>
</details>

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Given values
# -----------------------------
eps0 = 8.85e-12    # F/m
A = 0.0200         # m^2
d0 = 1.00e-3       # m
d1 = 3.00e-3       # m
dV0 = 500.0        # V

# -----------------------------
# Physics:
# C = eps0 * A / d
#
# Scenario A (battery connected): ΔV fixed
#   Q = C ΔV
#   E = ΔV / d
#   ΔU = (1/2) C (ΔV)^2
#
# Scenario B (battery disconnected): Q fixed (equals initial Q0)
#   ΔV = Q / C
#   E = ΔV / d
#   ΔU = Q^2 / (2C)   (convenient when Q fixed)
# -----------------------------

# Capacitances
C0 = eps0 * A / d0
C1 = eps0 * A / d1

# Scenario A: ΔV fixed
Q0_A = C0 * dV0
Q1_A = C1 * dV0
E0_A = dV0 / d0
E1_A = dV0 / d1
dU0_A = 0.5 * C0 * dV0**2
dU1_A = 0.5 * C1 * dV0**2

# Scenario B: Q fixed (equal to initial charge when connected at ΔV0)
Q0_B = Q0_A
Q1_B = Q0_B
dV1_B = Q0_B / C1
E0_B = dV0 / d0
E1_B = dV1_B / d1
dU0_B = Q0_B**2 / (2 * C0)
dU1_B = Q0_B**2 / (2 * C1)

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Parallel-plate capacitor: battery connected vs disconnected (plate spacing changes)\n")
print("Given:")
print(f"  Plate area:        A = {A:.4f} m^2")
print(f"  Initial spacing:   d0 = {d0:.3e} m")
print(f"  Final spacing:     d1 = {d1:.3e} m")
print(f"  Battery voltage:  ΔV0 = {dV0:.3g} V")
print(f"  Vacuum permittivity: eps0 = {eps0:.3e} F/m\n")

print("Capacitance (geometry only):")
print(f"  C0 = {C0:.3e} F")
print(f"  C1 = {C1:.3e} F\n")

print("Scenario A: Battery connected (ΔV fixed)")
print(f"  Q0 = {Q0_A:.3e} C")
print(f"  Q1 = {Q1_A:.3e} C")
print(f"  E0 = {E0_A:.3e} V/m")
print(f"  E1 = {E1_A:.3e} V/m")
print(f"  ΔU0 = {dU0_A:.3e} J")
print(f"  ΔU1 = {dU1_A:.3e} J\n")

print("Scenario B: Battery disconnected before moving plates (Q fixed)")
print(f"  Q0 = {Q0_B:.3e} C (fixed)")
print(f"  Q1 = {Q1_B:.3e} C (fixed)")
print(f"  ΔV0 = {dV0:.3g} V")
print(f"  ΔV1 = {dV1_B:.3e} V")
print(f"  E0 = {E0_B:.3e} V/m")
print(f"  E1 = {E1_B:.3e} V/m")
print(f"  ΔU0 = {dU0_B:.3e} J")
print(f"  ΔU1 = {dU1_B:.3e} J\n")

print("Physical interpretation:")
print("• Scenario A: Battery holds ΔV constant, so increasing d decreases C and therefore decreases Q and E.")
print("• Scenario B: Charge is trapped (fixed Q). Increasing d decreases C, so ΔV rises; E stays the same for parallel plates.")
print("• Energy changes differently depending on what is held fixed (ΔV vs Q).")
```

## Problem D3.3 — Energy and Work to Charge a Capacitor (Calculus)

A capacitor has capacitance

$$
C = 4.00\ \mu\text{F}.
$$

It is initially uncharged. The capacitor is slowly charged until the final potential difference across it is

$$
\Delta V_f = 12.0\ \text{V}.
$$

Answer the following:

1. Find the final charge $Q_f$ on the capacitor.
2. Use a calculus argument to compute the total work done by an external agent during the charging process, and show that this equals the change in stored electric potential energy $\Delta U$.
3. Compute $\Delta U$ numerically.
4. Briefly explain (in one or two sentences) why the factor of $\tfrac{1}{2}$ appears in the energy expression.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Capacitance relationship:

$$
Q = C\,\Delta V
$$

During charging, the instantaneous potential difference when the capacitor holds charge $q$ is

$$
\Delta V(q) = \frac{q}{C}.
$$

Energy/work connection for adding a small amount of charge $dq$:

$$
dW_{\text{ext}} = dU = \Delta V(q)\,dq.
$$

---

**Given**

- $C = 4.00\times 10^{-6}\ \text{F}$
- $\Delta V_f = 12.0\ \text{V}$

---

**1) Final charge**

$$
Q_f = C\,\Delta V_f
= (4.00\times 10^{-6})(12.0)
= 4.80\times 10^{-5}\ \text{C}.
$$

$$
\boxed{Q_f = 4.80\times 10^{-5}\ \text{C}}
$$

---

**2) Work done during charging (calculus)**

At an intermediate stage with charge $q$, the potential difference is

$$
\Delta V(q) = \frac{q}{C}.
$$

The incremental work to move charge $dq$ is

$$
dW_{\text{ext}} = \Delta V(q)\,dq = \frac{q}{C}\,dq.
$$

Total work from $q=0$ to $q=Q_f$:

$$
W_{\text{ext}} = \int_0^{Q_f}\frac{q}{C}\,dq
= \frac{1}{C}\left[\frac{q^2}{2}\right]_0^{Q_f}
= \frac{Q_f^2}{2C}.
$$

Since the capacitor starts uncharged, this work equals the change in stored energy:

$$
\Delta U = W_{\text{ext}} = \frac{Q_f^2}{2C}.
$$

Using $Q_f = C\Delta V_f$ gives the familiar form

$$
\Delta U = \frac{1}{2}C(\Delta V_f)^2.
$$

---

**3) Numerical value of stored energy**

$$
\Delta U = \frac{1}{2}C(\Delta V_f)^2
= \frac{1}{2}(4.00\times 10^{-6})(12.0)^2
= 2.88\times 10^{-4}\ \text{J}.
$$

$$
\boxed{\Delta U = 2.88\times 10^{-4}\ \text{J}}
$$

---

**4) Why the factor $\tfrac{1}{2}$ appears**

The potential difference starts at $0$ and increases linearly to $\Delta V_f$ as charge builds up. The total work is the “area under” this linear increase, which is half of the rectangle you would get if the voltage were constant at $\Delta V_f$ for the entire process.

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Given values
# -----------------------------
C = 4.00e-6       # F
dVf = 12.0        # V

# -----------------------------
# Physics:
# Final charge: Qf = C ΔVf
#
# During charging: ΔV(q) = q/C
# Incremental work (and energy increase): dW_ext = dU = ΔV(q) dq
# Total: ΔU = ∫_0^Qf (q/C) dq = Qf^2/(2C) = (1/2) C (ΔVf)^2
# -----------------------------
Qf = C * dVf
dU = 0.5 * C * dVf**2
Wext = dU  # since capacitor starts uncharged, ΔU = W_ext

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Charging a capacitor: charge, work, and stored potential energy\n")
print("Given:")
print(f"  Capacitance:        C = {C:.3e} F")
print(f"  Final voltage:     ΔVf = {dVf:.3g} V\n")

print("Results (3 significant figures):")
print(f"  Final charge:       Qf = {Qf:.3e} C")
print(f"  Stored energy:     ΔU = {dU:.3e} J")
print(f"  Work by external:  W_ext = {Wext:.3e} J\n")

print("Physical interpretation:")
print("• The voltage rises from 0 to ΔVf as charge accumulates, so the average voltage during charging is (1/2)ΔVf.")
print("• That is why ΔU = (1/2) Qf ΔVf = (1/2) C (ΔVf)^2.")
print("• The work done by the external agent becomes stored electric potential energy in the capacitor's electric field.")
```

## Problem D3.4 — Stud Finder as a Capacitive Sensor

Many electronic stud finders work by detecting changes in **capacitance**. A simplified model is that the device contains a sensing plate facing the wall. The wall acts like the other “plate,” so the system behaves approximately like a parallel-plate capacitor.

Assume the sensor can be modeled as a parallel-plate capacitor with:

$$
A = 8.0\ \text{cm}^2,
\qquad
d = 2.0\ \text{mm},
\qquad
\varepsilon_0 = 8.85\times10^{-12}\ \text{F/m}.
$$

When the sensor is over dry air behind thin drywall, take the effective permittivity as approximately $\varepsilon_0$.

When the sensor is over a wooden stud, the effective permittivity increases (more polarization), and for this problem model it as

$$
\varepsilon_{\text{stud}} = 3.0\,\varepsilon_0.
$$

The drywall acts as a **dielectric**, which we will discuss in Phase B.

The electronics hold the sensor at a constant potential difference of

$$
\Delta V = 9.0\ \text{V}.
$$

Answer the following:

1. Compute the capacitance when the sensor is over air, $C_{\text{air}}$.
2. Compute the capacitance when the sensor is over a stud, $C_{\text{stud}}$.
3. With $\Delta V$ held constant, compute the stored charge in each case: $Q_{\text{air}}$ and $Q_{\text{stud}}$.
4. Compute the stored energy in each case: $\Delta U_{\text{air}}$ and $\Delta U_{\text{stud}}$.
5. In one or two sentences, explain why detecting a change in $Q$ (or $\Delta U$) allows the stud finder to detect the stud.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Parallel-plate capacitance:

$$
C = \varepsilon \frac{A}{d}
$$

Charge at fixed voltage:

$$
Q = C\,\Delta V
$$

Energy at fixed voltage:

$$
\Delta U = \frac{1}{2}C(\Delta V)^2
$$

---

**Given**

- $A = 8.0~\text{cm}^2 = 8.0\times10^{-4}~\text{m}^2$
- $d = 2.0~\text{mm} = 2.0\times10^{-3}~\text{m}$
- $\varepsilon_0 = 8.85\times10^{-12}~\text{F/m}$
- $\varepsilon_{\text{air}} = \varepsilon_0$
- $\varepsilon_{\text{stud}} = 3.0\,\varepsilon_0$
- $\Delta V = 9.0~\text{V}$

---

**1) Capacitance over air**

$$
C_{\text{air}} = \varepsilon_0 \frac{A}{d}
= (8.85\times10^{-12})\frac{8.0\times10^{-4}}{2.0\times10^{-3}}
= 3.54\times10^{-12}~\text{F}.
$$

$$
\boxed{C_{\text{air}} = 3.54~\text{pF}}
$$

---

**2) Capacitance over a stud**

$$
C_{\text{stud}} = (3.0\varepsilon_0)\frac{A}{d}
= 3.0\,C_{\text{air}}
= 1.06\times10^{-11}~\text{F}.
$$

$$
\boxed{C_{\text{stud}} = 10.6~\text{pF}}
$$

---

**3) Stored charge (fixed voltage)**

$$
Q_{\text{air}} = C_{\text{air}}\Delta V
= (3.54\times10^{-12})(9.0)
= 3.19\times10^{-11}~\text{C}
$$

$$
Q_{\text{stud}} = C_{\text{stud}}\Delta V
= (1.06\times10^{-11})(9.0)
= 9.56\times10^{-11}~\text{C}
$$

---

**4) Stored energy (fixed voltage)**

$$
\Delta U_{\text{air}} = \frac{1}{2}C_{\text{air}}(\Delta V)^2
= \frac{1}{2}(3.54\times10^{-12})(9.0)^2
= 1.43\times10^{-10}~\text{J}
$$

$$
\Delta U_{\text{stud}} = \frac{1}{2}C_{\text{stud}}(\Delta V)^2
= \frac{1}{2}(1.06\times10^{-11})(9.0)^2
= 4.29\times10^{-10}~\text{J}
$$

---

**5) Physical interpretation**

A stud increases the effective permittivity, which increases the capacitance. With the same applied voltage, the capacitor stores more charge and more energy. The stud finder detects this change in capacitance (often by measuring how quickly the sensor charges/discharges or how an oscillator frequency shifts), allowing it to locate the stud.

## Problem D3.5 — Thunderstorm–Ground as a Giant Capacitor

In fair-weather conditions, the electric field near Earth’s surface is relatively small. During a thunderstorm, however, charge separation in the clouds can create very large electric fields near the ground. A simple model is to treat the cloud base and the ground as the two “plates” of a gigantic parallel-plate capacitor.

Assume the following idealized thunderstorm scenario:

  R = 1.20\ \text{km}.

## Problem D3.7 — Substation Capacitor Accident: High Voltage, Large Capacitance, Severe Burns

Substation capacitor banks use many individual capacitor units to provide reactive power support. Even a single unit can store a dangerous amount of energy.

In this problem, model one capacitor unit as having:

$$
C = 100\ \mu\text{F}
\qquad \text{charged to} \qquad
\Delta V = 20.0\ \text{kV}.
$$

During an accident, the capacitor discharges through an arc and dumps energy into a region of tissue. To estimate burn severity, model the affected tissue as water with:

- affected tissue mass:

  $$
  m = 0.100\ \text{kg}
  $$
  
- specific heat of water:

  $$
  c = 4186\ \mathrm{J/(kg\cdot^\circ C)}.
  $$

Assume (as a worst-case estimate) that **all** stored electrical energy becomes thermal energy in that tissue region and that heat loss during the short event is negligible.

Answer the following:

1. Compute the stored energy in the capacitor.
2. Compute the temperature increase $\Delta T$ of the tissue.
3. If the tissue started at $33^\circ\text{C}$ (skin temperature), what final temperature does your model predict?
4. In one sentence, explain why this can still be lethal even if the discharge lasts only milliseconds.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Energy stored in a capacitor:

$$
\Delta U = \frac{1}{2}C(\Delta V)^2
$$

Thermal energy and temperature change:

$$
Q_{\text{thermal}} = mc\,\Delta T
\qquad \Rightarrow \qquad
\Delta T = \frac{Q_{\text{thermal}}}{mc}
$$

Worst-case model:

$$
Q_{\text{thermal}} = \Delta U.
$$

---

**Given**

- $C = 100~\mu\text{F} = 1.00\times10^{-4}\ \text{F}$
- $\Delta V = 20.0~\text{kV} = 2.00\times10^{4}\ \text{V}$
- $m = 0.100\ \text{kg}$
- $c = 4186\ \mathrm{J/(kg\cdot^\circ C)}$

---

**1) Stored energy**

$$
\Delta U = \frac{1}{2}(1.00\times10^{-4})(2.00\times10^{4})^2.
$$

Since $(2.00\times10^{4})^2 = 4.00\times10^{8}$,

$$
\Delta U = \frac{1}{2}(1.00\times10^{-4})(4.00\times10^{8})
= 2.00\times10^{4}\ \text{J}.
$$

$$
\boxed{\Delta U = 2.00\times10^{4}\ \text{J}}
$$

---

**2) Temperature increase**

$$
\Delta T = \frac{\Delta U}{mc}
= \frac{2.00\times10^{4}}{(0.100)(4186)}
= 4.78\times10^{1}\ ^\circ\text{C}.
$$

$$
\boxed{\Delta T \approx 47.8^\circ\text{C}}
$$

---

**3) Final temperature**

$$
T_f = 33^\circ\text{C} + 47.8^\circ\text{C} = 80.8^\circ\text{C}.
$$

$$
\boxed{T_f \approx 81^\circ\text{C}}
$$

This is well into the range of severe burns and tissue damage.

---

**4) Why this is still dangerous**

Even if the discharge lasts only milliseconds, the capacitor can deliver very large current and power, causing intense localized heating (arc burns) and potentially driving dangerous current through the body.

**Safety note:** This is a simplified “worst-case” thermal estimate; real injuries depend strongly on the discharge path, arc duration, and how the energy is distributed (some may go into the metal, air plasma, sound, and light). The key point is that the stored energy scale can be enormous.

</div>
</details>

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Given values
# -----------------------------
C = 100e-6          # F
dV = 20.0e3         # V

m = 0.100           # kg
c = 4186.0          # J/(kg·°C)
T_initial = 33.0    # °C

# -----------------------------
# Physics:
# Energy stored: ΔU = (1/2) C (ΔV)^2
# Thermal model: Q_thermal = m c ΔT
# Worst-case: Q_thermal = ΔU  =>  ΔT = ΔU / (m c)
# Final temperature: T_f = T_initial + ΔT
# -----------------------------
dU = 0.5 * C * dV**2
dT = dU / (m * c)
T_final = T_initial + dT

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Substation capacitor accident: energy and temperature rise estimate\n")
print("Given:")
print(f"  Capacitance:       C = {C:.3e} F")
print(f"  Voltage:          ΔV = {dV:.3e} V")
print(f"  Tissue mass:       m = {m:.3f} kg")
print(f"  Specific heat:     c = {c:.0f} J/(kg·°C)")
print(f"  Initial temp:      T_i = {T_initial:.1f} °C\n")

print("Results (3 significant figures):")
print(f"  Stored energy:    ΔU = {dU:.3e} J")
print(f"  Temp increase:    ΔT = {dT:.3g} °C")
print(f"  Final temp:       T_f = {T_final:.3g} °C\n")

print("Physical interpretation:")
print("• Even a single high-voltage capacitor can store tens of kilojoules.")
print("• If that energy is deposited locally, the temperature rise can be large enough to cause severe burns.")
print("• High voltage also creates additional hazards: arcing, high current, and dangerous current paths through the body.")
```

## Problem D3.8 — Capacitor Discharge Welding (CDW)

In **capacitor discharge welding (CDW)**, a capacitor bank is charged and then discharged rapidly through a small metal region at the weld interface. The goal is to deliver energy quickly enough to melt a thin layer locally without heating the entire part.

A CDW unit uses a capacitor bank with

$$
C = 12.0\ \text{mF}
\qquad \text{charged to} \qquad
\Delta V = 200\ \text{V}.
$$

During a weld pulse, assume **60%** of the stored capacitor energy ends up as thermal energy in a small region of steel at the weld interface (the rest goes into the arc/plasma, electrodes, sound, light, etc.).

Model the affected steel region as:

- mass:  

  $$
  m = 2.50\ \text{g}
  $$
  
- specific heat (steel):  

  $$
  c = 450\ \mathrm{J/(kg\cdot^\circ C)}
  $$

  
- initial temperature:  

  $$
  T_i = 20^\circ\text{C}
  $$

Ignore phase change (melting) for this estimate and treat all absorbed energy as heating.

Answer the following:

1. Compute the stored energy in the capacitor bank, $\Delta U$.
2. Compute the thermal energy delivered to the steel region, $Q_{\text{steel}}$.
3. Compute the temperature increase $\Delta T$ and the predicted final temperature $T_f$.
4. Based on your result, comment on whether melting is plausible (steel melts at roughly $1500^\circ\text{C}$).

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Energy stored in a capacitor:

$$
\Delta U = \frac{1}{2}C(\Delta V)^2
$$

Thermal energy and temperature change:

$$
Q_{\text{steel}} = mc\,\Delta T
\qquad \Rightarrow \qquad
\Delta T = \frac{Q_{\text{steel}}}{mc}
$$

Energy transfer assumption:

$$
Q_{\text{steel}} = 0.60\,\Delta U.
$$

---

**Given**

- $C = 12.0~\text{mF} = 1.20\times10^{-2}\ \text{F}$
- $\Delta V = 200\ \text{V}$
- $m = 2.50~\text{g} = 2.50\times10^{-3}\ \text{kg}$
- $c = 450\ \mathrm{J/(kg\cdot^\circ C)}$
- $T_i = 20^\circ\text{C}$

---

**1) Stored capacitor energy**

$$
\Delta U = \frac{1}{2}C(\Delta V)^2
= \frac{1}{2}(1.20\times10^{-2})(200)^2.
$$

Since $200^2 = 4.00\times10^{4}$,

$$
\Delta U = 0.5(1.20\times10^{-2})(4.00\times10^{4})
= 2.40\times10^{2}\ \text{J}.
$$

$$
\boxed{\Delta U = 240\ \text{J}}
$$

---

**2) Thermal energy to the steel**

$$
Q_{\text{steel}} = 0.60\,\Delta U = 0.60(240) = 144\ \text{J}.
$$

$$
\boxed{Q_{\text{steel}} = 144\ \text{J}}
$$

---

**3) Temperature rise and final temperature**

$$
\Delta T = \frac{Q_{\text{steel}}}{mc}
= \frac{144}{(2.50\times10^{-3})(450)}
= \frac{144}{1.125}
= 1.28\times10^{2}\ ^\circ\text{C}.
$$

$$
\boxed{\Delta T \approx 128^\circ\text{C}}
$$

Final temperature:

$$
T_f = T_i + \Delta T = 20^\circ\text{C} + 128^\circ\text{C} = 148^\circ\text{C}.
$$

$$
\boxed{T_f \approx 148^\circ\text{C}}
$$

---

**4) Is melting plausible?**

Steel melts near $1500^\circ\text{C}$, and this simple heating-only estimate predicts a final temperature of only about $150^\circ\text{C}$.

So **melting is not plausible** under these assumptions.

**Important note:** In real CDW, the heated mass is often far smaller than a few grams, heating occurs extremely locally, and phase change (latent heat) must be included for realistic melt estimates.

## Problem D3.9 — Capacitive Touchscreen: Detecting a Finger by a Capacitance Change

A simplified model of a capacitive touchscreen pixel is a tiny capacitor. When no finger is present, the pixel has capacitance

$$
C_0 = 40.0\ \text{pF}.
$$

When a finger touches near the pixel, the electric field configuration changes and the measured capacitance increases to

$$
C_1 = 55.0\ \text{pF}.
$$

Assume the touchscreen electronics operate the pixel at a constant potential difference

$$
\Delta V = 3.30\ \text{V}.
$$

Answer the following:

1. Compute the stored charge on the pixel before touch, $Q_0$.
2. Compute the stored charge on the pixel during touch, $Q_1$.
3. Compute the change in stored charge, $\Delta Q = Q_1 - Q_0$.
4. Compute the stored energy before touch, $\Delta U_0$, and during touch, $\Delta U_1$.
5. In one or two sentences, explain why a touchscreen can detect a finger by measuring changes in charge (or an RC timing change) even though the capacitance values are tiny.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Charge at fixed voltage:

$$
Q = C\,\Delta V
$$

Energy at fixed voltage:

$$
\Delta U = \frac{1}{2}C(\Delta V)^2
$$

---

**Given**

- $C_0 = 40.0~\text{pF} = 40.0\times10^{-12}~\text{F}$
- $C_1 = 55.0~\text{pF} = 55.0\times10^{-12}~\text{F}$
- $\Delta V = 3.30~\text{V}$

---

**1) Charge before touch**

$$
Q_0 = C_0\Delta V
= (40.0\times10^{-12})(3.30)
= 1.32\times10^{-10}~\text{C}.
$$

$$
\boxed{Q_0 = 1.32\times10^{-10}\ \text{C}}
$$

---

**2) Charge during touch**

$$
Q_1 = C_1\Delta V
= (55.0\times10^{-12})(3.30)
= 1.82\times10^{-10}~\text{C}.
$$

$$
\boxed{Q_1 = 1.82\times10^{-10}\ \text{C}}
$$

---

**3) Change in charge**

$$
\Delta Q = Q_1 - Q_0
= (1.82 - 1.32)\times10^{-10}
= 5.0\times10^{-11}~\text{C}.
$$

$$
\boxed{\Delta Q = 5.0\times10^{-11}\ \text{C}}
$$

---

**4) Energy before and during touch**

Before touch:

$$
\Delta U_0 = \frac{1}{2}C_0(\Delta V)^2
= \frac{1}{2}(40.0\times10^{-12})(3.30)^2
= 2.18\times10^{-10}~\text{J}.
$$

During touch:

$$
\Delta U_1 = \frac{1}{2}C_1(\Delta V)^2
= \frac{1}{2}(55.0\times10^{-12})(3.30)^2
= 2.99\times10^{-10}~\text{J}.
$$

So,

$$
\boxed{\Delta U_0 = 2.18\times10^{-10}\ \text{J}}, 
\qquad
\boxed{\Delta U_1 = 2.99\times10^{-10}\ \text{J}}.
$$

---

**5) Physical interpretation**

Even though the charges and energies are tiny, the electronics can measure very small changes in charge or detect the change indirectly through a timing/oscillator circuit. A finger modifies the electric field geometry (and effective permittivity), changing $C$, which changes $Q = C\Delta V$ at a fixed voltage.

</div>
</details>

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Given values
# -----------------------------
C0 = 40.0e-12      # F
C1 = 55.0e-12      # F
dV = 3.30          # V

# -----------------------------
# Physics:
# At fixed voltage: Q = C ΔV
# Stored energy: ΔU = (1/2) C (ΔV)^2
# Changes: ΔQ = Q1 - Q0, ΔU = U1 - U0
# -----------------------------
Q0 = C0 * dV
Q1 = C1 * dV
dQ = Q1 - Q0

U0 = 0.5 * C0 * dV**2
U1 = 0.5 * C1 * dV**2
dU = U1 - U0

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Capacitive touchscreen pixel: before touch vs touch\n")
print("Given:")
print(f"  C0 = {C0:.3e} F")
print(f"  C1 = {C1:.3e} F")
print(f"  ΔV = {dV:.2f} V\n")

print("Results (3 significant figures):")
print(f"  Q0 = {Q0:.3e} C")
print(f"  Q1 = {Q1:.3e} C")
print(f"  ΔQ = {dQ:.3e} C\n")

print(f"  ΔU0 = {U0:.3e} J")
print(f"  ΔU1 = {U1:.3e} J")
print(f"  ΔU  = {dU:.3e} J\n")

print("Physical interpretation:")
print("• A finger changes the electric field geometry and effective permittivity, increasing the pixel capacitance.")
print("• With ΔV held constant by the electronics, a larger C means more stored charge (Q = CΔV).")
print("• Touchscreens detect this change directly or indirectly using timing/oscillator circuits.")
```
