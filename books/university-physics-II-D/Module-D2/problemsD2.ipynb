## Problem D2.1 - Work Done by a Uniform Electric Field

Between the plates of a **parallel-plate capacitor** there exists a **uniform electric field** with magnitude

$$
E = 5.00\times10^{9}\ \text{N/C},
$$

directed from the **positive plate** to the **negative plate**.  
The spacing (gap) between the plates is

$$
d = 2.00\ \text{mm}.
$$

Answer the following:

1. How much **work** is done by the electric force in moving a **proton** from the positive plate to the negative plate?  
2. How much **work** is done by the electric force in moving an **electron** from the positive plate to the negative plate?

Be careful with **signs**, **units**, and **physical interpretation**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

For motion parallel to a uniform electric field, the work done by the electric force is

$$
W = \vec{F}\cdot\vec{d} = q E d.
$$

---

**Given**

- Electric field:  

  $$E = 5.00\times10^{9}\ \text{N/C}$$

- Plate separation:  

  $$d = 2.00\times10^{-3}\ \text{m}$$

- Elementary charge magnitude:  

  $$e = 1.60\times10^{-19}\ \text{C}$$

---

### Part 1 — Proton

For a proton,

$$
q = +e.
$$

**Work done**

$$
W_p = qEd
= (1.60\times10^{-19})(5.00\times10^{9})(2.00\times10^{-3}).
$$

$$
W_p = 1.60\times10^{-12}\ \text{J}.
$$

**Result**

$$
\boxed{W_p = +1.60\times10^{-12}\ \text{J}}
$$

The work is **positive** because the proton moves **in the direction of the electric force**.

---

### Part 2 — Electron

For an electron,

$$
q = -e.
$$

**Work done**

$$
W_e = qEd
= (-1.60\times10^{-19})(5.00\times10^{9})(2.00\times10^{-3}).
$$

$$
W_e = -1.60\times10^{-12}\ \text{J}.
$$

**Result**

$$
\boxed{W_e = -1.60\times10^{-12}\ \text{J}}
$$

The work is **negative** because the electric force on the electron points **opposite** the direction of motion.

---

**Final Interpretation**

- The **same electric field** does equal-magnitude work over the same distance.
- The **sign of the work** depends on the **sign of the charge**.
- Positive work means the field **adds kinetic energy** to the particle.
- Negative work means the field **removes kinetic energy** from the particle.

This example highlights why **electric potential energy depends on charge sign**, while the **electric potential does not**.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Given values
# -----------------------------
E = 5.00e9        # N/C
d = 2.00e-3       # m
e = 1.60e-19      # C

q_proton = +e
q_electron = -e

# -----------------------------
# Physics: work by uniform E-field
# W = q E d   (motion parallel to E)
# -----------------------------
W_proton = q_proton * E * d
W_electron = q_electron * E * d

# -----------------------------
# Nicely formatted output
# -----------------------------
print("Work done by a uniform electric field (motion along the field)\n")
print("Given:")
print(f"  Electric field:  E = {E:.2e} N/C")
print(f"  Plate spacing:   d = {d:.2e} m")
print(f"  Charge magnitude: e = {e:.2e} C\n")

print("Results (3 significant figures):")
print(f"  Proton   (q = +e):  W = {W_proton:.2e} J")
print(f"  Electron (q = –e):  W = {W_electron:.2e} J\n")

print("Physical interpretation:")
print("• The magnitude of the work is the same for both particles because |q|, E, and d are the same.")
print("• The sign of the work depends on the sign of the charge.")
print("• The proton moves in the direction of the electric force, so the field does positive work.")
print("• The electron experiences a force opposite its direction of motion, so the work is negative.")
print("• This illustrates why electric potential energy depends on charge sign, while electric potential does not.")
```

## Problem D2.2 - Electric Potential Energy in a Uniform Electric Field

Between the plates of a **parallel-plate capacitor** there exists a **uniform electric field** directed from the **positive plate** toward the **negative plate**.

The magnitude of the electric field is

$$
E = 5.0\times10^{9}\ \text{N/C},
$$

and the separation between the plates is

$$
d = 2.0\ \text{mm}.
$$

A charged particle is moved from the **positive plate** to the **negative plate**.

Answer the following:

1. Does the **electric potential energy** of a **proton** increase or decrease? Explain your reasoning.  
2. By how much does the electric potential energy of the proton change?  
3. Repeat parts (1) and (2) for an **electron**.

Be careful with **signs**, **units**, and **physical interpretation**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant relationships**

For motion parallel to a uniform electric field,

$$
\Delta V = -Ed,
$$

and the change in electric potential energy is

$$
\Delta U = q\,\Delta V.
$$

---

**Given**

- Electric field:  

  $$
  E = 5.0\times10^{9}\ \text{N/C}
  $$

- Plate separation:  

  $$
  d = 2.0\times10^{-3}\ \text{m}
  $$

- Elementary charge magnitude:  

  $$
  e = 1.60\times10^{-19}\ \text{C}
  $$

---

**Part 1 — Proton: qualitative behavior**

A proton has charge

$$
q = +e.
$$

The electric field points from the positive plate to the negative plate, so the **electric potential decreases** in the direction of motion.

Because

$$
\Delta U = q\,\Delta V,
$$

and both $q$ and $\Delta V$ are positive and negative respectively, the change in potential energy is **negative**.

**Conclusion:**  
The electric potential energy of the proton **decreases**.

---

**Part 2 — Proton: quantitative change**

First compute the potential difference:

$$
\Delta V = -Ed = -(5.0\times10^{9})(2.0\times10^{-3})
= -1.0\times10^{7}\ \text{V}.
$$

Now compute the change in potential energy:

$$
\Delta U_p = q\,\Delta V
= (1.60\times10^{-19})(-1.0\times10^{7})
= -1.6\times10^{-12}\ \text{J}.
$$

**Result**

$$
\boxed{\Delta U_p = -1.6\times10^{-12}\ \text{J}}
$$

---

**Part 3 — Electron: qualitative and quantitative behavior**

An electron has charge

$$
q = -e.
$$

The potential difference $\Delta V$ is the same as before, but the charge has the **opposite sign**.

**Qualitative conclusion:**  
The electric potential energy of the electron **increases**, even though it moves in the same direction as the proton.

Now compute the change:

$$
\Delta U_e = q\,\Delta V
= (-1.60\times10^{-19})(-1.0\times10^{7})
= +1.6\times10^{-12}\ \text{J}.
$$

**Result**

$$
\boxed{\Delta U_e = +1.6\times10^{-12}\ \text{J}}
$$

---

**Final interpretation**

- The **electric potential change** depends only on the field and geometry.
- The **electric potential energy change** depends on the **sign of the charge**.
- Positive and negative charges behave oppositely in the same electric potential.

This example highlights why it is essential to distinguish clearly between **electric potential** and **electric potential energy**.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

# -----------------------------
# Given values
# -----------------------------
E = 5.0e9          # N/C
d = 2.0e-3         # m
e = 1.60e-19       # C

q_proton = +e
q_electron = -e

# -----------------------------
# Physics
# -----------------------------
# Potential difference for motion along the electric field
# ΔV = -E d
dV = -E * d

# Change in potential energy
dU_proton = q_proton * dV
dU_electron = q_electron * dV

# -----------------------------
# Output
# -----------------------------
print("Electric potential energy change in a uniform electric field\n")

print("Given:")
print(f"  Electric field:      E = {E:.2e} N/C")
print(f"  Plate separation:    d = {d:.2e} m")
print(f"  Potential difference ΔV = {dV:.2e} V\n")

print("Results:")
print(f"  Proton:")
print(f"    Charge q = +e")
print(f"    ΔU = {dU_proton:.2e} J  (potential energy decreases)\n")

print(f"  Electron:")
print(f"    Charge q = −e")
print(f"    ΔU = {dU_electron:.2e} J  (potential energy increases)\n")

print("Physical interpretation:")
print("• The electric potential change ΔV depends only on the field and geometry.")
print("• The potential energy change ΔU = qΔV depends on the sign of the charge.")
print("• Positive and negative charges respond oppositely in the same electric field.")
print("• This is why electric potential is charge-independent, but potential energy is not.")
```

## Problem D2.3 - Work Done by a Non-Uniform Electric Field (1D)

An electric field along the $x$-axis is given by

$$
E_x(x) = (1.5x \times 10^{9})\ \text{N/C},
$$

where $x$ is measured in meters.

An **electron** is moved along the $x$-axis from

$$
x_1 = 2.5\times10^{-3}\ \text{m}
\quad \text{to} \quad
x_2 = 4.5\times10^{-3}\ \text{m}.
$$

Calculate the **work done by the electric force** on the electron.

Be careful with the **position dependence** of the electric field and the **sign of the charge**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

The work done by the electric force in one dimension is

$$
W = \int \vec{F}\cdot d\vec{x}
= \int qE_x(x)\,dx.
$$

---

**Given**

- Electric field:  

  $$
  E_x(x) = 1.5\times10^{9}x\ \text{N/C}
  $$

- Charge of an electron:  

  $$
  q = -1.60\times10^{-19}\ \text{C}
  $$

- Initial position:  

  $$
  x_1 = 2.5\times10^{-3}\ \text{m}
  $$

- Final position:  

  $$
  x_2 = 4.5\times10^{-3}\ \text{m}
  $$

---

**Set up the integral**

$$
W = \int_{x_1}^{x_2} qE_x(x)\,dx
= \int_{x_1}^{x_2} (-1.60\times10^{-19})(1.5\times10^{9}x)\,dx.
$$

---

**Evaluate the integral**

$$
W = -1.60\times10^{-19}\,(1.5\times10^{9})
\int_{x_1}^{x_2} x\,dx
$$

$$
W = -2.40\times10^{-10}
\left[\frac{x^2}{2}\right]_{x_1}^{x_2}.
$$

Substitute the limits:

$$
W = -2.40\times10^{-10}
\left[
\frac{(4.5\times10^{-3})^2 - (2.5\times10^{-3})^2}{2}
\right].
$$

$$
W = -1.68\times10^{-15}\ \text{J}.
$$

---

**Result**

$$
\boxed{W = -1.68\times10^{-15}\ \text{J}}
$$

---

**Physical interpretation**

- The **negative sign** indicates that the electric force does **negative work** on the electron.
- The electron’s force is opposite the direction of the electric field.
- This result reflects the charge dependence of **electric potential energy**, even though the electric field itself is independent of the test charge.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols and given functions
# -----------------------------
x = sp.Symbol('x', real=True)

E = 1.5e9 * x                  # E_x(x) in N/C (x in meters)
q = -1.60e-19                  # electron charge in C
x1 = 2.5e-3                    # m
x2 = 4.5e-3                    # m

# -----------------------------
# Work integral: W = ∫ q E_x(x) dx
# -----------------------------
display(Math(r"W = \int_{x_1}^{x_2} q\,E_x(x)\,dx"))
display(Math(rf"E_x(x) = {sp.latex(1.5e9*x)}\ \text{{N/C}}"))
display(Math(rf"q = {sp.latex(q)}\ \text{{C}},\quad x_1 = {sp.latex(x1)}\ \text{{m}},\quad x_2 = {sp.latex(x2)}\ \text{{m}}"))

integrand = q * E
display(Math(rf"W = \int_{{{sp.latex(x1)}}}^{{{sp.latex(x2)}}} \left({sp.latex(integrand)}\right)\,dx"))

W_expr = sp.integrate(integrand, (x, x1, x2))
display(Math(rf"W = {sp.latex(W_expr)}\ \text{{J}}"))

# -----------------------------
# Numeric evaluation
# -----------------------------
W_val = float(sp.N(W_expr, 10))

print(f"Work done by the electric force on the electron: W = {W_val:.1e} J")
```

## Problem D2.4 - Work and Potential Energy in a Proton–Electron System

Consider a **proton** and an **electron** separated by an initial distance

$$
r_i = 1.0\times10^{-3}\ \text{m}.
$$

The separation distance between the two particles is changed to

$$
r_f = 1.0\times10^{-6}\ \text{m}.
$$

Answer the following:

1. How much **work is done by the electric force** during this change in separation?  
2. Does the **electric potential energy** of the proton–electron system increase or decrease, and by how much? Explain your reasoning.

Be careful with **signs**, **units**, and **physical interpretation**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant relationships**

For two point charges interacting electrostatically, the electric potential energy is

$$
U(r) = k\,\frac{q_1 q_2}{r},
$$

where

$$
k = 8.99\times10^{9}\ \text{N·m}^2/\text{C}^2.
$$

The work done by the electric force is related to the change in potential energy by

$$
W = -\Delta U.
$$

---

**Given**

- Proton charge:  

  $$
  q_p = +e = +1.60\times10^{-19}\ \text{C}
  $$

- Electron charge:  

  $$
  q_e = -e = -1.60\times10^{-19}\ \text{C}
  $$

- Initial separation:  

  $$
  r_i = 1.0\times10^{-3}\ \text{m}
  $$

- Final separation:  

  $$
  r_f = 1.0\times10^{-6}\ \text{m}
  $$

---

**Part 1 — Change in electric potential energy**

The product of the charges is

$$
q_p q_e = -(1.60\times10^{-19})^2 = -2.56\times10^{-38}\ \text{C}^2.
$$

Compute the initial and final potential energies:

$$
U_i = k\,\frac{q_p q_e}{r_i},
\qquad
U_f = k\,\frac{q_p q_e}{r_f}.
$$

Substitute values:

$$
U_i = (8.99\times10^{9})\frac{-2.56\times10^{-38}}{1.0\times10^{-3}}
= -2.30\times10^{-25}\ \text{J},
$$

$$
U_f = (8.99\times10^{9})\frac{-2.56\times10^{-38}}{1.0\times10^{-6}}
= -2.30\times10^{-22}\ \text{J}.
$$

The change in potential energy is

$$
\Delta U = U_f - U_i
= -2.30\times10^{-22} - (-2.30\times10^{-25})
\approx -2.30\times10^{-22}\ \text{J}.
$$

---

**Part 2 — Work done by the electric force**

The work done by the electric force is

$$
W = -\Delta U.
$$

Therefore,

$$
W = +2.30\times10^{-22}\ \text{J}.
$$

**Result**

$$
\boxed{W = +2.30\times10^{-22}\ \text{J}}
$$

---

**Physical interpretation**

- The proton and electron attract each other.
- As the separation decreases, the electric potential energy becomes **more negative**.
- Therefore, the **potential energy decreases** by approximately

  $$
  \boxed{2.30\times10^{-22}\ \text{J}}.
  $$

- The electric force does **positive work**, converting electric potential energy into kinetic energy of the particles.

This behavior is characteristic of **attractive interactions** between opposite charges.

</div>
</details>

```python
# DIY Cell
```

```python
%reset -f

# -----------------------------
# Given constants and values
# -----------------------------
k = 8.99e9            # N·m^2/C^2
e = 1.60e-19          # C

q_p = +e              # proton charge
q_e = -e              # electron charge

r_i = 1.0e-3          # m
r_f = 1.0e-6          # m

# -----------------------------
# Electric potential energy function
# U(r) = k q1 q2 / r
# -----------------------------
U_i = k * q_p * q_e / r_i
U_f = k * q_p * q_e / r_f

# Change in potential energy
delta_U = U_f - U_i

# Work done by the electric force
W = -delta_U

# -----------------------------
# Output
# -----------------------------
print("Proton–Electron System: Work and Potential Energy\n")

print("Given:")
print(f"  Initial separation r_i = {r_i:.2e} m")
print(f"  Final separation   r_f = {r_f:.2e} m")
print(f"  Charges: q_p = +e, q_e = −e\n")

print("Electric potential energy:")
print(f"  Initial U_i = {U_i:.2e} J")
print(f"  Final   U_f = {U_f:.2e} J\n")

print("Results:")
print(f"  Change in potential energy ΔU = {delta_U:.2e} J")
print(f"  Work done by electric force W = {W:.2e} J\n")

print("Physical interpretation:")
print("• Proton and electron attract each other.")
print("• As separation decreases, potential energy becomes more negative.")
print("• The potential energy of the system decreases.")
print("• The electric force does positive work, converting potential energy into kinetic energy.")
```

## Problem D2.5 - Electric Field from a Potential Difference

Between the plates of a **parallel-plate capacitor** there exists an electric potential difference of

$$
\Delta V = 1.0\times10^{3}\ \text{J/C}.
$$

The separation between the plates is

$$
d = 0.1\times10^{-3}\ \text{mm}.
$$

Assuming the electric field between the plates is **uniform**, answer the following:

1. What is the **magnitude of the electric field** between the plates?

Be careful with **unit conversions** and **orders of magnitude**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant relationship**

For a uniform electric field between parallel plates, the magnitude of the electric field is

$$
E = \frac{\Delta V}{d}.
$$

---

**Given**

- Potential difference:

  $$
  \Delta V = 1.0\times10^{3}\ \text{V}
  $$

- Plate separation:

  $$
  d = 0.1\times10^{-3}\ \text{mm}
  = 1.0\times10^{-7}\ \text{m}
  $$

---

**Compute the electric field**

$$
E = \frac{1.0\times10^{3}}{1.0\times10^{-7}}
= 1.0\times10^{10}\ \text{V/m}.
$$

---

**Result**

$$
\boxed{E = 1.0\times10^{10}\ \text{V/m}}
$$

---

**Physical interpretation**

- A very small plate separation combined with a modest potential difference produces an extremely large electric field.
- This magnitude is comparable to fields near electrical breakdown in air.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

# -----------------------------
# Given values
# -----------------------------
delta_V = 1.0e3          # V (since 1 J/C = 1 V)
d_mm = 0.1e-3            # mm

# Convert plate separation to meters
d = d_mm * 1e-3          # mm → m

# -----------------------------
# Physics
# -----------------------------
# Uniform electric field between parallel plates
# E = ΔV / d
E = delta_V / d

# -----------------------------
# Output
# -----------------------------
print("Electric Field Between Parallel Plates\n")

print("Given:")
print(f"  Potential difference ΔV = {delta_V:.2e} V")
print(f"  Plate separation d = {d:.2e} m\n")

print("Result:")
print(f"  Electric field magnitude E = {E:.2e} V/m\n")

print("Physical interpretation:")
print("• For a uniform field, the electric field equals the potential difference divided by the plate separation.")
print("• A very small separation leads to a very large electric field.")
print("• Such fields approach the scale where electrical breakdown can occur.")
```

## Problem D2.6 - Electric Potential Change in a Uniform Electric Field

Between the plates of a **parallel-plate capacitor** there exists a **uniform electric field** directed from the **positive plate** toward the **negative plate**.

The magnitude of the electric field is

$$
E = 5.0\times10^{9}\ \text{N/C},
$$

and the separation between the plates is

$$
d = 2.0\ \text{mm}.
$$

Answer the following:

1. For a **proton** moving from the positive plate to the negative plate, what is the **change in electric potential**?  
2. For an **electron** moving from the positive plate to the negative plate, what is the **change in electric potential**?

Explain your reasoning.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Key idea**

Electric potential is a property of the **electric field and geometry**, not of the particle moving in the field.

For motion parallel to a uniform electric field,

$$
\Delta V = -E\,d.
$$

---

**Given**

- Electric field:

  $$
  E = 5.0\times10^{9}\ \text{N/C}
  $$

- Plate separation:

  $$
  d = 2.0\times10^{-3}\ \text{m}
  $$

---

**Compute the potential change**

$$
\Delta V = -(5.0\times10^{9})(2.0\times10^{-3})
= -1.0\times10^{7}\ \text{V}.
$$

---

**Answers**

**1. Proton**

The change in electric potential experienced by the proton is

$$
\boxed{\Delta V = -1.0\times10^{7}\ \text{V}}.
$$

**2. Electron**

The change in electric potential experienced by the electron is

$$
\boxed{\Delta V = -1.0\times10^{7}\ \text{V}}.
$$

---

**Physical interpretation**

- The electric potential **decreases** in the direction of the electric field.
- The change in electric potential depends only on the **field and distance**, not on the charge.
- Although the proton and electron experience the **same change in potential**, their **potential energy changes differ** because their charges have opposite signs.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

# -----------------------------
# Given values
# -----------------------------
E = 5.0e9          # N/C
d = 2.0e-3         # m

# -----------------------------
# Physics
# -----------------------------
# For motion parallel to a uniform electric field:
# ΔV = -E d
delta_V = -E * d

# -----------------------------
# Output
# -----------------------------
print("Electric Potential Change in a Uniform Electric Field\n")

print("Given:")
print(f"  Electric field magnitude E = {E:.2e} N/C")
print(f"  Plate separation d = {d:.2e} m\n")

print("Calculation:")
print("  ΔV = -E d\n")

print("Results:")
print(f"  Change in electric potential for a proton   ΔV = {delta_V:.2e} V")
print(f"  Change in electric potential for an electron ΔV = {delta_V:.2e} V\n")

print("Physical interpretation:")
print("• Electric potential is a property of the field and geometry, not the particle.")
print("• Both proton and electron experience the same potential change.")
print("• The potential decreases when moving in the direction of the electric field.")
print("• The difference between proton and electron appears only in ΔU = qΔV, not in ΔV itself.")
```

## Problem D2.7 - Electric Potential Energy and Motion of a Point Charge

A point charge

$$
q_1 = +4.0~\mu\text{C}
$$

is fixed in place. A second charge

$$
q_2 = +2.0~\mu\text{C}
$$

with mass

$$
m = 5.0~\mu\text{g}
$$

is free to move in the surrounding space.

Answer the following:

1. Evaluate the **electric potential energy** of $q_2$ when it is located **3.0 cm** from $q_1$.  
2. If $q_2$ starts from rest at **3.0 cm** from $q_1$, what will be its **speed** when it reaches a distance of **6.0 cm** from $q_1$?  
   Assume that $q_1$ remains fixed and that only electric forces act.

Be careful with **units**, **signs**, and **energy conservation**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant relationships**

Electric potential energy of two point charges separated by distance $r$:

$$
U(r) = k\,\frac{q_1 q_2}{r}.
$$

If $q_2$ starts from rest and only the electric force acts, mechanical energy is conserved:

$$
U_i + K_i = U_f + K_f,
\qquad K_i = 0.
$$

So,

$$
K_f = U_i - U_f.
$$

Relate kinetic energy to speed:

$$
K_f = \frac{1}{2}mv^2
\quad \Rightarrow \quad
v = \sqrt{\frac{2(U_i - U_f)}{m}}.
$$

---

**Given**

$$
q_1 = +4.0\times10^{-6}\ \text{C},\quad
q_2 = +2.0\times10^{-6}\ \text{C},\quad
m = 5.0\times10^{-9}\ \text{kg}
$$

$$
r_i = 3.0\times10^{-2}\ \text{m},\quad
r_f = 6.0\times10^{-2}\ \text{m},\quad
k = 8.99\times10^{9}\ \text{N·m}^2/\text{C}^2
$$

---

**Compute $U_i$ and $U_f$**

$$
U_i = k\,\frac{q_1 q_2}{r_i},
\qquad
U_f = k\,\frac{q_1 q_2}{r_f}.
$$

---

**Compute the speed**

$$
v = \sqrt{\frac{2\left(k\frac{q_1 q_2}{r_i} - k\frac{q_1 q_2}{r_f}\right)}{m}}.
$$

Because both charges are positive, the interaction is repulsive, so $U$ decreases as $r$ increases and the difference $U_i-U_f$ becomes kinetic energy.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
k, q1, q2, m, r, r_i, r_f = sp.symbols('k q1 q2 m r r_i r_f', positive=True)

# -----------------------------
# Physics (symbolic)
# -----------------------------
# Electric potential energy
U = k * q1 * q2 / r

U_i = U.subs(r, r_i)
U_f = U.subs(r, r_f)

# Energy conservation (starts from rest)
K_f = U_i - U_f

# Speed from kinetic energy
v = sp.sqrt(2 * K_f / m)

# -----------------------------
# Display equations (symbolic)
# -----------------------------
display(Math(r"U(r) = k\,\frac{q_1 q_2}{r}"))

display(Math(
    r"U_i = k\,\frac{q_1 q_2}{r_i}, \qquad "
    r"U_f = k\,\frac{q_1 q_2}{r_f}"
))

display(Math(
    r"\text{Energy conservation: }"
    r"U_i = U_f + K_f"
))

display(Math(
    r"K_f = U_i - U_f"
))

display(Math(
    r"\frac{1}{2}mv^2 = K_f"
))

display(Math(
    r"v = \sqrt{\frac{2(U_i - U_f)}{m}}"
))

# -----------------------------
# Substitute numerical values
# -----------------------------
values = {
    k: 8.99e9,
    q1: 4.0e-6,
    q2: 2.0e-6,
    m: 5.0e-9,
    r_i: 3.0e-2,
    r_f: 6.0e-2
}

U_i_val = float(sp.N(U_i.subs(values)))
U_f_val = float(sp.N(U_f.subs(values)))
K_f_val = float(sp.N(K_f.subs(values)))
v_val   = float(sp.N(v.subs(values)))

# -----------------------------
# Numeric results (plain text)
# -----------------------------
print("\nNumeric results:")
print(f"  Initial potential energy U_i = {U_i_val:.3e} J")
print(f"  Final potential energy   U_f = {U_f_val:.3e} J")
print(f"  Kinetic energy gained    K_f = {K_f_val:.3e} J")
print(f"  Speed of q2              v   = {v_val:.2e} m/s")

print("\nInterpretation:")
print("• Both charges are positive → repulsive interaction.")
print("• Increasing separation lowers electric potential energy.")
print("• The loss in potential energy appears as kinetic energy.")
print("• Energy methods avoid computing forces or acceleration.")
```

## Problem D2.8 - Electron Acceleration in a Vacuum Tube

In a classic **vacuum tube**, electrons are emitted from a heated **cathode** and accelerated toward a positively charged **anode** by an electric potential difference.

Consider a vacuum tube in which electrons start from rest at the cathode and are accelerated through a potential difference of

$$
\Delta V = 2.50\times10^{3}\ \text{V}.
$$

Assume the region between the cathode and anode is evacuated (no collisions), and ignore relativistic effects.

Answer the following:

1. How much **electric potential energy** does an electron lose as it moves from the cathode to the anode?
2. What is the **kinetic energy** of the electron when it reaches the anode?
3. What is the **speed** of the electron upon reaching the anode?
4. Briefly explain why **electric potential**, rather than electric force, is the most convenient quantity to use in this situation.

Be careful with **signs**, **units**, and **physical interpretation**.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Key relationships**

The change in electric potential energy is related to the potential difference by

$$
\Delta U = q\,\Delta V.
$$

If the electron starts from rest and only electric forces act, energy conservation gives

$$
\Delta K = -\Delta U.
$$

Kinetic energy is related to speed by

$$
K = \frac{1}{2}mv^2.
$$

---

**Given**

- Electron charge:

  $$
  q = -1.60\times10^{-19}\ \text{C}
  $$

- Electron mass:

  $$
  m = 9.11\times10^{-31}\ \text{kg}
  $$

- Potential difference:

  $$
  \Delta V = +2.50\times10^{3}\ \text{V}
  $$

---

**1. Change in electric potential energy**

$$
\Delta U = q\,\Delta V
= (-1.60\times10^{-19})(2.50\times10^{3})
= -4.00\times10^{-16}\ \text{J}.
$$

**Interpretation:**  
The electric potential energy of the electron **decreases**.

---

**2. Kinetic energy at the anode**

Since the initial kinetic energy is zero, 

$$
K_f = -\Delta U = 4.00\times10^{-16}\ \text{J}.
$$

---

**3. Speed of the electron**

$$
\frac{1}{2}mv^2 = 4.00\times10^{-16}
$$

$$
v = \sqrt{\frac{2(4.00\times10^{-16})}{9.11\times10^{-31}}}
= 2.97\times10^{7}\ \text{m/s}.
$$

---

**4. Why electric potential is useful**

- The electron’s energy gain depends only on the **potential difference**, not on the detailed electric field structure.
- No force, acceleration, or time calculation is required.
- This makes electric potential the natural tool for analyzing vacuum tubes, electron guns, and particle accelerators.

</div>
</details>


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
q, dV, m = sp.symbols('q dV m', real=True)
dU, K, v = sp.symbols('dU K v', real=True)

# -----------------------------
# Equations (symbolic)
# -----------------------------
# ΔU = q ΔV
dU_expr = sp.Eq(dU, q*dV)

# If starting from rest: K = -ΔU
K_expr = sp.Eq(K, -dU)

# K = (1/2) m v^2
Kv_expr = sp.Eq(K, sp.Rational(1, 2)*m*v**2)

# Solve for v in terms of ΔU (or ΔV)
v_expr = sp.solve(Kv_expr.subs(K, -q*dV), v)[1]  # positive root

# -----------------------------
# Display equations (LaTeX)
# -----------------------------
display(Math(sp.latex(dU_expr)))
display(Math(r"\text{(starts from rest)}\quad \Delta K = -\Delta U \;\Rightarrow\; K = -\Delta U"))
display(Math(sp.latex(Kv_expr)))
display(Math(r"v = " + sp.latex(v_expr)))

# -----------------------------
# Numerical values
# -----------------------------
vals = {
    q: -1.60e-19,     # C (electron)
    dV: 2.50e3,       # V
    m: 9.11e-31       # kg
}

dU_val = float(sp.N((q*dV).subs(vals)))
K_val  = float(sp.N((-q*dV).subs(vals)))
v_val  = float(sp.N(v_expr.subs(vals)))

# -----------------------------
# Plain numeric answers (nice output)
# -----------------------------
print("\nNumeric results:")
print(f"  ΔU = qΔV = {dU_val:.2e} J")
print(f"  K  = -ΔU = {K_val:.2e} J")
print(f"  v  = {v_val:.2e} m/s")

print("\nInterpretation:")
print("• The electron has negative charge, so ΔU = qΔV is negative for a positive accelerating voltage.")
print("• The lost electric potential energy becomes kinetic energy.")
print("• Using ΔV gives the energy change directly, without computing force or acceleration.")
```

## Problem D2.9 Thundercloud–Ground as a Giant Capacitor

A simplified model of a thunderstorm treats the **base of a charged thundercloud** and the **ground** as two large conducting “plates” with air in between.

Assume a storm in which the potential difference between the cloud base and the ground is

\Delta V = 1.0\times10^{8}\ \text{V},
$$

and the cloud base is approximately

$$
d = 1.0\times10^{3}\ \text{m}
$$

above the ground.

Answer the following:

1. Estimate the magnitude of the electric field between the cloud and ground (assume it is uniform).  
2. Compare your result to the approximate ionization (breakdown) field of air,

   $$
   E_{\text{break}} \approx 3.0\times10^{6}\ \text{V/m}.
   $$

3. Based on your comparison, explain why lightning often requires **local field enhancement** (sharp tips, raindrops, dust, leaders) rather than a perfectly uniform field.

Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant relationship**

For a uniform field between “plates,” the field magnitude is

E = \frac{|\Delta V|}{d}.
$$

---

**1. Electric field magnitude**

Convert directly using the given values:

$$
E = \frac{1.0\times10^{8}}{1.0\times10^{3}}

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
dV, d, E = sp.symbols('dV d E', positive=True)

# -----------------------------
# Equation (uniform field magnitude)
# -----------------------------
E_expr = sp.Eq(E, dV/d)
display(Math(sp.latex(E_expr)))

# -----------------------------
# Given values
# -----------------------------
dV_val = 1.0e8          # V
d_val  = 1.0e3          # m
E_break = 3.0e6         # V/m  (approx breakdown field of air)

# -----------------------------
# Compute
# -----------------------------
E_val = dV_val / d_val
ratio = E_val / E_break

# -----------------------------
# Output (plain text for final numbers)
# -----------------------------
print("\nNumeric results:")
print(f"  ΔV = {dV_val:.2e} V")
print(f"  d  = {d_val:.2e} m")
print(f"  E  = ΔV/d = {E_val:.2e} V/m")
print(f"  Breakdown field (air) E_break ≈ {E_break:.2e} V/m")
print(f"  Ratio E/E_break = {ratio:.2e}")

print("\nInterpretation:")
print("• The uniform-field estimate is far below the breakdown field of air.")
print("• Lightning initiation typically requires local field enhancement (non-uniform fields).")
print("• Sharp tips, droplets, and stepped leaders can boost the local E-field to breakdown values.")
```

Two large, parallel conducting plates are separated by a distance of

**Given**

- Plate separation:

  $$
  d = 6.00\ \text{cm} = 0.0600\ \text{m}
  $$

- Distance from zero-volt plate:

  $$
  V(x) = 180\ \text{V}
  $$

---

**1. Electric field strength**

Because the potential changes linearly,

$$
E = \frac{V(x) - 0}{x}

$$

$$

**Result**

$$
\boxed{E = 7.20\times10^{3}\ \text{V/m}}
$$

---

**2. Voltage between the plates**

The total potential difference across the plates is

$$
\Delta V_{\text{plates}} = E\,d.
$$

$$
\Delta V_{\text{plates}} = (7.20\times10^{3})(0.0600)
= 4.32\times10^{2}\ \text{V}.
$$

**Result**

$$
\boxed{\Delta V_{\text{plates}} = 432\ \text{V}}
$$

---

**Physical interpretation**

- A constant electric field produces a **linear change in electric potential** with distance.
- Measuring the potential at a single point is sufficient to determine the field everywhere between the plates.

## Problem D2.11 - Finding Electric Potential from a Position-Dependent Force

A particle moves along the **x-axis** under the influence of a one-dimensional electric force given by

$$
F_x(x) = -\alpha x^2
$$

where  
$\alpha = 4.00\times10^{-6}\ \mathrm{N/m^2}$ is a positive constant.

Assume the force is conservative and acts only along the x-direction.

The electric potential is defined such that

$$
V(0) = 0.
$$

Answer the following:

1. Using the definition of work, write an expression for the **work done by the force** as the particle moves from $x=0$ to a general position $x$.  
2. Use calculus to find the **electric potential $V(x)$** corresponding to this force.  
3. Evaluate the **potential difference** between $x = 0$ and  

   $$
   x = 0.50\ \mathrm{m}.
   $$
   
4. State whether the force is **attractive or repulsive** for $x>0$, and explain how this is reflected in the shape of the potential.

You must explicitly show the **integral** used to obtain the potential.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Key idea (1D only)**  

In one dimension, the electric potential difference is defined by

$$
\Delta V = -\int_{x_i}^{x_f} \frac{F_x(x)}{q}\,dx.
$$

Equivalently, for a conservative force, potential energy satisfies

$$
F_x(x) = -\frac{dU}{dx},
\qquad V(x)=\frac{U(x)}{q}.
$$

---

**1. Work done from $0$ to $x$**

$$
W = \int_0^x F_x(x')\,dx'
= \int_0^x (-\alpha x'^2)\,dx'.
$$

---

**2. Electric potential**

Because $V(0)=0$,

$$
V(x) = -\frac{1}{q}\int_0^x F_x(x')\,dx'
= \frac{\alpha}{q}\int_0^x x'^2\,dx'.
$$

Evaluate the integral:

$$
V(x) = \frac{\alpha}{q}\left(\frac{x^3}{3}\right).
$$

So,

$$
\boxed{V(x) = \frac{\alpha}{3q}x^3}
$$

---

**3. Potential difference to $x=0.50\ \mathrm{m}$**

$$
\Delta V = V(0.50) = \frac{\alpha}{3q}(0.50)^3.
$$

(Substitute numerical values if desired.)

---

**4. Physical interpretation**

- For $x>0$, the force $F_x=-\alpha x^2$ points in the **negative x-direction**, pulling the particle back toward the origin.
- The force is therefore **attractive** toward $x=0$.
- The potential increases with $x$, meaning the particle must climb “uphill” in potential to move away from the origin.

This is a one-dimensional example where the **shape of the force determines the shape of the potential through an integral**, illustrating why calculus is essential in electrostatics.

</div>
</details>


```python
# DIY Cell
```

```python
# Problem X — Python/SymPy solution
# Finding electric potential from a position-dependent force (1D, calculus-based)

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
x, xp = sp.symbols('x xp', real=True)
alpha, q = sp.symbols('alpha q', positive=True)

# -----------------------------
# Given force
# -----------------------------
Fx = -alpha * x**2

display(Math(r"F_x(x) = " + sp.latex(Fx)))

# -----------------------------
# Definition of potential (1D)
# V(0) = 0
# V(x) = -(1/q) ∫_0^x F(x') dx'
# -----------------------------
Vx = -(1/q) * sp.integrate(-alpha * xp**2, (xp, 0, x))
Vx = sp.simplify(Vx)

display(Math(r"V(x) = -\frac{1}{q}\int_0^x F_x(x')\,dx'"))
display(Math(r"V(x) = " + sp.latex(Vx)))

# -----------------------------
# Evaluate potential at x = 0.50 m
# -----------------------------
x_val = 0.50

alpha_val = 4.00e-6      # N/m^2
q_val = 1.60e-19         # C (proton, for example)

V_numeric = sp.N(Vx.subs({x: x_val, alpha: alpha_val, q: q_val}))

# -----------------------------
# Output (plain text)
# -----------------------------
print("\nNumeric evaluation:")
print(f"  alpha = {alpha_val:.2e} N/m^2")
print(f"  q     = {q_val:.2e} C")
print(f"  x     = {x_val:.2f} m\n")

print(f"  V({x_val:.2f} m) = {V_numeric:.3e} V")

print("\nPhysical interpretation:")
print("• The force varies with position, so calculus is required to find the potential.")
print("• The negative sign in V(x) comes from the work–potential relationship.")
print("• Because the force points toward negative x for x > 0, the potential increases with x.")
print("• The shape of V(x) is determined directly by integrating the force.")
```

## Problem D2.12 - Electric Field from a Given Potential Energy Function

A particle with charge

$$
q = +2.00\times10^{-9}\ \mathrm{C}
$$

moves along the x-axis. Its electric potential energy is given by

$$
U(x) = (3.00\times10^{-6})x^3 - (2.00\times10^{-5})x^2 + (6.00\times10^{-6})x
$$

with $U$ in joules when $x$ is in meters.

Answer the following:

1. Use calculus to find the force $F_x(x)$ from the potential energy function.  
2. Use $E_x(x)=F_x(x)/q$ to find the electric field $E_x(x)$.  
3. Evaluate $E_x$ at

   $$

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Key relationships (1D)**

$$
F_x(x) = -\frac{dU}{dx}
$$

$$
E_x(x)=\frac{F_x(x)}{q}

Start with

$$
U(x) = (3.00\times10^{-6})x^3 - (2.00\times10^{-5})x^2 + (6.00\times10^{-6})x

Differentiate term-by-term:

$$
\frac{dU}{dx}
= 3(3.00\times10^{-6})x^2 - 2(2.00\times10^{-5})x + (6.00\times10^{-6})
$$

Simplify:

$$
\frac{dU}{dx}
= (9.00\times10^{-6})x^2 - (4.00\times10^{-5})x + (6.00\times10^{-6})
$$

Now apply the negative sign:

$$
F_x(x) = -\frac{dU}{dx}
= -(9.00\times10^{-6})x^2 + (4.00\times10^{-5})x - (6.00\times10^{-6})

\boxed{F_x(x)=-(9.00\times10^{-6})x^2 + (4.00\times10^{-5})x - (6.00\times10^{-6})}

$$
E_x(x)=\frac{F_x(x)}{q}
=\frac{-(9.00\times10^{-6})x^2 + (4.00\times10^{-5})x - (6.00\times10^{-6})}{2.00\times10^{-9}}
$$

So,

$$
\boxed{
E_x(x)=\frac{-(9.00\times10^{-6})x^2 + (4.00\times10^{-5})x - (6.00\times10^{-6})}{2.00\times10^{-9}}
}
$$

---

**3. Evaluate at $x=0.50\ \mathrm{m}$**

$$
F_x(0.50)
= -(9.00\times10^{-6})(0.50)^2 + (4.00\times10^{-5})(0.50) - (6.00\times10^{-6})
$$

$$
(0.50)^2 = 0.25
$$

$$
-(9.00\times10^{-6})(0.25) = -2.25\times10^{-6}
$$

$$
(4.00\times10^{-5})(0.50) = 2.00\times10^{-5}
$$

$$
F_x(0.50) = (-2.25\times10^{-6}) + (2.00\times10^{-5}) - (6.00\times10^{-6})

F_x(0.50) = 1.175\times10^{-5}\ \mathrm{N}
$$

$$
E_x(0.50)=\frac{F_x(0.50)}{q}
=\frac{1.175\times10^{-5}}{2.00\times10^{-9}}
=5.88\times10^{3}\ \mathrm{N/C}

$$
\boxed{E_x(0.50)=5.88\times10^{3}\ \mathrm{N/C}}

**4. Direction at $x=0.50\ \mathrm{m}$**

Because $E_x(0.50)>0$, the electric field points in the **$+x$ direction**.

```python
# Problem X — Python/SymPy solution
# Electric field from a given potential energy function (1D, calculus-based)

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
x = sp.symbols('x', real=True)
q = sp.symbols('q', positive=True)

# -----------------------------
# Given potential energy function U(x)
# -----------------------------
U = (3.00e-6)*x**3 - (2.00e-5)*x**2 + (6.00e-6)*x

display(Math(r"U(x) = " + sp.latex(U)))

# -----------------------------
# Force from potential energy
# F_x = - dU/dx
# -----------------------------
Fx = -sp.diff(U, x)
Fx = sp.simplify(Fx)

display(Math(r"F_x(x) = -\frac{dU}{dx} = " + sp.latex(Fx)))

# -----------------------------
# Electric field from force
# E_x = F_x / q
# -----------------------------
Ex = sp.simplify(Fx / q)

display(Math(r"E_x(x) = \frac{F_x(x)}{q} = " + sp.latex(Ex)))

# -----------------------------
# Numerical evaluation at x = 0.50 m
# -----------------------------
vals = {
    x: 0.50,
    q: 2.00e-9     # C
}

Fx_val = float(sp.N(Fx.subs(vals)))
Ex_val = float(sp.N(Ex.subs(vals)))

# -----------------------------
# Output (plain text for final answers)
# -----------------------------
print("\nNumeric evaluation at x = 0.50 m:")
print(f"  Force F_x = {Fx_val:.3e} N")
print(f"  Electric field E_x = {Ex_val:.3e} N/C")

if Ex_val > 0:
    print("\nDirection:")
    print("• Electric field points in the +x direction.")
else:
    print("\nDirection:")
    print("• Electric field points in the −x direction.")

print("\nPhysical interpretation:")
print("• The force is obtained by differentiating the potential energy.")
print("• The electric field is force per unit charge.")
print("• Calculus directly links potential energy, force, and field in 1D.")
print("• The sign of E_x determines the field direction.")
```

## D2.13 - Challenge Problem 🐍💻 Proton Deflection Between Parallel Plates

A proton enters the region between two large **parallel conducting plates** from the left side with an initial velocity that is purely horizontal.

- Bottom plate: **positive**
- Top plate: **negative**
- Plate separation:  

  $$d = 1.00\ \text{cm}$$

- Plate length (in the direction of motion):  

  $$L = 0.200\ \text{m}$$

- Potential difference (top relative to bottom):  

  $$\Delta V = 200\ \text{V}$$

The proton enters exactly halfway between the plates (so the initial vertical position is at the midline), and it enters with horizontal speed

$$
v_0 = 2.00\times10^{6}\ \text{m/s}.
$$

Assume the field between the plates is uniform and edge effects are negligible.

---

Answer the following. **You must use symbolic Python (SymPy) to derive the expressions before substituting numbers.**

1. **Gravity check (justification)**  
   Justify ignoring gravity by comparing the magnitudes of the electric and gravitational forces.
   Evaluate the ratio between the two forces for a proton and explain your conclusion in one or two sentences.

3. **Does it make it through?**  
   
   Determine whether the proton exits the plates at $x=L$, or hits the top plate first.  
   *(Hint: the top plate is at $y=+d/2$ if you choose the midline as $y=0$.)*

4. **Minimum entrance speed to exit**  
   Derive an expression for the minimum horizontal entrance speed $v_{\min}$ required for the proton to **just barely** avoid hitting the top plate and **exit** at $x=L$.  
   Then compute its numerical value.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution guide (symbolic roadmap)
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Key field relationship**

Because the plates are parallel and conducting, the field between them is uniform:

$$
E=\frac{|\Delta V|}{d}.
$$

---

**(1) Gravity check**

Compute

$$
\frac{F_E}{F_g}=\frac{qE}{mg}
=\frac{q}{mg}\left(\frac{|\Delta V|}{d}\right).
$$

If this ratio is extremely large, gravity is negligible compared with the electric interaction.

---

**(2) Exit vs. collision**

Time to traverse the plate length:

$$
t_{\text{exit}}=\frac{L}{v_0}.
$$

Vertical displacement at that time:

$$
y_{\text{exit}}=\frac{1}{2}a t_{\text{exit}}^2
=\frac{1}{2}\left(\frac{qE}{m}\right)\left(\frac{L}{v_0}\right)^2
=\frac{1}{2}\left(\frac{q}{m}\frac{|\Delta V|}{d}\right)\left(\frac{L}{v_0}\right)^2.
$$

Compare $y_{\text{exit}}$ to $d/2$:
- If $y_{\text{exit}} \ge d/2$, the proton hits the **top** plate before exiting.
- If $y_{\text{exit}} < d/2$, it exits successfully.

---

**(3) Minimum speed to just exit**

Set the boundary condition “just grazes the top plate at the exit”:

$$
\frac{d}{2}=\frac{1}{2}\left(\frac{q}{m}\frac{|\Delta V|}{d}\right)\left(\frac{L}{v_{\min}}\right)^2.
$$

Solving gives

$$
v_{\min}=L\sqrt{\frac{q|\Delta V|}{m d^2}}.
$$

After you derive this symbolically, substitute the numerical values.

</div>
</details>

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
q, m, g, d, L, V, v0 = sp.symbols('q m g d L V v0', positive=True, real=True)

# -----------------------------
# Electric field from potential difference
# -----------------------------
E = V/d

# -----------------------------
# 1) Gravity check
# -----------------------------
ratio = sp.simplify((q*E)/(m*g))
display(Math(r"\frac{F_E}{F_g} = " + sp.latex(ratio)))

# -----------------------------
# 2) Vertical motion
# -----------------------------
a = q*E/m
t_exit = L/v0
y_exit = sp.simplify(sp.Rational(1,2)*a*t_exit**2)
display(Math(r"y_{\text{exit}} = " + sp.latex(y_exit)))

# -----------------------------
# 3) Minimum speed to just exit
# -----------------------------
vmin = sp.solve(sp.Eq(y_exit, d/2), v0)[0]
vmin = sp.simplify(vmin)
display(Math(r"v_{\min} = " + sp.latex(vmin)))

# -----------------------------
# Numerical values (proton)
# -----------------------------
vals = {
    q: 1.60e-19,      # C
    m: 1.67e-27,      # kg
    g: 9.80,          # m/s^2
    d: 1.00e-2,       # m
    L: 2.00e-1,       # m
    V: 200.0,         # V
    v0: 2.00e6        # m/s (given entrance speed)
}

ratio_val = float(sp.N(ratio.subs(vals)))
y_exit_val = float(sp.N(y_exit.subs(vals)))
vmin_val = float(sp.N(vmin.subs(vals)))
half_gap = 0.5 * vals[d]

# -----------------------------
# Explicit outcome check
# -----------------------------
print("\nNumeric evaluation:")
print(f"  F_E/F_g = {ratio_val:.2e}")
print(f"  Vertical displacement at exit y_exit = {y_exit_val:.3e} m")
print(f"  Half plate separation d/2 = {half_gap:.3e} m")
print(f"  Minimum speed to exit v_min = {vmin_val:.2e} m/s\n")

if y_exit_val < half_gap:
    print("Outcome:")
    print("• The proton EXITS the plates without hitting the top plate.")
else:
    print("Outcome:")
    print("• The proton DOES NOT exit the plates.")
    print("• It hits the TOP plate before reaching the far edge.")

# -----------------------------
# Interpretation
# -----------------------------
print("\nPhysical interpretation:")
print("• The electric force dominates gravity by many orders of magnitude.")
print("• Vertical deflection grows with time spent between the plates.")
print("• A sufficiently large entrance speed is required to avoid collision.")
print("• This is the same physics used in particle deflectors and beam steering.")
```

## D2.14 - Challenge Problem Energy, Power, and Heating in a Medical X-ray Tube

In a medical **X-ray tube**, electrons are accelerated through a large electric potential difference and strike a metal target.  
Most of the electrical energy delivered to the electrons is converted into **thermal energy**, while a smaller fraction becomes X-rays.

Assume a pulse of electrons with total charge

$$
Q = 3.00\times10^{-6}\ \text{C}
$$

is accelerated through a potential difference of

$$
\Delta V = 1.50\times10^{5}\ \text{V}.
$$

The charge is delivered in a time interval of

$$
\Delta t = 0.0800\ \text{s}.
$$

The target is a metal block with mass

$$
m = 0.40\ \text{kg}
$$

and specific heat capacity

$$
c = 380\ \mathrm{J/(kg\cdot^\circ C)}.
$$

Assume that **80%** of the electrical energy becomes thermal energy in the target.

Answer the following:

1. How much **electric potential energy** is transferred to the electrons?  
2. What is the **average power** delivered during this pulse?  
3. How much **thermal energy** is deposited in the target?  
4. By how much does the **temperature of the target increase**?

Be careful with **units** and clearly state assumptions.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Key relationships**

Electric potential difference relates energy and charge:

$$
\Delta U = Q\,\Delta V.
$$

Average power:

$$
P_{\text{avg}} = \frac{\Delta U}{\Delta t}.
$$

Thermal energy deposited:

$$
Q_{\text{thermal}} = 0.80\,\Delta U.
$$

Thermal energy and temperature change:

$$
Q_{\text{thermal}} = mc\,\Delta T.
$$

---

**Given**

$$
Q = 3.00\times10^{-6}\ \text{C},\quad
\Delta V = 1.50\times10^{5}\ \text{V},\quad
\Delta t = 0.0800\ \text{s}
$$

$$
m = 0.40\ \text{kg},\quad
c = 380\ \mathrm{J/(kg\cdot^\circ C)}
$$

---

**1. Electric potential energy transferred**

$$
\Delta U = Q\Delta V
= (3.00\times10^{-6})(1.50\times10^{5})
= 4.50\times10^{-1}\ \text{J}.
$$

---

**2. Average power delivered**

$$
P_{\text{avg}} = \frac{4.50\times10^{-1}}{0.0800}
= 5.63\ \text{W}.
$$

---

**3. Thermal energy deposited**

$$
Q_{\text{thermal}} = 0.80(4.50\times10^{-1})
= 3.60\times10^{-1}\ \text{J}.
$$

---

**4. Temperature increase of the target**

Solve for $\Delta T$:

$$
\Delta T = \frac{Q_{\text{thermal}}}{mc}
= \frac{3.60\times10^{-1}}{(0.40)(380)}
= 2.37\times10^{-3}\ ^\circ\text{C}.
$$

---

**Results**

$$
\boxed{\Delta U = 0.450\ \text{J}},\quad
\boxed{P_{\text{avg}} = 5.63\ \text{W}},\quad
\boxed{Q_{\text{thermal}} = 0.360\ \text{J}},\quad
\boxed{\Delta T = 2.37\times10^{-3}\ ^\circ\text{C}}
$$

---

**Physical interpretation**

- Electric potential difference sets the **energy per charge** delivered.
- Power depends on how quickly charge is delivered.
- Most energy in X-ray tubes becomes **heat**, not radiation.
- Even small temperature rises matter in medical devices because pulses repeat rapidly.

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
Q, dV, dt, m, c, eta = sp.symbols('Q dV dt m c eta', positive=True, real=True)

# -----------------------------
# Equations (symbolic)
# -----------------------------
dU = Q*dV                          # electric potential energy transferred
Pavg = dU/dt                       # average power
Qth = eta*dU                       # thermal energy deposited
dT = Qth/(m*c)                     # temperature increase

# -----------------------------
# Display equations (math only; no \text or \cdot)
# -----------------------------
display(Math(r"\Delta U = Q\,\Delta V"))
display(Math(r"P_{\mathrm{avg}} = \frac{\Delta U}{\Delta t}"))
display(Math(r"Q_{\mathrm{thermal}} = \eta\,\Delta U"))
display(Math(r"\Delta T = \frac{Q_{\mathrm{thermal}}}{mc}"))

# -----------------------------
# Substitute numerical values
# -----------------------------
vals = {
    Q: 3.00e-6,      # C
    dV: 1.50e5,      # V
    dt: 0.0800,      # s
    m: 0.40,         # kg
    c: 380.0,        # J/(kg*C)
    eta: 0.80        # fraction to thermal
}

dU_val = float(sp.N(dU.subs(vals)))
P_val  = float(sp.N(Pavg.subs(vals)))
Qth_val = float(sp.N(Qth.subs(vals)))
dT_val  = float(sp.N(dT.subs(vals)))

# -----------------------------
# Plain numeric answers
# -----------------------------
print("\nNumeric results:")
print(f"  ΔU = {dU_val:.3e} J")
print(f"  P_avg = {P_val:.3e} W")
print(f"  Q_thermal = {Qth_val:.3e} J")
print(f"  ΔT = {dT_val:.3e} °C")

print("\nInterpretation:")
print("• The potential difference sets energy per charge: ΔU = QΔV.")
print("• Average power depends on how quickly the charge is delivered: P_avg = ΔU/Δt.")
print("• Only a fraction becomes heat (η); that thermal energy raises the target temperature.")
```
