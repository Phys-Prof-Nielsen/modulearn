## Why this module matters

So far in this course, we have focused on **forces, fields, and particle motion**. In this module, we shift perspective and study **light** — not as a wave or quantum object yet, but as something that travels along well-defined **paths**.

Ray optics provides a powerful and intuitive framework for understanding:
- how light **reflects** from surfaces,
- how it **refracts** when entering new materials,
- why **total internal reflection** occurs,
- and how lenses and optical systems form images.

Beyond practical optics, this module also introduces a profound idea: light does not merely follow rules — it follows an **extremum principle**. Through **Fermat’s Principle**, we glimpse how nature often selects paths that extremize a quantity, foreshadowing **variational physics** and later ideas such as the **Lagrangian formulation** of mechanics.

At the D-level, the emphasis is on:
- definitions and geometry,
- physical interpretation rather than heavy derivation,
- and building intuition that will support wave optics, modern optics, and advanced physics later on.

By the end of this module, you will be able to:

---

### A. Light Rays and the Ray Optics Approximation  
*Defines what we mean by a “ray” and when the model applies.*

- **A1.** Define a **light ray** as an idealized straight-line path indicating the direction of light propagation.
- **A2.** State the assumptions underlying **ray optics**, including when wavelength effects can be neglected.
- **A3.** Distinguish between rays as **geometric tools** and light as a physical phenomenon.
- **A4.** Identify situations where the ray model is expected to break down (qualitative only).

---

### B. Reflection of Light  
*Describes how rays interact with reflecting surfaces.*

- **B1.** State the **law of reflection**: the angle of incidence equals the angle of reflection.
- **B2.** Identify the **normal line** and measure angles relative to it.
- **B3.** Distinguish between **specular reflection** and **diffuse reflection**.
- **B4.** Construct simple ray diagrams for reflection from flat surfaces.

---

### C. Refraction and Snell’s Law  
*Explains how light changes direction in different media.*

- **C1.** Define the **index of refraction** and relate it to the speed of light in a medium.
- **C2.** State **Snell’s law** and identify the physical meaning of each term.
- **C3.** Predict qualitatively how a ray bends when entering a medium with higher or lower refractive index.
- **C4.** Explain refraction using changes in light speed (conceptual, not wave-based).

---

### D. Total Internal Reflection  
*Reveals a limiting case of refraction.*

- **D1.** State the conditions required for **total internal reflection**.
- **D2.** Define and identify the **critical angle**.
- **D3.** Predict when light will be fully reflected rather than refracted.
- **D4.** Describe real-world applications of total internal reflection (qualitative).

---

### E. Fermat’s Principle and a First Look at Variational Physics  
*Introduces an extremum principle governing light paths.*

- **E1.** State **Fermat’s Principle**: light follows the path that extremizes travel time.
- **E2.** Explain how reflection and refraction can be understood using this principle (conceptual).
- **E3.** Recognize Fermat’s Principle as an example of a **variational principle**.
- **E4.** Identify the conceptual connection between Fermat’s Principle and later ideas in **Lagrangian mechanics** (no formal mathematics).

---

### F. Thin Lenses and Image Formation  
*Applies ray optics to optical systems.*

- **F1.** Distinguish between **converging** and **diverging** thin lenses.
- **F2.** Identify the **focal point** and **focal length** of a lens.
- **F3.** Use principal rays to construct **ray diagrams** for image formation.
- **F4.** Describe images as **real or virtual**, **upright or inverted**, and **magnified or reduced**.
- **F5.** Interpret lens behavior using geometry rather than memorized rules.

---

### Performance Criteria

- Light rays are treated as **geometric approximations**, not physical objects.
- Reflection and refraction laws are stated clearly and applied correctly.
- Total internal reflection is explained using physical reasoning and geometry.
- Fermat’s Principle is understood conceptually as an extremum principle.
- Thin-lens behavior is interpreted using ray diagrams and physical insight.
- Connections between optics and broader variational ideas are explicitly recognized.
