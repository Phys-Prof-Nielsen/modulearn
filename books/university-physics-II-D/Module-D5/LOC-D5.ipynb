# D5.0 Learning Outcomes — Magnetic Force on a Moving Charge (Lorentz Force)
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this module matters

Up to now, we have described how **electric fields** exert forces on charges, whether those charges are at rest or in motion.  
However, many important physical systems involve **moving charges interacting with magnetic fields** — from electric motors and particle accelerators to Earth’s magnetosphere and astrophysical plasmas.

This module introduces the **magnetic force on a single charged particle** and combines it with the electric force into a single, unified description known as the **Lorentz force**. This force law is one of the most fundamental relationships in electromagnetism.

At the D-level, the focus is on **definitions, geometry, and physical interpretation**:
- how the magnetic force depends on velocity,
- why magnetic forces change direction but not speed,
- and how electric and magnetic effects coexist in real systems.

These ideas lay the conceptual foundation for later topics such as circular motion in magnetic fields, currents in magnetic fields, and electromagnetic devices.

By the end of this module, you will be able to:

---

### A. Experimental Definition of the Magnetic Force  
*Introduces the magnetic force from observation.*

- **A1.** State that experiments show a force acts on a **moving charged particle** in the presence of a magnetic field.
- **A2.** Identify the experimentally observed properties of the magnetic force:
  - proportional to charge magnitude,
  - proportional to particle speed,
  - dependent on the angle between velocity and magnetic field,
  - zero for stationary charges.
- **A3.** Write the experimentally determined magnitude of the magnetic force,

  $$
  F_B = |q|\,v\,B\,\sin\theta,
  $$

  where $\theta$ is the angle between $\vec{v}$ and $\vec{B}$.
- **A4.** State that the magnetic force is always **perpendicular** to both the velocity and the magnetic field.

---

### B. Direction of the Magnetic Force and the Cross Product  
*Encodes magnitude and direction in a single mathematical form.*

- **B1.** Use the right-hand rule to determine the direction of the magnetic force on a **positive charge**.
- **B2.** Correctly reverse the force direction for a **negative charge**.
- **B3.** Recognize that the perpendicular nature of the force motivates the use of a **cross product**.
- **B4.** Write the magnetic force in vector form,

  $$
  \vec{F}_B = q\,\vec{v} \times \vec{B}.
  $$

- **B5.** Interpret the cross product as a compact way to encode both **direction** and **$\sin\theta$ dependence**.

---

### C. The Lorentz Force  
*Combines electric and magnetic forces into a single law.*

- **C1.** Write the Lorentz force law,

  $$
  \vec{F} = q(\vec{E} + \vec{v} \times \vec{B}).
  $$

- **C2.** Identify the electric force as acting on charges whether or not they are moving.
- **C3.** Identify the magnetic force as acting only on **moving charges**.
- **C4.** Explain qualitatively when electric or magnetic effects dominate particle motion.

---

### D. Motion of a Particle in a Magnetic Field  
*Develops physical intuition for trajectories.*

- **D1.** Explain why magnetic forces do **no work** on a charged particle.
- **D2.** State that magnetic forces can change the **direction** but not the **speed** of a particle.
- **D3.** Describe qualitatively how circular or curved motion arises in a uniform magnetic field.
- **D4.** Relate magnetic force to centripetal acceleration conceptually (no full derivations).

---

### Performance Criteria

- Magnetic force is introduced from **experimental observation**, not assumed field behavior.
- Force magnitude and direction are interpreted geometrically and physically.
- The cross product form $q\,\vec{v}\times\vec{B}$ is understood as encoding experimental facts.
- Electric and magnetic forces are clearly distinguished and correctly combined.
- Particle motion in magnetic fields is explained using physical reasoning.
