## D2.3.1 Motivation — From Force to Energy to Potential in Electrostatics

In electrostatics, forces between charges can vary strongly with position. Tracking motion using forces alone quickly becomes cumbersome.

As in mechanics, energy methods provide often a more powerful and general approach.

Electric **potential energy** allows us to describe how energy is stored due to the interaction between charges.  Electric **potential** goes one step further by describing how energy varies with position due to the **electric field itself**, independent of any specific test charge.

This mirrors the progression used earlier for gravity:

- force → potential energy → potential  
- object-based description → field-based description  

By introducing electric potential, we gain a tool that simplifies calculations, clarifies physical interpretation, and naturally connects to later topics such as **capacitors, circuits, and equipotentials**.

<h3 style="margin-top:0; color:#000000;">Example — Electric Potential Energy and Potential from a Uniform Electric Field</h3>

Consider a charge $q$ moving along the $x$-axis in a uniform electric field $E_x$ directed in the positive $x$-direction.

The electric force on the charge is

$$
F_x = qE_x.
$$

---

**Step 1: Relate force and potential energy**

For a conservative force in one dimension,

$$
F_x = -\frac{dU}{dx}.
$$

Substituting the electric force,

$$
qE_x = -\frac{dU}{dx}.
$$

This gives

$$
\frac{dU}{dx} = -qE_x.
$$

Integrating,

$$
U(x) = -qE_{x}x + C.
$$

---

**Step 2: Define the electric potential**

The electric potential is defined as potential energy per unit charge:

$$
V(x) = \frac{U(x)}{q}.
$$

Substituting the expression for $U(x)$,

$$
V(x) = -E_{x}x + \frac{C}{q}.
$$

Choosing the reference so that $C = 0$,

$$
V(x) = -E_{x}x.
$$

---

**Physical interpretation**

- The electric potential depends only on the **electric field**, not on the charge.
- The potential energy depends on both the **field** and the **sign and magnitude of the charge**.
- A positive charge naturally moves toward lower electric potential.

The electric force on a test charge $q$ due to a source charge $Q$ at distance $r$ is given by Coulomb’s law:

4. **From Energy to Potential**

   $$
   V(r) = \frac{U(r)}{q}.
   $$

5. **Reference Choice**  
   Choose a convenient reference level for the potential.

<h3 style="margin-top:0; color:#000000;">Danger Zone — Electric Potential vs. Electric Potential Energy</h3>

In electrostatics, **electric charge carries a sign**. Mass does not.

This creates an important distinction:

- **Electric potential $V$** depends only on the **source charges and field configuration**.
- **Electric potential energy $U = qV$** depends on the **sign and magnitude of the test charge**.

As a result:

- A region of **high electric potential** does not automatically mean **high potential energy**.
- Positive and negative charges behave **oppositely** in the same electric potential.
- Forgetting the sign of $q$ is one of the most common errors in electrostatics.

Always separate these ideas clearly:
- Use **$V$** to describe the field.
- Use **$U$** to describe how a *specific charge* interacts with that field.

## D2.3.2 Units of Electric Potential

Electric potential is defined as **potential energy per unit charge**:

$$
V = \frac{U}{q}.
$$

Because potential energy is measured in **joules (J)** and electric charge is measured in **coulombs (C)**, the SI unit of electric potential is

$$
1~\text{volt (V)} = 1~\text{joule per coulomb} = 1~\frac{\text{J}}{\text{C}}.
$$

---

### Physical meaning of the volt

A potential difference of **1 volt** means that **1 joule of energy** is transferred when **1 coulomb of charge** moves between two points:

$$
\Delta U = q\,\Delta V.
$$

---

### An alternative energy unit: the electron volt

In atomic, nuclear, and particle physics, energies are often extremely small when expressed in joules.  
For this reason, a more convenient unit is commonly used: the **electron volt (eV)**.

An **electron volt** is defined as the amount of energy gained by a charge with magnitude equal to the **elementary charge** when it moves through a potential difference of **1 volt**:

$$
1~\text{eV} = (1.602 \times 10^{-19}\,\text{C})(1~\text{V})
= 1.602 \times 10^{-19}\,\text{J}.
$$

---

### Why the electron volt is useful

- It directly connects **energy** to **electric potential difference**.
- It avoids repeatedly writing extremely small numbers in joules.
- It is the natural energy unit for electrons, ions, and subatomic particles.

## D2.3.3 How Electric Potential Is Used in Practice

Most of us have already encountered **electric potential** in everyday life, even if they have not thought about it explicitly.

In **household electricity**, electric potential differences are measured in **volts**. Wall outlets, batteries, and power supplies are all characterized by the potential difference they provide:

- A **battery** maintains a fixed potential difference between its terminals.
- A **wall outlet** provides an oscillating potential difference that drives electric current.
- Devices operate by converting electric potential energy into other forms of energy such as light, heat, or motion.

In these contexts, electric potential is preferred over force because it directly tells us **how much energy per charge is available**.

---

### Using potential to predict motion

Physicists frequently use electric potential to determine how fast charged particles move after being accelerated.

When a charge moves through a potential difference $\Delta V$, its electric potential energy changes by

$$
\Delta U = q\,\Delta V.
$$

If this energy is converted into kinetic energy, then

$$
q\,\Delta V = \Delta K.
$$

This allows particle speeds to be calculated **without ever computing the force or acceleration explicitly**.

---

### Applications in physics and technology

This energy-based use of potential is fundamental in:

- **Particle accelerators**, where charged particles gain enormous kinetic energy by passing through large potential differences.
- **Electron guns** and **cathode-ray tubes**, where electron speeds are set by applied voltages.
- **Vacuum tubes and detectors**, where particle energies determine how matter and radiation interact.

In all of these cases, electric potential provides a direct bridge between **fields** and **observable motion**.

---

### Key idea

> Electric potential is a measure of energy availability.  
> By knowing the potential difference alone, we can predict energy changes, speeds, and outcomes—often more easily than by using forces.

---

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example — Accelerating Electrons in a Particle Accelerator</h3>

In particle physics experiments, charged particles are often accelerated using **electric potential differences** rather than mechanical forces.

Consider an **electron** that starts from rest and is accelerated through a potential difference of

$$
\Delta V = 5.0 \times 10^6~\text{V} \quad (5~\text{MV}).
$$

This is a typical scale used in early stages of particle accelerators and electron guns.

---

**Step 1: Relate potential difference to energy**

The change in electric potential energy is

$$
\Delta U = q\,\Delta V.
$$

For an electron, $q = -e$, so the **magnitude** of the energy change is

$$
|\Delta U| = e\,\Delta V.
$$

Using electron volts,

$$
|\Delta U| = 5.0~\text{MeV}.
$$

---

**Step 2: Convert potential energy to kinetic energy**

If the electron starts from rest, the electric potential energy is converted into kinetic energy:

$$
\Delta K = |\Delta U|.
$$

Thus, the electron emerges with a kinetic energy of

$$
K = 5.0~\text{MeV}.
$$

---

**Step 3: Physical interpretation**

- The result depends **only on the potential difference**, not on the details of the electric field.
- No force or acceleration calculation is required.
- The sign of the electron’s charge determines the **direction of motion**, but not the energy gained.

This energy scale is large enough that relativistic effects become important, which is why particle physicists often work directly with **energy units (MeV, GeV)** rather than speeds.

---

**Why this matters in particle physics**

Modern accelerators—such as those used to study fundamental particles—operate by stacking many such potential differences to give particles enormous kinetic energies.  
Electric potential is therefore the **primary control knob** for particle energy in experimental physics.

<h3 style="margin-top:0; color:#000000;">Box Activity 2 — Speed of an Exotic Particle Accelerated in a Supernova</h3>

High-energy particles observed on Earth—such as **muons** produced by cosmic rays—often originate from extreme astrophysical environments like **supernova remnants**.

One important mechanism responsible for their enormous energies is **Fermi acceleration**, in which particles gain energy through repeated interactions with moving magnetic fields and shock fronts.

Rather than modeling each interaction in detail, physicists often describe the net effect as an **effective potential difference**.

---

Consider a muon (mass  
$m_\mu = 1.88\times10^{-28}\,\text{kg}$, charge magnitude $e$) that gains energy equivalent to being accelerated through an effective potential difference of

$$
\Delta V = 5.1 \times 10^{7}\,\text{V} \quad (51~\text{MV})
$$

in a supernova shock region.

---

Work through the following:

1. **Energy Gain**  
   Use

   $$
   \Delta U = q\,\Delta V
   $$

   to determine the energy gained by the muon.

2. **Energy Conversion**  
   Assume the muon starts from rest and that this energy becomes kinetic energy.

3. **Speed Estimate**  
   Use the classical expression

   $$
   K = \frac{1}{2}mv^2
   $$

   to estimate the muon’s speed.

4. **Interpretation**  
   Compare your result to the speed of light

   $$
   c = 3.0\times10^8\,\text{m/s}.
   $$

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The energy gained by the muon is

$$
|\Delta U| = e\,\Delta V
= (1.60\times10^{-19})(5.1\times10^{7})
\approx 8.2\times10^{-12}\,\text{J}.
$$

Setting this equal to kinetic energy,

$$
\frac{1}{2}m_\mu v^2 = 8.2\times10^{-12}.
$$

Solving for $v$,

$$
v = \sqrt{\frac{2(8.2\times10^{-12})}{1.88\times10^{-28}}}
\approx 2.94\times10^{8}\,\text{m/s}.
$$

Expressed as a fraction of the speed of light,

$$
v \approx 0.98\,c.
$$

---

**Important physical note**

At speeds approaching the speed of light, the **classical kinetic energy expression**

$$
K = \frac{1}{2}mv^2
$$

is no longer valid.

In this regime, **special relativity must be used** to relate energy, momentum, and speed.  
The relationship between electric potential and energy,

$$
\Delta U = q\,\Delta V,
$$

**remains correct**, but the way that energy translates into motion changes.

As particles gain energy:
- Their **speed approaches but never reaches** the speed of light.
- Additional energy goes into increasing **relativistic energy**, not speed.
- Physicists therefore describe particles primarily by their **energy (MeV, GeV, TeV)** rather than their velocity.

This example marks the boundary where **classical physics gives way to relativistic physics**, a transition explored later in the course.
