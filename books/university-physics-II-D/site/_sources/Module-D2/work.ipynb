## D2.1.1 Work as a Dot Product

In mechanics, **work** measures how much energy is transferred by a force as an object is displaces. The most general definition for 1D motion expresses work as a **dot product** between the force and an infinitesimal displacement in 1D:

$$
W = \int \vec{F} \cdot d\vec{x}.
$$

This definition emphasizes an important physical idea: **only the component of the force along the direction of motion can do work**.

Even though we restrict motion to one dimension for now, this vector form makes clear that work is not simply “force times distance,” but rather a measure of how force and displacement align.

### Geometric Interpretation of Work

The dot product can be written in its **geometric form** as

$$
\vec{F} \cdot d\vec{x} = F\,dx\,\cos\theta,
$$

where $\theta$ is the angle between the force vector and the displacement vector.

This form highlights the physics:

- If $\theta = 0^\circ$, the force is parallel to the motion and does **maximum work**, the force **adds** energy to the object.
- If $\theta = 90^\circ$, the force is perpendicular to the motion and does **no work**.
- If $\theta = 180^\circ$, the force is anti-parallel to the motion and does **maximum work**, the force **removes** energy from the object.

Work is therefore not determined by force alone, but by **how the force is oriented relative to the motion**.

### Component Form of Work in One Dimension

When motion is restricted to the $x$-axis, the dot product simplifies to a **component form**:

$$
\vec{F} \cdot d\vec{x} = F_x\,dx.
$$

The total work becomes

$$
W = \int F_x(x)\,dx.
$$

This is the form used most often in one-dimensional problems. It shows that work can be interpreted as the **area under a force–position graph**.

Although this expression looks simple, it retains the full physical meaning of the dot product: only the force component along the direction of motion contributes to energy transfer.

## D2.1.2 Conservative Forces

A **conservative force** is one for which the work done depends only on the **initial and final positions**, not on the path taken between them.

Key properties of conservative forces:

- Work around any closed path is zero.
- A potential energy function can be defined.
- Energy methods can replace force-based motion analysis.

Examples include gravitational forces, spring forces, and (in electrostatics) electric forces due to fixed charges.

This property allows energy to be treated as a **stored quantity**, rather than something that must be tracked continuously through motion.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example — Work Done by the Gravitational Force</h3>

Consider an object of mass $m$ moving vertically along the $z$-axis near Earth’s surface.  
The gravitational force acting on the object is constant and given by

$$
\vec{F}_g = -mg\,\hat{z}.
$$

Suppose the object moves from an initial height $z_i$ to a final height $z_f$.

---

**Step 1: Write the work as a dot product**

The work done by a force is defined as

$$
W = \int \vec{F} \cdot d\vec{r}.
$$

Because the motion is purely vertical,

$$
d\vec{r} = dz\,\hat{z}.
$$

---

**Step 2: Evaluate the dot product**

$$
\vec{F}_g \cdot d\vec{r}
= (-mg\,\hat{z}) \cdot (dz\,\hat{z})
= -mg\,dz.
$$

---

**Step 3: Integrate to find the work**

$$
W_g = \int_{z_i}^{z_f} (-mg)\,dz
= -mg(z_f - z_i).
$$

---

**Physical interpretation**

- If the object moves **downward** ($z_f < z_i$), the work done by gravity is **positive**.
- If the object moves **upward** ($z_f > z_i$), the work done by gravity is **negative**.

This result shows explicitly that gravity does work only through the component of force **along the displacement**, and that the work depends only on the **initial and final heights**, not on how the object moves between them.

This path independence is the defining feature of a **conservative force**.

<h3 style="margin-top:0; color:#000000;">Box Activity 1 — Work Done by Newton’s Law of Gravity</h3>

Now consider an object of mass $m$ moving **radially** in the gravitational field of Earth, treated as a point mass $M$. The gravitational force on the object is given by **Newton’s law of gravity**,

$$
\vec{F}_g = -\,G\,\frac{Mm}{r^2}\,\hat{r},
$$

where $r$ is the distance from Earth’s center.

---

Work through the following:

1. **Observation**  
   The gravitational force always points **toward Earth’s center** and its magnitude depends on the distance $r$.

2. **Question**  
   *How much work is done by gravity when the object moves from an initial distance $r_i$ to a final distance $r_f$?*

3. **Hypothesis**  
   Because gravity is a conservative force, the work done should depend only on $r_i$ and $r_f$, not on the path taken.

4. **Prediction**  
   Predict whether gravity does **positive or negative work** when the object moves:
   - inward ($r_f < r_i$),
   - outward ($r_f > r_i$).

5. **Reasoning**  
   Use the dot product between $\vec{F}_g$ and $d\vec{r}$ to determine how the sign and magnitude of the work arise.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Because the motion is radial,

$$
d\vec{r} = dr\,\hat{r}.
$$

The dot product becomes

$$
\vec{F}_g \cdot d\vec{r}
= \left(-G\frac{Mm}{r^2}\hat{r}\right)\cdot (dr\,\hat{r})
= -G\frac{Mm}{r^2}\,dr.
$$

The work done by gravity is therefore

$$
W_g = \int_{r_i}^{r_f} -G\frac{Mm}{r^2}\,dr
= GMm\left(\frac{1}{r_f} - \frac{1}{r_i}\right).
$$

**Interpretation:**
- If $r_f < r_i$ (object moves inward), gravity does **positive work**.
- If $r_f > r_i$ (object moves outward), gravity does **negative work**.

This result depends only on the initial and final distances, confirming that gravity is a **conservative force** and motivating the definition of **gravitational potential energy**.

<h3 style="margin-top:0; color:#000000;">Box Activity 2 — Work Done by a Constant Electric Field Along a Closed Path</h3>

A proton is immersed in a **constant electric field** given by

\vec{E} = \left[(2.5\times10^{12})\hat{i} + (5.0\times10^{12})\hat{j}\right]\ \text{N/C}.

The proton starts at the origin $(x,y)=(0,0)\ \text{m}$ and is moved along **straight-line segments only**, forming a rectangular path.

---

Work through the following:

1. **Observation**  
   The electric field is uniform and has both $x$- and $y$-components.

2. **Key Relationship**  
   For a constant electric field, the work done by the electric force over a displacement $\Delta\vec{r}$ is

   $$
   W = q\,\vec{E}\cdot\Delta\vec{r}.
   $$

3. **Segment-by-Segment Analysis**  
   Compute the work done by the electric force on the proton for each segment:
   - (a) $(0,0) \rightarrow (2.00,0)$  
   - (b) $(2.00,0) \rightarrow (2.00,2.00)$  
   - (c) $(2.00,2.00) \rightarrow (0,2.00)$  
   - (d) $(0,2.00) \rightarrow (0,0)$  

4. **Closed-Path Question**  
   Add the work from all four segments to find the **total work** done around the closed path.

5. **Conceptual Reasoning**  
   Based on your result, determine whether the electric force in this situation is **conservative**, and explain why.

**Charge of the proton**

$$
q = +e = 1.60\times10^{-19}\ \text{C}.
$$

---

**Segment (a): $(0,0)\rightarrow(2.00,0)$**

$$
\Delta\vec{r} = (2.00)\hat{i}.

$$
W_a = q(E_x\Delta x) 
= (1.60\times10^{-19})(2.5\times10^{12})(2.00)
= 8.00\times10^{-7}\ \text{J}.
$$

---

**Segment (b): $(2.00,0)\rightarrow(2.00,2.00)$**

$$
\Delta\vec{r} = (2.00)\hat{j}.
$$

$$
W_b = q(E_y\Delta y) 
= (1.60\times10^{-19})(5.0\times10^{12})(2.00)
= 1.60\times10^{-6}\ \text{J}.
$$

---

**Segment (c): $(2.00,2.00)\rightarrow(0,2.00)$**

$$
\Delta\vec{r} = (-2.00)\hat{i}.
$$

$$
W_c = q(E_x\Delta x) 
= -8.00\times10^{-7}\ \text{J}.
$$

---

**Segment (d): $(0,2.00)\rightarrow(0,0)$**

$$
\Delta\vec{r} = (-2.00)\hat{j}.
$$

$$
W_d = q(E_y\Delta y) 
= -1.60\times10^{-6}\ \text{J}.
$$

---

**Total work around the loop**

$$
W_{\text{total}} = W_a + W_b + W_c + W_d = 0.
$$

---

**Interpretation**

- The net work done by the electric force around a closed path is **zero**.
- This indicates that the electric force in this electrostatic situation is **conservative**.
- The result is independent of the path taken and depends only on the endpoints.

This is a defining feature of electrostatic forces and motivates the use of **electric potential**.

## D2.1.3 Definition of Potential Energy

For a conservative force, **potential energy** $U$ is defined through the work done by the force:

$$
\Delta U = -W.
$$

In one dimension, this leads to the relationship

$$
F_x = -\frac{dU}{dx}.
$$

This equation states that force points in the direction of **decreasing potential energy**.

Potential energy is not an absolute quantity—only **changes** in potential energy have physical meaning.  
The zero level of potential energy can be chosen for convenience without affecting observable results.

---


<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example — Obtaining Gravitational Potential Energy</h3>

Consider an object of mass $m$ moving vertically along the $z$-axis near Earth’s surface. The gravitational force acting on the object is constant and given by

$$
\vec{F}_g = -mg\,\hat{z}.
$$

From the previous example, the work done by gravity as the object moves from $z_i$ to $z_f$ is

$$
W_g = -mg\,(z_f - z_i).
$$

---

**Step 1: Relate work to potential energy**

For a conservative force, the change in potential energy is defined as

$$
\Delta U = -W.
$$

Applying this to gravity gives

$$
\Delta U = mg\,(z_f - z_i).
$$

---

**Step 2: Identify the potential energy function**

Because $\Delta U = U(z_f) - U(z_i)$, this result implies

$$
U(z) = mgz + C,
$$

where $C$ is an arbitrary constant that sets the reference level.

---

**Step 3: Interpret the result physically**

- As the object moves **upward**, its gravitational potential energy **increases**.
- As the object moves **downward**, its gravitational potential energy **decreases**.
- Only **changes** in $U$ affect the motion; the constant $C$ has no physical consequence.

This expression shows that gravitational potential energy near Earth’s surface is a **linear function of height**, reflecting the fact that the gravitational force is constant in this region.

</div>

<h3 style="margin-top:0; color:#000000;">Box Activity 2 — Gravitational Potential Energy from Newton’s Law</h3>

Refer back to **Box Activity 3 — Work Done by Newton’s Law of Gravity**, where you found that the work done by gravity for radial motion from $r_i$ to $r_f$ is

$$
W_g = GMm\left(\frac{1}{r_f} - \frac{1}{r_i}\right).
$$

Now use that result to obtain the **gravitational potential energy function**.

---

Work through the following:

1. **Observation**  
   Gravity is a **conservative force**, so a potential energy function $U(r)$ can be defined.

2. **Key Definition**  
   For a conservative force,

   $$
   \Delta U = -W.
   $$

3. **Question**  
   *Using the work result above, what is $\Delta U = U(r_f)-U(r_i)$?*

4. **Goal**  
   Find a function $U(r)$ such that

   $$
   U(r_f)-U(r_i) = -\,W_g.
   $$

5. **Reference Choice**  
   A common and convenient choice is to set

   $$
   U(\infty) = 0.
   $$

   Use this to determine the constant in your expression.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Using the definition $\Delta U = -W_g$,

$$
U(r_f) - U(r_i)
= -GMm\left(\frac{1}{r_f} - \frac{1}{r_i}\right)
= GMm\left(\frac{1}{r_i} - \frac{1}{r_f}\right).
$$

A function that produces this difference is

$$
U(r) = -G\frac{Mm}{r} + C.
$$

To choose the reference level, set $U(\infty)=0$.  
Since $1/r \to 0$ as $r \to \infty$, this gives $C=0$, so

$$
U(r) = -G\frac{Mm}{r}.
$$

**Interpretation:**
- $U(r)$ is **negative**, meaning the object–Earth system is bound.
- As $r$ increases, $U(r)$ increases toward $0$ (it becomes “less negative”).
- Only changes in $U$ affect motion; the reference choice $U(\infty)=0$ is a convention.
