## D2.2.1 Motivation — From Potential Energy to Potential

In previous sections, **potential energy** was introduced as a way to describe energy stored due to the **interaction between objects**. For example, gravitational potential energy depends on both the mass of the object and the gravitational interaction it experiences.

However, this immediately raises a conceptual issue:

- Potential energy depends on the **test object** (its mass or charge).
- The underlying **environment** responsible for the interaction exists whether or not the test object is present.

This motivates a shift in perspective.

Instead of attaching energy directly to the object–interaction pair, we introduce a quantity that belongs to the **field itself**. This quantity is called the **potential**.

The potential describes the energy *available per unit mass or per unit charge* at each point in space, independent of what object might later be placed there.


<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D2.2.2 Interaction vs. Field Viewpoint

It is useful to distinguish clearly between **potential energy** and **potential**:

- **Potential energy** describes the energy of a **specific system**  
  (for example, Earth + object, or source charge + test charge).

- **Potential** describes a property of the **field configuration in space**.

Mathematically, this distinction is expressed as

$$
\text{potential energy} = (\text{property of object}) \times (\text{potential}).
$$

For gravity near Earth:
- Object property: mass $m$
- Field property: gravitational potential

For electricity:
- Object property: charge $q$
- Field property: electric potential

This separation allows us to describe the field once, and then analyze the behavior of *any* object placed within it.


### Why Use Potential?

Using potential instead of potential energy offers several important advantages:

- The potential depends **only on the field**, not on the test object.
- The same potential applies to **all masses or charges** placed in that region.
- Energy changes can be computed immediately using

  $$
  \Delta U = (\text{object property})\,\Delta(\text{potential}).
  $$

This approach simplifies both computation and interpretation:

- Complex interactions are encoded into a single scalar function.
- Motion and energy changes can be predicted without tracking forces directly.
- Potentials naturally connect to later topics such as **equipotential surfaces**, **circuits**, and **energy landscapes**.

For these reasons, potentials play a central role in electromagnetism and modern physics.

### Relationship Between Force, Energy, Field, and Potential (1D)

For **conservative interactions**, force and potential energy are not independent ideas. They are **derived from one another** through how energy changes with position.

In one dimension, this relationship is

$$
F_x(x) = -\frac{dU(x)}{dx}.
$$

This equation states that the force points in the direction of **decreasing potential energy**. Where the potential energy changes rapidly with position, the force is large. Where the potential energy is constant, the force is zero.

---

The same structure applies at the **field level**. Instead of attaching energy and force to a specific object, we define quantities **per unit object property**:

- The **field** describes force per unit mass or per unit charge.
- The **potential** describes potential energy per unit mass or per unit charge.

In one dimension, these relationships are (for gravitational and electric fields):

$$
g(x) = -\frac{d\Phi(x)}{dx}
\quad \text{and} \quad
E(x) = -\frac{dV(x)}{dx},
$$

where:
- $g(x)$ is the gravitational field,
- $\Phi(x)$ is the gravitational potential,
- $E(x)$ is the electric field,
- $V(x)$ is the electric potential.

---

Force and potential energy describe the **object’s response**. Field and potential describe the **structure of space** that causes that response.

## D2.2.3 Definition of Potential

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Definition — Potential</h3>

The **potential** is a scalar quantity that describes how **potential energy varies with position** due to a field.

It is defined as **potential energy per unit object property**:

$$
\text{potential} = \frac{\text{potential energy}}{\text{object property}}.
$$

For the interactions encountered in this course, the relevant object properties are:

- **mass** for gravitational interactions  
- **electric charge** for electric interactions  

In one dimension, this gives:

- **Gravitational potential**
  $$
  \Phi(x) = \frac{U(x)}{m}
  $$

- **Electric potential**
  $$
  V(x) = \frac{U(x)}{q}
  $$

In field theory, mass and electric charge play the role of **coupling strengths**: they determine how strongly an object responds to a given field.  

The potential itself depends only on the **field configuration**. Once the potential is known, the potential energy of any object is found by multiplying by its mass or charge.

</div>

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example — From Force to Potential Energy to Potential </h3>

Near Earth’s surface, the gravitational force on an object of mass $m$ moving along the vertical $z$-axis is

$$
F_z = -mg.
$$

---

**Step 1: Relate force and potential energy**

For a conservative force in one dimension,

$$
F_z = -\frac{dU}{dz}.
$$

Substituting the gravitational force gives

$$
-mg = -\frac{dU}{dz}.
$$

Solving for $U(z)$,

$$
\frac{dU}{dz} = mg.
$$

Integrating,

$$
U(z) = mgz + C,
$$

where $C$ is an arbitrary reference constant.

---

**Step 2: Define the gravitational potential**

The **gravitational potential** is defined as potential energy **per unit mass**:

$$
\Phi(z) = \frac{U(z)}{m}.
$$

Substituting the expression for $U(z)$,

$$
\Phi(z) = gz + \frac{C}{m}.
$$

Choosing the reference so that $C = 0$,

$$
\Phi(z) = gz.
$$

---

**Physical interpretation**

- The **force** tells how an object accelerates.
- The **potential energy** tells how much energy the object–Earth system has due to position.
- The **potential** tells how energy varies with position in the gravitational field itself.

At the same height $z$, all objects experience the **same gravitational potential**, even though heavier objects have more gravitational potential energy.

**Gravitational Potential Beyond Intro Physics**

We often work with **gravitational potential energy**, but the closely related concept of **gravitational potential** appears far beyond introductory mechanics.

In **satellite motion and spaceflight**, gravitational potential provides the natural language for describing bound orbits, escape, and orbital energy. Satellite trajectories are understood most clearly by how potential varies with distance from Earth, rather than by tracking forces directly.

In **geodesy and Earth science**, gravitational potential is a primary quantity. Variations in Earth’s mass distribution—mountain ranges, ocean trenches, and density changes deep inside the planet—produce small but measurable changes in gravitational potential. Modern satellite missions map these variations to study Earth’s structure and evolution.

Gravitational potential also plays a crucial role in **global positioning systems (GPS)**. Differences in gravitational potential between Earth’s surface and satellite orbits cause clocks to tick at different rates. Correcting for these potential differences is essential for GPS to function accurately.

At the most fundamental level, in **Einstein’s theory of general relativity**, gravity is no longer described as a force. Instead, gravitational potential is closely connected to how **time flows differently** in different regions of space. In this view, objects move along paths determined by the structure of spacetime itself.

**Key idea:**  
Gravitational potential is not just a mathematical convenience. It is a foundational concept connecting energy methods, Earth science, satellite technology, and modern theories of gravity.

<h3 style="margin-top:0; color:#000000;">Box Activity 1 — Finding the Gravitational Potential from Newton’s Law</h3>

In this activity, you will derive the **gravitational potential** starting from **Newton’s law of gravity**. The gravitational force between Earth (mass $M$) and an object (mass $m$) separated by a distance $r$ is

$$
F_r = -G\,\frac{Mm}{r^2}.
$$

---

Work through the following:

1. **Observation**  
   The force depends only on the distance $r$ and always points toward Earth’s center.

2. **Key Relationship**  
   For a conservative force in one dimension,

   $$
   F_r = -\frac{dU}{dr}.
   $$

3. **Question**  
   *What potential energy function $U(r)$ is consistent with this force?*

4. **From Potential Energy to Potential**  
   Once $U(r)$ is known, define the **gravitational potential**

   $$
   \Phi(r) = \frac{U(r)}{m}.
   $$

5. **Reference Choice**  
   Choose a convenient reference level for the potential and explain your choice.

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Starting from

$$
F_r = -\frac{dU}{dr},
$$

substitute the gravitational force:

$$
-\,G\frac{Mm}{r^2} = -\frac{dU}{dr}.
$$

This gives

$$
\frac{dU}{dr} = G\frac{Mm}{r^2}.
$$

Integrating,

$$
U(r) = -G\frac{Mm}{r} + C.
$$

A common and convenient choice is to set

$$
U(\infty) = 0,
$$

which gives $C = 0$.

The gravitational potential is therefore

$$
\Phi(r) = \frac{U(r)}{m} = -G\frac{M}{r}.
$$

**Interpretation:**
- The potential depends only on the **gravitational field**, not on the object.
- The negative sign indicates a **bound interaction**.
- Once $\Phi(r)$ is known, the energy of any mass placed at distance $r$ is found from $U = m\Phi$.
