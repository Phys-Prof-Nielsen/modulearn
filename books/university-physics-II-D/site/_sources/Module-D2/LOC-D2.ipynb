# D2.0 Learning Outcomes — Work, Energy, and Electric Potential
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this module matters

In *University Physics I*, the **work–energy framework** provided an alternative to force-based analysis. Rather than tracking acceleration and motion directly, physical processes were described in terms of **energy transfer**, **potential energy**, and **conservation principles**.

This module begins with a focused review of that framework using **vector notation**, while restricting motion to **one dimension**.  
Work is defined using the general expression

$$
W = \int \vec{F} \cdot d\vec{x},
$$

which in one-dimensional motion reduces to an integral along the direction of motion.  

This review establishes the mathematical and conceptual foundation needed for electrostatics.

The framework is then extended to **electric forces**. We will compute the work done on a charge by an electric field, define **electric potential energy**, and finally introduce **electric potential** as energy per unit charge. This progression mirrors the development of gravitational and elastic potential energy in mechanics.

The culmination of this module is the transition from force-based descriptions to **energy-based field quantities**, which play a central role in later topics such as **capacitors, circuits, and continuous charge distributions**.

By the end of this module, you will be able to:

---

### A. Work and Energy Review (Vector Form, 1D Motion)
*Reconnects mechanical energy concepts using vector notation.*

- **A1.** Define work using the vector line integral

  $$
  W = \int \vec{F} \cdot d\vec{x}.
  $$

- **A2.** Reduce the vector expression for work to one-dimensional motion and interpret its meaning physically.
- **A3.** Interpret work graphically as the **area under a force–position curve**.
- **A4.** Apply the **work–energy theorem** to relate net work to changes in kinetic energy.

---

### B. Conservative Forces and Potential Energy
*Establishes energy as a property of position.*

- **B1.** Identify **conservative forces** and explain why they allow the definition of potential energy.
- **B2.** Relate force and potential energy using the one-dimensional relation

  $$
  F_x = -\frac{dU}{dx}.
  $$

- **B3.** Explain the physical meaning of the negative gradient of potential energy.
- **B4.** Distinguish between **changes in potential energy** and the choice of reference level.

---

### C. Work Done by the Electric Force
*Applies the work–energy framework to electrostatics.*

- **C1.** Describe the electric force as a **conservative force** for static charge configurations.
- **C2.** Write Coulomb’s force in vector form for a point charge interaction,

  $$
  \vec{F}_e = k\,\frac{q_1 q_2}{r^2}\,\hat{r}.
  $$

- **C3.** Compute the work done by the electric force using

  $$
  W = \int \vec{F}_e \cdot d\vec{r}.
  $$

- **C4.** Explain why the work done by the electric force depends only on the initial and final positions.

---

### D. Electric Potential Energy
*Defines energy stored in electric interactions.*

- **D1.** Define electric potential energy in terms of work done by the electric force,

  $$
  \Delta U = - W_{\text{electric}}.
  $$

- **D2.** Calculate changes in electric potential energy for a charge moving under Coulomb’s force.
- **D3.** Interpret electric potential energy as a property of the **charge–field system**, not the test charge alone.
- **D4.** Compare electric potential energy to gravitational potential energy to emphasize structural similarity.

---

### E. Electric Potential
*Introduces energy per unit charge as a field quantity.*

- **E1.** Define electric potential as

  $$
  V = \frac{U}{q}.
  $$

- **E2.** Relate changes in electric potential energy to changes in electric potential,

  $$
  \Delta U = q\,\Delta V.
  $$

- **E3.** Explain electric potential as a **scalar field** defined at every point in space.
- **E4.** Interpret electric potential as an energy landscape created by charge distributions.

---

### Performance Criteria

- Work is computed using **vector dot products** and interpreted physically.
- Electric interactions are analyzed using **energy methods**, not force tracking alone.
- Coulomb’s force is correctly incorporated into work and energy calculations.
- Electric potential is treated as a **field quantity**, independent of test charge.
- Connections between mechanical and electric energy are articulated clearly.
