# D1.0 Learning Outcomes — Electric Charge, Forces, and Fields

## Why this module matters

Electricity and magnetism describe **field interactions** that act **through space**, without direct contact. While these interactions are often introduced in terms of **forces between charges**, this viewpoint is ultimately limited. A deeper and more powerful description emerges when interactions are understood in terms of **fields**—physical quantities that exist at every point in space and time, independent of whether a test object is present.

In this module, you will begin with familiar ideas—**electric charge**, how objects become charged, and how charged objects exert forces on one another. The empirical Coulomb’s law provides a precise quantitative description of these forces for simple systems of point charges. However, forces are treated here as a **derived concept**, not the most fundamental one.

The central goal of this module is to establish the **electric field** as the primary physical object. By first introducing the **gravitational field** as an analogue, you will see how interactions can be described as properties of space itself rather than as direct particle–particle influences. This shift in perspective allows forces to be understood as the result of a charge responding to a pre-existing field.

This emphasis is intentional. In later phases—especially Phase A—electromagnetism is described entirely in terms of **fields**, most notably through **Maxwell’s equations**, which make no direct reference to forces at all. Those equations describe how electric and magnetic fields are generated, and how they evolve. Phase D lays the conceptual groundwork for that transition by reframing electrostatics around fields rather than forces.

Throughout the module, you will practice interpreting mathematical expressions as statements about **field structure, symmetry, and superposition**, building intuition that will carry forward into more advanced treatments of electromagnetism and modern physics.

By the end of this module, you will be able to:

---

### A. Electric Charge and Matter  
*Establishes the physical properties of charge and how matter responds to it.*

- **A1.** Describe electric charge as a fundamental property of matter and state the principles of **charge conservation** and **charge quantization**.  
- **A2.** Distinguish between **conductors** and **insulators** based on the mobility of charge.  
- **A3.** Explain common **charging mechanisms**, including conduction, induction, and polarization.  
- **A4.** Predict qualitative charge distributions and interactions based on material properties and charging processes.

---

### B. Coulomb’s Law and Electric Forces  
*Introduces the quantitative law governing electrostatic interactions.*

- **B1.** State the observation-based **Coulomb’s law** for the magnitude of the electric force between two point charges,

  $$
  F = k \frac{|q_1 q_2|}{r^2}.
  $$

- **B2.** Write Coulomb’s law in **vector form**, identifying both magnitude and direction of the force.  
- **B3.** Determine the direction of electric forces based on charge sign and relative position.  
- **B4.** Apply Coulomb’s law and **vector superposition** to compute the net force on a charge due to a **discrete set of point charges**.

---

### C. The Concept of a Field  
*Reframes interactions as properties of space.*

- **C1.** Explain the motivation for introducing fields rather than treating forces as direct particle–particle interactions.  
- **C2.** Define the **gravitational field** and interpret it as force per unit mass.  
- **C3.** Use the gravitational field as an analogue to motivate the definition of the **electric field**.  
- **C4.** Interpret fields as vector quantities that exist independently of the presence of a test object.

---

### D. Electric Field of Point Charges  
*Connects Coulomb’s law to field descriptions.*

- **D1.** Define the electric field as force per unit charge,

  $$
  \vec{E} = \frac{\vec{F}}{q}.
  $$

- **D2.** Determine the electric field produced by a **single point charge** at a specified location.  
- **D3.** Use **superposition** to calculate the net electric field due to a **system of discrete point charges**.  
- **D4.** Relate electric field direction and magnitude to the underlying charge configuration.

---

### E. Vector Representation and Physical Interpretation  
*Emphasizes meaning over formalism.*

- **E1.** Represent forces and fields using vector notation and component methods in two and three dimensions.  
- **E2.** Translate between mathematical expressions, vector diagrams, and physical descriptions of electric interactions.  
- **E3.** Explain how electric fields provide predictive power independent of the motion or presence of test charges.

---

### Performance Criteria

- Electric interactions are explained using **clear physical reasoning**, not memorized formulas.  
- Forces and fields are treated consistently as **vector quantities**.  
- Superposition is applied correctly to both forces and fields from multiple charges.  
- Mathematical expressions are interpreted in terms of **physical meaning and assumptions**.  
- The electric field is understood as a **foundational concept**, not merely a computational tool.
