```python
import numpy as np

k = 8.99e9
q1 = 2.1e-3
q2 = 1.5e-3

r1 = np.array([2.0, -2.0, 0.5])
r2 = np.array([-1.5, 2.3, -1.4])

r21 = r2 - r1
r21_mag = np.sqrt(np.dot(r21,r21))

#r21_mag = np.linalg.norm(r21)
#r21_mag = np.sqrt(r21 @ r21)

F = k*q1*q2 * r21 / (r21_mag**3)
F_mag = np.sqrt(np.dot(F,F))

print(f'Coulomb force on particle 2: {F} N')
print(f'Magnitude of Coulomb force on particle 2: {F_mag:.2e} N')
```
