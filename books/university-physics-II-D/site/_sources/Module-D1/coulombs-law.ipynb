## D1.4.1 Coulomb's Experimental Discovery

Coulomb established his law through careful experiments using a **torsion balance**, studying the interaction between two charged objects. While the objects themselves were not truly point-like, they were small conducting spheres whose charge could be carefully controlled and redistributed. By systematically adding and removing charge and measuring the resulting forces, Coulomb identified several key features of the electric interaction:

- The force is **proportional to the product of the charges** on the two objects.  
- The force is **inversely proportional to the square of the distance** between them.  
- The force acts **along the straight line** connecting the two objects.  
- The force is **attractive for opposite charges** and **repulsive for like charges**.  
- The strength of the interaction is set by a proportionality constant, now called **Coulomb’s constant**, $k$.

These observations are summarized mathematically by **Coulomb’s Law**.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Coulomb’s Law (Magnitude)</h3>

The magnitude of the electric force acting on charge $q_2$ due to charge $q_1$ is

$$
F = \frac{k\,|q_1|\,|q_2|}{r^2},
$$

where $r$ is the distance between the charges.

</div>

The proportionality constant is given by

$$
k = \frac{1}{4\pi\varepsilon_0},
$$

with numerical values

$$
\varepsilon_0 = 8.85\times10^{-12}\ \mathrm{N^{-1}\,m^{-2}\,C^2},
\qquad
k = 8.99\times10^9\ \mathrm{N\,m^2\,C^{-2}}.
$$

> **Note:** Sometimes the separation distance is written as $d$ instead of $r$ to emphasize that it represents a physical distance. We will use $r$ because it connects naturally to later topics such as **Gauss’s law**, introduced later in the course.

<div style="background-color:#ffe6e6; border-left:6px solid #990000; padding:14px; border-radius:4px; margin-top:16px;">
<h3 style="margin-top:0; color:#000000;">Danger Zone — When Coulomb’s Law Applies</h3>

**Coulomb’s law applies only to point charges or objects that can be treated as point-like.**

If the size, shape, or charge distribution of an object matters—such as for extended conductors, surfaces, or continuous charge distributions—Coulomb’s law in its simple form cannot be applied directly. In those cases, the interaction must be analyzed using **superposition**, **electric fields**, or more advanced tools.

</div>

<h3 style="margin-top:0; color:#000000;">Box Activity 1 — Electric Force in the Hydrogen Atom (Naive Model)</h3>

A ground-state electron in the hydrogen atom is most likely to be found at a distance of

$$
r = 5.29\times10^{-11}\,\mathrm{m}
$$

from the nucleus (a single proton).

Assume:
- the proton is a **point charge** with charge  

  $$
  q_p = +1.602\times10^{-19}\,\mathrm{C},
  $$

- the electron has charge  

  $$
  q_e = -1.602\times10^{-19}\,\mathrm{C},
  $$

- and that Coulomb’s law applies in this simplified, classical picture.

---

Work through the following:

1. **Observation**  
   The electron and proton carry opposite charges and are separated by a very small distance.

2. **Question**  
   *What electric force does the proton exert on the electron?*

3. **Hypothesis**  
   Predict whether the force on the electron is attractive or repulsive, and which direction it points.

4. **Calculation**  
   Use Coulomb’s law to compute the **magnitude** of the force.

5. **Conclusion**  
   State both the **magnitude** and **direction** of the force acting on the electron.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1: Direction**

Because the proton and electron have **opposite charges**, the electric force is **attractive**.  
The force on the electron points **toward the proton**.

---

**Step 2: Magnitude**

Using Coulomb’s law,

$$
F = \frac{k\,|q_p|\,|q_e|}{r^2},
$$

with

$$
k = 8.99\times10^9\ \mathrm{N\,m^2\,C^{-2}},
$$

we find

$$
F = \frac{(8.99\times10^9)(1.602\times10^{-19})^2}{(5.29\times10^{-11})^2}.
$$

Evaluating,

$$
F \approx 8.2\times10^{-8}\,\mathrm{N}.
$$

---

**Final Answer**

The electron experiences an electric force of magnitude

$$
\boxed{F \approx 8.2\times10^{-8}\,\mathrm{N}},
$$

directed **toward the proton**.

---

**Important Note**

This calculation treats the hydrogen atom using a **classical, point-particle model**. In reality, the electron does not follow a classical orbit, and a full description requires **quantum mechanics**. This example is meant only to illustrate the *scale* of electric forces at atomic distances.

</div>
</details>

</div>

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

# -----------------------------
# Given values
# -----------------------------
k = sp.Float("8.99e9")        # N·m^2/C^2
q = sp.Float("1.602e-19")     # C
r = sp.Float("5.29e-11")      # m

# -----------------------------
# Coulomb's law (magnitude)
# -----------------------------
F = k*q**2/r**2

# Nice symbolic display
display(sp.Eq(sp.Symbol("F"), sp.Symbol("k")*sp.Symbol("q")**2/sp.Symbol("r")**2))

# Substitute and display numerical result
display(sp.Eq(sp.Symbol("F"), sp.N(F, 6)))

# Optional: scientific notation explicitly
display(sp.Eq(sp.Symbol("F"), sp.Float(F).evalf(3)))
```

<h3 style="margin-top:0; color:#000000;">Box Activity 2 — Electron Speed in a Classical Circular Orbit (Hydrogen)</h3>

A ground-state electron in the hydrogen atom is most likely to be found at a distance of

$$
r = 5.29\times10^{-11}\,\mathrm{m}
$$

from the nucleus (a single proton).

Assume:
- the proton is a **point charge** with charge  

  $$
  q_p = +1.602\times10^{-19}\,\mathrm{C},
  $$

- the electron has charge  

  $$
  q_e = -1.602\times10^{-19}\,\mathrm{C},
  $$

- and we use a **classical circular-orbit** model.

**HINT:** Think polar coordinates.

---

Work through the following:

1. **Observation**  
   The electron experiences an attractive electric force toward the proton.

2. **Question**  
   *If the electron moves in a circular orbit of radius $r$, what must its speed $v$ be?*

3. **Hypothesis**  
   For a stable circular orbit, the inward (radial) electric force must supply the required centripetal acceleration.

4. **Setup**  
   Newton's 2nd law in polar coordinates (radial motion):

   $$
   \sum f_r = ma_r.
   $$

5. **Conclusion**  
   Solve for $v$ and report a numerical value.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**1) Physics setup (radial force balance)**

For a circular orbit, the radial (inward) net force must be

$$
F_r = -m_e\frac{v^2}{r}.
$$

The attractive electric force is

$$
F_e = -\frac{k\,e^2}{r^2}.
$$

Hence, we have

$$
-\frac{k\,e^2}{r^2} = -\frac{m_e v^2}{r}.
$$

Solve for $v$:

$$
v = \sqrt{\frac{k e^2}{m_e r}}.
$$

---

**2) Numerical evaluation**

Using
- $k = 8.99\times10^9\ \mathrm{N\,m^2\,C^{-2}}$
- $e = 1.602\times10^{-19}\ \mathrm{C}$
- $m_e = 9.11\times10^{-31}\ \mathrm{kg}$
- $r = 5.29\times10^{-11}\ \mathrm{m}$

gives

$$
v \approx 2.2\times10^{6}\ \mathrm{m/s}.
$$

</div>
</details>
</div>
<br>

```python
# DIY Cell
```

```python
%reset -f
import sympy as sp

# -----------------------------
# Define symbols
# -----------------------------
k, e, m_e, r = sp.symbols('k e m_e r', positive=True)

# -----------------------------
# Physics expression
# -----------------------------
# From centripetal force = Coulomb force:
# m_e v^2 / r = k e^2 / r^2
v = sp.sqrt(k*e**2/(m_e*r))

display(sp.Eq(sp.Symbol("v"), v))

# -----------------------------
# Numerical values with units (in comments)
# -----------------------------
values = {
    k: sp.Float("8.99e9"),      # N·m^2/C^2
    e: sp.Float("1.602e-19"),   # C
    m_e: sp.Float("9.11e-31"),  # kg
    r: sp.Float("5.29e-11"),    # m
}

# -----------------------------
# Evaluate speed
# -----------------------------
v_val = v.subs(values)
v_val_num = sp.N(v_val, 3)

display(sp.Eq(sp.Symbol("v"), v_val_num))

print(f"Electron speed ≈ {v_val_num} m/s")
print("Direction: tangential to the circular orbit (perpendicular to radius)")
```

## 1.4.2 Coulomb’s Law in Vector Form

Depending on the physical situation, it is often necessary to use **Coulomb’s law in vector form** rather than relying only on magnitudes. The vector form automatically accounts for the **direction** of the force and correctly handles both attractive and repulsive interactions.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Coulomb’s Law (Vector Form)</h3>

The electric force exerted on charge $q_2$ by charge $q_1$ is

$$
\vec{F} = \frac{k\,q_1 q_2}{r_{21}^2}\,\hat{r}_{21}.
$$

</div>

Here, $\hat{r}_{21}$ is the **unit vector** in the direction of $\vec{r}_{21}$, which is defined as the **directed line segment from the source charge $q_1$ to the test charge $q_2$**:

$$
\vec{r}_{21} = \vec{r}_2 - \vec{r}_1.
$$

This formulation ensures that the force points in the correct direction for both positive and negative charge combinations.

---

We can make the directional dependence even more explicit by using the definition of the unit vector:

$$
\hat{r}_{21} = \frac{\vec{r}_{21}}{r_{21}}.
$$

Substituting this into Coulomb’s law gives an equivalent—but often more useful—expression.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Coulomb’s Law (Vector Form — Expanded)</h3>

$$
\vec{F} = \frac{k\,q_1 q_2}{r_{21}^3}\,\vec{r}_{21}.
$$

</div>

Finally, writing $\vec{r}_{21}$ explicitly in terms of the position vectors of the two charges yields the most general form.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Coulomb’s Law (Position-Vector Form)</h3>

$$
\vec{F}
=
k\,q_1 q_2
\frac{\vec{r}_2 - \vec{r}_1}{\lvert \vec{r}_2 - \vec{r}_1 \rvert^3}.
$$

</div>

This final form is especially powerful when working in two or three dimensions, as it allows forces to be computed directly from known charge positions using vector algebra, as well as when we are dealing with >2 electrical charges.

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">

<h3 style="margin-top:0; color:#000000;">Box Activity 3 — Electric Force Between Two Point Charges (Vector Form)</h3>

A **point charge** of

$$
q_1 = 5.00\,\mathrm{mC}
$$

is located at

$$
(x_1, y_1) = (2.00\,\mathrm{m},\,3.00\,\mathrm{m}).
$$

A second **point charge** of

$$
q_2 = 2.00\,\mathrm{mC}
$$

is located at

$$
(x_2, y_2) = (3.00\,\mathrm{m},\,1.50\,\mathrm{m}).
$$

---

Work through the following:

1. **Observation**  
   Both charges are positive and separated in two-dimensional space.

2. **Question**  
   *What is the electric force acting on the $2.00\,\mathrm{mC}$ charge due to the $5.00\,\mathrm{mC}$ charge?*

3. **Hypothesis**  
   Predict whether the force on the $2.00\,\mathrm{mC}$ charge is attractive or repulsive, and in which general direction it points.

4. **Vector Setup**  
   - Write the position vectors $\vec r_1$ and $\vec r_2$.  
   - Construct the displacement vector  

     $$
     \vec r_{21} = \vec r_2 - \vec r_1.
     $$

5. **Calculation**  
   Use **Coulomb’s law in vector form**

   $$
   \vec F_{21}
   =
   k\,q_1 q_2
   \frac{\vec r_{21}}{|\vec r_{21}|^3}
   $$

   to compute the force on charge $q_2$.

6. **Conclusion**  
   Report the **vector force**, including magnitude, direction, and units.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1: Direction (qualitative)**  
Because both charges are **positive**, the force on $q_2$ is **repulsive** and points *away from* $q_1$.

---

**Step 2: Position and displacement vectors**

$$
\vec r_1 = \langle 2.00,\;3.00\rangle\,\mathrm{m},
\qquad
\vec r_2 = \langle 3.00,\;1.50\rangle\,\mathrm{m}
$$

$$
\vec r_{21} = \vec r_2 - \vec r_1
= \langle 1.00,\;-1.50\rangle\,\mathrm{m}
$$

Magnitude:

$$
|\vec r_{21}| = \sqrt{(1.00)^2 + (-1.50)^2}
= 1.80\,\mathrm{m}.
$$

---

**Step 3: Apply Coulomb’s law**

$$
\vec F_{21}
=
k\,q_1 q_2
\frac{\vec r_{21}}{|\vec r_{21}|^3}
$$

Substitute values:

$$
\vec F_{21}
=
(8.99\times10^9)
(5.00\times10^{-3})
(2.00\times10^{-3})
\frac{\langle 1.00,\,-1.50\rangle}
{(1.80)^3}.
$$

Evaluating:

$$
\vec F_{21}
\approx
\langle 15.3\times 10^3,\,-23.1\times 10^3\rangle\ \mathrm{N}.
$$

---

**Final Answer**

The force on the $2.00\,\mathrm{mC}$ charge due to the $5.00\,\mathrm{mC}$ charge is

$$
\boxed{
\vec F
\approx
(15.4\times 10^3\,\hat{\imath} - 23.0\times 10^3\,\hat{\jmath})\ \mathrm{N}
}
$$

This vector points away from the source charge, consistent with repulsion between like charges.

</div>
</details>

</div>

```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# Input parameters
# ---------------
q1 = 5.00E-3
q2 = 2.00E-3
k = 8.99E9

x1 = 2.00
y1 = 3.00
x2 = 3.00
y2 = 1.50

# components of directed line segment from source charge (1) to test charge (2)
rx = x2 - x1
ry = y2 - y1
# magnitiude of directed line segment from source charge (1) to test charge (2)
r = np.sqrt(rx**2 + ry**2)

# component form of Coulomb's law (Equation 3 in notes above)
Fx = k*q1*q2*rx/r**3
Fy = k*q1*q2*ry/r**3

# combine the components into an array (easier to print, that's all)
F = [Fx,Fy]

print('The force components are: [Fx, Fy] = '+str(F)+' N')
```
