## D4.3.1 Motivation: Why Do We Need Ohm’s Law?

In the previous section, we developed a **local, microscopic description** of electrical conduction. We related the electric field inside a material to the current density through intrinsic material properties such as resistivity and conductivity. While this description is fundamental, it is not always the most practical way to analyze real electrical systems.

In circuits, we are usually not interested in the electric field or current density at every point inside a wire or device. Instead, we want to know **how much current flows**, **what voltage is required**, and **how energy is transferred** between circuit elements.

Ohm’s law provides the essential bridge between these two viewpoints. It connects:
- the **microscopic response of materials** to electric fields, and
- the **macroscopic quantities** used in circuit analysis: voltage, current, and resistance.

By introducing the concept of **resistance**, Ohm’s law allows us to replace complicated internal details of a conductor with a simple, measurable parameter that captures the combined effects of material properties and geometry. This makes it possible to analyze circuits using straightforward algebra rather than field theory.

It is important to recognize that Ohm’s law is **not a universal law of nature**, but a useful approximation that applies when materials respond linearly to applied electric fields. Understanding where Ohm’s law comes from—and when it applies—will help prevent common misconceptions and prepare us for more complex circuit behavior later in the course.

In this section, we derive Ohm’s law from the generalized microscopic description and show how it naturally emerges in simple DC circuits.

## D4.3.2 Ohm’s Law

In the previous section, we introduced the **local** relationship between electric field and current density for materials that respond linearly to an applied field,

$$
\vec{E} = \rho\,\vec{J}.
$$

This equation applies **point by point** inside a material. To use it for circuit analysis, we now connect this microscopic description to the **macroscopic quantities** measured in a circuit.

---

### From Local Fields to Circuit Quantities

Consider a straight, uniform conducting wire segment of length $\Delta x$ and constant cross-sectional area $A$. If the electric field inside the wire is approximately uniform along its length, the **average electric field** can be written as

$$
E = \frac{\Delta V}{\Delta x},
$$

where $\Delta V$ is the potential difference across the segment.

Similarly, if the current density is uniform across the cross-section, we can write

$$
J = \frac{I}{A}.
$$

Substituting these average quantities into the resistivity relation gives

$$
\frac{\Delta V}{\Delta x} = \rho \frac{I}{A}.
$$

Solving for the potential difference,

$$
\Delta V = \frac{\rho\,\Delta x}{A}\, I.
$$

---

### Definition of Resistance

The combination of material and geometric factors appearing in this expression motivates the definition of **resistance**:

$$
R \equiv \frac{\rho\,\Delta x}{A}.
$$

Resistance depends on:
- the **material** (through $\rho$),
- the **length** of the conductor,
- and its **cross-sectional area**.

---

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Ohm’s Law</h3>

For a uniform conducting element made of an ohmic material, the potential difference across the element is related to the current through it by

$$
\Delta V = RI.
$$

**Conceptual Meaning**

Ohm’s law states that when a current **passes through** a resistive element, a potential difference exists **across** that element.  
The voltage represents the energy lost per unit charge as charge moves through the material.
</div>

Materials that exhibit this linear relationship between voltage and current are called **ohmic materials**. Many metals behave approximately ohmically under ordinary conditions, though deviations can occur at high fields, high temperatures, or in non-metallic materials.

---


<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example — Resistance from Material Properties and Current Flow</h3>

A uniform copper wire has length $L = 1.50~\text{m}$ and a circular cross-section of radius $r = 1.0~\text{mm}$. The resistivity of copper is $\rho = 1.7 \times 10^{-8}~\Omega\cdot\text{m}$. The wire is connected across a potential difference of $\Delta V = 6.0~\text{V}$.

Answer the following:

1. Determine the **resistance** of the wire.  
2. Determine the **current** flowing through the wire.  
3. Interpret the **energy lost per coulomb of charge** in the wire.

---

**Step 1: Identify the relevant relationships**

The resistance of a uniform conductor is given by

$$
R = \frac{\rho L}{A},
$$

and the cross-sectional area of the wire is

$$
A = \pi r^2.
$$

Once the resistance is known, the current follows from Ohm’s law,

$$
\Delta V = RI.
$$

---

**Step 2: Compute the cross-sectional area**

$$
A = \pi (1.0 \times 10^{-3}~\text{m})^2
  = 3.14 \times 10^{-6}~\text{m}^2.
$$

---

**Step 3: Compute the resistance**

Substitute the values into the resistance formula:

$$
R = \frac{(1.7 \times 10^{-8}~\Omega\cdot\text{m})(1.50~\text{m})}
         {3.14 \times 10^{-6}~\text{m}^2}
  = 8.1 \times 10^{-3}~\Omega.
$$

---

**Step 4: Compute the current**

Rearrange Ohm’s law to solve for the current:

$$
I = \frac{\Delta V}{R}.
$$

Substitute the values:

$$
I = \frac{6.0~\text{V}}{8.1 \times 10^{-3}~\Omega}
  \approx 7.4 \times 10^{2}~\text{A}.
$$

---

**Step 5: Interpret the energy loss per charge**

A potential difference of $6.0~\text{V}$ means that each coulomb of charge loses

$$
6.0~\text{J}
$$

of electric potential energy as it passes through the wire. This energy is transferred to the lattice as thermal energy due to resistive interactions.

---

**Physical interpretation**

- The resistance depends on both the **material** and the **geometry** of the wire.
- Even a small resistance can lead to a large current if the applied voltage is sufficient.
- The voltage across a resistive element represents **energy lost per unit charge**, independent of the amount of current flowing.

This example highlights how microscopic material properties determine macroscopic circuit behavior.
