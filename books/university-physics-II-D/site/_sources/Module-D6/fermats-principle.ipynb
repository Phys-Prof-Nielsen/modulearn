# D6.5 Fermat’s Principle of Least Time

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.5.1 Motivation: A Deeper Pattern Behind Light’s Path

So far, we have described how light reflects and refracts using **geometric rules**:
- equal angles for reflection,
- Snell’s law for refraction.

But these rules raise a natural question:

**Why does light follow these particular paths?**

Fermat’s Principle provides a single, unifying answer. Rather than treating reflection and refraction as separate laws, it tells us that light behaves in a way that **optimizes** something — specifically, **travel time**.

This idea reveals that ray optics is not just geometry. It is an early example of a much broader theme in physics: **nature follows extremal paths**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.5.2 What Is Fermat’s Principle?

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Fermat’s Principle of Least Time</h3>

The path taken by a ray of light between two points is the path that takes the **least time**.

</div>

Fermat’s Principle is a **variational principle**.

Rather than describing how light moves moment by moment, it:
- considers **all possible paths** between two points,
- and selects the one for which the total travel time is minimized (or more precisely, *extremized*).

This single idea explains:
- straight-line motion in uniform media,
- reflection from surfaces,
- refraction at boundaries between materials.

Fermat’s Principle shows that the rules of ray optics are not arbitrary — they arise from a deeper optimization principle.


<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.5.3 Reflection from the Principle of Least Time

Consider light traveling from point A to point B, reflecting off a flat mirror.

- Light could reflect from **any point** on the mirror.
- Each possible reflection point corresponds to a different total travel time.

Because the speed of light is the same before and after reflection, minimizing time is equivalent to minimizing **total path length**.

When this minimization is carried out geometrically, the result is:

$$
\theta_i = \theta_r
$$

That is, **the angle of incidence equals the angle of reflection**.

➡️ Fermat’s Principle explains *why* the law of reflection exists — not just how to apply it.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.5.4 Refraction from the Principle of Least Time

Now consider light traveling from point A in one medium to point B in another.

- The speed of light is different in the two media:
  - $v_1$ in medium 1,
  - $v_2$ in medium 2.
- The total travel time is the sum of the times spent in each medium.

Since time equals distance divided by speed:

- $t_1 = \dfrac{\ell_1}{v_1}$
- $t_2 = \dfrac{\ell_2}{v_2}$
- $T = t_1 + t_2$

Fermat’s Principle selects the path that minimizes this total time.

Carrying out this minimization leads directly to:

$$
n_1 \sin\theta_1 = n_2 \sin\theta_2
$$

This is **Snell’s Law**.

➡️ Light bends at boundaries because the **fastest path is not the shortest path** when speeds differ.

## D6.5.5 A Glimpse Ahead: Least Action in Mechanics

Fermat’s Principle is one example of a broader class of ideas in physics known as **variational principles**.

In **classical mechanics**, a closely related concept is the **Principle of Least Action**:

> A particle moves between two points in such a way that the **action** is extremized.

The action is defined as:

$$
S = \int L\,dt, \qquad \text{with } L = T - V
$$

where:
- $T$ is kinetic energy,
- $V$ is potential energy.

Just as light minimizes travel time, particles follow paths that extremize the action.

## D6.5.6 The Deepest Connection: All Possible Paths

At an even deeper level, Fermat’s Principle is not fundamental.

<div style="background-color:#FFF7D6; border-left:6px solid #8C6D1F; padding:14px; border-radius:4px;">

<h3 style="margin-top:0; color:#000000;">A Light-Hearted Look at Precision</h3>
In **quantum physics**, light does not choose a single path.

According to the **path-integral picture**:
- a photon explores **all possible paths** between two points,
- each path contributes a quantum amplitude,
- paths far from the least-time path interfere destructively,
- paths near the least-time path interfere constructively.

The familiar ray path of least time emerges as the **dominant classical behavior**.

➡️ Fermat’s Principle is an emergent result of **quantum interference**.

</div>

## D6.5.7 Summary

- Fermat’s Principle states that light follows the path that minimizes travel time.
- From this principle, we obtain:
  - the law of reflection,
  - Snell’s law of refraction.
- Fermat’s Principle is an example of a **variational principle**.
- It connects ray optics to:
  - classical mechanics (least action),
  - quantum physics (sum over all paths).

Ray optics, through Fermat’s Principle, provides a first glimpse into the deep unity underlying physical laws.
