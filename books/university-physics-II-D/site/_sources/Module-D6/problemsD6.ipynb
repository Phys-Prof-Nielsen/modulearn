## Problem D6.1 — Wavelength, Frequency, and Refraction

A monochromatic beam of light travels through **air** and then enters a block of **glass**. The light has frequency

$$
f = 5.00\times 10^{14}\ \text{Hz}
$$

and the index of refraction of the glass is

$$
n = 1.50.
$$

Assume the speed of light in vacuum is

$$
c = 3.00\times 10^{8}\ \text{m/s}.
$$

Answer the following:

1. Compute the **wavelength of the light in air**.  
2. Determine the **speed of light in the glass**.  
3. Compute the **wavelength of the light inside the glass**.  
4. State whether the **frequency**, **wavelength**, and **speed** of the light change when the light:
   - reflects from a mirror,
   - refracts into the glass.
5. Provide a short **physical interpretation** explaining *why* refraction changes some quantities but not others.

Be careful with units.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant relationships**

Speed–frequency–wavelength relation:

$$
v = f\lambda
$$

Index of refraction:

$$
n = \frac{c}{v}
$$

---

**Given**

- $f = 5.00\times 10^{14}\ \text{Hz}$  
- $n = 1.50$  
- $c = 3.00\times 10^{8}\ \text{m/s}$  

---

**1) Wavelength in air**

In air (approximately vacuum), $v \approx c$:

$$
\lambda_{\text{air}} = \frac{c}{f}
= \frac{3.00\times 10^{8}}{5.00\times 10^{14}}
= 6.00\times 10^{-7}\ \text{m}.
$$

$$
\boxed{\lambda_{\text{air}} = 600\ \text{nm}}
$$

---

**2) Speed of light in the glass**

Using the index of refraction:

$$
v_{\text{glass}} = \frac{c}{n}
= \frac{3.00\times 10^{8}}{1.50}
= 2.00\times 10^{8}\ \text{m/s}.
$$

$$
\boxed{v_{\text{glass}} = 2.00\times 10^{8}\ \text{m/s}}
$$

---

**3) Wavelength in the glass**

The frequency does **not** change during refraction, so:

$$
\lambda_{\text{glass}} = \frac{v_{\text{glass}}}{f}
= \frac{2.00\times 10^{8}}{5.00\times 10^{14}}
= 4.00\times 10^{-7}\ \text{m}.
$$

$$
\boxed{\lambda_{\text{glass}} = 400\ \text{nm}}
$$

---

**4) What changes during reflection and refraction?**

- **Reflection from a mirror**
  - Frequency: **unchanged**
  - Speed: **unchanged**
  - Wavelength: **unchanged**

- **Refraction into glass**
  - Frequency: **unchanged**
  - Speed: **decreases**
  - Wavelength: **decreases**

---

**5) Physical interpretation**

- The **frequency** of light is set by the source and must remain continuous across boundaries.
- When light enters a material, interactions with electrons cause the wave to propagate more slowly.
- Because the speed changes while the frequency stays the same, the **wavelength must change**.
- Reflection does not involve entry into a new medium, so no change in speed or wavelength occurs.

This is why refraction bends light and reflection does not.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
c, f, n = sp.symbols('c f n', positive=True)
v, lam_air, lam_glass = sp.symbols('v lambda_air lambda_glass', positive=True)

# -----------------------------
# Given values (as SymPy objects for clean LaTeX)
# -----------------------------
c_val = sp.Float('3.00') * 10**8          # m/s
f_val = sp.Float('5.00') * 10**14         # Hz
n_val = sp.Float('1.50')                  # glass

# -----------------------------
# Equations (symbolic)
# -----------------------------
lam_air_eq   = sp.Eq(lam_air, c / f)
v_glass_eq   = sp.Eq(v, c / n)
lam_glass_eq = sp.Eq(lam_glass, v / f)

# -----------------------------
# Substitute values
# -----------------------------
lam_air_val   = sp.simplify(lam_air_eq.rhs.subs({c: c_val, f: f_val}))
v_glass_val   = sp.simplify(v_glass_eq.rhs.subs({c: c_val, n: n_val}))
lam_glass_val = sp.simplify(lam_glass_eq.rhs.subs({v: v_glass_val, f: f_val}))

# -----------------------------
# Display: symbolic setup + substituted equations (clean LaTeX)
# -----------------------------
display(Math(r"\textbf{Step 1: Wavelength of light in air}"))
display(Math(sp.latex(lam_air_eq)))
display(Math(rf"\lambda_{{\text{{air}}}} = \frac{{{sp.latex(c_val)}}}{{{sp.latex(f_val)}}}"))
display(Math(rf"\lambda_{{\text{{air}}}} = {sp.latex(sp.nsimplify(lam_air_val))}\ \text{{m}}"))

display(Math(r"\textbf{Step 2: Speed of light in glass}"))
display(Math(sp.latex(v_glass_eq)))
display(Math(rf"v_{{\text{{glass}}}} = \frac{{{sp.latex(c_val)}}}{{{sp.latex(n_val)}}}"))
display(Math(rf"v_{{\text{{glass}}}} = {sp.latex(sp.nsimplify(v_glass_val))}\ \text{{m/s}}"))

display(Math(r"\textbf{Step 3: Wavelength of light in glass}"))
display(Math(sp.latex(lam_glass_eq)))
display(Math(rf"\lambda_{{\text{{glass}}}} = \frac{{{sp.latex(v_glass_val)}}}{{{sp.latex(f_val)}}}"))
display(Math(rf"\lambda_{{\text{{glass}}}} = {sp.latex(sp.nsimplify(lam_glass_val))}\ \text{{m}}"))

# -----------------------------
# Numerical results
# -----------------------------
lam_air_num   = float(lam_air_val.evalf())
v_glass_num   = float(v_glass_val.evalf())
lam_glass_num = float(lam_glass_val.evalf())

print("\nResults:")
print(f"  Wavelength in air:     λ_air = {lam_air_num:.2e} m")
print(f"  Speed in glass:        v_glass = {v_glass_num:.2e} m/s")
print(f"  Wavelength in glass:   λ_glass = {lam_glass_num:.2e} m")

print("\nReflection vs Refraction:")
print("• Reflection: frequency, speed, and wavelength remain unchanged.")
print("• Refraction: frequency remains unchanged, but speed and wavelength change.")

print("\nPhysical interpretation:")
print("• The frequency of light is fixed by the source and must remain continuous across boundaries.")
print("• Inside matter, electromagnetic interactions reduce the wave’s propagation speed.")
print("• Because v = fλ and f is constant, a lower speed implies a shorter wavelength.")
print("• Reflection does not involve entering a new medium, so no such changes occur.")
```

## Problem D6.2 — Reflection and Refraction at a Flat Surface

A ray of monochromatic light travels in **air** and strikes the flat surface of a block of **glass**.

- Index of refraction of air:  

  $$
  n_1 = 1.00
  $$

- Index of refraction of glass:  

  $$
  n_2 = 1.50
  $$

The ray strikes the surface at an angle of

$$
25^\circ
$$

**measured with respect to the surface** (not the normal).

Answer the following:

1. Determine the **angle of incidence** (measured from the normal).
2. Determine the **angle of reflection**.
3. Determine the **angle of refraction** inside the glass.
4. State whether the refracted ray bends **toward** or **away from** the normal.
5. Provide a short **physical interpretation** of the result.

Be careful with angle definitions.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1: Convert the given angle to an angle from the normal**

The given angle is measured relative to the **surface**.

Since the normal is perpendicular to the surface:

$$
\theta_i = 90^\circ - 25^\circ = 65^\circ.
$$

---

**Step 2: Angle of reflection**

The law of reflection states:

$$
\theta_i = \theta_r.
$$

So,

$$
\theta_r = 65^\circ.
$$

---

**Step 3: Angle of refraction (Snell’s law)**

Apply Snell’s law:

$$
n_1 \sin\theta_i = n_2 \sin\theta_2.
$$

Substitute values:

$$
(1.00)\sin(65^\circ) = (1.50)\sin\theta_2.
$$

$$
\sin\theta_2 = \frac{\sin(65^\circ)}{1.50}.
$$

Using $\sin(65^\circ) \approx 0.906$:

$$
\sin\theta_2 \approx \frac{0.906}{1.50} \approx 0.604.
$$

So,

$$
\theta_2 = \sin^{-1}(0.604) \approx 37^\circ.
$$

---

**Step 4: Direction of bending**

The refracted angle is **smaller** than the incident angle:

$$
\theta_2 < \theta_i.
$$

Therefore, the ray bends **toward the normal**.

---

**Step 5: Physical interpretation**

- Light enters a medium with a **higher index of refraction**, so its speed decreases.
- Because frequency remains constant, the wavelength decreases.
- The change in speed causes the wavefront to rotate, which bends the ray toward the normal.
- Reflection obeys geometry alone, while refraction reflects a change in propagation speed.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
theta_surface, theta_i, theta_r, theta_t = sp.symbols(
    'theta_surface theta_i theta_r theta_t', real=True
)
n1, n2 = sp.symbols('n1 n2', positive=True)

# -----------------------------
# Given values
# -----------------------------
theta_surface_deg = sp.Integer(25)   # degrees (measured from surface)
n1_val = sp.Float('1.00')            # air
n2_val = sp.Float('1.50')            # glass

# Convert to radians where needed
theta_surface = sp.pi * theta_surface_deg / 180

# -----------------------------
# Step 1: Angle of incidence (from the normal)
# -----------------------------
theta_i_expr = sp.pi/2 - theta_surface
theta_i_val = sp.simplify(theta_i_expr)

# -----------------------------
# Step 2: Angle of reflection
# -----------------------------
theta_r_expr = theta_i_expr

# -----------------------------
# Step 3: Snell’s law for refraction
# -----------------------------
snell_eq = sp.Eq(n1 * sp.sin(theta_i), n2 * sp.sin(theta_t))

theta_t_expr = sp.asin((n1/n2) * sp.sin(theta_i))

theta_t_val = theta_t_expr.subs({
    n1: n1_val,
    n2: n2_val,
    theta_i: theta_i_val
})

# -----------------------------
# Display symbolic steps
# -----------------------------
display(Math(r"\textbf{Step 1: Convert angle from surface to normal}"))
display(Math(
    rf"\theta_i = 90^\circ - {theta_surface_deg}^\circ"
))
display(Math(
    rf"\theta_i = {sp.latex(theta_i_val * 180 / sp.pi)}^\circ"
))

display(Math(r"\textbf{Step 2: Angle of reflection}"))
display(Math(
    r"\theta_r = \theta_i"
))
display(Math(
    rf"\theta_r = {sp.latex(theta_i_val * 180 / sp.pi)}^\circ"
))

display(Math(r"\textbf{Step 3: Angle of refraction (Snell's Law)}"))
display(Math(sp.latex(snell_eq)))

display(Math(
    rf"\sin\theta_t = \frac{{{sp.latex(n1_val)}\sin({sp.latex(theta_i_val)})}}{{{sp.latex(n2_val)}}}"
))

display(Math(
    rf"\theta_t = {sp.latex(sp.nsimplify(theta_t_val * 180 / sp.pi))}^\circ"
))

# -----------------------------
# Numerical results
# -----------------------------
theta_i_num = float(theta_i_val * 180 / sp.pi)
theta_t_num = float(theta_t_val * 180 / sp.pi)

print("\nResults:")
print(f"  Angle of incidence:    θ_i = {theta_i_num:.1f}°")
print(f"  Angle of reflection:   θ_r = {theta_i_num:.1f}°")
print(f"  Angle of refraction:   θ_t = {theta_t_num:.1f}°")

print("\nPhysical interpretation:")
print("• The given angle was measured from the surface, so it must be converted to an angle from the normal.")
print("• Reflection preserves the angle relative to the normal.")
print("• Entering a higher-index medium slows the light, causing it to bend toward the normal.")
print("• This behavior is captured quantitatively by Snell’s law.")
```

## Problem D6.3 — Critical Angle in an Optical Fiber

A plastic optical fiber is designed with:

$$
n_{\text{core}} = 1.48
$$

and a surrounding cladding with:

$$
n_{\text{clad}} = 1.40.
$$

A light ray travels inside the core and strikes the core–cladding boundary at an angle of:

$$
\theta = 68^\circ
$$

measured **with respect to the normal**.

Answer the following:

1. Calculate the **critical angle** $\theta_c$ for the core–cladding interface.
2. Determine whether the ray undergoes **total internal reflection** at $\theta = 68^\circ$.
3. If the ray does **not** undergo total internal reflection, state what happens to the light and why this reduces signal quality in fiber optics.
4. Provide a short **physical interpretation** of why a critical angle exists only when light goes from higher $n$ to lower $n$.

Be careful with angle definitions.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equation**

The critical angle for light going from a higher-index medium to a lower-index medium is:

$$
\sin\theta_c = \frac{n_{\text{clad}}}{n_{\text{core}}}.
$$

---

**Given**

- $n_{\text{core}} = 1.48$  
- $n_{\text{clad}} = 1.40$  
- $\theta = 68^\circ$  

---

**1) Critical angle**

$$
\sin\theta_c = \frac{1.40}{1.48} = 0.946.
$$

$$
\theta_c = \sin^{-1}(0.946) \approx 71^\circ.
$$

$$
\boxed{\theta_c \approx 71^\circ}
$$

---

**2) Check for total internal reflection**

Total internal reflection occurs when:

$$
\theta > \theta_c.
$$

Here:

$$
68^\circ < 71^\circ
$$

So the ray does **not** undergo total internal reflection.

$$
\boxed{\text{No total internal reflection occurs at } 68^\circ.}
$$

---

**3) What happens instead?**

Because $\theta$ is below the critical angle:

- the ray **refracts into the cladding**,
- light energy leaks out of the core,
- the beam weakens as it travels.

This reduces signal quality because less light remains guided in the fiber.

---

**4) Physical interpretation**

A critical angle exists only when light tries to go from **higher index to lower index** ($n_1 > n_2$).

In that case, Snell’s law predicts that increasing $\theta_1$ eventually demands:

$$
\sin\theta_2 > 1,
$$

which is impossible. At that point, refraction cannot occur, so the light reflects entirely.

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
n_core, n_clad = sp.symbols('n_core n_clad', positive=True)
theta_c = sp.symbols('theta_c', real=True)

# -----------------------------
# Given values
# -----------------------------
n_core_val = sp.Float('1.48')
n_clad_val = sp.Float('1.40')
theta_inc_deg = sp.Integer(68)   # incident angle (from normal)

# -----------------------------
# Critical angle equation
# -----------------------------
crit_eq = sp.Eq(sp.sin(theta_c), n_clad / n_core)

theta_c_expr = sp.asin(n_clad / n_core)

theta_c_val = theta_c_expr.subs({
    n_clad: n_clad_val,
    n_core: n_core_val
})

# -----------------------------
# Display symbolic steps
# -----------------------------
display(Math(r"\textbf{Step 1: Critical angle condition}"))
display(Math(sp.latex(crit_eq)))

display(Math(
    rf"\sin\theta_c = \frac{{{sp.latex(n_clad_val)}}}{{{sp.latex(n_core_val)}}}"
))

display(Math(
    rf"\theta_c = {sp.latex(sp.nsimplify(theta_c_val * 180 / sp.pi))}^\circ"
))

# -----------------------------
# Numerical comparison
# -----------------------------
theta_c_num = float(theta_c_val * 180 / sp.pi)
theta_inc_num = float(theta_inc_deg)

print("\nResults:")
print(f"  Critical angle:        θ_c = {theta_c_num:.1f}°")
print(f"  Incident angle:        θ = {theta_inc_num:.1f}°")

# -----------------------------
# Check for total internal reflection
# -----------------------------
print("\nTotal Internal Reflection Check:")

if theta_inc_num > theta_c_num:
    print("• θ > θ_c  →  Total internal reflection occurs.")
else:
    print("• θ < θ_c  →  No total internal reflection.")
    print("• Light refracts into the cladding and leaks out of the core.")

# -----------------------------
# Physical interpretation
# -----------------------------
print("\nPhysical interpretation:")
print("• A critical angle exists only when light travels from a higher-index medium to a lower-index medium.")
print("• For large incident angles, Snell’s law would require sin(θ₂) > 1, which is impossible.")
print("• In that case, refraction cannot occur and all light is reflected back into the original medium.")
print("• This principle allows optical fibers to guide light efficiently.")
```

## Problem D6.4 — Lateral Shift of Different Wavelengths in a Glass Plate

A flat, parallel-sided **glass plate** of thickness

$$
t = 6.0~\text{mm}
$$

is used in an optical instrument to slightly separate different wavelengths of light.

A beam containing **red light** and **blue light** strikes the plate from air at an angle of

$$
\theta_i = 50^\circ
$$

(measured from the normal).

The refractive indices of the glass are:
- Red light: $n_r = 1.50$
- Blue light: $n_b = 1.53$

Assume the plate is surrounded by air ($n = 1.00$).

Answer the following:

1. Calculate the **angle of refraction** inside the glass for the red and blue light.
2. Determine the **lateral shift** experienced by each wavelength as it passes through the plate.
3. Compute the **difference in lateral shift** between the red and blue light.
4. Explain why this effect can be used to **separate wavelengths** in real optical systems.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Given**

- Plate thickness: $t = 6.0~\text{mm} = 6.0\times 10^{-3}~\text{m}$
- Angle of incidence: $\theta_i = 50^\circ$
- Refractive indices:
  - Red light: $n_r = 1.50$
  - Blue light: $n_b = 1.53$
- Surrounding medium: air ($n = 1.00$)

---

**1) Angles of refraction (Snell’s law)**

At the first surface, light refracts according to Snell’s law:

$$
n_{\text{air}}\sin\theta_i = n\sin\theta_t
$$

For **red light**:

$$
\sin\theta_{t,r} = \frac{\sin(50^\circ)}{1.50} \approx 0.511
\quad \Rightarrow \quad
\theta_{t,r} \approx 30.7^\circ
$$

For **blue light**:

$$
\sin\theta_{t,b} = \frac{\sin(50^\circ)}{1.53} \approx 0.501
\quad \Rightarrow \quad
\theta_{t,b} \approx 30.1^\circ
$$

---

**2) Why a lateral shift occurs (geometric reasoning)**

As the ray enters the glass, it bends and travels through the slab at an angle $\theta_t$ relative to the normal.

At the second surface, the ray refracts again and exits back into air.  
Because the two surfaces are **parallel**, the exiting ray is **parallel** to the incident ray.

However, the ray does **not** exit at the same position it would have if no refraction had occurred.  
Instead, it is displaced sideways by a distance called the **lateral shift**.

This occurs because:
- the ray travels at an angle inside the glass,
- the slab has a finite thickness $t$,
- and the internal path is longer than the straight-through path.

---

**3) Determining the lateral shift from geometry**

Inside the glass, the ray travels a straight-line path at angle $\theta_t$ through a thickness $t$.

Using right-triangle geometry:
- the vertical distance traveled inside the glass is fixed by $t$,
- the horizontal displacement depends on the difference between the incident and refracted directions.

From this geometry, the sideways displacement of the emerging ray is:

$$
\delta = t\,\frac{\sin(\theta_i - \theta_t)}{\cos\theta_t}
$$

This expression follows directly from projecting the internal ray path onto the direction perpendicular to the incident ray.

---

**4) Lateral shift for each wavelength**

For **red light**:

$$
\delta_r = (6.0\times 10^{-3})
\frac{\sin(50^\circ - 30.7^\circ)}{\cos(30.7^\circ)}
\approx 2.41~\text{mm}
$$

For **blue light**:

$$
\delta_b = (6.0\times 10^{-3})
\frac{\sin(50^\circ - 30.1^\circ)}{\cos(30.1^\circ)}
\approx 2.48~\text{mm}
$$

---

**5) Difference in lateral shift**

$$
\Delta\delta = \delta_b - \delta_r \approx 0.07~\text{mm}
$$

---

**6) Physical interpretation**

- Glass exhibits **dispersion**: its refractive index depends on wavelength.
- Blue light refracts more strongly than red light.
- These small angle differences accumulate into measurable lateral shifts.
- All wavelengths exit the slab **parallel**, but at different positions.

This effect is used in optical systems for wavelength separation, beam alignment, and dispersion measurements.

</div>
</details>


---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
theta_i, theta_t = sp.symbols('theta_i theta_t', real=True)
t, n = sp.symbols('t n', positive=True)

# -----------------------------
# Given values
# -----------------------------
theta_i_deg = sp.Integer(50)                  # incident angle (degrees, from normal)
theta_i_val = sp.pi * theta_i_deg / 180       # radians

t_val = sp.Float('6.0e-3')                    # slab thickness (m)

n_red_val  = sp.Float('1.50')
n_blue_val = sp.Float('1.53')

# -----------------------------
# Snell's law (air -> glass)
# -----------------------------
theta_t_expr = sp.asin(sp.sin(theta_i) / n)

theta_t_red  = theta_t_expr.subs({theta_i: theta_i_val, n: n_red_val})
theta_t_blue = theta_t_expr.subs({theta_i: theta_i_val, n: n_blue_val})

# -----------------------------
# Lateral shift formula
# δ = t * sin(θi − θt) / cos(θt)
# -----------------------------
delta_expr = t * sp.sin(theta_i - theta_t) / sp.cos(theta_t)

delta_red  = sp.simplify(delta_expr.subs({
    t: t_val, theta_i: theta_i_val, theta_t: theta_t_red
}))

delta_blue = sp.simplify(delta_expr.subs({
    t: t_val, theta_i: theta_i_val, theta_t: theta_t_blue
}))

delta_diff = sp.simplify(delta_blue - delta_red)

# -----------------------------
# Display symbolic steps
# -----------------------------
display(Math(r"\textbf{Step 1: Angles of refraction (Snell's Law)}"))
display(Math(r"n\sin\theta_t = \sin\theta_i"))

display(Math(rf"\theta_{{t,\text{{red}}}} = {sp.latex(theta_t_red * 180 / sp.pi)}^\circ"))
display(Math(rf"\theta_{{t,\text{{blue}}}} = {sp.latex(theta_t_blue * 180 / sp.pi)}^\circ"))

display(Math(r"\textbf{Step 2: Lateral shift formula}"))
display(Math(r"\delta = t\,\frac{\sin(\theta_i - \theta_t)}{\cos\theta_t}"))

display(Math(rf"\delta_{{\text{{red}}}} = {sp.latex(delta_red)}\ \text{{m}}"))
display(Math(rf"\delta_{{\text{{blue}}}} = {sp.latex(delta_blue)}\ \text{{m}}"))

display(Math(r"\textbf{Step 3: Difference in lateral shift}"))
display(Math(rf"\Delta\delta = {sp.latex(delta_diff)}\ \text{{m}}"))

# -----------------------------
# Numerical results
# -----------------------------
theta_tr_num  = float(theta_t_red.evalf() * 180 / sp.pi)
theta_tb_num  = float(theta_t_blue.evalf() * 180 / sp.pi)

delta_r_num   = float(delta_red.evalf())
delta_b_num   = float(delta_blue.evalf())
delta_diff_num = float(delta_diff.evalf())

print("\nResults:")
print(f"  Red refraction angle:   θ_t,r = {theta_tr_num:.2f}°")
print(f"  Blue refraction angle:  θ_t,b = {theta_tb_num:.2f}°")

print(f"\n  Red lateral shift:      δ_r = {delta_r_num*1e3:.2f} mm")
print(f"  Blue lateral shift:     δ_b = {delta_b_num*1e3:.2f} mm")

print(f"\n  Difference in shift:    Δδ = {delta_diff_num*1e3:.2f} mm")

print("\nPhysical interpretation:")
print("• Glass has a wavelength-dependent refractive index (dispersion).")
print("• Blue light (higher n) bends more and travels a slightly different path.")
print("• Both beams exit parallel, but with different lateral shifts.")
print("• This effect is used for wavelength separation and precision beam steering.")
```

## Problem D6.5 — Imaging an Object onto a Camera Sensor

A digital camera uses a **thin converging lens** to form an image on its CCD/CMOS sensor.

The lens has a focal length of

$$
f = 50~\text{mm}.
$$

A distant object is placed

$$
d_o = 2.5~\text{m}
$$

in front of the camera lens. The object has a height of

$$
h_o = 1.80~\text{m}.
$$

The camera’s image sensor (film/CCD) is fixed inside the camera body.

Answer the following:

1. Determine the **image distance** $d_i$ required to bring the object into sharp focus on the sensor.
2. Calculate the **magnification** of the image.
3. Determine the **height of the image** formed on the sensor.

**Given**

- Focal length: $f = 50~\text{mm} = 0.050~\text{m}$
- Object distance: $d_o = 2.5~\text{m}$
- Object height: $h_o = 1.80~\text{m}$

---

**1) Image distance**

Use the thin-lens equation:

$$
\frac{1}{f} = \frac{1}{d_o} + \frac{1}{d_i}
$$

Substitute values:

$$
\frac{1}{0.050} = \frac{1}{2.5} + \frac{1}{d_i}
$$

$$
\frac{1}{d_i} = 20.0 - 0.40 = 19.6
$$

$$
d_i \approx 0.051~\text{m} = 51~\text{mm}.
$$

---

**2) Magnification**

$$
M = -\frac{d_i}{d_o} = -\frac{0.051}{2.5} \approx -0.020.
$$

---

**3) Image height**

$$
h_i = M\,h_o = (-0.020)(1.80) \approx -0.036~\text{m} = -3.6~\text{cm}.
$$

---

**4) Image characteristics**

- **Image type:** Real (positive $d_i$)
- **Orientation:** Inverted (negative magnification)

This is exactly how images are formed on camera sensors.

---

**5) Physical interpretation**

- Only one specific image distance satisfies the thin-lens equation for a given object distance.
- If the sensor is closer or farther than this distance, light rays do not converge to a point and the image becomes blurred.
- Autofocus systems work by adjusting the lens position so that the image plane coincides precisely with the sensor.

This explains why precise lens positioning is critical in cameras.

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
f, d_o, d_i = sp.symbols('f d_o d_i', positive=True)
h_o, h_i, M = sp.symbols('h_o h_i M', real=True)

# -----------------------------
# Given values (SI units)
# -----------------------------
f_val  = sp.Float('50e-3')     # 50 mm -> m
do_val = sp.Float('2.5')       # m
ho_val = sp.Float('1.80')      # m

# -----------------------------
# Equations (symbolic)
# -----------------------------
thin_lens_eq = sp.Eq(1/f, 1/d_o + 1/d_i)
mag_eq       = sp.Eq(M, -d_i/d_o)
hi_eq        = sp.Eq(h_i, M*h_o)

# -----------------------------
# Solve for image distance
# -----------------------------
di_expr = sp.solve(thin_lens_eq.subs({f: f_val, d_o: do_val}), d_i)[0]
M_expr  = mag_eq.rhs.subs({d_i: di_expr, d_o: do_val})
hi_expr = hi_eq.rhs.subs({M: M_expr, h_o: ho_val})

# -----------------------------
# Display symbolic steps (LaTeX)
# -----------------------------
display(Math(r"\textbf{Step 1: Thin lens equation}"))
display(Math(sp.latex(thin_lens_eq)))

display(Math(rf"\frac{{1}}{{f}} = \frac{{1}}{{d_o}} + \frac{{1}}{{d_i}}"))
display(Math(rf"\frac{{1}}{{{sp.latex(f_val)}}} = \frac{{1}}{{{sp.latex(do_val)}}} + \frac{{1}}{{d_i}}"))

display(Math(r"\textbf{Step 2: Solve for the image distance}"))
display(Math(rf"d_i = {sp.latex(sp.nsimplify(di_expr))}\ \text{{m}}"))

display(Math(r"\textbf{Step 3: Magnification}"))
display(Math(sp.latex(mag_eq)))
display(Math(rf"M = -\frac{{{sp.latex(sp.nsimplify(di_expr))}}}{{{sp.latex(do_val)}}} = {sp.latex(sp.nsimplify(M_expr))}"))

display(Math(r"\textbf{Step 4: Image height}"))
display(Math(sp.latex(hi_eq)))
display(Math(rf"h_i = ({sp.latex(sp.nsimplify(M_expr))})({sp.latex(ho_val)}) = {sp.latex(sp.nsimplify(hi_expr))}\ \text{{m}}"))

# -----------------------------
# Numerical results
# -----------------------------
di_num = float(di_expr.evalf())
M_num  = float(M_expr.evalf())
hi_num = float(hi_expr.evalf())

print("\nResults:")
print(f"  Image distance:   d_i = {di_num*1e3:.2f} mm")
print(f"  Magnification:    M = {M_num:.4f}")
print(f"  Image height:     h_i = {hi_num*1e2:.2f} cm")

print("\nImage characteristics:")
if di_num > 0:
    print("• d_i > 0 → image is REAL (forms on the sensor side of the lens).")
else:
    print("• d_i < 0 → image is VIRTUAL (would form on the object side).")

if M_num < 0:
    print("• M < 0 → image is INVERTED on the sensor.")
else:
    print("• M > 0 → image is UPRIGHT.")

print("\nPhysical interpretation:")
print("• A camera forms a real image when the object is farther than the focal length.")
print("• The sensor must be located at the image distance d_i for sharp focus.")
print("• Autofocus adjusts lens position so the image plane coincides with the sensor.")
```
