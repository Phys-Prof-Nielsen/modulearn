# D6.6 Thin Lenses

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.6.1 Motivation: Why Thin Lenses Matter

So far, ray optics has focused on how light behaves at **single surfaces** — reflecting from mirrors or refracting at boundaries.  
Thin lenses take the next step by combining **multiple refractions** into a single, powerful optical element.

A thin lens allows us to:
- **redirect and focus light**,
- form **real and virtual images**,
- and control where light converges or diverges.

Thin lenses are everywhere:
- eyeglasses and contact lenses,
- cameras and smartphone lenses,
- microscopes and telescopes,
- the human eye itself.

What makes thin lenses especially valuable is that they let us replace complex wave behavior with **simple ray diagrams and equations**, while still making accurate predictions about image position, size, and orientation.

By studying thin lenses, we learn:
- how curved surfaces shape light paths,
- how multiple refractions work together,
- and how optical systems are designed using geometry rather than trial and error.

Thin-lens optics is where ray optics becomes a practical **tool for imaging**, connecting abstract laws of reflection and refraction to devices we rely on every day.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.6.2 What Are Thin Lenses?

A **lens** is a transparent optical element that bends (refracts) light to form images.

A **thin lens** is one where the lens thickness is much smaller than the radii of curvature of its surfaces — so we can treat all refraction as occurring at a single plane.

There are two main types:
- **Converging lens (convex)**: brings parallel rays **to a point** (focus)
- **Diverging lens (concave)**: spreads parallel rays **away from a point**

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Focal Point</h3>

The **focal point** is the point at which rays initially **parallel to the optical axis** converge (for a **converging lens**) or appear to diverge from (for a **diverging lens**) after passing through the lens.

The **optical axis** is an imaginary line that passes through the center of the lens and is perpendicular to the lens surface. It represents the axis of symmetry of the lens and serves as the reference direction for defining ray angles and image positions.

The focal point lies on the optical axis at a distance equal to the **focal length** from the lens and represents where the lens focuses (or appears to focus) light.
</div>



<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.6.3 Principal Rays and Ray Diagrams

To construct image diagrams, we use three **principal rays**:

### For a converging lens:
1. **Parallel ray**: enters parallel to the axis, exits through the far focal point.
2. **Central ray**: goes through the center of the lens, continues straight.
3. **Focal ray**: enters through the near focal point, exits parallel to the axis.

### For a diverging lens:
1. **Parallel ray**: enters parallel, exits as if it came from the near focal point.
2. **Central ray**: goes through the center, continues straight.
3. **Focal ray**: heads toward the far focal point, exits parallel to the axis.

Where the rays **intersect** (or appear to intersect), the **image** is formed.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.6.4 The Thin Lens Equation

The relationship between focal length, object distance, and image distance is:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px; text-align:center;">
<h3 style="margin-top:0; color:#000000;">Thin Lens Equation</h3>

$$
\frac{1}{f} = \frac{1}{d_o} + \frac{1}{d_i}
$$

</div>

    
Where:
- $f$: focal length (positive for converging, negative for diverging)
- $d_o$: object distance (positive if object is on the incoming side)
- $d_i$: image distance (positive if image is on the outgoing side)

Use **sign conventions** carefully:
- **Converging lens**: $f > 0$
- **Diverging lens**: $f < 0$
- **Real image**: $d_i > 0$
- **Virtual image**: $d_i < 0$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.6.5 Magnification

The size and orientation of the image is given by:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px; text-align:center;">
<h3 style="margin-top:0; color:#000000;">Magnification</h3>

$$
M = \frac{h_i}{h_o} = -\frac{d_i}{d_o}
$$

</div>


Where:
- $M$: magnification
- $h_i$, $h_o$: image and object heights
- $M > 0$: image is **upright**
- $M < 0$: image is **inverted**

If $|M| > 1$, the image is **larger** than the object.  
If $|M| < 1$, the image is **smaller**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.6.6 Real vs. Virtual Images

- **Real image**: formed when rays **actually converge**; can be projected on a screen.
- **Virtual image**: formed when rays **appear to diverge** from a point; cannot be projected.

Examples:
- Converging lens: real image when object is **outside** focal length; virtual if **inside**.
- Diverging lens: always forms **virtual**, **upright**, **smaller** image.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.6.7 Applications of Thin Lenses

- Cameras  
- Microscopes  
- Telescopes  
- Magnifying glasses  
- Eyeglasses and contact lenses

These devices use combinations of converging and diverging lenses to control image formation, size, and position.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">


## D6.6.8 Summary

- **Converging lenses** focus light to a point; **diverging lenses** spread it out.
- The **thin lens equation** relates object distance, image distance, and focal length.
- Use sign conventions and ray diagrams to determine image type.
- Magnification tells us how **big and in what orientation** the image is.

<h3 style="margin-top:0; color:#000000;">Box Activity 1 — Converging Lens: Image Location and Magnification</h3>

An object is placed $d_o = 30~\text{cm}$ in front of a **converging lens** with focal length $f = 10~\text{cm}$.

---

Work through the following:

1. **Image location**  
   Use the thin-lens equation to determine where the image is formed.

2. **Magnification**  
   Calculate the magnification of the image.

3. **Image type**  
   Decide whether the image is **real or virtual**.

4. **Orientation**  
   Determine whether the image is **upright or inverted**.

---

<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1: Identify the relevant relationships**

Thin-lens equation:

$$
\frac{1}{f} = \frac{1}{d_o} + \frac{1}{d_i}
$$

Magnification:

$$
M = -\frac{d_i}{d_o}
$$

---

**Step 2: Solve for the image distance**

Substitute the given values:

$$
\frac{1}{10} = \frac{1}{30} + \frac{1}{d_i}
$$

$$
\frac{1}{d_i} = \frac{1}{10} - \frac{1}{30} = \frac{2}{30}
$$

So,

$$
d_i = 15~\text{cm}.
$$

---

**Step 3: Calculate the magnification**

$$
M = -\frac{15}{30} = -0.50.
$$

---

**Step 4: Interpret the results**

- **Image location:** $15~\text{cm}$ on the opposite side of the lens from the object  
- **Magnification:** $M = -0.50$  
- **Image type:** **Real** (positive image distance)  
- **Orientation:** **Inverted** (negative magnification)

---

**Physical interpretation**

Because the object is placed **outside the focal length** of a converging lens:
- the image forms on the opposite side of the lens,
- the image is real and inverted,
- and the image is smaller than the object.

<h3 style="margin-top:0; color:#000000;">Box Activity 2 — Diverging Lens: Image Location and Characteristics</h3>

An object is placed $d_o = 5~\text{cm}$ in front of a **diverging lens** with focal length $f = -15~\text{cm}$.

---

Work through the following:

1. **Image location**  
   Use the thin-lens equation to determine where the image is formed.

2. **Magnification**  
   Calculate the magnification of the image.

3. **Image type**  
   Decide whether the image is **real or virtual**.

4. **Orientation**  
   Determine whether the image is **upright or inverted**.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1: Identify the relevant relationships**

Thin-lens equation:

$$
\frac{1}{f} = \frac{1}{d_o} + \frac{1}{d_i}
$$

Magnification:

$$
M = -\frac{d_i}{d_o}
$$

---

**Step 2: Solve for the image distance**

Substitute the given values:

$$
\frac{1}{-15} = \frac{1}{5} + \frac{1}{d_i}
$$

$$
\frac{1}{d_i} = -\frac{1}{15} - \frac{1}{5} = -\frac{4}{15}
$$

So,

$$
d_i = -3.75~\text{cm}.
$$

---

**Step 3: Calculate the magnification**

$$
M = -\frac{-3.75}{5} = +0.75.
$$

---

**Step 4: Interpret the results**

- **Image location:** $3.75~\text{cm}$ on the **same side of the lens** as the object  
- **Magnification:** $M = +0.75$  
- **Image type:** **Virtual** (negative image distance)  
- **Orientation:** **Upright** (positive magnification)

---

**Physical interpretation**

For a diverging lens:
- images are always **virtual** for real objects,
- images are **upright**,
- and images are **smaller** than the object.

This behavior is why diverging lenses are used in applications such as corrective lenses for nearsightedness.
