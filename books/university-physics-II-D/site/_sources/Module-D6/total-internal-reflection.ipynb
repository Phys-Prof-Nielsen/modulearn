# D6.4 Total Internal Reflection

## D6.4.1 Motivation: When Light Cannot Escape

In many optical systems, light does not simply refract out of a material — instead, it can become **completely trapped** inside. This phenomenon is called **total internal reflection** (TIR).

Total internal reflection explains:
- why optical fibers can guide light over long distances,
- how prisms redirect light with high efficiency,
- why diamonds sparkle so intensely,
- and how modern technologies transmit information using light.

Unlike ordinary reflection, total internal reflection:
- occurs **only under specific conditions**,
- involves **no loss of light energy** at the boundary,

In this module, we will see how:
- refraction can fail entirely,
- geometry and wave behavior predict this failure,
- and reflection becomes the only possible outcome.

Total internal reflection is one of the most powerful and elegant results of ray optics — and a perfect example of how simple rules lead to striking physical behavior.


<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.4.2 The Critical Angle

Total internal reflection does not occur for all angles.  
It begins only when the angle of incidence exceeds a specific value called the **critical angle**.

---

### When Does Total Internal Reflection Occur?

Total internal reflection can occur **only if**:

1. Light is traveling from a **higher-index medium** to a **lower-index medium**  
   (for example, glass → air).

2. The angle of incidence is **sufficiently large**.

As the angle of incidence increases, the refracted ray bends farther **away from the normal**.  
Eventually, the refracted ray would have to travel *along the boundary itself*.  
This limiting case defines the critical angle.

---

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Critical Angle</h3>

The **critical angle** $\theta_c$ is the angle of incidence (in the higher-index medium) for which the angle of refraction is $90^\circ$.

Using Snell’s law:

$$
n_1 \sin\theta_c = n_2 \sin(90^\circ)
$$

Since $\sin(90^\circ) = 1$, this simplifies to:

$$
\sin\theta_c = \frac{n_2}{n_1}
$$

</div>

---

### Interpreting the Formula

- $n_1$ is the index of refraction of the **initial (denser) medium**
- $n_2$ is the index of refraction of the **second (less dense) medium**
- The formula applies **only when $n_1 > n_2$**

If $n_2 \ge n_1$, the ratio $n_2 / n_1 \ge 1$ and **no critical angle exists**.

---

### What Happens Beyond the Critical Angle?

- For $\theta_1 < \theta_c$:  
  Light refracts out of the material.

- For $\theta_1 = \theta_c$:  
  The refracted ray travels **along the boundary**.

- For $\theta_1 > \theta_c$:  
  Refraction is impossible, and **total internal reflection** occurs.

All of the light reflects back into the original medium.

---

### Key Takeaways

- The critical angle marks the **boundary between refraction and total internal reflection**.
- It arises naturally from **Snell’s law**.
- Total internal reflection is not an approximation — it is a **strict physical result**.

## D6.4.3 Consequences and Applications of Total Internal Reflection

Total internal reflection is not just a theoretical result — it is widely used in science, technology, and everyday devices because it allows light to be redirected with **nearly 100% efficiency**.

When total internal reflection occurs:
- no light energy is transmitted into the second medium,
- the reflection is much more efficient than reflection from ordinary mirrors,
- and the behavior is completely predictable from geometry.

---

### Optical Fibers

One of the most important applications of total internal reflection is **optical fiber technology**.

- Light enters a fiber made of glass or plastic.
- The fiber has a **higher refractive index** than the surrounding material.
- Light strikes the boundary at angles greater than the critical angle.
- As a result, light is repeatedly **totally internally reflected** and guided along the fiber.

This principle allows:
- high-speed internet communication,
- medical imaging (endoscopy),
- and precise light delivery in sensors.

---

### Prisms and Beam Steering

Prisms often use total internal reflection instead of mirrors.

- A prism can redirect light by $90^\circ$ or $180^\circ$.
- The reflection is extremely efficient and does not degrade over time like metallic coatings.
- This is why prisms are used in binoculars, periscopes, and optical instruments.

---

### Diamonds and Sparkle

Diamonds are cut to maximize total internal reflection.

- Diamond has a very high refractive index.
- Light entering a diamond undergoes multiple total internal reflections.
- This traps light inside longer, increasing brightness and sparkle before it exits.

The brilliance of a diamond is therefore a direct consequence of **geometry and refractive index**, not just material beauty.

---

### Beyond Ray Optics

Although total internal reflection appears complete in ray optics, wave optics reveals additional subtleties:

- An **evanescent wave** exists just beyond the boundary.
- No net energy is transmitted across the interface.
- These effects become important in advanced optics and sensing techniques.

For now, ray optics provides an accurate and powerful framework for understanding when and why light becomes trapped.

A laser beam enters the end of a **plastic optical fiber**. The fiber consists of:

Light is guided along the fiber by repeated reflections at the **core–cladding interface**.

**Step 1: Identify the relevant relationship**

The critical angle $\theta_c$ for light traveling from a higher-index medium to a lower-index medium is given by:

$$
\sin\theta_c = \frac{n_{\text{clad}}}{n_{\text{core}}}.
$$

---

**Step 2: Calculate the critical angle**

Substitute the given values:

$$
\sin\theta_c = \frac{1.40}{1.48} \approx 0.946.
$$

So,

$$
\theta_c = \sin^{-1}(0.946) \approx 71^\circ.
$$

---

**Step 3: Determine whether total internal reflection occurs**

The given angle of incidence is:

$$
\theta = 72^\circ.
$$

Since

$$
\theta > \theta_c,
$$

the ray **does undergo total internal reflection** at the core–cladding interface.

---

**Step 4: What if the angle is smaller than the critical angle?**

If $\theta < \theta_c$:

- refraction into the cladding becomes possible,
- light leaks out of the core,
- and the ray is no longer guided along the fiber.

---

**Physical interpretation**

- Optical fibers work because the core has a **higher refractive index** than the cladding.
- Rays that strike the boundary above the critical angle remain trapped by **total internal reflection**.
- This allows light to travel long distances through the fiber with minimal loss.

**Step 1: Calculate the critical angle**

The critical angle for light going from a higher-index medium to a lower-index medium is given by:

$$
\sin\theta_c = \frac{n_{\text{air}}}{n_{\text{diamond}}}.
$$

Substitute the given values:

$$
\sin\theta_c = \frac{1.00}{2.42} \approx 0.413.
$$

So,

$$
\theta_c = \sin^{-1}(0.413) \approx 24.4^\circ.
$$

---

**Step 2: Why light remains trapped**

Because the critical angle is relatively small:

- many internal rays strike the diamond–air boundary at angles **greater than** $\theta_c$,
- refraction becomes impossible,
- and total internal reflection occurs.

This traps light inside the diamond for multiple reflections.

---

**Step 3: Effect of diamond cutting**

A well-cut diamond is designed so that:

- internal rays hit the surfaces at **steep angles** relative to the normal,
- those angles exceed the critical angle,
- light is reflected back toward the top of the diamond rather than escaping through the sides or bottom.

This increases the amount of light that eventually exits through the top, enhancing sparkle and brightness.

---

**Step 4: Physical interpretation**

- Diamonds sparkle not because they generate light, but because they **trap and redirect** it efficiently.
- Total internal reflection allows light to remain inside long enough to undergo many reflections.
- Geometry and refractive index work together to maximize brilliance.
