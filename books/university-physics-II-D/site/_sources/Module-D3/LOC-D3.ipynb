# D3.0 Learning Outcomes — Capacitance and the Parallel-Plate Capacitor
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this module matters

In earlier modules, electric fields and electric potential were introduced as ways to describe how charges interact and how energy is stored in electric fields. These ideas come together naturally in the concept of a **capacitor**.

A capacitor is a physical system designed to **store electric energy** by separating charge. Rather than focusing on individual forces on charges, capacitors emphasize how **geometry and electric fields** determine the relationship between charge, voltage, and stored energy.

This module introduces **capacitance** as a quantitative measure of a system’s ability to store charge per unit potential difference. We begin with the definition of capacitance, then analyze the **simplest and most fundamental example**: the **parallel-plate capacitor**.

By studying this idealized system, we develop clear physical intuition for how electric fields, potential difference, plate separation, and plate area work together. These ideas form the foundation for later topics such as **circuits, dielectrics, energy storage, and electromagnetic devices**.

By the end of this module, you will be able to:

---

### A. Definition of Capacitance  
*Introduces capacitance as a system property.*

- **A1.** Define capacitance as the ratio of charge stored to potential difference,

  $$
  C = \frac{Q}{\Delta V}.
  $$

- **A2.** Explain why capacitance depends on **geometry and materials**, not on the amount of charge stored.
- **A3.** Identify the SI unit of capacitance (the farad) and interpret its physical meaning.
- **A4.** Distinguish between charge, voltage, and capacitance conceptually and mathematically.

---

### B. Parallel-Plate Capacitor Geometry  
*Applies the definition to an idealized physical system.*

- **B1.** Describe the structure of an ideal parallel-plate capacitor and the assumptions used in its analysis.
- **B2.** Explain why the electric field between large, closely spaced plates is approximately uniform.
- **B3.** Relate surface charge density on the plates to the electric field magnitude.
- **B4.** Identify the direction of the electric field relative to plate charge.

---

### C. Electric Field and Potential Difference  
*Connects field concepts to voltage.*

- **C1.** Relate electric field and potential difference for a uniform field using

  $$
  \Delta V = - \int \vec{E} \cdot d\vec{r}.
  $$

- **C2.** Derive the magnitude of the potential difference between parallel plates,

  $$
  \Delta V = E\,d.
  $$

- **C3.** Explain physically how increasing plate separation affects the potential difference.
- **C4.** Distinguish between situations where a capacitor **is** and **is not** connected to a battery.

---

### D. Capacitance of a Parallel-Plate Capacitor  
*Derives capacitance from field relationships.*

- **D1.** Combine field and potential relations to obtain the capacitance of a parallel-plate capacitor,

  $$
  C = \varepsilon_0 \frac{A}{d}.
  $$

- **D2.** Explain how capacitance depends on plate area and separation.
- **D3.** Predict qualitatively how changes in geometry affect stored charge for a fixed voltage.
- **D4.** Interpret capacitance as a measure of a system’s ability to store electric energy.

---

### E. Energy Storage in Capacitors  
*Connects capacitance to energy concepts.*

- **E1.** Express the energy stored in a capacitor in terms of charge and voltage,

  $$
  \Delta U = \frac{1}{2} Q \Delta V.
  $$

- **E2.** Rewrite the stored energy in terms of capacitance and voltage.
- **E3.** Interpret capacitor energy as energy stored in the **electric field**.
- **E4.** Explain how energy storage changes when capacitance or voltage is altered.

---

### Performance Criteria

- Capacitance is correctly defined and interpreted as a **system property**.
- Electric field and potential difference are clearly connected for uniform fields.
- The parallel-plate capacitor formula is derived, not memorized.
- Geometric dependencies of capacitance are explained physically.
- Energy storage in capacitors is described using both mathematical expressions and conceptual reasoning.
