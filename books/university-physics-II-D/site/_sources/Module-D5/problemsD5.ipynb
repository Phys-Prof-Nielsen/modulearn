## Problem D5.1 — Magnetic Force on a Moving Charge

A particle with charge

$$
q = +3.20\times 10^{-19}\ \text{C}
$$

moves through a region of uniform magnetic field with speed

$$
v = 2.50\times 10^{6}\ \text{m/s}.
$$

The magnetic field has magnitude

$$
B = 0.40\ \text{T},
$$

and the particle’s velocity makes an angle of

$$
\theta = 60^\circ
$$

with the direction of the magnetic field.

Answer the following:

1. Compute the **magnitude** of the magnetic force acting on the particle.  
2. State whether the magnetic force would be **zero, maximum, or intermediate** if the particle instead moved parallel to the magnetic field.  
3. Provide a short **physical interpretation** of what this force does to the particle’s motion.

Be careful with units.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Relevant equation</strong>

The magnitude of the magnetic force on a moving charge is

$$
F_B = |q|\,v\,B\,\sin\theta.
$$

---

<strong>Given</strong>

- $q = 3.20\times 10^{-19}\ \text{C}$  
- $v = 2.50\times 10^{6}\ \text{m/s}$  
- $B = 0.40\ \text{T}$  
- $\theta = 60^\circ$  

---

<strong>1) Magnitude of the magnetic force</strong>

Substitute into the formula:

$$
F_B
= (3.20\times 10^{-19})
  (2.50\times 10^{6})
  (0.40)
  \sin(60^\circ).
$$

Using $\sin(60^\circ)=0.866$,

$$
F_B
= 2.77\times 10^{-13}\ \text{N}.
$$

$$
\boxed{F_B = 2.8\times 10^{-13}\ \text{N}}
$$

---

<strong>2) Special case: motion parallel to the field</strong>

If the particle moves parallel to the magnetic field,

$$
\theta = 0^\circ \quad \Rightarrow \quad \sin\theta = 0,
$$

so

$$
\boxed{F_B = 0}.
$$

The magnetic force would be zero.

---

<strong>3) Physical interpretation</strong>

- The magnetic force depends only on the component of velocity **perpendicular** to the magnetic field.
- Because the force is perpendicular to the particle’s velocity, it can change the **direction** of motion but not the **speed**.
- This is why magnetic fields cause charged particles to curve rather than speed up or slow down.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
q, v, B, theta = sp.symbols('q v B theta', positive=True)
FB = sp.symbols('F_B', positive=True)

# -----------------------------
# Given values (as SymPy objects for clean LaTeX)
# -----------------------------
q_val     = sp.Float('3.20') * 10**(-19)     # C
v_val     = sp.Float('2.50') * 10**(6)       # m/s
B_val     = sp.Float('0.40')                # T
theta_deg = sp.Integer(60)                  # degrees

# Convert degrees -> radians for SymPy trig
theta_val = sp.pi * theta_deg / 180

# -----------------------------
# Equations (symbolic)
# -----------------------------
FB_eq = sp.Eq(FB, sp.Abs(q) * v * B * sp.sin(theta))

# -----------------------------
# Substitute values
# -----------------------------
FB_val = sp.simplify(FB_eq.rhs.subs({q: q_val, v: v_val, B: B_val, theta: theta_val}))

# -----------------------------
# Display: symbolic setup + substituted equations (clean LaTeX)
# -----------------------------
display(Math(r"\textbf{Step 1: Magnetic force magnitude}"))
display(Math(sp.latex(FB_eq)))

display(Math(rf"F_B = ({sp.latex(q_val)})({sp.latex(v_val)})({sp.latex(B_val)})\sin({theta_deg}^\circ)"))

# Show exact trig value if it's nice (sin 60 = sqrt(3)/2)
display(Math(rf"\sin({theta_deg}^\circ) = {sp.latex(sp.sin(theta_val))}"))

display(Math(rf"F_B = {sp.latex(sp.nsimplify(FB_val))}\ \text{{N}}"))

# -----------------------------
# Numerical results (formatted cleanly)
# -----------------------------
FB_num = float(FB_val.evalf())

print("\nResults:")
print(f"  Magnetic force magnitude:  F_B = {FB_num:.2e} N")

print("\nSpecial case check:")
print("• If the particle moves parallel to the magnetic field, then θ = 0° and sin(θ) = 0, so F_B = 0.")

print("\nExplanation:")
print("• The magnetic force magnitude is |q| v B sin(θ).")
print("• Only the component of velocity perpendicular to B contributes to the force.")
print("• Because the force is perpendicular to the velocity, it changes direction of motion but not speed.")
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
q, v, B, theta = sp.symbols('q v B theta', positive=True)
FB = sp.symbols('F_B', positive=True)

# -----------------------------
# Given values (SymPy objects for clean LaTeX)
# -----------------------------
q_val = sp.Float('1.50') * 10**(-19)   # C
v_val = sp.Float('3.00') * 10**(6)     # m/s
B_val = sp.Float('0.50')              # T

# Geometry: v is 30° above +x, B is along +y (90° from +x)
# Angle between them is theta = 90° - 30° = 60°
theta_deg = sp.Integer(60)
theta_val = sp.pi * theta_deg / 180   # radians

# -----------------------------
# Equation (symbolic)
# -----------------------------
FB_eq = sp.Eq(FB, sp.Abs(q) * v * B * sp.sin(theta))

# -----------------------------
# Substitute values
# -----------------------------
FB_val = sp.simplify(FB_eq.rhs.subs({q: q_val, v: v_val, B: B_val, theta: theta_val}))

# -----------------------------
# Display: geometry and symbolic setup
# -----------------------------
display(Math(r"\textbf{Step 1: Determine the angle between } \vec{v} \textbf{ and } \vec{B}"))
display(Math(r"\vec{v}\text{ is }30^\circ\text{ above }+x,\quad \vec{B}\text{ is along }+y\text{ (i.e., }90^\circ\text{ from }+x\text{)}"))
display(Math(r"\theta = 90^\circ - 30^\circ = 60^\circ"))

display(Math(r"\textbf{Step 2: Magnetic force magnitude}"))
display(Math(sp.latex(FB_eq)))
display(Math(rf"F_B = ({sp.latex(q_val)})({sp.latex(v_val)})({sp.latex(B_val)})\sin({theta_deg}^\circ)"))
display(Math(rf"\sin({theta_deg}^\circ) = {sp.latex(sp.sin(theta_val))}"))
display(Math(rf"F_B = {sp.latex(sp.nsimplify(FB_val))}\ \text{{N}}"))

# -----------------------------
# Numerical result
# -----------------------------
FB_num = float(FB_val.evalf())
print("\nResults:")
print(f"  Angle: θ = {int(theta_deg)}°")
print(f"  Magnetic force magnitude: F_B = {FB_num:.2e} N")

# -----------------------------
# Direction (RHR explanation)
# -----------------------------
print("\nDirection (Right-Hand Rule):")
print("• In the xy-plane, rotate from v to B: from 30° above +x toward +y (90°).")
print("• This rotation is counterclockwise, so v × B points along +z (out of the page).")
print("• Since q is positive, F_B points along +z.")
print("• If q were negative, the force direction would reverse to −z (into the page).")
```

```python
%reset -f

import numpy as np

def cross_product(a, b):
    """
    Compute the cross product of two 3D vectors a and b.
    """
    return np.cross(a, b)

# Define the vectors
A = np.array([2, -1, 3])
B = np.array([-4, 0, 1])

# Compute the cross product A × B
A_cross_B = cross_product(A, B)

print("A × B =", A_cross_B)
```

## Problem D5.6 — Auroral Electron Spiraling Along a Magnetic Field Line

Auroral electrons often spiral (move in a helix) along Earth’s magnetic field lines. The spiraling comes from the component of the electron’s velocity **perpendicular** to the magnetic field.

An auroral electron has kinetic energy

$$
K = 1.00\ \text{keV}
$$

and moves in a region where the magnetic field magnitude is

$$
B = 5.0\times 10^{-5}\ \text{T}.
$$

At the instant shown, the electron’s velocity makes a **pitch angle** of

$$
\alpha = 30^\circ
$$

with the magnetic field direction. (So $\alpha$ is the angle between $\vec{v}$ and $\vec{B}$.)

Use:
- electron mass $m_e = 9.11\times 10^{-31}\ \text{kg}$
- electron charge magnitude $e = 1.60\times 10^{-19}\ \text{C}$

Answer the following:

1. Compute the electron’s speed $v$ from its kinetic energy.
2. Compute the perpendicular component $v_\perp = v\sin\alpha$.
3. Find the radius $r$ of the spiral (the “gyroradius”).

---

### Hint (polar-coordinate reminder)

For motion in a plane, the **full acceleration** in polar coordinates is

$$
\vec{a} =
(\ddot r - r\dot\theta^2)\,\hat{r}
+
(r\ddot\theta + 2\dot r \dot\theta)\,\hat{\theta}.
$$

For uniform circular motion in the perpendicular plane:
- $r$ is constant $\Rightarrow \dot r = 0$ and $\ddot r = 0$
- $\dot\theta$ is constant $\Rightarrow \ddot\theta = 0$

so the acceleration becomes purely radial:

$$
\vec{a} = -r\dot\theta^2\,\hat{r}.
$$

That corresponds to centripetal acceleration with magnitude

$$
a = \frac{v_\perp^2}{r}.
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

*Step 1: Speed from kinetic energy*

Convert $1.00\ \text{keV}$ to joules:

$$
K = (1.00\times 10^{3}\ \text{eV})(1.60\times 10^{-19}\ \text{J/eV})
  = 1.60\times 10^{-16}\ \text{J}.
$$

Use $K=\tfrac12 m v^2$:

$$
v = \sqrt{\frac{2K}{m_e}}
  = \sqrt{\frac{2(1.60\times 10^{-16})}{9.11\times 10^{-31}}}
  \approx 1.87\times 10^{7}\ \text{m/s}.
$$

$$
\boxed{v \approx 1.87\times 10^{7}\ \text{m/s}}
$$

---

*Step 2: Perpendicular speed*

$$
v_\perp = v\sin\alpha
= (1.87\times 10^{7})\sin(30^\circ)
= (1.87\times 10^{7})(0.500)
\approx 9.37\times 10^{6}\ \text{m/s}.
$$

$$
\boxed{v_\perp \approx 9.37\times 10^{6}\ \text{m/s}}
$$

---

*Step 3: Use magnetic force as the centripetal force*

Magnetic force magnitude (perpendicular part):

$$
F_B = e\,v_\perp B.
$$

Centripetal force needed for circular motion:

$$
F_c = m_e\frac{v_\perp^2}{r}.
$$

Set $F_B = F_c$:

$$
e\,v_\perp B = m_e\frac{v_\perp^2}{r}.
$$

Cancel one factor of $v_\perp$:

$$
eB = m_e\frac{v_\perp}{r}
\quad\Rightarrow\quad
r = \frac{m_e v_\perp}{eB}.
$$

Now substitute numbers:

$$
r = \frac{(9.11\times 10^{-31})(9.37\times 10^{6})}{(1.60\times 10^{-19})(5.0\times 10^{-5})}
\approx 1.07\ \text{m}.
$$

$$
\boxed{r \approx 1.1\ \text{m}}
$$

---

*Physical interpretation*

- The electron’s path is a helix: motion along $\vec{B}$ continues forward, while the perpendicular motion produces a circle.
- Only $v_\perp$ determines the spiral radius.
- A stronger magnetic field makes the spiral tighter (smaller $r$).

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
K, m, e, B, alpha = sp.symbols('K m e B alpha', positive=True)
v, v_perp, r = sp.symbols('v v_perp r', positive=True)

# -----------------------------
# Given values
# -----------------------------
K_val     = sp.Float('1.00e3') * 1.60e-19      # 1.00 keV -> J
m_val     = sp.Float('9.11e-31')               # kg
e_val     = sp.Float('1.60e-19')               # C
B_val     = sp.Float('5.0e-5')                 # T
alpha_deg = sp.Integer(30)
alpha_val = sp.pi * alpha_deg / 180            # radians

# -----------------------------
# Physics equations
# -----------------------------
v_eq      = sp.Eq(v, sp.sqrt(2*K/m))
vperp_eq  = sp.Eq(v_perp, v*sp.sin(alpha))
r_eq      = sp.Eq(r, m*v_perp/(e*B))

# -----------------------------
# Substitute values step-by-step
# -----------------------------
v_val = sp.simplify(v_eq.rhs.subs({K: K_val, m: m_val}))
vperp_val = sp.simplify(vperp_eq.rhs.subs({v: v_val, alpha: alpha_val}))
r_val = sp.simplify(r_eq.rhs.subs({m: m_val, v_perp: vperp_val, e: e_val, B: B_val}))

# -----------------------------
# Display symbolic setup
# -----------------------------
display(Math(r"\textbf{Step 1: Speed from kinetic energy}"))
display(Math(sp.latex(v_eq)))
display(Math(rf"v = {sp.latex(v_val)}\ \text{{m/s}}"))

display(Math(r"\textbf{Step 2: Perpendicular velocity component}"))
display(Math(sp.latex(vperp_eq)))
display(Math(rf"v_\perp = {sp.latex(vperp_val)}\ \text{{m/s}}"))

display(Math(r"\textbf{Step 3: Spiral (gyroradius)}"))
display(Math(sp.latex(r_eq)))
display(Math(rf"r = {sp.latex(sp.nsimplify(r_val))}\ \text{{m}}"))

# -----------------------------
# Numerical results
# -----------------------------
v_num = float(v_val.evalf())
vperp_num = float(vperp_val.evalf())
r_num = float(r_val.evalf())

print("\nNumerical results:")
print(f"  Speed:            v = {v_num:.3e} m/s")
print(f"  Perpendicular v:  v_perp = {vperp_num:.3e} m/s")
print(f"  Spiral radius:    r = {r_num:.2f} m")

print("\nExplanation:")
print("• Only the velocity component perpendicular to the magnetic field contributes to circular motion.")
print("• The magnetic force provides the centripetal acceleration.")
print("• Motion along the field line is unaffected, producing a helical (spiral) trajectory.")
```

## Problem D5.7 — Separation of Uranium Isotopes

### Historical note — The Calutron

The calutron (a contraction of California University Cyclotron) was developed during World War II as part of the Manhattan Project to separate uranium isotopes. It was based directly on earlier mass-spectrometer designs, using magnetic fields to separate ions according to their mass-to-charge ratio. At the Y-12 facility in Oak Ridge, Tennessee, abouth thousand calutrons were operated to enrich uranium-235 from natural uranium, despite the process being energy-intensive and inefficient. Although later replaced by more economical methods such as gas centrifuges, the calutron played a crucial early role in demonstrating that large-scale electromagnetic isotope separation was possible.

A *calutron* separates isotopes by sending **ionized atoms** through a uniform magnetic field. Because different isotopes have different masses, they follow circular paths with slightly different radii.

Suppose uranium atoms are **singly ionized** (charge magnitude $q = e$) and accelerated from rest through a potential difference

$$
\Delta V = 10.0\ \text{kV}.
$$

They then enter a region with a uniform magnetic field (perpendicular to their velocity) of magnitude

$$
B = 0.50\ \text{T}.
$$

Use:
- $e = 1.60\times 10^{-19}\ \text{C}$
- $1\ \text{u} = 1.66\times 10^{-27}\ \text{kg}$
- $m_{235} = 235\ \text{u}$ and $m_{238} = 238\ \text{u}$

Answer the following:

1. Find the speed of a singly ionized uranium ion after acceleration through $\Delta V$.
2. Show that the radius of the circular path in the magnetic field can be written as

   $$
   r = \sqrt{\frac{2m\,\Delta V}{q\,B^2}}.
   $$

3. Compute the radii $r_{235}$ and $r_{238}$.
4. If the ions travel through a **half-circle** and hit a collector plate, estimate the separation distance between the two isotopes on the plate.  
   *(Hint: after a half-circle, the horizontal displacement is approximately the diameter $2r$.)*

Be careful with units.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">


**1) Speed after acceleration through $\Delta V$**

A singly ionized uranium ion has charge magnitude $q=e$. Accelerating from rest through a potential difference $\Delta V$ gives kinetic energy:

$$
q\Delta V = \frac{1}{2}mv^2
\quad\Rightarrow\quad
v = \sqrt{\frac{2q\Delta V}{m}}.
$$

Because $m$ differs slightly between isotopes, the speeds differ slightly as well. In practice, we combine this step with the radius step below.

---

**2) Derive the radius formula**

In the magnetic field region, the velocity is perpendicular to $\vec{B}$, so the magnetic force magnitude is

$$
F_B = qvB.
$$

This force provides the centripetal force for circular motion:

$$
F_c = \frac{mv^2}{r}.
$$

Set $F_B = F_c$:

$$
qvB = \frac{mv^2}{r}
\quad\Rightarrow\quad
r = \frac{mv}{qB}.
$$

Substitute $v = \sqrt{\frac{2q\Delta V}{m}}$:

$$
r = \frac{m}{qB}\sqrt{\frac{2q\Delta V}{m}}
= \sqrt{\frac{2m\Delta V}{qB^2}}.
$$

So,

$$
\boxed{r = \sqrt{\frac{2m\Delta V}{qB^2}}.}
$$

---

**3) Compute $r_{235}$ and $r_{238}$**

Convert masses to kg:

$$
m_{235} = 235\,\text{u}, \qquad m_{238} = 238\,\text{u},
\qquad 1\,\text{u} = 1.66\times 10^{-27}\ \text{kg}.
$$

Use $q=e$, $\Delta V = 10.0\ \text{kV} = 1.00\times 10^{4}\ \text{V}$, and $B=0.50\ \text{T}$ in

$$
r = \sqrt{\frac{2m\Delta V}{qB^2}}.
$$

Script outputs:

$$
\boxed{r_{235} \approx 0.4416\ \text{m}}
\qquad
\boxed{r_{238} \approx 0.4445\ \text{m}}.
$$

---

**4) Separation on the collector plate after a half-circle**

After a half-circle, each isotope lands roughly a horizontal distance of about $2r$ from where it entered, so the separation is approximately

$$
\Delta x \approx 2(r_{238}-r_{235}).
$$

Using the script radii:

$$
\Delta x \approx 2(0.4445 - 0.4416)\ \text{m}
\approx 0.00562\ \text{m}
= 5.62\ \text{mm}.
$$

So,

$$
\boxed{\Delta x \approx 5.62\ \text{mm}}.
$$

---

**Physical interpretation**

- The heavier isotope (U-238) has a slightly larger radius because $r\propto \sqrt{m}$.
- Even a small radius difference can lead to measurable separation after traveling a curved path.
- This is the core idea behind electromagnetic isotope separation in a calutron.

</div>
</details>

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
m, q, V, B = sp.symbols('m q V B', positive=True)
r = sp.symbols('r', positive=True)

# -----------------------------
# Constants / given values
# -----------------------------
e_val = sp.Float('1.60e-19')      # C
u_val = sp.Float('1.66e-27')      # kg
V_val = sp.Float('1.00e4')        # V  (10.0 kV)
B_val = sp.Float('0.50')          # T

m235_val = sp.Integer(235) * u_val
m238_val = sp.Integer(238) * u_val

# -----------------------------
# Key equation: radius after accelerating through ΔV, then entering B ⟂ v
# r = sqrt(2 m ΔV / (q B^2))
# -----------------------------
r_eq = sp.Eq(r, sp.sqrt(2*m*V/(q*B**2)))

display(Math(r"\textbf{Radius formula for a calutron (after acceleration through }\Delta V\textbf{)}"))
display(Math(sp.latex(r_eq)))

# -----------------------------
# Compute radii for U-235 and U-238
# -----------------------------
r235_expr = sp.simplify(r_eq.rhs.subs({m: m235_val, V: V_val, q: e_val, B: B_val}))
r238_expr = sp.simplify(r_eq.rhs.subs({m: m238_val, V: V_val, q: e_val, B: B_val}))

display(Math(r"\textbf{U-235 radius}"))
display(Math(rf"m_{{235}} = {sp.latex(m235_val)}\ \text{{kg}}"))
display(Math(rf"r_{{235}} = {sp.latex(r235_expr)}\ \text{{m}}"))

display(Math(r"\textbf{U-238 radius}"))
display(Math(rf"m_{{238}} = {sp.latex(m238_val)}\ \text{{kg}}"))
display(Math(rf"r_{{238}} = {sp.latex(r238_expr)}\ \text{{m}}"))

# -----------------------------
# Separation after a half-circle
# Approximate landing displacement ≈ 2r, so separation ≈ 2(r238 - r235)
# -----------------------------
sep_expr = sp.simplify(2*(r238_expr - r235_expr))

display(Math(r"\textbf{Collector separation after a half-circle (approx.)}"))
display(Math(rf"\Delta x \approx 2(r_{{238}}-r_{{235}}) = {sp.latex(sep_expr)}\ \text{{m}}"))

# -----------------------------
# Numerical results
# -----------------------------
r235_num = float(r235_expr.evalf())
r238_num = float(r238_expr.evalf())
sep_num  = float(sep_expr.evalf())

print("\nNumerical results:")
print(f"  r_235 = {r235_num:.4f} m")
print(f"  r_238 = {r238_num:.4f} m")
print(f"  separation after half-circle: Δx ≈ {sep_num*1000:.2f} mm")
```

## Problem D5.8 — Belle II Track Curvature: Speed and Equivalent Accelerating Potential

**Belle II (at the SuperKEKB electron–positron collider in Tsukuba, Japan)** measures the momentum of charged particles by observing how their tracks curve in a strong solenoidal magnetic field. In the tracking region, a charged particle moving perpendicular to the magnetic field follows (approximately) **circular motion** in the plane perpendicular to $\vec{B}$.

In a reconstructed event, a **proton** track in the transverse ($xy$) plane is observed to have a curvature radius

$$
R = 5.0\ \text{cm}.
$$

The magnetic field is approximately uniform with magnitude

$$
B = 1.5\ \text{T},
$$

and is directed along the $+z$ axis. Assume the proton’s velocity is perpendicular to $\vec{B}$.

Use:
- proton mass $m_p = 1.67\times 10^{-27}\ \text{kg}$
- proton charge magnitude $e = 1.60\times 10^{-19}\ \text{C}$

Answer the following:

1. Use the magnetic force as the centripetal force to find the proton’s speed $v$.
2. Compute the proton’s kinetic energy $K$.
3. Find the **equivalent accelerating potential** $\Delta V$ that would be needed to accelerate a proton from rest to this kinetic energy (so that $e\Delta V = K$).

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

*Step 1: Use magnetic force causing centripetal acceleration*

Because $\vec{v}\perp\vec{B}$, the magnetic force magnitude is

$$
F_B = e\,vB.
$$

For circular motion of radius $R$, the required centripetal force magnitude is

$$
F_c = m_p\frac{v^2}{R}.
$$

Set $F_B = F_c$:

$$
e\,vB = m_p\frac{v^2}{R}.
$$

Cancel one factor of $v$:

$$
eB = m_p\frac{v}{R}
\quad\Rightarrow\quad
v = \frac{eBR}{m_p}.
$$

Now substitute $R = 5.0\ \text{cm} = 5.0\times 10^{-2}\ \text{m}$:

$$
v =
\frac{(1.60\times 10^{-19})(1.5)(5.0\times 10^{-2})}{1.67\times 10^{-27}}
\approx 7.19\times 10^{6}\ \text{m/s}.
$$

$$
\boxed{v \approx 7.19\times 10^{6}\ \text{m/s}}
$$

---

*Step 2: Kinetic energy*

$$
K = \frac{1}{2}m_p v^2
= \frac{1}{2}(1.67\times 10^{-27})(7.19\times 10^{6})^2
\approx 4.31\times 10^{-14}\ \text{J}.
$$

$$
\boxed{K \approx 4.31\times 10^{-14}\ \text{J}}
$$

---

*Step 3: Equivalent accelerating potential*

If the proton were accelerated from rest through a potential difference $\Delta V$,

$$
e\Delta V = K
\quad\Rightarrow\quad
\Delta V = \frac{K}{e}.
$$

$$
\Delta V =
\frac{4.31\times 10^{-14}}{1.60\times 10^{-19}}
\approx 2.69\times 10^{5}\ \text{V}.
$$

$$
\boxed{\Delta V \approx 2.69\times 10^{5}\ \text{V} \;=\; 269\ \text{kV}}
$$

---

*Physical interpretation*

- The magnetic force bends the proton’s path into a circle in the plane perpendicular to $\vec{B}$.
- A tighter curve (smaller $R$) corresponds to a smaller momentum and lower speed for the same particle.
- The “equivalent accelerating potential” is a convenient way to express how much kinetic energy the proton has.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
e, B, R, mp = sp.symbols('e B R mp', positive=True)
v, K, dV = sp.symbols('v K dV', positive=True)

# -----------------------------
# Given values
# -----------------------------
e_val  = sp.Float('1.60e-19')     # C
B_val  = sp.Float('1.5')          # T
R_val  = sp.Float('5.0e-2')       # m (5.0 cm)
mp_val = sp.Float('1.67e-27')     # kg

# -----------------------------
# Equations
# -----------------------------
v_eq  = sp.Eq(v, e*B*R/mp)
K_eq  = sp.Eq(K, sp.Rational(1,2)*mp*v**2)
dV_eq = sp.Eq(dV, K/e)

# -----------------------------
# Substitute values
# -----------------------------
v_val  = sp.simplify(v_eq.rhs.subs({e: e_val, B: B_val, R: R_val, mp: mp_val}))
K_val  = sp.simplify(K_eq.rhs.subs({mp: mp_val, v: v_val}))
dV_val = sp.simplify(dV_eq.rhs.subs({K: K_val, e: e_val}))

# -----------------------------
# Display symbolic steps
# -----------------------------
display(Math(r"\textbf{Step 1: Speed from magnetic curvature}"))
display(Math(sp.latex(v_eq)))
display(Math(rf"v = {sp.latex(v_val)}\ \text{{m/s}}"))

display(Math(r"\textbf{Step 2: Kinetic energy}"))
display(Math(sp.latex(K_eq)))
display(Math(rf"K = {sp.latex(K_val)}\ \text{{J}}"))

display(Math(r"\textbf{Step 3: Equivalent accelerating potential}"))
display(Math(sp.latex(dV_eq)))
display(Math(rf"\Delta V = {sp.latex(dV_val)}\ \text{{V}}"))

# -----------------------------
# Numerical outputs
# -----------------------------
v_num  = float(v_val.evalf())
K_num  = float(K_val.evalf())
dV_num = float(dV_val.evalf())

print("\nNumerical results:")
print(f"  v  = {v_num:.3e} m/s")
print(f"  K  = {K_num:.3e} J")
print(f"  ΔV = {dV_num:.3e} V  ({dV_num/1000:.1f} kV)")
```
