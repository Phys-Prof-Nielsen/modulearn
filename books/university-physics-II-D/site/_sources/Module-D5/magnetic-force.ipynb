# D5.1 Magnetic Force on a Moving Charge
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.1.1 Motivation

In earlier modules, we learned that **electric fields** exert forces on electric charges. Those forces explain phenomena such as electric acceleration, electric potential energy, and electric current.

However, many important physical systems cannot be explained using electric forces alone:
- charged particles curve through space,
- currents interact with magnets,
- forces appear even when electric fields are absent.

These observations point to the existence of a **second type of force** that acts only when charges are **moving**.  
This module introduces that force: the **magnetic force on a charged particle**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.1.2 Historical Context: What Experiments Revealed

Before modern electromagnetic field theory existed, scientists made several key experimental observations:

- A **stationary charge** experiences **no magnetic force**
- A **moving charge** experiences a force near magnets or currents
- The force depends on:
  - the **amount of charge**
  - the **speed of the particle**
  - the **orientation of motion**
- The force is **always perpendicular** to the particle’s velocity

From repeated measurements, the **magnitude** of the magnetic force was found to be:

$$
F_B = |q|\,v\,B\,\sin\theta
$$

where:
- $q$ is the particle’s charge,
- $v$ is its speed,
- $B$ is the magnetic field strength,
- $\theta$ is the angle between the velocity and the magnetic field.

The **direction** of the force was found experimentally using right-hand rules.

These results came **before** the idea of magnetic fields as physical entities — the force was defined first by observation.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.1.3 Defining the Magnetic Force Using the Cross Product

The experimentally observed properties of the magnetic force can be summarized compactly using a **cross product**.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Magnetic Force on a Moving Charge</h3>

The magnetic force on a charged particle moving in a magnetic field is

$$
\vec{F}_B = q\,\vec{v} \times \vec{B}
$$

</div>

This definition automatically encodes:
- the **$\sin\theta$ dependence** of the force,
- the fact that the force is **perpendicular** to both $\vec{v}$ and $\vec{B}$,
- the **direction reversal** for negative charges.

The cross product is not introduced as an abstraction — it is chosen because it matches what experiments show.

### Special Casess of the Magnetic Force

The cross-product form makes important cases immediately clear:

- **Particle at rest ($v = 0$):**

  $$
  \vec{F}_B = 0
  $$

- **Velocity parallel/anti-parallel to the field ($\theta = 0^\circ$ or $180^\circ$):**

  $$
  \vec{F}_B = 0
  $$

- **Velocity perpendicular to the field ($\theta = 90^\circ$):**

  $$
  F_B = |q|vB \quad \text{(maximum force)}
  $$

Because the force is always perpendicular to velocity, the magnetic force:
- can change the **direction** of motion,
- but **cannot change the speed** of the particle.

This immediately explains why magnetic fields cause **curved motion** rather than speeding particles up or slowing them down.

<h3 style="margin-top:0; color:#000000;">Example — Magnetic Force from Geometry and Direction (RHR)</h3>

A proton with charge $q = +e = 1.60 \times 10^{-19}\,\text{C}$ moves through a uniform magnetic field of magnitude $B = 0.80~\text{T}$.

The proton’s speed is $v = 3.0 \times 10^{6}~\text{m/s}$, and its velocity makes an angle of $\theta = 30^\circ$ CCW with the direction of the magnetic field.

Answer the following:

1. Determine the **magnitude** of the magnetic force on the proton.  
2. Determine the **direction** of the magnetic force using the right-hand rule.  

---

<strong>Step 1: Identify the relevant relationship</strong>

The magnitude of the magnetic force on a moving charge is

$$
F_B = |q|\,v\,B\,\sin\theta.
$$

---

<strong>Step 2: Substitute the given values</strong>

$$
F_B = (1.60 \times 10^{-19}\,\text{C})
      (3.0 \times 10^{6}\,\text{m/s})
      (0.80\,\text{T})
      \sin(30^\circ).
$$

Since $\sin(30^\circ) = 0.50$,

$$
F_B = 1.9 \times 10^{-13}\,\text{N}.
$$

---

<strong>Step 3: Determine the direction (Right-Hand Rule)</strong>

- Point your fingers in the direction of the proton’s velocity $\vec{v}$
- Curl your fingers toward the magnetic field direction $\vec{B}$
- Your thumb points in the direction of $\vec{v} \times \vec{B}$

Because the proton is **positively charged**, the magnetic force points in the **same direction** as $\vec{v} \times \vec{B}$.

(If the particle were an electron, the force direction would be opposite.)

Let the velocity and magnetic field vectors reside in plane of the screen you are looking at. Then the RHR dictates that **$\vec{v}\times\vec{B}~$ is into the plane**.

---

<strong>Step 4: Physical interpretation</strong>

- The magnetic force depends on the **geometry** of the motion through $\sin\theta$
- Only the velocity component **perpendicular** to the magnetic field contributes
- The force is perpendicular to the motion, so it can change the particle’s **direction** but not its **speed**

This explains why magnetic fields bend particle trajectories rather than speeding particles up or slowing them down.

<h3 style="margin-top:0; color:#000000;">Example — Magnetic Force Using Vector Components</h3>

An electron moves through a region of uniform magnetic field.

The particle has:
- charge $q = -e = -1.60 \times 10^{-19}~\text{C}$,
- velocity  

  $$
  \vec{v} = (2.0 \times 10^{6})\,\hat{i} + (1.0 \times 10^{6})\,\hat{j}\;\text{m/s},
  $$
  
- magnetic field  

   $$
  \vec{B} = (0.40)\,\hat{k}\;\text{T}.
  $$

Answer the following:

1. Compute the **magnetic force vector** on the electron.  
2. Determine the **direction** of the force in words.

---

<strong>Step 1: Write the magnetic force law</strong>

The magnetic force is given by

$$
\vec{F}_B = q\,\vec{v} \times \vec{B}.
$$

---

<strong>Step 2: Compute the cross product</strong>

Using the determinant form,

$$
\vec{v} \times \vec{B}
=
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
2.0 \times 10^{6} & 1.0 \times 10^{6} & 0 \\
0 & 0 & 0.40
\end{vmatrix}.
$$

Evaluating,

$$
\vec{v} \times \vec{B}
=
(1.0 \times 10^{6})(0.40)\,\hat{i}
-
(2.0 \times 10^{6})(0.40)\,\hat{j}.
$$

So,

$$
\vec{v} \times \vec{B}
=
(4.0 \times 10^{5})\,\hat{i}
-
(8.0 \times 10^{5})\,\hat{j}.
$$

---

<strong>Step 3: Multiply by the charge</strong>

Now multiply by $q = -1.60 \times 10^{-19}~\text{C}$:

$$
\vec{F}_B
=
(-1.60 \times 10^{-19})
\left[
(4.0 \times 10^{5})\,\hat{i}
-
(8.0 \times 10^{5})\,\hat{j}
\right].
$$

This gives

$$
\vec{F}_B
=
(-6.4 \times 10^{-14})\,\hat{i}
+
(1.28 \times 10^{-13})\,\hat{j}\;\text{N}.
$$

---

<strong>Step 4: Interpret the direction</strong>

- The force lies entirely in the **$xy$-plane**.
- The force has a **negative $x$-component** and a **positive $y$-component**.
- Because the charge is negative, the force direction is opposite what a right-hand rule would predict for a positive charge.

---

<strong>Physical interpretation</strong>

- The cross product automatically enforces that the force is perpendicular to the magnetic field.
- Component-based calculations are especially useful when velocity and field are not aligned with coordinate axes.
- Even though the electron’s speed is large, the force remains small because the charge is very small.

This example shows how the Lorentz force law works cleanly and systematically in vector-component form.

  = 1.1 \times 10^{6}
  $$

Thus,

$$
\vec{v} \times \vec{B}
=
0\,\hat{i}
-
(5.5 \times 10^{5})\,\hat{j}
+
(1.1 \times 10^{6})\,\hat{k}.

<strong>Step 3: Multiply by the charge</strong>

$$
q = -4.0~\text{nC} = -4.0 \times 10^{-9}~\text{C}.

$$
\vec{F}_B
=
(-4.0 \times 10^{-9})
\left[

-
(5.5 \times 10^{5})\,\hat{j}
+
(1.1 \times 10^{6})\,\hat{k}

\vec{F}_B
=

-
(4.4 \times 10^{-3})\,\hat{k}
\;\text{N}.

- The magnetic force has **no $x$-component**, but has both $y$- and $z$-components.
- The force points toward **positive $y$** and **negative $z$**.
- If the particle were **positively charged**, the entire force vector would reverse direction.

<strong>Physical interpretation</strong>

- Even when both $\vec{v}$ and $\vec{B}$ have components along all three axes, the cross product enforces perpendicularity.
- Component-by-component evaluation removes ambiguity from right-hand-rule reasoning.
- This approach generalizes directly to arbitrary 3D electromagnetic configurations.

## D5.1.4 Combining Electric and Magnetic Forces: The Lorentz Force

In real physical systems, both electric and magnetic fields may be present simultaneously.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Lorentz Force</h3>

The total force on a charged particle in electric and magnetic fields is

$$
\vec{F} = q(\vec{E} + \vec{v} \times \vec{B})
$$

</div>

This is the **Lorentz force law**, one of the most fundamental equations in electromagnetism.

- The **electric force** acts on charges whether or not they are moving.
- The **magnetic force** acts only on moving charges.
- Together, they determine how charged particles move in all electromagnetic systems.

Everything that follows in magnetism builds on this single force law.

## D5.1.5 Work Done by the Magnetic Force

An important and somewhat surprising property of the magnetic force is that it does **no work** on a charged particle.

To see why, recall the definition of work done by a force:

$$
W = \int \vec{F}\cdot d\vec{r}.
$$

For magnetic forces, the force on a charged particle is

$$
\vec{F}_B = q\,\vec{v} \times \vec{B}.
$$

---

### Geometric argument

- The magnetic force $\vec{F}_B$ is always **perpendicular** to the particle’s velocity $\vec{v}$.
- The particle’s instantaneous displacement $d\vec{r}$ is always **parallel** to $\vec{v}$.

Because the force is perpendicular to the displacement,

$$
\vec{F}_B \cdot d\vec{r} = 0
$$

at every instant.

Therefore,

$$
\boxed{W_B = 0}.
$$

---

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px; margin-top:16px; margin-bottom:16px;">
<h3 style="margin-top:0; color:#000000;">Work Done by a Magnetic Force</h3>

A magnetic force does **no work** on a charged particle because the force is always perpendicular to the particle’s velocity.

</div>

---

### Physical interpretation

- Since no work is done, the magnetic force **cannot change the particle’s kinetic energy**.
- This means the particle’s **speed remains constant**.
- The magnetic force can only change the **direction** of motion.

This explains why magnetic fields:
- bend particle trajectories,
- produce circular or helical motion,
- but never speed particles up or slow them down.

Any change in a charged particle’s speed must come from an **electric force**, not a magnetic one.

---

### Key takeaway

- Electric forces can do work and change kinetic energy.
- Magnetic forces cannot do work and only redirect motion.

This distinction becomes crucial when analyzing particle motion in combined electric and magnetic fields.

## D5.1.6 Important Real-World Applications of the Magnetic Force

The magnetic force on moving charges is not an abstract idea — it explains and enables many real technologies and natural phenomena.

### 1. Electric Motors  
In electric motors, **moving charges in current-carrying wires** experience magnetic forces.  
These forces produce **torques** that cause coils to rotate, converting electrical energy into mechanical motion.

### 2. Particle Accelerators  
Devices such as cyclotrons and synchrotrons use magnetic fields to **bend and guide charged particles** along precise paths while electric fields increase their energy.  
Without magnetic forces, particle beams could not be controlled.

### 3. Mass Spectrometers  
Charged particles moving through magnetic fields follow curved paths whose radii depend on:
- particle mass,
- charge,
- speed.

This allows scientists to **identify chemical elements and isotopes** with extreme precision.

### 4. Velocity Selectors  
By combining electric and magnetic forces, devices can select particles moving at a **specific speed**.

### 5. Medical Imaging and Therapy  
Magnetic forces play a role in:
- guiding charged particles in radiation therapy,
- controlling particle beams in medical accelerators,
- understanding charged-particle motion in strong magnetic fields.

### 6. Earth’s Magnetosphere and Space Physics  
Charged particles from the Sun interact with Earth’s magnetic field, producing:
- auroras,
- radiation belts,
- protection from solar wind.

In all of these cases, particle motion is governed directly by the **Lorentz force**.

**Key takeaway:**  
Any time moving charges curve, rotate, or are guided without changing speed, the magnetic force is at work.
