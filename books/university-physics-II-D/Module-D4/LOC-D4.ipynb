# D4.0 Learning Outcomes — Simple DC Circuits

## Why this module matters

Electric fields and electric potential describe how charges interact and how energy is stored.  
In real physical systems, however, charges are not static — they **move**, transferring energy from one part of a system to another.

This module introduces **electric current** as a quantitative description of charge flow and develops the microscopic and macroscopic ideas needed to understand how materials respond to electric fields. We begin by defining current in terms of moving charges, then introduce **resistivity and conductivity** as intrinsic material properties that govern how easily charges flow.

Building on this foundation, we define **resistance** and derive **Ohm’s law** from the generalized microscopic description. Finally, we assemble these ideas into the **simplest possible circuit**: a single resistive element connected to a source of electromotive force (EMF).

By the end of this module, you will understand how **energy is supplied by a source**, **transported by moving charges**, and **dissipated by resistive elements**, as well as how to correctly interpret **power** in electrical systems. These ideas form the backbone of all later circuit analysis.

By the end of this module, you will be able to:

---

### A. Electric Current  
*Introduces current as charge flow.*

- **A1.** Define average electric current as the rate of charge flow,

  $$
  I = \frac{\Delta Q}{\Delta t}.
  $$

- **A2.** Define instantaneous current as the time derivative of charge.
- **A3.** Explain the concept of **drift velocity** and distinguish it from random thermal motion.
- **A4.** Relate current to microscopic charge motion using

  $$
  I = nqAv_d.
  $$

- **A5.** Define **current density** and relate it to current and cross-sectional area.

---

### B. Resistivity and Conductivity  
*Introduces material response to electric fields.*

- **B1.** Define resistivity $\rho$ and conductivity $\sigma$ as intrinsic material properties.
- **B2.** Explain the physical origin of resistivity in terms of charge–lattice interactions.
- **B3.** Relate current density to electric field using the generalized Ohm’s law,

  $$
  \vec{J} = \sigma \vec{E}.
  $$

- **B4.** Distinguish clearly between **material properties** and **geometric properties**.
- **B5.** Identify SI units of resistivity and conductivity and interpret their physical meaning.

---

### C. Resistance and Ohm’s Law  
*Connects microscopic behavior to circuit-level quantities.*

- **C1.** Define electrical resistance for a uniform conductor,

  $$
  R = \rho \frac{L}{A}.
  $$

- **C2.** Derive the macroscopic form of Ohm’s law,

  $$
  V = IR,
  $$

  from the generalized Ohm’s law.
- **C3.** Explain the conditions under which Ohm’s law is valid.
- **C4.** Distinguish between resistance as a **property of a component** and voltage/current as **circuit variables**.

---

### D. Single-Resistor DC Circuits  
*Applies definitions to a complete physical system.*

- **D1.** Define electromotive force (EMF) as energy supplied per unit charge.
- **D2.** Identify energy gains from sources and energy losses in resistive elements.
- **D3.** Apply Kirchhoff’s loop principle to a single-loop circuit.
- **D4.** Explain physically why current is the same everywhere in a single-series circuit.

---

### E. Power and Energy in Circuits  
*Interprets energy transfer and dissipation.*

- **E1.** Define electrical power as the rate of energy transfer,

  $$
  P = IV.
  $$

- **E2.** Derive equivalent expressions for resistive (dissipative) power,

  $$
  P = I^2 R \quad \text{and} \quad P = \frac{V^2}{R}.
  $$

- **E3.** Distinguish between **power delivered by a source** and **power dissipated by a resistor**.
- **E4.** Interpret resistive power as thermal energy transfer to the surroundings.

---

### Performance Criteria

- Current is correctly defined at both macroscopic and microscopic levels.
- Resistivity and conductivity are identified as intrinsic material properties.
- Ohm’s law is understood as a consequence of microscopic physics, not a definition.
- EMF, voltage, and resistance are clearly distinguished conceptually.
- Power and energy flow in simple DC circuits are described accurately and consistently.
