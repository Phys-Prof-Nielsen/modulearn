## Problem D4.1 — Number of Electrons Passing a Point in a Wire

A steady current of

$$
I = 2.40\ \text{A}
$$

flows through a wire for a time interval of

$$
\Delta t = 35.0\ \text{s}.
$$

Assume the charge of one electron has magnitude

$$
e = 1.60\times 10^{-19}\ \text{C}.
$$

Answer the following:

1. Compute the total charge $\Delta Q$ that passes a fixed point in the wire during this time.
2. Compute the number of electrons $N$ that pass that point during this time.
3. Provide a short physical interpretation of what this result means.

Be careful with units.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Current definition:

$$
I = \frac{\Delta Q}{\Delta t}
\quad \Rightarrow \quad
\Delta Q = I\,\Delta t
$$

Number of electrons:

$$
N = \frac{\Delta Q}{e}
$$

---

**Given**

- $I = 2.40\ \text{A} = 2.40\ \text{C/s}$
- $\Delta t = 35.0\ \text{s}$
- $e = 1.60\times 10^{-19}\ \text{C}$

---

**1) Total charge passing the point**

$$
\Delta Q = I\Delta t = (2.40)(35.0) = 84.0\ \text{C}
$$

$$
\boxed{\Delta Q = 84.0\ \text{C}}
$$

---

**2) Number of electrons**

$$
N = \frac{\Delta Q}{e}
= \frac{84.0}{1.60\times 10^{-19}}
= 5.25\times 10^{20}
$$

$$
\boxed{N = 5.25\times 10^{20}\ \text{electrons}}
$$

---

**3) Physical interpretation**

- A current of only a few amperes corresponds to an enormous number of electrons passing a point each second.
- This does **not** mean the electrons travel quickly along the wire; it means that many charge carriers are already present throughout the wire and drift collectively.
- The current measures a **rate of charge flow** through a cross-section, not the speed of a single electron from the battery to the device.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
I, dt, e = sp.symbols('I dt e', positive=True)
dQ, N = sp.symbols('dQ N', positive=True)

# -----------------------------
# Given values (as SymPy objects for clean LaTeX)
# -----------------------------
I_val  = sp.Float('2.40')            # A
dt_val = sp.Float('35.0')            # s
e_val  = sp.Float('1.60') * 10**(-19)  # C

# -----------------------------
# Equations (symbolic)
# -----------------------------
dQ_eq = sp.Eq(dQ, I*dt)
N_eq  = sp.Eq(N, dQ/e)

# -----------------------------
# Substitute values
# -----------------------------
dQ_val = sp.simplify(dQ_eq.rhs.subs({I: I_val, dt: dt_val}))
N_val  = sp.simplify(N_eq.rhs.subs({dQ: dQ_val, e: e_val}))

# -----------------------------
# Display: symbolic setup + substituted equations (clean LaTeX)
# -----------------------------
display(Math(r"\textbf{Step 1: Total charge passing a point}"))
display(Math(sp.latex(dQ_eq)))
display(Math(rf"\Delta Q = ({sp.latex(I_val)})({sp.latex(dt_val)})"))

display(Math(r"\textbf{Step 2: Number of electrons}"))
display(Math(sp.latex(N_eq)))
display(Math(rf"N = \frac{{\Delta Q}}{{e}} = \frac{{{sp.latex(dQ_val)}}}{{{sp.latex(e_val)}}}"))

# -----------------------------
# Numerical results (formatted cleanly)
# -----------------------------
dQ_num = float(dQ_val.evalf())
N_num  = float(N_val.evalf())

print("\nResults:")
print(f"  Total charge:  ΔQ = {dQ_num:.3g} C")
print(f"  Electrons:       N = {N_num:.3e} electrons")

print("\nExplanation:")
print("• Current is charge flow rate: ΔQ = IΔt.")
print("• The number of electrons is N = ΔQ/e, where e is the charge per electron (magnitude).")
print("• A few amperes corresponds to an enormous number of electrons passing a point each second.")
print("• This does not mean electrons move quickly from the battery to the device—drift speeds are typically very small.")
```

## Problem D4.2 — Current from a Time-Varying Charge

The total charge that has passed through a fixed cross-section of a wire is described by the function

$$
Q(t) = (3.0t^3 - 4.0t^2 + 2.0t)\ \text{C},
$$

where time $t$ is measured in seconds.

Answer the following:

1. Determine the **SI units** of each numerical coefficient in the expression for $Q(t)$.  
2. Find an expression for the **instantaneous current** $I(t)$.  
3. Compute the current at $t = 2.0~\text{s}$.  
4. State whether the current is increasing or decreasing at $t = 2.0~\text{s}$.

Provide brief explanations where appropriate.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant definition**

Instantaneous current is defined as

$$
I(t) = \frac{dQ}{dt}.
$$

---

**1) Units of the coefficients**

Each term in $Q(t)$ must have units of coulombs.

- For $3.0t^3$:

$$
[3.0] = \text{C/s}^3
$$

- For $-4.0t^2$:

$$
[4.0] = \text{C/s}^2
$$

- For $2.0t$:

$$
[2.0] = \text{C/s}
$$

---

**2) Instantaneous current**

Differentiate $Q(t)$ with respect to time:

$$
I(t) = \frac{d}{dt}(3.0t^3 - 4.0t^2 + 2.0t)
$$

$$
I(t) = 9.0t^2 - 8.0t + 2.0
$$

So,

$$
\boxed{I(t) = (9.0t^2 - 8.0t + 2.0)\ \text{A}}
$$

---

**3) Current at $t = 2.0~\text{s}$**

$$
I(2.0) = 9.0(2.0)^2 - 8.0(2.0) + 2.0
$$

$$
I(2.0) = 36.0 - 16.0 + 2.0 = 22.0~\text{A}
$$

---

**4) Is the current increasing or decreasing?**

Compute the time derivative of the current:

$$
\frac{dI}{dt} = \frac{d}{dt}(9.0t^2 - 8.0t + 2.0)
$$

$$
\frac{dI}{dt} = 18.0t - 8.0
$$

At $t = 2.0~\text{s}$,

$$
\frac{dI}{dt} = 18.0(2.0) - 8.0 = 28.0~\text{A/s} > 0
$$

So the current is **increasing** at $t = 2.0~\text{s}$.

---

**Physical interpretation**

- A polynomial charge function implies that charge is accumulating at a changing rate.
- The current is the **slope** of the $Q$ vs. $t$ graph.
- A positive $dI/dt$ means the current itself is increasing with time.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Define symbols
# -----------------------------
t = sp.symbols('t', real=True)
Q, I = sp.symbols('Q I')

# -----------------------------
# Define the charge function
# -----------------------------
Q_t = 3.0*t**3 - 4.0*t**2 + 2.0*t   # coulombs

# -----------------------------
# Physics:
# Instantaneous current:
# I(t) = dQ/dt
# -----------------------------
I_t = sp.diff(Q_t, t)

# -----------------------------
# Evaluate at a specific time
# -----------------------------
t_val = 2.0
I_at_t = I_t.subs(t, t_val)

# -----------------------------
# Display equations cleanly
# -----------------------------
display(Math(r"\textbf{Given charge as a function of time:}"))
display(Math(sp.latex(sp.Eq(Q, Q_t))))

display(Math(r"\textbf{Instantaneous current definition:}"))
display(Math(r"I(t) = \frac{dQ}{dt}"))

display(Math(r"\textbf{Current as a function of time:}"))
display(Math(sp.latex(sp.Eq(I, I_t))))

display(Math(rf"\textbf{{Current evaluated at }} t = {t_val}\ \text{{s}}:"))
display(Math(sp.latex(sp.Eq(I.subs(t, t_val), I_at_t))))

# -----------------------------
# Numerical output (plain text)
# -----------------------------
print("\nNumerical result:")
print(f"  I({t_val:.1f} s) = {float(I_at_t):.3g} A")

print("\nExplanation:")
print("• The instantaneous current is the time derivative of the charge function.")
print("• A polynomial Q(t) leads to a polynomial I(t) of one lower degree.")
print("• Evaluating I(t) at a specific time gives the current at that instant.")
print("• The current corresponds to the slope of the Q vs. t graph at that time.")
```

## Problem D4.3 — Charge from a Time-Varying Current

The electric current through a wire varies with time according to

$$
I(t) = (6.0t^2 - 4.0t + 1.0)\ \text{A},
$$

where time $t$ is measured in seconds.

Assume that the total charge that has passed through the wire is zero at $t = 0$.

Answer the following:

1. Determine the **SI units** of each numerical coefficient in the expression for $I(t)$.  
2. Find an expression for the **total charge** $Q(t)$ that has passed through the wire as a function of time.  
3. Compute the total charge that has passed through the wire by $t = 3.0~\text{s}$.  
4. Briefly interpret the physical meaning of the integration constant.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant definition**

Electric current is defined as

$$
I(t) = \frac{dQ}{dt}.
$$

---

**1) Units of the coefficients**

Each term in $I(t)$ must have units of amperes (C/s).

- For $6.0t^2$:

$$
[6.0] = \text{A/s}^2
$$

- For $-4.0t$:

$$
[4.0] = \text{A/s}
$$

- For $1.0$:

$$
[1.0] = \text{A}
$$

---

**2) Charge as a function of time**

Integrate the current:

$$
Q(t) = \int I(t)\,dt
     = \int (6.0t^2 - 4.0t + 1.0)\,dt
$$

$$
Q(t) = 2.0t^3 - 2.0t^2 + 1.0t + \text{Constant}.
$$

Apply the initial condition $Q(0) = 0$:

$$
\text{Constant} = 0.
$$

So,

$$
\boxed{Q(t) = (2.0t^3 - 2.0t^2 + t)\ \text{C}}
$$

Here, we used $\text{Constant}$ for the integration constant to avoiud confusion with the unit for charge: C - Coulomb.

---

**3) Charge at $t = 3.0~\text{s}$**

$$
Q(3.0) = 2.0(3.0)^3 - 2.0(3.0)^2 + 3.0
$$

$$
Q(3.0) = 54 - 18 + 3 = 39~\text{C}.
$$

---

**4) Physical interpretation**

- The charge is the **area under the $I$ vs. $t$ curve**.
- The integration constant represents the initial amount of charge that has already passed the point.
- Setting $Q(0)=0$ establishes a clear physical reference point for charge flow.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
t = sp.symbols('t', real=True)
I, Q = sp.symbols('I Q')

# -----------------------------
# Given current function
# -----------------------------
I_t = 6.0*t**2 - 4.0*t + 1.0   # amperes

# -----------------------------
# Physics:
# I(t) = dQ/dt  →  Q(t) = ∫ I(t) dt
# -----------------------------
Q_t = sp.integrate(I_t, t)

# Apply initial condition Q(0) = 0
C = sp.solve(Q_t.subs(t, 0), sp.Symbol('C')) if False else 0
Q_t = Q_t.subs(sp.Symbol('C'), 0)

# -----------------------------
# Evaluate at specific time
# -----------------------------
t_val = 3.0
Q_at_t = Q_t.subs(t, t_val)

# -----------------------------
# Display equations cleanly
# -----------------------------
display(Math(r"\textbf{Given current as a function of time:}"))
display(Math(sp.latex(sp.Eq(I, I_t))))

display(Math(r"\textbf{Charge from current:}"))
display(Math(r"I(t) = \frac{dQ}{dt} \quad \Rightarrow \quad Q(t) = \int I(t)\,dt"))

display(Math(sp.latex(sp.Eq(Q, Q_t))))

display(Math(rf"\textbf{{Charge evaluated at }} t = {t_val}\ \text{{s}}:"))
display(Math(sp.latex(sp.Eq(Q.subs(t, t_val), Q_at_t))))

# -----------------------------
# Numerical output
# -----------------------------
print("\nNumerical result:")
print(f"  Q({t_val:.1f} s) = {float(Q_at_t):.3g} C")

print("\nExplanation:")
print("• Charge is obtained by integrating current over time.")
print("• The result represents the area under the I–t curve.")
print("• The integration constant sets the reference for zero charge flow.")
```

## Problem D4.4 — Oscillating Charge on a Circuit Element (Finding the Current)

In a circuit element (for example, the plates of a capacitor), the charge stored on one side is observed to oscillate in time according to

$$
Q(t) = Q_0\cos(\omega t),
$$

with

$$
Q_0 = 12.0~\mu\text{C}
\qquad \text{and} \qquad
\omega = 500~\text{rad/s}.
$$

Assume $t$ is measured in seconds.

Answer the following:

1. Determine the **SI units** of $Q_0$ and $\omega$.
2. Find an expression for the **instantaneous current** $I(t)$ in the circuit element.
3. Determine the **maximum current** $I_{\max}$.
4. Compute the current at $t = 4.00~\text{ms}$.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant definition**

Current is the time rate of change of charge:

$$
I(t) = \frac{dQ}{dt}.
$$

---

**1) Units**

- $Q_0$ is a charge amplitude, so its SI unit is coulombs:

$$
[Q_0] = \text{C}.
$$

- The argument of cosine must be dimensionless, so $\omega t$ is unitless. Since $t$ is seconds:

$$
[\omega] = \text{s}^{-1} \quad (\text{often written as rad/s}).
$$

---

**2) Current as a function of time**

Differentiate $Q(t)$:

$$
I(t) = \frac{d}{dt}\left(Q_0\cos(\omega t)\right)
     = -Q_0\omega \sin(\omega t).
$$

Substitute the given values. First convert $Q_0$:

$$
12.0~\mu\text{C} = 12.0\times 10^{-6}\ \text{C}.
$$

So,

$$
\boxed{I(t) = -(12.0\times 10^{-6})(500)\sin(500t)\ \text{A}}
$$

or

$$
\boxed{I(t) = -(6.00\times 10^{-3})\sin(500t)\ \text{A}.}
$$

---

**3) Maximum current**

Since $|\sin(\omega t)| \le 1$, the maximum magnitude of the current is the amplitude:

$$
\boxed{I_{\max} = Q_0\omega = (12.0\times 10^{-6})(500) = 6.00\times 10^{-3}\ \text{A}.}
$$

---

**4) Current at $t = 4.00~\text{ms}$**

Convert time:

$$
4.00~\text{ms} = 4.00\times 10^{-3}\ \text{s}.
$$

Compute:

$$
I(0.00400) = -(6.00\times 10^{-3})\sin\big(500(0.00400)\big)
          = -(6.00\times 10^{-3})\sin(2.00).
$$

Numerically, $\sin(2.00) \approx 0.909$, so

$$
\boxed{I(4.00~\text{ms}) \approx -5.45\times 10^{-3}\ \text{A}.}
$$

---

**Physical interpretation**

- When the stored charge oscillates, the current oscillates as well because current measures how rapidly charge changes with time.
- The current is **zero** when the charge is at a maximum or minimum (slope of $Q(t)$ is zero).
- The current is **largest** when the charge passes through zero (slope is steepest).

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
t = sp.symbols('t', real=True)
Q0, omega = sp.symbols('Q_0 \omega', positive=True)
Q, I = sp.symbols('Q I')

# -----------------------------
# Given oscillating charge
# -----------------------------
Q_t = Q0 * sp.cos(omega * t)

# -----------------------------
# Physics:
# I(t) = dQ/dt
# -----------------------------
I_t = sp.diff(Q_t, t)

# -----------------------------
# Substitute numerical values
# -----------------------------
Q0_val = 12.0e-6      # C
omega_val = 500       # rad/s
t_val = 4.00e-3       # s

I_t_num = I_t.subs({Q0: Q0_val, omega: omega_val})
I_at_t = I_t_num.subs(t, t_val)

# -----------------------------
# Display symbolic equations only
# -----------------------------
display(Math(r"\textbf{Given oscillating charge:}"))
display(Math(sp.latex(sp.Eq(Q, Q_t))))

display(Math(r"\textbf{Instantaneous current definition:}"))
display(Math(r"I(t) = \frac{dQ}{dt}"))

display(Math(r"\textbf{Current as a function of time:}"))
display(Math(sp.latex(sp.Eq(I, I_t))))

display(Math(r"\textbf{Maximum current magnitude:}"))
display(Math(sp.latex(sp.Eq(sp.Symbol('I_{max}'), Q0*omega))))

# -----------------------------
# Numerical results (plain text)
# -----------------------------
print("\nNumerical results:")
print(f"  I_max = {Q0_val * omega_val:.3e} A")
print(f"  I({t_val:.3e} s) = {float(I_at_t):.3e} A")

print("\nExplanation:")
print("• An oscillating charge implies an oscillating current.")
print("• Current is the time derivative of charge.")
print("• The current amplitude equals Q0·ω.")
print("• The current is largest when the charge changes most rapidly.")
```

## Problem D4.5 — Charge from an Oscillating Current

In a circuit element, the electric current varies with time according to

$$
I(t) = I_0 \sin(\omega t),
$$

where

$$
I_0 = 8.0~\text{mA}
\qquad \text{and} \qquad
\omega = 400~\text{rad/s}.
$$

Assume time $t$ is measured in seconds and that the total charge that has passed through the element is zero at $t = 0$.

Answer the following:

1. Determine the **SI units** of $I_0$ and $\omega$.
2. Find an expression for the **total charge** $Q(t)$ that has passed through the element as a function of time.
3. Determine the **maximum magnitude** of the charge.
4. Compute the charge at $t = 5.00~\text{ms}$.
5. Briefly explain the **phase relationship** between current and charge.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant definition**

Electric current is defined as

$$
I(t) = \frac{dQ}{dt}.
$$

---

**1) Units**

- $I_0$ is a current amplitude:

$$
[I_0] = \text{A}.
$$

- The argument of the sine function must be dimensionless, so

$$
[\omega] = \text{s}^{-1} \quad (\text{rad/s}).
$$

---

**2) Charge as a function of time**

Integrate the current:

$$
Q(t) = \int I(t)\,dt
     = \int I_0 \sin(\omega t)\,dt.
$$

$$
Q(t) = -\frac{I_0}{\omega}\cos(\omega t) + C.
$$

Apply the initial condition $Q(0) = 0$:

$$
0 = -\frac{I_0}{\omega}\cos(0) + C
\quad \Rightarrow \quad
C = \frac{I_0}{\omega}.
$$

Thus,

$$
\boxed{Q(t) = \frac{I_0}{\omega}\big[1 - \cos(\omega t)\big].}
$$

---

**3) Maximum charge**

Since $0 \le 1 - \cos(\omega t) \le 2$,

$$
\boxed{Q_{\max} = \frac{2I_0}{\omega}.}
$$

---

**4) Charge at $t = 5.00~\text{ms}$**

Convert time:

$$
5.00~\text{ms} = 5.00\times 10^{-3}\ \text{s}.
$$

Substitute values:

$$
Q(0.005) = \frac{8.0\times 10^{-3}}{400}
           \big[1 - \cos(2.00)\big].
$$

Numerically,

$$
\boxed{Q(0.005~\text{s}) \approx 2.83\times 10^{-5}\ \text{C}.}
$$

---

**5) Physical interpretation**

- Charge is the **time integral** of current.
- The charge oscillates because the current oscillates.
- The charge lags the current by a phase of $\pi/2$ (90°), since integration shifts phase.
- The amplitude of charge depends on both the current amplitude and the oscillation frequency.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
t = sp.symbols('t', real=True)
I0, omega = sp.symbols('I_0 \omega', positive=True)
Q, I = sp.symbols('Q I')

# -----------------------------
# Given oscillating current
# -----------------------------
I_t = I0 * sp.sin(omega * t)

# -----------------------------
# Physics:
# I(t) = dQ/dt  →  Q(t) = ∫ I(t) dt
# -----------------------------
Q_t_general = sp.integrate(I_t, t)

# Apply initial condition: Q(0) = 0
Q_t = sp.simplify(Q_t_general - Q_t_general.subs(t, 0))

# -----------------------------
# Display symbolic results (clean equations only)
# -----------------------------
display(Math(r"\textbf{Given current as a function of time:}"))
display(Math(sp.latex(sp.Eq(I, I_t))))

display(Math(r"\textbf{Definition of current:}"))
display(Math(r"I(t) = \frac{dQ}{dt}"))

display(Math(r"\textbf{Charge as a function of time (with } Q(0)=0 \textbf{):}"))
display(Math(sp.latex(sp.Eq(Q, Q_t))))

display(Math(r"\textbf{Maximum charge magnitude:}"))
display(Math(sp.latex(sp.Eq(sp.Symbol('Q_{\max}'), 2*I0/omega))))

# -----------------------------
# Numerical evaluation (plain text only)
# -----------------------------
I0_val = 8.0e-3      # A
omega_val = 400      # rad/s
t_val = 5.00e-3      # s

Q_t_num = Q_t.subs({I0: I0_val, omega: omega_val})
Q_at_t = sp.N(Q_t_num.subs(t, t_val), 6)

print("\nNumerical results:")
print("  Q(t) = (I0/ω)[1 − cos(ωt)]")
print(f"  Q({t_val:.3e} s) = {float(Q_at_t):.3e} C")
print(f"  Q_max = {2*I0_val/omega_val:.3e} C")

print("\nExplanation:")
print("• Charge is obtained by integrating current over time.")
print("• An oscillating current produces an oscillating charge.")
print("• The integration constant is fixed by Q(0)=0.")
print("• Charge lags current by 90 degrees (π/2 phase shift).")
```

## Problem D4.6 — Daily Life Circuit: Phone Charger and Cable Heating

You plug your phone into a wall charger. The charger provides an output voltage of

$$
\Delta V = 5.0\ \text{V}.
$$

While charging, the phone draws a current of

$$
I = 2.0\ \text{A}.
$$

The charging cable (both wires together) has a total resistance of

$$
R_{\text{cable}} = 0.30\ \Omega.
$$

Assume steady-state DC behavior.

Answer the following:

1. Compute the **power delivered** by the charger.
2. Compute the **voltage drop** across the cable.
3. Compute the **power loss** (thermal power) in the cable.
4. What fraction (percent) of the charger’s power is lost as heat in the cable?
5. Provide a short physical interpretation: why do higher-quality (thicker) cables tend to run cooler and charge faster?

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Power delivered:

$$
P = I\Delta V
$$

Voltage drop across a resistor (Ohm’s law):

$$
\Delta V_R = IR
$$

Resistive power loss:

$$
P_{\text{loss}} = I^2R
$$

---

**Given**

- $\Delta V = 5.0\ \text{V}$
- $I = 2.0\ \text{A}$
- $R_{\text{cable}} = 0.30\ \Omega$

---

**1) Power delivered by the charger**

$$
P_{\text{charger}} = I\Delta V = (2.0)(5.0) = 10\ \text{W}
$$

$$
\boxed{P_{\text{charger}} = 10\ \text{W}}
$$

---

**2) Voltage drop across the cable**

$$
\Delta V_{\text{cable}} = IR_{\text{cable}} = (2.0)(0.30) = 0.60\ \text{V}
$$

$$
\boxed{\Delta V_{\text{cable}} = 0.60\ \text{V}}
$$

This means the phone effectively receives about $5.0 - 0.60 = 4.4\ \text{V}$.

---

**3) Power loss in the cable**

$$
P_{\text{loss,cable}} = I^2R_{\text{cable}} = (2.0)^2(0.30) = 1.2\ \text{W}
$$

$$
\boxed{P_{\text{loss,cable}} = 1.2\ \text{W}}
$$

---

**4) Percent power lost in the cable**

$$
\%\ \text{lost} = \frac{P_{\text{loss,cable}}}{P_{\text{charger}}}\times 100\%
= \frac{1.2}{10}\times 100\%
= 12\%
$$

$$
\boxed{12\% \text{ of the power is lost as heat in the cable.}}
$$

---

**5) Physical interpretation**

- The cable behaves like a small resistor in series with the phone.
- A larger cable resistance produces a larger voltage drop and larger power loss.
- Thicker, higher-quality cables have lower resistance, so they waste less power as heat and deliver more voltage to the phone, allowing faster charging.

</div>
</details>

---

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
V, I, R = sp.symbols('dV I R', positive=True)
P_charger, V_drop, P_loss, frac = sp.symbols('P_{charger} dV_{cable} P_{loss} f')

# -----------------------------
# Given values
# -----------------------------
V_val = 5.0     # V
I_val = 2.0     # A
R_val = 0.30    # ohm

# -----------------------------
# Physics equations
# -----------------------------
P_charger_expr = sp.Eq(P_charger, I*V)
V_drop_expr    = sp.Eq(V_drop, I*R)
P_loss_expr    = sp.Eq(P_loss, I**2 * R)
frac_expr      = sp.Eq(frac, P_loss / P_charger)

# -----------------------------
# Display symbolic equations (clean)
# -----------------------------
display(Math(r"\textbf{Equations}"))
display(Math(sp.latex(P_charger_expr)))
display(Math(sp.latex(V_drop_expr)))
display(Math(sp.latex(P_loss_expr)))
display(Math(sp.latex(frac_expr)))

# -----------------------------
# Numerical evaluation (plain text only)
# -----------------------------
P_charger_num = float(P_charger_expr.rhs.subs({I: I_val, V: V_val}))
V_drop_num    = float(V_drop_expr.rhs.subs({I: I_val, R: R_val}))
P_loss_num    = float(P_loss_expr.rhs.subs({I: I_val, R: R_val}))
frac_num      = P_loss_num / P_charger_num

print("\nGiven:")
print(f"  ΔV = {V_val:.3g} V")
print(f"  I  = {I_val:.3g} A")
print(f"  R_cable = {R_val:.3g} Ω")

print("\nResults:")
print(f"  Power delivered by charger: P = {P_charger_num:.3g} W")
print(f"  Voltage drop across cable:  ΔV_cable = {V_drop_num:.3g} V")
print(f"  Power loss in cable:        P_loss = {P_loss_num:.3g} W")
print(f"  Fraction lost:              f = {frac_num:.3g}  ({100*frac_num:.2f}%)")

print("\nExplanation:")
print("• The charger delivers power P = IΔV.")
print("• The cable behaves like a resistor: ΔV_drop = IR.")
print("• Heat loss in the cable is P_loss = I^2R, which reduces voltage available to the phone.")
print("• Lower cable resistance means less heating and more efficient charging.")
```

## Problem D4.7 — Resistance of a Copper Wire

Household wiring, extension cords, and device cables use **standard wire gauges** rather than specifying the wire diameter directly. In this problem, you will determine the electrical resistance of a copper wire using its **gauge**, length, and material properties.

---

### A note on wire gauge

In the **American Wire Gauge (AWG)** system:

- The gauge number specifies the **diameter** of the wire.
- **Smaller gauge number → thicker wire → lower resistance**.
- The diameter (or radius) is **not** encoded in a simple formula and is normally obtained from a **standard AWG table**.

👉 **For this problem, you must look up the diameter (or radius) of the given gauge** using a reliable reference (textbook table, manufacturer chart, or trusted online source).

---

### Given

A straight copper wire has:

- Gauge: **AWG 18**
- Length:
  
  $$
  L = 12.0\ \text{m}
  $$

- Resistivity of copper:
  
  $$
  \rho = 1.68\times 10^{-8}\ \Omega\cdot\text{m}
  $$

Assume the wire is uniform and at room temperature.

---

### Tasks

1. Look up the **diameter or radius** of an **AWG 18** copper wire.  
2. Compute the **cross-sectional area** of the wire.  
3. Compute the **electrical resistance** of the wire.  
4. Briefly explain why lower-gauge wires are used for high-current applications.

Be careful with units.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equation**

For a uniform cylindrical wire, resistance is given by

$$
R = \rho\,\frac{L}{A},
$$

where $A$ is the cross-sectional area.

---

**1) Wire diameter (from AWG table)**

For **AWG 18**, the diameter is approximately

$$
d = 1.02\ \text{mm}.
$$

Thus the radius is

$$
r = 0.510\ \text{mm} = 5.10\times 10^{-4}\ \text{m}.
$$

---

**2) Cross-sectional area**

$$
A = \pi r^2
  = \pi(5.10\times 10^{-4})^2
  = 8.17\times 10^{-7}\ \text{m}^2.
$$

---

**3) Electrical resistance**

$$
R = \rho\frac{L}{A}
  = (1.68\times 10^{-8})\frac{12.0}{8.17\times 10^{-7}}
  \approx 0.247\ \Omega.
$$

$$
\boxed{R \approx 0.25\ \Omega}
$$

---

**4) Physical interpretation**

- Thicker (lower-gauge) wires have **larger cross-sectional area**.
- Larger area means **lower resistance**, which reduces:
  - voltage drop,
  - power loss ($I^2R$ heating),
  - overheating and fire risk.
- This is why high-current devices (heaters, power tools, appliances) require **low-gauge wiring**.

</div>
</details>

---


```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
rho, L, r = sp.symbols(r'\rho L r', positive=True)
A, R = sp.symbols('A R', positive=True)

# -----------------------------
# Physics relationships
# -----------------------------
A_expr = sp.Eq(A, sp.pi * r**2)
R_expr = sp.Eq(R, rho * L / A)

# -----------------------------
# Display symbolic equations (clean, general)
# -----------------------------
display(Math(r"\textbf{Wire resistance model}"))
display(Math(sp.latex(A_expr)))
display(Math(sp.latex(R_expr)))

# -----------------------------
# Numerical values
# -----------------------------
rho_val = 1.68e-8        # ohm·m (copper)
L_val   = 12.0           # m

# AWG 18 lookup (students look this up)
d_val = 1.02e-3          # m  (diameter)
r_val = d_val / 2        # m

# -----------------------------
# Substitute numerical values
# -----------------------------
A_val = A_expr.rhs.subs(r, r_val)
R_val = R_expr.rhs.subs({rho: rho_val, L: L_val, A: A_val})

# -----------------------------
# Display substituted expressions (still clean)
# -----------------------------
display(Math(r"\textbf{Substituting wire radius from AWG table:}"))
display(Math(rf"r = {sp.latex(sp.N(r_val,3))}\ \text{{m}}"))

display(Math(r"\textbf{Cross-sectional area:}"))
display(Math(sp.latex(sp.Eq(A, A_val))))

display(Math(r"\textbf{Wire resistance:}"))
display(Math(sp.latex(sp.Eq(R, R_val))))

# -----------------------------
# Numerical output (plain text only)
# -----------------------------
print("\nNumerical results:")
print(f"  Wire radius (AWG 18): r = {r_val:.3e} m")
print(f"  Cross-sectional area: A = {float(A_val):.3e} m^2")
print(f"  Wire resistance:      R = {float(R_val):.3f} Ω")

print("\nExplanation:")
print("• Wire gauge specifies diameter; smaller gauge means thicker wire.")
print("• Resistance decreases with increasing cross-sectional area.")
print("• Lower-gauge (thicker) wires are used for high-current applications")
print("  to reduce voltage drop and I²R heating.")
```

A single-cup coffee machine heats water using an internal **resistive heating element**.

of water (about one cup) from room temperature

T_i = 20^\circ\text{C}

to the brewing temperature

T_f = 95^\circ\text{C}.

Assume:
- The specific heat capacity of water is  

  $$
  c = 4186\ \mathrm{J/(kg·K)}.
  $$
  
- All electrical energy delivered to the heating element goes into heating the water.
- The heating element operates at constant resistance.

---

### Tasks

2. Compute the **power** dissipated by the heating element.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Ohm’s law:

$$
I = \frac{\Delta V}{R}
$$

Electrical power:

$$
P = I\Delta V
$$

Thermal energy:

$$
Q = mc\Delta T
$$

Heating time:

$$
t = \frac{Q}{P}

---

**1) Current through the heating element**

I = \frac{120}{12.0} = 10.0\ \text{A}

$$
\boxed{I = 10.0\ \text{A}}

**2) Power dissipated**

$$
P = I\Delta V = (10.0)(120) = 1200\ \text{W}
$$

$$
\boxed{P = 1.20\times 10^{3}\ \text{W}}
$$

---

**3) Thermal energy required**

Temperature change:

$$
\Delta T = 95 - 20 = 75^\circ\text{C}

$$

\boxed{Q = 7.85\times 10^{4}\ \text{J}}

t = \frac{Q}{P} = \frac{7.85\times 10^{4}}{1200}

t \approx 65.4\ \text{s}
$$

$$
\boxed{t \approx 1.1\ \text{min}}
$$

---

**5) Physical interpretation**

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
V, R, I, P = sp.symbols('V R I P', positive=True)
m, c, Ti, Tf, Q, t = sp.symbols('m c T_i T_f Q t', positive=True)

# -----------------------------
# Given values
# -----------------------------
V_val  = 120.0        # V
R_val  = 12.0         # ohm
m_val  = 0.250        # kg
c_val  = 4186.0       # J/(kg·K)
Ti_val = 20.0         # °C
Tf_val = 95.0         # °C

# -----------------------------
# Physics relationships
# -----------------------------
I_expr = sp.Eq(I, V/R)
P_expr = sp.Eq(P, I*V)
Q_expr = sp.Eq(Q, m*c*(Tf - Ti))
t_expr = sp.Eq(t, Q/P)

# -----------------------------
# Display symbolic equations (clean)
# -----------------------------
display(Math(r"\textbf{Ohm's law (current through heater):}"))
display(Math(sp.latex(I_expr)))

display(Math(r"\textbf{Electrical power dissipated:}"))
display(Math(sp.latex(P_expr)))

display(Math(r"\textbf{Thermal energy required to heat water:}"))
display(Math(sp.latex(Q_expr)))

display(Math(r"\textbf{Heating time:}"))
display(Math(sp.latex(t_expr)))

# -----------------------------
# Numerical evaluation (plain text only)
# -----------------------------
I_num = float(I_expr.rhs.subs({V: V_val, R: R_val}))
P_num = float(P_expr.rhs.subs({I: I_num, V: V_val}))
Q_num = float(Q_expr.rhs.subs({m: m_val, c: c_val, Ti: Ti_val, Tf: Tf_val}))
t_num = Q_num / P_num

print("\nGiven:")
print(f"  Voltage:           V = {V_val:.3g} V")
print(f"  Heater resistance: R = {R_val:.3g} Ω")
print(f"  Water mass:        m = {m_val:.3g} kg")
print(f"  Initial temp:      T_i = {Ti_val:.3g} °C")
print(f"  Final temp:        T_f = {Tf_val:.3g} °C")

print("\nResults:")
print(f"  Current:           I = {I_num:.3g} A")
print(f"  Power:             P = {P_num:.3g} W")
print(f"  Energy required:   Q = {Q_num:.3e} J")
print(f"  Heating time:      t = {t_num:.3g} s  ({t_num/60:.2f} min)")

print("\nExplanation:")
print("• The heating element behaves as a resistor connected to household voltage.")
print("• Electrical power P = IV is converted into thermal energy.")
print("• The water temperature rises according to Q = mcΔT.")
print("• Heating time follows from energy conservation: t = Q/P.")
print("• High-power appliances require large currents at fixed line voltage.")
```

## Problem D4.10 — Real Life Electricity Cost: Incandescent vs. LED

A common household upgrade is replacing old incandescent light bulbs with LED bulbs that produce the same brightness with much less power. A 100-W incandescent bulb can be replaced by a 16-W LED bulb. Both produce about **1600 lumens** of light (roughly the brightness of a traditional “100 W bulb”).

with $P$ in kilowatts and $t$ in hours gives $E$ in kWh.

\text{Cost} = (\text{rate})\times E.

E_{\text{inc}} = (0.100)(1825) = 182.5~\text{kWh}

</div>
</details>

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
from IPython.display import display, Math

# -----------------------------
# Symbols
# -----------------------------
P_inc, P_led, rate, h_day, days = sp.symbols(
    'P_{inc} P_{LED} r h d', positive=True
)

E_inc, E_led = sp.symbols('E_{inc} E_{LED}', positive=True)
C_inc, C_led, S = sp.symbols('C_{inc} C_{LED} S', positive=True)

# -----------------------------
# Given values
# -----------------------------
P_inc_val = 0.100      # kW (100 W)
P_led_val = 0.016      # kW (16 W)
rate_val  = 0.12       # $/kWh
h_day_val = 5.0        # h/day
days_val  = 365        # days/year

# -----------------------------
# Physics relationships
# -----------------------------
E_inc_expr = sp.Eq(E_inc, P_inc * h_day * days)
E_led_expr = sp.Eq(E_led, P_led * h_day * days)

C_inc_expr = sp.Eq(C_inc, rate * E_inc)
C_led_expr = sp.Eq(C_led, rate * E_led)

S_expr = sp.Eq(S, C_inc - C_led)

# -----------------------------
# Display symbolic equations (clean)
# -----------------------------
display(Math(r"\textbf{Annual energy use}"))
display(Math(sp.latex(E_inc_expr)))
display(Math(sp.latex(E_led_expr)))

display(Math(r"\textbf{Annual cost}"))
display(Math(sp.latex(C_inc_expr)))
display(Math(sp.latex(C_led_expr)))

display(Math(r"\textbf{Annual savings}"))
display(Math(sp.latex(S_expr)))

# -----------------------------
# Numerical evaluation (plain text only)
# -----------------------------
E_inc_num = float(E_inc_expr.rhs.subs({
    P_inc: P_inc_val, h_day: h_day_val, days: days_val
}))
E_led_num = float(E_led_expr.rhs.subs({
    P_led: P_led_val, h_day: h_day_val, days: days_val
}))

C_inc_num = rate_val * E_inc_num
C_led_num = rate_val * E_led_num
S_num = C_inc_num - C_led_num

print("\nGiven:")
print(f"  Incandescent power: {P_inc_val:.3f} kW")
print(f"  LED power:          {P_led_val:.3f} kW")
print(f"  Usage:              {h_day_val:.1f} h/day × {days_val} days")
print(f"  Electricity rate:   ${rate_val:.2f}/kWh")

print("\nResults:")
print(f"  Incandescent energy: {E_inc_num:.1f} kWh/year")
print(f"  LED energy:          {E_led_num:.1f} kWh/year")
print(f"  Incandescent cost:   ${C_inc_num:.2f}/year")
print(f"  LED cost:            ${C_led_num:.2f}/year")
print(f"  Annual savings:      ${S_num:.2f}/year")

print("\nExplanation:")
print("• Energy use is power multiplied by total operating time.")
print("• Cost is proportional to energy consumed (kWh).")
print("• LEDs provide the same light output using far less power.")
print("• Most energy in incandescent bulbs is lost as heat, not light.")
```

  I_{\text{avg}} = \frac{\Delta Q}{\Delta t}.
  $$
  
- **Instantaneous current** at a time is the slope of the **tangent line**:

  $$
  I(t) = \frac{dQ}{dt}.
  $$

---

**2) Average current from $t=1~\text{s}$ to $t=3~\text{s}$**

From the graph (read to the nearest whole coulomb):

- At $t = 1~\text{s}$, $Q \approx 2~\text{C}$
- At $t = 3~\text{s}$, $Q \approx 8~\text{C}$

Then

$$
I_{\text{avg}} \approx \frac{8 - 2}{3 - 1}
= \frac{6}{2}
= 3~\text{A}.
$$

$$
\boxed{I_{\text{avg}} \approx 3~\text{A}}
$$

Approximate slope:

$$
I(4~\text{s}) \approx \frac{27 - 15}{4.5 - 3.5}
= \frac{12}{1}
\approx 12~\text{A}.
$$

$$
\boxed{I(4~\text{s}) \approx 10\text{–}15~\text{A}}
$$

(Any estimate in this range is reasonable.)

**4) Where is the current largest?**

The current is largest where the $Q(t)$ curve is **steepest**, since slope corresponds to current.

From the graph, the curve becomes steepest near the **right end of the plot**, roughly:

$$
t \gtrsim 5~\text{s}.
$$

**Reason:** the charge is increasing most rapidly there, so $dQ/dt$ is largest.

- A gently sloped curve corresponds to small current.
- A steep curve corresponds to large current.

Q(t) = (2.0t^2 + 1.0t)\ \text{C},

**2) Current**

$$
I(t)=\frac{d}{dt}(2.0t^2+1.0t)=4.0t+1.0

Evaluate:

```python
%reset -f

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Math

# ============================================================
# Quadratic Charge Model — Full Solution Script
# ============================================================

# -----------------------------
# Symbols
# -----------------------------
t = sp.symbols('t', real=True)

# -----------------------------
# Given charge function Q(t)
# -----------------------------
Q = 2.0*t**2 + 1.0*t   # C

# -----------------------------
# Current from derivative
# -----------------------------
I = sp.diff(Q, t)

# -----------------------------
# Charge transfer on [1, 5]
# -----------------------------
t1, t2 = 1.0, 5.0
dQ_int  = sp.integrate(I, (t, t1, t2))
dQ_diff = Q.subs(t, t2) - Q.subs(t, t1)

# Average current on [1, 5]
I_avg = dQ_diff / (t2 - t1)

# -----------------------------
# Display equations (clean)
# -----------------------------
display(Math(r"\textbf{Given charge function}"))
display(Math(sp.latex(sp.Eq(sp.Symbol('Q(t)'), Q))))

display(Math(r"\textbf{Current definition } I(t)=\frac{dQ}{dt}"))
display(Math(sp.latex(sp.Eq(sp.Symbol('I(t)'), I))))

display(Math(r"\textbf{Charge transfer check}"))
display(Math(r"\Delta Q_{1\to5}=\int_{1}^{5} I(t)\,dt = Q(5)-Q(1)"))

display(Math(r"\textbf{Average current on }[1,5]"))
display(Math(sp.latex(sp.Eq(sp.Symbol('I_{avg}'), I_avg))))

# -----------------------------
# Numerical results (plain text)
# -----------------------------
print("\n--- Numerical results ---")
print(f"  Q(1) = {float(Q.subs(t, t1)):.3g} C")
print(f"  Q(5) = {float(Q.subs(t, t2)):.3g} C")
print(f"  ΔQ (1→5) from integral = {float(dQ_int):.3g} C")
print(f"  ΔQ (1→5) from Q diff  = {float(dQ_diff):.3g} C")
print(f"  Average current       = {float(I_avg):.3g} A")

print("\nExplanation:")
print("• Current is the time derivative of charge.")
print("• The integral of I(t) over time equals the net charge transferred.")
print("• For a quadratic Q(t), the current varies linearly with time.")

# -----------------------------
# Side-by-side plots
# -----------------------------
Q_func = sp.lambdify(t, Q, 'numpy')
I_func = sp.lambdify(t, I, 'numpy')

t_plot = np.linspace(0, 6, 400)
Q_plot = Q_func(t_plot)
I_plot = I_func(t_plot)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

# Q(t) plot
axes[0].plot(t_plot, Q_plot, linewidth=2)
axes[0].set_xlabel("Time (s)")
axes[0].set_ylabel("Charge Q (C)")
axes[0].set_title("Charge vs Time")
axes[0].grid(True)

# I(t) plot
axes[1].plot(t_plot, I_plot, linewidth=2)
axes[1].set_xlabel("Time (s)")
axes[1].set_ylabel("Current I (A)")
axes[1].set_title("Current vs Time")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```
