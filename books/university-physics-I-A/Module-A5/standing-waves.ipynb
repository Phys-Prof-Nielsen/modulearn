## Standing Waves

One of the most important consequences of the **superposition principle** is the formation of **standing waves**.

A standing wave occurs when two waves with the same frequency and amplitude travel in **opposite directions** through the same medium. This situation commonly arises when a wave reflects from a boundary and interferes with the incoming wave.

Consider two waves moving in opposite directions:

$$
y_1(x,t) = A \sin(kx - \omega t)
$$

$$
y_2(x,t) = A \sin(kx + \omega t)
$$

When these waves combine, the total displacement is

$$
y(x,t) = y_1 + y_2
$$

Using trigonometric identities, this becomes

$$
y(x,t) = 2A \sin(kx)\cos(\omega t)
$$

This expression describes a **standing wave**.

$$

Identify $\alpha$ and $\beta$.

\frac{\alpha+\beta}{2}
$$

$$
\frac{\alpha-\beta}{2}
$$

---

### Step 5 — Write the final standing wave

Substitute the results into the identity to obtain the standing wave expression.

## Physical Interpretation

A standing wave behaves very differently from a traveling wave.

- The wave pattern **does not move through space**.
- Instead, the pattern remains fixed while the medium **oscillates in place**.

The mathematical form

$$
y(x,t) = 2A \sin(kx)\cos(\omega t)
$$

contains two separate factors:

- $\sin(kx)$ controls **where motion occurs in space**
- $\cos(\omega t)$ controls **how the system oscillates in time**

Thus every point oscillates with the **same frequency**, but the **amplitude depends on position**.

<strong>Definition — Standing Wave</strong>

A **standing wave** is a wave pattern formed by the superposition of two waves with equal frequency and amplitude traveling in opposite directions.

The resulting pattern does not propagate through space but instead oscillates with a fixed spatial structure.

## Nodes and Antinodes

The spatial factor

$$
\sin(kx)
$$

determines where motion occurs.

### Nodes

Nodes occur where

$$
\sin(kx) = 0
$$

At these positions

$$
y(x,t) = 0
$$

for all times.

Thus **nodes are points that never move**.

### Antinodes

Antinodes occur where

$$
|\sin(kx)| = 1
$$

At these locations the oscillation amplitude is maximal:

$$
y_{\text{max}} = 2A
$$

Thus **antinodes are points of maximum motion**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

<strong>Example — Standing Wave on a String</strong>

A string fixed at both ends reflects waves at its boundaries. The reflected wave travels back along the string and interferes with the incoming wave.

The superposition of these two waves produces a **standing wave pattern** consisting of nodes and antinodes.

The endpoints of the string must always be nodes because the displacement at the boundaries is zero.

This restriction will determine the **allowed wavelengths** of the system, which leads to the concept of **normal modes** in the next section.

</div>

3. Find the distance between two neighboring nodes.  
4. Compare this distance to the **wavelength** $\lambda$.

<details>
<summary style="color:#006633; font-weight:bold;">Show Solution</summary>

Nodes occur where

$$
\sin(kx) = 0
$$

which gives

$$
kx = n\pi
$$

Thus

$$
x = \frac{n\pi}{k}
$$

Using

$$
k = \frac{2\pi}{\lambda}
$$

we obtain

$$
x = \frac{n\lambda}{2}
$$

Therefore nodes are separated by

$$
\frac{\lambda}{2}
$$

Antinodes occur halfway between nodes.

## Key Idea

Standing waves are a direct consequence of the **superposition principle**.  

Whenever waves reflect from boundaries, they can interfere with incoming waves to produce **stationary patterns of oscillation**.  

These patterns form the foundation for understanding **normal modes, resonance, and sound production** in many physical systems.
