## Waves Transport Energy

A key feature of waves is that they **transport energy through a medium without transporting matter over long distances**.

Consider a transverse wave traveling along a string. Each small element of the string oscillates up and down about its equilibrium position. While the individual particles do not move along the string, the **wave disturbance moves**, carrying energy from one location to another.

This is why waves can transmit signals and energy across large distances. Examples include:

- ocean waves transferring energy across the surface of the water  
- sound waves carrying energy through air  
- seismic waves transporting energy through the Earth  

Thus, waves act as **energy carriers** within physical systems.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Energy in a Vibrating String

When a wave travels along a string, each small element of the string possesses two types of energy:

### Kinetic Energy

Because the string elements are moving, they have kinetic energy.

For a small element of mass $dm$ moving with transverse velocity $v_y$:

$$
dK = \frac{1}{2} dm\, v_y^2
$$

### Elastic Potential Energy

As the string bends, tension in the string stores elastic potential energy.

The curvature of the string produces stretching, which stores energy in the string’s tension.

Thus, energy in a wave alternates between:

- **kinetic energy** of moving string elements  
- **potential energy** stored in the stretched string

As the wave travels, energy continuously transfers between these two forms.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Energy Transport Along the String

Although each piece of the string oscillates in place, the **wave disturbance moves with speed $v$**.

As the wave moves, the energy associated with the oscillations moves with it.

The **rate at which energy is transported** along the string is called the **power of the wave**.

For a sinusoidal wave on a stretched string, the average power transmitted by the wave is

$$
P = \frac{1}{2}\mu \omega^2 A^2 v
$$

where

- $\mu$ = mass per unit length of the string  
- $A$ = amplitude of the wave  
- $\omega$ = angular frequency  
- $v$ = wave speed

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

<strong>Definition — Wave Intensity</strong>

The **intensity** of a wave is the rate of energy transport per unit area.

$$
I = \frac{P}{A}
$$

where

- $P$ is the power carried by the wave
- $A$ is the area through which the wave passes.

Intensity measures how strongly a wave delivers energy to its surroundings.

</div>

## Amplitude and Energy

One of the most important results for wave energy is that the transported energy depends strongly on the **wave amplitude**.

From the power expression,

$$
P \propto A^2
$$

Thus:

- doubling the amplitude increases the transported energy by a factor of **four**
- small changes in amplitude can produce large changes in energy transport

This is why large ocean waves carry far more energy than small ripples.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

<strong>Example — Energy Transport in Ocean Waves</strong>

Ocean waves can travel across entire oceans while transporting energy.

Even though water molecules mainly oscillate up and down locally, the **wave energy moves horizontally** across the ocean surface.

When waves reach the shore, this transported energy is transferred to the coastline, producing the breaking waves we observe.

<strong>Box Activity — Wave Amplitude and Energy Transport</strong>

A sinusoidal wave travels along a string.

1. The wave amplitude is doubled while all other parameters remain unchanged.  
2. Determine how the transported power changes.  
3. If the amplitude is tripled, how does the transported power change?  
4. Explain why small increases in amplitude can dramatically increase wave energy.

<details>
<summary style="color:#006633; font-weight:bold;">Show Solution</summary>

The power carried by a sinusoidal wave is

$$
P = \frac{1}{2}\mu \omega^2 A^2 v
$$

This shows that

$$
P \propto A^2
$$

If the amplitude doubles:

$$
A \rightarrow 2A
$$

then

$$
P \rightarrow (2A)^2 = 4A^2
$$

so the power becomes **four times larger**.

If the amplitude triples:

$$
A \rightarrow 3A
$$

then

$$
P \rightarrow 9A^2
$$

so the power becomes **nine times larger**.

Thus wave energy increases rapidly with amplitude because the transported energy depends on the **square of the amplitude**.

## Key Idea

Mechanical waves transport energy through a medium while the individual particles of the medium oscillate around equilibrium positions.

The amount of energy transported depends strongly on the **wave amplitude**, with the transported power proportional to $A^2$.
