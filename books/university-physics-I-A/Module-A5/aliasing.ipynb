# A5.9 Sampling, Measurement, and Aliasing
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Measuring Waves

In modern science and technology, waves are often measured using **digital sensors**. Microphones, cameras, oscilloscopes, and scientific detectors all convert physical signals into sequences of numbers that computers can analyze.

To do this, the signal is **sampled** at discrete moments in time.

Instead of recording a continuous signal

$$
y(t),
$$

a digital measurement records values at specific times

$$
t_0,\; t_1,\; t_2,\; t_3,\; \dots
$$

with a fixed spacing between measurements.

If the time between samples is $\Delta t$, then the **sampling frequency** is

$$
f_s = \frac{1}{\Delta t}.
$$

Sampling is essential for modern technologies such as:

- digital audio recording  
- video capture  
- medical imaging  
- scientific data acquisition  

However, sampling introduces an important limitation.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## The Nyquist Criterion

If the sampling rate is too low, the recorded signal can become **distorted or misinterpreted**.

The key principle that determines whether a signal can be measured accurately is the **Nyquist criterion**.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

<strong>Definition — Nyquist Criterion</strong>

To accurately measure a signal with maximum frequency $f_{\text{signal}}$, the sampling frequency must satisfy

$$
f_s \ge 2f_{\text{signal}}.
$$

This limit is known as the **Nyquist frequency**.

</div>

If this condition is not satisfied, the recorded signal can appear to have a completely different frequency.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Aliasing

When the sampling frequency is too low, the measured signal can appear to oscillate at the **wrong frequency**. This phenomenon is called **aliasing**.

Aliasing occurs because the sampled points can be interpreted as belonging to a different sinusoidal wave.

For example, if a high-frequency wave is sampled too slowly, the measured points might align with a lower-frequency oscillation.

The resulting signal is an **alias** of the true signal.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

<strong>Danger Zone — Why Aliasing Happens</strong>

Sampling does not record the entire continuous wave. It only records isolated points.

If those points happen to line up with a different sinusoidal curve, the measurement system cannot distinguish between the true signal and the lower-frequency alias.

Thus the recorded frequency may be completely incorrect.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Real-World Examples of Aliasing

Aliasing appears in many familiar situations.

### Wagon Wheel Effect

In movies, rotating wheels sometimes appear to:

- rotate slowly  
- stand still  
- rotate backward  

This occurs because the camera captures frames at discrete intervals. If the wheel rotates too fast relative to the frame rate, the sampled images produce an **alias frequency**.

### Digital Audio

Audio recordings use high sampling rates to avoid aliasing.

For example:

- CDs use a sampling rate of **44.1 kHz**

This allows accurate reproduction of frequencies up to about

$$
22\,\text{kHz},
$$

which covers the range of human hearing.

### Scientific Instruments

Aliasing can distort data in experiments when sensors sample signals too slowly.

Scientists must therefore ensure that measurement systems satisfy the **Nyquist criterion**.

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

<strong>Box Activity — Predicting Aliasing</strong>

A sinusoidal signal has frequency

$$
f = 1200\ \text{Hz}.
$$

A measurement system samples the signal at

$$
f_s = 1500\ \text{Hz}.
$$

1. Determine the Nyquist frequency for this sampling rate.  
2. Determine whether the sampling satisfies the Nyquist criterion.  
3. Predict whether aliasing will occur.  
4. Explain why increasing the sampling rate reduces aliasing.

<details>
<summary style="color:#006633; font-weight:bold;">Show Solution</summary>

The Nyquist frequency is half the sampling frequency:

$$
f_{\text{Nyquist}} = \frac{f_s}{2}.
$$

Thus

$$
f_{\text{Nyquist}} = \frac{1500}{2} = 750\ \text{Hz}.
$$

The signal frequency is

$$
1200\ \text{Hz},
$$

which is greater than the Nyquist frequency.

Therefore the Nyquist criterion

$$
f_s \ge 2f
$$

is **not satisfied**.

As a result, aliasing will occur and the recorded signal will appear to have an incorrect frequency.

Increasing the sampling rate increases the Nyquist frequency, allowing the system to correctly capture higher-frequency signals.

</details>

</div>

## Key Idea

Digital measurement systems record waves by sampling signals at discrete times.

If the sampling rate is too low, **aliasing** can occur, causing the recorded signal to appear to have the wrong frequency.

The **Nyquist criterion** ensures that signals can be measured accurately by requiring the sampling frequency to be at least twice the highest frequency present in the signal.
