# A4.2 Driven Oscillations

## Motivation

In the previous section we saw that **damping removes energy** from an oscillator, causing the amplitude of motion to gradually decrease. If no additional energy is supplied, oscillations eventually die out.

However, many real systems are continually supplied with energy by **external forces**. Examples include:

- pushing a child on a swing  
- alternating currents driving electrical circuits  
- periodic forces acting on buildings during earthquakes  
- electromagnetic waves driving electrons in antennas  

When a system experiences a **periodic driving force**, the oscillator can maintain motion even in the presence of damping. The competition between **energy input from the driving force** and **energy loss due to damping** leads to steady oscillations and one of the most important phenomena in physics: **resonance**.

In this section we explore how oscillators behave when driven by an external periodic force.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Driven Oscillator**

A **driven oscillator** is an oscillatory system that experiences an external time-dependent force supplying energy to the system.

For a mass–spring oscillator with damping and a periodic driving force, the equation of motion is

$$
m\ddot{x} + b\dot{x} + kx = F_0 \cos(\omega t)
$$

where

- $m$ is the mass  
- $b$ is the damping coefficient  
- $k$ is the spring constant  
- $F_0$ is the amplitude of the driving force  
- $\omega$ is the **driving angular frequency**

</div>

---

## Natural Frequency of the Oscillator

Before introducing driving forces, recall that an undamped oscillator has a **natural angular frequency**

$$
\omega_0 = \sqrt{\frac{k}{m}}
$$

This frequency characterizes how the system oscillates when disturbed and then left alone.

When a driving force is applied, the system can respond at the **driving frequency**, but the amplitude of the response depends strongly on the relationship between $\omega$ and $\omega_0$.

---

## Two Types of Motion

The motion of a driven oscillator generally consists of two parts.

### Transient Motion

Immediately after the driving force begins acting, the system exhibits **transient motion**.

This part of the motion:

- depends on the initial conditions  
- resembles the damped oscillations discussed in the previous section  
- gradually disappears due to damping

Mathematically, the transient part behaves like

$$
x(t) \propto e^{-\gamma t}
$$

where

$$
\gamma = \frac{b}{2m}.
$$

Because of damping, this contribution eventually becomes negligible.

---

### Steady-State Motion

After the transient motion dies away, the oscillator settles into **steady-state motion**.

In this regime:

- the oscillator moves at the **driving frequency**
- the amplitude becomes constant
- the driving force continuously replaces the energy lost to damping

The motion can be written as

$$
x(t) = A(\omega)\cos(\omega t - \delta)
$$

where

- $A(\omega)$ is the **frequency-dependent amplitude**
- $\delta$ is the **phase difference** between the force and the motion.

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Phase Difference**

The **phase difference** $\delta$ describes how the oscillator’s motion is shifted in time relative to the driving force.

- At low driving frequencies, the oscillator moves **in phase** with the force.  
- At high frequencies, the motion becomes **out of phase** with the force.  

Near resonance, the phase shift is approximately

$$
\delta \approx \frac{\pi}{2}.
$$

</div>

---

## Energy Flow in a Driven Oscillator

A driven oscillator reaches a steady state when

> **Energy added by the driving force equals energy lost due to damping.**

In this situation:

- energy flows continuously **into the system**
- the oscillator converts that energy into motion
- damping removes the same amount of energy each cycle

Thus, the amplitude becomes **constant in time**, even though energy is continually moving through the system.

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Pushing a Swing**

Consider pushing a playground swing.

If you push randomly, the motion may remain small. But if you push **periodically at just the right rhythm**, the amplitude of the swing increases dramatically.

In this situation:

- the child and swing act as an oscillator  
- your pushes provide the **driving force**  
- air resistance and friction provide **damping**

When the pushing frequency matches the natural oscillation frequency, the system approaches **resonance**, producing large oscillations.

## Looking Ahead

The amplitude of steady-state oscillations depends strongly on the relationship between the **driving frequency** and the **natural frequency** of the system.

When these frequencies align, oscillations can grow dramatically. This phenomenon is known as **resonance**, and it plays a central role in physics, engineering, and many natural systems.

In the next section we will investigate **resonance curves** and learn how oscillators respond to different driving frequencies.
