## Motivation

In the previous section we introduced **driven oscillations**, where an external periodic force supplies energy to a damped oscillator. We found that after transient motion dies out, the system settles into **steady-state oscillations** at the **driving frequency**.

However, the **amplitude of this motion is not the same for all driving frequencies**. Instead, the system responds much more strongly when the driving frequency is close to the oscillator’s **natural frequency**.

This dramatic increase in amplitude is called **resonance**. Resonance plays a fundamental role throughout physics and engineering:

- bridges and buildings responding to vibrations  
- musical instruments amplifying sound  
- radio receivers selecting specific frequencies  
- atoms absorbing electromagnetic radiation  

Understanding resonance helps us predict when oscillations become **large and potentially dangerous** or **useful and efficient**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Resonance**

**Resonance** occurs when a driven oscillator responds with a **maximum amplitude** because the driving frequency is close to the system’s natural frequency.

In a damped mass–spring oscillator driven by a periodic force,

$$
m\ddot{x} + b\dot{x} + kx = F_0 \cos(\omega t),
$$

the steady-state oscillation amplitude depends strongly on the **driving frequency $\omega$**.

</div>

---

## Frequency Response of an Oscillator

The amplitude of the steady-state oscillation is

$$
A(\omega) =
\frac{F_0/m}
{\sqrt{(\omega_0^2-\omega^2)^2 + (2\gamma\omega)^2}}
$$

where

$$
\omega_0 = \sqrt{\frac{k}{m}}
\qquad
\gamma = \frac{b}{2m}.
$$

This expression shows that the amplitude depends on the **difference between the driving frequency and the natural frequency**.

When the denominator becomes small, the amplitude becomes large.

---

## Resonance Curve

A **resonance curve** plots oscillation amplitude versus driving frequency.

Typical features:

- **Small amplitude** at very low frequencies  
- **Large amplitude near resonance**  
- **Small amplitude again at very high frequencies**

The peak of this curve occurs near the **resonant frequency**.

---

### Physical Interpretation of Frequency Limits

**Low driving frequency ($\omega \ll \omega_0$)**

The oscillator moves slowly and has time to follow the force almost exactly.

- motion is nearly **in phase**
- amplitude is moderate

---

**High driving frequency ($\omega \gg \omega_0$)**

The force changes direction too quickly for the mass to respond effectively.

- inertia prevents large motion  
- amplitude becomes small  

---

**Near resonance ($\omega \approx \omega_0$)**

The driving force pushes the system **in step with its natural motion**, continually adding energy.

- amplitude becomes large  
- energy accumulates over many cycles

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Tacoma Narrows Bridge**

One of the most famous examples of resonance occurred in 1940 when the **Tacoma Narrows Bridge** collapsed.

Steady winds created oscillations in the bridge structure. Because the driving frequency of the wind matched one of the bridge’s natural vibration modes, oscillations grew dramatically.

This resonance eventually caused structural failure.

</div>

---

## Role of Damping

Damping strongly influences resonance.

- **Weak damping**
  - produces a very **sharp resonance peak**
  - oscillations can become extremely large

- **Strong damping**
  - reduces the peak amplitude
  - broadens the resonance curve

In real systems, damping prevents amplitudes from becoming infinitely large.

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Resonant Frequency**

The **resonant frequency** is the driving frequency that produces the **maximum oscillation amplitude**.

For weak damping, it is approximately

$$
\omega_{\text{res}} \approx \omega_0.
$$

</div>

**Box Activity 1 — Predicting Resonance Behavior**

Consider three oscillators with identical masses and spring constants but different damping strengths.

1. Which oscillator will have the **largest resonance amplitude**?
2. Which will have the **sharpest resonance peak**?
3. Which will respond over the **widest range of frequencies**?

<details>
<summary>Discussion</summary>

1. The **least damped oscillator** produces the largest amplitude.  
2. Weak damping creates the **sharpest resonance peak**.  
3. Strong damping broadens the resonance curve, allowing response over a wider range of frequencies.

</details>

## Key Insight

Resonance occurs when energy supplied by a periodic force **constructively reinforces the natural oscillations of a system**.

> When the driving frequency matches the natural rhythm of the system, energy accumulates cycle after cycle, producing large motion.

This principle explains a wide variety of phenomena across physics, from vibrating strings to electromagnetic radiation.
