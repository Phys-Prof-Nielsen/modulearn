## Motivation

Oscillations describe motion that repeats in time. **Waves** extend this idea further: they describe how oscillatory motion **propagates through space**.

In earlier modules we introduced the mathematical description of **traveling waves** and the principle of **superposition**, which allow multiple waves to combine and propagate through space and time. These ideas provide the foundation for understanding many physical phenomena, from vibrations in strings to sound waves in air.

When waves interact with **boundaries**, they can combine with their reflections to produce **standing waves**. These standing patterns form the basis of **resonance**, determining the characteristic frequencies of musical instruments, acoustic cavities, and many mechanical systems.

In this module we explore several important wave phenomena. We examine how standing waves arise from superposition, how **normal modes** emerge from boundary conditions, how **sound waves** propagate through matter, and how relative motion between sources and observers produces the **Doppler effect**.

Finally, we explore how waves are **measured and recorded**, leading to the concept of **aliasing**, which plays an essential role in modern digital technologies such as audio recording, video capture, and scientific instrumentation.

Together, these ideas show how waves provide a powerful framework for understanding how **energy and information propagate through physical systems**.

## Learning Outcomes

By the end of this module, you should be able to:

### Conceptual Understanding

1. Describe how **standing waves** arise from the superposition of waves traveling in opposite directions.  
2. Identify **nodes** and **antinodes** in standing wave patterns and explain their physical origin.  
3. Explain how **boundary conditions** restrict the allowed wavelengths and frequencies of waves in confined systems.  
4. Describe the concept of **normal modes** and explain how they determine the resonant behavior of vibrating systems.  
5. Explain how **sound waves** arise from oscillations of pressure and density in a medium.  
6. Interpret how wave intensity relates to **energy transport** in mechanical waves.  
7. Explain the physical origin of the **Doppler effect** and how relative motion changes the observed frequency of waves.  
8. Describe how waves are recorded through **discrete sampling** and explain how insufficient sampling leads to **aliasing**.

### Analytical and Mathematical Skills

9. Analyze the mathematical form of a **standing wave** produced by superposition.

$$
y(x,t) = 2A \sin(kx)\cos(\omega t)
$$

10. Determine the allowed wavelengths and frequencies for waves in systems with fixed boundaries, such as a vibrating string.

$$
\lambda_n = \frac{2L}{n}, \qquad f_n = \frac{nv}{2L}
$$

11. Apply the definition of wave **intensity** to analyze energy transport.

$$
I = \frac{P}{A}
$$

12. Interpret sound levels using the **decibel scale**.

$$
\beta = 10 \log_{10}\left(\frac{I}{I_0}\right)
$$

13. Apply the **Doppler effect** to determine the observed frequency when sources or observers are in motion.

$$
f' = f \frac{v \pm v_o}{v \mp v_s}
$$

14. Apply the **Nyquist sampling condition** to determine whether a signal can be accurately measured.

$$
f_{\text{sample}} \ge 2 f_{\text{signal}}
$$

### Experimental and Computational Skills

15. Interpret graphical representations of wave motion in space and time.  
16. Analyze experimental or simulated data to determine **wave properties** such as wavelength, frequency, and wave speed.  
17. Identify **standing wave patterns** and extract normal mode frequencies from measurements.  
18. Interpret resonance patterns observed in vibrating strings, air columns, or acoustic systems.  
19. Analyze sampled wave data to identify **aliasing effects** and understand their physical origin.
