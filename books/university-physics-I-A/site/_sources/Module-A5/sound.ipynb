## Sound as a Mechanical Wave

Sound is a **mechanical wave** that propagates through a material medium such as air, water, or solids. Unlike the transverse waves on a string that we studied earlier, sound waves in fluids are primarily **longitudinal waves**.

In a longitudinal wave, the motion of the particles in the medium is **parallel to the direction in which the wave travels**.

As sound propagates through a medium, regions of **compression** and **rarefaction** move through the material.

- **Compression:** region of higher pressure and density  
- **Rarefaction:** region of lower pressure and density  

These alternating pressure variations travel through the medium at the **speed of sound**.

<strong>Definition — Longitudinal Wave</strong>

A **longitudinal wave** is a wave in which the motion of the medium is parallel to the direction of wave propagation.

Sound waves in gases and liquids are longitudinal waves.

</div>

## Pressure Oscillations

Sound waves can be described as oscillations of **pressure** and **density** in the medium.

If the equilibrium pressure of the medium is $P_0$, the pressure variation caused by the sound wave can be written as

$$
P(x,t) = P_0 + \Delta P \sin(kx - \omega t)
$$

where

- $\Delta P$ is the amplitude of the pressure variation  
- $k$ is the wave number  
- $\omega$ is the angular frequency  

These pressure variations push on nearby air molecules, causing the disturbance to propagate through the medium.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Speed of Sound

The speed at which sound travels depends on the **properties of the medium**.

For a fluid such as air, the speed of sound is

$$
v = \sqrt{\frac{B}{\rho}}
$$

where

- $B$ is the **bulk modulus** of the medium  
- $\rho$ is the **density**

The bulk modulus measures how resistant the medium is to compression.

In air at room temperature, the speed of sound is approximately

$$
v \approx 343\ \text{m/s}
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

<strong>Example — Sound Traveling Through Different Media</strong>

Sound travels at different speeds depending on the material through which it propagates.

Typical values include:

| Medium | Speed of Sound |
|------|------|
| Air (20°C) | $\approx 343$ m/s |
| Water | $\approx 1480$ m/s |
| Steel | $\approx 5000$ m/s |

In general, sound travels faster in materials that are **stiffer (large bulk modulus)** and slower in materials that are **more compressible**.

</div>

## Frequency and Pitch

The **frequency** of a sound wave determines the **pitch** that we perceive.

- Higher frequency → higher pitch  
- Lower frequency → lower pitch  

The range of frequencies that humans can hear is approximately

$$
20\ \text{Hz} \le f \le 20\,000\ \text{Hz}
$$

This range varies between individuals and generally decreases with age.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

<strong>Danger Zone — Particle Motion vs Wave Motion</strong>

It is common to imagine that air particles move from the speaker to your ear when sound travels.

This is **not correct**.

The air molecules mainly oscillate back and forth about their equilibrium positions. What actually travels through the air is the **wave disturbance**, not the individual particles.

Thus sound transports **energy and pressure variations**, not the air itself.

</div>

<strong>Box Activity — Visualizing Sound Waves</strong>

Imagine a long row of air molecules in a tube.

1. A speaker at one end begins oscillating back and forth.  
2. Describe how this motion produces regions of **compression** and **rarefaction**.  
3. Explain why the disturbance moves through the air even though the molecules themselves only oscillate locally.  
4. Predict what happens to the wave if the frequency of the speaker increases.

<details>
<summary style="color:#006633; font-weight:bold;">Show Solution</summary>

When the speaker moves forward, it pushes nearby air molecules closer together, creating a **compression** region with higher pressure.

When the speaker moves backward, the air spreads out slightly, producing a **rarefaction** with lower pressure.

These pressure differences push on neighboring molecules, causing the disturbance to propagate through the air.

Although the disturbance travels through the medium, individual air molecules mainly oscillate around their equilibrium positions.

If the speaker oscillates faster (higher frequency), the compressions and rarefactions occur more frequently, producing a sound with **higher pitch**.

</details>

## Key Idea

Sound waves are **longitudinal mechanical waves** consisting of alternating compressions and rarefactions that propagate through a medium.

They transport energy through pressure oscillations while the particles of the medium oscillate locally around equilibrium positions.
