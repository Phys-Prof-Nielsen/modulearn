# A5.7 Sound Intensity and the Decibel Scale
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Sound Energy and Intensity

Sound waves transport energy through a medium. As the wave propagates, energy flows through space in the form of pressure oscillations and particle motion.

To quantify how strongly a sound delivers energy, we introduce the concept of **intensity**.

The **intensity of a sound wave** is defined as the power carried by the wave per unit area perpendicular to the direction of propagation:

$$
I = \frac{P}{A}
$$

where

- $P$ is the power transported by the wave
- $A$ is the area through which the energy passes.

Intensity tells us **how much energy arrives at a location each second per unit area**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Intensity and Distance from a Source

Most sound sources emit waves approximately **in all directions**. As the sound spreads outward, the energy is distributed over a larger and larger spherical surface.

The surface area of a sphere is

$$
A = 4\pi r^2
$$

Therefore, if a source emits power $P$ uniformly, the intensity at distance $r$ becomes

$$
I = \frac{P}{4\pi r^2}
$$

This is known as the **inverse-square law**.

It shows that sound intensity decreases rapidly with distance from the source.

For example:

- doubling the distance reduces intensity by a factor of **4**
- tripling the distance reduces intensity by a factor of **9**

<strong>Definition — Sound Intensity Level</strong>

Because the range of audible sound intensities is extremely large, sound levels are often expressed using a **logarithmic scale** called the **decibel scale**.

The sound intensity level $\beta$ is defined as

$$
\beta = 10\log_{10}\!\left(\frac{I}{I_0}\right)
$$

where

$$
I_0 = 10^{-12}\ \text{W/m}^2
$$

is the **reference intensity**, approximately the threshold of human hearing.

Sound intensity level $\beta$ is measured in **decibels (dB)**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why a Logarithmic Scale?

The human ear can detect an enormous range of sound intensities—from extremely faint sounds to very loud noises.

Typical intensities range from

$$
10^{-12}\ \text{W/m}^2
$$

to

$$
1\ \text{W/m}^2
$$

This corresponds to a range of **twelve orders of magnitude**.

A logarithmic scale compresses this range into manageable numbers.

For example:

| Sound | Intensity Level |
|------|------|
| Threshold of hearing | 0 dB |
| Whisper | 30 dB |
| Conversation | 60 dB |
| Busy street | 80–90 dB |
| Rock concert | 110 dB |
| Pain threshold | ~120 dB |

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

<strong>Danger Zone — Decibels Are Logarithmic</strong>

The decibel scale is **not linear**.

An increase of **10 dB** corresponds to a **tenfold increase in intensity**:

$$
I_2 = 10 I_1
$$

An increase of **20 dB** corresponds to a **hundredfold increase** in intensity.

This means that sounds that differ by only a few tens of decibels can differ enormously in the amount of energy they carry.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

<strong>Example — Comparing Two Sound Levels</strong>

Suppose one sound has an intensity level of

$$
\beta_1 = 60\ \text{dB}
$$

and another sound has

$$
\beta_2 = 80\ \text{dB}.
$$

The difference is

$$
\Delta \beta = 20\ \text{dB}.
$$

Because each 10 dB corresponds to a factor of 10 in intensity, a 20 dB increase corresponds to

$$
I_2 = 100 I_1.
$$

Thus the louder sound carries **100 times more energy per unit area** than the quieter one.

<strong>Box Activity — Intensity Changes with Distance</strong>

A loudspeaker emits sound power uniformly in all directions.

1. Use the inverse-square law to determine how intensity changes with distance from the source.  
2. If the distance from the speaker doubles, determine the new intensity relative to the original value.  
3. If the distance triples, determine the new intensity.  
4. Explain why people sitting far from a stage hear much quieter sound than people near the speakers.

<details>
<summary style="color:#006633; font-weight:bold;">Show Solution</summary>

The intensity of sound from an isotropic source is

$$
I = \frac{P}{4\pi r^2}.
$$

If the distance doubles,

$$
r \rightarrow 2r,
$$

then

$$
I' = \frac{P}{4\pi (2r)^2}
     = \frac{P}{16\pi r^2}
     = \frac{1}{4} I.
$$

Thus the intensity becomes **one-quarter** of the original value.

If the distance triples,

$$
r \rightarrow 3r,
$$

then

$$
I' = \frac{P}{4\pi (3r)^2}
     = \frac{1}{9} I.
$$

Therefore intensity decreases rapidly as sound spreads over a larger spherical area.

## Key Idea

Sound waves transport energy through pressure oscillations.  

The **intensity** of a sound wave measures the rate at which this energy passes through a given area, while the **decibel scale** provides a convenient logarithmic way to represent the enormous range of sound intensities detectable by the human ear.
