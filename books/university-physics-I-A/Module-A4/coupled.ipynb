# A4.5 Coupled Oscillators and Normal Modes
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Motivation

So far we have studied oscillations of **a single oscillator**. Many real systems, however, contain **multiple oscillators connected together**. When oscillators interact, energy can move between them, producing more complex motion.

Examples include:

- atoms vibrating in a molecule  
- masses connected by springs  
- buildings connected by structural supports  
- coupled electrical circuits  
- vibrating strings and crystal lattices  

Despite this complexity, many coupled systems exhibit a remarkable property:

> Their motion can be described as combinations of simple collective oscillations called **normal modes**.

Understanding these modes allows us to analyze complex systems in a surprisingly simple way.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Coupled Oscillators**

**Coupled oscillators** are oscillating systems connected in such a way that the motion of one oscillator affects the motion of another.

The coupling allows **energy to transfer between the oscillators**.

</div>

---

## Example System: Two Masses and Three Springs

Consider a simple mechanical example.

Two masses are connected by springs:

wall — spring — mass 1 — spring — mass 2 — spring — wall

If one mass is displaced, the springs exert forces on both masses, causing the motion of each mass to influence the other.

The equations of motion become

$$
m\ddot{x}_1 = -k x_1 + k(x_2 - x_1)
$$

$$
m\ddot{x}_2 = -k x_2 + k(x_1 - x_2)
$$

Because the equations are **linked**, the motions of $x_1$ and $x_2$ cannot be solved independently.

---

## Energy Transfer Between Oscillators

If only one mass is initially displaced, something interesting happens:

- The first oscillator begins moving  
- Energy gradually transfers to the second oscillator  
- The first oscillator slows down while the second speeds up  
- Energy then transfers back again

This process continues, producing **periodic energy exchange** between the oscillators.

This behavior is sometimes called **beats** or **energy sloshing** between oscillators.

---

## Normal Modes

Although the motion of coupled oscillators may look complicated, there exist special patterns of motion called **normal modes**.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Normal Mode**

A **normal mode** is a pattern of motion in which all parts of the system oscillate at the **same frequency** with a fixed relationship between their amplitudes.

In a normal mode, the entire system behaves like a **single effective oscillator**.

</div>

---

## Normal Modes of the Two-Mass System

For the two-mass system above, two normal modes exist.

### Mode 1 — In-Phase Motion

Both masses move **together in the same direction**.

$$
x_1 = x_2
$$

Because the middle spring does not stretch in this case, the system behaves like a single oscillator.

This mode has the **lower frequency**.

---

### Mode 2 — Out-of-Phase Motion

The masses move **in opposite directions**.

$$
x_1 = -x_2
$$

The middle spring stretches strongly, producing a larger restoring force.

This mode has a **higher frequency**.

---

## Why Normal Modes Matter

Normal modes provide a powerful way to understand complicated oscillatory systems.

Any motion of the system can be written as a **combination of its normal modes**.

For example:

- starting with one mass displaced produces a mixture of both modes  
- the interference of these modes produces the complex motion we observe

This idea is extremely important throughout physics.

Normal modes explain:

- vibrations of molecules  
- sound waves in air  
- vibrations of strings and membranes  
- oscillations of atoms in solids  

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Molecular Vibrations**

Atoms in a molecule behave like masses connected by springs.

Each molecule has characteristic **normal mode vibrations**, which correspond to specific frequencies.

These frequencies determine how molecules interact with **infrared radiation**, allowing scientists to identify molecules using spectroscopy.

</div>

**Box Activity 1 — Predicting Normal Modes**

Two identical masses are connected by three identical springs between rigid walls.

1. Describe the motion in which both masses move **in the same direction**.  
2. Describe the motion in which the masses move **in opposite directions**.  
3. Which motion has the **higher frequency**? Explain why.

<details>
<summary>Discussion</summary>

1. When both masses move together, the middle spring does not stretch significantly.  
2. When the masses move oppositely, the middle spring stretches strongly.  
3. The out-of-phase motion has the higher frequency because the restoring force is larger.

</details>

## Key Insight

Even very complicated oscillatory systems can be understood in terms of **simple collective motions**.

> Complex oscillations are combinations of a system’s **normal modes**.

This idea forms the basis for understanding **waves**, where large numbers of coupled oscillators interact.

## Looking Ahead

When many oscillators are coupled together — such as elements of a string or atoms in a solid — normal modes combine to form **wave motion**.
