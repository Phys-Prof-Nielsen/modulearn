# A4.4 Quality Factor and Resonance Width
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Motivation

In the previous section we studied **resonance**, where a driven oscillator responds strongly when the driving frequency is close to the system’s natural frequency. We saw that the amplitude of oscillation reaches a peak near resonance.

However, not all resonance peaks look the same. Some systems respond **very sharply**, oscillating strongly only within a narrow range of frequencies. Other systems respond **more broadly**, oscillating moderately over a wider range of frequencies.

To quantify how sharp or broad a resonance is, we introduce an important quantity called the **quality factor**, or **$Q$ factor**. The $Q$ factor tells us how effectively an oscillator stores energy compared to how quickly it loses energy through damping.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Definition — Quality Factor**

The **quality factor** $Q$ measures how weakly damped an oscillator is. It is defined as

$$
Q = \frac{\text{energy stored}}{\text{energy lost per cycle}}.
$$

A large value of $Q$ means the oscillator stores energy efficiently and loses only a small fraction of energy each cycle.

</div>

---

## Mechanical Expression for the Quality Factor

For a damped mass–spring oscillator

$$
m\ddot{x} + b\dot{x} + kx = F_0 \cos(\omega t),
$$

the quality factor can be written as

$$
Q = \frac{m\omega_0}{b},
$$

where

$$
\omega_0 = \sqrt{\frac{k}{m}}
$$

is the natural angular frequency.

This expression shows that

- **larger damping ($b$)** → smaller $Q$  
- **smaller damping** → larger $Q$

---

## Physical Meaning of the Quality Factor

The quality factor provides a direct measure of how long oscillations persist.

- **Large $Q$**
  - weak damping  
  - oscillations decay slowly  
  - very sharp resonance peak

- **Small $Q$**
  - strong damping  
  - oscillations decay quickly  
  - broad resonance curve

Many familiar systems illustrate this idea:

| System | Typical Behavior |
|------|------|
| tuning fork | very high $Q$ |
| car suspension | moderate $Q$ |
| door closer | very low $Q$ |

---

## Resonance Width

Another way to understand the $Q$ factor is through the **width of the resonance curve**.

The resonance curve plots oscillation amplitude versus driving frequency.

The **resonance width** $\Delta \omega$ is the range of frequencies over which the oscillator responds strongly.

The quality factor is related to this width by

$$
Q = \frac{\omega_0}{\Delta \omega}.
$$

This relationship shows that

- **large $Q$ → narrow resonance**
- **small $Q$ → broad resonance**

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Musical Instruments**

A tuning fork has a very large quality factor.

- It vibrates for a long time after being struck  
- It produces a very pure tone at a specific frequency

This occurs because the tuning fork loses very little energy each oscillation, creating a **sharp resonance peak**.

**Box Activity 1 — Comparing Quality Factors**

Two oscillators have the same natural frequency but different damping coefficients.

- Oscillator A: small damping  
- Oscillator B: large damping  

1. Which oscillator has the **larger $Q$ factor**?  
2. Which oscillator will have the **sharper resonance peak**?  
3. Which oscillator will continue oscillating longer after being disturbed?

<details>
<summary>Discussion</summary>

1. Oscillator A (smaller damping) has the larger $Q$.  
2. Oscillator A produces the sharper resonance peak.  
3. Oscillator A continues oscillating longer because it loses energy more slowly.

</details>

## Key Insight

The quality factor connects **energy loss, resonance behavior, and damping strength**.

> A system with a large $Q$ stores energy efficiently and responds strongly to a narrow range of driving frequencies.

This concept appears throughout physics, including

- mechanical oscillators  
- electrical circuits  
- optical cavities  
- quantum systems
