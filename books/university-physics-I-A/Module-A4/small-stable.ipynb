# A4.6 Small Oscillations About Stable Equilibrium
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Motivation

In earlier modules we saw that many physical systems can oscillate around an **equilibrium position**. In simple cases—such as a mass attached to a spring—the restoring force is already linear and the motion is exactly **simple harmonic motion (SHM)**.

However, many real systems do **not** have perfectly linear restoring forces. The forces may depend on position in complicated ways, and the potential energy may have a complex shape.

Despite this complexity, something remarkable happens:

> **Near a stable equilibrium, almost any system behaves like a harmonic oscillator.**

This result allows us to approximate many physical systems as SHM when the motion involves **small displacements from equilibrium**.

This idea forms one of the most powerful tools in physics and is widely used in mechanics, molecular physics, and quantum mechanics.

**Definition — Equilibrium Point**

An **equilibrium point** occurs where the net force on a system is zero.

In terms of potential energy $U(x)$,

$$
F(x) = -\frac{dU}{dx}.
$$

Therefore equilibrium occurs where

$$
\frac{dU}{dx} = 0.
$$

</div>

---

## Stable Equilibrium

Not all equilibrium points produce oscillations.

For oscillations to occur, the equilibrium must be **stable**.

A stable equilibrium occurs when the potential energy has a **minimum** at the equilibrium position.

Mathematically,

$$
\frac{d^2 U}{dx^2} > 0.
$$

If the system is displaced slightly from this point, the restoring force pushes it back toward equilibrium.

---

## Approximating the Potential Energy

To understand the motion near equilibrium, we expand the potential energy function around the equilibrium position $x_0$.

Using a Taylor expansion,

$$
U(x) \approx U(x_0)
+ \left.\frac{dU}{dx}\right|_{x_0}(x-x_0)
+ \frac{1}{2}\left.\frac{d^2U}{dx^2}\right|_{x_0}(x-x_0)^2 + \dots
$$

At equilibrium,

$$
\left.\frac{dU}{dx}\right|_{x_0} = 0.
$$

Therefore the leading term governing motion becomes

$$
U(x) \approx U(x_0) + \frac{1}{2}k_{\text{eff}}(x-x_0)^2
$$

where

$$
k_{\text{eff}} =
\left.\frac{d^2U}{dx^2}\right|_{x_0}.
$$

This quadratic form is exactly the **potential energy of a harmonic oscillator**.

---

## Resulting Equation of Motion

Using the force relation

$$
F = -\frac{dU}{dx},
$$

the approximate restoring force becomes

$$
F \approx -k_{\text{eff}}(x-x_0).
$$

Applying Newton’s Second Law,

$$
m\ddot{x} = -k_{\text{eff}}(x-x_0).
$$

Thus the motion obeys the equation for **simple harmonic motion**.

The angular frequency of oscillation becomes

$$
\omega = \sqrt{\frac{k_{\text{eff}}}{m}}.
$$

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Small-Angle Pendulum**

The exact restoring torque for a pendulum depends on $\sin\theta$:

$$
\tau = -mgL \sin\theta.
$$

For small angles,

$$
\sin\theta \approx \theta.
$$

This approximation makes the restoring torque proportional to displacement, producing the familiar SHM result

$$
\omega = \sqrt{\frac{g}{L}}.
$$

</div>

---

## Why Small Oscillations Matter

This result explains why SHM appears so frequently in physics.

Even when the exact forces are complicated:

- near equilibrium the potential energy is approximately quadratic  
- the restoring force becomes approximately linear  
- the motion becomes approximately harmonic  

This approximation allows physicists to analyze systems ranging from

- vibrating molecules  
- atoms in crystal lattices  
- mechanical structures  
- trapped particles  

using the simple mathematics of SHM.

## Key Insight

The principle of small oscillations shows that:

> **Any system near a stable equilibrium behaves approximately like a harmonic oscillator.**

This insight explains why SHM appears across so many areas of physics and why oscillations provide a foundation for understanding **waves, molecular vibrations, and quantum systems**.
