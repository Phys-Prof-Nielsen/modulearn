# Google Colab for Physics Labs: From Template to HTML Submission
---

## Why We Use Google Colab for Lab Reports

In this course, lab reports are provided as **Jupyter Notebook (`.ipynb`) templates** that combine written explanations, data analysis, figures, and code in a single, organized document. We use **Google Colab** because it allows you to open these templates **without installing any software**, save your work directly to **your own Google Drive**, and modify the notebook from any device with a web browser. Colab also makes it easy to **share your notebook with lab partners**, supporting collaboration while keeping your work version-controlled and accessible. Once complete, your lab report can be **exported as an HTML file** for submission, ensuring that formatting, figures, and calculations appear exactly as intended.

---

## From Jupyter Book Template to HTML Submission

Lab report templates are provided directly inside the course **Jupyter Book**. You will open the template in **Google Colab**, save a copy to **your own Google Drive**, complete the lab there, and then export a final **HTML file** for submission. Follow the steps below in order.

---

## Step 1: Open the Lab Template in Google Colab

1. Navigate to the lab report template page in the Jupyter Book  
2. Click the **Open in Colab** (rocket) button at the top of the page  
3. The lab template will open in a new Google Colab tab

At this point, you are viewing a **temporary copy** of the notebook. You must save it to your own Drive before editing.

---

## Step 2: Download the Lab Template from Colab

1. In Colab, go to **File → Download → Download `.ipynb`**  
2. The notebook will be saved to your computer

---

## Step 3: Save a Copy to Your Google Drive
 
1. Open **https://drive.google.com**
2. Create a folder named `Physics-I-Lab` (if you do not already have one)
3. Move the copied notebook into this folder. You can either do drag-drop or Click **New → File upload**
4. Rename the file  
   `lab1_template.ipynb` → `lab1.ipynb`

All edits you make from this point forward will be saved automatically to your Drive.

---

## Step 4: Open the Notebook in Google Colab from Google Drive

1. In **Google Drive**, locate `lab1.ipynb` inside your `Physics-I-Lab` folder
2. Right-click (or click the three dots) on the file
3. Select **Open with → Google Colaboratory**

If **Google Colaboratory does not appear** in the “Open with” menu:

1. Click **Open with → Connect more apps**
2. Search for **Google Colaboratory**
3. Click **Install**
4. Return to your notebook and open it using **Open with → Google Colaboratory**

Once opened, the notebook will run directly from your Google Drive, and all changes will be saved automatically.

---

## Step 5: Mount Google Drive in Colab (Do This First)

In order for Colab to **read data files** (such as CSV files) and **save exported HTML files** to your Google Drive, you must mount your Drive.  
This should be done **at the very top of the notebook**, before running any other code cells.

### Add a Code Cell at the Top of the Notebook

1. In Colab, click on the **first cell** in the notebook
2. From the top menu, choose **Insert → Code cell**
   - This inserts a new code cell **above** the current cell
3. Drag the new code cell to the very top if needed

---

### Mount Google Drive

In the new top code cell, run:

`from google.colab import drive`  
`drive.mount('/content/drive')`

When prompted, follow the authorization steps to grant access.

Once mounted, your Google Drive will be available in Colab at:

`/content/drive/MyDrive/`

> **Important:** If you skip this step, Colab will not be able to find your data files or save exported HTML files.

---

## Step 6: Complete the Lab in Google Colab

Work through the notebook as instructed:
- Enter written explanations in Markdown cells
- Use LaTeX for equations 
- Run and modify code cells as needed  
- Generate figures, tables, and numerical results  

You may share the notebook with lab partners using Google Drive sharing tools.

---

## Step 7: Export the Notebook as an HTML File

### Convert the Notebook to HTML

In a new code cell, run the command below.  
Make sure the folder name and filename match your Google Drive path exactly:

`!jupyter nbconvert --to html --no-input --embed-images "/content/drive/MyDrive/Physics-I-Lab/lab1.ipynb"`

This will generate `lab1.html` in the same Google Drive folder.

---

## Step 8: Download the HTML File

1. Locate `lab1.html` in Google Drive  
2. Right-click the file → **Download**

---

You now have a **static HTML version** of your lab report that can be opened in any web browser and submitted to Canvas.
