# Markdown for Lab Reports — Quick Guide & Cheat Sheet

In this course, you will write lab reports using **Markdown** inside Jupyter Notebooks. Markdown lets you format text to make your reports clean, readable, and professional when exported to HTML.

---

## Why Use Markdown?

- Organize your report into clear sections  
- Highlight important results or ideas  
- Make tables, lists, and structure look professional  
- Display code and Python outputs clearly

---

## What is Markdown?

Markdown is a lightweight markup language that allows users to create formatted text using a plain-text editor. Originally created by John Gruber in 2004, Markdown was designed to be simple and readable in its raw form while also being easily convertible to structured formats like HTML. It uses intuitive symbols—such as `#` for headings, `*` for bold or italic text, and `-` for bullet points—to apply formatting without relying on complex code. Because it focuses on clarity and ease of use, Markdown is widely used in documentation, blogging, software development, and education. In environments like Jupyter Notebooks, Markdown provides a convenient way for students and researchers to mix explanations, equations, and data analysis in a clean and organized manner.

For lab reports in particular, Markdown is a powerful tool that enhances readability and professionalism. Instead of using word processors, students can write structured documents directly within the Jupyter Notebook interface, seamlessly integrating formatted text with code and data visualizations. Using Markdown, a student can create section headers, summarize experimental methods, insert tables for data, and emphasize key results—all within the same environment they use to run their Python code. When exported to HTML or PDF, the final result is a polished report suitable for academic or professional submission. This not only reinforces technical communication skills but also encourages good habits in combining documentation with computational analysis.

## Common Markdown Syntax

### Headings

Use `#` to create headings:

    # Title (Level 1)  
    ## Section Heading (Level 2)  
    ### Subsection (Level 3)

---

### Italic and Bold

Use asterisks `*` or underscores `_`:

    *italic* or _italic_  
    **bold** or __bold__  
    ***bold and italic***

---

### Horizontal Line (Divider)

Use three dashes on a new line:

    ---

---

### Bullet Lists

Use `-` or `*` followed by a space:

    - First item  
    - Second item  
      - Sub-item

---

### Numbered Lists

Use numbers followed by a period:

    1. Step one  
    2. Step two  
    3. Step three

---

### Tables

Use `|` to separate columns and `-` to define the header row:

    | Quantity | Value | Unit |  
    |----------|-------|------|  
    | Mass     | 1.2   | kg   |  
    | Time     | 3.5   | s    |  
    | Force    | 7.0   | N    |

---

### Links

To add a hyperlink:

    [Link Text](https://example.com)

---

### Inline Code

Wrap code or commands with single backticks:

    Use the `print()` function to display output.

---

### Code Block (Multiline)

Use triple backticks to show multiple lines of code:

    ```
    for i in range(5):
        print(i)
    ```

---

## ✅ Summary

Use Markdown to:

- Structure your report with `#` headings  
- Emphasize using `*italic*` and `**bold**`  
- Present data in tables  
- Add bullet or numbered lists  
- Insert code using backticks  
- Separate sections with `---`  

Good Markdown = Clean, Professional Reports!
