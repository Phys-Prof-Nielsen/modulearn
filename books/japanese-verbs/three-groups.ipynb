# The Three Verb Groups: The Structural Key
---

Before we begin conjugating verbs, we need to answer one essential question:

> What kind of verb is it?

Japanese verbs fall into **three main groups**.  
Once you can identify the group, conjugation becomes predictable.

Think of this as learning three mechanical systems.  
Once you understand how each system operates, you can apply it to dozens of verbs.

---

## 1️⃣ る-Verbs (Ichidan Verbs)

These are often called **Ichidan verbs** (一段動詞), meaning “one-step verbs.”

They typically end in:

- **〜る (ru)**
- And the syllable before る usually contains an **e-sound** or **i-sound**

Examples:

- 食べる (taberu) – to eat  
- 見る (miru) – to see  
- 起きる (okiru) – to wake up  
- 教える (oshieru) – to teach  

### Why are they called “one-step”?

Because conjugation is simple:

You remove る and add the new ending.

食べる → 食べ + ます  
食べる → 食べ + ない  

The stem stays stable.

This group is the most regular and easiest to work with.

---

## 2️⃣ う-Verbs (Godan Verbs)

These are called **Godan verbs** (五段動詞), meaning “five-step verbs.”

They do not follow the simple drop-る pattern.

Instead, their final syllable changes across the five vowel sounds:

**a / i / u / e / o**

Examples:

- 書く (kaku) – to write  
- 話す (hanasu) – to speak  
- 飲む (nomu) – to drink  
- 行く (iku) – to go  
- 帰る (kaeru) – to return  

Notice something important:

帰る (kaeru) ends in る but is **NOT** a る-verb.

This is why classification matters.

### Why “five-step”?

Because the final consonant shifts depending on the conjugation:

書く  
書かない (kakanai)  
書きます (kakimasu)  
書ける (kakeru)  
書こう (kakō)

The ending moves through vowel rows.

This group is slightly more complex, but extremely systematic once understood.

---

## 3️⃣ Irregular Verbs

Fortunately, there are only **two common irregular verbs** in modern Japanese:

- する (suru) – to do  
- 来る (kuru) – to come  

These verbs do not follow normal patterns and must simply be memorized.

However, する appears everywhere:

- 勉強する (benkyō suru) – to study  
- 運動する (undō suru) – to exercise  
- 電話する (denwa suru) – to call  

Many “nouns + する” combinations form verbs.

Because of this, mastering する is extremely important.

---

## Why This System Exists  
### A Glimpse Into 1,000 Years of Linguistic History

The three-group system we use today — る-verbs, う-verbs, and irregular verbs — is actually a modern simplification of a much older and more complex system.

To understand why Japanese verbs behave the way they do, we need to briefly look back in time.

---

### Old Japanese: A More Complex Beginning

In the earliest recorded stage of the language (8th century, during the Nara period), verbs were classified into multiple conjugation classes based on how their stems changed.

Rather than just three groups, there were several distinct patterns.  
These patterns reflected natural sound shifts in the language — especially vowel alternations.

Over time, some distinctions disappeared, others merged, and pronunciation evolved.  
What we now call **Godan verbs** preserve traces of those older vowel shifts.

---

### The Five-Step System (Godan) and Vowel Rows

The term 五段 (*Godan*) literally means:

> “Five levels” or “five steps.”

This refers to how the final syllable of the verb shifts across the five Japanese vowel sounds:

- a
- i
- u
- e
- o

For example, consider 書く (kaku):

- 書かない (kakanai) – negative (a-row)
- 書きます (kakimasu) – polite (i-row)
- 書く (kaku) – dictionary (u-row)
- 書ける (kakeru) – potential (e-row)
- 書こう (kakō) – volitional (o-row)

These vowel shifts are not arbitrary rules invented by textbooks.

They reflect deep phonological patterns inherited from earlier stages of Japanese.

When you conjugate a Godan verb, you are moving along a historical vowel grid that has existed for over a millennium.

---

### Ichidan Verbs: A Simplification Over Time

Ichidan (一段) means:

> “One level” or “one step.”

These verbs do not shift across vowel rows.  
Their stem remains stable.

食べる → 食べ + ます  
見 る → 見 + ない  

Historically, these verbs developed from older verb types whose stems gradually became phonologically stable. Over centuries, many verbs migrated into this simpler pattern.

Languages naturally move toward simplification in common speech.

Ichidan verbs represent that simplification.

---

### Why Only Two Irregular Verbs?

Modern Japanese has remarkably few irregular verbs compared to many Indo-European languages.

English has:
- go → went
- see → saw
- be → am/is/are/was/were

Japanese largely avoided this type of irregularity.

The two major irregular verbs today — する (to do) and 来る (to come) — are extremely ancient and extremely common. Because they were used constantly, they preserved older forms.

High-frequency words often resist regularization in many languages.

In that sense, する and 来る are linguistic fossils — living remnants of older Japanese.

---

### Influence of Writing and Social Structure

Another factor in the development of modern conjugation was the introduction of Chinese characters (kanji).

Japanese originally had no writing system.  
When Chinese writing was adopted (around the 5th–6th century), Japanese scholars had to adapt it to represent a completely different grammar.

This led to:

- The development of kana (hiragana and katakana)
- Clear separation between verb stems (often written in kanji)
- And grammatical endings (written in hiragana)

This visual separation reinforced the layered structure of verbs:

Kanji = core meaning  
Hiragana = grammatical transformation  

Over time, this reinforced the idea of a stable stem plus attached endings.

---

### A Language of Layers

Japanese grammar is often described as *agglutinative*.

That means:

> Meaning is built by attaching layers.

Instead of changing the core word dramatically, Japanese adds endings that stack:

食べ  
食べます  
食べました  
食べさせられました  

Each layer adds nuance without destroying the base.

This layered structure reflects a broader aesthetic seen throughout Japanese culture:

- Architecture built in modular layers  
- Seasonal clothing layered with intention  
- Poetry structured in measured syllabic units  

The language itself mirrors this design philosophy.

---

### The Modern Three-Group System: A Teaching Tool

The three-group classification used today is not ancient.

It is a modern pedagogical system created to make learning easier.

Historically, grammar books used more technical classifications based on classical Japanese. But for learners, the simplified model works beautifully:

1. Stable stem verbs (Ichidan)
2. Vowel-shifting verbs (Godan)
3. Two ancient irregular verbs

This is a map, not the territory.

But it is a very effective map.

---

### What This Means for You

When you conjugate Japanese verbs, you are not memorizing random endings.

You are participating in:

- Phonological patterns over 1,000 years old  
- Cultural norms of politeness embedded in grammar  
- A writing system that visually separates meaning and transformation  

Understanding this makes conjugation feel less mechanical —  
and more like entering a living system.

Now that you understand *why* the system exists,  
we can begin using it.

---

## The Golden Rule Moving Forward

Before conjugating any verb, always ask:

> Is it a る-verb?  
> Is it a う-verb?  
> Or is it irregular?

Once you answer that, everything else becomes methodical.

In the next section, we will begin with the simplest transformations:
- Dictionary form
- Polite form (〜ます)

And you will see how quickly patterns emerge.
