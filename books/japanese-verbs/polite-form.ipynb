# The First Transformation: Dictionary Form → Polite Form (〜ます)

---

Now we begin actually conjugating.

The form listed in dictionaries is called the **dictionary form**:

- 食べる（たべる / taberu）– to eat  
- 書く（かく / kaku）– to write  
- 飲む（のむ / nomu）– to drink  
- する（する / suru）– to do  

This is the neutral, plain form.

In everyday conversation — especially when speaking politely — Japanese uses the **〜ます form**.

---

## Step 1: Find the Stem

Everything begins with the **stem**.

The stem is what remains after removing or modifying the final syllable according to the verb group.

---

## る-Verbs (Ichidan Verbs)

Remove る.

食べる（たべる / taberu）  
→ 食べ（たべ / tabe）  

見る（みる / miru）  
→ 見（み / mi）  

起きる（おきる / okiru）  
→ 起き（おき / oki）  

Then add ます:

食べます（たべます / tabemasu）  
見ます（みます / mimasu）  
起きます（おきます / okimasu）  

The stem stays stable. Conjugation is straightforward.

---

## う-Verbs (Godan Verbs)

Change the final syllable to the **i-row** of the same consonant.

書く（かく / kaku）  
→ 書き（かき / kaki）  
→ 書きます（かきます / kakimasu）  

飲む（のむ / nomu）  
→ 飲み（のみ / nomi）  
→ 飲みます（のみます / nomimasu）  

話す（はなす / hanasu）  
→ 話し（はなし / hanashi）  
→ 話します（はなします / hanashimasu）  

帰る（かえる / kaeru）  
→ 帰り（かえり / kaeri）  
→ 帰ります（かえります / kaerimasu）  

Notice:

The final vowel shifts from **u → i**.

This vowel shift is the defining feature of Godan verbs.

---

## Irregular Verbs

There are only two common irregular verbs:

する（する / suru）  
→ します（します / shimasu）  

来る（くる / kuru）  
→ 来ます（きます / kimasu）  

These must be memorized.

---

## What Just Happened?

We did not “change the verb” randomly.

We:

1. Identified its group.
2. Formed the stem.
3. Attached ます.

Japanese verbs work in layers:

**Core meaning (often kanji) + grammatical ending (hiragana).**

For example:

食べます  
食 (kanji: meaning “eat/food”)  
べます (hiragana: grammatical ending)

This visual separation reinforces the layered structure of the language.

---

## Try Saying These Out Loud

私は食べます。  
（わたしは たべます / watashi wa tabemasu）  
I eat / I will eat.

彼は書きます。  
（かれは かきます / kare wa kakimasu）  
He writes.

Even with just this transformation, your expressive ability increases dramatically.
