# The Negative Form (〜ない)
---
So far, we can say:

食べる（たべる / taberu） – to eat  
食べた（たべた / tabeta） – ate  

But how do we say:

“I do not eat”?

Japanese forms the negative by attaching:

ない

However, how we attach ない depends on the verb group.

---

## る-Verbs (Ichidan)

Remove る and add ない.

食べる（たべる / taberu）  
→ 食べない（たべない / tabenai）

見る（みる / miru）  
→ 見ない（みない / minai）

起きる（おきる / okiru）  
→ 起きない（おきない / okinai）

Very simple. The stem stays stable.

---

## う-Verbs (Godan)

Here we see the vowel grid again.

To form the negative:

Change the final syllable to the **a-row**, then add ない.

---

### Examples

書く（かく / kaku）  
→ 書かない（かかない / kakanai）

飲む（のむ / nomu）  
→ 飲まない（のまない / nomanai）

話す（はなす / hanasu）  
→ 話さない（はなさない / hanasanai）

帰る（かえる / kaeru）  
→ 帰らない（かえらない / kaeranai）

待つ（まつ / matsu）  
→ 待たない（またない / matanai）

Notice:

u → a

This is the defining shift for the negative form.

---

### Special Case: う

買う（かう / kau）  
→ 買わない（かわない / kawanai）

The “u” changes to “wa” before ない.

---

## Irregular Verbs

する（する / suru）  
→ しない（しない / shinai）

来る（くる / kuru）  
→ 来ない（こない / konai）

These must be memorized.

---

## Something Important: ない Is an Adjective

Historically and grammatically, ない is actually an adjective.

That is why it can conjugate like one:

食べない → 食べなかった  
（たべなかった / tabenakatta）  
“Did not eat”

Notice:

ない → なかった

This behaves like an い-adjective:

高い → 高かった  
（たかい → たかかった）

Understanding this makes the system feel logical instead of memorized.

---

## Compare the Vowel Shifts

Polite form → i-row  
Negative form → a-row  

書く  
書きます（i-row）  
書かない（a-row）

The vowel grid is not random.  
It is structural.

---

## Usage Examples

私は寿司を食べない。  
（わたしは すしを たべない / watashi wa sushi o tabenai）  
I do not eat sushi.

彼は来ない。  
（かれは こない / kare wa konai）  
He does not come.
