# The て-Form (〜て)
---

The て-form is one of the most important forms in Japanese.

It is used to:

- Connect actions (“do A and then do B”)
- Make requests
- Form progressive tense (〜ている)
- Give permission
- Express prohibition

It does not indicate tense by itself.  
It simply connects or extends meaning.

---

## The Key Insight

If you already learned the plain past (〜た),

You already know most of the て-form.

In almost every case:

た → て  
だ → で

The sound patterns are the same.

---

## る-Verbs (Ichidan)

Remove る and add て.

食べる（たべる / taberu）  
→ 食べて（たべて / tabete）

見る（みる / miru）  
→ 見て（みて / mite）

起きる（おきる / okiru）  
→ 起きて（おきて / okite）

Very simple and stable.

---

## う-Verbs (Godan)

The patterns mirror the た-form.

---

### く → いて

書く（かく / kaku）  
→ 書いて（かいて / kaite）

※ Exception: 行く（いく / iku） → 行って（いって / itte）

---

### ぐ → いで

泳ぐ（およぐ / oyogu）  
→ 泳いで（およいで / oyoide）

---

### す → して

話す（はなす / hanasu）  
→ 話して（はなして / hanashite）

---

### む・ぶ・ぬ → んで

飲む（のむ / nomu）  
→ 飲んで（のんで / nonde）

遊ぶ（あそぶ / asobu）  
→ 遊んで（あそんで / asonde）

死ぬ（しぬ / shinu）  
→ 死んで（しんで / shinde）

---

### る・う・つ → って

帰る（かえる / kaeru）  
→ 帰って（かえって / kaette）

買う（かう / kau）  
→ 買って（かって / katte）

待つ（まつ / matsu）  
→ 待って（まって / matte）

---

## Irregular Verbs

する（する / suru）  
→ して（して / shite）

来る（くる / kuru）  
→ 来て（きて / kite）

---

## Why Does This Form Matter So Much?

Because it builds sentences.

For example:

朝ごはんを食べて、学校へ行く。  
（あさごはんを たべて、がっこうへ いく）  
I eat breakfast and go to school.

First action → second action.

The て-form connects them.

---

## Making Requests

Add ください:

食べてください。  
（たべてください / tabete kudasai）  
Please eat.

待ってください。  
（まってください / matte kudasai）  
Please wait.

---

## Progressive Tense (Preview)

Add いる:

食べている。  
（たべている / tabete iru）  
Is eating / eating (in progress)

We will explore this more later.

---

## Structural Insight

Notice again:

た-form → て-form  
書いた → 書いて  
飲んだ → 飲んで  

The only difference is:

た ↔ て  
だ ↔ で  

You are not learning a new system.  
You are extending an existing one.
