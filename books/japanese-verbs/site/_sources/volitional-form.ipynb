# The Volitional Form (〜よう / 〜ましょう)
---

The volitional form expresses:

- Suggestion (“Let’s…”)
- Intention (“I will…”)
- Invitation (“Shall we…?”)

There are two versions:

Plain volitional  
Polite volitional

---

## 1️⃣ Plain Volitional (〜よう)

---

### る-Verbs (Ichidan)

Remove る and add よう.

食べる（たべる / taberu）  
→ 食べよう（たべよう / tabeyō）

見る（みる / miru）  
→ 見よう（みよう / miyō）

起きる（おきる / okiru）  
→ 起きよう（おきよう / okiyō）

Simple and predictable.

---

### う-Verbs (Godan)

Change the final syllable to the **o-row**, then add う.

書く（かく / kaku）  
→ 書こう（かこう / kakō）

飲む（のむ / nomu）  
→ 飲もう（のもう / nomō）

話す（はなす / hanasu）  
→ 話そう（はなそう / hanasō）

帰る（かえる / kaeru）  
→ 帰ろう（かえろう / kaerō）

待つ（まつ / matsu）  
→ 待とう（まとう / matō）

Notice:

u → o

This completes the vowel grid pattern.

---

### Irregular Verbs

する（する / suru）  
→ しよう（しよう / shiyō）

来る（くる / kuru）  
→ 来よう（こよう / koyō）

---

## 2️⃣ Polite Volitional (〜ましょう)

This form is easier.

Simply change:

ます → ましょう

食べます（たべます / tabemasu）  
→ 食べましょう（たべましょう / tabemashō）

書きます（かきます / kakimasu）  
→ 書きましょう（かきましょう / kakimashō）

します（します / shimasu）  
→ しましょう（しましょう / shimashō）

来ます（きます / kimasu）  
→ 来ましょう（きましょう / kimashō）

---

### Usage Examples

一緒に食べよう。  
（いっしょに たべよう / issho ni tabeyō）  
Let’s eat together.

映画を見よう。  
（えいがを みよう / eiga o miyō）  
Let’s watch a movie.

帰りましょう。  
（かえりましょう / kaerimashō）  
Let’s go home. (polite)

---

### Structural Insight

Look at what you now know:

書く  
書きます（i-row）  
書かない（a-row）  
書こう（o-row）

The vowel shifts are not random.

They form a structural grid:

a – negative  
i – polite  
u – dictionary  
o – volitional  

You now understand the architecture of Japanese verbs.
