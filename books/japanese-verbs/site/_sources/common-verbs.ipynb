### Important Notes

• 勝つ (katsu) = to win  
• 負ける (makeru) = to lose  
• 頑張る (ganbaru) = to try hard / do one’s best  

頑張る is extremely common in Japanese culture.  
It carries deeper meaning than simply “try hard.”  
It implies perseverance, effort, and endurance.

### Useful Phrases in Context

野菜を切る。  
（やさいを きる / yasai o kiru）  
Cut the vegetables.

水を沸かす。  
（みずを わかす / mizu o wakasu）  
Boil the water.

スープを混ぜている。  
（すーぷを まぜている / sūpu o mazete iru）  
I am stirring the soup.

塩を加えます。  
（しおを くわえます / shio o kuwaemasu）  
I add salt.

---

### Cultural Note

Japanese cooking verbs often distinguish:

- 焼く (yaku) – grill, bake, pan-fry  
- 揚げる (ageru) – deep fry  
- 茹でる (yuderu) – boil  

Precision matters in Japanese cuisine, and the language reflects that.
