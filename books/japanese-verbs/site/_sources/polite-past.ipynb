# Adding Time: The Polite Past (〜ました)
---

So far, we can say:

食べます（たべます / tabemasu）  
“I eat” / “I will eat”

But how do we say:

“I ate”?

Japanese expresses past tense by modifying the ending.

The polite ending ます changes to:

ました

That’s it.

---

## The Pattern

ます → ました

You do not change the stem again.  
You simply replace ます with ました.

---

## る-Verbs (Ichidan)

食べる（たべる / taberu）  
→ 食べます（たべます / tabemasu）  
→ 食べました（たべました / tabemashita）  

見る（みる / miru）  
→ 見ます（みます / mimasu）  
→ 見ました（みました / mimashita）  

起きる（おきる / okiru）  
→ 起きます（おきます / okimasu）  
→ 起きました（おきました / okimashita）  

---

## う-Verbs (Godan)

書く（かく / kaku）  
→ 書きます（かきます / kakimasu）  
→ 書きました（かきました / kakimashita）  

飲む（のむ / nomu）  
→ 飲みます（のみます / nomimasu）  
→ 飲みました（のみました / nomimashita）  

話す（はなす / hanasu）  
→ 話します（はなします / hanashimasu）  
→ 話しました（はなしました / hanashimashita）  

帰る（かえる / kaeru）  
→ 帰ります（かえります / kaerimasu）  
→ 帰りました（かえりました / kaerimashita）  

---

## Irregular Verbs

する（する / suru）  
→ します（します / shimasu）  
→ しました（しました / shimashita）  

来る（くる / kuru）  
→ 来ます（きます / kimasu）  
→ 来ました（きました / kimashita）  

---

## What Did We Learn?

We did not create a new tense system.

We simply replaced:

ます → ました

The structure remains layered:

Verb stem + polite marker + past marker

For example:

食べました  
食 (meaning core)  
べ (stem)  
ました (polite past)

---

## Important Cultural Note

In Japanese, tense is simpler than in English.

There are only two primary distinctions:

- Non-past (present/future)
- Past

食べます can mean:
- I eat
- I will eat

Context determines which meaning applies.

Japanese relies heavily on shared context rather than strict grammatical time markers.

---

## Try These

私は飲みました。  
（わたしは のみました / watashi wa nomimashita）  
I drank.

彼は来ました。  
（かれは きました / kare wa kimashita）  
He came.

Now you can speak about both present and past events politely.
