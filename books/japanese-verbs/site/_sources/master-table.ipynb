# Master Verb Conjugation Table
---

Below is a complete overview of the core verb system.

We use three representative verbs:

る-verb (Ichidan): 食べる（たべる / taberu） – to eat  
う-verb (Godan): 書く（かく / kaku） – to write  
Irregular: する（する / suru） – to do  

---

## 1️⃣ Dictionary Form (Non-Past)

| Type | Kanji | Hiragana | Romaji |
|------|--------|-----------|---------|
| る-verb | 食べる | たべる | taberu |
| う-verb | 書く | かく | kaku |
| Irregular | する | する | suru |

## Structural Pattern Summary

For う-verbs like 書く (kaku):

かかない  → a-row (negative)  
かきます  → i-row (polite)  
かく      → u-row (dictionary)  
かこう    → o-row (volitional)

The vowel grid forms the structural backbone of Japanese verbs.
