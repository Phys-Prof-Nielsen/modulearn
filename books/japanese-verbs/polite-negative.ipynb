# Completing the Polite System: 〜ません and 〜ませんでした
---

So far, we have:

食べます（たべます / tabemasu） – eat (polite)  
食べました（たべました / tabemashita） – ate (polite)

But how do we say:

“I do not eat.”  
“I did not eat.”

In polite Japanese, we do not use ない.  
Instead, we replace ます with:

ません

---

# 1️⃣ Polite Negative (Non-Past)

ます → ません

---

## る-Verbs (Ichidan)

食べる（たべる / taberu）  
→ 食べます（たべます / tabemasu）  
→ 食べません（たべません / tabemasen）

見る（みる / miru）  
→ 見ます（みます / mimasu）  
→ 見ません（みません / mimasen）

起きる（おきる / okiru）  
→ 起きます（おきます / okimasu）  
→ 起きません（おきません / okimasen）

---

## う-Verbs (Godan)

書く（かく / kaku）  
→ 書きます（かきます / kakimasu）  
→ 書きません（かきません / kakimasen）

飲む（のむ / nomu）  
→ 飲みます（のみます / nomimasu）  
→ 飲みません（のみません / nomimasen）

話す（はなす / hanasu）  
→ 話します（はなします / hanashimasu）  
→ 話しません（はなしません / hanashimasen）

帰る（かえる / kaeru）  
→ 帰ります（かえります / kaerimasu）  
→ 帰りません（かえりません / kaerimasen）

---

## Irregular Verbs

する（する / suru）  
→ します（します / shimasu）  
→ しません（しません / shimasen）

来る（くる / kuru）  
→ 来ます（きます / kimasu）  
→ 来ません（きません / kimasen）

---

# 2️⃣ Polite Past Negative

To say “did not,” simply replace:

ません → ませんでした

食べませんでした  
（たべませんでした / tabemasen deshita）  
Did not eat.

書きませんでした  
（かきませんでした / kakimasen deshita）  
Did not write.

来ませんでした  
（きませんでした / kimasen deshita）  
Did not come.

---

## Structural Insight

Notice how stable the polite system is:

ます  
ました  
ません  
ませんでした  

Nothing changes in the stem.

This makes polite Japanese extremely regular and predictable.

Compare that to plain forms:

食べない  
食べなかった  

Here, ない behaves like an adjective.

In polite speech, the system becomes simpler.

---

## Full Polite System Overview

Using 食べる（たべる / taberu）:

食べます – eat  
食べました – ate  
食べません – do not eat  
食べませんでした – did not eat  

This symmetry makes polite Japanese very powerful for beginners.

---

## Usage Examples

私は肉を食べません。  
（わたしは にくを たべません）  
I do not eat meat.

昨日、映画を見ませんでした。  
（きのう、えいがを みませんでした）  
Yesterday, I did not watch a movie.
