# The Plain Past Form (〜た)
---

So far, we have worked inside the polite system:

食べます → 食べました  
（たべます → たべました / tabemasu → tabemashita）

Now we step into the **plain form**.

The plain past form is used:

- In casual speech
- In writing
- In narratives
- In subordinate clauses
- Before particles like けど (but), から (because), etc.

This is an extremely important form.

---

## る-Verbs (Ichidan)

For る-verbs, it is very simple:

Remove る and add た.

食べる（たべる / taberu）  
→ 食べた（たべた / tabeta）

見る（みる / miru）  
→ 見た（みた / mita）

起きる（おきる / okiru）  
→ 起きた（おきた / okita）

This is clean and predictable.

---

## う-Verbs (Godan)

Here the sound changes begin.

Instead of simply removing る, the final syllable transforms.

These patterns follow consistent sound rules.

---

### く → いた

書く（かく / kaku）  
→ 書いた（かいた / kaita）

※ Exception: 行く（いく / iku） → 行った（いった / itta）

---

### ぐ → いだ

泳ぐ（およぐ / oyogu）  
→ 泳いだ（およいだ / oyoida）

---

### す → した

話す（はなす / hanasu）  
→ 話した（はなした / hanashita）

---

### む・ぶ・ぬ → んだ

飲む（のむ / nomu）  
→ 飲んだ（のんだ / nonda）

遊ぶ（あそぶ / asobu）  
→ 遊んだ（あそんだ / asonda）

死ぬ（しぬ / shinu）  
→ 死んだ（しんだ / shinda）

---

### る・う・つ → った

帰る（かえる / kaeru）  
→ 帰った（かえった / kaetta）

買う（かう / kau）  
→ 買った（かった / katta）

待つ（まつ / matsu）  
→ 待った（まった / matta）

---

## Irregular Verbs

する（する / suru）  
→ した（した / shita）

来る（くる / kuru）  
→ 来た（きた / kita）

---

## Why So Many Sound Changes?

These are not random.

Historically, Japanese verbs attached a past suffix that evolved over time.  
As sounds merged, consonants softened or doubled.

For example:

書く + た  
Originally produced a sound closer to “kakita,”  
which gradually shifted to 書いた (kaita).

These patterns reflect natural sound evolution.

What looks complicated is actually phonological efficiency.

---

## Compare Polite and Plain

食べました（たべました / tabemashita）  
食べた（たべた / tabeta）

書きました（かきました / kakimashita）  
書いた（かいた / kaita）

Notice:

Polite past stays mechanically stable.  
Plain past reveals sound shifts.

---

## Usage Example

昨日、寿司を食べた。  
（きのう、すしをたべた / kinō, sushi o tabeta）  
Yesterday, I ate sushi.

彼は帰った。  
（かれは かえった / kare wa kaetta）  
He went home.
