# The Progressive Form (〜ている)
---

Now we build on the て-form.

To express an action that is happening right now, Japanese uses:

て-form + いる

This corresponds roughly to English “-ing”:

I am eating.  
He is studying.  
They are working.

---

## 1️⃣ Basic Structure

Verb (て-form) + いる

---

### Examples (Plain Form)

食べる（たべる / taberu）  
→ 食べている（たべている / tabete iru）  
→ Is eating / Am eating

書く（かく / kaku）  
→ 書いている（かいている / kaite iru）  
→ Is writing

飲む（のむ / nomu）  
→ 飲んでいる（のんでいる / nonde iru）  
→ Is drinking

話す（はなす / hanasu）  
→ 話している（はなしている / hanashite iru）  
→ Is speaking

来る（くる / kuru）  
→ 来ている（きている / kite iru）  
→ Is coming / Has come (context dependent)

する（する / suru）  
→ している（している / shite iru）  
→ Is doing

---

## 2️⃣ Polite Progressive

Now we conjugate いる using ます.

いる → います

So:

食べている  
→ 食べています（たべています / tabete imasu）  
Is eating (polite)

書いている  
→ 書いています（かいています / kaite imasu）

飲んでいる  
→ 飲んでいます（のんでいます / nonde imasu）

---

## 3️⃣ Negative Progressive

Plain:

食べていない  
（たべていない / tabete inai）  
Is not eating

Polite:

食べていません  
（たべていません / tabete imasen）  
Is not eating (polite)

Past:

食べていました  
（たべていました / tabete imashita）  
Was eating

食べていませんでした  
（たべていませんでした / tabete imasen deshita）  
Was not eating

Notice:

We are only conjugating いる.  
The main verb stays in て-form.

---

## 4️⃣ Usage Examples

私は今ご飯を食べています。  
（わたしは いま ごはんを たべています）  
I am eating now.

彼は勉強している。  
（かれは べんきょうしている）  
He is studying.

昨日、雨が降っていました。  
（きのう、あめが ふっていました）  
It was raining yesterday.

---

## 5️⃣ Important Note: Ongoing Action vs Resulting State

In English, “-ing” always suggests ongoing action.

In Japanese, 〜ている can also describe a resulting state.

Example:

結婚している  
（けっこんしている / kekkon shite iru）  
Is married

This does not mean “currently getting married.”  
It describes the state resulting from the action.

For now, remember the simple idea:

て-form + いる  
= action in progress (or state resulting from action)

---

## What You Can Now Express

You can now say:

- I eat.
- I ate.
- I do not eat.
- I did not eat.
- I am eating.
- I was eating.
- Let’s eat.

You now control tense, negation, and aspect in both polite and plain styles.
