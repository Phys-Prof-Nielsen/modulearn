# D9.1 Vectors and Index Notation
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D9.1.1 Motivation: A More Compact Way to Write Vector Equations

In mechanics we frequently work with **vectors** such as position, velocity, force, and momentum. In introductory physics these vectors are usually written using Cartesian components:

$$
\mathbf{a} = (a_x, a_y, a_z).
$$

Vector operations are then written explicitly in terms of these components. For example, the dot product of two vectors is

$$
\mathbf{a}\cdot\mathbf{b} = a_x b_x + a_y b_y + a_z b_z.
$$

While this expression is easy to understand, it becomes cumbersome when equations contain many vectors or when we want to emphasize the **structure of the equations rather than the specific components**.

Physicists often use a compact notation called **index notation**, also known as the **Einstein summation convention**, which simplifies many vector expressions.

In this notation the dot product above becomes

$$
\mathbf{a}\cdot\mathbf{b} = a_i b_i.
$$

This single expression automatically represents the sum over all spatial components.

Index notation will allow us to write vector equations in a way that is:

- **compact**
- **general**
- **easier to manipulate algebraically**

This notation will also prepare us for later topics in the course such as **rotational dynamics**, where quantities like the **moment of inertia tensor** naturally appear.

## D9.1.2 Vector Components

A vector in three-dimensional space can be written as

$$
\mathbf{a} = (a_x, a_y, a_z).
$$

These represent the components of the vector along the Cartesian coordinate axes.

In index notation we write these components as

$$
a_i
$$

where the index $i$ identifies the coordinate direction.

Typically

| Index | Cartesian Direction |
|------|------|
| $i = 1$ | $x$ |
| $i = 2$ | $y$ |
| $i = 3$ | $z$ |

Thus

$$
a_1 = a_x, \qquad a_2 = a_y, \qquad a_3 = a_z.
$$

The vector itself may be thought of as the collection of these components.

In **index notation**, the components of a vector are written using a subscript index:

$$
a_i
$$

where the index $i$ labels the coordinate direction.

For three-dimensional space

$$
i = 1,2,3.
$$

## D9.1.3 Dot Product Using Index Notation

The dot product of two vectors

$$
\mathbf{a}\cdot\mathbf{b}
$$

is normally written

$$
a_x b_x + a_y b_y + a_z b_z.
$$

Using index notation, we write this as

$$
\mathbf{a}\cdot\mathbf{b} = a_i b_i.
$$

Here the repeated index $i$ indicates that we sum over the three spatial components.

Later we will make this rule explicit as the **Einstein summation convention**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example — Writing a Dot Product with Indices</h3>

<strong>Problem</strong>

Let

$$
\mathbf{a} = (2,-1,4), \qquad \mathbf{b} = (3,5,1).
$$

Compute $\mathbf{a}\cdot\mathbf{b}$ using index notation.

---

<strong>Solution</strong>

Using

$$
\mathbf{a}\cdot\mathbf{b} = a_i b_i
$$

we compute

$$
a_i b_i =
(2)(3) + (-1)(5) + (4)(1).
$$

Thus

$$
\mathbf{a}\cdot\mathbf{b} = 6 - 5 + 4 = 5.
$$

## D9.1.4 Why This Notation Is Useful

Index notation allows us to write vector equations in a way that is independent of the coordinate system. Instead of writing separate equations for $x$, $y$, and $z$, we can write a single compact equation.

For example, Newton’s second law can be written as

$$
F_i = m a_i.
$$

This equation represents three component equations simultaneously.

Later in the course we will extend this notation to more complex objects such as **tensors**, which naturally appear in rotational dynamics and many areas of advanced physics.

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Curiosity — Upper and Lower Indices</h3>

In this module we write vector components using indices such as

$$
a_i.
$$

In more advanced physics you will sometimes see components written with **upper indices**

$$
a^i.
$$

These represent related but slightly different mathematical objects called **vectors** and **covectors**.

In ordinary Cartesian coordinates the distinction is not important because the two behave the same way. However, in more advanced theories — such as **general relativity and quantum fields** — the distinction becomes essential. 

For now, we will simply treat indexed quantities like $a_i$ as ordinary vector components.
