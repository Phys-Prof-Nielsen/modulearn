# D9.2 Einstein Summation Convention

## D9.2.1 Motivation: Writing Vector Equations More Efficiently

In the previous section we introduced **index notation** as a compact way to represent vector components. We saw that the dot product

$$
\mathbf{a}\cdot\mathbf{b} = a_x b_x + a_y b_y + a_z b_z
$$

can be written more compactly as

$$
\mathbf{a}\cdot\mathbf{b} = a_i b_i.
$$

This expression implicitly represents a sum over the three spatial components. Instead of writing the summation symbol explicitly, physicists often adopt a rule called the **Einstein summation convention**.

This convention makes many equations in mechanics and physics significantly shorter and easier to manipulate.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Definition — Einstein Summation Convention</h3>

When an index appears **twice in a single term**, it is understood to be **summed over all possible values** of that index.

Thus

$$
a_i b_i
$$

means

$$
a_1 b_1 + a_2 b_2 + a_3 b_3.
$$

The summation symbol

$$
\sum_{i=1}^{3}
$$

is therefore omitted.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D9.2.2 Explicit vs Implicit Summation

Using the summation symbol explicitly, we could write the dot product as

$$
\mathbf{a}\cdot\mathbf{b} =
\sum_{i=1}^{3} a_i b_i.
$$

Under the Einstein summation convention, this becomes

$$
\mathbf{a}\cdot\mathbf{b} = a_i b_i.
$$

Both expressions represent exactly the same quantity. The index notation simply **suppresses the summation symbol**.

This convention will be used throughout the remainder of the course.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D9.2.3 Example: Magnitude of a Vector

The magnitude squared of a vector

$$
\mathbf{r} = (x,y,z)
$$

is normally written as

$$
|\mathbf{r}|^2 = x^2 + y^2 + z^2.
$$

Using index notation,

$$
|\mathbf{r}|^2 = r_i r_i.
$$

Expanding the summation gives

$$
r_i r_i = r_1 r_1 + r_2 r_2 + r_3 r_3.
$$

Thus

$$
r_i r_i = x^2 + y^2 + z^2.
$$

## D9.2.4 Free Indices vs Summed Indices

It is important to distinguish between two types of indices.

### Summed (dummy) index

An index that appears **twice** in a term is summed over.

Example:

$$
a_i b_i
$$

The index $i$ is a **dummy index**, meaning it represents a summation over $1,2,3$.

---

### Free index

An index that appears **only once** in an equation is called a **free index**. It identifies a particular component of a vector.

Example:

$$
F_i = m a_i.
$$

Here $i$ is a **free index**, meaning this equation represents three separate component equations:

$$
F_1 = m a_1, \qquad
F_2 = m a_2, \qquad
F_3 = m a_3.
$$

Thus a single indexed equation can represent multiple component equations simultaneously.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example — Interpreting an Indexed Equation</h3>

<strong>Problem</strong>

Interpret the equation

$$
F_i = m a_i.
$$

What does this equation represent in terms of Cartesian components?

---

<strong>Solution</strong>

The index $i$ is a **free index**, meaning the equation represents the three component equations

$$
F_1 = m a_1
$$

$$
F_2 = m a_2
$$

$$
F_3 = m a_3.
$$

In Cartesian coordinates these correspond to

$$
F_x = m a_x,
$$

$$
F_y = m a_y,
$$

$$
F_z = m a_z.
$$

Thus the single indexed equation represents Newton’s second law in all three directions.

## D9.2.5 Why the Summation Convention Is Powerful

The Einstein summation convention allows complicated vector relationships to be written in a very compact form. Instead of writing three separate component equations, we can write a single indexed equation that represents all components simultaneously.

This becomes especially powerful when dealing with more complicated quantities such as **tensors**, where many components must be handled simultaneously.

In later modules we will use this notation when working with objects such as the **moment of inertia tensor**, where index notation greatly simplifies the mathematical expressions.
