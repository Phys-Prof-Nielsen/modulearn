# D8.1 Constraints and Degrees of Freedom in Classical Mechanics

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.1.1 Motivation: Why Constraints?

In introductory physics, we often solved problems by implicitly using **constraints**, even if we did not explicitly identify them. For example:

- A block on a ramp moves only along the slope, not perpendicular to it.  
- A pendulum bob is forced to remain a fixed distance from its pivot.  
- A bead sliding on a wire must stay on the wire at all times.

In each of these cases, the motion of the system is **restricted** by a geometric or physical condition.

In classical mechanics—particularly in the **Lagrangian and Hamiltonian formulations**—it becomes essential to **identify these restrictions explicitly** and express them mathematically.

Constraints play two central roles:

1. They restrict the motion of a system.
2. They reduce the number of **independent variables** required to describe that motion.

This leads naturally to the concept of **degrees of freedom**, which tells us how many independent coordinates are required to specify the configuration of a system.

Determining the degrees of freedom is therefore the **first step in constructing a systematic description of motion** in analytical mechanics.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.1.2 Degrees of Freedom

The **degrees of freedom (DOF)** of a mechanical system is the number of independent coordinates required to uniquely specify its configuration.

For a system of $N$ particles in three-dimensional space, the unconstrained configuration requires

$$
3N
$$

coordinates:

$$
(x_1,y_1,z_1), (x_2,y_2,z_2), \dots, (x_N,y_N,z_N)
$$

Each independent constraint reduces the number of independent coordinates by one.

Thus, if the system has $k$ independent constraints, the number of degrees of freedom is

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Degrees of Freedom</h3>

$$
\text{DOF} = 3N - k
$$

</div>

This relation tells us the **minimum number of independent variables needed to describe the motion**.

>Once the degrees of freedom are known, we can choose generalized coordinates that describe the system using only these independent variables.

### Examples

**Particle in free space**

- Coordinates: $(x,y,z)$  
- Constraints: none  

DOF:

$$
3
$$

---

**Particle constrained to a plane**

Constraint:

$$
z = 0
$$

DOF:

$$
3 - 1 = 2
$$

---

**Particle constrained to a circle**

Constraint:

$$
x^2 + y^2 = R^2
$$

DOF:

$$
2 - 1 = 1
$$

The motion can then be described by a single coordinate such as the angle $\theta$.

Identifying the degrees of freedom allows us to introduce **generalized coordinates** that describe the system using only the independent variables.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.1.3 Examples from Introductory Physics

Let us revisit familiar systems using the language of **constraints and degrees of freedom**.

### Pendulum (rigid rod of length $l$)

Coordinates of the bob:

$$
(x,y)
$$

Constraint:

$$
x^2 + y^2 - l^2 = 0
$$

Originally two coordinates are needed to specify the position in the plane.  
The constraint reduces this to

$$
\text{DOF} = 2 - 1 = 1
$$

A convenient coordinate describing the motion is the **angle** $\theta$.

In polar coordinates the constraint becomes

$$
r = l
$$

which automatically satisfies the constraint.

---

### Block sliding on a ramp

Coordinates of the block:

$$
(x,y)
$$

Constraint (equation of the ramp):

$$
y = 0
$$

The motion is restricted to one direction along the incline, so

$$
\text{DOF} = 2 - 1 = 1
$$

Thus the motion can be described using a single coordinate along the ramp.

---

### Bead on a circular hoop

Position coordinates:

$$
(x,y)
$$

Constraint:

$$
x^2 + y^2 = R^2
$$

Degrees of freedom:

$$
1
$$

The motion is fully described by the angular coordinate $\theta$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.1.4 Constraint Equations: General Form

In classical mechanics, the coordinates describing a system are often written as

$$
q_1, q_2, \dots, q_n
$$

These coordinates may represent Cartesian coordinates or more general variables suited to the geometry of the system.

A **constraint** is any condition that restricts the allowed values of these coordinates.

Many constraints can be written in the form

$$
f(q_1, q_2, \dots, q_n, t) = 0
$$

Each independent equation of this type reduces the number of independent coordinates by one.

Thus constraints determine the **true degrees of freedom of the system**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.1.5 Holonomic Constraints

A **holonomic constraint** is one that can be written as

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Holonomic Constraint Equation</h3>

$$
f(q_1, q_2, \dots, q_n, t) = 0
$$

</div>

These constraints depend only on the coordinates and possibly time.

Example:

Pendulum:

$$
x^2 + y^2 - l^2 = 0
$$

Each independent holonomic constraint **reduces the number of degrees of freedom by one**.

Holonomic constraints are particularly useful because they allow us to introduce **generalized coordinates** that automatically satisfy the constraints.

For example, instead of $(x,y)$ we may use

$$
q = \theta
$$

which uniquely specifies the configuration of the pendulum.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.1.6 Non-Holonomic Constraints

A **non-holonomic constraint** cannot be written purely as a relation among the coordinates.

Instead, it may involve

- velocities,
- differential relations, or
- inequalities.

Example:

**Rolling without slipping**

For a wheel of radius $R$,

$$
v_{\text{cm}} = R\omega
$$

This relates translational velocity and angular velocity rather than positions alone.

Another example is an inequality constraint describing a particle confined inside a sphere:

$$
x^2 + y^2 + z^2 \le R^2
$$

Such constraints cannot generally be eliminated by simply redefining coordinates and require more advanced methods.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.1.7 Why Constraints Matter

Constraints are fundamental in analytical mechanics because they allow us to

- determine the **degrees of freedom**,
- reduce the number of coordinates needed to describe motion,
- introduce **generalized coordinates** adapted to the geometry of the system,
- simplify the equations of motion.

Rather than describing motion using redundant Cartesian coordinates, we focus only on the **independent variables that truly describe the system**.

This idea will become central when we develop **Lagrangian mechanics**, where the equations of motion are written directly in terms of generalized coordinates.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.1.8 Summary

- Constraints restrict the motion of a system.
- They reduce the number of independent coordinates needed to describe motion.
- The **degrees of freedom** of a system is the number of independent coordinates required to specify its configuration.
- Holonomic constraints can be written as coordinate equations

$$
f(q_1,q_2,\dots,t)=0
$$

- Non-holonomic constraints involve velocities or inequalities and require more advanced treatment.

Understanding constraints and degrees of freedom is the first step toward formulating the equations of motion using **generalized coordinates and the Lagrangian formalism**.

<h3 style="margin-top:0; color:#000000;">Example – Degrees of Freedom of a Pendulum</h3>

<strong>Problem</strong>  

A particle of mass $m$ is attached to a rigid, massless rod of length $l$ and allowed to swing freely in a vertical plane.

1. Determine the number of **degrees of freedom** of the system if the position of the mass is described using Cartesian coordinates $(x,y)$.
2. Write the **constraint equation** imposed by the rigid rod.
3. Show how the constraint reduces the system to one independent coordinate.

---

<strong>Solution</strong>

<strong>1. Coordinates of the system</strong>

If we describe the position of the mass using Cartesian coordinates

$$
(x,y)
$$

then two coordinates are required to specify the position of the particle in the plane.

Thus, before considering constraints, the system has

$$
2
$$

possible coordinates.

---

<strong>2. Constraint equation</strong>

The rigid rod forces the particle to remain at a constant distance $l$ from the pivot.

This gives the constraint

$$
x^2 + y^2 = l^2.
$$

This equation restricts the motion of the particle to a **circle of radius $l$** centered at the pivot.

---

<strong>3. Determining the degrees of freedom</strong>

The number of degrees of freedom is

$$
\text{DOF} = \text{number of coordinates} - \text{number of constraints}.
$$

Here we have

$$
\text{DOF} = 2 - 1 = 1.
$$

Thus the motion of the pendulum can be completely described using a **single independent variable**.

---

<strong>4. Choosing a generalized coordinate</strong>

A convenient coordinate describing the motion is the angle $\theta$ measured from the vertical.

The Cartesian coordinates can then be written as

$$
x = l\sin\theta, \qquad y = -l\cos\theta.
$$

Using this coordinate automatically satisfies the constraint $x^2 + y^2 = l^2$.

---

<strong>Conclusion</strong>

Although the particle appears to require two coordinates $(x,y)$, the constraint imposed by the rigid rod reduces the motion to

$$
\text{DOF} = 1.
$$

This single degree of freedom is naturally described by the generalized coordinate $\theta$.

<h3 style="margin-top:0; color:#000000;">Example – Constraint Equation for a Simple Pulley</h3>

<strong>Problem</strong>  

Consider a simple pulley system:

- A frictionless pulley with a massless rope passes over it.  
- On the left side of the rope hangs mass $m_1$, and on the right side hangs mass $m_2$.  
- The rope is inextensible, so its total length remains constant.

Consider the following figure:  
<img src="D1.1-E1.png" width="400"/>

1. Write down the **constraint equation** relating the coordinates of $m_1$ and $m_2$.  
2. Derive the relationship between their **velocities** and **accelerations**.

---

<strong>Solution</strong>

<strong>1. Choosing coordinates</strong>

Let $y_1$ and $y_2$ be the vertical displacements of masses $m_1$ and $m_2$, each measured **downward** from a convenient reference level near the pulley.

Because the rope is massless and inextensible, its total length must remain constant during the motion.

The total rope length is made up of:

- the segment supporting $m_1$,
- the segment supporting $m_2$,
- and the small portion wrapped around the pulley.

The wrapped portion does not change, so it can be absorbed into a constant. Therefore the changing part of the rope length is simply

$$
y_1 + y_2.
$$

Since the total length is fixed, we must have

$$
y_1 + y_2 = L,
$$

or equivalently,

$$
y_1 + y_2 - L = 0.
$$

This is the **constraint equation**.

---

<strong>2. Why this is a holonomic constraint</strong>

This constraint involves only the coordinates $y_1$ and $y_2$ and a constant $L$. It does not involve velocities or accelerations.

Therefore it is a **holonomic constraint**.

It also tells us that the two coordinates are not independent. Once $y_1$ is known, $y_2$ is determined:

$$
y_2 = L - y_1.
$$

So the system has only **one independent degree of freedom**.

---

<strong>3. Relationship between the velocities</strong>

To find how the velocities are related, differentiate the constraint equation with respect to time:

$$
\frac{d}{dt}(y_1 + y_2 - L) = 0.
$$

Since $L$ is constant, its derivative is zero, so we obtain

$$
\dot{y}_1 + \dot{y}_2 = 0.
$$

Thus,

$$
\dot{y}_1 = -\dot{y}_2.
$$

This means the two masses always move with equal speed in opposite directions.

If one mass moves downward, the other must move upward at the same rate.

---

<strong>4. Relationship between the accelerations</strong>

Differentiate the velocity relation once more with respect to time:

$$
\frac{d}{dt}(\dot{y}_1 + \dot{y}_2) = 0.
$$

This gives

$$
\ddot{y}_1 + \ddot{y}_2 = 0.
$$

Therefore,

$$
\ddot{y}_1 = -\ddot{y}_2.
$$

So the accelerations are also equal in magnitude and opposite in direction.

---

<strong>Conclusion</strong>

The inextensible rope imposes the holonomic constraint

$$
y_1 + y_2 - L = 0.
$$

Differentiating this gives the velocity relation

$$
\dot{y}_1 = -\dot{y}_2,
$$

and differentiating again gives the acceleration relation

$$
\ddot{y}_1 = -\ddot{y}_2.
$$

Thus, the two masses cannot move independently: when one goes down, the other must go up with equal speed and equal-magnitude opposite acceleration.

<h3 style="margin-top:0; color:#000000;">Box Activity 1 – Constraint on a Block Sliding Down an Inclined Plane</h3>

In this activity, you will identify how a **constraint restricts motion** and see exactly **where that constraint enters the equations of motion**.

Consider a block sliding down a rough inclined plane.

An observer is positioned at the **top of the incline**, and the coordinate system is chosen so that:

- The **$x$-axis is aligned with the surface of the incline**, pointing **down the slope**.
- The **$y$-axis is perpendicular to the surface** of the incline.

This coordinate choice naturally reflects the geometry of the constraint imposed by the surface.

---

**Task:** Complete the following steps.

---

**1. Physical setup**

A block of mass $m$ slides down an incline of angle $\alpha$ with coefficient of kinetic friction $\mu_k$.

Forces acting on the block:

- gravitational force $mg$
- normal force $N$
- kinetic friction force $f_k$

---

**2. Identify the constraint**

Because the block remains in contact with the surface of the incline, its motion **cannot occur perpendicular to the plane**.

Write the constraint equation describing this condition.

(Hint: Think about the coordinate describing motion perpendicular to the surface.)

---

**3. Determine the degrees of freedom**

Using the coordinates $(x,y)$ defined above, determine:

- how many coordinates are required to describe the motion without constraints
- how the constraint reduces the system to a single independent coordinate

---

**4. Apply Newton’s Second Law perpendicular to the plane**

Write Newton’s Second Law in the $y$ direction.

Use the constraint you identified in Step 2 to simplify the equation.

Solve for the normal force $N$.

---

**5. Determine the friction force**

Using

$$
f_k = \mu_k N,
$$

write an expression for the kinetic friction force acting on the block.

Be careful with the **direction** of the force relative to the chosen coordinate system.

---

**6. Apply Newton’s Second Law along the incline**

Write Newton’s Second Law along the $x$ direction.

Include the gravitational component along the incline and the friction force.

Solve for the acceleration of the block.

---

**7. Reflection**

- Where exactly did the **constraint equation** enter the solution?  
- Why does the constraint eliminate acceleration perpendicular to the plane?  
- How did choosing coordinates aligned with the incline simplify the problem?

---

<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 2: Constraint equation**

The block remains on the surface of the incline, so its perpendicular displacement from the surface is zero.

Thus,

$$
y = 0.
$$

This is a **holonomic constraint**.

---

**Step 3: Degrees of freedom**

Without constraints, the block would require two coordinates:

$$
(x,y).
$$

The constraint $y=0$ removes one coordinate.

Therefore,

$$
\text{DOF} = 2 - 1 = 1.
$$

The motion is fully described by the single coordinate

$$
x.
$$

---

**Step 4: Newton’s Second Law perpendicular to the plane**

Forces perpendicular to the surface:

- normal force $N$ (outward)
- component of gravity $mg\cos\alpha$ (into the surface)

Newton’s Second Law:

$$
\sum F_y = ma_y.
$$

But the constraint implies

$$
y = 0 \quad \Rightarrow \quad \dot y = 0 \quad \Rightarrow \quad \ddot y = 0.
$$

Thus

$$
a_y = 0.
$$

Therefore

$$
N - mg\cos\alpha = 0.
$$

Solving for the normal force:

$$
N = mg\cos\alpha.
$$

This is the **exact step where the constraint is used**.

---

**Step 5: Friction force**

The kinetic friction force is

$$
f_k = \mu_k N.
$$

Substituting the normal force:

$$
f_k = \mu_k mg\cos\alpha.
$$

The friction force acts **up the incline**, opposite the motion.

---

**Step 6: Newton’s Second Law along the incline**

Forces along the incline:

- gravitational component down the plane: $mg\sin\alpha$
- friction up the plane: $f_k$

Newton’s Second Law:

$$
mg\sin\alpha - f_k = ma.
$$

Substitute the friction force:

$$
mg\sin\alpha - \mu_k mg\cos\alpha = ma.
$$

Divide by $m$:

$$
a = g(\sin\alpha - \mu_k\cos\alpha).
$$

---

**Interpretation**

The constraint $y=0$ ensures that the block cannot accelerate perpendicular to the surface.

This allowed us to set

$$
a_y = 0,
$$

which determines the normal force and ultimately the friction force.

The entire dynamics of the system is therefore governed by the **single coordinate $x$**, the distance along the incline.

<h3 style="margin-top:0; color:#000000;">Box Activity 2 – Constraint on a Block Sliding Down an Inclined Plane (Horizontal–Vertical Frame)</h3>

In the previous activity, we analyzed a block sliding down an incline using a coordinate system **aligned with the incline**.

In this activity, we analyze the **same physical system**, but we describe it using the **standard Cartesian coordinate system**:

- The **$x$-axis is horizontal**
- The **$y$-axis is vertical** (positive upward)

The top of the incline is at height $H$ above the base and the block starts at the point $(0,H)$.

---

**Problem**

A block of mass $m$ slides down a rough incline of angle $\alpha$ with coefficient of kinetic friction $\mu_k$.

Using the coordinate system described above:

1. Determine the **constraint equation** relating the coordinates $x$ and $y$ that ensures the block remains on the incline.

2. Show that the constraint reduces the motion to **one degree of freedom**.

3. Apply Newton’s Second Law in the horizontal and vertical directions and use the constraint to determine:
   - the **acceleration components relative to the observer**, $(a_x, a_y)$,
   - the **acceleration along the incline**.

Follow the **same reasoning used in Box Activity 1**, but now using the horizontal–vertical coordinate system.

---

**Important observation**

Constraint equations are not unique.  
We are free to **add constants or rearrange terms** without changing the physical constraint.

For example

$$
H - y = x\tan\alpha
$$

and

$$
y = H - x\tan\alpha
$$

represent the **same constraint**.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1: Constraint equation**

The block remains on the surface of the incline.

At the top of the ramp the coordinates are

$$
(x,y) = (0,H).
$$

As the block moves horizontally a distance $x$, its vertical height decreases according to the slope of the incline.

Thus the geometric relation between $x$ and $y$ is

$$
H - y = x\tan\alpha.
$$

or equivalently

$$
H - y - x\tan\alpha = 0.
$$

This is a **holonomic constraint**.

---

**Step 2: Degrees of freedom**

Without constraints the system requires

$$
(x,y)
$$

so there are two coordinates.

The constraint reduces this to

$$
\text{DOF} = 2 - 1 = 1.
$$

Thus the motion can be described using a single independent coordinate.

---

**Step 3: Velocity and acceleration relations**

Differentiate the constraint equation.

Starting from

$$
H - y = x\tan\alpha
$$

First derivative:

$$
-\dot y = \dot x\tan\alpha
$$

so

$$
\dot y = -\dot x\tan\alpha.
$$

Second derivative:

$$
\ddot y = -\ddot x\tan\alpha.
$$

Thus the acceleration components are related by

$$
a_y = -a_x\tan\alpha.
$$

---

**Step 4: Forces acting on the block**

Forces acting on the block:

- gravitational force $mg$ (downward)
- normal force $N$
- kinetic friction $f_k$

with

$$
f_k = \mu_k N.
$$

---

**Step 5: Force components**

The normal force is perpendicular to the incline.

Its components in the horizontal–vertical frame are

Horizontal component

$$
N_x = N\sin\alpha
$$

Vertical component

$$
N_y = N\cos\alpha
$$

The friction force acts **up the incline**.

Its components are

Horizontal

$$
f_x = -f_k\cos\alpha
$$

Vertical

$$
f_y = f_k\sin\alpha.
$$

---

**Step 6: Newton’s Second Law**

Horizontal direction

$$
\sum F_x = m a_x
$$

$$
N\sin\alpha - f_k\cos\alpha = m a_x.
$$

Vertical direction

$$
\sum F_y = m a_y
$$

$$
N\cos\alpha + f_k\sin\alpha - mg = m a_y.
$$

---

**Step 7: Use the constraint relation**

From the constraint

$$
a_y = -a_x\tan\alpha.
$$

Substitute into the vertical equation

$$
N\cos\alpha + f_k\sin\alpha - mg = -m a_x\tan\alpha.
$$

Using

$$
f_k = \mu_k N,
$$

solve together with the horizontal equation.

---

**Step 8: Acceleration relative to the observer**

Solving the equations yields

$$
a_x = g(\sin\alpha\cos\alpha - \mu_k\cos^2\alpha)
$$

and using the constraint

$$
a_y = -a_x\tan\alpha.
$$

These are the **horizontal and vertical acceleration components** observed in this frame.

---

**Step 9: Acceleration along the incline**

The acceleration along the incline is obtained by projecting the acceleration vector onto the direction of the incline.

This gives

$$
a = g(\sin\alpha - \mu_k\cos\alpha).
$$

---

**Interpretation**

Although the coordinate system is different, the physical acceleration along the incline is the **same result obtained in Box Activity 1**.

This illustrates a fundamental principle:

> The laws of mechanics are independent of the coordinate system used to describe the motion.
