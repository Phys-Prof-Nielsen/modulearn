# D8.3 Coordinate Systems and Coordinate Transformations
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.3.1 Motivation: Many Ways to Describe the Same Motion

In classical mechanics, the **physical motion of a system does not depend on the coordinates we choose**. What changes is only how we describe that motion mathematically.

For example, the position of a particle in the plane can be described using

- **Cartesian coordinates:** $(x,y)$  
- **Polar coordinates:** $(r,\theta)$  

Both coordinate systems describe the **same physical location**. The choice simply changes how we represent the position mathematically.

This freedom is extremely important in analytical mechanics. A good coordinate choice

- simplifies the equations,
- automatically satisfies constraints,
- reveals the symmetry of the problem.

Generalized coordinates allow us to select **whatever variables are most natural for the system**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.3.2 Coordinate Transformations

Different coordinate systems are related through **coordinate transformations**.

For example, the transformation between Cartesian and polar coordinates is

$$
x = r\cos\theta
$$

$$
y = r\sin\theta
$$

and the inverse relations are

$$
r = \sqrt{x^2 + y^2}
$$

$$
\theta = \tan^{-1}\!\left(\frac{y}{x}\right).
$$

These equations show that the same physical point can be described using either coordinate system.

In analytical mechanics we write this idea more generally as

$$
x_i = x_i(q_1, q_2, \dots, q_N, t)
$$

where

- $x_i$ are Cartesian coordinates  
- $q_i$ are generalized coordinates.

Thus the Cartesian position of a particle can always be expressed as a function of the generalized coordinates.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.3.3 Transformations of Velocity

If the position depends on generalized coordinates

$$
\mathbf{r} = \mathbf{r}(q_1,q_2,\dots,q_N,t),
$$

then the velocity is obtained by differentiating with respect to time (total derivative):

$$
\mathbf{v} = \frac{d\mathbf{r}}{dt}.
$$

Since $\mathbf{r}$ depends on several variables, we apply the **chain rule**:

$$
\mathbf{v} =
\sum_{i=1}^{N}
\frac{\partial \mathbf{r}}{\partial q_i}\dot q_i
+
\frac{\partial \mathbf{r}}{\partial t}.
$$

This relation is fundamental in analytical mechanics because it allows us to express the velocity entirely in terms of

$$
q_i \quad \text{and} \quad \dot q_i .
$$

Once the velocity is known, we can write the **kinetic energy**

$$
T = \frac{1}{2} m v^2
$$

directly in generalized coordinates.

>The chain rule provides the bridge between coordinate choice and dynamics, allowing velocities and kinetic energy to be expressed in generalized coordinates.

This step is essential when constructing the **Lagrangian**, where the dynamics of the system are expressed in terms of generalized coordinates and generalized velocities.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.3.4 Choosing Good Coordinates

Although any set of independent coordinates can describe a system, some choices make the physics much simpler.

Good coordinate choices typically

- match the **geometry of the constraints**,  
- reflect **symmetries of the system**,  
- reduce the number of variables needed.

Examples:

| System | Good Coordinate Choice |
|------|------|
| particle in plane | $(x,y)$ or $(r,\theta)$ |
| pendulum | $\theta$ |
| bead on circular hoop | $\theta$ |
| block on incline | distance along plane $s$ |
| rigid body rotation | angular coordinates |

Choosing coordinates adapted to the system often makes the mathematics **dramatically simpler**.

For example, describing a pendulum in Cartesian coordinates requires solving a constraint equation, whereas using $\theta$ automatically satisfies the constraint.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.3.5 Example: Pendulum Coordinate Transformation

Consider a pendulum of length $l$.

Using the generalized coordinate

$$
q = \theta
$$

the Cartesian coordinates of the mass are

$$
x = l\sin\theta
$$

$$
y = -l\cos\theta.
$$

Differentiating with respect to time gives the velocity components

$$
\dot{x} = l\cos\theta\,\dot{\theta}
$$

$$
\dot{y} = l\sin\theta\,\dot{\theta}.
$$

The speed of the mass is therefore

$$
v = \sqrt{\dot{x}^2 + \dot{y}^2}.
$$

Substituting the velocity components,

$$
v = \sqrt{l^2\cos^2\theta\,\dot{\theta}^2 + l^2\sin^2\theta\,\dot{\theta}^2}.
$$

Using

$$
\cos^2\theta + \sin^2\theta = 1,
$$

we obtain

$$
v = l\dot{\theta}.
$$

Thus the kinetic energy becomes

$$
T = \frac{1}{2} m l^2 \dot{\theta}^2.
$$

This simple expression illustrates why generalized coordinates are so powerful:  
the constraint has been incorporated directly into the coordinate choice.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.3.6 Summary

- The same physical motion can be described using many coordinate systems.
- Coordinate transformations relate Cartesian coordinates to generalized coordinates.
- Velocities are obtained from coordinate transformations using the **chain rule**.
- Choosing coordinates that match the geometry and symmetry of a system simplifies mechanics.

These ideas prepare us for the next step in analytical mechanics: writing the equations of motion using the **Lagrangian formulation**, where the dynamics are expressed entirely in terms of

$$
q_i \quad \text{and} \quad \dot q_i .
$$

<h3 style="margin-top:0; color:#000000;">Example – Coordinate Transformation and Velocity for a Particle in Polar Coordinates</h3>

<strong>Problem</strong>  

A particle moves in the plane. Instead of describing the motion using Cartesian coordinates $(x,y)$, we use **polar coordinates** $(r,\theta)$.

1. Write the **coordinate transformation** between Cartesian and polar coordinates.
2. Determine the **velocity components** $(\dot x,\dot y)$ in terms of $r$, $\theta$, $\dot r$, and $\dot\theta$.
3. Compute the **speed of the particle**.
4. Write the **kinetic energy** in polar coordinates.

---

<strong>Solution</strong>

<strong>1. Coordinate transformation</strong>

In polar coordinates, the position of the particle is described by:

- $r$: the distance from the origin  
- $\theta$: the angle measured from the $x$-axis

The Cartesian coordinates are related to polar coordinates by

$$
x = r\cos\theta
$$

$$
y = r\sin\theta
$$

These equations represent a **coordinate transformation**, expressing the same physical position using different variables.

---

<strong>2. Velocity components</strong>

To determine the velocity, differentiate the coordinates with respect to time.

Horizontal velocity:

$$
\dot x = \frac{d}{dt}(r\cos\theta)
$$

Using the product rule,

$$
\dot x = \dot r\cos\theta - r\sin\theta\,\dot\theta.
$$

Vertical velocity:

$$
\dot y = \frac{d}{dt}(r\sin\theta)
$$

$$
\dot y = \dot r\sin\theta + r\cos\theta\,\dot\theta.
$$

Thus the Cartesian velocity components are

$$
\dot x = \dot r\cos\theta - r\sin\theta\,\dot\theta
$$

$$
\dot y = \dot r\sin\theta + r\cos\theta\,\dot\theta.
$$

---

<strong>3. Speed of the particle</strong>

The speed is

$$
v = \sqrt{\dot x^2 + \dot y^2}.
$$

Substituting the expressions for $\dot x$ and $\dot y$ gives

$$
v^2 =
(\dot r\cos\theta - r\sin\theta\,\dot\theta)^2
+
(\dot r\sin\theta + r\cos\theta\,\dot\theta)^2.
$$

Expand the squares:

$$
v^2 =
\dot r^2\cos^2\theta
-2r\dot r\sin\theta\cos\theta\,\dot\theta
+r^2\sin^2\theta\,\dot\theta^2
$$

$$
+
\dot r^2\sin^2\theta
+2r\dot r\sin\theta\cos\theta\,\dot\theta
+r^2\cos^2\theta\,\dot\theta^2.
$$

The cross terms cancel, leaving

$$
v^2 =
\dot r^2(\cos^2\theta+\sin^2\theta)
+
r^2\dot\theta^2(\sin^2\theta+\cos^2\theta).
$$

Using

$$
\sin^2\theta + \cos^2\theta = 1,
$$

we obtain

$$
v^2 = \dot r^2 + r^2\dot\theta^2.
$$

Thus the speed is

$$
v = \sqrt{\dot r^2 + r^2\dot\theta^2}.
$$

---

<strong>4. Kinetic energy</strong>

The kinetic energy of the particle is

$$
T = \frac{1}{2}mv^2.
$$

Substituting the expression for $v^2$,

$$
T = \frac{1}{2}m(\dot r^2 + r^2\dot\theta^2).
$$

---

<strong>Conclusion</strong>

The Cartesian and polar coordinate systems describe the **same physical motion**, but the variables used to represent that motion differ.

The coordinate transformation

$$
x = r\cos\theta, \qquad y = r\sin\theta
$$

leads to the velocity relation

$$
v^2 = \dot r^2 + r^2\dot\theta^2.
$$

This example illustrates how changing coordinates alters the **mathematical description** of motion while leaving the underlying physics unchanged.

<h3 style="margin-top:0; color:#000000;">Box Activity – Coordinate Transformation Using Parabolic Coordinates</h3>

In the previous example we used the **chain rule** to determine how velocities transform when switching from Cartesian coordinates to polar coordinates.

In this activity we apply the **same idea to a different coordinate system**: **parabolic coordinates**.

This reinforces the key idea:

> When coordinates change, velocities follow from the **chain rule**.

---

**Problem**

A particle moves in the plane. Instead of using Cartesian coordinates $(x,y)$, we describe its position using **parabolic coordinates** $(u,v)$ defined by

$$
x = uv
$$

$$
y = \frac{1}{2}(u^2 - v^2).
$$

---

**Task**

1. **Compute the velocity components**

Using the chain rule, determine

$$
\dot{x} = \frac{dx}{dt}
\qquad \text{and} \qquad
\dot{y} = \frac{dy}{dt}
$$

in terms of $u$, $v$, $\dot u$, and $\dot v$.

---

2. **Compute the speed**

Use

$$
v^2 = \dot{x}^2 + \dot{y}^2
$$

to show that

$$
v^2 = (u^2 + v^2)(\dot{u}^2 + \dot{v}^2).
$$

---

3. **Interpretation**

- Which factors depend on the **coordinates**?
- Which factors depend on the **generalized velocities**?

Why does the coordinate system introduce the factor

$$
u^2 + v^2
$$

in the expression for the speed?

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1 – Differentiate $x$**

Starting with

$$
x = uv
$$

apply the product rule:

$$
\dot{x} = u\dot{v} + v\dot{u}.
$$

---

**Step 2 – Differentiate $y$**

Starting with

$$
y = \frac{1}{2}(u^2 - v^2)
$$

differentiate with respect to time:

$$
\dot{y} = u\dot{u} - v\dot{v}.
$$

---

**Step 3 – Compute the speed**

The speed of the particle is

$$
v^2 = \dot{x}^2 + \dot{y}^2.
$$

Substitute the velocity components:

$$
v^2 =
(u\dot{v} + v\dot{u})^2
+
(u\dot{u} - v\dot{v})^2.
$$

Expand the squares:

$$
v^2 =
u^2\dot{v}^2
+
2uv\dot{u}\dot{v}
+
v^2\dot{u}^2
+
u^2\dot{u}^2
-
2uv\dot{u}\dot{v}
+
v^2\dot{v}^2.
$$

The cross terms cancel, leaving

$$
v^2 =
(u^2 + v^2)\dot{u}^2
+
(u^2 + v^2)\dot{v}^2.
$$

Factor the common term:

$$
v^2 =
(u^2 + v^2)(\dot{u}^2 + \dot{v}^2).
$$

---

**Interpretation**

The velocity depends on both

- the **generalized velocities** $(\dot{u},\dot{v})$
- the **coordinates** $(u,v)$

This happens because the geometry of the coordinate system changes the way distances are measured.

Thus the kinetic energy becomes

$$
T = \frac{1}{2}m (u^2 + v^2)(\dot{u}^2 + \dot{v}^2).
$$

This illustrates how the **chain rule naturally produces the kinetic energy expression in generalized coordinates**.
