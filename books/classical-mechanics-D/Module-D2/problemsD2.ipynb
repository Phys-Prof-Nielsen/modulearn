# D2.P Problems — Index Notation Practice
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

These problems help you practice translating familiar vector equations from **standard vector notation** into **index notation** using the Einstein summation convention.

Throughout these problems, assume that indices run over

$$
i = 1,2,3.
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 1: Translating Mechanics Equations into Index Notation

Many equations from introductory physics involve vectors. Rewrite the following equations using **index notation**.

1. Newton's second law  

$$
\mathbf{F} = m\mathbf{a}
$$

2. Momentum  

$$
\mathbf{p} = m\mathbf{v}
$$

3. Dot product  

$$
\mathbf{a}\cdot\mathbf{b}
$$

4. Kinetic energy  

$$
T = \frac{1}{2} m v^2
$$

5. Angular momentum  

$$
\mathbf{L} = \mathbf{r}\times\mathbf{p}
$$

6. Torque  

$$
\boldsymbol{\tau} = \mathbf{r}\times\mathbf{F}
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

1. Newton's second law  

The vector equation represents three component equations. In index notation

$$
F_i = m a_i.
$$

---

2. Momentum  

$$
p_i = m v_i.
$$

---

3. Dot product  

$$
\mathbf{a}\cdot\mathbf{b} = a_i b_i.
$$

---

4. Kinetic energy  

Since

$$
v^2 = \mathbf{v}\cdot\mathbf{v},
$$

we obtain

$$
T = \frac{1}{2} m v_i v_i.
$$

---

5. Angular momentum  

Using the Levi–Civita symbol,

$$
L_i = \epsilon_{ijk} r_j p_k.
$$

---

6. Torque  

Similarly,

$$
\tau_i = \epsilon_{ijk} r_j F_k.
$$

---

### Interpretation

Index notation allows familiar vector equations to be written in a compact form that automatically represents all components simultaneously. This notation will become especially useful when working with rotational dynamics and tensor quantities later in the course.

## Problem 2: Magnitudes and Dot Products

Rewrite the following expressions using **index notation**.

1. Magnitude squared of position  

$$
r^2 = x^2 + y^2 + z^2
$$

2. Speed squared  

$$
v^2 = v_x^2 + v_y^2 + v_z^2
$$

3. Work done by a constant force  

$$
W = \mathbf{F}\cdot\Delta\mathbf{r}
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

1.

$$
r^2 = r_i r_i
$$

---

2.

$$
v^2 = v_i v_i
$$

---

3.

$$
W = F_i \Delta r_i
$$

</div>
</details>

## Problem 3: Component Interpretation

The indexed equation

$$
F_i = m a_i
$$

represents three equations simultaneously.

1. Write out the three equations explicitly using Cartesian components.

2. Explain why index notation allows us to write all three equations using a single compact expression.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The free index $i$ represents three separate equations:

For $i=1$

$$
F_1 = m a_1
$$

For $i=2$

$$
F_2 = m a_2
$$

For $i=3$

$$
F_3 = m a_3
$$

In Cartesian coordinates these correspond to

$$
F_x = m a_x
$$

$$
F_y = m a_y
$$

$$
F_z = m a_z.
$$

Thus the indexed equation

$$
F_i = m a_i
$$

represents all three component equations simultaneously.

</div>
</details>
