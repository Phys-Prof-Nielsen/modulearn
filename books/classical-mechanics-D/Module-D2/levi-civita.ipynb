# D2.4 The Levi–Civita Symbol and the Cross Product
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D2.4.1 Motivation: Writing the Cross Product Compactly

In three-dimensional mechanics we frequently encounter the **cross product** of two vectors. For example:

- **angular momentum**

$$
\mathbf{L} = \mathbf{r}\times\mathbf{p}
$$

- **torque**

$$
\boldsymbol{\tau} = \mathbf{r}\times\mathbf{F}
$$

In Cartesian components the cross product is written as

$$
\mathbf{a}\times\mathbf{b}
=
\begin{pmatrix}
a_y b_z - a_z b_y \\
a_z b_x - a_x b_z \\
a_x b_y - a_y b_x
\end{pmatrix}.
$$

While this expression is familiar, it becomes cumbersome in more complicated calculations. Index notation provides a compact way to write the cross product using a special object called the **Levi–Civita symbol**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Definition — Levi–Civita Symbol</h3>

The **Levi–Civita symbol** is written

$$
\epsilon_{ijk}.
$$

It is defined by

$$
\epsilon_{ijk} =
\begin{cases}
+1 & \text{if } (i,j,k) \text{ is an even permutation of } (1,2,3) \\
-1 & \text{if } (i,j,k) \text{ is an odd permutation of } (1,2,3) \\
0 & \text{if any two indices are equal}
\end{cases}
$$

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D2.4.2 Even and Odd Permutations

The Levi–Civita symbol depends on the ordering of the indices.

### Even permutations

$$
\epsilon_{123} = 1
$$

$$
\epsilon_{231} = 1
$$

$$
\epsilon_{312} = 1
$$

### Odd permutations

$$
\epsilon_{132} = -1
$$

$$
\epsilon_{213} = -1
$$

$$
\epsilon_{321} = -1
$$

If any two indices are equal, the value is zero:

$$
\epsilon_{112} = 0
$$

because two indices are the same.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D2.4.3 Cross Product in Index Notation

Using the Levi–Civita symbol, the cross product can be written compactly as

$$
(\mathbf{a}\times\mathbf{b})_i =
\epsilon_{ijk} a_j b_k.
$$

Here:

- $i$ is a **free index**, specifying the component of the resulting vector.
- $j$ and $k$ are **summed indices**.

This single equation represents all three components of the cross product simultaneously.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D2.4.4 Example: Computing a Cross Product

Let

$$
\mathbf{a} = (a_1,a_2,a_3)
$$

and

$$
\mathbf{b} = (b_1,b_2,b_3).
$$

Using

$$
(\mathbf{a}\times\mathbf{b})_i =
\epsilon_{ijk} a_j b_k
$$

we compute the first component.

For $i=1$,

$$
(\mathbf{a}\times\mathbf{b})_1 =
\epsilon_{1jk} a_j b_k.
$$

Expanding the possible combinations gives

$$
\epsilon_{123} a_2 b_3
+
\epsilon_{132} a_3 b_2.
$$

Using the definition of $\epsilon_{ijk}$,

$$
\epsilon_{123}=1,
\qquad
\epsilon_{132}=-1,
$$

so

$$
(\mathbf{a}\times\mathbf{b})_1 =
a_2 b_3 - a_3 b_2.
$$

This matches the familiar formula for the $x$ component of the cross product.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Example — Verifying the Second Component</h3>

<strong>Problem</strong>

Use the expression

$$
(\mathbf{a}\times\mathbf{b})_i =
\epsilon_{ijk} a_j b_k
$$

to compute the second component of the cross product.

---

<strong>Solution</strong>

For $i=2$,

$$
(\mathbf{a}\times\mathbf{b})_2 =
\epsilon_{2jk} a_j b_k.
$$

Expanding the nonzero terms gives

$$
\epsilon_{231} a_3 b_1
+
\epsilon_{213} a_1 b_3.
$$

Using the Levi–Civita definition,

$$
\epsilon_{231}=1,
\qquad
\epsilon_{213}=-1.
$$

Thus

$$
(\mathbf{a}\times\mathbf{b})_2 =
a_3 b_1 - a_1 b_3.
$$

This is the familiar expression for the $y$ component of the cross product.

For $i=3$,

$$
(\mathbf{a}\times\mathbf{b})_3 =
\epsilon_{3jk} a_j b_k.
$$

The nonzero terms occur for

$$
(j,k) = (1,2) \quad \text{and} \quad (2,1).
$$

Thus

$$
(\mathbf{a}\times\mathbf{b})_3 =
\epsilon_{312} a_1 b_2
+
\epsilon_{321} a_2 b_1.
$$

Using the Levi–Civita values,

$$
\epsilon_{312} = 1,
\qquad
\epsilon_{321} = -1,
$$

we obtain

$$
(\mathbf{a}\times\mathbf{b})_3 =
a_1 b_2 - a_2 b_1.
$$

Thus the Levi–Civita expression correctly reproduces the standard cross product formula.

## D2.4.5 Why the Levi–Civita Symbol Is Important

The Levi–Civita symbol provides a compact way to express operations involving **antisymmetry**, such as the cross product.

It also allows many vector identities to be derived in a systematic way. Later in the course we will use expressions involving $\epsilon_{ijk}$ together with the **Kronecker delta** to simplify calculations in rotational dynamics.

These tools form the foundation of the **tensor notation** widely used throughout advanced physics.
