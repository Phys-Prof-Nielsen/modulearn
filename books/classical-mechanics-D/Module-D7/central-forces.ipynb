# D7.1 Central Forces and Angular Momentum
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.1.1 Motivation: Motion Under Radial Forces

In many physical systems, the force acting on a particle depends only on its **distance from a central point**.

Examples include:

- gravitational attraction between two masses  
- electrostatic forces between charged particles  
- motion of planets and satellites  
- particles constrained to move around a center  

In these situations, the force always points **toward or away from a fixed point**.

This naturally suggests using **polar coordinates**, where motion is already separated into:

- radial direction  
- angular direction  

---

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Key Idea</h3>

Central forces are best described in coordinate systems that match their symmetry.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.1.2 Definition — Central Force

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Definition — Central Force</h3>

A **central force** is a force that:

1. depends only on the distance $r$ from a fixed point  
2. points along the radial direction  

$$
\vec{F}(r) = F(r)\,\hat{r}
$$

</div>

---

### Key properties

- No angular component:

$$
F_\theta = 0
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.1.3 Newton’s Second Law For Central Forces

Recall:

$$
\vec{a} =
(\ddot{r} - r\dot{\theta}^2)\hat{r}
+
(2\dot{r}\dot{\theta} + r\ddot{\theta})\hat{\theta}
$$

Applying Newton’s second law:

$$
\vec{F} = m\vec{a}
$$

---

### Radial equation

$$
F(r) = m(\ddot{r} - r\dot{\theta}^2)
$$

---

### Angular equation

Since $F_\theta = 0$:

$$
0 = m(2\dot{r}\dot{\theta} + r\ddot{\theta})
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.1.4 Angular Equation → Conservation of Angular Momentum

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<h3 style="margin-top:0;">Box Activity — Show that Orbital Angular Momentum is Conserved</h3>

Consider a particle moving under a **central force**

$$
\vec{F}(r) = F(r)\,\hat{r}.
$$

---

**Task**

Using Newton’s second law in polar coordinates, show that

$$
L = m r^2 \dot{\theta}
$$

is constant in time.

---

**Steps to guide your reasoning**

1. Start with the angular component of acceleration:

$$
a_\theta = 2\dot{r}\dot{\theta} + r\ddot{\theta}
$$

2. Apply Newton’s second law in the $\hat{\theta}$ direction

3. Use the fact that a central force has **no angular component**

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

From Newton’s second law:

$$
F_\theta = m a_\theta
$$

For a central force:

$$
F_\theta = 0
$$

Thus,

$$
0 = m(2\dot{r}\dot{\theta} + r\ddot{\theta})
$$

Divide by $m$:

$$
2\dot{r}\dot{\theta} + r\ddot{\theta} = 0
$$

---

Multiply by $r$:

$$
2r\dot{r}\dot{\theta} + r^2\ddot{\theta} = 0
$$

Recognize this as a derivative:

$$
\frac{d}{dt}(r^2 \dot{\theta}) = 0
$$

---

Therefore,

$$
r^2 \dot{\theta} = \text{constant}
$$

Multiplying by $m$:

$$
\boxed{
L = m r^2 \dot{\theta} = \text{constant}
}
$$

which we recognize as the **orbital angular momentum for a point particle**.

---

**Interpretation**

- Orbital Angular momentum is conserved  
- This follows directly from the absence of angular force  
- The result is purely a consequence of the **symmetry of the force**

</div>
</details>

</div>

---

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Result — Angular Momentum Conservation</h3>

$$
L = m r^2 \dot{\theta} = \text{constant}
$$

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.1.5 Physical Meaning of Angular Momentum Conservation

From the previous activity, we found:

$$
L = m r^2 \dot{\theta} = \text{constant}.
$$

This result has two powerful physical consequences.

---

### Motion stays in a plane

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Key Idea — No Torque, No Change in Direction of $\vec{L}$</h3>

A central force always points along $\hat{r}$:

$$
\vec{F} = F(r)\hat{r}.
$$

This means the torque is

$$
\vec{\tau} = \vec{r} \times \vec{F} = 0.
$$

</div>

---

#### What this implies

- No torque → angular momentum vector $\vec{L}$ is **constant in direction and magnitude**
- Since

$$
\vec{L} = \vec{r} \times \vec{p},
$$

the vectors $\vec{r}$ and $\vec{v}$ must always lie in a plane **perpendicular to $\vec{L}$**

👉 Therefore, the motion is confined to a **single plane**

---

#### Physical picture

- The particle may move in complicated ways within the plane  
- But it can never “tilt” out of that plane  
- The plane is fixed by the initial position and velocity  

---

### Equal areas in equal times (Kepler’s Second Law)

From

$$
L = m r^2 \dot{\theta},
$$

we can interpret this geometrically.

#### Area of a Small “Pie Slice”

Over a small time interval $dt$, the particle moves through a small angle $d\theta$ at a distance $r$ from the origin.

The area swept out is approximately a **sector of a circle** (a “pie slice”):

$$
dA = \frac{1}{2} r^2 d\theta
$$

---

#### Why this formula?

The area of a full circle is:

$$
A = \pi r^2
$$

A full circle corresponds to an angle of $2\pi$, so a fraction of the circle with angle $d\theta$ has area:

$$
dA = \frac{d\theta}{2\pi} \cdot \pi r^2 = \frac{1}{2} r^2 d\theta
$$

---

#### Areal velocity

Divide by time:

$$
\frac{dA}{dt} = \frac{1}{2} r^2 \dot{\theta}
$$

---

#### Connection to angular momentum

Since

$$
L = m r^2 \dot{\theta},
$$

we get:

$$
\frac{dA}{dt} = \frac{L}{2m} = \text{constant}
$$

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Key Idea — Equal Areas in Equal Time</h3>

$$
\frac{dA}{dt} = \text{constant}
$$

</div>

---

#### Physical meaning

- When the particle is **closer to the center**:
  - $r$ is small  
  - $\dot{\theta}$ must increase  
  → motion speeds up  

- When the particle is **farther away**:
  - $r$ is large  
  - $\dot{\theta}$ decreases  
  → motion slows down  

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Curiosity — Why This Happens</h3>

Angular momentum conservation forces a tradeoff:


small radius → faster rotation
large radius → slower rotation

</div>

👉 This is **Kepler’s Second Law**

## D7.1.6 Example — Gravitational Force

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Inverse-Square Central Force</h3>

The gravitational force between two masses is:

$$
\vec{F}(r) = -\frac{GMm}{r^2}\hat{r}
$$

Substitute into the radial equation:

$$
-\frac{GMm}{r^2} = m(\ddot{r} - r\dot{\theta}^2)
$$

Cancel $m$:

$$
\ddot{r} - r\dot{\theta}^2 = -\frac{GM}{r^2}
$$

This equation governs planetary motion.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.1.7 Why This Matters

Central force problems are fundamental because they lead to:

- planetary motion  
- orbital mechanics  
- conservation laws  
- deeper symmetry principles

## D7.1.8 Energy in Central Force Motion

The kinetic energy in polar coordinates is:

$$
T = \frac{1}{2}m\dot{r}^2 + \frac{1}{2}m r^2 \dot{\theta}^2.
$$

Using angular momentum:

$$
\dot{\theta} = \frac{L}{mr^2},
$$

we get:

$$
T = \frac{1}{2}m\dot{r}^2 + \frac{L^2}{2mr^2}.
$$

---

### Total Energy

$$
E = \frac{1}{2}m\dot{r}^2 + \frac{L^2}{2mr^2} + U(r).
$$

---
<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Inverse-Square Central Force</h3>
    
The motion separates into:
<ul>
<li>radial motion: $\frac{1}{2}m\dot{r}^2$</li>
<li>angular contribution: $\frac{L^2}{2mr^2}$</li>
</ul>
</div>

Central forces give us:

- Conservation of angular momentum  
- Planar motion

These results allow us to:

- reduce a 2D problem to a **1D radial problem**
