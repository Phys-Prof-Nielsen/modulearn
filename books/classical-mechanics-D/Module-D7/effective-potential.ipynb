## D7.2.1 Why Effective Potential?

From D7.1, we found that motion under a central force:

- is confined to a plane  
- conserves angular momentum  

We also found that the total energy can be written as:

$$
E = \frac{1}{2}m\dot{r}^2 + \frac{L^2}{2mr^2} + U(r).
$$

👉 This suggests something powerful:

> The motion can be treated as a **one-dimensional problem in $r$**.

---

### Key Idea

We combine terms to define a new quantity:

$$
U_{\text{eff}}(r) = U(r) + \frac{L^2}{2mr^2}.
$$

This allows us to rewrite the energy as:

$$
E = \frac{1}{2}m\dot{r}^2 + U_{\text{eff}}(r).
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.2.2 Definition: Effective Potential

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<strong>Definition: Effective Potential</strong>

The **effective potential** is defined as

$$
U_{\text{eff}}(r) = U(r) + \frac{L^2}{2mr^2}.
$$

It combines:
<ul>
<li>The actual potential energy $U(r)$</li>
<li>A term arising from angular motion</li>
</ul>

</div>

---

### Physical Meaning

- $U(r)$ → physical interaction (gravity, electrostatics, etc.)  
- $\frac{L^2}{2mr^2}$ → **centrifugal barrier**

👉 This term becomes very large as $r \to 0$, preventing collapse unless $L = 0$.

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Centrifugal Barrier — Physical Interpretation</h3>

The **centrifugal barrier** is the term

$$
\frac{L^2}{2mr^2}
$$

in the effective potential

$$
U_{\text{eff}}(r) = U(r) + \frac{L^2}{2mr^2}.
$$

---

### Where it comes from

Angular momentum is conserved:

$$
L = m r^2 \dot{\theta}.
$$

Solving for angular speed:

$$
\dot{\theta} = \frac{L}{m r^2}.
$$

As the particle moves closer to the center ($r \downarrow$), the angular speed must increase rapidly.

---

### Why it acts like a barrier

The kinetic energy associated with angular motion is:

$$
T_{\text{angular}} = \frac{1}{2} m r^2 \dot{\theta}^2 = \frac{L^2}{2mr^2}.
$$

As $r \to 0$:

$$
\frac{L^2}{2mr^2} \to \infty.
$$

This means:

- The energy required to reach small $r$ becomes extremely large  
- The particle cannot reach the center  

👉 This creates an **effective barrier** at small $r$.

---

### Effective force

Taking the derivative:

$$
F_{\text{eff}} = -\frac{d}{dr}\left(\frac{L^2}{2mr^2}\right)
= \frac{L^2}{mr^3}.
$$

This force is:

- **positive (outward)**  
- very large at small $r$  

👉 It behaves like a **repulsive force**, even if the actual force is attractive.

---

### Key insight

Even in an attractive potential like gravity,

$$
U(r) = -\frac{GMm}{r},
$$

the effective potential becomes:

$$
U_{\text{eff}}(r) = -\frac{GMm}{r} + \frac{L^2}{2mr^2}.
$$

So:

- Gravity pulls inward  
- Angular momentum pushes outward  

👉 Their competition creates stable orbits and prevents collapse.

---

### Special case

If $L = 0$:

- No centrifugal barrier  
- Motion is purely radial  
- The particle can fall directly inward  

---

<strong>Big idea:</strong>  
The centrifugal barrier is not a real force—it is a consequence of conserving angular momentum. It allows us to treat orbital motion as a one-dimensional energy problem in $r$.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.2.3 Radial Motion as a 1D Problem

The radial motion now satisfies:

$$
E = \frac{1}{2}m\dot{r}^2 + U_{\text{eff}}(r).
$$

---

### Interpretation

This is mathematically identical to a **1D particle moving in a potential**.

We can now use all tools from D6:

- turning points  
- energy diagrams  
- stability  
- oscillations  

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Insight:</strong><br>
A two-dimensional orbital problem has been reduced to a one-dimensional energy problem.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.2.4 Turning Points

Turning points occur when:

$$
E = U_{\text{eff}}(r).
$$

At these points:

- $\dot{r} = 0$  
- radial motion reverses direction  

---

### Allowed Motion

- Allowed region:
$$
E \ge U_{\text{eff}}(r)
$$

- Forbidden region:
$$
E < U_{\text{eff}}(r)
$$

---

### Interpretation

The particle oscillates between two radii:

$$
r_{\min}, \quad r_{\max}.
$$

These define the **radial extent of the orbit**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D7.2.5 Example: Gravitational Potential

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Effective Potential for Gravity</h3>

For gravity:

$$
U(r) = -\frac{GMm}{r}.
$$

The effective potential is:

$$
U_{\text{eff}}(r) = -\frac{GMm}{r} + \frac{L^2}{2mr^2}.
$$

</div>

---


```python
%reset -f
# Effective Potential for Gravity
# Circular, Bound/Elliptical, and Unbound Motion

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq

# -----------------------------
# Parameters (dimensionless units)
# -----------------------------
G = 1.0
M = 1.0
m = 1.0
L = 1.0

# Radial grid
r = np.linspace(0.15, 8.0, 2000)

# Effective potential
U_grav = -G * M * m / r
U_cent = L**2 / (2 * m * r**2)
U_eff = U_grav + U_cent

# Circular orbit occurs at minimum of U_eff
r_circ = L**2 / (G * M * m**2)
E_circ = -G * M * m / (2 * r_circ)

# Example energy levels
E_bound = -0.25     # bound elliptical-type orbit
E_unbound = 0.12    # unbound scattering trajectory

# -----------------------------
# Find turning points
# -----------------------------
def Ueff(x):
    return -G*M*m/x + L**2/(2*m*x**2)

def f_bound(x):
    return Ueff(x) - E_bound

def f_unbound(x):
    return Ueff(x) - E_unbound

# Bound orbit has two turning points
r1_bound = brentq(f_bound, 0.2, r_circ)
r2_bound = brentq(f_bound, r_circ, 8.0)

# Unbound orbit has one inner turning point
r_turn_unbound = brentq(f_unbound, 0.15, r_circ)

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(9, 6))

plt.plot(r, U_eff, linewidth=2, label=r"$U_{\rm eff}(r)$")
plt.plot(r, U_grav, linestyle="--", label=r"$U(r)=-GMm/r$")
plt.plot(r, U_cent, linestyle=":", label=r"$L^2/(2mr^2)$")

# Energy lines
plt.axhline(E_circ, linestyle="-.", label="Circular orbit energy")
plt.axhline(E_bound, linestyle="-.", label="Bound / elliptical orbit energy")
plt.axhline(E_unbound, linestyle="-.", label="Unbound orbit energy")

# Mark circular orbit
plt.scatter([r_circ], [E_circ], s=80, zorder=5)
plt.text(r_circ + 0.15, E_circ - 0.08, "circular orbit\nstable equilibrium")

# Mark bound turning points
plt.scatter([r1_bound, r2_bound], [E_bound, E_bound], s=60, zorder=5)
plt.text(r1_bound - 0.05, E_bound + 0.08, r"$r_{\min}$")
plt.text(r2_bound - 0.15, E_bound + 0.08, r"$r_{\max}$")
plt.text((r1_bound + r2_bound)/2 - 0.35, E_bound + 0.08, "bound / elliptical\nradial motion")

# Mark unbound turning point
plt.scatter([r_turn_unbound], [E_unbound], s=60, zorder=5)
plt.text(r_turn_unbound + 0.12, E_unbound + 0.06, "unbound\nturning point")

# Shade forbidden regions for the bound-energy case
plt.fill_between(
    r, E_bound, U_eff,
    where=(U_eff > E_bound),
    alpha=0.25,
    label="Forbidden for bound energy"
)

# Shade forbidden region for the unbound-energy case near small r
plt.fill_between(
    r, E_unbound, U_eff,
    where=(U_eff > E_unbound),
    alpha=0.15,
    label="Forbidden for unbound energy"
)

# Labels
plt.text(0.28, 1.2, "Forbidden\nregion", fontsize=11)
plt.text(5.5, E_bound - 0.18, "Forbidden for\nbound orbit", fontsize=11)

plt.xlabel(r"Radius $r$")
plt.ylabel("Energy")
plt.title("Effective Potential for Gravity")
plt.ylim(-0.7, 1.6)
plt.xlim(0.15, 8.0)
plt.grid(True)
plt.legend()
plt.show()
```

## D7.2.6 Interpreting the Effective Potential Diagram

The effective potential diagram provides a complete qualitative description of the motion.

We plot:

- the curve $U_{\text{eff}}(r)$  
- horizontal lines representing different total energies $E$  

The motion is determined by where:

$$
E \ge U_{\text{eff}}(r)
$$

---

### Circular Orbit (Stable Equilibrium)

At the minimum of the effective potential:

- $E = U_{\text{eff}}(r_0)$  
- $\frac{dU_{\text{eff}}}{dr} = 0$  

This corresponds to:

- constant radius $r = r_0$  
- a **circular orbit**  

Since the minimum is stable:

- small perturbations lead to oscillations about $r_0$  

---

### Bound (Elliptical-Type) Motion

For energies slightly above the minimum:

- the energy line intersects $U_{\text{eff}}(r)$ at two points  

$$
r_{\min}, \quad r_{\max}
$$

These are the **turning points**.

The motion:

- oscillates between $r_{\min}$ and $r_{\max}$  
- remains confined to a finite region  

👉 This corresponds to a **bound elliptical orbit**.

---

### Unbound Motion

For higher energies:

- the particle has only one turning point  
- it can escape to large $r$  

👉 This corresponds to **unbound motion** (scattering trajectory).

---

## Why Bound Orbits Have Negative Energy

For gravitational systems:

$$
U(r) = -\frac{GMm}{r}
$$

---

### Reference Point

By convention:

$$
U(\infty) = 0
$$

So:

- At large distance: $E \to 0$  
- Negative energy means the system is **below the escape threshold**

---

### Physical Interpretation

If the total energy is negative:

$$
E < 0
$$

then:

- the particle does not have enough energy to reach infinity  
- it is **trapped** in a finite region  

👉 This is exactly what we mean by a **bound orbit**.

---

### Escape Condition

To escape to infinity, the particle must satisfy:

$$
E \ge 0
$$

- $E = 0$ → marginal escape (parabolic trajectory)  
- $E > 0$ → unbound motion (hyperbolic trajectory)  

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Insight:</strong><br>
Negative energy does not mean “low energy”—it means the system is bound relative to the zero of energy at infinity.

</div>

---

## Key Takeaway

The effective potential diagram tells us:

- where motion is allowed  
- where turning points occur  
- whether the motion is bound or unbound  

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Big Picture:</strong><br>
Bound orbits correspond to energies below the escape threshold, while unbound motion occurs when the system has enough energy to reach infinity.

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Big Picture:</strong><br>
Effective potential transforms orbital motion into a familiar one-dimensional problem, making complex motion much easier to understand.<br><br>

This idea is not limited to gravitational systems. It also appears in **electrostatic (Coulomb) systems**, such as the attraction between an electron and a proton in the hydrogen atom.<br><br>

In early quantum mechanics, the electron was modeled using **Bohr orbits**, where only certain circular orbits were allowed. These orbits correspond to specific energies—still negative, meaning the electron is bound—but now **quantized** rather than continuous.<br><br>

Later, this was understood more deeply through wave mechanics: the electron behaves like a wave, and only those orbits that form **standing waves** are allowed. This imposes constraints similar to boundary conditions, leading to discrete energy levels.<br><br>

So while classical mechanics allows a continuous range of bound orbits (all with $E<0$), quantum mechanics restricts these to a discrete set determined by wave behavior.<br><br>

👉 The underlying idea remains the same:  
<strong>attractive potentials create bound states</strong>, but quantum mechanics adds structure by allowing only specific standing-wave solutions.

</div>
