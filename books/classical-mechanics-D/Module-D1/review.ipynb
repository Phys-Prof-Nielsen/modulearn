## D1.1 Impulse–Momentum: Review and Foundations
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D1.1.1 Impulse and Momentum - Point Particle

From Physics I, we recall that **motion** or **momentum** is defined as

$$
\mathbf{p} = m\mathbf{v}
$$

and **transfer of momentum** or **impulse** is

$$
\mathbf{J} = \int_{t_1}^{t_2} \mathbf{F}\,dt.
$$

The key result is the **impulse–momentum theorem**:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Impulse–Momentum Theorem</strong>

$$
\mathbf{J}_{\text{net}} = \Delta \mathbf{p}
$$

</div>

> Impulse represents **momentum transfer**, and only the **net impulse** determines the change in momentum.

---

<div style="background-color:#e0f7fa; border-left:6px solid #00838f; padding:12px; border-radius:4px;">
<strong>Example — Finding Final Velocity from a Time-Dependent Force</strong>

A particle of mass $m = 2.0~\text{kg}$ is initially at rest. A net force

$$
F(t) = 4t~\text{N}
$$

acts in the $+x$ direction from

$$
t = 0 \quad \text{to} \quad t = 3.0~\text{s}.
$$

Find the final velocity.

---

**Solution**

Compute the net impulse:

$$
J_net = \int_0^3 4t\,dt
= 4\left[\frac{t^2}{2}\right]_0^3
= 18~\text{N·s}.
$$

Using

$$
J_net  = \Delta p = m v_f - mv_i,
$$

we get

$$
18 = (2.0)v_f
\quad \Rightarrow \quad
v_f = 9.0~\text{m/s}.
$$

---

**Result**

$$
\boxed{
v_f = 9.0~\text{m/s}
}
$$

</div>

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<strong>Box Activity 1 — From Impulse to $F = ma$</strong>

Starting from

$$
\mathbf{J}_{\text{net}} = \int \mathbf{F}_{\text{net}}\,dt = \Delta \mathbf{p},
$$

derive

$$
\mathbf{F}_{\text{net}} = m\mathbf{a}
$$

for a point particle.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

We begin with

$$
\int \mathbf{F}_{\text{net}}\,dt = \Delta \mathbf{p}.
$$

Divide by the time interval:

$$
\frac{1}{\Delta t}\int \mathbf{F}_{\text{net}}\,dt
=
\frac{\Delta \mathbf{p}}{\Delta t}.
$$

Taking the limit $\Delta t \to 0$ gives

$$
\mathbf{F}_{\text{net}} = \frac{d\mathbf{p}}{dt}.
$$

A point particle is an idealized object whose size, shape, and internal structure are not relevant to the motion being studied; in that model, the object is treated as a single body with constant mass, so changes in mass are outside the point-particle approximation. Using the momentum,

$$
\mathbf{p} = m\mathbf{v},
$$

we obtain

$$
\mathbf{F} = \frac{d}{dt}(m\mathbf{v})
$$

$$
\mathbf{F} = m\frac{d\mathbf{v}}{dt} + \mathbf{v}\frac{dm}{dt}
$$

$$
\mathbf{F} = m\mathbf{a} + \mathbf{v}\,\cancel{\frac{dm}{dt}}
$$

$$
\mathbf{F} = m\mathbf{a}
$$

Thus,

$$
\boxed{
\mathbf{F}_{\text{net}} = m\mathbf{a}
}
$$

---

**Insight**

Newton’s second law emerges as the **local (instantaneous)** form of the impulse–momentum theorem.

</div>
</details>

</div>

## D1.1.2 Systems of Particles

For a system of particles, we define the total momentum as

$$
\mathbf{P} = \sum_i \mathbf{p}_i.
$$

---

### Impulse–Momentum Theorem

Integrating over time:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Impulse–Momentum Theorem — System</strong>

$$
\int_{t_1}^{t_2} \sum \mathbf{F}\,dt = \Delta \mathbf{P}
$$

</div>

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<strong>Box Activity 2 — Deriving the Impulse–Momentum Theorem for a System</strong>

Starting from the **impulse–momentum theorem for a point particle**

$$
\int \mathbf{F}_i\,dt = \Delta \mathbf{p}_i,
$$

derive the impulse–momentum theorem for a system of particles:

$$
\int \sum \mathbf{F}\,dt = \Delta \mathbf{P}.
$$

Then, use the definition of the center of mass to verify that

$$
\mathbf{P} = M\mathbf{V}_{\text{CM}}.
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Start with the impulse–momentum theorem for each particle:

$$
\int \mathbf{F}_i\,dt = \Delta \mathbf{p}_i.
$$

Sum over all particles:

$$
\sum_i \int \mathbf{F}_i\,dt = \sum_i \Delta \mathbf{p}_i.
$$

Since summation and integration commute:

$$
\int \sum_i \mathbf{F}_i\,dt = \Delta \left(\sum_i \mathbf{p}_i\right).
$$

Define the total momentum of the system:

$$
\mathbf{P} = \sum_i \mathbf{p}_i.
$$

Thus,

$$
\boxed{
\int \sum \mathbf{F}\,dt = \Delta \mathbf{P}
}
$$

---

Now connect to the center of mass.

The center of mass is defined as

$$
\mathbf{R}_{\text{CM}} = \frac{1}{M}\sum_i m_i \mathbf{r}_i.
$$

Differentiate with respect to time:

$$
\mathbf{V}_{\text{CM}} = \frac{1}{M}\sum_i m_i \mathbf{v}_i.
$$

Multiply both sides by $M$:

$$
M\mathbf{V}_{\text{CM}} = \sum_i m_i \mathbf{v}_i.
$$

Recognizing that

$$
\mathbf{p}_i = m_i \mathbf{v}_i,
$$

we obtain

$$
\boxed{
\mathbf{P} = M\mathbf{V}_{\text{CM}}
}
$$

---

**Final Result**

$$
\boxed{
\int \sum \mathbf{F}\,dt = \Delta (M\mathbf{V}_{\text{CM}})
}
$$

---

**Insight**

The impulse–momentum theorem for a system shows that the motion of a system can be described entirely in terms of its center of mass, just as if all mass were concentrated at that point.

</div>
</details>

</div>

### Newton’s Second Law 

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Newton’s Second Law — System</strong>

$$
\sum \mathbf{F} = \frac{d\mathbf{P}}{dt}
$$

</div>

---

### Interpretation

- The **total force** acting on a system determines how its **total momentum changes**  
- This is the direct extension of the point-particle result  

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<strong>Box Activity — Why Only External Forces Matter</strong>

Starting from

$$
\sum \mathbf{F} = \frac{d\mathbf{P}}{dt},
$$

show that only external forces can change the total momentum of a system.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Now we decompose the total force:

$$
\sum \mathbf{F} = \sum \mathbf{F}_{\text{ext}} + \sum \mathbf{F}_{\text{int}}.
$$

By Newton’s third law,

$$
\mathbf{F}_{ij} = -\mathbf{F}_{ji}.
$$

Thus, internal forces occur in equal and opposite pairs, so

$$
\sum \mathbf{F}_{\text{int}} = 0.
$$

Therefore,

$$
\sum \mathbf{F} = \sum \mathbf{F}_{\text{ext}}.
$$

Substitute into Newton’s second law:

$$
\boxed{
\frac{d\mathbf{P}}{dt} = \sum \mathbf{F}_{\text{ext}}
}
$$

---

**Insight**

Even though internal forces may be large, they do not affect the motion of the system as a whole.

</div>
</details>

</div>

## D1.1.6 Connection to Work–Energy

We will discuss Work-Energy in next module, but we may recognize the analogy:

| Work–Energy | Impulse–Momentum |
|------|------|
| $\int \mathbf{F}\cdot d\mathbf{r}$ | $\int \mathbf{F}\,dt$ |
| $\Delta K$ | $\Delta \mathbf{P}$ |
| scalar | vector |
| accumulation over space | accumulation over time |

---

<div style="background-color:#ffebee; border-left:6px solid #c62828; padding:12px; border-radius:4px;">
<strong>Danger — Momentum Conservation vs Energy Conservation</strong>

If

$$
\sum \mathbf{F}_{\text{ext}} = 0,
$$

then

$$
\mathbf{P} = \text{constant}.
$$

However, this does <b>not</b> imply that kinetic energy is conserved.

For example, in collisions:

- momentum is always conserved (if no external forces act)  
- kinetic energy may decrease (inelastic collision) or remain constant (elastic collision)  

---

<strong>Key Point</strong>

$$
\sum \mathbf{F}_{\text{ext}} = 0 \;\Rightarrow\; \mathbf{P} = \text{constant}
$$

does <b>not</b> imply

$$
K = \text{constant}.
$$

</div>

---

### Big Idea

> Work changes energy.  
> Impulse changes momentum.  

These are parallel, global descriptions of motion.
