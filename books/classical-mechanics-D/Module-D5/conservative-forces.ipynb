# D5.4 Conservative Forces
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.4.1 Why this matters

In the previous module, we saw that the work done by a force can depend on the **path taken** between two points. This makes problems more complex, because the full trajectory must be known.

However, there exists a special class of forces for which this complication disappears.

For these forces:

> The work depends only on the starting and ending points—not on the path.

These are called **conservative forces**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.4.2 A Global Definition

A force $\vec{F}$ is called <strong>conservative</strong> if the work done between two points $A$ and $B$ is independent of the path taken.

Mathematically:

$$
\int_{L_1} \vec{F}\cdot d\vec{l}
=
\int_{L_2} \vec{F}\cdot d\vec{l}
$$

for <strong>any</strong> two paths $L_1$ and $L_2$ connecting the same points.

### Why this is a global statement

This definition is **global**:

- it must hold for **all possible paths** between $A$ and $B$  
- it is not enough to check a single path, or even a few paths  
- the statement applies to the entire region of space under consideration  

This makes the definition powerful—but also difficult to verify directly.

> In practice, we cannot test every possible path.

---

### Equivalent Closed-Loop Condition

An equivalent formulation is:

> The work done around any closed path is zero.

$$
\oint \vec{F}\cdot d\vec{l} = 0.
$$

This is also a **global condition**:

- it must hold for every closed loop  
- it encodes the same idea of path independence  

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.4.3 A Local Definition — Curl of $\vec{F}$

The global definition of a conservative force requires that the work be independent of path, or equivalently that the work around any closed loop is zero.

While this is conceptually clear, it is not practical—we cannot test every possible path.

Fortunately, there exists a <strong>local condition</strong> that is equivalent (under appropriate conditions):

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<strong>Definition — Local Test for Conservative Forces</strong>

$$
\nabla \times \vec{F} = \vec{0}
$$

</div>

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<strong>Box Activity — From Global to Local (Using Stokes’ Theorem)</strong>

<strong>Goal</strong>  

Use Stokes’ theorem to show that the global condition

$$
\oint \vec{F}\cdot d\vec{l} = 0
$$

implies the local condition

$$
\nabla \times \vec{F} = \vec{0}.
$$

---

<strong>Guided Steps</strong>  

1. Start with the closed-loop condition for a conservative force.  
2. Apply Stokes’ theorem to convert the line integral into a surface integral.  
3. Argue why the result must hold for any surface.  
4. Conclude what this implies about the integrand.  

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Step 1 — Start with the global condition</strong>  

For a conservative force:

$$
\oint \vec{F}\cdot d\vec{l} = 0
$$

for any closed loop.

---

<strong>Step 2 — Apply Stokes’ theorem</strong>  

Stokes’ theorem states:

$$
\oint \vec{F}\cdot d\vec{l}
=
\iint_S (\nabla \times \vec{F}) \cdot d\vec{A}.
$$

Thus, the global condition becomes:

$$
\iint_S (\nabla \times \vec{F}) \cdot d\vec{A} = 0.
$$

---

<strong>Step 3 — Use arbitrariness of the surface</strong>  

The surface $S$ is arbitrary (as long as it is bounded by the loop).  

For the integral to be zero for <strong>all possible surfaces</strong>, the integrand itself must vanish everywhere:

$$
\nabla \times \vec{F} = \vec{0}.
$$

---

<strong>Conclusion</strong>  

The local condition

$$
\nabla \times \vec{F} = 0
$$

follows directly from the global condition using Stokes’ theorem.

## D5.4.4 Existence of Potential Energy

If a force is conservative, then there exists a scalar function $U(x,y,z)$ such that

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<strong>Definition — Potential Energy</strong>

$$
W_{A \to B} = -\left(U(B) - U(A)\right)
$$

</div>

## What this really means

This statement is much stronger than it first appears.

It says that instead of computing work by integrating along a path, we can:

- assign a number $U(x,y,z)$ to every point in space  
- compute work using only the values at the endpoints  

Thus, the entire force field can be encoded in a **single scalar function**.

---

## Connection to the force

If such a function exists, then the force must be related to it through derivatives:

$$
F_x = -\frac{\partial U}{\partial x},
\qquad
F_y = -\frac{\partial U}{\partial y},
\qquad
F_z = -\frac{\partial U}{\partial z}.
$$

This can be written compactly as

$$
\vec{F} = -\nabla U.
$$

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:14px; border-radius:4px;">
<strong>Curiosity — Where does the “−” sign come from?</strong>  

At first glance, the minus sign in

$$
\vec{F} = -\nabla U
$$

may seem arbitrary. After all, only **differences in energy** matter, so we could define energy in many ways.

In fact, we <em>could</em> define a function

$$
\tilde{U}(\vec{r}) = +\int \vec{F}\cdot d\vec{l}
$$

instead of

$$
U(\vec{r}) = -\int \vec{F}\cdot d\vec{l}.
$$

Both would differ only by a sign convention.

---

<strong>So why do we choose the minus sign?</strong>  

The answer comes from how we define **work done by the force** from the objects point of view:

$$
W_{A \to B} = \int_A^B \vec{F}\cdot d\vec{l}.
$$

If we want energy to behave in a useful and intuitive way, we define potential energy such that

$$
W_{A \to B} = -\left(U(B) - U(A)\right).
$$

This choice ensures:

- when the force does <strong>positive work</strong>, the potential energy <strong>decreases</strong>  
- when the potential energy increases, the force must be doing <strong>negative work</strong>  

---

<strong>Why this convention is useful</strong>  

With this choice, energy conservation takes the clean and familiar form:

$$
K + U = \text{constant}.
$$

If we had chosen the opposite sign, we would instead get:

$$
K - \tilde{U} = \text{constant},
$$

which is mathematically valid but less intuitive.

> The minus sign is not forced by mathematics—it is a **convention** chosen to make energy accounting natural and consistent with how forces do work.

</div>

---

## Why this is powerful

This result allows us to:

- replace vector fields with scalar functions  
- avoid path integrals entirely  
- use energy conservation instead of Newton’s laws  

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<strong>Box Activity — Proving the Existence of a Scalar Potential</strong>

<strong>Goal</strong>  

Show that if a vector field is curl-free,

$$
\nabla \times \vec{F} = \vec{0},
$$

then, under suitable conditions, there exists a scalar function $\phi(x,y,z)$ such that

$$
\vec{F} = \nabla \phi.
$$

---

<strong>Guided Steps</strong>  

1. Assume the field is defined in a simply connected region.  
2. Define a scalar function using a line integral from a fixed reference point $O$ to a point $P$.  
3. Use path independence to show that this definition is well-defined.  
4. Compute the change in $\phi$ under a small displacement.  
5. Show that this implies $\vec{F} = \nabla \phi$.  

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Step 1 — Define a scalar function</strong>  

Choose a fixed reference point $O$. For any point $P$, define

$$
\phi(P) = \int_O^P \vec{F}\cdot d\vec{l}.
$$

Because

$$
\nabla \times \vec{F} = \vec{0},
$$

and the region is simply connected, the integral is path independent. Therefore, $\phi(P)$ is well-defined.

---

<strong>Step 2 — Move from $P$ to a nearby point</strong>  

Let the point $P$ have position $\vec{r}$, and consider a nearby point

$$
P + dP
$$

with displacement

$$
d\vec{r}.
$$

The change in $\phi$ is

$$
d\phi = \phi(\vec{r}+d\vec{r}) - \phi(\vec{r}).
$$

Using the definition of $\phi$,

$$
d\phi = \int_{\vec{r}}^{\vec{r}+d\vec{r}} \vec{F}\cdot d\vec{l}.
$$

For an infinitesimal displacement,

$$
d\phi = \vec{F}\cdot d\vec{r}.
$$

---

<strong>Step 3 — Write both sides in components</strong>  

The right-hand side is

$$
\vec{F}\cdot d\vec{r}
=
F_x\,dx + F_y\,dy + F_z\,dz.
$$

But the total differential of $\phi$ is

$$
d\phi =
\frac{\partial \phi}{\partial x}dx
+
\frac{\partial \phi}{\partial y}dy
+
\frac{\partial \phi}{\partial z}dz.
$$

---

<strong>Step 4 — Compare coefficients</strong>  

Since this must hold for arbitrary infinitesimal displacements $dx$, $dy$, and $dz$, we compare coefficients:

$$
F_x = \frac{\partial \phi}{\partial x},
\qquad
F_y = \frac{\partial \phi}{\partial y},
\qquad
F_z = \frac{\partial \phi}{\partial z}.
$$

Therefore,

$$
\boxed{
\vec{F} = \nabla \phi
}
$$

---

<strong>Step 5 — Connection to potential energy</strong>  

In mechanics, we define potential energy with the opposite sign:

$$
U = -\phi.
$$

Therefore,

$$
\vec{F} = \nabla \phi = -\nabla U.
$$

---

<strong>Conclusion</strong>  

A curl-free vector field in a simply connected region can be written as the gradient of a scalar function. In mechanics, that scalar function is related to potential energy by a minus sign.

</div>
</details>

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<strong>Example — Checking Whether a Force is Conservative</strong>

<strong>Problem</strong>  

Consider the force field

$$
\vec{F}(x,y) =
(2xy)\,\hat{\mathbf{i}} + (x^2 + 3y^2)\,\hat{\mathbf{j}}.
$$

1. Show that this force is conservative.  
2. Find a potential energy function $U(x,y)$ such that

$$
\vec{F} = -\nabla U.
$$

---

<strong>Solution</strong>  

Write the components as

$$
F_x = 2xy,
\qquad
F_y = x^2 + 3y^2.
$$

For a two-dimensional force field, the local test for a conservative force is

$$
\frac{\partial F_y}{\partial x}
-
\frac{\partial F_x}{\partial y}
=0.
$$

Compute each derivative:

$$
\frac{\partial F_y}{\partial x}
=
\frac{\partial}{\partial x}(x^2+3y^2)
=
2x
$$

and

$$
\frac{\partial F_x}{\partial y}
=
\frac{\partial}{\partial y}(2xy)
=
2x.
$$

Thus,

$$
\frac{\partial F_y}{\partial x}
-
\frac{\partial F_x}{\partial y}
=
2x-2x=0.
$$

Therefore, the force is conservative.

---

<strong>Finding the potential energy</strong>  

Since

$$
\vec{F} = -\nabla U,
$$

we have

$$
F_x = -\frac{\partial U}{\partial x},
\qquad
F_y = -\frac{\partial U}{\partial y}.
$$

Start with the $x$-component:

$$
2xy = -\frac{\partial U}{\partial x}.
$$

So

$$
\frac{\partial U}{\partial x} = -2xy.
$$

Integrate with respect to $x$:

$$
U(x,y) = -x^2y + C(y).
$$

The “constant” of integration may depend on $y$ because we integrated only with respect to $x$.

Now use the $y$-component:

$$
F_y = -\frac{\partial U}{\partial y}.
$$

Differentiate $U$ with respect to $y$:

$$
\frac{\partial U}{\partial y}
=
-x^2 + C'(y).
$$

Therefore,

$$
-\frac{\partial U}{\partial y}
=
x^2 - C'(y).
$$

Set this equal to $F_y$:

$$
x^2 - C'(y) = x^2 + 3y^2.
$$

Thus,

$$
-C'(y)=3y^2,
$$

so

$$
C'(y)=-3y^2.
$$

Integrate:

$$
C(y) = -y^3 + C_0.
$$

Therefore,

$$
U(x,y) = -x^2y - y^3 + C_0.
$$

---

<strong>Final Result</strong>  

$$
\boxed{
\vec{F}(x,y) =
(2xy)\,\hat{\mathbf{i}} + (x^2 + 3y^2)\,\hat{\mathbf{j}}
\text{ is conservative}
}
$$

and one possible potential energy function is

$$
\boxed{
U(x,y) = -x^2y - y^3
}
$$

where an arbitrary additive constant may be included.

---

<strong>Check</strong>  

$$
-\nabla U
=
-\left(
\frac{\partial U}{\partial x}\hat{\mathbf{i}}
+
\frac{\partial U}{\partial y}\hat{\mathbf{j}}
\right)
$$

$$
-\nabla U
=
-\left(
-2xy\,\hat{\mathbf{i}}
+
(-x^2-3y^2)\,\hat{\mathbf{j}}
\right)
$$

$$
-\nabla U
=
(2xy)\,\hat{\mathbf{i}}
+
(x^2+3y^2)\,\hat{\mathbf{j}}
=
\vec{F}.
$$

</div>

## D5.4.5 A Practical Test — Equality of Mixed Partials

We introduced the local condition for a conservative force:

$$
\nabla \times \vec{F} = \vec{0}.
$$

In the earlier example, the force depended only on $x$ and $y$, making it a **two-dimensional force field**. In that case, the curl simplifies significantly:

- there is only one nonzero component (the $z$-component)  
- the test reduces to a single expression  

$$
(\nabla \times \vec{F})_z =
\frac{\partial F_y}{\partial x}
-
\frac{\partial F_x}{\partial y}.
$$

This makes the calculation relatively straightforward.

However, in general, a force may depend on all three spatial variables:

$$
\vec{F}(x,y,z).
$$

In three dimensions, the curl becomes

$$
\nabla \times \vec{F}
=
\begin{pmatrix}
\frac{\partial F_z}{\partial y} - \frac{\partial F_y}{\partial z} \\
\frac{\partial F_x}{\partial z} - \frac{\partial F_z}{\partial x} \\
\frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y}
\end{pmatrix}.
$$

Now we must compute **three separate expressions**, each involving partial derivatives of different components. This makes the calculation more involved and potentially more error-prone.

There is an equivalent—and often more intuitive—way to express this condition using **mixed partial derivatives**, which avoids explicitly computing the full curl and instead checks for consistency between component derivatives.

---

### The Theorem

Consider a force field

$$
\vec{F}(x,y,z) =
F_x(x,y,z)\,\hat{\mathbf{i}} +
F_y(x,y,z)\,\hat{\mathbf{j}} +
F_z(x,y,z)\,\hat{\mathbf{k}}.
$$

If the components have continuous partial derivatives, then the force is conservative **if and only if**

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<strong>Definition — Equality of Mixed Partials</strong>

$$
\frac{\partial F_y}{\partial x} = \frac{\partial F_x}{\partial y}
$$

$$
\frac{\partial F_z}{\partial x} = \frac{\partial F_x}{\partial z}
$$

$$
\frac{\partial F_z}{\partial y} = \frac{\partial F_y}{\partial z}
$$

</div>

---

### Why this works

Recall the curl in 3D:

$$
\nabla \times \vec{F}
=
\left(
\frac{\partial F_z}{\partial y} - \frac{\partial F_y}{\partial z}
\right)\hat{\mathbf{i}}
+
\left(
\frac{\partial F_x}{\partial z} - \frac{\partial F_z}{\partial x}
\right)\hat{\mathbf{j}}
+
\left(
\frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y}
\right)\hat{\mathbf{k}}.
$$

Thus,

$$
\nabla \times \vec{F} = \vec{0}
$$

is equivalent to the three equalities above.

---

### The underlying mathematics

These equalities ultimately come from a fundamental result in multivariable calculus:

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:14px; border-radius:4px;">
<strong>Clairaut’s Theorem (Schwarz’s Theorem)</strong>  

If a scalar function $U(x,y,z)$ has continuous second partial derivatives, then

$$
\frac{\partial^2 U}{\partial x \partial y}
=
\frac{\partial^2 U}{\partial y \partial x},
\quad
\frac{\partial^2 U}{\partial x \partial z}
=
\frac{\partial^2 U}{\partial z \partial x},
\quad
\frac{\partial^2 U}{\partial y \partial z}
=
\frac{\partial^2 U}{\partial z \partial y}.
$$

</div>

---

### Why this is powerful

In 3D, this test becomes especially valuable:

- it avoids computing the full curl determinant  
- it reduces the problem to checking pairwise derivative equality  
- it directly connects to the existence of a scalar potential  

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<strong>Box Activity — Proving the 3D Condition</strong>

<strong>Goal</strong>  

Show that if

$$
\vec{F} = -\nabla U,
$$

then all three mixed partial equalities must hold.

---

<strong>Guided Steps</strong>  

1. Write the components:

   $$
   F_x = -\frac{\partial U}{\partial x}, \quad
   F_y = -\frac{\partial U}{\partial y}, \quad
   F_z = -\frac{\partial U}{\partial z}.
   $$

2. Compute:

   $$
   \frac{\partial F_y}{\partial x}, \quad
   \frac{\partial F_x}{\partial y}, \quad
   \frac{\partial F_z}{\partial x}, \quad
   \frac{\partial F_x}{\partial z}, \quad
   \frac{\partial F_z}{\partial y}, \quad
   \frac{\partial F_y}{\partial z}.
   $$

3. Apply Clairaut’s theorem.

4. Show that all required equalities follow.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Start with

$$
F_x = -\frac{\partial U}{\partial x}, \quad
F_y = -\frac{\partial U}{\partial y}, \quad
F_z = -\frac{\partial U}{\partial z}.
$$

Compute:

$$
\frac{\partial F_y}{\partial x}
=
-\frac{\partial^2 U}{\partial x \partial y},
\quad
\frac{\partial F_x}{\partial y}
=
-\frac{\partial^2 U}{\partial y \partial x}.
$$

By Clairaut:

$$
\frac{\partial F_y}{\partial x}
=
\frac{\partial F_x}{\partial y}.
$$

Similarly,

$$
\frac{\partial F_z}{\partial x}
=
-\frac{\partial^2 U}{\partial x \partial z},
\quad
\frac{\partial F_x}{\partial z}
=
-\frac{\partial^2 U}{\partial z \partial x},
$$

so

$$
\frac{\partial F_z}{\partial x}
=
\frac{\partial F_x}{\partial z}.
$$

And,

$$
\frac{\partial F_z}{\partial y}
=
-\frac{\partial^2 U}{\partial y \partial z},
\quad
\frac{\partial F_y}{\partial z}
=
-\frac{\partial^2 U}{\partial z \partial y},
$$

so

$$
\frac{\partial F_z}{\partial y}
=
\frac{\partial F_y}{\partial z}.
$$

---

<strong>Conclusion</strong>  

All three conditions are satisfied, showing that any force derived from a scalar potential must satisfy the mixed partial equalities.

</div>
</details>

</div>

---

### Big Picture

In three dimensions, the equality of mixed partial derivatives provides a **practical and systematic way** to verify whether a force is conservative.

> Instead of computing the curl, we check whether all cross-partial derivatives match.

## D5.4.6 When Curl = 0 is Not Enough

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:14px; border-radius:4px;">
<strong>Curiosity — A Subtle Exception</strong>  

We have stated that

$$
\nabla \times \vec{F} = 0
$$

implies that a force is conservative.

However, this statement is only true under an important condition:

> The region of space must be <strong>simply connected</strong> (no holes).

---

<strong>What can go wrong?</strong>  

Consider the vector field

$$
\vec{F} =
\left(
\frac{-y}{x^2+y^2}
\right)\hat{\mathbf{i}}
+
\left(
\frac{x}{x^2+y^2}
\right)\hat{\mathbf{j}}.
$$

This field has

$$
\nabla \times \vec{F} = 0
$$

everywhere except at the origin, where it is undefined.

---

<strong>Now consider a closed loop</strong>  

Take a circular path around the origin.

Even though the curl is zero everywhere along the path,

$$
\oint \vec{F}\cdot d\vec{l} \ne 0.
$$

---

<strong>What does this mean?</strong>  

- Locally (away from the origin), the field looks conservative  
- Globally, it is <strong>not conservative</strong>  

The issue is that the origin acts like a “hole” in the domain.

---

<strong>Big Idea</strong>  

> Curl = 0 guarantees a conservative field only in regions without holes.

This is why we must always consider the **domain of the force field**, not just the derivatives.

</div>

## D5.4.7 Physical Interpretation of Curl

The curl of a vector field is often introduced as a mathematical operation, but it has a clear physical meaning.

---

### Local rotation

The curl measures the tendency of a force field to produce **local rotation**.

- If $\nabla \times \vec{F} = 0$, the field has no local “twisting”  
- If $\nabla \times \vec{F} \ne 0$, the field induces rotational motion  

---

### Connection to circulation

Recall from Stokes’ theorem:

$$
\oint \vec{F}\cdot d\vec{l}
=
\iint (\nabla \times \vec{F}) \cdot d\vec{A}.
$$

This tells us:

- curl represents **circulation per unit area**  
- it measures how much the field “wants” to push around a loop  

---

### Visual picture

Imagine placing a tiny paddle wheel in the field:

- If the wheel spins → curl is nonzero  
- If the wheel does not rotate → curl is zero  

---

### Why this matters for work

If a force field has rotation:

- different paths can accumulate different amounts of work  
- energy depends on the path  

If there is no rotation:

- work cancels out around loops  
- energy depends only on endpoints  

---

### Big Idea

> Curl measures the local tendency of a force field to generate path dependence.

- Zero curl → no local rotation → path independence  
- Nonzero curl → local rotation → path dependence  

---

This connects the mathematics directly to the physics:

- curl is not just a derivative  
- it encodes how forces “circulate” in space

## D5.4.8 Important Subtlety — Non-local nature

The definition of a conservative force is inherently **non-local**:

- it does not depend on the force at a single point  
- it depends on how the force behaves over entire paths  

Later, we will introduce a **local test** that allows us to determine whether a force is conservative without checking every path.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.4.9 Big Picture

Conservative forces represent a deep simplification:

- global path independence  
- existence of potential energy  
- conservation of mechanical energy  

> Instead of tracking motion along a path, we can compare energy between states.
