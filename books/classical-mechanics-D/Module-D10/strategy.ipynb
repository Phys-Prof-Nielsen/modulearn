# D10.2 Solving Strategy Using the Lagrangian
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D10.2.1 Why a Strategy Matters

The Lagrangian method is powerful, but students often get lost in the complexity of a problem instead of focusing on the small steps towards the solution (the same can be said about problems involving Newton's 2nd law).

👉 The goal is to build good skills by starting with a **consistent workflow**:

> Build $L = T - U$ → apply Lagrange’s equation → obtain the equation of motion.

This section provides a **repeatable procedure** you can use on any problem.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D10.2.2 General Solving Strategy

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">

<strong>Solving Strategy: Lagrangian Mechanics</strong>

Given a system:

---

**1. Sketch the system**

- Draw the physical setup  
- Indicate:
  - coordinates  
  - velocities  
  - constraints  
- This step is essential for organizing the problem

---

**2. Choose coordinates**

- Select appropriate generalized coordinate(s) $q$  
- Identify:
  $$
  q, \quad \dot{q}
  $$
- Choose coordinates that match the geometry (e.g., $x$, $r$, $\theta$)

---

**3. Write the kinetic energy**

Start from:

$$
T = \frac{1}{2}m v^2.
$$

If motion is not purely Cartesian:

- Express position in terms of $q$
- Then compute velocity using:

$$
\dot{x} = \frac{dx}{dq}\dot{q}
$$

(or more generally using chain rule)

$$
\mathbf{v} =
\sum_{i=1}^{N}
\frac{\partial \mathbf{r}}{\partial q_i}\dot q_i
+
\frac{\partial \mathbf{r}}{\partial t}.
$$

---

**4. Write the potential energy**

- Identify all potential energy contributions:

$$
U(q)
$$

Examples:
- spring: $\frac{1}{2}kq^2$  
- gravity: $mg h(q)$  

👉 Express everything in terms of the chosen coordinate(s)

---

**5. Construct the Lagrangian**

$$
L(q,\dot{q}) = T - U
$$

---

**6. Apply constraint equations (see Danger Box below for more info)**

- If the system has constraints, write them as:

$$
f(q_1, q_2, \dots) = 0
$$

- Use constraints to:
  - eliminate variables  
  - reduce the number of coordinates  

👉 In this module, constraints are typically used for substitution (not multipliers)

---

**7. Compute derivatives**

$$
\frac{\partial L}{\partial q},
\qquad
\frac{\partial L}{\partial \dot{q}}
$$

---

**8. Take the time derivative**

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{q}}
\right)
$$

---

**9. Apply Lagrange’s equation**

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{q}}
\right)
-
\frac{\partial L}{\partial q}
= 0
$$

---

**10. Simplify and solve**

- Rearrange into a familiar equation of motion  
- Solve if needed  

</div>

---

<div style="background-color:#ffe6e6; border-left:6px solid #cc0000; padding:14px; border-radius:4px;">

<strong>Danger:</strong><br>

It is important to understand <strong>when you are allowed to apply constraints</strong> in Lagrangian mechanics.

---

### Case 1 — No Lagrange multipliers (this module)

If you are <strong>not using Lagrange multipliers</strong>, then:

- You may apply constraints <strong>before constructing the Lagrangian</strong>  
- You can reduce the system to the <strong>minimum number of coordinates</strong>  
- The Lagrangian is written directly in terms of independent coordinates  

👉 In this case, it does <strong>not matter</strong> whether you think of the constraint as being applied “before” or “after”—because you have already eliminated it.

---

### Case 2 — With Lagrange multipliers (later modules)

If you are using <strong>Lagrange multipliers</strong>, then:

- You must <strong>keep all coordinates independent</strong>  
- You must <strong>NOT substitute constraints into $L$</strong>  
- You apply constraints only <strong>after</strong> deriving the equations of motion  

👉 Substituting constraints too early will not provide you with the essential information about the <strong>forces of constraint</strong>.

---

### Key distinction

- Without multipliers → eliminate constraints early  
- With multipliers → enforce constraints later  

---

<strong>Big idea:</strong><br>
The timing of when you apply constraints depends on whether you treat them as substitutions (early) or as equations enforced through multipliers (late).

</div>

## D10.2.3 Example

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Block Sliding Horizontally at Constant Height</h3>

A block of mass $m$ slides with constant velocity along the $+x$ direction.  
Find the equation of motion for the block in the $x$-direction using the Lagrangian method.

Assume the block moves on a horizontal surface at constant height

$$
y = h.
$$

---

**Step 1 — Sketch the system**

The block moves horizontally:

$$
x = x(t)
$$

while its vertical position remains fixed:

$$
y = h.
$$

So the motion is constrained to a horizontal line.

---

**Step 2 — Choose coordinates**

Start with Cartesian coordinates:

$$
(x,y).
$$

The velocity components are

$$
v_x = \dot{x},
\qquad
v_y = \dot{y}.
$$

---

**Step 3 — Write the kinetic energy**

In two dimensions,

$$
v^2 = v_x^2 + v_y^2.
$$

Therefore,

$$
T = \frac{1}{2}m(v_x^2+v_y^2).
$$

Using

$$
v_x = \dot{x},
\qquad
v_y = \dot{y},
$$

we have

$$
T = \frac{1}{2}m(\dot{x}^2+\dot{y}^2).
$$

---

**Step 4 — Write the potential energy**

The gravitational potential energy is

$$
U = mgy.
$$

---

**Step 5 — Apply the constraint**

The constraint is

$$
y = h.
$$

Therefore,

$$
\dot{y}=0.
$$

So the kinetic energy becomes

$$
T = \frac{1}{2}m\dot{x}^2.
$$

The potential energy becomes

$$
U = mgh.
$$

---

**Step 6 — Construct the Lagrangian**

$$
L = T - U.
$$

Thus,

$$
L = \frac{1}{2}m\dot{x}^2 - mgh.
$$


<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Insight:</strong><br>
By applying the constraint

$$
y = h,
$$

we reduced the number of degrees of freedom from two $(x,y)$ to one $(x)$.  

This means the system is now effectively a **one-dimensional problem** with generalized coordinate

$$
q = x.
$$

👉 This illustrates a key advantage of applying constraints early:  
the Lagrangian becomes simpler, and we only work with independent coordinates before applying Lagrange’s equation.

</div>

---

**Step 7 — Compute the derivatives**

Lagrange’s equation for $x$ is

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{x}}
\right)
-
\frac{\partial L}{\partial x}
=0.
$$

First,

$$
\frac{\partial L}{\partial \dot{x}}
=
m\dot{x}.
$$

Then,

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{x}}
\right)
=
m\ddot{x}.
$$

Also,

$$
\frac{\partial L}{\partial x}=0.
$$

Notice that the constant term $mgh$ disappears because it does not depend on $x$.

---

**Step 8 — Apply Lagrange’s equation**

$$
m\ddot{x} - 0 = 0.
$$

Therefore,

$$
m\ddot{x}=0.
$$

So,

$$
\boxed{
\ddot{x}=0
}
$$

---

**Step 9 — Interpret the result**

The equation

$$
\ddot{x}=0
$$

means that the horizontal velocity is constant:

$$
\dot{x}=v=\text{constant}.
$$

Thus, the motion is

$$
x(t)=x_0+vt.
$$

This agrees with the original description: the block slides with constant velocity in the $+x$ direction.

</div>

<h3 style="margin-top:0;">Box Activity — Mass on a Spring Using the Lagrangian</h3>

A block of mass $m$ is attached to a horizontal spring with spring constant $k$. The block moves without friction along a horizontal surface.

The spring is unstretched when

$$
x=0.
$$

Use the Lagrangian method to find the equation of motion.

Determine:

1. The starting coordinates  
2. The kinetic energy  
3. The potential energy  
4. The equation of constraint  
5. The reduced generalized coordinate  
6. The Lagrangian  
7. The equation of motion  

</div>

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Activity solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1 — Sketch the system**

The block moves horizontally while remaining on a fixed horizontal surface.

We begin by describing the motion in the plane using coordinates

$$
(x,y).
$$

At this stage, we do **not** reduce the system yet.

---

**Step 2 — Write the kinetic energy**

In two dimensions,

$$
v^2 = v_x^2 + v_y^2.
$$

Since

$$
v_x=\dot{x},
\qquad
v_y=\dot{y},
$$

the kinetic energy is

$$
T=
\frac{1}{2}m(\dot{x}^2+\dot{y}^2).
$$

---

**Step 3 — Write the potential energy**

There are two potential energy contributions.

The spring potential energy is

$$
U_s=\frac{1}{2}kx^2.
$$

The gravitational potential energy is

$$
U_g=mgy.
$$

Therefore,

$$
U=
\frac{1}{2}kx^2+mgy.
$$

---

**Step 4 — State the equation of constraint**

The block is constrained to remain on the horizontal surface.

Choose the surface to be located at

$$
y=0.
$$

So the constraint is

$$
y=0.
$$

Since the block remains on this surface,

$$
\dot{y}=0.
$$

---

**Step 5 — Construct the Lagrangian**

Before applying the constraint, the Lagrangian would be

$$
L
=
\frac{1}{2}m(\dot{x}^2+\dot{y}^2)
-
\left(
\frac{1}{2}kx^2+mgy
\right).
$$

Now apply

$$
y=0,
\qquad
\dot{y}=0.
$$

This gives

$$
L
=
\frac{1}{2}m\dot{x}^2
-
\frac{1}{2}kx^2.
$$

The gravitational term disappears because the height is constant and reduces the original coordinate description from

$$
(x,y)
$$

to a single independent coordinate:

$$
q=x.
$$

The system is now a one-dimensional problem.

---

**Step 6 — Compute the derivatives**

Lagrange’s equation for the remaining coordinate $x$ is

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{x}}
\right)
-
\frac{\partial L}{\partial x}
=0.
$$

First,

$$
\frac{\partial L}{\partial \dot{x}}
=
m\dot{x}.
$$

Then,

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{x}}
\right)
=
m\ddot{x}.
$$

Also,

$$
\frac{\partial L}{\partial x}
=
-kx.
$$

---

**Step 7 — Apply Lagrange’s equation**

Substitute into Lagrange’s equation:

$$
m\ddot{x}-(-kx)=0.
$$

Thus,

$$
m\ddot{x}+kx=0.
$$

---

**Final Result**

$$
\boxed{
m\ddot{x}+kx=0
}
$$

or

$$
\boxed{
\ddot{x}+\frac{k}{m}x=0.
}
$$

---

**Interpretation**

The block undergoes simple harmonic motion:

$$
x(t)=A\cos(\omega t+\phi),
$$

where

$$
\omega=\sqrt{\frac{k}{m}}.
$$

The important habit is that we first wrote the system using the full coordinate description $(x,y)$, included both kinetic and potential energy terms, and only then applied the constraint.

This helps build the correct workflow for later problems involving Lagrange multipliers, where constraints must be handled more carefully.

## D10.2.4 Key Observations

- Forces do not appear explicitly  
- Everything comes from energy  
- The method works even in non-Cartesian coordinates  
- The structure of the equation is always the same  

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Big Picture:</strong><br>
The Lagrangian provides a systematic way to generate equations of motion using energy, making it especially powerful for more complex systems.
