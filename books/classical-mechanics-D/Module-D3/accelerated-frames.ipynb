# D3.5 Transformation of Acceleration Between Linearly Accelerating Coordinate Systems

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.5.1 Motivation
The study of motion in different coordinate systems is fundamental to classical mechanics. When transitioning between inertial and non-inertial reference frames, acceleration transforms in a manner distinct from the classical Galilean transformations applied to velocity and position. In this section, we analyze how acceleration changes between coordinate systems that accelerate linearly relative to each other, elucidate why acceleration does not transform according to Galilean transformations, and discuss the implications for Newton’s second law, particularly in the context of fictitious forces.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.5.2 Review: Galilean Transformations and Their Limitations
Galilean transformations describe how position and velocity transform between two reference frames moving at a constant velocity relative to each other. Given an inertial frame $S$ and another frame $S'$ moving with constant velocity $\mathbf{v}_0$ relative to $S$, the transformations are:

$$
\mathbf{r}' = \mathbf{r} - \mathbf{v}_0 t
$$
$$
\mathbf{v}' = \mathbf{v} - \mathbf{v}_0
$$

where $\mathbf{r}$ and $\mathbf{r}'$ are the position vectors in $S$ and $S'$, respectively, and $\mathbf{v}$ and $\mathbf{v}'$ are the corresponding velocities. Importantly, the acceleration remains invariant under Galilean transformations:

$$
\mathbf{a}' = \mathbf{a}.
$$

However, this invariance does not hold when the transformation involves a frame that is accelerating relative to another.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.5.3 Transformation of Acceleration Between Linearly Accelerating Frames
Consider two coordinate systems: an inertial frame $S$ and a non-inertial frame $S'$ accelerating linearly with acceleration $\mathbf{a}_0$ relative to $S$. If an object has acceleration $\mathbf{a}$ in $S$, its acceleration in $S'$ is given by:

$$
\mathbf{a}' = \mathbf{a} - \mathbf{a}_0.
$$

This transformation differs from the Galilean case because acceleration is explicitly dependent on the reference frame’s acceleration. **Consequently, acceleration is not invariant under transformations between inertial and non-inertial frames**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.5.4 Application to Newton’s Second Law and Fictitious Forces
Newton’s second law states that for a mass $m$:

$$
\sum\mathbf{F} = m \mathbf{a} \tag{1}.
$$

It is tempting to state that Newton's second law is also true for the observer in $S'$, that is, we wish to write:

$$
\sum\mathbf{F}' = m \mathbf{a}' \tag{2}.
$$

We know from Equation 1, that $\sum\mathbf{F}$ represents our **true forces**, so what is the meaning of $\sum\mathbf{F}'$?

Substituting the acceleration transformation into equation 1:

$$
\sum\mathbf{F} = m(\mathbf{a'} + \mathbf{a}_0).
$$

Rearranging, we obtain:

$$
\sum\mathbf{F} - m \mathbf{a}_0 = m\mathbf{a'}.
$$

We can now see that **Equation 2 will be true as long as we define**

$$
\sum\mathbf{F}' = \sum\mathbf{F}_{True} + \mathbf{F}_{Fictitious}.
$$

The additional term $\mathbf{F}_{Fictitious}=  -m \mathbf{a}_0 $ is recognized as a **fictitious force** (or inertial force or pseudo force), which appears in the non-inertial frame but has no physical origin in interactions with other objects. This force is necessary to account for the acceleration of the reference frame itself.

When working within a non-inertial frame, fictitious forces must be introduced to correctly apply Newton’s second law. Without these forces, predictions made using Newton’s laws would not match observed behavior. While fictitious forces are not actual physical interactions, they allow Newtonian mechanics to remain valid in accelerating reference frames.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.5.5 Conclusion
Unlike velocity and position, acceleration does not transform under Galilean transformations when moving between frames that accelerate linearly relative to each other. Instead, it acquires an additional term dependent on the acceleration of the reference frame. This distinction leads directly to the concept of fictitious forces when applying Newton’s second law in non-inertial frames. Understanding this transformation is crucial in mechanics, particularly in analyzing motion within accelerating systems such as rotating frames, accelerating vehicles, and planetary dynamics.
