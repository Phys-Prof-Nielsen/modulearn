## Problem 1: Velocity and Acceleration Magnitudes in Polar Coordinates

A particle moves in a plane using polar coordinates. At a particular instant, the particle has

$$
r = 4.0~\text{m},
$$

$$
\dot{r} = 3.0~\text{m/s},
$$

$$
\dot{\theta} = 2.0~\text{rad/s}.
$$

Assume that both $\dot{r}$ and $\dot{\theta}$ are constant, so

$$
\ddot{r} = 0,
\qquad
\ddot{\theta} = 0.
$$

Determine:

1. The magnitude of the velocity  
2. The magnitude of the acceleration  

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

Velocity in polar coordinates:

$$
\vec{v} =
\dot{r}\hat{r}
+
r\dot{\theta}\hat{\theta}
$$

Acceleration in polar coordinates:

$$
\vec{a} =
(\ddot{r}-r\dot{\theta}^2)\hat{r}
+
(2\dot{r}\dot{\theta}+r\ddot{\theta})\hat{\theta}
$$

---

### Step 1 — Velocity components

Radial velocity:

$$
v_r = \dot{r} = 3.0~\text{m/s}
$$

Tangential velocity:

$$
v_\theta = r\dot{\theta}
= (4.0)(2.0)
= 8.0~\text{m/s}
$$

Thus,

$$
\vec{v}
=
3.0\hat{r}
+
8.0\hat{\theta}
\quad \text{m/s}.
$$

The magnitude is

$$
|\vec{v}|
=
\sqrt{v_r^2+v_\theta^2}
=
\sqrt{(3.0)^2+(8.0)^2}.
$$

$$
|\vec{v}|=
\sqrt{73}
=
8.5~\text{m/s}.
$$

---

### Step 2 — Acceleration components

Since $\dot{r}$ and $\dot{\theta}$ are constant,

$$
\ddot{r}=0,
\qquad
\ddot{\theta}=0.
$$

Radial acceleration:

$$
a_r
=
\ddot{r}-r\dot{\theta}^2
=
0-(4.0)(2.0)^2
=
-16~\text{m/s}^2.
$$

Angular acceleration component:

$$
a_\theta
=
2\dot{r}\dot{\theta}+r\ddot{\theta}
=
2(3.0)(2.0)+(4.0)(0)
=
12~\text{m/s}^2.
$$

Thus,

$$
\vec{a}
=
-16\hat{r}
+
12\hat{\theta}
\quad \text{m/s}^2.
$$

The magnitude is

$$
|\vec{a}|
=
\sqrt{a_r^2+a_\theta^2}
=
\sqrt{(-16)^2+(12)^2}.
$$

$$
|\vec{a}|
=
\sqrt{400}
=
20~\text{m/s}^2.
$$

---

### Final Results

$$
\boxed{
|\vec{v}| = 8.5~\text{m/s}
}
$$

$$
\boxed{
|\vec{a}| = 20~\text{m/s}^2
}
$$

---

### Interpretation

Even though $\dot{r}$ and $\dot{\theta}$ are constant, the particle still has acceleration because the polar unit vectors change direction as the particle moves.

The radial term

$$
-r\dot{\theta}^2
$$

comes from rotational motion, while the angular term

$$
2\dot{r}\dot{\theta}
$$

comes from the coupling between radial and angular motion.

</div>
</details>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# Given values
r = 4.0              # m
r_dot = 3.0          # m/s
theta_dot = 2.0      # rad/s

# Since given constant
r_ddot = 0.0
theta_ddot = 0.0

# -----------------------------------------
# Velocity components
# -----------------------------------------
v_r = r_dot
v_theta = r * theta_dot

v_mag = np.sqrt(v_r**2 + v_theta**2)

# -----------------------------------------
# Acceleration components
# -----------------------------------------
a_r = r_ddot - r * theta_dot**2
a_theta = 2 * r_dot * theta_dot + r * theta_ddot

a_mag = np.sqrt(a_r**2 + a_theta**2)

# -----------------------------------------
# Output
# -----------------------------------------
print("Velocity components:")
print(f"v_r = {v_r:.2f} m/s")
print(f"v_theta = {v_theta:.2f} m/s")
print(f"|v| = {v_mag:.2f} m/s")

print("\nAcceleration components:")
print(f"a_r = {a_r:.2f} m/s^2")
print(f"a_theta = {a_theta:.2f} m/s^2")
print(f"|a| = {a_mag:.2f} m/s^2")
```

## Problem 2: Time-Dependent Polar Motion and Trajectory Plots

A particle moves in a plane using polar coordinates. Its position is described by

$$
r(t) = 2.0 + t^2
$$

and

$$
\theta(t) = 0.50t^2.
$$

At time

$$
t = 2.0~\text{s},
$$

determine:

1. The velocity vector  
2. The magnitude of the velocity  
3. The acceleration vector  
4. The magnitude of the acceleration  
5. A plot of the trajectory in **polar coordinates**  
6. A plot of the same trajectory in **Cartesian coordinates**

+

(\ddot{r}-r\dot{\theta}^2)\hat{r}

v_r = \dot r = 4.0~\text{m/s}
$$

Tangential component:

$$
v_\theta = r\dot\theta = (6.0)(2.0)=12.0~\text{m/s}

\vec{v}

$$
|\vec v|
=
\sqrt{v_r^2+v_\theta^2}

|\vec v|

a_r

a_\theta

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------------
# Define time
# -----------------------------------------
t = np.linspace(0, 3, 500)

# -----------------------------------------
# Define motion
# -----------------------------------------
r = 2.0 + t**2
theta = 0.50 * t**2

# Derivatives
r_dot = 2*t
r_ddot = 2*np.ones_like(t)

theta_dot = t
theta_ddot = np.ones_like(t)

# -----------------------------------------
# Evaluate at t0 = 2.0 s
# -----------------------------------------
t0 = 2.0

r0 = 2.0 + t0**2
theta0 = 0.50 * t0**2

r_dot0 = 2*t0
r_ddot0 = 2.0

theta_dot0 = t0
theta_ddot0 = 1.0

# Velocity components
v_r = r_dot0
v_theta = r0 * theta_dot0

v_mag = np.sqrt(v_r**2 + v_theta**2)

# Acceleration components
a_r = r_ddot0 - r0 * theta_dot0**2
a_theta = 2 * r_dot0 * theta_dot0 + r0 * theta_ddot0

a_mag = np.sqrt(a_r**2 + a_theta**2)

# -----------------------------------------
# Print results
# -----------------------------------------
print("At t = 2.0 s:\n")

print("Velocity components:")
print(f"v_r = {v_r:.2f} m/s")
print(f"v_theta = {v_theta:.2f} m/s")
print(f"|v| = {v_mag:.2f} m/s\n")

print("Acceleration components:")
print(f"a_r = {a_r:.2f} m/s^2")
print(f"a_theta = {a_theta:.2f} m/s^2")
print(f"|a| = {a_mag:.2f} m/s^2")

# -----------------------------------------
# Convert to Cartesian for plotting
# -----------------------------------------
x = r * np.cos(theta)
y = r * np.sin(theta)

x0 = r0 * np.cos(theta0)
y0 = r0 * np.sin(theta0)

# -----------------------------------------
# Create side-by-side plots
# -----------------------------------------
fig = plt.figure(figsize=(12,5))

# Polar plot
ax1 = fig.add_subplot(1,2,1, projection='polar')
ax1.plot(theta, r, linewidth=2)
ax1.scatter(theta0, r0, zorder=5)

ax1.annotate(
    r"$t=2.0\,\mathrm{s}$",
    xy=(theta0, r0),
    xytext=(theta0, r0 + 2.0),
    arrowprops=dict(arrowstyle="->"),
    fontsize=12
)

ax1.set_title("Polar Representation")
ax1.grid(True)

# Cartesian plot
ax2 = fig.add_subplot(1,2,2)
ax2.plot(x, y, linewidth=2)
ax2.scatter(x0, y0, zorder=5)

ax2.annotate(
    r"$t=2.0\,\mathrm{s}$",
    xy=(x0, y0),
    xytext=(x0, y0 - 2.0),
    arrowprops=dict(arrowstyle="->"),
    fontsize=12
)

ax2.set_title("Cartesian Representation")
ax2.set_xlabel("x (m)")
ax2.set_ylabel("y (m)")
ax2.axis("equal")
ax2.grid(True)

plt.tight_layout()
plt.show()
```

## Problem 3: Test Pilot Centrifuge Training
A test pilot trains in a human-rated centrifuge. The pilot sits in a capsule at the end of a rotating arm of radius

$$
R = 8.0~\text{m}.
$$

During one part of the training, the centrifuge rotates at a constant angular speed so that the pilot experiences a radial acceleration equal to

$$
6.0g.
$$

Assume

$$
g = 9.81~\text{m/s}^2.
$$

Determine:

1. The required centripetal acceleration in $\text{m/s}^2$  
2. The tangential speed of the capsule  
3. The angular speed $\omega$ in $\text{rad/s}$  
4. The rotation rate in revolutions per minute (rpm)

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

$$
a_c = R\omega^2 = \frac{v^2}{R}
$$

$$
v = R\omega
$$

---

### Step 1 — Required acceleration

$$
a_c = 6.0g = 6.0(9.81) = 58.9~\text{m/s}^2
$$

---

### Step 2 — Tangential speed

$$
v = \sqrt{a_c R}
= \sqrt{(58.9)(8.0)}
$$

$$
v = 21.7~\text{m/s}
$$

---

### Step 3 — Angular speed

$$
\omega = \frac{v}{R}
= \frac{21.7}{8.0}
$$

$$
\omega = 2.71~\text{rad/s}
$$

---

### Step 4 — Convert to rpm

$$
f = \frac{\omega}{2\pi} = \frac{2.71}{2\pi} = 0.431~\text{rev/s}
$$

$$
\text{rpm} = 60f = 25.9~\text{rpm}
$$

---

### Final Results

$$
\boxed{a_c = 58.9~\text{m/s}^2}
$$

$$
\boxed{v = 21.7~\text{m/s}}
$$

$$
\boxed{\omega = 2.71~\text{rad/s}}
$$

$$
\boxed{\text{rpm} = 25.9}
$$

---

### Interpretation

The large inward acceleration is required to continuously change the direction of the pilot’s velocity.

In polar coordinates:

$$
\vec{a} = -R\omega^2 \hat{r}
$$

The force producing this acceleration comes from the seat and structure of the centrifuge—not from a separate “centripetal force.”

</div>
</details>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# Given values
R = 8.0          # radius (m)
g = 9.81         # gravitational acceleration (m/s^2)
g_force = 6.0    # multiples of g

# -----------------------------------------
# Step 1 — Centripetal acceleration
# -----------------------------------------
a_c = g_force * g

# -----------------------------------------
# Step 2 — Tangential speed
# -----------------------------------------
v = np.sqrt(a_c * R)

# -----------------------------------------
# Step 3 — Angular speed
# -----------------------------------------
omega = v / R

# -----------------------------------------
# Step 4 — Convert to rpm
# -----------------------------------------
f = omega / (2 * np.pi)   # rev/s
rpm = 60 * f

# -----------------------------------------
# Output results
# -----------------------------------------
print("Centrifuge Results:\n")

print(f"Centripetal acceleration: {a_c:.2f} m/s^2")
print(f"Tangential speed: {v:.2f} m/s")
print(f"Angular speed: {omega:.2f} rad/s")
print(f"Rotation rate: {rpm:.2f} rpm")
```

## Problem 4: Designing a Banked Interstate Highway Turn  

A civil engineer is designing a curved section of interstate highway. The curve has radius

$$
R = 250~\text{m}.
$$

The design speed is

$$
v = 31~\text{m/s},
$$

approximately $70~\text{mph}$.

Assume the road is banked so that **no friction is required** for a car to safely follow the curve at the design speed.

Determine the required banking angle $\alpha$.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

### Step 1 — Coordinate system (cylindrical)

Use cylindrical coordinates centered at the center of curvature:

- $\hat r$: radially outward  
- $\hat \theta$: tangential  
- $\hat z$: vertical upward  

For circular motion:

$$
r = R = \text{constant}, \quad \dot r = 0, \quad \ddot r = 0
$$

Thus the radial acceleration is

$$
a_r = \ddot r - r\dot\theta^2 = -R\dot\theta^2
$$

Using $v = R\dot\theta$:

$$
a_r = -\frac{v^2}{R}
$$

---

### Step 2 — Forces

Gravity:

$$
\vec F_g = -mg\,\hat z
$$

Normal force (decomposed):

$$
\vec F_N = -F_N\sin\alpha\,\hat r + F_N\cos\alpha\,\hat z
$$

Note: the radial component is **negative** because it points inward.

---

### Step 3 — Newton’s Second Law in $z$

No vertical acceleration:

$$
\sum F_z = ma_z = 0
$$

$$
F_N\cos\alpha - mg = 0
$$

$$
F_N\cos\alpha = mg
$$

---

### Step 4 — Newton’s Second Law in $r$

$$
\sum F_r = ma_r
$$

$$
- F_N\sin\alpha = m\left(-\frac{v^2}{R}\right)
$$

$$
F_N\sin\alpha = m\frac{v^2}{R}
$$

---

### Step 5 — Solve for $\alpha$

Divide equations:

$$
\frac{F_N\sin\alpha}{F_N\cos\alpha}
=
\frac{m v^2/R}{mg}
$$

$$
\tan\alpha = \frac{v^2}{Rg}
$$

Substitute:

$$
\tan\alpha =
\frac{(31)^2}{(250)(9.81)} = 0.392
$$

$$
\alpha = \tan^{-1}(0.392) \approx 21.4^\circ
$$

---

### Final Result

$$
\boxed{\alpha \approx 21^\circ}
$$

---

### Interpretation

The acceleration is

$$
\vec a = -\frac{v^2}{R}\hat r
$$

and is produced entirely by the inward radial component of the normal force:

$$
- F_N\sin\alpha\,\hat r
$$

while the vertical component balances gravity:

$$
F_N\cos\alpha = mg
$$

</div>
</details>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
m, g, R, v, alpha, FN = sp.symbols("m g R v alpha F_N", positive=True)

# Cylindrical-coordinate acceleration components
a_r = -v**2 / R
a_z = 0

# Force components
# r-hat points outward, so inward normal component is negative
F_r = -FN * sp.sin(alpha)

# z-hat points upward
F_z = FN * sp.cos(alpha) - m*g

# Newton's second law components
eq_r = sp.Eq(F_r, m*a_r)
eq_z = sp.Eq(F_z, m*a_z)

print("Radial equation:")
display(eq_r)

print("Vertical equation:")
display(eq_z)

# --------------------------------------------------
# Solve symbolically
# --------------------------------------------------
solution = sp.solve([eq_r, eq_z], (FN, alpha), dict=True)

print("Symbolic solution:")
display(solution[0])

# Extract alpha solution
alpha_solution = solution[0][alpha]

print("Banking angle:")
display(sp.Eq(alpha, alpha_solution))

# --------------------------------------------------
# Substitute numerical values
# --------------------------------------------------
values = {
    R: 250.0,      # m
    v: 31.0,       # m/s
    g: 9.81        # m/s^2
}

alpha_numeric_rad = alpha_solution.subs(values)
alpha_numeric_deg = sp.deg(alpha_numeric_rad).evalf()

print("Numerical banking angle:")
display(sp.Eq(alpha, alpha_numeric_deg))
```

## Problem 5: Banked Interstate Highway Turn with Friction  

A civil engineer is designing a curved section of interstate highway. The curve has radius

$$
R = 250~\text{m}.
$$

The road is banked at

$$
\alpha = 15^\circ.
$$

The coefficient of static friction between the tires and road is

$$
\mu_s = 0.40.
$$

Use cylindrical coordinates centered at the center of curvature, with:

- $\hat r$ pointing radially outward  
- $\hat \theta$ tangent to the curve  
- $\hat z$ vertically upward  

Determine the range of speeds for which a car can safely travel through the curve without slipping.

Assume

$$
g = 9.81~\text{m/s}^2.
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

### Step 1 — Radial acceleration

For circular motion:

$$
r = R, \quad \dot r = 0, \quad \ddot r = 0
$$

Thus

$$
a_r = -\frac{v^2}{R}.
$$

---

### Step 2 — Forces

Gravity:

$$
\vec F_g = -mg\hat z
$$

Normal force:

$$
\vec F_N =
-F_N\sin\alpha\,\hat r
+
F_N\cos\alpha\,\hat z
$$

Friction depends on motion direction and its magnitude given by

$$ 
F_f = \mu_s F_N
$$

---

## Case 1 — Minimum speed

At low speeds, the car tends to slide down the bank → friction acts up the slope:

$$
\vec f_s =
f_s\cos\alpha\,\hat r
+
f_s\sin\alpha\,\hat z
$$

with

$$
f_s = \mu_s F_N
$$

---

### Equations

Radial:

$$
-F_N\sin\alpha + \mu_s F_N\cos\alpha = -m\frac{v_{\min}^2}{R}
$$

Vertical:

$$
F_N\cos\alpha + \mu_s F_N\sin\alpha = mg
$$

---

### Result

$$
v_{\min}^2 =
Rg\frac{\sin\alpha - \mu_s\cos\alpha}{\cos\alpha + \mu_s\sin\alpha}
$$

---

## Case 2 — Maximum speed

At high speeds, the car tends to slide up the bank → friction acts down the slope:

$$
\vec f_s =
- f_s\cos\alpha\,\hat r
-
f_s\sin\alpha\,\hat z
$$

---

### Equations

Radial:

$$
-F_N\sin\alpha - \mu_s F_N\cos\alpha = -m\frac{v_{\max}^2}{R}
$$

Vertical:

$$
F_N\cos\alpha - \mu_s F_N\sin\alpha = mg
$$

---

### Result

$$
v_{\max}^2 =
Rg\frac{\sin\alpha + \mu_s\cos\alpha}{\cos\alpha - \mu_s\sin\alpha}
$$

---

### Step 3 — Numerical evaluation

$$
\sin 15^\circ - 0.40\cos 15^\circ < 0
$$

Thus

$$
\boxed{v_{\min} = 0}
$$

For maximum speed:

$$
v_{\max} =
\sqrt{
(250)(9.81)
\frac{\sin 15^\circ + 0.40\cos 15^\circ}
{\cos 15^\circ - 0.40\sin 15^\circ}
}
$$

$$
\boxed{v_{\max} \approx 41.7~\text{m/s}}
$$

---

### Final Result

$$
\boxed{0 \le v \le 41.7~\text{m/s}}
$$

---

### Interpretation

Without friction:

$$
\tan\alpha = \frac{v^2}{Rg}
$$

→ only one design speed works.

With friction:

- low speeds → friction acts up the slope  
- high speeds → friction acts down the slope  

>Friction expands one allowed speed into a range of speeds

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
m, g, R, v, alpha, mu, FN = sp.symbols("m g R v alpha mu F_N", positive=True)

# Cylindrical-coordinate acceleration
# r-hat points outward, so radial acceleration is inward
a_r = -v**2 / R
a_z = 0

# --------------------------------------------------
# Case 1: Minimum speed
# Friction acts up the bank
# --------------------------------------------------
f = mu * FN

# Force components
F_r_min = -FN*sp.sin(alpha) + f*sp.cos(alpha)
F_z_min = FN*sp.cos(alpha) + f*sp.sin(alpha) - m*g

# Newton's second law
eq_r_min = sp.Eq(F_r_min, m*a_r)
eq_z_min = sp.Eq(F_z_min, m*a_z)

print("Minimum-speed case:")
print("Radial equation:")
display(eq_r_min)

print("Vertical equation:")
display(eq_z_min)

# Solve for FN and v^2
solution_min = sp.solve([eq_r_min, eq_z_min], (FN, v**2), dict=True)[0]

print("Symbolic solution for minimum-speed case:")
display(solution_min)

vmin_squared = sp.simplify(solution_min[v**2])

print("Minimum speed squared:")
display(sp.Eq(sp.Symbol("v_min")**2, vmin_squared))

# --------------------------------------------------
# Case 2: Maximum speed
# Friction acts down the bank
# --------------------------------------------------
F_r_max = -FN*sp.sin(alpha) - f*sp.cos(alpha)
F_z_max = FN*sp.cos(alpha) - f*sp.sin(alpha) - m*g

eq_r_max = sp.Eq(F_r_max, m*a_r)
eq_z_max = sp.Eq(F_z_max, m*a_z)

print("\nMaximum-speed case:")
print("Radial equation:")
display(eq_r_max)

print("Vertical equation:")
display(eq_z_max)

solution_max = sp.solve([eq_r_max, eq_z_max], (FN, v**2), dict=True)[0]

print("Symbolic solution for maximum-speed case:")
display(solution_max)

vmax_squared = sp.simplify(solution_max[v**2])

print("Maximum speed squared:")
display(sp.Eq(sp.Symbol("v_max")**2, vmax_squared))

# --------------------------------------------------
# Numerical substitution
# --------------------------------------------------
values = {
    R: 250.0,
    g: 9.81,
    alpha: sp.rad(15),
    mu: 0.40
}

vmin_sq_num = vmin_squared.subs(values).evalf()
vmax_sq_num = vmax_squared.subs(values).evalf()

# If v_min^2 is negative, minimum physical speed is zero
vmin_num = 0 if vmin_sq_num < 0 else sp.sqrt(vmin_sq_num).evalf()
vmax_num = sp.sqrt(vmax_sq_num).evalf()

# Convert to mph
mps_to_mph = 2.23694
vmin_mph = vmin_num * mps_to_mph
vmax_mph = vmax_num * mps_to_mph

print("\nNumerical results:")
display(sp.Eq(sp.Symbol("v_min")**2, vmin_sq_num))
display(sp.Eq(sp.Symbol("v_max")**2, vmax_sq_num))

print(f"v_min = {float(vmin_num):.2f} m/s = {float(vmin_mph):.1f} mph")
print(f"v_max = {float(vmax_num):.2f} m/s = {float(vmax_mph):.1f} mph")
```

## Problem 6: Pendulum Using Radial and Transverse Components  

A simple pendulum consists of a mass $m$ attached to a massless string of length $L$. The pendulum swings in a vertical plane.

At some instant, the pendulum makes an angle $\theta$ from the vertical and has angular speed $\dot{\theta}$.

Use polar coordinates centered at the pivot, with:

- $\hat{r}$ pointing outward along the string  
- $\hat{\theta}$ pointing in the direction of increasing $\theta$  

Determine:

1. The radial component of Newton’s second law  
2. The transverse component of Newton’s second law  
3. An expression for the tension $T$ in the string  
4. The equation of motion for $\theta(t)$  

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

### Step 1 — Constraint

The string has fixed length, so

$$
r = L.
$$

Therefore,

$$
\dot r = 0,
\qquad
\ddot r = 0.
$$

---

### Step 2 — Acceleration components

The polar acceleration is

$$
\vec a =
(\ddot r-r\dot{\theta}^2)\hat r
+
(2\dot r\dot{\theta}+r\ddot{\theta})\hat{\theta}.
$$

Using $r=L$, $\dot r=0$, and $\ddot r=0$:

$$
a_r = -L\dot{\theta}^2
$$

and

$$
a_\theta = L\ddot{\theta}.
$$

---

### Step 3 — Force components

There are two forces:

- tension $T$, directed inward along the string  
- gravity $mg$, directed downward  

Since $\hat r$ points outward from the pivot, tension is

$$
\vec T = -T\hat r.
$$

The gravitational force has components:

$$
F_{g,r} = mg\cos\theta
$$

and

$$
F_{g,\theta} = -mg\sin\theta.
$$

Thus,

$$
\sum F_r = mg\cos\theta - T
$$

and

$$
\sum F_\theta = -mg\sin\theta.
$$

---

### Step 4 — Radial equation

Newton’s second law in the radial direction gives

$$
\sum F_r = ma_r.
$$

Therefore,

$$
mg\cos\theta - T = m(-L\dot{\theta}^2).
$$

So

$$
mg\cos\theta - T = -mL\dot{\theta}^2.
$$

Solving for tension:

$$
T = mg\cos\theta + mL\dot{\theta}^2.
$$

---

### Step 5 — Transverse equation

Newton’s second law in the $\hat{\theta}$ direction gives

$$
\sum F_\theta = ma_\theta.
$$

Thus,

$$
-mg\sin\theta = mL\ddot{\theta}.
$$

Cancel $m$:

$$
L\ddot{\theta} = -g\sin\theta.
$$

Therefore,

$$
\boxed{
\ddot{\theta} + \frac{g}{L}\sin\theta = 0
}
$$

---

### Final Results

Radial equation:

$$
\boxed{
mg\cos\theta - T = -mL\dot{\theta}^2
}
$$

Tension:

$$
\boxed{
T = mg\cos\theta + mL\dot{\theta}^2
}
$$

Transverse equation:

$$
\boxed{
mL\ddot{\theta} = -mg\sin\theta
}
$$

Equation of motion:

$$
\boxed{
\ddot{\theta} + \frac{g}{L}\sin\theta = 0
}
$$

---

### Interpretation

The radial equation determines the **constraint force** $T$ needed to keep the mass moving in a circle.

The transverse equation determines the **actual angular motion** of the pendulum, which we should recognize as the equation for **Simple Harmonic Motion**.

This separation is important:

- radial direction → constraint force  
- transverse direction → equation of motion  

</div>
</details>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
m, g, L, theta = sp.symbols("m g L theta", positive=True)
theta_dot, theta_ddot, T = sp.symbols(r"\dot{\theta} \ddot{\theta} T")

# --------------------------------------------------
# Constraint: r = L
# --------------------------------------------------
r = L
r_dot = 0
r_ddot = 0

# --------------------------------------------------
# Polar acceleration components
# --------------------------------------------------
a_r = r_ddot - r*theta_dot**2
a_theta = 2*r_dot*theta_dot + r*theta_ddot

print("Acceleration components:")
display(sp.Eq(sp.Symbol("a_r"), a_r))
display(sp.Eq(sp.Symbol("a_theta"), a_theta))

# --------------------------------------------------
# Force components
# r-hat points outward from pivot
# theta measured from vertical
# --------------------------------------------------
F_r = m*g*sp.cos(theta) - T
F_theta = -m*g*sp.sin(theta)

print("Force components:")
display(sp.Eq(sp.Symbol("F_r"), F_r))
display(sp.Eq(sp.Symbol("F_theta"), F_theta))

# --------------------------------------------------
# Newton's second law components
# --------------------------------------------------
eq_r = sp.Eq(F_r, m*a_r)
eq_theta = sp.Eq(F_theta, m*a_theta)

print("Radial equation:")
display(eq_r)

print("Transverse equation:")
display(eq_theta)

# --------------------------------------------------
# Solve radial equation for tension
# --------------------------------------------------
T_solution = sp.solve(eq_r, T)[0]

print("Tension:")
display(sp.Eq(T, T_solution))

# --------------------------------------------------
# Solve transverse equation for theta_ddot
# --------------------------------------------------
theta_ddot_solution = sp.solve(eq_theta, theta_ddot)[0]

print("Equation of motion:")
display(sp.Eq(theta_ddot, theta_ddot_solution))

# Standard form
print("Standard pendulum equation:")
display(sp.Eq(theta_ddot + (g/L)*sp.sin(theta), 0))
```

a_r = -\frac{v_0^2}{r}, \qquad a_z = 0.

---

### Step 2 — Forces

Gravity:

$$
\vec F_g = -mg\,\hat z
$$

Normal force (perpendicular to cone surface):

$$
\vec F_N = -F_N\cos\theta\,\hat r + F_N\sin\theta\,\hat z
$$

---

### Step 3 — Newton’s Second Law in $z$

No vertical acceleration:

$$
\sum F_z = 0
$$

$$
F_N\sin\theta - mg = 0
$$

$$
F_N\sin\theta = mg
$$

---

### Step 4 — Newton’s Second Law in $r$

$$
\sum F_r = ma_r
$$

- F_N\cos\theta = -m\frac{v_0^2}{r}

F_N\cos\theta = m\frac{v_0^2}{r}

Divide radial by vertical equations:

$$
\frac{F_N\cos\theta}{F_N\sin\theta}
=
\frac{m v_0^2/r}{mg}

$$
\cot\theta = \frac{v_0^2}{rg}
$$

Solve for $r$:

$$
r = \frac{v_0^2}{g}\tan\theta
$$

---

### Final Result

$$
\boxed{
r = \frac{v_0^2}{g}\tan\theta
}
$$

---

### Interpretation

The normal force simultaneously:

- balances gravity in the vertical direction  
- provides the inward radial force required for circular motion  

The geometry of the cone determines how these components split, leading to a unique radius for a given speed.

## Problem 8

This is Problem 3.4 in Kleppner and Kolenkow

<details>

Problem solution

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

### Step 1 — Physical setup

For a circular orbit, gravity supplies the required inward radial acceleration.

Using cylindrical/polar coordinates centered at Earth:

$$
a_r = -\frac{v^2}{r}.
$$

Newton’s second law in the radial direction is

$$
-\frac{GMm}{r^2} = -m\frac{v^2}{r}.
$$

Canceling $m$ and simplifying:

$$
\frac{GM}{r^2} = \frac{v^2}{r}.
$$

---

### Step 2 — Relate speed to orbital period

For a circular orbit,

$$
v = \frac{2\pi r}{T}.
$$

Substitute into the radial equation:

$$
\frac{GM}{r^2}
=
\frac{1}{r}
\left(\frac{2\pi r}{T}\right)^2.
$$

Simplify:

$$
\frac{GM}{r^2}
=
\frac{4\pi^2 r}{T^2}.
$$

Thus,

$$
r^3 = \frac{GMT^2}{4\pi^2}.
$$

So,

$$
r =
\left(
\frac{GMT^2}{4\pi^2}
\right)^{1/3}.
$$

---

### Step 3 — Express using Earth’s radius

At Earth’s surface,

$$
g = \frac{GM}{R_E^2}.
$$

Thus,

$$
GM = gR_E^2.
$$

Substitute:

$$
r =
\left(
\frac{gR_E^2T^2}{4\pi^2}
\right)^{1/3}.
$$

To express this in units of $R_E$, divide by $R_E$:

$$
\frac{r}{R_E}
=
\left(
\frac{gT^2}{4\pi^2 R_E}
\right)^{1/3}.
$$

---

### Step 4 — Numerical evaluation

Use

$$
T = 24~\text{h} = 86400~\text{s},
$$

$$
g = 9.81~\text{m/s}^2,
$$

and

$$
R_E = 6.37\times 10^6~\text{m}.
$$

Then

$$
\frac{r}{R_E}
=
\left(
\frac{(9.81)(86400)^2}{4\pi^2(6.37\times 10^6)}
\right)^{1/3}.
$$

This gives

$$
\frac{r}{R_E} \approx 6.6.
$$

---

### Final Result

$$
\boxed{
r \approx 6.6 R_E
}
$$

This is the distance from Earth’s center. The altitude above Earth’s surface is therefore

$$
h = r - R_E \approx 5.6R_E.
$$

---

### Interpretation

A synchronous satellite must orbit far above Earth’s surface because it must have a long orbital period of $24~\text{h}$.

The orbit radius is approximately

$$
r \approx 4.2\times 10^7~\text{m},
$$

measured from Earth’s center.

</div>
</details>

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
G, M, r, T = sp.symbols("G M r T", positive=True)

# Orbital speed for circular motion
v = 2 * sp.pi * r / T

# Radial force balance: gravity provides centripetal acceleration
eq = sp.Eq(G*M/r**2, v**2/r)

print("Radial equation:")
display(eq)

# --------------------------------------------------
# Solve symbolically for r
# --------------------------------------------------
r_solution = sp.solve(eq, r)[0]

print("Orbital radius (symbolic):")
display(sp.Eq(r, r_solution))

# --------------------------------------------------
# Substitute numerical values (Earth + synchronous orbit)
# --------------------------------------------------
values = {
    G: 6.67430e-11,      # gravitational constant (m^3 kg^-1 s^-2)
    M: 5.972e24,         # mass of Earth (kg)
    T: 24 * 3600         # orbital period (s)
}

# Evaluate radius
r_numeric = r_solution.subs(values).evalf()

# Earth's radius
R_E = 6.37e6

# Ratio and altitude
ratio = (r_numeric / R_E).evalf()
h = (r_numeric - R_E).evalf()

# --------------------------------------------------
# Output
# --------------------------------------------------
print("\nNumerical results:")
display(sp.Eq(r, r_numeric))
display(sp.Eq(sp.Symbol("r/R_E"), ratio))
display(sp.Eq(sp.Symbol("h"), h))

print(f"r = {float(r_numeric):.3e} m")
print(f"r / R_E = {float(ratio):.2f}")
print(f"Altitude h = {float(h):.3e} m")
```

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
m, g, l, omega = sp.symbols("m g l omega", positive=True)
T_up, T_low = sp.symbols("T_up T_low", real=True)

angle = sp.pi/4

# Radius of circular motion
R = l * sp.sin(angle)

# Cylindrical acceleration components
a_r = -R * omega**2
a_z = 0

# Force components
F_r = -T_up*sp.sin(angle) - T_low*sp.sin(angle)
F_z = T_up*sp.cos(angle) - T_low*sp.cos(angle) - m*g

# Newton's second law
eq_r = sp.Eq(F_r, m*a_r)
eq_z = sp.Eq(F_z, m*a_z)

print("Radial equation:")
display(eq_r)

print("Vertical equation:")
display(eq_z)

# Solve symbolically
solution = sp.solve([eq_r, eq_z], (T_up, T_low), dict=True)[0]

print("Symbolic solution:")
display(solution)

# Display tensions clearly
T_up_solution = sp.simplify(solution[T_up])
T_low_solution = sp.simplify(solution[T_low])

print("Upper tension:")
display(sp.Eq(T_up, T_up_solution))

print("Lower tension:")
display(sp.Eq(T_low, T_low_solution))

# Condition for lower string to remain taut
condition = sp.solve_univariate_inequality(T_low_solution > 0, omega)

print("Condition for lower string to remain taut:")
display(condition)
```

## Problem 10

Problem 3.15 in Kleppner and Kolenkow  

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

### Step 1 — Mass enclosed inside radius $r$

Assume Earth has uniform density $\rho$.  

The mass enclosed inside radius $r$ is

$$
M(r)=\rho \left(\frac{4}{3}\pi r^3\right).
$$

The total mass of Earth is

$$
M_E=\rho \left(\frac{4}{3}\pi R_E^3\right).
$$

Therefore,

$$
M(r)=M_E\frac{r^3}{R_E^3}.
$$

---

### Step 2 — Gravitational force inside Earth (Gauss' Law for Gravity)

Only the mass inside radius $r$ contributes to the gravitational force. Thus,

$$
F(r)=-\frac{GmM(r)}{r^2}.
$$

Substitute $M(r)$:

$$
F(r)
=
-\frac{Gm}{r^2}
\left(
M_E\frac{r^3}{R_E^3}
\right).
$$

Simplify:

$$
F(r)
=
-\frac{GmM_E}{R_E^3}r.
$$

This force is proportional to $-r$, which is the form of a restoring force.

---

### Step 3 — Equation of motion

Newton’s second law gives

$$
m\ddot r = -\frac{GmM_E}{R_E^3}r.
$$

Cancel $m$:

$$
\ddot r = -\frac{GM_E}{R_E^3}r.
$$

Rewrite:

$$
\ddot r + \frac{GM_E}{R_E^3}r = 0.
$$

This is the equation for simple harmonic motion:

$$
\ddot r + \omega^2 r = 0,
$$

with

$$
\omega^2 = \frac{GM_E}{R_E^3}.
$$

Thus,

$$
\omega = \sqrt{\frac{GM_E}{R_E^3}}.
$$

---

### Step 4 — Period of the motion

The period of simple harmonic motion is

$$
T = \frac{2\pi}{\omega}.
$$

Therefore,

$$
T =
2\pi\sqrt{\frac{R_E^3}{GM_E}}.
$$

This is the time required to return to the starting point at the surface.

The time to reach the center is one-quarter period, and the time to reach the opposite side is one-half period.

---

### Step 5 — Compare with a low circular orbit

For a satellite in a circular orbit near Earth’s surface,

$$
\frac{GM_E m}{R_E^2} = m\frac{v^2}{R_E}.
$$

Using

$$
v = \frac{2\pi R_E}{T_{\text{orbit}}},
$$

we have

$$
\frac{GM_E}{R_E^2}
=
\frac{1}{R_E}
\left(\frac{2\pi R_E}{T_{\text{orbit}}}\right)^2.
$$

Simplify:

$$
\frac{GM_E}{R_E^2}
=
\frac{4\pi^2 R_E}{T_{\text{orbit}}^2}.
$$

Solve for $T_{\text{orbit}}$:

$$
T_{\text{orbit}}
=
2\pi\sqrt{\frac{R_E^3}{GM_E}}.
$$

Thus,

$$
T_{\text{orbit}} = T.
$$

---

### Final Result

$$
\boxed{
T =
2\pi\sqrt{\frac{R_E^3}{GM_E}}
}
$$

This is both:

- the time to return to the original point in the tunnel  
- the orbital period of a low satellite near Earth’s surface  

Using $g = GM_E/R_E^2$, we can also write

$$
T =
2\pi\sqrt{\frac{R_E}{g}}.
$$

---

### Interpretation

Inside a uniform Earth, gravity decreases linearly with distance from the center:

$$
F(r) \propto -r.
$$

Therefore, falling through a tunnel through Earth is mathematically the same as simple harmonic motion.

The period is independent of the traveler’s mass and, in this idealized model, independent of the tunnel direction.

### Deeper Insight — Why the Periods Match

This result is not a coincidence but reflects a deeper connection in how gravity governs motion. In both cases—the motion through a tunnel and circular orbit near Earth’s surface—the dynamics are controlled by the same characteristic quantity

$$
\frac{GM_E}{R_E^3},
$$

which sets the natural timescale of motion in Earth’s gravitational field. Inside the Earth, the gravitational force varies linearly with distance from the center, producing simple harmonic motion with frequency

$$
\omega^2 = \frac{GM_E}{R_E^3}.
$$

For a satellite in low orbit, gravity provides the centripetal acceleration required for circular motion, leading to the same angular frequency. Thus, both motions share the same period because they are governed by the same underlying gravitational “stiffness” of the system. In this sense, orbiting the Earth is closely related to falling through it—each represents a different manifestation of motion under the same gravitational field.

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
G, M_E, R_E, m, r, T, g = sp.symbols("G M_E R_E m r T g", positive=True)

# --------------------------------------------------
# Mass enclosed inside radius r for uniform Earth
# --------------------------------------------------
M_r = M_E * r**3 / R_E**3

print("Mass enclosed inside radius r:")
display(sp.Eq(sp.Symbol("M(r)"), M_r))

# --------------------------------------------------
# Gravitational force inside Earth
# --------------------------------------------------
F_r = -G*m*M_r/r**2
F_r_simplified = sp.simplify(F_r)

print("Force inside Earth:")
display(sp.Eq(sp.Symbol("F_r"), F_r_simplified))

# --------------------------------------------------
# Equation of motion: m r_ddot = F_r
# --------------------------------------------------
omega_sq = sp.simplify(G*M_E/R_E**3)

print("Equation of motion:")
display(sp.Eq(sp.Symbol(r"\ddot{r}"), -omega_sq*r))

print("Simple harmonic oscillator form:")
display(sp.Eq(sp.Symbol(r"\ddot{r}") + omega_sq*r, 0))

print("Angular frequency:")
display(sp.Eq(sp.Symbol(r"\omega^2"), omega_sq))

# --------------------------------------------------
# Period of oscillation through Earth
# --------------------------------------------------
T_tunnel = sp.simplify(2*sp.pi/sp.sqrt(omega_sq))

print("Tunnel oscillation period:")
display(sp.Eq(sp.Symbol("T_tunnel"), T_tunnel))

# --------------------------------------------------
# Low-orbit satellite period
# --------------------------------------------------
v = 2*sp.pi*R_E/T

# Gravity supplies centripetal acceleration
eq_orbit = sp.Eq(G*M_E/R_E**2, v**2/R_E)

print("Low-orbit radial equation:")
display(eq_orbit)

T_orbit_solution = sp.solve(eq_orbit, T)[0]

print("Low-orbit period:")
display(sp.Eq(sp.Symbol("T_orbit"), T_orbit_solution))

# --------------------------------------------------
# Verify equality
# --------------------------------------------------
difference = sp.simplify(T_tunnel - T_orbit_solution)

print("Verification that periods match:")
display(sp.Eq(T_tunnel - T_orbit_solution, difference))

# --------------------------------------------------
# Alternative form using g = GM_E/R_E^2
# --------------------------------------------------
T_using_g = sp.simplify(T_tunnel.subs(G*M_E, g*R_E**2))

print("Period using g:")
display(sp.Eq(sp.Symbol("T"), T_using_g))

# --------------------------------------------------
# Numerical evaluation
# --------------------------------------------------
values = {
    G: 6.67430e-11,      # m^3 kg^-1 s^-2
    M_E: 5.972e24,       # kg
    R_E: 6.37e6          # m
}

T_num = T_tunnel.subs(values).evalf()
T_minutes = (T_num/60).evalf()

print("Numerical result:")
display(sp.Eq(sp.Symbol("T"), T_num))
display(sp.Eq(sp.Symbol("T_minutes"), T_minutes))

print(f"T = {float(T_num):.2f} s")
print(f"T = {float(T_minutes):.2f} min")
```

## Problem 11

Problem 3.22 in Kleppner and Kolenkow  

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

### Step 1 — Coordinate system and constraint

Use polar coordinates centered at the ring.

$$
r(t) = r_0 - Vt
$$

$$
\dot r = -V, \qquad \ddot r = 0
$$

---

### Step 2 — Force diagram

Only force is tension:

$$
\vec T = -T\,\hat r
$$

No transverse force:

$$
F_\theta = 0
$$

---

### Step 3 — Transverse equation

$$
a_\theta = 2\dot r\,\omega + r\dot\omega
$$

Newton’s second law:

$$
0 = m(2\dot r\,\omega + r\dot\omega)
$$

Cancel $m$:

$$
2\dot r\,\omega + r\dot\omega = 0
$$

Insert $\dot r = -V$:

$$
-2V\omega + r\dot\omega = 0
$$

$$
\boxed{
\dot\omega = \frac{2V}{r}\omega
}
$$

---

### Step 4 — Solve for $\omega(t)$

We can rewrite

$$
2\dot r\,\omega + r\dot\omega = 0
$$

as

$$
\frac{d}{dt}(r^2\omega)=0
$$

since $$
\frac{d}{dt}(r^2\omega)= 2r\dot r\,\omega + r^2\dot\omega
$$

Dividing through by $r$ gives us the desired relation.

Thus

$$
r^2\omega = \text{constant} = r_0^2\omega_0
$$

$$
\boxed{
\omega(t)=\omega_0\left(\frac{r_0}{r_0-Vt}\right)^2
}
$$

---

### Step 5 — Radial equation

$$
a_r = -r\omega^2
$$

Newton’s second law:

$$
-T = m(-r\omega^2)
$$

$$
T = mr\omega^2
$$

---

### Step 6 — Substitute $\omega(t)$

$$
T(t)
=
m(r_0-Vt)
\left[
\omega_0\left(\frac{r_0}{r_0-Vt}\right)^2
\right]^2
$$

$$
\boxed{
T(t)=\frac{m\omega_0^2 r_0^4}{(r_0-Vt)^3}
}
$$

---

### Final Results

$$
\boxed{
\omega(t)=\omega_0\left(\frac{r_0}{r_0-Vt}\right)^2
}
$$

$$
\boxed{
F_{\text{pull}}(t)=\frac{m\omega_0^2 r_0^4}{(r_0-Vt)^3}
}
$$

---

### Interpretation

No transverse force ⇒ angular momentum is conserved:

$$
mr^2\omega = \text{constant}
$$

As $r$ decreases, $\omega$ increases rapidly:

$$
\omega \propto \frac{1}{r^2}
$$

The required pulling force grows even faster:

$$
T \propto \frac{1}{r^3}
$$

</div>
</details>

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
m, r0, V, t, omega0 = sp.symbols("m r_0 V t omega_0", positive=True)
omega = sp.Function("omega")

# Radial constraint
r = r0 - V*t
r_dot = sp.diff(r, t)
r_ddot = sp.diff(r_dot, t)

print("Radial constraint:")
display(sp.Eq(sp.Symbol("r(t)"), r))
display(sp.Eq(sp.Symbol(r"\dot r"), r_dot))
display(sp.Eq(sp.Symbol(r"\ddot r"), r_ddot))

# --------------------------------------------------
# Transverse equation
# --------------------------------------------------
# a_theta = 2 r_dot omega + r omega_dot
omega_t = omega(t)
omega_dot = sp.diff(omega_t, t)

eq_theta = sp.Eq(2*r_dot*omega_t + r*omega_dot, 0)

print("Transverse equation:")
display(eq_theta)

# --------------------------------------------------
# Solve differential equation for omega(t)
# --------------------------------------------------
omega_solution = sp.dsolve(eq_theta, omega_t)

print("General solution:")
display(omega_solution)

# Apply initial condition omega(0) = omega0
C1 = list(omega_solution.rhs.free_symbols - {r0, V, t})[0]

C1_solution = sp.solve(
    sp.Eq(omega_solution.rhs.subs(t, 0), omega0),
    C1
)[0]

omega_t_solution = sp.simplify(omega_solution.rhs.subs(C1, C1_solution))

print("Angular velocity:")
display(sp.Eq(sp.Symbol(r"\omega(t)"), omega_t_solution))

# --------------------------------------------------
# Radial equation and tension
# --------------------------------------------------
# a_r = r_ddot - r omega^2
a_r = r_ddot - r*omega_t_solution**2

# Only radial force is tension inward: F_r = -T
T = sp.symbols("T", positive=True)

eq_r = sp.Eq(-T, m*a_r)

print("Radial equation:")
display(eq_r)

T_solution = sp.solve(eq_r, T)[0]
T_solution = sp.simplify(T_solution)

print("Tension / Pulling force:")
display(sp.Eq(T, T_solution))

# --------------------------------------------------
# Angular momentum check
# --------------------------------------------------
L = sp.simplify(m*r**2*omega_t_solution)

print("Angular momentum:")
display(sp.Eq(sp.Symbol("L"), L))

print("Time derivative of angular momentum:")
display(sp.Eq(sp.diff(L, t), sp.simplify(sp.diff(L, t))))
```

## Problem 12: Speed in Parabolic Coordinates Using Index Notation

A particle moves in the plane. Its Cartesian coordinates are related to parabolic coordinates $(u,v)$ by

$$
x = uv
$$

$$
y = \frac{1}{2}(u^2 - v^2).
$$

Using index notation, determine an expression for the speed squared

$$
v_{\text{speed}}^2
$$

in terms of $u$, $v$, $\dot u$, and $\dot v$.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

We use

$$
v_{\text{speed}}^2 = g_{jk}\dot q_j \dot q_k,
$$

where

$$
g_{jk}
=
\frac{\partial x_i}{\partial q_j}
\frac{\partial x_i}{\partial q_k}.
$$

Here,

$$
q_1 = u, \qquad q_2 = v.
$$

---

<strong>Step 1 — Compute partial derivatives</strong>

For

$$
x_1 = x = uv,
$$

$$
\frac{\partial x}{\partial u} = v,
\qquad
\frac{\partial x}{\partial v} = u.
$$

For

$$
x_2 = y = \frac{1}{2}(u^2-v^2),
$$

$$
\frac{\partial y}{\partial u} = u,
\qquad
\frac{\partial y}{\partial v} = -v.
$$

---

<strong>Step 2 — Compute metric components</strong>

$$
g_{uu}
=
\frac{\partial x_i}{\partial u}
\frac{\partial x_i}{\partial u}
=
\frac{\partial x_1}{\partial u}
\frac{\partial x_1}{\partial u}
+
\frac{\partial x_2}{\partial u}
\frac{\partial x_2}{\partial u}
$$

$$
g_{uu}
=
v^2 + u^2.
$$

---

$$
g_{uv}
=
\frac{\partial x_i}{\partial u}
\frac{\partial x_i}{\partial v}
=
\frac{\partial x_1}{\partial u}
\frac{\partial x_1}{\partial v}
+
\frac{\partial x_2}{\partial u}
\frac{\partial x_2}{\partial v}
$$

$$
g_{uv}
=
(v)(u) + (u)(-v) = 0.
$$

---
Similarly, 

$$
g_{vv}
=
\frac{\partial x_i}{\partial v}
\frac{\partial x_i}{\partial v}
$$

$$
g_{vv}
=
u^2 + v^2.
$$

Thus,

$$
g_{uu}=u^2+v^2,
\qquad
g_{uv}=0,
\qquad
g_{vv}=u^2+v^2.
$$

---

<strong>Step 3 — Compute speed squared</strong>

Using

$$
v_{\text{speed}}^2 = g_{jk}\dot q_j\dot q_k,
$$

we get

$$
v_{\text{speed}}^2
=
g_{uu}\dot u^2
+
2g_{uv}\dot u\dot v
+
g_{vv}\dot v^2.
$$

Since $g_{uv}=0$,

$$
v_{\text{speed}}^2
=
(u^2+v^2)\dot u^2
+
(u^2+v^2)\dot v^2.
$$

Therefore,

$$
\boxed{
v_{\text{speed}}^2
=
(u^2+v^2)(\dot u^2+\dot v^2)
}
$$

and

$$
\boxed{
v_{\text{speed}}
=
\sqrt{(u^2+v^2)(\dot u^2+\dot v^2)}
}
$$

---

<strong>Interpretation</strong>

The metric tells us how coordinate changes in $(u,v)$ correspond to physical distance in the plane. Here the coordinates are orthogonal because

$$
g_{uv}=0.
$$

Both coordinate directions have the same scale factor:

$$
h_u = h_v = \sqrt{u^2+v^2}.
$$

Thus, motion in either coordinate direction contributes to speed through the same geometric factor.
