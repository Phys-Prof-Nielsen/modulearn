## Problem D3.1

A $45^\circ$ wedge is pushed along a horizontal table with constant acceleration $A$. A block of mass $m$ slides without friction on the incline part of the wedge. Optimus is a stationary observer on the floor and see the wedge moving away from him while Optimus Prime is at rest at the bottom of the wedge. 

1. Find the block’s acceleration as observed by Optimus.
2. Find the block’s acceleration as observed by Optimus Prime.

*This is a modified version of Problem 2.11 in Kleppner and Kolenkow*.

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
import numpy as np

theta = np.radians(45)

AX, ax, ay, fn, m, g = sp.symbols('AX, ax, ay, fn, m, g')

#-----------------------

# PART 1
# N2 in x-direction
eq_x = m*ax - fn*sp.sin(theta)
eq_y = m*ay - fn*sp.cos(theta) + m*g

#constraint equation
#con = (H-y) - sp.tan(theta)*(x - X)
con_ddot = -ay - sp.tan(theta)*(ax - AX)

sol = sp.solve((eq_x, eq_y, con_ddot),(ax,ay,fn))
print(sol)

#-----------------------

# PART 2

ax_prime, ay_prime, fn, m, g = sp.symbols('ax_prime, ay_prime, fn, m, g')
f_fict_x = -m*AX
f_fict_y = 0
# N2 in x_prime-direction
eq_x_prime = m*ax_prime - fn*sp.sin(theta) - f_fict_x
eq_y_prime = m*ay_prime - fn*sp.cos(theta) + m*g - 0

#constraint equation
#con = (H-y_prime) - sp.tan(theta)*x_prime
con_ddot = -ay_prime - sp.tan(theta)*ax_prime

sol = sp.solve((eq_x_prime, eq_y_prime, con_ddot),(ax_prime,ay_prime,fn))
print(sol)
```

## Problem D3.2

accelerates at rate $A$ along a horizontal road. What is the static angle of the string from the
vertical, and what is the tension in the string?

Solve it for both an observer stationary on the road (Optimus)  and an observer at rest in the car (Optimus Prime).

## Problem D3.3

A solid cylinder of mass $M$ and radius $R$ rolls without slipping on a plank
that is accelerated at rate $A$. Find the acceleration of the cylinder:

1. For an observer stationary on the plank.
2. For an observer stationary on the ground (an inertial system).

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp
import numpy as np

M, R, A, ax_prime, ff, alpha = sp.symbols('M, R, A, ax_prime, ff, alpha')

# Optimus Prime
# -----
# Linear N2
f_fict_x = -M*A

eqx_prime = M*ax_prime - ff - f_fict_x

#rotational N2
I = 0.5*M*R**2
eq_torque = I*alpha - R*ff

# constraint
con = ax_prime + R*alpha

sol = sp.solve((eqx_prime, eq_torque, con),(ax_prime, alpha, ff))
print('Acceleration observed by Optimus Prime: '+str(sol[ax_prime]))

# Optimus
ax_prime = sol[ax_prime]
ax = A + ax_prime
print('Acceleration observed by Optimus : '+str(ax))
```

## Problem D3.4

**Scenario.**  
Inside a cart accelerating **to the right** with constant acceleration $a$, a block of mass $m$ slides without friction on a **horizontal guide rail**. The block is attached to a wall at the **left end** of the rail by a **light horizontal spring** of stiffness $k$ and natural length $L$. Let the block’s coordinate be $x$ measured to the right from the wall anchor point. 

**Constraint (geometry of motion).**  
The block is constrained to the horizontal rail:

$$
y=0 \quad\text{(holonomic constraint)}.
$$

Thus the motion is 1D along $x$.

**Tasks.**
1. Using Newton’s second law along $x$ in the accelerating frame, find the **equilibrium position** $x^\ast$ where the block rests relative to the cart.  
   *Use the spring force $F_s = -k\,(x-L)\,\hat{\mathbf{x}}$.*  
2. If the rail has hard stops $x\in[0,X_{\max}]$, state the **existence condition** for the interior equilibrium $x^\ast\in[0,X_{\max}]$.  
3. Report the **normal force** from the rail on the block at equilibrium.

```python
# SymPy solution: horizontal spring in a linearly accelerating frame (Newtonian only)
import sympy as sp

# Symbols
m, a, k, L, Xmax = sp.symbols('m a k L Xmax', positive=True, real=True)
x = sp.symbols('x', real=True)

# Forces along x:
# Spring force on block: F_sx = -k (x - L)
F_sx = -k*(x - L)

# Fictitious force (cart accelerating to the right): F_fict_x = -m a  (to the left)
F_fx = -m*a

# Equilibrium (block at rest in cart frame): sum of forces along x = 0
eq_x = sp.Eq(F_sx + F_fx, 0)

# Solve for equilibrium position x*
x_star = sp.solve(eq_x, x)[0]
x_star_simplified = sp.simplify(x_star)

# Normal force along y balances weight (no vertical motion, frictionless rail)
g = sp.symbols('g', positive=True, real=True)
N = m*g

# Existence condition for interior equilibrium if there are hard stops x in [0, Xmax]
existence_condition = sp.And(sp.Ge(x_star_simplified, 0), sp.Le(x_star_simplified, Xmax))

print("Equilibrium position:")
sp.pprint(sp.Eq(sp.Symbol('x*'), x_star_simplified))   # x* = L - (m a)/k

print("\nNormal force at equilibrium:")
sp.pprint(sp.Eq(sp.Symbol('N'), N))

print("\nExistence condition for interior equilibrium (with hard stops 0 ≤ x ≤ X_max):")
sp.pprint(existence_condition)

# Quick physical notes for sanity:
# If a -> 0: x* -> L  (spring at natural length in an inertial frame)
# If a increases: x* < L (spring compresses to push right and balance leftward fictitious force)
print("\nSanity checks:")
print("a -> 0: x* =", sp.simplify(x_star_simplified.subs(a, 0)))
```

## Problem D3.5

**Scenario.** Inside a subway car accelerating **horizontally to the right** with constant acceleration $a$, a **straight, rigid rod** of length $L$ is fixed to the car so that it makes a fixed angle $\phi$ with the **vertical downward** direction (so the rod leans to the right if $\phi>0$). A small **frictionless bead** of mass $m$ can slide along the rod. The bead is attached to the top end of the rod (the pivot point) by a **light spring** of stiffness $k$ that runs **along the rod**; when the bead is at the pivot the spring is unstretched.

We analyze everything **in the non-inertial car frame**, where a single fictitious (inertial) force $\mathbf{F}_\text{fict} = -m\,\mathbf{a}$ acts on the bead (to the **left**).

**Given.**
- Coordinates: $x$ to the right, $y$ downward.  
- The rod direction (unit vector $\hat{\mathbf{u}}$) has components $\hat{\mathbf{u}}=(\sin\phi)\,\hat{\mathbf{x}}+(\cos\phi)\,\hat{\mathbf{y}}$.
- **Equation of constraint (geometry of motion along the rod):** the bead’s position is constrained to the rod, so its coordinates satisfy
- 
 $$
  (x,y) = \big(s\sin\phi,\; s\cos\phi\big),\qquad 0\le s\le L,
  $$
  
  where $s$ is the **scalar coordinate along the rod** measured from the pivot (where the spring is anchored).

**Tasks.**
1. Using **Newton’s second law along the rod**, find the **equilibrium position** $s^\ast$ of the bead (where it comes to rest relative to the car).  
2. Find the **normal reaction N** of the rod on the bead (perpendicular to the rod) at equilibrium.  
3. State the **existence condition** for equilibrium (i.e., when $0\le s^\ast\le L$) and interpret the limiting cases $a\to 0$ and $g\to 0$.

**Hints.**
- Resolve real forces (gravity $m\mathbf g$, spring $-k s\,\hat{\mathbf{u}}$) and the **single** fictitious force $-m a\,\hat{\mathbf{x}}$ **along** and **perpendicular** to $\hat{\mathbf{u}}$.  
- Along the rod at equilibrium: sum of components $=0$.  
- Perpendicular to the rod at equilibrium: the normal force $N$ balances the perpendicular components of gravity and the fictitious force.


```python
%reset -f

import sympy as sp

# Symbols
m, g, a, k, L, phi, s = sp.symbols('m g a k L phi s', positive=True, real=True)

# Unit vectors: we don't need explicit vectors; just project along and perpendicular to the rod.
# Along-rod unit: u = (sinφ, cosφ)
sinφ, cosφ = sp.sin(phi), sp.cos(phi)

# Forces:
# Gravity: m g in +y (downward). Component along rod: m g * cosφ
F_g_along = m*g*cosφ

# Fictitious force: -m a in -x (to the left). Component along rod: (-m a)·u = -m a * sinφ
F_fict_along = -m*a*sinφ

# Spring force along rod: -k s (toward the pivot), i.e., opposite to +u
F_spring_along = -k*s

# Equilibrium along the rod: sum = 0
eq_along = sp.Eq(F_g_along + F_fict_along + F_spring_along, 0)

# Solve for equilibrium s*
s_star = sp.simplify(sp.solve(eq_along, s)[0])

# Now compute the normal reaction N (perpendicular to rod).
# Perpendicular-to-rod unit (rotate u clockwise): n = (cosφ, -sinφ)
# Perpendicular components:
#   Gravity ⟂: m g * (projection on n) = m g * (-sinφ)  (since n·(0,1) = -sinφ)
#   Fictitious ⟂: (-m a, 0)·n = -m a * cosφ
# The normal N (from rod on bead) must balance these at equilibrium:
#   N + m g*(-sinφ) + (-m a*cosφ) = 0  =>  N = m(g sinφ + a cosφ)
N = sp.simplify(m*(g*sp.sin(phi) + a*sp.cos(phi)))

# Existence condition for equilibrium: 0 <= s* <= L
s_star_conditions = sp.And(sp.Ge(s_star, 0), sp.Le(s_star, L))

# Nice forms
s_star_simplified = sp.simplify(s_star)
N_simplified = sp.simplify(N)

print("Equilibrium position along the rod:")
sp.pprint(sp.Eq(sp.Symbol('s*'), s_star_simplified))
print("\nNormal reaction (perpendicular to the rod) at equilibrium:")
sp.pprint(sp.Eq(sp.Symbol('N'), N_simplified))

# Interpretations and limits
print("\nExistence condition for equilibrium (on-rod): 0 <= s* <= L  ->", s_star_conditions)
print("\nChecks / limits:")
print("a -> 0: s* =", sp.simplify(s_star_simplified.subs(a, 0)), " (pure gravity + spring along rod)")
print("g -> 0: s* =", sp.simplify(s_star_simplified.subs(g, 0)), " (pure fictitious + spring along rod)")
```
