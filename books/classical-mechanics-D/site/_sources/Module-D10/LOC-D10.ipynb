## Why this module matters

So far, we have described motion using **forces and Newton’s laws**. While this approach is powerful, it can become complicated when dealing with:

- multiple coordinates  
- constrained motion  
- non-Cartesian systems  

The **Lagrangian formulation** provides a new and often simpler way to describe motion based on **energy rather than force**.

In this module, we introduce the Lagrangian and the **Lagrange equations**

The goal at this stage is not to derive these equations, but to **learn how to use them** in familiar Physics I systems.

This sets the stage for later phases, where we will:
- derive the equations from deeper principles  
- apply them to constrained systems  
- extend them to more advanced problems

By the end of this module, you will be able to:

---

### A. The Lagrangian  
*Defines a new way to describe physical systems.*

- **A1.** Define the Lagrangian as  

$$
L = T - U.
$$

- **A2.** Identify kinetic energy $T$ and potential energy $U$ for simple systems.  
- **A3.** Express the Lagrangian in terms of a generalized coordinate $q$ and velocity $\dot{q}$.  

---

### B. Lagrange’s Equation  
*Introduces the equation of motion in Lagrangian form.*

- **B1.** State Lagrange’s equation  

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\right)
-
\frac{\partial L}{\partial q}
= 0.
$$

- **B2.** Identify each term in the equation:
  - $\frac{\partial L}{\partial \dot{q}}$  
  - $\frac{\partial L}{\partial q}$  
  - time derivative  

- **B3.** Recognize that this equation produces the system’s equation of motion.  

---

### C. Solving Problems with the Lagrangian  
*Applies the Lagrangian to familiar systems.*

- **C1.** Construct the Lagrangian for simple systems (e.g., particle, spring, pendulum).  
- **C2.** Apply Lagrange’s equation step-by-step to obtain the equation of motion.  
- **C3.** Verify that the result matches the corresponding Newtonian equation.  

---

### D. Interpretation and Connection  
*Builds intuition for the new formulation.*

- **D1.** Interpret the Lagrangian as a function of position and velocity.  
- **D2.** Understand that Lagrange’s equation replaces Newton’s second law.  
- **D3.** Recognize that this approach naturally extends to more complex systems.  

---

### Performance Criteria

- The Lagrangian is correctly constructed from kinetic and potential energy.  
- Lagrange’s equation is applied correctly to obtain equations of motion.  
- Results are consistent with known Physics I systems.  
- Students demonstrate fluency in using the Lagrangian without relying on force-based reasoning.
