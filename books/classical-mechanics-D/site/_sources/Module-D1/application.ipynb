# D1.2 Using Impulse–Momentum Instead of $F = ma$
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D1.2.1 Motivation

In Physics I, most problems were solved using Newton’s second law:

$$
\mathbf{F} = m\mathbf{a}.
$$

This approach focuses on the **instantaneous relationship** between force and motion and typically leads to solving differential equations.

However, as we have emphasized, the more fundamental relationship is

$$
d\mathbf{p} = \mathbf{F}\,dt,
$$

which leads to the **impulse–momentum theorem**

$$
\int \mathbf{F}\,dt = \Delta \mathbf{p}.
$$

---

### Why use impulse–momentum?

- It describes motion through **accumulated effects over time**, rather than instantaneous changes  
- It avoids solving differential equations directly  
- It naturally extends to **time-dependent forces**  
- It is essential for understanding **variable mass systems** (where $m$ is not constant)  

---

### Key Perspective

> Physics is fundamentally expressed in **differential form**

$$
d\mathbf{p} = \mathbf{F}\,dt,
$$

>not as pre-packaged derivative equations like

$$
\mathbf{F} = m\frac{d\mathbf{v}}{dt}.
$$

---

In this section, we will solve problems using the impulse–momentum approach instead of $F = ma$.

## D1.2.2 Example — Block on an Incline with Friction

<div style="background-color:#e0f7fa; border-left:6px solid #00838f; padding:12px; border-radius:4px;">
<strong>Example — Velocity from Impulse–Momentum</strong>

A block of mass $m = 2.0~\text{kg}$ is released from rest on an incline of angle $\theta = 30^\circ$. The coefficient of kinetic friction is $\mu_k = 0.20$.

Find the velocity of the block after $t = 3.0~\text{s}$ using the impulse–momentum theorem.

---

**Solution**

The forces along the incline are:

- Gravity component:  

$$
F_g = mg\sin\theta
$$

- Friction (opposing motion):  

$$
F_f = \mu_k mg\cos\theta
$$

Thus, the net force is

$$
F_{\text{net}} = mg\sin\theta - \mu_k mg\cos\theta.
$$

---

Apply impulse–momentum:

$$
\int_0^t F_{\text{net}}\,dt = \Delta p.
$$

Since the force is constant and the block starts from rest,

$$
F_{\text{net}} t = m v.
$$

Substitute:

$$
\left[ mg(\sin\theta - \mu_k \cos\theta) \right] t = m v.
$$

Cancel $m$:

$$
g(\sin\theta - \mu_k \cos\theta)\, t = v.
$$

---

Insert values:

$$
v = (9.8)\left(\sin 30^\circ - 0.20 \cos 30^\circ\right)(3.0).
$$

$$
v = 9.8\left(0.5 - 0.20(0.866)\right)(3.0).
$$

$$
v = 9.8(0.5 - 0.173)(3.0).
$$

$$
v = 9.8(0.327)(3.0).
$$

$$
v \approx 9.6~\text{m/s}.
$$

---

**Result**

$$
\boxed{
v \approx 9.6~\text{m/s}
}
$$

---

**Insight**

We solved this problem without ever writing

$$
a = \frac{F}{m}.
$$

Instead, we directly accumulated the effect of the force over time.

<h3 style="margin-top:0;">Box Activity — Atwood Machine Acceleration from Impulse–Momentum</h3>

An Atwood machine has two blocks connected by a light string over a frictionless pulley. Block $a$ has mass $m_a$, and block $b$ has mass $m_b$, with $m_b>m_a$.

A ground-based observer stands below the Atwood machine and defines upward as the positive direction for both blocks. At the instant we begin observing, the blocks are already moving.

Use the impulse–momentum theorem, not $\sum F=ma$, to determine the acceleration of each block.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Activity solution
</summary>

**Step 1 — Choose the reference frame and sign convention**

The observer is fixed on the ground, so we use the ground frame. This is an inertial reference frame.

Upward is positive for both blocks.

Because $m_b>m_a$, we expect block $a$ to accelerate upward and block $b$ to accelerate downward.

---

**Step 2 — Apply impulse–momentum to each block separately**

The impulse–momentum theorem with constant force is

$$
\Delta p = F_{\text{net}}\Delta t
$$

Since each block has constant mass,

$$
\Delta p = m\Delta v
$$

For block $a$, do not assume the tension has the same value as on block $b$ yet. Call the tension force on block $a$ by the string $T_a$.

$$
\boxed{m_a\Delta v_a = (T_a-m_ag)\Delta t}
$$

For block $b$, call the tension force on block $b$ by the string $T_b$.

$$
\boxed{m_b\Delta v_b = (T_b-m_bg)\Delta t}
$$

These are the impulse–momentum equations for the two blocks.

---

**Step 3 — Use the pulley information**

The pulley is frictionless, and the string is light.

A frictionless pulley only redirects the tension force; it does not change its magnitude.

Therefore,

$$
T_a=T_b=T
$$

So the boxed impulse–momentum equations become

$$
m_a\Delta v_a = (T-m_ag)\Delta t
$$

$$
m_b\Delta v_b = (T-m_bg)\Delta t
$$

---

**Step 4 — Equation of Constraint**

Because the string length is fixed, when block $a$ moves up, block $b$ moves down by the same amount.

Using upward as positive for both blocks,

$$
\Delta y_a=-\Delta y_b
$$

Divide by $\Delta t$:

$$
\frac{\Delta y_a}{\Delta t}=-\frac{\Delta y_b}{\Delta t}
$$

So the velocity changes must also be equal in magnitude and opposite in direction:

$$
\Delta v_a=-\Delta v_b
$$

Let block $a$ have acceleration $+a$. Then over a time interval $\Delta t$,

$$
\Delta v_a=a\Delta t
$$

and therefore

$$
\Delta v_b=-a\Delta t
$$

---

**Step 5 — Substitute the constraint into the impulse–momentum equations**

For block $a$:

$$
m_a(a\Delta t)=(T-m_ag)\Delta t
$$

Cancel $\Delta t$:

$$
m_aa=T-m_ag
$$

For block $b$:

$$
m_b(-a\Delta t)=(T-m_bg)\Delta t
$$

Cancel $\Delta t$:

$$
-m_ba=T-m_bg
$$

---

**Step 6 — Solve for the acceleration**

We now have

$$
m_aa=T-m_ag
$$

$$
-m_ba=T-m_bg
$$

Solve each equation for $T$:

$$
T=m_aa+m_ag
$$

$$
T=m_bg-m_ba
$$

Set them equal:

$$
m_aa+m_ag=m_bg-m_ba
$$

Collect the acceleration terms:

$$
m_aa+m_ba=m_bg-m_ag
$$

$$
(m_a+m_b)a=(m_b-m_a)g
$$

Therefore,

$$
\boxed{a=\frac{(m_b-m_a)g}{m_a+m_b}}
$$

So block $a$ accelerates upward:

$$
\boxed{a_a=+\frac{(m_b-m_a)g}{m_a+m_b}}
$$

and block $b$ accelerates downward:

$$
\boxed{a_b=-\frac{(m_b-m_a)g}{m_a+m_b}}
$$

The blocks did not need to start from rest. The impulse–momentum theorem only requires the change in momentum, and the string constraint determines how the two velocity changes are related.

</details>

---

### Big Idea

> Impulse–momentum allows us to go directly from **force → velocity**, without solving for acceleration.
