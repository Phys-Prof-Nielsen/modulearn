# D1.0 Learning Outcomes — Constraints and Generalized Coordinates
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this module matters

In introductory mechanics, motion is typically described using **Cartesian coordinates** and analyzed directly through Newton’s laws. While this approach works well for simple systems, it becomes inefficient when motion is restricted by **geometric constraints** such as rods, surfaces, or connected objects.

In classical mechanics we instead begin by identifying the **degrees of freedom** of a system and describing motion using **coordinates that respect the constraints**. These variables, called **generalized coordinates**, allow us to describe motion using only the independent variables that determine the configuration of the system.

This module introduces the fundamental ideas that make this approach possible. We begin by identifying **constraints** and learning how they restrict motion and reduce the number of independent variables. This leads naturally to the concept of **degrees of freedom**, which determines how many coordinates are required to specify the configuration of a system.

We then introduce **generalized coordinates** and **generalized velocities**, which provide a flexible way to describe motion that is adapted to the geometry of the system. We also examine how coordinates are transformed and how **velocities arise from coordinate transformations using the chain rule**.

Finally, we explore an important conceptual idea: many physical laws remain unchanged under certain transformations. These **symmetries or invariances** reveal that some parts of our mathematical description are arbitrary choices rather than physical properties. Later in the course, we will see that these symmetries lead directly to **conservation laws through Noether’s theorem**.

These ideas form the coordinate-based language used throughout **analytical mechanics**, where the equations of motion will be expressed in terms of generalized coordinates rather than Cartesian components of force and acceleration.

By the end of this module, you will be able to:

---

### A. Constraints and Restricted Motion  
*Introduces the role of constraints in mechanical systems.*

- **A1.** Describe **constraints** as physical conditions that restrict the motion of a system.  
- **A2.** Write **constraint equations** that mathematically represent geometric restrictions on motion.  
- **A3.** Distinguish between **holonomic constraints** and **non-holonomic constraints**.  
- **A4.** Explain how constraints reduce the number of independent variables describing motion.

---

### B. Degrees of Freedom  
*Establishes how many independent coordinates describe a system.*

- **B1.** Define **degrees of freedom** as the number of independent coordinates required to specify the configuration of a system.  
- **B2.** Determine the degrees of freedom of simple mechanical systems.  
- **B3.** Use the relation  

$$
\text{DOF} = 3N - k
$$

to determine the number of independent variables describing a system of particles with constraints.  

- **B4.** Interpret how constraints reduce the dimensionality of motion.

---

### C. Generalized Coordinates  
*Introduces coordinates adapted to the geometry of constrained systems.*

- **C1.** Define **generalized coordinates** as independent variables that uniquely describe the configuration of a system.  
- **C2.** Choose generalized coordinates that simplify the description of constrained motion.  
- **C3.** Express Cartesian coordinates as functions of generalized coordinates.  
- **C4.** Recognize that generalized coordinates may represent **distances, angles, or other parameters**, not necessarily Cartesian positions.

---

### D. Generalized Velocities  
*Connects generalized coordinates to motion.*

- **D1.** Define **generalized velocities**

$$
\dot q_i = \frac{dq_i}{dt}.
$$

- **D2.** Interpret generalized velocities as rates of change of the generalized coordinates.  
- **D3.** Distinguish generalized velocities from Cartesian velocity components.

---

### E. Configuration Space  
*Provides a geometric interpretation of system configurations.*

- **E1.** Describe **configuration space** as the set of all possible configurations of a system.  
- **E2.** Interpret the coordinates  

$$
(q_1, q_2, \dots, q_N)
$$

as defining a point in configuration space.  
- **E3.** Relate the dimensionality of configuration space to the number of degrees of freedom.

---

### F. Coordinate Transformations and the Chain Rule  
*Relates coordinate choice to velocity and kinetic energy.*

- **F1.** Write coordinate transformations expressing Cartesian coordinates as functions of generalized coordinates,

$$
x_i = x_i(q_1,q_2,\dots,t).
$$

- **F2.** Apply the **chain rule** to compute velocities when coordinates are transformed,

$$
\mathbf{v} =
\sum_i
\frac{\partial \mathbf{r}}{\partial q_i}\dot q_i
+
\frac{\partial \mathbf{r}}{\partial t}.
$$

- **F3.** Compute velocity components and speeds using coordinate transformations.

---

### G. Invariance and Symmetry  
*Introduces the idea that physical laws may remain unchanged under certain transformations.*

- **G1.** Define **invariance** as the property that a physical law remains unchanged under a transformation.  
- **G2.** Distinguish between **symmetry of description** and **symmetry of a physical system**.  
- **G3.** Show that shifting the coordinate origin  

$$
x \rightarrow x + a
$$

does not change velocity, acceleration, or momentum.  

- **G4.** Explain why adding a constant to potential energy  

$$
U \rightarrow U + C
$$

does not change the force or motion.  

- **G5.** Recognize that invariances of physical laws are connected to **conservation laws**, a relationship formalized later by **Noether’s theorem**.

---

### Performance Criteria

- Constraints are clearly identified and expressed using **mathematical relations**.  
- Degrees of freedom are determined systematically from the number of coordinates and constraints.  
- Generalized coordinates are chosen to **match the geometry and constraints of the system**.  
- Velocities are computed correctly using the **chain rule**.  
- Motion is consistently described using **generalized coordinates and generalized velocities**.  
- Students recognize when **different mathematical descriptions represent the same physical system**.
