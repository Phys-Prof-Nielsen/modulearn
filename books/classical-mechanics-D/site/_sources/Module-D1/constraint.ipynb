# D1.1 Constraints in Classical Mechanics

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D1.1.1 Motivation: Why Constraints?

In introductory physics, we often solved problems by implicitly using **constraints**, even if we didn’t call them that. For example:

- A block on a ramp only moves along the slope, not perpendicular to it.  
- A pendulum bob is forced to stay at a fixed distance from its pivot.  

In each of these cases, the motion of the system is **restricted** by some physical condition.  
In classical mechanics, especially when we move into **Lagrangian and Hamiltonian formulations**, it is crucial to **explicitly state and classify** these restrictions.  
Constraints allow us to reduce the number of independent coordinates, and they determine the mathematical form of the equations of motion.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D1.1.2 Examples from Introductory Physics

Let us revisit a few familiar systems, but now with the language of **constraints**:

1. **Pendulum (rigid rod of length $l$):**
   - Coordinates of the bob: $(x, y)$  
   - Constraint:
     
     $$
     x^2 + y^2 - l^2 = 0
     $$
     
   - This reduces the motion to one independent variable (the angle $\theta$).
   - Alternatively, we could  write this as $r = R$ in polar coordinates.

2. **Block sliding on a ramp with angle $\alpha$:**
   - Coordinates of the block: $(x, y)$  
   - Constraint (equation of the ramp) with coordinate system x-axis aligned with the ramp:
     
     $$
     y = 0
     $$
     
   - Motion is restricted to one dimension along the incline.

In all of these, the constraint reduces the system to fewer **degrees of freedom**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D1.1.3 Constraint Equations: General Form

In mechanics, the coordinates of a system are denoted by

$$
q_1, q_2, \dots, q_n
$$

where $n$ is the number of coordinates needed to describe all possible motions in an unconstrained system.

A **constraint** is any condition that restricts these coordinates.  
Mathematically, a constraint can often be written as an equation:

$$
f(q_1, q_2, \dots, q_n, t) = 0
$$

- If the function $f$ involves only the coordinates and possibly time, it is called a **holonomic constraint**.  
- If it involves velocities or cannot be expressed as a simple coordinate relation, it is called a **non-holonomic constraint**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D1.1.4 Holonomic Constraints

A **holonomic constraint** is one that can be expressed as:

$$
f(q_1, q_2, \dots, q_n, t) = 0
$$

Example:

- **Pendulum:** $x^2 + y^2 - l^2 = 0$  

Holonomic constraints reduce the number of independent coordinates by one for each independent equation.  
They allow us to define new **generalized coordinates** that automatically satisfy the constraints.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D1.1.5 Non-Holonomic Constraints

A **non-holonomic constraint** cannot be expressed purely in terms of coordinates.  
Instead, it may involve inequalities or differential relations.

Examples:
1. **Rolling without slipping (sphere of radius $R$ on a plane):**
   - Condition:
      
     $$
     v_{\text{cm}} = R \, \omega
     $$
     
   - This relates velocity of the center of mass to angular velocity, not just positions.

2. **Particle constrained to stay inside a sphere of radius $R$:**
   - Inequality constraint:
      
     $$
     x^2 + y^2 + z^2 \leq R^2
     $$
     
   - This is not an equation, but an inequality.

Such constraints **do not reduce the coordinates in a straightforward way** and require special treatment in mechanics (e.g., Lagrange multipliers with non-holonomic conditions).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D1.1.6 Summary

- Constraints are conditions that restrict motion.  
- They are essential to understand in order to correctly count **degrees of freedom** and formulate the **Lagrangian equations of motion**.  
- **Holonomic constraints:**  
  - Can be written as equations involving only coordinates and time.  
  - Allow the introduction of generalized coordinates.  
- **Non-holonomic constraints:**  
  - Cannot be reduced to coordinate equations alone.  
  - Often involve velocities or inequalities.  
  - Require more advanced treatment.

<h3> Example 1 - Pulley Constraint</h3>
    </header>
    
**Problem:**  
Consider a simple pulley system:  
- A frictionless pulley with a massless rope passes over it.  
- On the left side of the rope hangs mass $m_1$, and on the right side hangs mass $m_2$.  
- The rope is inextensible (fixed length).

1. Write down the **constraint equation** that relates the coordinates of $m_1$ and $m_2$.  
2. Derive the relationship between their **velocities** and **accelerations**.  
 
</div>

```{admonition} Solution - Example 1
:class: tip, dropdown

### ✅ Solution

Consider the following figure:
<img src="D1.1-E1.png"  width="400"/>

1. **Constraint Equation**

   Let $y_1$ and $y_2$ be the vertical displacements of masses $m_1$ and $m_2$ measured downward from a reference point centered at the pulley.

   Since the rope length is constant:
   
   $$
   y_1 + y_2 - L = 0
   $$
   
   where $L$ is the total length of rope (excluding the part wrapped around the pulley).

   This is a **holonomic constraint**, since it relates only the coordinates.

---

2. **Velocity and Acceleration Relations**

   Differentiate the constraint once:
   
   $$
   \dot{y}_1 + \dot{y}_2 = 0
   $$
   
   which means:
   
   $$
   \dot{y}_1 = -\dot{y}_2
   $$

   Differentiate again:
   
   $$
   \ddot{y}_1 + \ddot{y}_2 = 0
   $$
   
   or
   
   $$
   \ddot{y}_1 = -\ddot{y}_2
   $$

   Thus, when one mass moves down, the other moves up with equal speed and opposite acceleration.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">
