# D6.3 Linearization About Stable Equilibrium
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why Linearization?

Most physical systems are **not exactly solvable**. However, near a **stable equilibrium**, their behavior simplifies dramatically.

👉 The key idea:
> Replace a complicated system with a **simpler, approximate one** that captures the local behavior.

This process is called **linearization**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.3.1 Linearization of the Force

Suppose a system is governed by a force $F(x)$, and let $x_0$ be an equilibrium point:

$$
F(x_0) = 0.
$$

We expand the force in a Taylor series about $x_0$:

$$
F(x) \approx F(x_0) + (x - x_0) F'(x_0) + \cdots
$$

Since $F(x_0) = 0$, this becomes

$$
F(x) \approx (x - x_0) F'(x_0).
$$

---

### Stable Equilibrium

From D6.2, stability requires:

$$
F'(x_0) < 0.
$$

Define

$$
k \equiv -F'(x_0) > 0.
$$

Then

$$
F(x) \approx -k(x - x_0).
$$

👉 This is exactly **Hooke’s Law**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.3.2 Equation of Motion Near Equilibrium

Applying Newton’s second law:

$$
m\ddot{x} = -k(x - x_0).
$$

Define displacement from equilibrium:

$$
\xi = x - x_0.
$$

Then

$$
m\ddot{\xi} + k\xi = 0.
$$

---

<div style="background-color:#e7f3ff; padding: 15px; border-left: 6px solid #2196F3;">

<strong>Definition: Linearized Equation of Motion</strong>

Near a stable equilibrium, the motion of a system is approximately governed by

$$
m\ddot{\xi} + k\xi = 0,
$$

which is the equation of **simple harmonic motion** with

$$
\omega = \sqrt{\frac{k}{m}}.
$$

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.3.3 Linearization from the Potential Energy

Often it is easier to work with the **potential energy** $U(x)$.

Equilibrium occurs at

$$
U'(x_0) = 0.
$$

Expand $U(x)$ about $x_0$:

$$
U(x) \approx U(x_0) + \frac{1}{2} U''(x_0)(x - x_0)^2.
$$

---

### Interpretation

- The constant term $U(x_0)$ does not affect motion  
- The linear term vanishes at equilibrium  
- The quadratic term dominates nearby  

Define

$$
k = U''(x_0).
$$

Then

$$
U(x) \approx \frac{1}{2}k(x - x_0)^2.
$$

👉 The system behaves like a **spring near equilibrium**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">


## D6.3.4 Emergence of SHM

Using

$$
F = -\frac{dU}{dx},
$$

we get

$$
F(x) \approx -k(x - x_0).
$$

Thus the equation of motion becomes

$$
m\ddot{\xi} + k\xi = 0.
$$

---

### Final Result

The angular frequency is

$$
\omega = \sqrt{\frac{k}{m}} = \sqrt{\frac{U''(x_0)}{m}}.
$$

---

<div style="background-color:#fff3cd; padding: 15px; border-left: 6px solid #ffecb5;">
<strong>Key Insight:</strong><br>
<strong>All stable equilibrium points behave like harmonic oscillators when viewed locally.</strong><br><br>
This is why SHM appears everywhere in physics—it is not a special case, but a universal approximation.
</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D6.3.5 When Linearization Fails

Linearization is an **approximation**, and it only works when:

- Displacements are **small**  
- Higher-order terms are negligible  

---

### Limitations

- Large oscillations → nonlinear behavior  
- Asymmetry in the potential → non-sinusoidal motion  
- Frequency may depend on amplitude  

---

<div style="background-color:#f8d7da; padding: 15px; border-left: 6px solid #f5c2c7;">
<strong>Danger:</strong><br>
Do not assume all oscillations are simple harmonic.  
SHM is only valid when the system remains close to equilibrium.
</div>
