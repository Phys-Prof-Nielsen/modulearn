## D5.2.1 Motivation

In Physics I, work was typically calculated in situations where motion occurred along a **single direction**, such as along the $x$-axis. In those cases, we could think of work as the force acting along the direction of motion, accumulated over distance.

This worked well because:

- motion was effectively **one-dimensional**  
- forces were aligned with the motion  
- the path taken was simple and often implicit  

---

## What changes in two dimensions?

In more realistic situations:

- objects can move in **multiple directions**  
- forces may act **at angles** or have components in different directions  
- motion may occur along paths that are not straight lines  

This leads to an important question:

> How do we think about work when motion is not confined to a single direction?

---

## A first step: motion along straight segments

Before developing a full mathematical framework, we begin with a simpler idea:

- break the motion into **straight segments**  
- for example: move along $x$, then along $y$  

This allows us to:

- extend our understanding of work to **two dimensions**  
- isolate how different components of force contribute  
- build intuition without introducing new mathematical tools  

---

## Why this matters

This approach reveals something new:

- work depends on **how the object moves**, not just where it starts and ends  
- changing the path between the same two points can change the work  

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:10px; border-radius:4px;">
<strong>Key Insight</strong>  

In one dimension, the path is essentially fixed.

In two dimensions, the path becomes a choice—and that choice can affect the work.

This is the first hint that work may depend on the path taken, a concept that will become central in the next module.

</div>

## D5.2.2 Line Integral of Vector Fields

From Calculus I, we learned how to evaluate integrals of functions that depend on a **single variable** (in physics, often position $x$ or time $t$).  

When integrating with respect to position $x$, we are implicitly considering motion along a **straight line segment**. For example, the work done by a force $\vec{F}$ acting on an object moving from $\vec{x}_1$ to $\vec{x}_2$ can be written as

$$
W = \int_{x_1}^{x_2} \vec{F} \cdot d\vec{x}.
$$

We can complete the dot product in either geometric or component forms:

$$
W = \int_{x_1}^{x_2} F \cos\theta \, dx,
$$

where:

- $F$ and $dx$ are the magnitudes of the force and differential displacement  
- $\theta$ is the angle between $\vec{F}$ and $d\vec{x}$,

or

$$
W = \int_{x_1}^{x_2} F_x \, dx,
$$

where:

- $F_x$ and $dx$ are the components of the force and differential displacement

Both expressions show that **only the component of the force along the direction of motion contributes to the work**.

---

### What this really represents

Although this looks like a standard single-variable integral, it is actually a special case of something more general:

> It is a **line integral of a vector field**, evaluated along a straight path.

In other words, even in Physics I, we were already doing line integrals—we just didn’t call them that because the path was simple.

---

### Generalizing to arbitrary paths

In general, motion does not occur along a straight line. Instead, an object may follow some path $C$ through space.

If we take a small segment of that path, we can represent it as a **directional displacement vector** $d\vec{l}$. The work done along the entire path is then

$$
W = \int_C \vec{F} \cdot d\vec{l},
$$

where the subscript $C$ indicates that we are integrating **along the path**.

---

### Making this computable

Since we are not yet comfortable integrating vectors directly, we express the dot product in terms of components.

In Cartesian coordinates:

$$
W = \int_C (F_x\,dx + F_y\,dy + F_z\,dz).
$$

This is a crucial step:

> We have converted a vector integral into a sum of scalar contributions.

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:10px; border-radius:4px;">
<strong>Key Insight</strong>  

A line integral is built from familiar one-dimensional integrals—but applied along a path in space.

The key idea is simple:

> Work is accumulated from the component of the force along each small segment of the path.

</div>

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Work Along a Piecewise Path</h3>

<strong>Problem</strong>  

Consider a massive physics textbook with a mass of $2.18~\text{kg}$. Its initial position is

$$
\vec{r}_0 = (0,0).
$$

A constant push force of magnitude $5.0~\text{N}$ is applied. The book is first moved along the $+x$ direction, and then along the $+y$ direction, ending at

$$
\vec{r} = (0.3~\text{m},\,0.2~\text{m}).
$$

Thus, the path $L$ consists of two straight segments forming two sides of a rectangle.

Find the total work done by the applied force.

---

<strong>Solution</strong>  

We begin with the definition of work:

$$
W = \int_C \vec{F}\cdot d\vec{l}.
$$

---

<strong>Step 1 — Split the path</strong>  

The motion occurs in two distinct stages:

- Path 1: motion along the $x$-direction from $0 \to 0.3$  
- Path 2: motion along the $y$-direction from $0 \to 0.2$  

Thus, the total work is the sum of the work along each segment:

$$
W = W_1 + W_2.
$$

---

<strong>Step 2 — Work along each segment</strong>  

Along the $x$-direction:

- $d\vec{l} = dx\,\hat{\mathbf{i}}$  
- Only the $x$-component of the force contributes  

$$
W_1 = \int_0^{0.3} F_x\,dx.
$$

Along the $y$-direction:

- $d\vec{l} = dy\,\hat{\mathbf{j}}$  
- Only the $y$-component of the force contributes  

$$
W_2 = \int_0^{0.2} F_y\,dy.
$$

---

<strong>Step 3 — Insert values</strong>  

With a constant force of magnitude $5.0~\text{N}$ along each segment:

$$
W = \int_0^{0.3} 5.0\,dx + \int_0^{0.2} 5.0\,dy.
$$

---

<strong>Step 4 — Evaluate</strong>  

$$
W = (5.0)(0.3) + (5.0)(0.2)
$$

$$
W = 1.5 + 1.0
$$

$$
\boxed{W = 2.5~\text{J}}
$$

---

<strong>Interpretation</strong>  

The key idea is that we do not split the integral arbitrarily—we split it because the **path itself is piecewise**.

Each segment contributes work based on the component of the force along that segment.

</div>

---

<div style="background-color:#ffe6e6; border-left:6px solid #990000; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Danger Zone — Components Are Not the Same as Path Segments</h3>

A common mistake is to take a straight-line displacement from

$$
(0,0) \to (0.3,0.2)
$$

and treat it as if the object moved first along $x$ and then along $y$.

These are **not the same path**.

---

<strong>Incorrect approach — treating components as path segments</strong>

Suppose we incorrectly write the work as

$$
W =
\int_0^{0.3} F_x\,dx
+
\int_0^{0.2} F_y\,dy.
$$

If we insert

$$
F_x = 5.0~\text{N},
\qquad
F_y = 5.0~\text{N},
$$

then

$$
W =
\int_0^{0.3} 5.0\,dx
+
\int_0^{0.2} 5.0\,dy.
$$

This gives

$$
W = (5.0)(0.3) + (5.0)(0.2)
$$

$$
W = 1.5 + 1.0
$$

$$
W = 2.5~\text{J}.
$$

But this result corresponds to a **two-segment path**, not the straight-line path.

---

<strong>Correct idea — straight-line motion has a shorter path length</strong>

The straight-line distance from $(0,0)$ to $(0.3,0.2)$ is

$$
s =
\sqrt{(0.3)^2 + (0.2)^2}.
$$

Therefore,

$$
s =
\sqrt{0.09 + 0.04}
=
\sqrt{0.13}
\approx 0.36~\text{m}.
$$

If the applied force has constant magnitude

$$
F = 5.0~\text{N}
$$

and points along the straight-line displacement, then the actual work along the straight path is

$$
W_{\text{straight}} = Fs.
$$

Thus,

$$
W_{\text{straight}} = (5.0)(0.36)
$$

$$
W_{\text{straight}} \approx 1.8~\text{J}.
$$

---

<strong>Comparison</strong>

Two-segment calculation:

$$
W_{\text{x-then-y}} = 2.5~\text{J}
$$

Straight-line path:

$$
W_{\text{straight}} \approx 1.8~\text{J}
$$

These are not the same because they correspond to **different paths**.

</div>

---

<div style="background-color:#ffe6e6; border-left:6px solid #990000; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Danger Zone — Infinitesimal vs Finite Motion</h3>

The expression

$$
d\vec{l} = dx\,\hat{\mathbf{i}} + dy\,\hat{\mathbf{j}}
$$

does <strong>not</strong> mean the object first moves in $x$ and then moves in $y$.

It means that each infinitesimal displacement along the actual path has components in the $x$ and $y$ directions.

---

**What this really means**

At any given point along a path:

- the motion has a specific direction  
- that direction can be decomposed into $x$ and $y$ components  

Thus,

$$
d\vec{l}
$$

represents a **single tiny step along the path**, not two separate steps.

---

**The subtle but important distinction**

At the **infinitesimal level**:

- any small segment of a curve behaves like a straight line  
- that tiny displacement can always be written as  

$$
d\vec{l} = dx\,\hat{\mathbf{i}} + dy\,\hat{\mathbf{j}}
$$

So locally, it *looks like* motion in both $x$ and $y$ directions.

---

At the **finite level**:

- moving along a curve is not the same as moving first in $x$ and then in $y$  
- different paths between the same points can give **different work**  

</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:10px; border-radius:4px;">
<strong>Key Insight</strong>  

Breaking a displacement into components is not the same as breaking the motion into separate path segments.

> Components describe direction.  
> Path segments describe the actual route taken.

</div>

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<h3 style="margin-top:0;">Box Activity — Path Dependence of Work</h3>

<strong>Problem</strong>  

Consider a force given by

$$
\vec{F} = 5.0\,(xy\,\hat{\mathbf{i}} + y^2\,\hat{\mathbf{j}})
$$

acting on an object.

In two separate experiments, the object moves from

$$
\vec{r}_i = (0,0)
\quad \text{to} \quad
\vec{r}_f = (1,1)
$$

along two different paths:

- <strong>Path 1:</strong> along $+x$ followed by $+y$  
- <strong>Path 2:</strong> along $+y$ followed by $+x$  

Compute the work done by the force along each path and compare the results.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Path 1 — Move along $x$, then $y$</strong>  

First segment: $(0,0) \to (1,0)$  

$$
W_1^{(x)} =
\int_{0}^{1} F_x(x,0)\,dx
=
\int_{0}^{1} 5.0\,(x\cdot 0)\,dx
$$

$$
W_1^{(x)} = 0
$$

Second segment: $(1,0) \to (1,1)$  

$$
W_1^{(y)} =
\int_{0}^{1} F_y(1,y)\,dy
=
\int_{0}^{1} 5.0\,y^2\,dy
$$

$$
W_1^{(y)} = \frac{5}{3}
$$

Total work:

$$
W_{\text{Path 1}} = \frac{5}{3}
$$

---

<strong>Path 2 — Move along $y$, then $x$</strong>  

First segment: $(0,0) \to (0,1)$  

$$
W_2^{(y)} =
\int_{0}^{1} F_y(0,y)\,dy
=
\int_{0}^{1} 5.0\,y^2\,dy
$$

$$
W_2^{(y)} = \frac{5}{3}
$$

Second segment: $(0,1) \to (1,1)$  

$$
W_2^{(x)} =
\int_{0}^{1} F_x(x,1)\,dx
=
\int_{0}^{1} 5.0\,x\,dx
$$

$$
W_2^{(x)} = \frac{5}{2}
$$

Total work:

$$
W_{\text{Path 2}} = \frac{5}{3} + \frac{5}{2} = \frac{25}{6}
$$

---

<strong>Comparison</strong>  

$$
W_{\text{Path 1}} = \frac{5}{3},
\qquad
W_{\text{Path 2}} = \frac{25}{6}
$$

Thus,

$$
W_{\text{Path 1}} \ne W_{\text{Path 2}}
$$

<h3 style="margin-top:0;">Conservative vs Non-Conservative Forces</h3>

From the previous example, we observed that the work done in moving an object from point $A$ to point $B$ depended on the <strong>path taken</strong>.

This observation leads to an important classification of forces:

---

<strong>Non-conservative forces</strong>  

If the work done by a force depends on the path taken between two points, the force is called a <strong>non-conservative force</strong>.

Examples include:

- applied pushing or pulling forces  
- friction  

These forces are typically associated with energy being transferred <strong>into or out of a system</strong>.

---

<strong>Conservative forces</strong>  

If the work done by a force depends only on the initial and final positions—and not on the path taken—the force is called a <strong>conservative force</strong>.

Examples include:

- gravitational forces  
- spring forces  

These forces do not add or remove energy from the system. Instead, they <strong>transfer energy between different forms</strong>, such as kinetic and potential energy.

---

<strong>Why this matters</strong>  

Conservative forces allow us to define a <strong>potential energy function</strong>, which leads to powerful tools such as energy conservation:

$$
K + U = \text{constant}.
$$

Non-conservative forces, on the other hand, require us to explicitly account for energy entering or leaving the system.

---

<strong>Big picture</strong>  

> Conservative forces reshape energy within a system.  
> Non-conservative forces exchange energy with the surroundings.

</div>
