## D5.1.1 Why this module matters

In Physics I, we developed the concepts of **work and energy** primarily in one dimension or along simple paths. These ideas provide a powerful alternative to Newton’s second law by relating forces to motion through scalar quantities.

However, most real physical systems involve motion in **two or three dimensions**, where paths are not straight and forces may vary with position. Before extending these ideas, we revisit the foundational definitions, including **conservative forces and potential energy**, and clarify their assumptions and limitations.

This review prepares us to develop a more general and rigorous framework using **line integrals** and **local criteria for conservative forces**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.1.2 Work

Work is defined as the integral of the force along the path of motion:

$$
W = \int \mathbf{F} \cdot d\mathbf{r}.
$$

For motion along a straight line, this reduces to familiar expressions from Physics I. The key idea is that **only the component of the force along the displacement contributes to work**.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Work When Force and Displacement Are Not Aligned</h3>

<strong>Problem</strong>  

A particle moves along a straight horizontal path from $x=0$ to $x=5~\text{m}$.  
A constant force acts on the particle:

$$
\mathbf{F} = (3\,\hat{\mathbf{i}} + 4\,\hat{\mathbf{j}})\ \text{N}.
$$

Find the work done by the force.

---

<strong>Step 1 — Describe the displacement</strong>  

The motion is purely along the $x$-direction:

$$
d\mathbf{r} = dx\,\hat{\mathbf{i}}.
$$

---

<strong>Step 2 — Compute the dot product</strong>  

The differential work is

$$
dW = \mathbf{F} \cdot d\mathbf{r}.
$$

Substitute:

$$
dW = (3\,\hat{\mathbf{i}} + 4\,\hat{\mathbf{j}})\cdot(dx\,\hat{\mathbf{i}}).
$$

Using the dot product:

$$
\hat{\mathbf{i}}\cdot\hat{\mathbf{i}} = 1,
\qquad
\hat{\mathbf{j}}\cdot\hat{\mathbf{i}} = 0,
$$

we obtain

$$
dW = 3\,dx.
$$

---

<strong>Step 3 — Integrate</strong>  

$$
W = \int_0^5 3\,dx = 15~\text{J}.
$$

---

<strong>Final Answer</strong>  

$$
\boxed{W = 15~\text{J}}
$$

---

<strong>Interpretation</strong>  

Even though the force has both $x$ and $y$ components, only the component **along the direction of motion** contributes to work.

- The $3~\text{N}$ (horizontal) component does work  
- The $4~\text{N}$ (vertical) component does **no work**  

---

**Key Insight**

Work depends on the projection of the force along the displacement:

$$
W = \int \mathbf{F} \cdot d\mathbf{r}.
$$

Perpendicular components of force do not contribute to work.

## D5.1.3 Kinetic Energy

Kinetic energy is defined as

$$
K = \frac{1}{2} m\,\mathbf{v}\cdot\mathbf{v}.
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.1.4 Work–Energy Theorem

The net work done on a particle is equal to the change in kinetic energy:

$$
W_{\text{net}} = \Delta K.
$$

This provides an alternative to solving Newton’s second law directly by relating initial and final states.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.1.5 Work for Systems (Center of Mass)

For a system of particles, we define **pseudo work**:

$$
W_{\text{pseudo}} = \mathbf{F}_{\text{net}} \cdot d\mathbf{r}_{\text{cm}}.
$$

This describes how external forces influence the motion of the center of mass, and must be distinguished from the actual work done on individual particles.

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Pseudo Work vs Actual Work for Two Connected Masses</h3>

<strong>Problem</strong>  

Two identical masses, each of mass $m$, are connected by a light string on a frictionless horizontal surface. Initially, the two masses are aligned along the $y$-axis.

A constant external force

$$
\mathbf{F} = F\hat{\mathbf{i}}
$$

is applied by pulling the string in the $+x$ direction at a point between the two masses.

The point where the force is applied moves a distance

$$
D
$$

in the $+x$ direction.

Because the two masses remain separated by the string, the center of mass does not move as far as the point where the string is pulled. Let the center of mass move a distance

$$
D_{\text{cm}},
$$

where

$$
D_{\text{cm}} < D.
$$

Compare:

1. The actual work done by the applied force  
2. The pseudo work associated with the center-of-mass motion  
3. The physical meaning of the difference  

---

<strong>Step 1 — Actual work done on the system</strong>  

Actual work is computed using the displacement of the point where the force is applied:

$$
W_{\text{actual}} =
\int \mathbf{F}\cdot d\mathbf{r}_{\text{apply}}.
$$

Since the applied force is constant and the point of application moves a distance $D$ in the direction of the force,

$$
W_{\text{actual}} = FD.
$$

Thus,

$$
\boxed{
W_{\text{actual}} = FD
}
$$

---

<strong>Step 2 — Pseudo work of the center of mass</strong>  

Pseudo work uses the displacement of the center of mass:

$$
W_{\text{pseudo}}
=
\int \mathbf{F}\cdot d\mathbf{r}_{\text{cm}}
$$

The center of mass moves

$$
\Delta \mathbf{r}_{\text{cm}} = D_{\text{cm}}\hat{\mathbf{i}}.
$$

Thus,

$$
\boxed{
W_{\text{pseudo}} = FD_{\text{cm}}
}
$$

Since

$$
D_{\text{cm}} < D,
$$

we have

$$
\boxed{
W_{\text{pseudo}} < W_{\text{actual}}.
}
$$

---

<strong>Step 3 — Interpretation</strong>  

The actual work done by the applied force is

$$
W_{\text{actual}} = FD.
$$

This is the energy transferred into the full two-mass system.

The pseudo work is

$$
W_{\text{pseudo}} = FD_{\text{cm}},
$$

which only tracks the change in center-of-mass kinetic energy:

$$
W_{\text{pseudo}} = \Delta K_{\text{cm}}.
$$

The difference is

$$
W_{\text{actual}} - W_{\text{pseudo}}
=
F(D-D_{\text{cm}}).
$$

This energy goes into motion relative to the center of mass:

$$
K_{\text{total}}
=
K_{\text{cm}}
+
K_{\text{relative}}.
$$

This should be clear since the individual particles travels a longer distance than the CM during the interaction, and hence their individual speed **must** be greater than the speed of the CM. Therefore the total kinetic energy of the system **must** be greater than the CM kinetic energy.

</div>

---

<div style="background-color:#ffe6e6; border-left:6px solid #990000; padding:10px; border-radius:4px;">
<strong>Danger Zone — Pseudo Work Is Not Actual Work</strong>  

For a system of particles,

$$
W_{\text{pseudo}} = \int \mathbf{F_{\text{net}}}\cdot d\mathbf{r}_{\text{CM}}
$$

tracks only the center-of-mass motion.

But actual work depends on where the force is applied:

$$
W_{\text{actual}} =
\int \mathbf{F}\cdot d\mathbf{r}_{\text{apply}}.
$$

These are not generally the same.

</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:10px; border-radius:4px;">
<strong>Key Insight</strong>  

The point where the force is applied can move farther than the center of mass.

Therefore:

$$
W_{\text{actual}} > W_{\text{pseudo}}.
$$

The extra energy goes into internal or relative motion of the masses, not into motion of the center of mass.

</div>

## D5.1.6 Conservative Forces and Potential Energy

A **conservative force** is one for which the work done depends only on the initial and final positions, not on the path taken.

For such forces, we define a **potential energy function**:

$$
W = -\Delta U.
$$

In one dimension, we have

$$
F_x = -\frac{dU}{dx}.
$$

If we define the mechanical energy as

$$
E = K + U
$$

then we can rewrite the work–energy theorem between two states $A$ and $B$ as:

$$
E_A + W_{NC} = E_B.
$$

where $W_{NC}$ is the work done but non-conservative forces.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Highlight — Role of Non-Conservative Forces</strong>

From the energy relation

$$
E_A + W_{\text{nc}} = E_B,
$$

we observe that non-conservative forces can change the total mechanical energy of a system.

- If $W_{\text{nc}} > 0$, energy is added to the system  
- If $W_{\text{nc}} < 0$, energy is removed from the system  

Thus, unlike conservative forces, non-conservative forces allow energy to be transferred into or out of the system.

</div>

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<h3 style="margin-top:0;">Box Activity — Deriving Simple Harmonic Motion from Energy</h3>

<strong>Problem</strong>  

Consider a block of mass $m$ attached to a spring with spring constant $k$ on a frictionless horizontal surface.  

Using the <strong>work–energy theorem</strong>, derive an expression for the position $x(t)$ of the block.

---

<strong>Hint</strong>  

Use conservation of energy:

$$
K + U = \text{constant}
$$

with

$$
K = \frac{1}{2}m\,\dot{x}^2, 
\qquad
U = \frac{1}{2}kx^2.
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Step 1 — Energy conservation</strong>  

$$
\frac{1}{2}m\,\dot{x}^2 + \frac{1}{2}kx^2 = E.
$$

---

<strong>Step 2 — Solve for $\dot{x}$</strong>  

$$
\dot{x}^2 = \frac{2E}{m} - \frac{k}{m}x^2.
$$

$$
\dot{x} = \sqrt{\frac{2E}{m} - \frac{k}{m}x^2}.
$$

---

<strong>Step 3 — Separate variables</strong>  

$$
\frac{dx}{\sqrt{\frac{2E}{m} - \frac{k}{m}x^2}} = dt.
$$

Let

$$
A^2 = \frac{2E}{k},
\qquad
\omega = \sqrt{\frac{k}{m}}.
$$

Then

$$
\frac{dx}{\sqrt{A^2 - x^2}} = \omega\,dt.
$$

---

<strong>Step 4 — Integrate</strong>  

$$
\int \frac{dx}{\sqrt{A^2 - x^2}} = \int \omega\,dt.
$$

This gives

$$
\sin^{-1}\left(\frac{x}{A}\right) = \omega t + \phi.
$$

---

<strong>Step 5 — Solve for $x(t)$</strong>  

$$
x(t) = A \sin(\omega t + \phi).
$$

---

<strong>Final Result</strong>  

$$
\boxed{
x(t) = A \sin(\omega t + \phi), 
\qquad 
\omega = \sqrt{\frac{k}{m}}
}
$$

---

<strong>Interpretation</strong>  

We derived **simple harmonic motion** using energy alone, without directly solving Newton’s second law.

## D5.1.7 Conservation of Kinetic Energy — A Subtle but Important Limitation

In previous module, we emphasized that for **momentum**, internal forces cancel due to Newton’s third law, leading to conservation of total momentum in an isolated system.

It is tempting to make a similar argument for **kinetic energy**, but this is **not valid**.

---

### Why momentum and energy behave differently

Momentum is governed by

$$
\Delta \mathbf{P} = \int \sum \mathbf{F}\,dt.
$$

Since internal forces occur in equal and opposite pairs and act over the same time interval $dt$, their contributions cancel.

---

Work, however, is given by

$$
W = \int \mathbf{F}\cdot d\mathbf{r}.
$$

Here is the crucial difference:

- Internal forces are equal and opposite  
- But the **displacements $d\mathbf{r}$ of interacting particles are not necessarily equal and opposite**  

This means internal forces can do **net work** on a system.

---

<div style="background-color:#ffebee; border-left:6px solid #c62828; padding:12px; border-radius:4px;">
<strong>Danger — Internal Forces and Kinetic Energy</strong>

Even in an **isolated system**,

$$
\sum \mathbf{F}_{\text{ext}} = 0,
$$

it does <b>not</b> follow that kinetic energy is conserved.

---

<strong>Why?</strong>

For two interacting particles:

$$
\mathbf{F}_{12} = -\mathbf{F}_{21},
$$

but their displacements may differ:

$$
d\mathbf{r}_1 \neq -d\mathbf{r}_2.
$$

Thus,

$$
\mathbf{F}_{12}\cdot d\mathbf{r}_1
+
\mathbf{F}_{21}\cdot d\mathbf{r}_2
\neq 0.
$$

Internal forces can therefore do **net work**.

---

<strong>Physical consequences</strong>

This leads to:

- **deformation** of objects  
- **heat generation**  
- **sound production**  
- **internal energy changes**

---

<strong>Key takeaway</strong>

$$
\sum \mathbf{F}_{\text{ext}} = 0
\quad \Rightarrow \quad
\mathbf{P} = \text{constant}
$$

but

$$
\sum \mathbf{F}_{\text{ext}} = 0
\;\;\not\Rightarrow\;\;
K = \text{constant}.
$$

---

<strong>Elastic vs Inelastic Collisions</strong>

- **Elastic collision** → kinetic energy conserved  
- **Inelastic collision** → kinetic energy not conserved  

In both cases, momentum is conserved (if isolated), but kinetic energy may not be.

</div>

---

### Big Idea

> Momentum conservation follows from force symmetry.  
> Kinetic energy conservation requires stronger conditions.

## D5.1.8 Limitations of the Physics I Approach

In many Physics I problems:

- Motion is effectively one-dimensional  
- Paths are simple or implicitly assumed  
- Conservative forces are identified based on known examples (e.g., gravity, springs)  
- Work is often evaluated along a single path  

However, these approaches have important limitations:

- In multiple dimensions, work must be computed along **arbitrary paths**  
- Work may depend on the **path taken**, not just endpoints  
- Evaluating work along a single path is **not sufficient** to determine whether a force is conservative  
- There is no general method (yet) for determining whether a force is conservative  

This means we need a more systematic and rigorous framework.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D5.1.9 Looking Ahead

To overcome these limitations, we will:

- express work along curves using **line integrals**  
- parameterize motion using $\mathbf{r}(t)$  
- distinguish clearly between path-dependent and path-independent work  
- develop a **local test for conservative forces** using the curl  

This will allow us to determine, in a general and practical way, whether a force admits a potential energy description.
