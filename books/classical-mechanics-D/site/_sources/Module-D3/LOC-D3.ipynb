# D3.0 Learning Outcomes — Polar Coordinates and Central Forces

## Why this module matters

Many physical systems exhibit **rotational symmetry** or are naturally described relative to a **central point**. In such cases, Cartesian coordinates are often inefficient because they do not align with the geometry of the motion.

**Polar coordinates** provide a more natural framework for describing motion in a plane when distances and angles are the key variables. By expressing position, velocity, and acceleration in terms of

$$
(r, \theta),
$$

we can reveal the structure of motion more clearly—especially in problems involving circular motion, orbital motion, and central forces.

In this module, we develop the expressions for **velocity and acceleration in polar coordinates**, and then apply **Newton’s second law** in this coordinate system. This allows us to analyze systems where forces depend on distance from a central point, setting the stage for later topics such as **orbital motion, effective potentials, and angular momentum**.

By the end of this module, you will be able to:

---

### A. Position in Polar Coordinates  
*Describes motion using radial and angular variables.*

- **A1.** Express the position vector in polar coordinates

$$
\mathbf{r} = r\,\hat{\mathbf{r}}.
$$

- **A2.** Interpret the meaning of the unit vectors  

$$
\hat{\mathbf{r}}, \quad \hat{\boldsymbol{\theta}}.
$$

- **A3.** Recognize that polar coordinates are adapted to problems with **radial symmetry**.

---

### B. Velocity in Polar Coordinates  
*Derives velocity using time-dependent unit vectors.*

- **B1.** Compute velocity as the total derivative of position  

$$
\mathbf{v} = \frac{d\mathbf{r}}{dt}.
$$

- **B2.** Apply the chain rule to account for the time dependence of both $r$ and $\theta$.  
- **B3.** Derive the velocity expression  

$$
\mathbf{v} = \dot{r}\,\hat{\mathbf{r}} + r\dot{\theta}\,\hat{\boldsymbol{\theta}}.
$$

- **B4.** Interpret radial and tangential components of velocity.

---

### C. Acceleration in Polar Coordinates  
*Develops the full acceleration expression including centripetal terms.*

- **C1.** Compute acceleration as  

$$
\mathbf{a} = \frac{d\mathbf{v}}{dt}.
$$

- **C2.** Account for the time dependence of the unit vectors  

$$
\hat{\mathbf{r}}, \quad \hat{\boldsymbol{\theta}}.
$$

- **C3.** Derive the acceleration expression  

$$
\mathbf{a} =
(\ddot{r} - r\dot{\theta}^2)\,\hat{\mathbf{r}}
+
(r\ddot{\theta} + 2\dot{r}\dot{\theta})\,\hat{\boldsymbol{\theta}}.
$$

- **C4.** Interpret the physical meaning of each term (radial, centripetal, and Coriolis-like contributions).

---

### D. Newton’s Second Law in Polar Coordinates  
*Applies dynamics in a coordinate system aligned with motion.*

- **D1.** Apply Newton’s second law component-wise  

$$
\mathbf{F} = m\mathbf{a}.
$$

- **D2.** Write the radial equation  

$$
F_r = m(\ddot{r} - r\dot{\theta}^2).
$$

- **D3.** Write the angular equation  

$$
F_\theta = m(r\ddot{\theta} + 2\dot{r}\dot{\theta}).
$$

- **D4.** Interpret how forces determine motion in radial and angular directions.

---

### E. Central Forces  
*Introduces forces that depend only on distance from a central point.*

- **E1.** Define a **central force** as a force of the form  

$$
\mathbf{F} = F(r)\,\hat{\mathbf{r}}.
$$

- **E2.** Recognize that central forces produce **no tangential force component**  

$$
F_\theta = 0.
$$

- **E3.** Use this condition to simplify the angular equation of motion.  
- **E4.** Identify implications for motion under central forces (e.g., constrained angular behavior).

---

### F. Symmetry and Coordinate Choice  
*Connects coordinate systems to symmetry of physical systems.*

- **F1.** Recognize that polar coordinates are natural for systems with **rotational symmetry**.  
- **F2.** Identify when a coordinate system simplifies the equations of motion.  
- **F3.** Connect symmetry of a system to simplifications in its mathematical description.

---

### Performance Criteria

- Position, velocity, and acceleration are correctly expressed in **polar coordinates**.  
- Unit vectors $\hat{\mathbf{r}}$ and $\hat{\boldsymbol{\theta}}$ are handled correctly, including their **time dependence**.  
- Newton’s second law is applied correctly in radial and angular components.  
- Central forces are correctly identified and used to simplify equations of motion.  
- Students recognize when **polar coordinates provide a more natural description** than Cartesian coordinates.
