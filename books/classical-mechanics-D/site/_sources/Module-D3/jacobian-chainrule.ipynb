## D3.4 Coordinate Transformations, Jacobians, and the Chain Rule  
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.4.1 Motivation: Why Return to Coordinate Transformations?

Earlier, we introduced generalized coordinates as a way to describe motion using variables that match the geometry of a system. In polar coordinates, for example, we describe motion using

$$
(r,\theta)
$$

instead of

$$
(x,y).
$$

The connection between these descriptions is a coordinate transformation:

$$
x = r\cos\theta, \qquad y = r\sin\theta.
$$

In this module, we take a more general view. Instead of treating polar coordinates as a special case, we ask:

> How do velocities and differentials transform when one set of coordinates depends on another?

This leads to the **Jacobian matrix**, the **chain rule**, and a compact formulation using **Einstein notation**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.4.2 General Coordinate Transformation

Let two coordinate systems describe the same point:

$$
(x_1,x_2,\dots,x_n), \qquad (q_1,q_2,\dots,q_n).
$$

A transformation relates them:

$$
x_i = x_i(q_1,q_2,\dots,q_n).
$$

---

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Definition — Coordinate Transformation</h3>

A **coordinate transformation** expresses one set of coordinates in terms of another:

$$
x_i = x_i(q_j).
$$

Both coordinate systems describe the same physical point.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.4.3 The Jacobian Matrix (Matrix Form)

The **Jacobian matrix** is defined as the matrix of partial derivatives:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Definition — Jacobian Matrix</h3>


$$
\mathbf{J} =
\begin{pmatrix}
\dfrac{\partial x_1}{\partial q_1} & \dfrac{\partial x_1}{\partial q_2} & \cdots \\
\dfrac{\partial x_2}{\partial q_1} & \dfrac{\partial x_2}{\partial q_2} & \cdots \\
\vdots & \vdots & \ddots
\end{pmatrix}.
$$

</div>

Each row corresponds to a Cartesian coordinate, and each column corresponds to a generalized coordinate.

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<h3 style="margin-top:0;">Box Activity — Jacobian Matrix for Polar Coordinates</h3>

<strong>Problem</strong>  

Given the transformation from polar to Cartesian coordinates:

$$
x = r\cos\theta, \qquad y = r\sin\theta,
$$

construct the Jacobian matrix

$$
\mathbf{J} =
\begin{pmatrix}
\dfrac{\partial x}{\partial r} & \dfrac{\partial x}{\partial \theta} \\
\dfrac{\partial y}{\partial r} & \dfrac{\partial y}{\partial \theta}
\end{pmatrix}.
$$

Then interpret the geometric meaning of each column.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Step 1 — Compute partial derivatives</strong>

For $x = r\cos\theta$:

$$
\frac{\partial x}{\partial r} = \cos\theta,
\qquad
\frac{\partial x}{\partial \theta} = -r\sin\theta
$$

For $y = r\sin\theta$:

$$
\frac{\partial y}{\partial r} = \sin\theta,
\qquad
\frac{\partial y}{\partial \theta} = r\cos\theta
$$

---

<strong>Step 2 — Construct the Jacobian matrix</strong>

$$
\mathbf{J} =
\begin{pmatrix}
\cos\theta & -r\sin\theta \\
\sin\theta & r\cos\theta
\end{pmatrix}.
$$

---

<strong>Step 3 — Interpret the columns</strong>

First column:

$$
\frac{\partial \vec x}{\partial r}
=
(\cos\theta,\sin\theta)
$$

→ This vector points in the **radial direction**.

Second column:

$$
\frac{\partial \vec x}{\partial \theta}
=
(-r\sin\theta, r\cos\theta)
$$

→ This vector points in the **angular direction**.

<strong>Conclusion</strong>  

The columns of the Jacobian are the coordinate basis vectors expressed in Cartesian components. This is why the Jacobian naturally appears in velocity transformations and in defining distances.

</div>
</details>

</div>

## D3.4.4 Index Notation (Einstein Notation)

The same object can be written using index notation:

$$
J_{ij} = \frac{\partial x_i}{\partial q_j}.
$$

Here:

- $i$ labels the row (which $x_i$)  
- $j$ labels the column (which $q_j$)

This is **not the Jacobian itself**, but the **component representation** of the Jacobian matrix.

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Curiosity — Matrix vs Index Notation</h3>

- $\mathbf{J}$ is the matrix  
- $J_{ij}$ are its components  

Index notation allows us to write matrix operations compactly without explicitly writing matrices.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.4.5 Differentials and the Chain Rule

A small change in the generalized coordinates produces a change in Cartesian coordinates:

$$
dx_i =
\frac{\partial x_i}{\partial q_1}dq_1
+
\frac{\partial x_i}{\partial q_2}dq_2
+\cdots
$$

In index notation:

$$
dx_i = J_{ij} dq_j.
$$

This is the chain rule written compactly.

---

## D3.4.6 Velocity Transformation

If $q_j = q_j(t)$, then

$$
\dot{x}_i = \frac{dx_i}{dt}.
$$

Applying the chain rule:

$$
\dot{x}_i =
\frac{\partial x_i}{\partial q_j}\dot{q}_j.
$$

Thus,

$$
\dot{x}_i = J_{ij}\dot{q}_j.
$$

---

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Definition — Velocity Transformation</h3>

$$
\dot{x}_i = J_{ij}\dot{q}_j
$$

This is the chain rule expressed in index notation.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D3.4.7 Example — Polar Velocity via Jacobian

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Velocity Transformation</h3>

$$
\begin{pmatrix}
\dot x \\
\dot y
\end{pmatrix}
=
\mathbf{J}
\begin{pmatrix}
\dot r \\
\dot\theta
\end{pmatrix}
$$

Using

$$
\mathbf{J} =
\begin{pmatrix}
\cos\theta & -r\sin\theta \\
\sin\theta & r\cos\theta
\end{pmatrix},
$$

we obtain

$$
\dot x = \dot r\cos\theta - r\dot\theta\sin\theta
$$

$$
\dot y = \dot r\sin\theta + r\dot\theta\cos\theta.
$$

</div>

---

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Velocity Transformation and Polar Velocity</h3>

<strong>Key Question</strong>  

How does the Jacobian result connect to the familiar polar velocity expression

$$
\vec v = \dot r\,\hat r + r\dot\theta\,\hat\theta \, ?
$$

---

<strong>Step 1 — Interpret the Jacobian columns</strong>

From the previous result,

$$
\mathbf{J} =
\begin{pmatrix}
\cos\theta & -r\sin\theta \\
\sin\theta & r\cos\theta
\end{pmatrix}.
$$

We can rewrite this as

$$
\mathbf{J} =
\begin{pmatrix}
\;|\; & \;|\; \\
\displaystyle \frac{\partial \vec x}{\partial r} & \displaystyle \frac{\partial \vec x}{\partial \theta} \\
\;|\; & \;|\;
\end{pmatrix}.
$$

Thus,

$$
\frac{\partial \vec x}{\partial r}
=
(\cos\theta,\sin\theta)
=
\hat r
$$

and

$$
\frac{\partial \vec x}{\partial \theta}
=
(-r\sin\theta, r\cos\theta)
=
r\,\hat\theta.
$$

---

<strong>Step 2 — Apply the velocity transformation</strong>

The Jacobian transformation gives

$$
\vec v =
\frac{\partial \vec x}{\partial r}\dot r
+
\frac{\partial \vec x}{\partial \theta}\dot\theta.
$$

Substitute the results above:

$$
\vec v =
\hat r\,\dot r + (r\hat\theta)\dot\theta.
$$

Thus,

$$
\boxed{
\vec v = \dot r\,\hat r + r\dot\theta\,\hat\theta
}
$$

---

<strong>Step 3 — Connection to Cartesian components</strong>

If we expand the unit vectors:

$$
\hat r = \cos\theta\,\hat i + \sin\theta\,\hat j,
$$

$$
\hat\theta = -\sin\theta\,\hat i + \cos\theta\,\hat j,
$$

then

$$
\vec v =
(\dot r\cos\theta - r\dot\theta\sin\theta)\hat i
+
(\dot r\sin\theta + r\dot\theta\cos\theta)\hat j,
$$

which matches exactly:

$$
\dot x, \dot y.
$$

</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:10px; border-radius:4px;">
<strong>Key Insight</strong>  

The Jacobian columns are not just derivatives—they are the **basis vectors of the coordinate system**:

$$
\frac{\partial \vec x}{\partial r} = \hat r,
\qquad
\frac{\partial \vec x}{\partial \theta} = r\hat\theta.
$$

So the Jacobian transformation

$$
\dot{x}_i = J_{ij}\dot q_j
$$

is exactly the polar velocity formula written in disguise.

</div>

</div>

## D3.4.8 Metric from the Jacobian 

### Why introduce the metric?

So far, we have learned how to transform **coordinates** and **velocities** using the Jacobian. But physics ultimately depends on measurable quantities such as:

- distance  
- speed  
- kinetic energy  

This raises a fundamental question:

> How do we measure distances and speeds when using generalized coordinates?

---

### Step 1 — Start from physical space

In Cartesian coordinates, the squared speed is

$$
v^2 = \dot{x}_i \dot{x}_i.
$$

This is not a definition—it comes from the geometry of space (dot product).

---

### Step 2 — Express velocity using the Jacobian

From the chain rule:

$$
\dot{x}_i = J_{ij}\dot{q}_j.
$$

Substitute:

$$
v^2 =
(J_{ij}\dot q_j)(J_{ik}\dot q_k).
$$

---

### Step 3 — Rearranging the expression

Group terms:

$$
v^2 =
(J_{ij}J_{ik})\dot q_j \dot q_k.
$$

This suggests defining a new object:

$$
g_{jk} = J_{ij}J_{ik}.
$$

Thus,

$$
\boxed{
v^2 = g_{jk}\dot q_j \dot q_k
}
$$

---

### Step 4 — What is the metric?

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Definition — Metric</h3>

$$
g_{jk} =
\frac{\partial x_i}{\partial q_j}
\frac{\partial x_i}{\partial q_k}
$$

The **metric** determines how distances and speeds are measured in generalized coordinates.

</div>

---

### Step 5 — Where does distance come in?

A small displacement in Cartesian coordinates is

$$
dx_i = \frac{\partial x_i}{\partial q_j} dq_j.
$$

The squared distance is defined by the Euclidean metric:

$$
ds^2 = dx_i\,dx_i.
$$

Substitute the transformation:

$$
ds^2 =
\left(\frac{\partial x_i}{\partial q_j} dq_j\right)
\left(\frac{\partial x_i}{\partial q_k} dq_k\right).
$$

Rearranging:

$$
ds^2 =
\left(
\frac{\partial x_i}{\partial q_j}
\frac{\partial x_i}{\partial q_k}
\right)
dq_j dq_k.
$$

Define

$$
g_{jk} =
\frac{\partial x_i}{\partial q_j}
\frac{\partial x_i}{\partial q_k}.
$$

Thus,

$$
\boxed{
ds^2 = g_{jk}\,dq_j dq_k
}
$$

---

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<h3 style="margin-top:0;">Box Activity — Metric in Polar Coordinates</h3>

<strong>Problem</strong>

Using the transformation

$$
x = r\cos\theta, \qquad y = r\sin\theta,
$$

compute the metric components

$$
g_{jk} =
\frac{\partial x_i}{\partial q_j}
\frac{\partial x_i}{\partial q_k},
$$

where

$$
q_1 = r, \qquad q_2 = \theta.
$$

Then:

1. Derive the expression for $ds^2$
2. Derive the expression for $v^2$
3. Interpret the physical meaning of the $r^2$ term

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Step 1 — Compute partial derivatives</strong>

$$
\frac{\partial x}{\partial r} = \cos\theta,
\qquad
\frac{\partial x}{\partial \theta} = -r\sin\theta
$$

$$
\frac{\partial y}{\partial r} = \sin\theta,
\qquad
\frac{\partial y}{\partial \theta} = r\cos\theta
$$

---

<strong>Step 2 — Compute metric components</strong>

$$
g_{rr} =
(\cos\theta)^2 + (\sin\theta)^2 = 1
$$

$$
g_{r\theta} =
\cos\theta(-r\sin\theta) + \sin\theta(r\cos\theta) = 0
$$

$$
g_{\theta\theta} =
r^2\sin^2\theta + r^2\cos^2\theta = r^2
$$

---

<strong>Step 3 — Distance formula</strong>

$$
ds^2 = g_{jk} dq_j dq_k
$$

$$
\boxed{
ds^2 = dr^2 + r^2 d\theta^2
}
$$

---

<strong>Step 4 — Speed formula</strong>

$$
v^2 = g_{jk}\dot q_j \dot q_k
$$

$$
\boxed{
v^2 = \dot r^2 + r^2 \dot\theta^2
}
$$

---

<strong>Step 5 — Interpretation</strong>

The factor $r^2$ appears because a change in angle corresponds to an arc length:

$$
ds = rd\theta.
$$

Thus:

> Angular motion translates into physical distance only after scaling by $r$

This is why the metric is essential—it converts coordinate changes into real distances.

</div>
</details>

</div>


---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:10px; border-radius:4px;">
<strong>Key Insight</strong>  

The Jacobian tells us **how coordinates map into space**, but the metric tells us **how to measure geometry in those coordinates**.

- Jacobian → directions (how coordinates change position)  
- Metric → distances (how we measure length and speed)  

This is why the metric is essential: without it, coordinates alone have no physical meaning.

</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffc20e; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Do We Really Need the Jacobian?</h3>

In this module, we introduced the Jacobian matrix as a way to transform velocities and derive the metric. However, strictly speaking, the Jacobian is **not necessary**.

Everything we did follows directly from the chain rule written in index notation:

$$
\dot{x}_i = \frac{\partial x_i}{\partial q_j}\dot{q}_j.
$$

From this single relation, we can derive:

- velocity transformations  
- expressions for $v^2$  
- the metric $g_{jk}$  

without ever assembling a matrix.

---

So why introduce the Jacobian?

Most students are already familiar with matrices from Calculus III and/or linear algebra, so writing

$$
\dot{x}_i = J_{ij}\dot{q}_j
$$

or in matrix form

$$
\dot{\mathbf{x}} = \mathbf{J}\,\dot{\mathbf{q}}
$$

provides a **clear and structured way** to organize the transformation.

---

<strong>Key Insight</strong>  

> The Jacobian is not new physics—it is simply a convenient way of organizing the chain rule.

---

In more advanced settings (such as differential geometry and relativity), the **index notation becomes primary**, and the Jacobian is understood as a representation of the transformation rather than the fundamental object.

So:

- Chain rule → fundamental  
- Jacobian → organizational tool  

Both describe the same underlying idea.

</div>

## D3.4.9 Summary

- The Jacobian is a **matrix** describing how coordinates transform  
- Its components are written as $J_{ij}$ in index notation  
- The chain rule gives velocity transformation  
- The metric arises naturally from the Jacobian  

> Coordinates are labels, but the Jacobian tells us how those labels connect to physical space.
