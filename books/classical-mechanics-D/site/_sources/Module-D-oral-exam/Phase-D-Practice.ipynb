
# Phase D — Preparing for the Oral Exam: Learn to Explain Physics

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why an Oral Exam Is Different

In a written exam, you may be able to solve a problem by recognizing a familiar equation, substituting numbers, and following a procedure you have practiced.

In an oral exam, that is not enough.

You must demonstrate that you **understand the physics behind your mathematical work**. You should be able to explain:

- **Concepts:** What physical principles govern the system?
- **Strategy:** Why did you choose this particular method?
- **Equations:** What do the terms mean, and why does the equation apply?
- **Mathematics:** How do you move from one step to the next?
- **Interpretation:** What does your result tell you about the physical system?

You are not expected to give a perfectly polished lecture. You are expected to think, reason, explain, and demonstrate genuine understanding.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">

<strong>The Fundamental Difference</strong>

Being able to recognize a solution when you see it is not the same as being able to construct and explain that solution yourself.

**Your goal is not simply to know how to solve a problem. Your goal is to understand it well enough that you can teach someone else how and why the solution works.**

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## The Three-Explanation Study Method

One of the most effective ways to prepare for an oral exam is to work with a small study group of two to four students.

Choose a problem, and take turns explaining it to one another.

But do not explain it just once. Explain it **at least three times**.

### First Explanation — Discover What You Do Not Know

Solve the problem and explain your reasoning aloud, without reading from a completed solution.

Your group members should interrupt with questions such as:

- Why did you choose that equation?
- What does that term physically represent?
- Why is that quantity zero?
- What assumption are you making?
- Could you solve this using another method?
- How do you know your answer makes physical sense?

You will probably discover gaps in your understanding. **That is exactly what you want during preparation.**

### Second Explanation — Build Understanding

Review the concepts and mathematical steps that caused difficulty.

Now explain the problem again, preferably without looking at your previous work.

This time, concentrate on explaining *why* each step follows from the previous one.

Your explanation should become clearer, more structured, and more physically meaningful.

### Third Explanation — Demonstrate Mastery

Explain the entire problem again, starting with a blank sheet of paper or whiteboard.

Your goal is to communicate the solution as though you were teaching it to another student.

Try to anticipate questions before your group members ask them.

You should now be able to explain not only what you are doing but also why you are doing it.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Why Three Times?</strong>

The first explanation reveals what you do not understand.

The second helps you repair and connect your understanding.

The third tests whether you can retrieve, organize, and communicate that understanding independently.

You may discover something new every time you explain the same problem. That is not wasted repetition. It is part of learning physics.

If the third explanation still reveals important gaps, repeat the process.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## How to Be an Effective Study Partner

When someone else is explaining a problem, do not simply listen and agree.

**Your job is to help reveal and improve their understanding.**

Ask questions. Challenge unclear statements. Request physical interpretations. Ask what would happen if a condition changed.

Do not immediately provide the correct answer when someone struggles. Give them time to reason through the difficulty.

Then switch roles.

Everyone should have opportunities to be both the person explaining and the person asking questions.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## A Strategy for Explaining Any Physics Problem

Practice organizing your oral explanation around five stages:

1. **Physical model:** Describe the system, assumptions, and relevant physical principles.
2. **Strategy:** Explain which framework you will use and why it is appropriate.
3. **Mathematical formulation:** Write the governing equations and explain their terms.
4. **Execution:** Carry out the mathematics while explaining the important steps.
5. **Interpretation:** Check units, signs, limiting cases, and physical meaning.

<div style="background-color:#ffe6e6; border-left:6px solid #cc0000; padding:14px; border-radius:4px;">

<strong>Danger — The Illusion of Understanding</strong>

Reading a worked solution and thinking "That makes sense" is not reliable evidence that you can reproduce or explain it independently.

Likewise, memorizing a sequence of mathematical steps is not the same as understanding the underlying physics.

**Close the solution. Start with a blank page. Explain it aloud.**

If you cannot explain a step, you have identified something worth studying.

</div>

## D1 — Impulse, Momentum, and Momentum Methods

### Comprehensive Problem — Two Particles and an External Impulse

Two particles, $A$ and $B$, move along a frictionless horizontal track.

Their masses are

$$
m_A=2.0~\mathrm{kg},
\qquad
m_B=3.0~\mathrm{kg}.
$$

Initially, their velocities are

$$
v_{A,i}=4.0~\mathrm{m/s},
\qquad
v_{B,i}=-2.0~\mathrm{m/s}.
$$

Particle $A$ experiences an external force

$$
F_{\mathrm{ext}}(t)=6t,
\qquad 0\leq t\leq2.0~\mathrm{s},
$$

directed along the positive $x$-axis.

During this interval, the two particles may also exert forces on one another, but these internal forces are unknown.

**Part A — Momentum and Impulse**

1. Calculate the initial total momentum of the two-particle system.
2. Calculate the net external impulse over the time interval.
3. Determine the final total momentum of the system.

**Part B — Center-of-Mass Motion**

4. Determine the initial and final velocities of the center of mass.
5. Explain why the unknown internal forces are not needed to determine the final center-of-mass velocity.

**Part C — Newton's Second Law**

6. Starting from Newton's second law in momentum form, derive the impulse–momentum theorem.
7. Explain the physical difference between the instantaneous and accumulated descriptions of motion.

**Part D — Conceptual Understanding**

8. Could you determine the final velocity of each particle individually from the information given? Explain.
9. If the external force were removed, what quantity would remain constant? Would the individual particle velocities necessarily remain constant?
10. Explain why impulse is a vector quantity and how this differs from work.

Detailed solution and conceptual explanations
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Part A — Momentum and Impulse**

The total momentum is the vector sum of the individual momenta:

$$
P=m_Av_A+m_Bv_B.
$$

Initially,

$$
P_i=(2)(4)+(3)(-2)=2~\mathrm{kg\,m/s}.
$$

Therefore,

$$
\boxed{P_i=2~\mathrm{kg\,m/s}}.
$$

The net external impulse is

$$
J_{\mathrm{ext}}=\int_0^2 6t\,dt.
$$

Evaluating,

$$
J_{\mathrm{ext}}
=\left[3t^2\right]_0^2
=12~\mathrm{N\,s}.
$$

The impulse–momentum theorem gives

$$
J_{\mathrm{ext}}=P_f-P_i.
$$

Therefore,

$$
P_f=2+12.
$$

$$
\boxed{P_f=14~\mathrm{kg\,m/s}}.
$$

**Part B — Center-of-Mass Motion**

The total mass is

$$
M=m_A+m_B=5.0~\mathrm{kg}.
$$

Since

$$
P=Mv_{\mathrm{CM}},
$$

the initial center-of-mass velocity is

$$
v_{\mathrm{CM},i}=\frac{2}{5}
=\boxed{0.40~\mathrm{m/s}}.
$$

The final velocity is

$$
v_{\mathrm{CM},f}=\frac{14}{5}
=\boxed{2.8~\mathrm{m/s}}.
$$

The internal forces cancel when calculating the rate of change of total momentum, assuming the internal interactions obey Newton's third law.

Thus,

$$
\boxed{\mathbf F_{\mathrm{ext}}=\frac{d\mathbf P}{dt}}.
$$

The center-of-mass motion is determined by the net external force.

**Part C — Local and Global Descriptions**

Newton's second law in momentum form is

$$
\mathbf F_{\mathrm{net}}=\frac{d\mathbf p}{dt}.
$$

Multiplying by $dt$ and integrating,

$$
\int_{t_i}^{t_f}\mathbf F_{\mathrm{net}}\,dt
=
\int_{\mathbf p_i}^{\mathbf p_f}d\mathbf p.
$$

Therefore,

$$
\boxed{\mathbf J_{\mathrm{net}}=\Delta\mathbf p}.
$$

Newton's second law describes the instantaneous rate of momentum change.

The impulse–momentum theorem describes the accumulated change over a finite time interval.

They are not competing laws; they are differential and integral forms of the same relationship.

**Part D — Conceptual Understanding**

**Can we determine both final particle velocities?**

No. We know only that

$$
m_Av_{A,f}+m_Bv_{B,f}=14.
$$

This is one equation with two unknown final velocities. Additional information about their interaction would be required.

**What if the external force were zero?**

Then

$$
\frac{dP}{dt}=0,
$$

so

$$
\boxed{P=\text{constant}}.
$$

Individual momenta may still change because of internal forces, but the total momentum remains constant.

**Why is impulse a vector?**

Impulse is the time integral of force:

$$
\mathbf J=\int\mathbf F\,dt.
$$

Because force has direction, impulse also has direction.

Work is instead a scalar:

$$
W=\int\mathbf F\cdot d\mathbf r.
$$

The dot product produces a scalar, while impulse remains a vector.

**Key oral-exam explanation:** The net external impulse changes the total momentum, even when the internal dynamics are unknown.

## D2 — Work, Energy, and Conservative Forces

### Comprehensive Problem — Work Along a Path in a Two-Dimensional Force Field

A particle of mass $m=2.0~\mathrm{kg}$ moves in the $xy$-plane under the force field

$$
\mathbf F(x,y)
=
(2xy+y^2)\hat{\mathbf i}
+
(x^2+2xy+4y)\hat{\mathbf j}.
$$

The particle moves from $A=(1,2)$ to $B=(3,4)$ along a straight-line path. Its initial speed is $v_A=2.0~\mathrm{m/s}$.

Assume this is the only force doing work.

**Part A — Work Along a Path**

1. Parameterize the straight-line path from $A$ to $B$ using $0\leq t\leq1$. Explain how you constructed the parameterization.
2. Determine the differential displacement $d\mathbf r$.
3. Set up and evaluate the work integral

$$
W_{A\rightarrow B}=\int_C\mathbf F\cdot d\mathbf r.
$$

**Part B — Conservative Forces**

4. Determine whether the force field is conservative using the local criterion.
5. Explain the relationship between the local condition and the work around a closed path.

**Part C — Potential Energy**

6. Determine a potential energy function $U(x,y)$ such that

$$
\mathbf F=-\nabla U.
$$

7. Choose $U(1,2)=0$ and use the potential to verify the work calculated in Part A.

**Part D — Energy and Interpretation**

8. Use the work–energy theorem to determine the particle's final speed.
9. Explain why the work does not depend on the path taken between $A$ and $B$.
10. Explain why adding a constant to $U$ does not change the motion.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed solution and conceptual explanations
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Part A — Parameterization and Work**

A straight line connecting two position vectors can be parameterized as

$$
\mathbf r(t)=\mathbf r_A+t(\mathbf r_B-\mathbf r_A),
\qquad 0\leq t\leq1.
$$

The displacement from $A$ to $B$ is

$$
\mathbf r_B-\mathbf r_A
=
(3-1)\hat{\mathbf i}+(4-2)\hat{\mathbf j}
=
2\hat{\mathbf i}+2\hat{\mathbf j}.
$$

Starting at $A$ and adding a fraction $t$ of this displacement,

$$
\boxed{
\mathbf r(t)
=
(1+2t)\hat{\mathbf i}+(2+2t)\hat{\mathbf j}
}.
$$

At $t=0$, the position is $A=(1,2)$, and at $t=1$, it is $B=(3,4)$.

Thus,

$$
x=1+2t,\qquad y=2+2t.
$$

Differentiating,

$$
\boxed{
d\mathbf r=(2\hat{\mathbf i}+2\hat{\mathbf j})\,dt
}.
$$

Substitute the parameterization into the force field:

$$
F_x=2(1+2t)(2+2t)+(2+2t)^2.
$$

Expanding,

$$
F_x=8+20t+12t^2.
$$

Similarly,

$$
F_y=(1+2t)^2+2(1+2t)(2+2t)+4(2+2t).
$$

Expanding,

$$
F_y=13+24t+12t^2.
$$

The work is

$$
W=\int_0^1\mathbf F(t)\cdot\frac{d\mathbf r}{dt}\,dt.
$$

Therefore,

$$
W=\int_0^1[2F_x+2F_y]\,dt.
$$

Substituting,

$$
W=\int_0^1(42+88t+48t^2)\,dt.
$$

Evaluating,

$$
W=\left[42t+44t^2+16t^3\right]_0^1.
$$

Thus,

$$
\boxed{W_{A\rightarrow B}=102~\mathrm{J}}.
$$

**Part B — Conservative-Force Test**

For a continuously differentiable force field in a simply connected region, the local conservative-force criterion in two dimensions is

$$
\frac{\partial F_y}{\partial x}
-
\frac{\partial F_x}{\partial y}=0.
$$

Here,

$$
\frac{\partial F_y}{\partial x}=2x+2y,
$$

and

$$
\frac{\partial F_x}{\partial y}=2x+2y.
$$

Therefore,

$$
\boxed{\nabla\times\mathbf F=\mathbf 0}.
$$

Since the field is defined throughout the plane, it is conservative.

For a conservative force, the work around any closed path is zero:

$$
\boxed{\oint_C\mathbf F\cdot d\mathbf r=0}.
$$

This is the global condition corresponding to path independence.

**Part C — Potential Energy**

A conservative force can be expressed as

$$
\mathbf F=-\nabla U.
$$

Starting with the $x$ component,

$$
\frac{\partial U}{\partial x}=-(2xy+y^2).
$$

Integrating with respect to $x$,

$$
U(x,y)=-x^2y-xy^2+g(y).
$$

The function $g(y)$ appears because differentiation with respect to $x$ eliminates any function depending only on $y$.

Now use the $y$ component:

$$
F_y=-\frac{\partial U}{\partial y}.
$$

Since

$$
\frac{\partial U}{\partial y}
=
-x^2-2xy+g'(y),
$$

we obtain

$$
x^2+2xy+4y
=
x^2+2xy-g'(y).
$$

Therefore,

$$
g'(y)=-4y,
$$

so

$$
g(y)=-2y^2+C.
$$

The potential energy is

$$
\boxed{
U(x,y)=-x^2y-xy^2-2y^2+C
}.
$$

We choose $U(1,2)=0$.

Substituting,

$$
0=-(1)^2(2)-(1)(2)^2-2(2)^2+C.
$$

Therefore,

$$
0=-2-4-8+C,
$$

giving

$$
C=14.
$$

Thus,

$$
\boxed{
U(x,y)=-x^2y-xy^2-2y^2+14
}.
$$

At $A$,

$$
U_A=0.
$$

At $B=(3,4)$,

$$
U_B=-(3)^2(4)-(3)(4)^2-2(4)^2+14.
$$

Therefore,

$$
U_B=-36-48-32+14=-102~\mathrm{J}.
$$

Using

$$
W_{A\rightarrow B}=-(U_B-U_A),
$$

we obtain

$$
\boxed{W_{A\rightarrow B}=102~\mathrm{J}},
$$

in agreement with the line integral.

**Part D — Work–Energy Theorem**

The initial kinetic energy is

$$
K_A=\frac12mv_A^2.
$$

Substituting,

$$
K_A=\frac12(2)(2)^2=4~\mathrm{J}.
$$

The work–energy theorem gives

$$
W_{\mathrm{net}}=K_B-K_A.
$$

Thus,

$$
K_B=4+102=106~\mathrm{J}.
$$

Since

$$
K_B=\frac12mv_B^2,
$$

we have

$$
106=\frac12(2)v_B^2.
$$

Therefore,

$$
\boxed{
v_B=\sqrt{106}\approx10.30~\mathrm{m/s}
}.
$$

**Conceptual Interpretation**

For a conservative force, the work depends only on the initial and final positions:

$$
W_{A\rightarrow B}=U_A-U_B.
$$

This means that different paths connecting the same two points produce the same work, even if the force varies along those paths.

Adding a constant to the potential energy does not change the force because

$$
-\nabla(U+C)=-\nabla U.
$$

Likewise, the potential energy difference is unchanged:

$$
(U_B+C)-(U_A+C)=U_B-U_A.
$$

Therefore, the motion is unaffected by the choice of potential energy zero.

**Key oral-exam explanation:** Work is accumulated along a path through a line integral. For a conservative force, the same work can be obtained from the potential energy difference between the endpoints, independently of the path.

## D4 — Polar Coordinates and Central Forces

### Comprehensive Problem — Motion Under a Central Force

A particle of mass $m$ moves in the $xy$-plane under an attractive central force

$$
\mathbf F(r)=-\frac{k}{r^2}\hat{\mathbf r},
$$

where $k>0$.

The position is described using polar coordinates $(r,\theta)$.

**Part A — Polar Kinematics**

1. Write the position vector in polar coordinates.
2. Explain why the polar unit vectors change with time even though their magnitudes remain constant.
3. Derive the velocity and acceleration vectors in polar coordinates.

**Part B — Newton's Second Law**

4. Apply Newton's second law separately in the radial and angular directions.
5. Explain the physical meaning of each term in the acceleration expression.

**Part C — Angular Momentum**

6. Starting from the angular equation of motion, show that

$$
mr^2\dot\theta=\text{constant}.
$$

7. Explain why angular momentum is conserved for a central force.

**Part D — Circular Motion**

8. Suppose the particle moves in a circular orbit of radius $R$. Determine the angular speed $\omega$.
9. Explain why the particle accelerates even though its speed is constant.
10. If the particle instead moves inward while conserving angular momentum, what happens to its angular speed?

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed solution and conceptual explanations
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Part A — Polar Kinematics**

The position vector is

$$
\boxed{\mathbf r=r\hat{\mathbf r}}.
$$

The polar unit vectors are

$$
\hat{\mathbf r}
=
\cos\theta\,\hat{\mathbf i}
+
\sin\theta\,\hat{\mathbf j},
$$

$$
\hat{\boldsymbol\theta}
=
-\sin\theta\,\hat{\mathbf i}
+
\cos\theta\,\hat{\mathbf j}.
$$

Differentiating with respect to time gives

$$
\boxed{
\frac{d\hat{\mathbf r}}{dt}
=
\dot\theta\hat{\boldsymbol\theta}
}
$$

and

$$
\boxed{
\frac{d\hat{\boldsymbol\theta}}{dt}
=
-\dot\theta\hat{\mathbf r}
}.
$$

These derivatives are nonzero because the directions of the unit vectors change as $\theta$ changes.

Differentiating position,

$$
\mathbf v=\frac{d}{dt}(r\hat{\mathbf r}),
$$

and applying the product rule gives

$$
\boxed{
\mathbf v=
\dot r\hat{\mathbf r}
+
r\dot\theta\hat{\boldsymbol\theta}
}.
$$

Differentiating again,

$$
\mathbf a
=
\frac{d}{dt}
\left(
\dot r\hat{\mathbf r}
+
r\dot\theta\hat{\boldsymbol\theta}
\right).
$$

Expanding,

$$
\mathbf a=
\ddot r\hat{\mathbf r}
+
\dot r\dot\theta\hat{\boldsymbol\theta}
+
(\dot r\dot\theta+r\ddot\theta)\hat{\boldsymbol\theta}
-
r\dot\theta^2\hat{\mathbf r}.
$$

Collecting components,

$$
\boxed{
\mathbf a=
(\ddot r-r\dot\theta^2)\hat{\mathbf r}
+
(r\ddot\theta+2\dot r\dot\theta)\hat{\boldsymbol\theta}
}.
$$

**Part B — Newton's Second Law**

The central force has only a radial component:

$$
F_r=-\frac{k}{r^2},
\qquad F_\theta=0.
$$

The radial equation is

$$
\boxed{
m(\ddot r-r\dot\theta^2)
=
-\frac{k}{r^2}
}.
$$

The angular equation is

$$
\boxed{
m(r\ddot\theta+2\dot r\dot\theta)=0
}.
$$

The terms in the acceleration have different physical meanings:

- $\ddot r$: radial acceleration associated with changing radial motion.
- $-r\dot\theta^2$: inward centripetal contribution.
- $r\ddot\theta$: tangential acceleration associated with changing angular speed.
- $2\dot r\dot\theta$: contribution arising when radial motion and angular motion occur simultaneously.

The last term arises from the changing polar basis and is sometimes called a Coriolis-like term.

**Part C — Angular Momentum Conservation**

From the angular equation,

$$
r\ddot\theta+2\dot r\dot\theta=0.
$$

Multiply by $mr$:

$$
mr^2\ddot\theta+2mr\dot r\dot\theta=0.
$$

Recognize that

$$
\frac{d}{dt}(mr^2\dot\theta)
=
mr^2\ddot\theta+2mr\dot r\dot\theta.
$$

Therefore,

$$
\boxed{
\frac{d}{dt}(mr^2\dot\theta)=0
}.
$$

Hence,

$$
\boxed{
L_z=mr^2\dot\theta=\text{constant}
}.
$$

A central force points parallel or antiparallel to the position vector, so

$$
\boldsymbol\tau=\mathbf r\times\mathbf F=\mathbf 0.
$$

Since

$$
\frac{d\mathbf L}{dt}=\boldsymbol\tau,
$$

angular momentum is conserved.

**Part D — Circular Motion**

For a circular orbit,

$$
r=R,\qquad \dot r=0,\qquad\ddot r=0,
$$

and

$$
\dot\theta=\omega=\text{constant}.
$$

The radial equation becomes

$$
-mR\omega^2=-\frac{k}{R^2}.
$$

Therefore,

$$
\boxed{
\omega=\sqrt{\frac{k}{mR^3}}
}.
$$

The particle accelerates because its velocity direction continually changes, even when its speed is constant.

Finally, angular momentum conservation gives

$$
mr^2\dot\theta=L_z.
$$

Thus,

$$
\dot\theta=\frac{L_z}{mr^2}.
$$

For fixed nonzero angular momentum, the magnitude of the angular speed increases as the radius decreases.

**Key oral-exam explanation:** Polar coordinates simplify central-force problems because the coordinate directions naturally align with the geometry of the force and motion.

## D6 — Small Oscillations and Stability

### Comprehensive Problem — Particle in a Periodic Potential

A particle of mass $m$ moves along the $x$-axis under the conservative potential

$$
U(x)=U_0\left[1-\cos\left(\frac{x}{a}\right)\right],
$$

where $U_0>0$ is a constant with units of energy and $a>0$ is a characteristic length.

The particle is initially located near $x=0$.

**Part A — Equilibrium and Stability**

1. Determine the force $F(x)$ acting on the particle.
2. Find all equilibrium positions.
3. Classify the equilibrium points as stable or unstable.
4. Sketch the potential energy function and identify its important features.

**Part B — Linearization and Small Oscillations**

5. Expand the potential about the stable equilibrium at $x=0$ using a Taylor series through second order.
6. Show that the equation of motion reduces to that of a simple harmonic oscillator for sufficiently small displacements.
7. Determine the angular frequency and period of small oscillations.

**Part C — Energy and Turning Points**

8. Suppose the total mechanical energy satisfies

$$
0<E<2U_0.
$$

Determine the turning points for motion about $x=0$.

9. Starting from conservation of energy, derive an integral expression for the oscillation period.
10. Explain why the period approaches the simple harmonic oscillator period as the amplitude becomes small.

**Part D — Phase Space and Conceptual Understanding**

11. Sketch or describe the phase-space trajectory for small oscillations about $x=0$.
12. Explain how the phase-space trajectory changes as the energy increases toward $2U_0$.
13. What happens to the motion when $E>2U_0$?
14. Explain why many nonlinear systems behave approximately like harmonic oscillators near stable equilibria.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed solution and conceptual explanations
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Part A — Equilibrium and Stability**

The force is related to the potential energy by

$$
F(x)=-\frac{dU}{dx}.
$$

Differentiating,

$$
\boxed{
F(x)=-\frac{U_0}{a}\sin\left(\frac{x}{a}\right)
}.
$$

Equilibrium occurs where

$$
F(x)=0.
$$

Therefore,

$$
\sin\left(\frac{x}{a}\right)=0,
$$

giving

$$
\boxed{x_n=n\pi a,\qquad n=0,\pm1,\pm2,\ldots}.
$$

To classify stability, calculate the second derivative:

$$
U''(x)=\frac{U_0}{a^2}\cos\left(\frac{x}{a}\right).
$$

At equilibrium,

$$
U''(x_n)=\frac{U_0}{a^2}(-1)^n.
$$

For even $n$,

$$
U''(x_n)>0,
$$

so these are stable equilibria:

$$
\boxed{x=2n\pi a}.
$$

For odd multiples of $\pi a$,

$$
U''(x_n)<0,
$$

so these are unstable equilibria:

$$
\boxed{x=(2n+1)\pi a}.
$$

The potential is periodic, with minima at $U=0$ and maxima at $U=2U_0$.

**Part B — Linearization**

Near $x=0$, use the Taylor expansion

$$
\cos\left(\frac{x}{a}\right)
\approx
1-\frac{x^2}{2a^2}.
$$

Substituting,

$$
U(x)\approx
U_0\left[1-\left(1-\frac{x^2}{2a^2}\right)\right].
$$

Thus,

$$
\boxed{
U(x)\approx\frac{U_0}{2a^2}x^2
}.
$$

This has the form of a harmonic oscillator potential,

$$
U_{\mathrm{SHM}}=\frac12k_{\mathrm{eff}}x^2,
$$

where

$$
\boxed{k_{\mathrm{eff}}=\frac{U_0}{a^2}}.
$$

Newton's second law becomes

$$
m\ddot x=-\frac{U_0}{a^2}x.
$$

Therefore,

$$
\boxed{
\ddot x+\frac{U_0}{ma^2}x=0
}.
$$

Comparing with

$$
\ddot x+\omega_0^2x=0,
$$

we obtain

$$
\boxed{
\omega_0=\sqrt{\frac{U_0}{ma^2}}
}.
$$

The small-oscillation period is

$$
\boxed{
T_0=2\pi\sqrt{\frac{ma^2}{U_0}}
}.
$$

**Part C — Energy and Turning Points**

The total energy is

$$
E=\frac12m\dot x^2+U(x).
$$

At a turning point, the velocity is zero:

$$
E=U(x).
$$

Therefore,

$$
E=U_0\left[1-\cos\left(\frac{x}{a}\right)\right].
$$

Rearranging,

$$
\cos\left(\frac{x}{a}\right)=1-\frac{E}{U_0}.
$$

For oscillations around $x=0$, the turning points are

$$
\boxed{
x_{\pm}=\pm a\cos^{-1}\left(1-\frac{E}{U_0}\right)
}.
$$

These exist within the central potential well for

$$
0<E<2U_0.
$$

From energy conservation,

$$
\frac12m\dot x^2=E-U(x).
$$

Thus,

$$
\dot x=\pm\sqrt{\frac{2}{m}[E-U(x)]}.
$$

Since

$$
dt=\frac{dx}{\dot x},
$$

the full oscillation period is

$$
T=2\int_{x_-}^{x_+}
\frac{dx}{\sqrt{\frac{2}{m}[E-U(x)]}}.
$$

Therefore,

$$
\boxed{
T=
2\sqrt{\frac{m}{2}}
\int_{x_-}^{x_+}
\frac{dx}{
\sqrt{E-U_0[1-\cos(x/a)]}
}
}.
$$

For small amplitudes, the potential becomes approximately quadratic:

$$
U(x)\approx\frac12k_{\mathrm{eff}}x^2.
$$

Therefore, the period approaches

$$
\boxed{T\rightarrow T_0}
$$

as the amplitude approaches zero.

**Part D — Phase Space and Interpretation**

The phase space is described by the variables

$$
(x,v).
$$

For small oscillations about $x=0$, the energy equation becomes

$$
\frac12mv^2+\frac12k_{\mathrm{eff}}x^2=E.
$$

This is an ellipse in phase space.

As the energy increases, the trajectories become increasingly distorted from ellipses because the potential is nonlinear.

As

$$
E\rightarrow2U_0^-,
$$

the particle approaches the unstable equilibrium positions at $x=\pm\pi a$.

The motion becomes increasingly slow near these positions, and the oscillation period grows without bound.

At the critical energy $E=2U_0$, the phase-space trajectory forms a separatrix between oscillatory and unbounded motion.

For

$$
E>2U_0,
$$

the particle can cross the potential barriers and move through successive potential wells.

The motion is no longer confined to a single well. Its velocity remains either positive or negative because the kinetic energy never reaches zero.

**Why does linearization work?**

Near a smooth stable equilibrium $x_0$,

$$
U'(x_0)=0,
\qquad U''(x_0)>0.
$$

The Taylor expansion gives

$$
U(x)\approx U(x_0)+\frac12U''(x_0)(x-x_0)^2.
$$

Thus, the restoring force becomes approximately linear:

$$
F(x)\approx-U''(x_0)(x-x_0).
$$

This is mathematically equivalent to Hooke's law.

Therefore, many nonlinear systems exhibit approximately simple harmonic motion when displaced only slightly from a stable equilibrium.

**Key oral-exam explanation:** The harmonic oscillator is not restricted to springs. It emerges naturally as the local approximation to many nonlinear systems near stable equilibrium.

## D8 — Constraints and Generalized Coordinates

### Comprehensive Problem — A Pendulum with a Moving Support

A particle of mass $m$ is attached to a massless rigid rod of length $\ell$. The rod is connected to a pivot that moves horizontally according to

$$
x_p(t)=vt,
$$

where $v$ is constant.

The pendulum swings in a vertical plane. Choose the $x$-axis horizontally to the right and the $y$-axis vertically upward.

Let $\theta$ be the angle the rod makes with the downward vertical.

**Part A — Constraints and Degrees of Freedom**

1. Write the geometric constraint equation in Cartesian coordinates.
2. Determine the number of degrees of freedom.
3. Is the constraint holonomic? Is it time-dependent? Explain.

**Part B — Generalized Coordinates**

4. Choose a suitable generalized coordinate.
5. Express the Cartesian coordinates $x$ and $y$ in terms of the generalized coordinate and time.
6. Use the chain rule to derive the Cartesian velocity components.

**Part C — Kinetic Energy and Configuration Space**

7. Determine the speed squared and the kinetic energy.
8. Explain why the generalized velocity $\dot\theta$ is not itself the Cartesian speed.
9. Describe the configuration space of the pendulum.

**Part D — Invariance and Interpretation**

10. Suppose we introduce a new horizontal coordinate $x'=x-vt$. What does this coordinate represent, and how does it simplify the position equations?
11. Does this transformation change the physical motion?
12. Explain why generalized coordinates are useful for describing constrained systems.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed solution and conceptual explanations
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Part A — Constraint and Degrees of Freedom**

The particle must remain a fixed distance $\ell$ from the moving pivot.

Therefore,

$$
\boxed{(x-vt)^2+y^2=\ell^2}.
$$

The particle moves in a plane and would ordinarily require two independent Cartesian coordinates.

The rigid rod introduces one independent constraint:

$$
\boxed{\mathrm{DOF}=2-1=1}.
$$

The constraint is holonomic because it can be written as an equation involving coordinates and time:

$$
g(x,y,t)=0.
$$

It is explicitly time-dependent because the pivot moves.

Such a constraint is called a rheonomous constraint.

**Part B — Generalized Coordinates and Velocities**

Choose $\theta$ as the generalized coordinate.

The Cartesian position is

$$
\boxed{x=vt+\ell\sin\theta}
$$

and

$$
\boxed{y=-\ell\cos\theta}.
$$

Differentiating using the chain rule,

$$
\dot x=
\frac{\partial x}{\partial\theta}\dot\theta
+
\frac{\partial x}{\partial t}.
$$

Therefore,

$$
\boxed{\dot x=v+\ell\dot\theta\cos\theta}.
$$

Similarly,

$$
\boxed{\dot y=\ell\dot\theta\sin\theta}.
$$

The velocity contains contributions from both the motion of the pivot and the rotation of the pendulum.

**Part C — Kinetic Energy and Configuration Space**

The speed squared is

$$
v_{\mathrm{particle}}^2=\dot x^2+\dot y^2.
$$

Substituting,

$$
v_{\mathrm{particle}}^2
=
(v+\ell\dot\theta\cos\theta)^2
+
(\ell\dot\theta\sin\theta)^2.
$$

Expanding and simplifying,

$$
\boxed{
v_{\mathrm{particle}}^2
=
v^2+2v\ell\dot\theta\cos\theta+\ell^2\dot\theta^2
}.
$$

The kinetic energy is therefore

$$
\boxed{
T=
\frac12m
\left(
v^2+2v\ell\dot\theta\cos\theta+\ell^2\dot\theta^2
\right)
}.
$$

The generalized velocity $\dot\theta$ is an angular rate, with units of radians per second.

It is not a Cartesian velocity component or the particle's speed.

The configuration space is one-dimensional because one generalized coordinate completely specifies the pendulum's configuration at a given time.

For a pendulum free to make complete revolutions, its configuration space has the topology of a circle, since $\theta$ and $\theta+2\pi$ describe the same configuration.

**Part D — Coordinate Transformations and Invariance**

Introduce

$$
x'=x-vt.
$$

This coordinate measures horizontal position relative to the moving pivot.

The position equations become

$$
x'=\ell\sin\theta,
\qquad
y=-\ell\cos\theta.
$$

The constraint becomes

$$
\boxed{x'^2+y^2=\ell^2}.
$$

The description is simpler because the pivot is stationary in the new coordinates.

Since $v$ is constant, this is a uniformly translating reference frame, and its acceleration is zero relative to the original inertial frame.

The physical motion has not changed. Only the coordinate description has changed.

This illustrates an important principle: different coordinate systems can describe the same physical system.

**Key oral-exam explanation:** Generalized coordinates are selected to describe the independent configurations of a constrained system, and the chain rule connects those coordinates to physical velocities.

## D10 — Introduction to the Lagrangian

### Comprehensive Problem — The Simple Pendulum Using Lagrange's Equation

A particle of mass $m$ is attached to a massless rigid rod of length $\ell$. The rod is suspended from a fixed pivot, and the particle swings freely in a vertical plane under gravity.

Choose the pivot as the origin, with the positive $y$-axis upward. Let $\theta$ be measured from the downward vertical.

Neglect friction and air resistance.

**Part A — Generalized Coordinates and Energy**

1. Explain why $\theta$ is a suitable generalized coordinate.
2. Express the Cartesian coordinates $x$ and $y$ in terms of $\theta$.
3. Determine the kinetic energy $T$ and gravitational potential energy $U$, using $U=0$ at the pivot.

**Part B — The Lagrangian**

4. Construct the Lagrangian

$$
L=T-U.
$$

5. Explain why the Lagrangian is expressed in terms of $\theta$ and $\dot\theta$ rather than $x$, $y$, $\dot x$, and $\dot y$.

**Part C — Lagrange's Equation**

6. Apply

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial\dot\theta}
\right)
-
\frac{\partial L}{\partial\theta}
=0
$$

to derive the equation of motion.

7. Explain the physical meaning of the two terms in Lagrange's equation.

**Part D — Interpretation and Connections**

8. Verify that the equation of motion agrees with Newton's second law in the tangential direction.
9. Use the small-angle approximation to determine the angular frequency of small oscillations.
10. Explain how the Lagrangian approach handles the constraint imposed by the rigid rod without explicitly calculating the tension.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed solution and conceptual explanations
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Part A — Generalized Coordinates and Energy**

The rod fixes the distance from the pivot:

$$
x^2+y^2=\ell^2.
$$

Therefore, the pendulum has one degree of freedom.

Choose $\theta$ as the generalized coordinate.

The Cartesian position is

$$
x=\ell\sin\theta,
\qquad
y=-\ell\cos\theta.
$$

Differentiating,

$$
\dot x=\ell\dot\theta\cos\theta,
$$

$$
\dot y=\ell\dot\theta\sin\theta.
$$

The speed squared is

$$
v^2=\dot x^2+\dot y^2.
$$

Thus,

$$
v^2=\ell^2\dot\theta^2(\cos^2\theta+\sin^2\theta),
$$

giving

$$
\boxed{v^2=\ell^2\dot\theta^2}.
$$

The kinetic energy is

$$
\boxed{T=\frac12m\ell^2\dot\theta^2}.
$$

Using the pivot as the zero of gravitational potential energy,

$$
U=mgy.
$$

Therefore,

$$
\boxed{U=-mg\ell\cos\theta}.
$$

**Part B — Constructing the Lagrangian**

The Lagrangian is

$$
L=T-U.
$$

Substituting,

$$
\boxed{
L=\frac12m\ell^2\dot\theta^2+mg\ell\cos\theta
}.
$$

The Lagrangian depends on the generalized coordinate and generalized velocity.

Using $\theta$ automatically satisfies the fixed-length constraint, so we do not need to treat $x$ and $y$ as independent variables.

**Part C — Lagrange's Equation**

The equation is

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial\dot\theta}
\right)
-
\frac{\partial L}{\partial\theta}
=0.
$$

First,

$$
\frac{\partial L}{\partial\dot\theta}
=
m\ell^2\dot\theta.
$$

Taking the time derivative,

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial\dot\theta}
\right)
=
m\ell^2\ddot\theta.
$$

Next,

$$
\frac{\partial L}{\partial\theta}
=
-mg\ell\sin\theta.
$$

Substituting into Lagrange's equation,

$$
m\ell^2\ddot\theta+mg\ell\sin\theta=0.
$$

Dividing by $m\ell^2$,

$$
\boxed{
\ddot\theta+\frac{g}{\ell}\sin\theta=0
}.
$$

This is the nonlinear equation of motion for the simple pendulum.

**Physical interpretation of Lagrange's equation**

The quantity

$$
\frac{\partial L}{\partial\dot\theta}
=
m\ell^2\dot\theta
$$

is the generalized momentum associated with $\theta$. In this case, it is the angular momentum about the pivot.

Its time derivative represents the rate of change of angular momentum.

The quantity

$$
\frac{\partial L}{\partial\theta}
=
-mg\ell\sin\theta
$$

is the generalized force associated with $\theta$. Here, it is the gravitational torque about the pivot.

Thus, Lagrange's equation expresses the same angular dynamics as the torque equation.

**Part D — Newtonian Verification**

The tangential component of gravity is

$$
F_\theta=-mg\sin\theta.
$$

Newton's second law in the tangential direction gives

$$
m\ell\ddot\theta=-mg\sin\theta.
$$

Multiplying by $\ell$,

$$
m\ell^2\ddot\theta=-mg\ell\sin\theta.
$$

Therefore,

$$
\boxed{
\ddot\theta+\frac{g}{\ell}\sin\theta=0
}.
$$

This agrees with the Lagrangian result.

**Small-Angle Approximation**

For sufficiently small angles measured in radians,

$$
\sin\theta\approx\theta.
$$

The equation becomes

$$
\ddot\theta+\frac{g}{\ell}\theta=0.
$$

Comparing with the simple harmonic oscillator equation,

$$
\ddot\theta+\omega_0^2\theta=0,
$$

we identify

$$
\boxed{\omega_0=\sqrt{\frac{g}{\ell}}}.
$$

The corresponding period is

$$
\boxed{T_0=2\pi\sqrt{\frac{\ell}{g}}}.
$$

**Why does tension not appear?**

The rigid rod constrains the particle to move along a circular path.

The tension acts radially, while the allowed instantaneous motion is tangential.

Thus, tension does no work along the allowed motion.

By choosing $\theta$ as the generalized coordinate, the constraint is already incorporated into the description.

The Lagrangian method therefore produces the equation of motion without explicitly calculating the tension.

This does not mean tension is physically absent. It means that the chosen generalized coordinate allows us to determine the motion without solving for that constraint force.

**Key oral-exam explanation:** The Lagrangian formulation uses kinetic and potential energy to obtain the equation of motion in coordinates adapted to the system, rather than beginning with individual force components.
