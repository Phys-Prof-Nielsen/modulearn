## D2.3.1 Motivation

In the *Introduction to Line Integrals*, we have been so fortunate that our paths were aligned well with the $xy$-directions in a cartesian coordinate system. However, often the world is not as nice and will serve us some humble pie when applying the approach above. For example, what if the path was part of a circle? Already then it appears messy to evaluate the dot product between our force and the line element. This is where **parameterization** comes in and saves us from mathematical humiliation (well, we may still be humiliated but then it is at least because of our own inability to do math). The approach of parameterization is a general approach to solving line integrals (or higher orders such surface and volume integrals as well).

## D2.3.2 Parameterization

For simplicity, we introduce parameterization for a path in the $xy$-plane. The same idea extends naturally to three dimensions.

The starting point is that many paths can be described by an equation of the form

$$
g(x,y)=0.
$$

For example:

- Straight line:

$$
y - ax - b = 0
$$

- Circle:

$$
g(x,y)=x^2+y^2-R^2=0
$$

This equation describes the full curve, but it does not always tell us how an object moves along the curve. For work, we need to follow the path point by point.

---

### Parameterizing a path

To describe motion along a path, we introduce a parameter $s$. Each point on the path is then written as

$$
x=x(s), \qquad y=y(s).
$$

---

### Differentials along the path

As the parameter changes by a small amount $ds$, the position changes by

$$
dx=\frac{dx}{ds}ds
$$

and

$$
dy=\frac{dy}{ds}ds.
$$

These expressions tell us how changes in the parameter produce changes in position along the path.

---

### Work along a parameterized path

The work done by a force is

$$
W=\int_C \vec{F}\cdot d\vec{l}.
$$

In Cartesian components,

$$
W=\int_C (F_x\,dx+F_y\,dy).
$$

Substituting the parameterized differentials gives

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Definition — Work Along a Parameterized Path</h3>

If a path is parameterized by

$$
x=x(s), \qquad y=y(s),
$$

then the work is

$$
\boxed{
W =
\int_{s_1}^{s_2}
\left[
F_x(x(s),y(s))\frac{dx}{ds}
+
F_y(x(s),y(s))\frac{dy}{ds}
\right]ds
}
$$

</div>

This formula is the bridge between the piecewise straight-line examples and fully general line integrals.

---

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<h3 style="margin-top:0;">Example — Parameterization from Physics I</h3>

<strong>Problem</strong>  

In Physics I, we often described motion along a straight line using time as a parameter.  

Consider motion with constant velocity $v_x$ along the $x$-axis, starting from position $x_0$.

Show how this is an example of **parameterization**, and relate it to the differential form used in line integrals.

---

<strong>Solution</strong>  

In Physics I, we wrote the position as

$$
x = x_0 + v_x t.
$$

This means that the position is expressed as a function of a parameter:

$$
x = x(t).
$$

Here, time $t$ plays the role of the parameter that traces out the path.

---

<strong>How the parameter evolves</strong>  

As the particle moves, the parameter changes by a small amount $dt$. This produces a small change in position:

$$
dx = \frac{dx}{dt}dt.
$$

Since the velocity is constant,

$$
\frac{dx}{dt} = v_x,
$$

so

$$
dx = v_x\,dt.
$$

---

<strong>Interpretation</strong>  

This is exactly the same structure used in parameterization:

- The path is described by a function $x(t)$  
- Motion along the path is described by how $t$ changes  
- Small displacements are obtained through the derivative  

---

**Key Insight**

We have already been using parameterization:

> Time is just one possible parameter.

In general, we replace $t$ with a parameter $s$ to describe motion along any path, not just motion in time.

</div>

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">
<strong>Example — Work Along a Parameterized Half-Circle</strong>

<strong>Problem</strong>  

Consider the force

$$
\vec{F} = 5.0(xy\,\hat{\mathbf{i}} + y^2\,\hat{\mathbf{j}})
$$

acting on an object. The object moves along a half-circle, counterclockwise from $(x,y)=(R,0)$ to $(x,y)=(-R,0)$.

Find the work done by this force.

---

<strong>Step 1 — Choose a parameter</strong>  

The geometry suggests using the angle $\theta$ as the parameter:

$$
s \rightarrow \theta
$$

<div style="text-align:center;">
<img src="fig1.png" width="300">
</div>

---

<strong>Step 2 — Parameterize the path</strong>  

$$
x = R\cos\theta,
\qquad
y = R\sin\theta
$$

---

<strong>Step 3 — Compute derivatives</strong>  

$$
\frac{dx}{d\theta} = -R\sin\theta,
\qquad
\frac{dy}{d\theta} = R\cos\theta
$$

---

<strong>Step 4 — Express the force in terms of $\theta$</strong>  

$$
F_x = 5.0xy = 5.0R^2\sin\theta\cos\theta
$$

$$
F_y = 5.0y^2 = 5.0R^2\sin^2\theta
$$

---

<strong>Step 5 — Set up the line integral</strong>  

$$
W = \int_{\theta=0}^{\theta=\pi}
\left[
F_x(\theta)\frac{dx}{d\theta}
+
F_y(\theta)\frac{dy}{d\theta}
\right] d\theta
$$

$$
W = \int_{0}^{\pi}
\left[
(5.0R^2\sin\theta\cos\theta)(-R\sin\theta)
+
(5.0R^2\sin^2\theta)(R\cos\theta)
\right] d\theta
$$

---

<strong>Step 6 — Simplify and evaluate</strong>  

$$
W = 5.0R^3 \int_{0}^{\pi}
\left[
-\sin^2\theta\cos\theta
+
\sin^2\theta\cos\theta
\right] d\theta
$$

The integrand cancels:

$$
W = 0
$$

---

<strong>Result</strong>  

$$
\boxed{W = 0~\text{J}}
$$

---

<strong>Interpretation</strong>  

The contributions from different parts of the path cancel exactly.

This example highlights two key ideas:

- Parameterization reduces a multi-variable problem to a **single-variable integral**  
- Even nontrivial paths can sometimes produce **zero net work** due to symmetry  

</div>

## D2.3.3 Common Parameterizations

In practice, choosing a good parameter is often the most important step in evaluating a line integral.  

While any path can, in principle, be parameterized, there are a few standard cases that appear repeatedly in physics. In this course, we will focus on three important types of paths:

- straight lines  
- circles  
- ellipses  

These capture a large number of physically relevant situations.

---

### Straight Line

A straight line between two points

$$
(x_1,y_1) \rightarrow (x_2,y_2)
$$

can be parameterized using a parameter $s$ that runs from $0$ to $1$:

$$
x(s) = x_1 + (x_2 - x_1)s
$$

$$
y(s) = y_1 + (y_2 - y_1)s
$$

with

$$
0 \le s \le 1.
$$

This parameterization has a clear interpretation:

- $s=0$ corresponds to the starting point  
- $s=1$ corresponds to the ending point  
- intermediate values trace the path smoothly between them  

---

### Circle

A circle of radius $R$ centered at the origin is naturally parameterized using an angle $\theta$:

$$
x(\theta) = R\cos\theta
$$

$$
y(\theta) = R\sin\theta
$$

The parameter $\theta$ determines how far around the circle we move:

- $\theta = 0$ corresponds to $(R,0)$  
- $\theta = \pi/2$ corresponds to $(0,R)$  
- $\theta = 2\pi$ completes a full revolution  

To describe a portion of a circle, we simply restrict the limits of $\theta$.

---

### Ellipse

An ellipse with semi-major axis $a$ and semi-minor axis $b$ can also be parameterized using an angle:

$$
x(\theta) = a\cos\theta
$$

$$
y(\theta) = b\sin\theta
$$

Even though the shape is not circular, the parameter $\theta$ still “sweeps out” the curve.

The limits on $\theta$ determine how much of the ellipse is traced:

- $0 \le \theta \le 2\pi$ gives the full ellipse  
- restricted intervals give arcs  

---

### Why these matter

These three parameterizations are especially useful because:

- they match common physical geometries  
- they lead to manageable derivatives $dx/ds$ and $dy/ds$  
- they allow us to reduce line integrals to **single-variable integrals**  

In most problems in this course, identifying one of these standard forms will significantly simplify the work.
