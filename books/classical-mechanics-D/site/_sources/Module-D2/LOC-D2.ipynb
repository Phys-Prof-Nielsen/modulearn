## Why this module matters

In the previous module, we revisited **momentum and impulse**, developing a global view of motion in which forces accumulate their effects over **time**:

$$
\int \mathbf{F}\,dt \;\rightarrow\; \Delta \mathbf{p}.
$$

In this module, we develop a closely related and equally powerful framework based on **work and energy**. Instead of tracking how forces change momentum through time, we now examine how forces transfer energy through **displacement**.

This leads to the concept of **work**, defined as

$$
W = \int \mathbf{F} \cdot d\mathbf{r},
$$

and the **work–energy theorem**

$$
W_{\text{net}} = \Delta K.
$$

Just as impulse represents **momentum transfer**, work represents **energy transfer**.

A key structural parallel emerges:

- **Impulse–Momentum**: accumulation of force over **time** → change in momentum  
- **Work–Energy**: accumulation of force over **space** → change in energy  

While impulse–momentum provides a **vector description** of motion, work–energy provides a **scalar description**, which often simplifies analysis—especially in situations where forces depend on position.

In this module, we revisit work from Physics I and extend it to motion along **general paths** using **line integrals**. This leads naturally to the concept of **conservative forces**, where work becomes independent of path and can be expressed in terms of a **potential energy function**.

Together, impulse–momentum and work–energy form two complementary global frameworks for analyzing motion, each emphasizing a different aspect of how forces act on a system.

By the end of this module, you will be able to:

---

### A. Work and Energy (Review)  
*Reconnects with Physics I concepts and identifies their limitations.*

- **A1.** Define work as the integral of force along a path  

$$
W = \int \mathbf{F} \cdot d\mathbf{r}.
$$

- **A2.** Compute work for motion along straight-line paths.  
- **A3.** State and apply the **work–energy theorem**  

$$
W_{\text{net}} = \Delta K.
$$

- **A4.** Define kinetic energy  

$$
K = \frac{1}{2}mv^2.
$$

- **A5.** Distinguish between **work on a particle** and **pseudo work on a system (center of mass)**.  
- **A6.** Recognize the limitations of treating work in only one dimension or along simple paths.

---

### B. Work Along General Paths  
*Extends work to motion in two and three dimensions.*

- **B1.** Define differential work  

$$
dW = \mathbf{F} \cdot d\mathbf{r}.
$$

- **B2.** Interpret work as the accumulation of local contributions along a path.  
- **B3.** Recognize that work may depend on the **path taken**.  
- **B4.** Distinguish between path-dependent and path-independent work conceptually.

---

### C. Line Integrals and Parameterization  
*Introduces the mathematical framework for computing work along curves.*

- **C1.** Represent a path using a parameterization  

$$
\mathbf{r}(t).
$$

- **C2.** Express differential displacement as  

$$
d\mathbf{r} = \frac{d\mathbf{r}}{dt} dt.
$$

- **C3.** Compute work as a line integral  

$$
W = \int_C \mathbf{F} \cdot d\mathbf{r}.
$$

- **C4.** Evaluate work integrals using parameterized paths.  
- **C5.** Recognize that coordinate choice (Cartesian, polar, etc.) can simplify parameterization.

---

### D. Conservative Forces (Global and Local)  
*Identifies when work becomes path-independent.*

- **D1.** Define a **conservative force** as one for which work is path-independent.  
- **D2.** State the global condition  

$$
\oint \mathbf{F} \cdot d\mathbf{r} = 0.
$$

- **D3.** Recognize that testing all paths is impractical.  
- **D4.** Use the local condition for conservative forces  

$$
\nabla \times \mathbf{F} = 0.
$$

- **D5.** Interpret the physical meaning of zero curl.

---

### E. Potential Energy and Force  
*Connects conservative forces to scalar potential functions.*

- **E1.** Define potential energy through work  

$$
W = -\Delta U.
$$

- **E2.** Relate force to potential energy  

$$
\mathbf{F} = -\nabla U.
$$

- **E3.** Connect this to the one-dimensional result  

$$
F_x = -\frac{dU}{dx}.
$$

- **E4.** Interpret potential energy as a scalar field encoding the force.  
- **E5.** Use potential energy to analyze motion without directly solving Newton’s second law.

---

### F. Energy Methods and Physical Insight  
*Highlights the advantages of energy-based analysis.*

- **F1.** Use energy conservation when forces are conservative  

$$
K + U = \text{constant}.
$$

- **F2.** Recognize when energy methods are simpler than force-based methods.  
- **F3.** Interpret motion in terms of energy exchange and constraints.  
- **F4.** Connect energy methods to symmetry and invariance in physical systems.

---

### Performance Criteria

- Work is correctly computed as a **line integral** along a path.  
- Students correctly distinguish between **path-dependent and path-independent** work.  
- Conservative forces are identified using both **global and local criteria**.  
- Potential energy is correctly defined and used to compute forces.  
- Energy conservation is applied appropriately to analyze motion.  
- Students recognize when **energy methods provide a simpler approach** than Newton’s laws.
