# D2.3 The Kronecker Delta
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D2.3.1 Motivation: An Identity Operator for Components

In index notation we often need a compact way to express operations that **select specific components of vectors**. For example, suppose we want to write an expression that simply returns the $i$-th component of a vector.

In standard notation this would be trivial, but in index notation it is convenient to introduce a special symbol that behaves like an **identity operator**.

This object is called the **Kronecker delta**, and it plays an important role in index notation and tensor algebra.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:14px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Definition — Kronecker Delta</h3>

The **Kronecker delta** is defined as

$$
\delta_{ij} =
\begin{cases}
1 & i = j \\
0 & i \ne j
\end{cases}
$$

Thus the Kronecker delta equals **1 when the indices are equal** and **0 otherwise**.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D2.3.2 Acting on Vector Components

The Kronecker delta acts like the **identity operator** on vector components.

For example,

$$
\delta_{ij} a_j = a_i.
$$

To see why this works, expand the summation explicitly:

$$
\delta_{ij} a_j =
\delta_{i1}a_1 + \delta_{i2}a_2 + \delta_{i3}a_3.
$$

Because $\delta_{ij}$ equals 1 only when the indices match, only one term survives in the sum.

For example, if $i=2$,

$$
\delta_{2j} a_j =
\delta_{21}a_1 + \delta_{22}a_2 + \delta_{23}a_3.
$$

Using the definition of the Kronecker delta,

$$
\delta_{21}=0, \quad
\delta_{22}=1, \quad
\delta_{23}=0,
$$

so the expression becomes

$$
\delta_{2j} a_j = a_2.
$$

Thus the Kronecker delta simply **selects the component corresponding to the free index**.

## D2.3.3 Dot Products with the Kronecker Delta

The Kronecker delta can also be used to express dot products in index notation.

Recall that

$$
\mathbf{a}\cdot\mathbf{b} = a_i b_i.
$$

This can also be written as

$$
\mathbf{a}\cdot\mathbf{b} = \delta_{ij} a_i b_j.
$$

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">
<h3 style="margin-top:0; color:#000000;">Box Activity — Showing that $\delta_{ij}a_i b_j$ Produces the Dot Product</h3>

In index notation, the dot product of two vectors can be written as

$$
\mathbf{a}\cdot\mathbf{b} = \delta_{ij} a_i b_j.
$$

In this activity you will expand the indices explicitly to see how the **Kronecker delta selects the matching components**.

Start by expanding the indices explicitly.

$$
\delta_{ij} a_i b_j
=
\delta_{11} a_1 b_1
+
\delta_{12} a_1 b_2
+
\delta_{13} a_1 b_3
$$

$$
+
\delta_{21} a_2 b_1
+
\delta_{22} a_2 b_2
+
\delta_{23} a_2 b_3
$$

$$
+
\delta_{31} a_3 b_1
+
\delta_{32} a_3 b_2
+
\delta_{33} a_3 b_3.
$$

Now apply the definition of the Kronecker delta:

$$
\delta_{11}=\delta_{22}=\delta_{33}=1
$$

and

$$
\delta_{12}=\delta_{13}=\delta_{21}=\delta_{23}=\delta_{31}=\delta_{32}=0.
$$

All cross terms therefore vanish, leaving

$$
\delta_{ij} a_i b_j
=
a_1 b_1 + a_2 b_2 + a_3 b_3.
$$

Thus

$$
\delta_{ij} a_i b_j = \mathbf{a}\cdot\mathbf{b}.
$$

---

### Interpretation

The Kronecker delta acts like an **identity operator that forces the indices to match**.  
It removes all cross terms and leaves only products of corresponding components.

<h3 style="margin-top:0; color:#000000;">Example — Using the Kronecker Delta</h3>

<strong>Problem</strong>

Let

$$
\mathbf{a} = (2,-3,4).
$$

Compute

$$
\delta_{ij} a_j
$$

for $i=1,2,3$.

---

<strong>Solution</strong>

Using

$$
\delta_{ij} a_j =
\delta_{i1}a_1 +
\delta_{i2}a_2 +
\delta_{i3}a_3
$$

we evaluate each component.

For $i=1$

$$
\delta_{1j}a_j =
\delta_{11}a_1 +
\delta_{12}a_2 +
\delta_{13}a_3
=
a_1 = 2.
$$

For $i=2$

$$
\delta_{2j}a_j =
a_2 = -3.
$$

For $i=3$

$$
\delta_{3j}a_j =
a_3 = 4.
$$

Thus

$$
\delta_{ij}a_j = a_i.
$$

## D2.3.4 Why the Kronecker Delta Is Important

The Kronecker delta appears throughout physics whenever we need a compact way to represent identity relationships between components.

Later in the course it will appear in expressions such as

- the **moment of inertia tensor**
- vector identities involving cross products
- tensor equations in advanced mechanics

Learning to manipulate expressions involving $\delta_{ij}$ is an important step toward understanding the more general mathematical structures used in modern physics.
