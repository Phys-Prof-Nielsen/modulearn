## D2.5.1 Motivation: Why Use Index Notation?

In earlier sections we introduced two important objects:

- the **Kronecker delta** $\delta_{ij}$
- the **Levi–Civita symbol** $\epsilon_{ijk}$

Together these objects allow us to express vector operations compactly and derive vector identities systematically.

Many vector identities that appear complicated in component form become much easier to manipulate using index notation.

These tools will be especially useful later when we study **rotational dynamics**, where quantities such as angular momentum and torque naturally involve cross products.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D2.5.2 Key Identity

One of the most important identities involving the Levi–Civita symbol is

$$
\epsilon_{ijk}\epsilon_{ilm}
=
\delta_{jl}\delta_{km}
-
\delta_{jm}\delta_{kl}.
$$

This identity allows products of Levi–Civita symbols to be rewritten in terms of Kronecker deltas.

This conversion is extremely useful because the Kronecker delta acts like an **identity operator** on vector components.

## D2.5.3 Example: Triple Cross Product

Consider the vector identity

$$
\mathbf{a}\times(\mathbf{b}\times\mathbf{c})
$$

Using index notation,

$$
(\mathbf{b}\times\mathbf{c})_k =
\epsilon_{klm} b_l c_m
$$

Thus

$$
(\mathbf{a}\times(\mathbf{b}\times\mathbf{c}))_i
=
\epsilon_{ijk} a_j (\mathbf{b}\times\mathbf{c})_k
$$

Substituting the previous expression gives

$$
(\mathbf{a}\times(\mathbf{b}\times\mathbf{c}))_i
=
\epsilon_{ijk} a_j \epsilon_{klm} b_l c_m.
$$

Now apply the identity

$$
\epsilon_{ijk}\epsilon_{klm}
=
\delta_{il}\delta_{jm}
-
\delta_{im}\delta_{jl}.
$$

This gives

$$
(\mathbf{a}\times(\mathbf{b}\times\mathbf{c}))_i
=
(\delta_{il}\delta_{jm}-\delta_{im}\delta_{jl})
a_j b_l c_m.
$$

Applying the Kronecker deltas simplifies the expression to

$$
(\mathbf{a}\times(\mathbf{b}\times\mathbf{c}))_i
=
b_i(\mathbf{a}\cdot\mathbf{c})
-
c_i(\mathbf{a}\cdot\mathbf{b}).
$$

Thus we obtain the familiar identity

$$
\mathbf{a}\times(\mathbf{b}\times\mathbf{c})
=
\mathbf{b}(\mathbf{a}\cdot\mathbf{c})
-
\mathbf{c}(\mathbf{a}\cdot\mathbf{b}).
$$

The scalar triple product can be written as

$$
\mathbf{a}\cdot(\mathbf{b}\times\mathbf{c})
=
a_i (\mathbf{b}\times\mathbf{c})_i.
$$

Using

$$
(\mathbf{b}\times\mathbf{c})_i
=
\epsilon_{ijk} b_j c_k
$$

we obtain

$$
\mathbf{a}\cdot(\mathbf{b}\times\mathbf{c})
=
\epsilon_{ijk} a_i b_j c_k.
$$

Because the Levi–Civita symbol is **cyclically symmetric**, the expression remains unchanged under cyclic permutations of the vectors. Thus

$$
\mathbf{a}\cdot(\mathbf{b}\times\mathbf{c})
=
\mathbf{b}\cdot(\mathbf{c}\times\mathbf{a})
=
\mathbf{c}\cdot(\mathbf{a}\times\mathbf{b}).
$$

## D2.5.4 Why These Identities Matter

Index notation allows vector identities to be derived in a systematic and compact way. Instead of memorizing many different formulas, we can often derive them using a few basic identities involving the Kronecker delta and Levi–Civita symbol.

These tools become especially valuable in topics such as

- **rotational dynamics**
- **angular momentum**
- **rigid body motion**
- **electromagnetism**

where cross products appear frequently.
