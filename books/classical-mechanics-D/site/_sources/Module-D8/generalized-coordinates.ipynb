# D8.2 Generalized Coordinates

## D8.2.1 Motivation: Why Generalized Coordinates?

In introductory physics, motion is usually described using **Cartesian coordinates** $(x, y, z)$.  
However, when constraints are present, not all coordinates are independent.

For example:

- A pendulum bob in the plane moves along a circle of fixed radius.  
- A bead on a wire must follow the shape of the wire.  
- A particle moving in polar coordinates is naturally described by $(r,\theta)$ rather than $(x,y)$.

Describing such systems with **too many coordinates** leads to redundant variables and unnecessarily complicated equations.

**Generalized coordinates** provide a systematic way to describe motion:

- They automatically respect the constraints of the system.
- They reduce the system to the **minimum number of independent variables** (the degrees of freedom).
- They allow us to choose variables that match the **geometry of the problem**.

The guiding idea is simple:

> Choose coordinates that match the geometry and constraints of the system.

This approach becomes the foundation of **analytical mechanics**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.2.2 Coordinate Choice

A key advantage of generalized coordinates is that **we are not restricted to Cartesian coordinates**.  
Instead, we can choose coordinates that simplify the description of the system.

Good coordinate choices typically

- automatically satisfy constraints,
- reduce the number of variables,
- reflect the symmetry or geometry of the system.

### Example 1: Simple Pendulum

Cartesian coordinates:

$$
(x,y)
$$

Constraint:

$$
x^2 + y^2 = l^2
$$

Instead, introduce a single coordinate

$$
q = \theta
$$

The position becomes

$$
x = l\sin\theta, \qquad y = -l\cos\theta
$$

The constraint is now automatically satisfied, and the motion is described by **one coordinate instead of two**.

---

### Example 2: Block on an Inclined Plane

Cartesian description:

$$
(x,y)
$$

Constraint:

$$
y - x\tan\alpha = 0
$$

A more natural coordinate is the **distance along the incline**

$$
q = s
$$

The position becomes

$$
x = s\cos\alpha, \qquad y = s\sin\alpha
$$

Again, the motion is described by **one coordinate instead of two**.

## D8.2.3 Generalized Velocities

Once generalized coordinates are introduced, we also define their **time derivatives**, called **generalized velocities**.

If

$$
q_i
$$

is a generalized coordinate, the generalized velocity is

$$
\dot q_i = \frac{dq_i}{dt}.
$$

These quantities describe **how quickly the coordinates change with time**.

Generalized velocities are not necessarily ordinary Cartesian velocities.

Example:

Pendulum

Generalized coordinate:

$$
q = \theta
$$

Generalized velocity:

$$
\dot q = \dot\theta
$$

The physical speed of the mass is

$$
v = l\dot\theta.
$$

Thus generalized velocities describe **rates of change of coordinates**, not necessarily physical speeds.

In general, the Cartesian position of a particle can be written as a function of generalized coordinates:

$$
\mathbf{r} = \mathbf{r}(q_1,q_2,\dots,q_N,t)
$$

The velocity then follows from the **chain rule**:

$$
\mathbf{v} =
\sum_i
\frac{\partial \mathbf{r}}{\partial q_i}\dot q_i
+
\frac{\partial \mathbf{r}}{\partial t}
$$

This relation allows us to express velocities — and therefore kinetic energy — entirely in terms of generalized coordinates and generalized velocities.

>This chain-rule relationship between Cartesian and generalized velocities will be used repeatedly when expressing kinetic energy in new coordinate systems.

## D8.2.4 Configuration Space

When we describe a system using generalized coordinates

$$
(q_1, q_2, \dots, q_N)
$$

each possible set of coordinate values corresponds to a **specific configuration of the system**.

The set of all possible configurations forms an abstract space called **configuration space**.

Important idea:

> A point in configuration space represents the complete configuration of the system.

Examples:

**Pendulum**

Configuration described by

$$
\theta
$$

Configuration space: a **circle**.

---

**Particle moving in a plane**

Coordinates:

$$
(x,y)
$$

Configuration space: a **two-dimensional plane**.

---

**Double pendulum**

Coordinates:

$$
(\theta_1,\theta_2)
$$

Configuration space: a **two-dimensional surface** (topologically a torus).

In analytical mechanics, the motion of a system can be viewed as a **trajectory through configuration space**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.2.5 Why Generalized Coordinates Are Powerful

Using generalized coordinates allows us to

- eliminate constraint equations,
- describe systems using the **minimum number of independent variables**,
- adapt coordinates to the geometry of the system,
- systematically construct equations of motion.

These ideas become essential when developing **Lagrangian mechanics**, where the dynamics are expressed entirely in terms of

$$
q_i \quad \text{and} \quad \dot q_i.
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## D8.2.6 Summary

- **Generalized coordinates** are independent variables that uniquely describe the configuration of a system.
- The number of generalized coordinates equals the **degrees of freedom**.
- Coordinates should be chosen to match the **geometry and constraints** of the system.
- The time derivatives of generalized coordinates are called **generalized velocities**.
- The set of all possible configurations forms **configuration space**.

These concepts provide the coordinate-based language used throughout **analytical mechanics**.

<h3 style="margin-top:0; color:#000000;">Example – Generalized Coordinates for a Bead Sliding on a Circular Hoop</h3>

<strong>Problem</strong>  

A small bead of mass $m$ slides without friction on a rigid circular hoop of radius $R$.  
The hoop is fixed in a vertical plane.

1. Determine the **degrees of freedom** of the system.
2. Choose a convenient **generalized coordinate**.
3. Write the **Cartesian coordinates** of the bead in terms of the generalized coordinate.
4. Determine the **generalized velocity** and the bead’s speed.
5. Write the **kinetic energy** in terms of the generalized coordinate.

---

<strong>Solution</strong>

<strong>1. Degrees of freedom</strong>

If we describe the bead using Cartesian coordinates in the plane,

$$
(x,y),
$$

two coordinates are required.

However, the bead is constrained to remain on the circular hoop.

The geometric constraint is

$$
x^2 + y^2 = R^2.
$$

Thus there is one constraint relating the two coordinates.

The number of degrees of freedom is therefore

$$
\text{DOF} = 2 - 1 = 1.
$$

So the motion can be described using **one independent coordinate**.

---

<strong>2. Choosing a generalized coordinate</strong>

The natural coordinate describing motion around the hoop is the **angle** $\theta$ measured from the vertical.

Thus we define the generalized coordinate

$$
q = \theta.
$$

Using this coordinate automatically satisfies the circular constraint.

---

<strong>3. Cartesian coordinates in terms of the generalized coordinate</strong>

The position of the bead on the hoop is

$$
x = R\sin\theta
$$

$$
y = -R\cos\theta.
$$

Substituting these into the constraint equation gives

$$
x^2 + y^2 = R^2,
$$

so the constraint is automatically satisfied.

This illustrates one of the main advantages of generalized coordinates:  
the constraint is **built into the coordinate choice**.

---

<strong>4. Generalized velocity</strong>

The generalized velocity is the time derivative of the generalized coordinate:

$$
\dot q = \dot\theta.
$$

To determine the physical velocity of the bead, differentiate the Cartesian coordinates.

Horizontal velocity:

$$
\dot x = R\cos\theta\,\dot\theta
$$

Vertical velocity:

$$
\dot y = R\sin\theta\,\dot\theta.
$$

---

<strong>5. Speed of the bead</strong>

The speed is

$$
v = \sqrt{\dot x^2 + \dot y^2}.
$$

Substituting the expressions above gives

$$
v =
\sqrt{
R^2\cos^2\theta\,\dot\theta^2
+
R^2\sin^2\theta\,\dot\theta^2
}.
$$

Factor out common terms:

$$
v =
\sqrt{
R^2\dot\theta^2(\cos^2\theta+\sin^2\theta)
}.
$$

Using

$$
\cos^2\theta + \sin^2\theta = 1,
$$

we obtain

$$
v = R\dot\theta.
$$

---

<strong>6. Kinetic energy in generalized coordinates</strong>

The kinetic energy of the bead is

$$
T = \frac{1}{2}mv^2.
$$

Substituting the expression for the speed,

$$
T = \frac{1}{2}mR^2\dot\theta^2.
$$

Thus the kinetic energy is written entirely in terms of the **generalized coordinate** $\theta$ and the **generalized velocity** $\dot\theta$.

---

<strong>Conclusion</strong>

The bead initially appears to require two Cartesian coordinates $(x,y)$, but the circular constraint reduces the motion to one degree of freedom.

A natural generalized coordinate is

$$
q = \theta.
$$

The Cartesian coordinates are

$$
x = R\sin\theta, \qquad y = -R\cos\theta,
$$

the speed of the bead is

$$
v = R\dot\theta,
$$

and the kinetic energy becomes

$$
T = \frac{1}{2}mR^2\dot\theta^2.
$$

This example shows how generalized coordinates simplify the description of constrained motion.

<h3 style="margin-top:0; color:#000000;">Box Activity – Generalized Coordinate for a Block Sliding on an Inclined Plane</h3>

In the previous module, we described a block sliding on an incline using the **Cartesian coordinates** $(x,y)$ together with a **constraint equation**.

In analytical mechanics, a more efficient approach is to choose a **generalized coordinate that automatically satisfies the constraint**.

In this activity you will replace the redundant coordinates $(x,y)$ with a **single generalized coordinate describing motion along the incline**.

---

**Problem**

A block of mass $m$ slides down a rough incline of angle $\alpha$ with coefficient of kinetic friction $\mu_k$.

Choose a coordinate system with

- the **$x$-axis horizontal**
- the **$y$-axis vertical (positive upward)**

The top of the ramp is at height $H$.

The block moves along the straight surface of the incline.

---

**Task**

1. **Constraint equation**

Using the horizontal–vertical coordinate system, write the constraint equation relating $x$ and $y$ that ensures the block remains on the surface of the incline.

---

2. **Introduce a generalized coordinate**

Instead of using $(x,y)$, introduce a generalized coordinate

$$
q = s
$$

where $s$ represents the **distance measured along the incline from the top of the ramp**.

Explain why this coordinate automatically satisfies the constraint.

---

3. **Express Cartesian coordinates in terms of the generalized coordinate**

Write the Cartesian coordinates of the block as functions of $s$.

(Hint: the motion follows the direction of the incline.)

---

4. **Generalized velocity**

Determine the generalized velocity

$$
\dot q = \dot s
$$

and compute the **Cartesian velocity components** $(\dot x,\dot y)$.

---

5. **Speed of the block**

Use

$$
v = \sqrt{\dot x^2 + \dot y^2}
$$

to determine the speed of the block in terms of $\dot s$.

---

6. **Reflection**

- Why does the generalized coordinate eliminate the need for the constraint equation?
- Why is the motion described by **one coordinate instead of two**?
- Why are generalized coordinates often chosen to follow the **geometry of the system**?

---

<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
A Possible Solution Guide

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1 – Constraint equation**

The block remains on the incline.

Using horizontal–vertical coordinates, the surface of the incline satisfies

$$
H - y = x\tan\alpha.
$$

This equation relates $x$ and $y$, restricting the motion to the surface.

---

**Step 2 – Generalized coordinate**

Define the generalized coordinate

$$
q = s,
$$

where $s$ is the distance measured **along the incline** from the top.

Since the block moves only along the surface, this coordinate uniquely determines its position.

Thus the system has

$$
\text{DOF} = 1.
$$

---

**Step 3 – Cartesian coordinates**

The block moves along a line inclined at angle $\alpha$.

Thus

$$
x = s\cos\alpha
$$

$$
y = H - s\sin\alpha.
$$

Substituting these into the constraint equation confirms it is automatically satisfied.

---

**Step 4 – Velocity components**

Differentiate with respect to time.

Horizontal velocity

$$
\dot x = \dot s\cos\alpha
$$

Vertical velocity

$$
\dot y = -\dot s\sin\alpha.
$$

---

**Step 5 – Speed of the block**

The speed is

$$
v = \sqrt{\dot x^2 + \dot y^2}.
$$

Substituting

$$
v =
\sqrt{
(\dot s\cos\alpha)^2 + (-\dot s\sin\alpha)^2
}.
$$

Factor out $\dot s^2$:

$$
v =
\sqrt{
\dot s^2(\cos^2\alpha + \sin^2\alpha)
}.
$$

Using

$$
\cos^2\alpha + \sin^2\alpha = 1,
$$

we obtain

$$
v = \dot s.
$$

---

**Interpretation**

The generalized coordinate $s$ directly measures the **distance traveled along the incline**, so its time derivative is exactly the **speed of the block**.

By choosing a coordinate aligned with the motion, the constraint is automatically satisfied and the dynamics become simpler.
