## Problem 1: Effective Potential for Earth Orbiting the Sun

The Earth moves around the Sun under the gravitational potential

$$
U(r) = -\frac{GM_\odot m_E}{r}.
$$

For central-force motion, the effective potential is

$$
U_{\text{eff}}(r)
= -\frac{GM_\odot m_E}{r} + \frac{L^2}{2m_E r^2},
$$

where $L$ is Earth’s angular momentum about the Sun.

Use:

$$
G = 6.674\times 10^{-11}~\text{N m}^2/\text{kg}^2,
$$

$$
M_\odot = 1.989\times 10^{30}~\text{kg},
$$

$$
m_E = 5.972\times 10^{24}~\text{kg},
$$

$$
r_E = 1.496\times 10^{11}~\text{m}.
$$

Assume Earth’s orbit is approximately circular.

Determine:

1. Earth’s orbital speed $v_E$  
2. Earth’s angular momentum $L$  
3. The circular orbit radius from the effective potential  
4. The circular orbit energy  
5. Whether the circular orbit is stable  
6. A plot of $U_{\text{eff}}(r)$ showing Earth’s circular orbit  

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

For a circular orbit,

$$
v_E = \sqrt{\frac{GM_\odot}{r_E}}.
$$

$$
L = m_E r_E v_E.
$$

$$
U_{\text{eff}}(r)
=
-\frac{GM_\odot m_E}{r}
+
\frac{L^2}{2m_E r^2}.
$$

---

**Step 1 — Orbital speed**

$$
v_E \approx 2.98\times 10^4~\text{m/s}
$$

---

**Step 2 — Angular momentum**

$$
L \approx 2.66\times 10^{40}~\text{kg m}^2/\text{s}
$$

---

**Step 3 — Circular orbit radius**

$$
r_0 = \frac{L^2}{GM_\odot m_E^2}
$$

$$
r_0 \approx 1.496\times 10^{11}~\text{m}
$$

---

**Step 4 — Circular orbit energy**

$$
E_{\text{circ}} = -\frac{GM_\odot m_E}{2r_0}
$$

$$
E_{\text{circ}} \approx -2.65\times 10^{33}~\text{J}
$$

---

**Step 5 — Stability**

$$
\frac{d^2U_{\text{eff}}}{dr^2} > 0 \Rightarrow \text{stable orbit}
$$

---

**Python plot See Below**

---

**Final Results and Interpretation**

$$
v_E \approx 2.98\times 10^4~\text{m/s}
$$

$$
L \approx 2.66\times 10^{40}~\text{kg m}^2/\text{s}
$$

$$
r_0 \approx 1.496\times 10^{11}~\text{m}
$$

$$
E_{\text{circ}} \approx -2.65\times 10^{33}~\text{J}
$$

---

The minimum of the effective potential corresponds to Earth’s circular orbit.

At this radius, gravitational attraction and the centrifugal barrier balance.

Because the minimum is stable, small radial perturbations lead to oscillations rather than escape or collapse.

</div>
</details>

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Constants
G = 6.674e-11
M_sun = 1.989e30
m_E = 5.972e24
r_E = 1.496e11

# Circular orbital speed
v_E = np.sqrt(G * M_sun / r_E)

# Angular momentum
L = m_E * r_E * v_E

# Effective potential
def U_eff(r):
    return -G * M_sun * m_E / r + L**2 / (2 * m_E * r**2)

# Circular orbit
r0 = L**2 / (G * M_sun * m_E**2)
E_circ = U_eff(r0)

# Plot range
r = np.linspace(0.4*r_E, 2.5*r_E, 1500)
U = U_eff(r)

plt.figure(figsize=(9, 6))

plt.plot(r/r_E, U, label=r"$U_{\mathrm{eff}}(r)$")
plt.axhline(E_circ, linestyle="--", label="Circular orbit energy")
plt.axvline(r0/r_E, linestyle="--")

plt.scatter([r0/r_E], [E_circ], s=80, zorder=5, label="Earth circular orbit")
plt.text(r0/r_E + 0.05, E_circ, "Earth orbit\nstable equilibrium")

plt.xlabel(r"$r/r_E$")
plt.ylabel("Energy (J)")
plt.title("Effective Potential for Earth Orbiting the Sun")
plt.grid(True)
plt.legend()
plt.show()

print(f"Orbital speed v_E = {v_E:.3e} m/s")
print(f"Angular momentum L = {L:.3e} kg m^2/s")
print(f"Circular orbit radius r0 = {r0:.3e} m")
print(f"Earth orbital radius r_E = {r_E:.3e} m")
print(f"Circular orbit energy E = {E_circ:.3e} J")
```

## Problem 2: Effective Potential Near a Schwarzschild Black Hole

A test particle moves near a stationary, non-rotating black hole described by the Schwarzschild geometry.

In geometric units,

$$
G = c = M = m = 1,
$$

the effective potential for a massive particle can be written as

$$
V_{\text{eff}}(r)
=
\sqrt{
\left(1-\frac{2}{r}\right)
\left(1+\frac{\ell^2}{r^2}\right)
},
$$

where $\ell$ is the specific angular momentum of the particle.

Use

$$
\ell = 4.
$$

For a Schwarzschild black hole, the event horizon (Schwarzschild radius) is

$$
r_s = 2M
$$

in geometric units ($G = c = 1$).

Determine:

1. The Schwarzschild radius  
2. The circular orbit radii  
3. Which circular orbit is stable and which is unstable  
4. The energy values at the circular orbits  
5. A plot of $V_{\text{eff}}(r)$ showing the horizon, circular orbits, and forbidden regions for  

$$
E = 0.98.
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equations**

The effective potential is

$$
V_{\text{eff}}(r)
=
\sqrt{
\left(1-\frac{2}{r}\right)
\left(1+\frac{\ell^2}{r^2}\right)
}.
$$

Radial motion is allowed where

$$
E \ge V_{\text{eff}}(r).
$$

Circular orbits occur where

$$
\frac{dV_{\text{eff}}}{dr}=0.
$$

For Schwarzschild massive-particle orbits, this gives

$$
r^2-\ell^2 r+3\ell^2=0.
$$

---

**Step 1 — Schwarzschild radius**

In geometric units with $M=1$, the Schwarzschild radius is

$$
r_s = 2M = 2.
$$

So,

$$
\boxed{r_s = 2}
$$

---

**Step 2 — Circular orbit radii**

Using

$$
r^2-\ell^2 r+3\ell^2=0,
$$

and

$$
\ell=4,
$$

we get

$$
r^2 - 16r + 48 = 0.
$$

Solving,

$$
r =
\frac{16 \pm \sqrt{16^2-4(48)}}{2}.
$$

$$
r =
\frac{16 \pm \sqrt{64}}{2}
=
\frac{16 \pm 8}{2}.
$$

Thus,

$$
r=4
$$

and

$$
r=12.
$$

So the circular orbit radii are

$$
\boxed{r=4}
$$

and

$$
\boxed{r=12}.
$$

---

**Step 3 — Stability of the circular orbits**

Stability is determined by the curvature of the effective potential.

For circular orbits:

$$
\frac{dV_{\text{eff}}}{dr}=0.
$$

A circular orbit is:

$$
\begin{cases}
\text{stable} & \text{if } \displaystyle \frac{d^2V_{\text{eff}}}{dr^2} > 0, \\[6pt]
\text{unstable} & \text{if } \displaystyle \frac{d^2V_{\text{eff}}}{dr^2} < 0.
\end{cases}
$$

Instead of differentiating the square root directly, we can analyze

$$
V_{\text{eff}}^2(r)
=
\left(1-\frac{2}{r}\right)
\left(1+\frac{\ell^2}{r^2}\right).
$$

For $\ell=4$,

$$
V_{\text{eff}}^2(r)
=
\left(1-\frac{2}{r}\right)
\left(1+\frac{16}{r^2}\right).
$$

The circular orbits occur at

$$
r=4
\qquad \text{and} \qquad
r=12.
$$

Evaluating the curvature at these points gives:

$$
\frac{d^2}{dr^2}V_{\text{eff}}^2\bigg|_{r=4} < 0,
$$

so

$$
r=4
$$

is an **unstable circular orbit**.

Meanwhile,

$$
\frac{d^2}{dr^2}V_{\text{eff}}^2\bigg|_{r=12} > 0,
$$

so

$$
r=12
$$

is a **stable circular orbit**.

---


**Step 4 — Energies at the circular orbits**

Evaluate the effective potential at each circular orbit.

For $r=4$:

$$
V_{\text{eff}}(4)
=
\sqrt{
\left(1-\frac{2}{4}\right)
\left(1+\frac{16}{4^2}\right)
}
$$

$$
V_{\text{eff}}(4)
=
\sqrt{
\left(\frac{1}{2}\right)(2)
}
=
1.
$$

So,

$$
\boxed{E_{\text{unstable}} = 1}
$$

For $r=12$:

$$
V_{\text{eff}}(12)
=
\sqrt{
\left(1-\frac{2}{12}\right)
\left(1+\frac{16}{12^2}\right)
}.
$$

$$
V_{\text{eff}}(12)
=
\sqrt{
\left(\frac{5}{6}\right)
\left(\frac{10}{9}\right)
}
=
\sqrt{\frac{25}{27}}.
$$

Thus,

$$
\boxed{
E_{\text{stable}} =
\sqrt{\frac{25}{27}}
\approx 0.962
}
$$

---

**Python plot Below**

---

**Final results**

$$
\boxed{r_s = 2}
$$

$$
\boxed{r_{\text{unstable}} = 4}
$$

$$
\boxed{r_{\text{stable}} = 12}
$$

$$
\boxed{r_{\text{ISCO}} = 6}
$$

$$
\boxed{E_{\text{unstable}} = 1}
$$

$$
\boxed{E_{\text{stable}} \approx 0.962}
$$

---

**Interpretation**

The Schwarzschild effective potential looks similar to the Newtonian effective potential, but with an important new feature: not all circular orbits are stable.

The orbit at

$$
r=12
$$

occurs at a minimum of the effective potential, so it is stable. A small radial disturbance produces oscillations about that radius.

The orbit at

$$
r=4
$$

occurs at a maximum of the effective potential, so it is unstable. A small perturbation can cause the particle either to move outward or plunge inward toward the black hole.

The radius

$$
r=6
$$

is the **innermost stable circular orbit**, or ISCO. Stable circular orbits exist only outside this radius.

For the chosen energy

$$
E=0.98,
$$

regions where

$$
E < V_{\text{eff}}(r)
$$

are forbidden. These regions would require negative radial kinetic energy.

---

<strong>Big Picture:</strong><br>
In Newtonian gravity, stable circular orbits can occur at many radii.  
Near a black hole, general relativity changes the effective potential so that there is an innermost stable circular orbit.  
Inside this radius, circular motion becomes unstable and particles can plunge toward the black hole.

</div>
</details>

```python
import numpy as np
import matplotlib.pyplot as plt

# Geometric units: G = c = M = m = 1
ell = 4.0
E = 0.85

# Effective potential for massive particle near Schwarzschild black hole
def V_eff(r):
    return np.sqrt((1 - 2/r) * (1 + ell**2/r**2))

# Circular orbit radii for ell = 4
r_unstable = 4.0
r_stable = 12.0

E_unstable = V_eff(r_unstable)
E_stable = V_eff(r_stable)

# Radial range outside horizon
r = np.linspace(2.02, 40, 3000)
V = V_eff(r)

plt.figure(figsize=(9, 6))

# Plot effective potential
plt.plot(r, V, label=r"$V_{\mathrm{eff}}(r)$")

# Circular orbits
plt.scatter([r_unstable], [E_unstable], s=80, zorder=5, label="unstable circular orbit")
plt.scatter([r_stable], [E_stable], s=80, zorder=5, label="stable circular orbit")

plt.text(r_unstable, E_unstable + 0.01, "unstable\ncircular orbit")
plt.text(r_stable, E_stable + 0.01, "stable\ncircular orbit")

# Forbidden regions: E < V_eff
plt.fill_between(
    r, E, V,
    where=(V > E),
    alpha=0.25,
    label="Forbidden region"
)

plt.text(23, 0.90, "Forbidden", fontsize=11)

plt.xlabel(r"Radius $r/M$")
plt.ylabel(r"Effective potential $V_{\mathrm{eff}}$")
plt.title("Effective Potential Near a Schwarzschild Black Hole")
plt.ylim(0.85, 1.08)
plt.xlim(2, 40)
plt.grid(True)
plt.legend()
plt.show()

print(f"Unstable circular orbit radius = {r_unstable:.2f}")
print(f"Stable circular orbit radius = {r_stable:.2f}")
print(f"Energy at unstable orbit = {E_unstable:.3f}")
print(f"Energy at stable orbit = {E_stable:.3f}")
```
