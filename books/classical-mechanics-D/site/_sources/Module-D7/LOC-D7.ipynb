## Why this module matters

Many of the most important systems in physics—from **planetary motion** to **atomic structure**—are governed by **central forces**, where the force depends only on the distance from a central point.

These systems exhibit rich behavior:
- circular motion  
- bound orbits  
- scattering trajectories  

By combining **Newton’s laws**, **conservation of energy**, and **conservation of angular momentum**, we can reduce a complicated two-dimensional problem to an effective one-dimensional problem in the radial coordinate.

This leads to the concept of an **effective potential**, which allows us to:
- analyze motion using **energy diagrams**  
- determine **stable and unstable orbits**  
- understand **orbital motion qualitatively and quantitatively**  

This module builds directly on earlier ideas:
- energy methods (D6.5)  
- stability and oscillations (D6.2–D6.3)  

and extends them to systems of major physical importance.

By the end of this module, you will be able to:

---

### A. Central Forces and Angular Momentum  
*Describes motion under forces that depend only on distance.*

- **A1.** Define a **central force**  

$$
\mathbf{F}(r) = F(r)\,\hat{\mathbf{r}}.
$$

- **A2.** Show that central forces exert **no torque** about the origin.  
- **A3.** Demonstrate conservation of **angular momentum**  

$$
\mathbf{L} = \mathbf{r} \times m\mathbf{v}.
$$

- **A4.** Interpret motion as confined to a **plane**.

---

### B. Equations of Motion in Polar Coordinates  
*Expresses dynamics in a coordinate system aligned with motion.*

- **B1.** Write velocity and acceleration in polar coordinates.  
- **B2.** Apply Newton’s second law in radial and angular components.  
- **B3.** Identify the role of angular momentum in simplifying the motion.  

---

### C. Energy and Effective Potential  
*Reduces the motion to an equivalent one-dimensional problem.*

- **C1.** Write the total energy as  

$$
E = \frac{1}{2}m\dot{r}^2 + \frac{L^2}{2mr^2} + U(r).
$$

- **C2.** Define the **effective potential**  

$$
U_{\text{eff}}(r) = U(r) + \frac{L^2}{2mr^2}.
$$

- **C3.** Interpret the **centrifugal term** as a repulsive contribution.  
- **C4.** Analyze motion using energy diagrams.

---

### D. Orbits and Types of Motion  
*Classifies motion using energy and effective potential.*

- **D1.** Identify **turning points** from  

$$
E = U_{\text{eff}}(r).
$$

- **D2.** Distinguish between:
  - bound orbits  
  - circular orbits  
  - unbound trajectories  

- **D3.** Interpret radial motion as motion in an effective potential.  

---

### E. Circular Orbits and Stability  
*Connects orbital motion to equilibrium and oscillations.*

- **E1.** Identify circular orbits as equilibrium points of $U_{\text{eff}}(r)$.  
- **E2.** Apply the condition  

$$
\frac{dU_{\text{eff}}}{dr} = 0.
$$

- **E3.** Determine stability using  

$$
\frac{d^2U_{\text{eff}}}{dr^2}.
$$

- **E4.** Interpret stable circular motion as a minimum of the effective potential.  

---

### F. Small Oscillations About Orbits  
*Applies oscillation concepts to orbital motion.*

- **F1.** Linearize motion about a circular orbit.  
- **F2.** Show that small radial oscillations behave like SHM.  
- **F3.** Interpret orbital motion as a combination of angular motion and radial oscillations.  

---

### Performance Criteria

- Central forces are correctly identified and interpreted.  
- Angular momentum conservation is applied correctly.  
- Effective potential is constructed and used to analyze motion.  
- Energy diagrams are used to classify motion and determine turning points.  
- Circular orbits and their stability are correctly identified.  
- Connections between orbital motion and oscillations are clearly understood.
