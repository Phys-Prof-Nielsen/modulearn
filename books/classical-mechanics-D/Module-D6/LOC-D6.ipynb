## Why this module matters

Many physical systems—ranging from **masses on springs** to **molecules, circuits, and planetary motion**—exhibit oscillatory behavior. While some of these systems are simple and exactly solvable, most real systems are more complex.

A powerful idea in physics is that **near a stable equilibrium**, many systems behave approximately like **simple harmonic oscillators**, regardless of their detailed structure. This allows us to analyze complicated systems using simpler mathematical tools.

In this module, we build on familiar ideas from Physics I (such as **SHM, damping, and resonance**) and extend them using tools from classical mechanics. We introduce **linearization**, develop a geometric understanding using **phase space**, compute periods using **energy methods**, and analyze motion using **effective potentials**.

These tools form a foundation for understanding more advanced topics such as **nonlinear motion, orbital dynamics, and stability in dynamical systems**.

By the end of this module, you will be able to:

---

### A. Review of Oscillatory Motion  
*Connects prior knowledge of oscillators to stability and dynamics.*

- **A1.** Recognize and describe **simple harmonic motion (SHM)**.  
- **A2.** Interpret the effects of **damping** on oscillatory motion.  
- **A3.** Describe **driven oscillations**, including resonance behavior.  
- **A4.** Define and interpret the **quality factor (Q-factor)**.  
- **A5.** Distinguish between **stable and unstable equilibrium**.

---

### B. Linearization About Stable Points  
*Approximates complex systems near equilibrium.*

- **B1.** Identify equilibrium points from a force or potential function.  
- **B2.** Expand a force or potential using a **Taylor series** near equilibrium.  
- **B3.** Show that near a stable equilibrium, motion reduces to  

$$
m\ddot{x} \approx -kx.
$$

- **B4.** Interpret why many systems behave like **harmonic oscillators near minima**.

---

### C. Phase Space Representation  
*Describes motion geometrically in terms of position and velocity.*

- **C1.** Define **phase space** for a one-dimensional system  

$$
(x, v).
$$

- **C2.** Sketch phase-space trajectories for oscillatory motion.  
- **C3.** Distinguish between **closed orbits (bounded motion)** and **open trajectories**.  
- **C4.** Interpret how damping and driving modify phase-space behavior.

---

### D. Period from the Energy Integral  
*Uses energy conservation to determine motion.*

- **D1.** Apply conservation of energy  

$$
E = \frac{1}{2}mv^2 + U(x).
$$

- **D2.** Express velocity in terms of position and energy  

$$
v = \frac{dx}{dt}.
$$

- **D3.** Derive the period as an integral  

$$
T = 2 \int_{x_{\min}}^{x_{\max}} \frac{dx}{v(x)}.
$$

- **D4.** Interpret turning points and allowed regions of motion.

---

### E. Effective Potential and Energy Diagrams  
*Applies energy methods to more complex systems, including orbits.*

- **E1.** Interpret motion using **potential energy diagrams**.  
- **E2.** Identify **equilibrium points** as extrema of the potential.  
- **E3.** Distinguish between **stable (minima)** and **unstable (maxima)** equilibria.  
- **E4.** Use the **effective potential** to analyze radial motion in central-force problems.  
- **E5.** Interpret circular motion as equilibrium in an effective potential and describe small oscillations about it.

---

### Performance Criteria

- Oscillatory systems are correctly classified as **stable or unstable**.  
- Linearization is used appropriately to approximate motion near equilibrium.  
- Phase-space diagrams correctly represent qualitative system behavior.  
- Energy methods are used to determine **turning points and periods**.  
- Effective potential diagrams are used to interpret **motion and stability**, including orbital motion.
