## Why this module matters

In Physics I, we introduced motion through **forces, acceleration, momentum, and impulse**, including both Newton’s second law and the **impulse–momentum theorem**.

In this module, we revisit these ideas with a more unified and deeper perspective. Rather than treating these results as separate formulas, we emphasize how they are connected and how they describe motion in complementary ways.

A key idea is that **impulse represents momentum transfer**. While this was introduced previously, an important point that is often underemphasized is that only the **net impulse** acting on a particle or system determines the change in momentum.

We will also reinterpret Newton’s second law. Instead of viewing

$$
\mathbf{F} = m\mathbf{a}
$$

as the starting point, we recognize it as a special case of the more general relationship

$$
\mathbf{F}_{\text{net}} = \frac{d\mathbf{p}}{dt},
$$

which itself is the **instantaneous (local)** form of the impulse–momentum theorem.

Thus:

- **Impulse–momentum** describes how forces accumulate their effects over **time**  
- **Newton’s second law** describes the **instantaneous rate of change** of momentum  

In this module, we extend these ideas to **systems of particles**, where additional structure emerges. In particular, we will see how the total momentum of a system evolves and under what conditions it is conserved.

This refined viewpoint will be especially useful for analyzing **collisions, time-dependent forces, and systems with changing mass**, and will later connect directly to **energy methods**, where forces accumulate their effects over space rather than time.

By the end of this module, you will be able to:

---

### A. Momentum and Impulse (Review)  
*Reconnects with Physics I concepts and establishes foundational definitions.*

- **A1.** Define linear momentum  

$$
\mathbf{p} = m\mathbf{v}.
$$

- **A2.** Define impulse as the accumulation of force over time  

$$
\mathbf{J} = \int \mathbf{F}\,dt.
$$

- **A3.** Interpret impulse as **momentum transfer** due to a force acting over time.  
- **A4.** Distinguish between individual impulses and **net impulse**.

---

### B. Impulse–Momentum Theorem  
*Connects force accumulation to changes in motion.*

- **B1.** State the impulse–momentum theorem  

$$
\mathbf{J}_{\text{net}} = \Delta \mathbf{p}.
$$

- **B2.** Recognize that only the **net impulse** determines the change in momentum.  
- **B3.** Apply the theorem to compute final velocities without solving differential equations.  
- **B4.** Interpret the vector nature of momentum changes.

---

### C. Connection to Newton’s Second Law  
*Relates differential and integral descriptions of motion.*

- **C1.** Recognize Newton’s second law as  

$$
\mathbf{F}_{\text{net}} = \frac{d\mathbf{p}}{dt}.
$$

- **C2.** Derive the impulse–momentum theorem by integrating over time.  
- **C3.** Interpret Newton’s second law as the **local (instantaneous)** description of motion.  
- **C4.** Interpret impulse–momentum as the **global (accumulated)** description.

---

### D. Impulse for Time-Dependent Forces  
*Applies the impulse framework to non-constant forces.*

- **D1.** Compute impulse for forces that vary in time.  
- **D2.** Evaluate definite integrals of force over time intervals.  
- **D3.** Relate time-dependent forces directly to changes in velocity.  
- **D4.** Recognize when impulse methods simplify analysis.

---

### E. Systems and Center of Mass  
*Extends impulse–momentum to multi-particle systems.*

- **E1.** Apply impulse–momentum to the center of mass  

$$
\mathbf{F}_{\text{ext}} = \frac{d\mathbf{P}_{\text{CM}}}{dt}.
$$

- **E2.** Compute changes in system momentum using **external impulse only**  

$$
\mathbf{J}_{\text{ext}} = \Delta \mathbf{P}_{\text{CM}}.
$$

- **E3.** Distinguish between internal and external forces.  
- **E4.** Recognize the analogy to pseudo-work in energy methods.

---

### F. Momentum Methods vs Energy Methods  
*Establishes structural connections and highlights advantages.*

- **F1.** Recognize the analogy  

$$
\int \mathbf{F}\cdot d\mathbf{r}
\quad \leftrightarrow \quad
\int \mathbf{F}\,dt.
$$

- **F2.** Interpret work as accumulation over **space**, and impulse as accumulation over **time**.  
- **F3.** Distinguish between scalar (energy) and vector (momentum) descriptions.  
- **F4.** Identify situations where impulse–momentum provides a simpler solution than force-based methods.

---

### Performance Criteria

- Momentum and impulse are correctly defined and interpreted as vector quantities.  
- The distinction between **impulse** and **net impulse** is clearly understood.  
- The impulse–momentum theorem is applied correctly to compute motion.  
- Time-dependent forces are handled using integration over time.  
- Impulse–momentum is correctly applied to systems and center-of-mass motion.  
- Students recognize when **momentum methods provide a simpler approach** than Newton’s laws.
