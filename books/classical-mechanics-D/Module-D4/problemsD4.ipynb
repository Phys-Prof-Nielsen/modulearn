## Problem 1: Work Done by a 2D Force Along a 1D Path

An object moves along the horizontal line

$$
y = 2.0~\text{m}
$$

from

$$
x = 0
$$

to

$$
x = 3.0~\text{m}.
$$

A position-dependent force acts on the object:

$$
\vec{F}(x,y)
=
(4xy)\hat{\mathbf{i}}
+
(3y^2)\hat{\mathbf{j}}
\quad \text{N}.
$$

Determine:

1. The work done by the force  
2. Which component of the force contributes to the work  
3. Why the $y$-component does not contribute  

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equation**

Work is

$$
W = \int_C \vec{F}\cdot d\vec{l}.
$$

In Cartesian components,

$$
W = \int_C (F_x\,dx + F_y\,dy).
$$

---

**Step 1 — Describe the path**

The object moves along

$$
y = 2.0~\text{m}.
$$

Since $y$ is constant,

$$
dy = 0.
$$

The displacement is only in the $x$-direction, so

$$
d\vec{l} = dx\,\hat{\mathbf{i}}.
$$

---

**Step 2 — Identify the contributing force component**

The force is

$$
\vec{F}(x,y)
=
(4xy)\hat{\mathbf{i}}
+
(3y^2)\hat{\mathbf{j}}.
$$

Thus,

$$
F_x = 4xy,
\qquad
F_y = 3y^2.
$$

Along the path $y=2.0$,

$$
F_x(x,2.0) = 4x(2.0)=8x.
$$

---

**Step 3 — Set up the work integral**

Since $dy=0$,

$$
W = \int_0^3 F_x(x,2.0)\,dx.
$$

Therefore,

$$
W = \int_0^3 8x\,dx.
$$

---

**Step 4 — Evaluate**

$$
W = 8\left[\frac{x^2}{2}\right]_0^3
$$

$$
W = 4(3.0)^2
$$

$$
W = 36~\text{J}.
$$

---

**Final Result**

$$
\boxed{
W = 36~\text{J}
}
$$

---

**Interpretation**

Even though the force has both $x$- and $y$-components, the object only moves in the $x$-direction.

Because

$$
dy=0,
$$

the $y$-component contributes no work:

$$
F_y\,dy = 0.
$$

Only the component of the force along the actual displacement contributes to the work.

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
x, y = sp.symbols("x y", real=True)

# Force components
F_x = 4*x*y
F_y = 3*y**2

# Path: y = 2.0, x from 0 to 3.0
y_path = 2
x_initial = 0
x_final = 3

# Along this path
dy = 0

# --------------------------------------------------
# Display force field
# --------------------------------------------------
print("Force field:")
display(sp.Eq(sp.Symbol("F_x"), F_x))
display(sp.Eq(sp.Symbol("F_y"), F_y))

# --------------------------------------------------
# Work integral
# --------------------------------------------------
print("Work integral in Cartesian components:")
display(sp.Eq(
    sp.Symbol("W"),
    sp.Integral(F_x, x) + sp.Integral(F_y, y)
))

print("Since the path is y = constant, dy = 0:")
display(sp.Eq(sp.Symbol("dy"), dy))

# Substitute path y = 2
F_x_path = F_x.subs(y, y_path)

print("Force component along the path:")
display(sp.Eq(sp.Symbol("F_x(x,2)"), F_x_path))

# Work along path
W_integral = sp.Integral(F_x_path, (x, x_initial, x_final))

print("Work along the path:")
display(sp.Eq(sp.Symbol("W"), W_integral))

# Evaluate
W = W_integral.doit()

print("Evaluated work:")
display(sp.Eq(sp.Symbol("W"), W))

print(f"Work done by the force: {float(W):.2f} J")
```

## Problem 2: Work with Time as a Parameter

A particle moves along the $x$-axis with position

$$
x(t)=t^2+1
$$

where $x$ is measured in meters and $t$ is measured in seconds.

The force acting on the particle is given as a function of time:

$$
F(t)=3t^2+2t
$$

where $F$ is measured in newtons.

Determine the work done by the force from

$$
t=0
$$

to

$$
t=2.0~\text{s}.
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant idea**

Work in one dimension is

$$
W = \int F\,dx.
$$

But the force and position are both given in terms of time, so we use

$$
dx = \frac{dx}{dt}dt = \dot{x}\,dt.
$$

Thus,

$$
W = \int_{t_i}^{t_f} F(t)\dot{x}(t)\,dt.
$$

---

**Step 1 — Find the velocity**

$$
x(t)=t^2+1
$$

so

$$
\dot{x}(t)=2t.
$$

---

**Step 2 — Set up the work integral**

$$
W = \int_0^2 F(t)\dot{x}(t)\,dt.
$$

Substitute:

$$
W = \int_0^2 (3t^2+2t)(2t)\,dt.
$$

---

**Step 3 — Simplify the integrand**

$$
(3t^2+2t)(2t)=6t^3+4t^2.
$$

So

$$
W = \int_0^2 (6t^3+4t^2)\,dt.
$$

---

**Step 4 — Evaluate**

$$
W =
\left[
\frac{6}{4}t^4+\frac{4}{3}t^3
\right]_0^2.
$$

$$
W =
\left[
\frac{3}{2}t^4+\frac{4}{3}t^3
\right]_0^2.
$$

$$
W =
\frac{3}{2}(16)+\frac{4}{3}(8).
$$

$$
W =
24+\frac{32}{3}.
$$

$$
W =
\frac{104}{3}~\text{J}.
$$

---

**Final Result**

$$
\boxed{
W = \frac{104}{3}~\text{J} \approx 34.7~\text{J}
}
$$

---

**Interpretation**

Even though this is still one-dimensional motion, the force was not given directly as a function of position. Because both $F$ and $x$ were given in terms of time, time acted as the parameter.

This is the same idea we will use later for more general paths:

$$
\text{work integral} \quad \rightarrow \quad \text{parameterized work integral}.
$$

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Define symbols
# --------------------------------------------------
t = sp.symbols('t', real=True)

# --------------------------------------------------
# Define functions
# --------------------------------------------------
x = t**2 + 1           # position
F = 3*t**2 + 2*t       # force

# Velocity
v = sp.diff(x, t)

# --------------------------------------------------
# Display given functions
# --------------------------------------------------
print("Given functions:")
display(sp.Eq(sp.Symbol("x(t)"), x))
display(sp.Eq(sp.Symbol("F(t)"), F))

print("Velocity:")
display(sp.Eq(sp.Symbol("dx/dt"), v))

# --------------------------------------------------
# Work integral
# --------------------------------------------------
W_integral = sp.Integral(F * v, (t, 0, 2))

print("Work integral:")
display(sp.Eq(sp.Symbol("W"), W_integral))

# --------------------------------------------------
# Evaluate
# --------------------------------------------------
W = sp.simplify(W_integral.doit())

print("Evaluated work:")
display(sp.Eq(sp.Symbol("W"), W))

print(f"Work done: {float(W):.2f} J")
```

## Problem 3: Work Along a Quarter-Circle Path

Consider a force given by

$$
\vec{F} = \left(x\hat{\mathbf{i}} + y^2\hat{\mathbf{j}}\right)\ \text{N}
$$

acting on an object. The object moves clockwise from

$$
\vec{r}_i = (0,3)
$$

to

$$
\vec{r}_f = (3,0)
$$

along a quarter circle centered at $(0,0)$.

What is the work done by the force along this path?

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equation**

$$
W = \int_C \vec{F}\cdot d\vec{l}
$$

In Cartesian components,

$$
W = \int_C (F_x\,dx + F_y\,dy).
$$

---

**Step 1 — Parameterize the path**

Since the object moves clockwise from $(0,3)$ to $(3,0)$ along a quarter circle of radius $3$, we use the standard polar parameterization

$$
x = 3\cos\theta
$$

$$
y = 3\sin\theta.
$$

The initial point $(0,3)$ corresponds to

$$
\theta = \frac{\pi}{2},
$$

and the final point $(3,0)$ corresponds to

$$
\theta = 0.
$$

Therefore, because the motion is clockwise,

$$
\theta: \frac{\pi}{2} \to 0.
$$

The derivatives are

$$
\frac{dx}{d\theta} = -3\sin\theta
$$

and

$$
\frac{dy}{d\theta} = 3\cos\theta.
$$

---

**Step 2 — Express the force components**

The force is

$$
\vec{F} = x\hat{\mathbf{i}} + y^2\hat{\mathbf{j}}.
$$

Thus,

$$
F_x = x = 3\cos\theta
$$

and

$$
F_y = y^2 = 9\sin^2\theta.
$$

---

**Step 3 — Set up the work integral**

Using

$$
W =
\int_{\theta_i}^{\theta_f}
\left[
F_x(\theta)\frac{dx}{d\theta}
+
F_y(\theta)\frac{dy}{d\theta}
\right]d\theta,
$$

we get

$$
W =
\int_{\pi/2}^{0}
\left[
(3\cos\theta)(-3\sin\theta)
+
(9\sin^2\theta)(3\cos\theta)
\right]d\theta.
$$

Simplify:

$$
W =
\int_{\pi/2}^{0}
\left[
-9\sin\theta\cos\theta
+
27\sin^2\theta\cos\theta
\right]d\theta.
$$

---

**Step 4 — Evaluate**

Separate the integral:

$$
W =
-9\int_{\pi/2}^{0}\sin\theta\cos\theta\,d\theta
+
27\int_{\pi/2}^{0}\sin^2\theta\cos\theta\,d\theta.
$$

For the first integral,

$$
\int_{\pi/2}^{0}\sin\theta\cos\theta\,d\theta
=
-\frac{1}{2}.
$$

For the second integral, let

$$
u=\sin\theta,
\qquad
du=\cos\theta\,d\theta.
$$

When

$$
\theta=\frac{\pi}{2}, \quad u=1,
$$

and when

$$
\theta=0, \quad u=0.
$$

Thus,

$$
\int_{\pi/2}^{0}\sin^2\theta\cos\theta\,d\theta
=
\int_1^0 u^2\,du
=
-\frac{1}{3}.
$$

Therefore,

$$
W =
-9\left(-\frac{1}{2}\right)
+
27\left(-\frac{1}{3}\right).
$$

$$
W =
\frac{9}{2}
-
9.
$$

$$
W =
-\frac{9}{2}.
$$

---

**Final Result**

$$
\boxed{
W = -\frac{9}{2}~\text{J}
}
$$

or

$$
\boxed{
W = -4.5~\text{J}
}
$$

---

**Interpretation**

The negative result means that, along this clockwise quarter-circle path, the force removes mechanical energy from the object.

The direction of the path matters: using the correct limits,

$$
\theta:\frac{\pi}{2}\to 0,
$$

ensures the integral follows the actual clockwise motion.

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
θ = sp.symbols('θ', real=True)

# --------------------------------------------------
# Parameterization (clockwise)
# x = 3 cosθ, y = 3 sinθ
# θ: π/2 → 0
# --------------------------------------------------
x = 3*sp.cos(θ)
y = 3*sp.sin(θ)

# Derivatives
dx_dθ = sp.diff(x, θ)
dy_dθ = sp.diff(y, θ)

# --------------------------------------------------
# Force components
# F = (x, y^2)
# --------------------------------------------------
F_x = x
F_y = y**2

# --------------------------------------------------
# Display expressions
# --------------------------------------------------
print("Parameterization:")
display(sp.Eq(sp.Symbol("x(θ)"), x))
display(sp.Eq(sp.Symbol("y(θ)"), y))

print("Derivatives:")
display(sp.Eq(sp.Symbol("dx/dθ"), dx_dθ))
display(sp.Eq(sp.Symbol("dy/dθ"), dy_dθ))

print("Force components:")
display(sp.Eq(sp.Symbol("F_x"), F_x))
display(sp.Eq(sp.Symbol("F_y"), F_y))

# --------------------------------------------------
# Work integral
# --------------------------------------------------
integrand = F_x*dx_dθ + F_y*dy_dθ

print("Work integrand:")
display(sp.simplify(integrand))

W = sp.integrate(integrand, (θ, sp.pi/2, 0))

print("Work:")
display(sp.Eq(sp.Symbol("W"), sp.simplify(W)))

print(f"Work done: {float(W):.2f} J")
```

## Problem 4: Work Along an Offset Half-Circle Path

Consider a force given by

$$
\vec{F} = 2.0\left(x^3\hat{\mathbf{i}} + xy^2\hat{\mathbf{j}}\right)
$$

acting on an object. The object moves counter-clockwise from

$$
\vec{r}_i = (0,0)
$$

to

$$
\vec{r}_f = (0,2)
$$

along a half-circle centered at

$$
(x,y)=(0,1).
$$

What is the work done by the force along this path?

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Relevant equation**

$$
W = \int_C \vec{F}\cdot d\vec{l}
$$

In Cartesian components,

$$
W = \int_C (F_x\,dx + F_y\,dy).
$$

---

**Step 1 — Parameterize the path**

The half-circle has center $(0,1)$ and radius

$$
R=1.
$$

A convenient parameterization is

$$
x = \cos\theta,
$$

$$
y = 1+\sin\theta.
$$

To move counter-clockwise from $(0,0)$ to $(0,2)$ along the right half of the circle, use

$$
\theta:-\frac{\pi}{2}\to \frac{\pi}{2}.
$$

The derivatives are

$$
\frac{dx}{d\theta}=-\sin\theta,
$$

$$
\frac{dy}{d\theta}=\cos\theta.
$$

---

**Step 2 — Express the force components**

The force is

$$
\vec{F}=2.0(x^3\hat{\mathbf{i}}+xy^2\hat{\mathbf{j}}).
$$

Thus,

$$
F_x = 2.0x^3,
$$

$$
F_y = 2.0xy^2.
$$

Substitute the parameterization:

$$
F_x(\theta)=2.0\cos^3\theta,
$$

$$
F_y(\theta)=2.0\cos\theta(1+\sin\theta)^2.
$$

---

**Step 3 — Set up the work integral**

Using

$$
W =
\int_{\theta_i}^{\theta_f}
\left[
F_x(\theta)\frac{dx}{d\theta}
+
F_y(\theta)\frac{dy}{d\theta}
\right]d\theta,
$$

we get

$$
W =
\int_{-\pi/2}^{\pi/2}
\left[
(2.0\cos^3\theta)(-\sin\theta)
+
\left(2.0\cos\theta(1+\sin\theta)^2\right)(\cos\theta)
\right]d\theta.
$$

---

**Step 4 — Simplify**

$$
W =
\int_{-\pi/2}^{\pi/2}
\left[
-2.0\cos^3\theta\sin\theta
+
2.0\cos^2\theta(1+\sin\theta)^2
\right]d\theta.
$$

---

**Step 5 — Evaluate**

$$
W =
2.0\int_{-\pi/2}^{\pi/2}
\left[
-\cos^3\theta\sin\theta
+
\cos^2\theta(1+\sin\theta)^2
\right]d\theta.
$$

Expanding,

$$
W =
2.0\int_{-\pi/2}^{\pi/2}
\left[
-\cos^3\theta\sin\theta
+
\cos^2\theta(1+2\sin\theta+\sin^2\theta)
\right]d\theta.
$$

The odd terms integrate to zero over the symmetric interval, leaving

$$
W =
2.0\int_{-\pi/2}^{\pi/2}
\left[
\cos^2\theta+\cos^2\theta\sin^2\theta
\right]d\theta.
$$

Using

$$
\int_{-\pi/2}^{\pi/2}\cos^2\theta\,d\theta=\frac{\pi}{2}
$$

and

$$
\int_{-\pi/2}^{\pi/2}\cos^2\theta\sin^2\theta\,d\theta=\frac{\pi}{8},
$$

we get

$$
W =
2.0\left(
\frac{\pi}{2}+\frac{\pi}{8}
\right).
$$

$$
W =
2.0\left(\frac{5\pi}{8}\right)
=
\frac{5\pi}{4}.
$$

---

**Final Result**

$$
\boxed{
W = \frac{5\pi}{4}~\text{J}
}
$$

or

$$
\boxed{
W \approx 3.93~\text{J}
}
$$

---

**Interpretation**

The work is positive, meaning the force adds mechanical energy to the object along this counter-clockwise half-circle path.

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f
import sympy as sp

# --------------------------------------------------
# Symbolic setup
# --------------------------------------------------
θ = sp.symbols('θ', real=True)

# --------------------------------------------------
# Parameterization (counter-clockwise half-circle)
# Center (0,1), radius 1
# θ: -π/2 → π/2
# --------------------------------------------------
x = sp.cos(θ)
y = 1 + sp.sin(θ)

dx_dθ = sp.diff(x, θ)
dy_dθ = sp.diff(y, θ)

# --------------------------------------------------
# Force components
# F = 2 (x^3, x y^2)
# --------------------------------------------------
F_x = 2 * x**3
F_y = 2 * x * y**2

# --------------------------------------------------
# Display parameterization
# --------------------------------------------------
print("Parameterization:")
display(sp.Eq(sp.Symbol("x(θ)"), x))
display(sp.Eq(sp.Symbol("y(θ)"), y))

print("Derivatives:")
display(sp.Eq(sp.Symbol("dx/dθ"), dx_dθ))
display(sp.Eq(sp.Symbol("dy/dθ"), dy_dθ))

print("Force components:")
display(sp.Eq(sp.Symbol("F_x(θ)"), F_x))
display(sp.Eq(sp.Symbol("F_y(θ)"), F_y))

# --------------------------------------------------
# Work integrand
# --------------------------------------------------
integrand = sp.simplify(F_x*dx_dθ + F_y*dy_dθ)

print("Work integrand:")
display(integrand)

# --------------------------------------------------
# Work integral
# --------------------------------------------------
W = sp.integrate(integrand, (θ, -sp.pi/2, sp.pi/2))

print("Work:")
display(sp.Eq(sp.Symbol("W"), sp.simplify(W)))

print(f"Work done: {float(W):.4f} J")
```

## Problem 5: Is the Force Conservative?

Consider the force field

$$
\vec{F}(x,y) =
\left(2xy + y\right)\hat{\mathbf{i}}
+
\left(x^2 + x\right)\hat{\mathbf{j}}.
$$

Determine:

1. Whether this force is conservative  
2. If it is conservative, find a potential energy function $U(x,y)$ such that  
   $$
   \vec{F} = -\nabla U
   $$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1 — Identify components**

$$
F_x = 2xy + y,
\qquad
F_y = x^2 + x.
$$

---

**Step 2 — Apply the test for conservative forces**

In 2D, the force is conservative if

$$
\frac{\partial F_y}{\partial x}
=
\frac{\partial F_x}{\partial y}.
$$

Compute:

$$
\frac{\partial F_y}{\partial x}
=
\frac{\partial}{\partial x}(x^2 + x)
=
2x + 1
$$

$$
\frac{\partial F_x}{\partial y}
=
\frac{\partial}{\partial y}(2xy + y)
=
2x + 1.
$$

Since

$$
\frac{\partial F_y}{\partial x}
=
\frac{\partial F_x}{\partial y},
$$

the force is conservative.

---

**Step 3 — Find the potential energy**

We use

$$
F_x = -\frac{\partial U}{\partial x}.
$$

So

$$
2xy + y = -\frac{\partial U}{\partial x}.
$$

Thus,

$$
\frac{\partial U}{\partial x} = -(2xy + y).
$$

Integrate with respect to $x$:

$$
U(x,y) = -x^2y - xy + C(y).
$$

---

**Step 4 — Determine $C(y)$**

Now use

$$
F_y = -\frac{\partial U}{\partial y}.
$$

Compute

$$
\frac{\partial U}{\partial y}
=
-x^2 - x + C'(y).
$$

Thus,

$$
-\frac{\partial U}{\partial y}
=
x^2 + x - C'(y).
$$

Set equal to $F_y$:

$$
x^2 + x - C'(y) = x^2 + x.
$$

So

$$
C'(y) = 0.
$$

Thus,

$$
C(y) = C_0.
$$

---

**Final Result**

$$
\boxed{
\vec{F} \text{ is conservative}
}
$$

$$
\boxed{
U(x,y) = -x^2y - xy
}
$$

(up to an arbitrary constant)

---

**Interpretation**

This force satisfies the mixed partial condition, meaning it has no rotational component (zero curl), and therefore a scalar potential exists.

</div>
</details>

---

```python
# DIY Cell
```

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Define symbols
# --------------------------------------------------
x, y = sp.symbols('x y', real=True)

# --------------------------------------------------
# Define force field
# --------------------------------------------------
F_x = 2*x*y + y
F_y = x**2 + x

print("Force field:")
display(sp.Eq(sp.Symbol("F_x"), F_x))
display(sp.Eq(sp.Symbol("F_y"), F_y))

# --------------------------------------------------
# Check if conservative (mixed partials)
# --------------------------------------------------
dFy_dx = sp.diff(F_y, x)
dFx_dy = sp.diff(F_x, y)

print("Mixed partial derivatives:")
display(sp.Eq(sp.Symbol("∂F_y/∂x"), dFy_dx))
display(sp.Eq(sp.Symbol("∂F_x/∂y"), dFx_dy))

print("Check condition:")
display(sp.Eq(dFy_dx, dFx_dy))

# --------------------------------------------------
# Find potential U such that F = -grad(U)
# --------------------------------------------------
# From F_x = -∂U/∂x
Ux = -F_x

U = sp.integrate(Ux, x)  # integrate w.r.t x
print("Intermediate potential U(x,y):")
display(sp.Eq(sp.Symbol("U(x,y)"), U))

# Add arbitrary function of y
C = sp.Function('C')(y)
U = U + C

print("Include unknown function C(y):")
display(sp.Eq(sp.Symbol("U(x,y)"), U))

# --------------------------------------------------
# Determine C(y) using F_y
# --------------------------------------------------
Uy = sp.diff(U, y)
Fy_expr = -Uy

print("Match with F_y:")
display(sp.Eq(F_y, Fy_expr))

# Solve for C'(y)
dC_dy = sp.solve(sp.simplify(Fy_expr - F_y), sp.diff(C, y))[0]

print("C'(y):")
display(sp.Eq(sp.Symbol("C'(y)"), dC_dy))

# Integrate to get C(y)
C_sol = sp.integrate(dC_dy, y)
U_final = sp.simplify(sp.integrate(Ux, x) + C_sol)

print("Final potential:")
display(sp.Eq(sp.Symbol("U(x,y)"), U_final))

print("Check: -grad(U) = F")
display(sp.Eq(-sp.diff(U_final, x), F_x))
display(sp.Eq(-sp.diff(U_final, y), F_y))
```

## Problem 6: A Non-Conservative Force in Three Dimensions

Consider the force field

$$
\vec{F}(x,y,z)
=
(yz)\hat{\mathbf{i}}
+
(xz)\hat{\mathbf{j}}
+
(xy+z^2+x)\hat{\mathbf{k}}.
$$

Verify that this force is non-conservative.

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

**Step 1 — Identify force components**

$$
F_x = yz,
\qquad
F_y = xz,
\qquad
F_z = xy+z^2+x.
$$

---

**Step 2 — Use the mixed partial test**

For a force in three dimensions to be conservative, all three conditions must be true:

$$
\frac{\partial F_y}{\partial x}
=
\frac{\partial F_x}{\partial y}
$$

$$
\frac{\partial F_z}{\partial x}
=
\frac{\partial F_x}{\partial z}
$$

$$
\frac{\partial F_z}{\partial y}
=
\frac{\partial F_y}{\partial z}.
$$

---

**Step 3 — Check the first condition**

$$
\frac{\partial F_y}{\partial x}
=
\frac{\partial}{\partial x}(xz)
=
z
$$

$$
\frac{\partial F_x}{\partial y}
=
\frac{\partial}{\partial y}(yz)
=
z
$$

So the first condition is satisfied.

---

**Step 4 — Check the second condition**

$$
\frac{\partial F_z}{\partial x}
=
\frac{\partial}{\partial x}(xy+z^2+x)
=
y+1
$$

$$
\frac{\partial F_x}{\partial z}
=
\frac{\partial}{\partial z}(yz)
=
y
$$

Since

$$
y+1 \ne y,
$$

the second condition is not satisfied.

Therefore, the force is not conservative.

---

**Final Result**

$$
\boxed{
\vec{F} \text{ is not conservative}
}
$$

---

**Interpretation**

In three dimensions, a force must satisfy all three mixed partial conditions to be conservative.

If even one condition fails, the force cannot be written as

$$
\vec{F}=-\nabla U.
$$

Therefore, no scalar potential energy function exists for this force field.

```python
%reset -f

import sympy as sp

# --------------------------------------------------
# Define symbols
# --------------------------------------------------
x, y, z = sp.symbols("x y z", real=True)

# --------------------------------------------------
# Define force components
# F = (yz)i + (xz)j + (xy + z^2 + x)k
# --------------------------------------------------
F_x = y*z
F_y = x*z
F_z = x*y + z**2 + x

print("Force field:")
display(sp.Eq(sp.Symbol("F_x"), F_x))
display(sp.Eq(sp.Symbol("F_y"), F_y))
display(sp.Eq(sp.Symbol("F_z"), F_z))

# --------------------------------------------------
# Mixed partial tests for conservative force in 3D
# --------------------------------------------------
test_1_left  = sp.diff(F_y, x)
test_1_right = sp.diff(F_x, y)

test_2_left  = sp.diff(F_z, x)
test_2_right = sp.diff(F_x, z)

test_3_left  = sp.diff(F_z, y)
test_3_right = sp.diff(F_y, z)

print("\nMixed partial tests:")

display(sp.Eq(sp.Symbol("∂F_y/∂x"), test_1_left))
display(sp.Eq(sp.Symbol("∂F_x/∂y"), test_1_right))

display(sp.Eq(sp.Symbol("∂F_z/∂x"), test_2_left))
display(sp.Eq(sp.Symbol("∂F_x/∂z"), test_2_right))

display(sp.Eq(sp.Symbol("∂F_z/∂y"), test_3_left))
display(sp.Eq(sp.Symbol("∂F_y/∂z"), test_3_right))

# --------------------------------------------------
# Check whether all conditions are satisfied
# --------------------------------------------------
condition_1 = sp.simplify(test_1_left - test_1_right)
condition_2 = sp.simplify(test_2_left - test_2_right)
condition_3 = sp.simplify(test_3_left - test_3_right)

print("\nDifferences:")
display(sp.Eq(sp.Symbol("∂F_{y}/∂x - ∂F_{x}/∂y"), condition_1))
display(sp.Eq(sp.Symbol("∂F_{z}/∂x - ∂F_{x}/∂z"), condition_2))
display(sp.Eq(sp.Symbol("∂F_{z}/∂y - ∂F_{y}/∂z"), condition_3))

# --------------------------------------------------
# Final conclusion
# --------------------------------------------------
is_conservative = (
    condition_1 == 0 and
    condition_2 == 0 and
    condition_3 == 0
)

print("\nConclusion:")
if is_conservative:
    print("The force is conservative.")
else:
    print("The force is not conservative.")
```
