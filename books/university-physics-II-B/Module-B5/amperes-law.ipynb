# B5.4 Ampère's Law

## B5.4.1 Motivation

In the previous sections, we studied the magnetic field of moving charges and derived expressions for currents using the Biot–Savart Law. The Biot–Savart law is a powerful but computationally heavy tool: for every current distribution, we must integrate over all current elements to find the magnetic field.

But nature often gives us symmetry, and when symmetry is present, we should exploit it. Just as Gauss’s Law allowed us to avoid complicated integrations for electric fields by relating them to enclosed charge, Ampère’s Law provides a similar shortcut for magnetic fields: it relates the circulation of the magnetic field around a closed loop to the total current passing through that loop.

### Why Do We Need It?

- **Simplification**:
When symmetry is present (straight wires, solenoids, toroids), Ampère’s Law gives the magnetic field in a single step rather than complex integrations.

- **Insight into Magnetism**:
It reveals a fundamental relationship between currents and magnetic fields: moving charges (currents) are the source of magnetic fields, just as charges are the source of electric fields.

- **Foundation for Electrodynamics**:
Ampère’s Law, when combined with Faraday’s Law and Gauss’s Laws, becomes one of Maxwell’s equations, the cornerstone of classical electromagnetism.

Just as **Gauss’s Law** provides a shortcut for electric fields when symmetry is present,  
**Ampère’s Law** does the same for magnetic fields:

- **Simplification:** Quick determination of magnetic fields for symmetric configurations.  
- **Physical Insight:** Shows that **currents are the source of magnetic fields**.  
- **Foundation of Electrodynamics:** Forms one of **Maxwell’s four equations** when extended to include displacement current.

## B5.4.2 Historical Context

- **1820 – Oersted:** Observed compass deflection near a current-carrying wire, linking electricity and magnetism.  
- **1820 – Biot & Savart:** Quantified the field of a current element.  
- **1826 – André-Marie Ampère:** Systematically studied forces between current-carrying wires and formulated the early version of **Ampère’s Circuital Law**.  
- **1860s – James Clerk Maxwell:** Added the **displacement current term**, making the law consistent with charge conservation and enabling electromagnetic wave theory.

Ampère thought primarily in terms of forces between wires, not fields. Maxwell later recast Ampère’s findings in the **field formulation** we use today.

## B5.4.3 Ampère’s Law – Statement

The **integral form** for steady currents is:

$$
\oint_{\mathcal{C}} \mathbf{B} \cdot d\mathbf{l} = \mu_0 I_{\text{enc}}
$$

Where:

- $\oint_{\mathcal{C}} \mathbf{B} \cdot d\mathbf{l}$ = circulation of the magnetic field around closed loop $\mathcal{C}$  
- $I_{\text{enc}}$ = total current enclosed by the loop  
- $\mu_0 = 4 \pi \times 10^{-7} \ \text{T}\cdot\text{m/A}$ (permeability of free space)

**Interpretation:** Currents generate a curling magnetic field.

---

### Differential Form

Using Stokes’ theorem:

$$
\nabla \times \mathbf{B} = \mu_0 \mathbf{J}
$$

with $\mathbf{J}$ = current density.

This explicitly shows that magnetic fields have **curl** in the presence of current (unlike electric fields in electrostatics, which are curl-free).

---

## B5.4.4 When and How to Use Ampère’s Law

Ampère’s Law is most effective when:

- **High symmetry is present**:
  - Cylindrical (straight wire)
  - Planar (infinite current sheet)
  - Toroidal (solenoids, toroids)

- The chosen **Ampèrian loop** follows the magnetic field lines:
  - $B$ is tangential and constant along the loop.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B5.4.5 Examples

#### (a) Infinite Straight Wire

Choose a circular Ampèrian loop of radius $r$:

$$
B (2\pi r) = \mu_0 I \quad \Rightarrow \quad B = \frac{\mu_0 I}{2\pi r}
$$

---

#### (b) Infinite Solenoid

Inside:

$$
B = \mu_0 n I
$$

Outside:

$$
B \approx 0
$$

where $n$ = turns per unit length.

---

#### (c) Toroid

Inside the toroid:

$$
B (2\pi r) = \mu_0 N I \quad \Rightarrow \quad B = \frac{\mu_0 N I}{2\pi r}
$$

Outside: $B \approx 0$

## B5.4.6 Calculated Example

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;">       
<header>
<h3>Example: Magnetic Field of an Infinite Solenoid</h3>
    </header>

Use **Ampère’s Law** to determine the magnetic field **inside** and **outside** an **ideal infinite solenoid**  
carrying a steady current $I$. The solenoid has:

- $n$ = number of turns per unit length  
- Current flows in tightly wound coils along the solenoid.

---

**Step 1: Choosing an Ampèrian Loop**

We choose a rectangular Ampèrian loop:

- One side **inside** the solenoid (parallel to the axis, length $l$).  
- One side **outside** the solenoid (also length $l$).  
- Two sides perpendicular to the axis (negligible contribution because $\mathbf{B} \perp d\mathbf{l}$).

**Why this loop?**

- The field **inside** is expected to be uniform and parallel to the solenoid axis.  
- The field **outside** is expected to be negligible for an *ideal* infinite solenoid (by symmetry).

---

**Step 2: Ampère’s Law**

Start with the integral form:

$$
\oint_{\mathcal{C}} \mathbf{B} \cdot d\mathbf{l} = \mu_0 I_{\text{enc}}
$$

Break the loop into four sides:

1. **Inside (length $l$):** $\mathbf{B}$ is parallel to $d\mathbf{l}$, constant in magnitude:
   $$
   \int_{\text{inside}} \mathbf{B} \cdot d\mathbf{l} = B_{\text{inside}} l
   $$

2. **Outside (length $l$):** $\mathbf{B}$ is essentially zero (ideal solenoid assumption):
   $$
   \int_{\text{outside}} \mathbf{B} \cdot d\mathbf{l} \approx 0
   $$

3. **Perpendicular sides:** $\mathbf{B} \perp d\mathbf{l}$, so:
   $$
   \int_{\perp} \mathbf{B} \cdot d\mathbf{l} = 0
   $$

Thus:

$$
B_{\text{inside}} l = \mu_0 I_{\text{enc}}
$$

---

**Step 3: Finding $I_{\text{enc}}$**

The enclosed current is the **total current through the area enclosed by the Ampèrian loop**:

- There are $n$ turns per unit length.  
- Over length $l$, there are $n l$ turns.  
- Each turn carries current $I$.

So:

$$
I_{\text{enc}} = (n l) I
$$

---

**Step 4: Solving for $B_{\text{inside}}$**

Substitute into Ampère’s Law:

$$
B_{\text{inside}} l = \mu_0 (n l) I
$$

Cancel $l$:

$$
\boxed{B_{\text{inside}} = \mu_0 n I}
$$

---

**Step 5: Field Outside**

For an **ideal infinite solenoid**, symmetry and field-line confinement imply:

$$
B_{\text{outside}} \approx 0
$$

---

**Final Answer**

- **Inside the solenoid:**

$$
B = \mu_0 n I \quad \text{(uniform and parallel to the axis)}
$$

- **Outside the solenoid:**

$$
B \approx 0
$$

---

**Physical Interpretation**

- The field strength depends only on the **turn density** $n$ and current $I$,  
  not on the radius of the solenoid.  
- The near-zero external field is why solenoids are useful in devices that need  
  a confined magnetic field region.

## B5.4.6 Importance in Physics & Engineering

- **Magnetic device design:** Solenoids, inductors, transformers  
- **Predicting steady-state magnetic fields**  
- **Key Maxwell equation** (forms the basis of electrodynamics)  
- **Symmetry-based problem solving:** Powerful analytical tool

## B5.4.7 Looking Ahead: Ampère–Maxwell Law

For time-varying situations:

$$
\oint_{\mathcal{C}} \mathbf{B} \cdot d\mathbf{l} = \mu_0 I_{\text{enc}} + \mu_0 \varepsilon_0 \frac{d\Phi_E}{dt}
$$

- The extra term, the **displacement current**, shows that changing electric fields generate magnetic fields.
- Crucial for understanding **electromagnetic waves**.
