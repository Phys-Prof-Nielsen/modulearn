# B5.2 Magnetic Fields and the Biot–Savart Law

## B5.2.1 Motivation

In the previous note, we introduced the **magnetic field from a single moving charge**:

$$
\mathbf{B} = \frac{\mu_0}{4\pi}\frac{q \mathbf{v}\times \hat{\mathbf{r}}}{r^2}
$$

However, most real-life situations involve **continuous distributions of moving charges** (currents in wires), not isolated particles.  

To find the magnetic field from these systems, we need a more general law: the **Biot–Savart Law**

### Historical Perspective (Empirical Origin)

The Biot–Savart law was originally an **empirical law**.  

- **Jean-Baptiste Biot** and **Félix Savart** (1820s) performed experiments to measure the magnetic field around current-carrying wires.  
- They found that the magnetic field $d\mathbf{B}$ produced by a small current element $I\,d\boldsymbol{\ell}$ is:

$$
d\mathbf{B} = \frac{\mu_0}{4\pi}\frac{I\, d\boldsymbol{\ell} \times \hat{\mathbf{r}}}{r^2}
$$

- This relationship was purely **phenomenological**—a description based on experimental observation rather than derived from theory.

At that time, electricity and magnetism were not yet unified into a single theory.

---

### Modern View (Theoretical Justification)

In modern electromagnetism, the Biot–Savart law is **no longer considered fundamental**.  
Instead, it can be derived from **Maxwell’s equations** under steady-state (time-independent) conditions.

#### From Maxwell’s Equations

For steady currents:

$$
\nabla \times \mathbf{B} = \mu_0 \mathbf{J}, \qquad \nabla \cdot \mathbf{B}=0
$$

Solving these using vector calculus (with the appropriate Green’s function) yields:

$$
\mathbf{B}(\mathbf{r}) = \frac{\mu_0}{4\pi}\int \frac{\mathbf{J}(\mathbf{r}')\times(\mathbf{r}-\mathbf{r}')}{|\mathbf{r}-\mathbf{r}'|^3} \, d^3r'
$$

For a thin current-carrying wire where $\mathbf{J}d^3r' \to I\,d\boldsymbol{\ell}$, this reduces exactly to the **Biot–Savart law**:

$$
\boxed{\mathbf{B} = \frac{\mu_0}{4\pi}\int \frac{I\,d\boldsymbol{\ell}\times \hat{\mathbf{r}}}{r^2}}
$$

## B5.2.2 The Biot–Savart Law

For a steady current $I$ flowing through a small element of wire of length $d\boldsymbol{\ell}$, the magnetic field at an observation point is:

$$
d\mathbf{B} = \frac{\mu_0}{4\pi} \frac{I\, d\boldsymbol{\ell} \times \hat{\mathbf{r}}}{r^2}
$$

To find the total field:

$$
\mathbf{B} = \frac{\mu_0}{4\pi}\int \frac{I\, d\boldsymbol{\ell} \times \hat{\mathbf{r}}}{r^2}
$$

where:

- $I$ = current (assumed steady),  
- $d\boldsymbol{\ell}$ = vector along the direction of current flow,  
- $\mathbf{r}$ = vector from the current element to the observation point,  
- $r = |\mathbf{r}|$,  
- $\hat{\mathbf{r}} = \frac{\mathbf{r}}{r}$,  
- $\mu_0 = 4\pi \times 10^{-7}\ \text{T·m/A}$.

---

### ✅ **Connection to a Moving Charge**

Recall: $I\, d\ell = n q v\, d\ell$, which is essentially many moving charges summed up. Thus, the Biot–Savart law is just the continuous version of the single-charge formula.

## B5.2.3 Direction of the Magnetic Field

- Determined by the **cross product $d\boldsymbol{\ell} \times \hat{\mathbf{r}}$**.  
- The field **circulates** around the wire (right-hand rule: thumb along $I$, fingers curl in direction of $\mathbf{B}$).  

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;"> 
<header>
<h3>Example 1: Long Straight Current-Carrying Wire</h3>
    </header>

**Problem:**

Find the magnetic field a perpendicular distance $R$ from a long straight wire carrying a steady current $I$.

---

**Solution:**

1. **Set Up Geometry:**

Place the wire along the $z$-axis, observation point at distance $R$ in the $x$–$y$ plane.  

- $d\boldsymbol{\ell} = dz\, \hat{\mathbf{z}}$  
- $\mathbf{r} = R\,\hat{\mathbf{x}} - z\,\hat{\mathbf{z}}$  
- $r = \sqrt{R^2 + z^2}$  
- $\hat{\mathbf{r}} = \frac{\mathbf{r}}{r}$

---

2. **Biot–Savart Differential Form:**

$$
d\mathbf{B} = \frac{\mu_0}{4\pi}\frac{I\, d\boldsymbol{\ell}\times \hat{\mathbf{r}}}{r^2}
$$

Substitute:

- $d\boldsymbol{\ell}\times \hat{\mathbf{r}} = dz\, \hat{\mathbf{z}}\times\left(\frac{R\,\hat{\mathbf{x}} - z\,\hat{\mathbf{z}}}{r}\right)$

Since $\hat{\mathbf{z}}\times \hat{\mathbf{x}} = -\hat{\mathbf{y}}$ and $\hat{\mathbf{z}}\times \hat{\mathbf{z}}=0$:

$$
d\boldsymbol{\ell}\times \hat{\mathbf{r}} = -\frac{R\,dz}{r}\hat{\mathbf{y}}
$$

Thus:

$$
dB = \frac{\mu_0}{4\pi}\frac{I(-R\,dz)\hat{\mathbf{y}}}{r^3}
$$

(Magnitude only: $dB = \frac{\mu_0 I R}{4\pi r^3}dz$, and direction is $\pm \hat{\mathbf{y}}$)

---

3. **Integrate Along the Entire Wire:**

For an infinitely long wire: $z \in (-\infty, +\infty)$, $r = \sqrt{R^2 + z^2}$:

$$
B = \frac{\mu_0 I R}{4\pi} \int_{-\infty}^{\infty}\frac{dz}{(R^2+z^2)^{3/2}}
$$

Use:

$$
\int_{-\infty}^\infty \frac{dz}{(R^2+z^2)^{3/2}} = \frac{2}{R^2}
$$

Thus:

$$
B = \frac{\mu_0 I R}{4\pi}\cdot \frac{2}{R^2} = \frac{\mu_0 I}{2\pi R}
$$

---

**Result:**

$$
\boxed{B = \frac{\mu_0 I}{2\pi R}}
$$

Direction: **tangential, following the right-hand rule** around the wire.

<header>
<h3>Example 2: Magnetic Field at the Center of a Circular Loop</h3>
    </header>

**Problem:**

A circular loop of radius $R$ carries a steady current $I$. Find the magnetic field at the center.

---

**Solution:**

1. **Symmetry:**

At the center, every current element contributes equally, and by symmetry $\mathbf{B}$ points **perpendicular to the loop**.

---

2. **Use Biot–Savart:**

For each element:

- $r = R$,  
- $d\ell \perp r$, so $\sin\theta = 1$.

Thus:

$$
dB = \frac{\mu_0}{4\pi} \frac{I\, d\ell}{R^2}
$$

Integrate around the full circle ($\ell = 2\pi R$):

$$
B = \frac{\mu_0 I}{4\pi R^2}(2\pi R) = \frac{\mu_0 I}{2R}
$$

---

**Result:**

$$
\boxed{B = \frac{\mu_0 I}{2R}}
$$

Direction: **given by right-hand rule** (thumb along $I$).

## B5.2.6 Physical Insight

1. The **Biot–Savart law** is fundamental but not always easy to use—often replaced by **Ampère’s Law** when symmetry allows.  
2. However, Biot–Savart works for **any arbitrary current geometry**, unlike Ampère’s law.  
3. The dependence on $1/r^2$ and the cross product is directly inherited from the moving charge formulation.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

##  ✅ Key Takeaways

1. **General law for steady currents:**

   $$
   \mathbf{B} = \frac{\mu_0}{4\pi}\int \frac{I\, d\boldsymbol{\ell}\times \hat{\mathbf{r}}}{r^2}
   $$

2. Reduces to simple, elegant forms for highly symmetric systems (wire, loop).

3. Forms the basis for understanding solenoids, toroids, and general current distributions.
