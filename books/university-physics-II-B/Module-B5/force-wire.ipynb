# B5.3 Force on a Current-Carrying Wire
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B5.3.1 Motivation

We have learned that:

1. **A moving charge experiences a magnetic force**:  
   $$
   \mathbf{F} = q\, \mathbf{v} \times \mathbf{B}
   $$

2. A current is simply **many moving charges** in a conductor.  

Thus, it’s natural to ask:  
**What is the net force on a current-carrying wire placed in a magnetic field?**

This is essential for understanding **motors, magnetic levitation, and electromechanical devices**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B5.3.2 Force on a Current Element

Consider a small segment of wire carrying current $I$ in a magnetic field $\mathbf{B}$.

A small charge element $dq$ moving with drift velocity $\mathbf{v}_d$ experiences:

$$
d\mathbf{F} = dq\, \mathbf{v}_d \times \mathbf{B}
$$

Since $dq = n q\, dV$ and $I = n q v_d A$ (with $A$ cross-sectional area):

- $dq \mathbf{v}_d = (n q v_d A) d\boldsymbol{\ell} = I\, d\boldsymbol{\ell}$

Thus:

$$
\boxed{d\mathbf{F} = I\, d\boldsymbol{\ell} \times \mathbf{B}}
$$

Integrating over the whole wire:

$$
\boxed{\mathbf{F} = I \int d\boldsymbol{\ell} \times \mathbf{B}}
$$

## B5.3.3 Uniform Magnetic Field

For a straight wire of total vector length $\mathbf{L}$ in a uniform $\mathbf{B}$:

$$
\mathbf{F} = I\, \mathbf{L} \times \mathbf{B}
$$

- Magnitude: $F = I L B \sin\theta$  
- Direction: Perpendicular to both $L$ and $B$ (right-hand rule).
