## Problem 1 – Polar Form & Euler’s Formula

Write the complex number:

$$
z = -3 + 3\sqrt{3}\, i
$$

in **polar form**:

1. Find the magnitude $ r $ and argument $ \theta $.
2. Express $ z $ as $ r e^{i\theta} $ using Euler’s formula.
3. Verify that converting back to rectangular form recovers the original $ z $.

```{admonition} Problem 1: Solution
:class: tip, dropdown

The magnitude:

$$
r = |z| = \sqrt{(\text{Re}\,z)^2 + (\text{Im}\,z)^2} =
\sqrt{(-3)^2 + (3\sqrt{3})^2} = \sqrt{9 + 27} = \sqrt{36} = 6
$$

---

**Step 2: Argument (Angle)**

The argument:

$$
\theta = \arctan\left(\frac{\text{Im}\,z}{\text{Re}\,z}\right)
= \arctan\left(\frac{3\sqrt{3}}{-3}\right)
= \arctan(-\sqrt{3})
$$

Since $z$ is in the **2nd quadrant** (negative real, positive imaginary):

$$
\theta = \pi - \frac{\pi}{3} = \frac{2\pi}{3}
$$

---

**Step 3: Polar Form**

Thus:

$$
z = r e^{i\theta} = 6 e^{i\,2\pi/3}
$$

---

**Step 4: Verification**

Convert back using Euler’s formula:

$$
6 e^{i\,2\pi/3} = 6\left(\cos\frac{2\pi}{3} + i\sin\frac{2\pi}{3}\right)
= 6\left(-\frac12 + i\frac{\sqrt{3}}{2}\right)
= -3 + 3\sqrt{3}\, i
$$

✔ Verified.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 2 – Complex Conjugates & Probability Amplitudes

A quantum state has a complex probability amplitude:

$$
\psi = \frac{1}{2}\left(1 + i\sqrt{3}\right)
$$

1. Find the **complex conjugate** $\psi^*$.  
2. Compute the **probability** $P = |\psi|^2 = \psi^* \psi$.  
3. Write $\psi$ in **polar form** $r e^{i\theta}$ and interpret $r$.

```{admonition} Problem 2: Solution
:class: tip, dropdown

**Step 1: Complex Conjugate**

Change $i \to -i$:

$$
\psi^* = \frac12 \left(1 - i\sqrt{3}\right)
$$

---

**Step 2: Probability**

Multiply:

$$
|\psi|^2 = \psi^*\psi
= \left(\frac12 (1 - i\sqrt{3})\right)\left(\frac12 (1 + i\sqrt{3})\right)
$$

Use $(a-b)(a+b)=a^2-b^2$:

$$
|\psi|^2 = \frac14(1^2 - (i\sqrt{3})^2)
= \frac14(1 - (-3)) = \frac14(4) = 1
$$

Thus, the state is **normalized**.

---

**Step 3: Polar Form**

Find $r$ and $\theta$:

- Magnitude:

$$
r = |\psi| = 1
$$

- Argument:

$$
\theta = \arctan\left(\frac{\sqrt{3}}{1}\right) = \frac{\pi}{3}
$$

So:

$$
\psi = e^{i\pi/3}
$$

---

**Step 4: Interpretation**

The magnitude $r=1$ confirms this is a **normalized probability amplitude**.  
The phase $\theta = \pi/3$ does not change the total probability but affects **interference** when combined with other amplitudes.

```

## Problem 3 – Building a Wave Packet from Two Waves

Two stationary matter waves of equal amplitude have slightly different wavenumbers:

$$
\psi_1(x) = A \cos(k_1 x), \quad \psi_2(x) = A \cos(k_2 x)
$$

where:

- $k_1 = k - \Delta k/2$  
- $k_2 = k + \Delta k/2$

1. Show that their sum can be written as a **modulated wave** (a "wave packet").  
2. Identify the **envelope size** in terms of $\Delta k$.

```{admonition} Problem 3: Solution
:class: tip, dropdown

**Step 1: Add the two waves**

Use the trig identity:

$$
\cos a + \cos b = 2\cos\left(\frac{a+b}{2}\right)\cos\left(\frac{a-b}{2}\right)
$$

Substitute:

- $a = k_1 x = (k - \Delta k/2)x$  
- $b = k_2 x = (k + \Delta k/2)x$

Then:

$$
\psi_1 + \psi_2 = 2A \cos\left(\frac{(k_1+k_2)x}{2}\right)
\cos\left(\frac{(k_1-k_2)x}{2}\right)
$$

---

**Step 2: Simplify**

- Average wavenumber:

$$
\frac{k_1+k_2}{2} = \frac{(k-\Delta k/2)+(k+\Delta k/2)}{2} = k
$$

- Difference:

$$
\frac{k_1-k_2}{2} = -\frac{\Delta k}{2}
$$

So:

$$
\psi(x) = 2A\cos(k x)\cos\left(-\frac{\Delta k}{2}x\right)
= 2A\cos(k x)\cos\left(\frac{\Delta k}{2}x\right)
$$

(The cosine is even, so the minus sign drops.)

---

**Step 3: Interpretation**

- The **carrier wave**: $\cos(k x)$  
- The **envelope (modulation)**: $\cos\left(\frac{\Delta k}{2} x\right)$

The envelope width is roughly:

$$
\Delta x \sim \frac{2\pi}{\Delta k}
$$

Thus, **smaller $\Delta k$ → larger $\Delta x$**, consistent with the uncertainty principle.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 4 – Identifying Components of a Signal

You are given a stationary wave:

$$
f(x) = \cos(2x) + \cos(5x)
$$

1. What are the two basic frequency components (wavenumbers) present in this signal?  
2. If this represented a particle’s wavefunction (ignoring normalization), what does the presence of two different wavenumbers imply about the particle’s momentum?

```{admonition} Problem 4: Solution
:class: tip, dropdown

**Step 1: Frequency components**

The signal is already written as a sum of two cosines:

- One with wavenumber $k_1 = 2$  
- One with wavenumber $k_2 = 5$

Thus, $f(x)$ is a superposition of two **plane waves**.

---

**Step 2: Physical interpretation**

In quantum mechanics:

- $p = \hbar k$  
- Having **two wavenumbers** means the particle does not have a single definite momentum.

The particle has two possible momentum components:

- $p_1 = \hbar k_1 = 2\hbar$  
- $p_2 = \hbar k_2 = 5\hbar$

Thus, a more complicated (localized) wave can be thought of as many simple momentum components added together — this is the basis of **Fourier analysis**.

```

## Problem 5 – Confined Particle (Order-of-Magnitude Energy Estimate)

An electron is confined inside a one-dimensional box of width:

$$
L = 1 \times 10^{-10} \ \text{m} \quad (1\ \text{Å})
$$

Use the Heisenberg Uncertainty Principle ($\Delta x \, \Delta p \gtrsim \frac{\hbar}{2}$) to estimate the electron’s minimum kinetic energy.


```{admonition} Problem 5: Solution
:class: tip, dropdown

**Step 1: Uncertainty in position**

The electron is roughly localized to:

$$
\Delta x \sim L = 1 \times 10^{-10} \ \text{m}
$$

---

**Step 2: Uncertainty in momentum**

Heisenberg’s principle:

$$
\Delta x \, \Delta p \gtrsim \frac{\hbar}{2}
\quad \Rightarrow \quad
\Delta p \sim \frac{\hbar}{2 \Delta x}
$$

Substitute:

- $\hbar = 1.055\times10^{-34}\ \text{J·s}$

$$
\Delta p \sim \frac{1.055\times10^{-34}}{2 \times 10^{-10}}
= 5.3 \times 10^{-25}\ \text{kg·m/s}
$$

Since we are estimating the **minimum kinetic energy**, take $p \sim \Delta p$.

---

**Step 3: Kinetic energy**

For a non-relativistic particle:

$$
E_{\text{kin}} \sim \frac{p^2}{2m_e}
$$

with $m_e = 9.11 \times 10^{-31}\ \text{kg}$:

$$
E_{\text{kin}} \sim \frac{(5.3\times10^{-25})^2}{2\times9.11\times10^{-31}}
\approx 1.5 \times 10^{-19}\ \text{J}
$$

Convert to electronvolts:

$$
1\ \text{eV} = 1.6\times10^{-19}\ \text{J}
\quad \Rightarrow \quad
E_{\text{kin}} \sim 1\ \text{eV}
$$

---

**Step 4: Interpretation**

The electron’s **ground-state energy** in this box is about **1 eV**, which agrees with the scale of electron energies in atoms.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 6 – Lifetime and Energy Width

An excited atomic state has a lifetime:

$$
\Delta t = 1.0 \times 10^{-8}\ \text{s}
$$

Estimate the minimum energy uncertainty (spectral line width) of the photon it emits. Use the quantum mechanical uncertainty relationship: $\Delta E \, \Delta t \gtrsim \frac{\hbar}{2}$.

```{admonition} Problem 6: Solution
:class: tip, dropdown

**Step 1: Energy-time uncertainty**

Use:

$$
\Delta E \, \Delta t \gtrsim \frac{\hbar}{2}
\quad \Rightarrow \quad
\Delta E \sim \frac{\hbar}{2\Delta t}
$$

---

**Step 2: Substitute values**

- $\hbar = 1.055 \times 10^{-34}\ \text{J·s}$

$$
\Delta E \sim \frac{1.055\times10^{-34}}{2\times10^{-8}}
= 5.3\times10^{-27}\ \text{J}
$$

Convert to electronvolts:

$$
\Delta E = \frac{5.3\times10^{-27}}{1.6\times10^{-19}}
\approx 3.3\times10^{-8}\ \text{eV}
$$

---

**Step 3: Interpretation**

- The energy uncertainty (spectral line width) is extremely small — typical of long-lived atomic states.
- Shorter lifetimes produce broader spectral lines.

```

## Problem 7 – Normalization and Expectation Value

A particle in one dimension is described by the (unnormalized) wavefunction:

$$
\psi(x) = A \, e^{-|x|}, \quad -\infty < x < \infty
$$

1. Find the normalization constant $A$.  
2. Calculate the expectation value of $x$, $\langle x \rangle$.  
3. Determine the uncertainty (standard deviation) $\Delta x$.

```{admonition} Problem 7: Solution
:class: tip, dropdown

**Step 1: Normalization**

The condition:

$$
\int_{-\infty}^\infty |\psi(x)|^2 dx = 1
$$

Substitute $|\psi|^2 = A^2 e^{-2|x|}$:

Because of symmetry:

$$
1 = 2 A^2 \int_0^\infty e^{-2x} dx
$$

Evaluate:

$$
\int_0^\infty e^{-2x} dx = \frac{1}{2}
\quad \Rightarrow \quad
1 = 2 A^2 \left(\frac12\right) = A^2
$$

Thus:

$$
A = 1
$$

---

**Step 2: Expectation value of $x$**

By symmetry, $\langle x \rangle = 0$ (even probability density).

---

**Step 3: Uncertainty**

The variance:

$$
\Delta x^2 = \langle x^2\rangle - \langle x\rangle^2 = \langle x^2\rangle
$$

since $\langle x\rangle=0$.

Compute:

$$
\langle x^2\rangle = \int_{-\infty}^\infty x^2 |\psi|^2 dx
= 2 \int_0^\infty x^2 e^{-2x} dx
$$

Use:

$$
\int_0^\infty x^n e^{-ax} dx = \frac{n!}{a^{n+1}}
$$

For $n=2$, $a=2$:

$$
\int_0^\infty x^2 e^{-2x} dx = \frac{2!}{2^3} = \frac{2}{8} = \frac14
$$

Thus:

$$
\Delta x = \sqrt{\frac12} \approx 0.71
$$

---

**Step 4: Interpretation**

The particle is localized around $x=0$ with a typical spread of $\Delta x \approx 0.7$ (in whatever units $x$ is measured).

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 8 – Measurement and Collapse

A particle is described by the normalized wavefunction:

$$
\psi(x) =
\begin{cases}
\sqrt{\frac{2}{L}} \sin\left(\frac{\pi x}{L}\right), & 0<x<L \\
0, & \text{otherwise}
\end{cases}
$$

1. What is the probability of finding the particle between $x = L/4$ and $x = 3L/4$?  
2. If a position measurement gives $x = 3L/4$, how does the wavefunction change immediately after the measurement?

```{admonition} Problem 8: Solution
:class: tip, dropdown

**Step 1: Probability between $L/4$ and $3L/4$**

The probability:

$$
P = \int_{L/4}^{3L/4} |\psi(x)|^2 dx
= \frac{2}{L}\int_{L/4}^{3L/4}\sin^2\left(\frac{\pi x}{L}\right)dx
$$

Use $\sin^2\theta = \frac12(1-\cos 2\theta)$:

$$
P = \frac{2}{L}\int_{L/4}^{3L/4} \frac12\left(1-\cos\frac{2\pi x}{L}\right)dx
= \frac{1}{L}\left[x - \frac{L}{2\pi}\sin\frac{2\pi x}{L}\right]_{L/4}^{3L/4}
$$

Substitute limits:

At $x=3L/4$:

- $x = 3L/4$  
- $\sin(2\pi x/L)=\sin(3\pi/2)=-1$

At $x=L/4$:

- $x = L/4$  
- $\sin(2\pi x/L)=\sin(\pi/2)=1$

So:

$$
P = \frac{1}{L}\left[\left(\frac{3L}{4} + \frac{L}{2\pi}\right) -
\left(\frac{L}{4} - \frac{L}{2\pi}\right)\right]
= \frac{1}{L}\left(\frac{L}{2} + \frac{L}{\pi}\right)
= \frac12 + \frac{1}{\pi}\approx 0.82
$$

---

**Step 2: Wavefunction after measurement**

Immediately after measuring $x \approx 3L/4$:

- The wavefunction **collapses** to be sharply peaked near that point.  
- In an idealized sense, it becomes a **delta function**:

$$
\psi(x) \to \delta\left(x - \frac{3L}{4}\right)
$$

In practice, it collapses to a very narrow wave packet centered at $x=3L/4$.

---

**Step 3: Interpretation**

- Before measurement: extended probability distribution.  
- After measurement: the particle is found at a definite position, and all future predictions must use the collapsed wavefunction as the new starting point.

```

## Problem 9 – Fringe Spacing Calculation

Electrons with de Broglie wavelength:

$$
\lambda = 0.5\ \text{nm}
$$

pass through a double-slit setup:

- Slit separation: $d = 1.0\ \mu\text{m}$  
- Screen distance: $L = 1.0\ \text{m}$

1. Calculate the spacing between adjacent bright fringes on the screen.  
2. If we halve the electron’s wavelength, what happens to the fringe spacing?

```{admonition} Problem 9: Solution
:class: tip, dropdown

**Step 1: Fringe spacing formula**

For small angles:

$$
\Delta y = \frac{\lambda L}{d}
$$

---

**Step 2: Substitute values**

Convert to SI:

- $\lambda = 0.5\ \text{nm} = 0.5\times10^{-9}\ \text{m}$  
- $d = 1.0\ \mu\text{m} = 1.0\times10^{-6}\ \text{m}$

$$
\Delta y = \frac{(0.5\times10^{-9})(1.0)}{1.0\times10^{-6}}

$$

---

**Step 3: Effect of halving the wavelength**

If $\lambda \to \lambda/2$, then:

$$
\Delta y \to \frac{\lambda}{2} \frac{L}{d} = \frac{\Delta y}{2}
$$

So the fringes become **twice as closely spaced**.

---

**Step 4: Interpretation**

Decreasing wavelength (increasing particle momentum) compresses the interference pattern.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 10 – Which-Path Detection and Uncertainty

A double-slit experiment has:

- Slit separation: $d = 10\ \mu\text{m}$  
- Electrons with de Broglie wavelength: $\lambda = 1\ \text{nm}$

Suppose we try to determine **which slit** an electron passes through.

1. Using the uncertainty principle, estimate the minimum transverse momentum uncertainty $\Delta p_x$ required to localize the electron to within $d$.  
2. Compare $\Delta p_x$ to the electron’s original transverse momentum $p_x$ that produces the interference fringes.  
3. Explain why the interference pattern disappears when we try to gain which-path information.

```{admonition} Problem 10: Solution
:class: tip, dropdown

**Step 1: Uncertainty principle**

To know which slit:

$$
\Delta x \sim d \quad \Rightarrow \quad \Delta p_x \gtrsim \frac{\hbar}{\Delta x} = \frac{\hbar}{d}
$$

Substitute:

- $\hbar = 1.055\times10^{-34}\ \text{J·s}$  
- $d = 10\ \mu\text{m} = 10^{-5}\ \text{m}$

So:

$$
\Delta p_x \gtrsim \frac{1.055\times10^{-34}}{10^{-5}}
\approx 1.1\times10^{-29}\ \text{kg·m/s}
$$

---

**Step 2: Compare to fringe momentum spacing**

The electron’s momentum is:

$$
p = \frac{h}{\lambda} = \frac{6.626\times10^{-34}}{10^{-9}}
\approx 6.6\times10^{-25}\ \text{kg·m/s}
$$

But we need the **transverse momentum** spacing for fringes:

For fringe spacing:

$$
\Delta p_x \sim \frac{h}{d} = \frac{6.626\times10^{-34}}{10^{-5}}
\approx 6.6\times10^{-29}\ \text{kg·m/s}
$$

This is of the **same order** as the uncertainty we introduce by measuring the path.

---

**Step 3: Interpretation**

Because the extra momentum uncertainty $\Delta p_x$ is comparable to the fringe momentum spacing, the electron’s path is blurred enough to wash out the interference.

Thus, simply trying to determine which slit the electron goes through destroys the interference pattern.

```

## Problem 11 – Electron Double-Slit Wave Packet

Electrons are sent one at a time through a double-slit apparatus:

- Electron kinetic energy: $E = 150\ \text{eV}$  
- Slit separation: $d = 1.0\ \mu\text{m}$  
- Screen distance: $L = 1.0\ \text{m}$  

Answer the following:

1. **De Broglie wavelength**: Find the electron’s wavelength.  
2. **Fringe spacing**: Estimate the spacing between adjacent bright fringes on the screen.  
3. **Wave packet width**: Suppose the electron wave packet in $x$-space has width $\sigma_x = 0.1\ \mu\text{m}$ as it reaches the slits.  
   - What is the approximate momentum spread $\Delta p_x$ using Fourier/uncertainty reasoning?  
4. **Which-path information**: Would trying to determine which slit the electron passed through (separated by $d$) destroy the interference pattern? Compare $\Delta p_x$ to the momentum spacing between fringes.  
5. **Probability interpretation**: Explain why, even though electrons arrive **one at a time**, we still see the interference pattern build up on the screen.

```{admonition} Problem 11: Solution
:class: tip, dropdown

**1. De Broglie Wavelength**

The electron momentum:

$$
p = \sqrt{2 m_e E}, \quad m_e = 9.11\times10^{-31}\ \text{kg}
$$

Convert energy:

$$
E = 150\ \text{eV} = 150\times1.6\times10^{-19}\ \text{J} \approx 2.4\times10^{-17}\ \text{J}
$$

Then:

$$
p = \sqrt{2(9.11\times10^{-31})(2.4\times10^{-17})}
\approx 6.6\times10^{-24}\ \text{kg·m/s}
$$

The de Broglie wavelength:

$$
\lambda = \frac{h}{p} \approx \frac{6.626\times10^{-34}}{6.6\times10^{-24}}
\approx 1.0\times10^{-10}\ \text{m} \ (0.1\ \text{nm})
$$

---

**2. Fringe Spacing**

Using:

$$
\Delta y = \frac{\lambda L}{d}
$$

Substitute:

- $\lambda = 1.0\times10^{-10}\ \text{m}$  
- $L = 1.0\ \text{m}$  
- $d = 1.0\times10^{-6}\ \text{m}$

Then:

$$
\Delta y = \frac{(1.0\times10^{-10})(1.0)}{10^{-6}}
= 10^{-4}\ \text{m} = 0.1\ \text{mm}
$$

---

**3. Wave Packet Width and Momentum Spread**

Fourier/uncertainty reasoning:

$$
\Delta x \sim \sigma_x = 0.1\ \mu\text{m} = 10^{-7}\ \text{m}
$$

Thus:

$$
\Delta p_x \sim \frac{\hbar}{2 \Delta x}
= \frac{1.055\times10^{-34}}{2\times10^{-7}}
\approx 5.3\times10^{-28}\ \text{kg·m/s}
$$

---

**4. Which-Path Information**

Fringe momentum spacing:

$$
\Delta p_{\text{fringe}} \sim \frac{h}{d}
= \frac{6.626\times10^{-34}}{10^{-6}}
\approx 6.6\times10^{-28}\ \text{kg·m/s}
$$

Since:

$$
\Delta p_x \approx 5\times10^{-28}\ \text{kg·m/s} \sim \Delta p_{\text{fringe}}
$$

trying to localize the electron to one slit ($\Delta x \sim d$) introduces enough momentum uncertainty to wash out the interference pattern.  
Thus, **which-path detection destroys fringes**.

---

**5. Probability Interpretation**

- Each electron is described by a **wavefunction** that goes through **both slits** (superposition).  
- The probability distribution on the screen is:

$$
P(x) = |\psi_1 + \psi_2|^2
$$

- Sending electrons one at a time produces **individual random hits**, but over time the distribution of hits follows $P(x)$, forming the interference pattern.

This is direct evidence of the **Born interpretation**: $|\psi|^2$ gives the probability density.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 12 – Localizing an Electron and Its Consequences

An electron in a double-slit experiment initially has a well-defined momentum and produces a clear interference pattern.  
Now, imagine we attempt to localize the electron just before it reaches the slits using a very narrow light beam.

Answer the following conceptually:

1. **De Broglie and Localization**: What happens to the electron’s wave packet in position space when we localize it more tightly?  
2. **Fourier/Uncertainty Reasoning**: How does this affect the spread of possible momenta ($\Delta p$)?  
3. **Double-Slit Consequences**: What happens to the interference pattern as $\Delta p$ increases? Why?  
4. **Probability & Measurement**: Why do single electrons still land randomly on the screen even if we do not try to measure which slit they pass through?  
5. **Wavefunction Collapse**: What fundamentally changes about the electron’s wavefunction if we successfully detect which slit it passes through?

```{admonition} Problem 12: Solution
:class: tip, dropdown

**1. De Broglie and Localization**

Localizing the electron in $x$-space forces its wave packet to become **narrower**.  
A perfectly localized particle would correspond to a $\delta(x)$-like function (completely confined).

---

**2. Fourier/Uncertainty Reasoning**

A narrower wave packet in $x$-space requires a **broader range of $k$ values** to build it (Fourier analysis).  
Since $p = \hbar k$:

$$
\Delta x \downarrow \quad \Rightarrow \quad \Delta p \uparrow
$$

The electron’s momentum becomes more uncertain.

---

**3. Double-Slit Consequences**

A large $\Delta p_x$ means:

- The electron’s transverse momentum components are spread out randomly.
- The carefully ordered phase relationships that produce **interference fringes** are smeared out.

Eventually, the pattern becomes just two overlapping **single-slit diffraction envelopes**.

---

**4. Probability & Measurement**

Even when we do not measure which slit:

- Each electron is described by a **wavefunction that goes through both slits**.
- The interference pattern arises from adding **amplitudes**, not classical probabilities.
- Each electron lands randomly, but after many electrons, the distribution follows $P(x) = |\psi|^2$.

---

**5. Wavefunction Collapse**

If we successfully measure which slit:

- The electron’s superposition collapses into either $\psi_1$ or $\psi_2$.  
- Probabilities add classically:

$$
P(x) = |\psi_1|^2 + |\psi_2|^2
$$

- The interference term vanishes, and only two single-slit patterns remain.

---

**Summary**

This thought experiment connects all major Phase B ideas:

- **De Broglie waves & Fourier analysis** → wave packets and $\Delta x$–$\Delta p$ tradeoff.  
- **Heisenberg uncertainty** → why localization destroys interference.  
- **Probability & measurement** → why $|\psi|^2$ gives the pattern, even for single electrons.  
- **Collapse** → why acquiring which-path information changes the pattern fundamentally.

```

## Problem 13 – Wave Packet Through a Double Slit

An electron wave packet approaches a double-slit apparatus:

- Central de Broglie wavelength: $\lambda = 0.2\ \text{nm}$  
- Slit separation: $d = 2.0\ \mu\text{m}$  
- Screen distance: $L = 1.0\ \text{m}$  
- The wave packet in position space (as it hits the slits) can be approximated by a Gaussian function:

$$
\psi(x) = A \, e^{-x^2/(4\sigma_x^2)}, \quad \sigma_x = 0.5\ \mu\text{m}
$$

Answer the following:

1. **Normalize the wavefunction** to find $A$.  
2. **Uncertainty in position and momentum**:  
   - Find $\Delta x$.  
   - Estimate $\Delta p_x$ using the Fourier/uncertainty relation.  
3. **Check interference visibility**: Compare $\Delta p_x$ to the fringe momentum spacing $\Delta p_{\text{fringe}} \sim h/d$.  
   Will the interference fringes remain visible?  
4. **Fringe spacing**: Calculate the expected spacing $\Delta y$ on the screen.  
5. **Probability interpretation**: What fraction of electrons will pass through the slits if they are $1\ \mu\text{m}$ wide (centered at $x = \pm d/2$)?

```{admonition} Problem 13: Solution
:class: tip, dropdown

**1. Normalize the wavefunction**

The normalization condition:

$$
\int_{-\infty}^\infty |\psi(x)|^2 dx = 1
$$

Substitute $|\psi|^2 = A^2 e^{-x^2/(2\sigma_x^2)}$:

$$
1 = A^2 \sqrt{2\pi}\sigma_x \quad \Rightarrow \quad
A = \frac{1}{(2\pi\sigma_x^2)^{1/4}}
$$

Substitute $\sigma_x = 0.5\ \mu\text{m} = 5\times10^{-7}\ \text{m}$:

$$
A \approx \frac{1}{(2\pi (5\times10^{-7})^2)^{1/4}}
\approx 670 \ \text{m}^{-1/2}
$$

---

### **2. Uncertainty in Position and Momentum**

(a) Position uncertainty

For a Gaussian:

$$
\Delta x = \sigma_x = 5\times10^{-7}\ \text{m}
$$

(b) Momentum uncertainty

Using $\Delta x \Delta p_x \sim \hbar/2$:

$$
\Delta p_x \sim \frac{\hbar}{2\Delta x}
= \frac{1.055\times10^{-34}}{2\times5\times10^{-7}}
\approx 1.1\times10^{-28}\ \text{kg·m/s}
$$

---

**3. Interference Visibility**

Fringe momentum spacing:

$$
\Delta p_{\text{fringe}} \sim \frac{h}{d}
= \frac{6.626\times10^{-34}}{2\times10^{-6}}
\approx 3.3\times10^{-28}\ \text{kg·m/s}
$$

Compare:

$$
\Delta p_x \approx 1.1\times10^{-28} \ll \Delta p_{\text{fringe}}
$$

Thus, the electron’s momentum uncertainty is **small compared to fringe spacing** → **interference remains clearly visible**.

---

**4. Fringe Spacing**

Using:

$$
\Delta y = \frac{\lambda L}{d}
$$

Substitute:

- $\lambda = 0.2\ \text{nm} = 2\times10^{-10}\ \text{m}$  
- $L = 1.0\ \text{m}$  
- $d = 2.0\ \mu\text{m} = 2\times10^{-6}\ \text{m}$

Then:

$$
\Delta y = \frac{(2\times10^{-10})(1.0)}{2\times10^{-6}}
= 10^{-4}\ \text{m} = 0.1\ \text{mm}
$$

---

**5. Probability of Passing Through the Slits**

Each slit has width $w = 1\ \mu\text{m} = 10^{-6}\ \text{m}$, centered at $x = \pm d/2 = \pm 10^{-6}\ \text{m}$.

The total probability:

$$
P = 2\int_{x_0 - w/2}^{x_0 + w/2} |\psi(x)|^2 dx,
\quad x_0 = d/2 = 10^{-6}\ \text{m}
$$

Substitute $|\psi|^2 = \frac{1}{\sqrt{2\pi}\sigma_x} e^{-x^2/(2\sigma_x^2)}$:

Use symmetry (Gaussian tails small, so we can integrate directly):

Convert to dimensionless $u = x/\sqrt{2}\sigma_x$:

$$
P = 2 \times \frac{1}{\sqrt{\pi}}\int_{u_1}^{u_2} e^{-u^2} du
= 2\left[\text{erf}(u_2)-\text{erf}(u_1)\right]
$$

where:

- $u_1 = \frac{x_0 - w/2}{\sqrt{2}\sigma_x} = \frac{(10^{-6}-0.5\times10^{-6})}{\sqrt{2}(5\times10^{-7})}\approx 0.71$  
- $u_2 = \frac{x_0 + w/2}{\sqrt{2}\sigma_x} \approx 2.12$

Using $\text{erf}(0.71)\approx0.68$, $\text{erf}(2.12)\approx0.997$:

$$
P \approx 2(0.997 - 0.68) \approx 0.63
$$

So about **63% of electrons pass through the slits**.

---

**Summary**

- **Fringes visible** because $\Delta p_x \ll \Delta p_{\text{fringe}}$.  
- **Fringe spacing** $\Delta y \approx 0.1\ \text{mm}$.  
- **Probability through slits** $\approx 63\%$ of electrons.

This single problem ties together: **normalization, probability interpretation, Fourier/uncertainty reasoning, and double-slit interference physics**.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 14 – Electron Confined in a Nanowire

An electron is confined to move freely along a very narrow, 1D nanowire of width:

- Effective confinement region: $L = 5.0\ \text{nm}$

The electron is modeled as a **Gaussian wave packet** centered in the wire:

$$
\psi(x) = A \, e^{-(x-L/2)^2/(4\sigma_x^2)}, \quad 0 < x < L
$$

with $\sigma_x = 1.0\ \text{nm}$.

Answer the following:

---

1. **de Broglie wavelength & momentum**:  
   If the electron has an average kinetic energy of $E = 0.5\ \text{eV}$, find its de Broglie wavelength and average momentum.  

2. **Normalization**: Find the normalization constant $A$.  

3. **Uncertainty**:

   For the given Gaussian wave packet:

   
   - Determine $\Delta x$ and from that estimate $\Delta p$ using Heisenberg.  
   - Compare $\Delta p$ to the average momentum $p$ — does it behave more like a classical particle or a strongly quantum one?  

5. **Minimum Energy from Confinement**:  
   Estimate the minimum kinetic energy due to confinement using the uncertainty principle alone. Compare it to the given $E$.  

6. **Probability Interpretation**:  
   What is the probability of finding the electron in the middle third of the wire ($L/3 < x < 2L/3$)?



<hr style="height:2px;border-width:0;color:gray;background-color:gray">

```{admonition} Problem 14: Solution
:class: tip, dropdown

✅ **Solution**

---

**1. de Broglie Wavelength & Momentum**

Convert energy to SI:
    
$$
E = 0.5\ \text{eV} = 0.5 \times 1.6\times10^{-19} = 8.0\times10^{-20}\ \text{J}
$$

Momentum:

$$
p = \sqrt{2 m_e E}, \quad m_e = 9.11\times10^{-31}\ \text{kg}
$$

Substitute:

$$
p = \sqrt{2(9.11\times10^{-31})(8.0\times10^{-20})}
\approx 3.8\times10^{-25}\ \text{kg·m/s}
$$

de Broglie wavelength:

$$
\lambda = \frac{h}{p} = \frac{6.626\times10^{-34}}{3.8\times10^{-24}}
\approx 1.7\ \text{nm}
$$

---

**2. Normalization (Expanded)**

We need to ensure the total probability of finding the electron anywhere in the wire is **1**:

$$
\int_0^L |\psi(x)|^2 \, dx = 1
$$

Substitute the given wavefunction:

$$
|\psi(x)|^2 = A^2 e^{-\frac{(x - L/2)^2}{2\sigma_x^2}}
$$

So:

$$
1 = A^2 \int_0^L e^{-\frac{(x - L/2)^2}{2\sigma_x^2}} \, dx
$$

---

**2.1 Approximating the Integral**

Since $\sigma_x = 1\ \text{nm}$ and $L = 5\ \text{nm}$:

- The Gaussian is strongly localized near $x = L/2$.  
- The tails at $x \approx 0$ and $x \approx L$ contribute negligibly.  

Thus, we can approximate the limits to $(-\infty, +\infty)$ for easier integration (look up online):

$$
1 \approx A^2 \int_{-\infty}^{\infty} e^{-\frac{(x - L/2)^2}{2\sigma_x^2}} \, dx
$$

---

**2.2 Using the Gaussian Integral**

A standard Gaussian integral:

$$
\int_{-\infty}^\infty e^{-\frac{(x-\mu)^2}{2\sigma^2}} dx = \sqrt{2\pi}\,\sigma
$$

Here:

- $\mu = L/2$ (center of the Gaussian)  
- $\sigma = \sigma_x$

So:

$$
1 = A^2 \sqrt{2\pi}\,\sigma_x
$$

---

**2.3 Solving for $A$**

Rearrange:

$$
A = \frac{1}{(2\pi\sigma_x^2)^{1/4}}
$$

Substitute:

- $\sigma_x = 1.0\ \text{nm} = 10^{-9}\ \text{m}$

Then:

$$
A = \frac{1}{(2\pi (10^{-9})^2)^{1/4}}
\approx 1.26\times10^4\ \text{m}^{-1/2}
$$

---

**2.4 Physical Meaning**

- $A$ has units of $1/\sqrt{\text{length}}$ (as required for 1D probability densities).  
- A larger $\sigma_x$ (more spread-out packet) would make $A$ smaller because the probability is spread over a larger region.

---

**3. Uncertainty**

(a) Position uncertainty

This is the true physical width of the wave packet. For a Gaussian wave packet that is the standard deviation:

$$
\Delta x = \sigma_x = 1.0\times10^{-9}\ \text{m}
$$

(b) Momentum uncertainty

Heisenberg:

$$
\Delta p \sim \frac{\hbar}{2 \Delta x}
= \frac{1.055\times10^{-34}}{2\times10^{-9}}
\approx 5.3\times10^{-26}\ \text{kg·m/s}
$$

This momentum tells us how the "quantum smear" of the actual particle.

(c) Compare to real momentum

Ratio:

$$
\frac{\Delta p}{p} \approx \frac{5.3\times10^{-26}}{1.2\times10^{-24}}\approx 0.044
$$

So $\Delta p \ll p$ → the electron behaves **mostly classically** in this nanowire.

---

**4. Minimum Energy from Confinement**

Here, we are asked: “Even if I know nothing about the exact shape of the wavefunction, what is the minimum momentum uncertainty forced by the fact that the particle is confined to a region of width $L$? This is a purely Heisenberg-based lower bound and is not associated with the specific wave packet.

If we only know the particle is confined to $L = 5\ \text{nm}$:

$$
\Delta x \sim L \quad \Rightarrow \quad \Delta p \sim \frac{\hbar}{2L}
$$

Thus:

$$
E_{\min} \sim \frac{(\Delta p)^2}{2m_e}
= \frac{\hbar^2}{8 m_e L^2}
$$

Substitute:

- $\hbar = 1.055\times10^{-34}$  
- $m_e = 9.11\times10^{-31}$  
- $L = 5\times10^{-9}$

Then:

$$
E_{\min} \approx \frac{(1.055\times10^{-34})^2}{8(9.11\times10^{-31})(25\times10^{-18})}
\approx 6.1\times10^{-22}\ \text{J} \approx 0.004\ \text{eV}
$$

Since $E = 0.5\ \text{eV} \gg E_{\min}$ → the electron’s energy is dominated by its actual motion, not just confinement.

---

**5. Probability in the Middle Third**

Middle third:

- Lower bound: $x_1 = L/3 \approx 1.67\times10^{-9}$  
- Upper bound: $x_2 = 2L/3 \approx 3.33\times10^{-9}$

Probability:

$$
P = \int_{x_1}^{x_2} |\psi(x)|^2 dx
= \frac{1}{\sqrt{2\pi}\sigma_x}\int_{x_1}^{x_2} e^{-(x-L/2)^2/(2\sigma_x^2)}dx
$$

Substitute $u = (x-L/2)/(\sqrt{2}\sigma_x)$ and then look up the integral:

$$
P = \frac12\left[\text{erf}\left(\frac{x_2 - L/2}{\sqrt{2}\sigma_x}\right)
-\text{erf}\left(\frac{x_1 - L/2}{\sqrt{2}\sigma_x}\right)\right]
$$

- $x_2 - L/2 = +0.83\times10^{-9}$ → $u_2 \approx 0.59$  
- $x_1 - L/2 = -0.83\times10^{-9}$ → $u_1 \approx -0.59$

Use $\text{erf}(0.59)\approx0.60$, $\text{erf}(-0.59)\approx-0.60$:

$$
P \approx \frac12(0.60 - (-0.60)) \approx 0.60
$$

So about **60% of electrons are found in the middle third of the wire**.

---

✅ **Summary**

This problem integrates all Phase B quantum ideas **without using a double slit**:

1. **de Broglie relations** → $\lambda$, $p$.  
2. **Wave packets & Fourier reasoning** → $\Delta x$, $\Delta p$.  
3. **Heisenberg principle** → $E_{\min}$ from confinement.  
4. **Probability interpretation** → normalization and region probability.  
5. **Classical vs quantum** → compare $\Delta p$ to $p$.

```
