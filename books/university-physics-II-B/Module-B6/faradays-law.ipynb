# B6.2 Faraday’s Law of Electromagnetic Induction
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.2.1 Motivation

We just learned about **magnetic flux**, which measures how many magnetic field lines pass through a surface.  

But why do we care about magnetic flux?

- In the 1800s, **Michael Faraday** discovered that **a changing magnetic environment can create (induce) an electric current**.  
- This principle is the foundation of **electric generators, transformers, and almost all modern electrical power production**.

The key idea:

> **A changing magnetic flux induces an electric field, which can drive currents in conductors.**

This is the physical mechanism behind electromagnetic induction.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.2.2 Faraday’s Law (Integral Form)

### Statement

Faraday’s Law relates the **induced electromotive force (emf)** in a loop to the rate of change of magnetic flux through that loop:

$$
\boxed{\mathcal{E} = -\frac{d\Phi_B}{dt}}
$$

Where:

- $\mathcal{E}$ = induced emf (voltage) around the loop,
- $\Phi_B = \displaystyle\int_S \mathbf{B}\cdot d\mathbf{A}$ = magnetic flux through the surface bounded by the loop,
- $-\dfrac{d\Phi_B}{dt}$ = negative time rate of change of flux.

The negative sign encodes **Lenz’s Law** (explained below).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.2.3 Lenz’s Law (Direction of Induced Current)

Faraday gave the magnitude of the induced emf, but what about its **direction**?

**Lenz’s Law** tells us:

> The induced current always flows in a direction that **opposes the change** in magnetic flux.

This is a consequence of **energy conservation**:  
The induced current resists the change causing it (similar to inertia).

Examples:

1. **Increasing flux into the page** → Induced current produces a magnetic field **out of the page** to oppose the increase.  
2. **Decreasing flux into the page** → Induced current produces a magnetic field **into the page** to resist the decrease.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.2.4 Faraday’s Law (Maxwell–Faraday Equation)

In terms of electric fields, Faraday’s law becomes:

$$
\oint_{S} \mathbf{E} \cdot d\mathbf{s} = -\,\frac{d}{dt}\int_S \mathbf{B}\cdot d\mathbf{A}
$$

This shows that:

- A **changing magnetic flux** creates a **non-conservative electric field**,
- Which can exist even **without wires** (e.g., in empty space—basis of electromagnetic waves).

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.2.5 Why Faraday’s Law Matters

1. **Electric Power Generation**  
   - Rotating coils in a magnetic field changes flux → induces current → electricity!

2. **Transformers and Induction Coils**  
   - Alternating currents produce changing magnetic flux → transfer energy between circuits.

3. **Magnetic Braking & Induction Cooking**  
   - Induced currents in metals produce resistive heating or opposing forces.

4. **Fundamental to Maxwell’s Equations**  
   - Describes one of the fundamental ways electric and magnetic fields are linked.

## B6.2.6 Example

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;"> 
<header>
<h3>Example</h3>
    </header>
    
A circular loop of radius $r = 0.10\ \text{m}$ is in a uniform magnetic field increasing at a rate of $dB/dt = 0.50\ \text{T/s}$, perpendicular to the loop.

**Find the induced emf.**

---

**Solution**

1. **Magnetic flux:**

$$
\Phi_B = B A = B\,(\pi r^2)
$$

Differentiate:

$$
\mathcal{E} = -\frac{d\Phi_B}{dt} = -\frac{d}{dt}(B \pi r^2) = -\pi r^2 \frac{dB}{dt}
$$

2. **Substitute values:**

$$
\mathcal{E} = -\pi(0.10)^2 (0.50) \\
\mathcal{E} = -1.57\times10^{-2}\ \text{V}
$$

(The negative sign indicates the **direction** opposes the increasing flux.)

---

**Answer**

The magnitude of the induced emf is:

$$
|\mathcal{E}| = 15.7\ \text{mV}
$$
