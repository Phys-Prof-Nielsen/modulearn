# B6.3 Motional EMF
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.3.1 Motivation

So far, we’ve discussed **changing magnetic flux** due to varying $B$, $A$, or orientation.  

But what if a conductor **moves through a magnetic field**?  

- A moving conductor cuts through magnetic field lines, and charges inside the conductor experience a **magnetic force**.
- This generates a **potential difference** (emf) even if the magnetic field itself is constant.

This is called **Motional EMF** and is a direct, tangible application of Faraday’s law.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.3.2 The Physics of Motional EMF

### Force on Charges

A charge $q$ moving with velocity $\mathbf{v}$ in a magnetic field experiences the **Lorentz force**:

$$
\mathbf{F} = q\,\mathbf{v} \times \mathbf{B}
$$

Inside a conductor moving through a magnetic field:

- Electrons experience a sideways force and accumulate at one end, creating an **electric field $E$** and a potential difference.

Equilibrium occurs when $qE = q v B$ (for perpendicular motion).

Thus:

$$
E = v B
$$

And for a conductor of length $L$ oriented perpendicular to both $\mathbf{v}$ and $\mathbf{B}$:

$$
\mathcal{E}_{\text{motional}} = E L = v B L
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.3.3: Alternative: Motional EMF and Faraday’s Law

This result is completely consistent with Faraday’s law:

- As the rod moves, the **area** of the loop formed by the rod and rails changes with time.
- This changes the **magnetic flux**:

$$
\Phi_B = B A = B (L x)
$$

with $dx/dt = v$.

Differentiating:

$$
\mathcal{E} = -\frac{d\Phi_B}{dt} = - B L \frac{dx}{dt} = - B L v
$$

The minus sign is Lenz’s law.

## B6.3.4 Worked Example


<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;"> 
<header>
<h3>Sliding Rod on Conducting Rails</h3>
    </header>

A conducting rod of length $L = 0.30\ \text{m}$ slides to the right on two parallel conducting rails at a constant speed $v = 5.0\ \text{m/s}$. The entire setup is in a uniform magnetic field $B = 0.80\ \text{T}$ pointing **into the page**.

Find:

1. The magnitude of the motional emf in the rod.
2. The direction of the induced current if the rails are connected to form a closed circuit.

---

**Solution**

**Step 1: Use the motional emf formula**

$$
\mathcal{E} = B\,L\,v \\
\mathcal{E} = (0.80)(0.30)(5.0) = 1.2\ \text{V}
$$

---

**Step 2: Direction of the current (Lenz’s law)**

- The area of the loop **increases** as the rod moves to the right.
- Magnetic flux **into the page** increases (since $B$ is into the page).
- The induced current must **oppose the increase** → it will generate a field **out of the page**.
- By the right-hand rule, the current flows **counterclockwise**.

---

**Answer**

1. $\boxed{\mathcal{E} = 1.2\ \text{V}}$  
2. The induced current flows **counterclockwise** around the loop.

</div>

## B6.3.5 Applications

- **Electric generators**: rotating coils in a magnetic field are essentially continuous motional emf systems.
- **Railguns**: use motional emf and magnetic forces to accelerate projectiles.
- **Magnetic braking**: induced currents oppose motion.
