## B3.1.2 Work as a Line Integral

### Definition of Work

For a particle moving in a force field $\mathbf{F}(\mathbf{r})$, the **work** done by the force along a path $C$ from point $A$ to $B$ is:

$$
W_{A \to B} = \int_C \mathbf{F} \cdot d\mathbf{r}
$$

with

$$
d\mathbf{r} = dx \, \hat{\mathbf{i}} + dy \, \hat{\mathbf{j}} + dz \, \hat{\mathbf{k}},
$$


if working in cartesian coordinates.

This is a **line integral** — something we are now equipped to handle after our recent work on line integrals.

For an **electric force** on a charge $q$:

$$
\mathbf{F} = q\mathbf{E}
\quad \Rightarrow \quad
W_{A \to B} = q\int_C \mathbf{E} \cdot d\mathbf{r}.
$$

### Path Dependence?

In general, the work **could** depend on the path taken from $A$ to $B$. But for certain special forces — **conservative forces** — the work depends only on the endpoints, not the path.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.1.3 Conservative Forces

### Definition

A force field $\mathbf{F}(\mathbf{r})$ is **conservative** if any (and therefore all) of the following equivalent conditions hold:

1. **Path Independence:**

   $$
   \int_{A}^{B} \mathbf{F} \cdot d\mathbf{r} \quad \text{is independent of the path between A and B}.
   $$

2. **Zero Work Around Closed Loops:**

   $$
   \oint \mathbf{F} \cdot d\mathbf{r} = 0
   $$

   for any closed path.

3. **Existence of a Potential Function:**

   There exists a scalar function $U(\mathbf{r})$ (potential energy) such that:

   $$
   \mathbf{F}(\mathbf{r}) = -\nabla U(\mathbf{r}).
   $$

4. **Curl Condition (from vector calculus):**

   For regions where the field is continuous,

   $$
   \nabla \times \mathbf{F} = 0.
   $$

### Connection to Partial Derivatives

Using $\mathbf{F} = F_x \hat{\mathbf{i}} + F_y \hat{\mathbf{j}} + F_z \hat{\mathbf{k}}$:

If $F_x = -\frac{\partial U}{\partial x}$,  
$F_y = -\frac{\partial U}{\partial y}$,  
$F_z = -\frac{\partial U}{\partial z}$,  

then mixed partial derivative equality (Clairaut’s theorem) must hold:

$$
\frac{\partial F_x}{\partial y} = \frac{\partial F_y}{\partial x}, \quad
\frac{\partial F_x}{\partial z} = \frac{\partial F_z}{\partial x}, \quad
\frac{\partial F_y}{\partial z} = \frac{\partial F_z}{\partial y}.
$$

This is often the easiest way to check conservativeness in 2D or 3D.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.1.4 Potential Energy

For a conservative force:

$$
\mathbf{F}(\mathbf{r}) = -\nabla U(\mathbf{r}).
$$

The **potential energy difference** between points $A$ and $B$ is:

$$
\Delta U = U(B) - U(A) = - \int_{A}^{B} \mathbf{F} \cdot d\mathbf{r}.
$$

Notice the negative sign:  
- Work done **by** the force decreases potential energy.  
- Work done **against** the force increases potential energy.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.1.5 Electric Potential Energy and Electric Potential

### Electric Potential Energy

For a charge $q$ in an electric field:

$$
\mathbf{F} = q\mathbf{E} \implies
\Delta U = -\int_{A}^{B} q\mathbf{E} \cdot d\mathbf{r}.
$$

So

$$
\Delta U = q \, \Delta V
$$

if we define the **electric potential** $V$ as:

$$
\Delta V = - \int_{A}^{B} \mathbf{E} \cdot d\mathbf{r}.
$$

### Electric Potential

Thus:

$$
\mathbf{E} = - \nabla V.
$$

The electric potential is **potential energy per unit charge**:

$$
V(\mathbf{r}) = \frac{U(\mathbf{r})}{q}.
$$

### Why We Use Electric Potential

- **Scalar field**: much easier to work with than a vector field.  
- **Superposition**: potentials add algebraically.  
- **Measurements**: we measure potential differences directly with voltmeters.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.1.6 Physical Interpretation and Example

### Point Charge

For a point charge $Q$ at the origin:

1. **Electric Field**:

   $$
   \mathbf{E} = \frac{1}{4\pi\varepsilon_0} \frac{Q}{r^2}\hat{\mathbf{r}}.
   $$

2. **Potential** (choosing $V(\infty)=0$):

   $$
   V(r) = - \int_{\infty}^{r} \mathbf{E}\cdot d\mathbf{r} = 
   \frac{1}{4\pi\varepsilon_0} \frac{Q}{r}.
   $$

3. **Potential Energy** of a test charge $q$:

   $$
   U(r) = qV(r) = \frac{1}{4\pi\varepsilon_0} \frac{qQ}{r}.
   $$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B3.1.7 Conceptual Payoff

- **Conservative electric fields** mean we can switch between descriptions:
  - **Field perspective** ($\mathbf{E}$ does work on charges)
  - **Energy perspective** (charges "roll downhill" in $V$)
- **Why partial derivatives matter**: The entire concept of potential energy relies on the fact that mixed partial derivatives are equal (ensuring $\nabla \times \mathbf{E}=0$ in electrostatics).
- This framework is universal — gravity, springs, electric fields… all obey the same mathematics when conservative.

<h3>Example 1: Using the Curl Test (Clairaut’s Theorem) to Determine if a Force is Conservative</h3>
    </header>
    
**Problem**

Consider a force field in the $x$–$y$ plane:

$$
\mathbf{F}(x, y) = (3x^2y + 2)\,\hat{\mathbf{i}} + (x^3 + 4y)\,\hat{\mathbf{j}}.
$$

1. Determine whether this force is **conservative** by using **Clairaut’s theorem** (equality of mixed partial derivatives).  
2. If it is conservative, find the corresponding **potential energy function** $U(x, y)$ (up to an arbitrary constant).

---

**Solution**

**1. Test for Conservativeness**

For 2D vector fields:

$$
\text{If conservative:}\quad \frac{\partial F_x}{\partial y} = \frac{\partial F_y}{\partial x}.
$$

Compute:

- $F_x = 3x^2y + 2 \implies \dfrac{\partial F_x}{\partial y} = 3x^2$.  

- $F_y = x^3 + 4y \implies \dfrac{\partial F_y}{\partial x} = 3x^2$.  

Since:

$$
\frac{\partial F_x}{\partial y} = \frac{\partial F_y}{\partial x} = 3x^2,
$$

the force is **conservative** in the entire $x$–$y$ plane.

*(Equivalently, in vector calculus form, $\nabla \times \mathbf{F} = 0$.)*

---

**2. Finding the Potential Energy Function**

For a conservative force:

$$
\mathbf{F} = -\nabla U \implies 
\begin{cases}
F_x = -\frac{\partial U}{\partial x} \\
F_y = -\frac{\partial U}{\partial y}
\end{cases}
$$

Thus:

$$
-\frac{\partial U}{\partial x} = 3x^2y + 2 \\
\implies \frac{\partial U}{\partial x} = -(3x^2y + 2).
$$

Integrate with respect to $x$:

$$
U(x, y) = -\int (3x^2y + 2)\, dx = -x^3y - 2x + g(y),
$$

where $g(y)$ is an unknown function of $y$ only.

---

Now use the $F_y$ condition:

$$
-\frac{\partial U}{\partial y} = x^3 + 4y.
$$

But:

$$
-\frac{\partial U}{\partial y} = -\left(-x^3 - g'(y)\right) = x^3 + g'(y).
$$

Set equal to $F_y$:

$$
x^3 + g'(y) = x^3 + 4y \implies g'(y) = 4y.
$$

Integrate:

$$
g(y) = 2y^2 + C.
$$

---

**3. Final Potential Energy Function**

Substitute back:

$$
U(x, y) = -x^3y - 2x + 2y^2 + C.
$$

The constant $C$ is arbitrary and can be chosen based on reference potential energy.

---

**Answer**

1. **The force is conservative.**  
2. **Potential energy function** (up to an arbitrary constant):

$$
U(x, y) = -x^3y - 2x + 2y^2.
$$
