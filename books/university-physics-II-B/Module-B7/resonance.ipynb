## B7.9 Driven Circuits and Resonance

So far, we have studied RLC circuits that evolve freely after being set into motion.  
In many real systems, however, circuits are **continuously driven by an external source**, typically an alternating (AC) voltage.

This introduces a new question:

> How does a circuit respond when it is forced to oscillate at a particular frequency?

The answer leads to one of the most important ideas in physics and engineering: **resonance**.

---

### B7.9.1 Driven RLC Circuits

We now consider an RLC circuit connected to an AC source:

$$
V(t) = V_0 \cos(\omega t)
$$

Applying Kirchhoff’s loop rule gives:

$$
L \frac{d^2Q}{dt^2} + R \frac{dQ}{dt} + \frac{1}{C} Q = V_0 \cos(\omega t)
$$

This is no longer a homogeneous equation — it includes a **driving term**.

---

Before analyzing the full solution, it is useful to understand what kind of behavior we expect.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Forced Oscillations**

A driven circuit exhibits two types of behavior:

1. **Transient response**
   - Depends on initial conditions  
   - Decays over time due to resistance  

2. **Steady-state response**
   - Matches the driving frequency  
   - Persists indefinitely  

👉 After a long time, only the **steady-state motion** remains.

</div>

---

### B7.9.2 Steady-State Solution and Frequency Response

We focus on the steady-state behavior, where the system oscillates at the same frequency as the driving source.

We assume a solution of the form:

$$
Q(t) = Q_0 \cos(\omega t - \phi)
$$

which leads to a current:

$$
I(t) = I_0 \cos(\omega t - \phi)
$$

The key question becomes:

> How do the amplitude $I_0$ and phase $\phi$ depend on the driving frequency $\omega$?

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Impedance**

The circuit’s opposition to current in AC is called **impedance**:

$$
Z = \sqrt{R^2 + \left(\omega L - \frac{1}{\omega C}\right)^2}
$$

The current amplitude is:

$$
I_0 = \frac{V_0}{Z}
$$

👉 Unlike DC circuits, the response depends on frequency.

</div>

---

To interpret this expression, it is helpful to examine how each component contributes.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Competing Effects**

- Resistor:
  - Always opposes current  
  - Independent of frequency  

- Inductor:
  - Opposes high-frequency changes  
  - Contribution increases with $\omega$  

- Capacitor:
  - Opposes low-frequency changes  
  - Contribution decreases with $\omega$  

👉 The system’s behavior depends on the **balance** between these effects.

</div>

---

### B7.9.3 Resonance

As the driving frequency changes, the current amplitude changes. A particularly important condition occurs when the response is maximized.

---

Before defining it mathematically, consider the physical idea.

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Curiosity — When Does the System Respond Most Strongly?**

If energy is supplied at just the right rate, it can reinforce the system’s natural motion.

👉 This leads to a maximum response called **resonance**.

</div>

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Resonance Condition**

Resonance occurs when:

$$
\omega L = \frac{1}{\omega C}
$$

Solving:

$$
\omega_0 = \frac{1}{\sqrt{LC}}
$$

At this frequency:

$$
Z = R
$$

and the current amplitude is maximized:

$$
I_0 = \frac{V_0}{R}
$$

</div>

---

### B7.9.4 Phase Relationship

In addition to amplitude, the phase difference between voltage and current is important.

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Phase Angle**

The phase angle is given by:

$$
\tan \phi = \frac{\omega L - \frac{1}{\omega C}}{R}
$$

This leads to three regimes:

- Low frequency ($\omega \ll \omega_0$):
  - Capacitor dominates  
  - Current leads voltage  

- High frequency ($\omega \gg \omega_0$):
  - Inductor dominates  
  - Current lags voltage  

- At resonance ($\omega = \omega_0$):
  - $\phi = 0$  
  - Current and voltage are in phase  

</div>

---

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — Resonance Is Not Infinite**

In an ideal circuit with no resistance:

- $Z \to 0$ at resonance  
- Current would become arbitrarily large  

In real systems:

- Resistance limits the amplitude  
- Energy is continuously dissipated  

👉 Resonance produces a **maximum response**, not infinite response.

</div>

---

### B7.9.5 Bandwidth and Selectivity

Resonance is not just about a single frequency — it also determines how sharply the system responds to nearby frequencies.

---

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Bandwidth and Quality Factor**

The **bandwidth** $\Delta \omega$ measures the range of frequencies over which the system responds strongly.

The **quality factor** is:

$$
Q = \frac{\omega_0}{\Delta \omega}
$$

Interpretation:

- Large $Q$:
  - Sharp resonance  
  - Weak damping  

- Small $Q$:
  - Broad response  
  - Strong damping  

</div>

---

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Real-World Systems**

- Radios:
  - Select a narrow range of frequencies (high $Q$)

- Audio systems:
  - Broader response (lower $Q$)

- Sensors:
  - Use resonance to detect small signals  

👉 Resonance allows systems to **select and amplify specific frequencies**.

</div>

---

### B7.9.6 Big Picture

Driven RLC circuits reveal a deep principle:

> Systems respond most strongly when driven at their natural frequency.

This idea extends far beyond circuits:

- Mechanical resonance (bridges, buildings)  
- Optical resonance (lasers, cavities)  
- Quantum systems (energy transitions)  

Understanding resonance provides a unified framework for analyzing how systems **absorb, store, and transfer energy** in response to external driving.
