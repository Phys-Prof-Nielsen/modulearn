## B7.1 RC Circuits — Charging and Discharging

In earlier modules, we introduced capacitors as devices that store energy in electric fields. We now place a capacitor into a **complete circuit** with a resistor and a source, allowing us to study how circuits **evolve in time**.

An RC circuit is the simplest system that combines:
- **Energy storage** (capacitor)
- **Energy dissipation** (resistor)

This combination leads to time-dependent behavior that is fundamentally different from static circuits.

---

### B7.1.1 Circuit Structure and Governing Equation

To analyze an RC circuit, we combine the relationships for capacitors and resistors with Kirchhoff’s loop rule.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — What Is an RC Circuit?**

An RC circuit consists of:
- A resistor ($R$)
- A capacitor ($C$)
- A source of EMF (for charging)

When the circuit is closed, charge begins to flow and the capacitor **gradually accumulates charge**.

Key relationships:

$$
V_C = \frac{Q}{C}, \qquad V_R = IR
$$

Applying Kirchhoff’s loop rule:

$$
\mathcal{E} = IR + \frac{Q}{C}
$$

</div>

---

### B7.1.2 Physical Picture of Charging

Before solving the equations, it is important to understand the physical behavior of the system as it evolves in time.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Physical Picture of Charging**

When the switch is first closed:

- The capacitor is initially **uncharged** ($Q = 0$)
- The voltage across the capacitor is zero
- Current is **maximum**

As time progresses:

- Charge accumulates on the capacitor plates  
- The capacitor voltage increases  
- The current gradually **decreases**

Eventually:

- The capacitor behaves like an **open circuit**  
- Current goes to zero  

</div>

---

### B7.1.3 Exponential Behavior

Solving the governing equation reveals a characteristic type of time evolution.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Exponential Behavior**

Solving the circuit equation leads to:

- Charging:

$$
Q(t) = Q_{\text{max}} \left(1 - e^{-t/RC}\right)
$$

$$
I(t) = \frac{\mathcal{E}}{R} e^{-t/RC}
$$

- Discharging:

$$
Q(t) = Q_0 e^{-t/RC}
$$

These expressions show that RC circuits evolve **exponentially in time**.

</div>

---

### B7.1.4 Why RC Circuits Do Not Oscillate

At this stage, it is important to understand a fundamental limitation of RC circuits.

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — No Oscillations**

RC circuits **cannot oscillate**.

- There is only **one energy storage mechanism** (the capacitor)
- Energy flows in one direction:
  - Stored in the capacitor  
  - Dissipated in the resistor  

👉 Once energy begins to dissipate, it cannot return

Result:

$$
\text{All behavior is exponential decay toward equilibrium}
$$

</div>

---

### B7.1.5 The Time Constant

The exponential behavior is governed by a single parameter that sets the timescale of the system.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — The Time Constant**

A key parameter of RC circuits is the **time constant**:

$$
\tau = RC
$$

This determines how quickly the system evolves:

- At $t = \tau$:
  - Charging: $Q \approx 0.63\, Q_{\text{max}}$
  - Discharging: $Q \approx 0.37\, Q_0$

👉 Larger $R$ or $C$ → slower response  
👉 Smaller $R$ or $C$ → faster response  

</div>

---

### B7.1.6 Energy Flow in RC Circuits

Understanding how energy moves through the circuit provides deeper physical insight.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Energy Flow**

Initially:
- Energy is supplied by the source

As the capacitor charges:
- Energy is stored in the electric field:

$$
U_C = \frac{1}{2} C V^2
$$

At the same time:
- Energy is continuously **dissipated** in the resistor as thermal energy

👉 Not all supplied energy is stored — some is always lost

</div>

### B7.1.7 Checking Understanding

These activities help reinforce the physical interpretation of the circuit behavior.

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Early-Time Behavior**

Right after closing the switch:

1. What is the charge on the capacitor?  
2. What is the voltage across the capacitor?  
3. Is the current large or small?  
4. Where is the energy primarily going?

<details>
<summary>Solution</summary>

1. $Q \approx 0$  
2. $V_C \approx 0$  
3. Current is large (maximum)  
4. Energy is being delivered by the source, with some stored and some dissipated  

</details>

</div>

---

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 2 — Long-Time Behavior**

After a long time:

1. What happens to the current?  
2. What is the voltage across the capacitor?  
3. Is energy still being dissipated in the resistor?  

<details>
<summary>Solution</summary>

1. Current goes to zero  
2. $V_C = \mathcal{E}$  
3. No — with no current, no power is dissipated  

</details>

</div>

---

RC circuits introduce a crucial idea:

> Circuits can **evolve in time**, but with only one energy storage mechanism, they cannot oscillate.

In the next section, we will explore how **magnetic effects of currents** lead to a second energy storage mechanism, making oscillations possible.
