# B4.2 Gauss’s Law in Dielectrics

## B4.2.1 Motivation

In vacuum, Gauss’s law relates electric flux to total charge enclosed:

$$
\oint_S \mathbf{E} \cdot d\mathbf{A} = \frac{Q_{\text{enc}}}{\varepsilon_0}.
$$

But in materials, the situation changes:

- **The dielectric weakens the electric field.**  
- The relationship between field and enclosed charge must be modified to account for the material.

We introduce the **electric displacement field** $\mathbf{D}$ to make Gauss’s law easier to apply with dielectrics.

## B4.2.2 The Electric Displacement Field

In a linear, uniform dielectric:

$$
\mathbf{D} \equiv \varepsilon \mathbf{E} = \kappa \varepsilon_0 \mathbf{E}
$$

where:

- $\varepsilon = \kappa \varepsilon_0$ = permittivity of the dielectric,  
- $\kappa$ = dielectric constant,  
- $\mathbf{E}$ = electric field inside the dielectric.

Using $\mathbf{D}$, **Gauss’s law in dielectrics** becomes:

$$
\oint_S \mathbf{D} \cdot d\mathbf{A} = Q_{\text{free}}
$$

Only the **free charge** (charge placed on conductors or introduced externally) appears in this law.  
The effect of the dielectric is already built into $\varepsilon$ (or $\mathbf{D}$).

---

### **Important Notes**

1. If the dielectric is uniform and fills the region completely, you can simply replace $\varepsilon_0$ with $\varepsilon$ in the usual Gauss’s law:

$$
\oint_S \mathbf{E} \cdot d\mathbf{A} = \frac{Q_{\text{free}}}{\varepsilon}
$$

2. If different dielectrics are present, use $\mathbf{D}$ because it stays continuous across boundaries (except where free surface charge is present).

## B4.2.3 Using Gauss’s Law in Dielectrics

### **Steps**

1. **Choose a Gaussian surface** based on symmetry (spherical, cylindrical, planar).  
2. **Write Gauss’s law in terms of $\mathbf{D}$:**

$$
\oint_S \mathbf{D} \cdot d\mathbf{A} = Q_{\text{free}}.
$$

3. Solve for $\mathbf{D}$ first:

$$
D = \frac{Q_{\text{free, enclosed}}}{\text{surface area}}.
$$

4. Convert to $\mathbf{E}$ using:

$$
\mathbf{E} = \frac{\mathbf{D}}{\varepsilon}.
$$

## B4.2.4 Examples

### Spherical Charge in a Dielectric

**Problem:**  
A point charge $+Q$ is placed at the center of a large uniform dielectric with permittivity $\varepsilon$.  
Find the electric field at distance $r$.

**Solution:**

1. Symmetry → choose a spherical Gaussian surface of radius $r$.  
2. Gauss’s law with $\mathbf{D}$:

$$
\oint_S \mathbf{D} \cdot d\mathbf{A} = Q_{\text{free}} \\
D (4\pi r^2) = Q \\
D = \frac{Q}{4\pi r^2}
$$

3. Convert to $E$:

$$
E = \frac{D}{\varepsilon} = \frac{Q}{4\pi \varepsilon r^2} = \frac{Q}{4\pi \kappa \varepsilon_0 r^2}
$$

So the dielectric **reduces** the electric field by a factor of $\kappa$ compared to vacuum.

---

### Parallel-Plate Capacitor Filled with Dielectric

**Problem:**  
A parallel-plate capacitor is filled completely with a dielectric of permittivity $\varepsilon$.  
Find the electric field between the plates for free charge $+Q$ on the top plate.

**Solution:**

1. Gaussian surface: a rectangular box enclosing one plate (area $A$).  
2. Gauss’s law with $\mathbf{D}$:

$$
\oint_S \mathbf{D} \cdot d\mathbf{A} = Q_{\text{free}} \\
D A = Q \\
D = \frac{Q}{A}
$$

3. Convert to $E$:

$$
E = \frac{D}{\varepsilon} = \frac{Q}{\varepsilon A}
$$

4. Voltage between plates ($d =$ separation):

$$
V = E d = \frac{Q d}{\varepsilon A}
$$

5. Capacitance:

$$
C = \frac{Q}{V} = \frac{\varepsilon A}{d} = \kappa \frac{\varepsilon_0 A}{d}
$$

as expected.
