## B2.1.1 Motivation

So far, we learned about fields: gravitational fields and electric fields.  
Often, we are not just interested in the field at a single point but in how much of this field "passes through" a surface.

- **Examples to motivate the concept:**
    - **Water analogy:** Imagine a rain gauge that detects the amount of rain passing a sensor. The gauge do not collect the water, simply measure certain amount passing through. The sensor sits inside a cavity (often a vertical tube). On a non-windy day, the maximum amount of water measured is when the tube is vertical and zero water is measured if the tube is horizontal.   
    - Similarly, a field can be thought of as a "flow" of lines through an area.
    
> **Gravitational or electric fields:** The number of field lines passing through a surface tells us how much "source" (mass or charge) is enclosed.

This leads us to the concept of **flux**, which is fundamental to understanding **Gauss’s law**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.1.2 What is Flux?

**Intuitive Definition:**

Flux measures how much of a vector field "flows through" a given surface.

- A large flux means many field lines pass through the surface.  
- If the field is tangent to the surface, no lines pass through — **flux = 0**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.1.3 Mathematical Definition (Differential Form)

Consider a small surface element $dA$ with an associated **normal vector** $\mathbf{\hat{n}}$ (perpendicular to the surface).

### Differential Flux

For a vector field $\mathbf{F}$ (this could be $\mathbf{g}$ for gravitational field or $\mathbf{E}$ for electric field):

$$
d\Phi = \mathbf{F} \cdot d\mathbf{A}
$$

where:

- $d\mathbf{A} = \mathbf{\hat{n}}\,dA$  
- $dA$ is the scalar area of the infinitesimal patch  
- $\mathbf{\hat{n}}$ is the **unit normal** pointing outward (for closed surfaces)  
- The dot product naturally encodes **how tilted** the field is relative to the surface.

Expanding the dot product:

$$
d\Phi = F\,dA\,\cos\theta
$$

with $\theta$ being the angle between $\mathbf{F}$ and the surface normal.

---

### Total Flux

The total flux through a finite surface $S$ is obtained by summing all the differential contributions:

$$
\Phi = \iint_S \mathbf{F} \cdot d\mathbf{A}
$$

This is the **surface integral** form.  

- For **open surfaces**: $S$ is just the surface of interest, and the orientation is chosen by convention.  
- For **closed surfaces**: The normal vector points **outward** by convention.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.1.4 Visualizing Flux

### Orientation and Sign

- If the field lines pass **outward**, the flux is **positive**.
- If the field lines pass **inward** (opposite to the chosen normal), the flux is **negative**.

Think of the flux as "net number of field lines going out."

---

### Special Cases

1. **Constant Field Perpendicular to Surface:**

   $$
   \Phi = F \cdot A
   $$


2. **Field Parallel to Surface:**

   $$
   \Phi = 0
   $$

   since $\cos 90^\circ = 0$.

3. **Tilted Surface with constant force and orientation:**

   $$
   \Phi = F\,A\,\cos\theta
   $$

## B2.1.5 Flux Through a Curved Surface

For a non-flat surface, $d\mathbf{A}$ changes direction at each point. Example: flux through a sphere.

For a **radial field** $\mathbf{F} = k\,\frac{\mathbf{\hat{r}}}{r^2}$:

- Magnitude of $F$: $F = k/r^2$  
- $d\mathbf{A} = \mathbf{\hat{r}}\,dA$ (points radially outward on a sphere)

Thus:

$$
\Phi = \iint_S F\,dA = \iint_S \frac{k}{r^2}dA = \frac{k}{r^2}\,(4\pi r^2)=4\pi k
$$

The flux is **independent of radius** — a key observation leading to Gauss’s law.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B2.1.6 Connection to Gauss’s Law

- For gravitational field:

$$
\oint_S \mathbf{g}\cdot d\mathbf{A} = -4\pi G\,M_{\text{enc}}
$$

- For electric field:

$$
\oint_S \mathbf{E}\cdot d\mathbf{A} = \frac{Q_{\text{enc}}}{\varepsilon_0}
$$

## B2.1.7 Summary

- Flux quantifies how much of a field passes through a surface.
- Orientation and closed vs. open surfaces matter.
- The concept naturally leads to **Gauss’s law**, where flux is directly proportional to the enclosed source (mass or charge).
