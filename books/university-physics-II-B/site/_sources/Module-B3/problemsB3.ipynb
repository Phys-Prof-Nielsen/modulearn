## Problem 1: Verifying a Force Field is Conservative and Finding its Potential

A biomedical device called **magnetic tweezers** creates force fields to manipulate cells. A simplified model of the force (in N) in the $x$–$y$ plane is:

$$
\mathbf{F}(x, y) = (2xy)\,\hat{\mathbf{i}} + (x^2 + y)\,\hat{\mathbf{j}}.
$$

1. Determine whether this force field is **conservative** using **Clairaut’s theorem**.  
2. If it is conservative, find the corresponding **potential energy function** $U(x, y)$ (up to an arbitrary constant).

```{admonition} Problem 1: Solution
:class: tip, dropdown

**Solution**

**1. Check Conservativeness (Clairaut’s Theorem)**

For 2D fields:

$$
\text{Conservative if: } \frac{\partial F_x}{\partial y} = \frac{\partial F_y}{\partial x}.
$$

Compute:

- $F_x = 2xy \implies \dfrac{\partial F_x}{\partial y} = 2x$.  
- $F_y = x^2 + y \implies \dfrac{\partial F_y}{\partial x} = 2x$.

Since they are equal, the field is **conservative**.

---

**2. Find Potential Energy Function**

For conservative fields:

$$
\mathbf{F} = -\nabla U \implies
\begin{cases}
F_x = -\frac{\partial U}{\partial x} \\
F_y = -\frac{\partial U}{\partial y}
\end{cases}
$$

From $F_x$:

$$
-\frac{\partial U}{\partial x} = 2xy \implies
\frac{\partial U}{\partial x} = -2xy.
$$

Integrate w.r.t. $x$:

$$
U(x, y) = -x^2y + g(y).
$$

Differentiate w.r.t. $y$ and compare with $F_y$:

$$
-\frac{\partial U}{\partial y} = -(-x^2 - g'(y)) = x^2 + g'(y) = F_y = x^2 + y.
$$

Thus:

$$
g'(y)=y \implies g(y)=\frac{1}{2}y^2 + C.
$$

---

**Final Answer**

The potential energy function is:

$$
U(x, y) = -x^2y + \frac{1}{2}y^2 + C
$$

(up to an arbitrary constant $C$).

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 2: Determining if a Force Field is Conservative

A charged particle moves in a force field in the $x$–$y$ plane given by:

$$
\mathbf{F}(x, y) = (-y)\,\hat{\mathbf{i}} + (x)\,\hat{\mathbf{j}}.
$$

This type of field can arise in charged particles moving in a magnetic field.

1. Determine whether the field is conservative.  
2. Physically interpret what your result means for **work done around closed paths**.


```{admonition} Problem 2: Solution
:class: tip, dropdown

**Solution**

**1. Conservativeness Test**

Use Clairaut’s theorem:

- $F_x = -y \implies \frac{\partial F_x}{\partial y} = -1$  
- $F_y = x \implies \frac{\partial F_y}{\partial x} = 1$

Since:

$$
\frac{\partial F_x}{\partial y} \neq \frac{\partial F_y}{\partial x},
$$

the field is **NOT conservative**.

Alternatively, $\nabla \times \mathbf{F} \neq 0$.

---

**2. Physical Interpretation**

Because the field is non-conservative:

$$
\oint \mathbf{F}\cdot d\mathbf{r} \neq 0.
$$

The work done around a closed loop depends on the path, typical of **magnetic forces**, which do not have a potential energy function.

---

**Final Answer**

- The field is **non-conservative**.  
- **Work around closed paths is nonzero** — energy can circulate (as in magnetic induction).

```

## Problem 3: Deriving Electric Potential from a Non-Uniform Electric Field

Inside a charged vacuum tube, the electric field is:

$$
\mathbf{E}(x, y) = (4x)\,\hat{\mathbf{i}} + (2y)\,\hat{\mathbf{j}} \quad [\text{N/C}].
$$

1. Verify the electric field is conservative.  
2. Find the **electric potential** $V(x, y)$ (take $V=0$ at the origin).  
3. Find the **potential difference** $V(2,3) - V(0,0)$.

```{admonition} Problem 3: Solution
:class: tip, dropdown

**Solution**

**1. Conservativeness Test**

For electrostatics, we expect:

$$
\frac{\partial E_x}{\partial y} = \frac{\partial E_y}{\partial x}.
$$

- $E_x=4x \implies \frac{\partial E_x}{\partial y}=0$.  
- $E_y=2y \implies \frac{\partial E_y}{\partial x}=0$.

They are equal → field is **conservative**.

---

**2. Find $V(x, y)$**

By definition:

$$
\mathbf{E} = -\nabla V \implies
\begin{cases}
E_x = -\frac{\partial V}{\partial x} \\
E_y = -\frac{\partial V}{\partial y}
\end{cases}
$$

From $E_x$:

$$
- \frac{\partial V}{\partial x} = 4x \implies
\frac{\partial V}{\partial x} = -4x.
$$

Integrate w.r.t. $x$:

$$
V(x, y) = -2x^2 + h(y).
$$

Now use $E_y$:

$$
- \frac{\partial V}{\partial y} = 2y \implies
-\left(h'(y)\right)=2y \implies h'(y)=-2y.
$$

Integrate:

$$
h(y) = -y^2 + C.
$$

Thus:

$$
V(x, y) = -2x^2 - y^2 + C.
$$

Take $C=0$ for $V(0,0)=0$.

---

**3. Potential Difference**

At $(2,3)$:

$$
V(2,3) = -2(2)^2 - (3)^2 = -8 -9 = -17\,\text{V}.
$$

Thus:

$$
V(2,3) - V(0,0) = -17\,\text{V}.
$$

Point $(2,3)$ is at **lower potential**.

---

**Final Answer**

- **Electric potential**: $V(x, y) = -2x^2 - y^2$.  
- **Potential difference**: $V(2,3) = -17\,\text{V}$.

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 4: Work Done Along Two Different Paths (Demonstrating Path Independence)

The electric field between two charged plates is:

$$
\mathbf{E}(x, y) = (6x)\,\hat{\mathbf{i}} + (0)\,\hat{\mathbf{j}} \quad [\text{N/C}].
$$

A $+2.0\,\mu\text{C}$ charge moves from $A(0,0)$ to $B(2,3)$.

1. Show that the work is **independent of path** by calculating it along:
   - Path 1: straight line from $A$ to $B$  
   - Path 2: move first along $x$-axis, then along $y$-axis  
2. Find the change in electric potential energy.

```{admonition} Problem 4: Solution
:class: tip, dropdown

### **Solution**

**1. Work Along Any Path**

Work:

$$
W = q\int_A^B \mathbf{E}\cdot d\mathbf{r} = q\int_A^B (E_x dx + E_y dy).
$$

Since $E_y=0$:

$$
W = q\int_A^B E_x dx = q\int_{x=0}^{x=2} 6x \, dx = q[3x^2]_0^2 = q(12).
$$

Regardless of path, only $x$ changes matter.

Substitute:

$$
W = (2\times10^{-6})(12)=2.4\times10^{-5}\,\text{J}.
$$

---

**2. Change in Potential Energy**

$$
\Delta U = -W = -2.4\times10^{-5}\,\text{J}.
$$

Potential energy decreases as charge moves to higher $x$.

---

**Final Answer**

- **Work (any path)**: $2.4\times10^{-5}\,\text{J}$  
- **Change in potential energy**: $-2.4\times10^{-5}\,\text{J}$

```

## Problem 5: Deriving Potential Energy Function from a Conservative Force in 3D

A particle moves in a force field:

$$
\mathbf{F}(x,y,z) = (10x)\,\hat{\mathbf{i}} + (10y)\,\hat{\mathbf{j}} + (10z)\,\hat{\mathbf{k}} \quad [\text{N}].
$$

1. Show the force is **conservative**.  
2. Find the **potential energy function** $U(x, y, z)$ with $U(0,0,0)=0$.  
3. Find the **work done** by the force when moving from $(0,0,0)$ to $(2,1,3)$.

```{admonition} Problem 5: Solution
:class: tip, dropdown

**Solution**

**1. Check Conservativeness**

Conservative if:

$$
\frac{\partial F_x}{\partial y} = \frac{\partial F_y}{\partial x}, \quad
\frac{\partial F_x}{\partial z} = \frac{\partial F_z}{\partial x}, \quad
\frac{\partial F_y}{\partial z} = \frac{\partial F_z}{\partial y}.
$$

All are zero → **conservative**.

---

**2. Find Potential Energy Function**

$\mathbf{F} = -\nabla U$:

- $F_x = -\frac{\partial U}{\partial x}=10x \implies \frac{\partial U}{\partial x} = -10x$.  
Integrate:

$$
U(x,y,z) = -5x^2 + f(y,z).
$$

- $F_y = -\frac{\partial U}{\partial y}=10y \implies
-\frac{\partial f}{\partial y}=10y \implies f(y,z)=-5y^2+g(z).$

- $F_z = -\frac{\partial U}{\partial z}=10z \implies
-\frac{dg}{dz}=10z \implies g(z)=-5z^2 + C.$

With $U(0,0,0)=0 \implies C=0$.

Thus:

$$
U(x,y,z)=-5x^2 -5y^2 -5z^2.
$$

---

**3. Work Done**

For conservative forces:

$$
W = -\Delta U = -[U(B)-U(A)].
$$

At $B(2,1,3)$:

$$
U_B=-5(4+1+9)=-70\,\text{J}, \quad U_A=0
$$

Thus:

$$
W = -(-70-0)=+70\,\text{J}.
$$

The force does positive work as the particle moves away from the origin.

---

**Final Answer**

- **Potential energy**: $U=-5x^2 -5y^2 -5z^2$  
- **Work** (0,0,0 → 2,1,3): $70\,\text{J}$

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Problem 6: Work Along a Curved Path in an Electric Field (Parameterizing a Path)

A charged particle ($q = +1.0\,\mu\text{C}$) moves in an electric field created by a long charged rod.  
In the region of interest, the electric field is approximately:

$$
\mathbf{E}(x, y) = \frac{k}{x^2 + y^2}\,(x\,\hat{\mathbf{i}} + y\,\hat{\mathbf{j}})
$$

where $k = 200\,\text{N}\cdot\text{m}^2/\text{C}$.  

The particle moves from point $A(1,0)$ to point $B(0,1)$ **along a quarter-circle path** of radius $r=1\,\text{m}$, parameterized as:

$$
x(\theta) = \cos\theta,\quad y(\theta)=\sin\theta,\quad 0\le \theta\le\frac{\pi}{2}.
$$

1. **Compute the work done by the electric field** along this path using the **line integral explicitly parameterized**.  
2. **Find the potential difference** $V_B - V_A$ and compare it to the work (check for path independence).

```{admonition} Problem 6: Solution
:class: tip, dropdown

**Solution**

---

**1. Work Along the Parameterized Path**

Work:

$$
W = q\int_C \mathbf{E}\cdot d\mathbf{r} = q\int_{\theta=0}^{\pi/2} \mathbf{E}(x(\theta), y(\theta)) \cdot \frac{d\mathbf{r}}{d\theta}\,d\theta
$$

---

**(a) Compute $d\mathbf{r}$**

$$
d\mathbf{r} = dx\,\hat{\mathbf{i}} + dy\,\hat{\mathbf{j}}
= \frac{dx}{d\theta}d\theta\hat{\mathbf{i}} + \frac{dy}{d\theta}d\theta\hat{\mathbf{j}}.
$$

With $x(\theta)=\cos\theta, y(\theta)=\sin\theta$:

- $\frac{dx}{d\theta} = -\sin\theta$  
- $\frac{dy}{d\theta} = \cos\theta$

So:

$$
d\mathbf{r} = (-\sin\theta\,\hat{\mathbf{i}} + \cos\theta\,\hat{\mathbf{j}})\,d\theta.
$$

---

**(b) Substitute $\mathbf{E}$**

The electric field:

$$
\mathbf{E} = \frac{k}{x^2 + y^2}(x\,\hat{\mathbf{i}} + y\,\hat{\mathbf{j}}).
$$

On the path, $x^2 + y^2 = (\cos^2\theta + \sin^2\theta)=1$ (constant radius!).  

Thus:

$$
\mathbf{E} = k(\cos\theta\,\hat{\mathbf{i}} + \sin\theta\,\hat{\mathbf{j}}).
$$

---

**(c) Dot product**

$$
\mathbf{E}\cdot d\mathbf{r} = k(\cos\theta\,\hat{\mathbf{i}} + \sin\theta\,\hat{\mathbf{j}})\cdot(-\sin\theta\,\hat{\mathbf{i}} + \cos\theta\,\hat{\mathbf{j}})d\theta
$$

Compute:

$$
= k(-\cos\theta\sin\theta + \sin\theta\cos\theta)\,d\theta = 0.
$$

Thus:

$$
W = q\int_0^{\pi/2}0\,d\theta = 0.
$$

**No work is done** along the circular arc.

---

**2. Potential Difference $V_B - V_A$**

For a conservative electric field:

$$
V_B - V_A = -\int_A^B \mathbf{E}\cdot d\mathbf{r}.
$$

But we already computed $W = q\int \mathbf{E}\cdot d\mathbf{r}=0$, so:

$$
\Delta V = V_B - V_A = -\frac{W}{q}=0.
$$

Thus, points $A$ and $B$ are at the **same potential** — expected because they lie on the same circle of radius $r=1$ around the charge distribution (equipotential).

---

**Final Answer**

1. **Work**: $W=0\,\text{J}$ (no work along the arc).  
2. **Potential difference**: $V_B - V_A =0\,\text{V}$.  
3. **Physical Meaning**: The particle moved along an **equipotential path** — consistent with electrostatics.

```

## Problem 7: Work Along a Non-Radial Path in a Conservative Electric Field

A particle with charge $q = +2.0\,\mu\text{C}$ moves in an electric field created by a point charge.  
In the region of interest, the electric field can be approximated as:

$$
\mathbf{E}(x, y) = \frac{k}{(x^2 + y^2)^{3/2}}(x\,\hat{\mathbf{i}} + y\,\hat{\mathbf{j}}),
$$

where $k = 9.0\times10^9\,\text{N}\cdot\text{m}^2/\text{C}$.  

The particle moves from $A(1,0)$ to $B(0,2)$ along the **straight-line path** parameterized by:

$$
x(t)=1- t,\quad y(t)=2t,\quad 0\le t\le1.
$$

1. **Compute the work** done by the electric field along this path via the parameterized line integral.  
2. **Find the potential difference** $V_B - V_A$ using the potential function (showing path independence).  
3. Confirm the two results are consistent.

```{admonition} Problem 6: Solution
:class: tip, dropdown

**Solution**

---

**1. Parameterized Work Calculation**

The work is:

$$
W = q\int_C \mathbf{E}\cdot d\mathbf{r}
= q\int_0^1 \mathbf{E}(x(t), y(t))\cdot\frac{d\mathbf{r}}{dt}\,dt.
$$

---

**(a) Compute $d\mathbf{r}/dt$**

From $x(t), y(t)$:

- $dx/dt = -1$  
- $dy/dt = 2$

Thus:

$$
d\mathbf{r} = (-\hat{\mathbf{i}} + 2\hat{\mathbf{j}})\, dt.
$$

---

**(b) Substitute Electric Field Components**

On the path:

- $x(t)=1-t$  
- $y(t)=2t$  
- $r(t)=\sqrt{x^2+y^2} = \sqrt{(1-t)^2+(2t)^2}$

So:

$$
\mathbf{E}(t) = \frac{k}{r^3}(x\,\hat{\mathbf{i}}+y\,\hat{\mathbf{j}})=
\frac{k}{\left[(1-t)^2+4t^2\right]^{3/2}}\big((1-t)\hat{\mathbf{i}}+2t\hat{\mathbf{j}}\big).
$$

---

**(c) Dot Product**

Compute:

$$
\mathbf{E}\cdot d\mathbf{r} = \frac{k}{\left[(1-t)^2+4t^2\right]^{3/2}}\Big((1-t)(-1) + (2t)(2)\Big)dt
$$

Simplify numerator:

$$
(-1+t +4t) = (-1 +5t).
$$

So:

$$
\mathbf{E}\cdot d\mathbf{r} = \frac{k(-1+5t)}{\left[(1-t)^2+4t^2\right]^{3/2}}\, dt.
$$

---

**(d) Work Integral**

Thus:

$$
W = q\int_0^1 \frac{k(-1+5t)}{\left[(1-t)^2+4t^2\right]^{3/2}}\, dt.
$$

Substitute $q=2.0\times10^{-6}$, $k=9.0\times10^9$ later; first evaluate the integral.

Let’s evaluate analytically.

---

**(e) Analytical Simplification**

Notice $r(t) = \sqrt{(1-t)^2+4t^2}$ and:

$$
\frac{dr}{dt} = \frac{(1-t)(-1) + (2)(2t)}{r} = \frac{-1+t+4t}{r} = \frac{-1+5t}{r}.
$$

So:

$$
\frac{-1+5t}{r^3} = \frac{1}{r^2}\frac{dr}{dt}.
$$

Therefore:

$$
W = qk\int_0^1 \frac{1}{r^2}\frac{dr}{dt}\, dt = qk\int_{r(0)}^{r(1)}\frac{dr}{r^2}.
$$

At endpoints:

- $r(0)=\sqrt{1^2+0^2}=1$  
- $r(1)=\sqrt{0^2+2^2}=2$

Integrate:

$$
\int \frac{dr}{r^2} = -\frac{1}{r}
\implies W = qk\left[-\frac{1}{r}\right]_1^2 = qk\left(-\frac{1}{2}+\frac{1}{1}\right)=qk\left(\frac{1}{2}\right).
$$

Substitute:

$$
W = (2.0\times10^{-6})(9.0\times10^9)(0.5)\approx9.0\times10^3\,\text{J}.
$$

---

**2. Potential Difference (Path Independence Check)**

The electric potential for this field (point charge-like):

$$
V(r) = \frac{k}{r}.
$$

So:

$$
V_B - V_A = \frac{k}{r_B}-\frac{k}{r_A}= \frac{k}{2}-\frac{k}{1}=-\frac{k}{2}.
$$

Thus:

$$
\Delta U = q(V_B - V_A) = q\left(-\frac{k}{2}\right)=-qk\frac{1}{2}.
$$

And:

$$
W = -\Delta U = +qk\frac{1}{2},
$$

exactly matching the work we found.

---

**Final Answer**

1. **Work (explicit path integral)**: $W \approx 9.0\times10^3\,\text{J}$.  
2. **Potential difference**: $V_B - V_A = -k/2$, consistent with $W$.  
3. **Interpretation**: Even though we integrated along a **non-radial straight path**, the result depends only on endpoints (conservative field).

```

<hr style="height:2px;border-width:0;color:gray;background-color:gray">
