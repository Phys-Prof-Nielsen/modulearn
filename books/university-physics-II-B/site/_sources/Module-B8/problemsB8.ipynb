## Problem 1: Bradley’s Stellar Aberration

The Earth moves in its orbit around the Sun at $v = 30\,\text{km/s}$ ($\beta = v/c$).  
Consider a star that is located **perpendicular to the Earth's velocity vector** in the **star’s rest frame**.

We define our coordinate system as:

- The **+x direction** is along Earth’s velocity (the direction of motion of the observer in the star’s frame).  
- $\theta$ is measured **from the +x direction** in the **star’s rest frame**.

Thus, the star is at:

- $\theta = 90^\circ$ (true position in the source frame).  
- The observer is moving **in the +x direction**, as assumed in the standard aberration formula.

**Find:**

1. The apparent angle $\theta'$ in the observer’s frame.  
2. The angular shift in arcseconds.

```python
import numpy as np

# Given
v = 30e3
c = 3e8
beta = v/c
theta = np.deg2rad(90)

gamma = 1/np.sqrt(1-beta**2)
theta_prime = np.arccos( (np.cos(theta)-beta)/((1 - beta*np.cos(theta))) )

theta_prime_deg = np.rad2deg(theta_prime)
angular_shift_arcsec = abs(theta_prime_deg - 90) * 3600

theta_prime_deg, angular_shift_arcsec
```

## Problem 3: Approaching Star – Relativistic Doppler Shift

A star emits light with a rest wavelength of $\lambda_{\text{source}} = 500\,\text{nm}$.  
The star is moving **toward** Earth at $v = 0.2c$.

**Find:**

1. The observed wavelength $\lambda_{\text{obs}}$ on Earth.  
2. The frequency shift relative to the source.

```python
import numpy as np

# Given
c = 2.998e8
lambda_source = 500e-9  # m
v = 0.2 * c
beta = v / c

# Doppler formulas
doppler_factor = np.sqrt((1 + beta) / (1 - beta))
lambda_obs = lambda_source / doppler_factor
f_source = c / lambda_source
f_obs = c / lambda_obs

lambda_obs_nm = lambda_obs * 1e9  # convert to nm
freq_shift_percent = ((f_obs - f_source) / f_source) * 100

lambda_obs_nm, freq_shift_percent
```

## Problem 4: Receding Galaxy – Extreme Redshift

A distant galaxy emits light at a rest-frame wavelength of $\lambda_{\text{source}} = 400\,\text{nm}$.  
The galaxy is receding from Earth at $v = 0.9c$.

**Find:**

1. The observed wavelength $\lambda_{\text{obs}}$.  
2. Into which part of the electromagnetic spectrum this light is shifted.

```python
# Given
lambda_source = 400e-9  # m
v = 0.9 * c
beta = v / c

# Doppler formulas
doppler_factor = np.sqrt((1 + beta) / (1 - beta))
lambda_obs = lambda_source * doppler_factor
lambda_obs_nm = lambda_obs * 1e9  # nm

# Spectrum classification
if lambda_obs_nm < 380:
    spectrum = "Ultraviolet"
elif 380 <= lambda_obs_nm <= 750:
    spectrum = "Visible"
elif 750 < lambda_obs_nm <= 1000:
    spectrum = "Near Infrared"
else:
    spectrum = "Infrared"

lambda_obs_nm, spectrum
```

## Problem 5: Beaming Angle of Relativistic Electrons

Electrons in a synchrotron move with a Lorentz factor of $\gamma = 50$.

Estimate the **characteristic beaming angle** $\theta'_{\text{beam}}$ (in degrees) into which most radiation is confined.  

Use the headlight-effect approximation:

$$
\theta'_{\text{beam}} \sim \frac{1}{\gamma}
$$


```python
import numpy as np

# Given
gamma = 50
theta_beam_rad = 1 / gamma
theta_beam_deg = np.rad2deg(theta_beam_rad)
theta_beam_deg
```

## Problem 6: Intensity Boost of a Quasar Jet

A quasar jet is moving with $\gamma = 10$ directly toward Earth.

1. Calculate the **forward intensity boost** relative to the rest frame, assuming emission is concentrated near $\theta' \approx 0$.  
2. Express the boost as a factor of $I'/I$.

Use the headlight-effect formula:

$$
I'(\theta' \approx 0) \propto (1 - \beta)^{-3} \approx (2\gamma^2)^3 = 8\gamma^6
$$


```python
gamma = 10
intensity_boost = 8 * gamma**6
intensity_boost
```
