## B7.1.1 Motivation

### What is Aberration of Light?

Aberration of light is the **apparent change in the direction of incoming light** due to the **relative motion** between an observer and the light source.  

In simple terms:  
- You are moving through space, so the light rays seem to be coming from a slightly different direction than they actually are.  
- **Analogy:** Running in the rain — even if the rain falls vertically, you tilt your umbrella forward because of your motion.

---

### Historical Context

- **James Bradley (1728):**  
  While trying to measure stellar parallax, Bradley discovered that stars appear to trace small ellipses in the sky over a year.  
  - He correctly interpreted this as an effect of Earth's orbital motion around the Sun.  
  - The measured angular shift (~20.5 arcseconds) provided **indirect evidence for the finite speed of light**.

- **Pre-Relativity Explanation:**  
  Classical physics explained it approximately as a vector addition of the Earth’s velocity and the light's velocity.  
  - But for high velocities (close to $c$), the classical picture fails.

- **Relativistic Insight:**  
  Special relativity provides an **exact description** using **Lorentz transformations** and the **relativistic velocity addition law**.

---

### Why Study It in Special Relativity?

1. **Direct Application of Lorentz Transformations:**  
   Aberration is one of the first truly physical predictions after learning the mathematics of Lorentz transformations.  

2. **Links to Other Relativistic Phenomena:**  
   - **Relativistic Doppler Shift:** Frequency changes are derived in a similar way.  
   - **Headlight Effect:** At high velocities, aberration explains why light is beamed forward (jets of quasars, synchrotron radiation).  

3. **Astrophysical Relevance:**  
   - Measurement of stellar aberration gave one of the earliest proofs of Earth's motion.  
   - Modern astronomy uses aberration corrections for high-precision measurements.  
   - In high-energy astrophysics, understanding beamed radiation is crucial for interpreting observations of relativistic jets.

4. **Conceptual Understanding:**  
   Aberration provides an intuitive visualization of **how space and time mix** under motion — angles get distorted just as lengths and times do.

---

### Everyday / Visual Intuition

- **Running-in-the-rain analogy:**  
  If you run at velocity $v$, the rain seems to hit you from a forward angle, even if it falls straight down. Light behaves similarly, but relativistic effects make the shift stronger at high $v$.

- **Forward Beaming (Relativistic):**  
  At speeds close to $c$, most incoming light gets concentrated into a narrow forward cone — a key to understanding **relativistic jets** and the **blazing brightness** of certain astrophysical objects.

## B7.1.2 Derivation of the Aberration Formula

Before we can find how light appears shifted to a moving observer, we must recall **how velocities transform between frames** in special relativity.

###  Setup of the Two Reference Frames

- **Frame S (Source Frame):** The star is at rest. Light propagates with components:

  $$
  u_x = c \cos\theta, \qquad u_y = c \sin\theta
  $$

- **Frame S′ (Observer Frame):** The observer moves with constant velocity $v$ along the $+x$ direction relative to the star. 

- We are looking for the light velocity components $u'_x, u'_y$ in the moving frame S′.

---

### General Relativistic Velocity Transformation

From Lorentz transformations of space and time:

- For any particle (or photon):

  $$
  u'_x = \frac{u_x - v}{1 - \frac{v u_x}{c^2}}, \tag{1}
  $$
  $$
  u'_y = \frac{u_y}{\gamma\left(1 - \frac{v u_x}{c^2}\right)}, \tag{2}
  $$
  
  where $\gamma = 1/\sqrt{1-\beta^2}$ and $\beta = v/c$.

- These formulas reduce to classical Galilean addition if $v \ll c$.

---

### Specializing to Light (Photon Case)

Because light always moves at speed $c$ in every inertial frame:

- Substitute $u_x = c\cos\theta$ and $u_y = c\sin\theta$:

$$
u'_x = \frac{c\cos\theta - v}{1 - \beta \cos\theta}, \tag{3}
$$

$$
u'_y = \frac{c\sin\theta}{\gamma(1 - \beta \cos\theta)}. \tag{4}
$$

Notice:

1. The speed remains $c$ after transformation (you can verify that $u'^2_x + u'^2_y = c^2$ as a check of consistency).

2. The **direction**, however, **changes** — this is the aberration we are about to compute.

---

### Solving for the Observed Angle

The angle $\theta'$ in the observer’s frame S′ is determined by:

$$
\tan\theta' = \frac{u'_y}{u'_x}
$$

Substitute Equations (3) and (4):

$$
\tan\theta' = 
\frac{\frac{c\sin\theta}{\gamma(1 - \beta \cos\theta)}}
{\frac{c\cos\theta - v}{1 - \beta \cos\theta}}
$$

The $(1 - \beta\cos\theta)$ cancels immediately, leaving:

$$
\tan\theta' = \frac{\sin\theta}{\gamma(\cos\theta - \beta)} \tag{5}
$$

Equation (5) is the **relativistic aberration formula in tangent form**.

---

### Alternative Forms

For many practical applications, we may prefer **sine** or **cosine** forms.

---

#### (a) Cosine Form

Use $u'_x = c\cos\theta'$, then we have directly from (1):

$$
\cos\theta' = \frac{u'_x}{c} =
\frac{c\cos\theta - v}{c(1 - \beta \cos\theta)} =
\frac{\cos\theta - \beta}{1 - \beta\cos\theta} \tag{6}
$$

This form is often the **most compact** and commonly used in astrophysics.

---

#### (b) Sine Form

Similarly, from $u'_y = c\sin\theta'$ and using (Eq. 2):

$$
\sin\theta' = \frac{u'_y}{c} =
\frac{c\sin\theta}{c\gamma(1 - \beta\cos\theta)} =
\frac{\sin\theta}{\gamma(1 - \beta\cos\theta)} \tag{7}
$$


---

### Physical Interpretation

- **Forward Beaming:**  
  As $v \to c$ ($\beta \to 1$), $\cos\theta' \to 1$ for most $\theta$, meaning almost all light is observed near $\theta' = 0^\circ$ (forward direction).  

- **Low-Velocity Limit:**  
  Expanding (7) for $\beta \ll 1$ gives:

  $$
  \theta' \approx \theta + \beta \sin\theta
  $$

**This is Bradley’s classical stellar aberration formula**.

## B7.1.3 Worked Example
<br>
<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;"> 
<header>
<h3>Example: Stellar Aberration</h3>
    </header>


**Problem:**  
A telescope is moving at $30\,\text{km/s}$ ($\beta \approx 10^{-4}$) relative to a star. The star appears at $\theta = 90^\circ$ in the star’s rest frame. What is the apparent angle $\theta'$?  

---

**Solution:**

Given:

$$
\beta = 10^{-4}, \quad \cos\theta = 0, \quad \sin\theta =1
$$

Use the low-speed limit:

$$
\theta' \approx \theta + \beta \sin\theta = 90^\circ + 10^{-4}\ \text{rad}
$$

Convert:

$$
10^{-4}\ \text{rad} \approx 20.6''\ (\text{arcseconds})
$$

This is almost exactly Bradley’s observed value (~20.5"). In the moving observer’s frame (S'), even though the star is actually perpendicular in the source frame, the star appears slightly shifted forward (aberrated). This is exactly like the running-in-the-rain analogy:

* In the ground frame, rain falls vertically ($90^\circ$ to your direction of motion).
* But because you’re moving, you tilt the umbrella forward — you see the rain coming at an angle.

**In this case, the telescope must be tilted slightly forward**.

## B7.1.4 Summary (Key Formulas)

- **Cosine form** 

  $$\cos\theta' = \frac{\cos\theta - \beta}{1 - \beta\cos\theta}$$

- **Tangent form (from velocity addition)** 

  $$\tan\theta' = \frac{\sin\theta}{\gamma(\cos\theta - \beta)}$$

- **Sine form**  

  $$\sin\theta' = \frac{\sin\theta}{\gamma(1 - \beta\cos\theta)}$$

- **Low-$\beta$ (Bradley’s stellar aberration)**  

  $$\theta' \approx \theta + \beta \sin\theta$$
  
>The standard aberration formula are derived for light rays (photons).
So: $\theta$ (source frame) = angle of the photon’s direction of travel relative to the observer’s velocity (+x) in the source rest frame. $\theta '$ (observer frame) = angle of the same photon’s direction as measured by the moving observer. Thus, the formulas do not directly tell you where you “see” the star — they tell you the direction from which the light arrives.

### ✅ Photon Direction vs Star Position

When you look at a star, you are looking **back along the photon’s path**.

So:

$$
\theta_{\text{star}} = 180^\circ - \theta_{\text{photon}}
$$

$$
\theta'_{\text{star}} = 180^\circ - \theta'_{\text{photon}}
$$

---

#### **Example**

- **Star directly in front of you:**  
  - Star position: $\theta_{\text{star}} = 0^\circ$  
  - Photon direction: $\theta_{\text{photon}} = 180^\circ$

- **Star to your right at $90^\circ$:**  
  - Star position: $90^\circ$  
  - Photon direction: $90^\circ$ (same because the ray is perpendicular)

- **Star behind you:**  
  - Star position: $\theta_{\text{star}} = 180^\circ$  
  - Photon direction: $\theta_{\text{photon}} = 0^\circ$ (the photons come from straight ahead, toward you)
