# D7.0 Learning Outcomes — RC, LC, and RLC Circuits  
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this module matters

Electric fields and electric potential describe how energy is stored in systems of charges. In real circuits, however, energy is not only stored — it is **transferred, transformed, and dissipated over time**.

In this module, we move beyond static circuits and study how circuits **evolve dynamically**. We begin with **RC circuits**, where energy stored in a capacitor is gradually dissipated through a resistor, leading to exponential behavior.

This naturally raises an important question:

> Can a circuit store energy in more than one way and allow it to move back and forth instead of simply decaying?

To answer this, we introduce **inductors**, which store energy in magnetic fields. Combining capacitors and inductors leads to **LC circuits**, where energy oscillates between electric and magnetic forms. Adding resistance produces **RLC circuits**, which model real systems where oscillations gradually decay.

These ideas are central to physics and engineering, appearing in:

- Signal processing and electronic filters  
- Wireless communication and radio tuning  
- Power systems and energy transfer  
- Resonance phenomena across physics  

By the end of this module, you will understand how circuits can **decay, oscillate, and resonate**, and how energy flows between different forms in time-dependent systems.

By the end of this module, you will be able to:

### A. RC Circuits and Exponential Behavior  
*Introduces time-dependent circuits with dissipation.*

- **A1.** Describe qualitatively how a capacitor charges and discharges in an RC circuit.
- **A2.** Apply Kirchhoff’s loop rule to write the governing equation,

  $$
  \mathcal{E} = IR + \frac{Q}{C}.
  $$

- **A3.** Identify exponential behavior in charging and discharging processes.
- **A4.** Define the time constant,

  $$
  \tau = RC,
  $$

  and interpret its physical meaning.
- **A5.** Explain why RC circuits do **not** oscillate.

---

### B. Magnetic Effects and Inductance  
*Introduces the physical origin of inductors.*

- **B1.** Describe how electric currents produce magnetic fields.
- **B2.** Explain qualitatively why changing currents lead to opposing effects.
- **B3.** Recognize that this effect depends on $\frac{dI}{dt}$ rather than $I$.
- **B4.** Distinguish this phenomenon from resistive effects.
- **B5.** Motivate the need for a new circuit element that stores magnetic energy.

---

### C. Inductors as Energy Storage Elements  
*Defines inductors and their properties.*

- **C1.** Define inductance and write the relationship,

  $$
  V_L = L \frac{dI}{dt}.
  $$

- **C2.** Explain physically how inductors oppose changes in current.
- **C3.** Define energy stored in an inductor,

  $$
  U_L = \frac{1}{2} L I^2.
  $$

- **C4.** Compare inductors and capacitors as complementary energy storage elements.
- **C5.** Use mechanical analogies to interpret inductive behavior.

---

### D. LC Circuits and Oscillations  
*Introduces undamped oscillatory systems.*

- **D1.** Describe energy exchange between capacitor and inductor.
- **D2.** Apply Kirchhoff’s loop rule to derive,

  $$
  L \frac{d^2Q}{dt^2} + \frac{1}{C} Q = 0.
  $$

- **D3.** Recognize this equation as analogous to simple harmonic motion.
- **D4.** Identify the natural angular frequency,

  $$
  \omega_0 = \frac{1}{\sqrt{LC}}.
  $$

- **D5.** Explain why LC circuits can oscillate while RC circuits cannot.

---

### E. RLC Circuits and Damping  
*Introduces realistic systems with energy loss.*

- **E1.** Extend the circuit equation to include resistance,

  $$
  L \frac{d^2Q}{dt^2} + R \frac{dQ}{dt} + \frac{1}{C} Q = 0.
  $$

- **E2.** Explain the role of resistance as energy dissipation.
- **E3.** Distinguish between underdamped, critically damped, and overdamped behavior.
- **E4.** Describe how damping affects amplitude and energy over time.
- **E5.** Connect electrical damping to mechanical analogs.

---

### F. Driven Circuits and Resonance  
*Introduces frequency-dependent behavior.*

- **F1.** Describe a driven RLC circuit with an AC source.
- **F2.** Define impedance and its role in circuit response.
- **F3.** Identify the resonance condition and explain its significance.
- **F4.** Relate resonant frequency to system parameters.
- **F5.** Explain how amplitude depends on driving frequency.

---

### Performance Criteria

- RC circuits are correctly interpreted as **time-dependent dissipative systems**.
- Inductance is understood as arising from **magnetic effects of currents**.
- LC circuits are recognized as **oscillatory systems with energy exchange**.
- RLC circuits are understood as **damped oscillators**.
- Resonance is explained in terms of **frequency-dependent energy transfer**.
