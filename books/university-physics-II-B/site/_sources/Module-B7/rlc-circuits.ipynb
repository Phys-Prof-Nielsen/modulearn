## B7.7 RLC Circuits — Damping and Real Systems

In an ideal LC circuit, energy oscillates indefinitely between electric and magnetic forms.  
However, real circuits always include **resistance**, which causes energy to be **dissipated** over time.

When resistance is included, the circuit becomes an **RLC circuit**, and the oscillations are no longer sustained — they are **damped**.

---

### B7.7.1 The RLC Circuit Equation

To describe this system quantitatively, we apply Kirchhoff’s loop rule to a circuit containing all three elements.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — The RLC Circuit Equation**

An RLC circuit contains:

- A resistor ($R$)  
- An inductor ($L$)  
- A capacitor ($C$)

Applying Kirchhoff’s loop rule:

$$
V_R + V_L + V_C = 0
$$

Using:

$$
V_R = IR, \qquad V_L = L \frac{dI}{dt}, \qquad V_C = \frac{Q}{C}, \qquad I = \frac{dQ}{dt}
$$

we obtain:

$$
L \frac{d^2Q}{dt^2} + R \frac{dQ}{dt} + \frac{1}{C} Q = 0
$$

</div>

---

### B7.7.2 Why Oscillations Decay

The addition of resistance fundamentally changes the behavior of the system.

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — Why Oscillations Decay**

Unlike LC circuits:

- Resistance converts electrical energy into **thermal energy**  
- Energy is continuously **lost from the system**

👉 As a result, oscillations decrease in amplitude over time

</div>

---

### B7.7.3 Mechanical Analogy

The structure of the RLC equation reveals a direct analogy with damped mechanical systems.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Mechanical Analogy**

The RLC circuit is analogous to a **mass–spring system with friction**:

- Inductor ↔ mass  
- Capacitor ↔ spring  
- Resistor ↔ friction  

The equation:

$$
L \frac{d^2Q}{dt^2} + R \frac{dQ}{dt} + \frac{1}{C} Q = 0
$$

is equivalent to:

$$
m \frac{d^2x}{dt^2} + b \frac{dx}{dt} + kx = 0
$$

👉 The system loses energy over time due to damping.

</div>

---

### B7.7.4 Types of Damping

The behavior of the system depends on how strongly resistance affects the motion.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Types of Damping**

The behavior of the system depends on the relative size of $R$, $L$, and $C$.

1. **Underdamped ($R$ small)**  
   - Oscillations occur  
   - Amplitude gradually decreases  

2. **Critically damped (specific value of $R$)**  
   - No oscillations  
   - Fastest return to equilibrium  

3. **Overdamped ($R$ large)**  
   - No oscillations  
   - Slow return to equilibrium  

👉 These behaviors arise from how strongly energy is dissipated.

</div>

---

### B7.7.5 Energy Perspective

We can also interpret damping in terms of how energy evolves in the system.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Energy Perspective**

At each cycle:

- Energy flows between capacitor and inductor  
- Some energy is lost in the resistor  

This leads to:

$$
U_{\text{total}} \text{ decreases over time}
$$

👉 The system “runs out of energy,” causing oscillations to fade.

</div>

---

### B7.7.6 Identifying the Regime

We can now analyze how different values of resistance affect the system.

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Identifying the Regime**

Consider three circuits with identical $L$ and $C$, but different resistances.

1. Which circuit would oscillate the longest?  
2. Which circuit would not oscillate at all?  
3. Which circuit returns to equilibrium fastest without oscillating?

<details>
<summary>Solution</summary>

1. The circuit with the smallest $R$ (underdamped)  
2. The circuit with large $R$ (overdamped)  
3. The critically damped circuit  

</details>

</div>

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 2 — Comparing RC, LC, and RLC**

Compare the three types of circuits:

1. Which system always loses energy without oscillation?  
2. Which system conserves energy and oscillates indefinitely?  
3. Which system both oscillates and loses energy?

<details>
<summary>Solution</summary>

1. RC circuits  
2. LC circuits  
3. RLC circuits  

</details>

</div>
