## B7.6 Energy Perspective of LC Oscillations

In the previous section, we saw that LC circuits exhibit oscillatory behavior.  
We now examine this behavior from an **energy perspective**, which provides the deepest physical insight.

Rather than focusing only on charge and current, we track how energy moves between:

- The **electric field** of the capacitor  
- The **magnetic field** of the inductor  

---

### B7.6.1 Energy Stored in Each Element

To understand the system, we begin by identifying how energy is stored in each component.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Energy Stored in Each Element**

The total energy in an LC circuit is the sum of:

- Energy in the capacitor:

$$
U_C = \frac{1}{2} C V^2 = \frac{Q^2}{2C}
$$

- Energy in the inductor:

$$
U_L = \frac{1}{2} L I^2
$$

Total energy:

$$
U_{\text{total}} = U_C + U_L
$$

</div>

---

### B7.6.2 Energy Conservation

A key feature of the LC circuit is that no energy is lost.

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — No Energy Loss**

In an ideal LC circuit:

- There is **no resistance**  
- No energy is dissipated  

👉 Therefore:

$$
U_{\text{total}} = \text{constant}
$$

If resistance were present, energy would decrease over time.

</div>

---

### B7.6.3 Energy Exchange Over a Cycle

We now examine how energy moves between the capacitor and inductor during oscillation.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Energy Over One Cycle**

Consider one full oscillation:

1. **Maximum charge ($Q = Q_{\max}$)**
   - $I = 0$  
   - $U_C$ is maximum  
   - $U_L = 0$

2. **Discharging phase**
   - $Q$ decreases  
   - $I$ increases  
   - Energy transfers from capacitor to inductor  

3. **Zero charge ($Q = 0$)**
   - $I$ is maximum  
   - $U_L$ is maximum  
   - $U_C = 0$

4. **Recharging phase**
   - Energy flows back into capacitor (opposite polarity)

👉 Energy continuously **oscillates between electric and magnetic forms**.

</div>

---

### B7.6.4 Mathematical Expression of Energy Conservation

We can verify energy conservation mathematically using the solutions for $Q(t)$ and $I(t)$.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Mathematical Expression of Energy Conservation**

Using:

$$
Q(t) = Q_0 \cos(\omega_0 t)
$$

$$
I(t) = -\omega_0 Q_0 \sin(\omega_0 t)
$$

we find:

$$
U_C = \frac{Q_0^2}{2C} \cos^2(\omega_0 t)
$$

$$
U_L = \frac{1}{2} L \omega_0^2 Q_0^2 \sin^2(\omega_0 t)
$$

Since:

$$
\omega_0^2 = \frac{1}{LC}
$$

we obtain:

$$
U_L = \frac{Q_0^2}{2C} \sin^2(\omega_0 t)
$$

Thus:

$$
U_{\text{total}} = \frac{Q_0^2}{2C}
$$

👉 Total energy remains constant in time.

</div>

---

### B7.6.5 Connection to Mechanics

This energy exchange mirrors behavior seen in mechanical systems.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Analogy to Mechanics**

Compare with a mass–spring system:

- Potential energy:

$$
U = \frac{1}{2} k x^2
$$

- Kinetic energy:

$$
K = \frac{1}{2} m v^2
$$

Total energy:

$$
E = U + K = \text{constant}
$$

Mapping:

- Capacitor energy ↔ potential energy  
- Inductor energy ↔ kinetic energy  

👉 LC circuits are **energy oscillators**, just like mechanical systems.

</div>

---

### B7.6.6 Interpreting Energy at Key Moments

We can now analyze specific points in the oscillation.

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Energy at Key Moments**

At the moment when $Q = 0$:

1. What is the energy in the capacitor?  
2. What is the energy in the inductor?  
3. What is the total energy?

<details>
<summary>Solution</summary>

1. $U_C = 0$  
2. $U_L$ is maximum  
3. Total energy remains constant  

</details>

</div>

---

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 2 — Time Dependence of Energy**

At what times is the energy equally shared between capacitor and inductor?

<details>
<summary>Solution</summary>

Energy is equally shared when:

$$
\cos^2(\omega_0 t) = \sin^2(\omega_0 t)
$$

This occurs when:

$$
\omega_0 t = \frac{\pi}{4}, \; \frac{3\pi}{4}, \; \dots
$$

At these times:

$$
U_C = U_L = \frac{1}{2} U_{\text{total}}
$$

</details>

</div>

---

The energy perspective reveals the true nature of LC oscillations:

> The circuit is not simply “moving charge,” but continuously exchanging energy between electric and magnetic fields.

In the next section, we will introduce resistance and explore how this ideal energy exchange is modified in **real circuits**, leading to damping.
