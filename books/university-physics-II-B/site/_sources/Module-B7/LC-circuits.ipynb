## B7.5 LC Circuits — Pure Oscillations

We now combine the two energy storage elements we have introduced:

- The **capacitor**, which stores energy in an electric field  
- The **inductor**, which stores energy in a magnetic field  

When these two elements are connected together (with no resistance), the circuit can support a new type of behavior:

> **Oscillations**, where energy moves back and forth between electric and magnetic forms.

---

### B7.5.1 Constructing the LC Circuit

To understand how oscillations arise, we begin by writing the governing equation for a circuit containing only an inductor and a capacitor.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — The LC Circuit**

An LC circuit consists of:

- An inductor ($L$)  
- A capacitor ($C$)  
- No resistance  

Applying Kirchhoff’s loop rule:

$$
V_L + V_C = 0
$$

Using:

$$
V_L = L \frac{dI}{dt}, \qquad V_C = \frac{Q}{C}, \qquad I = \frac{dQ}{dt}
$$

we obtain:

$$
L \frac{d^2Q}{dt^2} + \frac{1}{C} Q = 0
$$

</div>

---

### B7.5.2 Connection to Simple Harmonic Motion

The structure of this equation reveals a deep connection to mechanics.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Mechanical Analogy**

This equation is mathematically identical to that of a mass–spring system:

$$
m \frac{d^2x}{dt^2} + kx = 0
$$

Mapping the variables:

- $Q \leftrightarrow x$ (position)  
- $L \leftrightarrow m$ (mass)  
- $\frac{1}{C} \leftrightarrow k$ (spring constant)  

👉 The LC circuit behaves like a **simple harmonic oscillator**.

</div>

---

### B7.5.3 Natural Frequency of Oscillation

The mathematical form of the equation leads directly to oscillatory solutions.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Natural Frequency of Oscillation**

The solution to the LC equation is oscillatory:

$$
Q(t) = Q_0 \cos(\omega_0 t + \phi)
$$

where the natural angular frequency is:

$$
\omega_0 = \frac{1}{\sqrt{LC}}
$$

👉 This frequency depends only on the circuit properties $L$ and $C$.

</div>

---

### B7.5.4 Why Oscillations Occur

We can now understand physically why oscillations are possible in this system.

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — Why Oscillations Are Possible**

Oscillations occur because:

- Energy is stored in **two forms**:
  - Electric field (capacitor)  
  - Magnetic field (inductor)

- Energy can be transferred back and forth without loss

👉 If resistance were present, energy would be dissipated and oscillations would decay.

</div>

---

### B7.5.5 Energy Exchange Over Time

The oscillatory motion can be understood as a continuous exchange of energy between the two elements.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Energy Exchange**

Consider the motion over one cycle:

1. **Capacitor fully charged**
   - Energy stored in electric field  
   - Current is zero  

2. **Discharge begins**
   - Current increases  
   - Magnetic energy builds in the inductor  

3. **Capacitor empty**
   - Energy entirely in magnetic field  
   - Current is maximum  

4. **Current decreases**
   - Capacitor recharges with opposite polarity  

👉 The process repeats, producing oscillations.

</div>

---

### B7.5.6 Interpreting the Oscillation

To reinforce understanding, consider how energy is distributed at key moments.

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Identifying Energy Forms**

At the moment when current is maximum:

1. Is the capacitor charged or uncharged?  
2. Where is the energy stored?  
3. What is happening to the current at that instant?

<details>
<summary>Solution</summary>

1. The capacitor is uncharged ($Q = 0$)  
2. All energy is stored in the magnetic field of the inductor  
3. The current is momentarily at its maximum before beginning to decrease  

</details>

</div>

---

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 2 — Comparing to RC Circuits**

Compare LC and RC circuits:

1. Which system allows energy to return after being stored?  
2. Which system dissipates energy continuously?  
3. Which system can oscillate?

<details>
<summary>Solution</summary>

1. LC circuits allow energy to move back and forth  
2. RC circuits dissipate energy through the resistor  
3. Only LC circuits can oscillate  

</details>

</div>

---

The LC circuit represents the simplest system capable of **sustained oscillations**.

In the next section, we will introduce resistance into the system, leading to **RLC circuits**, where oscillations gradually decay due to energy loss.
