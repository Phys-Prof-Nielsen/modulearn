## B7.3 Magnetic Effects of Currents — A Bridge to Inductance

In previous sections, we focused on electric fields and electric potential.  
However, when charges move, they introduce an additional physical effect: **magnetism**.

To understand inductors, we must first understand how **currents and magnetic fields are connected**.

---

### B7.3.1 Currents and Magnetic Fields

Electric current is not just a flow of charge — it also generates a magnetic field in the surrounding space.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Currents Create Magnetic Fields**

We have focused primarily on electric fields and electric potential. However, moving charges — that is, **electric currents** — produce **magnetic fields**.

- A steady current creates a steady magnetic field  
- A changing current produces a **changing magnetic field**

This introduces a new physical connection:

> Electric phenomena (currents) are directly linked to magnetic phenomena (fields)

</div>

---

### B7.3.2 Visualizing the Magnetic Field

To build intuition, we consider a simple and common example.

<div style="background-color:#e0f7fa; padding:15px; border-left:6px solid #006a80; border-radius:5px;">

**Example — Magnetic Field Around a Wire**

When current flows through a straight wire:

- Magnetic field lines form **closed loops** around the wire  
- The direction is given by the **right-hand rule**

If the current increases or decreases:

- The magnetic field strength also changes

</div>

---

### B7.3.3 Response to Changing Current

The key insight comes when the current is not constant.

<div style="background-color:#fff8e1; padding:15px; border-left:6px solid #ffc20e; border-radius:5px;">

**Curiosity — Nature Resists Change**

Experiments show something surprising:

> When the current in a circuit changes, the system **opposes that change**

- Increasing current → opposition to increase  
- Decreasing current → opposition to decrease  

This is not due to resistance (which dissipates energy), but something different:

👉 The changing magnetic field itself plays a role

</div>

---

### B7.3.4 Distinguishing from Resistance

It is important to clearly separate this new behavior from what we already know about resistors.

<div style="background-color:#ffe6e6; padding:15px; border-left:6px solid #990000; border-radius:5px;">

**Danger Zone — Not the Same as Resistance**

It is important to distinguish:

- **Resistance** opposes current and converts energy to heat  
- This new effect opposes **changes in current** and does **not immediately dissipate energy**

👉 This means energy is being temporarily **stored**, not lost

</div>

---

### B7.3.5 Induced Voltage from Changing Current

We can now describe this effect mathematically.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Induced Voltage from Changing Current**

When current changes, the circuit experiences a voltage that opposes the change.

This behavior can be described mathematically as:

$$
V \propto \frac{dI}{dt}
$$

The faster the current changes, the stronger the opposing effect.

This is fundamentally different from Ohm’s law:

$$
V = IR
$$

which depends on the current itself, not how it changes.

</div>

---

### B7.3.6 Interpreting the Effect

To deepen understanding, consider how this behavior appears in a simple circuit.

<div style="background-color:#e8f5e9; padding:15px; border-left:6px solid #006633; border-radius:5px;">

**Box Activity 1 — Interpreting the Physics**

Consider turning on a circuit with a switch.

1. When the switch is first closed, is the current changing rapidly or slowly?  
2. Based on $V \propto \frac{dI}{dt}$, when would the opposing effect be strongest?  
3. As time passes and current becomes steady, what happens to $\frac{dI}{dt}$?  
4. What does this imply about the opposing effect?

<details>
<summary>Solution</summary>

1. The current is changing rapidly at the moment the switch is closed  
2. The opposing effect is strongest when $\frac{dI}{dt}$ is largest  
3. $\frac{dI}{dt} \to 0$ as the current stabilizes  
4. The opposing effect disappears when the current becomes constant  

</details>

</div>

---

### B7.3.7 Big Idea

We are now ready to introduce a new circuit element that captures this behavior.

<div style="background-color:#e6f0ff; padding:15px; border-left:6px solid #003366; border-radius:5px;">

**Concept — Big Idea**

A changing current produces a changing magnetic field, and this leads to an effect that:

- Opposes changes in current  
- Depends on how quickly the current changes  
- Stores energy temporarily rather than dissipating it  

This phenomenon leads to a new circuit element:

> The **inductor**, which we will define next as a device that stores energy in a magnetic field and resists changes in current.

</div>
