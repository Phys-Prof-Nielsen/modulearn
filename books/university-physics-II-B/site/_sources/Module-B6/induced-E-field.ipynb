# B6.4 Induced Electric Field: What Sets It Apart from the Static Electric Field?
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.4.1 Motivation

Up to now:

- **Electrostatics**: Electric fields arise from charges ($\mathbf{E}$ due to Coulomb’s law or Gauss’s law).
- **Faraday’s Law**: A **changing magnetic flux** induces an electric field, even in regions **with no charges**.

This **induced electric field** has properties fundamentally different from static electric fields. Understanding this distinction is critical for grasping **electromagnetic induction** and eventually **electromagnetic waves**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.4.2 Faraday’s Law in Differential Form

The induced electric field is expressed by Faraday’s law:

$$
\oint_{\partial S} \mathbf{E}\cdot d\mathbf{s} \;=\; -\frac{d}{dt}\int_S \mathbf{B}\cdot d\mathbf{A}
$$

or, in differential form:

$$
\nabla \times \mathbf{E} \;=\; -\,\frac{\partial \mathbf{B}}{\partial t}
$$

This tells us:

- A **time-varying magnetic field** generates a circulating electric field.  
- No charges are required; it is a property of space itself.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B6.4.3 Comparing **Electrostatic** and **Induced** Electric Fields

| **Property** | **Electrostatic (Coulomb)** | **Induced (Faraday)** |
|--------------|-----------------------------|------------------------|
| **Source** | Stationary charges ($\rho$) | Changing magnetic field ($\partial B/\partial t$) |
| **Conservativeness** | **Conservative**: $\nabla \times \mathbf{E} = 0$ | **Non-conservative**: $\nabla \times \mathbf{E} = -\partial B/\partial t \neq 0$ |
| **Work around a closed loop** | $ \oint \mathbf{E}\cdot d\mathbf{s} = 0$ | $ \oint \mathbf{E}\cdot d\mathbf{s} = -\dfrac{d\Phi_B}{dt} \neq 0$ |
| **Potential Function** | Can define a single-valued potential $V$: $\mathbf{E} = -\nabla V$ | Cannot define a single-valued scalar potential (must use vector potential or time-varying potentials) |
| **Field Lines** | Begin and end on charges | Form **closed loops** around regions of changing magnetic flux |
| **Exists in vacuum without charges?** | No (needs charges) | Yes, purely field-driven |

### Physical Interpretation

- In electrostatics, energy is stored in the **potential energy of charges**.  
- In induced fields, energy is transferred via **electromagnetic induction** — the field does real work moving charges around a loop.

This is why **electric generators work**: no battery is needed; the induced field itself pushes charges.

## B6.4.4 Visualizing the Induced Electric Field

Imagine a region where a magnetic field into the page is **increasing**:

- By Faraday’s law, an electric field is induced in **closed loops around the changing magnetic flux**.
- Use the **right-hand rule with Lenz’s law** to find the circulation direction.

Unlike electrostatic fields:

- Field lines do **not** start or end on charges.  
- They form **continuous closed loops** (similar to magnetic field lines).

## B6.4.5 Conceptual Example

### **Problem**

A circular loop sits in a region where the magnetic field into the page is **increasing**.  

**(a)** Is there an induced electric field in the space **outside** the loop?  
**(b)** Can you find a scalar potential that explains the current?

---

### **Solution**

**(a)** **Yes.** The induced electric field exists everywhere surrounding the region of changing flux, not just in the conductor. Even if you remove the loop, a test charge placed nearby would feel a force.  

**(b)** **No single-valued scalar potential exists.** The electric field is non-conservative, so $\mathbf{E}\neq -\nabla V$. Instead, you would need to describe it using the **vector potential** or directly from Faraday’s law.

## B6.4.6 Why This Matters

1. **Fundamental Difference in Electrodynamics**  
   - Static electricity is governed by Coulomb’s law.  
   - Time-varying magnetic fields give rise to completely new behavior — this is what unifies electric and magnetic fields in Maxwell’s theory.

2. **Technology**  
   - **Induced electric fields** are what make generators, transformers, and wireless power transfer work.  
   - You can induce currents in conductors **without any battery**.

3. **Electromagnetic Waves**  
   - Changing electric and magnetic fields continuously generate each other in space, forming self-propagating waves.
