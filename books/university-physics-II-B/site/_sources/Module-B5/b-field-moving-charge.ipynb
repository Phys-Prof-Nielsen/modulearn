# B5.1 Magnetic Field from a Moving Charge
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B5.1.1 Motivation

So far, we have studied electric fields produced by stationary charges. But what happens when charges move?  

Experiments (starting with **Ørsted** in 1820) showed that **moving charges produce magnetic effects**—a compass needle deflects when placed near a current-carrying wire. This led to the understanding that:

- **Electric fields are created by stationary charges.**  
- **Magnetic fields arise from moving charges (currents).**

In modern physics, we now understand that electricity and magnetism are unified under **electromagnetism**—what we call a magnetic field is simply the relativistic effect of electric fields when charges are in motion.  

In this lecture, we will derive and understand the **magnetic field due to a single moving charge**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B5.1.2 The Magnetic Field from a Point Charge

For a **point charge $q$ moving with constant velocity $\mathbf{v}$**, the magnetic field at some observation point is given by the **Biot–Savart law** for a point charge:

$$
\mathbf{B} = \frac{\mu_0}{4\pi} \, \frac{q \, \mathbf{v} \times \hat{\mathbf{r}}}{r^2}
$$

where:

- $\mu_0 = 4\pi \times 10^{-7}\ \text{T·m/A}$ is the permeability of free space,  
- $\mathbf{r}$ is the vector from the charge to the observation point,  
- $r = |\mathbf{r}|$,  
- $\hat{\mathbf{r}} = \frac{\mathbf{r}}{r}$ is the unit vector pointing from the charge to the observation point,  
- $\mathbf{v}$ is the **instantaneous velocity** of the charge.

⚠ **Important Assumptions:**

1. The speed of the charge is much less than $c$ (non-relativistic).  
2. The field is computed at an instant when the charge is at a given position.  
3. Retardation effects (finite speed of EM propagation) are neglected here.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B5.1.3 Direction of the Magnetic Field

The **cross product** tells us the direction:  

- $\mathbf{B}$ is **perpendicular** to both $\mathbf{v}$ and $\mathbf{r}$.  
- The right-hand rule applies: point fingers along $\mathbf{v}$, curl towards $\mathbf{r}$, and your thumb points along $\mathbf{B}$.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## B5.1.4 Comparison to the Electric Field

The electric field from a point charge is:

$$
\mathbf{E} = \frac{1}{4\pi \epsilon_0}\frac{q}{r^2}\hat{\mathbf{r}}
$$

Key differences:

- $\mathbf{E}$ points **radially** away or toward the charge.  
- $\mathbf{B}$ points **circulates around** the direction of motion (tangential direction).  
- Magnitude ratio (for non-relativistic speeds):  

$$
\frac{B}{E} \sim \frac{v}{c^2}
$$

Thus, magnetic effects are much weaker at low speeds.

<header>
<h3>Example 1</h3>
    </header>
    
**Problem:**  
A proton moves with a speed $v = 2\times10^6\ \text{m/s}$ along the $+x$-axis. Find the magnetic field at a point $P$ located $0.05\ \text{m}$ directly above the proton (along the $+y$-axis).  

---

**Solution**:

1. **Geometry:**  
   - $\mathbf{v} = v \hat{\mathbf{x}}$  
   - $\mathbf{r} = 0.05\, \hat{\mathbf{y}}$  
   - $r = 0.05\ \text{m}$, $\hat{\mathbf{r}} = \hat{\mathbf{y}}$

2. **Cross product:**

$$
\mathbf{v} \times \hat{\mathbf{r}} = (v\hat{\mathbf{x}}) \times (\hat{\mathbf{y}}) = v \hat{\mathbf{z}}
$$

3. **Substitute into Biot–Savart:**

$$
\vec{B}= \frac{\mu_0}{4\pi}\frac{qv}{r^2}\hat{z}
$$

Substitute:

- $q = 1.6\times10^{-19}\ \text{C}$,  
- $v = 2\times10^6\ \text{m/s}$,  
- $r = 0.05\ \text{m}$,  
- $\mu_0/4\pi = 10^{-7}\ \text{T·m/A}$:

$$
B = 10^{-7}\frac{(1.6\times10^{-19})(2\times10^6)}{(0.05)^2}
$$

$$
B \approx 1.28\times10^{-17}\ \text{T}
$$

4. **Direction:**  
Since $v$ is along $+x$ and $r$ along $+y$, $B$ points along $+z$.

**Answer:**  

$$
\boxed{\mathbf{B} \approx 1.3\times10^{-17}\ \text{T}\ \hat{\mathbf{z}}}
$$

## ✅ **Key Takeaways**

1. A moving charge creates a magnetic field given by:

   $$
   \mathbf{B} = \frac{\mu_0}{4\pi} \frac{q \mathbf{v}\times \hat{\mathbf{r}}}{r^2}
   $$

2. $\mathbf{B}$ is perpendicular to both $\mathbf{v}$ and $\mathbf{r}$.

3. The magnitude is small for non-relativistic motion, scaling as $v/c^2$ relative to the electric field.

4. This result is a stepping stone to understanding **magnetic fields of currents** and eventually **Ampère’s law**.
