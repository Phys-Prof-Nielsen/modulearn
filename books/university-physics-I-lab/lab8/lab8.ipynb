## **Introduction**
At night, most solar effects are absent, which makes it easier to isolate **conduction**, **convection**, and **radiation** between objects and their surroundings. The FLIR C-series thermal camera measures **infrared radiation** emitted from surfaces and converts it into a surface temperature, assuming a given **emissivity** and **reflected ambient temperature**.

In this lab, you will use the camera around campus at night to identify and analyze examples of heat transfer. Common locations include:
- Heated pipes or radiators
- Warm air vents (HVAC systems)
- Laptop chargers or power adapters
- Building walls, doors, and window frames
- Hot beverages or containers cooling over time

You will perform both **qualitative imaging** (seeing patterns of heat transfer) and **quantitative analysis** (using simplified models to estimate parameters).

## **Theory and Models**

### 1) **Conduction**
**Conduction** is the transfer of heat through a material due to a **temperature gradient**. It occurs when faster-moving particles transfer energy to slower ones by collisions or vibrations.

The rate of conductive heat transfer is given by **Fourier’s Law**:

$$
\dot Q_{\text{cond}} = -kA\,\frac{dT}{dx}
$$

**Where:**
- $\dot Q_{\text{cond}}$ = rate of heat transfer (W)  
- $k$ = thermal conductivity of the material (W/m·K)  
- $A$ = cross-sectional area through which heat flows (m²)  
- $\frac{dT}{dx}$ = temperature gradient (K/m)

A **negative sign** indicates heat flows from higher to lower temperature.  
For a uniform slab of thickness $L$ with steady temperatures $T_{\text{hot}}$ and $T_{\text{cold}}$:

$$
\dot Q_{\text{cond}} = kA\,\frac{T_{\text{hot}} - T_{\text{cold}}}{L}.
$$

You will observe this along conductive paths such as metal handrails or pipes, where heat spreads out from a warm region.

---

### 2) **Convection**
**Convection** occurs between a surface and a moving fluid (such as air) that carries energy away or toward the surface.

**Newton’s Law of Cooling** describes this process:

$$
\dot Q_{\text{conv}} = hA\,(T_s - T_\infty)
$$

**Where:**
- $\dot Q_{\text{conv}}$ = rate of convective heat transfer (W)  
- $h$ = convective heat transfer coefficient (W/m²·K)  
- $A$ = surface area exposed to the fluid (m²)  
- $T_s$ = surface temperature (K or °C)  
- $T_\infty$ = temperature of the surrounding air (K or °C)

If the object cools uniformly (lumped system):

$$
m c \frac{dT}{dt} = -hA\,(T - T_\infty),
$$

leading to the exponential cooling model:

$$
T(t) = T_\infty + (T_0 - T_\infty)e^{-t/\tau}, \qquad \tau = \frac{m c}{hA}.
$$

**Parameters:**
- $m$ = mass of object (kg)  
- $c$ = specific heat (J/kg·K)  
- $\tau$ = time constant for cooling (s)

By measuring the surface temperature over time, $h$ can be estimated from $\tau$.

---

### 3) **Thermal Radiation**

All surfaces continuously emit **infrared radiation**, and the rate of this emission increases with the **fourth power** of their absolute temperature.  
The net rate of radiative heat transfer between a surface and its surroundings is given by the **Stefan–Boltzmann law**:

$$
\dot Q_{\text{rad}} = \epsilon \sigma A (T_s^4 - T_{\text{sur}}^4)
$$

**Where:**
- $\dot Q_{\text{rad}}$ — total radiative heat transfer rate (W)  
- $\epsilon$ — emissivity (0–1), describing how effectively the surface emits radiation compared to a perfect blackbody  
- $\sigma$ — Stefan–Boltzmann constant ($5.670 \times 10^{-8}$ W/m²·K⁴)  
- $A$ — surface area of emission (m²)  
- $T_s$ — surface temperature (K)  
- $T_{\text{sur}}$ — surrounding (environmental) temperature (K)

It is often convenient to express this as a **heat flux**, which is the rate of heat transfer per unit area:

$$
q''_{\text{rad}} = \frac{\dot Q_{\text{rad}}}{A} = \epsilon \sigma (T_s^4 - T_{\text{sur}}^4)
$$

Here \( q''_{\text{rad}} \) (read as *q double prime*) represents how much thermal energy is radiated **per square meter** of surface.  
This makes it easier to compare radiation with the other modes of heat transfer, which can also be written in flux form:

$$
q''_{\text{cond}} = -k\,\frac{dT}{dx}, \qquad q''_{\text{conv}} = h\,(T_s - T_\infty)
$$

Thus, whether heat is conducted through a solid, convected to air, or radiated to the surroundings, \( q'' \) provides a **common measure of energy flow intensity** at the surface.

For **small temperature differences**, radiation can be approximated as a linear process with an **effective heat transfer coefficient**:

$$
q''_{\text{rad}} = h_{\text{rad}} (T_s - T_{\text{sur}}), \qquad 
h_{\text{rad}} = 4 \epsilon \sigma T_m^3
$$

where \( T_m \) is the mean absolute temperature (K).  
In this linearized form, radiation behaves similarly to convection and can be combined with it when both are present.

---

### 4) **Infrared Thermography Concepts**
Thermal cameras do not measure temperature directly—they detect **infrared radiation** emitted by surfaces. The apparent temperature depends on:
- **Emissivity ($\epsilon$):** fraction of emitted radiation relative to an ideal blackbody.  
  - Matte materials (tape, paper, skin): $\epsilon \approx 0.9$–$0.98$  
  - Shiny metals: $\epsilon < 0.2$ (poor emitters, strong reflections)
- **Reflected apparent temperature ($T_{\text{refl}}$):** infrared radiation reflected from the environment.
- **Distance and viewing angle.**

The **FLIR C-series** allows you to input emissivity and reflected temperature.  
For accurate readings:
- Use **black electrical tape** on shiny surfaces as a calibration marker ($\epsilon = 0.95$).
- Avoid reflections (windows, metal surfaces).
- Work indoors or in stable air to reduce noise and convection variability.

### **Part A — Radiation and Emissivity Effects**
**Goal:** Examine how emissivity and reflection affect apparent temperature readings.

1. In a room or hallway, place a piece of **black tape** and **aluminum foil** side-by-side on a wall.
2. Set emissivity in the camera to $\epsilon = 0.95$.
3. Record an image of both materials.
4. Note how the foil may appear much colder or hotter than its surroundings due to reflection of other warm surfaces.
5. Adjust the reflected temperature $T_{\text{refl}}$ in the camera and note how readings change.

**Deliverable:** Annotated image showing both surfaces and a short explanation of how emissivity affects apparent temperature.

---

### **Part B — Conduction Along a Metal Object**
**Goal:** Visualize conduction through a solid.

1. Locate a **metal handrail**, **door hinge**, **window frame**, or **pipe** connected between a warm indoor surface and cooler air.  
2. Mark points along the metal (use tape dots if needed) at equal distances.
3. Record the temperature $T(x)$ at each point.
4. Plot $T$ vs. $x$ to determine the temperature gradient $\frac{dT}{dx}$.
5. Estimate $\dot Q_{\text{cond}}$ using:
   
   $$
   \dot Q_{\text{cond}} = kA\,\frac{T_{\text{warm}} - T_{\text{cool}}}{L},
   $$
   
   assuming typical $k$ for aluminum (≈ 200 W/m·K) or steel (≈ 50 W/m·K).

**Deliverable:** Temperature profile and discussion of how geometry and materials influence conduction.

---

### **Part C — Convective Cooling**
**Goal:** Estimate the convective heat transfer coefficient $h$ using a cooling curve.

1. Fill a **paper cup** or **ceramic mug** with hot water (~60–70 °C).  
   Apply a small piece of black tape to the cup’s side for temperature measurement.
2. Record the **initial temperature** $T_0$.
3. Leave the cup in still air and record $T(t)$ every 30 s for 5–10 min.
4. Repeat once with **gentle airflow** (fan or waving hand).
5. Fit data to:
   
   $$
   T(t) = T_\infty + (T_0 - T_\infty)e^{-t/\tau}.
   $$
   
   From $\tau$, estimate
   
   $$
   h = \frac{m c}{A\,\tau}.
   $$

**Deliverable:** Plot of $\ln(T - T_\infty)$ vs. $t$, slope $=-1/\tau$, and comparison of $h$ (still vs. air moving).

---

### **Part D — Radiation vs. Convection**
**Goal:** Separate radiative and convective contributions.

1. Repeat Part C’s test twice:
   - Case 1: high-emissivity (bare surface with tape marker)
   - Case 2: wrap the cup loosely with **aluminum foil** (low emissivity)
2. Record and analyze both cooling curves.
3. Compute $h_{\text{eff}}$ for each case and estimate $h_{\text{rad}}$:
   
   $$
   h_{\text{rad}} = h_{\text{eff,high}\,\epsilon} - h_{\text{eff,foil}}.
   $$

**Deliverable:** Table comparing time constants $\tau$, $h_{\text{eff}}$, and the estimated radiation contribution.
