# Newton’s Second Law Lab: Friction-Limited Acceleration of a Cart

---

## Purpose & Learning Goals
- Use Newton’s second law to model motion with kinetic friction on horizontal surfaces.  
- Predict the cart’s acceleration under specified friction conditions.  
- Measure acceleration with a smartphone (phyphox app) and estimate the coefficient of kinetic friction, $\mu_k$, for different surfaces.  
- Practice data quality and statistics: collect multiple trials per surface; compute the sample mean $\bar{a}$, sample standard deviation $s$, and standard error of the mean $s_{\bar{a}} = s/\sqrt{n}$.  
- Compare how added mass affects (or does not affect) friction-limited acceleration.  

---

## Background
On a level surface with **kinetic** friction opposing motion, the horizontal force model:

$$
\sum F_x = ma_x
$$

The only horizontal force is friction:

$$
-f_k = ma_x
$$

The vertical model is the balance between normal force and force of gravity:

$$
f_N = mg
$$

The magnitude of the friction force is related to the magnitude of the normal force through the **coefficient of friction ($\mu$)**: 

$$f_f = \mu f_N$$

so that  

$$
a = -\mu_k g.
$$

Key predictions:  
- During a **coast phase** (after a brief push, hands off), the cart experiences approximately constant acceleration $a \approx -\mu_k g$.  
- $a$ should be independent of mass (since $m$ cancels out), provided wheels/bearings and contact conditions don’t change.

## Experimental Design Overview

1. **Phone orientation & axis check (1–2 min)**
   - Mount the phone securely on the cart with one axis aligned along the motion direction.
   - Start **Acceleration without g** in Phyphox.  
   - Gently push the cart forward/backward while holding it in place. Identify which axis shows the response.  
     Choose that axis as your **analysis axis**.

2. **Baseline at rest (5–10 s)**
   - With the cart at rest, record a short segment.  
   - In this mode, all axes should read close to $0$ when stationary.  
   - Confirm your chosen analysis axis is stable near zero.

3. **Coast trials (repeat $n \ge 6$ per surface)**
   - **Start recording**, give a **short push**, then **hands off** immediately.  
   - Let the cart coast and slow down due to kinetic friction.  
   - **Stop recording** after it slows significantly.  
   - Try to keep the push similar across trials, but analysis will focus on the coast phase only.

4. **Export & column selection**
   - Export the file (tab-separated CSV). Header line typically:  
     `time (s) [tab] ax [tab] ay [tab] az [tab] |a|` (all in m/s²).  
   - Sampling is **100 Hz** (that is 100 recordings per second: $\Delta t \approx 0.01$ s).  
   - Select the column corresponding to your chosen analysis axis.

5. **Identify coast phase from the data**
   - In your acceleration data:  
     - The **push** appears as a sharp impulse (large positive or negative spike).  
     - The **coast phase** follows: a region of nearly constant acceleration at a negative value ($a \approx -\mu_k g$).  
   - Select a **steady segment** after the push spike but before the cart nearly stops.  
   - Use the **same duration** across all trials on a given surface.

6. **Compute one acceleration per trial**
   - Average acceleration over the coast segment to obtain $a_i$ (m/s²).  
   
7. **Repeat across conditions**
   - **Surface A**: collect $n \ge 6$ trials.  
   - **Surface B**: collect $n \ge 6$ trials.  
   
8. **Statistics & coefficient of friction**
   - For each surface, compute:  

     $$
     \bar{a} = \frac{1}{n}\sum_{i=1}^n a_i, \qquad
     s_a = \sqrt{\frac{1}{n-1}\sum_{i=1}^n (a_i - \bar{a})^2}, \qquad
     s_{\bar{a}} = \frac{s_a}{\sqrt{n}}
     $$

   - Report mean acceleration as $\bar{a} \pm s_{\bar{a}}$ (m/s²).  
   - Convert to coefficient of friction:  

     $$
     \mu_k = \frac{|\bar{a}|}{g}, \qquad
     u_\mu \approx \frac{s_{\bar{a}}}{g}.
     $$

**Notes & tips**
- Orientation must remain fixed for all trials on a surface. If the phone shifts, discard the trial.  
- Coast interval should look “flat” on the graph. If not, shorten the interval.  
- Use consistent units: time in seconds, acceleration in m/s². Report $\mu_k$ without units.

**Sub-note: Standard deviation within vs. across trials**  
- Each coast window in a single trial contains many samples (e.g., 50 data points in 0.5 s at 100 Hz). You will average these to get one acceleration value $a_i$ for that trial.  
- The **standard deviation within that window** tells you how noisy the sensor data is (random fluctuations, bumps, vibrations). This is mainly a measure of short-term noise, not the physics we are testing.  
- The **spread of averaged accelerations across multiple trials** ($a_1, a_2, \dots, a_n$) is what matters for the experiment. This variation reflects differences in pushes, cart alignment, and surface variability.  
- Therefore:  
  - You may calculate the within-trial standard deviation as a “noise check.”  
  - But the uncertainty you report for $\bar{a}$ and $\mu_k$ should come from the **across-trials standard deviation** $s$ and the standard error $s_{\bar{a}}$.

**Sub-note: Two equivalent paths to $\mu_k$ — what happens to the standard deviation?**

You can compute $\mu_k$ in either order, and (if you keep the sign) they are mathematically equivalent:

**Path A (acceleration first, then divide by $g$):**
- Compute across-trials stats of acceleration:
  
  $$
  \bar{a},\quad s_a,\quad s_{\bar{a}}=\frac{s_a}{\sqrt{n}}
  $$
  
- Convert to friction:

  $$
  \mu_k = -\frac{\bar{a}}{g},\qquad
  s_{\mu}=\frac{s_a}{g},\qquad
  s_{\bar{\mu}}=\frac{s_{\bar{a}}}{g}
  $$

  (Use the minus sign because $a\!\approx\!-\mu_k g$ on level ground; you may report $|\mu_k|$ at the end.)

**Path B (trial-wise $\mu$ first, then average):**
- For each trial, map acceleration to friction:
  
  $$
  \mu_i = -\frac{a_i}{g}
  $$
  
- Then compute across-trials stats on $\{\mu_i\}$:

  $$
  \bar{\mu}=\frac{1}{n}\sum \mu_i,\qquad
  s_{\mu}=\sqrt{\frac{1}{n-1}\sum(\mu_i-\bar{\mu})^2},\qquad
  s_{\bar{\mu}}=\frac{s_{\mu}}{\sqrt{n}}
  $$

**Key point:** Because division by $g$ is a *linear* transformation, the mean and standard deviations simply scale by $1/g$. Thus Path A and Path B give the **same** $\bar{\mu}$, $s_{\mu}$, and $s_{\bar{\mu}}$ (up to rounding), **provided you keep the sign** (i.e., use $\mu=-a/g$).

**Avoid a subtle pitfall:** If you take **magnitudes** trial-by-trial ($\mu_i = |a_i|/g$), the absolute value is *nonlinear*. If any $a_i$ crosses zero due to noise, $|\cdot|$ can bias the mean/variance slightly. Safer practice:
- Compute with **signed** values first: $\mu_i=-a_i/g$,
- Do your statistics,
- Then report $|\bar{\mu}|$ (and keep $s_{\mu}$, $s_{\bar{\mu}}$ unchanged).

**Recommendation for this lab:** Use **Path B** (trial-wise $\mu_i=-a_i/g$ then average). It reinforces the idea that each trial yields a physical estimate of $\mu_k$, and your reported uncertainty naturally reflects the *between-trial* variability.
