# Lab 11: Polar Coordinates with a Simple Pendulum — Measuring $v$ from $\omega$

**Goal:**  
To understand how the description of motion using polar coordinates $(r, \theta)$ naturally captures circular and oscillatory motion, and to measure the linear velocity of a pendulum bob using $v = R\,\omega$.

---

## Learning Objectives
- Express the position, velocity, and acceleration of a particle in polar coordinates.  
- Derive the velocity components $\dot r$ and $r\dot\theta$ and understand their directions.  
- Apply the theory to a simple pendulum and interpret physical quantities in the $r$ and $\theta$ directions.  
- Record and analyze experimental data to compare measured and theoretical velocities.  

---

## Theory and Model

### 1. Position in Polar Coordinates

For planar motion, the position vector of a particle relative to a fixed origin $O$ can be written as:

$$
\vec{r} = r\,\hat{e}_r,
$$

where:
- $r$ is the **radial distance** from the origin,
- $\hat{e}_r$ is the **radial unit vector** pointing from $O$ to the particle,
- $\hat{e}_\theta$ is the **transverse unit vector** obtained by rotating $\hat{e}_r$ by $+90^\circ$ counterclockwise.

In terms of Cartesian coordinates $(x, y)$:

$$
x = r\cos\theta, \qquad y = r\sin\theta.
$$

and the unit vectors are:

$$
\hat{e}_r = \cos\theta\,\hat{i} + \sin\theta\,\hat{j}, \qquad
\hat{e}_\theta = -\sin\theta\,\hat{i} + \cos\theta\,\hat{j}.
$$

These unit vectors **change direction** as the particle moves (because $\theta$ changes with time).

---

### 2. Time Derivatives of the Unit Vectors

Taking the time derivative of $\hat{e}_r$ and $\hat{e}_\theta$ gives:

$$
\dot{\vec{\hat{e}}}_r= \dot{\theta}\,\hat{e}_\theta, \qquad
\dot{\vec{\hat{e}}}_\theta= -\dot{\theta}\,\hat{e}_r.
$$

These relations show that as the object moves, the direction of $\hat{e}_r$ rotates with angular speed $\dot{\theta}$, and $\hat{e}_\theta$ rotates the same way but in the opposite direction.

---

### 3. Velocity in Polar Coordinates

Differentiating the position vector:

$$
\vec{v} = \dot{\vec{r}}
= \frac{d}{dt}\big(r\,\hat{e}_r\big)
= \dot{r}\,\hat{e}_r + r\,\dot{\hat{e}}_r.
$$

Substituting the derivative of $\hat{e}_r$:

$$
\vec{v} = \dot{r}\,\hat{e}_r + r\dot{\theta}\,\hat{e}_\theta.
$$

Thus, the velocity vector has two components:

- **Radial component:** $\dot{r}$ — represents the rate of change of the radius (motion toward or away from the origin).  
- **Transverse (angular) component:** $r\dot{\theta}$ — represents the tangential speed due to rotation around the origin.

The **magnitude** of velocity is:

$$
v = \sqrt{(\dot{r})^2 + (r\dot{\theta})^2}.
$$

---

### 4. Acceleration in Polar Coordinates

Differentiating again to find acceleration:

$$
\vec{a} = \dot{\vec{v}}
= \left(\ddot{r} - r\dot{\theta}^2\right)\hat{e}_r 
+ \left(r\ddot{\theta} + 2\dot{r}\dot{\theta}\right)\hat{e}_\theta.
$$

Here:
- $\ddot{r} - r\dot{\theta}^2$ is the **radial acceleration** (centripetal when $\dot{r}=0$),
- $r\ddot{\theta} + 2\dot{r}\dot{\theta}$ is the **tangential acceleration** (responsible for changing the angular speed).

---

### 5. Application to a Simple Pendulum

A pendulum of length $L$ and bob mass $m$ is modeled as a point mass moving along a **circular arc** of radius $R$. 

Thus:

$$
r = R \quad \Rightarrow \quad \dot{r} = 0, \quad \ddot{r} = 0.
$$

The motion occurs entirely through changing $\theta(t)$, so:

$$
\vec{v} = R\dot{\theta}\,\hat{e}_\theta, \qquad
\vec{a} = (-R\dot{\theta}^2)\hat{e}_r + (R\ddot{\theta})\hat{e}_\theta.
$$

At the lowest point of the swing ($\theta = 0$), $\hat{e}_r$ is vertical and $\hat{e}_\theta$ is horizontal.  
Hence, the **velocity** is purely horizontal, and the **acceleration** is momentarily directed upward (centripetal).

The angular speed $\dot\theta$ is often written as $\omega$.The magnitude of the transverse speed is then:

$$
v = R\omega.
$$

---

### 6. Energy Consistency Check

At release from rest at angle $\theta_0$:

$$
m g R(1 - \cos\theta_0) = \tfrac{1}{2} m v_{\max}^2.
$$

Solving for $v_{\max}$ gives:

$$
v_{\max} = \sqrt{2 g R (1 - \cos\theta_0)}.
$$

This provides a theoretical value for the maximum velocity at the bottom of the swing, which can be compared to the measured $v_{\max} = R\dot{\theta}_{\max}$ from experimental data.

## Equipment
- String and small pendulum bob (metal washer or dense sphere)  
- Rigid support or clamp stand  
- Meter stick or measuring tape  
- Protractor (or printed angular scale)  
- Stopwatch or smartphone (slow-motion camera preferred)  

---

## Setup
1. Suspend the bob from a fixed point; measure the pendulum length $L$ from pivot to bob center.  
2. Set up a protractor behind the pendulum with $\theta=0$ at the vertical equilibrium position.  
3. Mark angular displacements in $5^\circ$ intervals for easy reference.  

---

## Procedure
1. Pull the bob aside to an initial small angle ($\theta_0 \approx 10^\circ$).  
2. Release it without pushing and start recording.  
3. Using slow-motion video or timing marks, record the angular position $\theta(t)$ over several oscillations.  
4. Convert all angles to radians.  
5. Compute $\dot{\theta}(t)$ using finite differences and obtain $v(t) = R\dot{\theta}(t)$.  

---

## Analysis
1. Plot $\theta(t)$, $\dot{\theta}(t)$, and $v(t)$.  
2. Identify the times when $\theta = 0$ (pendulum passes equilibrium) and determine $v_{\max}$.  
3. Compare measured $v_{\max}$ to the theoretical value $\sqrt{2gR(1 - \cos\theta_0)}$.  
4. Discuss sources of error such as timing precision, parallax, and amplitude reduction due to air resistance.  

---

## Discussion Questions
- Why are $\hat{e}_r$ and $\hat{e}_\theta$ not constant in time even when $r$ is fixed?  
- How does the polar coordinate system simplify the analysis compared to using $x$ and $y$?  
- At what point is $\vec{v}$ perpendicular to $\vec{r}$? Why?  
- What would change if the string could stretch (i.e., $r$ not constant)?  

---

## Conclusion
Polar coordinates naturally describe the pendulum’s motion because only the angle $\theta$ changes while the radius $r=R$ is constant. The velocity direction and magnitude follow directly from the time rate of change of $\theta$, leading to $v = R\dot{\theta}$. Experimentally measuring $\theta(t)$ provides a clear connection between rotational and linear motion, reinforcing how circular motion is simply a special case of two-dimensional motion in polar coordinates.
