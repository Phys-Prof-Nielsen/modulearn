## Introduction and Theory

In this lab we will determine the launch speed of a projectile using a ballistic pendulum, then use that speed to predict where the ball will land when fired horizontally from a table. This connects **conservation of momentum**, **work done by gravity**, and **projectile motion**.

---

### Part I: Ballistic Pendulum

- A steel ball of mass $m$ is fired into a pendulum of mass $M$.  
- The ball embeds in the pendulum (perfectly inelastic collision).  
- Momentum conservation during the collision:
  
$$
m v = (M + m)V
$$

where $v$ = ball speed before collision, $V$ = pendulum+ball speed after collision.

- After collision, pendulum+ball swing upward. Kinetic energy is transferred into energy in the pendulum+ball system (known as potential energy) through work done against gravity:

$$
\frac{1}{2}(M+m)V^2 = (M+m) g h
$$

which gives

$$
V = \sqrt{2 g h}
$$

- Combining both:
  
$$
v = \frac{M+m}{m} \sqrt{2 g h}
$$

Thus, by measuring the rise height $h$, we determine the launch speed $v$.

---

### Part II: Projectile Prediction Challenge

Now, the launcher is aimed horizontally from a tabletop of height $H$. Using the speed $v$ derived above, predict the horizontal distance $x$ where the ball will land.

#### General Kinematic Equations

For a projectile launched from the origin at $t=0$ with initial velocity components (ignoring air drag):
- $v_x = v \cos \theta$
- $v_y = v \sin \theta$

the position as a function of time is:

$$
x(t) = v_x t
$$

$$
y(t) = v_y t - \tfrac{1}{2} g t^2
$$

---

#### Special Case: Horizontal Launch

For a horizontal launch, $\theta = 0^\circ$, so:
- $v_x = v$
- $v_y = 0$

Equations simplify to:

$$
x(t) = v t
$$

$$
y(t) = - \tfrac{1}{2} g t^2
$$

If the ball is launched from height $H$ above the floor, then the ball hits the floor when:

$$
H = \tfrac{1}{2} g t^2
$$

$$
t = \sqrt{\tfrac{2H}{g}}
$$

At this time, the horizontal displacement is:

$$
x = v \, t = v \sqrt{\tfrac{2H}{g}}
$$

Procedure:
1. Measure the table height $H$.  
2. Use your calculated $v$ from Part I.  
3. Compute $t$ and $x$.  
4. Mark the predicted spot $x$ from the table edge on the floor.  
5. Place carbon paper target ("bulls eye").  
6. Fire **one shot**:
   - If bulls eye is hit → class wins pizza!  
   - If missed, record data, adjust prediction, and try again (no pizza).  

Careful measurement and analysis are crucial!
