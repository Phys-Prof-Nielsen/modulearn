## Experimental Description
You will estimate your reaction time by catching a vertically dropped ruler **as fast as you can**. When the ruler is released **without warning**, your fingers close and catch it after some delay. The distance the ruler falls before being caught corresponds to the time it takes you to react.

---

## Methodology 

1. **Setup & Calibration**
   - One person (the **catcher**) places their thumb and forefinger slightly apart at a fixed height.  
   - The other person (**dropper**) holds the ruler vertically with the **0 mark just above** the catcher’s fingers, ensuring the scale increases downward.
   - Confirm a consistent grip and release orientation every time (control systematic effects).

2. **Measurement Protocol**
   - Without a verbal countdown, the dropper releases the ruler at random intervals.  
   - The catcher closes their fingers as quickly as possible to catch the ruler.  
   - Read the mark at the **top edge** of the catcher’s fingers (be consistent!) to the appropriate resolution. Record **distance $d$**.
   - **Repeat for $n=10$ trials** per student (more if time allows).

3. **Recording Data**
   - Use the data table template below. Include **units** and note the **instrument resolution**.
   - If a trial is clearly flawed (e.g., you **anticipated** the drop or fumbled), mark it and justify whether to exclude it. Do **not** delete data without a reason; document all decisions.

4. **Converting to Time**
   - For each trial, convert distance to time via
     
     $$
     t_i = \sqrt{\frac{2d_i}{g}}\,.
     $$
     
     Remember to convert $d_i$ to **meters** before computing $t_i$ in **seconds**.

5. **Statistical Analysis**
   - Compute $\bar{t}$, $s_t$, and $s_{\bar{t}}=s_t/\sqrt{n}$.  
   - Make a simple plot testing for consistency.

6. **Uncertainty Discussion**
   - Compare **instrument/readout resolution** to your observed **spread** $s_d$. Which dominates?  (this will require to you to also calculate $\bar{d}$ and $s_d$)
   - Identify potential **systematic** effects (e.g., consistent reading offset, anticipation, different droppers).  

7. **Optional Checks**
   - Swap roles (new dropper vs catcher) to see if mean reaction time changes—possible systematic shift.  
   - Repeat a subset of trials later to check for **learning/fatigue** effects.

## Common Pitfalls & How to Avoid Them
- **Unit slips:** Always convert $d$ to meters before computing $t$ in seconds.  
- **Inconsistent read point:** Always read at the **same edge** of the fingers.  
- **Anticipation bias:** Vary release times; avoid countdowns.  
- **Over-rounding:** Keep adequate significant figures during calculation; round only when **reporting** final results.
