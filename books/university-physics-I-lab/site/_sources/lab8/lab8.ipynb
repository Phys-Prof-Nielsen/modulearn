## **Introduction**

In this lab, we explore one of the most fundamental relationships in mechanics — **Hooke’s Law**, which describes how springs respond to applied forces. Springs are simple yet powerful models for understanding oscillations, restoring forces, and energy storage in physical systems.  

When a force $F$ is applied to stretch or compress a spring, the spring resists deformation with a restoring force that is proportional to the displacement $x$ from its equilibrium position:

$$
F = -kx
$$

The proportionality constant $k$ is the **spring constant**, which measures the stiffness of the spring. A large $k$ indicates a stiffer spring, while a small $k$ corresponds to a more flexible spring.

In this experiment, you will first determine the spring constant of several individual springs, then explore how springs behave when combined in **series** and **parallel** configurations. This provides insight into how complex mechanical systems distribute forces and displacements — a principle that has direct analogies in electrical circuits.

---

## **Theory and Model**

### **1. Hooke’s Law**

When a spring is stretched or compressed by a displacement $x$, the restoring force it exerts is given by:

$$
F = kx
$$

In equilibrium, the applied force and the spring’s restoring force are equal in magnitude:

$$
F_{\text{applied}} = kx
$$

If we hang a mass $m$ from a vertical spring, the equilibrium position occurs when the gravitational force balances the spring force:

$$
mg = kx
$$

Thus, the spring constant can be found from the slope of a graph of **force ($F = mg$)** versus **displacement ($x$)**.

---

### **2. Springs in Series**

When two springs are connected **in series**, the same force acts through each spring. We can understand that by considering static equilibrium (when the system is at rest), the net force on every point in the system must be zero.
Consider the junction point where the two springs meet:

The upper spring pulls upward on the junction with a restoring force $F_1 = k_1 x_1$.

The lower spring pulls downward on the same junction with a force $F_2 = k_2 x_2$.

For the junction not to accelerate (Newton’s 2nd law), these two forces must balance. That means the tension (or force) transmitted through the series of springs is the same everywhere — just like the tension in a rope holding a mass.

The total displacement is the sum of the displacements of each spring:

$$
x_{\text{total}} = x_1 + x_2 = \frac{F}{k_1} + \frac{F}{k_2}
$$

Rearranging gives the **equivalent spring constant** $k_{\text{series}}$:

$$
\frac{1}{k_{\text{series}}} = \frac{1}{k_1} + \frac{1}{k_2}
$$

This configuration results in a **softer system** — the equivalent spring constant is **smaller** than either individual spring.

---

### **3. Springs in Parallel**

For springs in **parallel**, each spring resists that same displacement with its own restoring force according to Hooke’s law. Because both springs pull upward on the same mass, the total upward restoring force is the sum of these individual forces:

$$
F_{\text{total}} = F_1 + F_2 = k_1x + k_2x
$$

Therefore, the **equivalent spring constant** $k_{\text{parallel}}$ is:

$$
k_{\text{parallel}} = k_1 + k_2
$$

This configuration creates a **stiffer system**, since the equivalent spring constant is **larger** than that of either individual spring.

---

### **4. Electrical Analogy**

The relationships above mirror the way **resistors** and **capacitors** combine in electrical circuits:

| System Type | Series Relationship | Parallel Relationship |
|--------------|--------------------|------------------------|
| **Springs** | $\dfrac{1}{k_{\text{eq}}} = \dfrac{1}{k_1} + \dfrac{1}{k_2}$ | $k_{\text{eq}} = k_1 + k_2$ |
| **Resistors** | $R_{\text{eq}} = R_1 + R_2$ | $\dfrac{1}{R_{\text{eq}}} = \dfrac{1}{R_1} + \dfrac{1}{R_2}$ |
| **Capacitors** | $\dfrac{1}{C_{\text{eq}}} = \dfrac{1}{C_1} + \dfrac{1}{C_2}$ | $C_{\text{eq}} = C_1 + C_2$ |

This analogy helps bridge mechanical and electrical systems — both obey similar mathematical structures that govern energy storage and transfer.

## **Methodology**

### **Equipment**
- 3–4 springs of varying stiffness  
- Stand and clamps  
- Mass hanger and a set of masses  
- Ruler or meter stick  
- Digital balance (optional, to verify mass)  

---

### **Part 1: Determining the Spring Constant of Individual Springs**

1. Hang a spring vertically and measure its **initial length** (no mass attached).  
2. Add masses incrementally (e.g., 50 g, 100 g, 150 g, …) and record the **elongation $x$** after each addition.  
3. For each trial, calculate the **applied force** using $F = mg$.  
4. Plot $F$ (y-axis) vs. $x$ (x-axis). The slope of the linear fit gives the spring constant $k$.  
5. Repeat for all springs and tabulate results.

---

### **Part 2: Springs in Series**

1. Connect two springs in series and hang them vertically.  
2. Repeat the same measurement process as in Part 1.  
3. Determine $k_{\text{series}}$ experimentally from the slope of the $F$ vs. $x$ plot.  
4. Compare the result to the theoretical prediction:

$$
\frac{1}{k_{\text{series,th}}} = \frac{1}{k_1} + \frac{1}{k_2}
$$

---

### **Part 3: Springs in Parallel**

1. Connect two springs side by side (in parallel), both supporting the same load.  
2. Record elongations for different applied forces.  
3. Determine $k_{\text{parallel}}$ experimentally.  
4. Compare to theory:

$$
k_{\text{parallel,th}} = k_1 + k_2
$$

---

- Finally, discuss the analogy to electrical networks and how similar mathematical structures appear in physics across different domains.
