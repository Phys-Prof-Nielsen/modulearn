## Why LaTeX?

LaTeX is a document preparation system used for creating high-quality, professionally formatted documents, particularly in fields that require mathematical expressions such as physics, mathematics, engineering, and computer science. Unlike traditional word processors that rely on point-and-click formatting, LaTeX uses plain text and markup commands to define the structure and content of a document. This allows for precise control over layout, consistent styling, and seamless integration of complex elements like equations, references, and figures. It was originally developed by Leslie Lamport as a user-friendly layer on top of TeX, the typesetting engine created by Donald Knuth.

In scientific writing, LaTeX is widely regarded as the gold standard for typesetting equations and technical content. It enables authors to produce clean, readable documents that meet the formatting requirements of academic journals and conferences. With LaTeX, users can easily write expressions like fractions, summations, integrals, and Greek symbols, and the results look polished and publication-ready. Beyond math, LaTeX also excels at handling bibliographies, cross-references, tables, and figures in a logical and consistent way. While it has a learning curve, mastering LaTeX equips students and researchers with a skill that is respected and expected across many scientific disciplines.

## Tips

- Always wrap math in `$...$` or `$$...$$`
- Use `{}` to group multi-character subscripts or exponents
- Use `\text{}` to write words inside math mode
- Use `\quad` or `\;` to add space between terms
