# Instructor Only — Pre-Inventory Scoring Guide

**Do not release this cell until students have submitted the pre-inventory.**

## Answer Key

| Question | Correct Answer | Target Concept |
|---|:---:|---|
| Q1 | B | A measurement is an uncertain estimate. |
| Q2 | C | Variability informs measurement repeatability. |
| Q3 | B | Repetition can reduce random uncertainty in the mean. |
| Q4 | C | Multiple inputs contribute to output uncertainty. |
| Q5 | B | Absolute uncertainty scales with sensitivity. |
| Q6 | B | Independent variances combine additively. |
| Q7 | B | A single measurement can be described by a probability distribution. |
| Q8 | C | Nonlinear transformations can alter distribution shape. |
| Q9 | B | Interpretation of probability intervals. |
| Q10 | B | Simulated samples are not new experimental observations. |
| Q11 | Rubric | Conceptual description of computational propagation. |
| Q12 | C | Agreement within uncertainty is not proof of accuracy. |

## Scoring

- Q1–Q10 and Q12: 1 point each for the correct choice; otherwise 0.
- Q11: 0–2 points, using the rubric below.
- **Maximum total: 13 points.**
- Background and confidence responses are not included in the concept score.

### Q11 Short-Response Rubric

| Score | Criterion |
|---:|---|
| 0 | No meaningful explanation, or treats simulation as new physical measurements. |
| 1 | Recognizes that possible input values can be generated and used in calculations, but does not explain how the resulting uncertainty is determined. |
| 2 | Describes generating possible values of both inputs, evaluating the physical model for each pair, and examining the distribution of calculated results. |

## Suggested Reporting

Retain item-level responses and the Q11 rubric score.

Report descriptive pre/post scores and matched item-level changes where feasible.

Treat the S1–S8 confidence responses as a separate dataset; do not combine them with the conceptual score.
