## Part A — Measurement and Uncertainty

### Q1. A Single Measurement

A student measures the length of a pendulum using a ruler and obtains

$$
L=0.850\ \mathrm{m}.
$$

Which statement best describes this measurement?

**A.** The true length must be exactly 0.850 m because the ruler provides a numerical value.

**B.** The measured value is an estimate of the length, and some uncertainty remains even if the measurement is performed carefully.

**C.** The measurement has no uncertainty unless the student repeats it.

**D.** The uncertainty is always equal to the smallest marking on the ruler.

### Q2. Repeated Measurements

A student measures the period of a pendulum ten times and obtains slightly different values.

What is the most appropriate interpretation of this variation?

**A.** The pendulum's gravitational acceleration changes between measurements.

**B.** The measurements are invalid because a constant physical quantity should always produce identical results.

**C.** The variation provides information about the repeatability of the measurement process.

**D.** The largest measured period must be closest to the true value.

### Q3. Improving Measurement Precision

A student measures the period of a pendulum using ten independent timing trials. The student then performs another ten trials under identical conditions.

Which statement is most appropriate?

**A.** Increasing the number of trials automatically eliminates all measurement uncertainty.

**B.** Additional trials can reduce the random uncertainty associated with the estimated mean period, but systematic effects may remain.

**C.** The uncertainty in the mean period must remain unchanged.

**D.** The uncertainty in each individual timing measurement is reduced by half.

## Part B — Propagation of Uncertainty

### Q4. Uncertainty in a Calculated Quantity

The area of a rectangular plate is calculated using

$$
A=LW,
$$

where both length $L$ and width $W$ have measurement uncertainty.

Which statement is correct?

**A.** The area has no uncertainty because it is calculated mathematically.

**B.** Only the measurement with the larger numerical uncertainty contributes to the uncertainty in area.

**C.** Uncertainty in both length and width can contribute to uncertainty in the calculated area.

**D.** The uncertainty in area must equal the sum of the numerical uncertainties in length and width.

### Q5. Sensitivity to Uncertainty

Two quantities are calculated using the same measured variable $x$:

$$
A=2x,\qquad B=10x.
$$

If the standard uncertainty in $x$ is $\sigma_x$, which statement is correct?

**A.** Both calculated quantities have the same absolute uncertainty.

**B.** Quantity $B$ has five times the absolute uncertainty of quantity $A$.

**C.** Quantity $A$ has five times the absolute uncertainty of quantity $B$.

**D.** Neither calculated quantity has uncertainty because multiplication by a constant is exact.

### Q6. Combining Independent Uncertainties

Two independent measurements contribute to the uncertainty in a calculated physical quantity.

Why is it generally inappropriate to assume that the resulting standard uncertainty is simply the arithmetic sum of their individual contributions?

**A.** Measurement uncertainties always cancel each other.

**B.** Independent random contributions can have different signs, so their variances—not their standard uncertainties—combine additively in first-order propagation.

**C.** Only the largest uncertainty should be included.

**D.** The arithmetic sum is always correct, regardless of the physical relationship.

## Part C — Probability Distributions and Nonlinear Relationships

### Q7. Representing Uncertainty

A student measures the length of an object once and estimates that its actual length is equally plausible anywhere between 19.8 cm and 20.2 cm.

Which representation best describes this information?

**A.** A single exact value of 20.0 cm with no uncertainty.

**B.** A uniform probability distribution between 19.8 cm and 20.2 cm.

**C.** A distribution in which values outside the stated interval are more probable than those inside.

**D.** Ten additional measurements that have not actually been performed.

### Q8. Nonlinear Transformations

Suppose a measured quantity $x$ has a symmetric uncertainty distribution, and we calculate

$$
y=x^2.
$$

Which statement is most appropriate?

**A.** The distribution of $y$ must always have exactly the same shape as that of $x$.

**B.** The uncertainty in $y$ must be zero.

**C.** A nonlinear transformation can change the shape and symmetry of the resulting distribution.

**D.** The calculated values of $y$ must be uniformly distributed.

## Part E — Physical Interpretation

### Q12. Agreement Between Theory and Experiment

A student calculates a physical quantity and obtains

$$
q_{\mathrm{exp}}=9.7\pm0.3.
$$

$$
q_{\mathrm{theory}}=9.8.
$$

**B.** The experiment is necessarily incorrect because the central values differ.

**C.** The experimental result is consistent with the theoretical value at the reported uncertainty level, although this does not establish that all systematic errors have been eliminated.

**D.** The experimental uncertainty must be zero because the accepted value lies within the reported interval.
