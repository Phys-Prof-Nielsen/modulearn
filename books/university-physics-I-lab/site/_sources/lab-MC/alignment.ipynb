## Methodological Notes for the Manuscript

- Describe the instrument as **researcher-developed**, not as a previously validated inventory.
- Obtain expert feedback on content alignment, clarity, and distractors before administration.
- Pilot the questions, if feasible, to identify ambiguity and unexpected interpretations.
- Keep conceptual understanding, mathematics/programming preparedness, and self-reported confidence as distinct constructs.
- Retain anonymous matched identifiers if paired pre/post analysis is planned and permitted under the approved research protocol.
- Distinguish conceptual learning from implementation proficiency; recognizing what a simulation does is not the same as independently writing code.
- MC propagates an assigned input uncertainty model. It does not automatically correct biased measurements or an incomplete physical model.

### Interpretation Caution

Any pre/post gain should be described in relation to the instructional intervention and study design.

Without an appropriate comparison condition, do not attribute the entire change uniquely to the Monte Carlo method rather than to instruction, repeated exposure, or the associated experimental activity.
