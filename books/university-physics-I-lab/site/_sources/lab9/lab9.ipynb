## **Introduction**

When a gas is compressed rapidly, its temperature increases. If compression happens quickly enough that there is negligible heat exchange with the surroundings, the process is approximately **adiabatic**.

The First Law of Thermodynamics states:

$$
\Delta U = Q - W_{\text{by}}
$$

For rapid compression:

$$
Q \approx 0
$$

so

$$
\Delta U \approx W_{\text{on}}
$$

The work done *on* the gas increases its internal energy, which raises its temperature.

In this experiment, we use a fire syringe to compress air rapidly. If the temperature rise is sufficient, a small piece of cotton ignites.

---

## **Theory and Model**

### 1. Ideal Gas Law

$$
PV = nRT
$$

---

### 2. Adiabatic Compression

For an adiabatic process:

$$
PV^{\gamma} = \text{constant}
$$

For air:

$$
\gamma \approx 1.4
$$

The temperature after compression is:

$$
T_2 = T_1 \left(\frac{V_1}{V_2}\right)^{\gamma - 1}
$$

The pressure after compression is:

$$
P_2 = P_1 \left(\frac{V_1}{V_2}\right)^{\gamma}
$$

---

### 3. Isothermal Compression

For slow compression:

$$
T_2 = T_1
$$

and

$$
P_2 = P_1 \left(\frac{V_1}{V_2}\right)
$$

## **Part 1: Measuring the Minimum Volume Using High-Speed Video**

### Setup

1. Mount phone securely.
2. Set camera to **Slo-Mo (240 fps if available)**.
3. Position camera perpendicular to the syringe barrel.
4. Ensure volume markings are clearly visible.

### Procedure

1. Measure the initial volume $V_1$.
2. Insert a very small piece of cotton.
3. Start recording in slow motion.
4. Rapidly compress the plunger.
5. Review the video frame-by-frame.
6. Identify the frame where the plunger reaches minimum position.
7. Record $V_2$ from the scale.
8. Record whether ignition occurred.
9. Repeat 3–5 times.

## **Part 4: Estimating the Ignition Temperature Assuming Adiabatic Compression**

In this part, we assume the compression is approximately **adiabatic**.  
This allows us to use the ideal adiabatic model to estimate the ignition temperature of the cotton.

---

### **Step 1: Determine the Ignition Compression Ratio**

From your experimental data, identify the smallest compression ratio

$$
r = \frac{V_1}{V_2}
$$

that reliably produces ignition.

---

### **Step 2: Use the Adiabatic Model**

For an adiabatic process:

$$
T_2 = T_1 r^{\gamma - 1}
$$

where:

- $T_1$ = initial room temperature (in Kelvin)
- $\gamma = 1.4$ for air

At the ignition threshold, we assume:

$$
T_{\text{ignite}} \approx T_2
$$

Therefore:

$$
T_{\text{ignite}} \approx T_1 r^{\gamma - 1}
$$

This gives an **experimental estimate** of the ignition temperature.

### **Conceptual Interpretation**

If the real system loses heat during compression, then the true peak temperature is lower than the ideal adiabatic prediction.

This means:

- The ideal adiabatic model likely **overestimates** the true temperature.
- Therefore, your calculated ignition temperature may be **higher than the true material ignition temperature**.

This comparison allows you to evaluate how closely your real system approaches ideal thermodynamic behavior.
