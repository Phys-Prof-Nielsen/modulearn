## Background

### Sound as a Function
A sound wave can be represented as a function of time:

$$
x(t)
$$

This could be your voice, a tuning fork, or a musical note. Every sound can be thought of as a combination of simple sine and cosine waves of different frequencies — just like any vector can be built from the sum of basis vectors.

---

### Dot Product, Orthogonality, and the Idea Behind Fourier Analysis

In physics and math, we often describe an object by how much it “lines up” with a set of **basis vectors**.  
For example, a 3D vector $\vec{v}$ can be written as

$$
\vec{v} = v_x \hat{i} + v_y \hat{j} + v_z \hat{k},
$$

where the unit vectors $\hat{i}, \hat{j}, \hat{k}$ are **orthogonal** — they point in perpendicular directions.  
The **dot product** tells us how much of $\vec{v}$ points in a given direction:

$$
v_x = \vec{v}\cdot \hat{i}, \quad v_y = \vec{v}\cdot \hat{j}, \quad v_z = \vec{v}\cdot \hat{k}.
$$

We can do the **same thing with functions** instead of vectors.

---

#### (a) Dot Product for Functions
The dot product of two functions over one full period $T$ is defined as

$$
\langle f, g \rangle = \frac{1}{T} \int_0^T f(t)\,g(t)\,dt.
$$

If this equals zero, the two functions are **orthogonal** (just like perpendicular vectors):

$$
\langle f, g \rangle = 0.
$$

---

#### (b) Orthogonality of Sine and Cosine Functions
The sine and cosine functions of different frequencies are orthogonal over a full number of periods:

$$
\frac{1}{T}\int_0^T \sin(2\pi n t/T)\,\sin(2\pi m t/T)\,dt = 0 \quad \text{for } n \neq m,
$$

and similarly for cosines.  
Also,

$$
\frac{1}{T}\int_0^T \sin(2\pi n t/T)\,\cos(2\pi m t/T)\,dt = 0.
$$

This means each sine or cosine of a different frequency points in its own “direction” in function space.

---

#### (c) Fourier Series as Expansion in Orthogonal Functions
Any reasonably smooth, periodic signal $x(t)$ with period $T$ can be represented as a **Fourier series**:

$$
x(t) = a_0 + \sum_{n=1}^{\infty}\big[a_n \cos(2\pi n t/T) + b_n \sin(2\pi n t/T)\big].
$$

The coefficients $a_n$ and $b_n$ are found by taking **dot products** of the signal with the basis functions:

$$
a_n = \frac{2}{T}\int_0^T x(t)\cos(2\pi n t/T)\,dt,
$$

$$
b_n = \frac{2}{T}\int_0^T x(t)\sin(2\pi n t/T)\,dt.
$$

So, to find “how much” of each frequency $n/T$ is in the signal, we project $x(t)$ onto each sine and cosine — just like finding $v_x$, $v_y$, and $v_z$ for a vector.

---

#### (d) A Simple Example

Suppose we have a signal made of two pure tones:

$$
x(t) = \sin(2\pi f_1 t) + 0.5\sin(2\pi f_2 t),
$$

where $f_1=200$ Hz and $f_2=400$ Hz.

If we take the dot product of $x(t)$ with $\sin(2\pi f t)$ for different $f$, we’ll get:
- a **large value** when $f=f_1$ or $f_2$ (strong match),  
- and nearly **zero** when $f$ is different (no overlap).

That’s exactly what the **Fourier Transform** does — it measures how much each pure sine and cosine contributes to the sound.

---

**Summary Idea:**  
- Think of sines and cosines as **unit vectors** in a space of functions.  
- The Fourier coefficients $a_n$ and $b_n$ are the **dot products** between your signal and each of those “basis waves.”  
- The **FFT plot** simply shows the length (magnitude) of these projections at every frequency.




---

#### Fourier Transform (FFT in phyphox)
The FFT (Fast Fourier Transform) is just a computer version of this projection idea.  
You’ll see a graph with frequency on the horizontal axis and amplitude (strength) on the vertical axis.  
- Peaks show the **frequencies that make up your sound**.  
- The largest peak is the **fundamental frequency**, which gives the **pitch**.  
- Smaller peaks at multiples of that frequency are **harmonics** that shape the tone or timbre.

Think of FFT as taking your sound and asking:
> “How strongly does this sound match each pure tone?”

---

### Autocorrelation — Comparing a Function with Itself

So far, the Fourier approach compared your signal $x(t)$ with a set of **different basis functions** (sines and cosines).  
Now we’ll compare the signal with **itself**, but shifted in time — that’s called **autocorrelation**.

---

#### (a) The Basic Idea

Imagine you slide a copy of the signal over itself and, at each shift, measure how well they overlap.  
When the peaks of one line up with the peaks of the other, the overlap (dot product) is large.  
When they don’t line up, the overlap is small or even negative.

Mathematically, the autocorrelation function is

$$
R_{xx}(\tau) = \frac{1}{T}\int_0^T x(t)\,x(t+\tau)\,dt
$$

where $\tau$ is the **time shift** (sometimes called “lag”).  
This is another kind of **dot product**, but between the signal and a shifted version of itself.

---

#### (b) Physical Meaning

- At $\tau = 0$, the signal lines up perfectly with itself, so $R_{xx}(0)$ is the **maximum value**.
- As we increase $\tau$, the two signals start to misalign.  
  When they line up again (for example, after one full period), $R_{xx}(\tau)$ becomes large again.

If the signal repeats every $T_0$ seconds, then $R_{xx}(\tau)$ will have peaks at

$$
\tau = T_0,\; 2T_0,\; 3T_0,\;\dots
$$

From the first peak (after $\tau=0$), we can find the period $T_0$ and therefore the frequency

$$
f_0 = \frac{1}{T_0}.
$$

---

#### (c) Example — A Simple Sine Wave

Let’s take a pure tone:

$$
x(t) = \sin(2\pi f_0 t),
$$

where $f_0$ is its frequency and $T_0 = 1/f_0$ is its period.

The **autocorrelation** is defined as

$$
R_{xx}(\tau) = \frac{1}{T_0}\int_0^{T_0} x(t)\,x(t+\tau)\,dt.
$$

Substitute the sine wave:

$$
R_{xx}(\tau) = \frac{1}{T_0}\int_0^{T_0} \sin(2\pi f_0 t)\,\sin[2\pi f_0(t+\tau)]\,dt.
$$

To simplify, we use the trigonometric identity:

$$
\sin A\,\sin B = \tfrac{1}{2}[\cos(A-B) - \cos(A+B)].
$$

Then:

$$
R_{xx}(\tau)
= \frac{1}{2T_0}\int_0^{T_0} \Big[\cos(2\pi f_0 \tau) - \cos(4\pi f_0 t + 2\pi f_0 \tau)\Big]\,dt.
$$

The first term, $\cos(2\pi f_0 \tau)$, is constant with respect to $t$, so its integral gives

$$
\frac{1}{2T_0}\int_0^{T_0}\cos(2\pi f_0 \tau)\,dt
= \frac{1}{2}\cos(2\pi f_0 \tau).
$$

The second term,

$$
\int_0^{T_0}\cos(4\pi f_0 t + 2\pi f_0 \tau)\,dt,
$$

contains two parts inside the cosine:

- The $4\pi f_0 t$ part depends on time and completes **two full oscillations** during the integration from $t=0$ to $t=T_0$, because the cosine argument changes by $4\pi f_0 T_0 = 4\pi$.
- The $2\pi f_0\tau$ part is just a **constant phase shift** — it slides the cosine curve left or right, but it does not change the area under one full number of cycles.

Since $\cos(4\pi f_0 t + \text{constant})$ spends equal time positive and negative over complete cycles, its average value over any integer number of periods is zero:

$$
\int_0^{T_0}\cos(4\pi f_0 t + 2\pi f_0 \tau)\,dt = 0.
$$

Therefore, the entire second term vanishes, and only the constant term
$\tfrac{1}{2}\cos(2\pi f_0 \tau)$ remains.


---

#### Interpretation

- At $\tau = 0$: $R_{xx}(0) = \tfrac{1}{2}$ → perfect alignment (maximum similarity).  
- At $\tau = T_0/2$: $R_{xx}(T_0/2) = \tfrac{1}{2}\cos(\pi) = -\tfrac{1}{2}$ → peaks of one line up with valleys of the other (opposite phase).  
- At $\tau = T_0$: $R_{xx}(T_0) = \tfrac{1}{2}\cos(2\pi) = \tfrac{1}{2}$ → perfect alignment again.

So $R_{xx}(\tau)$ repeats every $T_0$, showing that the **autocorrelation of a sine wave is a cosine function** of the same frequency.  
The “lag” $\tau$ where $R_{xx}$ reaches its first positive maximum (after zero) tells us the **period**, and therefore the **frequency** of the signal:

$$
f_0 = \frac{1}{T_0}.
$$

This simple result explains why the **autocorrelation plot** in phyphox shows evenly spaced peaks for any repeating sound — each peak corresponds to the signal lining up with itself again.


---

#### (d) Why It Works for Real Sounds

Most real sounds (voice, instruments, whistles) are *not* perfect sine waves — they are combinations of many frequencies.  
Even so, if the sound repeats regularly (has a pitch), its overall shape still lines up again after roughly the same time $T_0$, and we see that in $R_{xx}(\tau)$ as repeating peaks.

Autocorrelation therefore tells us:

- **Pitch or fundamental period** of a sound, even when the FFT has many harmonics.  
- **Beat patterns** when two similar frequencies interfere.  

---

#### (e) Relationship to Fourier Transform

Both Fourier and autocorrelation are based on **overlap or similarity**, but they ask different questions:

| Method | What’s compared | What it tells you | Analogy |
|:--|:--|:--|:--|
| **Fourier Transform** | Compare $x(t)$ with many sine/cosine waves | Which frequencies are present | “How much of each tone?” |
| **Autocorrelation** | Compare $x(t)$ with itself shifted by $\tau$ | How the signal repeats in time | “When does it sound the same again?” |

---

#### (f) Summary of the Intuition

- The Fourier Transform measures **how much your signal overlaps** with *different* sine and cosine waves.
- The Autocorrelation measures **how much your signal overlaps** with *itself*, shifted in time.

Both use the same central concept:  
> “Compare two functions using a dot product to see how similar they are.”



---

### 5) Comparison of the Two Methods

| Method | What it compares | What it shows best | When to use |
|:--|:--|:--|:--|
| **Fourier Transform** | The signal with pure sine/cosine waves | Frequencies and harmonics | Identifying tones and timbre |
| **Autocorrelation** | The signal with a delayed copy of itself | Periods, repetition, echoes, and beats | Finding pitch and delays |

They are two sides of the same coin:
- Fourier looks at **what frequencies** are inside the sound.  
- Autocorrelation looks at **how often** the sound repeats in time.

## 4) Key Takeaway

Both FFT and Autocorrelation are ways of comparing one function with another:
- FFT compares your sound with **mathematically perfect tones** (sines and cosines).  
- Autocorrelation compares your sound with **itself shifted in time**.

Both rely on the idea that functions can be **orthogonal**, and that we can find out “how much” one function overlaps with another — just like taking a **dot product** between vectors.
