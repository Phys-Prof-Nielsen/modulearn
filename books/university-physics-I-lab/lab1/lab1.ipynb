## Physics background (short)
Hooke’s Law states that the linear spring force is (in N)

$$
F_s = kx,
$$

where $x$ is the displacement (or extension) from spring equilibrium (m), and $k$ is the spring constant (N/m). In our data set, an applied force due to weight of various masses will stretch the spring from equilibrium. The force due to gravity is

$$
F_g = mg
$$

where $m$ is the applied mass and $g = 9.89665~\text{m/s}^2$ is the acceleration of gravity. Measurement noise and anb outlier have been applied to the dataset to emulate real experiments.
