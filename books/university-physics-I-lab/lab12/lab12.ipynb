## **Background Theory**

### **1. Simple Harmonic Motion**

A system is in simple harmonic motion when the restoring influence is proportional to displacement and directed toward equilibrium.

A common mathematical model for SHM is

$$
x(t) = A\cos(\omega t + \phi)
$$

where:

- $A$ is the amplitude  
- $\omega$ is the angular frequency  
- $\phi$ is the phase constant  

The angular frequency is related to the period $T$ by

$$
\omega = \frac{2\pi}{T}
$$

and the ordinary frequency is

$$
f = \frac{1}{T}
$$

so that

$$
\omega = 2\pi f
$$

---

### **2. Pendulum Motion**

A simple pendulum consists of a mass suspended from a pivot by a light string. When the pendulum is displaced slightly and released, it oscillates back and forth.

For small angles, the pendulum behaves approximately like an SHM system. Its angular frequency is

$$
\omega_{\text{theory}} = \sqrt{\frac{g}{L}}
$$

where:

- $g = 9.80\ \text{m/s}^2$ is the acceleration due to gravity  
- $L$ is the effective length of the pendulum  

The period is therefore

$$
T = 2\pi\sqrt{\frac{L}{g}}
$$

---

### **3. Effective Length of the Phone Pendulum**

In an ideal pendulum, the entire mass is concentrated at a single point. For a real phone pendulum, the mass is distributed across the phone. Therefore, the effective pendulum length should be measured from the pivot point to the approximate **center of mass** of the phone.

If the string length is $L_s$ and the distance from the top attachment point on the phone to the phone’s center of mass is estimated as $d$, then the effective pendulum length is

$$
L = L_s + d
$$

In many cases, a reasonable estimate for the center of mass is near the geometric center of the phone.

---

### **4. Percent Difference**

To compare theory and experiment, we will calculate the percent difference between the measured and predicted angular frequencies:

$$
\%\text{ difference} = \frac{|\omega_{\text{measured}} - \omega_{\text{theory}}|}{\omega_{\text{theory}}}\times 100\%
$$

## **Experimental Setup**

Attach the phone securely to a string so that it can swing like a pendulum. The setup should allow the phone to oscillate freely in a vertical plane.

Make sure:

- the phone is securely attached  
- the pivot point is stable  
- the oscillation angle remains small

Use the **Pendulum** experiment in Phyphox to record the motion.

## **Procedure**

### **Part A — Measure the Pendulum Geometry**

1. Measure the length of the string from the pivot point to the point where the string attaches to the phone. Record this as $L_s$.

2. Measure the dimensions of the phone and estimate the location of its center of mass.

3. Estimate the distance from the string attachment point on the phone to the center of mass. Record this as $d$.

4. Compute the effective pendulum length:

$$
L = L_s + d
$$

5. Record all measurements with units and reasonable uncertainty estimates.

---

### **Part B — Predict the Angular Frequency**

Using your measured effective length, calculate the theoretical angular frequency:

$$
\omega_{\text{theory}} = \sqrt{\frac{g}{L}}
$$

Then calculate the theoretical period:

$$
T_{\text{theory}} = 2\pi\sqrt{\frac{L}{g}}
$$

Record both values.

---

### **Part C — Measure the Motion with Phyphox**

1. Open the **Phyphox** app and select the pendulum experiment.

2. Set the pendulum swinging with a **small initial angle**. Try to keep the release gentle so that the phone swings in a single plane.

3. Start the measurement and allow the pendulum to oscillate for many cycles.

4. Record the measured period $T_{\text{measured}}$ from Phyphox.

5. Use the measured period to compute the experimental angular frequency:

$$
\omega_{\text{measured}} = \frac{2\pi}{T_{\text{measured}}}
$$

6. If Phyphox reports frequency directly, you may also use

$$
\omega_{\text{measured}} = 2\pi f
$$

7. Repeat the measurement at least **three times** and average your results.

---

### **Part D — Compare Theory and Experiment**

1. Compare your theoretical and measured values of angular frequency.

2. Calculate the percent difference:

$$
\%\text{ difference} = \frac{|\omega_{\text{measured}} - \omega_{\text{theory}}|}{\omega_{\text{theory}}}\times 100\%
$$

3. Decide whether your results agree reasonably well.

## **Sources of Uncertainty**

Some common uncertainties in this lab include:

- uncertainty in locating the phone’s center of mass  
- slight motion outside a single plane  
- large release angle  
- reaction or sensor limitations  
- stretch in the string  
- imperfect pivot point  
- damping from air resistance
