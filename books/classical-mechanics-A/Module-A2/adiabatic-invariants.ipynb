## **3.X.5 Adiabatic Invariants**

Adiabatic invariants arise when a system changes **slowly** compared to the natural timescale of its motion.  
If a parameter of the Hamiltonian varies on a timescale much longer than a period of the motion, the action variables $J_i$ remain **approximately constant**.

This principle appears in plasma physics, accelerator physics, astrophysics, molecular physics, and quantum mechanics.  
It is one of the deepest consequences of action–angle variables.

---

## **3.X.5.1 The Physical Setting**

Consider a Hamiltonian depending on a slowly varying parameter $\lambda(t)$:

$$
H(q,p;\lambda(t)).
$$

If $\dot{\lambda}/\lambda \ll \omega$  
(where $\omega$ is the system’s natural frequency), the motion is “locally periodic,” and we can define instantaneous action variables:

$$
J_i(\lambda(t)) = \frac{1}{2\pi}\oint p_i\,dq_i.
$$

The **adiabatic invariance theorem** states:

$$
\frac{dJ_i}{dt} \approx 0
\qquad \text{when } \lambda(t) \text{ varies slowly}.
$$

The action changes only by exponentially small terms (beyond all orders in $\dot{\lambda}$).

---

## **3.X.5.2 Why Action Variables Are Adiabatic Invariants**

In action–angle coordinates, the Hamiltonian becomes

$$
H = H(J_1,\ldots,J_n;\lambda(t)).
$$

The equations of motion are

$$
\dot{J}_i = -\frac{\partial H}{\partial \theta_i} = 0,
\qquad
\dot{\theta}_i = \omega_i(J,\lambda).
$$

When $\lambda$ varies slowly:

- $\dot{J}_i$ picks up corrections of order $\dot{\lambda}$,
- but on averaging over many cycles, these corrections **average to zero**.

Thus the action changes extremely slowly:

$$
\langle \dot{J}_i \rangle = 0 + \mathcal{O}(\dot{\lambda}^2).
$$

Geometrically:

- each instantaneous orbit is a closed curve in the $(q,p)$ plane,
- the area of the curve changes only by a negligible amount per cycle,
- therefore the area (the action) is preserved.

---

## **3.X.5.3 Adiabatic Invariant of the 1D Harmonic Oscillator**

Consider a harmonic oscillator with slowly varying frequency $\omega(t)$:

$$
H = \frac{p^2}{2m} + \frac12 m\omega^2(t)\, q^2.
$$

The instantaneous action is

$$
J = \frac{E}{\omega(t)}.
$$

Adiabatic invariance implies

$$
\frac{d}{dt}\left(\frac{E}{\omega(t)}\right) \approx 0.
$$

Thus the energy varies proportionally with $\omega$:

$$
E(t) \propto \omega(t).
$$

This important result appears in:

- quantum harmonic oscillator level spacings,  
- molecular vibration spectra,  
- slowly varying magnetic traps in plasma physics.

## **Teal Worked Example: Pendulum with Slowly Varying Length**

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">

### **Example: Simple Pendulum with Slowly Changing Length**

A small-angle pendulum has frequency

$$
\omega(t) = \sqrt{\frac{g}{L(t)}}.
$$

Assume $L(t)$ changes slowly compared to the oscillation period.

### **1. Instantaneous energy**

$$
E = \frac12 m\omega^2(t)\, A^2.
$$

### **2. Action**

For a harmonic oscillator,  
$$
J = \frac{E}{\omega} = \frac12 m \omega(t) A^2.
$$

Adiabatic invariance implies  
$$
\frac12 m\omega(t) A^2 = \text{constant}.
$$

### **3. Amplitude scaling**

Thus  
$$
A^2 \propto \frac{1}{\omega(t)}
          \propto \sqrt{\frac{L(t)}{g}}.
$$

So

$$
A(t) \propto \left(L(t)\right)^{1/4}.
$$

As the length increases, the amplitude **increases** slightly; as the length shortens, the amplitude decreases.

This counterintuitive result matches experiments with slowly lengthened pendulums.

### **Activity: Adiabatic Invariance of Magnetic Moment**

A charged particle gyrating in a magnetic field $B(t)$ has transverse motion equivalent to a harmonic oscillator with frequency

$$
\omega_c(t) = \frac{qB(t)}{m}.
$$

1. Write down the energy of the cyclotron motion.  
2. Compute the instantaneous action $J$.  
3. Show that $J$ is proportional to the magnetic moment  
   $$
   \mu = \frac{p_\perp^2}{2mB}.
   $$
4. Conclude that $\mu$ is an adiabatic invariant when $B(t)$ varies slowly.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

1. Cyclotron energy:

   $$
   E_\perp = \frac{p_\perp^2}{2m}.
   $$

2. Instantaneous action:

   $$
   J = \frac{E_\perp}{\omega_c}
     = \frac{p_\perp^2/2m}{qB/m}
     = \frac{p_\perp^2}{2qB}.
   $$

3. Magnetic moment:

   $$
   \mu = \frac{p_\perp^2}{2mB}.
   $$

   Thus

   $$
   J = \frac{m}{q}\,\mu.
   $$

4. Since $J$ is an adiabatic invariant, so is $\mu$.

This result is fundamental in plasma physics and charged-particle dynamics.

### **Concept Highlight: Why Adiabatic Invariants Matter**

Adiabatic invariants express the idea that when a system’s parameters change slowly compared to its natural oscillation frequency, the area of its phase–space orbit is conserved.  
This principle links action–angle variables to real physical systems, explains energy changes in slowly driven oscillators, and forms the foundation of canonical perturbation theory and KAM theory.

# **Teal Worked Example: The Three Adiabatic Motions of Charged Particles in Earth’s Magnetic Field — With Real Numbers**

Charged particles trapped in Earth’s dipole magnetic field execute **three hierarchical periodic motions**:

1. **Gyromotion** (fastest)  
2. **Bounce motion** (medium)  
3. **Azimuthal drift motion** (slowest)

Each motion corresponds to one of the **three adiabatic invariants**, and their approximate conservation is central to radiation-belt physics and the formation of the **aurora borealis**.

We now examine all three motions **with real numerical estimates** relevant to Earth’s magnetosphere.

---

# **1. Gyromotion (Fastest Motion, ~Milliseconds)**

Consider a **10 keV electron** near Earth’s equator where the magnetic field is about:

$$
B \approx 3\times10^{-5}\ \text{T}.
$$

### **Cyclotron frequency**
For an electron:

$$
\omega_c = \frac{qB}{m_e}
       = \frac{1.6\times10^{-19}(3\times10^{-5})}{9.11\times10^{-31}}
       \approx 5.3\times10^6\ \text{s}^{-1}.
$$

### **Cyclotron period**
$$
T_c = \frac{2\pi}{\omega_c}
    \approx 1.2\ \mu\text{s}.
$$

So electrons spiral around magnetic field lines **about a million times per second**.

### **Gyro-radius**
For a 10 keV electron, the perpendicular speed is roughly:

$$
v_\perp \approx 0.2c.
$$

Thus:

$$
r_g = \frac{v_\perp}{\omega_c}
    \approx \frac{0.2(3\times10^8)}{5.3\times10^6}
    \approx 11\ \text{m}.
$$

Even high-energy electrons have gyroradii of only **meters**, so they are very tightly magnetized.

### **1st Adiabatic Invariant**
$$
\mu = \frac{p_\perp^2}{2mB}
\quad\text{(approximately constant).}
$$

As $B$ increases toward the poles, $\mu$ forces the particle to mirror upward — a key mechanism in auroral precipitation.

---

# **2. Bounce Motion (Intermediate Timescale, ~Seconds)**

Because $B$ strengthens toward Earth’s poles, particles reflect at **mirror points**, oscillating between northern and southern hemispheres.

For the same 10 keV electron with pitch angle $\alpha \approx 60^\circ$, the typical **bounce period** is:

$$
T_b \approx 0.7\ \text{s}
\quad\text{(varies with pitch angle and L-shell).}
$$

So the electron bounces back and forth between hemispheres **about 1–2 times per second**.

### **2nd Adiabatic Invariant**
$$
J_b = \oint p_\parallel\, ds
\quad \text{(approximately conserved).}
$$

This invariance keeps particles confined to the same magnetic field line unless strong disturbances occur (e.g., wave–particle interactions).

---

# **3. Drift Motion (Slowest Timescale, Minutes)**

Curvature and gradient drifts cause trapped electrons and ions to **drift around the Earth**:

- electrons → **eastward**
- ions → **westward**

A 10 keV electron at L ≈ 5 has an approximate drift period:

$$
T_d \approx 6\ \text{minutes}.
$$

Thus an electron may circle Earth **10–20 times per hour**.

### **3rd Adiabatic Invariant**
$$
J_d = \oint \mathbf{p}\cdot d\mathbf{r}
\quad\text{(constant if magnetic field changes slowly).}
$$

This invariant determines the particle’s **drift shell** (L-shell), which is crucial for radiation-belt confinement.

---

# **Connection to the Aurora (Northern Lights)**

Auroral precipitation occurs when one or more adiabatic invariants are **broken**:

### **Examples:**

- **Solar storms** or **magnetic reconnection** abruptly change $B$ →  
  **breaks the 1st invariant $\mu$** → mirror point drops → particles plunge into the atmosphere.

- **Plasma waves (EMIC, whistler-mode)** scatter pitch angles →  
  **break the 2nd invariant $J_b$** → electrons enter the “loss cone.”

- **Strong electric fields / storm-time drifts** change drift shells →  
  **break the 3rd invariant $J_d$** → particles injected toward lower L-shells.

When electrons (1–100 keV) enter the dense atmosphere:

- collisions excite atomic oxygen → green (557.7 nm) and red (630.0 nm) aurora  
- collisions excite nitrogen → blue and purple aurora  

### **Numerical example:**
A typical auroral electron (~5–20 keV):

- enters atmosphere at ~100 km altitude  
- deposits energy over ~1 ms  
- produces ~20,000 photons / electron  

The breathtaking curtains and arcs of the aurora are the macroscopic result of **billions of particles whose adiabatic invariants have been violated**.

# **Summary (Now With Timescales)**

| Motion | Timescale | Radius/Distance | Invariant | Physical Role |
|-------|-----------|-----------------|-----------|----------------|
| **Gyromotion** | ~1 μs | ~10 m | $\mu$ | Magnetization, mirroring |
| **Bounce Motion** | ~0.5–2 s | thousands of km | $J_b$ | Confinement along field line |
| **Drift Motion** | ~minutes | entire Earth | $J_d$ | Drift shells, Van Allen belts |

These three motions and their associated invariants are foundational for understanding:

- radiation belts  
- storm-time injections  
- substorms  
- auroral precipitation  
- space weather  
- spacecraft charging  
- magnetospheric dynamics  

They also beautifully illustrate the physical meaning of **action variables** and **adiabatic invariants** in real-world physics.
