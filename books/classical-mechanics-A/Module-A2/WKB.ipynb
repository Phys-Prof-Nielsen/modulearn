## **3.X.6 Hamilton–Jacobi Theory and Quantum Mechanics (WKB Connection)**

Hamilton–Jacobi theory is more than a classical method:  
it is the **classical backbone of wave mechanics**.  

A remarkable fact is that the Hamilton–Jacobi equation emerges naturally when applying the **WKB approximation** (Wentzel–Kramers–Brillouin) to the Schrödinger equation in the limit $\hbar \to 0$.

This creates a deep conceptual bridge:

- Classical mechanics ↔ geometrical optics  
- Principal function $S$ ↔ wave phase  
- Action integrals $J$ ↔ quantization conditions  
- Canonical transformations ↔ unitary transformations  

Understanding these connections completes the Hamilton–Jacobi module.

---

## **3.X.6.1 Substituting a Phase Ansatz into the Schrödinger Equation**

Consider the 1D time-dependent Schrödinger equation:

$$
i\hbar\,\frac{\partial \psi}{\partial t}
=
-\frac{\hbar^2}{2m}\frac{\partial^2 \psi}{\partial q^2}
+ V(q)\,\psi.
$$

Use the **WKB ansatz**:

$$
\psi(q,t) = A(q,t)\,e^{\frac{i}{\hbar}S(q,t)}.
$$

Here:

- $S$ is a **rapidly varying phase**,
- $A$ is a slowly varying amplitude.

Plugging this into Schrödinger’s equation yields terms of different powers of $\hbar$.

---

## **3.X.6.2 Leading Order in $\hbar$: Recovering Hamilton–Jacobi**

Collect the $\hbar^0$ (leading-order) terms.  
They produce:

$$
\frac{\partial S}{\partial t}
+ \frac{1}{2m}\left(\frac{\partial S}{\partial q}\right)^2
+ V(q) = 0.
$$

This is precisely the **Hamilton–Jacobi equation**:

$$
H\!\left(q,\frac{\partial S}{\partial q}\right)
+ \frac{\partial S}{\partial t}
=0.
$$

Thus the HJ equation describes the leading-order behavior of the quantum phase for a wavepacket.

### Interpretation:
- The phase $S(q,t)$ is the **classical action**.  
- The trajectories are rays in the geometrical-optics sense.  
- Quantum mechanics reduces to classical mechanics when $\hbar$ is small.

---

## **3.X.6.3 Next Order in $\hbar$: Amplitude Transport**

The next-order terms ($\hbar^1$) give a **transport equation** for $A$:

$$
\frac{\partial A}{\partial t}
+
\frac{1}{m}\frac{\partial S}{\partial q}
\frac{\partial A}{\partial q}
+
\frac{1}{2m}\frac{\partial^2 S}{\partial q^2}\,A
=0.
$$

This determines how the amplitude evolves along classical trajectories.

The amplitude encodes the **probability flow**, just as Liouville’s theorem describes classical phase-space flow.

---

## **3.X.6.4 Action Integrals and Semiclassical Quantization**

For bound systems (periodic in $q$), the WKB method yields the quantization condition:

$$
\oint p\, dq = 2\pi\hbar\left(n + \frac{1}{2}\right).
$$

Since

$$
J = \frac{1}{2\pi} \oint p\, dq,
$$

WKB gives approximately:

$$
J \approx \hbar\left(n + \frac{1}{2}\right).
$$

Thus the **action variable is quantized in units of $\hbar$**.

This directly connects classical action–angle variables to energy-level structure in quantum mechanics.

# **Teal Worked Example: WKB Approximation for the Harmonic Oscillator**

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:16px; border-radius:4px;">

### **Example: Energy Levels from WKB**

For the harmonic oscillator:

$$
J = \frac{E}{\omega}.
$$

WKB quantization gives:

$$
J = \hbar\left(n + \frac{1}{2}\right).
$$

Thus:

$$
E = \hbar\omega\left(n + \frac{1}{2}\right),
$$

which is exactly the correct quantum result —  
a rare case where WKB is *exact*.

This example shows explicitly how action variables and Hamilton–Jacobi theory link to quantum energy levels.

# **Green Activity: WKB for a Particle in a 1D Box**

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">

### **Activity: Energy Quantization in the 1D Box**

For a particle of mass $m$ confined to $0<q<L$:

1. Write the classical momentum $p(q)$ in each region.  
2. Compute the action integral  
   $$
   \oint p\, dq.
   $$
3. Apply the WKB quantization rule  
   $$
   \oint p\, dq = 2\pi\hbar\left(n + \frac{1}{2}\right).
   $$
4. Solve for the energy levels $E_n$.  
5. Compare to the exact quantum result.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

1. For fixed energy $E$:

   $$
   p = \sqrt{2mE}.
   $$

2. Over one full cycle (left → right → left):

   $$
   \oint p\, dq = 2L\sqrt{2mE}.
   $$

3. Apply WKB:

   $$
   2L\sqrt{2mE}
   = 2\pi\hbar\left(n+\frac{1}{2}\right).
   $$

4. Solve:

   $$
   E_n = \frac{\pi^2\hbar^2}{2mL^2}
         \left(n+\frac{1}{2}\right)^2.
   $$

5. Compare:

   - Exact quantum:  
     $$
     E_n^\text{exact}
     = \frac{n^2\pi^2\hbar^2}{2mL^2}.
     $$

   - WKB agrees for large $n$ (semiclassical limit),  
     but differs for small $n$ due to boundary conditions (hard walls).

# **Gold Highlight Box: Classical Phase → Quantum Wavefunction**

<div style="background-color:#fff8e1; border-left:5px solid #ffc20e; padding:12px; border-radius:4px;">

### **Concept Highlight: The Principal Function $S$ as Quantum Phase**

In the WKB approximation, the quantum wavefunction takes the form  
$$
\psi = A\, e^{iS/\hbar},
$$  
and the leading term of the Schrödinger equation reduces exactly to the Hamilton–Jacobi equation.  
Thus the principal function $S$ is not just classical—it is the **phase** of the quantum wave in the semiclassical limit.  
This connection explains why action integrals become quantized and why classical trajectories serve as the skeleton of quantum motion.
