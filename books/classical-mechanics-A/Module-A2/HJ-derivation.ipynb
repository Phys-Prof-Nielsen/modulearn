## **3.X.2 Derivation of the Hamilton–Jacobi Equation**

Hamilton–Jacobi theory begins with a simple but powerful demand:

> **Find a canonical transformation that makes the new Hamiltonian $K$ constant.**

If such a transformation exists, the motion in the new variables becomes trivial.  
To construct this transformation, we use a **type-2 generating function**

$$
F_2(q,P,t),
$$

which relates the old and new variables by

$$
p_i = \frac{\partial F_2}{\partial q_i}, 
\qquad
Q_i = \frac{\partial F_2}{\partial P_i},
$$

and modifies the Hamiltonian according to

$$
K(Q,P,t)
=
H(q,p,t) + \frac{\partial F_2}{\partial t}.
$$

Our goal is to choose $F_2$ so that $K$ becomes a constant.  
When this happens, the new equations of motion reduce to

$$
\dot{Q}_i = \frac{\partial K}{\partial P_i} = \text{constant}, 
\qquad
\dot{P}_i = -\frac{\partial K}{\partial Q_i} = 0,
$$

and the dynamics are solved.

---

### **3.X.2.1 Choosing the Generating Function**

We look for a generating function whose time derivative cancels the old Hamiltonian:

$$
K = \text{constant} \equiv E.
$$

Thus,

$$
H\left(q,p,t\right)
+
\frac{\partial F_2}{\partial t}
= E.
$$

Now substitute  
$$
p_i = \frac{\partial F_2}{\partial q_i}.
$$  

This gives

$$
H\!\left(q,\frac{\partial F_2}{\partial q},t\right)
+
\frac{\partial F_2}{\partial t}
= E.
$$

To emphasize the role of this generating function, we rename it:

$$
S(q,t) \equiv F_2(q,P,t)
$$

and call it **Hamilton’s principal function**.

---

### **3.X.2.2 The Hamilton–Jacobi Equation**

Dropping the irrelevant constant $E$ to the left-hand side gives the **Hamilton–Jacobi equation**:

$$
\boxed{
H\!\left(q,\frac{\partial S}{\partial q},t\right)
+
\frac{\partial S}{\partial t}
= 0
}
$$

This *single* scalar partial differential equation contains the full dynamics.

Once $S(q,t)$ is known, the motion follows from

$$
p_i = \frac{\partial S}{\partial q_i}, 
\qquad
P_i = \text{constants}, 
\qquad
Q_i = \frac{\partial S}{\partial P_i}.
$$

Thus Hamilton–Jacobi theory replaces solving $2n$ first-order ODEs with solving one nonlinear PDE.

## **Teal Worked Example: Deriving HJ for a Free Particle**

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">

### **Example: Free Particle in 1D**

The Hamiltonian is

$$
H = \frac{p^2}{2m}.
$$

Insert $p = \partial S/\partial q$ into the Hamilton–Jacobi equation:

$$
\frac{1}{2m}\left(\frac{\partial S}{\partial q}\right)^2
+ \frac{\partial S}{\partial t}
= 0.
$$

This is a first-order PDE.  
Try a separable solution:

$$
S(q,t) = W(q) - Et.
$$

Then

$$
\frac{1}{2m}\left(\frac{dW}{dq}\right)^2
- E = 0.
$$

Solve for $dW/dq$:

$$
\frac{dW}{dq} = \pm \sqrt{2mE}.
$$

Integrate:

$$
W(q) = \pm \sqrt{2mE}\, q + \text{constant}.
$$

So the principal function is

$$
S(q,t) = \pm \sqrt{2mE}\, q - Et.
$$

This solution encodes all possible free-particle trajectories.

## **Green Activity Box: Derive the HJ Equation for a Harmonic Oscillator**

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">

### **Activity: Solve the Hamilton–Jacobi Equation for the 1D Harmonic Oscillator**

The Hamiltonian is

$$
H = \frac{p^2}{2m} + \frac{1}{2}m\omega^2 q^2.
$$

1. Write the Hamilton–Jacobi equation.  
2. Try a separable solution of the form  
   $$
   S(q,t) = W(q) - Et.
   $$
3. Obtain an expression for $W(q)$ up to an integral.  
4. Interpret what the separation constant $E$ must represent.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

1. Start from  
   $$
   H\!\left(q,\frac{\partial S}{\partial q}\right)
   + \frac{\partial S}{\partial t} = 0.
   $$

2. Substitute  
   $$
   S(q,t) = W(q) - Et
   $$
   to get  
   $$
   \frac{1}{2m}\left(\frac{dW}{dq}\right)^2
   + \frac{1}{2}m\omega^2 q^2 
   - E = 0.
   $$

3. Solve for $dW/dq$:

   $$
   \frac{dW}{dq} 
   = \pm \sqrt{2mE - m^2\omega^2 q^2}.
   $$

   Thus

   $$
   W(q)
   = \pm \int \sqrt{2mE - m^2\omega^2 q^2}\, dq.
   $$

4. For the harmonic oscillator, the constant $E$ is just the **energy**.  
   This matches the physical interpretation: constant $P$ after the canonical transformation corresponds to conserved energy in the original variables.

## **Gold Highlight Box: Why the HJ Equation Is a Canonical Transformation in Disguise**

<div style="background-color:#fff8e1; border-left:5px solid #ffc20e; padding:12px; border-radius:4px;">

### **Concept Highlight: The HJ Equation Encodes a Global Canonical Transformation**

The principal function $S(q,t)$ is a generating function whose derivatives define a canonical transformation that renders the new Hamiltonian constant.  
Thus the Hamilton–Jacobi equation is not just another form of mechanics — it is the condition required for a global symmetry-preserving transformation that reduces the dynamics to a trivial flow.  
In this sense, solving the HJ equation is equivalent to “solving” the entire mechanics problem.
