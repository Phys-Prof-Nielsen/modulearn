## **3.X.4 Action–Angle Variables**

Action–angle variables provide one of the most elegant coordinate systems in mechanics.  
They arise naturally when the Hamilton–Jacobi equation admits a separable (complete) integral.  
In action–angle coordinates $(J_i, \theta_i)$:

- the **actions** $J_i$ are constants of motion,  
- the **angles** $\theta_i$ increase uniformly in time,  
- and the Hamiltonian depends only on $J_i$, not on $\theta_i$.

As a result, trajectories become **straight lines** on an $n$–dimensional torus.

---

## **3.X.4.1 From the Principal Function to Actions**

Suppose the HJ equation is separable so that the complete integral can be written as

$$
S(q_1,\ldots,q_n;\ \alpha_1,\ldots,\alpha_n)
= \sum_{i=1}^n S_i(q_i;\ \alpha_1,\ldots,\alpha_n).
$$

For periodic motion, we define the **action** associated with coordinate $q_i$ by the closed-loop integral

$$
J_i = \frac{1}{2\pi} \oint p_i\, dq_i,
$$

where

$$
p_i = \frac{\partial S}{\partial q_i}.
$$

### Properties of action variables

- They depend only on energy and system parameters.  
- They are constants of motion (once constructed).  
- They measure the “area enclosed by a trajectory” in each $(q_i,p_i)$ plane.  
- They are invariant under canonical transformations.  
- They are the key objects in adiabatic physics.

---

## **3.X.4.2 Constructing Angle Variables**

Once the actions are known, the corresponding **angle variables** are defined by

$$
\theta_i = \frac{\partial S}{\partial J_i}.
$$

Because $S$ is a generating function, this transformation is canonical.

In these coordinates:

- the Hamiltonian becomes a function only of the actions:
  $$
  H = H(J_1,\ldots,J_n),
  $$

- and Hamilton’s equations reduce to

  $$
  \dot{J}_i = 0,
  \qquad
  \dot{\theta}_i = \omega_i(J),
  $$

  where the frequencies are

  $$
  \omega_i(J) = \frac{\partial H}{\partial J_i}.
  $$

Thus the dynamics are **quasiperiodic**, with motion on an $n$–torus.

---

## **3.X.4.3 Geometric Interpretation: Motion on the Torus**

In $(q,p)$ space, the motion of an integrable system can be complicated to visualize.  
But in action–angle variables:

- $J_i$ remain constant → they specify a fixed $n$–torus in phase space.  
- $\theta_i$ grow linearly → the motion winds around this torus.

If the frequencies $\omega_i$ are rationally related, the trajectory is periodic.  
If they are irrationally related, the trajectory fills the torus densely.

## **Teal Worked Example: Action–Angle Variables for the 1D Harmonic Oscillator**

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">

### **Example: Harmonic Oscillator**

The Hamiltonian is

$$
H = \frac{p^2}{2m} + \frac12 m\omega^2 q^2.
$$

### **1. Compute the action $J$**

Using the standard ellipse in phase space:

$$
p^2 + (m\omega q)^2 = 2mE,
$$

the area enclosed is

$$
A = \pi \left(\sqrt{2mE}\right) \left(\frac{\sqrt{2mE}}{m\omega}\right)
  = \frac{2\pi E}{\omega}.
$$

Thus

$$
J = \frac{1}{2\pi} A = \frac{E}{\omega}.
$$

So the energy is

$$
E = \omega J.
$$

### **2. Hamiltonian in action variables**

Thus

$$
H(J) = \omega J.
$$

### **3. Angle variable**

The frequency is

$$
\dot{\theta} = \frac{\partial H}{\partial J} = \omega.
$$

Thus

$$
\theta(t) = \omega t + \theta_0.
$$

The solution in original variables becomes

$$
q(t) = \sqrt{\frac{2J}{m\omega}} \sin(\theta),
\qquad
p(t) = \sqrt{2m\omega J}\,\cos(\theta).
$$

This demonstrates the simplicity of action–angle coordinates.

## **Green Activity: Actions for the Particle in a Box**

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">

### **Activity: Compute the Action for a Particle in a 1D Box**

A particle of energy $E$ moves freely between two walls at $q=0$ and $q=L$, reflecting elastically.

1. Write down $p(q)$ for this system.  
2. Compute the action  
   $$
   J = \frac{1}{2\pi} \oint p\,dq.
   $$
3. Explain why the result is proportional to $\sqrt{E}$.  
4. Describe qualitatively what the angle variable $\theta$ represents for this system.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

1. Since the particle moves freely,

   $$
   p = \sqrt{2mE}.
   $$

   This is constant except for sign changes at the walls.

2. Over a full cycle (left wall → right wall → left wall):

   $$
   \oint p\,dq = 2\sqrt{2mE}\, L.
   $$

   Thus

   $$
   J = \frac{1}{2\pi} (2\sqrt{2mE}\,L)
     = \frac{L}{\pi}\sqrt{2mE}.
   $$

3. So indeed

   $$
   J \propto \sqrt{E}.
   $$

4. The angle variable $\theta$ parametrizes the position of the particle along its periodic back-and-forth motion.  
   It increases uniformly with time, mapping the motion on a line to motion on a circle.

## **Gold Highlight Box: Why Action–Angle Variables Matter**

<div style="background-color:#fff8e1; border-left:5px solid #ffc20e; padding:12px; border-radius:4px;">

### **Concept Highlight: The Geometry of Integrability**

Action–angle variables reveal the geometric structure of integrable systems.  
The actions $J_i$ fix an invariant $n$–torus in phase space, and the angles $\theta_i$ describe uniform linear motion on that torus.  
This viewpoint is the foundation of canonical perturbation theory, adiabatic invariants, and modern approaches to chaos and KAM theory.

</div>
