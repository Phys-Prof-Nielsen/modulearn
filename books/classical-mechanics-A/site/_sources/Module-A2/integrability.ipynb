## **3.X.3 Complete Integral and Integrability**

The Hamilton–Jacobi equation is a first-order nonlinear PDE for the principal function $S(q,t)$:

$$
H\!\left(q, \frac{\partial S}{\partial q}, t\right)
+ \frac{\partial S}{\partial t} = 0.
$$

To reconstruct the full motion of an $n$–degree-of-freedom system, we need a solution that contains **$n$ independent constants**, just like the general solution of $n$ first-order ODEs.  
A solution with this structure is called a **complete integral**.

---

## **3.X.3.1 The Complete Integral**

A function  
$$
S(q_1,\ldots,q_n;\ \alpha_1,\ldots,\alpha_n;\ t)
$$
is a **complete integral** of the Hamilton–Jacobi equation if:

1. It satisfies the HJ equation.  
2. It contains **$n$ arbitrary parameters** $\alpha_i$.  
3. The matrix of mixed second derivatives
   $$
   \det\!\left(
   \frac{\partial^2 S}{\partial q_i \,\partial \alpha_j}
   \right)
   \neq 0
   $$
   (invertibility condition).

This ensures that the $\alpha_i$ are truly independent and that the transformation generated by $S$ is valid.

From this $S$, the dynamics follow from

$$
p_i = \frac{\partial S}{\partial q_i},
\qquad
Q_i = \frac{\partial S}{\partial \alpha_i}.
$$

The parameters $\alpha_i$ turn into **constants of motion** (the new momenta), and the $Q_i$ become the new coordinates.

---

## **3.X.3.2 Why the Invertibility Condition Matters**

The canonical transformation constructed from $S$ must be non-singular.  
This requires:

$$
\det\!\left(
\frac{\partial^2 S}{\partial q_i \,\partial \alpha_j}
\right) \neq 0.
$$

Geometrically:

- $S$ defines a family of surfaces in configuration space.
- Varying $\alpha_j$ moves us between these surfaces.
- The determinant condition ensures that these surfaces intersect properly and define a valid coordinate system.

If the determinant vanishes, $S$ does not encode enough independent information to reconstruct the motion.

---

## **3.X.3.3 Integrability**

A system is **integrable** if it admits a complete integral.  
Equivalently:

- it has **$n$ independent, mutually commuting constants of motion**,  
- the Hamilton–Jacobi equation separates in some coordinate system,  
- and trajectories lie on **$n$–dimensional tori** in phase space.

Integrability is the setting where Hamilton–Jacobi theory shines most clearly.

## **Teal Example: Complete Integral for a Free Particle in 2D**

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">

### **Example: 2D Free Particle**

The Hamiltonian is

$$
H = \frac{p_x^2 + p_y^2}{2m}.
$$

Use the ansatz

$$
S(x,y,t) = \alpha_x x + \alpha_y y - \frac{\alpha_x^2 + \alpha_y^2}{2m}\, t.
$$

1. This satisfies the HJ equation:

   $$
   \frac{1}{2m}\left(\alpha_x^2 + \alpha_y^2\right)
   - \frac{\alpha_x^2 + \alpha_y^2}{2m} = 0.
   $$

2. The constants $\alpha_x,\alpha_y$ represent the new momenta.

3. The matrix of mixed derivatives is

   $$
   \frac{\partial^2 S}{\partial q_i \partial \alpha_j}
   =
   \begin{pmatrix}
   \frac{\partial x}{\partial \alpha_x} & 0\\[4pt]
   0 & \frac{\partial y}{\partial \alpha_y}
   \end{pmatrix}
   =
   \begin{pmatrix}
   x & 0\\
   0 & y
   \end{pmatrix}.
   $$

   The determinant is $xy$, which is nonzero except on measure-zero sets.

Thus $S(x,y,t)$ is a complete integral.

</div>

## **Green Activity: Check a Complete Integral for the 1D Harmonic Oscillator**

<div style="background-color:#e8f5e9; border-left:5px solid #006633; padding:12px; border-radius:4px;">

### **Activity: Verify Completeness of the Harmonic Oscillator Solution**

Consider the partial solution to the harmonic oscillator HJ equation:

$$
S(q,t) = \pm \int \sqrt{2mE - m^2\omega^2 q^2}\, dq - Et.
$$

1. Identify the separation constant $\alpha$.  
2. Show explicitly that  
   $$
   p = \frac{\partial S}{\partial q}
   $$  
   reproduces the correct momentum–coordinate relation.  
3. Compute  
   $$
   \frac{\partial^2 S}{\partial q\,\partial E}
   $$  
   and verify that it is nonzero except at turning points.  
4. Explain why this implies $S$ is a complete integral.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

1. The separation constant is  
   $$
   \alpha = E,
   $$  
   the total energy.

2. The derivative gives  

   $$
   p = \pm\sqrt{2mE - m^2\omega^2 q^2},
   $$

   which is exactly the correct harmonic oscillator momentum.

3. Compute the mixed derivative:

   $$
   \frac{\partial^2 S}{\partial q \,\partial E}
   = \frac{m}{\sqrt{2mE - m^2\omega^2 q^2}}.
   $$

   This quantity is nonzero everywhere except at the turning points where $p=0$.

4. Because the mixed derivative is nonzero almost everywhere, the mapping between $(q,E)$ and $(p,Q)$ is invertible.  
   Therefore $S(q,E,t)$ is a complete integral.

## **Gold Highlight Box: Complete Integrals and Integrability**

<div style="background-color:#fff8e1; border-left:5px solid #ffc20e; padding:12px; border-radius:4px;">

### **Concept Highlight: Integrability as the Existence of a Complete Integral**

A system is integrable if and only if its Hamilton–Jacobi equation admits a complete integral.  
In this case, the principal function $S$ generates a canonical transformation that converts the dynamics into straight-line motion with constant momenta.  
This is the deepest and most geometric meaning of “solving” a mechanical system.
