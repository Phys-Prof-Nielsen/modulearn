# **Module 3.X — Hamilton–Jacobi Theory**

## **Learning Outcomes (LOC)**

### **General Motivation**
Hamilton–Jacobi theory provides a powerful alternative formulation of mechanics in which the entire motion of a system is encoded in a single scalar function $S(q,t)$.  
It transforms the problem of solving differential equations into solving a first-order PDE and leads naturally to integrable systems, action–angle variables, and adiabatic invariants.  
This module builds directly on canonical transformations and Poisson brackets, pushing towards the most global and elegant form of Hamiltonian mechanics.

### **Learning Outcomes**

By the end of this module, you will be able to:

1. Explain the purpose of Hamilton–Jacobi theory and how it fits into the broader Hamiltonian framework.
2. Derive the Hamilton–Jacobi equation from a type-2 generating function.
3. Solve simple systems using Hamilton’s principal function $S(q,t)$ or characteristic function $W(q,E)$.
4. Identify the structure of a complete integral and use it to reconstruct the dynamics.
5. Determine whether a system is integrable based on the separability and parameter structure of the HJ equation.
6. Construct action–angle variables for integrable systems and interpret their geometric meaning.
7. Compute action variables using closed-loop integrals $J_i = \frac{1}{2\pi} \oint p_i\,dq_i$.
8. Explain and apply the concept of adiabatic invariants in slowly varying systems.
9. Describe the connection between Hamilton–Jacobi theory and wave mechanics (WKB), and interpret $S$ as a classical phase.
10. Apply the HJ method to representative systems (free particle, harmonic oscillator, central forces) and interpret the results physically.
