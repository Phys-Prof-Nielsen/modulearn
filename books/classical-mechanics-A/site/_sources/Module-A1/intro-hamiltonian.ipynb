## A1.1.1 Motivation

### Introduction to the Hamiltonian  

The Hamiltonian is one of the most profound ideas in physics, reshaping how we understand motion and laying the groundwork for modern theories in mechanics, quantum physics, and beyond. Its origins trace back to the 19th century, when Sir William Rowan Hamilton sought a more elegant and general way to describe mechanical systems. Building on the work of Joseph-Louis Lagrange, Hamilton reformulated classical mechanics in terms of energy rather than force, providing a new perspective that would prove to be incredibly powerful.  

### Comparison with the Lagrangian  

Both the Hamiltonian and the Lagrangian formulations provide alternatives to Newton’s laws, but they differ in focus and application. The Lagrangian approach, introduced by Lagrange in the 18th century, is based on the principle of least action, where the equations of motion are derived by finding paths that minimize (or extremize) a quantity called the action. The Lagrangian is typically expressed as the difference between kinetic and potential energy, and it naturally leads to equations of motion that depend only on generalized coordinates and their time derivatives. This makes it particularly useful when dealing with complex constraints and symmetries, as seen in fields like theoretical mechanics and field theory.  

The Hamiltonian, in contrast, recasts mechanics in terms of energy and phase space, introducing a framework where positions and momenta evolve according to a set of first-order differential equations. Unlike the Lagrangian formulation, which focuses on trajectories in configuration space, the Hamiltonian approach provides a deeper insight into conserved quantities, stability, and dynamical evolution. It also establishes a direct link between classical and quantum mechanics, as the Hamiltonian function in classical physics corresponds to the Hamiltonian operator in quantum mechanics.  

One of the key conceptual shifts in Hamiltonian mechanics is that it treats positions and momenta on equal footing, allowing for a more symmetrical and general description of motion. This makes it particularly valuable in areas like statistical mechanics, where ensembles of systems are studied in phase space, and in celestial mechanics, where long-term behavior of planetary motion can be analyzed efficiently. While the Lagrangian is often more convenient for deriving equations of motion, the Hamiltonian provides a clearer picture of energy conservation, stability, and deeper theoretical structures, such as Poisson brackets and canonical transformations.  

### Importance of the Hamiltonian  

By introducing the Hamiltonian early in a student’s physics education, we gain not only a powerful method for solving classical problems but also a deeper appreciation for the unity of physics. The same function that describes a simple pendulum also governs the behavior of electrons in an atom and even the evolution of the universe itself. This broader perspective is what makes the Hamiltonian such a crucial and enduring concept in physics.

## A1.1.2 Hamiltonian

## Definition of the Hamiltonian and How to Find It  

The Hamiltonian, typically denoted as $\mathcal{H}$, is a function that represents the total energy of a system in terms of its generalized coordinates $q_i$ and conjugate momenta $p_i$. It is defined as  

<div class="alert alert-success">
   
$$ \mathcal{H}(q_i, p_i) = \sum_i p_i \dot{q}_i - \mathcal{L} $$
</div>


where $\mathcal{L}$ is the Lagrangian of the system, given by the difference between kinetic and potential energy:  

$$
\mathcal{L} = K - U.
$$ 

The Hamiltonian is often just the total energy, $ \mathcal{H} = K+ U$, but this is not always the case, especially in more advanced systems.  

### Steps to Find the Hamiltonian  

To construct the Hamiltonian for a given system, follow these steps:  

1. **Write down the Lagrangian**  
   - Identify the generalized coordinates $q_i$ that describe the system.  
   - Express the kinetic energy $K$ and potential energy $U$ in terms of these coordinates and their time derivatives.  
   - Form the Lagrangian as $\mathcal{L} = K - U$.  

2. **Determine the conjugate momenta**  
   - Compute the conjugate momentum for each coordinate:  

     $$
     p_i = \frac{\partial \mathcal{L}}{\partial \dot{q}_i}.
     $$

   - If the system consists of a single particle, the momentum is usually given by $ p = m \dot{q} $, but more complex systems may have different expressions.  

3. **Construct the Hamiltonian**  
   - Express $ \dot{q}_i $ in terms of $ p_i $ if necessary.  
   - Substitute the momenta into the Hamiltonian definition:  

     $$
     \mathcal{H} = \sum_i p_i \dot{q}_i - \mathcal{L}.
     $$ 

   - Rewrite $ \mathcal{H} $ entirely in terms of $ q_i$ and $p_i $, eliminating $ \dot{q}_i $ where possible.  
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;">       
<header>
<h3> Example 1</h3>
    </header>

For a simple 1D harmonic oscillator on a horizontal surface without friction with mass $ m $ and spring constant $ k $, the generalized coordinate is $x$ .

**Solution**
    
1. **Write the Lagrangian:**  

   $$
   \mathcal{L} = \frac{1}{2} m \dot{x}^2 - \frac{1}{2} k x^2.
   $$  

2. **Find the conjugate momentum:**  

   $$
   p = \frac{\partial \mathcal{L}}{\partial \dot{x}} = m \dot{x}.
  $$  

3. **Construct the Hamiltonian:**  

   $$
   \mathcal{H} = p \dot{x} - \mathcal{L} = p \frac{p}{m} - \left(\frac{1}{2} m \dot{x}^2 - \frac{1}{2} k x^2 \right).
   $$ 

   Substituting $ \dot{x} = p/m $:  

   $$
   \mathcal{H} = \frac{p^2}{2m} + \frac{1}{2} k x^2.
   $$  

This is the familiar total energy expression for a harmonic oscillator.  

By following this systematic approach, we can construct the Hamiltonian for a wide range of physical systems, allowing us to apply Hamilton’s equations and gain deeper insights into the dynamics of the system.

## A1.1.2 Hamilton's Equations and How to Apply Them

Once we have the Hamiltonian of a system, we can use **Hamilton’s equations** to determine how the system evolves over time. These equations describe how the generalized coordinates $q_i$ and conjugate momenta $p_i$ change and are given by:  

<div class="alert alert-success">
    
$$\dot{q_i} = \frac{\partial H}{\partial p_i}, \quad \dot{p_i} = -\frac{\partial H}{\partial q_i}.$$
</div>

These two first-order differential equations replace Newton’s second law, providing an equivalent but often more powerful method for solving problems in mechanics. The first equation tells us how the generalized coordinates evolve, while the second describes how the momenta change over time.  

### Steps to Apply Hamilton’s Equations  

1. **Write down the Hamiltonian**  
   - Express $ H(q_i, p_i) $ in terms of generalized coordinates and momenta.  

2. **Compute Hamilton’s equations**  
 
3. **Solve the resulting system of equations**  
   - The equations typically form a set of coupled differential equations that can be solved analytically or numerically.  

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

<div style="background-color: #D0D0D0 ; padding: 10px; border: 1px solid black;">       
<header>
<h3> Example 2</h3>
    </header>

For a simple harmonic oscillator with mass $ m $ and spring constant $ k $, we previously found that the Hamiltonian is  

$$
H = \frac{p^2}{2m} + \frac{1}{2} k x^2.
$$  

#### Compute Hamilton’s Equations  

Using the general form:  

- **Equation for $ \dot{x} $**  

  $$
  \dot{x} = \frac{\partial H}{\partial p} = \frac{p}{m}.
  $$ 

- **Equation for $ \dot{p} $**  

  $$
  \dot{p} = -\frac{\partial H}{\partial x} = -kx.
  $$  

#### Solve the System  

We now have the system:  

$$
\dot{x} = \frac{p}{m}, \quad \dot{x} = -kx.
$$ 

Taking the derivative of the first equation:  

$$
\ddot{x} = \frac{\dot{p}}{m} = -\frac{k}{m} x.
$$ 

This is the familiar equation for simple harmonic motion:  

$$
\ddot{x} + \frac{k}{m} x = 0.
$$ 

The general solution is:  

$$
x(t) = A \cos(\omega t) + B \sin(\omega t),
$$ 

where $ \omega = \sqrt{k/m} $ is the angular frequency of oscillation. 
    
Since $ p = m \dot{x} $, we find  

$$
p(t) = -m A \omega \sin(\omega t) + m B \omega \cos(\omega t).
$$ 

#### Interpretation  

These equations describe simple harmonic motion, just as we would expect from Newtonian mechanics. However, the Hamiltonian approach provides a systematic and generalizable framework that extends to more complex systems, including those in quantum mechanics and statistical physics.  

By using Hamilton’s equations, we can describe motion in terms of energy, revealing deeper insights into conserved quantities, symmetries, and phase space dynamics.
