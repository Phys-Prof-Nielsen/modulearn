# A1.3 From the Lagrangian to the Hamiltonian: The Legendre Transform
<hr style="height:5px;border-width:0;color:gray;background-color:gray">

The Hamiltonian formulation of mechanics emerges naturally when we “switch variables’’ from velocities $\dot{q}_i$ to momenta $p_i$.  
This switch is mathematically implemented using a **Legendre transform**, a tool that converts functions of one variable into functions of its conjugate variable.

In mechanics, the Lagrangian $L(q,\dot{q},t)$ depends on **generalized positions** and **generalized velocities**. But the Hamiltonian $H(q,p,t)$ must instead depend on **positions** and **generalized momenta**. So we must convert from a function of $(q,\dot{q})$ to a function of $(q,p)$.

The Legendre transform is the precise tool that makes this possible.

## Why the Lagrangian → Hamiltonian step must be a Legendre transform

The short answer is:

>Because the Legendre transform is the unique transformation that exchanges a function’s dependence on one set of variables (velocities) with the conjugate set (momenta), while preserving the variational structure of mechanics. *The Legendre transform is by mathematical definition the process of exchanging one variable for its conjugate variable*.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.3.1 Why a transformation is needed

The Lagrangian formalism gives us the equations of motion via

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\right) - \frac{\partial L}{\partial q_i} = 0.
$$

However, these equations are **second-order** and involve accelerations. In contrast, Hamilton’s formulation gives a **first-order system**:

$$
\dot{q}_i = \frac{\partial H}{\partial p_i}, 
\qquad
\dot{p}_i = -\,\frac{\partial H}{\partial q_i}.
$$

To get there, we must express the dynamics in terms of $(q_i,p_i)$ only. But the Lagrangian is written in terms of $(q_i,\dot{q}_i)$. Thus we need a variable change:

$$
(q_i,\dot{q}_i) \quad \longrightarrow \quad (q_i,p_i),
$$

and that is exactly the purpose of the Legendre transform.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.3.2 The mathematical Legendre transform

For a function of one variable, $f(x)$, its Legendre transform is 

$$
g(p) = p x - f(x), \qquad p = f'(x).
$$

This construct does three things:

1. Defines a new variable $p = f'(x)$  
2. Solves for $x$ in terms of $p$  
3. Rewrites the function using the new variable via  
   $g = p x - f$

In classical mechanics, we simply apply this idea **one coordinate at a time**.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.3.3 Applying the Legendre transform to the Lagrangian

### Step 1: Define the generalized momentum
For each coordinate $q_i$:

$$
p_i = \frac{\partial L}{\partial \dot{q}_i}.
$$

---

### Step 2: Solve for $\dot{q}_i$ in terms of $(q,p)$ — and why the Hessian matters

After defining the momenta,

$$
p_i = \frac{\partial L}{\partial \dot{q}_i},
$$

we must **invert these relations** to write the velocities $\dot{q}_i$ in terms of the new variables $(q_i,p_i,t)$.

This step is essential because the Hamiltonian must ultimately be a function

$$
H = H(q,p,t)
$$

with **no remaining $\dot{q}_i$**. Hamilton’s equations **cannot** be derived unless the Hamiltonian has this form.

---

#### Why we must invert the momentum–velocity map

The relations

$$
p_i = p_i(q,\dot{q},t)
$$

define a map  

$$
\dot{q} \;\longrightarrow\; p.
$$

To construct the Hamiltonian, we need the *inverse map*:  

$$
\dot{q}_i = \dot{q}_i(q,p,t).
$$

Without this, the expression 

$$
H = \sum_i p_i\dot{q}_i - L(q,\dot{q},t)
$$

would still contain $\dot{q}_i$, and **would not yet be a Hamiltonian**.

This inversion step is entirely algebraic — **no equations of motion are used**. 

Hamilton’s equation  

$$
\dot{q}_i = \frac{\partial H}{\partial p_i}
$$

only becomes true **after** $H(q,p,t)$ is constructed.

---

#### Why the Hessian must be invertible

The momentum definitions involve derivatives of $L$ with respect to the velocities:

$$
p_i = \frac{\partial L}{\partial \dot{q}_i}.
$$

To solve for the $\dot{q}_i$, the map must be locally invertible. By the **multivariable inverse function theorem**, this requires that the Jacobian determinant  

$$
\det\!\left[\frac{\partial^2 L}{\partial \dot{q}_i \partial \dot{q}_j}\right] \neq 0.
$$

This matrix is often called the **Hessian with respect to velocities** or the **mass matrix**.

---

#### Physical meaning: non-degenerate kinetic energy

For most mechanical systems,

$$
L = T(\dot{q}) - U(q),
$$

and the second derivative matrix of $T$ determines the effective mass matrix.

- If this matrix is **invertible**, then velocities uniquely determine momenta.  
- If this matrix is **singular**, then some momenta do **not** depend on velocities → constraints or gauge freedoms exist.

Examples where the Hessian is singular:
- Lagrangians with gauge symmetry (electromagnetism)  
- Systems with non-holonomic constraints  
- Reparameterization-invariant systems  
- General relativity in Arnowitt–Deser–Misner (ADM) form  

In such cases, the Legendre transform is **not** valid globally, and special methods (Dirac–Bergmann constrained Hamiltonian mechanics) are required.

But for ordinary mechanical systems in classical mechanics:
- masses are non-zero  
- kinetic energy is quadratic and positive definite  
- the Hessian is invertible  

Thus, the usual Legendre transform works smoothly.

---

#### Summary of why Step 2 is required

- The Hamiltonian must be a function of $(q,p,t)$, not $(q,\dot{q},t)$.  
- To eliminate $\dot{q}$, we must invert $p_i(q,\dot{q},t)$ for $\dot{q}_i(q,p,t)$.  
- This inversion requires the Hessian of $L$ with respect to velocities to be non-singular.  
- Only after this inversion can we build the Hamiltonian and derive Hamilton’s equations.

This is why Step 2 is not optional — it is the lynchpin that makes the Legendre transform work.

---

### Step 3: Construct the Hamiltonian

$$
\boxed{
H(q,p,t) = \sum_i p_i \dot{q}_i - L(q,\dot{q},t)
}
$$

The key point: 

>**$\dot{q}_i$ must be replaced by their expressions in terms of $(q,p)$** obtained in Step 2.

When this is done, $H$ becomes a function of $(q,p,t)$ only.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.3.4 Deriving Hamilton’s equations from the Legendre transform

Start from 

$$
H = \sum_i p_i \dot{q}_i - L.
$$

Take differentials:

$$
dH = \sum_i \dot{q}_i\,dp_i + \sum_i p_i\,d\dot{q}_i
 -\left(
\sum_i \frac{\partial L}{\partial q_i} dq_i 
+ \sum_i \frac{\partial L}{\partial \dot{q}_i} d\dot{q}_i
+ \frac{\partial L}{\partial t} dt
\right).
$$

Use $p_i = \partial L/\partial \dot{q}_i$ to cancel the $d\dot{q}_i$ terms:

$$
dH 
= \sum_i \dot{q}_i\,dp_i 
- \sum_i \frac{\partial L}{\partial q_i} dq_i 
- \frac{\partial L}{\partial t} dt.
$$

But Euler–Lagrange gives

$$
\frac{\partial L}{\partial q_i} = -\,\dot{p}_i.
$$

Thus:

$$
dH = \sum_i \dot{q}_i\,dp_i + \sum_i \dot{p}_i\,dq_i - \frac{\partial L}{\partial t}dt.
$$

Identify:

$$
\frac{\partial H}{\partial p_i} = \dot{q}_i,
\qquad
\frac{\partial H}{\partial q_i} = -\,\dot{p}_i,
\qquad
\frac{\partial H}{\partial t} = -\,\frac{\partial L}{\partial t}.
$$

And we arrive at **Hamilton’s equations**:

$$
\dot{q}_i = \frac{\partial H}{\partial p_i}, 
\qquad
\dot{p}_i = -\,\frac{\partial H}{\partial q_i}.
$$

**Step (a): Compute the differential**

Starting from 

$$
H = \sum_i p_i\dot{q}_i - L(q,\dot{q},t),
$$

differentiate:

$$
dH = \sum_i (\dot{q}_i\,dp_i + p_i\,d\dot{q}_i)
 - \left(
\sum_i \frac{\partial L}{\partial q_i} dq_i
+ \sum_i \frac{\partial L}{\partial \dot{q}_i} d\dot{q}_i
+ \frac{\partial L}{\partial t} dt
\right).
$$

---

**Step (b): Cancel $d\dot{q}_i$ terms**

Because  

$$
p_i = \frac{\partial L}{\partial \dot{q}_i},
$$

the terms  
- $p_i d\dot{q}_i$  
- $\left(\frac{\partial L}{\partial \dot{q}_i}\right)d\dot{q}_i$  

cancel exactly.

Result:

$$
dH = \sum_i \dot{q}_i\,dp_i 
     - \sum_i \frac{\partial L}{\partial q_i} dq_i 
     - \frac{\partial L}{\partial t} dt.
$$

---

**Step (c): Use Euler–Lagrange**

Euler–Lagrange gives

$$
\frac{\partial L}{\partial q_i} = -\dot{p}_i.
$$

Substitute:

$$
dH = \sum_i \dot{q}_i\,dp_i + \sum_i \dot{p}_i\,dq_i - \frac{\partial L}{\partial t} dt.
$$

---

**Step (d): Compare to the total differential of $H$**

For any function $H(q,p,t)$:

$$
dH = 
\sum_i \frac{\partial H}{\partial p_i} dp_i +
\sum_i \frac{\partial H}{\partial q_i} dq_i +
\frac{\partial H}{\partial t} dt.
$$

Match terms:

- Coefficient of $dp_i$: $\dot{q}_i = \frac{\partial H}{\partial p_i}$  
- Coefficient of $dq_i$: $\dot{p}_i = -\frac{\partial H}{\partial q_i}$  
- Coefficient of $dt$: $\frac{\partial H}{\partial t} = -\frac{\partial L}{\partial t}$

---

**Step (e): Hamilton’s Equations**

Thus,

$$
\boxed{\dot{q}_i = \frac{\partial H}{\partial p_i}}
\qquad\text{and}\qquad
\boxed{\dot{p}_i = -\frac{\partial H}{\partial q_i}}.
$$

These are **Hamilton’s equations**, obtained directly from the Legendre transform.

**Example: Legendre Transform for a 1D Free Particle**

Consider 

$$
L = \tfrac12 m \dot{x}^2.
$$

**1. Compute the momentum**

$$
p = \frac{\partial L}{\partial \dot{x}} = m\dot{x}.
$$

**2. Solve for the velocity**

$$
\dot{x} = \frac{p}{m}.
$$

**3. Build the Hamiltonian**

$$
H = p\dot{x} - L 
= p\left(\frac{p}{m}\right) - \frac12 m \left(\frac{p}{m}\right)^2
= \frac{p^2}{2m}.
$$

So the Hamiltonian is the familiar kinetic energy:

$$
H = \frac{p^2}{2m}.
$$

**(a) Generalized momentum**

Start from

$$
L = \frac12 m \dot{x}^2 - \frac12 k x^2.
$$

The momentum conjugate to $x$ is

$$
p = \frac{\partial L}{\partial \dot{x}} = m\dot{x}.
$$

---

**(b) Solve for $\dot{x}$ in terms of $p$**

From $p = m\dot{x}$:

$$
\dot{x} = \frac{p}{m}.
$$

---

**(c) Construct the Hamiltonian**

By definition:

$$
H = p\dot{x} - L.
$$

Substitute $\dot{x} = p/m$ and the expression for $L$:

$$
H = p\left(\frac{p}{m}\right)
   - \left[\frac12 m\left(\frac{p}{m}\right)^2 - \frac12 k x^2\right].
$$

Simplify term by term:

$$
p\left(\frac{p}{m}\right) = \frac{p^2}{m},
\qquad
\frac12 m\left(\frac{p}{m}\right)^2
= \frac12 m \frac{p^2}{m^2}
= \frac{p^2}{2m}.
$$

So

$$
H = \frac{p^2}{m} - \left( \frac{p^2}{2m} - \frac12 k x^2 \right)
  = \frac{p^2}{m} - \frac{p^2}{2m} + \frac12 k x^2
  = \frac{p^2}{2m} + \frac12 k x^2.
$$

Thus

$$
\boxed{
H(x,p) = \frac{p^2}{2m} + \frac12 k x^2
}
$$

which is the familiar sum of kinetic and potential energy.

---

**(d) Check Hamilton’s equations**

Hamilton’s equations are:

$$
\dot{x} = \frac{\partial H}{\partial p},
\qquad
\dot{p} = -\frac{\partial H}{\partial x}.
$$

Compute the derivatives:

1.  

$$
\frac{\partial H}{\partial p}
= \frac{\partial}{\partial p}\left( \frac{p^2}{2m} + \frac12 k x^2 \right)
= \frac{p}{m}.
$$

So

$$
\dot{x} = \frac{p}{m},
$$

which is consistent with $p = m\dot{x}$.

2.  

$$
\frac{\partial H}{\partial x}
= \frac{\partial}{\partial x}\left( \frac{p^2}{2m} + \frac12 k x^2 \right)
= kx.
$$

So

$$
\dot{p} = -\frac{\partial H}{\partial x} = -kx.
$$

Now differentiate $\dot{x}$ with respect to time:

$$
\ddot{x} = \frac{d}{dt}\left(\frac{p}{m}\right)
= \frac{1}{m}\dot{p}
= \frac{1}{m}(-kx)
= -\frac{k}{m}x.
$$

Define $\omega^2 = k/m$. Then

$$
\ddot{x} + \omega^2 x = 0,
\qquad
\omega = \sqrt{\frac{k}{m}}.
$$

This matches the standard harmonic oscillator equation, so the Hamiltonian formulation reproduces the expected dynamics.

## A1.3.5 Interpretation: Why the Legendre transform is “switching variables”

The Lagrangian uses velocities as its fundamental dynamical variables; the Hamiltonian uses momenta instead.  
The Legendre transform is the mathematical engine that replaces:

- slope with coordinate,  
- conjugate momentum with velocity,  
- tangent-space representation $(q,\dot{q})$ with cotangent-space representation $(q,p)$.

This shift turns second-order ODEs into first-order systems and reveals the geometric structure of dynamics in phase space.
