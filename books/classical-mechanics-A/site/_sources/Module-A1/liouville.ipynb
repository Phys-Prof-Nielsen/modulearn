# A1.5 Liouville’s Theorem: Volume Preservation in Phase Space
<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.5.1 Motivation

So far we have seen that Hamiltonian systems:

- move along contours of constant energy  
- have fixed points that are either **centers** or **saddles**  
- define a smooth vector field in phase space  

But we have not yet uncovered the most important structural property:

> **Hamiltonian flow preserves phase-space volume.**

This is known as **Liouville’s theorem**, and it has profound consequences:

- Hamiltonian systems can **never** have attractors or spirals.  
- Volume elements in phase space behave like **fluid particles in an incompressible flow**.  
- Probability densities evolve by **shearing**, not compression.  
- It is a key ingredient in **statistical mechanics** and **chaos theory**.  

Before proving it, we first observe the structure of Hamiltonian flow.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.5.2 The Hamiltonian Vector Field

A Hamiltonian system defines the vector field

$$
\mathbf{v}(q,p)
=
\left(
\frac{\partial H}{\partial p_1},
\ldots,
\frac{\partial H}{\partial p_n},
-\frac{\partial H}{\partial q_1},
\ldots,
-\frac{\partial H}{\partial q_n}
\right).
$$

A point in phase space flows along this vector field like a particle in a fluid.

To understand volume preservation, we compute the **divergence**:

$$
\nabla \cdot \mathbf{v}
=
\sum_{i=1}^n
\left(
\frac{\partial}{\partial q_i}\dot{q}_i
+
\frac{\partial}{\partial p_i}\dot{p}_i
\right).
$$

with the result that

$$
\boxed{
\nabla \cdot \mathbf{v} = 0
}
$$

→ **phase-space flow is incompressible**.

The divergence of this vector field is

$$
\nabla\!\cdot\!\mathbf{v}

**Your task:** Work through the steps to show that this divergence is **zero**, which proves that Hamiltonian flow preserves phase-space volume.

**Fill in the missing steps**

**(a)** Substitute the Hamiltonian expressions  
$\dot{q}_i = \frac{\partial H}{\partial p_i}$  
$\dot{p}_i = -\frac{\partial H}{\partial q_i}$

**(b)** Compute each partial derivative explicitly:

- $\displaystyle \frac{\partial}{\partial q_i}\left(\frac{\partial H}{\partial p_i}\right)$  
- $\displaystyle \frac{\partial}{\partial p_i}\left(-\frac{\partial H}{\partial q_i}\right)$

**(c)** Show that each pair of terms cancels because **mixed partial derivatives commute**:

$$
\frac{\partial^2 H}{\partial q_i \partial p_i}
=
\frac{\partial^2 H}{\partial p_i \partial q_i}.
$$

**(d)** Conclude that $\nabla\!\cdot\!\mathbf{v} = 0$.

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

**(a) Substitute Hamilton’s equations**

Start with

$$
\nabla\!\cdot\!\mathbf{v}
=
\sum_{i=1}^n
\left(
\frac{\partial}{\partial q_i}\dot{q}_i
+
\frac{\partial}{\partial p_i}\dot{p}_i
\right).
$$

Insert  
$\dot{q}_i = \frac{\partial H}{\partial p_i}$  
$\dot{p}_i = -\frac{\partial H}{\partial q_i}$:

$$
\nabla\!\cdot\!\mathbf{v}
=
\sum_{i=1}^n
\left(
\frac{\partial}{\partial q_i}
\frac{\partial H}{\partial p_i}
-
\frac{\partial}{\partial p_i}
\frac{\partial H}{\partial q_i}
\right).
$$

---

**(b) Compute each term**

These are mixed second partial derivatives:

$$
\frac{\partial}{\partial q_i}
\left(\frac{\partial H}{\partial p_i}\right)
=
\frac{\partial^2 H}{\partial q_i \partial p_i},
$$

$$
\frac{\partial}{\partial p_i}
\left(-\frac{\partial H}{\partial q_i}\right)
=
-
\frac{\partial^2 H}{\partial p_i \partial q_i}.
$$

---

**(c) Use equality of mixed partials**

For sufficiently smooth $H$ (all classical systems):

$$
\frac{\partial^2 H}{\partial q_i \partial p_i}
=
\frac{\partial^2 H}{\partial p_i \partial q_i}.
$$

Thus the two terms cancel *for each $i$*:

$$
\frac{\partial^2 H}{\partial q_i \partial p_i}
-
\frac{\partial^2 H}{\partial p_i \partial q_i}
= 0.
$$

---

**(d) Final result**

Adding over all $n$ degrees of freedom:

$$
\nabla\!\cdot\!\mathbf{v}
= \sum_{i=1}^n 0 = 0.
$$

Therefore:

$$
\boxed{
\nabla\!\cdot\!\mathbf{v} = 0
\quad\Longrightarrow\quad
\text{Hamiltonian flow preserves phase-space volume}.
}
$$

This is the core of **Liouville’s theorem**.

**Concept Highlight: Why Does Zero Divergence Mean Incompressible Flow?**

The result  

$$
\nabla\!\cdot\!\mathbf{v} = 0
$$  

means that the Hamiltonian velocity field has **no net expansion or contraction** at any point in phase space.

Here’s why that matters:

---

**1. Local Interpretation: No Squeezing or Stretching of Volume**

In fluid mechanics, $\nabla\!\cdot\!\mathbf{v}$ measures how much a tiny fluid element expands or contracts.

- If $\nabla\!\cdot\!\mathbf{v} > 0$ → local expansion  
- If $\nabla\!\cdot\!\mathbf{v} < 0$ → local compression  
- If $\nabla\!\cdot\!\mathbf{v} = 0$ → **incompressible flow**  

Hamiltonian phase-space flow behaves exactly like an incompressible fluid.

So a small phase-space “blob” cannot:

- shrink,  
- collapse,  
- or spiral inward.

It can only **stretch** and **shear**.

---

**2. Global Interpretation: Phase-Space Volume Is Preserved**

Consider any region $\Omega(t)$ carried along by the flow.  
The divergence theorem tells us:

$$
\frac{d}{dt} \text{Vol}[\Omega(t)]
=
\int_{\Omega(t)} \nabla\!\cdot\!\mathbf{v} \; d\Gamma.
$$

If $\nabla\!\cdot\!\mathbf{v}=0$ everywhere, then:

$$
\frac{d}{dt} \text{Vol}[\Omega(t)] = 0.
$$

This is exactly **Liouville’s theorem**.

---

**3. Consequence: Hamiltonian Systems Cannot Have Attractors**

Attractors require **volume contraction**, because trajectories must be “pulled together” over time.

But in Hamiltonian mechanics:

- no point can suck in nearby trajectories  
- no orbit can attract a neighborhood  
- no stable spirals can exist  

This rules out:

- sinks  
- stable spirals  
- limit cycles  
- damped oscillations  
- long-term collapse toward an equilibrium

The only possible fixed points are:

- **stable centers** (neutral, no volume change)  
- **saddles** (volume preserved but stretched + compressed equally)

**Hamiltonian Systems Can Have Unstable Fixed Points**

Liouville’s theorem forbids **attractors**, not **instability**.

- **Centers** (pure imaginary eigenvalues) are neutrally stable.  
- **Saddles** (real opposite-sign eigenvalues) are unstable.  

What Hamiltonian systems can *never* have are:

- sinks  
- stable spirals  
- limit cycles  
- any fixed point that attracts a region of initial conditions  

In other words:

> Hamiltonian fixed points may be **stable or unstable**,  
> but they are never **attracting**.

The instability of saddle points is fully compatible with volume preservation.

---

**4. Geometric Summary**

Hamiltonian evolution acts like:

> a perfectly incompressible, infinitely thin fluid moving in phase space,  
> twisting, folding, and stretching — but never changing volume.

This underlies:

- the structure of phase portraits  
- the onset of chaos via stretching & folding  
- the foundation of statistical mechanics  
- the reversibility of Hamiltonian dynamics

## A1.5.3 Statement of Liouville’s Theorem

> **The phase-space volume of any region is preserved under Hamiltonian time evolution.**

If $\Omega(t)$ is a region transported along the Hamiltonian flow, then

$$
\frac{d}{dt} \text{Vol}[\Omega(t)] = 0.
$$

In other words, blobs of initial conditions:

- stretch,  
- twist,  
- shear,  
- fold,  

but **never shrink or grow**.

---

<div style="background-color:#e3f2fd; border-left:5px solid #1565c0; padding:12px; border-radius:4px;">

**Concept Highlight: Why This Matters**

1. **No attractors**  
   Volume cannot contract → no trajectory can pull neighbors into a point or limit cycle.

2. **Centers, not spirals**  
   Linearized Hamiltonian systems only give centers or saddles:  
   spirals require contraction/expansion, which is forbidden.

3. **Reversibility of physics**  
   Volume preservation reflects the time-reversible nature of Hamilton’s equations.

4. **Chaos (later)**  
   Even chaotic Hamiltonian systems preserve volume globally.  
   Chaos arises from **stretching and folding**, not compression.

5. **Statistical mechanics**  
   Evolution of probability densities uses Liouville’s equation, a direct consequence of this theorem.

</div>

## A1.5.4 Geometric Picture: Phase Space as an Incompressible Fluid

Imagine filling phase space with an ideal fluid.  
Each “fluid element” represents an infinitesimal bundle of initial conditions.

Under Hamiltonian time evolution:

- fluid elements move along the flow lines,  
- shapes distort dramatically,  
- but **their volume never changes**.

Even though a distribution can become thinner than paper in one direction, it must grow correspondingly in others.

This stretching and folding is the root of **chaotic mixing**.

## A1.5.5 Liouville’s Equation as a Continuity Equation for Probability

Up to this point, Liouville’s theorem has been discussed in terms of **volume preservation**.  
But there is a deeper and more powerful interpretation:

> **Liouville’s theorem is the continuity equation for probability in phase space.**

In other words, it expresses **conservation of probability density** as initial conditions flow under Hamilton’s equations.

Let $\rho(q,p,t)$ be a probability density on phase space.  
This density satisfies the normalization:

$$
\int \rho(q,p,t)\, d\Gamma = 1,
$$

where $d\Gamma = dq_1 \cdots dq_n\, dp_1 \cdots dp_n$ is the volume element.

---

### Probability Conservation: The Continuity Equation

As $\rho$ moves along the Hamiltonian flow, probability cannot be created or destroyed.  
Thus $\rho$ must satisfy the **continuity equation**:

$$
\frac{\partial \rho}{\partial t}
+
\nabla\!\cdot(\rho\,\mathbf{v}) = 0,
$$

where  

$$
\mathbf{v}(q,p) = (\dot{q}_1,\ldots,\dot{q}_n,\;\dot{p}_1,\ldots,\dot{p}_n)
$$  

is the Hamiltonian vector field.

This equation says:

- the time-change of density inside a region  
- equals the flux of probability across its boundary  

Exactly the same structure holds for mass in a fluid or charge in electrodynamics.

---

### Using Incompressibility

We can expand:

$$
\nabla\!\cdot(\rho\,\mathbf{v})
=
(\nabla\rho)\cdot\mathbf{v}
+
\rho\,(\nabla\!\cdot\!\mathbf{v}).
$$

But Liouville’s theorem already told us:

$$
\nabla\!\cdot\mathbf{v} = 0.
$$

Thus the continuity equation simplifies to:

$$
\frac{\partial \rho}{\partial t}
+
\mathbf{v}\cdot\nabla\rho = 0.
$$

This has an important interpretation:

> **The probability density is constant along each trajectory.**

This is sometimes called **“convective invariance.”**

---

### Phase-Space Advection Form

Since along the flow,

$$
\frac{d}{dt} = \frac{\partial}{\partial t} + \mathbf{v}\cdot\nabla,
$$

we obtain:

$$
\boxed{
\frac{d\rho}{dt} = 0
}
$$

→ The density of initial conditions is simply **advected** by the Hamiltonian flow.

This is exactly what happens in an incompressible fluid:  
the density is carried along without compression or expansion.

---

### Poisson Bracket Form (standard in mechanics)

Using Hamilton’s equations one can show:

$$
\mathbf{v}\cdot\nabla\rho = \{\rho, H\},
$$

so the continuity equation becomes:

$$
\boxed{
\frac{\partial \rho}{\partial t} + \{\rho, H\} = 0
}
$$

This is the **Liouville equation** — the Hamiltonian version of the continuity equation.

---

### Interpretation

- $\rho$ behaves like the density of tiny tracers in an incompressible fluid.  
- They move and stretch and fold with the flow.  
- The number of tracers in any region is preserved.  
- No density pile-up is allowed (no attractors).  
- No density depletion is allowed (no repellers).  

This prepares us conceptually for:

- Hamilton–Jacobi theory  
- Poisson brackets  
- canonical transformations  
- phase mixing  
- and the onset of **Hamiltonian chaos**, where stretching & folding of probability densities becomes dramatic.

---

<div style="background-color:#e3f2fd; border-left:5px solid #1565c0; padding:12px; border-radius:4px;">

**Concept Highlight: How Liouville’s Equation Connects to Quantum Probability**

In classical mechanics, Liouville’s theorem tells us that a probability density  
$\rho(q,p,t)$ in phase space obeys the continuity equation

$$
\frac{\partial \rho}{\partial t} + \{\rho, H\} = 0,
$$

expressing that probability is **conserved** and simply **advected** by the Hamiltonian flow.

Quantum mechanics preserves probability too — but with very different mathematical objects.

---

**1. Classical: Probability Density in Phase Space**

- $\rho(q,p,t)$ is a *real-valued* probability density.  
- It evolves deterministically via Liouville's equation.  
- Probability flows like an **incompressible fluid** in phase space.  
- Volume preservation is guaranteed by Hamiltonian structure.

---

**2. Quantum: Probability Amplitude in Hilbert Space**

In quantum mechanics the wavefunction

$$
\psi(x,t)
$$

encodes probability through

$$
|\psi(x,t)|^2.
$$

The Schrödinger equation,

$$
i\hbar\frac{\partial \psi}{\partial t} = \hat{H}\psi,
$$

leads to the quantum continuity equation:

$$
\frac{\partial |\psi|^2}{\partial t}
+ \nabla\cdot \mathbf{j} = 0,
$$

where $\mathbf{j}$ is the **probability current**.

Thus probability is conserved — just like in the classical Liouville equation.

---

**3. Structural Bridge: Poisson Brackets → Commutators**

Classical Liouville equation:

$$
\frac{\partial \rho}{\partial t} = -\{\rho, H\}.
$$

Quantum von Neumann equation for the density matrix:

$$
\frac{\partial \hat{\rho}}{\partial t}
= -\frac{i}{\hbar} [\hat{H}, \hat{\rho}].
$$

These match under the correspondence:

$$
\{\, ,\,\} \quad \longrightarrow \quad \frac{1}{i\hbar}[\, ,\,].
$$

So:

> **Liouville’s equation is the classical limit of quantum probability evolution.**

---

**4. Wigner Function: Quantum Probability in Phase Space**

The **Wigner function** $W(q,p)$ is a quasi-probability distribution in phase space.

In the limit $\hbar \to 0$:

$$
W(q,p,t) \;\longrightarrow\; \rho(q,p,t),
$$

and its evolution reduces to Liouville’s equation.

This forms a direct bridge between:

- classical probability densities, and  
- quantum probability amplitudes.

---

**5. Summary**

- Both classical and quantum mechanics use **continuity equations** to express probability conservation.  
- Liouville’s equation ↔ Schrödinger/von Neumann equations share the same structural form.  
- Poisson brackets become commutators.  
- Wigner functions provide a phase-space formulation of quantum mechanics.  

Together these show that *probability conservation is a unifying structural principle across classical and quantum dynamics*.

## A1.5.6 Example: Harmonic Oscillator

For the harmonic oscillator:

$$
H = \frac{p^2}{2m} + \frac12 k x^2,
$$

a small circular blob of initial conditions:

- becomes an ellipse  
- rotates rigidly in phase space  
- never changes area  

This is the simplest visualization of Liouville’s theorem.
