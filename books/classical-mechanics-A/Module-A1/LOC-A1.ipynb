## Overall Motivation

Up to now, we have mastered the Lagrangian picture: write $L = T - U$, choose generalized coordinates, derive Euler–Lagrange equations, and solve the resulting dynamics.  
But the Lagrangian is only one way to understand classical mechanics.

A deeper structure emerges when we rewrite mechanics using:
- **generalized coordinates and momenta on equal footing**,  
- **phase space** rather than configuration space,  
- **a single scalar generator of motion — the Hamiltonian**,  
- **a symplectic geometric structure** that remains invariant under time evolution.

This module develops that symplectic viewpoint, establishes the meaning of phase-space flow, and builds the algebraic and geometric tools (canonical transformations and Poisson brackets) that connect classical mechanics to modern physics, including statistical mechanics, integrable systems, and the mathematical structure of quantum theory.

As a bridge between Hamiltonian flow and more advanced topics, we introduce **Liouville’s theorem**, which reveals that Hamiltonian evolution preserves phase-space volume.  
This idea becomes essential later when we examine integrability, chaos, and the Hamilton–Jacobi formalism.

## Learning Objectives

### After completing this module, you should be able to:

### 1. Derive the Hamiltonian from the Lagrangian
- Perform the Legendre transform to obtain  

  $$
  H(q,p,t) = \sum_i p_i \dot{q}_i - L(q,\dot{q},t)
  $$
  
  and derive Hamilton’s equations.  
- Explain why this construction requires that $\partial^2 L / \partial \dot{q}_i \partial \dot{q}_j$ be non-singular.

### 2. Describe phase space and interpret motion as a geometric flow
- Distinguish configuration space $Q$ from phase space $\Gamma$.  
- Represent a state as a point $(q_i,p_i)$ in a $2n$-dimensional space.  
- Interpret Hamilton’s equations as generating a **vector field** in phase space.  
- Construct and interpret phase portraits: fixed points, stability, separatrices, and closed orbits.

### 3. Explain and apply Liouville’s Theorem
- Demonstrate that the Hamiltonian flow has **zero divergence** in phase space.  
- Explain physically why **phase-space volume is conserved**.  
- Use the Liouville equation  

  $$
  \frac{\partial \rho}{\partial t} + \{\rho , H\} = 0
  $$
   
  to describe how probability densities evolve.  
- Connect volume preservation to the absence of attractors in Hamiltonian systems and its role as a precursor to chaos theory.

### 4. Use canonical transformations to simplify Hamiltonian systems
- Explain what makes a transformation “canonical.”  
- Apply generating functions to construct new variables $(Q,P)$.  
- Identify symplectic invariants preserved under canonical transformations.  
- Relate canonical transformations to symmetry, simplification of equations of motion, and the classical analogy of unitary transformations.

### 5. Compute and interpret Poisson brackets
- Compute brackets such as  

  $$
  \{q_i,p_j\} = \delta_{ij}, \qquad  
  \{q_i,q_j\} = \{p_i,p_j\} = 0.
  $$
   
- Express Hamilton’s equations in bracket form:  

  $$
  \dot{f} = \{f,H\} + \frac{\partial f}{\partial t}.
  $$

   
- Understand the algebra of observables and the classical limit of quantum commutators.

### 6. Appreciate the structure that leads into chaos and Hamilton–Jacobi theory
- Recognize how phase-space geometry and Liouville’s theorem constrain Hamiltonian chaos.  
- Understand why canonical transformations and Poisson brackets set the stage for Hamilton–Jacobi theory (covered in the next module).
