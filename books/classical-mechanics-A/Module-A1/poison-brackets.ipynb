# A1.7 Poisson Brackets
<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.7.1 Motivation
After studying canonical transformations and their generating functions, we now introduce the central algebraic structure of Hamiltonian mechanics: the **Poisson bracket**. It plays two crucial roles:

1. It provides the **compact expression for the time evolution** of any dynamical quantity.  
2. It encodes the **infinitesimal canonical transformations** that generate symmetries and conserved quantities.

In many ways, Poisson brackets are to classical mechanics what commutators are to quantum mechanics.

### Why We Need Poisson Brackets

Up to now, we have seen that:

- canonical transformations preserve the form of Hamilton’s equations,  
- generating functions $F$ produce these transformations,  
- Noether’s theorem links invariance under such transformations to conservation laws.

But we still lack a **single algebraic tool** that:

- expresses how any function of phase-space variables evolves in time,  
- tells us when two quantities are “canonically conjugate,”  
- generates infinitesimal canonical transformations,  
- encodes symmetries and conserved quantities,  
- mirrors the role of commutators in quantum mechanics.

The Poisson bracket accomplishes all of these in one structure.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.7.2 Definition of the Poisson Bracket

For any two sufficiently smooth functions $f(q,p,t)$ and $g(q,p,t)$ on phase space, the **Poisson bracket** is defined by:

$$
\{ f , g \}
=
\sum_{i=1}^n
\left(
\frac{\partial f}{\partial q_i} \frac{\partial g}{\partial p_i}
-
\frac{\partial f}{\partial p_i} \frac{\partial g}{\partial q_i}
\right).
$$

In vector-matrix form:

$$
\boxed{
\{f,g\} = (\nabla f)^{\mathrm{T}} J (\nabla g)
}
$$

where $J$ is the symplectic matrix introduced earlier.

This form makes the geometry of Poisson brackets explicit and shows that they arise naturally from the symplectic structure.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.7.3 Time Evolution via Poisson Brackets

Hamilton’s equations imply that the time evolution of any function $f(q,p,t)$ is given by:

$$
\frac{df}{dt}
= 
\{ f, H \}
+ \frac{\partial f}{\partial t}.
$$

This single formula contains all of classical mechanics:

- If $f = q_i$, it gives $\dot{q}_i = \partial H/\partial p_i$.  
- If $f = p_i$, it gives $\dot{p}_i = -\partial H/\partial q_i$.  
- If $f = H$, it gives $\dot{H} = \{H,H\} = 0$ (energy conservation for time-independent $H$).  
- If $f$ generates a symmetry, then $\{f,H\}=0$ is the condition for a conserved quantity.

Thus the Poisson bracket provides the **dynamical law** of Hamiltonian mechanics in its most compact and powerful form.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.7.4 Fundamental Poisson Brackets

Canonical coordinates satisfy:

$$
\{ q_i, p_j \} = \delta_{ij},
\qquad
\{ q_i, q_j \} = 0,
\qquad
\{ p_i, p_j \} = 0.
$$

These relations express the **canonical symplectic structure** in algebraic form and are preserved under all canonical transformations.

This mirrors the quantum commutators:

$$
[\hat{q}_i, \hat{p}_j] = i\hbar\,\delta_{ij}.
$$

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.7.5 Poisson Brackets as Generators of Canonical Transformations

If a function $G(q,p)$ is used as a generating function for an **infinitesimal** canonical transformation, then the change in any function $f$ is:

$$
\delta f = \epsilon \{f, G\},
$$

where $\epsilon$ is a small parameter.

This statement is the classical version of:

- unitary transformations in quantum mechanics  
- gauge transformations in electromagnetism  
- the Noether correspondence between symmetries and conserved quantities  

Thus the Poisson bracket provides the **mechanism** by which symmetries act on phase-space variables.

**Example: Poisson Brackets in the 2D Harmonic Oscillator**

Consider the Hamiltonian

$$
H = \frac{1}{2m}(p_1^2 + p_2^2) + \frac{k}{2}(q_1^2 + q_2^2).
$$

Compute:

1. $\{q_i , H\}$  
2. $\{p_i , H\}$  
3. $\{L_z , H\}$ where $L_z = q_1 p_2 - q_2 p_1$.

---

**1. Compute $\{q_i, H\}$**

Using the definition,

$$
\{q_i, H\} = \frac{\partial q_i}{\partial q_j}\frac{\partial H}{\partial p_j}
              -\frac{\partial q_i}{\partial p_j}\frac{\partial H}{\partial q_j}.
$$

Since  
$\partial q_i/\partial q_j = \delta_{ij}$  and $\partial q_i/\partial p_j = 0$:

$$
\{q_i, H\} = \frac{\partial H}{\partial p_i} = \frac{p_i}{m}.
$$

Which reproduces $\dot{q}_i$.

---

**2. Compute $\{p_i, H\}$**

Similarly,

$$
\{p_i, H\}
= - \frac{\partial H}{\partial q_i}
= - k q_i,
$$

which is exactly $\dot{p}_i$.

---

**3. Compute the conserved quantity $\{L_z, H\}$**

First compute derivatives:

$$
\frac{\partial L_z}{\partial q_1}=p_2,\qquad 
\frac{\partial L_z}{\partial q_2}=-p_1,
$$

$$
\frac{\partial L_z}{\partial p_1}=-q_2,\qquad
\frac{\partial L_z}{\partial p_2}=q_1.
$$

Now evaluate:

$$
\{L_z,H\}
=
\sum_{i=1}^2
\left(
\frac{\partial L_z}{\partial q_i}\frac{\partial H}{\partial p_i}
- 
\frac{\partial L_z}{\partial p_i}\frac{\partial H}{\partial q_i}
\right).
$$

Compute each term:

- For $i=1$:  

  $$
  p_2\frac{p_1}{m} - (-q_2)(kq_1)
  = \frac{p_1p_2}{m} + k q_1 q_2
  $$

- For $i=2$:  

  $$
  (-p_1)\frac{p_2}{m} - (q_1)(kq_2)
  = -\frac{p_1p_2}{m} - k q_1 q_2
  $$

Adding them:

$$
\{L_z,H\} = 0.
$$

Thus **$L_z$ is conserved** due to rotational symmetry.

**Box Activity 1: Show That Poisson Brackets Are Preserved by Canonical Transformations**

Let $z = (q,p)$ and $Z=(Q,P)$ be related by a canonical transformation.  
Show that for any functions $f$ and $g$,

$$
\{f,g\} = \{f,g\}' ,
$$

where the prime denotes Poisson brackets computed in $(Q,P)$.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Detailed Solution
</summary>

<div style="background-color:#e8f5e9; padding:12px; border-radius:4px; margin-top:6px;">

We use the matrix form:

$$
\{f,g\} = (\nabla f)^{\mathrm{T}} J (\nabla g).
$$

Under a canonical transformation:

- $\nabla f = M^{\mathrm{T}} \nabla_Z f$,  
- $\nabla g = M^{\mathrm{T}} \nabla_Z g$,  
- and the defining symplectic condition is  

  $$
  M^{\mathrm{T}} J M = J.
  $$

Compute:

$$
\{f,g\}
= (\nabla f)^{\mathrm{T}}J(\nabla g)
= (M^{\mathrm{T}}\nabla_Z f)^{\mathrm{T}}J(M^{\mathrm{T}}\nabla_Z g).
$$

Transpose the left term:

$$
= (\nabla_Z f)^{\mathrm{T}} M J M^{\mathrm{T}} \nabla_Z g.
$$

Use the symplectic identity $MJM^{\mathrm{T}} = J$:

$$
\{f,g\}
= (\nabla_Z f)^{\mathrm{T}} J (\nabla_Z g)
= \{f,g\}'.
$$

Thus Poisson brackets are **invariant** under canonical transformations.

**Poisson Brackets as Classical Commutators**

Poisson brackets encode how dynamical quantities generate transformations in phase space, exactly as commutators encode how observables generate transformations in quantum mechanics.  

The correspondence rule

$$
\{f,g\} \;\longrightarrow\; \frac{1}{i\hbar}[\hat{f},\hat{g}]
$$

shows that quantum mechanics preserves the same algebraic structure:  
**symmetry → generator → conservation law**, with commutators playing the role of quantum Poisson brackets.

## A1.7.6 Algebraic Properties of Poisson Brackets

Poisson brackets satisfy four key properties that make them a **Lie algebra**:

### 1. **Linearity**
$$
\{af + bg, h\} = a\{f,h\} + b\{g,h\}.
$$

### 2. **Antisymmetry**
$$
\{f,g\} = -\{g,f\}.
$$

### 3. **Leibniz Rule (Derivation Property)**
$$
\{fg, h\} = f\{g,h\} + g\{f,h\}.
$$

This shows PBs act like derivatives.

### 4. **Jacobi Identity**
$$
\{f,\{g,h\}\} + \{g,\{h,f\}\} + \{h,\{f,g\}\} = 0.
$$

The Jacobi identity is essential: it guarantees that Poisson brackets form a **Lie algebra**, enabling the elegant algebraic treatment of symmetries.

<hr style="height:5px;border-width:0;color:gray;background-color:gray">

## A1.7.7 Bridge to Hamilton–Jacobi Theory

Poisson brackets turn the space of dynamical quantities into a Lie algebra where each quantity generates a flow in phase space.  

Hamilton–Jacobi theory takes the next step: instead of tracking flows generated by all variables, it looks for a *single* function $S(q,t)$ whose derivatives generate a full canonical transformation sending the Hamiltonian to a constant.  

This transformation straightens out the flow, reducing the dynamics to a trivial evolution in the new coordinates.

In this way, Poisson brackets (infinitesimal canonical transformations) naturally lead to Hamilton–Jacobi theory (a global canonical transformation generated by the action function $S$).
