# 1.1 — Build the Sentence / 文を作ろう

<hr style="border:0; height:4px; background-color:#6b8e73; margin:32px 0;">

🌱 **Grow your English, one sentence at a time!**  
**１文ずつ、英語の力を育てよう！**

You have learned how to use **I am** and **You are**. Now, build your own sentences! Read the Japanese sentence, then drag the English word tiles into the correct order.

**I am** と **You are** を学びました。今度は自分で英文を作りましょう！日本語の文を読んで、英語の単語カードを正しい順番にドラッグしましょう。

Think about the three parts:  
３つのパーツを考えましょう。

**(Who? / だれが？) (am / are) (Description / どんな人・様子？)**

For example / 例：

**私はうれしいです。 → (I) (am) (happy).**

Remember: **I → am** and **You → are**. Some words will not be needed—choose the ones that match the Japanese sentence.

**I には am、You には are** を使います。使わない単語もあります。日本語の文に合う単語を選びましょう。

After checking your answer, listen and say the sentence aloud. Take your time, learn new words, and try again!

答えを確認したら、音声を聞いて、声に出して言ってみましょう。あせらず、新しい単語を覚えながら、何度でも挑戦しましょう！
