## How to use this notebook · このノートの使い方

- Words are in **A–Z order within each category**. 各グループの単語はアルファベット順です。
- **★** means the word appears in more than one category. Read the “Also used as” column. ★は、ほかの品詞でも使われる単語です。
- A word’s job depends on its sentence: **I cook.** (verb) / **He is a cook.** (noun). 文の中での役割を見てみましょう。
- Say a word, think of its meaning, and make a short sentence with Sanae. 単語を声に出し、意味を考え、早苗先生と短い文を作ってみましょう。

**For teachers / 先生へ：** All 436 supplied entries are retained. Common additional uses are included, so some uses go beyond Grade 5. This is a classroom classification, not a list of every rare dictionary sense or an official exam list. Categories can differ between dictionaries, especially for *same* and *little*. 大文字の月・曜日はそのまま残しています。すべてのまれな用法を網羅するものではありません。

**No Python needed / Pythonの実行は不要です。** Read the Markdown cells. Illustrations are embedded PNG attachments.

## Nouns
### 名詞

A noun names a person, animal, place, thing, or idea. Examples: teacher, rabbit, school, love. Days and months are also nouns.

名詞は、人・動物・場所・もの・考えなどを表す言葉です。例：teacher（先生）、rabbit（うさぎ）、school（学校）、love（愛）。曜日や月の名前も名詞です。

## Verbs
### 動詞

A verb tells us what someone or something does, or describes a state. Examples: swim, eat, know. “Wake up” is a phrasal verb: two words working together.

動詞は、人やものの動きや状態を表す言葉です。例：swim（泳ぐ）、eat（食べる）、know（知っている）。wake up（目を覚ます）は、２語で１つの意味を表す句動詞です。

## Teacher notes · 先生用メモ

- The main list preserves the supplied spelling: **shoes** and **socks** stay plural; **wake up** stays together. 元のつづり・複数形・句動詞を保っています。
- Colour words can name a colour (noun) or describe something (adjective): **I like blue. / a blue bag** 色の名前は名詞にも形容詞にもなります。
- Some added verb uses are extension vocabulary: **box, bus, coat, cloud, flower, garden, gift, house, milk, nurse, picture, salt, skirt, sugar, telephone**. Teach the familiar meaning first. 追加した動詞の用法には発展的なものがあります。まず身近な意味から学びましょう。
- **Live** has different pronunciations as a verb and adjective; **close** changes its final sound. **Read** is /riːd/ in the present and /red/ in the past. **Wind** is listed here as the noun /wɪnd/ (風); the verb /waɪnd/ (巻く) is a different-pronunciation extension and is not included in the main verb table.
- Capitalized **May** and **March** are months here. Lowercase **may** (modal verb) and **march** (noun/verb) are different entries and were not added.
