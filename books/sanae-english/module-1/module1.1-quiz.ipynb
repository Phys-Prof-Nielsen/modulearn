# 1.1 — Check Your Understanding / 理解度をチェックしよう

<hr style="border:0; height:4px; background-color:#6b8e73; margin:32px 0;">

🌱 **Let’s practice what you have learned!**  
**学んだことを練習しましょう！**

Try 10 questions using **I am** and **You are**. Choose the correct be verb, put words in order, and practice the short forms **I’m** and **You’re**.

**I am** と **You are** を使う10問に挑戦しましょう。正しいbe動詞を選んだり、単語を並べ替えたり、短縮形の **I’m** と **You’re** を練習したりします。

Remember / 覚えておこう：

- **I am → I’m**
- **You are → You’re**

The vocabulary changes each time you play, so you can practice sentence patterns while learning new words. There is no timer—take your time and think about each answer.

挑戦するたびに単語が変わるので、新しい単語を覚えながら文の形を練習できます。時間制限はありません。あせらず、１問ずつ考えましょう。

After checking your answer, listen and say the sentence aloud. Mistakes help you learn—try again and watch your English grow!

答えを確認したら、音声を聞いて、声に出して言ってみましょう。まちがいも学びの一歩です。何度も挑戦して、英語の力を育てましょう！
