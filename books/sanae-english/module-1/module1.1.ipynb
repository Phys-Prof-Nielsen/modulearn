# 1.1 — I am / You are
## 「私は〜です」「あなたは〜です」

We use **I am** to talk about ourselves and **You are** to talk about the person we are speaking to.

自分のことを言うときは **I am**、話している相手のことを言うときは **You are** を使います。

Let’s build a sentence with **three parts**.
３つのパーツで文を作りましょう。

**(Who? / だれは？) (am / are) (name or description / 名前・説明)**

| Who? / 主語 | Be verb / be動詞 | Name or description / 名前・説明 |
|---|---|---|
| (I) | (am) | (Haru). |
| (You) | (are) | (kind). |

**Remember / 覚えよう**

- **I → am**
- **You → are**
- Always write **I** with a capital letter. **I** はいつも大文字です。

The third part can contain more than one word:

**(I) (am) (a student).**

３つ目のパーツは、２語以上になることもあります。
かっこは文のしくみを見るためのものです。ふつうに書くときは、かっこをつけません。

**Short forms / 短縮形**

| Full form / 短縮しない形 | Short form / 短縮形 |
|---|---|
| I am | I'm |
| You are | You're |

The meaning stays the same. In a short form, the first two parts join together.

短縮形では、最初の２つのパーツがつながります。意味は変わりません。

## Example 1 — Introducing yourself / 自己紹介

<img src="fig-1.1.1.png" alt="A student introducing themselves by pointing gently to their own chest in a school garden." width="1000">

Haru is introducing themselves.
ハルさんが自己紹介をしています。

**(I) (am) (Haru).**

**I am Haru. → I'm Haru.**

私はハルです。

| Part / パーツ | Job / 役割 |
|---|---|
| (I) | The person speaking / 話している自分 |
| (am) | The be verb used with I / I と使うbe動詞 |
| (Haru) | The speaker’s name / 話している人の名前 |

**Say it / 言ってみよう**

Introduce yourself using your own name.
自分の名前で言ってみましょう。

**(I) (am) (________).**

## Example 2 — How are you feeling? / 今の様子を伝えよう

<img src="fig-1.1.2.png" alt="A student looking forward to eating curry rice at a school lunch table." width="1000">

Lunch is ready! The student is talking about how they feel.
お昼ごはんの時間です。自分の今の様子を伝えています。

**(I) (am) (hungry).**

**I am hungry. → I'm hungry.**

私はおなかがすいています。

The third part describes the speaker.
３つ目のパーツは、自分の様子を説明しています。

Keep **I am** and change the description!

**I am** はそのままで、最後の説明を変えてみましょう！

| Three parts / ３つのパーツ | Short form / 短縮形 | Meaning / 意味 |
|---|---|---|
| (I) (am) (happy). | I'm happy. | 私はうれしいです。 |
| (I) (am) (tired). | I'm tired. | 私は疲れています。 |

**Say it / 言ってみよう**

Choose a word that describes how you feel now.
今の自分の様子を表す言葉で言ってみましょう。

## Example 3 — Talking to a friend / 友だちに伝えよう

<img src="fig-1.1.3.png" alt="One student kindly returning a dropped notebook while their classmate thanks them." width="1000">

Your friend gives your notebook back to you. Speak directly to your friend.

友だちがノートを拾ってくれました。その友だちに話しかけましょう。

**(You) (are) (kind).**

**You are kind. → You're kind.**

あなたは親切ですね。

| Part / パーツ | Job / 役割 |
|---|---|
| (You) | The person you are speaking to / 話しかけている相手 |
| (are) | The be verb used with you / you と使うbe動詞 |
| (kind) | A description of that person / 相手の性格を表す説明 |

**Watch the spelling / つづりに注意**

**You're** means **you are**. The apostrophe **'** shows that letters have been left out.

**You're** は **you are** の短縮形です。アポストロフィー **'** は、文字が省略されていることを表します。

**Your** means「あなたの」and has a different job.

**Say it / 言ってみよう**

Tell a friend: **You're kind.**

The order is **who → be verb → description**.

**A student** stays together in the third part. We use **a** before the singular countable noun **student** here.

語順は **だれは → be動詞 → 説明** です。

**a student** は２語ですが、１つのパーツです。ここでは、数えられる名詞 **student** が１人を表すので、前に **a** をつけます。
