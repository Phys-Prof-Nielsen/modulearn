## Problem 1 - Numerical Methods for a Large-Amplitude Pendulum

A laboratory pendulum consists of a small bob of mass $m$ attached to a light, rigid rod of length $\ell$. The pivot is fixed, and friction is negligible.

Let $\theta$ measure the angle from the downward vertical, and let $\omega=\dot{\theta}$.

The pendulum is released from rest at

$$
\theta_0=60^\circ.
$$

You want to predict its motion without making the small-angle approximation.

1. Draw a position diagram and derive the differential equation for $\theta$.
2. Explain why the usual sinusoidal simple-harmonic-motion solution is not appropriate.
3. Starting from Taylor expansions, derive the forward Euler updates for $\theta$ and $\omega$.
4. Write the corresponding backward Euler and velocity-first Euler–Cromer updates. Explain the differences among the three methods.
5. Derive a central-difference approximation for $\ddot{\theta}$ and use it to obtain a position-based recurrence for $\theta$.
6. Explain how to initialize the central-difference calculation.
7. Identify physical and numerical checks you would use to assess the results.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Physical model</strong>

The tangential component of gravity is $-mg\sin\theta$. The bob's tangential acceleration is $\ell\ddot{\theta}$, so

$$
m\ell\ddot{\theta}=-mg\sin\theta.
$$

Therefore,

$$
\boxed{\ddot{\theta}=-\frac{g}{\ell}\sin\theta.}
$$

The rod's force is radial and does not enter the tangential equation.

Introduce angular velocity to write two first-order equations:

$$
\boxed{
\dot{\theta}=\omega,
\qquad
\dot{\omega}=-\frac{g}{\ell}\sin\theta.
}
$$

At $60^\circ$, replacing $\sin\theta$ by $\theta$ is not a sufficiently accurate description of the full motion. The standard sinusoidal solution solves the linearized equation, not this nonlinear equation.

Numerical integration provides a practical way to calculate the motion without linearization.

<strong>Deriving forward Euler</strong>

For time step $h=\Delta t$, Taylor expansion gives

$$
\theta(t+h)=\theta(t)+\dot{\theta}(t)h+O(h^2),
$$

$$
\omega(t+h)=\omega(t)+\dot{\omega}(t)h+O(h^2).
$$

Retaining the first-order terms,

$$
\boxed{
\begin{aligned}
\theta_{n+1}&=\theta_n+h\omega_n,\\
\omega_{n+1}&=\omega_n-h\frac{g}{\ell}\sin\theta_n.
\end{aligned}
}
$$

Both derivatives are evaluated at the beginning of the step. Both right-hand sides must use the old values.

Angles must be expressed in radians when evaluating these equations numerically.

<strong>Backward Euler</strong>

Backward Euler evaluates the derivatives at the end of the step:

$$
\boxed{
\begin{aligned}
\theta_{n+1}&=\theta_n+h\omega_{n+1},\\
\omega_{n+1}&=\omega_n-h\frac{g}{\ell}\sin\theta_{n+1}.
\end{aligned}
}
$$

These equations are implicit. Eliminating $\omega_{n+1}$ gives

$$
\boxed{
\theta_{n+1}
+h^2\frac{g}{\ell}\sin\theta_{n+1}
-\theta_n-h\omega_n=0.
}
$$

A root-finding method can determine $\theta_{n+1}$ at each step. Then calculate $\omega_{n+1}$.

<strong>Euler–Cromer</strong>

The velocity-first version updates angular velocity before angle:

$$
\boxed{
\begin{aligned}
\omega_{n+1}&=\omega_n-h\frac{g}{\ell}\sin\theta_n,\\
\theta_{n+1}&=\theta_n+h\omega_{n+1}.
\end{aligned}
}
$$

It is explicit: the first equation immediately gives the new angular velocity, which is then used in the second equation.

Unlike backward Euler, its angular acceleration is evaluated at $\theta_n$.

<strong>Deriving central difference</strong>

Expand forward and backward about $t_n$:

$$
\theta_{n+1}
=
\theta_n+h\dot{\theta}_n
+\frac{h^2}{2}\ddot{\theta}_n
+\frac{h^3}{6}\theta_n^{(3)}
+O(h^4),
$$

$$
\theta_{n-1}
=
\theta_n-h\dot{\theta}_n
+\frac{h^2}{2}\ddot{\theta}_n
-\frac{h^3}{6}\theta_n^{(3)}
+O(h^4).
$$

Adding cancels the odd-power terms:

$$
\boxed{
\ddot{\theta}_n
=
\frac{\theta_{n+1}-2\theta_n+\theta_{n-1}}{h^2}
+O(h^2).
}
$$

Substitute the pendulum equation and rearrange:

$$
\boxed{
\theta_{n+1}
=
2\theta_n-\theta_{n-1}
-h^2\frac{g}{\ell}\sin\theta_n.
}
$$

Subtracting the Taylor expansions also gives a centered angular-velocity estimate:

$$
\boxed{
\omega_n\approx\frac{\theta_{n+1}-\theta_{n-1}}{2h}.
}
$$

<strong>Initialization</strong>

The recurrence requires two angles. Given $\theta_0$ and $\omega_0=0$, use

$$
\theta_1
\approx
\theta_0+h\omega_0+\frac{h^2}{2}\ddot{\theta}_0.
$$

Thus,

$$
\boxed{
\theta_1
\approx
\theta_0-\frac{h^2g}{2\ell}\sin\theta_0.
}
$$

<strong>Checks and interpretation</strong>

The exact mechanical energy is constant:

$$
\boxed{
E=\frac12m\ell^2\omega^2+mg\ell(1-\cos\theta).
}
$$

Useful checks include:

- The initial angular acceleration is negative.
- The speed is greatest at the bottom.
- Turning points should remain near $\pm60^\circ$.
- Reducing $h$ should make the predicted trajectory and period converge.
- Energy should remain close to its initial value over the time interval of interest.

Forward Euler commonly introduces artificial energy growth in oscillatory motion. Backward Euler commonly introduces numerical damping. Euler–Cromer and central difference generally represent oscillations better with a sufficiently small step, although neither guarantees exact energy conservation.

In the small-amplitude limit, the period should approach

$$
T_{\mathrm{small}}=2\pi\sqrt{\frac{\ell}{g}}.
$$

The large-amplitude period should be longer.

## Problem 2 - A Moving Cart Collecting Sand

A cart moves along a straight, level, frictionless track. Initially, its mass is $M$ and its velocity is $U$ in the positive $x$-direction.

Sand falls vertically from a stationary hopper into the cart at constant mass rate $\mu>0$. The incoming sand has zero horizontal velocity in the ground frame and remains in the cart after landing.

A motor applies a constant horizontal force $F>0$ to the cart. Ignore any horizontal force from the hopper.

1. Identify the system and reference frame. Explain why the cart's changing mass requires careful momentum accounting.
2. Construct a horizontal momentum balance over a short time interval.
3. Derive the differential equation for the cart's velocity.
4. Determine its velocity as a function of time.
5. Find its velocity when the total mass reaches $2M$.
6. Explain when the cart accelerates, decelerates, or moves at constant velocity.
7. Explain why horizontal momentum can be conserved when $F=0$ even though kinetic energy is not conserved.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>System and incoming momentum</strong>

Use the ground frame, assumed inertial. The moving cart and collected sand form an open system whose mass is

$$
\boxed{m(t)=M+\mu t.}
$$

Incoming sand carries no horizontal momentum before landing.

For a short-time momentum balance, consider the cart, its contents, and the sand that joins during $dt$.

<strong>Momentum balance</strong>

Initially, horizontal momentum is

$$
p_{\mathrm{initial}}=mv+(\mu\,dt)(0).
$$

After collection, it is

$$
p_{\mathrm{final}}=(m+\mu\,dt)(v+dv).
$$

The external horizontal impulse is $F\,dt$, so

$$
mv+F\,dt=(m+\mu\,dt)(v+dv).
$$

Neglecting the product $\mu\,dt\,dv$,

$$
F\,dt=m\,dv+\mu v\,dt.
$$

Thus,

$$
\boxed{
m\frac{dv}{dt}=F-\mu v.
}
$$

The term $\mu v$ accounts for the horizontal momentum required to bring incoming sand up to the cart's speed. It does not represent friction with the track.

<strong>Solving for velocity</strong>

The equation can be written

$$
\frac{d}{dt}(mv)=F.
$$

Integrating and using $v(0)=U$,

$$
mv=MU+Ft.
$$

Therefore,

$$
\boxed{
v(t)=\frac{MU+Ft}{M+\mu t}.
}
$$

An informative alternative form is

$$
\boxed{
v(t)=\frac{F}{\mu}
+\frac{M}{M+\mu t}
\left(U-\frac{F}{\mu}\right).
}
$$

<strong>When the mass doubles</strong>

The condition $m=2M$ gives

$$
t=\frac{M}{\mu}.
$$

Therefore,

$$
\boxed{
v=\frac12\left(U+\frac{F}{\mu}\right).
}
$$

<strong>Acceleration and limiting behavior</strong>

Differentiate the velocity:

$$
\boxed{
a(t)=\frac{M(F-\mu U)}{(M+\mu t)^2}.
}
$$

Consequently:

- If $U<F/\mu$, the cart speeds up.
- If $U>F/\mu$, the cart slows down.
- If $U=F/\mu$, its velocity stays constant.

For continued sand collection,

$$
\boxed{v(t)\longrightarrow\frac{F}{\mu}.}
$$

At this limiting velocity, the applied force supplies exactly the rate of horizontal momentum needed by the incoming sand.

<strong>Momentum versus kinetic energy</strong>

If $F=0$,

$$
mv=MU,
$$

so the total horizontal momentum remains constant.

However,

$$
K=\frac12mv^2=\frac{M^2U^2}{2m}
$$

decreases as mass is collected. The collection is inelastic: mechanical energy is converted into internal energy and other forms.

More generally, the horizontal kinetic-energy rate is

$$
\boxed{
\frac{dK}{dt}=Fv-\frac12\mu v^2.
}
$$

The motor supplies power $Fv$, while inelastic collection removes horizontal kinetic energy at rate $\frac12\mu v^2$.

## Problem 3 - A Satellite's Rotation and a Ground-Test Gyroscope

A satellite has mass $100\ \mathrm{kg}$. At one instant, its center-of-mass position and velocity relative to an inertial origin $O$ are

$$
\mathbf{R}_C=7.0\times10^6\hat{\mathbf{x}}\ \mathrm{m},
\qquad
\mathbf{V}_C=7.5\times10^3\hat{\mathbf{y}}\ \mathrm{m/s}.
$$

At this instant, its inertia tensor about its center of mass, expressed in the same axes, is

$$
\mathbf{I}_C=
\begin{pmatrix}
6&2&0\\
2&6&0\\
0&0&10
\end{pmatrix}
\ \mathrm{kg\,m^2},
$$

and its angular velocity is

$$
\boldsymbol{\omega}
=
0.20\hat{\mathbf{x}}+0.10\hat{\mathbf{z}}
\ \mathrm{rad/s}.
$$

1. Explain the physical meaning of the angular velocity vector. Find the velocity relative to the center of mass of a point at
   $$
   \mathbf{r}_{P/C}=0.50\hat{\mathbf{y}}\ \mathrm{m}.
   $$
2. Calculate the spin angular momentum and determine whether it is parallel to $\boldsymbol{\omega}$.
3. Find the principal axes and principal moments of inertia.
4. Calculate the total angular momentum about $O$, distinguishing spin from orbital angular momentum.

The satellite also carries a small axisymmetric rotor. Before installation, the rotor is tested on Earth. It is attached to the far end of a uniform horizontal shaft whose other end is pivoted.

The rotor has mass $1.0\ \mathrm{kg}$ and spin moment of inertia $0.020\ \mathrm{kg\,m^2}$. The shaft has length $0.30\ \mathrm{m}$ and mass $0.20\ \mathrm{kg}$. The rotor's center is at the far end of the shaft.

The assembly spins about the shaft at $200\ \mathrm{rad/s}$. Treat the shaft as a thin rod and assume rapid spin with steady horizontal precession.

5. Determine the gravitational torque about the pivot and estimate the precession rate.
6. If the shaft and spin angular momentum initially point along $+\hat{\mathbf{x}}$, determine the precession direction. Take upward as $+\hat{\mathbf{z}}$.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Angular velocity and point velocity</strong>

The angular velocity vector gives the instantaneous rotation axis through the center of mass, its sense by the right-hand rule, and the rotation rate through its magnitude.

For a rigid body,

$$
\mathbf{v}_{P/C}=\boldsymbol{\omega}\times\mathbf{r}_{P/C}.
$$

Thus,

$$
\begin{aligned}
\mathbf{v}_{P/C}
&=
(0.20\hat{\mathbf{x}}+0.10\hat{\mathbf{z}})
\times(0.50\hat{\mathbf{y}})\\
&=
\boxed{-0.050\hat{\mathbf{x}}+0.100\hat{\mathbf{z}}\ \mathrm{m/s}.}
\end{aligned}
$$

The inertial velocity is $\mathbf{V}_C+\mathbf{v}_{P/C}$.

<strong>Spin angular momentum</strong>

$$
\mathbf{L}_C=\mathbf{I}_C\boldsymbol{\omega}.
$$

Therefore,

$$
\boxed{
\mathbf{L}_C
=
1.2\hat{\mathbf{x}}+0.4\hat{\mathbf{y}}+1.0\hat{\mathbf{z}}
\ \mathrm{kg\,m^2/s}.
}
$$

It is not parallel to $\boldsymbol{\omega}$: it has a nonzero $y$-component while $\boldsymbol{\omega}$ does not.

The inertia tensor relates a rotation direction and rate to the corresponding angular momentum. It generally changes the direction as well as the magnitude.

<strong>Principal axes</strong>

Solve

$$
\mathbf{I}_C\mathbf{e}_k=I_k\mathbf{e}_k.
$$

For the $xy$ block,

$$
\det
\begin{pmatrix}
6-I&2\\
2&6-I
\end{pmatrix}
=
(6-I)^2-4=0.
$$

This gives $I=4$ and $I=8$. The third moment is $10$.

A right-handed orthonormal principal basis is

$$
\boxed{
\begin{aligned}
\hat{\mathbf{e}}_1&=\frac{\hat{\mathbf{x}}-\hat{\mathbf{y}}}{\sqrt2},
& I_1&=4\ \mathrm{kg\,m^2},\\
\hat{\mathbf{e}}_2&=\frac{\hat{\mathbf{x}}+\hat{\mathbf{y}}}{\sqrt2},
& I_2&=8\ \mathrm{kg\,m^2},\\
\hat{\mathbf{e}}_3&=\hat{\mathbf{z}},
& I_3&=10\ \mathrm{kg\,m^2}.
\end{aligned}
}
$$

Rotation about any one principal axis gives $\mathbf{L}_C$ parallel to $\boldsymbol{\omega}$.

<strong>Orbital and total angular momentum</strong>

The orbital contribution is

$$
\mathbf{L}_{\mathrm{orb}}
=
\mathbf{R}_C\times M\mathbf{V}_C
=
\boxed{5.25\times10^{12}\hat{\mathbf{z}}\ \mathrm{kg\,m^2/s}.}
$$

Thus,

$$
\boxed{
\mathbf{L}_O
=
1.2\hat{\mathbf{x}}+0.4\hat{\mathbf{y}}
+\left(5.25\times10^{12}+1.0\right)\hat{\mathbf{z}}
\ \mathrm{kg\,m^2/s}.
}
$$

The center-of-mass inertia tensor gives the spin contribution, not the orbital contribution.

<strong>Ground test: center of mass and torque</strong>

The combined center-of-mass distance from the pivot is

$$
d_C=
\frac{(1.0)(0.30)+(0.20)(0.15)}{1.20}
=
\boxed{0.275\ \mathrm{m}}.
$$

The gravitational torque magnitude is therefore

$$
\tau=(1.20)g(0.275)
=
\boxed{3.24\ \mathrm{N\,m}}
$$

using $g=9.81\ \mathrm{m/s^2}$.

Equivalently,

$$
\tau=(1.0)g(0.30)+(0.20)g(0.15).
$$

The shaft's mass contributes to torque even though its axial spin inertia is negligible in the thin-rod model.

<strong>Precession</strong>

The dominant spin angular momentum is

$$
L_s=I_s\omega_s=(0.020)(200)
=
4.0\ \mathrm{kg\,m^2/s}.
$$

For steady precession,

$$
\boldsymbol{\tau}
\approx\boldsymbol{\Omega}_p\times\mathbf{L}_s.
$$

Therefore,

$$
\boxed{
\Omega_p=\frac{\tau}{L_s}
\approx0.809\ \mathrm{rad/s}.
}
$$

With the shaft along $+\hat{\mathbf{x}}$ and gravity downward,

$$
\boldsymbol{\tau}=+\tau\hat{\mathbf{y}}.
$$

Since $\hat{\mathbf{z}}\times\hat{\mathbf{x}}=\hat{\mathbf{y}}$,

$$
\boxed{\boldsymbol{\Omega}_p=+0.809\hat{\mathbf{z}}\ \mathrm{rad/s}.}
$$

The torque changes the direction of the dominant spin angular momentum. The approximation neglects nutation and the smaller angular momentum associated with precession.

## Problem 4 - A Slider in a Rotating Inspection Platform

An inspection platform rotates about a fixed vertical axis. Its angular velocity is $\boldsymbol{\Omega}(t)$.

A small instrument carriage of mass $m$ moves along a straight radial guide attached to the platform. A motor controls its radial motion, and the guide prevents sideways motion. All coordinates below refer to the platform frame.

1. Starting from the vector derivative transformation, derive the acceleration transformation between the inertial and rotating frames.
2. Write Newton's second law in the platform frame and identify the fictitious forces.
3. For carriage position $\mathbf{r}=r(t)\hat{\mathbf{r}}$, express its relative velocity and acceleration.
4. Find the radial and tangential components of the real horizontal force required to produce the prescribed motion.

At one instant,

$$
m=5.0\ \mathrm{kg},
\qquad
r=2.0\ \mathrm{m},
\qquad
\dot r=0.40\ \mathrm{m/s},
\qquad
\ddot r=0.10\ \mathrm{m/s^2},
$$

$$
\boldsymbol{\Omega}=0.60\hat{\mathbf{z}}\ \mathrm{rad/s},
\qquad
\dot{\boldsymbol{\Omega}}=0.20\hat{\mathbf{z}}\ \mathrm{rad/s^2}.
$$

5. Calculate the centrifugal, Coriolis, and Euler forces and the required real horizontal force.
6. Explain how the forces change if the platform rotates at constant angular velocity and the carriage remains fixed on the guide.

Ignore vertical motion; the platform supports the carriage's weight.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Derivative transformation</strong>

For any vector $\mathbf{Q}$,

$$
\left(\frac{d\mathbf{Q}}{dt}\right)_I
=
\left(\frac{d\mathbf{Q}}{dt}\right)_R
+\boldsymbol{\Omega}\times\mathbf{Q}.
$$

The platform origin is fixed, so the velocity transformation is

$$
\mathbf{v}_I=\mathbf{v}'+\boldsymbol{\Omega}\times\mathbf{r}.
$$

Differentiate in the inertial frame:

$$
\mathbf{a}_I
=
\left(\frac{d\mathbf{v}'}{dt}\right)_I
+\dot{\boldsymbol{\Omega}}\times\mathbf{r}
+\boldsymbol{\Omega}\times
\left(\frac{d\mathbf{r}}{dt}\right)_I.
$$

Using

$$
\left(\frac{d\mathbf{v}'}{dt}\right)_I
=
\mathbf{a}'+\boldsymbol{\Omega}\times\mathbf{v}'
$$

and

$$
\left(\frac{d\mathbf{r}}{dt}\right)_I
=
\mathbf{v}'+\boldsymbol{\Omega}\times\mathbf{r},
$$

gives

$$
\boxed{
\mathbf{a}_I
=
\mathbf{a}'
+2\boldsymbol{\Omega}\times\mathbf{v}'
+\dot{\boldsymbol{\Omega}}\times\mathbf{r}
+\boldsymbol{\Omega}\times(\boldsymbol{\Omega}\times\mathbf{r}).
}
$$

If the origin also accelerated, an additional $\mathbf{A}_{O'}$ would appear.

<strong>Newton's second law in the platform frame</strong>

Rearranging $m\mathbf{a}_I=\mathbf{F}_{\mathrm{real}}$,

$$
\boxed{
m\mathbf{a}'
=
\mathbf{F}_{\mathrm{real}}
+\mathbf{F}_{\mathrm{Cor}}
+\mathbf{F}_{\mathrm{Euler}}
+\mathbf{F}_{\mathrm{cf}},
}
$$

where

$$
\mathbf{F}_{\mathrm{Cor}}=-2m\boldsymbol{\Omega}\times\mathbf{v}',
$$

$$
\mathbf{F}_{\mathrm{Euler}}=-m\dot{\boldsymbol{\Omega}}\times\mathbf{r},
$$

$$
\mathbf{F}_{\mathrm{cf}}
=-m\boldsymbol{\Omega}\times(\boldsymbol{\Omega}\times\mathbf{r}).
$$

There is no translational fictitious force because the origin is fixed.

<strong>Relative motion along the guide</strong>

The radial guide has a fixed direction in the platform frame. Therefore,

$$
\boxed{
\mathbf{v}'=\dot r\hat{\mathbf{r}},
\qquad
\mathbf{a}'=\ddot r\hat{\mathbf{r}}.
}
$$

Do not add the platform's rotation again when calculating these relative derivatives.

<strong>Force components</strong>

With positive $\hat{\boldsymbol{\theta}}$ in the direction of rotation,

$$
\mathbf{F}_{\mathrm{cf}}=m\Omega^2r\hat{\mathbf{r}},
$$

$$
\mathbf{F}_{\mathrm{Cor}}=-2m\Omega\dot r\hat{\boldsymbol{\theta}},
$$

$$
\mathbf{F}_{\mathrm{Euler}}=-m\dot\Omega r\hat{\boldsymbol{\theta}}.
$$

Newton's second law gives

$$
m\ddot r=F_r+m\Omega^2r,
$$

$$
0=F_\theta-2m\Omega\dot r-m\dot\Omega r.
$$

Thus,

$$
\boxed{
F_r=m(\ddot r-\Omega^2r),
\qquad
F_\theta=m(2\Omega\dot r+\dot\Omega r).
}
$$

<strong>Numerical evaluation</strong>

The fictitious forces are

$$
\boxed{\mathbf{F}_{\mathrm{cf}}=3.6\hat{\mathbf{r}}\ \mathrm{N},}
$$

$$
\boxed{\mathbf{F}_{\mathrm{Cor}}=-2.4\hat{\boldsymbol{\theta}}\ \mathrm{N},}
$$

$$
\boxed{\mathbf{F}_{\mathrm{Euler}}=-2.0\hat{\boldsymbol{\theta}}\ \mathrm{N}.}
$$

The required real force is

$$
F_r=(5.0)[0.10-(0.60)^2(2.0)]=-3.1\ \mathrm{N},
$$

$$
F_\theta=(5.0)[2(0.60)(0.40)+(0.20)(2.0)]
=4.4\ \mathrm{N}.
$$

Therefore,

$$
\boxed{
\mathbf{F}_{\mathrm{real}}
=
-3.1\hat{\mathbf{r}}+4.4\hat{\boldsymbol{\theta}}\ \mathrm{N}.
}
$$

Its magnitude is approximately $5.38\ \mathrm{N}$.

The rotating-frame net force is

$$
(-3.1+3.6)\hat{\mathbf{r}}
+(4.4-2.4-2.0)\hat{\boldsymbol{\theta}}
=
0.50\hat{\mathbf{r}}\ \mathrm{N},
$$

which equals $m\mathbf{a}'$.

<strong>Fixed carriage and constant rotation</strong>

If $\dot r=\ddot r=0$ and $\dot\Omega=0$, only the centrifugal fictitious force remains.

The real horizontal force must be

$$
\boxed{\mathbf{F}_{\mathrm{real}}=-m\Omega^2r\hat{\mathbf{r}}.}
$$

In the platform frame, it balances the centrifugal force. In the inertial frame, it produces centripetal acceleration.

## Problem 5 - A Bead Sliding on a Curved Guide

A small bead of mass $m$ slides without friction on a fixed, smooth wire in a vertical plane. The wire has a parabolic shape.

Choose $x$ horizontally and $y$ vertically upward, with the vertex at the origin. The wire passes through the points

$$
(x,y)=(a,h)
\quad\text{and}\quad
(-a,h),
$$

where $a>0$ and $h>0$.

The bead is released from rest at $(a,h)$. Treat the bead as threaded onto the wire, so the wire can exert a normal force in either direction.

1. Draw a position diagram and use the given geometry to construct a constraint equation $g(x,y)=0$.
2. Construct the augmented Lagrangian
   $$
   L'=L+\lambda g.
   $$
3. Apply the Euler–Lagrange equations for $x$, $y$, and $\lambda$.
4. Differentiate the geometric constraint as needed and determine $\lambda$ in terms of $x$ and $\dot{x}$.
5. Find the wire's force on the bead using
   $$
   \mathbf{F}_{\mathrm{constraint}}=\lambda\nabla g.
   $$
6. Determine that force immediately after release and when the bead passes through the bottom.
7. Explain why the wire can change the bead's momentum without doing work.

Use $g_0$ for the gravitational acceleration to distinguish it from the constraint function $g(x,y)$.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Constructing the geometric constraint</strong>

A parabola with vertex at the origin and vertical symmetry axis has equation

$$
y=cx^2.
$$

Because it passes through $(a,h)$,

$$
c=\frac{h}{a^2}.
$$

Define

$$
b=\frac{h}{a^2},
$$

which has dimensions of inverse length. The constraint is

$$
\boxed{g(x,y)=y-bx^2=0.}
$$

<strong>Augmented Lagrangian</strong>

The kinetic and potential energies are

$$
T=\frac12m(\dot{x}^2+\dot{y}^2),
\qquad
V=mg_0y.
$$

Thus,

$$
\boxed{
L'=
\frac12m(\dot{x}^2+\dot{y}^2)-mg_0y
+\lambda(y-bx^2).
}
$$

<strong>Euler–Lagrange equations</strong>

For $x$,

$$
\boxed{m\ddot{x}+2bx\lambda=0.}
$$

For $y$,

$$
\boxed{m\ddot{y}+mg_0-\lambda=0.}
$$

For $\lambda$,

$$
\boxed{y-bx^2=0.}
$$

These equations determine the motion and the multiplier together.

<strong>Determining the multiplier</strong>

Differentiate the geometric constraint:

$$
\dot{y}=2bx\dot{x}.
$$

Differentiate again:

$$
\ddot{y}=2b\dot{x}^2+2bx\ddot{x}.
$$

From the Euler–Lagrange equations,

$$
\ddot{x}=-\frac{2bx\lambda}{m},
\qquad
\ddot{y}=\frac{\lambda}{m}-g_0.
$$

Substituting into the differentiated constraint,

$$
\frac{\lambda}{m}-g_0
=
2b\dot{x}^2-\frac{4b^2x^2\lambda}{m}.
$$

Therefore,

$$
\boxed{
\lambda=
\frac{m(g_0+2b\dot{x}^2)}{1+4b^2x^2}.
}
$$

For this upward-opening guide, $\lambda$ is positive.

<strong>Recovering the wire's force</strong>

The gradient of the constraint is

$$
\nabla g
=
-2bx\hat{\mathbf{x}}+\hat{\mathbf{y}}.
$$

Therefore,

$$
\boxed{
\mathbf{F}_{\mathrm{constraint}}
=
\frac{m(g_0+2b\dot{x}^2)}{1+4b^2x^2}
\left(-2bx\hat{\mathbf{x}}+\hat{\mathbf{y}}\right).
}
$$

A tangent to the wire is proportional to

$$
\hat{\mathbf{x}}+2bx\hat{\mathbf{y}}.
$$

Its dot product with $\nabla g$ is zero, confirming that the calculated force is normal to the wire.

The force magnitude is

$$
N=|\lambda|\,|\nabla g|.
$$

Thus,

$$
\boxed{
N=
\frac{m(g_0+2b\dot{x}^2)}
{\sqrt{1+4b^2x^2}}.
}
$$

Here $\lambda$ is the vertical component of the wire's force, not generally its magnitude.

<strong>Immediately after release</strong>

At $x=a$ and $\dot{x}=0$,

$$
\lambda_{\mathrm{release}}
=
\frac{mg_0}{1+4b^2a^2}.
$$

Using $b=h/a^2$,

$$
\boxed{
\mathbf{F}_{\mathrm{release}}
=
\frac{mg_0}{1+4h^2/a^2}
\left(-\frac{2h}{a}\hat{\mathbf{x}}+\hat{\mathbf{y}}\right).
}
$$

Its magnitude is

$$
\boxed{
N_{\mathrm{release}}
=
\frac{mg_0}{\sqrt{1+4h^2/a^2}}.
}
$$

The negative horizontal component helps accelerate the bead toward the bottom.

<strong>At the bottom</strong>

At $x=0$, the constraint gives $\dot{y}=0$, so the speed is $v_b=|\dot{x}|$.

Energy conservation gives

$$
\frac12mv_b^2=mg_0h,
$$

hence

$$
v_b^2=2g_0h.
$$

At the bottom, $\nabla g=\hat{\mathbf{y}}$, so the wire's force is upward:

$$
\mathbf{F}_{\mathrm{bottom}}
=
m(g_0+2bv_b^2)\hat{\mathbf{y}}.
$$

Therefore,

$$
\boxed{
\mathbf{F}_{\mathrm{bottom}}
=
mg_0\left(1+\frac{4h^2}{a^2}\right)\hat{\mathbf{y}}.
}
$$

This exceeds the weight because the bead's path curves upward at the bottom.

<strong>Momentum change without work</strong>

The instantaneous power delivered by the wire is

$$
P=\mathbf{F}_{\mathrm{constraint}}\cdot\mathbf{v}
=
\lambda\nabla g\cdot\mathbf{v}.
$$

Because the constraint is fixed,

$$
\frac{dg}{dt}=\nabla g\cdot\mathbf{v}=0.
$$

Thus,

$$
\boxed{P=0.}
$$

The wire's force changes the direction of the bead's motion without transferring energy. Gravity changes its speed.
