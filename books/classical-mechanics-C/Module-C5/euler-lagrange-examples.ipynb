# C5.5 Applications of the Euler–Lagrange Equation

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.5.1 Motivation

The Euler–Lagrange equation is one of the most important equations in mathematical physics.

Although it was derived from a purely mathematical problem involving functionals, it appears throughout science whenever nature seems to optimize a quantity.

Examples include:

- finding the shortest path between two points,
- determining the fastest path of descent under gravity,
- predicting how light travels through different media,
- determining the shape of hanging cables,
- deriving the equations of motion used in classical mechanics.

In this module, we explore several famous applications of the Euler–Lagrange equation and see how a single mathematical principle can explain a wide variety of physical phenomena.

<h3 style="margin-top:0;">Example — The Shortest Path Between Two Points</h3>

Suppose we wish to find the curve

$$
y(x)
$$

that connects two fixed points

$$
(x_1,y_1)
\qquad\text{and}\qquad
(x_2,y_2)
$$

while minimizing the total distance traveled.

The differential displacement vector is

$$
d\vec{s}
=
dx\,\hat{i}
+
dy\,\hat{j}.
$$

The magnitude of this displacement is

$$
ds
=
\sqrt{d\vec{s}\cdot d\vec{s}}
=
\sqrt{dx^2+dy^2}.
$$

Factoring out $dx$ gives

$$
ds
=
\sqrt{1+\left(\frac{dy}{dx}\right)^2}\,dx
=
\sqrt{1+(y')^2}\,dx.
$$

The total length of the curve is therefore

$$
J[y]
=
\int_{x_1}^{x_2}
\sqrt{1+(y')^2}\,dx.
$$

We identify the integrand as

$$
F(y,y',x)
=
\sqrt{1+(y')^2}.
$$

To determine the function that minimizes the path length, we apply the Euler–Lagrange equation:

$$
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
-
\frac{\partial F}{\partial y}
=
0.
$$

Since $F$ does not explicitly depend on $y$,

$$
\frac{\partial F}{\partial y}
=
0.
$$

Next,

$$
\frac{\partial F}{\partial y'}
=
\frac{y'}{\sqrt{1+(y')^2}}.
$$

Substituting into the Euler–Lagrange equation gives

$$
\frac{d}{dx}
\left(
\frac{y'}{\sqrt{1+(y')^2}}
\right)
=
0.
$$

A quantity whose derivative is zero must be constant, so

$$
\frac{y'}{\sqrt{1+(y')^2}}
=
C,
$$

where $C$ is a constant.

This implies that

$$
y'
=
\text{constant}.
$$

Integrating once yields

$$
y
=
mx+b,
$$

where the constants $m$ and $b$ are determined by the endpoints.

Therefore, the function that minimizes the distance between two points is a straight line.

$$
\boxed{
y=mx+b
}
$$

Thus, the Euler–Lagrange equation reproduces the familiar geometric result that the shortest distance between two points in a plane is a straight line.

<h3 style="margin-top:0;">Example — The Fastest Path Under Gravity</h3>

One of the most famous problems in the history of physics and mathematics is the **brachistochrone problem**.

The word *brachistochrone* comes from Greek and means *shortest time*.

Suppose a small bead slides without friction under the influence of gravity from one point to another. We wish to determine the path that allows the bead to reach its destination in the least possible time.

At first glance, one might guess that the shortest path—a straight line—would also be the fastest path.

Surprisingly, this is not true.

---

The travel time is given by

$$
t
=
\int \frac{ds}{v},
$$

where

- $ds$ is the differential path length,
- $v$ is the speed of the particle.

From geometry,

$$
ds
=
\sqrt{1+(y')^2}\,dx.
$$

Using conservation of energy, the speed of a particle released from rest is

$$
v
=
\sqrt{2gy},
$$

where $y$ measures the vertical distance fallen.

Substituting into the time integral gives

$$
J[y]
=
\int_{x_1}^{x_2}
\frac{\sqrt{1+(y')^2}}
{\sqrt{2gy}}
\,dx.
$$

This functional represents the total travel time.

The Euler–Lagrange equation can now be applied to determine the function $y(x)$ that minimizes this time.

---

After solving the resulting differential equation, one finds that the optimal curve is not a straight line but a **cycloid**:

$$
x
=
r(\theta-\sin\theta),
$$

$$
y
=
r(1-\cos\theta),
$$

where $r$ is a constant determined by the endpoints.

The curve is generated by tracking a point on the rim of a rolling wheel.

---

The cycloid initially drops steeply, allowing the particle to gain speed quickly. Although the path is longer than a straight line, the increased speed more than compensates for the additional distance.

Consequently, the cycloid produces a shorter travel time than any other path connecting the same two points.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Historical Note</strong>

The brachistochrone problem was posed in 1696 by the mathematician
Johann Bernoulli.

The challenge attracted solutions from many of the greatest mathematicians of the era, including Newton, Leibniz, and Jakob Bernoulli.

The problem played a major role in the development of the calculus of variations and ultimately helped pave the way for modern Lagrangian mechanics.

<h3 style="margin-top:0;">Example — How Light Chooses Its Path</h3>

One of the most remarkable applications of the calculus of variations occurs in optics.

Light traveling between two points appears to "choose" a path that minimizes travel time. This observation is known as **Fermat's Principle**.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Fermat's Principle</strong>

The path taken by light between two points is the path for which the travel time is stationary.

In many situations, this corresponds to the path of least time.
</div>

---

Suppose light travels through a medium whose refractive index varies with position.

The speed of light in the medium is

$$
v
=
\frac{c}{n(x)},
$$

where

- $c$ is the speed of light in vacuum,
- $n(x)$ is the refractive index.

The travel time along a path is

$$
t
=
\int \frac{ds}{v}.
$$

Substituting for $v$ gives

$$
t
=
\int \frac{n(x)}{c}\,ds.
$$

Since the constant factor $1/c$ does not affect the optimization problem, we may define the functional

$$
J[y]
=
\int_{x_1}^{x_2}
n(x)\sqrt{1+(y')^2}\,dx.
$$

The path taken by light is the function $y(x)$ that makes this functional stationary.

---

Applying the Euler–Lagrange equation to this functional ultimately leads to one of the most important results in optics:

$$
n_1\sin\theta_1
=
n_2\sin\theta_2.
$$

This relation is known as **Snell's Law**.

It describes how light bends when moving from one medium to another.

---

If light travels from air into water, then

$$
n_{\rm water}
>
n_{\rm air}.
$$

As a result, the light ray bends toward the normal.

Conversely, when light leaves water and enters air, it bends away from the normal.

The familiar bending of objects viewed through water is a direct consequence of this variational principle.

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Why This Is Amazing</strong>

Snell's law is often introduced in introductory physics as an experimental result.

The calculus of variations shows that it is actually a consequence of a much deeper principle:

$$
\text{Light follows the path of stationary travel time.}
$$

Thus, a simple law of refraction emerges naturally from an optimization principle.
</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>From Fermat to Feynman</strong>

Fermat's Principle suggests that light follows the path of stationary travel time. This raises a fascinating question:

<em>How does light "know" which path to choose?</em>

In the twentieth century, physicist Richard Feynman proposed a radically different viewpoint through his formulation of quantum mechanics known as the <strong>path integral</strong> approach.

Rather than assuming that a photon follows a single path, Feynman proposed that it explores <em>all possible paths</em> connecting the initial and final points.

Each possible path contributes a quantum amplitude, and these amplitudes combine through interference.

Most paths interfere destructively and cancel one another. However, paths near the stationary-action path interfere constructively and reinforce one another.

As a result, the path predicted by classical physics emerges naturally from the collective contribution of all possible paths.

In this picture:

$$
\text{Classical Path}
\quad\Longrightarrow\quad
\text{Stationary Action}
$$

and Fermat's Principle becomes a special case of a much deeper quantum principle.

This idea forms the foundation of <strong>Quantum Electrodynamics (QED)</strong>, one of the most successful scientific theories ever developed.
</div>

<h3 style="margin-top:0;">Example — The Catenary Curve</h3>

Suppose a flexible chain is suspended between two supports and allowed to hang under its own weight.

A natural question arises:

> What shape will the chain assume when it reaches equilibrium?

Many people guess that the chain forms a parabola.

Although the resulting curve looks similar to a parabola, the true solution is a different curve known as a **catenary**.

---

The chain seeks a configuration that minimizes its total gravitational potential energy.

If the chain has linear mass density $\lambda$, then a small segment of length $ds$ has mass

$$
dm=\lambda\,ds.
$$

The corresponding differential potential energy is

$$
dU
=
g\,y\,dm
=
\lambda g y\,ds.
$$

Using

$$
ds
=
\sqrt{1+(y')^2}\,dx,
$$

the total potential energy becomes

$$
U[y]
=
\lambda g
\int
y\sqrt{1+(y')^2}\,dx.
$$

The equilibrium shape is the function $y(x)$ that minimizes this functional.

---

Applying the Euler–Lagrange equation to this problem yields the differential equation describing the chain.

The resulting solution is

$$
y
=
a\cosh\!\left(\frac{x}{a}\right),
$$

where $a$ is a constant determined by the length of the chain and the locations of the supports.

This curve is called a **catenary**.

---

The hyperbolic cosine function is defined as

$$
\cosh x
=
\frac{e^x+e^{-x}}{2}.
$$

Unlike a parabola, the catenary becomes increasingly steep as one moves away from the lowest point.

The difference is often subtle near the center of the chain but becomes more noticeable farther from the midpoint.

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Engineering Application</strong>

The catenary appears throughout engineering and architecture.

Examples include:

<ul>
<li>Suspension bridge cables</li>
<li>Overhead power lines</li>
<li>Telephone and communication cables</li>
<li>Anchor chains suspended in water</li>
</ul>

Understanding the catenary allows engineers to predict cable tensions and structural loads accurately.

## C5.5.2 Final Thoughts

The examples in this module illustrate an important theme that appears throughout physics and mathematics:

> Many physical systems can be understood as solutions to optimization problems.

Using the Euler–Lagrange equation, we found that a single mathematical framework can be applied to a remarkably diverse collection of problems.

For example:

- The shortest path between two points is a straight line.
- The fastest path under gravity is a cycloid.
- The path taken by light leads to Snell's law.
- The equilibrium shape of a hanging chain is a catenary.

Although these systems appear completely unrelated, they all emerge from the same underlying variational principle.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Key Idea</strong>

The Euler–Lagrange equation provides a systematic method for determining the function that makes a functional stationary.

Whenever a physical problem can be written in the form

$$
J[y]
=
\int_a^b F(y,y',x)\,dx,
$$

the optimal function can often be found by solving

$$
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
-
\frac{\partial F}{\partial y}
=
0.
$$

This single equation lies at the heart of variational methods.
</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>From Euler to Einstein to Feynman</strong>

The ideas developed in this module extend far beyond the examples presented here.

In classical mechanics, the Euler–Lagrange equation becomes the foundation of Lagrangian mechanics through Hamilton's Principle.

In general relativity, freely falling objects follow paths through spacetime that make the proper time stationary.

In quantum mechanics, Richard Feynman's path integral formulation proposes that particles explore all possible paths, with the classical path emerging from constructive interference near the stationary-action path.

Thus, a mathematical idea that begins with finding the shortest curve between two points ultimately becomes part of our modern understanding of motion, gravity, light, and quantum phenomena.
