## Problem 1 - Tension in a Simple Pendulum

A mass $m$ is attached to the end of a massless rigid rod of length $\ell$. The rod is pivoted at one end, and the mass swings in a vertical plane under the influence of gravity.

Let $\theta$ be the angle the rod makes with the downward vertical direction.

Find the tension $T$ in the rod using:

1. a Newtonian force analysis,
2. the Lagrange multiplier method.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Newtonian Approach</strong>

Let the pendulum be described using polar coordinates

$$
(r,\theta),
$$

where the constraint requires

$$
r=\ell.
$$

The acceleration in polar coordinates is

$$
\vec a
=
(\ddot r-r\dot\theta^2)\hat e_r
+
(r\ddot\theta+2\dot r\dot\theta)\hat e_\theta.
$$

Since the rod has fixed length,

$$
r=\ell,
\qquad
\dot r=0,
\qquad
\ddot r=0.
$$

Therefore,

$$
\vec a
=
-\ell\dot\theta^2\hat e_r
+
\ell\ddot\theta\hat e_\theta.
$$

The radial component of the acceleration is thus

$$
a_r
=
-\ell\dot\theta^2.
$$

The negative sign indicates that the acceleration points toward the pivot.

---

Consider the forces in the radial direction.

The tension force acts toward the pivot and therefore contributes

$$
-T\,\hat e_r.
$$

The radial component of gravity is

$$
mg\cos\theta\,\hat e_r.
$$

Applying Newton's second law in the radial direction gives

$$
-T+mg\cos\theta
=
m(-\ell\dot\theta^2).
$$

Multiplying by $-1$ yields

$$
T-mg\cos\theta
=
m\ell\dot\theta^2.
$$

Therefore,

$$
\boxed{
T
=
m\ell\dot\theta^2
+
mg\cos\theta
}
$$

The tension consists of two contributions:

$$
m\ell\dot\theta^2
$$

which provides the centripetal acceleration required to keep the mass moving along a circular path, and

$$
mg\cos\theta
$$

which arises from the radial component of gravity.

---

<strong>Lagrange Multiplier Approach</strong>

Now solve the same problem using polar coordinates $(r,\theta)$.

The pendulum constraint is

$$
g(r,\theta)
=
\ell-r
=
0.
$$

We intentionally do **not** use this constraint to eliminate $r$ at the start. Instead, we keep $r$ as a coordinate and enforce the constraint with a Lagrange multiplier.

The kinetic energy is

$$
T_{\rm kinetic}
=
\frac12 m
\left(
\dot r^2+r^2\dot\theta^2
\right),
$$

and the gravitational potential energy is

$$
U
=
-mgr\cos\theta,
$$

where $\theta$ is measured from the downward vertical.

Thus,

$$
L
=
T_{\rm kinetic}-U
=
\frac12 m
\left(
\dot r^2+r^2\dot\theta^2
\right)
+
mgr\cos\theta.
$$

The augmented Lagrangian is

$$
L'
=
L+\lambda g,
$$

so

$$
L'
=
\frac12 m
\left(
\dot r^2+r^2\dot\theta^2
\right)
+
mgr\cos\theta
+
\lambda(\ell-r).
$$

Apply Lagrange's equation to $r$:

$$
\frac{d}{dt}
\left(
\frac{\partial L'}{\partial \dot r}
\right)
-
\frac{\partial L'}{\partial r}
=
0.
$$

We have

$$
\frac{\partial L'}{\partial \dot r}
=
m\dot r,
$$

so

$$
\frac{d}{dt}
\left(
\frac{\partial L'}{\partial \dot r}
\right)
=
m\ddot r.
$$

Also,

$$
\frac{\partial L'}{\partial r}
=
mr\dot\theta^2
+
mg\cos\theta
-
\lambda.
$$

Therefore,

$$
m\ddot r
-
\left(
mr\dot\theta^2
+
mg\cos\theta
-
\lambda
\right)
=
0.
$$

Hence,

$$
m\ddot r
=
mr\dot\theta^2
+
mg\cos\theta
-
\lambda.
$$

Now apply the constraint only after deriving the equation of motion:

$$
r=\ell,
\qquad
\dot r=0,
\qquad
\ddot r=0.
$$

This gives

$$
0
=
m\ell\dot\theta^2
+
mg\cos\theta
-
\lambda.
$$

Therefore,

$$
\boxed{
\lambda
=
m\ell\dot\theta^2
+
mg\cos\theta
}.
$$

Now compute the generalized constraint force in the radial direction:

$$
Q_r^{(\rm constraint)}
=
\lambda
\frac{\partial g}{\partial r}.
$$

Since

$$
g(r,\theta)=\ell-r,
$$

we have

$$
\frac{\partial g}{\partial r}
=
-1.
$$

This result tell us that the constrained force is opposite directed of the positive $\hat e_r$ direction.

Therefore,

$$
Q_r^{(\rm constraint)}
=
-\lambda.
$$

Substituting the result for $\lambda$ gives

$$
\boxed{
Q_r^{(\rm constraint)}
=
m\ell\dot\theta^2
+
mg\cos\theta
}.
$$

This agrees exactly with the Newtonian result.

---

<strong>Physical Interpretation</strong>

The sign of the constraint force was not inserted by hand. It emerged from the choice

$$
g(r,\theta)=\ell-r=0.
$$

Because

$$
\frac{\partial g}{\partial r}=-1,
$$

the generalized constraint force points in the negative radial direction:

$$
Q_r^{(\rm constraint)}=-\lambda.
$$

Since negative radial direction means inward toward the pivot, this is precisely the direction of the rod tension.

---

<strong>Physical Interpretation</strong>

The tension is larger when the mass moves quickly because the rod must provide the centripetal acceleration.

The tension is also affected by the radial component of gravity. At the bottom of the swing, where $\theta=0$, the tension is

$$
T
=
m\ell\dot{\theta}^2+mg.
$$

At the top of a full circular motion, where $\theta=\pi$, the gravitational contribution changes sign:

$$
T
=
m\ell\dot{\theta}^2-mg.
$$

The Lagrange multiplier method reproduces the Newtonian result while also showing how the constraint force arises from the fixed-length condition.

</div>
</details>

## Problem 2 - Rolling Disk on an Inclined Plane

A solid disk of mass $m$ and radius $R$ rolls without slipping down an inclined plane of angle $\theta$.

This is a standard University Physics I problem that is often solved using Newton's laws and rotational dynamics. In this problem, solve it using the method of Lagrange multipliers.

1. Determine the equation of motion for the center of mass.

2. Determine the constraint force associated with the rolling-without-slipping condition.

3. Determine the angular acceleration of the disk.

Assume the moment of inertia of a solid disk about its center is

$$
I=\frac12 mR^2.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Let $s$ measure the distance traveled by the center of mass **down the incline**.

Therefore, the positive $\hat e_s$ direction points down the incline.

Let $\phi$ be the rotation angle of the disk, chosen so that rolling down the incline corresponds to increasing $\phi$.

With these coordinate choices, rolling without slipping requires

$$
s=R\phi.
$$

Thus, the constraint equation is

$$
g(s,\phi)
=
s-R\phi
=
0.
$$

The kinetic energy is

$$
T
=
\frac12 m\dot{s}^2
+
\frac12 I\dot{\phi}^2.
$$

Since $s$ increases as the disk moves downward, the gravitational potential energy decreases as $s$ increases:

$$
U
=
-mgs\sin\theta.
$$

Thus, the Lagrangian is

$$
L
=
T-U
=
\frac12 m\dot{s}^2
+
\frac12 I\dot{\phi}^2
+
mgs\sin\theta.
$$

The augmented Lagrangian is

$$
L'
=
L+\lambda g,
$$

so

$$
L'
=
\frac12 m\dot{s}^2
+
\frac12 I\dot{\phi}^2
+
mgs\sin\theta
+
\lambda(s-R\phi).
$$

Apply Lagrange's equation to $s$:

$$
\frac{d}{dt}
\left(
\frac{\partial L'}{\partial \dot{s}}
\right)
-
\frac{\partial L'}{\partial s}
=
0.
$$

Since

$$
\frac{\partial L'}{\partial \dot{s}}
=
m\dot{s},
$$

and

$$
\frac{\partial L'}{\partial s}
=
mg\sin\theta+\lambda,
$$

we obtain

$$
m\ddot{s}
-
(mg\sin\theta+\lambda)
=
0.
$$

Therefore,

$$
m\ddot{s}
=
mg\sin\theta+\lambda.
$$

Now apply Lagrange's equation to $\phi$:

$$
\frac{d}{dt}
\left(
\frac{\partial L'}{\partial \dot{\phi}}
\right)
-
\frac{\partial L'}{\partial \phi}
=
0.
$$

Since

$$
\frac{\partial L'}{\partial \dot{\phi}}
=
I\dot{\phi},
$$

and

$$
\frac{\partial L'}{\partial \phi}
=
-\lambda R,
$$

we obtain

$$
I\ddot{\phi}
-
(-\lambda R)
=
0.
$$

Thus,

$$
I\ddot{\phi}
=
-\lambda R.
$$

Now use the constraint only after obtaining the equations of motion.

Since

$$
s=R\phi,
$$

differentiating twice gives

$$
\ddot{s}
=
R\ddot{\phi}.
$$

Therefore,

$$
\ddot{\phi}
=
\frac{\ddot{s}}{R}.
$$

Substitute this into the rotational equation:

$$
I\left(\frac{\ddot{s}}{R}\right)
=
-\lambda R.
$$

Solving for $\lambda$ gives

$$
\lambda
=
-\frac{I}{R^2}\ddot{s}.
$$

Substitute this into the translational equation:

$$
m\ddot{s}
=
mg\sin\theta
-
\frac{I}{R^2}\ddot{s}.
$$

Rearranging,

$$
\left(
m+\frac{I}{R^2}
\right)
\ddot{s}
=
mg\sin\theta.
$$

Using

$$
I
=
\frac12 mR^2,
$$

we obtain

$$
\left(
m+\frac12 m
\right)
\ddot{s}
=
mg\sin\theta.
$$

Thus,

$$
\frac32 m\ddot{s}
=
mg\sin\theta.
$$

Therefore, the center-of-mass acceleration is

$$
\boxed{
\ddot{s}
=
\frac23 g\sin\theta
}.
$$

This is the equation of motion for the center of mass:

$$
\boxed{
s(t)
=
s_0+v_0t+\frac12
\left(
\frac23 g\sin\theta
\right)t^2
}
$$

if $\theta$ is constant and the disk starts with initial position $s_0$ and velocity $v_0$.

The angular acceleration follows from

$$
\ddot{\phi}
=
\frac{\ddot{s}}{R}.
$$

Therefore,

$$
\boxed{
\ddot{\phi}
=
\frac{2g\sin\theta}{3R}
}.
$$

---

<strong>Interpreting the Constraint Force</strong>

To interpret the multiplier physically, we return to the coordinate system.

The generalized coordinate $s$ was chosen to increase down the incline. Therefore, a positive generalized force $Q_s$ acts down the incline.

The constraint equation is

$$
g(s,\phi)
=
s-R\phi.
$$

Thus,

$$
\frac{\partial g}{\partial s}
=
1.
$$

The generalized constraint force associated with $s$ is

$$
Q_s^{(\mathrm{constraint})}
=
\lambda
\frac{\partial g}{\partial s}.
$$

Therefore,

$$
Q_s^{(\mathrm{constraint})}
=
\lambda.
$$

We found

$$
\lambda
=
-\frac{I}{R^2}\ddot{s}.
$$

Using

$$
I=\frac12 mR^2
$$

and

$$
\ddot{s}
=
\frac23 g\sin\theta,
$$

we obtain

$$
\lambda
=
-\frac12 m
\left(
\frac23 g\sin\theta
\right).
$$

Thus,

$$
\boxed{
\lambda
=
-\frac13 mg\sin\theta
}.
$$

Since

$$
Q_s^{(\mathrm{constraint})}
=
\lambda,
$$

we have

$$
Q_s^{(\mathrm{constraint})}
=
-\frac13 mg\sin\theta.
$$

The negative sign means the constraint force acts opposite the positive $s$ direction.

Since positive $s$ points down the incline, the constraint force points **up the incline**.

Therefore, the magnitude of the constraint force is

$$
\boxed{
f_s
=
\frac13 mg\sin\theta
}
$$

acting up the incline.

This force is the static friction force required to enforce the rolling-without-slipping condition.

---

<strong>Final Results</strong>

The center-of-mass acceleration is

$$
\boxed{
\ddot{s}
=
\frac23 g\sin\theta
}.
$$

The angular acceleration is

$$
\boxed{
\ddot{\phi}
=
\frac{2g\sin\theta}{3R}
}.
$$

The constraint force is static friction, with magnitude

$$
\boxed{
f_s
=
\frac13 mg\sin\theta
}
$$

acting up the incline.

## Problem 3 - Normal Force on a Rolling Disk

Consider the same configuration as in Problem 2: a solid disk of mass $m$ and radius $R$ rolling without slipping down an inclined plane of angle $\theta$.

Use the method of Lagrange multipliers to determine the **normal force** exerted by the incline on the disk.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Let $n$ measure the perpendicular distance of the disk's center of mass away from the surface of the incline, with positive $n$ pointing outward from the surface.

Since the disk has radius $R$, the constraint is

$$
g(n)=n-R=0.
$$

The relevant part of the Lagrangian is

$$
L_n
=
\frac12 m\dot n^2
-
mg n\cos\theta.
$$

The augmented Lagrangian is

$$
L_n'
=
\frac12 m\dot n^2
-
mg n\cos\theta
+
\lambda_n(n-R).
$$

Applying Lagrange's equation to $n$ gives

$$
m\ddot n
=
-mg\cos\theta+\lambda_n.
$$

The constraint requires

$$
n=R,
\qquad
\dot n=0,
\qquad
\ddot n=0.
$$

Therefore,

$$
0
=
-mg\cos\theta+\lambda_n,
$$

so

$$
\lambda_n
=
mg\cos\theta.
$$

Since

$$
\frac{\partial g}{\partial n}=1,
$$

the generalized constraint force is

$$
Q_n^{(\mathrm{constraint})}
=
\lambda_n
\frac{\partial g}{\partial n}
=
\lambda_n.
$$

Because positive $n$ points outward from the incline, this generalized constraint force is the normal force.

Thus,

$$
\boxed{
F_N
=
mg\cos\theta
}.
$$

## Problem 4 - Tension in a Standard Atwood Machine

A standard Atwood machine consists of two masses, $m_1$ and $m_2$, connected by a light inextensible rope passing over a pulley.

Assume the pulley is a frictionless disk that only redirects the rope and does not rotate with the rope. In other words, there is no friction between the rope and the disk, so the tension is the same on both sides of the rope.

Use the method of Lagrange multipliers to determine the tension in the rope.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Let $y_1$ and $y_2$ measure the downward positions of masses $m_1$ and $m_2$, respectively.

Since the rope length is fixed, the constraint is

$$
g(y_1,y_2)
=
\ell-y_1-y_2
=
0,
$$

where $\ell$ is a constant.

The kinetic energy is

$$
T_{\rm kinetic}
=
\frac12 m_1\dot y_1^2
+
\frac12 m_2\dot y_2^2.
$$

Since downward coordinate is positive, the gravitational potential energy is

$$
U
=
-m_1gy_1-m_2gy_2.
$$

Therefore, the Lagrangian is

$$
L
=
T_{\rm kinetic}-U
=
\frac12 m_1\dot y_1^2
+
\frac12 m_2\dot y_2^2
+
m_1gy_1
+
m_2gy_2.
$$

The augmented Lagrangian is

$$
L'
=
L+\lambda g,
$$

so

$$
L'
=
\frac12 m_1\dot y_1^2
+
\frac12 m_2\dot y_2^2
+
m_1gy_1
+
m_2gy_2
+
\lambda(\ell-y_1-y_2).
$$

Apply Lagrange's equation to $y_1$:

$$
\frac{d}{dt}
\left(
\frac{\partial L'}{\partial \dot y_1}
\right)
-
\frac{\partial L'}{\partial y_1}
=
0.
$$

Since

$$
\frac{\partial L'}{\partial \dot y_1}
=
m_1\dot y_1,
$$

and

$$
\frac{\partial L'}{\partial y_1}
=
m_1g-\lambda,
$$

we obtain

$$
m_1\ddot y_1
=
m_1g-\lambda.
$$

Apply Lagrange's equation to $y_2$:

$$
\frac{d}{dt}
\left(
\frac{\partial L'}{\partial \dot y_2}
\right)
-
\frac{\partial L'}{\partial y_2}
=
0.
$$

Since

$$
\frac{\partial L'}{\partial \dot y_2}
=
m_2\dot y_2,
$$

and

$$
\frac{\partial L'}{\partial y_2}
=
m_2g-\lambda,
$$

we obtain

$$
m_2\ddot y_2
=
m_2g-\lambda.
$$

Now use the constraint only after obtaining the equations of motion.

From

$$
\ell-y_1-y_2=0,
$$

differentiating twice gives

$$
\ddot y_1+\ddot y_2=0.
$$

Thus,

$$
\ddot y_2=-\ddot y_1.
$$

Let

$$
\ddot y_1=a,
\qquad
\ddot y_2=-a.
$$

Then the two equations become

$$
m_1a
=
m_1g-\lambda
$$

and

$$
-m_2a
=
m_2g-\lambda.
$$

Solving each for $\lambda$ gives

$$
\lambda
=
m_1g-m_1a
$$

and

$$
\lambda
=
m_2g+m_2a.
$$

Set these equal:

$$
m_1g-m_1a
=
m_2g+m_2a.
$$

Therefore,

$$
(m_1-m_2)g
=
(m_1+m_2)a.
$$

So,

$$
a
=
\frac{m_1-m_2}{m_1+m_2}g.
$$

Substitute this into

$$
\lambda
=
m_1g-m_1a.
$$

This gives

$$
\lambda
=
m_1g
-
m_1
\left(
\frac{m_1-m_2}{m_1+m_2}g
\right).
$$

Therefore,

$$
\lambda
=
m_1g
\left[
1
-
\frac{m_1-m_2}{m_1+m_2}
\right].
$$

Simplifying,

$$
\lambda
=
m_1g
\left[
\frac{m_1+m_2-m_1+m_2}{m_1+m_2}
\right].
$$

Thus,

$$
\lambda
=
m_1g
\left(
\frac{2m_2}{m_1+m_2}
\right).
$$

Therefore,

$$
\boxed{
\lambda
=
\frac{2m_1m_2}{m_1+m_2}g
}.
$$

Now interpret the multiplier.

The generalized constraint force on mass $m_1$ is

$$
Q_1^{(\mathrm{constraint})}
=
\lambda
\frac{\partial g}{\partial y_1}.
$$

Since

$$
g(y_1,y_2)=\ell-y_1-y_2,
$$

we have

$$
\frac{\partial g}{\partial y_1}
=
-1.
$$

Therefore,

$$
Q_1^{(\mathrm{constraint})}
=
-\lambda.
$$

Since positive $y_1$ points downward, a negative generalized force points upward. This is exactly the direction of the rope tension on mass $m_1$.

Thus,

$$
T_{\rm rope}
=
\lambda.
$$

Hence,

$$
\boxed{
T_{\rm rope}
=
\frac{2m_1m_2}{m_1+m_2}g
}.
$$

The same reasoning applies to mass $m_2$, so the rope tension is the same on both sides, as expected for a massless rope passing over a frictionless pulley.

## Problem 5 - Particle Sliding on a Smooth Hemisphere

A small particle of mass $m$ is placed at rest on the top of a fixed smooth hemisphere of radius $a$.

The particle slides under the influence of gravity and remains in contact with the hemisphere until the surface can no longer provide a normal force.

Using the method of Lagrange multipliers:

1. Determine the constraint force exerted by the hemisphere on the particle as a function of $\theta$.

2. Determine the angle $\theta$ at which the particle leaves the hemisphere.

Assume $\theta$ is measured from the upward vertical direction, so that $\theta=0$ corresponds to the top of the hemisphere.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Use polar coordinates $(r,\theta)$ with origin at the center of the hemisphere.

The constraint is

$$
g(r,\theta)=r-a=0.
$$

The kinetic energy is

$$
T=
\frac12 m(\dot r^2+r^2\dot\theta^2),
$$

and the potential energy is

$$
U=mg r\cos\theta.
$$

Thus,

$$
L=
\frac12 m(\dot r^2+r^2\dot\theta^2)
-
mg r\cos\theta.
$$

The augmented Lagrangian is

$$
L'
=
\frac12 m(\dot r^2+r^2\dot\theta^2)
-
mg r\cos\theta
+
\lambda(r-a).
$$

Apply Lagrange's equation to $r$:

$$
\frac{d}{dt}
\left(
\frac{\partial L'}{\partial \dot r}
\right)
-
\frac{\partial L'}{\partial r}
=
0.
$$

This gives

$$
m\ddot r
=
mr\dot\theta^2
-
mg\cos\theta
+
\lambda.
$$

Now apply the constraint only after deriving the equation:

$$
r=a,
\qquad
\dot r=0,
\qquad
\ddot r=0.
$$

Therefore,

$$
0
=
ma\dot\theta^2
-
mg\cos\theta
+
\lambda.
$$

Solving for the multiplier,

$$
\lambda
=
mg\cos\theta
-
ma\dot\theta^2.
$$

Since

$$
g(r,\theta)=r-a,
$$

we have

$$
\frac{\partial g}{\partial r}=1.
$$

Thus the radial generalized constraint force is

$$
Q_r^{(\mathrm{constraint})}
=
\lambda.
$$

Because positive $r$ points outward from the center, this is the normal force exerted by the hemisphere on the particle:

$$
F_N
=
\lambda.
$$

So,

$$
F_N
=
mg\cos\theta
-
ma\dot\theta^2.
$$

To eliminate $\dot\theta$, use energy conservation.

The particle starts from rest at the top, so the initial energy is

$$
E_i=mga.
$$

At angle $\theta$,

$$
E
=
\frac12 ma^2\dot\theta^2
+
mga\cos\theta.
$$

Therefore,

$$
mga
=
\frac12 ma^2\dot\theta^2
+
mga\cos\theta.
$$

Solving,

$$
\frac12 ma^2\dot\theta^2
=
mga(1-\cos\theta),
$$

so

$$
a\dot\theta^2
=
2g(1-\cos\theta).
$$

Substitute this into the normal force expression:

$$
F_N
=
mg\cos\theta
-
m\left[2g(1-\cos\theta)\right].
$$

Thus,

$$
F_N
=
mg\cos\theta
-
2mg
+
2mg\cos\theta.
$$

Therefore,

$$
\boxed{
F_N
=
mg(3\cos\theta-2)
}.
$$

The particle leaves the hemisphere when the normal force becomes zero:

$$
F_N=0.
$$

Thus,

$$
mg(3\cos\theta-2)=0.
$$

So,

$$
3\cos\theta=2,
$$

and therefore

$$
\boxed{
\cos\theta=\frac23
}.
$$

Hence, the particle leaves the hemisphere at

$$
\boxed{
\theta
=
\cos^{-1}\left(\frac23\right)
}.
$$
