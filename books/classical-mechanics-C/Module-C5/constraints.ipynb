## C5.2.1 Motivation: Why Do Constraints Matter?
We introduced constraints in Phase D, but let us re-enforce it here.

In many real-world physical systems, objects are not completely free to move in space—they are **subject to constraints** that limit their possible motion. For example:
- A **pendulum** must remain at a fixed distance from its pivot.
- A **block on an inclined plane** cannot move freely in all directions; it must stay on the surface.
- A **rolling wheel without slipping** must maintain a specific relationship between its velocity and rotation.

Understanding constraints is crucial because they:
- **Reduce the number of independent degrees of freedom** in a system.
- **Influence the forces acting on a system**, sometimes introducing non-trivial constraint forces.
- **Require systematic mathematical tools**, such as **Lagrange multipliers**, to enforce them properly.

In classical mechanics, we categorize constraints into different types based on their mathematical form and physical nature.

## C5.2.2 Types of Constraints
Constraints can be classified into several categories, each with unique characteristics. The primary classifications are:

### **Holonomic vs. Non-Holonomic Constraints**
#### **Holonomic Constraints**
A **holonomic constraint** is one that can be expressed as a function of the **generalized coordinates** and time:

$$
g(q_1, q_2, ..., q_n, t) = 0
$$

These constraints purely restrict the possible **positions** of a system but do not directly involve velocities or momenta.

##### **Examples:**
- **Bead on a Wire:** A bead moving along a circular wire of radius $ R $ satisfies the constraint:

$$
x^2 + y^2 - R^2 = 0
$$
  
- **Double Pendulum:** The two masses must be fixed at distances $ L_1 $ and $ L_2 $ from their pivots, leading to constraints like:

$$
x_1^2 + y_1^2 - L_1^2 = 0, \quad (x_2 - x_1)^2 + (y_2 - y_1)^2 - L_2^2 = 0
$$

Holonomic constraints **reduce the number of independent coordinates**, making it easier to analyze motion.

#### **Non-Holonomic Constraints**
A **non-holonomic constraint** cannot be written purely in terms of generalized coordinates—it involves **velocities** or even higher-order derivatives:

$$
g(q, \dot{q}, t) = 0
$$

These constraints limit possible **paths** in configuration space but do not necessarily reduce the number of independent coordinates.

##### **Examples:**
- **Rolling Without Slipping:** A rolling wheel must satisfy:

  $$
  v = R\omega
  $$
  
  which relates translational velocity $v $ of the center of mass to angular velocity $ \omega $.
- **Constraint on a Car:** A car moving in a plane cannot move sideways instantaneously, enforcing a constraint on its velocity.

Non-holonomic constraints are generally **more difficult to handle** because they do not directly reduce the number of independent coordinates.

---

### **Scleronomic vs. Rheonomic Constraints**
#### **Scleronomic Constraints**
A **scleronomic constraint** does **not** depend explicitly on time:

$$
g(q_1, q_2, ..., q_n) = 0
$$

These are time-independent constraints that impose fixed geometric relationships.

##### **Examples:**
- **Bead on a Fixed Circular Wire:** The equation $ x^2 + y^2 - R^2 = 0 $ remains constant over time.
- **Pendulum with Fixed Length:** The length constraint $ x^2 + y^2 = L^2 $ does not change.

#### **Rheonomic Constraints**
A **rheonomic constraint** **explicitly** depends on time:

$$
g(q_1, q_2, ..., q_n, t) = 0
$$

These constraints evolve over time, making the system’s motion dependent on external factors.

##### **Examples:**
- **Bead on a Moving Wire:** If the wire itself is rotating, the constraint may change over time.
- **Pendulum with a Moving Support:** If the pivot point of a pendulum moves over time, the constraint is time-dependent.

---

### **Equality vs. Inequality Constraints**
#### **Equality Constraints**
Most constraints in classical mechanics are **equality constraints**, meaning they strictly define an allowable surface in configuration or phase space:

$$
g(q_1, q_2, ..., q_n, t) = 0
$$

##### **Example:**
- A mass sliding on a sphere must always satisfy $ x^2 + y^2 + z^2 - R^2 = 0 $.

#### **Inequality Constraints**
Some constraints **only restrict the motion to a certain region** without fixing it to a specific surface:

$$
g(q_1, q_2, ..., q_n, t) \geq 0
$$

##### **Example:**
- A **block on or above a floor** must satisfy $ y \geq 0 $, meaning it cannot go below the surface but can move freely above it.

Inequality constraints require additional methods like **complementary conditions** in mechanics.

## C5.2.3 Effects of Constraints on Degrees of Freedom
The presence of constraints **reduces the number of independent coordinates** in a system. If a system initially has $ N $ generalized coordinates and is subject to $ C $ **holonomic** constraints, the number of **degrees of freedom (DOF)** is:

$$
\text{DOF} = N - C
$$

### **Example:**
- A **single particle in 3D** has $ N = 3 $.
- If it is constrained to move on a plane, it has **one holonomic constraint** $ z = 0 $, reducing its degrees of freedom to **2**.

For **non-holonomic** constraints, the number of degrees of freedom is not always straightforward, as the constraints involve velocities rather than just positions.

## C5.2.4 Constraints in Lagrangian and Hamiltonian Mechanics
### **Constraints in Lagrangian Mechanics**
- **Holonomic constraints** are naturally handled in **Lagrangian mechanics** by choosing appropriate **generalized coordinates**.
- **Non-holonomic constraints** require additional methods, such as **Lagrange multipliers** or non-holonomic equations of motion.

### **Constraints in Hamiltonian Mechanics**
- **Constraints appear as conditions on momenta**, rather than just coordinates.
- The **Poisson bracket formalism** is often used to enforce constraints dynamically.
- **Primary and secondary constraints** appear naturally in phase space.

**Example:**
- A bead constrained to a wire in **Lagrangian mechanics** requires fewer coordinates.
- The same system in **Hamiltonian mechanics** requires enforcing the momentum constraint using **Poisson brackets**.

## C5.2.4 Final Thoughts
1. **Constraints limit the possible motion of a system**, reducing its degrees of freedom.
2. **Holonomic constraints** restrict motion through coordinate relations, while **non-holonomic constraints** involve velocities.
3. **Scleronomic constraints** do not depend on time, while **rheonomic constraints** do.
4. **Lagrangian mechanics handles constraints naturally** through coordinate choices, while **Hamiltonian mechanics** enforces constraints in phase space.
5. **Lagrange multipliers provide a powerful tool** for handling constraints and computing constraint forces.

Understanding constraints is crucial in **mechanical engineering, robotics, and quantum mechanics**, where constrained systems appear frequently. Mastering these concepts allows for **systematic problem-solving in complex physical systems**.
