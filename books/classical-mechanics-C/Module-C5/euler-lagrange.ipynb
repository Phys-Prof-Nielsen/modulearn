# C5.4 Calculus of Variations: Finding the Best Path - Euler-Lagrange Equation

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.4.1 Motivation

In University Physics and previous modules, we often asked questions such as:

- What force causes a particular motion?
- What acceleration results from a given force?
- How can we derive equations of motion?

These questions are the natural language of **Newtonian mechanics**.

However, nature often appears to follow a different organizing principle. Instead of asking what force acts at every instant, we can ask a more global question:

> Out of all possible paths connecting two points in space and time, which path does nature choose?

This seemingly simple question leads to one of the most powerful ideas in all of theoretical physics.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Curiosity:</strong><br>

Many of the most important theories in modern physics are built upon this idea of optimization.

Examples include:

<ul>
<li>Light choosing the path of least travel time.</li>
<li>Particles following paths that make the action stationary.</li>
<li>General relativity describing spacetime through an extremal principle.</li>
<li>Quantum mechanics through path integrals involving all possible paths.</li>
</ul>

The same mathematical framework underlies all of these seemingly unrelated subjects.
</div>

To motivate the problem, consider a particle traveling from point A to point B.

There are infinitely many possible curves connecting the two points.

Some curves are long.

Some curves are short.

Some are smooth.

Some are highly irregular.

If we wish to determine the shortest possible path, we are no longer trying to find a number that minimizes a function. Instead, we are trying to determine the **entire function itself**.

This is the central idea behind the **calculus of variations**.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Calculus of Variations</strong>

Ordinary calculus finds the value of a variable that minimizes or maximizes a function.

Calculus of variations finds the function that minimizes or maximizes a quantity called a functional.

A functional maps an entire function into a single number.
</div>

The methods developed in this section will ultimately lead to the **Euler–Lagrange equation**, which forms the mathematical foundation of Lagrangian mechanics.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.4.2 Deriving the Euler–Lagrange Equation

The Euler–Lagrange equation is the central result of the calculus of variations. Rather than presenting the derivation as a sequence of algebraic steps, we will derive it ourselves by examining how a functional changes when the underlying function is varied slightly.

Consider the functional

$$
J[y]
=
\int_a^b F(y,y',x)\,dx.
$$

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
    
The quantity

$$
F(y,y',x)
$$

is called the **integrand** of the functional.

In more advanced areas of physics, particularly field theory, analogous quantities are often called **Lagrangian densities** because they are integrated over space or spacetime to construct an action.

The functional itself is

$$
J[y]
=
\int_a^b F(y,y',x)\,dx,
$$

which assigns a single number to the entire function $y(x)$.
</div>

Our goal is to determine what condition the function $y(x)$ must satisfy in order for $J[y]$ to be stationary.

The derivation will proceed through a sequence of guided activities.

---

<div style="background-color:#e8f5e9; border-left:6px solid #006633; padding:14px; border-radius:4px;">

<h3 style="margin-top:0;">Box Activity — Constructing a Nearby Function</h3>

Suppose that $y(x)$ is the function that makes the functional stationary.

To investigate whether this is true, we compare it with a nearby function

$$
\tilde y(x)
=
y(x)+\epsilon\eta(x),
$$

where:

- $\epsilon$ is a small parameter,
- $\eta(x)$ is an arbitrary differentiable function,
- $\eta(a)=\eta(b)=0$.

Answer the following:

1. Why must $\eta(a)=\eta(b)=0$?

2. Show that

$$
\tilde y'(x)
=
y'(x)+\epsilon\eta'(x).
$$

3. Explain the physical meaning of varying a function by a small amount.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Activity solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The endpoints are fixed, so any allowable variation must leave the endpoints unchanged. Therefore

$$
\eta(a)=\eta(b)=0.
$$

Differentiating gives

$$
\tilde y'
=
\frac{d}{dx}
\left(
y+\epsilon\eta
\right)
=
y'
+
\epsilon\eta'.
$$

The variation function $\eta(x)$ represents an arbitrary deformation of the original path while keeping the endpoints fixed.

</div>
</details>

</div>

<div style="background-color:#e8f5e9; border-left:6px solid #006633; padding:14px; border-radius:4px;">

<h3 style="margin-top:0;">Box Activity — Varying the Functional</h3>

In the previous activity, we introduced a nearby function

$$
\tilde y(x)
=
y(x)+\epsilon\eta(x),
$$

with

$$
\tilde y'(x)
=
y'(x)+\epsilon\eta'(x).
$$

We now substitute this perturbed function into the functional

$$
J[y]
=
\int_a^b F(y,y',x)\,dx.
$$

The functional evaluated along the nearby path becomes

$$
J(\epsilon)
=
\int_a^b
F\!\left(
y+\epsilon\eta,
\,
y'+\epsilon\eta',
\,
x
\right)
dx.
$$

Notice that the functional is now an ordinary function of the parameter $\epsilon$.

Answer the following:

1. Explain why $J(\epsilon)$ is now an ordinary function rather than a functional.

2. If $y(x)$ truly makes the functional stationary, what condition must be satisfied by $J(\epsilon)$ at $\epsilon=0$?

3. Compare this condition with the familiar condition for extrema in ordinary calculus.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Activity solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The family of functions

$$
\tilde y(x)
=
y(x)+\epsilon\eta(x)
$$

represents infinitely many possible curves near the original function $y(x)$.

Once the variation function $\eta(x)$ has been chosen, the only quantity that can vary is the parameter $\epsilon$. Consequently,

$$
J(\epsilon)
=
\int_a^b
F\!\left(
y+\epsilon\eta,
\,
y'+\epsilon\eta',
\,
x
\right)
dx
$$

is no longer a functional. It is now an ordinary function whose value depends only on the single variable $\epsilon$.

Notice that when

$$
\epsilon=0,
$$

the variation disappears and we recover the original function:

$$
\tilde y(x)
=
y(x).
$$

Thus, the point $\epsilon=0$ corresponds to the original path whose optimality we wish to test.

If $y(x)$ truly produces a minimum or maximum value of the functional, then small positive or negative changes in $\epsilon$ should not produce a first-order change in the value of $J$.

Graphically, we can imagine $J(\epsilon)$ having a stationary point at $\epsilon=0$:

$$
\epsilon < 0
\qquad
\longleftarrow
\qquad
\epsilon=0
\qquad
\longrightarrow
\qquad
\epsilon > 0
$$

where nearby variations produce only higher-order changes in the functional.

Therefore, just as extrema of an ordinary function satisfy

$$
\frac{df}{dx}=0,
$$

the stationary condition for the functional becomes

$$
\left.
\frac{dJ}{d\epsilon}
\right|_{\epsilon=0}
=
0.
$$

This is the variational analogue of the familiar condition from ordinary calculus.

Our next goal is therefore to compute

$$
\frac{dJ}{d\epsilon}
$$

and determine the mathematical condition that must be satisfied for this derivative to vanish for every allowable variation $\eta(x)$.

</div>
</details>
</div>

<div style="background-color:#e8f5e9; border-left:6px solid #006633; padding:14px; border-radius:4px;">

<h3 style="margin-top:0;">Box Activity — Computing the First Variation</h3>

We have rewritten the functional as an ordinary function of the variation parameter $\epsilon$:

$$
J(\epsilon)
=
\int_a^b
F\!\left(
y+\epsilon\eta,
\,
y'+\epsilon\eta',
\,
x
\right)
dx.
$$

For the original function $y(x)$ to make the functional stationary, we require

$$
\left.
\frac{dJ}{d\epsilon}
\right|_{\epsilon=0}
=
0.
$$

Now compute this derivative.

Use the chain rule to show that

$$
\frac{dJ}{d\epsilon}
=
\int_a^b
\left[
\frac{\partial F}{\partial y}\eta
+
\frac{\partial F}{\partial y'}\eta'
\right]dx.
$$

Then evaluate the result at $\epsilon=0$.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Activity solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Since

$$
J(\epsilon)
=
\int_a^b
F\!\left(
y+\epsilon\eta,
\,
y'+\epsilon\eta',
\,
x
\right)
dx,
$$

we differentiate with respect to $\epsilon$.

The function $F$ depends on $\epsilon$ through two quantities:

$$
y+\epsilon\eta
$$

and

$$
y'+\epsilon\eta'.
$$

Using the chain rule,

$$
\frac{dF}{d\epsilon}
=
\frac{\partial F}{\partial y}
\frac{d}{d\epsilon}
\left(
y+\epsilon\eta
\right)
+
\frac{\partial F}{\partial y'}
\frac{d}{d\epsilon}
\left(
y'+\epsilon\eta'
\right).
$$

Since $y$, $y'$, $\eta$, and $\eta'$ do not depend on $\epsilon$,

$$
\frac{d}{d\epsilon}
\left(
y+\epsilon\eta
\right)
=
\eta
$$

and

$$
\frac{d}{d\epsilon}
\left(
y'+\epsilon\eta'
\right)
=
\eta'.
$$

Therefore,

$$
\frac{dF}{d\epsilon}
=
\frac{\partial F}{\partial y}\eta
+
\frac{\partial F}{\partial y'}\eta'.
$$

Thus,

$$
\frac{dJ}{d\epsilon}
=
\int_a^b
\left[
\frac{\partial F}{\partial y}\eta
+
\frac{\partial F}{\partial y'}\eta'
\right]dx.
$$

At the stationary path, we evaluate this expression at $\epsilon=0$:

$$
\left.
\frac{dJ}{d\epsilon}
\right|_{\epsilon=0}
=
\int_a^b
\left[
\frac{\partial F}{\partial y}\eta
+
\frac{\partial F}{\partial y'}\eta'
\right]dx
=
0.
$$

This expression is called the **first variation** of the functional.

The remaining difficulty is that the integral contains both $\eta$ and $\eta'$. To obtain a condition on $y(x)$ itself, we need to rewrite the integral so that only $\eta$ appears. That is the purpose of the next step.

</div>
</details>

</div>

<div style="background-color:#e8f5e9; border-left:6px solid #006633; padding:14px; border-radius:4px;">

<h3 style="margin-top:0;">Box Activity — Eliminating the Derivative of the Variation</h3>

In the previous activity, we obtained

$$
\left.
\frac{dJ}{d\epsilon}
\right|_{\epsilon=0}
=
\int_a^b
\left[
\frac{\partial F}{\partial y}\eta
+
\frac{\partial F}{\partial y'}\eta'
\right]dx
=
0.
$$

Our goal is to obtain an expression involving only the variation function $\eta(x)$ and not its derivative $\eta'(x)$.

To accomplish this, we apply integration by parts to the term

$$
\int_a^b
\frac{\partial F}{\partial y'}
\eta'
\,dx.
$$

Recall that

$$
\int_a^b u\,dv
=
\left.uv\right|_a^b
-
\int_a^b v\,du.
$$

Answer the following:

1. Let

$$
u=\frac{\partial F}{\partial y'},
\qquad
dv=\eta' dx.
$$

Determine $du$ and $v$.

2. Apply integration by parts to

$$
\int_a^b
\frac{\partial F}{\partial y'}
\eta'
\,dx.
$$

3. Use the endpoint conditions

$$
\eta(a)=\eta(b)=0
$$

to simplify the boundary term.

4. Substitute the result back into the expression for

$$
\left.\frac{dJ}{d\epsilon}\right|_{\epsilon=0}.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Activity solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Choosing

$$
u=\frac{\partial F}{\partial y'},
\qquad
dv=\eta' dx,
$$

gives

$$
v=\eta,
$$

and

$$
du
=
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
dx.
$$

Applying integration by parts,

$$
\int_a^b
\frac{\partial F}{\partial y'}
\eta'
\,dx
=
\left[
\frac{\partial F}{\partial y'}
\eta
\right]_a^b
-
\int_a^b
\eta
\,
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
dx.
$$

Because the endpoints are fixed,

$$
\eta(a)=\eta(b)=0,
$$

so the boundary term vanishes:

$$
\left[
\frac{\partial F}{\partial y'}
\eta
\right]_a^b
=
0.
$$

Therefore,

$$
\int_a^b
\frac{\partial F}{\partial y'}
\eta'
\,dx
=
-
\int_a^b
\eta
\,
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
dx.
$$

Substituting into the first variation yields

$$
\left.
\frac{dJ}{d\epsilon}
\right|_{\epsilon=0}
=
\int_a^b
\left[
\frac{\partial F}{\partial y}
-
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
\right]
\eta
\,dx
=
0.
$$

Notice that the derivative of the variation function has disappeared. The integral now contains only the arbitrary function $\eta(x)$.

This expression is only one step away from the Euler–Lagrange equation.

</div>
</details>

</div>

## C5.4.3 Summary of the Euler–Lagrange Equation

The calculus of variations provides a systematic method for determining the function that makes a functional stationary.

Starting from the functional

$$
J[y]
=
\int_a^b F(y,y',x)\,dx,
$$

we introduced a small variation of the function,

$$
\tilde y(x)
=
y(x)+\epsilon\eta(x),
$$

and required that the functional remain stationary under all allowable variations.

This procedure led to one of the most important equations in mathematical physics.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Euler–Lagrange Equation</strong>

A function $y(x)$ produces a stationary value of the functional

$$
J[y]
=
\int_a^b F(y,y',x)\,dx
$$

if it satisfies

$$
\boxed{
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
-
\frac{\partial F}{\partial y}
=
0
}
$$

This equation is called the Euler–Lagrange equation.
</div>

<br>

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Stationary Functional</strong>

A functional is said to be stationary when small variations of the underlying function produce no first-order change in the value of the functional.

Mathematically,

$$
\left.
\frac{dJ}{d\epsilon}
\right|_{\epsilon=0}
=
0.
$$

A stationary value may correspond to a minimum, maximum, or saddle point.
</div>

<br>

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Fundamental Idea of the Calculus of Variations</strong>

Ordinary calculus seeks the value of a variable that optimizes a function.

Calculus of variations seeks the function that optimizes a functional.

The Euler–Lagrange equation plays a role analogous to

$$
\frac{df}{dx}=0
$$

in ordinary calculus.
</div>

<br>

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Looking Ahead</strong>

The Euler–Lagrange equation appears throughout physics, engineering, mathematics, and optimization theory.

In the next section, we will apply the Euler–Lagrange equation to a special functional known as the <em>action</em>. This leads directly to the Lagrange equations of motion used throughout classical mechanics.
</div>
