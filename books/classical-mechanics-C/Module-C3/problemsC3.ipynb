## Problem 1 — Gyroscope with a Horizontal Axle

A bicycle-wheel gyroscope of mass $m=2.0~\mathrm{kg}$ and rim radius $R=0.20~\mathrm{m}$ is mounted on a horizontal axle and spins with angular speed

$$
\omega=160~\mathrm{rad/s}
$$

about its symmetry axis.

One end of the axle is supported at a fixed, frictionless pivot. The wheel+axle center of mass is a distance

$$
\ell=0.15~\mathrm{m}
$$

horizontally from the pivot.

Take

$$
g=9.81~\mathrm{m/s^2}.
$$

1. Treat the wheel as a thin hoop so that its spin moment of inertia about the axle is

$$
I=mR^2.
$$

2. Assuming steady precession about the vertical, find the precession rate

$$
\Omega
=
\frac{mg\ell}{I\omega}.
$$

3. Compute the rotational kinetic energy

$$
T
=
\frac{1}{2}I\omega^2.
$$

4. Briefly explain why precession occurs even though the axle is horizontal.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The moment of inertia of a thin hoop about its symmetry axis is

$$
I=mR^2.
$$

Substitute the given values:

$$
I
=
(2.0~\mathrm{kg})(0.20~\mathrm{m})^2
$$

$$
I
=
0.080~\mathrm{kg\,m^2}.
$$

The steady precession rate is

$$
\Omega
=
\frac{mg\ell}{I\omega}.
$$

Substitute:

$$
\Omega
=
\frac{(2.0)(9.81)(0.15)}
{(0.080)(160)}
$$

$$
\Omega
=
0.230~\mathrm{rad/s}.
$$

So the gyroscope precesses about the vertical at approximately

$$
\boxed{\Omega \approx 0.23~\mathrm{rad/s}}.
$$

The rotational kinetic energy is

$$
T
=
\frac{1}{2}I\omega^2.
$$

Substitute:

$$
T
=
\frac{1}{2}(0.080)(160)^2
$$

$$
T
=
1024~\mathrm{J}.
$$

Thus,

$$
\boxed{T=1.02\times 10^3~\mathrm{J}}.
$$

The axle is horizontal, so gravity produces a torque about the pivot. The torque is

$$
\tau
=
mg\ell.
$$

This torque does not mainly change the magnitude of the spin angular momentum. Instead, it changes the direction of the angular momentum vector.

Since torque satisfies

$$
\vec{\tau}
=
\frac{d\vec{L}}{dt},
$$

the angular momentum vector slowly turns about the vertical direction. This turning motion is the precession of the gyroscope.

</div>
</details>

## Problem 2 — Pivot Reaction Forces During Steady Precession

Use the same wheel and parameters from Problem 1. In steady precession at the rate $\Omega$ found previously, the wheel’s center of mass moves in a horizontal circle of radius

$$
\rho=\ell.
$$

1. Find the horizontal component $F_{\text{H}}$ of the pivot reaction needed to provide the centripetal acceleration.

2. Find the vertical component $F_{\text{V}}$ of the pivot reaction.

3. Compute the magnitude of the pivot reaction $F_{\text{pivot}}$ and its tilt angle $\phi$ above the horizontal:

$$
F_{\text{pivot}}
=
\sqrt{F_{\text{H}}^2+F_{\text{V}}^2},
\qquad
\phi
=
\tan^{-1}\left(\frac{F_{\text{V}}}{F_{\text{H}}}\right).
$$

Use

$$
\Omega=0.230~\mathrm{rad/s},
\qquad
\ell=0.15~\mathrm{m}.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The center of mass moves in a horizontal circle of radius

$$
\rho=\ell=0.15~\mathrm{m}.
$$

The required centripetal acceleration is

$$
a_c
=
\Omega^2\rho.
$$

Using $\Omega=0.230~\mathrm{rad/s}$,

$$
a_c
=
(0.230)^2(0.15)
$$

$$
a_c
=
7.94\times 10^{-3}~\mathrm{m/s^2}.
$$

The horizontal pivot force provides this centripetal acceleration:

$$
F_{\text{H}}
=
ma_c
=
m\Omega^2\ell.
$$

Substitute:

$$
F_{\text{H}}
=
(2.0)(0.230)^2(0.15)
$$

$$
F_{\text{H}}
=
1.59\times 10^{-2}~\mathrm{N}.
$$

So,

$$
\boxed{
F_{\text{H}}
\approx
1.6\times 10^{-2}~\mathrm{N}
}
$$

The vertical acceleration of the center of mass is zero during steady horizontal precession. Therefore, the vertical pivot force must balance the weight:

$$
F_{\text{V}}
=
mg.
$$

Substitute:

$$
F_{\text{V}}
=
(2.0)(9.81)
$$

$$
F_{\text{V}}
=
19.62~\mathrm{N}.
$$

So,

$$
\boxed{
F_{\text{V}}
=
19.6~\mathrm{N}
}
$$

The magnitude of the pivot reaction is

$$
F_{\text{pivot}}
=
\sqrt{F_{\text{H}}^2+F_{\text{V}}^2}.
$$

Substitute:

$$
F_{\text{pivot}}
=
\sqrt{(1.59\times 10^{-2})^2+(19.62)^2}
$$

$$
F_{\text{pivot}}
\approx
19.62~\mathrm{N}.
$$

Thus,

$$
\boxed{
F_{\text{pivot}}
\approx
19.6~\mathrm{N}
}
$$

The tilt angle above the horizontal is

$$
\phi
=
\tan^{-1}\left(\frac{F_{\text{V}}}{F_{\text{H}}}\right).
$$

Substitute:

$$
\phi
=
\tan^{-1}\left(
\frac{19.62}{1.59\times 10^{-2}}
\right)
$$

$$
\phi
\approx
89.95^\circ.
$$

Therefore,

$$
\boxed{
\phi
\approx
89.95^\circ
}
$$

The pivot reaction is almost entirely vertical because the centripetal force required for the slow precession is very small compared with the weight.

</div>
</details>

## Problem 3 — Quasi-Steady Precession During Spin-Down

The same horizontal-axle gyroscope from Problem 1 is spinning with initial angular speed

$$
\omega_0=160~\mathrm{rad/s}.
$$

Due to friction, the spin angular speed slowly decreases over time. The wheel continues to precess smoothly about the vertical, maintaining a constant tilt angle $\theta$, so its geometry of motion remains approximately steady. However, the precession rate $\Omega(t)$ changes as the spin decreases.

This situation is called **quasi-steady precession**, where the instantaneous precession rate is

$$
\Omega(t)
=
\frac{mg\ell}{I\omega(t)}.
$$

1. If the spin drops from

$$
\omega_0=160~\mathrm{rad/s}
$$

to

$$
\omega_1=80~\mathrm{rad/s},
$$

compute the corresponding change in precession rate.

2. If the spin decreases linearly in time,

$$
\omega(t)
=
\omega_0-\alpha t,
$$

with constant $\alpha>0$, show that the total precession angle $\Phi$ accumulated during this interval is

$$
\Phi
=
\frac{mg\ell}{I\alpha}
\ln\left(\frac{\omega_0}{\omega_1}\right),
$$

and evaluate $\Phi$ in degrees for

$$
\alpha=0.40~\mathrm{rad/s^2},
\qquad
\omega_0=160~\mathrm{rad/s},
\qquad
\omega_1=80~\mathrm{rad/s}.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

From Problem 1, the moment of inertia of the thin hoop is

$$
I
=
mR^2
=
0.080~\mathrm{kg\,m^2}.
$$

The precession rate is

$$
\Omega
=
\frac{mg\ell}{I\omega}.
$$

For $\omega_0=160~\mathrm{rad/s}$,

$$
\Omega_0
=
\frac{(2.0)(9.81)(0.15)}
{(0.080)(160)}
$$

$$
\Omega_0
=
0.230~\mathrm{rad/s}.
$$

For $\omega_1=80~\mathrm{rad/s}$,

$$
\Omega_1
=
\frac{(2.0)(9.81)(0.15)}
{(0.080)(80)}
$$

$$
\Omega_1
=
0.460~\mathrm{rad/s}.
$$

Therefore, when the spin angular speed is cut in half, the precession rate doubles:

$$
\boxed{
\Omega_0=0.230~\mathrm{rad/s}
}
$$

$$
\boxed{
\Omega_1=0.460~\mathrm{rad/s}
}
$$

and

$$
\boxed{
\Delta\Omega
=
\Omega_1-\Omega_0
=
0.230~\mathrm{rad/s}
}
$$

Now suppose

$$
\omega(t)=\omega_0-\alpha t.
$$

The total precession angle is

$$
\Phi
=
\int \Omega(t)\,dt.
$$

Using the quasi-steady precession rate,

$$
\Phi
=
\int_0^{t_1}
\frac{mg\ell}{I\omega(t)}\,dt.
$$

Substitute $\omega(t)=\omega_0-\alpha t$:

$$
\Phi
=
\frac{mg\ell}{I}
\int_0^{t_1}
\frac{dt}{\omega_0-\alpha t}.
$$

Let

$$
u=\omega_0-\alpha t.
$$

Then

$$
du=-\alpha dt,
\qquad
dt=-\frac{du}{\alpha}.
$$

When $t=0$,

$$
u=\omega_0.
$$

When $t=t_1$,

$$
u=\omega_1.
$$

Therefore,

$$
\Phi
=
\frac{mg\ell}{I}
\int_{\omega_0}^{\omega_1}
\frac{-du}{\alpha u}.
$$

So,

$$
\Phi
=
\frac{mg\ell}{I\alpha}
\int_{\omega_1}^{\omega_0}
\frac{du}{u}.
$$

This gives

$$
\Phi
=
\frac{mg\ell}{I\alpha}
\ln\left(\frac{\omega_0}{\omega_1}\right).
$$

Now evaluate this expression.

First compute

$$
mg\ell
=
(2.0)(9.81)(0.15)
=
2.943~\mathrm{N\,m}.
$$

Then

$$
I\alpha
=
(0.080)(0.40)
=
0.032.
$$

Thus,

$$
\Phi
=
\frac{2.943}{0.032}
\ln\left(\frac{160}{80}\right).
$$

Since

$$
\ln(2)=0.693,
$$

we get

$$
\Phi
=
91.97(0.693)
=
63.75~\mathrm{rad}.
$$

Convert to degrees:

$$
\Phi_{\deg}
=
63.75\left(\frac{180^\circ}{\pi}\right)
$$

$$
\Phi_{\deg}
\approx
3653^\circ.
$$

Therefore,

$$
\boxed{
\Phi
\approx
63.8~\mathrm{rad}
}
$$

or

$$
\boxed{
\Phi
\approx
3.65\times 10^3{}^\circ.
}
$$

This corresponds to about

$$
\frac{3653^\circ}{360^\circ}
\approx
10.1
$$

full revolutions of precession during the spin-down interval.

</div>
</details>
