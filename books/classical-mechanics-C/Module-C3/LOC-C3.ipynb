## Why this module matters

In earlier studies of rotational mechanics, we often restricted ourselves to rigid bodies rotating about a fixed axis. This simplified framework allowed us to develop intuition for angular displacement, angular velocity, torque, and rotational energy. However, many real physical systems are not constrained to rotate about a single unchanging axis.

Examples include:

- spinning tops  
- gyroscopes  
- rotating spacecraft  
- tumbling satellites  
- rotating molecules  
- free-falling cats reorienting mid-air  
- astrophysical rotating bodies  

In these systems, the axis of rotation may change continuously in time, and rotational motion must be treated fully in three dimensions.

To describe this richer behavior, we introduce the angular velocity vector:

$$
\vec{\omega}
$$

which characterizes both the magnitude and direction of rotational motion.

We also develop the moment of inertia tensor, which generalizes the scalar moment of inertia used in fixed-axis rotation. The inertia tensor describes how mass is distributed relative to all possible rotational axes and allows us to analyze arbitrary rigid-body motion.

This module introduces:

- angular velocity as a vector quantity  
- angular momentum in three dimensions  
- torque and rotational dynamics  
- gyroscopic motion and precession  
- conservation of angular momentum  
- the moment of inertia tensor  

These ideas form the foundation for later studies in:

- rigid body dynamics  
- spacecraft attitude control  
- gyroscopic stabilization  
- continuum mechanics  
- advanced classical mechanics  
- rotational motion in quantum systems

By the end of this module, you will be able to:

---

### A. General Rotation and Angular Velocity  
*Introduces the foundations of three-dimensional rotational motion.*

- **A1.** Explain why fixed-axis rotation is insufficient for describing many real rotating systems.

- **A2.** Define the angular velocity vector and interpret its magnitude and direction physically.

- **A3.** Relate angular velocity to the linear velocity of a point on a rigid body:

$$
\vec{v}
=
\vec{\omega}\times\vec{r}
$$

- **A4.** Apply vector methods to analyze rotational motion about arbitrary axes.

- **A5.** Distinguish between scalar rotational quantities and full vector descriptions of rotation.

---

### B. Angular Momentum and Torque in Three Dimensions  
*Develops the vector formulation of rotational dynamics.*

- **B1.** Calculate angular momentum for particles and rigid-body systems:

$$
\vec{L}
=
\vec{r}\times\vec{p}
$$

- **B2.** Relate torque to the rate of change of angular momentum:

$$
\vec{\tau}
=
\frac{d\vec{L}}{dt}
$$

- **B3.** Analyze rotational motion when angular momentum and angular velocity are not parallel.

- **B4.** Apply vector cross products correctly in rotational dynamics problems.

- **B5.** Interpret rotational motion geometrically in three dimensions.

---

### C. Gyroscopic Motion and Conservation Laws  
*Applies rotational dynamics to physically important systems.*

- **C1.** Explain the stability of gyroscopes using angular momentum conservation.

- **C2.** Describe the origin of gyroscopic precession under external torque.

- **C3.** Calculate precession rates for simple gyroscope models.

- **C4.** Apply conservation of angular momentum to isolated systems.

- **C5.** Analyze torque-free rotational motion qualitatively and quantitatively.

---

### D. The Moment of Inertia Tensor  
*Introduces the tensor description of rigid-body rotation.*

- **D1.** Define the moment of inertia tensor as a generalization of scalar moment of inertia.

- **D2.** Interpret diagonal and off-diagonal components of the inertia tensor physically.

- **D3.** Calculate angular momentum using the tensor relation:

$$
\vec{L}
=
\mathbf{I}\vec{\omega}
$$

- **D4.** Analyze rotational motion for arbitrary mass distributions.

- **D5.** Explain why angular momentum may not align with angular velocity.

---

### E. Computational and Analytical Skills  
*Develops mathematical and computational approaches to rotational dynamics.*

- **E1.** Solve rigid-body rotational problems using vector methods.

- **E2.** Compute inertia tensors for discrete and continuous mass distributions.

- **E3.** Implement computational models of rotational motion using Python.

- **E4.** Generate visualizations of:
  - angular velocity vectors  
  - angular momentum vectors  
  - gyroscopic precession  
  - rigid-body orientation changes  

- **E5.** Compare rotational motion about different axes and mass distributions.

---

### Performance Criteria

- Angular velocity and angular momentum are treated correctly as vector quantities.  
- Torque and rotational dynamics are analyzed consistently in three dimensions.  
- Conservation of angular momentum is applied appropriately to isolated systems.  
- The inertia tensor is interpreted and used correctly in rigid-body calculations.  
- Computational and analytical methods are used effectively to model rigid-body dynamics.
