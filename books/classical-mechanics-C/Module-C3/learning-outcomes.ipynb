## C2.1.1 Motivation

In the previous phase of our journey through classical mechanics, we reviewed the rotational motion of rigid bodies constrained to rotate about a fixed axis. This allowed us to develop an intuition for angular displacement, velocity, and acceleration in a simplified context, and introduced the concept of moment of inertia for rotation about a principal axis. However, many real-world systems—spinning tops, satellites, or tumbling books—are not confined to such simplifications. Their axes of rotation may shift, drift, or precess in space. To understand these richer motions, we must now lift the constraint of a fixed axis and treat rotation in its full generality.

In this section, we introduce the angular velocity vector, a geometric object that captures the true three-dimensional nature of a rigid body's rotation at any instant. Alongside this, we develop the moment of inertia tensor, which replaces the scalar moment of inertia and encodes how mass is distributed relative to all possible axes through the center of mass. These tools will equip us to analyze the complex and often unintuitive behavior of rotating bodies in free space. For example, they allow us to understand how a gyroscope resists changes to its orientation or why a falling cat can twist mid-air and still land on its feet. While Euler's angles offer a valuable framework for describing changing orientations, we will defer them to a later phase. For now, our focus is on the physics of rotation itself, grounded in vectors and tensors.

## C2.1.2 Learning Outcomes

### 1. Understand the Need for General Rotation

- Recognize the limitations of fixed-axis rotation and why it's insufficient for many real-world scenarios.
- Appreciate the importance of analyzing rotation in three dimensions for systems like gyroscopes, satellites, and tumbling bodies.
- Identify the need for introducing the angular velocity vector and the moment of inertia tensor.

---

### 2. Describe Angular Velocity as a Vector

- Define the angular velocity vector and explain its direction and magnitude.
- Relate the angular velocity vector to the linear velocity of a point on a rotating object using the cross product:

  $$
  \vec{v} = \vec{\omega} \times \vec{r}
  $$
    
- Solve problems involving rotation about arbitrary axes using vector representations.

---

### 3. Apply Angular Momentum and Torque to Point Systems

- Calculate angular momentum for a particle system:

  $$
  \vec{l} = \vec{r} \times \vec{p}
  $$
  
- Relate torque to the rate of change of angular momentum:

  $$
  \vec{\tau} = \frac{d\vec{l}}{dt}
  $$
  
- Analyze simple configurations (e.g., two-point systems) to build intuition.

---

### 4. Analyze Basic Gyroscope Dynamics

- Explain the role of angular momentum in the stability of gyroscopes.
- Describe how an external torque causes gyroscopic precession.
- Calculate precession rates for basic gyroscope models.

---

### 5. Use Conservation of Angular Momentum in 3D

- Identify conditions under which angular momentum is conserved.
- Apply conservation principles to analyze the motion of isolated systems.
- Solve problems involving conservation of angular momentum.

---

### 6. Use the Moment of Inertia Tensor to Describe General Rigid Body Rotation

- Define the moment of inertia tensor as a generalization of scalar moment of inertia.
- Calculate angular momentum of a rigid body using the moment of inertia tensor:

  $$
  \vec{L} = \mathbf{I}\vec{\omega}
  $$
  
- Analyze motion for bodies with arbitrary mass distributions and orientation.

---

### 7. Identify and Use Principal Axes of Rotation

- Define principal axes and explain their significance in simplifying rotation problems.
- Diagonalize the inertia tensor to find principal moments of inertia.
- Apply principal axes to reduce the complexity of angular momentum and torque equations.

---

### Overall Learning Outcomes

By the end of this tier, you should be able to:

- Represent and interpret the angular velocity of a rigid body as a vector.
- Calculate angular momentum and torque in general three-dimensional motion.
- Explain and predict the behavior of gyroscopes using physical principles.
- Apply conservation of angular momentum to analyze complex motion.
- Use the moment of inertia tensor and identify principal axes to simplify and solve problems involving general rotation.
