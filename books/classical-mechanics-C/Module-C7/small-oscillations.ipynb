# C7.1 Small Oscillations

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.1.1 Motivation

Many physical systems spend most of their time near equilibrium. A pendulum hangs nearly vertical, a spring oscillates near its rest length, and atoms in a solid vibrate about their equilibrium positions. When the displacement from equilibrium is small, the system’s behavior simplifies dramatically — complex forces and potentials can be replaced by simple linear approximations. This leads to **simple harmonic motion**, one of the most fundamental and widely applicable results in all of physics.

The power of studying small oscillations lies in this simplification. By expanding the potential energy around an equilibrium point, we find that almost any stable system behaves like a harmonic oscillator when slightly perturbed. This makes the concept of small oscillations not only a specific case of motion but also a **universal approximation technique** for understanding vibrations, stability, and wave phenomena across disciplines — from molecules and bridges to orbits and quantum systems.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.1.2 The Idea of Small Oscillations

Consider a one-dimensional system with coordinate $x$ and potential energy $U(x)$. The equation of motion follows from Newton’s second law:

$$
m\ddot{x} = -\frac{dU}{dx}.
$$

Let $x_0$ denote the equilibrium position, where the net force vanishes:

$$
\left.\frac{dU}{dx}\right|_{x_0} = 0.
$$

We now consider a small displacement $\xi = x - x_0$ from equilibrium. The potential energy can be expanded in a Taylor series about $x_0$:

$$
U(x) = U(x_0) + \frac{1}{2}U''(x_0)\xi^2 + \frac{1}{6}U'''(x_0)\xi^3 + \cdots
$$

Since the first derivative vanishes at equilibrium, the linear term disappears. For *small oscillations*, we keep only the quadratic term because higher-order terms are negligible when $\xi$ is small. Thus,

$$
U(x) \approx U(x_0) + \frac{1}{2}k\xi^2,
$$

where $k = U''(x_0)$ plays the role of an *effective spring constant*.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.1.3 Equation of Motion

Substituting this approximation into the force equation gives:

$$
m\ddot{\xi} = -\frac{dU}{d\xi} = -k\xi.
$$

Rearranging,

$$
\ddot{\xi} + \omega^2 \xi = 0,
\quad \text{where} \quad \omega = \sqrt{\frac{k}{m}}.
$$

This is the defining equation of **simple harmonic motion (SHM)**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.1.4 General Solution

The general solution to this linear differential equation is

$$
\xi(t) = A\cos(\omega t + \phi),
$$

where:
- $A$ is the amplitude of oscillation,
- $\phi$ is the phase constant determined by initial conditions.

The total energy of the oscillator is constant:

$$
E = \frac{1}{2}m\dot{\xi}^2 + \frac{1}{2}k\xi^2 = \frac{1}{2}kA^2.
$$

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.1.5 Physical Interpretation

At equilibrium, the potential energy curve has a local minimum. The curvature of the potential, represented by $U''(x_0)$, determines how “stiff” the restoring force is — and therefore the natural frequency $\omega$. A deeper or steeper potential well results in faster oscillations.

Thus, small oscillations serve as a **universal model** for motion near stable equilibria. No matter how complicated a system’s forces may be, near equilibrium the motion is governed by the same harmonic form.

## C7.1.6 Example

### The Simple Pendulum and the Small-Angle Approximation

A classic example of small oscillations is the simple pendulum — a mass $m$ suspended from a light, rigid rod or string of length $L$, swinging under the influence of gravity.

#### Setup and Equation of Motion

The position of the pendulum can be described by the angular displacement $\theta$ from the vertical.  
The restoring torque about the pivot is due to gravity:

$$
\tau = -mgL \sin\theta.
$$

Applying Newton’s second law for rotational motion:

$$
I \ddot{\theta} = \tau,
$$

where $I = mL^2$ is the moment of inertia of the mass about the pivot. Thus,

$$
mL^2 \ddot{\theta} = -mgL \sin\theta,
$$

or

$$
\ddot{\theta} + \frac{g}{L}\sin\theta = 0.
$$

This equation is **nonlinear** because of the sine term. Exact solutions exist but involve elliptic integrals. To obtain a simple, general form, we consider *small oscillations*.

#### Small-Angle Approximation

For small angular displacements ($\theta \ll 1$ rad), we can use the linear approximation:

$$
\sin\theta \approx \theta.
$$

Substituting this into the equation of motion gives:

$$
\ddot{\theta} + \frac{g}{L}\theta = 0.
$$

This is the familiar equation of **simple harmonic motion** with angular frequency

$$
\omega = \sqrt{\frac{g}{L}},
$$

and period

$$
T = 2\pi\sqrt{\frac{L}{g}}.
$$

#### Energy Interpretation

The potential energy of the pendulum (taking $U = 0$ at the lowest point) is

$$
U(\theta) = mgL(1 - \cos\theta).
$$

Expanding for small $\theta$ using the Taylor series $\cos\theta \approx 1 - \frac{1}{2}\theta^2$ gives:

$$
U(\theta) \approx \frac{1}{2}mgL\theta^2.
$$

Thus, near equilibrium, the potential is quadratic in $\theta$, confirming the harmonic nature of small oscillations.

#### Validity of the Approximation

The small-angle approximation is accurate as long as $\theta \lesssim 10^\circ$ (≈ 0.17 rad).  
For larger amplitudes, the restoring force becomes weaker than predicted by the linear model, and the period increases slightly. However, the linearized equation captures the essential behavior and is an excellent approximation for most practical oscillations.

#### Check Point

- The pendulum’s exact motion is nonlinear, but for small angles it behaves as a simple harmonic oscillator.  
- The angular frequency depends only on the ratio $g/L$, not on the amplitude or mass.  
- This approach demonstrates how *linearization near equilibrium* transforms complex systems into analytically solvable ones.  
- The same technique underlies the general theory of small oscillations applied later to multi-dimensional and coupled systems.

## C7.1.7 Key Takeaways

- Small oscillations occur when displacements from equilibrium are small enough to linearize the equations of motion.  
- Any stable equilibrium behaves approximately like a harmonic oscillator.  
- The frequency of oscillation is determined by the curvature of the potential at equilibrium.  
- This approach generalizes easily to systems with multiple degrees of freedom, leading to coupled oscillations and normal modes in the next sections.
