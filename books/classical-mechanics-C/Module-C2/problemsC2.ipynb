## Problem 1 — Sandcrawler Losing Cargo

A Jawa sandcrawler of mass $M$ is transporting loose scrap metal with total mass $m$ across the desert world of Tatooine.

At time $t=0$:

- the sandcrawler is initially at rest,
- its engines apply a constant horizontal driving force $F$,
- a cargo hatch underneath the vehicle accidentally opens,
- scrap metal begins falling vertically downward at a constant rate.

Let the magnitude of the cargo loss rate be

$$
k=-\frac{dm}{dt},
$$

where $k>0$.

Assume:
- the sandcrawler moves only horizontally,
- the falling scrap leaves vertically,
- the scrap has the same horizontal velocity as the sandcrawler at the instant it falls out,
- resistance forces are negligible.

Find the speed of the sandcrawler when all the scrap metal has fallen out.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

At time $t$, the sandcrawler plus remaining cargo has mass

$$
M(t)=M+m-kt
$$

and speed

$$
v.
$$

During a small time interval $dt$, a small amount of cargo leaves:

$$
dm_{\text{out}}=kdt.
$$

Because the cargo falls vertically, it leaves with the same horizontal speed $v$ as the sandcrawler at the instant it exits.

---

<strong>Momentum at Time $t$</strong>

Just before the small amount of cargo leaves, the horizontal momentum is

$$
P_x(t)
=
M(t)v.
$$

---

<strong>Momentum at Time $t+dt$</strong>

After the cargo leaves, the remaining sandcrawler mass is

$$
M(t)-dm_{\text{out}},
$$

and its speed is

$$
v+dv.
$$

The cargo that left has mass $dm_{\text{out}}$ and horizontal speed $v$. In principle, we could also say it left at a speed of $v + dv$, but since we ignore terms like $dm_{\text{out}}dv$ the answer remains the same.

Therefore,

$$
P_x(t+dt)
=
\left(M(t)-dm_{\text{out}}\right)(v+dv)
+
dm_{\text{out}}v.
$$

---

<strong>Compute the Change in Momentum</strong>

The change in horizontal momentum is

$$
dP_x
=
P_x(t+dt)-P_x(t).
$$

Substituting,

$$
dP_x
=
\left(M(t)-dm_{\text{out}}\right)(v+dv)
+
dm_{\text{out}}v
-
M(t)v.
$$

Expanding,

$$
dP_x
=
M(t)v
+
M(t)dv
-
dm_{\text{out}}v
-
dm_{\text{out}}dv
+
dm_{\text{out}}v
-
M(t)v.
$$

Canceling common terms gives

$$
dP_x
=
M(t)dv
-
dm_{\text{out}}dv.
$$

The term

$$
dm_{\text{out}}dv
$$

is second order and is neglected.

Thus,

$$
dP_x
=
M(t)dv.
$$

---

<strong>Apply the Impulse-Momentum Theorem</strong>

The external horizontal impulse is

$$
Fdt.
$$

Therefore,

$$
dP_x=Fdt.
$$

So,

$$
M(t)dv=Fdt.
$$

Dividing by $dt$,

$$
M(t)\frac{dv}{dt}=F.
$$

Using

$$
M(t)=M+m-kt,
$$

we get

$$
(M+m-kt)\frac{dv}{dt}=F.
$$

Thus,

$$
\frac{dv}{dt}
=
\frac{F}{M+m-kt}.
$$

---

<strong>Determine When the Cargo is Gone</strong>

All cargo is gone when

$$
m-kt_f=0.
$$

Thus,

$$
t_f=\frac{m}{k}.
$$

---

<strong>Integrate to Find the Final Speed</strong>

Since the sandcrawler starts from rest,

$$
v(0)=0.
$$

Therefore,

$$
\int_0^{v_f}dv
=
F\int_0^{m/k}
\frac{dt}{M+m-kt}.
$$

Let

$$
x=M+m-kt.
$$

Then

$$
dx=-kdt.
$$

So,

$$
v_f
=
-\frac{F}{k}
\int_{M+m}^{M}
\frac{dx}{x}.
$$

Evaluating,

$$
v_f
=
-\frac{F}{k}
\left[
\ln x
\right]_{M+m}^{M}.
$$

Therefore,

$$
v_f
=
\frac{F}{k}
\ln\left(\frac{M+m}{M}\right).
$$

Hence,

$$
\boxed{
v_f
=
\frac{F}{k}
\ln\left(\frac{M+m}{M}\right)
}
$$

</div>
</details>

## Problem 2 — Hoth Supply Pod Capture

During a supply operation near the ice planet Hoth, a Rebel transport flies directly behind an unmanned cargo pod along the same straight-line path. The two craft are not physically connected. To rapidly transfer supplies, the transport fires sealed supply capsules forward into the cargo pod using a magnetic launch system. The pilot adjusts the transport speed so that the separation distance between the transport and the cargo pod remains constant throughout the transfer. The supply capsules enter the cargo pod at a constant mass-transfer rate

$$
\frac{dm}{dt}
=
40.0~\text{kg/s}.
$$

The capsules enter with speed

$$
u
=
25.0~\text{m/s}
$$

relative to the cargo pod.

Initially, the cargo pod is at rest and has mass

$$
M_0
=
800~\text{kg}.
$$

Determine the speed of the cargo pod after

$$
t
=
30.0~\text{s}.
$$

Assume:
- one-dimensional motion,
- negligible external forces,
- all incoming capsules remain inside the cargo pod after impact,
- the launch speed relative to the pod remains constant.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Let

$$
k=\frac{dm}{dt}.
$$

At time $t$, the cargo pod has mass

$$
M(t)=M_0+kt
$$

and speed

$$
v.
$$

During a small time interval $dt$, the pod captures additional mass

$$
dm=kdt.
$$

The incoming capsules enter with speed $u$ relative to the pod, so their speed relative to the inertial frame is

$$
v+u.
$$

---

<strong>Momentum at Time $t$</strong>

Just before the small amount of mass $dm$ enters the pod, the total momentum of the pod plus incoming capsules is

$$
P_x(t)
=
M(t)v
+
dm(v+u).
$$

---

<strong>Momentum at Time $t+dt$</strong>

After the mass $dm$ has joined the pod, the new pod mass is

$$
M(t)+dm,
$$

and the pod speed is

$$
v+dv.
$$

Thus,

$$
P_x(t+dt)
=
\left(M(t)+dm\right)(v+dv).
$$

---

<strong>Compute the Change in Momentum</strong>

The change in momentum is

$$
dP_x
=
P_x(t+dt)-P_x(t).
$$

Substituting,

$$
dP_x
=
\left(M(t)+dm\right)(v+dv)
-
\left[
M(t)v
+
dm(v+u)
\right].
$$

Expand:

$$
dP_x
=
M(t)v
+
M(t)dv
+
dm\,v
+
dm\,dv
-
M(t)v
-
dm\,v
-
u\,dm.
$$

Cancel common terms:

$$
dP_x
=
M(t)dv
+
dm\,dv
-
u\,dm.
$$

The term

$$
dm\,dv
$$

is second order and is neglected.

Therefore,

$$
dP_x
=
M(t)dv
-
u\,dm.
$$

Since there are no external forces in the $x$-direction,

$$
dP_x=0.
$$

So,

$$
M(t)dv-u\,dm=0.
$$

Thus,

$$
M(t)dv=u\,dm.
$$

Dividing by $dt$ gives

$$
M(t)\frac{dv}{dt}
=
u\frac{dm}{dt}.
$$

Using

$$
M(t)=M_0+kt
$$

and

$$
\frac{dm}{dt}=k,
$$

we obtain

$$
(M_0+kt)\frac{dv}{dt}
=
uk.
$$

Therefore,

$$
\frac{dv}{dt}
=
\frac{uk}{M_0+kt}.
$$

---

<strong>Integrate to Find the Speed</strong>

Integrate from $v=0$ at $t=0$ to $v=v(t)$ at time $t$:

$$
\int_0^{v(t)}dv
=
uk
\int_0^t
\frac{dt'}{M_0+kt'}.
$$

This gives

$$
v(t)
=
u
\ln\left(
\frac{M_0+kt}{M_0}
\right).
$$

Now substitute the values:

$$
v(30.0)
=
25.0
\ln\left(
\frac{800+(40.0)(30.0)}{800}
\right).
$$

The final mass is

$$
M(30.0)=2000~\text{kg}.
$$

Thus,

$$
v(30.0)
=
25.0
\ln\left(
\frac{2000}{800}
\right).
$$

Since

$$
\ln(2.5)\approx0.916,
$$

we obtain

$$
v(30.0)
\approx
22.9~\text{m/s}.
$$

Therefore,

$$
\boxed{
v(30.0)
\approx
22.9~\text{m/s}
}
$$

</div>
</details>

## Problem 3 — Leaking Bucket Half-Atwood Machine

A block of mass $M$ rests on a horizontal table and is connected by a light string over a pulley to a hanging bucket. The block experiences kinetic friction with coefficient $\mu_k$. The hanging bucket initially has total mass

$$
m_0
$$

including the bucket and water.

The bucket has small holes in its side, so water leaks out horizontally at a constant rate

$$
k=-\frac{dm}{dt},
$$

where $k>0$.

Assume:
- the string and pulley are ideal,
- the system starts from rest,
- the water leaks horizontally, so it carries no extra vertical momentum relative to the bucket,
- the hanging mass decreases according to

$$
m(t)=m_0-kt.
$$

Find the greatest speed reached by the system.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Let the positive direction be:
- to the right for the block on the table,
- downward for the hanging bucket.

At time $t$, the bucket has mass

$$
m(t)=m_0-kt
$$

and the system has speed

$$
v.
$$

During a small time interval $dt$, a small amount of water leaks out:

$$
dm_{\text{out}}=kdt.
$$

Because the water leaks horizontally, it has the same vertical velocity as the bucket at the instant it leaves. Therefore, there is no rocket-like thrust term in the direction of motion.

---

<strong>Momentum at Time $t$</strong>

The total momentum along the direction of motion is

$$
P(t)
=
\left(M+m(t)\right)v.
$$

---

<strong>Momentum at Time $t+dt$</strong>

After the water leaks out:
- the block still has mass $M$,
- the bucket has mass $m(t)-dm_{\text{out}}$,
- the block and bucket have speed $v+dv$,
- the leaked water has mass $dm_{\text{out}}$ and speed $v$ along the direction of motion.

Therefore,

$$
P(t+dt)
=
\left[M+m(t)-dm_{\text{out}}\right](v+dv)
+
dm_{\text{out}}v.
$$

---

<strong>Compute the Change in Momentum</strong>

The change in momentum is

$$
dP
=
P(t+dt)-P(t).
$$

Substituting,

$$
dP
=
\left[M+m(t)-dm_{\text{out}}\right](v+dv)
+
dm_{\text{out}}v
-
\left[M+m(t)\right]v.
$$

Expanding,

$$
dP
=
\left[M+m(t)\right]v
+
\left[M+m(t)\right]dv
-
dm_{\text{out}}v
-
dm_{\text{out}}dv
+
dm_{\text{out}}v
-
\left[M+m(t)\right]v.
$$

Canceling common terms gives

$$
dP
=
\left[M+m(t)\right]dv
-
dm_{\text{out}}dv.
$$

The term

$$
dm_{\text{out}}dv
$$

is second order and is neglected.

Thus,

$$
dP
=
\left[M+m(t)\right]dv.
$$

---

<strong>Apply the Impulse-Momentum Theorem</strong>

The external force along the direction of motion is

$$
F_{\text{net}}
=
m(t)g-\mu_k Mg.
$$

Therefore,

$$
dP
=
F_{\text{net}}dt.
$$

So,

$$
\left[M+m(t)\right]dv
=
\left[m(t)g-\mu_k Mg\right]dt.
$$

Dividing by $dt$,

$$
\left[M+m(t)\right]\frac{dv}{dt}
=
m(t)g-\mu_k Mg.
$$

Using

$$
m(t)=m_0-kt,
$$

we get

$$
\left(M+m_0-kt\right)\frac{dv}{dt}
=
(m_0-kt)g-\mu_k Mg.
$$

Therefore,

$$
\frac{dv}{dt}
=
g\frac{m_0-kt-\mu_k M}{M+m_0-kt}.
$$

---

<strong>Find When the Greatest Speed Occurs</strong>

The speed is greatest when the acceleration changes from positive to zero:

$$
\frac{dv}{dt}=0.
$$

Therefore,

$$
m_0-kt_*-\mu_k M=0.
$$

Solving,

$$
t_*=
\frac{m_0-\mu_k M}{k}.
$$

This only makes sense if

$$
m_0>\mu_k M.
$$

---

<strong>Integrate to Find the Greatest Speed</strong>

The greatest speed is

$$
v_{\max}
=
\int_0^{t_*}
g\frac{m_0-kt-\mu_k M}{M+m_0-kt}\,dt.
$$

Use

$$
x=M+m_0-kt.
$$

Then

$$
dx=-kdt.
$$

Also,

$$
m_0-kt=x-M.
$$

So the numerator becomes

$$
m_0-kt-\mu_k M
=
x-M-\mu_k M
=
x-M(1+\mu_k).
$$

When $t=0$,

$$
x=M+m_0.
$$

When $t=t_*$,

$$
x=M+\mu_k M
=
M(1+\mu_k).
$$

Thus,

$$
v_{\max}
=
-\frac{g}{k}
\int_{M+m_0}^{M(1+\mu_k)}
\frac{x-M(1+\mu_k)}{x}\,dx.
$$

Reversing the limits,

$$
v_{\max}
=
\frac{g}{k}
\int_{M(1+\mu_k)}^{M+m_0}
\left[
1-\frac{M(1+\mu_k)}{x}
\right]dx.
$$

Evaluating,

$$
v_{\max}
=
\frac{g}{k}
\left[
x-M(1+\mu_k)\ln x
\right]_{M(1+\mu_k)}^{M+m_0}.
$$

Therefore,

$$
v_{\max}
=
\frac{g}{k}
\left[
m_0-\mu_k M
-
M(1+\mu_k)
\ln\left(
\frac{M+m_0}{M(1+\mu_k)}
\right)
\right].
$$

Hence,

$$
\boxed{
v_{\max}
=
\frac{g}{k}
\left[
m_0-\mu_k M
-
M(1+\mu_k)
\ln\left(
\frac{M+m_0}{M(1+\mu_k)}
\right)
\right]
}
$$

</div>
</details>

## Problem 4 — Rebel Rocket Launch Under Gravity

A Rebel escape rocket launches vertically upward from the surface of a remote moon.

At time $t=0$, the rocket starts from rest with initial mass

$$
M_i.
$$

The rocket burns fuel at a constant rate

$$
k=-\frac{dM}{dt},
$$

where $k>0$.

The exhaust leaves the rocket downward with constant speed

$$
u
$$

relative to the rocket.

Assume:
- the rocket moves vertically upward,
- gravity is constant with magnitude $g$,
- air resistance is negligible,
- the rocket burns fuel until its mass reaches $M_f$.

Find the speed of the rocket when the fuel burn ends.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

We use the rocket equation with gravity:

$$
\Delta v
=
u\ln\left(\frac{M_i}{M_f}\right)-g\Delta t.
$$

The burn time follows from the constant mass-loss rate:

$$
\Delta t
=
\frac{M_i-M_f}{k}.
$$

Substituting this into the rocket equation gives

$$
v_f-v_i
=
u\ln\left(\frac{M_i}{M_f}\right)
-
g\frac{M_i-M_f}{k}.
$$

Since the rocket starts from rest,

$$
v_i=0.
$$

Therefore,

$$
\boxed{
v_f
=
u\ln\left(\frac{M_i}{M_f}\right)
-
g\frac{M_i-M_f}{k}
}
$$

The first term,

$$
u\ln\left(\frac{M_i}{M_f}\right),
$$

is the ideal rocket contribution.

The second term,

$$
g\frac{M_i-M_f}{k},
$$

is the loss in speed due to gravity during the burn.

</div>
</details>

## Problem 5 — Rebel Rocket Launch with Atmospheric Drag

A Rebel escape rocket launches vertically upward from the surface of a planet with an atmosphere.

The rocket has initial mass

$$
M_i = 2.50\times 10^4~\text{kg}
$$

and burns fuel at a constant rate

$$
k = 120~\text{kg/s}.
$$

The exhaust speed relative to the rocket is

$$
u = 2.80\times 10^3~\text{m/s}.
$$

The rocket burns until its mass reaches

$$
M_f = 1.30\times 10^4~\text{kg}.
$$

Gravity has constant magnitude

$$
g = 9.81~\text{m/s}^2.
$$

Atmospheric drag is modeled as

$$
F_D = bv^2,
$$

where

$$
b = 0.45~\text{kg/m}.
$$

The rocket starts from rest.

Assume upward is positive, so drag acts downward while the rocket moves upward.

---

### Tasks

1. Write the differential equation for the rocket velocity during powered flight.

2. Explain why this problem generally requires a numerical method.

3. Use the Forward Euler method to estimate the rocket velocity during the burn.

4. Use the Euler-Cromer method to estimate both velocity and height during the burn.

5. Make plots of:
   - velocity vs time,
   - height vs time,
   - acceleration vs time.

6. Compare the result with the no-drag prediction:

$$
v_f
=
u\ln\left(\frac{M_i}{M_f}\right)
-
g\frac{M_i-M_f}{k}.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

During powered flight, the mass decreases according to

$$
M(t)=M_i-kt.
$$

The burn ends when

$$
M(t_f)=M_f.
$$

Therefore,

$$
t_f
=
\frac{M_i-M_f}{k}.
$$

Substituting the values,

$$
t_f
=
\frac{2.50\times 10^4-1.30\times 10^4}{120}
=
100~\text{s}.
$$

---

<strong>Differential Equation</strong>

The thrust force is

$$
F_T=uk.
$$

Gravity acts downward:

$$
F_g=-M(t)g.
$$

Drag also acts downward while the rocket moves upward:

$$
F_D=-bv^2.
$$

Therefore,

$$
M(t)\frac{dv}{dt}
=
uk-M(t)g-bv^2.
$$

So,

$$
\frac{dv}{dt}
=
\frac{uk}{M_i-kt}
-
g
-
\frac{b}{M_i-kt}v^2.
$$

This is the differential equation for the rocket velocity during powered flight.

---

<strong>Why Numerical Methods Are Needed</strong>

Without drag, the equation can be integrated directly and gives

$$
v_f
=
u\ln\left(\frac{M_i}{M_f}\right)
-
gt_f.
$$

With drag, the acceleration depends on both time and velocity:

$$
a(t,v)
=
\frac{uk}{M_i-kt}
-
g
-
\frac{b}{M_i-kt}v^2.
$$

Because $v^2$ appears in the differential equation, the velocity cannot be found by the simple rocket-equation integral used in the no-drag case.

A numerical update method is therefore appropriate.

---

<strong>Forward Euler Update</strong>

The Forward Euler method uses the current acceleration to update the velocity:

$$
v_{n+1}
=
v_n
+
a(t_n,v_n)\Delta t.
$$

where

$$
a(t_n,v_n)
=
\frac{uk}{M_i-kt_n}
-
g
-
\frac{b}{M_i-kt_n}v_n^2.
$$

Thus,

$$
v_{n+1}
=
v_n
+
\left[
\frac{uk}{M_i-kt_n}
-
g
-
\frac{b}{M_i-kt_n}v_n^2
\right]\Delta t.
$$

---

<strong>Euler-Cromer Update</strong>

For motion in one dimension, Euler-Cromer updates velocity first and then uses the updated velocity to update position:

$$
v_{n+1}
=
v_n
+
a(t_n,v_n)\Delta t
$$

and

$$
y_{n+1}
=
y_n
+
v_{n+1}\Delta t.
$$

This makes Euler-Cromer especially useful when tracking motion in time.

---

<strong>No-Drag Comparison</strong>

The no-drag prediction is

$$
v_f
=
u\ln\left(\frac{M_i}{M_f}\right)
-
g\frac{M_i-M_f}{k}.
$$

Substituting,

$$
v_f
=
(2.80\times10^3)
\ln\left(\frac{2.50\times10^4}{1.30\times10^4}\right)
-
(9.81)(100).
$$

Since

$$
\ln\left(\frac{2.50}{1.30}\right)\approx0.654,
$$

we get

$$
v_f
\approx
(2.80\times10^3)(0.654)-981.
$$

Therefore,

$$
v_f
\approx
850~\text{m/s}.
$$

The numerical solution with drag should produce a smaller final speed.

</div>
</details>

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Parameters
Mi = 2.50e4      # initial mass, kg
Mf = 1.30e4      # final mass, kg
k = 120.0        # mass loss rate, kg/s
u = 2.80e3       # exhaust speed, m/s
g = 9.81         # gravitational field strength, m/s^2
b = 0.45         # drag coefficient parameter, kg/m

# Burn time
tf = (Mi - Mf)/k

# Time step
dt = 0.05
N = int(tf/dt)

# Arrays
t = np.zeros(N+1)
v = np.zeros(N+1)
y = np.zeros(N+1)
a = np.zeros(N+1)

def mass(t):
    return Mi - k*t

def acceleration(t, v):
    M = mass(t)
    return (u*k)/M - g - (b/M)*v**2

# Euler-Cromer method
for n in range(N):
    t[n+1] = t[n] + dt
    a[n] = acceleration(t[n], v[n])
    v[n+1] = v[n] + a[n]*dt
    y[n+1] = y[n] + v[n+1]*dt

a[-1] = acceleration(t[-1], v[-1])

# No-drag prediction
v_no_drag = u*np.log(Mi/Mf) - g*tf

print(f"Burn time: {tf:.2f} s")
print(f"Final speed with drag: {v[-1]:.2f} m/s")
print(f"No-drag final speed: {v_no_drag:.2f} m/s")
print(f"Final height with drag: {y[-1]:.2f} m")

# Plot velocity vs time
plt.figure()
plt.plot(t, v, label="Euler-Cromer with drag")
plt.axhline(v_no_drag, linestyle="--", label="No-drag final speed")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Rocket Velocity vs Time")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()

# Plot height vs time
plt.figure()
plt.plot(t, y, label="Euler-Cromer with drag")
plt.xlabel("Time (s)")
plt.ylabel("Height (m)")
plt.title("Rocket Height vs Time")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()

# Plot acceleration vs time
plt.figure()
plt.plot(t, a, label="Acceleration with drag")
plt.xlabel("Time (s)")
plt.ylabel("Acceleration (m/s²)")
plt.title("Rocket Acceleration vs Time")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```
