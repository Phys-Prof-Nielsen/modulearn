## Problem 1 — Velocity from a Time-Dependent Acceleration

A particle moves along the x-axis with acceleration

$$
a(t)=vt+1
$$

in the positive x-direction.

At time

$$
t=0~\text{s},
$$

the velocity is

$$
v_0=+1.00~\text{m/s}.
$$

Use the Forward Euler method to estimate the velocity at

$$
t=3.00~\text{s}
$$

using a step size of

$$
h=0.50~\text{s}.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The Forward Euler velocity update equation is

$$
v_{n+1}=v_n+h a_n.
$$

The acceleration is

$$
a(t)=vt+1.
$$

We begin with

$$
t_0=0,
\qquad
v_0=1.00~\text{m/s}.
$$

The step size is

$$
h=0.50~\text{s}.
$$

---

### Step 1

At

$$
t_0=0,
$$

the acceleration is

$$
a_0=v(0)+1=1.
$$

Update velocity:

$$
v_1=v_0+h a_0.
$$

Substitute values:

$$
v_1=1.00+0.50(1).
$$

Thus,

$$
v_1=1.50~\text{m/s}.
$$

Advance time:

$$
t_1=0.50~\text{s}.
$$

---

### Step 2

Acceleration:

$$
a_1=v(0.50)+1.
$$

Thus,

$$
a_1=1.50v+1.
$$

Update velocity:

$$
v_2=v_1+0.50(1.50v+1).
$$

Continue this process until

$$
t=3.00~\text{s}.
$$

The repeated Euler updates produce the numerical estimate for the velocity at the final time.

</div>

</details>

```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# Parameters
h = 0.50          # step size (s)
t_final = 3.0     # final time (s)

# Initial conditions
t = 0.0
v = 1.0           # initial velocity (m/s)

# Lists for storing results
times = [t]
velocities = [v]

# Forward Euler loop
while t < t_final:
    
    # Acceleration function
    a = v*t + 1
    
    # Euler update
    v = v + h*a
    t = t + h
    
    # Store results
    times.append(t)
    velocities.append(v)

# Display results
print(" Forward Euler Results")
print("-------------------------")

for i in range(len(times)):
    print(f"t = {times[i]:4.2f} s    v = {velocities[i]:8.4f} m/s")

print("\nFinal estimate:")
print(f"v({t_final:.1f} s) ≈ {velocities[-1]:.4f} m/s")
```

## Problem 3 — Position from a Tangled Velocity Equation

The velocity and position of a particle traveling in the $+x$ direction are related by

$$
v+xt=t+t^2.
$$

At time

$$
t=0~\text{s},
$$

the position is

$$
x_0=0.00~\text{m}.
$$

Use the Forward Euler method to estimate the position at

$$
t=3.00~\text{s}
$$

using a step size of

$$
h=0.50~\text{s}.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The Forward Euler position update equation is

$$
x_{n+1}=x_n+h v_n.
$$

We are given the relationship

$$
v+xt=t+t^2.
$$

Solve for the velocity:

$$
v=t+t^2-xt.
$$

Therefore,

$$
v=t+t^2-x t.
$$

We begin with

$$
t_0=0,
\qquad
x_0=0.00~\text{m}.
$$

The step size is

$$
h=0.50~\text{s}.
$$

---

**Step 1**

At

$$
t_0=0,
\qquad
x_0=0,
$$

the velocity is

$$
v_0=0+0^2-(0)(0).
$$

Thus,

$$
v_0=0.
$$

Update position:

$$
x_1=x_0+h v_0.
$$

Substitute values:

$$
x_1=0+0.50(0).
$$

Thus,

$$
x_1=0.00~\text{m}.
$$

Advance time:

$$
t_1=0.50~\text{s}.
$$

---

**Step 2**

At

$$
t_1=0.50,
\qquad
x_1=0.00,
$$

the velocity is

$$
v_1=0.50+(0.50)^2-(0.00)(0.50).
$$

Thus,

$$
v_1=0.75~\text{m/s}.
$$

Update position:

$$
x_2=x_1+h v_1.
$$

Substitute values:

$$
x_2=0.00+0.50(0.75).
$$

Therefore,

$$
x_2=0.375~\text{m}.
$$

Continue this process until

$$
t=3.00~\text{s}.
$$

The repeated Euler updates produce the numerical estimate for the final position.

</div>

</details>

```python
# DIY Cell
```

```python
%reset -f
import numpy as np

# Parameters
h = 0.50
t_final = 3.0

# Initial conditions
t = 0.0
x = 0.0

# Lists for storing results
times = [t]
positions = [x]
velocities = []

# Forward Euler loop
while t < t_final:
    
    # Velocity from the given relationship:
    # v + x*t = t + t^2
    v = t + t**2 - x*t
    
    # Store velocity used for this step
    velocities.append(v)
    
    # Euler update
    x = x + h*v
    t = t + h
    
    # Store updated values
    times.append(t)
    positions.append(x)

# Compute final velocity value for table completeness
velocities.append(times[-1] + times[-1]**2 - positions[-1]*times[-1])

# Display results
print("Forward Euler Results")
print("-------------------------")

for i in range(len(times)):
    print(f"t = {times[i]:4.2f} s    x = {positions[i]:8.4f} m    v = {velocities[i]:8.4f} m/s")

print("\nFinal estimate:")
print(f"x({t_final:.1f} s) ≈ {positions[-1]:.4f} m")
```

Now also assume the initial position is

Use the Forward Euler method to estimate the position at

<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">

x_0=0.00~\text{m}.

The step size is

h=0.50~\text{s}.
$$

---

### Step 1

Initial acceleration:

a_0=v_0 t_0+1.
$$

Substitute values:

$$
a_0=(1.00)(0)+1.

Thus,

$$
a_0=1.00~\text{m/s}^2.

Update velocity:

v_1=v_0+h a_0.

```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# Parameters
h = 0.50
t_final = 3.0

# Initial conditions
t = 0.0
v = 1.0
x = 0.0

# Lists for storing results
times = [t]
velocities = [v]
positions = [x]

# Forward Euler loop
while t < t_final:
    
    # Acceleration
    a = v*t + 1
    
    # Store current velocity before updating
    v_old = v
    
    # Euler updates
    v = v + h*a
    x = x + h*v_old
    
    # Advance time
    t = t + h
    
    # Store results
    times.append(t)
    velocities.append(v)
    positions.append(x)

# Display results
print("Forward Euler Results")
print("-----------------------------------------")

for i in range(len(times)):
    print(f"t = {times[i]:4.2f} s    x = {positions[i]:8.4f} m    v = {velocities[i]:8.4f} m/s")

print("\nFinal estimate:")
print(f"x({t_final:.1f} s) ≈ {positions[-1]:.4f} m")
```

## Problem 5 — Combining Backward-Time Euler, Central Difference, and Forward Euler

A particle moves along the $x$-axis with velocity described by

$$
v(t,x)=t^2-x.
$$

At time

$$
t=1.00~\text{s},
$$

the position is

$$
x(1.00)=2.00~\text{m}.
$$

Use a step size of

$$
h=0.50~\text{s}.
$$

Estimate the position at

$$
t=2.00~\text{s}
$$

using the following required sequence:

1. Use **Forward Euler backward in time** to estimate $x(0.50)$.
2. Use the **central difference method** to estimate $x(1.50)$.
3. Use **Forward Euler forward in time** to estimate $x(2.00)$.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The velocity is

$$
v(t,x)=t^2-x.
$$

Since

$$
v=\frac{dx}{dt},
$$

we have

$$
\frac{dx}{dt}=t^2-x.
$$

We are given

$$
x(1.00)=2.00~\text{m}.
$$

The step size is

$$
h=0.50~\text{s}.
$$

---

**Step 1: Forward Euler backward in time**

Forward Euler is

$$
x_{n+1}=x_n+h f(t_n,x_n).
$$

To step backward in time, use

$$
h=-0.50~\text{s}.
$$

At

$$
t=1.00~\text{s},
\qquad
x=2.00~\text{m},
$$

the velocity is

$$
f(1.00,2.00)=(1.00)^2-2.00.
$$

Thus,

$$
f(1.00,2.00)=-1.00~\text{m/s}.
$$

Now step backward:

$$
x(0.50)=x(1.00)+(-0.50)(-1.00).
$$

Therefore,

$$
x(0.50)=2.50~\text{m}.
$$

---

**Step 2: Central difference**

The central difference update is

$$
x_{n+1}=x_{n-1}+2h f(t_n,x_n).
$$

Here,

$$
t_n=1.00~\text{s},
\qquad
x_n=x(1.00)=2.00~\text{m},
$$

and

$$
x_{n-1}=x(0.50)=2.50~\text{m}.
$$

Using

$$
f(1.00,2.00)=-1.00,
$$

we get

$$
x(1.50)=x(0.50)+2(0.50)(-1.00).
$$

Therefore,

$$
x(1.50)=1.50~\text{m}.
$$

---

**Step 3: Forward Euler forward in time**

Now use Forward Euler from $t=1.50~\text{s}$ to $t=2.00~\text{s}$.

First compute the velocity:

$$
f(1.50,1.50)=(1.50)^2-1.50.
$$

Thus,

$$
f(1.50,1.50)=2.25-1.50=0.75~\text{m/s}.
$$

Now update:

$$
x(2.00)=x(1.50)+0.50(0.75).
$$

Therefore,

$$
x(2.00)=1.875~\text{m}.
$$

</div>

</details>

```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# Step size
h = 0.50

# Differential equation: dx/dt = f(t,x)
def f(t, x):
    return t**2 - x

# Given value
t_center = 1.00
x_center = 2.00

# Step 1: Forward Euler backward in time
h_back = -h
x_back = x_center + h_back*f(t_center, x_center)
t_back = t_center + h_back

# Step 2: Central difference
# x_forward = x_back + 2h*f(t_center, x_center)
x_mid_forward = x_back + 2*h*f(t_center, x_center)
t_mid_forward = t_center + h

# Step 3: Forward Euler forward in time
x_final = x_mid_forward + h*f(t_mid_forward, x_mid_forward)
t_final = t_mid_forward + h

# Display results
print("Mixed Method Results")
print("-------------------------")
print(f"x({t_back:.2f} s) = {x_back:.4f} m   using Forward Euler backward in time")
print(f"x({t_mid_forward:.2f} s) = {x_mid_forward:.4f} m   using Central Difference")
print(f"x({t_final:.2f} s) = {x_final:.4f} m   using Forward Euler forward in time")
```

## Problem 6 — Projectile Motion with Air Drag Using Many Central Difference Steps

A small projectile is moving vertically through air. Choose upward as the positive direction.

The projectile experiences gravity and linear air drag, so its acceleration is

$$
a(v)=-g-\frac{b}{m}v.
$$

The constants are

$$
g=9.81~\text{m/s}^2,
\qquad
m=0.20~\text{kg},
\qquad
b=0.04~\text{kg/s}.
$$

At time

$$
t=1.00~\text{s},
$$

the projectile has velocity

$$
v(1.00)=8.00~\text{m/s}.
$$

Use a step size of

$$
h=0.01~\text{s}.
$$

Estimate the velocity at

$$
t=2.00~\text{s}
$$

using the following required sequence:

1. Use **Forward Euler backward in time** to estimate $v(0.99)$.
2. Use the **central difference method** to advance from $t=1.00~\text{s}$ to $t=1.99~\text{s}$.
3. Use **Forward Euler forward in time** to estimate $v(2.00)$.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The acceleration is

$$
a(v)=-g-\frac{b}{m}v.
$$

Since

$$
a=\frac{dv}{dt},
$$

we have

$$
\frac{dv}{dt}=-g-\frac{b}{m}v.
$$

We are given

$$
v(1.00)=8.00~\text{m/s},
$$

and

$$
h=0.01~\text{s}.
$$

First, compute

$$
\frac{b}{m}=\frac{0.04}{0.20}=0.20~\text{s}^{-1}.
$$

So the differential equation is

$$
\frac{dv}{dt}=-9.81-0.20v.
$$

---

**Step 1: Forward Euler backward in time**

Forward Euler is

$$
v_{n+1}=v_n+h a(v_n).
$$

To move backward in time, use a negative step size:

$$
h=-0.01~\text{s}.
$$

At

$$
v(1.00)=8.00~\text{m/s},
$$

the acceleration is

$$
a(8.00)=-9.81-0.20(8.00).
$$

Thus,

$$
a(8.00)=-11.41~\text{m/s}^2.
$$

Now step backward:

$$
v(0.99)=v(1.00)-0.01a(8.00).
$$

Therefore,

$$
v(0.99)=8.00-0.01(-11.41).
$$

So

$$
v(0.99)=8.1141~\text{m/s}.
$$

---

**Step 2: Central difference method**

The central difference update is

$$
v_{n+1}=v_{n-1}+2h a(v_n).
$$

This is repeated many times until reaching

$$
t=1.99~\text{s}.
$$

---

**Step 3: Forward Euler forward in time**

Once $v(1.99)$ is known, use Forward Euler one final time:

$$
v(2.00)=v(1.99)+h a(v(1.99)).
$$

Because this requires about 100 steps, the full calculation is best completed using Python.

</div>

</details>

```python
# DIY Cell
```

```python
%reset -f

import numpy as np

# Parameters
g = 9.81       # gravitational acceleration (m/s^2)
m = 0.20       # mass (kg)
b = 0.04       # linear drag coefficient (kg/s)
h = 0.01       # step size (s)

# Acceleration function
def acceleration(v):
    return -g - (b/m)*v

# Given value
t_current = 1.00
v_current = 8.00

# Step 1: Forward Euler backward in time
t_previous = t_current - h
v_previous = v_current - h*acceleration(v_current)

# Store values
times = [t_previous, t_current]
velocities = [v_previous, v_current]

# Step 2: Central difference steps from t = 1.00 to t = 1.99
while t_current < 1.99 - 1e-12:
    
    v_next = v_previous + 2*h*acceleration(v_current)
    t_next = t_current + h
    
    times.append(t_next)
    velocities.append(v_next)
    
    v_previous = v_current
    v_current = v_next
    
    t_previous = t_current
    t_current = t_next

# Step 3: Forward Euler forward in time from t = 1.99 to t = 2.00
v_final = v_current + h*acceleration(v_current)
t_final = t_current + h

times.append(t_final)
velocities.append(v_final)

# Display results
print("Mixed Method Results")
print("-------------------------")
print(f"v(0.99 s) = {velocities[0]:.6f} m/s   using Forward Euler backward in time")
print(f"v(1.00 s) = {velocities[1]:.6f} m/s   given")
print(f"v(1.99 s) = {v_current:.6f} m/s   after central difference steps")
print(f"v(2.00 s) = {v_final:.6f} m/s   using Forward Euler forward in time")

print("\nFinal estimate:")
print(f"v(2.00 s) ≈ {v_final:.6f} m/s")
```

## Problem 7 — Avalanche Control Shell with Air Drag

Explosive cannon shells launched from small howitzers are sometimes used to reduce avalanche danger by intentionally triggering smaller avalanches in areas prone to serious avalanche hazards.

x_0=0,

y_0=0.

---

**Objective 2**

Including quadratic air drag, determine:

- the launch angle that gives the maximum range,

---

**Objective 3**

The shell reaches a high altitude, so the changing density of the atmosphere may affect the drag force.

Model the atmospheric density using the adiabatic approximation

$$
\rho(y)=\rho_0\left(1-\frac{ay}{T_0}\right)^\alpha,
$$

where

$$
a=6.5\times 10^{-3}~\text{K/m},
$$

$$
\alpha=2.5,
$$

and use an appropriate sea-level temperature, such as

$$
T_0=288~\text{K}.
$$

The modified drag acceleration is

$$
a_{\text{drag}}(y)=\frac{\rho(y)}{\rho_0}a_{\text{drag},0}.
$$

Repeat the calculation with altitude-dependent air drag and determine:

- the launch angle that gives the maximum range,
- the maximum horizontal range,
- the maximum altitude

---

**Objective 4**

For each of the three cases, make a trajectory plot of the shell:

1. no air drag,
2. constant-density quadratic air drag,
3. altitude-dependent quadratic air drag.

Plot $y$ versus $x.$

Use the launch angle that gives the maximum range for each case.

Compare the three trajectories.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

Without air drag, the maximum range occurs at

$$
\theta=45^\circ.
$$

The ideal range is

$$
R=\frac{v_0^2\sin(2\theta)}{g}.
$$

At

$$
\theta=45^\circ,
$$

this becomes

$$
R_{\max}=\frac{v_0^2}{g}.
$$

Using

$$
v_0=1000.0~\text{m/s},
$$

we get

$$
R_{\max}=\frac{(1000.0)^2}{9.81}.
$$

Thus,

$$
R_{\max}\approx 1.02\times 10^5~\text{m}.
$$

The maximum altitude is

$$
H=\frac{v_0^2\sin^2\theta}{2g}.
$$

At

$$
\theta=45^\circ,
$$

this gives

$$
H\approx 2.55\times 10^4~\text{m}.
$$

So without drag, the shell travels unrealistically far and high.

With air drag, there is no simple closed-form range formula. We numerically integrate the equations of motion:

$$
\frac{dx}{dt}=v_x,
$$

$$
\frac{dy}{dt}=v_y,
$$

$$
\frac{dv_x}{dt}=-\frac{B}{m}v v_x,
$$

$$
\frac{dv_y}{dt}=-g-\frac{B}{m}v v_y.
$$

With altitude-dependent drag, we multiply the drag terms by

$$
\frac{\rho(y)}{\rho_0}
=
\left(1-\frac{ay}{T_0}\right)^\alpha.
$$

Because the shell eventually returns to the ground, the numerical simulation stops when

$$
y=0
$$

on the way down.

</div>

</details>

```python
# DIY Cell
```

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Constants
g = 9.81
v0 = 1000.0
B_over_m = 4.00e-5

# Atmosphere parameters
lapse_rate = 6.5e-3
T0 = 288.0
alpha = 2.5

# Time step
dt = 0.01

def density_ratio(y):
    factor = 1 - lapse_rate*y/T0
    
    if factor <= 0:
        return 0.0
    
    return factor**alpha

def simulate_projectile(theta_deg, drag=False, variable_density=False, store=False):
    theta = np.radians(theta_deg)
    
    x = 0.0
    y = 0.0
    vx = v0*np.cos(theta)
    vy = v0*np.sin(theta)
    
    max_y = y
    
    x_values = [x]
    y_values = [y]
    
    while True:
        x_old = x
        y_old = y
        
        v = np.sqrt(vx**2 + vy**2)
        
        if drag:
            density_factor = 1.0
            
            if variable_density:
                density_factor = density_ratio(y)
            
            ax = -density_factor*B_over_m*v*vx
            ay = -g - density_factor*B_over_m*v*vy
        else:
            ax = 0.0
            ay = -g
        
        # Euler-Cromer update
        vx = vx + ax*dt
        vy = vy + ay*dt
        
        x = x + vx*dt
        y = y + vy*dt
        
        if y > max_y:
            max_y = y
        
        x_values.append(x)
        y_values.append(y)
        
        if y < 0 and vy < 0:
            frac = y_old/(y_old - y)
            x_ground = x_old + frac*(x - x_old)
            
            if store:
                x_values[-1] = x_ground
                y_values[-1] = 0.0
                return x_ground, max_y, np.array(x_values), np.array(y_values)
            
            return x_ground, max_y

def scan_angles(drag=False, variable_density=False):
    angles = np.linspace(1, 89, 881)
    
    ranges = []
    heights = []
    
    for angle in angles:
        R, H = simulate_projectile(
            angle,
            drag=drag,
            variable_density=variable_density
        )
        ranges.append(R)
        heights.append(H)
    
    ranges = np.array(ranges)
    heights = np.array(heights)
    
    best_index = np.argmax(ranges)
    
    return angles[best_index], ranges[best_index], heights[best_index]

# Find best angles
angle_no_drag, range_no_drag, height_no_drag = scan_angles(
    drag=False,
    variable_density=False
)

angle_drag, range_drag, height_drag = scan_angles(
    drag=True,
    variable_density=False
)

angle_var_drag, range_var_drag, height_var_drag = scan_angles(
    drag=True,
    variable_density=True
)

# Generate trajectories using best angles
R1, H1, x1, y1 = simulate_projectile(
    angle_no_drag,
    drag=False,
    variable_density=False,
    store=True
)

R2, H2, x2, y2 = simulate_projectile(
    angle_drag,
    drag=True,
    variable_density=False,
    store=True
)

R3, H3, x3, y3 = simulate_projectile(
    angle_var_drag,
    drag=True,
    variable_density=True,
    store=True
)

# Display results
print("Projectile Results")
print("-----------------------------")

print("No air drag:")
print(f"Best angle:        {angle_no_drag:.2f} degrees")
print(f"Maximum range:     {range_no_drag/1000:.3f} km")
print(f"Maximum altitude:  {height_no_drag/1000:.3f} km")

print("\nConstant-density quadratic drag:")
print(f"Best angle:        {angle_drag:.2f} degrees")
print(f"Maximum range:     {range_drag/1000:.3f} km")
print(f"Maximum altitude:  {height_drag/1000:.3f} km")

print("\nAltitude-dependent quadratic drag:")
print(f"Best angle:        {angle_var_drag:.2f} degrees")
print(f"Maximum range:     {range_var_drag/1000:.3f} km")
print(f"Maximum altitude:  {height_var_drag/1000:.3f} km")

# Plot trajectories
plt.figure(figsize=(9,6))

plt.plot(x1/1000, y1/1000, label="No drag")
plt.plot(x2/1000, y2/1000, label="Constant-density drag")
plt.plot(x3/1000, y3/1000, label="Altitude-dependent drag")

plt.xlabel("Horizontal position x (km)")
plt.ylabel("Altitude y (km)")
plt.title("Howitzer Shell Trajectories")
plt.grid(True)
plt.legend()
plt.show()
```

## Problem 8 — Parachutist with Air Drag: Forward Euler vs Backward Euler

A parachutist is falling vertically downward. Choose downward as the positive direction.

The parachutist experiences gravity and linear air drag, so

$$
\frac{dv}{dt}=g-\frac{b}{m}v.
$$

The constants are

$$
g=9.81~\text{m/s}^2,
\qquad
m=80.0~\text{kg},
\qquad
b=24.0~\text{kg/s}.
$$

At time

$$
t=0,
$$

the parachutist starts from rest:

$$
v_0=0.
$$

Use a step size of

$$
h=1.00~\text{s}
$$

to estimate the velocity at

$$
t=10.0~\text{s}.
$$

Do this using both:

1. Forward Euler,
2. Backward Euler.

Then compare the two numerical results.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

For Forward Euler,

$$
v_{n+1}=v_n+h\left(g-\frac{b}{m}v_n\right).
$$

For Backward Euler,

$$
v_{n+1}=v_n+h\left(g-\frac{b}{m}v_{n+1}\right).
$$

Solving the Backward Euler equation for $v_{n+1}$ gives

$$
v_{n+1}
=
\frac{v_n+hg}{1+h\frac{b}{m}}.
$$

Both methods should approach the terminal velocity

$$
v_T=\frac{mg}{b}.
$$

</div>

</details>

```python
# DIY Cell
```

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Constants
g = 9.81
m = 80.0
b = 24.0

# Time settings
h = 1.00
t_final = 10.0

# Time array
times = np.arange(0, t_final + h, h)

# Arrays
v_forward = np.zeros(len(times))
v_backward = np.zeros(len(times))

# Initial condition
v_forward[0] = 0.0
v_backward[0] = 0.0

# Euler loops
for n in range(len(times)-1):
    
    # Forward Euler
    v_forward[n+1] = v_forward[n] + h*(g - (b/m)*v_forward[n])
    
    # Backward Euler
    v_backward[n+1] = (v_backward[n] + h*g)/(1 + h*b/m)

# Exact solution
v_terminal = m*g/b
v_exact = v_terminal*(1 - np.exp(-(b/m)*times))

# Display results
print("Forward Euler vs Backward Euler")
print("----------------------------------------")

for i in range(len(times)):
    print(
        f"t = {times[i]:4.1f} s"
        f"    Forward = {v_forward[i]:8.4f} m/s"
        f"    Backward = {v_backward[i]:8.4f} m/s"
        f"    Exact = {v_exact[i]:8.4f} m/s"
    )

print("\nFinal estimates:")
print(f"Forward Euler:   v({t_final:.1f} s) ≈ {v_forward[-1]:.4f} m/s")
print(f"Backward Euler:  v({t_final:.1f} s) ≈ {v_backward[-1]:.4f} m/s")
print(f"Exact:           v({t_final:.1f} s) = {v_exact[-1]:.4f} m/s")

print("\nTerminal velocity:")
print(f"v_T = {v_terminal:.4f} m/s")

# Plot
plt.figure(figsize=(8,5))
plt.plot(times, v_forward, "o-", label="Forward Euler")
plt.plot(times, v_backward, "s-", label="Backward Euler")
plt.plot(times, v_exact, "--", label="Exact solution")

plt.xlabel("Time (s)")
plt.ylabel("Downward velocity (m/s)")
plt.title("Parachutist with Linear Air Drag")
plt.grid(True)
plt.legend()
plt.show()
```

## Problem 9 — Muon Decay: Forward Euler vs Backward Euler

Muons are unstable particles that decay with a mean lifetime of approximately

$$
\tau = 2.20~\mu\text{s}.
$$

If $N(t)$ is the number of muons remaining at time $t$, then the decay equation is

$$
\frac{dN}{dt}=-\frac{1}{\tau}N.
$$

Suppose a detector initially contains

$$
N_0=10\,000
$$

muons at

$$
t=0.
$$

Use a step size of

$$
h=5.00~\mu\text{s}
$$

to estimate the number of muons remaining at

$$
t=25.0~\mu\text{s}.
$$

Do this using both:

1. Forward Euler,
2. Backward Euler.

Then compare the behavior of the two methods.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The decay equation is

$$
\frac{dN}{dt}=-\frac{1}{\tau}N.
$$

Forward Euler gives

$$
N_{n+1}=N_n-h\frac{1}{\tau}N_n.
$$

So

$$
N_{n+1}=\left(1-\frac{h}{\tau}\right)N_n.
$$

With

$$
h=5.00~\mu\text{s},
\qquad
\tau=2.20~\mu\text{s},
$$

we get

$$
1-\frac{h}{\tau}
=
1-\frac{5.00}{2.20}
\approx -1.27.
$$

This means Forward Euler predicts alternating positive and negative particle counts, which is unphysical.

Backward Euler gives

$$
N_{n+1}=N_n-h\frac{1}{\tau}N_{n+1}.
$$

Move all future-value terms to one side:

$$
N_{n+1}+\frac{h}{\tau}N_{n+1}=N_n.
$$

Factor:

$$
\left(1+\frac{h}{\tau}\right)N_{n+1}=N_n.
$$

Therefore,

$$
N_{n+1}=\frac{N_n}{1+h/\tau}.
$$

This remains positive and decays smoothly.

</div>

</details>

```python
# DIY Cell
```

```python
%reset -f

import numpy as np
import matplotlib.pyplot as plt

# Parameters
tau = 2.20      # muon lifetime in microseconds
h = 5.00        # step size in microseconds
t_final = 25.0  # final time in microseconds

# Initial number of muons
N0 = 10000.0

# Time array
times = np.arange(0, t_final + h, h)

# Arrays
N_forward = np.zeros(len(times))
N_backward = np.zeros(len(times))

# Initial values
N_forward[0] = N0
N_backward[0] = N0

# Numerical updates
for n in range(len(times)-1):
    
    # Forward Euler
    N_forward[n+1] = N_forward[n] - h*(N_forward[n]/tau)
    
    # Backward Euler
    N_backward[n+1] = N_backward[n]/(1 + h/tau)

# Exact solution
N_exact = N0*np.exp(-times/tau)

# Display results
print("Muon Decay: Forward Euler vs Backward Euler")
print("-------------------------------------------")

for i in range(len(times)):
    print(
        f"t = {times[i]:5.1f} microseconds"
        f"    Forward = {N_forward[i]:12.2f}"
        f"    Backward = {N_backward[i]:12.2f}"
        f"    Exact = {N_exact[i]:12.2f}"
    )

print("\nFinal estimates:")
print(f"Forward Euler:   N({t_final:.1f} microseconds) ≈ {N_forward[-1]:.2f}")
print(f"Backward Euler:  N({t_final:.1f} microseconds) ≈ {N_backward[-1]:.2f}")
print(f"Exact:           N({t_final:.1f} microseconds) = {N_exact[-1]:.2f}")

# Plot
plt.figure(figsize=(8,5))
plt.plot(times, N_forward, "o-", label="Forward Euler")
plt.plot(times, N_backward, "s-", label="Backward Euler")
plt.plot(times, N_exact, "--", label="Exact solution")

plt.xlabel("Time (microseconds)")
plt.ylabel("Number of muons")
plt.title("Muon Decay: Forward Euler vs Backward Euler")
plt.grid(True)
plt.legend()
plt.show()
```

## Problem 10 — Moon Orbit Around Earth: Forward Euler, Backward Euler, and Euler-Cromer

The Moon orbits Earth due to the gravitational force between them. In this problem, we model Earth as fixed at the origin and simulate the Moon's motion in two dimensions.

Use the gravitational acceleration

$$
\vec{a}
=
-\frac{GM}{r^3}\vec{r},
$$

where

$$
\vec{r}=x\hat{i}+y\hat{j},
$$

and

$$
r=\sqrt{x^2+y^2}.
$$

Use

$$
G=6.67430\times 10^{-11}~\text{N m}^2/\text{kg}^2,
$$

$$
M=5.972\times 10^{24}~\text{kg}.
$$

Assume the Moon begins at

$$
x_0=3.844\times 10^8~\text{m},
\qquad
y_0=0,
$$

with initial velocity

$$
v_{x,0}=0,
\qquad
v_{y,0}=1022~\text{m/s}.
$$

Use a time step of

$$
h=3600~\text{s},
$$

and simulate for

$$
30~\text{days}.
$$

Compare the orbit using:

1. Forward Euler,
2. Backward Euler,
3. Euler-Cromer.

Make plots of the Moon's trajectory for all three methods.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

The gravitational acceleration components are

$$
a_x=-\frac{GMx}{(x^2+y^2)^{3/2}},
$$

and

$$
a_y=-\frac{GMy}{(x^2+y^2)^{3/2}}.
$$

For Forward Euler, use

$$
v_{x,n+1}=v_{x,n}+h a_{x,n},
$$

$$
v_{y,n+1}=v_{y,n}+h a_{y,n},
$$

$$
x_{n+1}=x_n+h v_{x,n},
$$

$$
y_{n+1}=y_n+h v_{y,n}.
$$

For Euler-Cromer, update velocity first and then use the updated velocity:

$$
v_{x,n+1}=v_{x,n}+h a_{x,n},
$$

$$
v_{y,n+1}=v_{y,n}+h a_{y,n},
$$

$$
x_{n+1}=x_n+h v_{x,n+1},
$$

$$
y_{n+1}=y_n+h v_{y,n+1}.
$$

For Backward Euler, the acceleration must be evaluated at the future position:

$$
v_{x,n+1}=v_{x,n}+h a_x(x_{n+1},y_{n+1}),
$$

$$
v_{y,n+1}=v_{y,n}+h a_y(x_{n+1},y_{n+1}),
$$

$$
x_{n+1}=x_n+h v_{x,n+1},
$$

$$
y_{n+1}=y_n+h v_{y,n+1}.
$$

Because the future position appears inside the acceleration, Backward Euler requires solving an implicit system at every time step.

In the Python implementation below, we solve this implicit step using fixed-point iteration.

Forward Euler tends to add energy to the orbit.

Backward Euler tends to remove energy from the orbit.

Euler-Cromer usually gives the most physically reasonable long-term orbital behavior among these three methods.

</div>

</details>

```python
# DIY Cell
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
G = 6.67430e-11
M_earth = 5.972e24
mu = G*M_earth

# Initial conditions for Moon
x0 = 3.844e8        # m
y0 = 0.0            # m
vx0 = 0.0           # m/s
vy0 = 1022.0        # m/s

# Time settings
h = 3600.0          # one hour in seconds
days = 30
t_final = days*24*3600
times = np.arange(0, t_final + h, h)

def acceleration(x, y):
    r = np.sqrt(x**2 + y**2)
    ax = -mu*x/r**3
    ay = -mu*y/r**3
    return ax, ay

def energy(x, y, vx, vy):
    r = np.sqrt(x**2 + y**2)
    kinetic = 0.5*(vx**2 + vy**2)
    potential = -mu/r
    return kinetic + potential

def forward_euler():
    x = np.zeros(len(times))
    y = np.zeros(len(times))
    vx = np.zeros(len(times))
    vy = np.zeros(len(times))
    E = np.zeros(len(times))
    
    x[0], y[0] = x0, y0
    vx[0], vy[0] = vx0, vy0
    E[0] = energy(x[0], y[0], vx[0], vy[0])
    
    for n in range(len(times)-1):
        ax, ay = acceleration(x[n], y[n])
        
        vx[n+1] = vx[n] + h*ax
        vy[n+1] = vy[n] + h*ay
        
        x[n+1] = x[n] + h*vx[n]
        y[n+1] = y[n] + h*vy[n]
        
        E[n+1] = energy(x[n+1], y[n+1], vx[n+1], vy[n+1])
    
    return x, y, vx, vy, E

def euler_cromer():
    x = np.zeros(len(times))
    y = np.zeros(len(times))
    vx = np.zeros(len(times))
    vy = np.zeros(len(times))
    E = np.zeros(len(times))
    
    x[0], y[0] = x0, y0
    vx[0], vy[0] = vx0, vy0
    E[0] = energy(x[0], y[0], vx[0], vy[0])
    
    for n in range(len(times)-1):
        ax, ay = acceleration(x[n], y[n])
        
        vx[n+1] = vx[n] + h*ax
        vy[n+1] = vy[n] + h*ay
        
        x[n+1] = x[n] + h*vx[n+1]
        y[n+1] = y[n] + h*vy[n+1]
        
        E[n+1] = energy(x[n+1], y[n+1], vx[n+1], vy[n+1])
    
    return x, y, vx, vy, E

def backward_euler(iterations=20):
    x = np.zeros(len(times))
    y = np.zeros(len(times))
    vx = np.zeros(len(times))
    vy = np.zeros(len(times))
    E = np.zeros(len(times))
    
    x[0], y[0] = x0, y0
    vx[0], vy[0] = vx0, vy0
    E[0] = energy(x[0], y[0], vx[0], vy[0])
    
    for n in range(len(times)-1):
        
        # Initial guess using Euler-Cromer
        ax, ay = acceleration(x[n], y[n])
        vx_guess = vx[n] + h*ax
        vy_guess = vy[n] + h*ay
        x_guess = x[n] + h*vx_guess
        y_guess = y[n] + h*vy_guess
        
        # Fixed-point iteration for implicit Backward Euler
        for k in range(iterations):
            ax_guess, ay_guess = acceleration(x_guess, y_guess)
            
            vx_new = vx[n] + h*ax_guess
            vy_new = vy[n] + h*ay_guess
            
            x_new = x[n] + h*vx_new
            y_new = y[n] + h*vy_new
            
            x_guess = x_new
            y_guess = y_new
            vx_guess = vx_new
            vy_guess = vy_new
        
        x[n+1] = x_guess
        y[n+1] = y_guess
        vx[n+1] = vx_guess
        vy[n+1] = vy_guess
        
        E[n+1] = energy(x[n+1], y[n+1], vx[n+1], vy[n+1])
    
    return x, y, vx, vy, E

# Run simulations
x_fe, y_fe, vx_fe, vy_fe, E_fe = forward_euler()
x_ec, y_ec, vx_ec, vy_ec, E_ec = euler_cromer()
x_be, y_be, vx_be, vy_be, E_be = backward_euler()

# Plot trajectories
plt.figure(figsize=(8,8))

plt.plot(x_fe/1e8, y_fe/1e8, label="Forward Euler")
plt.plot(x_be/1e8, y_be/1e8, label="Backward Euler")
plt.plot(x_ec/1e8, y_ec/1e8, label="Euler-Cromer")

plt.plot(0, 0, "o", label="Earth")

plt.xlabel(r"$x$ position ($10^8$ m)")
plt.ylabel(r"$y$ position ($10^8$ m)")
plt.title("Moon Orbit Around Earth")
plt.axis("equal")
plt.grid(True)
plt.legend(loc="upper right")
plt.show()

# Plot energy behavior
plt.figure(figsize=(8,5))

plt.plot(times/(24*3600), (E_fe - E_fe[0])/abs(E_fe[0]), label="Forward Euler")
plt.plot(times/(24*3600), (E_be - E_be[0])/abs(E_be[0]), label="Backward Euler")
plt.plot(times/(24*3600), (E_ec - E_ec[0])/abs(E_ec[0]), label="Euler-Cromer")

plt.xlabel("Time (days)")
plt.ylabel("Relative energy change")
plt.title("Numerical Energy Behavior")
plt.grid(True)
plt.legend(loc="upper right")
plt.show()

# Print final distances
r_fe = np.sqrt(x_fe[-1]**2 + y_fe[-1]**2)
r_be = np.sqrt(x_be[-1]**2 + y_be[-1]**2)
r_ec = np.sqrt(x_ec[-1]**2 + y_ec[-1]**2)

print("Final Earth-Moon Distance")
print("-------------------------")
print(f"Forward Euler:   {r_fe/1e8:.4f} x 10^8 m")
print(f"Backward Euler:  {r_be/1e8:.4f} x 10^8 m")
print(f"Euler-Cromer:    {r_ec/1e8:.4f} x 10^8 m")
```
