# C1.0 Learning Outcomes — Numerical Methods and Finite Difference Techniques
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## Why this module matters

Many physical systems cannot be solved analytically. While exact solutions are elegant and powerful, real-world physics often involves:

- non-linear equations  
- coupled systems  
- drag forces  
- chaotic behavior  
- orbital dynamics  
- systems without closed-form solutions  

To study these systems, physicists rely on **numerical methods**.

In this module, we introduce the foundational finite difference methods used to transform differential equations into algebraic update rules that computers can solve step-by-step.

We will derive the basic finite difference approximations using Taylor series expansions and apply them to dynamical systems in physics.

The goal at this stage is not to develop advanced numerical algorithms, but to build intuition for:
- discretization  
- iterative solutions  
- numerical stability  
- computational modeling  

This module also serves as an introduction to **computational physics**, where physics, mathematics, and programming come together to study complex systems.

Later phases will extend these ideas to:
- higher-order numerical methods  
- coupled differential equations  
- partial differential equations  
- advanced simulations  
- computational research applications

By the end of this module, you will be able to:

---

### A. Finite Difference Approximations  
*Introduces the discretization of derivatives.*

- **A1.** Explain why numerical methods are needed in physics.  
- **A2.** Derive finite difference approximations using Taylor series expansions.  
- **A3.** State and apply:
  - forward difference  
  - backward difference  
  - central difference  

- **A4.** Recognize the role of timestep size in numerical approximations.  
- **A5.** Interpret finite differences as approximations to derivatives.  

---

### B. Euler-Type Numerical Methods  
*Develops iterative methods for solving differential equations.*

- **B1.** Derive the Explicit Euler method from the forward difference approximation.  

$$
y_{n+1}
=
y_n
+
h f(t_n,y_n)
$$

- **B2.** Derive the Implicit Euler method from the backward difference approximation.  

$$
y_{n+1}
=
y_n
+
h f(t_{n+1},y_{n+1})
$$

- **B3.** Apply the Central Difference approximation to estimate derivatives.  
- **B4.** State and apply the Euler-Cromer (Symplectic Euler) method.  
- **B5.** Compare explicit, implicit, and symplectic update schemes.  

---

### C. Numerical Solutions of Physical Systems  
*Applies numerical methods to physics problems.*

- **C1.** Numerically solve first-order differential equations.  
- **C2.** Numerically solve second-order equations of motion.  
- **C3.** Apply numerical methods to systems involving drag forces.  
- **C4.** Simulate orbital motion using iterative update equations.  
- **C5.** Compare numerical solutions with analytical solutions when available.  

---

### D. Stability, Accuracy, and Physical Interpretation  
*Builds intuition for numerical behavior.*

- **D1.** Explain the concept of numerical error and truncation error.  
- **D2.** Investigate how timestep size affects stability and accuracy.  
- **D3.** Recognize situations where numerical solutions become unstable.  
- **D4.** Explain why symplectic methods better preserve physical behavior in dynamical systems.  
- **D5.** Interpret numerical results using plots and physical reasoning.  

---

### E. Computational Physics Skills  
*Introduces computational modeling workflows.*

- **E1.** Implement numerical algorithms using Python.  
- **E2.** Use iterative loops to evolve physical systems in time.  
- **E3.** Generate plots of:
  - position vs time  
  - velocity vs time  
  - phase space trajectories  
  - orbital motion  

- **E4.** Analyze and interpret computational simulations.  
- **E5.** Recognize computational physics as an essential modern physics tool.  

---

### Performance Criteria

- Finite difference approximations are derived and applied correctly.  
- Numerical update equations are implemented correctly.  
- Physical systems are modeled using iterative numerical methods.  
- Numerical results are interpreted physically and computationally.  
- Students demonstrate understanding of numerical stability, timestep effects, and approximation error.
