# C4.1 Vector Transformation in Rotating Frames

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.1.1 Motivation
To describe motion from a rotating frame of reference, we need to understand how **vectors transform between reference frames**—specifically, how to relate time derivatives like velocity and acceleration in the inertial and rotating perspectives. This is the mathematical backbone that leads to fictitious forces in rotating systems.

Before diving into rotating frames, let’s quickly recall what happens in a linearly accelerating frame.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.1.2 Review: Vector Transformation in Linearly Accelerating Frames

In a linearly accelerating frame, the frame moves with acceleration $ \vec{a}_{\text{frame}} $ relative to an inertial frame. If we want to apply Newton’s second law inside this frame, we must introduce a **fictitious force** to account for the acceleration of the frame itself.

If $ \vec{a}_{\text{inertial}} $ is the true acceleration of a particle (from the inertial frame), and $ \vec{a}_{\text{non-inertial}} $ is the apparent acceleration as seen from the accelerating frame, the relation is:

$$
\vec{a}_{\text{inertial}} = \vec{a}_{\text{non-inertial}} + \vec{a}_{\text{frame}}
$$

Rewriting Newton’s second law from the perspective of the accelerating frame:

$$
m \vec{a}_{\text{non-inertial}} = \vec{F}_{\text{real}} - m \vec{a}_{\text{frame}}
$$

Here, $ -m \vec{a}_{\text{frame}} $ is the **fictitious force** that must be added to make the law work in the accelerating frame.

This concept generalizes to rotating frames, but the math becomes more intricate due to **changing direction** of the unit vectors in the rotational frame.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.1.3 Rotating Frames: A New Challenge

In rotating frames, the complication isn't just that the frame moves—it’s that the **orientation of the frame itself changes over time**. This means even if a vector's magnitude stays constant, its direction can change due to the frame's rotation.

To track how vectors like position, velocity, and acceleration change in rotating frames, we need to relate **time derivatives in the rotating frame** to **those in the inertial frame**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

### Time Derivatives in Rotating Frames

To understand motion from the perspective of a rotating observer, we need to carefully analyze how **vectors change over time** in a rotating reference frame. This involves comparing how quantities like velocity and acceleration appear in the inertial frame (fixed in space) versus a frame that is rotating with some angular velocity.

We begin by recalling a key result from previous module.


### Review from General Rigid Body Rotation

In the *General Rigid Body Rotation* module, we learned about the angular velocity vector $ \vec{\Omega} $, which describes the axis and rate of rotation of a rigid body.

One important result from that module was:

**If a vector is fixed in a rotating body (i.e. it rotates with the body), then its time derivative in the inertial frame is given by:**

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">

<strong>Key Result — Time Derivative of a Vector Fixed in a Rotating Frame</strong>

If a vector $\vec{A}$ is fixed relative to a rotating frame with angular velocity $\vec{\Omega}$, then its time derivative as observed from an inertial frame is

$$
\left( \frac{d\vec{A}}{dt} \right)_\text{inertial}
=
\vec{\Omega}\times\vec{A}.
$$

This result forms the foundation for deriving velocity and acceleration transformations in rotating reference frames and ultimately leads to the Coriolis and centrifugal forces.

</div>

This identity tells us that if you're observing a vector $ \vec{A} $ that stays fixed in the rotating frame (say, a direction marker attached to a spinning object), it will appear to rotate in the inertial frame. The direction and rate of that rotation is governed by the cross product with $ \vec{\Omega} $.

### Generalizing the Time Derivative

Now we want to generalize this idea to handle *any* vector $ \vec{A}(t) $ that may change over time, **even from the point of view of the rotating frame**.

Let’s define:

- $ \left( \frac{d\vec{A}}{dt} \right)_\text{inertial} $: the rate of change of $ \vec{A} $ as seen from an inertial frame.
- $ \left( \frac{d\vec{A}}{dt} \right)_\text{rot} $: the rate of change of $ \vec{A} $ as seen from the rotating frame (i.e. how a rotating observer sees it changing).

NOTE: The vector $\vec{A}$ is instantaneously the same in both reference frames, but its time derivative is not.

The full identity that connects these two views is:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">

<strong>Key Result — Time Derivative Transformation Between Frames</strong>

For any vector quantity $\vec{A}$ observed from a rotating frame with angular velocity $\vec{\Omega}$,

$$
\left( \frac{d\vec{A}}{dt} \right)_\text{inertial}
=
\left( \frac{d\vec{A}}{dt} \right)_\text{rot}
+
\vec{\Omega}\times\vec{A}.
$$

This fundamental relation connects the rate of change of a vector measured in an inertial frame to the rate of change measured by an observer in a rotating frame. It serves as the starting point for deriving the velocity and acceleration transformations that ultimately lead to the centrifugal and Coriolis forces.

</div>

This equation has two parts:

1. **The actual rate of change** of the vector in the rotating frame, $ \left( \frac{d\vec{A}}{dt} \right)_\text{rot} $.
2. **A correction term**, $ \vec{\Omega} \times \vec{A} $, that accounts for the rotation of the frame itself.

The second term is exactly what we saw in the rigid body module: it tells us how the *rotation of the coordinate system* contributes to the change in direction of any vector, even if the vector’s magnitude isn’t changing.

#### Why This Matters

This result is incredibly powerful because it applies to **any vector quantity**:

- Position vector $ \vec{r} $
- Velocity vector $ \vec{v} $
- Momentum, angular momentum, force vectors, etc.

It allows us to compare how these vectors behave when described from:

- A "stationary" point of view (inertial frame)
- A "spinning" point of view (rotating frame)

Let’s now apply this identity to the position vector $ \vec{r}(t) $.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.1.4 Velocity in a Rotating Frame

Let $ \vec{r} $ be the position of a particle. Then the velocity in the inertial frame is:

$$
\vec{v}_{\text{inertial}} = \left( \frac{d\vec{r}}{dt} \right)_\text{inertial}
$$

Using the identity above:

$$
\vec{v}_{\text{inertial}} = \left( \frac{d\vec{r}}{dt} \right)_\text{rot} + \vec{\Omega} \times \vec{r}
$$

So the velocity in the inertial frame is composed of:

- The velocity as seen in the rotating frame
- Plus an additional contribution from the motion of the rotating coordinate system

The second term, $ \vec{\Omega} \times \vec{r} $, is sometimes called the "rotation-induced velocity."

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.1.5 Acceleration in a Rotating Frame

We can go one step further and take the time derivative again to get acceleration. Doing that carefully, we get:

$$
\vec{a}_{\text{inertial}} = \vec{a}_{\text{rot}} + 2 \vec{\Omega} \times \vec{v}_{\text{rot}} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})
$$

This includes:

- The acceleration seen in the rotating frame
- A Coriolis term: $ 2 \vec{\Omega} \times \vec{v}_{\text{rot}} $
- A centrifugal term: $ \vec{\Omega} \times (\vec{\Omega} \times \vec{r}) $

We’ll explore these in depth in the next part.

Starting with

$$
\vec{v}_{\text{inertial}}
=
\vec{v}_{\text{rot}}
+
\vec{\Omega}\times\vec{r},
$$

we take the inertial-frame derivative:

$$
\vec{a}_{\text{inertial}}
=
\left(\frac{d\vec{v}_{\text{rot}}}{dt}\right)_{\text{inertial}}
+
\left(\frac{d}{dt}(\vec{\Omega}\times\vec{r})\right)_{\text{inertial}}.
$$

Applying the derivative transformation rule to the first term gives

$$
\left(\frac{d\vec{v}_{\text{rot}}}{dt}\right)_{\text{inertial}}
=
\left(\frac{d\vec{v}_{\text{rot}}}{dt}\right)_{\text{rot}}
+
\vec{\Omega}\times\vec{v}_{\text{rot}}
=
\vec{a}_{\text{rot}}
+
\vec{\Omega}\times\vec{v}_{\text{rot}}.
$$

For the second term,

$$
\left(\frac{d}{dt}(\vec{\Omega}\times\vec{r})\right)_{\text{inertial}}
=
\vec{\Omega}\times
\left(\frac{d\vec{r}}{dt}\right)_{\text{inertial}},
$$

because $\vec{\Omega}$ is constant.

Using the velocity transformation again,

$$
\left(\frac{d\vec{r}}{dt}\right)_{\text{inertial}}
=
\vec{v}_{\text{rot}}
+
\vec{\Omega}\times\vec{r},
$$

so

$$
\left(\frac{d}{dt}(\vec{\Omega}\times\vec{r})\right)_{\text{inertial}}
=
\vec{\Omega}\times\vec{v}_{\text{rot}}
+
\vec{\Omega}\times(\vec{\Omega}\times\vec{r}).
$$

Combining all terms gives

$$
\boxed{
\vec{a}_{\text{inertial}}
=
\vec{a}_{\text{rot}}
+
2\vec{\Omega}\times\vec{v}_{\text{rot}}
+
\vec{\Omega}\times(\vec{\Omega}\times\vec{r})
}
$$

which is the acceleration transformation between inertial and rotating frames.

## C4.1.6 Summary of Key Concepts

- **The angular velocity vector $ \vec{\Omega} $** captures how the rotating frame moves relative to the inertial frame.
- **Time derivatives of vectors** in rotating frames are not the same as in inertial frames, even if the vector is fixed in the rotating frame.
- The formula:

  $$
  \left( \frac{d\vec{A}}{dt} \right)_\text{inertial} = \left( \frac{d\vec{A}}{dt} \right)_\text{rot} + \vec{\Omega} \times \vec{A}
  $$
  
  gives us a precise way to relate changes seen from both frames.
- This forms the foundation for understanding how Newton’s laws must be modified when applied inside a rotating system.
