## Problem 1 - The Rotating Carnival Ride

A carnival ride consists of a cylindrical chamber of radius

$$
R=5.0~\text{m}
$$

rotating with a constant angular speed

$$
\Omega = 2.0~\text{rad/s}.
$$

A rider of mass

$$
m=70~\text{kg}
$$

stands against the wall of the ride.

Ignoring gravity, determine the force exerted by the wall on the rider using two different reference frames:

1. An inertial observer standing on the ground.
2. An observer rotating with the ride.

Show that both observers predict the same force, even though they use different forms of Newton's second law.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Inertial Observer</strong>

Let $\hat r$ point radially outward from the center of the ride. Since the rider remains pressed against the wall, the circular wall imposes the constraint

$$
r=R.
$$

Therefore,

$$
\dot r = 0,
\qquad
\ddot r = 0.
$$

In cylindrical coordinates, the acceleration is

$$
\vec a
=
(\ddot r-r\dot\theta^2)\hat r
+
(r\ddot\theta+2\dot r\dot\theta)\hat\theta.
$$

The ride rotates at constant angular speed, so

$$
\dot\theta = \Omega,
\qquad
\ddot\theta = 0.
$$

Using the constraint $r=R$, the acceleration becomes

$$
\vec a
=
-R\Omega^2\hat r.
$$

The wall exerts an inward normal force on the rider, so

$$
\vec F_N = -F_N\hat r.
$$

Newton's second law gives

$$
-F_N\hat r
=
m(-R\Omega^2\hat r).
$$

Therefore,

$$
F_N = mR\Omega^2.
$$

Substituting the values,

$$
F_N
=
(70)(5.0)(2.0)^2
=
1400~\text{N}.
$$

Thus, the wall exerts a force of magnitude

$$
\boxed{F_N=1400~\text{N}}
$$

directed toward the center of the ride.

<br>

<strong>Rotating Observer</strong>

In the frame rotating with the ride, the rider is stationary relative to the wall. Therefore,

$$
\vec a_{\text{rot}}=0,
\qquad
\vec v_{\text{rot}}=0.
$$

Since $\vec v_{\text{rot}}=0$, there is no Coriolis force.

Newton's second law in the rotating frame becomes

$$
m\vec a_{\text{rot}}
=
\vec F_N
-
m\vec\Omega\times(\vec\Omega\times\vec r).
$$

Since the rider is at rest in the rotating frame,

$$
0
=
\vec F_N
-
m\vec\Omega\times(\vec\Omega\times\vec r).
$$

For a point at radius $R$, the centrifugal term points outward:

$$
-m\vec\Omega\times(\vec\Omega\times\vec r)
=
m\Omega^2R\hat r.
$$

Therefore,

$$
0
=
-F_N\hat r
+
m\Omega^2R\hat r.
$$

So,

$$
F_N = m\Omega^2R.
$$

Substituting the values,

$$
F_N
=
(70)(2.0)^2(5.0)
=
1400~\text{N}.
$$

Thus,

$$
\boxed{F_N=1400~\text{N}}.
$$

<br>

<strong>Comparison</strong>

Both observers predict the same physical force exerted by the wall:

$$
\boxed{F_N=1400~\text{N}}.
$$

The inertial observer finds this force by applying Newton's second law after using the cylindrical-coordinate constraint $r=R$, which gives

$$
\vec a = -R\Omega^2\hat r.
$$

The rotating observer sees the rider at rest and introduces the outward centrifugal force,

$$
\vec F_{\text{centrifugal}}
=
-m\vec\Omega\times(\vec\Omega\times\vec r),
$$

which is balanced by the inward normal force from the wall.

Thus, the two descriptions are mathematically equivalent.

</div>

## Problem 2 - High-Speed Train on a Curved Track

A high-speed train travels around a level circular curve of radius

$$
R = 1200~\text{m}
$$

at a constant speed

$$
v = 70.0~\text{m/s}.
$$

A passenger car has mass

$$
m = 4.0\times 10^4~\text{kg}.
$$

Assume the track is level and that the train follows the circular constraint

$$
r = R.
$$

Let $\hat{r}$ point outward from the center of the curve. The rails exert an inward radial force on the train, so

$$
\vec{F}_{\text{rail}} = -F_{\text{rail}}\hat{r}.
$$

1. Analyze the radial motion from an inertial frame using cylindrical coordinates.
2. Analyze the same situation from the rotating frame of the train.
3. Explain why the outside rail may experience greater wear on a curve.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Inertial Observer</strong>

In cylindrical coordinates, the acceleration is

$$
\vec{a}
=
(\ddot r-r\dot\theta^2)\hat r
+
(r\ddot\theta+2\dot r\dot\theta)\hat\theta.
$$

The circular track imposes the constraint

$$
r=R.
$$

Therefore,

$$
\dot r = 0,
\qquad
\ddot r = 0.
$$

The train moves with constant speed

$$
v = R\dot\theta,
$$

so

$$
\dot\theta = \frac{v}{R}.
$$

The radial acceleration becomes

$$
a_r
=
-r\dot\theta^2
=
-R\left(\frac{v}{R}\right)^2
=
-\frac{v^2}{R}.
$$

Thus,

$$
\vec a
=
-\frac{v^2}{R}\hat r.
$$

Newton's second law in the radial direction gives

$$
-F_{\text{rail}}
=
m\left(-\frac{v^2}{R}\right).
$$

Therefore,

$$
F_{\text{rail}}
=
m\frac{v^2}{R}.
$$

Substituting the values,

$$
F_{\text{rail}}
=
(4.0\times 10^4)
\frac{(70.0)^2}{1200}
=
1.63\times 10^5~\text{N}.
$$

So the rails exert an inward radial force of magnitude

$$
\boxed{
F_{\text{rail}} = 1.63\times 10^5~\text{N}
}.
$$

<br>

<strong>Rotating Observer</strong>

Now analyze the train from the frame rotating with angular speed

$$
\Omega = \dot\theta = \frac{v}{R}.
$$

In this frame, the train is at rest relative to the track, so

$$
\vec a_{\text{rot}} = 0,
\qquad
\vec v_{\text{rot}} = 0.
$$

Because $\vec v_{\text{rot}}=0$, there is no Coriolis force.

Newton's second law in the rotating frame is

$$
m\vec a_{\text{rot}}
=
\vec F_{\text{rail}}
-
m\vec\Omega\times(\vec\Omega\times\vec r).
$$

The centrifugal term points outward. For circular motion,

$$
-m\vec\Omega\times(\vec\Omega\times\vec r)
=
m\Omega^2R\hat r.
$$

Therefore,

$$
0
=
-F_{\text{rail}}\hat r
+
m\Omega^2R\hat r.
$$

So

$$
F_{\text{rail}}
=
m\Omega^2R.
$$

Since

$$
\Omega = \frac{v}{R},
$$

we get

$$
F_{\text{rail}}
=
m\left(\frac{v}{R}\right)^2R
=
m\frac{v^2}{R}.
$$

Thus,

$$
\boxed{
F_{\text{rail}} = 1.63\times 10^5~\text{N}
}.
$$

<br>

<strong>Physical Interpretation</strong>

Both descriptions give the same physical rail force.

The inertial observer describes the train as accelerating inward because the circular constraint forces

$$
\vec a
=
-\frac{v^2}{R}\hat r.
$$

The rotating observer describes the train as stationary, but must introduce an outward centrifugal force. The rail force balances this fictitious force.

In a real track system, the sideways force is transmitted through the wheel-rail contact. On a curve, this can increase lateral contact and loading on one rail, especially the outside rail when the train speed is high relative to the curve design. This increased loading contributes to greater wear and tear.

</div>

</details>

## Problem 3 - Coriolis Force on a Japanese Maglev Train

A Japanese SCMaglev test train travels eastward along a nearly east-west section of track in Yamanashi, Japan. Approximate the latitude as

$$
\phi = 35.5^\circ
$$

and the train speed as

$$
v = 603~\text{km/h}.
$$

The angular speed of Earth is

$$
\Omega_E = 7.29\times 10^{-5}~\text{rad/s}.
$$

Assume the train has mass

$$
m = 4.0\times 10^4~\text{kg}.
$$

Find the Coriolis acceleration and Coriolis force on the train.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Convert the Speed</strong>

$$
v
=
603~\text{km/h}
=
603\left(\frac{1000}{3600}\right)
=
167.5~\text{m/s}.
$$

<br>

<strong>Coriolis Acceleration</strong>

The Coriolis acceleration is

$$
\vec a_{\text{Cor}}
=
-2\vec\Omega_E\times\vec v.
$$

At latitude $\phi$, for motion due east, the magnitude is

$$
a_{\text{Cor}}
=
2\Omega_E v.
$$

Substituting,

$$
a_{\text{Cor}}
=
2(7.29\times 10^{-5})(167.5)
=
2.44\times 10^{-2}~\text{m/s}^2.
$$

So,

$$
\boxed{
a_{\text{Cor}} = 2.44\times 10^{-2}~\text{m/s}^2
}
$$

This is only about

$$
\frac{a_{\text{Cor}}}{g}
=
\frac{2.44\times 10^{-2}}{9.81}
=
2.49\times 10^{-3},
$$

or about

$$
\boxed{0.25\% \text{ of } g}.
$$

<br>

<strong>Direction of the Coriolis Acceleration</strong>

Using local east-north-up coordinates, Earth's angular velocity can be written as

$$
\vec\Omega_E
=
\Omega_E\cos\phi\,\hat N
+
\Omega_E\sin\phi\,\hat U.
$$

For eastward motion,

$$
\vec v = v\hat E.
$$

Therefore,

$$
\vec a_{\text{Cor}}
=
2\Omega_E v\cos\phi\,\hat U
-
2\Omega_E v\sin\phi\,\hat N.
$$

So the Coriolis acceleration has:

- an upward component,
- a southward component.

The vertical component is

$$
a_U
=
2\Omega_E v\cos\phi
=
(2.44\times 10^{-2})\cos(35.5^\circ)
=
1.99\times 10^{-2}~\text{m/s}^2.
$$

The horizontal southward component is

$$
a_S
=
2\Omega_E v\sin\phi
=
(2.44\times 10^{-2})\sin(35.5^\circ)
=
1.42\times 10^{-2}~\text{m/s}^2.
$$

Thus,

$$
\boxed{
\vec a_{\text{Cor}}
=
(1.99\times 10^{-2})\hat U
-
(1.42\times 10^{-2})\hat N
}
$$

in $\text{m/s}^2$.

<br>

<strong>Coriolis Force</strong>

The Coriolis force is

$$
\vec F_{\text{Cor}}
=
m\vec a_{\text{Cor}}.
$$

The total magnitude is

$$
F_{\text{Cor}}
=
(4.0\times 10^4)(2.44\times 10^{-2})
=
9.76\times 10^2~\text{N}.
$$

So,

$$
\boxed{
F_{\text{Cor}} \approx 9.8\times 10^2~\text{N}
}
$$

The components are

$$
F_U
=
(4.0\times 10^4)(1.99\times 10^{-2})
=
7.96\times 10^2~\text{N},
$$

and

$$
F_S
=
(4.0\times 10^4)(1.42\times 10^{-2})
=
5.68\times 10^2~\text{N}.
$$

Therefore, for eastward motion,

$$
\boxed{
\vec F_{\text{Cor}}
=
(7.96\times 10^2)\hat U
-
(5.68\times 10^2)\hat N
}
$$

in newtons.

<br>

<strong>Physical Interpretation</strong>

For an eastward-moving train in Japan, the Coriolis force has an upward component and a southward component.

If the train were moving westward instead, the Coriolis force would reverse direction: it would point downward and northward.

The effect is small compared with the train's weight, but at very high speeds it is not mathematically negligible.

</div>

</details>

## Problem 4 - Coriolis Deflection of a High-Speed Projectile

A projectile is fired due north at speed

$$
v = 900~\text{m/s}
$$

toward a target located

$$
L = 1000~\text{m}
$$

away. The projectile is fired at latitude

$$
\phi = 40^\circ.
$$

Ignore air resistance, gravity drop, wind, and spin effects. Estimate the sideways Coriolis deflection before the projectile reaches the target.

Use

$$
\Omega_E = 7.29\times 10^{-5}~\text{rad/s}.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Time of Flight</strong>

Ignoring air resistance, the travel time is approximately

$$
t = \frac{L}{v}.
$$

Thus,

$$
t
=
\frac{1000}{900}
=
1.11~\text{s}.
$$

<br>

<strong>Coriolis Acceleration</strong>

For horizontal motion due north, the horizontal Coriolis acceleration has magnitude

$$
a_{\text{Cor}}
=
2\Omega_E v \sin\phi.
$$

Substituting,

$$
a_{\text{Cor}}
=
2(7.29\times 10^{-5})(900)\sin(40^\circ)
=
8.44\times 10^{-2}~\text{m/s}^2.
$$

In the Northern Hemisphere, a northward-moving projectile is deflected to the right, which means eastward.

<br>

<strong>Sideways Deflection</strong>

Assuming the Coriolis acceleration is approximately constant over the short flight time,

$$
\Delta x
=
\frac{1}{2}a_{\text{Cor}}t^2.
$$

Therefore,

$$
\Delta x
=
\frac{1}{2}
(8.44\times 10^{-2})
(1.11)^2
=
5.2\times 10^{-2}~\text{m}.
$$

So the deflection is approximately

$$
\boxed{
\Delta x \approx 5.2~\text{cm}
}
$$

to the east.

<br>

<strong>Interpretation</strong>

For a $1~\text{km}$ flight at $900~\text{m/s}$, Earth’s rotation produces a sideways deflection of only a few centimeters in this simplified model.

This estimate ignores effects that are often larger in real projectile motion, such as air resistance, wind, projectile spin, and vertical drop.

</div>

</details>

## Problem 5 - Coriolis Force and Hurricanes

A parcel of air in the Northern Hemisphere moves northward toward a low-pressure region. Suppose the air parcel has speed

$$
v = 25.0~\text{m/s}
$$

at latitude

$$
\phi = 25^\circ.
$$

The angular speed of Earth is

$$
\Omega_E = 7.29\times 10^{-5}~\text{rad/s}.
$$

Assume the air parcel has mass

$$
m = 1.00~\text{kg}.
$$

1. Find the Coriolis acceleration acting on the air parcel.
2. Find the Coriolis force acting on the air parcel.
3. Determine the direction of the Coriolis force.
4. Explain why hurricanes rotate counterclockwise in the Northern Hemisphere.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Coriolis Acceleration</strong>

For horizontal motion on Earth, the horizontal Coriolis acceleration has magnitude

$$
a_{\text{Cor}}
=
2\Omega_E v\sin\phi.
$$

Substituting the values,

$$
a_{\text{Cor}}
=
2(7.29\times 10^{-5})(25.0)\sin(25^\circ).
$$

Therefore,

$$
a_{\text{Cor}}
=
1.54\times 10^{-3}~\text{m/s}^2.
$$

So,

$$
\boxed{
a_{\text{Cor}}
=
1.54\times 10^{-3}~\text{m/s}^2
}
$$

<br>

<strong>Coriolis Force</strong>

The Coriolis force is

$$
F_{\text{Cor}}
=
ma_{\text{Cor}}.
$$

Since

$$
m=1.00~\text{kg},
$$

we get

$$
F_{\text{Cor}}
=
(1.00)(1.54\times 10^{-3})
=
1.54\times 10^{-3}~\text{N}.
$$

Thus,

$$
\boxed{
F_{\text{Cor}}
=
1.54\times 10^{-3}~\text{N}
}
$$

<br>

<strong>Direction</strong>

In the Northern Hemisphere, the Coriolis force deflects moving objects to the right of their motion.

Since the air parcel is moving northward, its right-hand side points eastward. Therefore, the Coriolis force points eastward.

$$
\boxed{
\text{Northward-moving air is deflected eastward.}
}
$$

<br>

<strong>Connection to Hurricanes</strong>

Air near the surface tends to flow inward toward a low-pressure region. However, as the air moves inward, the Coriolis force deflects it to the right in the Northern Hemisphere.

This means the air does not move straight into the low-pressure center. Instead, it curves around the center.

For a low-pressure system in the Northern Hemisphere, this rightward deflection produces a counterclockwise circulation.

In the Southern Hemisphere, the Coriolis force deflects motion to the left, so hurricanes rotate clockwise.

## Problem 6 - The Eötvös Effect: Flying East and West

A commercial airplane flies at constant speed

$$
v = 250~\text{m/s}
$$

at latitude

$$
\phi = 45^\circ.
$$

The airplane has mass

$$
m = 7.0\times 10^4~\text{kg}.
$$

Use Earth's angular speed

$$
\Omega_E = 7.29\times 10^{-5}~\text{rad/s}.
$$

In Earth's rotating frame, the airplane experiences a Coriolis force because it moves relative to Earth's surface.

1. Find the vertical component of the Coriolis force during eastward flight.
2. Determine the apparent weight of the airplane during eastward flight.
3. Repeat the calculation for westward flight.
4. Compare the apparent weight for eastward and westward motion.
5. Explain why this effect is called an apparent-weight effect.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Curiosity — The Eötvös Effect and Earth's Rotation</strong>

The Eötvös effect is named after the Hungarian physicist <strong>Loránd Eötvös</strong> (1848–1919), who studied extremely precise measurements of gravity. While analyzing gravity measurements made aboard moving ships, Eötvös discovered that the measured gravitational acceleration depended slightly on the direction of travel. Eastward-moving ships appeared slightly lighter, while westward-moving ships appeared slightly heavier.

Today we understand this effect as a consequence of the Coriolis force. For eastward motion, the Coriolis force has an upward component that slightly reduces apparent weight. For westward motion, the Coriolis force has a downward component that slightly increases apparent weight.

The same physics appears in many other situations:

<ul>

<li><strong>Aircraft:</strong> Eastward-flying aircraft are slightly lighter than westward-flying aircraft because the vertical component of the Coriolis force changes the required lift force.</li>

<li><strong>Ballistics:</strong> Eastward-moving projectiles experience a small upward Coriolis force and therefore drop slightly less than expected. Westward-moving projectiles experience a downward Coriolis force and therefore drop slightly more than expected. Long-range projectiles also experience horizontal Coriolis deflections.</li>

<li><strong>Rocket Launches:</strong> Rockets are typically launched eastward whenever possible. The primary benefit is that Earth's rotation provides a free eastward velocity boost before the engines even ignite. In addition, eastward motion produces a small upward Coriolis contribution that slightly reduces the required lift force.</li>

<li><strong>Spacecraft Navigation:</strong> Precise trajectory calculations often include corrections for Earth's rotation and the associated fictitious forces when working in Earth-fixed coordinates.</li>

<li><strong>Meteorology and Oceanography:</strong> The same Coriolis physics responsible for the Eötvös effect also contributes to the motion of hurricanes, large-scale weather systems, and ocean currents.</li>

</ul>

Although the Eötvös effect is usually small, it provides a striking example of how Earth's rotation can subtly influence the motion, trajectories, and apparent weight of moving objects.

</div>

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Step 1: Write Earth's angular velocity in local coordinates</strong>

Use local unit vectors:

- $\hat E$ points east,
- $\hat N$ points north,
- $\hat U$ points upward.

At latitude $\phi$, Earth's angular velocity is

$$
\vec\Omega_E
=
\Omega_E\cos\phi\,\hat N
+
\Omega_E\sin\phi\,\hat U.
$$

For eastward flight,

$$
\vec v_{\text{rot}}
=
v\hat E.
$$

<br>

<strong>Step 2: Calculate the Coriolis force for eastward flight</strong>

The Coriolis force is

$$
\vec F_{\text{Cor}}
=
-2m\vec\Omega_E\times\vec v_{\text{rot}}.
$$

Substituting,

$$
\vec F_{\text{Cor}}
=
-2m
\left(
\Omega_E\cos\phi\,\hat N
+
\Omega_E\sin\phi\,\hat U
\right)
\times
(v\hat E).
$$

Using

$$
\hat N\times\hat E=-\hat U,
\qquad
\hat U\times\hat E=\hat N,
$$

we obtain

$$
\vec F_{\text{Cor}}
=
2m\Omega_E v\cos\phi\,\hat U
-
2m\Omega_E v\sin\phi\,\hat N.
$$

The vertical component is therefore

$$
F_{\text{Cor},U}
=
2m\Omega_E v\cos\phi.
$$

Substituting the values,

$$
F_{\text{Cor},U}
=
2(7.0\times10^4)(7.29\times10^{-5})(250)\cos(45^\circ)
=
1.80\times10^3~\text{N}.
$$

Thus, during eastward flight,

$$
\boxed{
F_{\text{Cor},U}
=
1.80\times10^3~\text{N upward}
}
$$

<br>

<strong>Step 3: Apparent weight during eastward flight</strong>

The usual gravitational weight is

$$
mg
=
(7.0\times10^4)(9.81)
=
6.87\times10^5~\text{N}.
$$

During level eastward flight, the upward support force needed from lift is reduced because the Coriolis force has an upward component.

Thus,

$$
N_{\text{east}}
=
mg
-
2m\Omega_E v\cos\phi.
$$

Numerically,

$$
N_{\text{east}}
=
6.87\times10^5
-
1.80\times10^3
=
6.85\times10^5~\text{N}.
$$

Therefore,

$$
\boxed{
N_{\text{east}}
=
6.85\times10^5~\text{N}
}
$$

<br>

<strong>Step 4: Westward flight</strong>

For westward flight,

$$
\vec v_{\text{rot}}
=
-v\hat E.
$$

The Coriolis force reverses direction. The vertical component now points downward:

$$
F_{\text{Cor},U}
=
-1.80\times10^3~\text{N}.
$$

The required support force is therefore larger:

$$
N_{\text{west}}
=
mg
+
2m\Omega_E v\cos\phi.
$$

Numerically,

$$
N_{\text{west}}
=
6.87\times10^5
+
1.80\times10^3
=
6.89\times10^5~\text{N}.
$$

Thus,

$$
\boxed{
N_{\text{west}}
=
6.89\times10^5~\text{N}
}
$$

<br>

<strong>Step 5: Compare the apparent weights</strong>

The difference between westward and eastward apparent weight is

$$
N_{\text{west}}-N_{\text{east}}
=
2(1.80\times10^3)
=
3.60\times10^3~\text{N}.
$$

The fractional change for one direction relative to $mg$ is

$$
\frac{1.80\times10^3}{6.87\times10^5}
=
2.62\times10^{-3}.
$$

Therefore, the eastward or westward change is approximately

$$
\boxed{
0.262\%
}
$$

of the airplane's usual weight.

<br>

<strong>Step 6: Physical interpretation</strong>

For eastward flight, the airplane moves in the same direction as Earth's rotation. In Earth's rotating frame, the Coriolis force has an upward component, so the airplane is effectively slightly lighter.

For westward flight, the airplane moves opposite Earth's rotation. The Coriolis force has a downward component, so the airplane is effectively slightly heavier.

This is the Eötvös effect: an east-west motion relative to Earth changes the apparent weight of an object because of Earth's rotation.

</div>

</details>

## Problem 7 - Dropping a Ball from a Skyscraper

A ball is dropped from rest from the top of a skyscraper of height

$$
h = 300~\text{m}.
$$

The skyscraper is located at latitude

$$
\phi = 40^\circ.
$$

Ignore air resistance.

1. Determine the time required for the ball to reach the ground.
2. Using the Coriolis force, estimate the east-west deflection of the ball during its fall.
3. Determine whether the ball lands east or west of the base of the building.

Use

$$
g = 9.81~\text{m/s}^2,
\qquad
\Omega_E = 7.29\times10^{-5}~\text{rad/s}.
$$

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Time of Fall</strong>

The ball is released from rest, so

$$
h=\frac{1}{2}gt^2.
$$

Solving for the fall time,

$$
t
=
\sqrt{\frac{2h}{g}}
=
\sqrt{\frac{2(300)}{9.81}}
=
7.82~\text{s}.
$$

Thus,

$$
\boxed{
t = 7.82~\text{s}
}
$$

<br>

<strong>Coriolis Acceleration</strong>

Choose local coordinates:

- $\hat E$ = east
- $\hat N$ = north
- $\hat U$ = upward

Earth's angular velocity is

$$
\vec\Omega_E
=
\Omega_E\cos\phi\,\hat N
+
\Omega_E\sin\phi\,\hat U.
$$

The velocity of the falling ball is

$$
\vec v
=
-g t\,\hat U.
$$

The Coriolis acceleration is

$$
\vec a_{\rm Cor}
=
-2\vec\Omega_E\times\vec v.
$$

Substituting,

$$
\vec a_{\rm Cor}
=
2\Omega_E g t\cos\phi\,\hat E.
$$

Thus, the Coriolis acceleration points eastward.

<br>

<strong>Eastward Velocity</strong>

Integrating the acceleration,

$$
v_E
=
\int_0^t
2\Omega_E g t'\cos\phi\,dt'
=
\Omega_E g t^2\cos\phi.
$$

<br>

<strong>Eastward Deflection</strong>

Integrating once more,

$$
x_E
=
\int_0^t
\Omega_E g t'^2\cos\phi\,dt'
=
\frac{1}{3}
\Omega_E g t^3\cos\phi.
$$

Substituting the values,

$$
x_E
=
\frac{1}{3}
(7.29\times10^{-5})
(9.81)
(7.82)^3
\cos(40^\circ).
$$

This gives

$$
x_E
=
8.8\times10^{-2}\ \text{m}.
$$

Therefore,

$$
\boxed{
x_E \approx 8.8~\text{cm}
}
$$

<br>

<strong>Direction of Deflection</strong>

Since the Coriolis acceleration points eastward throughout the fall,

$$
\boxed{
\text{The ball lands east of the base of the building.}
}
$$

<br>

<strong>Physical Interpretation</strong>

As the ball falls, it retains the larger eastward speed it possessed at the top of the skyscraper. The ground below has a slightly smaller eastward speed because it is closer to Earth's rotation axis.

Consequently, the ball moves slightly ahead of the ground beneath it and lands east of the point directly below where it was released.

Historically, experiments of this type were among the first direct demonstrations that Earth rotates.

</div>

</details>

## Problem 8 - Walking Inside a Rotating Space Station

A rotating space station has radius

$$
R = 100~\text{m}
$$

and rotates with constant angular speed

$$
\Omega = 0.30~\text{rad/s}.
$$

An astronaut of mass

$$
m = 75~\text{kg}
$$

walks radially inward, from the outer rim toward the center, with speed relative to the station

$$
v_{\text{rot}} = 1.50~\text{m/s}.
$$

Assume the astronaut is initially near the outer rim.

1. Find the centrifugal force acting on the astronaut at $r=R$.
2. Find the Coriolis force acting on the astronaut while walking inward.
3. Determine the direction of each fictitious force.
4. Explain why walking straight toward the center would feel difficult.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Centrifugal Force</strong>

The centrifugal force in the rotating frame is

$$
\vec F_{\text{cent}}
=
-m\vec\Omega\times(\vec\Omega\times\vec r).
$$

At radius $r=R$, its magnitude is

$$
F_{\text{cent}}
=
m\Omega^2R.
$$

Substituting the values,

$$
F_{\text{cent}}
=
(75)(0.30)^2(100)
=
675~\text{N}.
$$

Therefore,

$$
\boxed{
F_{\text{cent}} = 675~\text{N}
}
$$

directed radially outward.

<br>

<strong>Coriolis Force</strong>

The Coriolis force in the rotating frame is

$$
\vec F_{\text{Cor}}
=
-2m\vec\Omega\times\vec v_{\text{rot}}.
$$

The astronaut walks radially inward, so

$$
\vec v_{\text{rot}}
=
-v_{\text{rot}}\hat r.
$$

Let the station rotate counterclockwise as viewed from above, so

$$
\vec\Omega = \Omega \hat z.
$$

Then

$$
\vec F_{\text{Cor}}
=
-2m(\Omega\hat z)\times(-v_{\text{rot}}\hat r).
$$

Since

$$
\hat z\times\hat r = \hat\theta,
$$

we get

$$
\vec F_{\text{Cor}}
=
2m\Omega v_{\text{rot}}\hat\theta.
$$

The magnitude is

$$
F_{\text{Cor}}
=
2m\Omega v_{\text{rot}}.
$$

Substituting,

$$
F_{\text{Cor}}
=
2(75)(0.30)(1.50)
=
67.5~\text{N}.
$$

Therefore,

$$
\boxed{
F_{\text{Cor}} = 67.5~\text{N}
}
$$

in the $+\hat\theta$ direction.

<br>

<strong>Physical Interpretation</strong>

The Coriolis force points sideways, perpendicular to the astronaut's radial motion. For inward walking in a counterclockwise-rotating station, the Coriolis force points in the $+\hat\theta$ direction.

The centrifugal force acts like artificial gravity, pressing the astronaut against the outer hull of the station, which serves as the floor. At the station rim, this force has magnitude

$$
675~\text{N},
$$

which is comparable to the astronaut's weight on Earth:

$$
mg = (75)(9.81)=736~\text{N}.
$$

While walking radially inward, the astronaut also experiences a Coriolis force directed perpendicular to the direction of motion. This force does not push the astronaut toward or away from the floor; instead, it acts sideways along the station's circumference.

Thus, the astronaut would feel:

- a centrifugal force pressing them against the floor,
- a sideways Coriolis force while walking toward the center of the station.

As a result, walking directly toward the hub would feel unnatural. The astronaut would experience a continual sideways deflection and would need to rely on friction and body adjustments to maintain a straight radial path.

</div>

</details>

## Problem 9 - Foucault Pendulum and the Coriolis Force

A Foucault pendulum is placed at latitude

$$
\phi = 40^\circ.
$$

The angular speed of Earth is

$$
\Omega_E = 7.29\times 10^{-5}~\text{rad/s}.
$$

The pendulum swings back and forth while Earth rotates underneath it.

In Earth's rotating frame, the pendulum bob experiences the Coriolis force

$$
\vec F_{\text{Cor}}
=
-2m\vec\Omega_E\times\vec v_{\text{rot}}.
$$

1. Explain why the Coriolis force is responsible for the slow rotation of the pendulum's swing plane.
2. Show why the relevant angular speed is the vertical component of Earth's angular velocity,

$$
\Omega_E\sin\phi.
$$

3. Find the angular precession rate of the pendulum's swing plane.
4. Determine the time required for the swing plane to complete one full rotation relative to the ground.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Step 1: Start with the Coriolis force</strong>

In Earth's rotating frame, the pendulum bob experiences the Coriolis force

$$
\vec F_{\text{Cor}}
=
-2m\vec\Omega_E\times\vec v_{\text{rot}}.
$$

This force is always perpendicular to the bob's velocity. Therefore, it does not primarily change the speed of the bob. Instead, it slowly changes the direction of the bob's motion.

For a pendulum, that slow sideways deflection causes the plane of oscillation to rotate relative to the ground.

<br>

<strong>Step 2: Decompose Earth's angular velocity</strong>

Use local unit vectors:

- $\hat U$ points upward along the local vertical direction.
- $\hat N$ points horizontally northward.

At latitude $\phi$, Earth's angular velocity can be written as

$$
\vec{\Omega}_E
=
\Omega_E\cos\phi\,\hat N
+
\Omega_E\sin\phi\,\hat U.
$$

The horizontal component $\Omega_E\cos\phi\,\hat N$ slightly modifies the detailed motion, but the component that rotates the local horizontal plane about the vertical axis is

$$
\Omega_E\sin\phi\,\hat U.
$$

Therefore, the part of Earth's rotation responsible for the observed turning of the pendulum's swing plane is

$$
\Omega_{\text{vertical}}
=
\Omega_E\sin\phi.
$$

<br>

<strong>Step 3: Connect the Coriolis force to precession</strong>

For horizontal pendulum motion, the dominant Coriolis effect associated with the vertical component of Earth's angular velocity is

$$
\vec a_{\text{Cor}}
=
-2(\Omega_E\sin\phi\,\hat U)\times\vec v_{\text{rot}}.
$$

This acceleration is horizontal and perpendicular to the bob's horizontal velocity.

That sideways acceleration continuously nudges the motion sideways. Over many oscillations, these tiny sideways deflections accumulate into a slow rotation of the swing plane.

The result is that the swing plane precesses relative to Earth's surface with angular speed

$$
\Omega_p
=
\Omega_E\sin\phi.
$$

Thus,

$$
\boxed{
\Omega_p
=
\Omega_E\sin\phi
}.
$$

<br>

<strong>Step 4: Calculate the precession rate</strong>

Substitute the given values:

$$
\Omega_p
=
(7.29\times10^{-5})\sin(40^\circ).
$$

Since

$$
\sin(40^\circ)=0.643,
$$

we get

$$
\Omega_p
=
4.69\times10^{-5}~\text{rad/s}.
$$

Therefore,

$$
\boxed{
\Omega_p
=
4.69\times10^{-5}~\text{rad/s}
}.
$$

<br>

<strong>Step 5: Find the time for one complete rotation</strong>

One complete rotation corresponds to

$$
2\pi~\text{rad}.
$$

Therefore,

$$
T_p
=
\frac{2\pi}{\Omega_p}.
$$

Substituting,

$$
T_p
=
\frac{2\pi}{4.69\times10^{-5}}
=
1.34\times10^5~\text{s}.
$$

Convert to hours:

$$
T_p
=
\frac{1.34\times10^5}{3600}
=
37.2~\text{hr}.
$$

Thus,

$$
\boxed{
T_p
\approx
37.2~\text{hr}
}.
$$

<br>

<strong>Step 6: Interpret the result</strong>

At latitude $40^\circ$, the pendulum precesses at about

$$
\sin(40^\circ)\approx 0.643
$$

of Earth's rotation rate.

This means the pendulum's swing plane rotates relative to the ground more slowly than it would at the North Pole.

At the North Pole,

$$
\phi=90^\circ,
$$

so

$$
\Omega_p=\Omega_E.
$$

The pendulum completes one rotation per day relative to the ground.

At the Equator,

$$
\phi=0^\circ,
$$

so

$$
\Omega_p=0.
$$

There is no precession because Earth's angular velocity has no component along the local vertical direction.

The Foucault pendulum therefore provides a direct demonstration of Earth's rotation through the Coriolis force.

</div>

</details>

## Problem 10 - North-South Maglev Train and Coriolis Loading

A Japanese superconducting maglev train travels due north along a straight section of guideway near Yamanashi, Japan. Approximate the latitude as

$$
\phi = 35.5^\circ.
$$

Assume the train moves at

$$
v = 500~\text{km/h},
$$

which is a realistic operating-scale speed for the Chuo Shinkansen maglev system. The Yamanashi maglev line has also been used for higher-speed testing, including the L0 series record of about $603~\text{km/h}$. 

Assume the train has mass

$$
m = 4.0\times 10^4~\text{kg}.
$$

Use Earth's angular speed

$$
\Omega_E = 7.29\times 10^{-5}~\text{rad/s}.
$$

1. Convert the train speed to $\text{m/s}$.
2. Find the Coriolis acceleration on the train.
3. Find the Coriolis force on the train.
4. Determine whether the force points east or west.
5. Explain how this could have engineering consequences for lateral loading of the guideway.

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Problem solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

<strong>Step 1: Convert the Speed</strong>

The train speed is

$$
v = 500~\text{km/h}.
$$

Converting to SI units,

$$
v
=
500\left(\frac{1000}{3600}\right)
=
138.9~\text{m/s}.
$$

Thus,

$$
\boxed{
v = 138.9~\text{m/s}
}
$$

<br>

<strong>Step 2: Set Up the Coriolis Acceleration</strong>

In Earth's rotating frame, the Coriolis acceleration is

$$
\vec a_{\text{Cor}}
=
-2\vec\Omega_E\times \vec v_{\text{rot}}.
$$

Use local unit vectors:

- $\hat E$ points east,
- $\hat N$ points north,
- $\hat U$ points upward.

At latitude $\phi$, Earth's angular velocity can be written as

$$
\vec\Omega_E
=
\Omega_E\cos\phi\,\hat N
+
\Omega_E\sin\phi\,\hat U.
$$

Since the train travels due north,

$$
\vec v_{\text{rot}}
=
v\hat N.
$$

Therefore,

$$
\vec a_{\text{Cor}}
=
-2
\left(
\Omega_E\cos\phi\,\hat N
+
\Omega_E\sin\phi\,\hat U
\right)
\times
(v\hat N).
$$

The first term is zero because

$$
\hat N\times \hat N = 0.
$$

The second term gives

$$
\vec a_{\text{Cor}}
=
-2\Omega_E v\sin\phi
\left(
\hat U\times \hat N
\right).
$$

Using the local right-handed system

$$
\hat E\times \hat N = \hat U,
$$

we have

$$
\hat U\times \hat N = -\hat E.
$$

Therefore,

$$
\vec a_{\text{Cor}}
=
2\Omega_E v\sin\phi\,\hat E.
$$

So a northward-moving train in the Northern Hemisphere is deflected eastward.

<br>

<strong>Step 3: Calculate the Coriolis Acceleration</strong>

The magnitude is

$$
a_{\text{Cor}}
=
2\Omega_E v\sin\phi.
$$

Substituting,

$$
a_{\text{Cor}}
=
2(7.29\times10^{-5})(138.9)\sin(35.5^\circ).
$$

Since

$$
\sin(35.5^\circ)=0.581,
$$

we get

$$
a_{\text{Cor}}
=
1.18\times10^{-2}~\text{m/s}^2.
$$

Thus,

$$
\boxed{
\vec a_{\text{Cor}}
=
1.18\times10^{-2}\hat E~\text{m/s}^2
}
$$

<br>

<strong>Step 4: Calculate the Coriolis Force</strong>

The Coriolis force is

$$
\vec F_{\text{Cor}}
=
m\vec a_{\text{Cor}}.
$$

Thus,

$$
F_{\text{Cor}}
=
(4.0\times10^4)(1.18\times10^{-2})
=
4.72\times10^2~\text{N}.
$$

Therefore,

$$
\boxed{
\vec F_{\text{Cor}}
=
4.72\times10^2\hat E~\text{N}
}
$$

<br>

<strong>Step 5: Engineering Interpretation</strong>

For northward travel in Japan, the Coriolis force points eastward. If the train travels southward instead, then

$$
\vec v_{\text{rot}} = -v\hat N,
$$

and the Coriolis force reverses direction. It would point westward.

The magnitude,

$$
F_{\text{Cor}}\approx 4.7\times10^2~\text{N},
$$

is very small compared with the train's weight,

$$
mg
=
(4.0\times10^4)(9.81)
=
3.92\times10^5~\text{N}.
$$

The ratio is

$$
\frac{F_{\text{Cor}}}{mg}
=
\frac{4.72\times10^2}{3.92\times10^5}
=
1.20\times10^{-3}.
$$

So the Coriolis force is only about

$$
\boxed{
0.12\%
}
$$

of the train's weight.

Even though this is small, it is a systematic lateral force. In a precision high-speed guideway system, a repeated sideways bias could slightly change the lateral loading on the guideway or suspension system. Over many trips, engineers may need to account for small asymmetric effects when designing tolerances, feedback control, alignment, and long-term wear patterns.

</div>
