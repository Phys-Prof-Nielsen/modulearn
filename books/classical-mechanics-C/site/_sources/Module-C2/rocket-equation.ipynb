# C2.2 — The Rocket Equation
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C2.2.1 — Motivation

Rockets are among the most remarkable applications of classical mechanics.

Unlike most systems studied in introductory mechanics, rockets continuously lose mass as fuel is expelled from the engine. This means that the rocket must be treated as a **variable mass system**.

The rocket equation describes how the velocity of a rocket changes as it ejects fuel.

This equation is fundamental in:
- spaceflight,
- satellite maneuvering,
- planetary landings,
- interplanetary missions,
- propulsion engineering.

One of the most surprising aspects of rocket motion is that rockets do not require air to push against.

Instead, propulsion occurs because momentum is carried away by the expelled exhaust gases.

By conservation of momentum, the rocket gains momentum in the opposite direction.

---

The standard rocket equation is often called the **Tsiolkovsky Rocket Equation**, named after Konstantin Tsiolkovsky, who first derived it in the late 1800s.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">

<strong>Curiosity</strong>

The rocket equation reveals one of the great engineering challenges of space travel:

To gain additional velocity, a rocket must carry more fuel.

But carrying more fuel increases the rocket’s mass, which itself requires even more fuel.

This exponential relationship is one reason why multi-stage rockets are commonly used.

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C2.2.2 — Physical Setup

We consider a rocket moving with velocity

$$
\vec{v}
$$

relative to an inertial reference frame.

At some instant, the rocket has mass

$$
M.
$$

During a small time interval $dt$, the rocket ejects a small amount of fuel with mass

$$
dm.
$$

The expelled exhaust moves with velocity

$$
\vec{v}+\vec{u},
$$

where:
- $\vec{u}$ is the exhaust velocity relative to the rocket,
- $\vec{v}$ is the rocket velocity relative to the inertial frame.

We also allow for a possible external force

$$
\vec{F}_{ext},
$$

which may represent:
- gravity,
- aerodynamic drag,
- or other environmental forces.

---

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">

<strong>Important Sign Convention</strong>

We define:
- $dM < 0$ for the rocket because the rocket loses mass,
- $dm > 0$ for the expelled exhaust mass.

Thus,

$$
dm=-dM.
$$

</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C2.2.3 — Derivation of the Rocket Equation

We begin with the differential form of the impulse-momentum theorem:

$$
d\vec{P}
=
\vec{F}_{ext}\,dt.
$$

---

### Initial Momentum

Before fuel is expelled, the momentum is

$$
\vec{P}_i
=
(M+dm)\vec{v}.
$$

---

### Final Momentum

After the fuel is expelled:
- the rocket mass is $M$,
- the rocket velocity becomes $\vec{v}+d\vec{v}$,
- the exhaust mass is $dm$,
- the exhaust velocity is $\vec{v}+d\vec{v}+\vec{u}$.

Thus,

$$
\vec{P}_f
=
M(\vec{v}+d\vec{v})
+
dm(\vec{v}+d\vec{v}+\vec{u}).
$$

---

### Change in Momentum

The momentum change is

$$
d\vec{P}
=
\vec{P}_f-\vec{P}_i.
$$

Substituting the expressions above,

$$
d\vec{P}
=
M(\vec{v}+d\vec{v})
+
dm(\vec{v}+d\vec{v}+\vec{u})
-
(M+dm)\vec{v}.
$$

Expanding,

$$
d\vec{P}
=
Md\vec{v}
+
\vec{u}\,dm
+
dm\,d\vec{v}.
$$

The term

$$
dm\,d\vec{v}
$$

is a second-order differential and is therefore neglected.

This leaves

$$
d\vec{P}
=
Md\vec{v}
+
\vec{u}\,dm.
$$

Applying the impulse-momentum theorem,

$$
\vec{F}_{ext}\,dt
=
Md\vec{v}
+
\vec{u}\,dm.
$$

Dividing by $dt$,

$$
\vec{F}_{ext}
=
M\frac{d\vec{v}}{dt}
+
\vec{u}\frac{dm}{dt}.
$$

Using

$$
dm=-dM,
$$

we obtain the general rocket equation:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">

<strong>General Rocket Equation</strong>

$$
\boxed{
\vec{F}_{ext}
=
M\frac{d\vec{v}}{dt}
-
\vec{u}\frac{dM}{dt}
}
$$

</div>

The term

$$
-\vec{u}\frac{dM}{dt}
$$

is called the **thrust force**.

<div style="background-color:#ffe6e6; border-left:6px solid #cc0000; padding:14px; border-radius:4px;">
<strong>Danger:</strong><br>

The exhaust velocity $\vec{u}$ is measured relative to the rocket, not relative to the ground.

This distinction is extremely important when applying the rocket equation correctly.
</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C2.2.4 — The Tsiolkovsky Rocket Equation

We now consider the idealized case where there are no external forces:

$$
\vec{F}_{ext}=0.
$$

The rocket equation becomes

$$
M\frac{d\vec{v}}{dt}
=
\vec{u}\frac{dM}{dt}.
$$

Rearranging,

$$
d\vec{v}
=
\vec{u}\frac{dM}{M}.
$$

Integrating from:
- initial mass $M_i$ to final mass $M_f$,
- initial velocity $\vec{v}_i$ to final velocity $\vec{v}_f$,

gives

$$
\int_{\vec{v}_i}^{\vec{v}_f} d\vec{v}
=
\vec{u}
\int_{M_i}^{M_f}
\frac{dM}{M}.
$$

Evaluating the integrals,

$$
\vec{v}_f-\vec{v}_i
=
\vec{u}
\ln\left(\frac{M_f}{M_i}\right).
$$

Rewriting,

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">

<strong>Tsiolkovsky Rocket Equation</strong>

$$
\boxed{
\Delta \vec{v}
=
\vec{u}
\ln\left(\frac{M_i}{M_f}\right)
}
$$

</div>

This equation predicts the total velocity change achievable from a given fuel mass ratio.

---

<div style="background-color:#e8f5e9; border-left:6px solid #006633; padding:14px; border-radius:4px;">

<h3 style="margin-top:0;">Box Activity — Escaping the Death Star</h3>

A Rebel pilot attempts to escape the Death Star using a small ion-drive starfighter.

The spacecraft initially has total mass

$$
M_i
=
4.0\times10^4~\text{kg},
$$

including fuel.

After a prolonged engine burn, the remaining mass is

$$
M_f
=
1.5\times10^4~\text{kg}.
$$

The ion exhaust leaves the spacecraft with constant speed

$$
u
=
2.8\times10^4~\text{m/s}
$$

relative to the spacecraft.

Assume:
- motion occurs in deep space,
- external forces are negligible,
- the spacecraft initially starts from rest.

---

### Tasks

1. Apply the Tsiolkovsky rocket equation to determine the final speed of the spacecraft.

2. Determine the total change in velocity.

3. If the spacecraft instead wanted to double its final speed, would it need:
   - twice as much fuel,
   - more than twice as much fuel,
   - or less than twice as much fuel?

Explain physically.

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Activity solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

### Step 1 — Apply the Rocket Equation

For motion without external forces, the rocket equation is

$$
\Delta v
=
u
\ln\left(\frac{M_i}{M_f}\right).
$$

Substituting the given values,

$$
\Delta v
=
(2.8\times10^4)
\ln\left(
\frac{4.0\times10^4}{1.5\times10^4}
\right).
$$

First compute the mass ratio:

$$
\frac{M_i}{M_f}
=
2.667.
$$

Then,

$$
\ln(2.667)\approx0.981.
$$

Thus,

$$
\Delta v
=
(2.8\times10^4)(0.981).
$$

Therefore,

$$
\boxed{
\Delta v
\approx
2.75\times10^4~\text{m/s}
}
$$

or approximately

$$
\boxed{
27.5~\text{km/s}
}
$$

---

### Step 2 — Determine the Final Speed

Since the spacecraft started from rest,

$$
v_i=0.
$$

Therefore,

$$
v_f=\Delta v.
$$

Hence,

$$
\boxed{
v_f
\approx
2.75\times10^4~\text{m/s}
}
$$

---

### Step 3 — Physical Interpretation

The rocket equation contains a logarithm:

$$
\Delta v
=
u
\ln\left(\frac{M_i}{M_f}\right).
$$

Because of this logarithmic dependence:
- doubling the desired speed does **not** require merely doubling the fuel,
- increasingly large fuel masses are required for additional velocity gains.

Therefore, to double the final speed, the spacecraft would require

$$
\boxed{\text{more than twice as much fuel}.}
$$

This exponential fuel requirement is one of the major challenges in rocket engineering and interstellar travel.

---

<strong>Curiosity</strong>

Real spacecraft often use multiple rocket stages because carrying enormous fuel masses becomes impractical.

As empty fuel tanks are discarded, the spacecraft mass decreases substantially, improving efficiency.

## C2.2.5 — Physical Interpretation

The rocket equation reveals several important physical ideas.

---

### Momentum Conservation Produces Thrust

Rockets accelerate because exhaust gases carry momentum away from the rocket.

As momentum leaves in one direction, the rocket gains momentum in the opposite direction.

No atmosphere is required.

This is why rockets function in space.

---

### The Mass Ratio is Critical

The quantity

$$
\frac{M_i}{M_f}
$$

is called the **mass ratio**.

Large velocity changes require large mass ratios.

This means rockets must carry enormous amounts of fuel to achieve high speeds.

---

### Velocity Depends Logarithmically on Fuel

Notice the logarithm:

$$
\ln\left(\frac{M_i}{M_f}\right).
$$

This means:
- doubling the fuel does not double the velocity,
- increasingly large fuel amounts produce diminishing returns.

This is one of the major engineering challenges in rocketry.

## C2.2.6 — Special Cases

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">

<h3 style="margin-top:0;">Special Case — Rocket Under Gravity</h3>

If the rocket moves vertically near Earth’s surface, then

$$
\vec{F}_{ext}
=
-Mg\,\hat{k}.
$$

The rocket equation becomes

$$
M\frac{d\vec{v}}{dt}
=
\vec{u}\frac{dM}{dt}
-
Mg\,\hat{k}.
$$

Gravity therefore reduces the rocket’s achievable velocity.

These losses are called **gravity losses**.

</div>

<br>

<div style="background-color:#e0f7fa; border-left:6px solid #006a80; padding:14px; border-radius:4px;">

<h3 style="margin-top:0;">Special Case — Atmospheric Drag</h3>

If aerodynamic drag is present, then the external force may depend on velocity:

$$
\vec{F}_{drag}
=
-c_d v^2 \hat{v}.
$$

In such situations, the rocket equation usually cannot be solved analytically and numerical methods become necessary.

</div>

<div style="background-color:#e8f5e9; border-left:6px solid #006633; padding:14px; border-radius:4px;">

<h3 style="margin-top:0;">Box Activity — Escaping a Death Star Tractor Field</h3>

A Rebel transport attempts to escape from the Death Star, but the Death Star projects a tractor field that pulls backward on the ship.

The transport moves in a straight line in the $+x$ direction.

The ship has initial mass

$$
M_i = 5.0\times 10^4~\text{kg},
$$

and the empty ship mass after all usable fuel is burned is

$$
M_f = 2.0\times 10^4~\text{kg}.
$$

The exhaust speed relative to the ship is

$$
u = 3.0\times 10^4~\text{m/s}.
$$

The Death Star tractor field exerts a constant backward force of magnitude

$$
F_T = 1.2\times 10^5~\text{N}.
$$

The Rebel ship must burn its engine for

$$
\Delta t = 60~\text{s}.
$$

Assume:
- straight-line motion,
- constant exhaust speed,
- constant mass-loss rate,
- the tractor field is the only external force,
- the ship starts from rest.

---

### Tasks

1. Determine the average mass-loss rate if the ship burns all usable fuel during the $60~\text{s}$ escape burn.

2. Determine the thrust force produced by the engine.

3. Determine the net force on the ship at the start of the burn.

4. Determine the velocity change using the rocket equation with an external force correction:

$$
\Delta v
=
u\ln\left(\frac{M_i}{M_f}\right)
+
\int_0^{\Delta t}\frac{F_{ext}}{M(t)}\,dt.
$$

5. Determine whether the ship escapes if it needs at least

$$
\Delta v_{\text{escape}} = 2.0\times 10^4~\text{m/s}.
$$

---

<details>
<summary style="background-color:#006633; color:white; padding:8px; border-radius:4px; cursor:pointer;">
Activity solution
</summary>

<div style="background-color:#e8f5e9; padding:10px; border-radius:4px; margin-top:6px;">

### Step 1 — Mass-Loss Rate

The ship loses mass

$$
\Delta M = M_i-M_f.
$$

Substituting,

$$
\Delta M
=
5.0\times10^4-2.0\times10^4
=
3.0\times10^4~\text{kg}.
$$

Since the burn lasts $60~\text{s}$,

$$
k
=
\frac{\Delta M}{\Delta t}
=
\frac{3.0\times10^4}{60}.
$$

Thus,

$$
\boxed{
k=500~\text{kg/s}
}
$$

---

### Step 2 — Thrust Force

The thrust force is

$$
F_{\text{thrust}}
=
uk.
$$

Substituting,

$$
F_{\text{thrust}}
=
(3.0\times10^4)(500).
$$

Therefore,

$$
\boxed{
F_{\text{thrust}}
=
1.5\times10^7~\text{N}
}
$$

---

### Step 3 — Net Force at the Start

The tractor field acts backward, so

$$
F_{ext}=-F_T.
$$

The net force at the start of the burn is

$$
F_{\text{net}}
=
F_{\text{thrust}}-F_T.
$$

Substituting,

$$
F_{\text{net}}
=
1.5\times10^7-1.2\times10^5.
$$

Thus,

$$
\boxed{
F_{\text{net}}
=
1.488\times10^7~\text{N}
}
$$

The initial acceleration is

$$
a_0
=
\frac{F_{\text{net}}}{M_i}.
$$

So,

$$
a_0
=
\frac{1.488\times10^7}{5.0\times10^4}.
$$

Therefore,

$$
\boxed{
a_0
\approx
298~\text{m/s}^2
}
$$

---

### Step 4 — Velocity Change with Tractor Field

The mass decreases linearly:

$$
M(t)=M_i-kt.
$$

The rocket equation with the external force correction is

$$
\Delta v
=
u\ln\left(\frac{M_i}{M_f}\right)
+
\int_0^{\Delta t}\frac{F_{ext}}{M(t)}\,dt.
$$

Since the tractor field points backward,

$$
F_{ext}=-F_T.
$$

Thus,

$$
\Delta v
=
u\ln\left(\frac{M_i}{M_f}\right)
-
\int_0^{\Delta t}\frac{F_T}{M_i-kt}\,dt.
$$

Evaluate the tractor-field correction:

$$
\int_0^{\Delta t}\frac{F_T}{M_i-kt}\,dt
=
-\frac{F_T}{k}
\left[
\ln(M_i-kt)
\right]_0^{\Delta t}.
$$

Since

$$
M_i-k\Delta t=M_f,
$$

we get

$$
\int_0^{\Delta t}\frac{F_T}{M_i-kt}\,dt
=
\frac{F_T}{k}
\ln\left(\frac{M_i}{M_f}\right).
$$

Therefore,

$$
\Delta v
=
\left(
u-\frac{F_T}{k}
\right)
\ln\left(\frac{M_i}{M_f}\right).
$$

Now compute each term:

$$
\frac{F_T}{k}
=
\frac{1.2\times10^5}{500}
=
240~\text{m/s}.
$$

The mass ratio is

$$
\frac{M_i}{M_f}
=
\frac{5.0\times10^4}{2.0\times10^4}
=
2.5.
$$

Thus,

$$
\ln(2.5)\approx0.916.
$$

Therefore,

$$
\Delta v
=
(3.0\times10^4-240)(0.916).
$$

So,

$$
\Delta v
\approx
(2.976\times10^4)(0.916).
$$

Hence,

$$
\boxed{
\Delta v
\approx
2.73\times10^4~\text{m/s}
}
$$

or

$$
\boxed{
\Delta v
\approx
27.3~\text{km/s}.
}
$$

---

### Step 5 — Escape Decision

The required velocity change is

$$
\Delta v_{\text{escape}}
=
2.0\times10^4~\text{m/s}.
$$

Since

$$
2.73\times10^4~\text{m/s}
>
2.0\times10^4~\text{m/s},
$$

the ship escapes.

Therefore,

$$
\boxed{
\text{The Rebel transport escapes the tractor field.}
}
$$

---

### Physical Interpretation

The tractor field reduces the final velocity by an amount determined by

$$
\frac{F_T}{k}
\ln\left(\frac{M_i}{M_f}\right).
$$

A stronger tractor field, a longer burn, or a smaller mass-loss rate would make escape harder.

The key lesson is that the fuel does two jobs:

- it produces thrust,
- and it must overcome the external force field.
