# C2.0 Learning Outcomes — Variable Mass Systems

## Why this module matters

Most introductory mechanics problems assume that mass remains constant. However, many important physical systems involve objects that gain or lose mass over time.

Examples include:

- rockets expelling fuel  
- leaking containers  
- raindrops accumulating water  
- conveyor systems  
- jet propulsion  
- astrophysical accretion processes  

In these systems, momentum changes not only because velocity changes, but also because mass changes.

This module introduces **variable mass systems** using the **impulse-momentum theorem** as the fundamental principle.

Rather than starting from

$$
\sum \vec{F}=m\vec{a},
$$

we begin with the momentum update idea:

$$
d\vec{p}
=
\vec{F}\,dt.
$$

This says that a small impulse changes the momentum of a system.

For variable mass systems, we must be careful about what we define as the system, because mass may enter or leave during the time interval $dt$.

We will examine:

- momentum conservation in systems with changing mass  
- mass flow rates  
- thrust forces  
- rocket propulsion  
- systems with incoming and outgoing mass  

This module also provides the foundation for later studies in:

- rocket dynamics  
- propulsion systems  
- fluid mechanics  
- astrophysics  
- advanced continuum mechanics

By the end of this module, you will be able to:

---

### A. Impulse, Momentum, and Variable Mass Concepts  
*Introduces the foundations of variable mass mechanics.*

- **A1.** Explain why the form  

$$
\sum \vec{F}=m\vec{a}
$$

is not generally sufficient for variable mass systems.

- **A2.** Use the impulse-momentum theorem as the starting point:

$$
d\vec{p}
=
\vec{F}\,dt
$$

- **A3.** Define and interpret:
  - mass flow rate  
  - exhaust velocity  
  - thrust  

- **A4.** Distinguish between:
  - closed systems  
  - open systems  
  - isolated systems  

- **A5.** Explain how momentum changes when mass enters or leaves a system.

---

### B. Derivation and Analysis of Rocket Motion  
*Develops the mathematical framework for rocket propulsion.*

- **B1.** Derive the rocket equation using impulse-momentum reasoning.

- **B2.** Interpret the physical meaning of exhaust velocity relative to the rocket.

- **B3.** Derive and apply the Tsiolkovsky rocket equation:

$$
\Delta v
=
v_e
\ln\left(\frac{m_i}{m_f}\right)
$$

- **B4.** Analyze how changing mass affects acceleration.

- **B5.** Explain why rockets can accelerate in vacuum.

---

### C. Applications of Variable Mass Systems  
*Applies momentum methods to realistic systems.*

- **C1.** Solve problems involving rockets and jet propulsion.

- **C2.** Analyze systems with incoming or outgoing mass flow.

- **C3.** Apply momentum methods to leaking or accumulating mass systems.

- **C4.** Determine thrust forces from mass ejection rates.

- **C5.** Apply variable mass analysis to vertical rocket motion under gravity.

---

### D. Physical Interpretation and Modeling  
*Builds conceptual understanding of variable mass dynamics.*

- **D1.** Interpret thrust as a momentum-transfer process.

- **D2.** Explain how external impulses affect variable mass systems.

- **D3.** Distinguish between system velocity and relative exhaust velocity.

- **D4.** Analyze limiting cases of rocket motion.

- **D5.** Interpret solutions physically using conservation laws and reference frames.

---

### E. Computational and Analytical Skills  
*Develops mathematical and computational modeling techniques.*

- **E1.** Solve variable mass differential equations analytically when possible.

- **E2.** Implement numerical models of rocket motion using Python.

- **E3.** Generate plots of:
  - velocity vs time  
  - mass vs time  
  - altitude vs time  
  - acceleration vs time  

- **E4.** Investigate how thrust, fuel consumption rate, and exhaust velocity affect motion.

- **E5.** Compare analytical and numerical solutions of rocket dynamics.

---

### Performance Criteria

- The impulse-momentum theorem is used correctly as the foundation for variable mass systems.  
- Momentum changes due to both velocity change and mass transfer are interpreted correctly.  
- Rocket equations are derived and interpreted correctly.  
- Variable mass systems are analyzed using appropriate system definitions.  
- Analytical and computational methods are used appropriately to model rocket motion.
