## Problem C2.1 — Gyroscope with a Horizontal Axle

A bicycle-wheel gyroscope of mass $m=2.0\ \mathrm{kg}$ and rim radius $R=0.20\ \mathrm{m}$ is mounted on a horizontal axle and spins with angular speed $\omega=160\ \mathrm{rad/s}$ about its symmetry axis. One end of the axle is supported at a fixed, frictionless pivot; the wheel+axle center of mass is a distance $\ell=0.15\ \mathrm{m}$ horizontally from the pivot. Take $g=9.81\ \mathrm{m/s^2}$.

1) Treat the wheel as a thin hoop so that its spin moment of inertia about the axle is $I=\;mR^2$.  
2) Assuming steady precession about the vertical, find the precession rate
   
$$
\Omega=\frac{m g \ell}{I\,\omega}.
$$

3) Compute the rotational kinetic energy $T=\tfrac12 I\omega^2$.  
4) Briefly explain *why* precession occurs even though the axle is horizontal.

```{admonition} Problem C2.1
:class: tip, dropdown

### Solution

**(1) Spin moment of inertia (thin hoop):**

$$
I=mR^2=(2.0)(0.20)^2=0.080\ \mathrm{kg\,m^2}.
$$

**(2) Steady precession rate:**

$$
\Omega=\frac{m g \ell}{I\,\omega}
=\frac{(2.0)(9.81)(0.15)}{(0.080)(160)}
\approx 0.23\ \mathrm{rad/s}
\;\;(\approx 13.1^\circ/\mathrm{s}).
$$

**(3) Rotational kinetic energy:**

$$
T=\tfrac12 I\omega^2=\tfrac12(0.080)(160)^2\approx 1.02\times 10^{3}\ \mathrm{J}.
$$

**(4) Why precession with a *horizontal* axle?**  
The lever arm from pivot to the center of mass is $\boldsymbol r=\ell\,\hat{\boldsymbol x}$ (horizontal), while the weight is $m\boldsymbol g=-mg\,\hat{\boldsymbol z}$. The gravitational torque

$$
\boldsymbol\tau=\boldsymbol r\times (m\boldsymbol g)=\ell\,\hat{\boldsymbol x}\times(-mg\,\hat{\boldsymbol z})
=+mg\ell\,\hat{\boldsymbol y}
$$

is **nonzero** and horizontal, so it turns the **direction** of the spin angular momentum $\boldsymbol L\approx I\omega\,\hat{\boldsymbol e}_{\text{axle}}$ around the vertical at rate $\Omega=\tau/L$.
```

## Problem C2.2 — Pivot reaction forces during steady precession (horizontal-axle wheel)**

Use the same wheel and parameters from **Problem C2.1**. In steady precession at the rate $\Omega$ you found, the wheel’s center of mass moves in a horizontal circle of radius $\rho=\ell$.

1) Find the **horizontal** component $F_{\text{H}}$ of the pivot reaction needed to provide the centripetal acceleration.  
2) Find the **vertical** component $F_{\text{V}}$ of the pivot reaction.  
3) Compute the **magnitude** of the pivot reaction $F_{\text{pivot}}$ and its **tilt angle** $\phi$ above the horizontal:
   
$$
F_{\text{pivot}}=\sqrt{F_{\text{H}}^2+F_{\text{V}}^2},\qquad
\phi=\tan^{-1}\!\left(\frac{F_{\text{V}}}{F_{\text{H}}}\right).
$$

Use $\Omega$ from Problem 1 and $\ell=0.15\ \mathrm{m}$.

```{admonition} Problem C2.2
:class: tip, dropdown

### Solution

**Geometry of the CM motion:**  
In steady precession with a horizontal axle, the center of mass executes uniform circular motion of radius $\rho=\ell$ at speed

$$
v=\Omega\,\rho=\Omega\,\ell.
$$

**(1) Horizontal reaction (centripetal):**

$$
F_{\text{H}}=\frac{m v^2}{\rho}=m\,\frac{(\Omega\ell)^2}{\ell}=m\,\Omega^2\,\ell.
$$

With $m=2.0\ \mathrm{kg}$, $\Omega\approx 0.23\ \mathrm{rad/s}$, and $\ell=0.15\ \mathrm{m}$,

$$
F_{\text{H}}\approx (2.0)(0.23)^2(0.15)\approx 0.016\ \mathrm{N}.
$$

**(2) Vertical reaction (weight support):**

$$
F_{\text{V}}=mg=(2.0)(9.81)=19.62\ \mathrm{N}.
$$

**(3) Magnitude and tilt of pivot reaction:**

$$
F_{\text{pivot}}=\sqrt{F_{\text{H}}^2+F_{\text{V}}^2}\approx \sqrt{(0.016)^2+(19.62)^2}\approx 19.62\ \mathrm{N},
$$

$$
\phi=\tan^{-1}\!\left(\frac{F_{\text{V}}}{F_{\text{H}}}\right)\approx \tan^{-1}\!\left(\frac{19.62}{0.016}\right)\approx 89.95^\circ.
$$

**Interpretation.** The pivot force is almost entirely **vertical**; the required centripetal force for the slow precession is tiny compared to the weight. (If $\Omega$ were larger, $F_{\text{H}}\propto\Omega^2$ would increase accordingly.)
```

## Problem C2.3 — Quasi-Steady Precession During Spin-Down

The same horizontal-axle gyroscope from **Problem C2.1** is spinning with angular speed $\omega_0 = 160\ \mathrm{rad/s}$, which slowly decreases over time due to friction. The wheel continues to precess smoothly about the vertical, maintaining a constant **tilt angle** $\theta$ (so its geometry of motion remains steady), but the **precession rate** $\Omega(t)$ changes as the spin decreases.

This situation is called **quasi-steady precession**, where the instantaneous precession rate is given by:

$$
\Omega(t) = \frac{m g \ell}{I\,\omega(t)}.
$$

1. If the spin drops from $\omega_0 = 160\ \mathrm{rad/s}$ to $\omega_1 = 80\ \mathrm{rad/s}$, compute the corresponding change in precession rate.  
2. If the spin decreases linearly in time,
 
   $$
   \omega(t) = \omega_0 - \alpha t,
   $$
   with constant $\alpha > 0$, show that the total precession angle $\Phi$ accumulated during this interval is
   
   $$
   \Phi = \frac{m g \ell}{I\,\alpha}\,\ln\!\left(\frac{\omega_0}{\omega_1}\right),
   $$
   and evaluate $\Phi$ (in degrees) for $\alpha = 0.40\ \mathrm{rad/s^2}$, $\omega_0 = 160\ \mathrm{rad/s}$, and $\omega_1 = 80\ \mathrm{rad/s}$.


````python
# DIY Cell
````

```{admonition} Problem C2.3
:class: tip, dropdown

### Solution

**(1) Change in precession rate.** 

Since $\Omega\propto 1/\omega$,

$$
\Omega_0=\frac{m g \ell}{I\,\omega_0},\qquad
\Omega_1=\frac{m g \ell}{I\,\omega_1}
= \left(\frac{\omega_0}{\omega_1}\right)\Omega_0.
$$

With $m=2.0,\ \ell=0.15,\ I=0.080,\ \omega_0=160,\ \omega_1=80$,

$$
\Omega_0=\frac{(2.0)(9.81)(0.15)}{(0.080)(160)}\approx 0.23\ \mathrm{rad/s},\qquad
\Omega_1=2\,\Omega_0\approx 0.46\ \mathrm{rad/s}.
$$

Thus, halving $\omega$ **doubles** $\Omega$.

**(2) Total precession angle.**

$$
\Phi=\int_{t_0}^{t_1}\Omega(t)\,dt=\frac{m g \ell}{I}\int_{t_0}^{t_1}\frac{dt}{\omega(t)}.
$$

Let $t_0=0$ and $t_1=(\omega_0-\omega_1)/\alpha$. Then

$$
\Phi=\frac{m g \ell}{I}\int_{0}^{t_1}\frac{dt}{\omega_0-\alpha t}
=\frac{m g \ell}{I\,\alpha}\,\ln\!\left(\frac{\omega_0}{\omega_1}\right).
$$

Numerically, with $\alpha=0.40\ \mathrm{rad/s^2}$ and $\omega_0/\omega_1=2$,

$$
\Phi=\frac{(2.0)(9.81)(0.15)}{(0.080)(0.40)}\ln 2
\approx \frac{7.3575}{0.032}\,(0.6931)
\approx 159.9\ \mathrm{rad}
\approx 9.16\times 10^{3}\ ^\circ.
$$

**Interpretation.** Even a modest increase in $\Omega$ during spin-down can integrate to a **large** total precession angle because the process lasts many seconds and $\Omega$ grows as $1/\omega$.
```
