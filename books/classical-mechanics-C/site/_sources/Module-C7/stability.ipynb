### LOC: Stability

#### Motivation

Oscillations occur only when an equilibrium can “hold” small disturbances. Stability analysis asks a precise question: if we slightly perturb a system from equilibrium, does it return (stable), drift away (unstable), or neither (neutral)? This matters across physics and engineering: a bridge should be stable under small gusts; a satellite’s attitude must not tip over; a pencil balanced on its tip is not stable. In this section we connect stability to small oscillations: the same linearization that gives harmonic motion also provides a clean criterion for stability.

#### Notions of stability (near equilibrium)

Consider generalized coordinates $q = (q_1,\dots,q_n)$ with an equilibrium at $q_0$. We say:
- Stable (Lyapunov): every sufficiently small perturbation remains small for all time.
- Asymptotically stable: trajectories not only remain near $q_0$ but return to it (requires dissipation).
- Unstable: some arbitrarily small perturbations eventually grow large.
- Neutrally (marginally) stable: perturbations neither grow nor decay; the system can “wander” on a flat direction.

For conservative mechanical systems, static stability is governed by the potential landscape near $q_0$.

#### One degree of freedom: curvature of the potential

Let $U(q)$ be the potential, with equilibrium at $q_0$ where $U'(q_0)=0$. Expand about $q_0$ using $\xi = q-q_0$:
$$
U(q_0+\xi) = U(q_0) + \tfrac{1}{2}U''(q_0)\,\xi^2 + \tfrac{1}{6}U^{(3)}(q_0)\,\xi^3 + \cdots.
$$
The linearized equation of motion is
$$
m\ddot{\xi} = -U'(q_0+\xi) \approx -U''(q_0)\,\xi
\quad\Rightarrow\quad
\ddot{\xi} + \omega^2 \xi = 0,\;\; \omega^2=\frac{U''(q_0)}{m}.
$$

Classification in 1D (conservative):
- $U''(q_0)>0$: local minimum, $\omega^2>0$ → small oscillations with frequency $\omega=\sqrt{U''/m}$ → stable.
- $U''(q_0)<0$: local maximum, $\omega^2<0$ → $\ddot{\xi}-|\omega|^2\xi=0$ → exponential growth/decay → unstable.
- $U''(q_0)=0$: inconclusive at quadratic order; stability depends on higher terms (e.g., flat inflection → often unstable; flat minimum with positive quartic term can be stable but with amplitude-dependent frequency).

Example snapshots:
- Simple pendulum at bottom: $U''>0$ → stable with small-angle oscillations.
- Inverted pendulum: $U''<0$ → unstable; linearized solutions grow exponentially.

#### Multiple degrees of freedom: the Hessian and normal modes

For $n$ coordinates $q$, with kinetic energy $T=\tfrac{1}{2}\dot{q}^T M \dot{q}$ (mass matrix $M$ symmetric positive definite) and potential $U(q)$, expand near $q_0$:
$$
U(q_0+\xi) \approx U(q_0) + \tfrac{1}{2}\,\xi^T K\,\xi,\qquad
K_{ij} = \left.\frac{\partial^2 U}{\partial q_i \partial q_j}\right|_{q_0}.
$$
Linearized equations:
$$
M\,\ddot{\xi} + K\,\xi = 0.
$$
Seek normal modes $\xi(t)=\Re\{\mathbf{A}\,e^{i\omega t}\}$, giving the eigenvalue problem
$$
(K - \omega^2 M)\,\mathbf{A}=0.
$$
Stability criterion (conservative, linearized):
- If $K$ is positive definite (all eigenvalues $>0$), then all $\omega^2>0$ and the equilibrium is linearly stable; motion decomposes into oscillatory normal modes.
- If $K$ has any negative eigenvalue, there exists a direction with $\omega^2<0$ → exponential growth → unstable.
- If $K$ has a zero eigenvalue (semidefinite), at least one mode has $\omega=0$ → marginal/neutral direction; higher-order terms decide.

Interpretation:
- Positive curvature in every independent direction → bowl-shaped potential well → stable oscillations.
- A saddle in the potential (some directions up, some down) → instability along the “downhill” directions.

#### Role of damping and driving

With light damping $C\,\dot{\xi}$:
$$
M\ddot{\xi} + C\dot{\xi} + K\xi = 0,
$$
eigenvalues acquire negative real parts when $K$ is positive definite and $C$ is positive semidefinite → asymptotic stability (trajectories spiral into equilibrium). External periodic driving can induce resonance even near stable equilibria but does not change the static stability classification; it alters response amplitudes.

#### Energy (Lyapunov) viewpoint

For conservative systems, the total energy near a stable equilibrium is
$$
E(\xi,\dot{\xi}) \approx \tfrac{1}{2}\dot{\xi}^T M \dot{\xi} + \tfrac{1}{2}\xi^T K \xi.
$$
If $M$ and $K$ are positive definite, $E$ is a positive definite quadratic form and is conserved → trajectories remain on bounded energy ellipses around $q_0$ (Lyapunov stability). If $K$ is not positive definite, $E$ is not bounded below in some direction → instability.

#### Quick diagnostic checklist

- 1D: compute $U''(q_0)$. Sign decides (quadratics suffice unless $U''=0$).
- Multi-DOF: compute the Hessian $K$ at equilibrium and check its definiteness (eigenvalues or Sylvester’s criteria). With $M\succ 0$, all generalized eigenvalues $\omega^2$ are positive iff $K\succ 0$.
- If any $\omega^2<0$ in the small-oscillation eigenproblem $(K-\omega^2M)\mathbf{A}=0$, expect exponential growth in the corresponding mode → unstable.

#### Example summary

- Pendulum bottom: stable, $\omega^2=g/\ell$.
- Pendulum top: unstable, $\omega^2=-g/\ell$ (imaginary frequency).
- Coupled masses with identical springs about their symmetric equilibrium: $K\succ 0$ → two real $\omega$ → stable; superpositions produce beats but remain bounded.

#### Key takeaways

- Stability near equilibrium is a statement about the local shape of $U$: minima are stable, maxima unstable; in $n$D, require the Hessian to be positive definite.
- Small oscillations and stability are two sides of the same linearization: real normal frequencies $\omega$ signal stability; imaginary $\omega$ signal instability.
- Damping converts linear stability into asymptotic stability by dissipating energy; driving changes response but not the static stability classification.
