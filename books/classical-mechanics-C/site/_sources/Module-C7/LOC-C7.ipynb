## Motivation

Everything that can move can also oscillate. From the vibration of a guitar string to the motion of atoms in a crystal, from the swing of a pendulum to the tiny ripples in the spacetime fabric that we call gravitational waves — oscillations are everywhere. They are the **fundamental language of motion near equilibrium**.

When a system is displaced slightly from its equilibrium configuration, it tends to experience a restoring influence. This interplay between displacement and restoring force gives rise to periodic motion — motion that repeats in time. Understanding such motion allows us to describe **how systems respond to disturbances**, how energy is exchanged between potential and kinetic forms, and how interactions between parts of a system lead to collective behaviors like **beats, resonances, and normal modes**.

Oscillations are not limited to mechanics:
- **Electrical circuits** exhibit oscillations of charge and current.  
- **Molecules and solids** oscillate at quantized frequencies, forming the basis of spectroscopy and thermal properties.  
- **Biological systems** show rhythmic oscillations — heartbeat, circadian cycles, neural patterns.  
- Even **stability analysis** in engineering, climate modeling, or economics often begins by asking: *If the system is disturbed slightly, will it oscillate or diverge?*

In short, oscillations form a **universal framework** for understanding how systems behave near equilibrium. The same mathematical patterns — sines, cosines, exponentials — emerge across all of physics. Recognizing this unifying structure is one of the most important conceptual transitions in classical mechanics.

## Learning Objectives

After completing this module, students should be able to:
1. Explain what defines *small oscillations* and why linearization is valid near equilibrium.  
2. Derive and solve the equation of motion for a simple harmonic oscillator.  
3. Describe energy exchange between kinetic and potential forms during oscillatory motion.  
4. Extend the analysis to *coupled oscillators* and identify *normal modes* and *normal frequencies*.  
5. Apply the method of small oscillations to test the *stability* of an equilibrium configuration.  
6. Recognize oscillatory behavior in diverse systems — mechanical, electrical, or quantum — and describe the shared mathematical framework.

## Broader Perspective

In physics, learning about oscillations marks a turning point: we move from describing *motion* to describing *patterns of motion*. Once a system is expressed in terms of oscillatory modes, we can predict how it will respond to any disturbance. This same framework later becomes the foundation of **wave physics**, **quantum mechanics**, and even **statistical mechanics**.
