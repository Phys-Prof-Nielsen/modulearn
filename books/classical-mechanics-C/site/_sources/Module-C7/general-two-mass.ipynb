# C7.3 General Two-Mass System with Unequal Spring Constants
<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.1 Motivation

So far, we have studied two identical masses connected by identical springs.  
In that symmetric case, the motion naturally separated into two normal modes with clear physical meaning: **in-phase** and **out-of-phase** motion.  

But what happens when the system is not symmetric—when the coupling spring is stiffer or weaker than the outer springs, or the two masses differ?  
Such asymmetry breaks the simple mode structure, and direct analytical solutions become cumbersome.  
This motivates a more systematic, compact formulation of the equations of motion using **matrix notation**, which will naturally lead us toward the ideas of **mass matrices** and **stiffness matrices**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.2 Setup

Consider two masses connected by three springs like in previous section, with possibly different spring constants:

Let $x_1$ and $x_2$ denote the displacements of $m_1$ and $m_2$ from their respective equilibrium positions (positive to the right).  
At equilibrium, all springs are unstretched.

---

### Equations of motion

For each mass, we apply Newton’s second law by summing the spring forces acting on it.

- For mass 1:

  $$
  m_1 \ddot{x}_1 = -k_1 x_1 + k_2(x_2 - x_1)
  $$

  
  or equivalently,

  $$
  m_1 \ddot{x}_1 = - (k_1 + k_2) x_1 + k_2 x_2
  $$

- For mass 2:

  $$
  m_2 \ddot{x}_2 = -k_3 x_2 + k_2(x_1 - x_2)
  $$

  
  or equivalently,

  $$
  m_2 \ddot{x}_2 = - (k_2 + k_3) x_2 + k_2 x_1
  $$

---

### Compact matrix form

We can write the two coupled equations together as:

$$
\begin{pmatrix}
m_1 & 0 \\
0 & m_2
\end{pmatrix}
\begin{pmatrix}
\ddot{x}_1 \\[4pt] \ddot{x}_2
\end{pmatrix}
+
\begin{pmatrix}
k_1 + k_2 & -k_2 \\[4pt]
 -k_2 & k_2 + k_3
\end{pmatrix}
\begin{pmatrix}
x_1 \\[4pt] x_2
\end{pmatrix}
= 
\begin{pmatrix}
0 \\[4pt] 0
\end{pmatrix}.
$$

This form highlights two key objects:

- The **mass matrix**

  $$
  M = 
  \begin{pmatrix}
  m_1 & 0 \\
  0 & m_2
  \end{pmatrix}
  $$

  
  which collects all the inertial properties of the system.

- The **stiffness (spring constant) matrix**

  $$
  K =
  \begin{pmatrix}
  k_1 + k_2 & -k_2 \\[4pt]
  -k_2 & k_2 + k_3
  \end{pmatrix}
  $$

  
  which encodes how the restoring forces depend linearly on the displacements.

The system can then be expressed compactly as:

$$
M\,\ddot{\mathbf{x}} + K\,\mathbf{x} = 0,
$$

where

$$
\mathbf{x} =
\begin{pmatrix}
x_1 \\[4pt] x_2
\end{pmatrix}.
$$

## C7.3.3 Discussion

The matrix form

$$
M\,\ddot{\mathbf{x}} + K\,\mathbf{x} = 0
$$

provides a compact and general way to represent both symmetric and asymmetric systems.  
It also allows us to discuss the **strength of coupling** between the two oscillators in a clear and quantitative way.

---

### Symmetric reference case

When $m_1 = m_2 = m$ and $k_1 = k_3 = k$, the system simplifies to

$$
M = mI, \qquad
K =
\begin{pmatrix}
k + k_2 & -k_2 \\[4pt]
 -k_2 & k + k_2
\end{pmatrix}.
$$

The parameter $k_2$ — the **coupling spring constant** — controls how strongly the two masses influence one another.

---

### Weak coupling: $k_2 \ll k$

When the coupling spring is much weaker than the outer springs:

- Each mass oscillates primarily with its **own local spring**, almost independently.
- The two natural frequencies are only slightly different:

  $$
  \omega_1 \approx \sqrt{\tfrac{k}{m}}, \qquad
  \omega_2 \approx \sqrt{\tfrac{k + 2k_2}{m}}.
  $$
  
- Because the frequencies are close, the superposition of the two modes leads to **slow beats**:

  $$
  \Delta \omega = \omega_2 - \omega_1 \propto k_2.
  $$
  
- Physically, energy transfers slowly back and forth between the oscillators.

This is the **weak-coupling limit**, typical of coupled pendulums or weakly interacting mechanical resonators.

---

### Strong coupling: $k_2 \gtrsim k$

As the coupling spring becomes comparable to or stiffer than the outer springs:

- The two oscillators become **strongly correlated** — the motion of one immediately affects the other.
- The two frequencies separate widely:

  $$
  \omega_1 \approx \sqrt{\tfrac{k}{m}} \quad (\text{in-phase mode}), \qquad
  \omega_2 \approx \sqrt{\tfrac{k + 3k_2}{m}} \quad (\text{out-of-phase mode}).
  $$
  
- The **in-phase mode** involves nearly equal displacements $x_1 \approx x_2$ — the coupling spring stretches little, so the frequency remains near $\sqrt{k/m}$.
- The **out-of-phase mode** stretches the coupling spring maximally, increasing the effective restoring force and producing a much higher frequency.

In this regime, the oscillators behave almost like a single rigid pair at high frequency and a single combined mass at low frequency.

---

### Asymmetry and generalization

If $m_1 \neq m_2$ or $k_1 \neq k_3$, the two oscillators no longer exchange energy symmetrically:
- The mode shapes $\mathbf{A}_1$ and $\mathbf{A}_2$ are no longer purely symmetric or antisymmetric.
- The eigenfrequencies are shifted differently for each side, depending on how inertia and stiffness are distributed.

Nonetheless, the matrix equation

$$
(K - \omega^2 M)\,\mathbf{A} = 0
$$

still governs the dynamics, and the eigenvalues of the generalized problem determine the system’s frequencies regardless of asymmetry.

---

### Summary of coupling strength effects

| Coupling regime | Typical feature | Frequency separation | Energy transfer |
|------------------|------------------|----------------------|-----------------|
| Weak coupling ($k_2 \ll k$) | Nearly independent oscillators | Small ($\omega_2 \approx \omega_1$) | Slow beats |
| Moderate coupling ($k_2 \sim k$) | Distinct in/out of phase modes | Noticeable | Regular energy exchange |
| Strong coupling ($k_2 \gg k$) | Nearly rigid pair motion | Large | Rapid, almost locked motion |

---

This framework not only clarifies weak vs. strong coupling but also sets up a **natural transition to the matrix eigenvalue problem**:

$$
\det(K - \omega^2 M) = 0,
$$

whose solutions give the normal mode frequencies for any combination of $k_i$ and $m_i$.

In the next section, we will rederive this same structure using the **Lagrangian formalism**,  
where the matrices $M$ and $K$ appear directly from the kinetic and potential energy expressions — a formulation that extends seamlessly to any number of degrees of freedom or coordinate system.

## C7.3.4 Physical interpretation

The two matrices have clear physical meaning:
- $M$ describes **how inertia is distributed** among the degrees of freedom.
- $K$ describes **how restoring forces couple** the coordinates.

The matrix formulation elegantly handles both symmetric and asymmetric cases and provides a unified structure that will generalize directly to larger systems.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C7.3.5 Transition

This formulation already anticipates what the **Lagrangian method** will give automatically:  
from the quadratic forms of kinetic and potential energy,

$$
T = \tfrac{1}{2}\dot{\mathbf{x}}^T M \dot{\mathbf{x}}, \qquad
U = \tfrac{1}{2}\mathbf{x}^T K \mathbf{x}.
$$

In the next section, we will derive $M$ and $K$ directly from the Lagrangian,  
which naturally generalizes to non-Cartesian coordinates and coordinate-dependent mass matrices.
