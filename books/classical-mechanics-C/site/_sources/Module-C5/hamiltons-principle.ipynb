# C5.6 Hamilton's Principle and Lagrange's Equations

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.6.1 Motivation

In the previous module, we developed the Euler–Lagrange equation,

$$
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
-
\frac{\partial F}{\partial y}
=
0,
$$

as a general condition for a function to make a functional stationary.

This result is mathematically powerful, but an important question remains:

> What functional does nature optimize in classical mechanics?

Surprisingly, the motion of a particle can be determined without directly applying Newton's laws. Instead, we define a special functional known as the **action** and require that it be stationary for the actual path taken by the system.

This idea is known as **Hamilton's Principle** and serves as the foundation of Lagrangian mechanics.

Rather than beginning with forces, we begin with energy.

Rather than asking how forces cause motion, we ask which path makes the action stationary.

Remarkably, this alternative viewpoint leads to equations that are completely equivalent to Newton's laws while often being easier to apply to complicated systems.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Historical Note</strong>

The variational principles developed by Euler and Lagrange were unified and expanded by William Rowan Hamilton during the nineteenth century.

Hamilton's formulation became one of the most influential frameworks in theoretical physics and ultimately played a major role in the development of quantum mechanics, field theory, and modern relativity.
</div>

In this module, we will define the action, introduce Hamilton's Principle, and show how the Euler–Lagrange equation naturally leads to the Lagrange equations of motion used throughout classical mechanics.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.6.2 The Action Functional

In the previous module, we encountered a general functional of the form

$$
J[y]
=
\int_a^b F(y,y',x)\,dx.
$$

The Euler–Lagrange equation tells us how to determine the function that makes such a functional stationary.

For classical mechanics, the corresponding functional is called the **action**.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Action</strong>

The action is a functional defined by

$$
S[q]
=
\int_{t_1}^{t_2}
L(q,\dot q,t)\,dt,
$$

where:

<ul>
<li>$q$ represents a generalized coordinate,</li>
<li>$\dot q$ represents the corresponding generalized velocity,</li>
<li>$L$ is the Lagrangian,</li>
<li>$t_1$ and $t_2$ are the initial and final times.</li>
</ul>

The action assigns a single number to an entire trajectory $q(t)$.
</div>

Notice the similarities between the action and the functional studied previously:

$$
J[y]
=
\int_a^b
F(y,y',x)\,dx
$$

and

$$
S[q]
=
\int_{t_1}^{t_2}
L(q,\dot q,t)\,dt.
$$

The correspondence is

$$
x
\longrightarrow
t,
$$

$$
y
\longrightarrow
q,
$$

$$
y'
\longrightarrow
\dot q,
$$

$$
F
\longrightarrow
L.
$$

The action is therefore a special type of functional whose integrand is the Lagrangian.

---

The action depends on the entire path followed by the system.

For example, consider a particle moving from point A to point B.

Many different trajectories are possible:

$$
q_1(t),
\qquad
q_2(t),
\qquad
q_3(t),
\qquad
\ldots
$$

Each trajectory produces a different value of the action:

$$
S[q_1],
\qquad
S[q_2],
\qquad
S[q_3],
\qquad
\ldots
$$

Nature somehow selects one of these paths.

The question is:

> Which path does nature choose?

The answer is provided by Hamilton's Principle.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Important Observation</strong>

Newtonian mechanics determines motion by solving differential equations involving forces.

Hamiltonian and Lagrangian mechanics determine motion by identifying which trajectory makes the action stationary.

Although these viewpoints appear very different, they ultimately produce the same physical predictions.
</div>

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C5.6.3 Hamilton's Principle

Now that we have defined the action, we are ready to state one of the most important principles in classical mechanics.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Hamilton's Principle</strong>

The actual path followed by a mechanical system between two fixed times makes the action stationary.

Mathematically,

$$
\delta S = 0.
$$

Equivalently,

$$
\delta
\int_{t_1}^{t_2}
L(q,\dot q,t)\,dt
=
0.
$$

</div>

This statement is sometimes called the **Principle of Stationary Action**.

The symbol

$$
\delta
$$

represents a variation, meaning a small change in the trajectory of the system.

Just as we varied a function

$$
y(x)
\rightarrow
y(x)+\epsilon\eta(x)
$$

in the previous module, we now vary a trajectory

$$
q(t)
\rightarrow
q(t)+\epsilon\eta(t).
$$

The actual motion of the system is the trajectory for which the action remains stationary under all allowable variations.

---

Historically, Hamilton's Principle was often called the **Principle of Least Action**.

Although this name is still widely used, it can sometimes be misleading.

<div style="background-color:#ffe6e6; border-left:6px solid #cc0000; padding:14px; border-radius:4px;">
<strong>Common Misconception</strong><br>

Hamilton's Principle does not necessarily require the action to be a minimum.

The action may correspond to:

<ul>
<li>a minimum,</li>
<li>a maximum, or</li>
<li>a stationary point that is neither.</li>
</ul>

For this reason, the term <strong>stationary action</strong> is generally more accurate than <strong>least action</strong>.
</div>

---

The remarkable consequence of Hamilton's Principle is that the equations of motion of classical mechanics emerge automatically from the requirement

$$
\delta S = 0.
$$

We already derived the general Euler–Lagrange equation in the previous module.

The next step is simply to apply that result to the action functional.

Doing so will lead directly to the Lagrange equations of motion.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Why This Is Remarkable</strong>

Newton's laws describe motion in terms of forces and accelerations.

Hamilton's Principle describes motion in terms of an optimization principle involving energy.

Although these viewpoints appear fundamentally different, they lead to exactly the same physical predictions.

Many physicists regard Hamilton's Principle as one of the deepest and most elegant principles in all of classical physics.
</div>

## C5.6.4 Why is the Lagrangian $L=T-U$?

In the previous section, we introduced the action

$$
S[q]
=
\int_{t_1}^{t_2}
L(q,\dot q,t)\,dt,
$$

and stated that the actual motion of a system makes the action stationary.

A natural question immediately arises:

> Why do we choose the Lagrangian to be
>
$$
L=T-U
$$
>
> rather than some other combination of physical quantities?

Historically, this form was not guessed arbitrarily. Physicists discovered that when the action is constructed using

$$
L=T-U,
$$

Hamilton's Principle reproduces the equations of motion obtained from Newton's laws.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Lagrangian</strong>

For many classical mechanical systems, the Lagrangian is defined as

$$
L=T-U,
$$

where

$$
T
=
\text{kinetic energy}
$$

and

$$
U
=
\text{potential energy}.
$$

The action is therefore

$$
S
=
\int_{t_1}^{t_2}
(T-U)\,dt.
$$

</div>

At first glance, this choice may seem unusual.

Why subtract the potential energy?

Why not use

$$
T+U,
$$

or perhaps

$$
T^2-U,
$$

or some entirely different expression?

The answer is that the difference

$$
T-U
$$

captures a balance between motion and stored energy.

Large kinetic energy tends to increase the Lagrangian, while large potential energy tends to decrease it. The actual trajectory of the system is the one that makes the action stationary when these competing effects are considered over the entire motion.

Although this idea may seem abstract, the remarkable result is that it reproduces the familiar laws of mechanics.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>An Important Observation</strong>

The quantity

$$
T+U
$$

is the total mechanical energy.

The Lagrangian is not the total energy.

Instead,

$$
L=T-U.
$$

The fact that the difference of kinetic and potential energy generates the correct equations of motion is one of the most remarkable discoveries in classical mechanics.
</div>

---

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>Can the Lagrangian Be Derived?</strong>

The form

$$
L=T-U
$$

is usually introduced as a postulate of classical mechanics because it correctly reproduces Newton's laws when combined with Hamilton's Principle.

However, more advanced treatments show that the possible forms of the Lagrangian are strongly constrained by fundamental physical principles and symmetries. For example, requirements such as spatial homogeneity, spatial isotropy, and Galilean invariance restrict the mathematical form that the Lagrangian can take.

In modern physics, symmetry considerations often play a central role in constructing Lagrangians. This idea culminates in Noether's Theorem, which establishes a profound connection between symmetries and conservation laws such as the conservation of energy, momentum, and angular momentum.
</div>

In the next section, we will substitute

$$
L=T-U
$$

into the action and apply the Euler–Lagrange equation. Doing so will lead directly to the Lagrange equations of motion.

## C5.6.5 From the Euler–Lagrange Equation to Lagrange's Equation

In Module C5.4, we derived the Euler–Lagrange equation for a general functional

$$
J[y]
=
\int_a^b F(y,y',x)\,dx.
$$

The resulting condition for a stationary functional was

$$
\frac{d}{dx}
\left(
\frac{\partial F}{\partial y'}
\right)
-
\frac{\partial F}{\partial y}
=
0.
$$

To apply this result to classical mechanics, we simply recognize that the action

$$
S[q]
=
\int_{t_1}^{t_2}
L(q,\dot q,t)\,dt
$$

has exactly the same mathematical structure.

The correspondence is

$$
x
\longrightarrow
t,
$$

$$
y
\longrightarrow
q,
$$

$$
y'
\longrightarrow
\dot q,
$$

$$
F
\longrightarrow
L.
$$

Substituting these quantities into the Euler–Lagrange equation immediately gives

$$
\boxed{
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot q}
\right)
-
\frac{\partial L}{\partial q}
=
0
}
$$

which is the fundamental equation of Lagrangian mechanics.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Lagrange's Equation</strong>

For a system described by a generalized coordinate $q$, the equation of motion is

$$
\boxed{
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot q}
\right)
-
\frac{\partial L}{\partial q}
=
0
}
$$

where

$$
L=T-U.
$$

</div>

The remarkable consequence of Hamilton's Principle is that Newton's laws can be replaced by this single equation.

Rather than beginning with forces, we begin with the action and require it to be stationary.

The resulting equations of motion are completely equivalent to those obtained using Newtonian mechanics.

<div style="background-color:#fff8e1; border-left:6px solid #ffcc00; padding:14px; border-radius:4px;">
<strong>A Different View of Mechanics</strong>

Newtonian mechanics begins with forces and derives motion.

Lagrangian mechanics begins with energy and derives motion.

Although the mathematical approaches are very different, both formulations describe the same physical world.
</div>

## C5.6.6 Lagrange's Equations for Multiple Coordinates

Many physical systems cannot be described by a single coordinate.

For example:

- A pendulum requires an angular coordinate $\theta$.
- A double pendulum requires two angular coordinates.
- A particle moving in three dimensions may require coordinates $(x,y,z)$.
- A rigid body may require several coordinates to specify its orientation.

In general, a mechanical system may require

$$
q_1,q_2,\ldots,q_n
$$

generalized coordinates.

The Lagrangian then becomes a function of all coordinates, velocities, and possibly time:

$$
L
=
L(q_1,q_2,\ldots,q_n,
\dot q_1,\dot q_2,\ldots,\dot q_n,
t).
$$

Since each generalized coordinate can vary independently, Hamilton's Principle must be satisfied for every coordinate.

The result is a separate Lagrange equation for each generalized coordinate.

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>Lagrange's Equations for Multiple Coordinates</strong>

For a system described by generalized coordinates

$$
q_1,q_2,\ldots,q_n,
$$

the equations of motion are

$$
\boxed{
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot q_i}
\right)
-
\frac{\partial L}{\partial q_i}
=
0
}
\qquad
(i=1,2,\ldots,n).
$$

There is one equation for each generalized coordinate.
</div>

For example, a system with two generalized coordinates

$$
q_1
\quad\text{and}\quad
q_2
$$

requires two equations:

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot q_1}
\right)
-
\frac{\partial L}{\partial q_1}
=
0,
$$

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot q_2}
\right)
-
\frac{\partial L}{\partial q_2}
=
0.
$$

These equations are often coupled, meaning that the motion of one coordinate influences the motion of the others.

---

The procedure for applying Lagrange's equations is straightforward:

<div style="background-color:#e6f0ff; border-left:6px solid #003366; padding:12px; border-radius:4px;">
<strong>General Procedure</strong>

1. Choose appropriate generalized coordinates.

2. Determine the kinetic energy

$$
T.
$$

3. Determine the potential energy

$$
U.
$$

4. Construct the Lagrangian

$$
L=T-U.
$$

5. Apply Lagrange's equation to each generalized coordinate.

6. Simplify the resulting equations of motion.

</div>

This procedure allows us to derive equations of motion without explicitly analyzing all of the individual forces acting on the system.

In the following examples, we will apply this procedure to several familiar mechanical systems.
