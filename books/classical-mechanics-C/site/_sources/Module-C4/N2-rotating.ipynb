# C4.2 Newton’s Second Law and Fictitious Forces

Now that you've seen how time derivatives of vectors behave in rotating frames, you're ready to apply this framework to Newton’s Second Law. In this section, you’ll see that **fictitious forces** are not arbitrary inventions—they emerge directly from the mathematics of rotation.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.2.1 Newton’s Second Law in an Inertial Frame

In an inertial frame (i.e. one that is not accelerating or rotating), Newton’s second law applies in its simplest form:

$$
\vec{F}_{\text{net}} = m \vec{a}_{\text{inertial}}
$$

This equation tells us that the net external force on a particle causes it to accelerate in proportion to its mass.

But what happens when you're in a **rotating reference frame**?

From the rotating observer’s point of view, the particle’s path often appears curved or accelerated even when no visible forces seem to be acting. This discrepancy leads us to rethink how Newton’s laws apply in such frames.

### Plugging in the Rotating Acceleration

From previous section, we know how to relate acceleration in a rotating frame to acceleration in the inertial frame:

$$
\vec{a}_{\text{inertial}} = \vec{a}_{\text{rot}} + 2 \vec{\Omega} \times \vec{v}_{\text{rot}} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})
$$

Substitute this into Newton’s law:

$$
\vec{F}_{\text{net}} = m \left( \vec{a}_{\text{rot}} + 2 \vec{\Omega} \times \vec{v}_{\text{rot}} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r}) \right)
$$

Now isolate $ \vec{a}_{\text{rot}} $:

<strong>Key Result — Newton's Second Law in a Rotating Frame</strong>

After accounting for the effects of rotation, Newton's second law can be written as

$$
m \vec{a}_{\text{rot}}
=
\vec{F}_{\text{net}}
+
\underbrace{
\left(- 2m \vec{\Omega} \times \vec{v}_{\text{rot}}\right)
}_{\text{Coriolis Force}}
+
\underbrace{
\left(- m \vec{\Omega} \times (\vec{\Omega} \times \vec{r})\right)
}_{\text{Centrifugal Force}}.
$$

This equation has the same form as Newton's second law in an inertial frame, provided we introduce the Coriolis and centrifugal forces. These forces are called <em>fictitious forces</em> because they do not arise from physical interactions between objects. Instead, they appear because the observer is analyzing motion from a rotating reference frame.

</div>

This is Newton’s second law **as seen from the rotating frame**. The right-hand side includes not just the real external forces (like gravity, tension, or contact forces), but also **two extra terms**. These extra terms are what we call **fictitious forces**.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.2.2 The Fictitious Forces

### Coriolis Force

$$
\vec{F}_{\text{Coriolis}} = -2m \vec{\Omega} \times \vec{v}_{\text{rot}}
$$

- Depends on the velocity of the particle *in the rotating frame*
- Always acts **perpendicular** to the velocity and to the axis of rotation
- Responsible for **deflection** of paths—such as wind currents on Earth or curved trajectories on a spinning disk
- **Direction reverses** depending on the direction of motion

### Centrifugal Force

$$
\vec{F}_{\text{centrifugal}} = -m \vec{\Omega} \times (\vec{\Omega} \times \vec{r})
$$

- Depends only on the **position** of the particle relative to the rotation axis
- Always points **outward**, away from the axis of rotation
- Explains why you feel "pushed outward" on a spinning ride, or why water is flung out of a washing machine
- Stronger the farther you are from the center, and the faster the system is rotating

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.2.3 Modified Newton’s Second Law in a Rotating Frame

Putting it all together, in the rotating frame, Newton’s second law becomes:

$$
m \vec{a}_{\text{rot}} = \vec{F}_{\text{real}} + \vec{F}_{\text{Coriolis}} + \vec{F}_{\text{centrifugal}}
$$

Where:

- $ \vec{F}_{\text{real}} $ includes all actual physical forces (gravity, friction, tension, etc.)
- $ \vec{F}_{\text{Coriolis}} $ and $ \vec{F}_{\text{centrifugal}} $ are fictitious forces added to make the law valid in the rotating frame

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.2.4 Important Notes on Fictitious Forces

- **They are not "fake" in effect**: You can feel them, measure them, and build machines that use them. But they don't arise from physical interactions with other objects.
- **They are frame-dependent**: They only appear when you're analyzing motion *from within a rotating frame*.
- **They preserve Newton’s laws**: By introducing these forces, we can use Newton’s second law just as we would in inertial frames—provided we account for the added effects of rotation.

### Physical Intuition

- The Coriolis force is what makes hurricanes spin and ocean currents bend.
- The centrifugal force is what pins you to the wall on a rotating ride or makes objects fly off a turntable.
- If you switched back to the inertial frame, **these forces disappear**—the curving or outward motion is explained by real forces and inertia alone.

<hr style="height:2px;border-width:0;color:gray;background-color:gray">

## C4.2.5 Summary

- Applying Newton’s second law in a rotating frame requires accounting for the non-inertial nature of the frame.
- Two additional fictitious forces—Coriolis and centrifugal—emerge naturally from the vector calculus of rotating frames.
- These forces are not physical interactions but are essential tools for analyzing motion from a rotating perspective.
